## De la teoría a la vida real: dos casos de un negocio de verdad 🏪

Hasta ahora vimos las "reglas del juego" (tipos de datos, gráficos). Ahora juguemos con dos casos como los que te encontrarías en tu primer trabajo de analista:

1. **"¿Quiénes son mis clientes?"** — un análisis demográfico.
2. **"¿Qué tan contentos están con mis productos?"** — un análisis de valoraciones (⭐).

En ambos casos usamos datos **inventados** (sintéticos) porque así aprendes sin necesitar acceso a una base de datos real — pero el proceso es idéntico al que harías con datos de verdad.

````python
# ============================================================
# Caso 1: "¿Quiénes son mis clientes?" — generamos datos de ejemplo
# ============================================================

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

plt.rcParams.update({'axes.spines.top': False, 'axes.spines.right': False})
np.random.seed(7)

n = 300
generos = np.random.choice(['Femenino', 'Masculino'], size=n, p=[0.55, 0.45])
segmento = np.random.choice(['Bronce', 'Plata', 'Oro'], size=n, p=[0.5, 0.35, 0.15])

df_clientes = pd.DataFrame({'genero': generos, 'segmento': segmento})

print("👥 300 clientes de ejemplo generados. Así se ven los primeros:")
print(df_clientes.head(6).to_string(index=False))
````

---
## Pregunta 1: ¿Cómo se distribuye el género de mis clientes? 🚻

Solo dos-tres categorías, sin orden entre ellas → es el caso perfecto para un **donut chart** (recuerda: torta/donut solo si hay pocas categorías).

````python
# ============================================================
# Donut chart de género — simple y directo
# ============================================================

conteo = df_clientes['genero'].value_counts()

fig, ax = plt.subplots(figsize=(5.5, 5.5))
wedges, texts, autotexts = ax.pie(
    conteo.values, labels=conteo.index, autopct='%1.0f%%',
    colors=['#3b82f6', '#10b981'], wedgeprops={'width': 0.5, 'edgecolor': 'white'},
    startangle=90
)
ax.set_title('¿Cómo se reparten mis clientes por género?', fontweight='bold')
plt.tight_layout()
plt.show()

print("👉 Con solo 2 categorías, el donut se lee perfecto de un vistazo.")
````

---
## Pregunta 2: ¿el segmento de cliente cambia algo? 💎

Aquí ya no es una sola variable — quiero cruzar género con segmento (Bronce/Plata/Oro, que SÍ tiene orden). La receta: **barras 100% apiladas**, porque me interesa la composición (%), no el total de clientes.

````python
# ============================================================
# Barras 100% apiladas: género x segmento
# ============================================================

df_clientes['segmento'] = pd.Categorical(df_clientes['segmento'], categories=['Bronce', 'Plata', 'Oro'], ordered=True)
tabla = pd.crosstab(df_clientes['genero'], df_clientes['segmento'], normalize='index') * 100

fig, ax = plt.subplots(figsize=(7, 4.5))
bottom = np.zeros(len(tabla))
colores_seg = ['#cd7f32', '#94a3b8', '#f59e0b']
for seg, color in zip(tabla.columns, colores_seg):
    ax.bar(tabla.index, tabla[seg], bottom=bottom, label=seg, color=color, edgecolor='white')
    bottom += tabla[seg].values
ax.set_title('Composición de Segmento por Género (100%)', fontweight='bold')
ax.set_ylabel('Proporción (%)')
ax.legend(title='Segmento')
plt.tight_layout()
plt.show()

print("👉 Cada barra suma 100%, así que comparas PROPORCIONES, no cantidades absolutas.")
````

## Caso 2: "¿Qué tan contentos están mis clientes?" ⭐

Ahora simulamos reseñas de producto (1 a 5 estrellas). Recuerda la regla de oro: **las estrellas son ordinales — jamás se reordenan por frecuencia.**

````python
# ============================================================
# Caso 2: generamos reseñas de producto de ejemplo
# ============================================================

np.random.seed(3)
n_resenas = 400
estrellas = np.random.choice([1,2,3,4,5], size=n_resenas, p=[0.05, 0.08, 0.17, 0.35, 0.35])
df_resenas = pd.DataFrame({'estrellas': pd.Categorical(estrellas, categories=[1,2,3,4,5], ordered=True)})

conteo_estrellas = df_resenas['estrellas'].value_counts().sort_index()

fig, ax = plt.subplots(figsize=(7, 4.5))
colores_est = ['#ef4444', '#f97316', '#f59e0b', '#84cc16', '#10b981']
bars = ax.bar(conteo_estrellas.index.astype(str) + ' ⭐', conteo_estrellas.values, color=colores_est, edgecolor='white')
for bar, val in zip(bars, conteo_estrellas.values):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 3, str(val), ha='center', fontweight='bold')
ax.set_title('¿Cuántas reseñas de cada calificación tengo?', fontweight='bold')
ax.set_ylabel('Número de reseñas')
plt.tight_layout()
plt.show()

promedio = df_resenas['estrellas'].astype(int).mean()
print(f"⭐ Promedio general: {promedio:.2f} / 5.0")
print("👉 Fíjate que el gráfico va de 1⭐ a 5⭐ en orden — así se debe leer siempre.")
````
