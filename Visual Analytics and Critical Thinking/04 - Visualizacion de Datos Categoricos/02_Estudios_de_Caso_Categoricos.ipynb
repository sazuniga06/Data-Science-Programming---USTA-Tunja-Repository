## 1. Caso 1: Perfil Demográfico de Clientes 👥

**Contexto de negocio:** el equipo de marketing de una tienda en línea necesita entender la composición de su base de clientes para diseñar campañas segmentadas. Se cuenta con un extracto del CRM con 4 variables categóricas: género (nominal), rango de edad (intervalo/bins), ciudad (nominal) y segmento de cliente (ordinal: Bronce, Plata, Oro, Platino).

````python
# ============================================================
# Caso 1 — Generación del dataset sintético de clientes
# ============================================================

import matplotlib.pyplot as plt
import plotly.express as px
import pandas as pd
import numpy as np

plt.rcParams.update({
    'font.family': 'sans-serif',
    'axes.spines.top': False,
    'axes.spines.right': False,
    'axes.grid': True,
    'grid.alpha': 0.3
})

np.random.seed(7)
n_clientes = 600

generos = np.random.choice(['Femenino', 'Masculino', 'No binario'], size=n_clientes, p=[0.52, 0.44, 0.04])
rangos_edad = np.random.choice(['18-25', '26-35', '36-50', '51+'], size=n_clientes, p=[0.22, 0.38, 0.28, 0.12])
ciudades = np.random.choice(['Tunja', 'Bogotá', 'Duitama', 'Sogamoso', 'Chiquinquirá'],
                             size=n_clientes, p=[0.30, 0.35, 0.15, 0.12, 0.08])
segmento = np.random.choice(['Bronce', 'Plata', 'Oro', 'Platino'], size=n_clientes, p=[0.45, 0.30, 0.18, 0.07])

df_clientes = pd.DataFrame({
    'cliente_id': range(1, n_clientes + 1),
    'genero': generos,
    'rango_edad': pd.Categorical(rangos_edad, categories=['18-25','26-35','36-50','51+'], ordered=True),
    'ciudad': ciudades,
    'segmento': pd.Categorical(segmento, categories=['Bronce','Plata','Oro','Platino'], ordered=True)
})

print("👥 Dataset sintético de clientes generado")
print(f"   Filas: {df_clientes.shape[0]} | Columnas: {df_clientes.shape[1]}")
df_clientes.head(8)
````

````python
# ============================================================
# Técnica 1: variable NOMINAL con pocas categorías → donut
# Pregunta: ¿cómo se distribuye el género de la base de clientes?
# ============================================================

conteo_genero = df_clientes['genero'].value_counts()

fig = px.pie(
    values=conteo_genero.values, names=conteo_genero.index,
    hole=0.55, title='Distribución de Clientes por Género',
    color_discrete_sequence=['#3b82f6', '#10b981', '#f59e0b'],
    template='plotly_white'
)
fig.update_traces(textinfo='label+percent', textposition='outside')
fig.update_layout(height=420, showlegend=False)
fig.show()

print(conteo_genero)
````

````python
# ============================================================
# Técnica 2: variable de INTERVALO (bins) × ORDINAL → barras 100% apiladas
# Pregunta: ¿cómo se distribuye el segmento de cliente dentro de cada rango de edad?
# ============================================================

tabla_cruzada = pd.crosstab(df_clientes['rango_edad'], df_clientes['segmento'], normalize='index') * 100
tabla_cruzada = tabla_cruzada[['Bronce', 'Plata', 'Oro', 'Platino']]

colores_segmento = ['#cd7f32', '#94a3b8', '#f59e0b', '#8b5cf6']

fig, ax = plt.subplots(figsize=(8, 4.8))
bottom = np.zeros(len(tabla_cruzada))
for seg, color in zip(tabla_cruzada.columns, colores_segmento):
    ax.bar(tabla_cruzada.index.astype(str), tabla_cruzada[seg], bottom=bottom,
           label=seg, color=color, edgecolor='white', linewidth=0.5)
    bottom += tabla_cruzada[seg].values
ax.set_title('Composición de Segmento de Cliente por Rango de Edad (100%)', fontweight='bold')
ax.set_ylabel('Proporción (%)')
ax.set_xlabel('Rango de edad')
ax.legend(title='Segmento', fontsize=8, bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()

print("💡 Hallazgo: observe en qué rango de edad se concentra el segmento 'Platino'.")
````

````python
# ============================================================
# Técnica 3: variable NOMINAL con varias categorías → barras horizontales ordenadas
# Pregunta: ¿en qué ciudades se concentra la base de clientes?
# ============================================================

conteo_ciudad = df_clientes['ciudad'].value_counts().sort_values()

fig, ax = plt.subplots(figsize=(8, 4.5))
colores_ciudad = ['#06b6d4', '#8b5cf6', '#ef4444', '#10b981', '#3b82f6']
bars = ax.barh(conteo_ciudad.index, conteo_ciudad.values, color=colores_ciudad, edgecolor='white')
for bar, val in zip(bars, conteo_ciudad.values):
    ax.text(bar.get_width() + 3, bar.get_y() + bar.get_height()/2, str(val), va='center', fontsize=9)
ax.set_title('Clientes por Ciudad', fontweight='bold')
ax.set_xlabel('Número de clientes')
plt.tight_layout()
plt.show()

print("💡 Hallazgo: Bogotá y Tunja concentran la mayoría de la base — foco natural de campañas locales.")
````

## 2. Caso 2: Valoraciones de Productos ⭐

**Contexto de negocio:** el equipo de producto quiere entender cómo se distribuyen las valoraciones (1 a 5 estrellas, escala **ordinal**) que los clientes dejan en el marketplace, y si esa distribución varía según la categoría de producto.

````python
# ============================================================
# Caso 2 — Generación del dataset sintético de valoraciones
# ============================================================

np.random.seed(11)
n_resenas = 1000

categorias_producto = ['Electrónica', 'Ropa', 'Hogar', 'Deportes']

# Simulamos distribuciones distintas de estrellas por categoría
probs_por_categoria = {
    'Electrónica': [0.05, 0.08, 0.15, 0.32, 0.40],
    'Ropa':        [0.03, 0.07, 0.20, 0.40, 0.30],
    'Hogar':       [0.08, 0.12, 0.22, 0.30, 0.28],
    'Deportes':    [0.04, 0.06, 0.15, 0.35, 0.40],
}

filas = []
cat_asignadas = np.random.choice(categorias_producto, size=n_resenas, p=[0.30, 0.28, 0.24, 0.18])
for cat in cat_asignadas:
    estrellas = np.random.choice([1, 2, 3, 4, 5], p=probs_por_categoria[cat])
    filas.append({'categoria_producto': cat, 'estrellas': estrellas})

df_resenas = pd.DataFrame(filas)
df_resenas['estrellas'] = pd.Categorical(df_resenas['estrellas'], categories=[1,2,3,4,5], ordered=True)

print("⭐ Dataset sintético de valoraciones generado")
print(f"   Filas: {df_resenas.shape[0]}")
df_resenas.head(8)
````

````python
# ============================================================
# Técnica 1: variable ORDINAL pura → barras ordenadas (NUNCA por frecuencia)
# Pregunta: ¿cómo se distribuyen las valoraciones en general?
# ============================================================

conteo_estrellas = df_resenas['estrellas'].value_counts().sort_index()

fig, ax = plt.subplots(figsize=(7.5, 4.5))
colores_estrellas = ['#ef4444', '#f97316', '#f59e0b', '#84cc16', '#10b981']
bars = ax.bar(conteo_estrellas.index.astype(str) + ' ⭐', conteo_estrellas.values,
              color=colores_estrellas, edgecolor='white')
for bar, val in zip(bars, conteo_estrellas.values):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 5, str(val), ha='center', fontweight='bold')
ax.set_title('Distribución General de Valoraciones (orden 1→5 obligatorio)', fontweight='bold')
ax.set_ylabel('Número de reseñas')
plt.tight_layout()
plt.show()

promedio = (df_resenas['estrellas'].astype(int)).mean()
print(f"⭐ Calificación promedio: {promedio:.2f} / 5.0")
````

````python
# ============================================================
# Técnica 2: ORDINAL × NOMINAL → barras 100% apiladas por categoría de producto
# Pregunta: ¿la satisfacción varía según la categoría de producto?
# ============================================================

tabla_resenas = pd.crosstab(df_resenas['categoria_producto'], df_resenas['estrellas'], normalize='index') * 100

fig, ax = plt.subplots(figsize=(8.5, 4.8))
bottom = np.zeros(len(tabla_resenas))
for estrella, color in zip(tabla_resenas.columns, colores_estrellas):
    ax.bar(tabla_resenas.index, tabla_resenas[estrella], bottom=bottom,
           label=f'{estrella} ⭐', color=color, edgecolor='white', linewidth=0.5)
    bottom += tabla_resenas[estrella].values
ax.set_title('Composición de Valoraciones por Categoría de Producto (100%)', fontweight='bold')
ax.set_ylabel('Proporción (%)')
ax.legend(title='Estrellas', fontsize=8, bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()

print("💡 Hallazgo: compare el color verde (5⭐) entre categorías — identifique la categoría con más reseñas negativas.")
````

## 3. Interpretación y Hallazgos Accionables 💡

| Caso | Técnica usada | Hallazgo típico | Acción de negocio sugerida |
|---|---|---|---|
| Demografía — género | Donut | Base mayoritariamente femenina | Ajustar tono de campañas de marketing |
| Demografía — edad × segmento | Barras 100% apiladas | Clientes "Platino" concentrados en 36-50 | Programas de fidelización dirigidos a ese rango |
| Demografía — ciudad | Barras horizontales | Bogotá y Tunja concentran clientes | Priorizar logística/eventos en esas ciudades |
| Reseñas — general | Barras ordinales | Mayoría de reseñas en 4-5 estrellas | Usar reseñas altas en testimonios de marketing |
| Reseñas — por categoría | Barras 100% apiladas | Una categoría con más 1-2 estrellas | Investigar calidad/logística de esa categoría específica |

Este es el patrón general que debe seguir todo estudio de caso con datos categóricos: **identificar la escala de cada variable → elegir la técnica correcta (Notebook 01) → visualizar → traducir el patrón visual en una decisión de negocio.**

## Resumen y Puntos Clave 🎯

- Un estudio de caso categórico real casi siempre combina **variables de distinta escala** (nominal, ordinal, intervalo) en el mismo dataset.
- El **cruce de dos variables categóricas** (por ejemplo, segmento × rango de edad, o estrellas × categoría de producto) casi siempre se visualiza mejor con **barras 100% apiladas**.
- Generar datos sintéticos con `numpy`/`pandas` es una práctica habitual para prototipar dashboards antes de tener acceso a datos reales de producción.
- El valor de una visualización categórica está en la **conclusión accionable** que permite extraer, no solo en el gráfico en sí mismo.
