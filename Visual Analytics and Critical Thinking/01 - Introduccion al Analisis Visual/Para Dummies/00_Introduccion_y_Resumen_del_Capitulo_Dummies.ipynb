## 1. Dos superpoderes que todo científico de datos necesita 🦸

Imagina que tienes dos superpoderes disponibles, pero solo puedes desarrollar uno:

- **Superpoder A:** puedes construir el modelo más preciso del mundo.
- **Superpoder B:** puedes explicarlo tan claro que hasta tu abuela entendería la decisión que hay que tomar, y además puedes detectar cuando alguien más te está mostrando datos engañosos.

Sin el Superpoder B, el Superpoder A **no sirve de nada** en el mundo real: nadie va a confiar en un modelo que no entiende, y tú mismo podrías caer en trampas visuales sin darte cuenta.

> 🧠 **En criollo:** este capítulo te enseña a "mostrar bien" (análisis visual) y a "no dejarte engañar" (pensamiento crítico). Son como los ojos y el filtro anti-engaños de un científico de datos.

````python
# Veamos en 10 segundos por que "como" se muestra un dato cambia la historia
import matplotlib.pyplot as plt

trimestres = ['T1', 'T2', 'T3', 'T4']
ingresos = [102, 104, 103, 106]  # el cambio real es chiquito: 4%

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].bar(trimestres, ingresos, color='#ef4444')
axes[0].set_ylim(100, 107)
axes[0].set_title('¡Mira qué "subidón"!\n(truco: el eje empieza en 100)', fontsize=10)

axes[1].bar(trimestres, ingresos, color='#3b82f6')
axes[1].set_ylim(0, 120)
axes[1].set_title('La verdad: casi no cambió\n(eje empieza en 0)', fontsize=10)

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
print("Mismos numeros. Dos historias distintas. Bienvenido al capitulo 1.")
````
