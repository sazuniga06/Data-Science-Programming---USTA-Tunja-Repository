## 1. ¿Qué es la Analítica Visual? 👁️

La **Analítica Visual** (*Visual Analytics*) es la ciencia del razonamiento analítico apoyado por interfaces visuales interactivas. Combina visualización de información, minería de datos y técnicas de interacción humano-computador para facilitar la extracción de conocimiento de datasets complejos.

> 📌 **Definición formal (Thomas & Cook, 2005):** *"Visual analytics is the science of analytical reasoning facilitated by interactive visual interfaces."*

### La diferencia entre visualización y analítica visual:

| Aspecto | Visualización Estática | Analítica Visual |
|---|---|---|
| **Interacción** | Ninguna o mínima | Filtros, drill-down, zoom, brushing |
| **Escala de datos** | Datasets pequeños-medianos | Big Data, streams en tiempo real |
| **Objetivo** | Comunicar un resultado ya conocido | Descubrir patrones desconocidos |
| **Herramientas** | Matplotlib estático, Excel | Power BI, Tableau, Plotly Dash, Streamlit |
| **Rol del analista** | Observador pasivo | Co-explorador activo |

### El sistema visual humano como ventaja competitiva:
- El cerebro humano procesa **información visual 60,000 veces más rápido** que texto.
- El **90% de la información** transmitida al cerebro es visual.
- La **memoria visual** (*iconic memory*) puede retener ~12 objetos durante 200 ms — mucho más que la memoria de trabajo verbal (~7 ± 2 elementos).

## 3. El Pipeline Cognitivo: Datos → Decisión 🔄

El proceso de análisis visual sigue un pipeline cognitivo compuesto por cuatro etapas:

```
┌─────────┐    ┌──────────────┐    ┌─────────────────┐    ┌───────────┐
│  DATOS  │ →  │  PERCEPCIÓN  │ →  │  COMPRENSIÓN    │ →  │ DECISIÓN  │
│         │    │ (preattentive│    │ (working memory │    │ (acción)  │
│ Raw data│    │  attributes) │    │  + reasoning)   │    │           │
└─────────┘    └──────────────┘    └─────────────────┘    └───────────┘
```

### Atributos Preatentivos (Preattentive Attributes):
Son características visuales procesadas por el cerebro **antes** de la atención consciente (~200 ms). Explotarlos es la clave del diseño visual eficiente:

| Atributo | Ejemplo de uso | Potencia |
|---|---|---|
| **Color (matiz)** | Categorías en un mapa | ⭐⭐⭐⭐⭐ |
| **Longitud** | Barras comparativas | ⭐⭐⭐⭐⭐ |
| **Posición** | Scatterplot X/Y | ⭐⭐⭐⭐⭐ |
| **Tamaño** | Bubble chart (magnitud) | ⭐⭐⭐⭐ |
| **Forma** | Marcadores en scatter | ⭐⭐⭐ |
| **Orientación** | Líneas de tendencia | ⭐⭐⭐ |
| **Textura** | Mapas de densidad | ⭐⭐ |
| **Movimiento** | Animaciones temporales | ⭐⭐⭐⭐ |

````python
# ============================================================
# Demostracion de Atributos Preatentivos en Python
# Libreria: Matplotlib + NumPy
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
fig.suptitle('Atributos Preatentivos: El Cerebro los Detecta Instantáneamente',
             fontsize=14, fontweight='bold', y=1.02)

# --- Panel 1: Color como atributo preatentivo ---
ax1 = axes[0]
n = 50
x = np.random.rand(n)
y = np.random.rand(n)
colors = ['#94a3b8'] * n  # todos grises
colors[25] = '#dc2626'    # uno rojo - lo veras INSTANTANEAMENTE
sizes = [60] * n
ax1.scatter(x, y, c=colors, s=sizes, edgecolors='white', linewidths=0.5)
ax1.set_title('① Color', fontweight='bold', fontsize=12)
ax1.set_xticks([])
ax1.set_yticks([])
ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)
ax1.text(0.5, -0.08, '¿Cuánto tardaste en ver el punto rojo?',
         ha='center', transform=ax1.transAxes, fontsize=9, color='#64748b')

# --- Panel 2: Tamaño como atributo preatentivo ---
ax2 = axes[1]
sizes2 = [40] * n
sizes2[10] = 400  # uno grande
ax2.scatter(x, y, c='#3b82f6', s=sizes2, alpha=0.7, edgecolors='white', linewidths=0.5)
ax2.set_title('② Tamaño', fontweight='bold', fontsize=12)
ax2.set_xticks([])
ax2.set_yticks([])
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)
ax2.text(0.5, -0.08, '¿Y el círculo grande?',
         ha='center', transform=ax2.transAxes, fontsize=9, color='#64748b')

# --- Panel 3: Forma como atributo preatentivo ---
ax3 = axes[2]
markers_list = ['o'] * n
markers_list[35] = 's'  # un cuadrado entre circulos
for i, (xi, yi, mi) in enumerate(zip(x, y, markers_list)):
    ax3.scatter(xi, yi, c='#10b981', s=60, marker=mi,
                alpha=0.8, edgecolors='white', linewidths=0.5)
ax3.set_title('③ Forma', fontweight='bold', fontsize=12)
ax3.set_xticks([])
ax3.set_yticks([])
ax3.spines['top'].set_visible(False)
ax3.spines['right'].set_visible(False)
ax3.text(0.5, -0.08, '¿Encontraste el cuadrado?',
         ha='center', transform=ax3.transAxes, fontsize=9, color='#64748b')

plt.tight_layout()
plt.show()
print("✅ Color, tamaño y forma son detectados antes de que 'pienses' en ello.")
````

## 4. Principios de Gestalt Aplicados a la Visualización 🎨

Los **principios de Gestalt** (escuela psicológica alemana, 1920s) describen cómo el cerebro humano organiza los elementos visuales en grupos y patrones coherentes. Son fundamentales para el diseño de visualizaciones efectivas.

| Principio | Descripción | Aplicación en dataviz |
|---|---|---|
| **Proximidad** | Elementos cercanos se perciben como grupo | Agrupar columnas relacionadas en barras agrupadas |
| **Similitud** | Elementos similares (color, forma) parecen relacionados | Misma paleta para la misma categoría en todos los gráficos |
| **Cierre** | El cerebro completa formas incompletas | Gráficos de área con relleno parcial |
| **Continuidad** | Se siguen líneas y curvas naturalmente | Líneas de tendencia en series temporales |
| **Figura-Fondo** | Se distingue un objeto de su contexto | Resaltar un segmento en un gráfico de barras |
| **Región común** | Elementos en la misma región parecen agrupados | Paneles y facetas en dashboards |

> 📌 **Regla práctica:** Antes de usar colores, formas o posiciones diferentes, pregúntate si realmente necesitas marcar una diferencia conceptual. Cada distinción visual debe tener un significado analítico claro.

## 5. Ecosistema de Herramientas: Python ↔ Power BI ↔ Tableau 🔺

El ecosistema actual de analítica visual se divide en tres grandes pilares. Comprenderlos y saber cuándo usar cada uno es una competencia crítica del científico de datos moderno.

### Comparativa General

| Criterio | 🐍 Python | 📊 Power BI | 📈 Tableau |
|---|---|---|---|
| **Tipo** | Lenguaje de programación | Herramienta BI (Low-code) | Herramienta BI (Low-code) |
| **Curva de aprendizaje** | Alta | Media | Media-Baja |
| **Flexibilidad** | Ilimitada | Alta (con DAX/M) | Alta |
| **Interactividad** | Con Dash/Streamlit | Nativa | Nativa |
| **Integración datos** | Cualquier fuente (código) | Connectors + Power Query | Connectors + Data Prep |
| **Colaboración** | Git, notebooks compartidos | Power BI Service | Tableau Server/Public |
| **Costo** | Gratuito (open-source) | Pro: ~$10 USD/mes | Creator: ~$70 USD/mes |
| **Mejor para** | Automatización, ML, custom viz | Reportes empresariales | Exploración visual rápida |

### Ecosistema Python para Visualización

| Librería | Nivel | Uso principal | Interactividad |
|---|---|---|---|
| `matplotlib` | Bajo nivel | Gráficos estáticos customizables | No (base) |
| `seaborn` | Medio | EDA estadístico | No |
| `plotly` | Medio-Alto | Gráficos interactivos web | Sí |
| `plotly.express` | Alto | Gráficos rápidos una línea | Sí |
| `altair` | Alto | Grammar of Graphics en Python | Sí |
| `bokeh` | Medio-Alto | Dashboards web | Sí |
| `dash` | Alto | Apps analytics completas | Sí (React) |
| `streamlit` | Alto | Prototipos rápidos de dashboards | Sí |

````python
# ============================================================
# Primer grafico real: comparativa de ecosistemas
# Matplotlib basico -> base de todo lo que viene
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

# Datos simulados: popularidad de herramientas de viz (encuesta hipotetica)
herramientas = ['Python\n(Matplotlib)', 'Python\n(Plotly)', 'Power BI', 'Tableau', 'Excel']
uso_analistas = [72, 68, 85, 61, 78]       # % uso en analistas de datos
uso_ciencias   = [91, 84, 45, 38, 32]       # % uso en cientificos de datos

x = np.arange(len(herramientas))
width = 0.35

fig, ax = plt.subplots(figsize=(12, 6))

bars1 = ax.bar(x - width/2, uso_analistas, width,
               label='Analistas de Datos', color='#3b82f6', alpha=0.85, edgecolor='white')
bars2 = ax.bar(x + width/2, uso_ciencias, width,
               label='Científicos de Datos', color='#10b981', alpha=0.85, edgecolor='white')

def autolabel(bars):
    for bar in bars:
        height = bar.get_height()
        ax.annotate(f'{height}%',
                    xy=(bar.get_x() + bar.get_width() / 2, height),
                    xytext=(0, 4), textcoords='offset points',
                    ha='center', va='bottom', fontsize=9, fontweight='bold')

autolabel(bars1)
autolabel(bars2)

ax.set_xlabel('Herramienta de Visualización', fontsize=11)
ax.set_ylabel('% de Profesionales que la usan regularmente', fontsize=11)
ax.set_title('Ecosistema de Herramientas de Visualización\n(Encuesta hipotética de uso profesional 2024)',
             fontsize=13, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(herramientas, fontsize=10)
ax.legend(fontsize=10)
ax.set_ylim(0, 105)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.yaxis.grid(True, alpha=0.3)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()

print("\n📊 Equivalente en Power BI: Gráfico de barras agrupadas (clustered bar chart)")
print("   → Insertar Visual > Gráfico de barras agrupadas")
print("   → Eje: herramienta | Valores: % analistas, % científicos | Leyenda: rol")
print("\n📈 Equivalente en Tableau: Bar Chart con dimensión adicional")
print("   → Arrastrar 'Herramienta' a Columns, 'Porcentaje' a Rows")
print("   → Arrastrar 'Rol' a Color en la tarjeta Marks")
````

print(f"Mayor brecha: {herramienta_resaltada} ({brecha_ord[i_max]:+.0f} pp).")
