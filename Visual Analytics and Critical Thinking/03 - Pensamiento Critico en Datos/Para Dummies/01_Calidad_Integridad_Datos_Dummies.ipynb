## ¿Qué vamos a aprender aquí? 🧹

Este cuaderno es la versión **"para no ingenieros"** del segundo cuaderno del Módulo 03. Piensa en un dataset como en una caja de verduras que acabas de comprar en el mercado. Antes de cocinar (analizar), necesitas revisar cada verdura:

- ¿Falta alguna? (**nulos**)
- ¿Compraste la misma dos veces sin darte cuenta? (**duplicados**)
- ¿Están mezcladas manzanas con "manzanas " (con espacio) y "MANZANA" (mayúscula)? (**inconsistencia**)
- ¿Hay alguna podrida o rarísima que no debería estar ahí? (**valores atípicos**)

Al terminar sabrás hacer esta "revisión de verduras" con código en Python, sin tener que mirar fila por fila a mano.

---
## 1. Nulos: las verduras que faltan 🕳️

Un **nulo** es simplemente una celda vacía — un dato que debería estar ahí, pero no está. No es un error en sí mismo, pero necesitas SABER cuántos hay antes de seguir, porque pueden arruinar un cálculo (por ejemplo, un promedio) sin que te des cuenta.

````python
# Creamos una mini tabla de clientes con algunos datos faltantes a propósito

import pandas as pd
import numpy as np

clientes = pd.DataFrame({
    "nombre": ["Ana", "Luis", "Marta", "Pedro", "Sofía", "Iván"],
    "edad": [28, np.nan, 35, 41, np.nan, 30],
    "ciudad": ["Tunja", "Bogotá", "Cali", np.nan, "Tunja", "Medellín"],
})

print("📋 Nuestra mini tabla de clientes:")
display(clientes)

print("\n🕳️ ¿Cuántos huecos (nulos) hay por columna?")
print(clientes.isna().sum())
````

---
## 2. Duplicados: la verdura repetida sin querer 👯

A veces el mismo registro aparece **dos veces** — quizás alguien lo cargó por error dos veces al sistema. Si no lo detectas, vas a contar a "Ana" dos veces en tus estadísticas.

````python
# Agregamos un cliente duplicado a propósito

clientes_con_duplicado = pd.concat([clientes, clientes.iloc[[0]]], ignore_index=True)

print("👯 ¿Hay filas duplicadas?")
print(clientes_con_duplicado.duplicated())

print(f"\nTotal de filas duplicadas: {clientes_con_duplicado.duplicated().sum()}")
print("💡 Con .duplicated() y .drop_duplicates() puedes encontrarlas y eliminarlas en una línea.")
````

---
## 3. Inconsistencias: la misma cosa escrita de formas distintas 🔤

Este es el problema más traicionero, porque **no genera ningún error** — el programa no se queja. Simplemente cuenta "Bogotá", "bogota" y "BOGOTÁ " como si fueran tres ciudades diferentes.

````python
# El mismo problema con nombres de ciudades escritos de formas distintas

ciudades_sucias = pd.Series(["Bogotá", "bogota", "BOGOTÁ ", "Tunja", "tunja", "Bogotá"])

print("😵 ANTES de limpiar — parece que hay muchas ciudades distintas:")
print(ciudades_sucias.value_counts())

ciudades_limpias = ciudades_sucias.str.strip().str.title()

print("\n✅ DESPUÉS de limpiar (.strip().title()) — solo hay 2 ciudades reales:")
print(ciudades_limpias.value_counts())
````

---
## 4. Valores atípicos: la verdura que no debería estar ahí 🚨

Un **valor atípico** (outlier) es un dato que se sale muchísimo del resto. Puede ser un error de digitación (edad = 200 años) o un caso real pero extremo. La computadora no decide solita cuál es cuál — pero sí puede **señalarlo** para que tú lo revises.

````python
# Detectamos edades imposibles con una regla simple: fuera de un rango razonable

import matplotlib.pyplot as plt

edades = pd.Series([25, 30, 28, 45, 130, 33, -5, 29, 31])

# Regla simple: nadie tiene menos de 0 ni más de 110 años
raras = (edades < 0) | (edades > 110)

print("🚨 Edades detectadas como raras (probablemente errores):")
print(edades[raras])

plt.figure(figsize=(7, 4))
plt.boxplot(edades, vert=False, patch_artist=True,
            boxprops=dict(facecolor="#fef3c7", color="#b45309"),
            medianprops=dict(color="#dc2626"))
plt.title("Boxplot de edades — los puntos fuera de los 'bigotes' son sospechosos")
plt.xlabel("Edad")
plt.tight_layout()
plt.show()
````
