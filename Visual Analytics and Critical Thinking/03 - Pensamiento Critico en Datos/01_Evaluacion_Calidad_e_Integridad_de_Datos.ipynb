## 1.5.2 Mejores Prácticas para Evaluar la Calidad e Integridad de los Datos 🧭

Antes de construir cualquier visualización o modelo, un analista riguroso **audita el dato crudo**. La frase *"garbage in, garbage out"* resume el riesgo: ningún gráfico, por bien diseñado que esté, puede corregir datos defectuosos en su origen.

### Las cinco dimensiones clásicas de la calidad del dato:

| Dimensión | Pregunta que responde | Ejemplo de violación |
|---|---|---|
| **Completitud** | ¿Faltan valores? | Una columna de "edad" con 30% de celdas vacías |
| **Unicidad** | ¿Hay registros duplicados? | El mismo cliente registrado dos veces con IDs distintos |
| **Consistencia** | ¿Los tipos y formatos son uniformes? | Una columna de fechas con "2024-01-15" y "15/01/2024" mezcladas |
| **Validez** | ¿Los valores están dentro de rangos posibles? | Una edad de -5 o de 200 años |
| **Exactitud** | ¿El dato refleja la realidad? | Un ingreso reportado 100 veces mayor al real por un error de digitación |

> 📌 **Buena práctica:** la auditoría de calidad no es un paso opcional "si sobra tiempo" — es el primer paso obligatorio de cualquier pipeline de análisis, y debería automatizarse para poder repetirse cada vez que llegan datos nuevos.

## 2. Construcción de un Dataset Sintético "Sucio" 🧪

Para practicar, construimos un dataset sintético de clientes con problemas de calidad **inyectados deliberadamente**: nulos, duplicados, tipos inconsistentes y valores atípicos — igual que encontrarías en un CRM real sin mantenimiento.

````python
# ============================================================
# Generación de un dataset sintético "sucio" para auditoría
# ============================================================

import numpy as np
import pandas as pd

rng = np.random.default_rng(11)
n = 200

df_raw = pd.DataFrame({
    "cliente_id": [f"C{1000+i}" for i in range(n)],
    "edad": rng.integers(18, 75, size=n).astype(float),
    "ingreso_mensual": rng.normal(2_800_000, 650_000, size=n).round(0),
    "fecha_registro": pd.date_range("2022-01-01", periods=n, freq="3D").astype(str),
    "ciudad": rng.choice(["Bogotá", "Tunja", "Medellín", "Cali", "Bucaramanga"], size=n),
})

# --- Inyección deliberada de problemas de calidad ---

# 1. Nulos en 'edad' e 'ingreso_mensual'
idx_nulos = rng.choice(n, size=18, replace=False)
df_raw.loc[idx_nulos[:10], "edad"] = np.nan
df_raw.loc[idx_nulos[10:], "ingreso_mensual"] = np.nan

# 2. Duplicados exactos (5 clientes repetidos)
duplicados = df_raw.sample(5, random_state=1)
df_raw = pd.concat([df_raw, duplicados], ignore_index=True)

# 3. Inconsistencia de formato de fecha (mezclando "YYYY-MM-DD" con "DD/MM/YYYY")
idx_fecha_rara = rng.choice(df_raw.index, size=12, replace=False)
df_raw.loc[idx_fecha_rara, "fecha_registro"] = pd.to_datetime(
    df_raw.loc[idx_fecha_rara, "fecha_registro"]
).dt.strftime("%d/%m/%Y")

# 4. Valores atípicos (outliers) imposibles/extremos
df_raw.loc[df_raw.sample(4, random_state=2).index, "edad"] = [130, -5, 145, 0]
df_raw.loc[df_raw.sample(3, random_state=3).index, "ingreso_mensual"] = [55_000_000, -200_000, 48_000_000]

# 5. Inconsistencia de categorías (mismo valor, distinta escritura)
df_raw.loc[df_raw.sample(6, random_state=4).index, "ciudad"] = ["bogota", "BOGOTÁ", "Tunja ", "medellin", "Cali", " cali"]

print(f"Dataset sintético generado: {df_raw.shape[0]} filas × {df_raw.shape[1]} columnas")
df_raw.head(8)
````

## 3. Auditoría Automatizada: Detección de Nulos y Duplicados 🔍

El primer paso de cualquier auditoría es cuantificar, columna por columna, cuántos valores faltan y cuántas filas están duplicadas.

````python
# ============================================================
# Auditoría de completitud (nulos) y unicidad (duplicados)
# ============================================================

def auditar_nulos_y_duplicados(df):
    reporte = pd.DataFrame({
        "nulos": df.isna().sum(),
        "% nulos": (df.isna().mean() * 100).round(2),
        "dtype": df.dtypes.astype(str),
    })
    n_duplicados_exactos = df.duplicated(keep=False).sum()
    n_duplicados_por_id = df.duplicated(subset=["cliente_id"], keep=False).sum() if "cliente_id" in df.columns else None
    return reporte, n_duplicados_exactos, n_duplicados_por_id

reporte_nulos, dup_exactos, dup_por_id = auditar_nulos_y_duplicados(df_raw)

print("📋 Reporte de completitud por columna:")
display(reporte_nulos)

print(f"\n🔁 Filas completamente duplicadas: {dup_exactos}")
print(f"🔁 Filas con 'cliente_id' repetido: {dup_por_id}")
````

````python
# ============================================================
# Detección de inconsistencias de formato y de categorías
# ============================================================

# --- Formatos de fecha mezclados ---
formatos_detectados = df_raw["fecha_registro"].str.contains(r"^\d{4}-\d{2}-\d{2}$", regex=True)
print(f"Fechas en formato ISO (YYYY-MM-DD): {formatos_detectados.sum()}")
print(f"Fechas en OTRO formato (probablemente DD/MM/YYYY): {(~formatos_detectados).sum()}")

# --- Inconsistencias de categorías por espacios/mayúsculas ---
ciudades_crudas = df_raw["ciudad"].value_counts()
print("\n🏙️ Valores únicos de 'ciudad' ANTES de normalizar (nótese la fragmentación):")
print(ciudades_crudas)

ciudad_normalizada = df_raw["ciudad"].str.strip().str.title()
print("\n✅ Valores únicos de 'ciudad' DESPUÉS de normalizar (.strip().title()):")
print(ciudad_normalizada.value_counts())

print(f"\n💡 Tras .strip().title() quedan {ciudad_normalizada.nunique()} categorías (el dato crudo tenía {df_raw['ciudad'].nunique()}), "
      "pero las ciudades reales son 5: 'Bogota' y 'Medellin' siguen separadas de 'Bogotá' y 'Medellín' "
      "porque .title() corrige espacios y mayúsculas, no tildes. La limpieza aún no termina (ver Práctica 2).")
````

## 5. Detección de Valores Atípicos (Outliers) 📈

Un valor atípico puede ser un **error de captura** (edad = -5) o una **observación real pero extrema** (un cliente con ingresos excepcionalmente altos). El método de auditoría no decide automáticamente cuál es cuál — pero sí debe **señalarlos** para que un analista los revise.

Dos métodos complementarios:
- **Regla del IQR (rango intercuartílico):** marca como atípico todo valor fuera de `[Q1 - 1.5·IQR, Q3 + 1.5·IQR]`.
- **Z-score:** marca como atípico todo valor a más de 3 desviaciones estándar de la media.

````python
# ============================================================
# Detección de outliers: IQR y Z-score, con visualización
# ============================================================

import matplotlib.pyplot as plt

def detectar_outliers_iqr(serie):
    q1, q3 = serie.quantile(0.25), serie.quantile(0.75)
    iqr = q3 - q1
    limite_inf, limite_sup = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    return (serie < limite_inf) | (serie > limite_sup)

def detectar_outliers_zscore(serie, umbral=3):
    z = (serie - serie.mean()) / serie.std()
    return z.abs() > umbral

edad_valida = df_raw["edad"].dropna()
ingreso_valido = df_raw["ingreso_mensual"].dropna()

outliers_edad_iqr = detectar_outliers_iqr(edad_valida)
outliers_ingreso_iqr = detectar_outliers_iqr(ingreso_valido)

print(f"🚨 Outliers en 'edad' (regla IQR): {outliers_edad_iqr.sum()} de {len(edad_valida)}")
print(df_raw.loc[edad_valida[outliers_edad_iqr].index, ["cliente_id", "edad"]])

print(f"\n🚨 Outliers en 'ingreso_mensual' (regla IQR): {outliers_ingreso_iqr.sum()} de {len(ingreso_valido)}")
print(df_raw.loc[ingreso_valido[outliers_ingreso_iqr].index, ["cliente_id", "ingreso_mensual"]])

fig, axes = plt.subplots(1, 2, figsize=(11, 5))
axes[0].boxplot(edad_valida, vert=True, patch_artist=True,
                 boxprops=dict(facecolor="#dbeafe", color="#1e3a8a"),
                 medianprops=dict(color="#dc2626"))
axes[0].set_title("Boxplot: Edad\n(outliers fuera de los bigotes)")
axes[0].set_ylabel("Años")

axes[1].boxplot(ingreso_valido, vert=True, patch_artist=True,
                 boxprops=dict(facecolor="#dbeafe", color="#1e3a8a"),
                 medianprops=dict(color="#dc2626"))
axes[1].set_title("Boxplot: Ingreso Mensual\n(outliers fuera de los bigotes)")
axes[1].set_ylabel("COP")

plt.tight_layout()
plt.show()
````

## 6. Buenas Prácticas: Checklist de Integridad de Datos ✅

Aplica este checklist **antes** de construir cualquier visualización o modelo sobre un dataset nuevo:

- [ ] ¿Se cuantificaron los nulos por columna y se decidió una estrategia (eliminar, imputar, marcar)?
- [ ] ¿Se identificaron y resolvieron los registros duplicados (exactos y por clave de negocio)?
- [ ] ¿Los tipos de dato de cada columna corresponden a lo esperado (fechas como fecha, no como texto)?
- [ ] ¿Las categorías fueron normalizadas (mayúsculas, espacios, sinónimos)?
- [ ] ¿Se aplicó una regla objetiva (IQR o z-score) para señalar outliers, en vez de "a simple vista"?
- [ ] ¿Los outliers detectados fueron revisados individualmente antes de eliminarlos (podrían ser datos reales)?
- [ ] ¿El reporte de auditoría queda documentado y es reproducible (código, no solo una revisión manual)?

> 📌 **Regla de oro:** nunca elimines un outlier o un duplicado silenciosamente. Documenta cuántos registros se descartaron y por qué — la trazabilidad es parte de la integridad.

## 7. Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Completitud y unicidad** | Cuantifica nulos y duplicados columna por columna — nunca asumas que el dataset está limpio |
| **Consistencia** | Los problemas de formato (mayúsculas, fechas mezcladas) son invisibles hasta que se agrupan o filtran los datos |
| **Outliers** | Usa reglas objetivas (IQR, z-score) para señalarlos — pero la decisión de eliminarlos requiere juicio humano |
| **Automatización** | Una auditoría que se puede volver a correr con un clic es mucho más confiable que una revisión manual única |
| **Trazabilidad** | Documenta siempre qué se limpió y por qué — la limpieza de datos también debe ser auditable |
