## 1. ¿Qué es Power Query? ⚡

**Power Query** es el motor de preparación y transformación de datos integrado en:
- **Microsoft Power BI** (para carga y transformación de datos antes de visualizar)
- **Microsoft Excel** (pestaña "Datos" → "Obtener y transformar")
- **Azure Data Factory** (para ETL en la nube)
- **SQL Server Integration Services (SSIS)**

### El rol de Power Query en el pipeline de Power BI:

```
FUENTES DE DATOS          POWER QUERY (M)           POWER BI
┌──────────────┐         ┌──────────────────┐       ┌──────────────┐
│ Excel / CSV  │──────→  │  Extraer         │       │              │
│ SQL Server   │──────→  │  Transformar     │──────→│  Modelo      │
│ API REST     │──────→  │  Limpiar         │       │  DAX         │
│ SharePoint   │──────→  │  Combinar        │       │  Visuales    │
│ Web          │──────→  │  Cargar          │       │              │
└──────────────┘         └──────────────────┘       └──────────────┘
   Extracción               Transformación             Carga + Viz
```

### Power Query vs SQL vs Pandas:

| Aspecto | Power Query (M) | SQL | Pandas (Python) |
|---|---|---|---|
| **Paradigma** | Funcional (lazy evaluation) | Declarativo | Imperativo/Funcional |
| **Interfaz** | GUI + Editor avanzado | Query text | Código Python |
| **Escala** | Hasta ~1M filas bien | Millones/billones | Millones (RAM-bound) |
| **Curva aprendizaje** | Baja (GUI) | Media | Alta |
| **Reproducibilidad** | Script M en el archivo | Script SQL | Código Python en Git |
| **Integración ML** | No nativa | No nativa | Total (scikit-learn, etc.) |

## 2. El Lenguaje M: Estructura y Sintaxis 📝

Power Query usa el **lenguaje M** (Power Query Formula Language). A diferencia de DAX (que se usa en el modelo de datos), M se usa en la fase de transformación.

### Estructura de una consulta M:

```m
let
    // Paso 1: Conectar a la fuente
    Fuente = Csv.Document(File.Contents("C:\datos\ventas.csv"),
                           [Delimiter=",", Encoding=65001]),

    // Paso 2: Promover primera fila como encabezados
    Encabezados = Table.PromoteHeaders(Fuente, [PromoteAllScalars=true]),

    // Paso 3: Cambiar tipos de datos
    TiposCambiados = Table.TransformColumnTypes(Encabezados, {
        {"Fecha", type date},
        {"Ventas", type number},
        {"Categoria", type text}
    }),

    // Paso 4: Filtrar filas
    FiltradoPorAño = Table.SelectRows(TiposCambiados,
                                      each Date.Year([Fecha]) >= 2022),

    // Paso 5: Agregar columna calculada
    ConMargen = Table.AddColumn(FiltradoPorAño, "Margen",
                                each [Ganancia] / [Ventas])
in
    ConMargen  // El resultado final de la consulta
```

### Características clave del lenguaje M:

| Característica | Descripción | Analogía en Python |
|---|---|---|
| **Case-sensitive** | `Table` ≠ `table` | Igual en Python |
| **Evaluación perezosa** | Solo evalúa lo necesario | Similar a generadores |
| **Inmutable** | Cada paso crea una nueva tabla | Como operaciones Pandas encadenadas |
| **`let...in`** | Define pasos y devuelve el último | Similar a funciones Python |
| **`each _`** | Función anónima por fila | Similar a `lambda x: ...` |

````python
# ============================================================
# Equivalencias M → Pandas: Operaciones fundamentales
# Este cuaderno se ejecuta en Python para mostrar los paralelos
# ============================================================

import pandas as pd
import numpy as np

# Crear dataset de ejemplo (simulando una carga desde CSV)
np.random.seed(42)
n = 1000
datos = {
    'Fecha': pd.date_range('2022-01-01', periods=n, freq='D')[:n],
    'Categoria': np.random.choice(['Electrónica', 'Ropa', 'Hogar', 'Alimentos'], n),
    'Region': np.random.choice(['Norte', 'Sur', 'Este', 'Oeste'], n),
    'Ventas': np.random.exponential(500, n).round(2),
    'Ganancia': np.random.normal(100, 80, n).round(2),
    'Cantidad': np.random.randint(1, 20, n)
}
df = pd.DataFrame(datos)

print("📊 Dataset cargado (simulación de CSV)")
print(f"   Shape: {df.shape}")
print(f"   Tipos: {dict(df.dtypes)}")
print("\nVista previa:")
print(df.head(3).to_string())

print("\n" + "="*60)
print("Equivalencias M → Pandas")
print("="*60)

# M: Csv.Document(...)
# Pandas: pd.read_csv('ventas.csv')
print("\n1. CARGAR DATOS")
print("   M:      Csv.Document(File.Contents('ventas.csv'), ...)")
print("   Pandas: df = pd.read_csv('ventas.csv')")

# M: Table.PromoteHeaders
# Pandas: ya incluido en pd.read_csv con header=0
print("\n2. PROMOVER ENCABEZADOS")
print("   M:      Table.PromoteHeaders(Fuente, ...)")
print("   Pandas: pd.read_csv('archivo.csv', header=0)  # Es el comportamiento por defecto")

# M: Table.TransformColumnTypes
print("\n3. CAMBIAR TIPOS DE DATOS")
print("   M:      Table.TransformColumnTypes(Tabla, {{\"Fecha\", type date}, {\"Ventas\", type number}})")
print("   Pandas:")
df['Fecha'] = pd.to_datetime(df['Fecha'])
df['Ventas'] = df['Ventas'].astype(float)
print(f"           df['Fecha'] = pd.to_datetime(df['Fecha'])")
print(f"           df['Ventas'] = df['Ventas'].astype(float)")
print(f"   Resultado: {dict(df[['Fecha', 'Ventas']].dtypes)}")
````

````python
# ============================================================
# Operaciones de filtrado y agrupación: M vs Pandas
# ============================================================

print("="*60)
print("Filtrado y Agrupación: M vs Pandas")
print("="*60)

# M: Table.SelectRows
print("\n4. FILTRAR FILAS")
print("   M:      Table.SelectRows(Tabla, each [Ventas] > 500)")
print("   Pandas: df_fil = df[df['Ventas'] > 500]")
df_fil = df[df['Ventas'] > 500]
print(f"   Resultado: {len(df_fil):,} filas (de {len(df):,})")

# M: Table.SelectColumns
print("\n5. SELECCIONAR COLUMNAS")
print("   M:      Table.SelectColumns(Tabla, {\"Fecha\", \"Ventas\", \"Categoria\"})")
print("   Pandas: df_sel = df[['Fecha', 'Ventas', 'Categoria']]")
df_sel = df[['Fecha', 'Ventas', 'Categoria']]
print(f"   Resultado: {df_sel.shape}")

# M: Table.AddColumn
print("\n6. AGREGAR COLUMNA CALCULADA")
print("   M:      Table.AddColumn(Tabla, \"Margen\", each [Ganancia] / [Ventas])")
print("   Pandas: df['Margen'] = df['Ganancia'] / df['Ventas']")
df['Margen'] = df['Ganancia'] / df['Ventas']
print(f"   Margen promedio: {df['Margen'].mean():.3f}")

# M: Table.Group
print("\n7. AGRUPAR Y AGREGAR")
print("   M:      Table.Group(Tabla, {\"Categoria\"}, {{\"Total\", each List.Sum([Ventas])}})")
print("   Pandas: df.groupby('Categoria')['Ventas'].sum().reset_index()")
resumen = df.groupby('Categoria').agg(
    Total_Ventas=('Ventas', 'sum'),
    Promedio_Ventas=('Ventas', 'mean'),
    Conteo=('Ventas', 'count')
).round(2)
print(f"   Resultado:")
print(resumen.to_string())
````

## 3. El Editor Avanzado de Power Query 🔧

En Power BI Desktop, el **Editor Avanzado** permite escribir código M directamente. Es el equivalente al editor de código en Python.

### Cómo acceder:
1. Abrir Power BI Desktop
2. **Inicio → Transformar datos** (abre el Editor de Power Query)
3. **Vista → Editor avanzado** (muestra el código M completo)

### Funciones M más utilizadas:

| Función M | Equivalente Pandas | Descripción |
|---|---|---|
| `Table.SelectRows(t, each ...)` | `df[condición]` | Filtrar filas |
| `Table.SelectColumns(t, lista)` | `df[['col1', 'col2']]` | Seleccionar columnas |
| `Table.RemoveColumns(t, lista)` | `df.drop(columns=[...])` | Eliminar columnas |
| `Table.AddColumn(t, "nombre", each ...)` | `df['nueva'] = ...` | Agregar columna |
| `Table.RenameColumns(t, {{"viejo", "nuevo"}})` | `df.rename(columns={...})` | Renombrar columnas |
| `Table.Group(t, keys, agregaciones)` | `df.groupby(...).agg(...)` | Agrupar y agregar |
| `Table.Join(t1, k1, t2, k2, modo)` | `pd.merge(df1, df2, on=...)` | Combinar tablas |
| `Table.Pivot(t, cols, val, "attr", agg)` | `df.pivot_table(...)` | Pivotear tabla |
| `Table.Unpivot(t, cols, attr, val)` | `df.melt(...)` | Despivotar (melt) |
| `Table.Sort(t, {{col, Order.Ascending}})` | `df.sort_values(...)` | Ordenar |
| `Table.RemoveDuplicates(t, cols)` | `df.drop_duplicates(subset=[...])` | Eliminar duplicados |
| `Table.FillDown(t, cols)` | `df[col].fillna(method='ffill')` | Rellenar hacia abajo |
| `Table.TransformColumns(t, {{col, transformación}})` | `df[col].apply(func)` | Aplicar función a columna |

> 📌 **Regla práctica:** Cualquier transformación que puedas hacer con el GUI de Power Query, se puede expresar en código M. El GUI **genera el código M automáticamente** — puedes aprender M observando lo que genera el GUI.
