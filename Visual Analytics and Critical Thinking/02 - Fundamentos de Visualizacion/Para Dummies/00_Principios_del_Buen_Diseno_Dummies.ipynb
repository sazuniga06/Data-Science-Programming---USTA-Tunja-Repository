## 1. Simplicidad: la regla del clóset ordenado 🧹

Imagina que estás armando la maleta para un viaje. Si metes ropa que no vas a usar "por si acaso", la maleta pesa más y es más difícil encontrar lo que sí necesitas.

Un gráfico funciona igual: **cada elemento que agregas (colores, sombras, bordes, cuadrículas) es peso extra**. Si ese elemento no representa un dato, sobra.

| Elemento | ¿Lo necesitas? | Motivo |
|---|---|---|
| Fondo oscuro o gris | ❌ No | Distrae de los datos |
| Borde grueso en cada barra | ❌ No | No agrega información |
| Sombras 3D | ❌ No | Distorsionan las alturas |
| Leyenda cuando el eje ya explica | ❌ No | Duplicación de información |
| Colores al azar | ❌ No | El color debe significar algo |

> 🧠 **Regla del abuelo Tufte:** "Por encima de todo, muestra los datos." Si puedes quitar algo sin perder información, quítalo.

````python
# Antes vs Despues de "vaciar la maleta" del grafico

import matplotlib.pyplot as plt
import numpy as np

meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun']
ventas = [120, 145, 138, 172, 168, 195]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Mismo dato — ¿Cuál preferirías presentar en una reunión?',
             fontsize=12, fontweight='bold')

# ANTES: maleta sobrecargada
ax1.set_facecolor('#1a1a2e')
colores_antes = ['#ff6b6b', '#ffd93d', '#6bcb77', '#4d96ff', '#ff922b', '#cc5de8']
bars1 = ax1.bar(meses, ventas, color=colores_antes, edgecolor='white', linewidth=2)
ax1.yaxis.grid(True, color='white', linewidth=1.5, alpha=0.5)
ax1.tick_params(colors='white', labelsize=11)
ax1.set_title('❌ Con ruido visual', fontsize=11, color='white', fontweight='bold')
for spine in ax1.spines.values():
    spine.set_edgecolor('white')
for bar, val in zip(bars1, ventas):
    ax1.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 3,
             str(val), ha='center', color='white', fontweight='bold', fontsize=10)

# DESPUES: maleta liviana
colores_despues = ['#94a3b8'] * len(ventas)
colores_despues[-1] = '#1e3a8a'
bars2 = ax2.bar(meses, ventas, color=colores_despues, edgecolor='white')
ax2.set_title('✅ Solo lo necesario', fontsize=11, fontweight='bold', color='#1e3a8a')
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)
ax2.spines['left'].set_visible(False)
ax2.tick_params(left=False)
ax2.yaxis.grid(True, alpha=0.2, linewidth=0.8)
ax2.set_axisbelow(True)
ax2.set_ylabel('Ventas (miles $)', fontsize=10, color='#64748b')
ax2.annotate('🏆 Mejor mes', xy=(5, 195), xytext=(3.5, 185),
             fontsize=9, color='#1e3a8a', fontweight='bold',
             arrowprops=dict(arrowstyle='->', color='#1e3a8a'))

plt.tight_layout()
plt.show()

print("\n💡 Quitamos: fondo oscuro, colores al azar, bordes gruesos, cuadricula pesada.")
print("   Dejamos: solo lo que ayuda a entender el dato mas importante (junio).")
````

## 2. Claridad: por qué tu cerebro "agrupa" cosas solo 🧲

Cuando ves puntos en una hoja, tu cerebro automáticamente intenta agruparlos, aunque nadie te lo pida. Eso es la **Gestalt**: reglas de percepción que usamos sin darnos cuenta, como cuando ves nubes con forma de animales.

| Si el diseñador hace esto... | Tu cerebro entiende... | Ejemplo cotidiano |
|---|---|---|
| Pone cosas juntas (**proximidad**) | "Estas van en el mismo grupo" | Los productos de una misma góndola en el súper |
| Usa el mismo color (**similitud**) | "Estas son del mismo tipo" | Todas las señales de "salida" son verdes |
| Deja una línea a medio dibujar (**cierre**) | "Ya sé cómo termina" | El logo de WWF (panda con líneas incompletas) |
| Ordena las cosas en secuencia (**continuidad**) | "Esto sigue un orden lógico" | Los meses del calendario en fila |

> 🧠 Un buen diseñador de gráficos no "decora" — **organiza para que tu cerebro no tenga que trabajar de más**.

````python
# Prueba tu propio cerebro: proximidad y color en accion

import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(7)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
fig.suptitle('¿En cuál ves los grupos más rápido?', fontsize=12, fontweight='bold')

x_random = rng.uniform(0, 10, 24)
y_random = rng.uniform(0, 10, 24)
axes[0].scatter(x_random, y_random, s=90, color='#94a3b8', edgecolor='white')
axes[0].set_title('❌ Todo mezclado', fontsize=11)
axes[0].set_xticks([]); axes[0].set_yticks([])
for sp in axes[0].spines.values(): sp.set_visible(False)

centros = [(2, 2), (7, 3), (4.5, 8)]
colores = ['#1e3a8a', '#10b981', '#f59e0b']
for (cx, cy), color in zip(centros, colores):
    xs = cx + rng.normal(0, 0.6, 8)
    ys = cy + rng.normal(0, 0.6, 8)
    axes[1].scatter(xs, ys, s=90, color=color, edgecolor='white')
axes[1].set_title('✅ Agrupado + coloreado', fontsize=11)
axes[1].set_xticks([]); axes[1].set_yticks([])
for sp in axes[1].spines.values(): sp.set_visible(False)

plt.tight_layout()
plt.show()

print("\n🎯 En el panel derecho, en menos de 1 segundo ya sabes que hay 3 grupos.")
print("   En el izquierdo, tuviste que 'pensarlo'. Esa es la diferencia entre un")
print("   grafico que se lee solo y uno que necesita explicacion.")
````

## 3. Consistencia: el semáforo no cambia de colores 🚦

Imagina si el semáforo de tu barrio significara "alto" en rojo, pero el de la siguiente cuadra significara "alto" en amarillo. Sería un caos. Por eso el semáforo es **consistente en todo el mundo**.

Con los gráficos pasa igual: si "Región Norte" es azul en el primer gráfico de tu informe, **tiene que seguir siendo azul en el segundo, el tercero y el décimo**. Cambiar de color a mitad de camino confunde a quien te lee, aunque el dato sea correcto.

### Reglas simples para no equivocarte:
- ✅ **Categorías** → colores distintos (máximo 6-7), siempre los mismos en todo el reporte
- ✅ **"Más" vs "menos"** → un color más oscuro/claro, no un color distinto
- ❌ **Nunca** cambies de paleta entre gráficos del mismo informe
- ❌ **Nunca** uses rojo-verde si tienes audiencia mixta (8% de hombres son daltónicos)

```python
from matplotlib.colors import to_rgba

regiones = ['Norte', 'Sur']
ventas_reg = [230, 190]
meses = ['Ene', 'Feb', 'Mar', 'Abr']
tendencia = {'Norte': [50, 55, 60, 65], 'Sur': [40, 38, 45, 47]}
color_de = {'Norte': '#0072B2', 'Sur': '#E69F00'}

# 1) Dos paneles
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# 2) Barras con el color de cada region
ax1.bar(regiones, ventas_reg, color=[color_de[r] for r in regiones])

# 3) Una linea por region, con el MISMO color
for r in regiones:
    ax2.plot(meses, tendencia[r], color=color_de[r], label=r, linewidth=2.5)

# 4) Titulos y leyenda
ax1.set_title('Ventas totales', fontweight='bold')
ax2.set_title('Tendencia mensual', fontweight='bold')
ax2.legend()
plt.tight_layout()
plt.show()

# 5) Comprobaciones
color_linea = {l.get_label(): to_rgba(l.get_color()) for l in ax2.lines}
for r, barra in zip(regiones, ax1.patches):
    assert to_rgba(barra.get_facecolor()) == color_linea[r], r
print("Norte siempre azul, Sur siempre naranja: como el semáforo.")
```

## Resumen en una página 🎯

| Principio | En una frase |
|---|---|
| **Simplicidad** | Si un elemento no representa un dato, quítalo (vacía la maleta) |
| **Claridad** | Agrupa con cercanía y color para que el cerebro no tenga que esforzarse |
| **Consistencia** | El mismo color siempre significa lo mismo, en todo el reporte (como el semáforo) |

**Las 3 reglas de oro:**
1. 🎯 Menos elementos, más significado.
2. 🧲 Agrupa lo relacionado, sepáralo de lo que no lo está.
3. 🚦 Un color, un significado, siempre igual.
