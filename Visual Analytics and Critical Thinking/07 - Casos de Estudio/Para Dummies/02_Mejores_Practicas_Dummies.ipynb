## La checklist del piloto de gráficos ✈️

| Pregunta simple | Analogía |
|---|---|
| ¿Para qué sirve este gráfico? | No subas al avión sin saber a dónde vas |
| ¿Sobra algo en el gráfico? | Quita el equipaje que no necesitas para el viaje |
| ¿El eje empieza en cero? | No exageres la altura del vuelo en el instrumento |
| ¿Elegí el tipo de gráfico correcto? | No uses un mapa de carretera para volar un avión |
| ¿Alguien podría malinterpretarlo? | Revisa que las señales sean claras para la torre de control |
| ¿Usé la herramienta correcta (Python/Power BI/Tableau)? | No uses un avión pequeño para un vuelo transatlántico |

Si respondes "sí" a todas, ¡tu gráfico está listo para despegar! 🛫

````python
# Demo simplificada: el "antes y despues" mas famoso de la visualizacion
import matplotlib.pyplot as plt

trimestres = ['T1', 'T2', 'T3', 'T4']
ventas = [820, 845, 810, 910]

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

# ANTES: eje recortado, parece un cambio ENORME
axes[0].bar(trimestres, ventas, color=['red', 'blue', 'green', 'purple'])
axes[0].set_ylim(780, 930)
axes[0].set_title('❌ Antes: parece un cambio gigante')

# DESPUES: eje completo, se ve el cambio real (mucho mas chico)
axes[1].bar(trimestres, ventas, color=['#94a3b8', '#94a3b8', '#94a3b8', '#1e3a8a'])
axes[1].set_ylim(0, 1000)
axes[1].set_title('✅ Despues: se ve el cambio real (~11%)')

plt.tight_layout()
plt.show()

print("Mismo dato, dos historias distintas. Por eso revisar el eje SIEMPRE es el primer paso.")
````
