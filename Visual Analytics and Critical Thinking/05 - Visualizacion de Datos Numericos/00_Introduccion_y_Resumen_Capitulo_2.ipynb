````python
# ============================================================
# Panorama visual: de datos categóricos (Cap. 1) a numéricos (Cap. 2)
# ============================================================
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
fig.suptitle('De lo categórico (Capítulo 1) a lo numérico (Capítulo 2)', fontsize=13, fontweight='bold')

# Izquierda: dato categórico típico del Cap. 1
categorias = ['Norte', 'Sur', 'Centro', 'Oriente']
ventas_cat = [320, 210, 450, 180]
axes[0].bar(categorias, ventas_cat, color='#6366f1', edgecolor='white')
axes[0].set_title('Cap. 1: Datos categóricos\n(¿Cuánto vendió cada región?)', fontsize=10, fontweight='bold')
axes[0].set_ylabel('Ventas (unidades)')
axes[0].spines['top'].set_visible(False)
axes[0].spines['right'].set_visible(False)

# Derecha: dato numérico continuo típico del Cap. 2
datos_num = np.random.normal(1000, 220, 500)
axes[1].hist(datos_num, bins=30, color='#1e3a8a', edgecolor='white', alpha=0.85)
axes[1].set_title('Cap. 2: Datos numéricos\n(¿Cómo se distribuyen los valores?)', fontsize=10, fontweight='bold')
axes[1].set_xlabel('Valor de la observación')
axes[1].set_ylabel('Frecuencia')
axes[1].spines['top'].set_visible(False)
axes[1].spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

print("Cap. 1 responde: '¿Qué grupo tiene más?'")
print("Cap. 2 responde: '¿Cómo se distribuyen, relacionan y comportan los valores?'")
````

## 3. ¿Por qué los datos numéricos necesitan técnicas distintas? 🔍

Un dato categórico (por ejemplo, "región de venta") solo puede compararse por **conteo o frecuencia** dentro de cada categoría. Un dato numérico, en cambio, tiene propiedades matemáticas mucho más ricas:

- **Orden natural:** 25 años es más que 20 años; "Norte" no es "más" que "Sur".
- **Distancias interpretables:** la diferencia entre 10°C y 20°C es la misma que entre 20°C y 30°C.
- **Operaciones aritméticas válidas:** se puede promediar, sumar, calcular varianza — cosas sin sentido en una categoría.
- **Distribución:** los valores numéricos se agrupan, se dispersan, tienen forma (simétrica, sesgada, bimodal) — un concepto que no aplica a categorías.

Estas propiedades son justamente las que explotan las técnicas del Capítulo 2: histogramas, KDE, box plots, scatter plots y gráficos de línea, que veremos en detalle en los siguientes notebooks de este módulo.

````python
# ============================================================
# Comparación rápida: pregunta categórica vs pregunta numérica
# ============================================================
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(7)
fig, ax = plt.subplots(figsize=(9, 4))

tiempos_respuesta = np.random.gamma(shape=2.0, scale=40, size=2000)  # ms
ax.hist(tiempos_respuesta, bins=40, color='#0ea5e9', edgecolor='white', alpha=0.85)
ax.axvline(np.median(tiempos_respuesta), color='#dc2626', linestyle='--', linewidth=2,
           label=f'Mediana: {np.median(tiempos_respuesta):.0f} ms')
ax.set_title('Pregunta numérica típica del Capítulo 2:\n"¿Cómo se distribuyen los tiempos de respuesta de un sistema?"',
             fontsize=11, fontweight='bold')
ax.set_xlabel('Tiempo de respuesta (ms)')
ax.set_ylabel('Frecuencia')
ax.legend()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
````
