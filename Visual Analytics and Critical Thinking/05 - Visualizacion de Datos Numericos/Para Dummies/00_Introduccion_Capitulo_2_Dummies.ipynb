Y dentro de **este módulo (05)** hay 3 "misiones":

2. Aprender las técnicas para dibujarlos (líneas, histogramas, scatter plots).
3. Resolver un caso completo con datos numéricos inventados (ventas, tiempos de respuesta).

````python
# ============================================================
# La diferencia de un vistazo: categorías vs números
# ============================================================
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(1)
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

# Categorías: solo se puede contar
sabores = ['Chocolate', 'Vainilla', 'Fresa']
ventas = [50, 30, 20]
axes[0].bar(sabores, ventas, color='#a78bfa')
axes[0].set_title('Categorías: "¿Cuál se vendió más?"', fontsize=10, fontweight='bold')

# Números: se pueden promediar, ver su forma
propinas = np.random.gamma(2, 3, 300)
axes[1].hist(propinas, bins=25, color='#f59e0b', edgecolor='white')
axes[1].set_title('Números: "¿Cómo se reparten los valores?"', fontsize=10, fontweight='bold')

for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
print("Con sabores de helado solo puedes contar. Con propinas en dólares, puedes")
print("promediar, ver si están 'sesgadas', encontrar valores raros (outliers), etc.")
````

ax.set_xlabel('Propina')
ax.set_ylabel('Cantidad de propinas')
ax.legend()
plt.show()

# 5) Comprobación
assert promedio > mediana
assert len(ax.lines) == 2
print(f"El promedio ({promedio:.1f}) es mayor que la mediana ({mediana:.1f}): las propinas grandes jalan el promedio.")
