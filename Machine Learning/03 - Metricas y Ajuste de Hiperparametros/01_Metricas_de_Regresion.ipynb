## 1. La Familia del Error Cuadrático Medio 📐

### 1.1 Residuos y definiciones

Dado un conjunto de $n$ ejemplos con valores reales $y_i$ y predicciones $\hat y_i$, el **residuo** (o error) es $e_i = y_i - \hat y_i$. Todas las métricas de esta sección resumen el vector $\mathbf{e}$ en un número:

| Métrica | Fórmula | Unidades | Rasgos |
|---|---|---|---|
| **MSE** (*Mean Squared Error*) | $\dfrac1n\sum_i e_i^2$ | **al cuadrado** ($\text{m}^2$, $\text{COP}^2$…) | Diferenciable, convexo; penaliza **cuadráticamente** los errores grandes |
| **RMSE** | $\sqrt{\text{MSE}}$ | las de $y$ | Misma ordenación que el MSE, pero **interpretable**; norma $\ell_2$ |
| **MAE** (*Mean Absolute Error*) | $\dfrac1n\sum_i \lvert e_i\rvert$ | las de $y$ | Penaliza **linealmente**; norma $\ell_1$; más robusta |
| **MedAE** (error absoluto mediano) | $\operatorname{mediana}_i \lvert e_i\rvert$ | las de $y$ | Muy robusta (ignora hasta el 50 % de errores extremos) |
| **Error máximo** | $\max_i \lvert e_i\rvert$ | las de $y$ | Peor caso; dependiente de un solo punto |
| **MAPE** | $\dfrac1n\sum_i \dfrac{\lvert e_i\rvert}{\lvert y_i\rvert}$ | adimensional (%) | Relativo; **trampas** (Sección 4) |
| **$R^2$** | $1-\dfrac{\sum_i e_i^2}{\sum_i (y_i-\bar y)^2}$ | adimensional | Fracción de varianza explicada; puede ser **negativo** (Sección 2) |

Entre RMSE y MAE hay una relación **siempre válida**: por la desigualdad de Jensen (o de Cauchy–Schwarz),

$$\text{MAE} \;\le\; \text{RMSE} \;\le\; \sqrt{n}\;\text{MAE},$$

con $\text{MAE}=\text{RMSE}$ **si y solo si todos los $\lvert e_i\rvert$ son iguales**. El cociente $\text{RMSE}/\text{MAE}$ es por tanto un **termómetro de la heterogeneidad** de los errores: crece cuando unos pocos errores son enormes (lo usaremos en la Sección 7).

### 1.2 Media vs. mediana: qué estima cada pérdida

Si el «modelo» es una **constante** $c$ (el predictor más simple), cada pérdida tiene un minimizador distinto:

$$\arg\min_c \sum_i (y_i - c)^2 = \bar y \quad(\text{media}), \qquad \arg\min_c \sum_i \lvert y_i - c\rvert = \operatorname{mediana}(y).$$

*Demostración (MSE):* derivando, $-2\sum_i (y_i - c) = 0 \Rightarrow c = \bar y$. *(MAE):* la derivada de $\sum_i\lvert y_i - c\rvert$ es $\#\{y_i < c\} - \#\{y_i > c\}$, que cambia de signo exactamente en la mediana. En general, **entrenar con MSE estima la media condicional** $\mathbb{E}[y\mid\mathbf{x}]$ y **entrenar con MAE la mediana condicional**: no son el mismo modelo cuando los errores son asimétricos o con *outliers*.

Implementamos todo **desde cero** y lo verificamos contra `sklearn.metrics` con el conjunto *Diabetes* (respuesta = progresión de la enfermedad, en **unidades originales**).

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas matplotlib seaborn scikit-learn scipy

import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.special import huber as huber_scipy
from scipy.optimize import minimize_scalar
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, HuberRegressor, QuantileRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.dummy import DummyRegressor
from sklearn import metrics as skm

warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (9, 5)
print("Entorno listo.")
````

````python
# ============================================================
# 1. Metricas de regresion DESDE CERO, verificadas contra scikit-learn
# ============================================================
EPS64 = np.finfo(np.float64).eps                                        # misma constante que usa sklearn en MAPE

def residuos(y, yh):  return np.asarray(y, float) - np.asarray(yh, float)
def mse(y, yh):       return np.mean(residuos(y, yh) ** 2)
def rmse(y, yh):      return np.sqrt(mse(y, yh))                        # (np.sqrt(mean_squared_error) funciona en todas las versiones)
def mae(y, yh):       return np.mean(np.abs(residuos(y, yh)))
def medae(y, yh):     return np.median(np.abs(residuos(y, yh)))
def err_max(y, yh):   return np.max(np.abs(residuos(y, yh)))
def mape(y, yh):      return np.mean(np.abs(residuos(y, yh)) / np.maximum(np.abs(np.asarray(y, float)), EPS64))   # fraccion, no %
def r2(y, yh):
    y = np.asarray(y, float)
    return 1 - np.sum(residuos(y, yh) ** 2) / np.sum((y - y.mean()) ** 2)
def var_explicada(y, yh):                                               # explained_variance_score: ignora el sesgo medio
    y = np.asarray(y, float)
    return 1 - np.var(residuos(y, yh)) / np.var(y)

# --- Datos: la respuesta esta en unidades originales (indice de progresion de la enfermedad) ---
dia = load_diabetes()
X, y = dia.data, dia.target
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=SEED)
lin = LinearRegression().fit(X_tr, y_tr)
y_hat = lin.predict(X_te)

pares = [("MSE", mse, skm.mean_squared_error), ("MAE", mae, skm.mean_absolute_error),
         ("MedAE", medae, skm.median_absolute_error), ("Error máximo", err_max, skm.max_error),
         ("MAPE", mape, skm.mean_absolute_percentage_error), ("R²", r2, skm.r2_score),
         ("Varianza explicada", var_explicada, skm.explained_variance_score)]
filas = []
for nombre, propia, de_sklearn in pares:
    a, b = propia(y_te, y_hat), de_sklearn(y_te, y_hat)
    assert np.isclose(a, b), (nombre, a, b)
    filas.append({"métrica": nombre, "desde cero": a, "scikit-learn": b})
assert np.isclose(rmse(y_te, y_hat), np.sqrt(skm.mean_squared_error(y_te, y_hat)))
filas.insert(1, {"métrica": "RMSE", "desde cero": rmse(y_te, y_hat), "scikit-learn": np.sqrt(skm.mean_squared_error(y_te, y_hat))})
display(pd.DataFrame(filas).set_index("métrica").round(4))
print("Todas las métricas propias coinciden con sklearn.metrics.\n")

# --- Desigualdades MAE <= RMSE <= sqrt(n) * MAE (en este y en 1000 vectores de error aleatorios) ---
g = np.random.default_rng(SEED)
for _ in range(1000):
    e = g.standard_t(df=2, size=g.integers(2, 60))                       # colas pesadas a proposito
    assert np.mean(np.abs(e)) <= np.sqrt(np.mean(e ** 2)) + 1e-12 <= np.sqrt(len(e)) * np.mean(np.abs(e)) + 1e-9
assert np.isclose(rmse([1, 2, 3], [2, 3, 4]), mae([1, 2, 3], [2, 3, 4]))     # errores todos iguales => RMSE = MAE

# --- Lectura en las unidades del problema ---
print(f"RMSE = {rmse(y_te, y_hat):.1f} y MAE = {mae(y_te, y_hat):.1f} unidades de progresión de la enfermedad; "
      f"la respuesta varía entre {y.min():.0f} y {y.max():.0f} (desv. estándar = {y.std():.1f}).")
print(f"Un error típico de ~{mae(y_te, y_hat):.0f} puntos sobre una escala con desviación {y.std():.0f} es modesto: el modelo explica R² = {r2(y_te, y_hat):.2f} de la varianza.")

# --- Media vs. mediana: la constante que minimiza cada perdida (con datos asimetricos) ---
datos = np.random.default_rng(SEED).lognormal(mean=0.0, sigma=0.8, size=2001)
cs = np.linspace(datos.min(), np.quantile(datos, 0.95), 4000)
c_mse = cs[np.argmin([mse(datos, np.full_like(datos, c)) for c in cs])]
c_mae = cs[np.argmin([mae(datos, np.full_like(datos, c)) for c in cs])]
assert abs(c_mse - datos.mean()) < 0.01 and abs(c_mae - np.median(datos)) < 0.01
print(f"\nPredictor constante sobre datos asimétricos: argmin MSE = {c_mse:.3f} ≈ media = {datos.mean():.3f} | argmin MAE = {c_mae:.3f} ≈ mediana = {np.median(datos):.3f}")
````

## 2. $R^2$: la Fracción de Varianza «Explicada» y por qué Puede ser Negativo 📉

### 2.1 Definición

El **coeficiente de determinación** compara el error del modelo con el de un **modelo de referencia** que siempre predice la media:

$$R^2 = 1 - \frac{\text{SSE}}{\text{SST}} = 1 - \frac{\sum_i (y_i-\hat y_i)^2}{\sum_i (y_i-\bar y)^2} = 1 - \frac{\text{MSE}}{\operatorname{Var}(y)}.$$

- $R^2=1$: predicciones perfectas; $R^2=0$: el modelo **no mejora** al de predecir la media; **$R^2<0$: es peor** que predecir la media.
- En **entrenamiento**, un modelo lineal con intercepto ajustado por mínimos cuadrados **nunca** puede dar $R^2<0$ (la constante $\bar y$ está dentro de la familia de modelos) y coincide con $\operatorname{corr}(y,\hat y)^2$.
- En **prueba**, nada lo garantiza: el modelo puede estar sobreajustado o sesgado y entonces $R^2<0$. Además `r2_score` usa la media **del propio conjunto evaluado** ($\bar y_{\text{prueba}}$), mientras que el modelo real de referencia que tenemos es la media **de entrenamiento**.

### 2.2 Relación exacta con el modelo base

Para cualquier constante $c$ se cumple la descomposición $\sum_i (y_i - c)^2 = \sum_i (y_i-\bar y)^2 + n(\bar y - c)^2$. Aplicada al **modelo base** «predecir la media de entrenamiento $\bar y_{\text{ent}}$» sobre la prueba:

$$R^2_{\text{base}} = 1 - \frac{\text{SST} + n(\bar y_{\text{prueba}} - \bar y_{\text{ent}})^2}{\text{SST}} = -\frac{(\bar y_{\text{prueba}} - \bar y_{\text{ent}})^2}{\operatorname{Var}(y_{\text{prueba}})} \;\le\; 0.$$

Es decir, **el baseline más tonto ya obtiene $R^2\le 0$ en prueba**, y se acerca a 0 solo si las medias de entrenamiento y prueba coinciden. Cualquier modelo útil debe superar ese piso.

### 2.3 Advertencias

- $R^2$ **no** mide si el modelo es *correcto* ni *bueno* en unidades: un $R^2=0.3$ puede ser excelente (ruido irreducible alto) o malo.
- $R^2$ **no baja** al agregar variables en entrenamiento (de ahí el $R^2$ ajustado); medirlo en prueba es lo honesto.
- $R^2$ **no es** $\operatorname{corr}^2$ fuera de entrenamiento: un modelo con **sesgo** puede correlacionar perfecto y aun así tener $R^2$ muy negativo.

Lo vemos en acción: (a) la identidad del baseline, (b) la diferencia $R^2$ vs. $\operatorname{corr}^2$ y (c) un modelo que **interpola** el entrenamiento (más variables que datos) y colapsa en prueba.

````python
# ============================================================
# 2. R2 negativo en prueba: baseline, sesgo y sobreajuste
# ============================================================
# (a) El modelo base "predecir la media de entrenamiento" y la identidad R2_base = -(diferencia de medias)^2 / Var(y_prueba)
base = DummyRegressor(strategy="mean").fit(X_tr, y_tr)
r2_base = skm.r2_score(y_te, base.predict(X_te))
identidad = -(y_te.mean() - y_tr.mean()) ** 2 / np.var(y_te)
assert np.isclose(r2_base, identidad)
print(f"(a) R² del baseline en prueba = {r2_base:.5f} = -(ȳ_prueba - ȳ_ent)²/Var(y_prueba) = {identidad:.5f}  (≤ 0 siempre)")

# (b) Correlacion perfecta pero R2 muy negativo: predicciones con sesgo y escala equivocados
y_sesgado = 2 * y_te + 80
print(f"(b) corr(y, ŷ)² = {np.corrcoef(y_te, y_sesgado)[0, 1] ** 2:.3f} pero R² = {skm.r2_score(y_te, y_sesgado):.2f}: la correlación no mide el error absoluto.")
assert np.isclose(np.corrcoef(y_te, y_sesgado)[0, 1] ** 2, 1.0) and skm.r2_score(y_te, y_sesgado) < -1

# (c) Mas variables irrelevantes que ejemplos => el modelo interpola el entrenamiento y falla en prueba
n_ent = 60
rng = np.random.default_rng(SEED)
idx = rng.permutation(len(y)); i_tr, i_te = idx[:n_ent], idx[n_ent:]
n_ruido = [0, 10, 25, 40, 50, 60, 70, 100, 150, 300]
r2_tr, r2_te = [], []
for k in n_ruido:
    Xk = np.c_[X, rng.normal(size=(len(y), k))] if k else X               # 10 variables reales + k de ruido puro
    m = LinearRegression().fit(Xk[i_tr], y[i_tr])
    r2_tr.append(skm.r2_score(y[i_tr], m.predict(Xk[i_tr]))); r2_te.append(skm.r2_score(y[i_te], m.predict(Xk[i_te])))
assert r2_tr[-1] > 0.999                                      # con p > n el ajuste interpola (R2 entrenamiento = 1)
assert r2_te[0] > 0 and min(r2_te) < 0                         # en prueba el R2 llega a ser negativo

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
axes[0].plot(n_ruido, r2_tr, "o-", color="#2563eb", label="$R^2$ entrenamiento (60 ejemplos)")
axes[0].plot(n_ruido, r2_te, "s--", color="#dc2626", label="$R^2$ prueba (382 ejemplos)")
axes[0].axhline(0, color="black", lw=0.8); axes[0].axvline(n_ent - 10, color="gray", linestyle=":", label="$p = n$ (ruido = 50)")
axes[0].set_ylim(-3, 1.05); axes[0].set_xlabel("Variables de ruido añadidas (10 variables reales)"); axes[0].set_ylabel("$R^2$ (eje recortado en -3)")
axes[0].set_title("Con más variables que datos: $R^2$ de prueba negativo", fontweight="bold"); axes[0].legend(fontsize=8)
axes[1].scatter(y_te, y_sesgado, s=18, color="#f59e0b", label="Predicciones con sesgo y escala erróneos")
axes[1].plot([y_te.min(), y_te.max()], [y_te.min(), y_te.max()], "k--", label="Predicción perfecta ($\\hat y = y$)")
axes[1].set_xlabel("Valor real $y$"); axes[1].set_ylabel("Predicción $\\hat y$")
axes[1].set_title("Correlación 1.0 y $R^2$ muy negativo", fontweight="bold"); axes[1].legend(fontsize=8)
plt.tight_layout()
plt.show()
print("R² de prueba por número de variables de ruido:", {k: round(float(v), 2) for k, v in zip(n_ruido, r2_te)})
print("Moraleja: calcula siempre R² con datos NO usados para ajustar; en entrenamiento puede valer 1 mientras en prueba es peor que la media.")
````

## 3. El Baseline Ingenuo: toda Métrica Necesita una Referencia 🧱

Un RMSE de 54 *no significa nada* por sí solo. Hay que compararlo con **lo que haría un modelo trivial**. `DummyRegressor` implementa los baselines clásicos:

| `strategy` | Predice | Es óptimo para |
|---|---|---|
| `"mean"` | la media de entrenamiento | pérdida cuadrática (MSE/RMSE) |
| `"median"` | la mediana de entrenamiento | pérdida absoluta (MAE) |
| `"quantile"` | un cuantil fijo | pérdida *pinball* (cuantiles) |
| `"constant"` | un valor dado por ti | casos con conocimiento del dominio |

La **habilidad relativa** (*skill score*) respecto al baseline es $\;1 - \dfrac{\text{RMSE}_{\text{modelo}}}{\text{RMSE}_{\text{base}}}\;$ (o con MAE). Fíjate que $1-\text{MSE}_{\text{modelo}}/\text{MSE}_{\text{base}}$ es **exactamente el $R^2$ respecto a la media de entrenamiento**. Si tu modelo no supera claramente al baseline, el problema no es la métrica: es el modelo (o los datos).

````python
# ============================================================
# 3. Baselines vs. modelos reales en Diabetes (RMSE, MAE y R2 en prueba, en unidades originales)
# ============================================================
candidatos = {"Baseline: media": DummyRegressor(strategy="mean"),
              "Baseline: mediana": DummyRegressor(strategy="median"),
              "Regresión lineal": LinearRegression(),
              "Random Forest (100 árboles, prof. 4)": RandomForestRegressor(n_estimators=100, max_depth=4, random_state=SEED)}
filas, preds = {}, {}
for nombre, mod in candidatos.items():
    mod.fit(X_tr, y_tr)
    p = mod.predict(X_te); preds[nombre] = p
    filas[nombre] = {"RMSE": rmse(y_te, p), "MAE": mae(y_te, p), "R²": r2(y_te, p)}
tab = pd.DataFrame(filas).T
rmse_base, mae_base = tab.loc["Baseline: media", "RMSE"], tab.loc["Baseline: mediana", "MAE"]
tab["habilidad vs. baseline (RMSE)"] = 1 - tab["RMSE"] / rmse_base
display(tab.round(3))

# El baseline "media" es el mejor baseline para RMSE y el baseline "mediana" para MAE (sobre entrenamiento)
assert np.isclose(candidatos["Baseline: media"].constant_[0, 0], y_tr.mean()) and np.isclose(candidatos["Baseline: mediana"].constant_[0, 0], np.median(y_tr))
assert tab.loc["Regresión lineal", "RMSE"] < rmse_base and tab.loc["Random Forest (100 árboles, prof. 4)", "RMSE"] < rmse_base
# 1 - MSE_modelo / MSE_base es el R2 respecto a la media de ENTRENAMIENTO
sse_base = np.sum((y_te - y_tr.mean()) ** 2)
assert np.isclose(1 - np.sum((y_te - preds["Regresión lineal"]) ** 2) / sse_base, 1 - mse(y_te, preds["Regresión lineal"]) / mse(y_te, preds["Baseline: media"]))

fig, ax = plt.subplots(figsize=(9, 4.6))
orden = tab.index
ax.barh(orden[::-1], tab["RMSE"][::-1], color=["#16a34a", "#2563eb", "#9ca3af", "#9ca3af"])
for i, v in enumerate(tab["RMSE"][::-1]): ax.text(v + 0.5, i, f"{v:.1f}", va="center")
ax.set_xlabel("RMSE en prueba (unidades de progresión de la enfermedad)"); ax.set_title("Un modelo solo vale lo que supera a su baseline", fontweight="bold")
plt.tight_layout()
plt.show()
print(f"El baseline de la media tiene RMSE = {rmse_base:.1f}; la regresión lineal lo reduce a {tab.loc['Regresión lineal', 'RMSE']:.1f} "
      f"({tab.loc['Regresión lineal', 'habilidad vs. baseline (RMSE)']:.0%} de mejora relativa).")
````

## 4. Errores Porcentuales: MAPE y sMAPE y sus Trampas ⚠️

Los errores **relativos** son atractivos: «nos equivocamos un 8 %» se entiende sin saber las unidades y permite comparar series de escalas distintas.

$$\text{MAPE} = \frac{100\,\%}{n}\sum_i \frac{\lvert y_i - \hat y_i\rvert}{\lvert y_i\rvert}, \qquad \text{sMAPE} = \frac{100\,\%}{n}\sum_i \frac{2\,\lvert y_i - \hat y_i\rvert}{\lvert y_i\rvert + \lvert \hat y_i\rvert}.$$

(La sMAPE tiene varias variantes en la literatura; esta es la más común, con rango $[0, 200\,\%]$. Hyndman & Koehler, 2006, discuten sus problemas.) Las **trampas** son serias:

1. **División por cero (o casi cero).** Si algún $y_i = 0$ el MAPE no está definido; si $y_i\approx 0$ ese único punto domina. `scikit-learn` evita el error dividiendo entre $\max(\lvert y_i\rvert, \varepsilon)$ con $\varepsilon \approx 2.2\times10^{-16}$, lo que **devuelve un número gigantesco** (¡no una advertencia!).
2. **Asimetría.** Con $y>0$ no se puede errar por más del 100 % *por defecto* (predecir 0), pero por *exceso* el error es ilimitado. Mismo error absoluto, distinto castigo: $y=100,\hat y=150\Rightarrow 50\,\%$, pero $y=150,\hat y=100\Rightarrow 33\,\%$.
3. **Sesgo hacia subestimar.** Como los errores por exceso pesan más, **minimizar MAPE favorece predicciones bajas**. Para una distribución log-normal $\ln y\sim\mathcal N(\mu,\sigma^2)$ los minimizadores de una **constante** son:

| Pérdida | Constante óptima | Con $\mu=\ln 100,\ \sigma=0.5$ |
|---|---|---|
| MSE | media $=e^{\mu+\sigma^2/2}$ | $113.3$ |
| MAE | mediana $=e^{\mu}$ | $100.0$ |
| MAPE | $e^{\mu-\sigma^2}$ | $\mathbf{77.9}$ |

   (el MAPE pondera cada punto por $1/y$, que desplaza el minimizador hacia valores pequeños).
4. **La sMAPE corrige a medias:** penaliza más la **subestimación** ($y=100,\hat y=50\to 66.7\,\%$; $\hat y=150\to 40\,\%$) y también explota cuando $y$ y $\hat y$ son ambos cercanos a 0.

**Alternativas** cuando hay ceros o escalas distintas: MAE/MedAE divididos por una escala de referencia (p. ej., el MAE del baseline, que da el **MASE**), el **RMSLE** (error sobre $\ln(1+y)$, que mide errores *relativos* sin dividir por cero) o simplemente reportar MAE junto con el rango de $y$.

````python
# ============================================================
# 4. MAPE / sMAPE: trampas numericas
# ============================================================
def smape(y, yh):
    """sMAPE (variante 2|e| / (|y| + |yh|)); 0/0 se define como 0."""
    y, yh = np.asarray(y, float), np.asarray(yh, float)
    den = np.abs(y) + np.abs(yh)
    return np.mean(np.divide(2 * np.abs(y - yh), den, out=np.zeros_like(den), where=den > 0))

# (1) Division por cero: sklearn y nuestra version dan el mismo numero gigantesco
y_cero, p_cero = np.array([0.0, 10.0, 20.0]), np.array([1.0, 11.0, 19.0])
print(f"(1) y = {y_cero}, ŷ = {p_cero}: MAPE sklearn = {skm.mean_absolute_percentage_error(y_cero, p_cero):.3e} | propio = {mape(y_cero, p_cero):.3e}  (¡un error de 1 unidad sobre y = 0!)")
assert np.isclose(mape(y_cero, p_cero), skm.mean_absolute_percentage_error(y_cero, p_cero)) and mape(y_cero, p_cero) > 1e10

# (2) Asimetria: mismo error absoluto, distinto porcentaje
print(f"(2) y = 100, ŷ = 150: MAPE = {mape([100], [150]):.0%} | y = 150, ŷ = 100: MAPE = {mape([150], [100]):.0%}")
print(f"    sMAPE: y = 100, ŷ = 50 → {smape([100], [50]):.1%} | y = 100, ŷ = 150 → {smape([100], [150]):.1%} (penaliza más la subestimación)")
assert np.isclose(mape([100], [150]), 0.5) and np.isclose(mape([150], [100]), 1 / 3)
assert smape([100], [50]) > smape([100], [150])

# (3) Sesgo hacia subestimar: argmin de cada perdida para una constante, con y ~ lognormal(ln 100, 0.5)
mu, sig = np.log(100.0), 0.5
ys = np.random.default_rng(SEED).lognormal(mu, sig, size=100_000)
cs = np.linspace(40, 180, 701)
curvas = {"MSE": np.array([mse(ys, np.full_like(ys, c)) for c in cs]), "MAE": np.array([mae(ys, np.full_like(ys, c)) for c in cs]),
          "MAPE": np.array([mape(ys, np.full_like(ys, c)) for c in cs]), "sMAPE": np.array([smape(ys, np.full_like(ys, c)) for c in cs])}
optimo = {k: cs[np.argmin(v)] for k, v in curvas.items()}
teorico = {"MSE": np.exp(mu + sig ** 2 / 2), "MAE": np.exp(mu), "MAPE": np.exp(mu - sig ** 2)}
for k in teorico:
    assert abs(optimo[k] - teorico[k]) / teorico[k] < 0.03, (k, optimo[k], teorico[k])     # coincide con la teoria
assert optimo["MAPE"] < optimo["MAE"] < optimo["MSE"]
display(pd.DataFrame({"óptimo empírico": optimo, "óptimo teórico": {**teorico, "sMAPE": np.nan}}).round(1))

fig, axes = plt.subplots(1, 2, figsize=(15, 4.8))
for (k, v), color in zip(curvas.items(), ["#2563eb", "#16a34a", "#dc2626", "#f59e0b"]):
    axes[0].plot(cs, v / v.min(), color=color, lw=2, label=f"{k} (mín. en {optimo[k]:.0f})")
axes[0].set_ylim(1, 3); axes[0].set_xlabel("Valor de la constante predicha $c$"); axes[0].set_ylabel("Pérdida / pérdida mínima")
axes[0].set_title("Cada pérdida elige una constante distinta", fontweight="bold"); axes[0].legend(fontsize=8)
yr = 100.0; pr = np.linspace(1, 300, 400)
axes[1].plot(pr, 100 * np.abs(yr - pr) / yr, color="#dc2626", lw=2, label="MAPE (y real = 100)")
axes[1].plot(pr, 100 * 2 * np.abs(yr - pr) / (yr + pr), color="#f59e0b", lw=2, label="sMAPE (y real = 100)")
axes[1].axvline(yr, color="gray", linestyle=":")
axes[1].set_xlabel("Predicción $\\hat y$"); axes[1].set_ylabel("Error porcentual (%)")
axes[1].set_title("Asimetría: acotado por defecto (MAPE), ilimitado por exceso", fontweight="bold"); axes[1].legend(fontsize=8)
plt.tight_layout()
plt.show()
print(f"El MAPE elige c = {optimo['MAPE']:.0f} (teoría {teorico['MAPE']:.1f}) frente a la mediana {optimo['MAE']:.0f} y la media {optimo['MSE']:.0f}: optimizar MAPE sesga a subestimar.")
````

## 5. Sensibilidad a *Outliers*: MSE, MAE y Huber 🎯

Al elevar el residuo al cuadrado, un error de 10 unidades pesa **100** veces lo que uno de 1; con valor absoluto pesa solo 10 veces. Por eso:

- el **MSE/RMSE** está **dominado por los errores grandes** (y su minimizador, la media, es arrastrado por los *outliers*);
- el **MAE** es proporcional al error y su minimizador (la mediana) tiene **punto de ruptura del 50 %**;
- la **pérdida de Huber** (Huber, 1964) es el compromiso: **cuadrática** para residuos pequeños ($\lvert e\rvert\le\delta$) y **lineal** para grandes,

$$L_\delta(e)=\begin{cases}\tfrac12 e^2, & \lvert e\rvert\le\delta,\\[2pt] \delta\big(\lvert e\rvert-\tfrac12\delta\big), & \lvert e\rvert>\delta.\end{cases}$$

Es diferenciable en todo punto (a diferencia del valor absoluto en 0) y no explota con *outliers*. Primero verificamos nuestra pérdida de Huber contra `scipy.special.huber`; luego hacemos dos experimentos: (a) cómo reaccionan las **métricas** a un solo error enorme y (b) cómo reaccionan los **modelos entrenados** con cada pérdida cuando los datos están contaminados.

````python
# ============================================================
# 5. Outliers: reaccion de las metricas y de los modelos entrenados con MSE, MAE y Huber
# ============================================================
def huber_perdida(e, delta=1.0):
    e = np.abs(np.asarray(e, float))
    return np.where(e <= delta, 0.5 * e ** 2, delta * (e - 0.5 * delta))

e_test = np.linspace(-6, 6, 241)
for d in (0.5, 1.0, 1.35, 3.0):
    assert np.allclose(huber_perdida(e_test, d), huber_scipy(d, e_test))          # coincide con scipy.special.huber
print("Pérdida de Huber propia == scipy.special.huber.\n")

# (a) Un solo error enorme: como se mueven RMSE, MAE y MedAE
rng = np.random.default_rng(SEED)
e0 = rng.normal(0, 1, 100)                                                         # 100 errores N(0,1)
tamanos = np.array([0, 2, 5, 10, 20, 50, 100])
res = {"RMSE": [], "MAE": [], "MedAE": []}
for k in tamanos:
    e = e0.copy(); e[0] = k                                                        # reemplazamos un error por uno de tamano k
    res["RMSE"].append(np.sqrt(np.mean(e ** 2))); res["MAE"].append(np.mean(np.abs(e))); res["MedAE"].append(np.median(np.abs(e)))
res = pd.DataFrame(res, index=tamanos); res.index.name = "tamaño del error atípico"
display(res.round(3).T)
assert res["RMSE"].iloc[-1] / res["RMSE"].iloc[0] > 4 > res["MAE"].iloc[-1] / res["MAE"].iloc[0] > 1.5 > res["MedAE"].iloc[-1] / res["MedAE"].iloc[0]

# (b) Modelos entrenados: y = 1 + 2x + ruido, con 10 % de datos contaminados (+40)
n = 200
x = rng.uniform(0, 10, n)
y_lim = 1 + 2 * x + rng.normal(0, 1, n)
cont = rng.random(n) < 0.10
y_sucio = y_lim + 40 * cont + rng.normal(0, 1, n) * cont
Xs = x.reshape(-1, 1)
modelos = {"MSE (mínimos cuadrados)": LinearRegression().fit(Xs, y_sucio),
           "Huber (δ = 1.35)": HuberRegressor(epsilon=1.35, alpha=0.0, max_iter=500).fit(Xs, y_sucio),
           "MAE (mediana, QuantileRegressor)": QuantileRegressor(quantile=0.5, alpha=0.0, solver="highs").fit(Xs, y_sucio)}
x_nuevo = rng.uniform(0, 10, 500); y_nuevo = 1 + 2 * x_nuevo + rng.normal(0, 1, 500)           # prueba LIMPIA
filas = {}
for nombre, m in modelos.items():
    filas[nombre] = {"intercepto (real = 1)": float(m.intercept_), "pendiente (real = 2)": float(m.coef_[0]),
                     "RMSE en prueba limpia": rmse(y_nuevo, m.predict(x_nuevo.reshape(-1, 1)))}
tabla = pd.DataFrame(filas).T
display(tabla.round(3))
err_int = {k: abs(v["intercepto (real = 1)"] - 1) for k, v in filas.items()}
assert err_int["Huber (δ = 1.35)"] < err_int["MSE (mínimos cuadrados)"] and err_int["MAE (mediana, QuantileRegressor)"] < err_int["MSE (mínimos cuadrados)"]

fig, axes = plt.subplots(1, 3, figsize=(19, 4.8))
axes[0].semilogy(tamanos[1:], res["RMSE"].iloc[1:] / res["RMSE"].iloc[0], "o-", color="#dc2626", label="RMSE")
axes[0].semilogy(tamanos[1:], res["MAE"].iloc[1:] / res["MAE"].iloc[0], "s-", color="#2563eb", label="MAE")
axes[0].semilogy(tamanos[1:], res["MedAE"].iloc[1:] / res["MedAE"].iloc[0], "^-", color="#16a34a", label="MedAE")
axes[0].set_xlabel("Tamaño de UN error atípico (los otros 99 ~ N(0,1))"); axes[0].set_ylabel("Valor / valor sin atípico (escala log)")
axes[0].set_title("Un solo error: el RMSE se dispara, la MedAE ni se entera", fontweight="bold"); axes[0].legend(fontsize=8)
for nombre, perdida, color in [("Cuadrática $\\frac{1}{2}e^2$", 0.5 * e_test ** 2, "#dc2626"), ("Absoluta $|e|$", np.abs(e_test), "#2563eb"),
                               ("Huber ($\\delta=1.35$)", huber_perdida(e_test, 1.35), "#16a34a")]:
    axes[1].plot(e_test, perdida, lw=2, label=nombre, color=color)
axes[1].set_ylim(0, 8); axes[1].set_xlabel("Residuo $e$"); axes[1].set_ylabel("Pérdida")
axes[1].set_title("Las tres pérdidas", fontweight="bold"); axes[1].legend(fontsize=8)
axes[2].scatter(x[~cont], y_sucio[~cont], s=14, color="#9ca3af", label="Datos limpios"); axes[2].scatter(x[cont], y_sucio[cont], s=22, color="#f59e0b", label="Contaminados (+40)")
xx = np.linspace(0, 10, 100).reshape(-1, 1)
for (nombre, m), color in zip(modelos.items(), ["#dc2626", "#16a34a", "#2563eb"]):
    axes[2].plot(xx, m.predict(xx), lw=2, color=color, label=nombre)
axes[2].plot(xx, 1 + 2 * xx, "k--", label="Verdad: $y=1+2x$")
axes[2].set_xlabel("$x$"); axes[2].set_ylabel("$y$"); axes[2].set_title("Ajuste con 10 % de datos contaminados", fontweight="bold"); axes[2].legend(fontsize=7)
plt.tight_layout()
plt.show()
print(f"Un único error de tamaño 100 multiplica el RMSE por {res['RMSE'].iloc[-1] / res['RMSE'].iloc[0]:.1f}, el MAE por {res['MAE'].iloc[-1] / res['MAE'].iloc[0]:.1f} y la MedAE por {res['MedAE'].iloc[-1] / res['MedAE'].iloc[0]:.2f}.")
print(f"Con datos contaminados, el error del intercepto es {err_int['MSE (mínimos cuadrados)']:.2f} (MSE) frente a {err_int['Huber (δ = 1.35)']:.2f} (Huber) y {err_int['MAE (mediana, QuantileRegressor)']:.2f} (MAE).")
````

## 6. La Mirada Probabilística: MSE ↔ Gaussiana, MAE ↔ Laplace 🔔

Elegir una pérdida equivale a **suponer una distribución del ruido**. Si $y_i = f(\mathbf{x}_i) + \varepsilon_i$ con $\varepsilon_i$ independientes:

**Ruido gaussiano**, $\varepsilon\sim\mathcal N(0,\sigma^2)$:

$$-\ln\mathcal L = \frac n2\ln(2\pi\sigma^2) + \frac{1}{2\sigma^2}\sum_i \big(y_i - f(\mathbf{x}_i)\big)^2.$$

Para $\sigma$ fijo, **maximizar la verosimilitud = minimizar el SSE (= minimizar el MSE)**. Además el estimador de máxima verosimilitud de la varianza es $\hat\sigma^2 = \text{MSE}$ (con divisor $n$), es decir, **$\hat\sigma = \text{RMSE}$**.

**Ruido de Laplace** (exponencial doble), $p(\varepsilon)=\frac{1}{2b}e^{-\lvert\varepsilon\rvert/b}$:

$$-\ln\mathcal L = n\ln(2b) + \frac1b\sum_i \big\lvert y_i - f(\mathbf{x}_i)\big\rvert.$$

**Maximizar la verosimilitud = minimizar el SAE (= minimizar el MAE)**, y $\hat b = \text{MAE}$ (evaluado en el ajuste óptimo, cuya posición es la **mediana**).

Las colas de Laplace son **más pesadas** que las gaussianas: asigna más probabilidad a errores grandes, por eso «se asusta menos» con *outliers*. (La pérdida de Huber corresponde a un ruido gaussiano en el centro con colas de Laplace; y la **pérdida *pinball*** da la regresión por cuantiles, ruido de Laplace asimétrico.) Lo verificamos numéricamente: ajustamos por máxima verosimilitud (con `scipy.stats`) y comparamos con la media, la mediana, el RMSE y el MAE.

````python
# ============================================================
# 6. MLE gaussiana -> media/RMSE ; MLE Laplace -> mediana/MAE (verificado numericamente)
# ============================================================
rng = np.random.default_rng(SEED)
n = 20000
ruido_g = rng.normal(loc=3.0, scale=2.0, size=n)                    # datos gaussianos
ruido_l = rng.laplace(loc=3.0, scale=2.0, size=n)                   # datos de Laplace

for nombre, d in [("gaussianos", ruido_g), ("de Laplace", ruido_l)]:
    mu_g, sd_g = stats.norm.fit(d)                                    # MLE gaussiana
    mu_l, b_l = stats.laplace.fit(d)                                  # MLE Laplace
    # Minimizacion numerica directa de la log-verosimilitud negativa (en la posicion), por si fit() cambiara de metodo
    nll_g = minimize_scalar(lambda m: np.sum((d - m) ** 2)).x
    nll_l = minimize_scalar(lambda m: np.sum(np.abs(d - m)), bounds=(d.min(), d.max()), method="bounded", options={"xatol": 1e-8}).x
    assert np.isclose(mu_g, d.mean()) and np.isclose(nll_g, d.mean(), atol=1e-6)           # posicion gaussiana = media
    assert np.isclose(sd_g, rmse(d, np.full_like(d, d.mean())))                             # sigma_MLE = RMSE alrededor de la media
    assert np.isclose(mu_l, np.median(d), atol=1e-3) and np.isclose(nll_l, np.median(d), atol=1e-3)   # posicion Laplace = mediana
    assert np.isclose(b_l, mae(d, np.full_like(d, np.median(d))), rtol=1e-3)                # b_MLE = MAE alrededor de la mediana
    ll_g = stats.norm.logpdf(d, mu_g, sd_g).sum(); ll_l = stats.laplace.logpdf(d, mu_l, b_l).sum()
    print(f"Datos {nombre:11s}: media = {d.mean():.3f}, mediana = {np.median(d):.3f}, RMSE = {sd_g:.3f}, MAE = {b_l:.3f} | "
          f"log-verosimilitud gaussiana = {ll_g:,.0f} | Laplace = {ll_l:,.0f} → mejor ajuste: {'gaussiana' if ll_g > ll_l else 'Laplace'}")
    assert (ll_g > ll_l) == (nombre == "gaussianos")
print("\nMLE gaussiana = (media, RMSE); MLE Laplace = (mediana, MAE): verificado. La familia que mejor ajusta los residuos sugiere la métrica natural.")

# El cociente RMSE/MAE del ruido identifica la forma: gaussiano -> sqrt(pi/2) = 1.2533 ; Laplace -> sqrt(2) = 1.4142
raz_g = np.sqrt(np.mean((ruido_g - ruido_g.mean()) ** 2)) / np.mean(np.abs(ruido_g - ruido_g.mean()))
raz_l = np.sqrt(np.mean((ruido_l - np.median(ruido_l)) ** 2)) / np.mean(np.abs(ruido_l - np.median(ruido_l)))
assert abs(raz_g - np.sqrt(np.pi / 2)) < 0.02 and abs(raz_l - np.sqrt(2)) < 0.04
print(f"RMSE/MAE: gaussiano = {raz_g:.4f} (teoría {np.sqrt(np.pi / 2):.4f}) | Laplace = {raz_l:.4f} (teoría {np.sqrt(2):.4f}).")

fig, axes = plt.subplots(1, 2, figsize=(14, 4.8))
xs = np.linspace(-7, 13, 500)
for ax, d, titulo in [(axes[0], ruido_g, "Datos gaussianos"), (axes[1], ruido_l, "Datos de Laplace")]:
    ax.hist(d, bins=80, density=True, color="#9ca3af", alpha=0.6, label="Datos")
    mg, sg = stats.norm.fit(d); ml, bl = stats.laplace.fit(d)
    ax.plot(xs, stats.norm.pdf(xs, mg, sg), color="#dc2626", lw=2, label=f"Gaussiana (MSE): $\\mu$={mg:.2f}, $\\sigma$={sg:.2f}")
    ax.plot(xs, stats.laplace.pdf(xs, ml, bl), color="#2563eb", lw=2, linestyle="--", label=f"Laplace (MAE): $\\mu$={ml:.2f}, $b$={bl:.2f}")
    ax.set_xlabel("Valor del error"); ax.set_ylabel("Densidad"); ax.set_title(titulo + ": ¿qué pérdida es natural?", fontweight="bold"); ax.legend(fontsize=8)
plt.tight_layout()
plt.show()
````

## 7. ¿Qué Métrica Elegir y Cómo Reportarla? 🧭

| Situación | Métrica recomendada | Razón |
|---|---|---|
| Errores grandes **desproporcionadamente** costosos (p. ej., pronóstico de demanda eléctrica) | **RMSE** (MSE para entrenar) | Penaliza cuadráticamente; estima la media |
| Costo **proporcional** al error; datos con *outliers* | **MAE** / **MedAE** | Robustas; estiman la mediana |
| Quieres lo mejor de ambos mundos | **Huber** | Cuadrática en el centro, lineal en las colas |
| Comparar series de **escalas** distintas, sin ceros ni valores cercanos a 0 | **MAPE** (con cuidado) / MAE relativo | Adimensional; sesgo a subestimar |
| Hay ceros o escalas muy distintas | **MASE**, **RMSLE** o MAE/rango | Evitan dividir entre $y\approx0$ |
| Decir «qué tan bien explica el modelo» | **$R^2$ en prueba** + baseline | Adimensional, pero **no** sustituye al error en unidades |
| Siempre | **RMSE o MAE en las unidades de $y$** y la comparación con el **baseline** | Es lo que el usuario del modelo entiende |

**Cómo reportar bien:**

1. **En unidades del problema**: «el error típico es de ≈ 43 puntos de progresión de la enfermedad (MAE), con escala de respuesta de 25 a 346».
2. **Contra un baseline** (`DummyRegressor`): el modelo gana X % de RMSE.
3. **Con incertidumbre**: una sola partición da una estimación ruidosa; un **intervalo bootstrap** sobre los errores de prueba (o la variabilidad entre pliegues de validación cruzada, cuaderno 02) dice cuánto confiar en la cifra.
4. **El diagnóstico $\text{RMSE}/\text{MAE}$**: ≈ 1.25 sugiere errores ≈ gaussianos; mucho más alto delata **colas pesadas** o *outliers* que conviene inspeccionar.
5. **Misma métrica para entrenar y evaluar** cuando sea posible, o justificar la diferencia (entrenas con MSE por conveniencia; evalúas con MAE porque el negocio lo pide).

````python
# ============================================================
# 7. Reporte completo: unidades, baseline, intervalo bootstrap y diagnostico RMSE/MAE
# ============================================================
def bootstrap_ic(y, yh, metrica, B=2000, seed=SEED):
    """Intervalo percentil 95 % remuestreando (con reemplazo) los pares (y, yh) del conjunto de prueba."""
    g = np.random.default_rng(seed); n = len(y)
    vals = np.empty(B)
    for b in range(B):
        i = g.integers(0, n, n)
        vals[b] = metrica(y[i], yh[i])
    return np.percentile(vals, [2.5, 97.5])

filas = {}
for nombre in ["Baseline: media", "Regresión lineal", "Random Forest (100 árboles, prof. 4)"]:
    p = preds[nombre]
    lo, hi = bootstrap_ic(y_te, p, rmse)
    filas[nombre] = {"RMSE": rmse(y_te, p), "IC95 RMSE": f"[{lo:.1f}, {hi:.1f}]", "MAE": mae(y_te, p), "RMSE/MAE": rmse(y_te, p) / mae(y_te, p), "R²": r2(y_te, p)}
reporte = pd.DataFrame(filas).T
display(reporte)
for nombre in filas:                                                   # el estimador puntual cae dentro de su intervalo y se cumple MAE <= RMSE
    lo, hi = bootstrap_ic(y_te, preds[nombre], rmse)
    assert lo <= filas[nombre]["RMSE"] <= hi and filas[nombre]["RMSE/MAE"] >= 1

lo_l, hi_l = bootstrap_ic(y_te, preds["Regresión lineal"], rmse)
lo_b, hi_b = bootstrap_ic(y_te, preds["Baseline: media"], rmse)
print(f"Regresión lineal: RMSE = {filas['Regresión lineal']['RMSE']:.1f} (IC95 % ≈ [{lo_l:.1f}, {hi_l:.1f}]) vs. baseline {rmse(y_te, preds['Baseline: media']):.1f} (IC95 % ≈ [{lo_b:.1f}, {hi_b:.1f}]).")
print("Como los intervalos no se solapan, la mejora sobre el baseline es real; entre regresión lineal y bosque suelen solaparse: la diferencia no es concluyente con un solo conjunto de prueba.")
print(f"RMSE/MAE de la regresión lineal = {filas['Regresión lineal']['RMSE/MAE']:.2f} (≈ 1.25 → errores casi gaussianos, sin outliers graves).")

fig, ax = plt.subplots(figsize=(9, 4.6))
etq = list(filas)
puntos = [filas[k]["RMSE"] for k in etq]
ics = [bootstrap_ic(y_te, preds[k], rmse) for k in etq]
ax.errorbar(puntos, np.arange(len(etq)), xerr=[[p - c[0] for p, c in zip(puntos, ics)], [c[1] - p for p, c in zip(puntos, ics)]],
            fmt="o", color="#1e3a8a", ecolor="#2563eb", capsize=5, lw=2)
ax.set_yticks(np.arange(len(etq))); ax.set_yticklabels(etq); ax.invert_yaxis()
ax.set_xlabel("RMSE en prueba (unidades de la respuesta) con IC95 % bootstrap"); ax.set_title("Reportar el error con su incertidumbre", fontweight="bold")
plt.tight_layout()
plt.show()
````

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **MSE / RMSE** | Pérdida cuadrática: dominada por errores grandes; estima la **media**; RMSE vuelve a las unidades de $y$ |
| **MAE / MedAE** | Pérdida lineal: robusta; estima la **mediana**; $\text{MAE}\le\text{RMSE}\le\sqrt n\,\text{MAE}$ |
| **$R^2$** | $1-\text{SSE}/\text{SST}$; **negativo** si el modelo es peor que predecir la media (típico con sobreajuste o sesgo en prueba); no es $\operatorname{corr}^2$ fuera de entrenamiento |
| **Baseline** | `DummyRegressor`: el RMSE de la media en prueba da $R^2_{\text{base}}=-(\bar y_{\text{prueba}}-\bar y_{\text{ent}})^2/\operatorname{Var}(y_{\text{prueba}})\le0$; sin comparación, una métrica no significa nada |
| **MAPE / sMAPE** | Dividen entre $y$: **explotan** con ceros, son **asimétricas** y MAPE **sesga a subestimar** (óptimo $e^{\mu-\sigma^2}$ en log-normal) |
| **Outliers** | Un error de tamaño 100 multiplica el RMSE por más de 10 y la MedAE casi nada; **Huber** es el compromiso diferenciable |
| **Probabilidad** | MSE ↔ MLE gaussiana ($\hat\sigma=$ RMSE); MAE ↔ MLE de Laplace ($\hat b=$ MAE); RMSE/MAE $\approx1.25$ (gaussiano) vs. $1.41$ (Laplace) |
| **Reporte** | Unidades del problema + baseline + intervalo de incertidumbre (*bootstrap* o validación cruzada) |
| **Implementación** | Todas las métricas desde cero coinciden con `sklearn.metrics` y `scipy.special.huber` (verificadas con `assert`) |
