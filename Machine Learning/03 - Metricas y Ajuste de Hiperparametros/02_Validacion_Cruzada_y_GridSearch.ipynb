## 1. Parámetros, Hiperparámetros y por qué no se Ajustan con el *Test* 🎛️

### 1.1 Dos tipos de números

| | **Parámetros** | **Hiperparámetros** |
|---|---|---|
| Quién los fija | El **algoritmo de entrenamiento**, optimizando la pérdida sobre los datos | **Tú**, **antes** de entrenar |
| Ejemplos | Coeficientes $\beta$ de la regresión, umbrales de un árbol, pesos de una red | $k$ en $k$-NN, `C` y `gamma` de un SVM, `max_depth`, tasa de aprendizaje, $\lambda$ de regularización |
| Se aprenden de | El conjunto de **entrenamiento** | Una **estimación del error de generalización** (validación) |

Los hiperparámetros **no pueden elegirse minimizando el error de entrenamiento**: este siempre premia la mayor complejidad (con $k=1$ el $k$-NN memoriza y acierta el 100 % del entrenamiento; un árbol sin límite de profundidad también). Hace falta una estimación del error **sobre datos no vistos**.

### 1.2 ¿Por qué no usar el conjunto de prueba?

Si probamos muchas configuraciones y **nos quedamos con la que mejor puntúa en prueba**, esa puntuación ya no es una estimación honesta: es el **máximo de muchas variables aleatorias**, que es **optimista** aunque cada una sea insesgada (**sesgo de selección** o *winner's curse*). La prueba pierde su virtud: ser datos que **no influyeron en ninguna decisión**.

La solución es una **tercera partición** (entrenamiento / validación / prueba) o, mejor, **validación cruzada** para reemplazar la partición de validación. La prueba se usa **una sola vez**, al final.

Dos demostraciones: (a) el error de entrenamiento siempre elige el modelo más complejo; (b) con datos de **ruido puro** (no hay nada que aprender: la exactitud real es 50 %), elegir la mejor de 300 configuraciones sobre el conjunto de prueba «descubre» una exactitud claramente mayor.

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas matplotlib seaborn scikit-learn scipy

import time
import itertools
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_breast_cancer, make_classification, load_diabetes
from sklearn.model_selection import (train_test_split, cross_val_score, KFold, StratifiedKFold, LeaveOneOut,
                                     GridSearchCV, ParameterGrid)
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import make_scorer, fbeta_score, average_precision_score

warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (9, 5)
print("Entorno listo.")
````

````python
# ============================================================
# 1. (a) El error de entrenamiento siempre elige la maxima complejidad
#    (b) Seleccionar sobre el conjunto de prueba produce una estimacion optimista (ruido puro)
# ============================================================
bc = load_breast_cancer()
Xb, yb = bc.data, bc.target

# (a) k-NN: precision en entrenamiento vs. validacion cruzada segun k
ks = list(range(1, 41, 2))
acc_tr, acc_cv = [], []
for k in ks:
    pipe = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=k))
    acc_tr.append(pipe.fit(Xb, yb).score(Xb, yb))
    acc_cv.append(cross_val_score(pipe, Xb, yb, cv=StratifiedKFold(5, shuffle=True, random_state=SEED)).mean())
assert ks[int(np.argmax(acc_tr))] == 1 and acc_tr[0] == 1.0                      # entrenamiento elige k = 1 (memoriza)
k_cv = ks[int(np.argmax(acc_cv))]

# (b) Etiquetas independientes de X: ninguna configuracion puede superar el 50 % en promedio
rng = np.random.default_rng(SEED)
X_ruido = rng.normal(size=(400, 30)); y_ruido = rng.integers(0, 2, 400)
X_tr, X_rest, y_tr, y_rest = train_test_split(X_ruido, y_ruido, test_size=0.5, random_state=SEED)
X_ta, X_tb, y_ta, y_tb = train_test_split(X_rest, y_rest, test_size=0.5, random_state=SEED)   # "prueba A" (se usa para elegir) y "prueba B" (datos frescos)
acc_A, acc_B = [], []
for _ in range(300):                                                              # 300 "configuraciones": subconjuntos aleatorios de 5 variables
    cols = rng.choice(30, size=5, replace=False)
    m = LogisticRegression().fit(X_tr[:, cols], y_tr)
    acc_A.append(m.score(X_ta[:, cols], y_ta)); acc_B.append(m.score(X_tb[:, cols], y_tb))
acc_A, acc_B = np.array(acc_A), np.array(acc_B)
mejor = int(np.argmax(acc_A))
assert acc_A[mejor] > 0.58 and abs(acc_A.mean() - 0.5) < 0.06                    # el maximo de A es optimista; el promedio ronda el azar (+-0.05 por n = 100)
assert acc_B[mejor] < acc_A[mejor] - 0.05                                          # en datos frescos la "ganadora" cae hacia 0.5

fig, axes = plt.subplots(1, 2, figsize=(15, 4.8))
axes[0].plot(ks, acc_tr, "o-", color="#dc2626", label="Entrenamiento")
axes[0].plot(ks, acc_cv, "s-", color="#2563eb", label="Validación cruzada (5 pliegues)")
axes[0].axvline(k_cv, color="#16a34a", linestyle=":", label=f"Mejor $k$ por validación: {k_cv}")
axes[0].set_xlabel("Vecinos $k$ (menor $k$ = más complejo)"); axes[0].set_ylabel("Accuracy")
axes[0].set_title("El entrenamiento siempre prefiere $k=1$", fontweight="bold"); axes[0].legend(fontsize=8)
axes[1].hist(acc_A, bins=18, color="#9ca3af", edgecolor="white", label="300 configuraciones en 'prueba A'")
axes[1].axvline(0.5, color="black", linestyle="--", label="Exactitud real (azar) = 0.5")
axes[1].axvline(acc_A[mejor], color="#dc2626", lw=2, label=f"Mejor en prueba A = {acc_A[mejor]:.2f}")
axes[1].axvline(acc_B[mejor], color="#16a34a", lw=2, label=f"Esa misma en datos frescos = {acc_B[mejor]:.2f}")
axes[1].set_xlabel("Accuracy sobre 100 ejemplos"); axes[1].set_ylabel("Número de configuraciones")
axes[1].set_title("Elegir sobre la prueba: el máximo es optimista", fontweight="bold"); axes[1].legend(fontsize=8)
plt.tight_layout()
plt.show()
print(f"(a) El entrenamiento elige k = 1 (accuracy {acc_tr[0]:.2f}, ¡memoriza!); la validación cruzada elige k = {k_cv} (accuracy {max(acc_cv):.3f}).")
print(f"(b) Con datos de ruido puro: promedio de las 300 configs = {acc_A.mean():.3f}; la 'mejor' en prueba A = {acc_A[mejor]:.2f}; esa misma en datos frescos = {acc_B[mejor]:.2f}.")
print("    Moraleja: la prueba solo sirve si NINGUNA decisión se tomó mirándola.")
````

## 2. *Hold-out* vs. *k-fold*: Validación Cruzada desde Cero 🔄

### 2.1 Definiciones

- ***Hold-out***: una sola partición (p. ej., 70/30). Barato, pero la estimación depende de **qué ejemplos cayeron en cada lado** (varianza alta) y **desperdicia** datos para entrenar.
- ***k-fold* (validación cruzada de $k$ pliegues)**: se parte el conjunto en $k$ pliegues $F_1,\dots,F_k$ disjuntos; para $j=1,\dots,k$ se entrena con los otros $k-1$ y se evalúa en $F_j$. Todo ejemplo se usa **exactamente una vez** para evaluar y $k-1$ para entrenar:

$$\widehat{\text{Err}}_{\text{CV}} = \frac1n\sum_{j=1}^{k}\sum_{i\in F_j} L\big(y_i,\;\hat f^{(-F_j)}(\mathbf{x}_i)\big),$$

donde $\hat f^{(-F_j)}$ es el modelo entrenado **sin** el pliegue $F_j$ (con pliegues del mismo tamaño es el promedio de las $k$ métricas por pliegue). Cuesta $k$ entrenamientos.
- ***Stratified k-fold***: igual, pero cada pliegue **conserva la proporción de clases** del total. Imprescindible con clases desbalanceadas (de lo contrario algún pliegue puede quedar sin ejemplos positivos).
- ***Leave-one-out* (LOO)**: $k=n$, un solo ejemplo por pliegue. Casi insesgado, pero cuesta $n$ entrenamientos.

Otras variantes, que **no** se desarrollan aquí: *repeated k-fold* (repite con distintas particiones aleatorias), `GroupKFold` (ejemplos de un mismo grupo —paciente, cliente— deben quedar juntos) y `TimeSeriesSplit` (series temporales: nunca entrenar con el futuro).

### 2.2 Implementación y verificación

Las construimos desde cero. `scikit-learn` asigna los tamaños de pliegue así: con $n = qk + r$, los primeros $r$ pliegues tienen $q+1$ elementos y el resto $q$. Con `shuffle=True` baraja los índices con `RandomState(random_state)`, algo que podemos reproducir **exactamente**. Para `StratifiedKFold` replicamos su reparto determinista por clase (los $k$ pliegues reciben la misma cantidad de cada clase, salvo $\pm 1$).

````python
# ============================================================
# 2. KFold y StratifiedKFold DESDE CERO, verificados contra scikit-learn
# ============================================================
def kfold_desde_cero(n, k, shuffle=False, seed=None):
    """Genera (indices_entrenamiento, indices_prueba). Primeros n % k pliegues con un elemento extra."""
    idx = np.arange(n)
    if shuffle:
        np.random.RandomState(seed).shuffle(idx)                                  # mismo mecanismo que scikit-learn
    tam = np.full(k, n // k); tam[: n % k] += 1
    inicio = 0
    for t in tam:
        prueba = idx[inicio: inicio + t]
        ent = np.concatenate([idx[:inicio], idx[inicio + t:]])
        yield np.sort(ent), np.sort(prueba)
        inicio += t

def stratified_kfold_desde_cero(y, k, shuffle=False, seed=None):
    """Asigna a cada ejemplo un pliegue: cada clase se reparte lo mas parejo posible entre los k pliegues."""
    y = np.asarray(y)
    _, primer, inv = np.unique(y, return_index=True, return_inverse=True)
    _, perm = np.unique(primer, return_inverse=True)
    y_cod = perm[inv]                                                              # clases codificadas por orden de aparicion
    n_clases = y_cod.max() + 1
    y_orden = np.sort(y_cod)
    reparto = np.array([np.bincount(y_orden[i::k], minlength=n_clases) for i in range(k)])   # reparto[j, c] = ejemplos de la clase c en el pliegue j
    rng = np.random.RandomState(seed) if shuffle else None
    pliegue = np.empty(len(y), dtype=int)
    for c in range(n_clases):
        asign = np.arange(k).repeat(reparto[:, c])
        if shuffle: rng.shuffle(asign)
        pliegue[y_cod == c] = asign
    for j in range(k):
        yield np.where(pliegue != j)[0], np.where(pliegue == j)[0]

# --- Verificacion de KFold (con y sin barajar, y con n no divisible entre k) ---
for n, k, sh in [(10, 3, False), (103, 5, False), (103, 5, True), (50, 10, True)]:
    X_fake = np.zeros((n, 1))
    mios = list(kfold_desde_cero(n, k, shuffle=sh, seed=SEED))
    sk = list(KFold(k, shuffle=sh, random_state=SEED if sh else None).split(X_fake))
    assert len(mios) == len(sk) == k
    for (a1, b1), (a2, b2) in zip(mios, sk):
        assert np.array_equal(a1, a2) and np.array_equal(b1, b2), (n, k, sh)
    # propiedad: los pliegues de prueba son una particion de {0..n-1}
    assert np.array_equal(np.sort(np.concatenate([b for _, b in mios])), np.arange(n))
print("KFold desde cero == sklearn.KFold (con y sin barajar; n no divisible entre k).")

# --- Verificacion de StratifiedKFold ---
y_des = np.r_[np.zeros(180, int), np.ones(20, int)]                               # 10 % de positivos, ORDENADOS (peor caso para KFold sin barajar)
for sh in (False, True):
    mios = list(stratified_kfold_desde_cero(y_des, 5, shuffle=sh, seed=SEED))
    sk = list(StratifiedKFold(5, shuffle=sh, random_state=SEED if sh else None).split(np.zeros((200, 1)), y_des))
    for (a1, b1), (a2, b2) in zip(mios, sk):
        assert np.array_equal(np.bincount(y_des[b1]), np.bincount(y_des[b2]))        # mismo numero de cada clase por pliegue
        assert len(a1) == len(a2) and np.array_equal(np.sort(np.concatenate([a1, b1])), np.arange(200))
        if not sh:
            assert np.array_equal(b1, b2) and np.array_equal(a1, a2)                  # sin barajar: pliegues IDENTICOS
print("StratifiedKFold desde cero == sklearn.StratifiedKFold (pliegues idénticos sin barajar; mismas proporciones al barajar).\n")

# --- Por que estratificar: positivos por pliegue con y sin estratificacion ---
tabla = pd.DataFrame({"KFold sin barajar": [y_des[b].sum() for _, b in kfold_desde_cero(200, 5)],
                      "KFold barajado": [y_des[b].sum() for _, b in kfold_desde_cero(200, 5, True, SEED)],
                      "Estratificado": [y_des[b].sum() for _, b in stratified_kfold_desde_cero(y_des, 5)]},
                     index=[f"pliegue {j + 1}" for j in range(5)])
display(tabla)
assert tabla["KFold sin barajar"].min() == 0 and tabla["Estratificado"].tolist() == [4] * 5
fig, ax = plt.subplots(figsize=(9, 4.2))
tabla.plot.bar(ax=ax, color=["#dc2626", "#f59e0b", "#16a34a"], edgecolor="white")
ax.axhline(4, color="black", linestyle=":", label="Reparto ideal (4 positivos)")
ax.set_ylabel("Positivos en el pliegue de prueba (de 20)"); ax.set_title("Sin estratificar, un pliegue puede no tener positivos", fontweight="bold")
ax.legend(fontsize=8); plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
print("Con los datos ordenados por clase, KFold sin barajar deja pliegues sin ningún positivo (métricas indefinidas); barajar ayuda pero solo la estratificación garantiza 4 por pliegue.")
````

````python
# ============================================================
# 2b. Leave-one-out desde cero y su atajo cerrado para regresion lineal (PRESS)
# ============================================================
# Regresion lineal en Diabetes (n = 442): LOO "a la fuerza" = 442 ajustes; atajo: e_loo_i = e_i / (1 - h_ii), con h_ii de la matriz sombrero
dia = load_diabetes()
Xd, yd = dia.data, dia.target
n = len(yd)

def loo_a_la_fuerza(X, y):
    errores = np.empty(len(y))
    for i in range(len(y)):
        mask = np.arange(len(y)) != i                                              # deja el ejemplo i fuera
        m = LinearRegression().fit(X[mask], y[mask])
        errores[i] = y[i] - m.predict(X[i:i + 1])[0]
    return errores

def loo_atajo(X, y):
    A = np.c_[np.ones(len(y)), X]                                                  # matriz de diseno con intercepto
    H = A @ np.linalg.pinv(A)                                                      # matriz sombrero H = A (A'A)^-1 A'
    residuos = y - H @ y
    return residuos / (1 - np.diag(H))                                             # error de dejar-uno-fuera

t0 = time.perf_counter(); e_fuerza = loo_a_la_fuerza(Xd, yd); t_fuerza = time.perf_counter() - t0
t0 = time.perf_counter(); e_atajo = loo_atajo(Xd, yd); t_atajo = time.perf_counter() - t0
assert np.allclose(e_fuerza, e_atajo, atol=1e-6)                                   # identicos
mse_loo = np.mean(e_atajo ** 2)
# Contra la API de scikit-learn
neg_mse_sk = cross_val_score(LinearRegression(), Xd, yd, cv=LeaveOneOut(), scoring="neg_mean_squared_error")
assert np.isclose(-neg_mse_sk.mean(), mse_loo)
assert len(list(LeaveOneOut().split(Xd))) == n == len(list(kfold_desde_cero(n, n)))      # LOO = k-fold con k = n
print(f"MSE leave-one-out = {mse_loo:.1f} (RMSE = {np.sqrt(mse_loo):.1f}): fuerza bruta == atajo cerrado == LeaveOneOut de sklearn.")
print(f"Tiempo: fuerza bruta ({n} ajustes) = {t_fuerza:.2f} s | atajo (1 ajuste) = {t_atajo * 1000:.1f} ms. "
      "Para modelos lineales, LOO es casi gratis; para el resto cuesta n entrenamientos.")
````

## 3. Sesgo y Varianza del Estimador de Validación Cruzada según $k$ 📊

La validación cruzada **estima** el error de generalización; como toda estimación tiene **sesgo** y **varianza**, y ambos dependen de $k$:

- **Sesgo.** Cada modelo de la validación cruzada se entrena con solo $\frac{k-1}{k}n$ ejemplos. Si la **curva de aprendizaje** todavía baja con más datos, el modelo con menos datos es **peor** que el modelo final y la validación cruzada **sobreestima el error** (sesgo pesimista). El sesgo **baja al aumentar $k$**; con LOO es casi nulo.
- **Varianza.** Al aumentar $k$ los conjuntos de entrenamiento de distintos pliegues **se solapan cada vez más** (en LOO difieren en un solo ejemplo), así que los $k$ modelos y sus errores están **muy correlacionados**, y promediar variables correlacionadas reduce poco la varianza (recuerda el $\rho\sigma^2$ de los [bosques](../02%20-%20Arboles%20Bosques%20y%20Redes%20Neuronales/01_Random_Forests.ipynb)). Por eso suele decirse que **LOO tiene varianza alta**, aunque el efecto depende del modelo (más marcado para modelos inestables; ver Hastie *et al.*, Cap. 7, y Arlot & Celisse, 2010).
- **Regla práctica:** $k=5$ o $k=10$ son un buen compromiso habitual entre sesgo, varianza y costo computacional (Kohavi, 1995; Hastie *et al.*). El efecto sobre la varianza es una **tendencia que depende del modelo y del tamaño de muestra**, no una ley: lo comprobamos a continuación.

Para verlo **sin teoría**, simulamos: generamos **muchos conjuntos de entrenamiento** de tamaño $n=60$ desde un mismo problema, calculamos el estimador de validación cruzada para distintos $k$ y comparamos con el **error verdadero** del modelo entrenado con los 60 ejemplos (evaluado en una población grande).

````python
# ============================================================
# 3. Simulacion: sesgo y desviacion del estimador CV segun k (k-NN, n = 60)
# ============================================================
N_ENT, REPS = 60, 30
X_pob, y_pob = make_classification(n_samples=20000, n_features=8, n_informative=4, n_redundant=0, n_clusters_per_class=2,
                                   class_sep=0.9, flip_y=0.05, random_state=SEED)
X_eval, y_eval = X_pob[:5000], y_pob[:5000]                                        # "poblacion" para medir el error verdadero
X_fuente, y_fuente = X_pob[5000:], y_pob[5000:]
ks_cv = [2, 3, 5, 10, 20, 60]                                                       # 60 = leave-one-out
modelo = lambda: KNeighborsClassifier(n_neighbors=3)

rng = np.random.default_rng(SEED)
err_verdadero, cv_est = [], {k: [] for k in ks_cv}
t0 = time.perf_counter()
for r in range(REPS):
    i = rng.choice(len(y_fuente), N_ENT, replace=False)
    Xr, yr = X_fuente[i], y_fuente[i]
    err_verdadero.append(1 - modelo().fit(Xr, yr).score(X_eval, y_eval))           # error del modelo final (entrenado con los 60)
    for k in ks_cv:
        errs = []
        for a, b in KFold(k, shuffle=True, random_state=r).split(Xr):
            errs.append(np.sum(modelo().fit(Xr[a], yr[a]).predict(Xr[b]) != yr[b]))
        cv_est[k].append(np.sum(errs) / N_ENT)                                       # tasa de error CV = errores totales / n
t_sim = time.perf_counter() - t0

objetivo = np.mean(err_verdadero)
res = pd.DataFrame({"sesgo (CV − error verdadero)": [np.mean(cv_est[k]) - objetivo for k in ks_cv],
                    "desv. estándar del estimador": [np.std(cv_est[k]) for k in ks_cv],
                    "error CV medio": [np.mean(cv_est[k]) for k in ks_cv]}, index=pd.Index(ks_cv, name="k"))
res["RMSE respecto al error verdadero"] = [np.sqrt(np.mean((np.array(cv_est[k]) - np.array(err_verdadero)) ** 2)) for k in ks_cv]
display(res.round(4))
print(f"Error verdadero medio del modelo entrenado con {N_ENT} ejemplos = {objetivo:.4f} | simulación: {REPS} conjuntos × {sum(ks_cv)} ajustes en {t_sim:.1f} s")
assert res["sesgo (CV − error verdadero)"].iloc[0] > res["sesgo (CV − error verdadero)"].iloc[-1]      # k = 2 mas pesimista que LOO

fig, axes = plt.subplots(1, 2, figsize=(14, 4.8))
axes[0].plot(ks_cv, res["sesgo (CV − error verdadero)"], "o-", color="#dc2626", lw=2)
axes[0].axhline(0, color="black", lw=0.8); axes[0].set_xscale("log")
axes[0].set_xticks(ks_cv); axes[0].set_xticklabels(ks_cv); axes[0].minorticks_off()
axes[0].set_xlabel("Número de pliegues $k$ (escala log; 60 = LOO)"); axes[0].set_ylabel("Sesgo del estimador de error")
axes[0].set_title("Sesgo pesimista: baja al crecer $k$", fontweight="bold")
axes[1].plot(ks_cv, res["desv. estándar del estimador"], "s-", color="#2563eb", lw=2, label="Desviación estándar del estimador CV")
axes[1].plot(ks_cv, res["RMSE respecto al error verdadero"], "^--", color="#16a34a", lw=2, label="RMSE respecto al error verdadero")
axes[1].set_xscale("log"); axes[1].set_xticks(ks_cv); axes[1].set_xticklabels(ks_cv); axes[1].minorticks_off()
axes[1].set_xlabel("Número de pliegues $k$ (escala log; 60 = LOO)"); axes[1].set_ylabel("Variabilidad del estimador")
axes[1].set_title("Variabilidad del estimador según $k$", fontweight="bold"); axes[1].legend(fontsize=8)
plt.tight_layout()
plt.show()
print("Lectura: con k = 2 el modelo se entrena con la mitad de los datos y el error se sobreestima; el sesgo baja al crecer k y se estabiliza hacia k = 10.")
print("En esta simulación (k-NN, un modelo bastante estable) la dispersión NO empeora con k grande: la fama de 'LOO tiene varianza alta' es una tendencia que depende del modelo y del problema.")
print("Lo robusto: a partir de k ≈ 5-10 ya no hay ganancia apreciable de sesgo ni de RMSE, y el costo (k entrenamientos) sigue creciendo: de ahí el compromiso habitual k = 5 o 10.")
````

## 4. Fuga de Datos dentro de la Validación Cruzada y el `Pipeline` 🚰

La validación cruzada solo es honesta si **cada pliegue de validación se comporta como datos nuevos**. Un error muy común: **ajustar un paso de preprocesamiento con TODOS los datos antes** de hacer la validación cruzada. Cualquier paso que **aprende de los datos** —escalado, imputación, selección de variables, PCA, codificación por objetivo, sobremuestreo— **debe aprenderse solo con el pliegue de entrenamiento**; si se aprende con todo, la información de los pliegues de validación **se filtra** al entrenamiento (*data leakage*) y la estimación sale **optimista**.

El caso más dramático es la **selección de variables**: si eliges las 20 variables más correlacionadas con $y$ **usando todos los datos** y luego validas, esas variables ya «vieron» las etiquetas de los pliegues de validación. Hastie, Tibshirani & Friedman (Cap. 7) lo presentan como **la forma incorrecta de hacer validación cruzada** *(sección citada de memoria)*.

La **solución** es envolver todos los pasos en un `Pipeline`: `cross_val_score` y `GridSearchCV` llaman a `fit` del `Pipeline` **solo con los datos de entrenamiento de cada pliegue** y aplican `transform` al de validación.

Lo demostramos con **ruido puro**: 60 ejemplos, 2 000 variables aleatorias y etiquetas aleatorias. La exactitud verdadera es 50 %.

````python
# ============================================================
# 4. Fuga de datos: seleccion de variables ANTES de la CV (mal) vs. DENTRO del Pipeline (bien)
# ============================================================
def experimento(seed):
    g = np.random.default_rng(seed)
    X = g.normal(size=(60, 2000)); y = g.integers(0, 2, 60)                            # ruido puro: no hay nada que aprender
    cv = StratifiedKFold(5, shuffle=True, random_state=seed)
    # MAL: seleccionar con TODOS los datos y luego validar
    X_sel = SelectKBest(f_classif, k=20).fit_transform(X, y)                           # <- fuga: usa las etiquetas de todos los pliegues
    mal = cross_val_score(LogisticRegression(), X_sel, y, cv=cv).mean()
    # BIEN: la seleccion ocurre dentro de cada pliegue de entrenamiento
    pipe = Pipeline([("sel", SelectKBest(f_classif, k=20)), ("clf", LogisticRegression())])
    bien = cross_val_score(pipe, X, y, cv=cv).mean()
    return mal, bien

resultados = np.array([experimento(s) for s in range(30)])                              # 30 repeticiones del experimento
mal, bien = resultados[:, 0], resultados[:, 1]
display(pd.DataFrame({"selección ANTES de la CV (fuga)": [mal.mean(), mal.std()], "selección DENTRO del Pipeline": [bien.mean(), bien.std()]},
                     index=["accuracy CV media", "desv. estándar (30 repeticiones)"]).round(3))
assert mal.mean() > 0.70 and abs(bien.mean() - 0.5) < 0.06                              # la fuga "descubre" señal inexistente

fig, ax = plt.subplots(figsize=(9, 4.6))
ax.boxplot([mal, bien], widths=0.5, patch_artist=True, boxprops=dict(facecolor="#bfdbfe"), medianprops=dict(color="#dc2626", lw=2))
ax.set_xticks([1, 2]); ax.set_xticklabels(["Selección ANTES de la CV\n(fuga de datos)", "Selección DENTRO\ndel Pipeline"])
ax.axhline(0.5, color="black", linestyle="--", label="Exactitud real (azar) = 0.5")
ax.set_ylabel("Accuracy estimada por validación cruzada"); ax.legend(fontsize=9)
ax.set_title("Ruido puro: la fuga de datos fabrica un 'modelo' que no existe", fontweight="bold")
plt.tight_layout()
plt.show()
print(f"Con la fuga, la validación cruzada estima ≈ {mal.mean():.0%} de accuracy sobre etiquetas completamente aleatorias; con Pipeline, {bien.mean():.0%} (el valor correcto).")
print("Regla: TODO lo que se ajusta con datos (escalado, imputación, selección, PCA...) va DENTRO del Pipeline.")
````

## 5. GridSearch: Producto Cartesiano × Validación Cruzada 🔎

**Búsqueda en rejilla** (*grid search*): se define un conjunto finito de valores para cada hiperparámetro y se prueban **todas las combinaciones** (producto cartesiano). Para cada combinación $\theta$ se estima el error por validación cruzada y se elige la mejor:

$$\hat\theta = \arg\min_{\theta\in\Theta_1\times\cdots\times\Theta_d}\ \widehat{\text{Err}}_{\text{CV}}(\theta), \qquad |\Theta| = \prod_{i=1}^{d} |\Theta_i|.$$

**Algoritmo:**

1. Separar una **prueba final** que no se toca.
2. Para cada $\theta\in\Theta$ y cada pliegue $j$: entrenar con los otros pliegues y puntuar en $F_j$. Promediar los $k$ puntajes.
3. Elegir el $\theta$ de mejor puntaje medio (empates: el primero de la rejilla).
4. **Reentrenar** (*refit*) con ese $\theta$ sobre **todo** el entrenamiento.
5. Evaluar **una sola vez** en la prueba.

Lo implementamos con `itertools.product` y nuestro `stratified_kfold_desde_cero`, sobre un `Pipeline` `StandardScaler` + `SVC` (núcleo RBF) con dos hiperparámetros: `C` (penalización de errores) y `gamma` (ancho del núcleo). Después lo comparamos con `GridSearchCV` **dándole exactamente los mismos pliegues**: `best_score_`, `best_params_` y **cada una** de las puntuaciones medias deben coincidir.

````python
# ============================================================
# 5. GridSearch DESDE CERO verificado contra GridSearchCV
# ============================================================
from sklearn.base import clone

def grid_search_desde_cero(estimador, rejilla, X, y, pliegues, scoring_fn):
    """Devuelve (mejores_params, mejor_score, tabla de resultados). 'pliegues' es una lista de (ent, val)."""
    claves = sorted(rejilla)                                                          # mismo orden que ParameterGrid: claves alfabeticas, la ultima varia mas rapido
    filas, lista = [], []
    for valores in itertools.product(*(rejilla[c] for c in claves)):                  # producto cartesiano
        params = dict(zip(claves, valores))
        puntajes = []
        for ent, val in pliegues:                                                     # validacion cruzada con CADA combinacion
            m = clone(estimador).set_params(**params).fit(X[ent], y[ent])
            puntajes.append(scoring_fn(m, X[val], y[val]))
        lista.append(params)
        filas.append({**params, "puntajes": puntajes, "media": np.mean(puntajes), "std": np.std(puntajes)})
    tabla = pd.DataFrame(filas)
    mejor = int(tabla["media"].values.argmax())                                       # argmax devuelve el PRIMERO en empates (igual que sklearn)
    return lista[mejor], tabla.loc[mejor, "media"], tabla

# Datos: Breast Cancer, con particion de prueba final
X_ent, X_pru, y_ent, y_pru = train_test_split(Xb, yb, test_size=0.25, stratify=yb, random_state=SEED)
pipe = Pipeline([("escala", StandardScaler()), ("svm", SVC(kernel="rbf"))])
rejilla = {"svm__C": [0.1, 1, 10, 100], "svm__gamma": [0.0001, 0.001, 0.01, 0.1, 1]}
pliegues = list(stratified_kfold_desde_cero(y_ent, 5))                                # NUESTROS pliegues (identicos a StratifiedKFold(5) sin barajar)
exactitud = lambda m, X, y: m.score(X, y)

t0 = time.perf_counter()
mejores, mejor_score, tabla_mia = grid_search_desde_cero(pipe, rejilla, X_ent, y_ent, pliegues, exactitud)
t_mio = time.perf_counter() - t0

gs = GridSearchCV(pipe, rejilla, cv=pliegues, scoring="accuracy", refit=True).fit(X_ent, y_ent)   # mismos pliegues
res_sk = pd.DataFrame(gs.cv_results_)

# Verificaciones
assert [dict(zip(sorted(rejilla), [r[c] for c in sorted(rejilla)])) for _, r in tabla_mia.iterrows()] == list(ParameterGrid(rejilla))   # mismas combinaciones, mismo orden
assert np.allclose(tabla_mia["media"].values, res_sk["mean_test_score"].values)                      # TODAS las medias coinciden
assert np.allclose(tabla_mia["std"].values, res_sk["std_test_score"].values)
assert np.isclose(mejor_score, gs.best_score_) and mejores == gs.best_params_
print(f"GridSearch desde cero == GridSearchCV: best_score_ = {mejor_score:.4f}, best_params_ = {mejores}")
print(f"Combinaciones = {len(tabla_mia)} × 5 pliegues = {len(tabla_mia) * 5} ajustes (+1 de reajuste final) | tiempo propio = {t_mio:.2f} s")

# Reajuste y evaluacion unica en prueba
final = clone(pipe).set_params(**mejores).fit(X_ent, y_ent)
assert np.isclose(final.score(X_pru, y_pru), gs.score(X_pru, y_pru))                                  # refit=True: best_estimator_ es el reajustado
print(f"Accuracy en validación cruzada (mejor combinación) = {mejor_score:.4f} | accuracy en la PRUEBA (una sola vez) = {final.score(X_pru, y_pru):.4f}")
print("Nota: el puntaje CV de la mejor combinación es ligeramente optimista (es el máximo sobre 20 candidatas); la prueba da la estimación honesta.")
````

````python
# ============================================================
# 5b. Analisis de cv_results_ con pandas y mapa de calor C x gamma
# ============================================================
res = res_sk[["param_svm__C", "param_svm__gamma", "mean_test_score", "std_test_score", "rank_test_score"]].copy()
res.columns = ["C", "gamma", "media CV", "desv. CV", "ranking"]
res[["C", "gamma"]] = res[["C", "gamma"]].astype(float)
display(res.sort_values("ranking").head(8).round(4).reset_index(drop=True))

media = res.pivot_table(index="C", columns="gamma", values="media CV")
desv = res.pivot_table(index="C", columns="gamma", values="desv. CV")
assert media.shape == (4, 5) and np.isclose(media.values.max(), gs.best_score_)

fig, axes = plt.subplots(1, 2, figsize=(15, 4.8))
sns.heatmap(media, annot=True, fmt=".3f", cmap="viridis", ax=axes[0], cbar_kws={"label": "Accuracy medio (CV)"})
axes[0].set_title("Accuracy de validación cruzada: `C` × `gamma`", fontweight="bold")
sns.heatmap(desv, annot=True, fmt=".3f", cmap="magma_r", ax=axes[1], cbar_kws={"label": "Desviación estándar entre pliegues"})
axes[1].set_title("Incertidumbre entre pliegues", fontweight="bold")
for ax in axes: ax.set_xlabel("gamma"); ax.set_ylabel("C"); ax.invert_yaxis()
plt.tight_layout()
plt.show()

# Regla de "una desviacion estandar": la combinacion de menor C (mas regularizada) estadisticamente indistinguible de la mejor
mejor_fila = res.loc[res["ranking"].idxmin()]
umbral = mejor_fila["media CV"] - mejor_fila["desv. CV"]
candidatas = res[res["media CV"] >= umbral].sort_values(["C", "gamma"])
print(f"Mejor: C = {mejor_fila['C']:g}, gamma = {mejor_fila['gamma']:g} (CV = {mejor_fila['media CV']:.4f} ± {mejor_fila['desv. CV']:.4f}).")
print(f"{len(candidatas)} combinaciones están a menos de una desviación estándar de la mejor; "
      f"la de menor `C` (modelo más regularizado, más simple) es C = {candidatas.iloc[0]['C']:g}, gamma = {candidatas.iloc[0]['gamma']:g}: criterio de parsimonia o «regla de una desviación estándar» (Hastie et al., Cap. 7, citado de memoria).")
print("Lectura del mapa: hay una 'cordillera' diagonal. gamma grande con C grande sobreajusta (memoriza); gamma muy pequeño subajusta. Conviene mirar la zona, no solo el máximo.")
````

## 6. `scoring` Personalizado y `refit` 🎯

Por defecto `GridSearchCV` maximiza el `score` del estimador (*accuracy* en clasificación, $R^2$ en regresión). Pero ya vimos en el [cuaderno 00](00_Metricas_de_Clasificacion.ipynb) que **el accuracy engaña con clases desbalanceadas**: la métrica con la que buscamos **determina qué modelo gana**. Con `scoring` puedes pasar:

- un **nombre** (`"f1"`, `"average_precision"`, `"neg_root_mean_squared_error"`...). Por convención, `scikit-learn` **maximiza** siempre, por eso las pérdidas llevan el prefijo `neg_`;
- un **objeto** construido con `make_scorer(funcion, **parametros)` (p. ej., `fbeta_score` con `beta=2`) o cualquier *callable* `f(estimador, X, y)`;
- un **diccionario** con **varias métricas** a la vez; entonces `refit` debe indicar **cuál** decide el mejor modelo (o ser `False`). `refit=True` (por defecto, con una sola métrica) **reentrena el mejor modelo con todo el entrenamiento** y lo deja en `best_estimator_`.

Lo vemos en un problema de fraude desbalanceado (≈ 5 % de positivos) con un bosque ligero y tres hiperparámetros (profundidad, tamaño de hoja y `class_weight`): evaluamos tres métricas a la vez y comprobamos que **distintas métricas eligen distintos hiperparámetros**.

````python
# ============================================================
# 6. scoring multiple y refit sobre datos desbalanceados (RandomForest ligero)
# ============================================================
Xf, yf = make_classification(n_samples=3000, n_features=12, n_informative=5, n_redundant=2, n_clusters_per_class=2,
                             weights=[0.95, 0.05], class_sep=1.0, flip_y=0.01, random_state=SEED)
Xf_tr, Xf_te, yf_tr, yf_te = train_test_split(Xf, yf, test_size=0.25, stratify=yf, random_state=SEED)
print(f"Positivos: {yf.mean():.1%} (entrenamiento: {yf_tr.sum()} fraudes de {len(yf_tr)})")

rejilla_rf = {"max_depth": [4, 8, None], "min_samples_leaf": [1, 10], "class_weight": [None, "balanced"]}
scoring = {"accuracy": "accuracy",
           "AP": "average_precision",                                          # PR-AUC (independiente del umbral)
           "F2": make_scorer(fbeta_score, beta=2)}                              # scorer personalizado: recall pesa mas que precision
t0 = time.perf_counter()
gs_rf = GridSearchCV(RandomForestClassifier(n_estimators=25, random_state=SEED, n_jobs=1), rejilla_rf,
                     cv=StratifiedKFold(4, shuffle=True, random_state=SEED), scoring=scoring, refit="AP").fit(Xf_tr, yf_tr)
t_rf = time.perf_counter() - t0
r = pd.DataFrame(gs_rf.cv_results_)
r["config"] = [f"prof={p['max_depth']}, hoja={p['min_samples_leaf']}, pesos={p['class_weight']}" for p in r["params"]]
tab = r.set_index("config")[["mean_test_accuracy", "mean_test_AP", "mean_test_F2"]]
tab.columns = ["accuracy", "AP (PR-AUC)", "F2"]
display(tab.round(3))

mejor_por = {m: tab[m].idxmax() for m in tab.columns}
print("Mejor configuración según cada métrica:")
for m, c in mejor_por.items(): print(f"  {m:12s} -> {c}  ({tab[m].max():.3f})")

# Comprobaciones: refit='AP' reentrena el mejor segun AP; best_score_ es el AP medio de CV
assert gs_rf.best_params_ == dict(r.loc[r["mean_test_AP"].idxmax(), "params"]) and np.isclose(gs_rf.best_score_, r["mean_test_AP"].max())
assert gs_rf.best_estimator_.get_params()["max_depth"] == gs_rf.best_params_["max_depth"]
# Un scorer es una funcion f(estimador, X, y); make_scorer(fbeta_score, beta=2) equivale a esto:
f2_a_mano = lambda est, X, y: fbeta_score(y, est.predict(X), beta=2)
assert np.isclose(f2_a_mano(gs_rf.best_estimator_, Xf_te, yf_te), scoring["F2"](gs_rf.best_estimator_, Xf_te, yf_te))
print(f"\nrefit='AP': best_estimator_ reentrenado con todo el entrenamiento. En prueba: AP = {average_precision_score(yf_te, gs_rf.predict_proba(Xf_te)[:, 1]):.3f}, "
      f"F2 = {f2_a_mano(gs_rf.best_estimator_, Xf_te, yf_te):.3f}, accuracy = {gs_rf.best_estimator_.score(Xf_te, yf_te):.3f}   (tiempo de la búsqueda: {t_rf:.1f} s)")
assert "balanced" in mejor_por["F2"] and "None" in mejor_por["accuracy"].split("pesos=")[1]            # accuracy y F2 eligen distinto
print("Lectura: el accuracy (todas ≈ 90-96 %, el piso del trivial es 94.3 %) prefiere NO reponderar las clases, mientras que F2 prefiere class_weight='balanced'.")
print("La métrica con la que buscas decide qué modelo gana: define primero la métrica según el costo de los errores (cuaderno 00) y solo después ajusta.")
````

## 7. Costo Computacional: la Explosión Combinatoria 💥

El número de **ajustes** de una búsqueda en rejilla es

$$N_{\text{ajustes}} = \underbrace{\prod_{i=1}^{d}|\Theta_i|}_{\text{combinaciones}} \times \underbrace{k}_{\text{pliegues}} \;(+\,1\ \text{reajuste final}),$$

**exponencial en el número $d$ de hiperparámetros**: con 5 valores por hiperparámetro y 5 pliegues, $d=2$ cuesta 125 ajustes, $d=5$ ya son 15 625 y $d=8$ pasan de **1.9 millones**. Y la rejilla además **desperdicia presupuesto**: si solo 1–2 hiperparámetros importan de verdad, probar 5 valores de cada uno de los demás repite el mismo valor útil muchas veces. Estimamos el costo real midiendo el tiempo de un ajuste.

**Alternativas** (Módulo 07, *no desarrolladas aquí*):

- ***Random Search*** (`RandomizedSearchCV`): muestrea combinaciones al azar con presupuesto fijo; suele superar a la rejilla con el mismo costo cuando pocos hiperparámetros importan (Bergstra & Bengio, 2012, *citado de memoria, verificar*). Géron, Cap. 2, lo presenta junto a *Grid Search*.
- ***Nested cross-validation*** (validación cruzada anidada): el puntaje de la mejor combinación es **optimista** porque se eligió **mirándolo** (Varma & Simon, 2006); el bucle externo estima honestamente el rendimiento de *todo el procedimiento* de selección (Raschka, Cap. 6).
- ***Optimización bayesiana*** (p. ej., Optuna): usa los resultados previos para decidir qué combinación probar a continuación.
- ***Successive halving***: descarta pronto las combinaciones malas con poco presupuesto.

````python
# ============================================================
# 7. Explosion combinatoria: numero de ajustes y tiempo estimado
# ============================================================
K = 5
def n_ajustes(d, valores=5, k=K): return valores ** d * k + 1

# Tiempo real de un ajuste (SVC en Breast Cancer, y bosque de 50 arboles en datos de fraude) promediando varios ajustes
def tiempo_ajuste(m, X, y, reps=5):
    t0 = time.perf_counter()
    for _ in range(reps): clone(m).fit(X, y)
    return (time.perf_counter() - t0) / reps
t_svm = tiempo_ajuste(pipe, X_ent, y_ent)
t_rf1 = tiempo_ajuste(RandomForestClassifier(n_estimators=50, random_state=SEED), Xf_tr, yf_tr, reps=3)

d = np.arange(1, 9)
aj = np.array([n_ajustes(i) for i in d])
tab = pd.DataFrame({"hiperparámetros d": d, "combinaciones (5^d)": 5 ** d, "ajustes (×5 pliegues + 1)": aj,
                    "tiempo con SVC": [f"{a * t_svm / 3600:.2f} h" if a * t_svm > 3600 else f"{a * t_svm:.0f} s" for a in aj],
                    "tiempo con bosque de 50 árboles": [f"{a * t_rf1 / 3600:.1f} h" if a * t_rf1 > 3600 else f"{a * t_rf1:.0f} s" for a in aj]}).set_index("hiperparámetros d")
display(tab)
assert n_ajustes(2) == 126 and n_ajustes(5) == 15626 and n_ajustes(8) > 1_900_000
assert len(gs.cv_results_["params"]) * 5 == 100                                   # la rejilla de la Seccion 5: 4 x 5 combinaciones x 5 pliegues
print(f"La rejilla de la Sección 5 tuvo {len(gs.cv_results_['params'])} combinaciones × 5 pliegues = {len(gs.cv_results_['params']) * 5} ajustes (+1 de reajuste).")
print(f"Tiempo medido por ajuste: SVC = {t_svm * 1000:.1f} ms; bosque de 50 árboles = {t_rf1 * 1000:.0f} ms.")

fig, ax = plt.subplots(figsize=(9, 4.8))
ax.semilogy(d, aj, "o-", color="#dc2626", lw=2, label="Ajustes totales ($5^d\\times5+1$)")
ax.semilogy(d, 5 ** d, "s--", color="#2563eb", lw=2, label="Combinaciones ($5^d$)")
ax.axhline(1e6, color="gray", linestyle=":", label="Un millón de ajustes")
ax.set_xlabel("Número de hiperparámetros $d$ (5 valores cada uno, 5 pliegues)"); ax.set_ylabel("Número de ajustes (escala log)")
ax.set_title("Explosión combinatoria de la búsqueda en rejilla", fontweight="bold"); ax.legend(fontsize=9)
plt.tight_layout()
plt.show()
print("Con 8 hiperparámetros la rejilla exhaustiva es inviable: ahí entran Random Search, optimización bayesiana y halving (Módulo 07).")
````

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Parámetros vs. hiperparámetros** | Los parámetros se aprenden del entrenamiento; los hiperparámetros los fijas tú y se eligen con una **estimación de generalización** (nunca con el error de entrenamiento) |
| **Prueba intocable** | Elegir sobre la prueba produce un máximo **optimista** (sesgo de selección); la prueba se usa **una sola vez**, al final |
| ***Hold-out* vs. *k-fold*** | $k$-fold usa todos los datos para evaluar y entrenar; `Stratified` conserva proporciones de clase; LOO es $k=n$ (con atajo cerrado PRESS en regresión lineal) |
| **Sesgo / varianza de CV** | $k$ pequeño → sesgo pesimista; $k$ grande → sesgo bajo pero pliegues muy correlacionados y mayor costo; la ganancia se estabiliza hacia $k=5$–$10$ (el efecto sobre la varianza depende del modelo) |
| **Fuga de datos** | Todo paso que aprende de los datos (escalado, selección, PCA…) va **dentro del `Pipeline`**; hacerlo antes fabrica rendimiento inexistente (ruido puro → ≈ 90 % en vez de 50 %) |
| **GridSearch** | Producto cartesiano × validación cruzada; reajuste con el mejor $\theta$; nuestra implementación coincide con `GridSearchCV` (puntajes, `best_params_`) |
| **`cv_results_`** | Analízalo con `pandas`: ranking, desviaciones y mapa de calor; mira la **región**, no solo el máximo (regla de una desviación estándar) |
| **`scoring` y `refit`** | La métrica decide qué modelo gana; con varias métricas, `refit='nombre'` indica cuál reajusta `best_estimator_` |
| **Costo** | $\prod|\Theta_i|\times k$: **exponencial en $d$**; alternativas (Random Search, validación anidada, optimización bayesiana) en el Módulo 07 |
