## 1. ¿Cuánto me equivoco, en promedio? 📏

Un amigo agente inmobiliario predice el precio de 6 apartamentos en Tunja (en millones de pesos). Para cada uno, el **error** es la **distancia** entre lo que predijo y lo que realmente costó:

- Si predijo 200 y costó 220, se equivocó en **20 millones**.
- Si predijo 300 y costó 290, se equivocó en **10 millones**.

El **error promedio** (MAE, por *Mean Absolute Error*) es simplemente el **promedio de esas distancias**. Su gran ventaja: viene **en las mismas unidades** que lo que predices: *«en promedio me equivoco por X millones»*. Se entiende sin ser experto.

````python
# Error promedio de un agente inmobiliario (precios en millones de pesos)
import numpy as np

real = np.array([220, 290, 150, 310, 180, 400])          # lo que realmente costaron
predicho = np.array([200, 300, 160, 290, 190, 380])      # lo que predijo el agente

errores = np.abs(real - predicho)                         # distancia de cada predicción a la realidad
print("Errores por apartamento (millones):", errores)
print(f"Error promedio: {errores.mean():.1f} millones de pesos")
print("\nLectura sencilla: 'en promedio, mi predicción se aleja unos 15 millones del precio real'.")
````

## 2. Los fallos grandes pesan más: el profesor exigente 🧑‍🏫

Ahora dos agentes con **el mismo error promedio**:

- **Agente A:** se equivoca **10 millones** en todos los apartamentos. Constante y predecible.
- **Agente B:** acierta casi todos, pero en uno se equivoca **por 50 millones**, un desastre.

Con el error promedio **parecen iguales**. Pero ¿querrías que B te venda tu casa? Existe otra nota, el **RMSE**, que funciona como un **profesor exigente**: castiga **mucho más los fallos grandes** que los pequeños (un error de 50 «duele» 25 veces más que uno de 10, no solo 5 veces). Si tu nota **RMSE es mucho mayor que el error promedio**, es señal de que hay **fallos graves escondidos**.

````python
# Dos agentes con el MISMO error promedio, pero uno tiene un fallo enorme
agente_a = np.array([10, 10, 10, 10, 10])                # siempre se equivoca 10
agente_b = np.array([0, 0, 0, 0, 50])                    # acierta casi todo y falla UNA vez por 50

for nombre, err in [("Agente A (constante)", agente_a), ("Agente B (un desastre)", agente_b)]:
    mae = err.mean()                                      # error promedio
    rmse = np.sqrt((err ** 2).mean())                     # el "profesor exigente": eleva al cuadrado, promedia, y saca la raiz
    print(f"{nombre:24s} error promedio = {mae:5.1f} | RMSE (exigente) = {rmse:5.1f}")

print("\nMismo error promedio (10), pero el RMSE del agente B es mucho mayor: delata el fallo grande.")
````

## 3. R²: ¿le gano al amigo que adivina el promedio? 🧪

Imagina un amigo **perezoso** que, sin mirar nada del apartamento, siempre dice: *«cuesta lo de siempre»* (el **precio promedio** de todos). Es la predicción más tonta posible, pero es una **referencia** excelente: **tu modelo debe ser mejor que ese amigo**.

El **R²** mide justo eso, en una escala fácil de leer:

- **R² = 1**: predicciones perfectas.
- **R² = 0**: tu modelo es **igual de bueno que el amigo perezoso** (no aporta nada).
- **R² negativo**: tu modelo es **peor que adivinar el promedio**. Suena imposible, pero pasa cuando el modelo **memoriza** y luego falla con datos nuevos (como estudiar de memoria las respuestas de un simulacro y perder el examen real).

Veámoslo con datos reales de **pacientes con diabetes** (predecir un índice de progresión de la enfermedad): comparamos al amigo perezoso, un modelo razonable y un modelo que memoriza.

````python
# R2: modelo razonable vs. amigo perezoso (promedio) vs. modelo que memoriza
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error

datos = load_diabetes()
rng = np.random.default_rng(42)
idx = rng.permutation(len(datos.target))
ent, exa = idx[:60], idx[60:]                                            # pocos pacientes para estudiar, muchos para el examen

X = datos.data
X_ruido = np.c_[X, rng.normal(size=(len(X), 100))]                       # el modelo "memorioso" recibe ademas 100 columnas de puro ruido
y = datos.target

modelos = [("Amigo perezoso (siempre el promedio)", DummyRegressor(), X),
           ("Modelo razonable (regresión lineal)", LinearRegression(), X),
           ("Modelo que memoriza (muchas columnas de ruido)", LinearRegression(), X_ruido)]
for nombre, modelo, datos_x in modelos:
    modelo.fit(datos_x[ent], y[ent])
    pred = modelo.predict(datos_x[exa])
    print(f"{nombre:48s} R² en el examen = {r2_score(y[exa], pred):7.2f} | error promedio = {mean_absolute_error(y[exa], pred):6.1f}")

print("\nEl modelo que memoriza sacaría nota perfecta con los pacientes que ya vio, pero en el examen real queda PEOR que el amigo perezoso (R² negativo).")
````

## 4. La trampa de los porcentajes 🌡️

Decir «me equivoco un **5 %**» suena muy cómodo. Pero los porcentajes tienen **trampas**:

1. **Cuando lo real es cero (o casi cero).** Si hoy la temperatura real es **0 °C** y predije 1 °C, ¿en qué porcentaje me equivoqué? Un grado sobre cero es... **infinito por ciento**. Un solo dato así arruina todo el promedio.
2. **No es simétrico.** Si predije 150 y era 100, me equivoqué un 50 %. Si predije 100 y era 150, me equivoqué solo un 33 %. ¡La **misma distancia** (50) sale con **porcentajes distintos**!
3. **Empuja a predecir «por debajo».** Como las predicciones muy altas se castigan más que las bajas, un modelo que se entrena para minimizar el porcentaje tiende a **quedarse corto**.

Por eso, además del porcentaje, conviene dar siempre el **error en las unidades del problema** (millones, grados, minutos).

````python
# Trampas del error en porcentaje
real = 100
print(f"Predije 150 y era 100  -> error = {abs(150 - real) / real:.0%}")
print(f"Predije 100 y era 150  -> error = {abs(100 - 150) / 150:.0%}   (misma distancia de 50, ¡distinto porcentaje!)")

temperaturas_reales = np.array([0.001, 10.0, 20.0])                      # una de las temperaturas esta casi en cero
temperaturas_predichas = np.array([1.0, 11.0, 19.0])                     # en las tres me equivoco por ~1 grado
porcentajes = np.abs(temperaturas_reales - temperaturas_predichas) / temperaturas_reales
print("\nErrores en grados:", np.abs(temperaturas_reales - temperaturas_predichas))
print("Errores en porcentaje:", [f"{p:.0%}" for p in porcentajes])
print(f"Promedio en porcentaje: {porcentajes.mean():.0%}  <- un solo dato casi en cero lo arruinó todo, aunque en grados el error es siempre ~1.")
````

## 5. ¿Cómo contarle bien el error a otras personas? 🗣️

Una buena forma de reportar tiene **tres partes**:

1. **En unidades que todos entiendan:** *«en promedio me equivoco por 15 millones de pesos»*, no «mi MSE es 340».
2. **Contra el amigo perezoso:** *«el amigo que adivina el promedio se equivoca por 60 millones; mi modelo, por 15»*. Sin esa comparación, un número no dice nada.
3. **Con humildad:** una sola prueba es una foto; con otros datos el número cambiaría un poco. Repetir la medición varias veces (lo que hace la *validación cruzada*, que verás en el siguiente cuaderno) te da una idea de **cuánto puede variar**.

> 💡 **Regla de oro:** un error de «20» puede ser **excelente** (si lo normal varía entre 0 y 10 000) o **terrible** (si varía entre 0 y 30). Siempre da el contexto.

## Resumen en una frase 🎯

> **Para medir cuánto se equivoca un modelo que predice números se usa el error promedio (en unidades del problema), el RMSE si quieres castigar los fallos grandes y el R² para saber si le ganas al amigo que adivina el promedio; los porcentajes tienen trampas y todo número se reporta con su contexto.**

### Lo que aprendiste hoy:
- ✅ El **error promedio** se lee directo en las unidades del problema.
- ✅ El **RMSE** castiga más los **fallos grandes**; si es mucho mayor que el error promedio, hay desastres escondidos.
- ✅ El **R²** compara con el amigo perezoso; si es **negativo**, el modelo es peor que adivinar el promedio.
- ✅ Los **porcentajes** fallan con valores reales cercanos a cero y no son simétricos.
- ✅ Reporta siempre **unidades, comparación y humildad**.
