## 1. La alarma de humo y los cuatro resultados 🔔

Imagina una **alarma de humo**. Cada día pueden pasar cuatro cosas:

| | **Suena la alarma** | **No suena** |
|---|---|---|
| **Sí hay incendio** | ✅ **Acierto** (detectó el fuego) | ❌ **Fallo grave**: ¡no avisó! (*falso negativo*) |
| **No hay incendio** | ❌ **Falsa alarma**: sonó por el tostador (*falso positivo*) | ✅ **Acierto** (todo tranquilo) |

Cualquier modelo que responde «sí / no» (¿es spam?, ¿está enfermo?, ¿es fraude?) tiene **esos mismos cuatro resultados**. Todas las notas que veremos salen de **contar** cuántas veces ocurrió cada uno. Contémoslos para una alarma imaginaria durante 1 000 días.

````python
# Una alarma de humo durante 1000 dias: contamos los cuatro resultados
import numpy as np

# 1000 dias: 20 con incendio real (los demas, tranquilos). Armamos lo que paso con cada dia.
hubo_incendio = np.array([1] * 20 + [0] * 980)
sono_alarma = np.array([1] * 16 + [0] * 4 + [1] * 30 + [0] * 950)       # detecto 16 de 20 incendios; dio 30 falsas alarmas

aciertos_fuego = np.sum((hubo_incendio == 1) & (sono_alarma == 1))
fallos_graves = np.sum((hubo_incendio == 1) & (sono_alarma == 0))
falsas_alarmas = np.sum((hubo_incendio == 0) & (sono_alarma == 1))
aciertos_calma = np.sum((hubo_incendio == 0) & (sono_alarma == 0))

print(f"Incendio y sonó la alarma (acierto)      : {aciertos_fuego}")
print(f"Incendio y NO sonó (fallo grave)         : {fallos_graves}")
print(f"Sin incendio y sonó (falsa alarma)       : {falsas_alarmas}")
print(f"Sin incendio y no sonó (acierto, calma)  : {aciertos_calma}")
print(f"Total de días                            : {aciertos_fuego + fallos_graves + falsas_alarmas + aciertos_calma}")
````

---
## 2. La trampa del 99 % ⚠️

Mira la alarma de arriba: acertó casi todos los días. Si cuentas **todos los aciertos** y divides entre los días, sale un número altísimo. A eso se le llama **exactitud** (*accuracy*). Pero hay un problema: **casi todos los días son tranquilos**, así que acertar esos días es muy fácil.

Ejemplo extremo: una «alarma» **rota**, que **nunca suena**, tendría **98 % de exactitud** en nuestros 1 000 días (porque solo 20 días tuvieron fuego)... ¡y no sirve para nada! Lo comprobamos.

````python
# Exactitud de la alarma real vs. una alarma ROTA que nunca suena
alarma_rota = np.zeros(1000, dtype=int)                                  # nunca suena

exactitud_real = np.mean(sono_alarma == hubo_incendio)
exactitud_rota = np.mean(alarma_rota == hubo_incendio)
incendios_detectados_real = np.sum((sono_alarma == 1) & (hubo_incendio == 1)) / 20
incendios_detectados_rota = np.sum((alarma_rota == 1) & (hubo_incendio == 1)) / 20

print(f"Alarma real : exactitud = {exactitud_real:.1%} | incendios detectados = {incendios_detectados_real:.0%}")
print(f"Alarma rota : exactitud = {exactitud_rota:.1%} | incendios detectados = {incendios_detectados_rota:.0%}")
print("\n¡La alarma rota tiene MÁS exactitud que la real, y no detecta NINGÚN incendio! Con pocos casos raros, la exactitud sola engaña.")
````

## 3. Precisión y cobertura: la red de pesca 🎣

Para no caer en la trampa, hacemos **dos preguntas distintas**. Piensa en un pescador que lanza una red al mar:

- **Cobertura** (*recall*): *de todos los peces que había en el mar, ¿cuántos atrapé?* Si el mar tenía 100 peces y saqué 80, mi cobertura es del 80 %.
- **Precisión** (*precision*): *de todo lo que saqué en la red, ¿cuánto eran peces de verdad?* Si saqué 100 cosas pero solo 80 eran peces (el resto, botas y latas), mi precisión es del 80 %.

Y aquí viene el **truco**: una red de **malla muy ancha** atrapa casi todos los peces (¡alta cobertura!) pero también mucha basura (baja precisión). Una red de **malla muy fina y exigente** saca solo peces seguros (alta precisión) pero se le escapan muchos (baja cobertura). **No se puede tener todo.**

En un modelo, esa «malla» es el **umbral**: *¿cuán seguro debe estar el modelo para decir «sí»?* Probémoslo con el conjunto de datos de **tumores** (positivo = tumor maligno): cambiamos qué tan exigente es el modelo y miramos cómo se mueven las dos notas.

````python
# Precision y cobertura segun que tan exigente (umbral) sea el modelo
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

datos = load_breast_cancer()
y_maligno = 1 - datos.target                                              # 1 = tumor maligno (el "pez" que queremos encontrar)
X_ent, X_exa, y_ent, y_exa = train_test_split(datos.data, y_maligno, test_size=0.3, stratify=y_maligno, random_state=42)
modelo = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)).fit(X_ent, y_ent)
seguridad = modelo.predict_proba(X_exa)[:, 1]                             # que tan seguro esta el modelo de que es maligno (0 a 1)

filas = []
for umbral in [0.05, 0.2, 0.5, 0.8, 0.95]:
    dice_si = seguridad >= umbral                                          # solo dice "maligno" si esta al menos tan seguro como el umbral
    pescados = np.sum(dice_si & (y_exa == 1))                              # peces atrapados
    precision = pescados / max(np.sum(dice_si), 1)                         # de lo que saque, cuanto era pez
    cobertura = pescados / np.sum(y_exa == 1)                              # de los peces que habia, cuantos saque
    filas.append({"exigencia (umbral)": umbral, "precisión": round(precision, 3), "cobertura": round(cobertura, 3), "alarmas dadas": int(dice_si.sum())})
print(pd.DataFrame(filas).to_string(index=False))

umbrales = np.linspace(0.02, 0.98, 97)
prec = [np.sum((seguridad >= u) & (y_exa == 1)) / max(np.sum(seguridad >= u), 1) for u in umbrales]
cob = [np.sum((seguridad >= u) & (y_exa == 1)) / np.sum(y_exa == 1) for u in umbrales]
plt.figure(figsize=(9, 4.8))
plt.plot(umbrales, prec, color="#2563eb", lw=2, label="Precisión (de lo que saqué, ¿cuánto era pez?)")
plt.plot(umbrales, cob, color="#f59e0b", lw=2, label="Cobertura (de los peces, ¿cuántos saqué?)")
plt.xlabel("Exigencia del modelo (umbral): más a la derecha = más exigente"); plt.ylabel("Valor entre 0 y 1")
plt.title("Más exigente: sube la precisión, baja la cobertura", fontweight="bold")
plt.legend(loc="lower center")
plt.show()
print("\nA mayor exigencia, el modelo acierta más cuando dice 'maligno', pero se le escapan más tumores malignos.")
````

## 4. F1: una nota que no se deja engañar 🎓

¿Hay una sola nota que junte precisión y cobertura? Sí: el **F1**. Podrías pensar: «sacamos el promedio de las dos». Pero fíjate en esta trampa con las **notas de un estudiante**:

- Matemáticas: **5.0**. Español: **0.1**. El promedio simple da **2.55**, ¡casi «regular»!, aunque en una materia está perdido.

El F1 funciona como un **promedio exigente**: si **una de las dos notas es muy baja, el F1 también lo es**. Para tener un F1 alto hay que ser bueno en **ambas**. Es como un equipo de dos remeros: la canoa solo avanza bien si **los dos** reman fuerte.

````python
# Promedio simple vs. F1 (promedio exigente) para varios casos de (precision, cobertura)
casos = [("Equilibrado", 0.80, 0.80), ("Precisión perfecta, casi no encuentra nada", 1.00, 0.01),
         ("Encuentra todo, pero casi todo es falsa alarma", 0.05, 1.00), ("Bueno en ambas", 0.95, 0.90)]
print(f"{'caso':50s} {'promedio simple':>16s} {'F1':>6s}")
for nombre, p, r in casos:
    simple = (p + r) / 2
    f1 = 2 * p * r / (p + r)
    print(f"{nombre:50s} {simple:16.2f} {f1:6.2f}")
print("\nEl promedio simple califica con ~0.5 a modelos que en realidad son malos; el F1 los castiga (~0.02 - 0.10).")
````

## 5. ¿Cuál nota me importa más? 🧭

No hay una «mejor nota» para todo. Depende de **qué error cuesta más caro**:

| Situación | Error más grave | Nota que importa más |
|---|---|---|
| **Detector de humo / pruebas médicas de cribado** | **No avisar** de un incendio o una enfermedad real | **Cobertura** (que no se escape ninguno) |
| **Filtro de correo basura (spam)** | Mandar a la basura un **correo importante** | **Precisión** (que lo marcado como spam sea spam) |
| **Antifraude de tarjetas** | Dejar pasar un fraude (cuesta dinero), pero bloquear mucho molesta al cliente | **Cobertura** con un límite de falsas alarmas |
| **Problema parejo y errores igual de graves** | Cualquiera | Exactitud o F1 |

La buena noticia: como el **umbral** (la exigencia) se puede mover **sin volver a entrenar**, puedes ajustar el modelo a lo que más le importa a tu problema.

> 💡 **Regla de oro:** antes de mirar números, pregúntate: *¿qué me duele más: un falso negativo o un falso positivo?* La respuesta te dice qué nota mirar.

## Resumen en una frase 🎯

> **Para evaluar un modelo de «sí / no» se cuentan los cuatro resultados posibles; la exactitud puede engañar cuando lo que buscas es raro, así que se mira la precisión (¿cuánto de lo que marqué era cierto?) y la cobertura (¿cuánto de lo cierto encontré?), se resumen con el F1 y se elige cuál priorizar según qué error cuesta más.**

### Lo que aprendiste hoy:
- ✅ Hay **cuatro resultados**: aciertos de ambos tipos, falsas alarmas y fallos graves.
- ✅ Un modelo que **nunca avisa** puede tener 98-99 % de exactitud y ser inútil.
- ✅ **Precisión** y **cobertura** se compensan al cambiar la exigencia (el umbral).
- ✅ El **F1** exige ser bueno en ambas a la vez.
- ✅ La nota a priorizar depende de **qué error cuesta más**.
