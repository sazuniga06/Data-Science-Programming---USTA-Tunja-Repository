## 3. Introducción al Machine Learning (1.3) 🤖

### 3.1 ¿Qué es "aprender"? La definición de Mitchell

Una de las definiciones más citadas de la disciplina se atribuye a **Arthur Samuel (1959)**, pionero del aprendizaje en juegos (su programa de damas en IBM): *el ML es el campo de estudio que da a las computadoras la capacidad de aprender sin ser programadas explícitamente*. Es una frase inspiradora pero imprecisa. La definición **operativa** y verificable es la de **Tom Mitchell (1997)**:

> 📌 *"Se dice que un programa de computador aprende de la experiencia $E$ respecto a una clase de tareas $T$ y una medida de desempeño $P$, si su desempeño en las tareas de $T$, medido por $P$, mejora con la experiencia $E$."*

Formalmente, para un conjunto de tareas $T$ y un programa con parámetros $\theta$, aprender significa que, a medida que aumenta la experiencia $E$ (datos), la medida $P(\theta_E; T)$ **mejora**. Esto convierte "aprender" en algo **medible**: si $P$ no mejora con más datos, el sistema no está aprendiendo.

| Sistema | Tarea $T$ | Medida de desempeño $P$ | Experiencia $E$ |
|---|---|---|---|
| Filtro de spam | Marcar correos nuevos como *spam* o *ham* | Proporción de correos bien clasificados | Correos marcados previamente por usuarios |
| Programa de damas (Samuel) | Jugar damas | Porcentaje de partidas ganadas | Partidas jugadas contra sí mismo |
| Estimador de precios de vivienda | Predecir el precio de un inmueble | Error absoluto medio (en pesos) | Ventas históricas con sus características |
| Segmentación de clientes | Agrupar clientes con comportamiento similar | Cohesión/separación de grupos (p. ej. *silhouette*) | Historial de compras sin etiquetas |

> 🔎 Observa que la **medida $P$ debe elegirse antes de entrenar** y que debe medirse sobre datos que el modelo **no usó para aprender** (lo retomaremos en la Sección 7 y en el Cuaderno 03).

### 3.2 ML vs. programación tradicional

En la programación clásica un humano **escribe las reglas**; en ML el humano entrega **datos y respuestas** y el algoritmo **descubre las reglas**:



El ML es especialmente útil cuando (i) las reglas son demasiado complejas o desconocidas (reconocer una cara), (ii) las reglas cambian con el tiempo (spam) o (iii) se necesita adaptarse a grandes volúmenes de datos. Veamos la diferencia con un experimento mínimo: la conversión de grados Celsius a Fahrenheit.

````python
# ============================================================
# Configuracion inicial del cuaderno
# (en Google Colab ya vienen instaladas estas librerias)
# ============================================================
# !pip install -q numpy pandas matplotlib seaborn scikit-learn

import warnings
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn

warnings.filterwarnings("ignore")

SEED = 42                       # semilla fija para reproducibilidad
np.random.seed(SEED)

sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (9, 5)

print("Entorno listo.")
print(f"numpy {np.__version__} | pandas {pd.__version__} | scikit-learn {sklearn.__version__}")
````

````python
# ============================================================
# Programacion tradicional vs Machine Learning
# ============================================================
from sklearn.linear_model import LinearRegression

# --- Enfoque 1: programacion tradicional (el humano conoce y escribe la regla) ---
def celsius_a_fahrenheit(c):
    return 1.8 * c + 32

# --- Enfoque 2: Machine Learning (el algoritmo descubre la regla desde ejemplos) ---
rng = np.random.default_rng(SEED)
celsius = rng.uniform(-10, 40, size=60).reshape(-1, 1)          # entrada X (matriz 2D)
fahrenheit = celsius_a_fahrenheit(celsius).ravel() + rng.normal(0, 0.5, size=60)  # respuestas y con ruido de medicion

modelo = LinearRegression().fit(celsius, fahrenheit)           # "aprender" = ajustar parametros

print("Regla real (escrita por humano): F = 1.8 * C + 32")
print(f"Regla aprendida (desde 60 ejemplos): F = {modelo.coef_[0]:.3f} * C + {modelo.intercept_:.3f}")

# Comparamos ambas predicciones para 25 C
x_nuevo = np.array([[25.0]])
print(f"\nPrediccion para 25 C -> humano: {celsius_a_fahrenheit(25):.2f} F | modelo: {modelo.predict(x_nuevo)[0]:.2f} F")
````

---
> 💡 **Lectura del experimento:** nunca le dijimos al modelo la fórmula; la **recuperó de los datos** (con pequeñas diferencias por el ruido). Si mañana cambiara la relación (otra escala de temperatura), bastaría **reentrenar** con nuevos datos en lugar de reescribir el programa. Esa es la esencia de $E \rightarrow$ mejora de $P$.

### 3.3 ML vs. Data Mining vs. Inteligencia Artificial

Los tres términos se usan a veces como sinónimos, pero tienen orígenes y énfasis distintos:



| Aspecto | Inteligencia Artificial | Machine Learning | Data Mining (KDD) |
|---|---|---|---|
| **Pregunta central** | ¿Cómo construir agentes que actúen con inteligencia? | ¿Cómo mejorar automáticamente el desempeño a partir de datos? | ¿Qué patrones válidos y útiles existen en estos datos? |
| **Énfasis** | Razonamiento, planificación, percepción, lenguaje | **Generalización**: predecir bien sobre datos *nuevos* | **Descubrimiento** y comprensión de patrones en datos *existentes* |
| **Salida típica** | Un agente o sistema inteligente | Un modelo predictivo o descriptivo | Conocimiento interpretable para decidir |
| **Comunidad de origen** | Informática, lógica, ciencias cognitivas | IA y estadística | Bases de datos y estadística |
| **Relación** | Disciplina más amplia | **Subcampo de la IA** | Se **solapa** con ML: usa sus algoritmos dentro del proceso KDD |

Conviene recordar que **no todo sistema de IA aprende** (los sistemas expertos de los años 80 se basaban en reglas escritas por humanos) y que **no todo uso de ML es Minería de Datos** (un robot que aprende a caminar no está "minando" una base de datos).

## 5. El Ciclo de Vida de un Proyecto de ML 🔄

Un modelo es apenas **una pieza** de un proyecto de ML. El flujo completo es iterativo: se retrocede a etapas previas cada vez que la evaluación no satisface el objetivo, y un modelo desplegado debe **monitorearse** porque el mundo cambia (*data drift*).



| Etapa | Pregunta guía | Herramienta típica | Equivalente CRISP-DM |
|---|---|---|---|
| 1. Definir | ¿Qué decisión mejora el modelo? ¿Cómo mediré el éxito? | Reunión con el negocio | Business Understanding |
| 2. Datos | ¿Qué datos existen y con qué calidad? | `pandas`, SQL | Data Understanding |
| 3. Explorar/limpiar | ¿Faltantes, outliers, errores? | `pandas`, `seaborn` | Data Understanding |
| 4. Preparar | ¿Cómo codifico y escalo las variables? | `ColumnTransformer` | Data Preparation |
| 5. Entrenar | ¿Qué familia de modelos pruebo? | `scikit-learn` | Modeling |
| 6. Evaluar | ¿Generaliza a datos nuevos? | `train_test_split`, validación cruzada, métricas | Evaluation |
| 7. Desplegar | ¿Cómo lo consume el usuario? | API, batch | Deployment |
| 8. Monitorear | ¿Se degrada con el tiempo? | Métricas en producción | (Fuera del ciclo clásico) |

> 🔗 Las metodologías formales (KDD, CRISP-DM y SEMMA) se estudian a fondo en [Data Mining — Cuaderno 01](../../Data%20Mining/00%20-%20Introduccion%20al%20Data%20Mining/01_Metodologias_Industriales_CRISP_DM_y_SEMMA.ipynb). Aquí nos quedamos con la versión de ML.

## 6. Ejemplo Mínimo *End-to-End* 🧪

Recorramos **todo el ciclo en pocas líneas** con el dataset *Breast Cancer Wisconsin (Diagnostic)* incluido en `scikit-learn` (569 tumores descritos por 30 medidas obtenidas de imágenes de biopsia; la variable objetivo indica si el tumor es **maligno** o **benigno**).

- **Tarea $T$:** clasificar un tumor como maligno o benigno.
- **Experiencia $E$:** casos con diagnóstico confirmado.
- **Desempeño $P$:** *accuracy* (proporción de aciertos) y, de forma más cuidadosa, la matriz de confusión.

### 6.1 Obtener y explorar los datos (etapas 2 y 3)

````python
# ============================================================
# Etapas 2 y 3: cargar y explorar los datos
# ============================================================
from sklearn.datasets import load_breast_cancer

datos = load_breast_cancer(as_frame=True)
X = datos.data                      # 30 variables numericas
y = datos.target                    # 0 = maligno, 1 = benigno (convencion de scikit-learn)
nombres_clase = {0: "maligno", 1: "benigno"}

print(f"Muestras: {X.shape[0]} | Variables: {X.shape[1]}")
print(f"Valores faltantes en total: {int(X.isna().sum().sum())}")
print("\nDistribucion de la variable objetivo:")
print(y.map(nombres_clase).value_counts().to_string())

display(X.iloc[:, :6].describe().T[["mean", "std", "min", "max"]].round(2))

# Grafico de balance de clases
ax = y.map(nombres_clase).value_counts().plot(kind="bar", color=["#2563eb", "#dc2626"], rot=0)
ax.set_title("Balance de clases en Breast Cancer Wisconsin", fontweight="bold")
ax.set_xlabel("Diagnóstico")
ax.set_ylabel("Número de pacientes")
for p in ax.patches:
    ax.annotate(int(p.get_height()), (p.get_x() + p.get_width() / 2, p.get_height()),
                ha="center", va="bottom", fontweight="bold")
plt.tight_layout()
plt.show()
````

---
### 6.2 Separar, entrenar y evaluar (etapas 4, 5 y 6)

Tres reglas de oro que aplicaremos desde ya:

1. **Separar antes de modelar:** reservamos un conjunto de **prueba** que el modelo nunca ve durante el entrenamiento; solo así $P$ mide la capacidad de generalizar.
2. **Tener una línea base (*baseline*):** un modelo trivial (predecir siempre la clase mayoritaria) nos dice cuánto vale realmente nuestro modelo.
3. **Preprocesar dentro de un `Pipeline`:** el escalado se ajusta solo con los datos de entrenamiento (la razón se profundiza en el Cuaderno 03).

````python
# ============================================================
# Etapas 4-6: separar, entrenar y evaluar
# ============================================================
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.dummy import DummyClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# 1) Separacion train/test estratificada (conserva la proporcion de clases)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=SEED
)
print(f"Entrenamiento: {X_train.shape[0]} filas | Prueba: {X_test.shape[0]} filas")

# 2) Linea base: siempre predice la clase mayoritaria
base = DummyClassifier(strategy="most_frequent").fit(X_train, y_train)
acc_base = accuracy_score(y_test, base.predict(X_test))

# 3) Modelo: escalado + regresion logistica dentro de un Pipeline
modelo = Pipeline([
    ("escalador", StandardScaler()),
    ("clasificador", LogisticRegression(max_iter=1000)),
])

# Validacion cruzada SOLO sobre el conjunto de entrenamiento (estimacion previa)
cv_scores = cross_val_score(modelo, X_train, y_train, cv=5, scoring="accuracy")
print(f"\nAccuracy en validacion cruzada (5 folds): {cv_scores.mean():.3f} +/- {cv_scores.std():.3f}")

# 4) Entrenamiento final y evaluacion en datos NO vistos
modelo.fit(X_train, y_train)
y_pred = modelo.predict(X_test)
acc_test = accuracy_score(y_test, y_pred)

print(f"Accuracy de la linea base (clase mayoritaria): {acc_base:.3f}")
print(f"Accuracy del modelo en prueba:                 {acc_test:.3f}")
print("\nReporte de clasificacion (conjunto de prueba):")
print(classification_report(y_test, y_pred, target_names=["maligno", "benigno"], digits=3))
````

````python
# ============================================================
# Visualizar el desempeno: matriz de confusion y variables mas influyentes
# ============================================================
from sklearn.metrics import ConfusionMatrixDisplay

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Panel 1: matriz de confusion
ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred, display_labels=["maligno", "benigno"], cmap="Blues", ax=axes[0], colorbar=False
)
axes[0].set_title("Matriz de confusión (conjunto de prueba)", fontweight="bold")
axes[0].set_xlabel("Clase predicha")
axes[0].set_ylabel("Clase real")
axes[0].grid(False)

# Panel 2: 10 coeficientes de mayor magnitud (variables estandarizadas => comparables)
coefs = pd.Series(modelo.named_steps["clasificador"].coef_[0], index=X.columns)
top = coefs.reindex(coefs.abs().sort_values(ascending=False).head(10).index).sort_values()
colores = ["#dc2626" if v < 0 else "#2563eb" for v in top.values]
axes[1].barh(top.index, top.values, color=colores)
axes[1].axvline(0, color="black", linewidth=0.8)
axes[1].set_title("Top 10 coeficientes del modelo\n(azul empuja a 'benigno', rojo a 'maligno')", fontweight="bold")
axes[1].set_xlabel("Coeficiente (variables estandarizadas)")

plt.tight_layout()
plt.show()
````

### 6.3 ¿Qué acabamos de hacer?

Con apenas unas líneas recorrimos el ciclo completo: **explorar → separar → establecer una línea base → entrenar → validar → evaluar en datos nuevos → interpretar**. Compara el *accuracy* del modelo con el de la línea base: esa diferencia es lo que el aprendizaje aportó. En medicina, además, no todos los errores cuestan lo mismo: un tumor maligno clasificado como benigno (falso negativo) es más grave que el error inverso. Por eso en el Módulo 01 estudiaremos métricas más ricas que el *accuracy*. Y una precaución sobre el gráfico de coeficientes: cuando las variables están muy correlacionadas entre sí (aquí lo están: radio, perímetro y área miden casi lo mismo), el signo y la magnitud de cada coeficiente individual deben interpretarse con cautela.

> ⚠️ **Advertencia:** este es un ejemplo didáctico. Un sistema clínico real exige validación externa, análisis de sesgos, aprobación regulatoria y supervisión médica.

from sklearn.metrics import confusion_matrix, recall_score
