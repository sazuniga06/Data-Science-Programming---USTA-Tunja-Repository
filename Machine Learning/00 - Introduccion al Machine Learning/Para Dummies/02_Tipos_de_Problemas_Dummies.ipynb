## 1. Los 4 oficios de la plaza de mercado 🏪

Imagina que el Machine Learning es un equipo de cuatro trabajadores en la plaza de mercado de Tunja:

| Oficio | Qué hace | Pregunta que responde | Ejemplo de Boyacá |
|---|---|---|---|
| **El Clasificador** 🏷️ | Mete cada cosa en una caja con nombre | "¿A cuál categoría pertenece?" (**Sí/No**, A/B/C) | ¿Esta papa es de primera o de segunda? |
| **El Adivino de Números** 🔢 | Calcula una cantidad exacta | "¿Cuánto?" | ¿Cuántas toneladas de papa saldrán de este lote? |
| **El Agrupador** 🧺 | Junta lo que se parece, sin que nadie le diga las categorías | "¿Qué tipos hay?" | ¿Qué tipos de clientes tiene la tienda? |
| **El Detective de Parejas** 🕵️ | Descubre qué cosas se compran juntas | "¿Qué va con qué?" | ¿Quién compra arepa suele llevar queso? |

## 2. La pregunta clave: ¿tengo respuestas anotadas? 🔑

Casi siempre basta con dos preguntas:

| Pregunta | Si la respuesta es... | Oficio |
|---|---|---|
| **1. ¿Tengo ejemplos pasados con la respuesta anotada?** | **Sí** → mira la forma de la respuesta | |
| ↳ ¿La respuesta es una **categoría** (sí/no, A/B/C)? | Sí | **Clasificador** 🏷️ |
| ↳ ¿La respuesta es un **número** (precio, cantidad)? | Sí | **Adivino de Números** 🔢 |
| **¿No tengo respuestas anotadas?** | ¿Quiero encontrar **grupos parecidos**? | **Agrupador** 🧺 |
| | ¿Quiero saber **qué cosas aparecen juntas**? | **Detective de Parejas** 🕵️ |

> 💡 Ojo con los nombres engañosos: la "regresión **logística**" suena a número, pero en realidad **clasifica**. Lo que manda es la **forma de la respuesta**, no el nombre del algoritmo.

Practiquemos con un pequeño juego: escribimos una función con esas dos preguntas y la probamos con varios casos.

````python
# JUEGO: ¿a que oficio llamo?
def elegir_oficio(tengo_respuestas, la_respuesta_es_un_numero=False, quiero_grupos=False):
    if tengo_respuestas:
        return "Adivino de Numeros" if la_respuesta_es_un_numero else "Clasificador"
    return "Agrupador" if quiero_grupos else "Detective de Parejas"

casos = [
    ("¿Este cliente pagará el crédito: sí o no?",          dict(tengo_respuestas=True)),
    ("¿Cuántos kilos de papa producirá este lote?",         dict(tengo_respuestas=True, la_respuesta_es_un_numero=True)),
    ("¿Qué tipos de clientes tiene mi tienda?",             dict(tengo_respuestas=False, quiero_grupos=True)),
    ("¿Qué productos se compran juntos en mi tienda?",      dict(tengo_respuestas=False)),
    ("¿Este correo es basura o normal?",                    dict(tengo_respuestas=True)),
    ("¿A qué precio se venderá este apartamento en Tunja?", dict(tengo_respuestas=True, la_respuesta_es_un_numero=True)),
]

for pregunta, datos in casos:
    print(f"{pregunta:<55} -> {elegir_oficio(**datos)}")
````

## 3. El Detective de Parejas en acción: el carrito de compras 🛒

Este oficio es el que menos suena conocido, así que veámoslo con calma. Tiene **tres medidas**, todas sencillas:

| Medida | Qué dice, en palabras simples | Ejemplo |
|---|---|---|
| **Soporte** | ¿En cuántas compras (de cada 100) aparece la combinación? | "En 40 de cada 100 compras hay arepa **y** queso" |
| **Confianza** | De los que compran A, ¿cuántos también compran B? | "De los que compran arepa, el 80 % lleva queso" |
| **Lift** | ¿Es una pareja de verdad o solo coincidencia porque B se compra mucho? | Si es **mayor que 1**, la pareja es real; si es cerca de 1, es pura casualidad |

Probémoslo con 10 compras de una tienda de barrio. ¡Tú puedes comprobar los números contando a mano!

````python
# DETECTIVE DE PAREJAS: soporte, confianza y lift contando a mano
compras = [
    {"arepa", "queso", "cafe"},
    {"arepa", "queso"},
    {"arepa", "queso", "huevos"},
    {"arepa", "huevos"},
    {"cafe", "panela"},
    {"cafe", "panela", "pan"},
    {"pan", "leche"},
    {"pan", "leche", "huevos"},
    {"arepa", "queso", "pan"},
    {"huevos", "leche"},
]
total = len(compras)

def contar(*productos):
    # en cuantas compras aparecen TODOS estos productos
    return sum(1 for c in compras if set(productos) <= c)

soporte = contar("arepa", "queso") / total                       # compras con arepa Y queso / total
confianza = contar("arepa", "queso") / contar("arepa")           # de las que llevan arepa, cuantas llevan queso
lift = confianza / (contar("queso") / total)                     # confianza comparada con lo normal del queso

print(f"Regla: quien compra AREPA tambien compra QUESO")
print(f"  Soporte:   {soporte * 100:.0f}% de todas las compras tienen arepa y queso")
print(f"  Confianza: {confianza * 100:.0f}% de quienes llevan arepa llevan queso")
print(f"  Lift:      {lift:.2f}  ->  {'pareja real (mayor que 1)' if lift > 1 else 'sin relacion especial'}")
````

## Resumen en una frase 🎯

> **Todo proyecto de datos cae casi siempre en uno de cuatro oficios: clasificar, predecir un número, agrupar o encontrar parejas; y se reconoce preguntando si hay respuestas anotadas y qué forma tienen.**

### Lo que aprendiste hoy:
- ✅ **Clasificador:** respuesta = categoría. **Adivino de Números:** respuesta = cantidad.
- ✅ **Agrupador:** junta lo parecido sin etiquetas. **Detective de Parejas:** descubre qué se compra junto.
- ✅ La pregunta clave: **¿tengo respuestas anotadas?** y **¿qué forma tienen?**
- ✅ En reglas de parejas: **soporte** (¿qué tan común?), **confianza** (¿qué tan seguro?) y **lift** (¿es pareja real?).
