## 1. Dos formas de cocinar: la receta vs. aprender probando 🍲

Imagina que quieres que un computador prepare el ajiaco perfecto.

| | **Programar (forma tradicional)** | **Machine Learning** |
|---|---|---|
| **¿Qué haces tú?** | Escribes la receta exacta: "2 papas, 1 mazorca, 20 minutos..." | Le muestras **cientos de ajiacos** ya preparados, cada uno con su calificación |
| **¿Quién descubre la receta?** | Tú | La máquina |
| **¿Funciona cuando la receta es muy difícil de escribir?** | No: ¡imagínate escribir las reglas para reconocer una cara! | Sí: aprende de los ejemplos |
| **Si cambian los gustos...** | Tienes que reescribir la receta | Le muestras ejemplos nuevos y se reajusta |

> 💡 **Para recordar:** en la programación tradicional entregas **reglas + datos** y obtienes respuestas. En Machine Learning entregas **datos + respuestas** y obtienes las **reglas**.

## 2. ¿Cuándo decimos que una máquina "aprendió"? 🎯

El investigador **Tom Mitchell** (1997) lo explicó con tres piezas. Te lo cuento con el **tejo**, nuestro deporte nacional:

| Pieza | En el tejo | En un filtro de correo basura |
|---|---|---|
| **La tarea** (¿qué debe hacer?) | Embocar el tejo en la mecha | Decidir si un correo es basura o no |
| **La experiencia** (¿de dónde aprende?) | Los lanzamientos que va practicando | Correos que otras personas ya marcaron |
| **La medida** (¿cómo sé si mejora?) | Cuántos aciertos hace de cada 10 lanzamientos | Cuántos correos clasifica bien |

Si con más práctica la medida **mejora**, hubo aprendizaje. Si no mejora, solo estuvo repitiendo sin aprender.

## 3. ¿Machine Learning, Inteligencia Artificial o Minería de Datos? 🪆

Se mezclan mucho, pero no son lo mismo. Piénsalo como una **cocina de restaurante**:

| Término | Es como... | Idea en una frase |
|---|---|---|
| **Inteligencia Artificial (IA)** | Toda la cocina | Máquinas que hacen cosas que parecen inteligentes |
| **Machine Learning (ML)** | La parte donde el cocinero **aprende probando** | La IA que aprende de los datos |
| **Minería de Datos** | El que revisa la despensa para descubrir qué se vende junto | Encontrar patrones útiles en datos que ya existen |

## 4. Tu primer modelo: ¿papa criolla o papa pastusa? 🥔

Tenemos 10 papas que ya pesamos, medimos y **etiquetamos** (los números están inventados para el ejemplo). Le enseñaremos a la máquina con esos ejemplos y luego le preguntaremos por papas **que nunca ha visto**.

````python
# Tu primer modelo de Machine Learning (datos inventados para el ejemplo)
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier

# 1) Los ejemplos con la respuesta ya conocida
papas = pd.DataFrame({
    "peso_gramos":  [35, 40, 50, 45, 55, 150, 180, 220, 200, 170],
    "diametro_cm":  [3.5, 3.8, 4.2, 4.0, 4.4, 6.5, 7.0, 8.2, 7.8, 6.8],
    "tipo":         ["criolla"] * 5 + ["pastusa"] * 5,
})

# 2) "Entrenar" = que la maquina descubra la regla a partir de los ejemplos
modelo = DecisionTreeClassifier(random_state=42)
modelo.fit(papas[["peso_gramos", "diametro_cm"]], papas["tipo"])

# 3) Preguntar por papas NUEVAS que la maquina no vio
nuevas = pd.DataFrame({"peso_gramos": [48, 190], "diametro_cm": [4.1, 7.5]})
respuestas = modelo.predict(nuevas)
for (_, fila), tipo in zip(nuevas.iterrows(), respuestas):
    print(f"Papa de {fila['peso_gramos']:.0f} g y {fila['diametro_cm']} cm -> el modelo dice: {tipo}")

# 4) Dibujo
colores = {"criolla": "#f59e0b", "pastusa": "#92400e"}
for tipo, grupo in papas.groupby("tipo"):
    plt.scatter(grupo["peso_gramos"], grupo["diametro_cm"], c=colores[tipo], s=90, label=f"{tipo} (ejemplos)")
plt.scatter(nuevas["peso_gramos"], nuevas["diametro_cm"], marker="*", s=300, c="#2563eb", label="Papas nuevas")
plt.xlabel("Peso (gramos)")
plt.ylabel("Diámetro (cm)")
plt.title("Aprendiendo a distinguir papas", fontweight="bold")
plt.legend()
plt.show()
````

---
### 🤔 ¿Qué acaba de pasar?
Nadie le dijo a la máquina "si pesa menos de tantos gramos es criolla". Ella **encontró la regla sola** mirando los ejemplos, y después la aplicó a papas nuevas. Eso es Machine Learning.

## 5. Los pasos de un proyecto de ML: como una cosecha 🌾

Un modelo no es todo el proyecto. Es **un paso** de varios, y se parece mucho a sembrar:

| Paso del proyecto | En la finca | Pregunta clave |
|---|---|---|
| **1. Definir el problema** | Decidir qué sembrar | ¿Qué quiero predecir y cómo sabré si salió bien? |
| **2. Conseguir los datos** | Conseguir semillas | ¿Qué datos tengo? |
| **3. Limpiar y explorar** | Preparar la tierra | ¿Hay datos dañados, vacíos o raros? |
| **4. Entrenar** | Sembrar | ¿Qué modelo pruebo? |
| **5. Evaluar** | Revisar la calidad de la cosecha | ¿Funciona con datos que **nunca vio**? |
| **6. Usarlo** | Llevar al mercado | ¿Cómo lo usará la gente? |
| **7. Vigilarlo** | Revisar cada temporada | ¿Sigue funcionando con el tiempo? |

Hagamos el paso 5: **un examen justo**. Guardamos algunos ejemplos sin enseñárselos a la máquina y después la examinamos con ellos.

````python
# El examen justo: ensenar con una parte de los datos y examinar con la otra
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

flores = load_iris()          # 150 flores de 3 especies, ya medidas
X_estudio, X_examen, y_estudio, y_examen = train_test_split(
    flores.data, flores.target, test_size=0.3, random_state=42)

modelo = KNeighborsClassifier().fit(X_estudio, y_estudio)    # la maquina estudia SOLO con X_estudio
aciertos = modelo.score(X_examen, y_examen)                  # y se examina con flores que nunca vio

print(f"Flores para estudiar: {len(X_estudio)} | Flores para el examen: {len(X_examen)}")
print(f"Nota del examen: {aciertos * 100:.0f}% de aciertos")
````

## Resumen en una frase 🎯

> **El Machine Learning es enseñar a una máquina con ejemplos, en lugar de escribirle las reglas, y comprobar con un examen justo que de verdad aprendió.**

### Lo que aprendiste hoy:
- ✅ Programar = tú escribes las reglas. ML = la máquina las descubre desde los datos.
- ✅ Aprender = con más experiencia, la medida de desempeño **mejora** (como en el tejo).
- ✅ IA es la cocina completa; ML es la parte que aprende probando; Minería de Datos revisa la despensa.
- ✅ Un proyecto tiene varios pasos (como una cosecha) y el modelo es solo uno de ellos.
- ✅ El examen debe hacerse con datos que la máquina **nunca vio**.
