## 1. Con profesor: el estudiante con el libro de respuestas 🎓

Es como estudiar para un examen con un libro que trae **la respuesta correcta al final de cada ejercicio**. Cada ejemplo viene con su "respuesta" (la **etiqueta**), y la máquina aprende a imitar esas respuestas.

- **Ejemplo de la vida real:** el filtro de correo basura. Miles de correos que personas ya marcaron como *basura* o *normal* le enseñan a la máquina.
- **Otros ejemplos:** decir si un tumor es benigno o maligno, o estimar el precio de un apartamento.

Probemos con 8 correos de juguete: contamos cuántas palabras de oferta ("gratis", "premio"...) y cuántos signos de exclamación traen.

````python
# APRENDER CON PROFESOR: cada ejemplo trae su respuesta
import pandas as pd
from sklearn.linear_model import LogisticRegression

correos = pd.DataFrame({
    "palabras_de_oferta": [0, 1, 0, 2, 7, 9, 6, 8],
    "signos_exclamacion": [0, 1, 1, 0, 8, 6, 9, 7],
    "etiqueta":           ["normal"] * 4 + ["basura"] * 4,     # <- la respuesta correcta
})

modelo = LogisticRegression().fit(correos[["palabras_de_oferta", "signos_exclamacion"]], correos["etiqueta"])

# Dos correos nuevos que la maquina no ha visto
nuevos = pd.DataFrame({"palabras_de_oferta": [1, 8], "signos_exclamacion": [0, 10]})
for fila, respuesta in zip(nuevos.values, modelo.predict(nuevos)):
    print(f"Correo con {fila[0]} palabras de oferta y {fila[1]} exclamaciones -> {respuesta}")
````

## 2. Sin profesor: ordenar el cajón sin que nadie te diga cómo 🧦

Ahora imagina que te dan un cajón lleno de ropa **sin ninguna etiqueta** y te piden ordenarlo. Tú solo agrupas "lo que se parece": camisetas con camisetas, medias con medias. Nadie te dijo cuántos grupos había ni cómo se llamaban.

- **Ejemplo de la vida real:** una tienda agrupa a sus clientes por comportamiento de compra para ofrecerles cosas distintas, sin que nadie haya inventado los grupos de antemano.
- A esta técnica de agrupar se le llama **clustering**, y el algoritmo más famoso es **K-Means**.

Probemos con 9 clientes de una tienda: cuánto gastan al mes (en miles de pesos) y cuántas veces visitan la tienda.

````python
# APRENDER SIN PROFESOR: la maquina agrupa por parecido, sin respuestas
import pandas as pd
from sklearn.cluster import KMeans

clientes = pd.DataFrame({
    "gasto_mes_miles": [30, 35, 28, 150, 160, 140, 80, 85, 90],
    "visitas_mes":     [1,  2,  1,  8,   9,   7,   4,  5,  4],
})

grupos = KMeans(n_clusters=3, n_init=10, random_state=42).fit_predict(clientes)   # le pedimos 3 grupos
clientes["grupo"] = grupos

print(clientes.sort_values("grupo").to_string(index=False))
print("\nPromedio de cada grupo (para ponerles nombre nosotros):")
print(clientes.groupby("grupo").mean().round(1))
````

---
> 🤔 **Ojo:** la máquina solo dijo "grupo 0, 1 y 2". **Los nombres** ("ocasionales", "fieles", "grandes compradores") se los ponemos nosotros mirando los promedios.

## 3. Por premios: entrenar a un perro con galletas 🐶

Aquí nadie te da respuestas. Pruebas algo, recibes un **premio** (o nada) y vas aprendiendo qué conviene repetir. Así se aprende a montar en bicicleta o se entrena a una mascota.

- **Ejemplo de la vida real:** programas que aprenden a jugar ajedrez o Go, o robots que aprenden a caminar.
- **El dilema:** ¿vuelves siempre al puesto que ya sabes que es bueno (**explotar**) o pruebas uno nuevo por si es mejor (**explorar**)? Hay que hacer un poco de las dos cosas.

**Juego:** hay 3 puestos de empanadas en el centro de Tunja. Cada uno tiene una probabilidad **secreta** de que la empanada te guste (30 %, 50 % y 80 %). No la conoces: solo la descubres comiendo. Durante 300 días, casi siempre vas al mejor puesto que conoces, pero 1 de cada 10 días pruebas uno al azar.

````python
# APRENDER POR PREMIOS: probar, recibir premio y ajustar
import numpy as np

rng = np.random.default_rng(42)
prob_secreta = [0.30, 0.50, 0.80]        # la maquina NO conoce estos numeros
visitas = np.zeros(3)                    # cuantas veces fue a cada puesto
premios = np.zeros(3)                    # cuantas veces le gusto la empanada

for dia in range(300):
    if rng.random() < 0.10 or visitas.min() == 0:    # explorar: probar un puesto al azar
        puesto = rng.integers(3)
    else:                                            # explotar: ir al de mejor promedio
        puesto = np.argmax(premios / visitas)
    gusto = rng.random() < prob_secreta[puesto]      # 1 = premio (le gusto), 0 = nada
    visitas[puesto] += 1
    premios[puesto] += gusto

for i in range(3):
    print(f"Puesto {i + 1}: lo visitó {visitas[i]:.0f} veces | le gustó el {premios[i] / visitas[i] * 100:.0f}% de las veces")
print(f"\nEl agente termino prefiriendo el puesto {np.argmax(visitas) + 1} (el mejor de verdad es el puesto 3).")
````

## 4. ¿Cuál uso yo? La tabla para decidir 🧭

| Si tienes... | Usas aprendizaje... | Es como... | Ejemplo |
|---|---|---|---|
| Ejemplos **con la respuesta** anotada | **Supervisado** (con profesor) | Estudiar con el libro de respuestas | Detectar correo basura |
| Datos **sin respuestas**, y quieres encontrar grupos o patrones | **No supervisado** (sin profesor) | Ordenar el cajón de ropa | Agrupar clientes parecidos |
| Un **agente que actúa** y recibe premios o castigos | **Por refuerzo** (por premios) | Entrenar un perro | Un robot que aprende a caminar |

> 💡 **Y cuando casi no hay respuestas:** conseguir respuestas (etiquetas) es caro, porque alguien tiene que anotarlas una por una. Por eso existen métodos **mixtos**: unos usan *pocas* respuestas y *muchos* datos sin respuesta (semi-supervisado), y otros **inventan sus propias respuestas** a partir de los datos (auto-supervisado), como cuando se le tapa una palabra a una frase para que la máquina la adivine. Así se entrenan los grandes modelos de lenguaje.

## Resumen en una frase 🎯

> **Una máquina puede aprender con profesor (respuestas dadas), sin profesor (buscando grupos por su cuenta) o por premios (probando y ajustando); el tipo de datos que tienes decide cuál usar.**

### Lo que aprendiste hoy:
- ✅ **Supervisado:** cada ejemplo trae su respuesta. Ejemplo: correo basura.

- ✅ **Por refuerzo:** aprende por ensayo y error con premios; debe equilibrar explorar y explotar.
- ✅ La pregunta clave para elegir: **¿tengo la respuesta anotada en mis datos?**
