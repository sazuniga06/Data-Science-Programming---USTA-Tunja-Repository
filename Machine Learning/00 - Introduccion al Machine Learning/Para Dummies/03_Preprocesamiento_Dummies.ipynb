## 1. Basura entra, basura sale 🗑️

Si cocinas con ingredientes dañados, la sopa sale mala aunque seas el mejor chef del mundo. Un modelo de Machine Learning es igual: **un buen algoritmo no arregla datos malos**.

Los datos suelen traer cuatro problemas típicos:

| Problema | Es como... | Ejemplo |
|---|---|---|
| **Datos vacíos** | Una receta con ingredientes que faltan | Un cliente que no declaró su ingreso |
| **Escalas distintas** | Comparar metros con kilómetros | Ingreso en millones de pesos junto a la edad en años |
| **Palabras en vez de números** | Una calculadora que no entiende letras | La columna "ciudad": Tunja, Duitama... |
| **Datos raros** | Una papa del tamaño de una sandía | Un ingreso con tres ceros de más por error |

Veamos cada uno, empezando por los datos vacíos.

````python
# PROBLEMA 1: datos vacios (NaN = "no hay dato")
import pandas as pd

clientes = pd.DataFrame({
    "cliente":  ["Ana", "Luis", "María", "Pedro", "Sofía"],
    "edad":     [34, None, 29, 41, None],
    "ingreso":  [2_500_000, 3_100_000, None, 4_000_000, 2_800_000],
})
print("ANTES (con huecos):")
print(clientes.to_string(index=False))

# Una solucion simple: rellenar con la mediana (el valor "del medio") de cada columna
rellenos = clientes[["edad", "ingreso"]].median()
clientes_limpios = clientes.fillna(rellenos)
print("\nDESPUES (rellenado con la mediana):")
print(clientes_limpios.to_string(index=False))
````

## 2. Escalas distintas: ¿metros o kilómetros? 📏

Muchos modelos deciden "qué tan parecidos son dos clientes" midiendo la **distancia** entre ellos. Si una columna tiene números enormes (el ingreso, en millones) y otra pequeños (la edad), **la grande se come a la pequeña** y la edad deja de importar.

Prueba con 6 clientes: ¿quién se parece más a **A** (30 años, $2.000.000)?

````python
# PROBLEMA 2: escalas distintas
import numpy as np
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
    "cliente": list("ABCDEF"),
    "edad":    [30, 60, 31, 45, 52, 38],
    "ingreso": [2_000_000, 2_100_000, 2_600_000, 3_900_000, 1_500_000, 3_100_000],
})

def mas_parecido_a_A(matriz):
    distancias = np.linalg.norm(matriz[1:] - matriz[0], axis=1)     # distancia de cada cliente a A
    return df["cliente"].iloc[1:].iloc[distancias.argmin()]

sin_ajustar = df[["edad", "ingreso"]].values
ajustado = StandardScaler().fit_transform(sin_ajustar)              # pone ambas columnas en la misma "unidad"

print("Sin ajustar la escala, el cliente mas parecido a A es:", mas_parecido_a_A(sin_ajustar))
print("   -> ¡Tiene 60 años y A tiene 30! Solo miro el ingreso.")
print("Con la escala ajustada, el cliente mas parecido a A es:", mas_parecido_a_A(ajustado))
print("   -> Tiene 31 años y un ingreso parecido. Ahora si tiene sentido.")
````

## 3. Palabras en vez de números 🔤

Los modelos solo entienden números. Para "traducir" una columna de palabras, se crea **una columna de sí/no por cada palabra** (esto se llama *one-hot*, o "codificación de una sola casilla encendida").

````python
# PROBLEMA 3: palabras -> numeros
ciudades = pd.DataFrame({"ciudad": ["Tunja", "Duitama", "Tunja", "Sogamoso"]})
print(pd.get_dummies(ciudades["ciudad"], prefix="ciudad", dtype=int))
````

---
> 💡 Fíjate: en cada fila solo **una** casilla vale 1 (la ciudad de ese cliente). Así la máquina no cree que "Tunja = 1" y "Sogamoso = 3" signifiquen que una vale tres veces la otra.

## 4. La trampa más peligrosa: ver el examen antes de presentarlo 🕵️

Imagina un estudiante que consigue **las preguntas del examen** de antemano y las estudia. Saca 5.0, pero **no aprendió nada**: el 5.0 no significa nada. En Machine Learning esto se llama **fuga de datos** (*data leakage*) y es sorprendentemente fácil de cometer sin darte cuenta.

| ❌ Mal hecho | ✅ Bien hecho |
|---|---|
| Rellenar huecos y ajustar escalas con **todos** los datos, y **después** separar los del examen | **Primero separar** los datos del examen y guardarlos bajo llave |
| Los datos del examen ya "influyeron" en la preparación | Preparar usando **solo** los datos de estudio |
| Nota alta pero falsa | Nota realista, la que tendrás en la vida real |

**El truco para no caer en la trampa:** usar un `Pipeline`, que es como una **línea de producción** que encadena todos los pasos de preparación y el modelo en uno solo. Cada vez que se entrena, la línea aprende **únicamente** con los datos de estudio.

solicitantes = pd.DataFrame({
    "edad":    [28, None, 35, 41],
    "ingreso": [2_100_000, 2_600_000, 3_000_000, 3_800_000],
    "ciudad":  ["Tunja", "Duitama", "Tunja", "Sogamoso"],

assert solicitantes_listos.isna().sum().sum() == 0
assert (ciudades_num.sum(axis=1) == 1).all()

from sklearn.model_selection import train_test_split

ingresos = pd.DataFrame({"ingreso_millones": [2.1, 2.5, None, 3.0, 2.8, 9.5, 2.4, None, 2.7, 3.2]})
estudio, examen = train_test_split(ingresos, test_size=0.3, random_state=42)   # primero separar

# La mediana se calcula SOLO con los datos de estudio
mediana_estudio = estudio["ingreso_millones"].median()
estudio_listo = estudio.fillna(mediana_estudio)
examen_listo = examen.fillna(mediana_estudio)        # el examen recibe la mediana del estudio, no la suya

# La trampa: calcular con todos los datos (el examen ya "influyo")
mediana_trampa = ingresos["ingreso_millones"].median()

print(f"Mediana honesta (solo estudio): {mediana_estudio}")
print(f"Mediana con trampa (todos):     {mediana_trampa}")

assert estudio_listo.isna().sum().sum() == 0 and examen_listo.isna().sum().sum() == 0
assert mediana_estudio == estudio["ingreso_millones"].median()
print("\nHuecos rellenados sin que el examen influyera en la preparacion.")

## Resumen en una frase 🎯

> **Antes de entrenar hay que "lavar las papas": llenar huecos, poner todo en la misma escala y pasar palabras a números, y siempre separar primero el examen para no hacer trampa sin querer.**

### Lo que aprendiste hoy:
- ✅ Datos malos = resultados malos (*Garbage In, Garbage Out*).
- ✅ Los datos vacíos se rellenan o se manejan con criterio.
- ✅ Columnas con números de tamaños muy distintos engañan a muchos modelos: hay que igualar escalas.
- ✅ Las palabras se convierten en columnas de sí/no.
- ✅ **Separar primero** el examen y usar un `Pipeline` evita la fuga de datos.
