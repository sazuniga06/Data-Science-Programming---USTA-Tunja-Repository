## 1. Por Qué el Desbalance Rompe el Entrenamiento y la Evaluación 💥

Sea $\pi=P(y=1)$ la **prevalencia** de la clase minoritaria (aquí $\pi\approx5\,\%$).

**En el entrenamiento.** La función de pérdida promedio es $\mathcal L=(1-\pi)\,\mathbb E[\ell\mid y=0]+\pi\,\mathbb E[\ell\mid y=1]$: la clase mayoritaria aporta **$(1-\pi)/\pi\approx19$ veces más** al gradiente. El modelo minimiza su error global **sacrificando** la minoría; la **regla de Bayes** con costos iguales, $\hat y=1\iff P(y=1\mid\mathbf x)>0.5$, **casi nunca** supera ese umbral aunque el modelo **rankee muy bien**: el problema suele estar en **dónde se corta**, no en cómo se ordena.

**En la evaluación.** Un clasificador trivial que siempre predice «negativo» tiene *accuracy* $=1-\pi=95\,\%$ y **recall = 0**: la **exactitud** es **engañosa**. El **ROC-AUC** también puede ser **optimista**: la tasa de falsos positivos $\text{FPR}=\text{FP}/\text{N}$ tiene un denominador enorme y **casi no se mueve** aunque la precisión sea pésima. La **curva precisión–recobrado** y su **PR-AUC** (o *average precision*) **sí** reflejan el problema, y su **línea base** no es 0.5 sino la **prevalencia** $\pi$ (Saito y Rehmsmeier, 2015).

Generamos «el mundo»: un problema sintético de 20 variables (5 informativas) con $\pi\approx5\,\%$ y **40 000 ejemplos** de los que apartamos la mitad como **prueba enorme** (la «verdad» contra la que medimos todo). El entrenamiento será una **muestra de 2 000 ejemplos** (≈ 100 minoritarios), como en la práctica.

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas matplotlib seaborn scikit-learn scipy
# !pip install -q imbalanced-learn        # OPCIONAL: solo para contrastar con la libreria imblearn

import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy
import sklearn
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_predict, cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import NearestNeighbors
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, precision_score, recall_score, f1_score, roc_auc_score,
                             average_precision_score, precision_recall_curve, roc_curve, brier_score_loss)

try:                                                   # imbalanced-learn es OPCIONAL: sin el solo se omiten las comparaciones con la libreria
    import imblearn
    from imblearn.over_sampling import SMOTE as SMOTE_IB, RandomOverSampler as ROS_IB
    from imblearn.under_sampling import RandomUnderSampler as RUS_IB
    from imblearn.pipeline import Pipeline as PipelineIB
except Exception:
    imblearn = None

warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (9, 5)
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"Entorno listo | numpy {np.__version__} | scipy {scipy.__version__} | scikit-learn {sklearn.__version__} | imbalanced-learn: {imblearn.__version__ if imblearn else 'NO instalado (opcional)'}")

# ============================================================
# 1. "El mundo" (40 000 ejemplos), prueba enorme y muestra de entrenamiento de 2 000
# ============================================================
PREV, N_ENT = 0.05, 2000
X_mundo, y_mundo = make_classification(40000, 20, n_informative=5, n_redundant=2, weights=[1 - PREV, PREV], class_sep=0.8, flip_y=0.01, random_state=SEED)
X_pool, X_prueba, y_pool, y_prueba = train_test_split(X_mundo, y_mundo, test_size=0.5, stratify=y_mundo, random_state=0)


def muestra_entrenamiento(semilla, n=N_ENT):
    """Una muestra aleatoria de n ejemplos del 'pool' (la prueba nunca se toca para entrenar ni para elegir nada)."""
    idx = np.random.default_rng(semilla).choice(len(X_pool), n, replace=False)
    return X_pool[idx], y_pool[idx]


def modelo_lr(**kw):
    return make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000, **kw))


X_ent, y_ent = muestra_entrenamiento(0)
print(f"Prevalencia en el mundo = {y_mundo.mean():.3f} | entrenamiento: {len(y_ent)} ejemplos, {int(y_ent.sum())} minoritarios ({y_ent.mean():.1%}) | prueba: {len(y_prueba)} ejemplos, {int(y_prueba.sum())} minoritarios")

# --- El clasificador trivial y la regresion logistica "ingenua" (sin tratar el desbalance) ---
ingenuo = modelo_lr().fit(X_ent, y_ent)
p_ing = ingenuo.predict_proba(X_prueba)[:, 1]
pred_ing = (p_ing > 0.5).astype(int)
pred_triv = np.zeros_like(y_prueba)
filas = {}
for nombre, pred, score in [("Siempre «negativo»", pred_triv, np.zeros(len(y_prueba))), ("Regresión logística (umbral 0.5)", pred_ing, p_ing)]:
    filas[nombre] = {"accuracy": accuracy_score(y_prueba, pred), "accuracy balanceado": balanced_accuracy_score(y_prueba, pred), "recall": recall_score(y_prueba, pred),
                     "precisión": precision_score(y_prueba, pred, zero_division=0), "F1": f1_score(y_prueba, pred), "ROC-AUC": roc_auc_score(y_prueba, score), "PR-AUC (AP)": average_precision_score(y_prueba, score)}
tabla = pd.DataFrame(filas).T
display(tabla.round(3))
assert tabla.loc["Siempre «negativo»", "accuracy"] > 0.94 and tabla.loc["Siempre «negativo»", "recall"] == 0       # el trivial: 95 % de exactitud y recall 0
assert np.isclose(tabla.loc["Siempre «negativo»", "PR-AUC (AP)"], y_prueba.mean())                                  # la linea base de la PR-AUC es la prevalencia

fig, axes = plt.subplots(1, 3, figsize=(17, 4.6))
pr, rc, _ = precision_recall_curve(y_prueba, p_ing)
axes[0].plot(rc, pr, color=PALETA[0], lw=2, label=f"Regresión logística (AP = {average_precision_score(y_prueba, p_ing):.2f})")
axes[0].axhline(y_prueba.mean(), color="black", ls="--", label=f"Línea base = prevalencia = {y_prueba.mean():.2f}")
axes[0].set_xlabel("Recall"); axes[0].set_ylabel("Precisión"); axes[0].legend(fontsize=8); axes[0].set_title("Curva precisión–recall: el problema se ve", fontweight="bold", fontsize=10)
fpr, tpr, _ = roc_curve(y_prueba, p_ing)
axes[1].plot(fpr, tpr, color=PALETA[1], lw=2, label=f"ROC-AUC = {roc_auc_score(y_prueba, p_ing):.2f}"); axes[1].plot([0, 1], [0, 1], "k--")
axes[1].set_xlabel("Tasa de falsos positivos"); axes[1].set_ylabel("Recall (TPR)"); axes[1].legend(fontsize=8); axes[1].set_title("ROC: se ve «bastante bien»", fontweight="bold", fontsize=10)
axes[2].hist(p_ing[y_prueba == 0], bins=40, alpha=0.7, color=PALETA[0], label="clase 0 (95 %)", density=True)
axes[2].hist(p_ing[y_prueba == 1], bins=40, alpha=0.7, color=PALETA[1], label="clase 1 (5 %)", density=True); axes[2].axvline(0.5, color="black", ls="--", label="umbral 0.5")
axes[2].set_xlabel("Probabilidad predicha de la clase 1"); axes[2].set_ylabel("Densidad"); axes[2].legend(fontsize=8); axes[2].set_title("Casi nada supera el umbral de 0.5", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
print(f"Con umbral 0.5 el modelo detecta solo el {tabla.loc['Regresión logística (umbral 0.5)', 'recall']:.0%} de los positivos; el accuracy ({tabla.loc['Regresión logística (umbral 0.5)', 'accuracy']:.3f}) apenas supera al del clasificador trivial ({tabla.loc['Siempre «negativo»', 'accuracy']:.3f}).")
````

## 2. Remuestreo Aleatorio y la Corrección de las Probabilidades ⚖️

Las dos estrategias más simples **cambian los datos de entrenamiento** para equilibrar las clases:

| Estrategia | Qué hace | Ventaja | Riesgo |
|---|---|---|---|
| **Submuestreo aleatorio** (*random undersampling*, RUS) | **Descarta** ejemplos de la mayoría hasta igualar la minoría | Rápido; entrena sobre menos datos | **Pierde información**; alta varianza |
| **Sobremuestreo aleatorio** (*random oversampling*, ROS) | **Duplica** ejemplos de la minoría (con reemplazo) hasta igualar | No pierde datos | **Copias exactas**: los modelos flexibles las **memorizan** (sobreajuste); los árboles pueden aislar réplicas |

**Efecto colateral sobre las probabilidades.** Remuestrear a 50/50 cambia la **prevalencia aparente** de $\pi$ a $\pi'=0.5$. Por la regla de Bayes, un modelo bien calibrado en los datos remuestreados da $p'$ tal que los *odds* quedan inflados por $\frac{1-\pi}{\pi}$ respecto a los reales. La **corrección por *prior*** recupera la probabilidad original (con $\pi'=0.5$):

$$\frac{p}{1-p}=\frac{p'}{1-p'}\cdot\frac{\pi}{1-\pi}\quad\Longleftrightarrow\quad p=\frac{\pi\,p'}{\pi\,p'+(1-\pi)(1-p')}.$$

Es una función **monótona**: **no cambia el orden** (la PR-AUC y el ROC-AUC quedan idénticos) pero **sí** las probabilidades, el umbral de 0.5 y la **calibración**. Implementamos RUS y ROS **desde cero** (devolviendo también **de qué fila original proviene** cada fila, para auditar), verificamos sus propiedades y la corrección por *prior*.

````python
# ============================================================
# 2. Submuestreo y sobremuestreo aleatorios desde cero (con procedencia) + correccion por prior
# ============================================================
def sub_aleatorio(X, y, semilla=0):
    """RUS: conserva TODOS los positivos y una muestra SIN reemplazo de negativos del mismo tamano. Devuelve (X, y, idx_original)."""
    rng = np.random.default_rng(semilla)
    i0, i1 = np.where(y == 0)[0], np.where(y == 1)[0]
    keep = np.r_[rng.choice(i0, len(i1), replace=False), i1]
    return X[keep], y[keep], keep


def sobre_aleatorio(X, y, semilla=0):
    """ROS: conserva todo y anade copias (CON reemplazo) de positivos hasta igualar a los negativos. Devuelve (X, y, idx_original)."""
    rng = np.random.default_rng(semilla)
    i1 = np.where(y == 1)[0]
    extra = rng.choice(i1, int((y == 0).sum() - len(i1)), replace=True)
    idx = np.r_[np.arange(len(y)), extra]
    return X[idx], y[idx], idx


n0, n1 = int((y_ent == 0).sum()), int((y_ent == 1).sum())
Xu, yu, iu = sub_aleatorio(X_ent, y_ent)
Xo, yo, io = sobre_aleatorio(X_ent, y_ent)
# --- Propiedades ---
assert (yu == 0).sum() == (yu == 1).sum() == n1 and len(np.unique(iu)) == len(iu)                      # RUS: clases iguales, sin repetir filas
assert np.array_equal(Xu, X_ent[iu]) and set(np.where(y_ent == 1)[0]) <= set(iu)                          # RUS: son filas originales y estan TODOS los positivos
assert (yo == 0).sum() == (yo == 1).sum() == n0 and np.array_equal(Xo, X_ent[io])                          # ROS: clases iguales y todas las filas son copias de originales
assert np.array_equal(Xo[:len(y_ent)], X_ent)                                                              # ROS: los originales se conservan intactos
print(f"Original: {n0} negativos / {n1} positivos  |  RUS: {int((yu == 0).sum())} / {int((yu == 1).sum())}  |  ROS: {int((yo == 0).sum())} / {int((yo == 1).sum())} "
      f"(cada positivo aparece en promedio {1 + (n0 - n1) / n1:.1f} veces)")
if imblearn is not None:                                                                                    # conteos identicos a imblearn (las filas elegidas dependen del azar)
    from collections import Counter
    assert Counter(RUS_IB(random_state=0).fit_resample(X_ent, y_ent)[1]) == Counter(yu) and Counter(ROS_IB(random_state=0).fit_resample(X_ent, y_ent)[1]) == Counter(yo)
    print("imbalanced-learn: los conteos de RandomUnderSampler y RandomOverSampler coinciden con los nuestros.")

# --- Entrenar con cada version y corregir las probabilidades por prior ---
def corregir_prior(p_bal, prev):
    """p = pi p' / (pi p' + (1-pi)(1-p')): devuelve la probabilidad con la prevalencia ORIGINAL a partir de la de un modelo entrenado a 50/50."""
    return prev * p_bal / (prev * p_bal + (1 - prev) * (1 - p_bal))


prev_ent = y_ent.mean()
filas = {}
for nombre, (Xr, yr) in {"Sin remuestrear": (X_ent, y_ent), "Submuestreo (RUS)": (Xu, yu), "Sobremuestreo (ROS)": (Xo, yo)}.items():
    p = modelo_lr().fit(Xr, yr).predict_proba(X_prueba)[:, 1]
    filas[nombre] = {"AP": average_precision_score(y_prueba, p), "recall @0.5": recall_score(y_prueba, p > 0.5), "precisión @0.5": precision_score(y_prueba, p > 0.5, zero_division=0),
                     "F1 @0.5": f1_score(y_prueba, p > 0.5), "prob. media predicha": p.mean(), "Brier": brier_score_loss(y_prueba, p)}
    if nombre != "Sin remuestrear":
        pc = corregir_prior(p, prev_ent)
        assert np.isclose(average_precision_score(y_prueba, pc), filas[nombre]["AP"])                      # monotona: el ranking (AP) NO cambia
        filas[nombre + " + corrección de prior"] = {"AP": average_precision_score(y_prueba, pc), "recall @0.5": recall_score(y_prueba, pc > 0.5), "precisión @0.5": precision_score(y_prueba, pc > 0.5, zero_division=0),
                                                      "F1 @0.5": f1_score(y_prueba, pc > 0.5), "prob. media predicha": pc.mean(), "Brier": brier_score_loss(y_prueba, pc)}
tabla_r = pd.DataFrame(filas).T
display(tabla_r.round(3))
print(f"Prevalencia real en la prueba = {y_prueba.mean():.3f}. Tras remuestrear a 50/50 las probabilidades medias suben de ≈ 0.05 a ≈ 0.3 (el modelo «cree» que los positivos son mucho más comunes); la corrección por prior las devuelve a la prevalencia y baja el Brier de ≈ 0.14 a ≈ 0.04.")
assert tabla_r.loc["Submuestreo (RUS)", "prob. media predicha"] > 0.2 and abs(tabla_r.loc["Submuestreo (RUS) + corrección de prior", "prob. media predicha"] - y_prueba.mean()) < 0.03
assert tabla_r.loc["Submuestreo (RUS) + corrección de prior", "Brier"] < tabla_r.loc["Submuestreo (RUS)", "Brier"]
````

##### 🛠️ Práctica 1: Submuestreo parcial y corrección de prior para una prevalencia que no es 50/50

La corrección de la Sección 2 supone un remuestreo a $\pi'=0.5$. Generalicémosla: si entrenas con una prevalencia aparente $\pi'$ y la real es $\pi$, los *odds* se corrigen con
$$\frac{p}{1-p}=\frac{p'}{1-p'}\cdot\frac{\pi/(1-\pi)}{\pi'/(1-\pi')}.$$
Con `X_ent`, `y_ent`, `modelo_lr`, `X_prueba`, `y_prueba`, `prev_ent` y las métricas ya importadas:

1. Haz un **submuestreo parcial**: conserva todos los positivos de `y_ent` y solo **tres negativos por positivo** (muestra sin reemplazo con `np.random.default_rng(0)`). Guarda las filas elegidas en `keep` y calcula la nueva prevalencia `pi_nuevo` $=0.25$.
2. Entrena `modelo_lr()` con esas filas y guarda en `p_par` sus probabilidades sobre `X_prueba`.

## 3. SMOTE desde Cero: Interpolar en Vez de Duplicar 🧬

El ***Synthetic Minority Over-sampling Technique*** (**SMOTE**; Chawla et al., 2002) evita las **copias exactas** del sobremuestreo aleatorio **creando ejemplos nuevos por interpolación**. Para generar un punto sintético:

1. Elige al azar un punto minoritario $\mathbf x_i$ (el **semilla**).
2. Busca sus $k$ **vecinos más cercanos entre los minoritarios** (por defecto $k=5$) y elige uno al azar, $\mathbf x_{\text{nn}}$.
3. Genera el punto
$$\mathbf x_{\text{nuevo}}=\mathbf x_i+\lambda\,(\mathbf x_{\text{nn}}-\mathbf x_i),\qquad \lambda\sim\mathcal U(0,1).$$

El punto cae **en el segmento** entre ambos, es decir, dentro de la **envoltura** de la minoría. Se repite hasta que el número de minoritarios **iguala** al de la mayoría ($N=n_{\text{may}}-n_{\text{min}}$ puntos sintéticos). Las **propiedades** que verificaremos:

- **(P1)** Cada punto sintético está **sobre un segmento** $[\mathbf x_i,\mathbf x_{\text{nn}}]$ con $\mathbf x_{\text{nn}}$ entre los $k$ vecinos minoritarios de $\mathbf x_i$ (se comprueba **por fuerza bruta**, sin usar la procedencia que guardó el generador).
- **(P2)** El conteo final **iguala** las clases y los **originales se conservan**.
- **(P3)** Los sintéticos quedan dentro de la **caja envolvente** de los minoritarios.
- **(Control negativo)** Puntos aleatorios en esa caja **no** pasan la prueba P1 (la prueba tiene poder de discriminación).

> 🔎 **Desempates:** si hay puntos minoritarios duplicados, «el vecino más cercano» no es único; `NearestNeighbors` rompe el empate por el orden de los datos. Como $\mathbf x_i$ **no** debe ser su propio vecino, tomamos los $k+1$ más cercanos y descartamos el primero (en datos **sin duplicados**, ese primero es $\mathbf x_i$ con distancia 0).

### 3.1 Variantes (a nivel conceptual)

| Variante | Idea | Para qué |
|---|---|---|
| ***Borderline-SMOTE*** (Han et al., 2005) | Interpola solo desde los minoritarios **cerca de la frontera** (los «en peligro») | Reforzar la zona donde se decide |
| ***ADASYN*** (He et al., 2008) | Genera **más** puntos donde la minoría es **más difícil** (más vecinos mayoritarios) | Adaptarse a la dificultad local |
| ***SMOTE-NC* / *SMOTENC*** | Maneja variables **categóricas** (la interpolación no tiene sentido en ellas) | Datos mixtos |
| ***Tomek links*** (Tomek, 1976) | **Elimina** pares (mayoría, minoría) que son **vecinos mutuos más cercanos** | **Limpiar** la frontera (submuestreo guiado) |
| ***ENN*** (Wilson, 1972) | **Elimina** los ejemplos cuya etiqueta **no coincide** con la de sus vecinos | Quitar ruido |
| ***SMOTE + Tomek / SMOTE + ENN*** (Batista et al., 2004) | Sobremuestrear y luego limpiar | Combinar |

**Limitaciones de SMOTE:** puede crear puntos **entre grupos separados** de la minoría (zonas sin datos reales), **amplifica el ruido** (si una semilla es una etiqueta errónea, replica el error), **ignora a la mayoría** al generar (puede invadir su región), **sufre en alta dimensión** (las distancias pierden sentido, [Módulo 04](../04%20-%20Clustering%20No%20Supervisado/00_Introduccion_Cap2_y_No_Supervisado.ipynb)) y **no define** la interpolación para variables categóricas.

````python
# ============================================================
# 3. SMOTE desde cero y verificacion de sus propiedades (P1, P2, P3 y control negativo)
# ============================================================
def vecinos_minoritarios(Xm, k):
    """Indices de los k vecinos MINORITARIOS mas cercanos de cada punto minoritario (se descarta el primero: el propio punto)."""
    idx = NearestNeighbors(n_neighbors=k + 1).fit(Xm).kneighbors(Xm, return_distance=False)
    return idx[:, 1:]


def smote_cero(X, y, k=5, semilla=0, info=False):
    """SMOTE: x_nuevo = x_i + lambda (x_nn - x_i); i al azar (con reemplazo), x_nn uno de sus k vecinos minoritarios, lambda ~ U(0,1).
    Genera n_may - n_min puntos. Devuelve (X, y, idx_origen) con idx = -1 en los sinteticos; con info=True anade (i, j, lambda)."""
    rng = np.random.default_rng(semilla)
    Xm = X[y == 1]
    n_min, n_gen = len(Xm), int((y == 0).sum() - len(Xm))
    vec = vecinos_minoritarios(Xm, k)
    i = rng.integers(0, n_min, n_gen)                          # punto semilla
    j = vec[i, rng.integers(0, k, n_gen)]                      # uno de sus k vecinos minoritarios
    lam = rng.random((n_gen, 1))
    S = Xm[i] + lam * (Xm[j] - Xm[i])                          # interpolacion
    Xr, yr = np.vstack([X, S]), np.r_[y, np.ones(n_gen, dtype=y.dtype)]
    idx = np.r_[np.arange(len(y)), -np.ones(n_gen, dtype=int)]
    return (Xr, yr, idx, (i, j, lam)) if info else (Xr, yr, idx)


def dist_a_segmentos(S, Xm, k):
    """Para cada punto de S, distancia minima a TODOS los segmentos [x_i, x_nn] (x_nn entre los k vecinos minoritarios de x_i). Fuerza bruta."""
    vec = vecinos_minoritarios(Xm, k)
    A = np.repeat(Xm, k, axis=0)
    B = Xm[vec.ravel()]
    AB = B - A
    den = (AB ** 2).sum(axis=1) + 1e-300
    out = np.empty(len(S))
    for t, s in enumerate(S):
        lam = np.clip(((s - A) * AB).sum(axis=1) / den, 0, 1)
        out[t] = np.sqrt((((s - (A + lam[:, None] * AB))) ** 2).sum(axis=1)).min()
    return out


K = 5
Xs, ys, ids, (ii, jj, ll) = smote_cero(X_ent, y_ent, k=K, semilla=0, info=True)
n_orig = len(y_ent)
S = Xs[n_orig:]
Xm = X_ent[y_ent == 1]
# P2: conteo final iguala las clases y los originales se conservan
assert (ys == 0).sum() == (ys == 1).sum() and np.array_equal(Xs[:n_orig], X_ent) and len(S) == n0 - n1
# P1: cada sintetico esta sobre algun segmento (verificado por fuerza bruta, independiente de la procedencia guardada)
d_sint = dist_a_segmentos(S, Xm, K)
assert d_sint.max() < 1e-9
# Procedencia guardada reproduce exactamente el punto: x = x_i + lambda (x_nn - x_i), con lambda en [0, 1)
assert np.allclose(S, Xm[ii] + ll * (Xm[jj] - Xm[ii])) and ((ll >= 0) & (ll < 1)).all()
# P3: dentro de la caja envolvente de los minoritarios
assert (S >= Xm.min(axis=0) - 1e-12).all() and (S <= Xm.max(axis=0) + 1e-12).all()
# Control negativo: puntos aleatorios dentro de la caja NO estan sobre segmentos
rng = np.random.default_rng(1)
alea = rng.uniform(Xm.min(axis=0), Xm.max(axis=0), (len(S) // 4, Xm.shape[1]))
d_alea = dist_a_segmentos(alea, Xm, K)
assert np.median(d_alea) > 0.5
print(f"SMOTE desde cero (k = {K}): {n1} minoritarios + {len(S)} sintéticos = {int((ys == 1).sum())} = {int((ys == 0).sum())} mayoritarios.")
print(f"P1 (sobre un segmento): distancia máxima al segmento más cercano = {d_sint.max():.1e}  |  control negativo (puntos aleatorios): mediana = {np.median(d_alea):.2f}")

# --- Contraste con imbalanced-learn (opcional): mismas propiedades; NO se exige igualdad exacta (depende del generador de azar) ---
if imblearn is not None:
    Xi, yi = SMOTE_IB(k_neighbors=K, random_state=0).fit_resample(X_ent, y_ent)
    Si = Xi[n_orig:]
    assert (yi == 0).sum() == (yi == 1).sum() and np.array_equal(Xi[:n_orig], X_ent)
    d_ib = dist_a_segmentos(Si, Xm, K)
    assert d_ib.max() < 1e-6
    print(f"imbalanced-learn SMOTE: clases iguales ({int((yi == 1).sum())} / {int((yi == 0).sum())}), originales intactos y los {len(Si)} sintéticos también caen sobre segmentos (distancia máx. {d_ib.max():.1e}). "
          "Los puntos concretos difieren (otro azar), las propiedades no.")
else:
    print("imbalanced-learn no está instalado: se omite el contraste con la librería (pip install imbalanced-learn).")

# --- Visualizacion en 2D: original, sobremuestreo aleatorio y SMOTE ---
X2, y2 = make_classification(300, 2, n_informative=2, n_redundant=0, n_clusters_per_class=1, weights=[0.92, 0.08], class_sep=1.0, flip_y=0, random_state=3)
Xo2, yo2, io2 = sobre_aleatorio(X2, y2, semilla=0)
Xs2, ys2, is2, (i2, j2, l2) = smote_cero(X2, y2, k=5, semilla=0, info=True)
Xm2 = X2[y2 == 1]
fig, axes = plt.subplots(1, 3, figsize=(17, 4.8), sharex=True, sharey=True)
for ax in axes:
    ax.scatter(*X2[y2 == 0].T, s=14, color=PALETA[0], alpha=0.4, label="mayoría")
    ax.scatter(*Xm2.T, s=40, color=PALETA[1], edgecolor="black", label="minoría original", zorder=3)
axes[0].set_title(f"Original ({int((y2 == 1).sum())} minoritarios)", fontweight="bold")
mult = np.bincount(io2[io2 < len(y2)], minlength=len(y2))[y2 == 1]
axes[1].scatter(*Xm2.T, s=40 + 25 * (mult - 1), color="none", edgecolor=PALETA[3], lw=1.5, zorder=2, label="tamaño ∝ nº de copias")
axes[1].set_title("Sobremuestreo aleatorio: copias EXACTAS (mismo punto)", fontweight="bold", fontsize=10)
for t in range(0, len(i2), 2):
    axes[2].plot(*np.c_[Xm2[i2[t]], Xm2[j2[t]]], color="#94a3b8", lw=0.5, alpha=0.6, zorder=1)
axes[2].scatter(*Xs2[len(y2):].T, s=14, color=PALETA[2], marker="x", zorder=4, label="sintéticos (SMOTE)")
axes[2].set_title("SMOTE: puntos NUEVOS sobre segmentos entre vecinos", fontweight="bold", fontsize=10)
for ax in axes:
    ax.set_xlabel("$x_1$"); ax.legend(fontsize=7, loc="lower right")
axes[0].set_ylabel("$x_2$")
plt.tight_layout(); plt.show()
````

## 4. Alternativas sin Remuestreo: Pesos, Umbral y Costo 🎚️

Remuestrear **no es la única forma** (ni siempre la mejor) de atender el desbalance. Tres alternativas que **no tocan los datos**:

**(1) Pesos por clase (`class_weight`).** Multiplicar la pérdida de cada ejemplo positivo por $w$: $\mathcal L=\sum_i s_i\,\ell_i$ con $s_i=w$ si $y_i=1$ y $1$ si no. Con `class_weight="balanced"` el peso de la clase $c$ es $n/(2\,n_c)$. Para pérdidas que **suman** sobre ejemplos, **ponderar por un entero $w$ es exactamente lo mismo que duplicar $w$ veces cada positivo** (el mismo objetivo), pero **sin duplicar filas en memoria** ni arriesgar el sobreajuste a copias de forma explícita. Lo verificamos con la regresión logística.

**(2) Mover el umbral de decisión.** Si el modelo **ordena bien** pero el corte de 0.5 es inadecuado, **elige otro umbral** $t$. **Se selecciona con datos de validación o con predicciones fuera de pliegue del entrenamiento, nunca con la prueba.** Con probabilidades **calibradas** y costos $C_{FN}$ y $C_{FP}$ el umbral de Bayes es $t^*=C_{FP}/(C_{FP}+C_{FN})$ (visto en el [Módulo 03](../03%20-%20Metricas%20y%20Ajuste%20de%20Hiperparametros/00_Metricas_de_Clasificacion.ipynb)); para maximizar el $F_1$ se busca el $t$ que lo maximiza en validación. **No cambia el ranking** (la PR-AUC queda igual): solo reubica el corte.

**(3) Aprendizaje sensible al costo.** La generalización de lo anterior: la función objetivo incorpora una **matriz de costos** por tipo de error (los pesos son un caso particular). Muchos algoritmos lo permiten con `class_weight` o `sample_weight`.

> 🌲 **A nivel de mención:** los **ensambles que equilibran internamente** (`BalancedRandomForest`, `EasyEnsemble` de `imbalanced-learn`) submuestrean la mayoría **en cada árbol/aprendiz**, de modo que no se pierde información global como en un único submuestreo; y los modelos de *boosting* ofrecen `scale_pos_weight` (XGBoost) o `class_weight`.

````python
# ============================================================
# 4. class_weight == duplicar filas (exacto en regresion logistica) y umbral elegido con predicciones fuera de pliegue
# ============================================================
# --- (1) Equivalencia exacta: ponderar por w  ==  duplicar w veces cada positivo (mismo objetivo de optimizacion) ---
Xe, ye = make_classification(600, 8, n_informative=4, weights=[0.85, 0.15], random_state=SEED)
w = 4
kw = dict(C=1.0, tol=1e-12, max_iter=20000)
con_pesos = LogisticRegression(class_weight={0: 1, 1: w}, **kw).fit(Xe, ye)
Xd = np.vstack([Xe] + [Xe[ye == 1]] * (w - 1)); yd = np.r_[ye, np.ones(int(ye.sum()) * (w - 1), dtype=int)]
duplicado = LogisticRegression(**kw).fit(Xd, yd)
dif_coef = np.abs(con_pesos.coef_ - duplicado.coef_).max(); dif_int = abs(con_pesos.intercept_[0] - duplicado.intercept_[0])
print(f"(1) class_weight={{1: {w}}} frente a duplicar {w} veces los positivos: diferencia máxima de coeficientes = {dif_coef:.1e}, del intercepto = {dif_int:.1e}")
assert dif_coef < 1e-5 and dif_int < 1e-5


# --- (2) Umbral elegido en el ENTRENAMIENTO con predicciones fuera de pliegue (no con la prueba) ---
def mejor_umbral_f1(y, p):
    pr, rc, th = precision_recall_curve(y, p)
    f1 = 2 * pr * rc / (pr + rc + 1e-12)
    return th[np.argmax(f1[:-1])]


oof = cross_val_predict(modelo_lr(), X_ent, y_ent, cv=StratifiedKFold(5, shuffle=True, random_state=0), method="predict_proba")[:, 1]
t_star = mejor_umbral_f1(y_ent, oof)
modelo = modelo_lr().fit(X_ent, y_ent)
p_te = modelo.predict_proba(X_prueba)[:, 1]
ts = np.linspace(0.02, 0.9, 89)
f1_test = np.array([f1_score(y_prueba, p_te > t) for t in ts])
f1_oof = np.array([f1_score(y_ent, oof > t) for t in ts])
print(f"(2) Umbral que maximiza F1 en el entrenamiento (fuera de pliegue): t = {t_star:.3f}. F1 en la prueba: con 0.5 = {f1_score(y_prueba, p_te > 0.5):.3f} → con t elegido = {f1_score(y_prueba, p_te > t_star):.3f} "
      f"(el mejor umbral posible mirando la prueba daría {f1_test.max():.3f}, pero eso sería hacer trampa).")
assert f1_score(y_prueba, p_te > t_star) > f1_score(y_prueba, p_te > 0.5)

fig, axes = plt.subplots(1, 2, figsize=(14, 4.4))
axes[0].plot(ts, f1_oof, color=PALETA[0], lw=2, label="F1 fuera de pliegue (entrenamiento)"); axes[0].plot(ts, f1_test, color=PALETA[1], lw=2, label="F1 en la prueba")
axes[0].axvline(0.5, color="black", ls=":", label="umbral 0.5"); axes[0].axvline(t_star, color=PALETA[2], ls="--", label=f"umbral elegido = {t_star:.2f}")
axes[0].set_xlabel("Umbral de decisión $t$"); axes[0].set_ylabel("$F_1$ de la clase minoritaria"); axes[0].legend(fontsize=8)
axes[0].set_title("Mover el umbral: el corte, no el ranking", fontweight="bold", fontsize=10)
axes[1].bar(["0.5", f"{t_star:.2f}\n(elegido en entrenamiento)", f"{ts[np.argmax(f1_test)]:.2f}\n(mirando la prueba:\ntrampa)"], [f1_score(y_prueba, p_te > 0.5), f1_score(y_prueba, p_te > t_star), f1_test.max()], color=[PALETA[9], PALETA[2], PALETA[3]])
axes[1].set_ylabel("$F_1$ en la prueba"); axes[1].set_title("La selección honesta consigue casi todo", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

```python
t_bayes = C_FP / (C_FP + C_FN)                                         # umbral teorico con probabilidades calibradas
rejilla = np.linspace(0.02, 0.9, 89)
t_oof = rejilla[np.argmin([costo(y_ent, oof > t) for t in rejilla])]   # elegido SIN mirar la prueba

costos_prueba = {t: 1000 * costo(y_prueba, p_te > t) / len(y_prueba) for t in [0.5, t_bayes, t_oof]}
for t, c in costos_prueba.items():
    print(f"umbral {t:.3f}: costo por 1000 ejemplos = {c:.0f}")

assert costos_prueba[t_bayes] < 0.8 * costos_prueba[0.5]
assert costos_prueba[t_oof] < 0.8 * costos_prueba[0.5]
print("Como perder un positivo cuesta 10 veces más, conviene alarmar con probabilidades bajas: el umbral óptimo ronda 0.1, muy lejos de 0.5.")
```

## 5. La Trampa Clave: Remuestrear **Antes** de Dividir 🚨

Es el error **más común** y el más costoso con datos desbalanceados. Si se sobremuestrea (o se aplica SMOTE) **sobre todo el conjunto** y **después** se hace la validación cruzada, ocurren **dos** problemas:

1. **Fuga de información (*leakage*).** Las **copias** (ROS) o los **puntos interpolados** (SMOTE) de un ejemplo original quedan repartidos entre los pliegues de entrenamiento y de validación: el modelo **ve en entrenamiento** una copia (o una mezcla) del mismo ejemplo que luego «predice» en validación. Los modelos que **memorizan** (bosques, 1-NN) lo aprovechan al máximo.
2. **Prevalencia falsa.** La validación ahora tiene **50 % de positivos** (en vez de 5 %): la PR-AUC y la precisión se **inflan** solo por eso, aunque el modelo no mejore.

La **solución correcta** es remuestrear **únicamente el conjunto de entrenamiento de cada pliegue** y evaluar sobre el pliegue de validación **intacto** (con su prevalencia real). En `imbalanced-learn` esto lo hace su `Pipeline` (que aplica el remuestreo **solo en `fit`**); sin la librería, se escribe el bucle a mano (lo hacemos aquí).

Para **aislar** los dos efectos medimos **cuatro** protocolos con **tres semillas** y dos modelos (regresión logística, que casi no memoriza, y un bosque, que sí), contra la **«verdad»** (la prueba enorme):

| Protocolo | Qué hace | Qué mide |
|---|---|---|
| **A. Remuestrear todo, validar en todo** | El error típico | Fuga **+** prevalencia falsa |
| **B. Remuestrear todo, validar solo en filas originales** | Quita el efecto de la prevalencia (la validación vuelve a ser 5 %) | **Solo la fuga** |
| **C. Remuestrear dentro del pliegue de entrenamiento** | **Correcto** | Estimación honesta |
| **D. Prueba enorme** | La «verdad» | Desempeño real |

````python
# ============================================================
# 5. Cuantificar la fuga: remuestrear antes de dividir (A, B) frente al protocolo correcto (C) y la verdad (D)
# ============================================================
t0 = time.time()
modelos = {"Regresión logística": lambda: modelo_lr(), "Bosque aleatorio (25 árboles)": lambda: RandomForestClassifier(25, random_state=0)}
remuestreos = {"ROS": lambda X, y, s: sobre_aleatorio(X, y, s)[:2], "SMOTE": lambda X, y, s: smote_cero(X, y, 5, s)[:2]}
registros = []
for semilla in range(3):
    X, y = muestra_entrenamiento(semilla)
    cv = StratifiedKFold(5, shuffle=True, random_state=semilla)
    for mn, mk in modelos.items():
        for rn, fn in remuestreos.items():
            Xr, yr = fn(X, y, semilla)
            original = np.r_[np.ones(len(y), bool), np.zeros(len(yr) - len(y), bool)]                    # False = copia o sintetico
            ap_a, ap_b, ap_c = [], [], []
            for tr, va in cv.split(Xr, yr):                                                              # A y B: el remuestreo ya ocurrio ANTES de dividir
                p = mk().fit(Xr[tr], yr[tr]).predict_proba(Xr[va])[:, 1]
                ap_a.append(average_precision_score(yr[va], p))
                ap_b.append(average_precision_score(yr[va][original[va]], p[original[va]]))
            for tr, va in cv.split(X, y):                                                                # C: se remuestrea SOLO el entrenamiento del pliegue
                Xt, yt = fn(X[tr], y[tr], semilla)
                ap_c.append(average_precision_score(y[va], mk().fit(Xt, yt).predict_proba(X[va])[:, 1]))
            p_d = mk().fit(*fn(X, y, semilla)).predict_proba(X_prueba)[:, 1]                              # D: verdad
            registros.append({"modelo": mn, "remuestreo": rn, "A. todo/todo": np.mean(ap_a), "B. solo originales": np.mean(ap_b), "C. correcto (en el pliegue)": np.mean(ap_c), "D. verdad (prueba enorme)": average_precision_score(y_prueba, p_d)})
df = pd.DataFrame(registros).groupby(["modelo", "remuestreo"]).mean()
display(df.round(3))
print(f"Línea base de la PR-AUC: prevalencia real = {y_prueba.mean():.3f} (en la validación remuestreada sería 0.5).  [{time.time() - t0:.0f} s]")

fig, ax = plt.subplots(figsize=(12, 4.6))
etiquetas = [f"{m.split(' (')[0]}\n+ {r}" for m, r in df.index]
ancho = 0.2
for k, (col, c) in enumerate(zip(df.columns, [PALETA[1], PALETA[3], PALETA[2], "black"])):
    ax.bar(np.arange(len(df)) + (k - 1.5) * ancho, df[col], ancho, color=c, label=col)
ax.set_xticks(np.arange(len(df))); ax.set_xticklabels(etiquetas, fontsize=9); ax.set_ylabel("PR-AUC (average precision)"); ax.legend(fontsize=8, ncol=2)
ax.set_title("Remuestrear antes de dividir infla la PR-AUC; el protocolo correcto (verde) acierta la verdad (negro)", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()

lr_rows, rf_rows = df.loc["Regresión logística"], df.loc["Bosque aleatorio (25 árboles)"]
assert (df["A. todo/todo"] > df["D. verdad (prueba enorme)"] + 0.15).all()                                  # A: siempre muy optimista (fuga + prevalencia)
assert (rf_rows["B. solo originales"] > rf_rows["D. verdad (prueba enorme)"] + 0.3).all()                    # B: con el bosque, la FUGA sola ya infla muchisimo
assert (lr_rows["B. solo originales"] - lr_rows["D. verdad (prueba enorme)"]).abs().max() < 0.06            # B: la regresion logistica casi no memoriza: su inflado era la PREVALENCIA
assert ((df["C. correcto (en el pliegue)"] - df["D. verdad (prueba enorme)"]).abs() < 0.1).all()             # C: estimacion honesta

if imblearn is not None:                                                                                    # el Pipeline de imblearn aplica el remuestreo SOLO al ajustar
    X, y = muestra_entrenamiento(0)
    pipe = PipelineIB([("smote", SMOTE_IB(random_state=0)), ("clf", modelo_lr())])
    ap_ib = cross_val_score(pipe, X, y, cv=StratifiedKFold(5, shuffle=True, random_state=0), scoring="average_precision").mean()
    ap_man = df.loc[("Regresión logística", "SMOTE"), "C. correcto (en el pliegue)"]
    print(f"imblearn.Pipeline(SMOTE → LR) en validación cruzada (semilla 0): AP = {ap_ib:.3f} | nuestro protocolo C promediado en 3 semillas: {ap_man:.3f} | verdad: {df.loc[('Regresión logística', 'SMOTE'), 'D. verdad (prueba enorme)']:.3f}")
    assert abs(ap_ib - df.loc[("Regresión logística", "SMOTE"), "D. verdad (prueba enorme)"]) < 0.12
else:
    print("imbalanced-learn no está instalado: usamos el protocolo manual (equivalente al Pipeline de imblearn).")
````

## 6. La Comparación Honesta: ¿Qué Estrategia Gana de Verdad? 🏁

Con **una** muestra de entrenamiento de 2 000 ejemplos, comparamos **siete** estrategias y **dos** modelos sobre la **prueba enorme**, repitiendo con **6 semillas** (6 muestras de entrenamiento distintas) para reportar **media ± desviación**. Las métricas, elegidas para el desbalance:

- **PR-AUC (AP):** no depende del umbral; mide el **ranking**. Si el remuestreo no mejora el ranking, **no mejora** el modelo, solo mueve el corte.
- **$F_1$ al umbral 0.5, *recall* y precisión** de la clase minoritaria: lo que se ve **al decidir** con el corte por defecto.
- **$F_1$ con umbral elegido en el entrenamiento** (predicciones fuera de pliegue): el costo de **mover el corte** con selección honesta.

Estrategias: **(1)** sin tratar, **(2)** `class_weight="balanced"`, **(3)** submuestreo aleatorio, **(4)** sobremuestreo aleatorio, **(5)** SMOTE, **(6)** **umbral elegido** (sin remuestrear). **Todas** remuestrean **solo el entrenamiento** (sin fuga).

````python
# ============================================================
# 6. Comparacion honesta: 6 estrategias x 2 modelos x 6 semillas, evaluadas en la prueba enorme
# ============================================================
t0 = time.time()
SEMILLAS = range(6)
fabricas = {"Regresión logística": lambda cw=None: modelo_lr(class_weight=cw),
            "Bosque aleatorio": lambda cw=None: RandomForestClassifier(40, random_state=0, class_weight=cw)}
ESTRATEGIAS = ["1. Sin tratar", "2. class_weight", "3. Submuestreo", "4. Sobremuestreo", "5. SMOTE", "6. Umbral elegido"]
registros = []
for semilla in SEMILLAS:
    X, y = muestra_entrenamiento(semilla)
    for mn, mk in fabricas.items():
        for est in ESTRATEGIAS:
            umbral = 0.5
            if est == "1. Sin tratar":      m = mk().fit(X, y)
            elif est == "2. class_weight":  m = mk("balanced").fit(X, y)
            elif est == "3. Submuestreo":   m = mk().fit(*sub_aleatorio(X, y, semilla)[:2])
            elif est == "4. Sobremuestreo": m = mk().fit(*sobre_aleatorio(X, y, semilla)[:2])
            elif est == "5. SMOTE":         m = mk().fit(*smote_cero(X, y, 5, semilla)[:2])
            else:
                m = mk().fit(X, y)
                oof = cross_val_predict(mk(), X, y, cv=StratifiedKFold(5, shuffle=True, random_state=semilla), method="predict_proba")[:, 1]
                umbral = mejor_umbral_f1(y, oof)
            p = m.predict_proba(X_prueba)[:, 1]
            pred = p > umbral
            registros.append({"modelo": mn, "estrategia": est, "semilla": semilla, "PR-AUC": average_precision_score(y_prueba, p), "F1": f1_score(y_prueba, pred),
                              "recall": recall_score(y_prueba, pred), "precisión": precision_score(y_prueba, pred, zero_division=0)})
res = pd.DataFrame(registros)
media = res.groupby(["modelo", "estrategia"])[["PR-AUC", "F1", "recall", "precisión"]].mean()
desv = res.groupby(["modelo", "estrategia"])[["PR-AUC", "F1", "recall", "precisión"]].std(ddof=0)
presentacion = media.round(3).astype(str) + " ± " + desv.round(3).astype(str)
display(presentacion)
print(f"[{time.time() - t0:.0f} s]  Línea base de la PR-AUC = prevalencia = {y_prueba.mean():.3f}")

fig, axes = plt.subplots(2, 2, figsize=(15, 8))
for fila, mn in enumerate(fabricas):
    for col, met in enumerate(["PR-AUC", "F1"]):
        ax = axes[fila, col]
        mm, dd = media.loc[mn][met], desv.loc[mn][met]
        ax.bar(np.arange(len(ESTRATEGIAS)), mm.loc[ESTRATEGIAS], yerr=dd.loc[ESTRATEGIAS], color=[PALETA[9], PALETA[0], PALETA[5], PALETA[3], PALETA[1], PALETA[2]], capsize=3)
        ax.set_xticks(np.arange(len(ESTRATEGIAS))); ax.set_xticklabels([e.split(". ")[1].replace(" ", "\n") for e in ESTRATEGIAS], fontsize=8)
        ax.set_ylabel(met); ax.set_title(f"{mn}: {met} (media ± desv., 6 semillas)", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()

ap = media["PR-AUC"].unstack()[ESTRATEGIAS]; f1 = media["F1"].unstack()[ESTRATEGIAS]
mejor_f1 = f1.idxmax(axis=1)
print("Mejor estrategia por F1 (umbral 0.5, salvo la 6):", mejor_f1.to_dict())
print(f"PR-AUC de la regresión logística: sin tratar = {ap.loc['Regresión logística', '1. Sin tratar']:.3f} | SMOTE = {ap.loc['Regresión logística', '5. SMOTE']:.3f} | class_weight = {ap.loc['Regresión logística', '2. class_weight']:.3f}")
assert np.isclose(ap.loc["Regresión logística", "6. Umbral elegido"], ap.loc["Regresión logística", "1. Sin tratar"])             # mover el umbral no toca el ranking
assert f1.loc["Regresión logística", "6. Umbral elegido"] > f1.loc["Regresión logística", "1. Sin tratar"] + 0.05                 # ... pero mejora mucho el F1
assert abs(ap.loc["Regresión logística", "2. class_weight"] - ap.loc["Regresión logística", "4. Sobremuestreo"]) < 0.02          # class_weight ~ sobremuestreo en LR
assert (ap["1. Sin tratar"] >= ap[["3. Submuestreo", "5. SMOTE"]].min(axis=1) - 0.01).all()                                      # remuestrear NO mejora el ranking (honesto)
````

---
**Lectura honesta de la comparación.** (1) **Para el *ranking* (PR-AUC), remuestrear no ayuda**: la regresión logística **sin tratar** tiene una PR-AUC **igual o mayor** que con submuestreo, sobremuestreo o SMOTE (el submuestreo es el peor: tira información); el bosque **sin tratar** y con `class_weight` o sobremuestreo quedan parecidos, y SMOTE **no lo mejora**. (2) Lo que **sí** cambia es el **corte**, y no igual en todos los casos: en la **regresión logística** `class_weight`, el submuestreo, el sobremuestreo y SMOTE suben el *recall* de ≈ 0.19 a ≈ 0.72–0.74 a costa de la precisión (de ≈ 0.63 a ≈ 0.19); en el **bosque** el efecto depende: el submuestreo lo sube mucho (≈ 0.71), SMOTE moderadamente (≈ 0.34) y `class_weight` y el sobremuestreo aleatorio **poco** (≈ 0.23 y ≈ 0.15), porque el bosque **memoriza las copias** y sigue dando probabilidades extremas. (3) **`class_weight` y el sobremuestreo aleatorio son casi lo mismo en la regresión logística** (ya lo habíamos probado: son el mismo objetivo). (4) **Elegir bien el umbral** con predicciones fuera de pliegue **consigue una mejora del $F_1$ comparable o superior a cualquier remuestreo, sin tocar los datos ni el ranking**. (5) **SMOTE sí ayuda en un caso**: el bosque con SMOTE mejora mucho su $F_1$ al 0.5 respecto a no tratar (porque desplaza el corte), pero **el umbral elegido logra lo mismo o más** sin crear datos sintéticos. **Conclusión:** antes de generar datos sintéticos, prueba **(a)** mejores métricas, **(b)** mover el umbral y **(c)** `class_weight`; el remuestreo es **una opción más**, no un remedio automático, y exige las precauciones de la sección 5.

## 7. Guía de Decisión para Datos Desbalanceados 🧭

| Situación | Qué hacer | Por qué |
|---|---|---|
| **Siempre** | Evalúa con **PR-AUC, $F_1$, recall y precisión** de la minoría y la **matriz de confusión**; estratifica las divisiones | La exactitud es engañosa; la PR-AUC tiene como línea base la prevalencia |
| **El modelo ordena bien (PR-AUC aceptable) pero no detecta positivos** | **Mueve el umbral** con validación o predicciones fuera de pliegue (y, si necesitas probabilidades, **calibra**) | El problema es el corte, no el ranking |
| **Modelo con pérdida aditiva (lineal, redes, árboles con `class_weight`)** | Prueba `class_weight="balanced"` o `sample_weight` | Equivale a sobremuestrear **sin duplicar filas** |
| **Muy pocos positivos (decenas)** | Cuidado con SMOTE: interpolar entre pocos puntos **amplifica el ruido**; prefiere pesos, regularización y más datos | Con pocos datos todo es varianza |
| **Muchos datos y la mayoría es redundante** | **Submuestreo** de la mayoría (o *BalancedRandomForest*) | Ahorra cómputo con poca pérdida |
| **Variables mixtas (categóricas y numéricas)** | `SMOTENC`, o evita la interpolación | La interpolación no tiene sentido en categorías |
| **Alta dimensión** | Reduce la dimensión **antes** o evita SMOTE | Las distancias pierden sentido |
| **Si remuestreas** | **Solo dentro del entrenamiento de cada pliegue** (`imblearn.Pipeline` o bucle propio); **nunca** valides sobre datos remuestreados; **corrige las probabilidades** si las usas | Evita la fuga y la prevalencia falsa |
| **Costos de error conocidos** | Usa **aprendizaje sensible al costo** y el umbral de Bayes $t^*=C_{FP}/(C_{FP}+C_{FN})$ con probabilidades calibradas | Optimiza lo que importa al negocio |

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Por qué rompe** | La mayoría domina el gradiente y el umbral de 0.5 casi nunca se supera. El clasificador trivial tiene 95 % de *accuracy* y recall 0; la **PR-AUC** (línea base = prevalencia) refleja el problema mejor que el ROC-AUC |
| **RUS / ROS** | Submuestreo (**pierde** información) y sobremuestreo (**copias exactas**, memorizables). Cambian la prevalencia aparente: la **corrección por prior** $p=\pi p'/(\pi p'+(1-\pi)(1-p'))$ restaura las probabilidades sin cambiar el ranking |
| **SMOTE** | $\mathbf x_{\text{nuevo}}=\mathbf x_i+\lambda(\mathbf x_{\text{nn}}-\mathbf x_i)$ con vecinos **minoritarios**; propiedades verificadas (cada sintético **sobre un segmento**, clases **iguales**, originales intactos; contraste opcional con `imblearn`). Variantes: *Borderline-SMOTE*, *ADASYN*, *Tomek*, *ENN* |
| **Sin remuestrear** | `class_weight` **equivale exactamente** a duplicar filas (regresión logística); **mover el umbral** (con predicciones fuera de pliegue) y el aprendizaje sensible al costo |
| **La trampa** | **Remuestrear antes de dividir** infla la PR-AUC por **dos** motivos: **prevalencia falsa** (en la regresión logística explica casi todo el inflado) y **fuga** (con el bosque la fuga sola lleva la PR-AUC a ≈ 1). La solución: remuestrear **solo el entrenamiento de cada pliegue** |
| **Comparación honesta** | Para el **ranking**, remuestrear **no mejora** (a veces empeora); lo que cambia es el **corte**. **Elegir bien el umbral** logra tanto o más que SMOTE sin crear datos. SMOTE ayuda con el bosque en el $F_1$ al 0.5, no en la PR-AUC |
