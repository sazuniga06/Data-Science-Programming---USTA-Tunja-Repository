## 2. Desplazamiento de Distribución (*Dataset Shift*): Cuando el Mundo Cambia 🌊

### 2.1 El supuesto que casi nunca se cumple

Todo el curso supone que los datos de **entrenamiento** y los de **uso** salen de la **misma** distribución conjunta. Formalmente, el modelo se entrena y se valida con $P_{\text{tr}}(\mathbf x, y)$ pero **se usa** sobre $P_{\text{te}}(\mathbf x, y)$, y el supuesto es $P_{\text{tr}} = P_{\text{te}}$. Cuando **no** lo es hablamos de ***dataset shift*** (Quiñonero-Candela et al., 2009). Como $P(\mathbf x, y)$ se factoriza de dos maneras,

$$P(\mathbf x, y) = P(y \mid \mathbf x)\,P(\mathbf x) = P(\mathbf x \mid y)\,P(y),$$

hay **tres** formas básicas de que algo cambie:

| Tipo | Qué cambia | Qué se mantiene | Ejemplo | Consecuencia típica |
|---|---|---|---|---|
| ***Covariate shift*** (de covariables) | $P(\mathbf x)$ | $P(y\mid\mathbf x)$ | Un modelo de crédito entrenado con clientes de una ciudad se usa en otra con ingresos distintos | Daño **grande si el modelo está mal especificado** o extrapola; **leve** si el modelo captura bien $P(y\mid\mathbf x)$ en la nueva región |
| ***Label* o *prior shift*** (de etiqueta) | $P(y)$ | $P(\mathbf x\mid y)$ | La **prevalencia** de una enfermedad sube en una epidemia | La **exactitud** casi no se mueve pero la **precisión** (VPP) **se desploma** o se infla; las probabilidades quedan **descalibradas** |
| ***Concept shift*** (de concepto, *concept drift*) | $P(y\mid\mathbf x)$ | $P(\mathbf x)$ | Cambia el comportamiento del fraude, o el significado de «correo no deseado» | El modelo **aprendió una regla que ya no vale**: el desempeño puede caer **por debajo del azar** |

Los tres pueden ocurrir **a la vez** y **no se distinguen** mirando solo la exactitud. Para ver su efecto, entrenamos modelos en un «laboratorio» sintético donde **conocemos la verdad**, y los evaluamos en el mismo laboratorio **después de cambiar una sola cosa**.

````python
# ============================================================
# 2. Tres tipos de desplazamiento sobre modelos ya entrenados (laboratorio sintetico con verdad conocida)
# ============================================================
def sig(z):
    return 1 / (1 + np.exp(-z))


# --- (a) Covariate shift: cambia P(x); P(y|x) es SIEMPRE la misma (no depende de mu) ---
def muestra_cov(n, mu, rng):
    """x1 ~ N(mu, 1), x2 ~ N(0, 1);  P(y=1|x) = sigmoide(4 (x2 - 0.8 sen(2 x1)))  -> frontera curva."""
    X = np.column_stack([rng.normal(mu, 1, n), rng.normal(0, 1, n)])
    y = (rng.random(n) < sig(4 * (X[:, 1] - 0.8 * np.sin(2 * X[:, 0])))).astype(int)
    return X, y


X_tr, y_tr = muestra_cov(1000, 0.0, np.random.default_rng(SEED))
modelos_cov = {"Regresión logística (lineal)": LogisticRegression().fit(X_tr, y_tr),
               "Bosque aleatorio": RandomForestClassifier(100, min_samples_leaf=5, random_state=SEED).fit(X_tr, y_tr)}
mus = [0.0, 0.5, 1.0, 1.5, 2.0]
acc_cov = {n: [] for n in modelos_cov}
for mu in mus:
    X_te, y_te = muestra_cov(4000, mu, np.random.default_rng(1))
    for n, m in modelos_cov.items():
        acc_cov[n].append(accuracy_score(y_te, m.predict(X_te)))
acc_cov = pd.DataFrame(acc_cov, index=pd.Index(mus, name="desplazamiento de x1 (mu)"))


# --- (b) Label shift: cambia la prevalencia P(y); P(x|y) fija (dos nubes gaussianas) ---
def muestra_prior(n, prev, rng):
    y = (rng.random(n) < prev).astype(int)
    X = np.where(y[:, None] == 1, 1.0, -1.0) + rng.normal(0, 1.3, (n, 2))
    return X, y


Xp, yp = muestra_prior(2000, 0.5, np.random.default_rng(SEED))
mod_prior = LogisticRegression().fit(Xp, yp)
filas = []
for prev in [0.5, 0.2, 0.1, 0.05, 0.02]:
    X_te, y_te = muestra_prior(20000, prev, np.random.default_rng(1))
    pred = mod_prior.predict(X_te)
    filas.append({"prevalencia": prev, "accuracy": accuracy_score(y_te, pred), "precisión": precision_score(y_te, pred),
                  "recall": recall_score(y_te, pred), "ROC-AUC": roc_auc_score(y_te, mod_prior.predict_proba(X_te)[:, 1])})
tab_prior = pd.DataFrame(filas).set_index("prevalencia")


# --- (c) Concept shift: gira la frontera P(y|x); P(x) no cambia ---
def muestra_concepto(n, theta, rng):
    X = rng.normal(0, 1.5, (n, 2))
    w = np.array([np.cos(theta), np.sin(theta)])
    y = ((X @ w + rng.normal(0, 0.5, n)) > 0).astype(int)
    return X, y


Xc, yc = muestra_concepto(2000, 0.0, np.random.default_rng(SEED))
mod_conc = LogisticRegression().fit(Xc, yc)
grados = [0, 15, 30, 45, 90, 135, 180]
acc_conc = []
for g in grados:
    X_te, y_te = muestra_concepto(20000, np.radians(g), np.random.default_rng(1))
    acc_conc.append(accuracy_score(y_te, mod_conc.predict(X_te)))

display(acc_cov.round(3))
display(tab_prior.round(3))
print("Concept shift (accuracy según el giro de la frontera):", dict(zip(grados, np.round(acc_conc, 3))))

fig, axes = plt.subplots(1, 3, figsize=(16, 4.4))
for (n, col), c in zip(acc_cov.items(), [PALETA[0], PALETA[2]]):
    axes[0].plot(acc_cov.index, col, "o-", color=c, lw=2, label=n)
axes[0].set_xlabel("Desplazamiento de $x_1$ en producción ($\\mu$)"); axes[0].set_ylabel("Accuracy"); axes[0].legend(fontsize=8)
axes[0].set_title("(a) Covariate shift: daña más al modelo lineal", fontweight="bold", fontsize=10)
for c, col in zip(PALETA, ["accuracy", "precisión", "recall", "ROC-AUC"]):
    axes[1].plot(100 * tab_prior.index, tab_prior[col], "o-", color=c, lw=2, label=col)
axes[1].set_xscale("log"); axes[1].set_xticks([2, 5, 10, 20, 50]); axes[1].set_xticklabels(["2", "5", "10", "20", "50"]); axes[1].minorticks_off(); axes[1].set_xlabel("Prevalencia de la clase positiva en producción (%)"); axes[1].set_ylabel("Métrica"); axes[1].legend(fontsize=8)
axes[1].set_title("(b) Label shift: la precisión se desploma", fontweight="bold", fontsize=10)
axes[2].plot(grados, acc_conc, "o-", color=PALETA[1], lw=2); axes[2].axhline(0.5, color="black", ls="--", label="Azar")
axes[2].set_xlabel("Giro de la frontera verdadera (grados)"); axes[2].set_ylabel("Accuracy"); axes[2].legend(fontsize=8)
axes[2].set_title("(c) Concept shift: por debajo del azar", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()

caida_lineal = acc_cov.iloc[0, 0] - acc_cov.iloc[-1, 0]
caida_bosque = acc_cov.iloc[0, 1] - acc_cov.iloc[-1, 1]
assert caida_lineal > 0.04 and caida_lineal > caida_bosque               # covariate shift: el modelo mal especificado sufre mas
assert abs(tab_prior["accuracy"].iloc[-1] - tab_prior["accuracy"].iloc[0]) < 0.03     # label shift: el accuracy casi no se mueve...
assert tab_prior["precisión"].iloc[-1] < 0.2 < tab_prior["precisión"].iloc[0]          # ... pero la precision se desploma
assert abs(tab_prior["ROC-AUC"].iloc[-1] - tab_prior["ROC-AUC"].iloc[0]) < 0.02       # ... y el AUC (que no depende de P(y)) sigue igual
assert acc_conc[0] > 0.85 and acc_conc[4] < 0.6 and acc_conc[-1] < 0.2                 # concept shift: del 90 % al azar y peor que el azar
print(f"\nCovariate shift: el lineal pierde {100 * caida_lineal:.1f} pts y el bosque {100 * caida_bosque:.1f} pts. "
      f"Label shift: con 2 % de prevalencia el accuracy sigue en {tab_prior['accuracy'].iloc[-1]:.2f} pero la precisión cae a {tab_prior['precisión'].iloc[-1]:.2f}.")
````

---
**Lecturas clave de la demostración:**

- **(a)** El *covariate shift* **no daña a todos los modelos igual**: el lineal está **mal especificado** (la frontera real es curva) y solo acierta cerca de donde entrenó, así que sufre cuando los datos se mudan; el bosque, más flexible, se degrada **menos** (aunque tampoco extrapola: fuera del rango visto repite lo que vio en el borde).
- **(b)** Un *label shift* es **traicionero**: la exactitud y el ROC-AUC (que no dependen de la prevalencia) **parecen sanos**, pero la **precisión** (que sí depende de ella, vía el teorema de Bayes) **se hunde**: con 2 % de positivos, la mayoría de las alarmas son falsas. Es el mismo fenómeno de la «paradoja de la exactitud» del [Módulo 03](../03%20-%20Metricas%20y%20Ajuste%20de%20Hiperparametros/00_Metricas_de_Clasificacion.ipynb), ahora causado por el **mundo** y no por el diseñador.
- **(c)** Un *concept shift* **invalida la regla** aprendida: con el giro de 180° el modelo acierta ~10 % (peor que lanzar una moneda).

## 3. Detectar y Corregir el *Covariate Shift*: Validación Adversaria y Reponderación 🔎⚖️

### 3.1 Validación adversaria: «¿puede un clasificador distinguir entrenamiento de producción?»

Si el desplazamiento afecta a $P(\mathbf x)$, **se puede detectar sin etiquetas**. Construimos un conjunto con **todas** las filas de entrenamiento (dominio $d=0$) y de producción (dominio $d=1$) y entrenamos un **clasificador de dominio** $\hat P(d=1\mid\mathbf x)$ con validación cruzada:

- **AUC ≈ 0.5**: no distingue los dominios → **no hay desplazamiento de covariables detectable**.
- **AUC ≫ 0.5**: sí los distingue → hay desplazamiento, y la **importancia de variables** del clasificador de dominio dice **cuáles variables** lo causan.

### 3.2 Reponderación por importancia

Si $P(y\mid\mathbf x)$ no cambia, el riesgo en producción se puede escribir con datos de entrenamiento **reponderados** (Shimodaira, 2000):

$$\mathbb E_{\text{te}}\big[L(y, f(\mathbf x))\big]=\mathbb E_{\text{tr}}\!\left[\,w(\mathbf x)\,L(y, f(\mathbf x))\right],\qquad w(\mathbf x)=\frac{P_{\text{te}}(\mathbf x)}{P_{\text{tr}}(\mathbf x)}=\underbrace{\frac{\hat P(d=1\mid\mathbf x)}{\hat P(d=0\mid\mathbf x)}}_{\text{del clasificador de dominio}}\cdot\frac{n_{\text{tr}}}{n_{\text{te}}}.$$

Es decir, los ejemplos de entrenamiento que **se parecen a los de producción** pesan más. Se aplica con `sample_weight`. Tiene un **costo**: si el desplazamiento es grande, **pocos** ejemplos concentran **casi todo** el peso. El **tamaño efectivo de muestra** (Kish) lo mide:

$$n_{\text{ef}}=\frac{\left(\sum_i w_i\right)^2}{\sum_i w_i^2},\qquad \frac{n_{\text{ef}}}{n}\in(0,1].$$

Con $n_{\text{ef}}/n\to 0$ la corrección es **inestable**. Probamos ambas ideas: validación adversaria en datos reales (*Breast Cancer*, con un desplazamiento **fabricado** por selección sobre el radio del tumor) y reponderación en el laboratorio sintético, donde **conocemos los pesos verdaderos** $w=\mathcal N(x_1;\mu,1)/\mathcal N(x_1;0,1)$ y podemos **verificar** los estimados.

````python
# ============================================================
# 3. Validacion adversaria (datos reales) y reponderacion por importancia (laboratorio con pesos verdaderos)
# ============================================================
bc = load_breast_cancer()
Xb = bc.data


def validacion_adversaria(X, dominio, nombres, seed=SEED):
    """AUC (validacion cruzada) de un clasificador que distingue dominio 0 vs 1, e importancia de variables."""
    rf = RandomForestClassifier(60, min_samples_leaf=3, random_state=seed)
    p = cross_val_predict(rf, X, dominio, cv=StratifiedKFold(5, shuffle=True, random_state=seed), method="predict_proba")[:, 1]
    imp = pd.Series(rf.fit(X, dominio).feature_importances_, index=nombres).sort_values(ascending=False)
    return roc_auc_score(dominio, p), imp


dom_azar = (np.random.default_rng(0).random(len(Xb)) < 0.5).astype(int)        # "produccion" = mitad al azar: SIN desplazamiento
dom_sel = (Xb[:, 0] > np.median(Xb[:, 0])).astype(int)                          # "produccion" = tumores de radio grande: CON desplazamiento
auc_azar, imp_azar = validacion_adversaria(Xb, dom_azar, bc.feature_names)
auc_sel, imp_sel = validacion_adversaria(Xb, dom_sel, bc.feature_names)
print(f"AUC adversario | división al azar: {auc_azar:.3f}  | selección por radio: {auc_sel:.3f}")
print("Variables que más delatan el desplazamiento (selección por radio):", list(imp_sel.index[:4]))
assert auc_azar < 0.6 and auc_sel > 0.95                                        # sin desplazamiento ~0.5; con desplazamiento ~1.0
assert {"mean radius", "mean perimeter", "mean area"} & set(imp_sel.index[:3])  # las culpables son las que miden tamano


# --- Reponderacion por importancia en el laboratorio (misma muestra_cov de la seccion 2) ---
def corrida_iw(mu, seed):
    rng = np.random.default_rng(seed)
    Xs, ys = muestra_cov(1000, 0.0, rng)                    # entrenamiento (con etiquetas)
    Xu, _ = muestra_cov(1000, mu, rng)                      # produccion SIN etiquetas (solo covariables)
    Xt, yt = muestra_cov(4000, mu, rng)                     # produccion con etiquetas, solo para evaluar
    dom = LogisticRegression(C=100).fit(np.vstack([Xs, Xu]), np.r_[np.zeros(len(Xs)), np.ones(len(Xu))])
    p = dom.predict_proba(Xs)[:, 1]
    w_est = (p / (1 - p)) * (len(Xs) / len(Xu))             # w(x) = P(d=1|x)/P(d=0|x) * n_tr/n_te
    w_ver = stats.norm.pdf(Xs[:, 0], mu, 1) / stats.norm.pdf(Xs[:, 0], 0, 1)    # densidad verdadera (solo posible en simulacion)
    Xo, yo = muestra_cov(1000, mu, rng)                     # "oraculo": entrenar con etiquetas de produccion
    return {"sin pesos": LogisticRegression().fit(Xs, ys).score(Xt, yt),
            "pesos estimados": LogisticRegression().fit(Xs, ys, sample_weight=w_est).score(Xt, yt),
            "pesos verdaderos": LogisticRegression().fit(Xs, ys, sample_weight=w_ver).score(Xt, yt),
            "oráculo (etiquetas de producción)": LogisticRegression().fit(Xo, yo).score(Xt, yt),
            "n_ef/n": w_est.sum() ** 2 / (w_est ** 2).sum() / len(w_est),
            "Spearman(est, verdad)": stats.spearmanr(w_est, w_ver)[0]}, (Xs[:, 0], w_est, w_ver)


mus_iw = [0.5, 1.0, 1.5, 2.0, 2.5]
res = {mu: [corrida_iw(mu, s)[0] for s in range(8)] for mu in mus_iw}
tab_iw = pd.DataFrame({mu: {k: np.mean([r[k] for r in rs]) for k in rs[0]} for mu, rs in res.items()}).T
tab_iw.index.name = "mu (desplazamiento)"
display(tab_iw.round(3))
assert (tab_iw["Spearman(est, verdad)"] > 0.95).all()                        # el clasificador de dominio recupera los pesos verdaderos
assert tab_iw.loc[1.5, "pesos estimados"] > tab_iw.loc[1.5, "sin pesos"] + 0.03     # la reponderacion recupera parte (casi todo) de lo perdido
assert tab_iw["n_ef/n"].is_monotonic_decreasing and tab_iw.loc[2.5, "n_ef/n"] < 0.05

_, (x1_s, w_e, w_v) = corrida_iw(1.5, 0)
fig, axes = plt.subplots(1, 3, figsize=(16, 4.4))
orden = np.argsort(x1_s)
axes[0].plot(x1_s[orden], w_v[orden], color="black", lw=2, label="peso verdadero")
axes[0].scatter(x1_s, w_e, s=6, color=PALETA[1], alpha=0.5, label="peso estimado (clasificador de dominio)")
axes[0].set_yscale("log"); axes[0].set_xlabel("$x_1$"); axes[0].set_ylabel("$w(x)$ (escala log)"); axes[0].legend(fontsize=8)
axes[0].set_title("Los pesos estimados siguen a los verdaderos ($\\mu=1.5$)", fontweight="bold", fontsize=10)
for c, col in zip(PALETA, ["sin pesos", "pesos estimados", "pesos verdaderos", "oráculo (etiquetas de producción)"]):
    axes[1].plot(tab_iw.index, tab_iw[col], "o-", color=c, lw=2, label=col)
axes[1].set_xlabel("Desplazamiento $\\mu$"); axes[1].set_ylabel("Accuracy en producción"); axes[1].legend(fontsize=7)
axes[1].set_title("La reponderación corrige parte de la caída", fontweight="bold", fontsize=10)
axes[2].plot(tab_iw.index, tab_iw["n_ef/n"], "o-", color=PALETA[3], lw=2)
axes[2].set_xlabel("Desplazamiento $\\mu$"); axes[2].set_ylabel("$n_{ef}/n$"); axes[2].set_yscale("log")
axes[2].set_title("Precio: el tamaño efectivo de muestra se hunde", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
print(f"Con mu = 1.5: sin pesos {tab_iw.loc[1.5, 'sin pesos']:.3f} → con pesos {tab_iw.loc[1.5, 'pesos estimados']:.3f} (oráculo {tab_iw.loc[1.5, 'oráculo (etiquetas de producción)']:.3f}). "
      f"Con mu = 2.5 queda un {100 * tab_iw.loc[2.5, 'n_ef/n']:.1f} % de muestra efectiva y la corrección ya es parcial.")
````

---
**Lectura.** La **validación adversaria** detecta el desplazamiento **sin una sola etiqueta** (AUC ≈ 0.5 con una división al azar frente a ≈ 1.0 con la selección por tamaño del tumor) y **señala las variables culpables** (las que miden tamaño). La **reponderación** recupera gran parte de lo perdido con desplazamientos moderados, y el clasificador de dominio estima pesos casi idénticos a los verdaderos (correlación de Spearman > 0.95). Pero hay tres **advertencias honestas**:

1. **Necesita solapamiento**: si en producción hay regiones donde el entrenamiento **no tiene datos**, el peso es infinito y no hay corrección posible (con $\mu=2.5$ ya casi no la hay: el $n_{\text{ef}}$ es < 5 %).
2. **Solo corrige el *covariate shift***: si $P(y\mid\mathbf x)$ cambió, reponderar **no ayuda**.
3. **Con modelos bien especificados casi no hace falta**: el bosque de la sección 2 se degradó bastante menos (≈ 4 puntos frente a ≈ 8 del modelo lineal). La reponderación es sobre todo un **remedio para modelos que no capturan bien la frontera** (lineales, por ejemplo).

## 4. ¿Por Qué la Validación Puede Mentir sobre la Producción? ⏳

Un buen número en validación es **condición necesaria, no suficiente**. Las causas más comunes de que la producción decepcione son:

| Causa | Qué pasa | Dónde se trata |
|---|---|---|
| **Fuga de datos** | La validación vio información que en producción no existirá | [Módulo 00](../00%20-%20Introduccion%20al%20Machine%20Learning/03_Importancia_del_Preprocesamiento.ipynb) |
| **Sesgo de selección** | Se eligió el mejor de muchos modelos con la misma validación | [Módulo 07](../07%20-%20Seleccion%20de%20Modelos%20y%20Optimizacion/01_Validacion_Cruzada_Avanzada.ipynb) (validación anidada) |
| **Divisiones que ignoran la estructura** | Mezclar el tiempo o repetir pacientes entre entrenamiento y prueba | [Módulo 07](../07%20-%20Seleccion%20de%20Modelos%20y%20Optimizacion/01_Validacion_Cruzada_Avanzada.ipynb) |
| **Desplazamiento de distribución** | Producción ≠ entrenamiento (sección 2) | **Aquí** |
| **Desfase de métricas y umbral** | Se optimizó lo que no importa al negocio | [Módulo 03](../03%20-%20Metricas%20y%20Ajuste%20de%20Hiperparametros/00_Metricas_de_Clasificacion.ipynb) |
| **Bucles de retroalimentación y etiquetas tardías** | El modelo influye en los datos futuros; las etiquetas llegan semanas después | Mención |

Géron (Cap. 1, *Data Mismatch*) propone el conjunto ***train-dev***: una parte **del entrenamiento** que no se usa para ajustar, de modo que **si el modelo falla en *train-dev* es sobreajuste y si falla en la validación de producción es *mismatch***.

Simulamos un **flujo de datos** de 12 periodos donde solo cambia $P(y\mid\mathbf x)$ (la frontera **gira 12° por periodo**) y $P(\mathbf x)$ permanece **idéntica**. Comparamos: **(1)** la validación cruzada que **mezcla los periodos**, **(2)** un modelo **estático** entrenado con los 3 primeros periodos, **(3)** un modelo **reentrenado** con una ventana de los 3 periodos más recientes, y **(4)** un **monitor de covariables**: la prueba de Kolmogórov–Smirnov entre el primer periodo y cada periodo posterior, **variable por variable**.

````python
# ============================================================
# 4. Flujo con concept drift: CV mezclada vs modelo estatico vs reentrenado, y monitor de covariables (KS)
# ============================================================
T, POR = 12, 500
rng = np.random.default_rng(SEED)
flujo = [muestra_concepto(POR, np.radians(12) * t, rng) for t in range(T)]      # periodo t: la frontera gira 12 t grados
X_all = np.vstack([d[0] for d in flujo]); y_all = np.concatenate([d[1] for d in flujo])

# (1) CV mezclando todos los periodos: da UN numero que no corresponde a ningun momento real
acc_cv = cross_val_score(LogisticRegression(), X_all, y_all, cv=StratifiedKFold(5, shuffle=True, random_state=0)).mean()

# (2) estatico (periodos 0-2) y (3) reentrenado con ventana de 3 periodos
def ajustar(periodos):
    return LogisticRegression().fit(np.vstack([flujo[t][0] for t in periodos]), np.concatenate([flujo[t][1] for t in periodos]))

estatico = ajustar(range(3))
acc_est, acc_vent, ks_p = [], [], []
for t in range(3, T):
    Xt, yt = flujo[t]
    acc_est.append(estatico.score(Xt, yt))
    acc_vent.append(ajustar(range(t - 3, t)).score(Xt, yt))
    # (4) monitor de covariables: KS entre el periodo 0 y el periodo t, por variable
    ks_p.append([stats.ks_2samp(flujo[0][0][:, j], Xt[:, j]).pvalue for j in range(2)])
ks_p = np.array(ks_p)
periodos = np.arange(3, T)

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
axes[0].plot(periodos, acc_est, "o-", color=PALETA[1], lw=2, label="Modelo estático (entrenado en periodos 0–2)")
axes[0].plot(periodos, acc_vent, "s-", color=PALETA[2], lw=2, label="Reentrenado con los 3 periodos más recientes")
axes[0].axhline(acc_cv, color="black", ls="--", label=f"CV que mezcla los periodos = {acc_cv:.3f}")
axes[0].set_xlabel("Periodo de producción"); axes[0].set_ylabel("Accuracy en el periodo"); axes[0].legend(fontsize=8)
axes[0].set_title("Concept drift: el modelo estático se degrada hasta fallar más que el azar", fontweight="bold", fontsize=9)
axes[1].plot(periodos, ks_p[:, 0], "o-", color=PALETA[0], label="$x_1$"); axes[1].plot(periodos, ks_p[:, 1], "s-", color=PALETA[4], label="$x_2$")
axes[1].axhline(0.01, color=PALETA[1], ls="--", label="umbral de alarma (p = 0.01)"); axes[1].set_yscale("log")
axes[1].set_xlabel("Periodo de producción"); axes[1].set_ylabel("p-valor de KS (periodo 0 vs t)"); axes[1].legend(fontsize=8)
axes[1].set_title("El monitor de covariables NO ve el concept drift", fontweight="bold", fontsize=9)
plt.tight_layout(); plt.show()

print(f"Estático: del periodo 3 al 11 pasa de {acc_est[0]:.3f} a {acc_est[-1]:.3f}. Reentrenado: se mantiene entre {min(acc_vent):.3f} y {max(acc_vent):.3f}. CV mezclada: {acc_cv:.3f}.")
print(f"Alarmas del monitor de covariables (p < 0.01): {(ks_p < 0.01).sum()} de {ks_p.size} pruebas.")
assert acc_est[-1] < acc_est[0] - 0.3 and acc_est[-1] < 0.5          # el estatico cae mucho y termina por debajo del azar
assert min(acc_vent) > 0.75                                         # reentrenar con datos recientes lo mantiene a flote
assert (ks_p < 0.01).sum() <= 1                                     # P(x) no cambio: el monitor de covariables casi no alarma
assert 0.6 < acc_cv < 0.9                                           # y la CV mezclada da un numero intermedio que no describe ningun periodo
````

---
**Lecciones del flujo con *drift*:**

- La **validación cruzada que mezcla el tiempo** devuelve un número **intermedio que no describe ningún momento real**: ni el pasado (≈ 0.9) ni el futuro (que puede ser peor que el azar). Para datos con el tiempo en juego, evalúa **hacia adelante** (*walk-forward*, [Módulo 07](../07%20-%20Seleccion%20de%20Modelos%20y%20Optimizacion/01_Validacion_Cruzada_Avanzada.ipynb)).
- El monitor de **covariables** (KS, PSI, validación adversaria) **solo ve cambios en $P(\mathbf x)$**; **no puede ver** un cambio en $P(y\mid\mathbf x)$ porque **las entradas se ven igual**. Para detectarlo hace falta **medir el desempeño con etiquetas** (aunque lleguen con retraso) o monitorear la **distribución de las predicciones**.
- **Reentrenar periódicamente** con una ventana reciente mitiga el *concept drift*, a costa de **olvidar** lo antiguo. La política (cada cuánto, con qué ventana, con qué *gatillo* de alarma) es un tema de **ingeniería de ML** (McMahon, Cap. 3, *Retraining required*; Pote, Cap. 14).

## 5. Ensambles Más Allá del *Bagging*: *Stacking* y *Blending* 🧱

El **bagging** (Módulo 02) promedia modelos **del mismo tipo** entrenados en remuestras. El ***stacking*** (Wolpert, 1992) combina modelos **de tipos distintos** con un **metamodelo** que **aprende** cómo ponderarlos:

1. Los **modelos base** $f_1,\dots,f_M$ se entrenan sobre los datos.
2. Se construyen **metacaracterísticas** $\mathbf z_i=\big(f_1^{(-k(i))}(\mathbf x_i),\dots,f_M^{(-k(i))}(\mathbf x_i)\big)$ donde $f_m^{(-k(i))}$ es el modelo base entrenado **sin el pliegue $k(i)$** que contiene al ejemplo $i$: son predicciones **fuera de pliegue** (*out-of-fold*).
3. Un **metamodelo** $g$ (típicamente una regresión logística simple) se entrena sobre $(\mathbf z_i, y_i)$.
4. En producción, los modelos base se **reentrenan con todos los datos** y $g$ combina sus predicciones.

> ⚠️ **Por qué *fuera de pliegue*:** si el metamodelo viera las predicciones de un modelo base **sobre los mismos ejemplos con los que se entrenó**, un modelo que **memoriza** (un bosque, un 1-NN) parecería perfecto y el metamodelo le daría **todo el peso**. Las predicciones fuera de pliegue **simulan** cómo se comportan los modelos con datos que no vieron.

El ***blending*** es la variante simple: en vez de validación cruzada se **aparta un conjunto de retención** (*hold-out*) y el metamodelo se entrena con las predicciones sobre ese conjunto; es más sencillo y rápido, pero **desperdicia datos** y es **más ruidoso**. En esta sección: **(a)** implementamos *stacking* **a mano** y lo verificamos **contra `StackingClassifier`**; **(b)** mostramos el **error de usar predicciones en muestra**; **(c)** comparamos **con honestidad** el *stacking* frente al mejor modelo base y frente a un promedio simple, en tres conjuntos de datos.

````python
# ============================================================
# 5. Stacking: (a) desde cero == StackingClassifier, (b) fuga con predicciones en muestra, (c) comparacion honesta
# ============================================================
def bases():
    return [("lr", make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))),
            ("rf", RandomForestClassifier(30, random_state=SEED)),
            ("knn", make_pipeline(StandardScaler(), KNeighborsClassifier(7))),
            ("nb", GaussianNB())]


def stacking_a_mano(X_a, y_a, X_b, fuera_de_pliegue=True):
    """Metacaracteristicas = P(clase 1) de cada base (OOF con StratifiedKFold(5) o, ingenuamente, en muestra); meta = regresion logistica."""
    if fuera_de_pliegue:
        Z = np.column_stack([cross_val_predict(m, X_a, y_a, cv=StratifiedKFold(5), method="predict_proba")[:, 1] for _, m in bases()])
    ajustados = [m.fit(X_a, y_a) for _, m in bases()]                           # las bases se reentrenan con TODO el entrenamiento
    if not fuera_de_pliegue:
        Z = np.column_stack([m.predict_proba(X_a)[:, 1] for m in ajustados])    # INGENUO: predicciones sobre lo mismo que vieron
    meta = LogisticRegression().fit(Z, y_a)
    Z_b = np.column_stack([m.predict_proba(X_b)[:, 1] for m in ajustados])
    return meta.predict_proba(Z_b)[:, 1], meta


# (a) a mano == StackingClassifier (binario: el metamodelo usa la columna de la clase positiva de cada base)
Xs_, ys_ = load_breast_cancer(return_X_y=True)
Xa, Xb_, ya, yb_ = train_test_split(Xs_, ys_, test_size=0.3, stratify=ys_, random_state=SEED)
p_mano, _ = stacking_a_mano(Xa, ya, Xb_)
sk = StackingClassifier(bases(), final_estimator=LogisticRegression(), cv=5).fit(Xa, ya)
dif = np.abs(p_mano - sk.predict_proba(Xb_)[:, 1]).max()
print(f"(a) Diferencia máxima entre el stacking a mano y StackingClassifier: {dif:.2e}")
assert dif < 1e-8

# (b) predicciones en muestra vs fuera de pliegue (datos ruidosos: un bosque memoriza y parece infalible)
Xn, yn = make_classification(1500, 20, n_informative=6, n_redundant=4, flip_y=0.1, class_sep=0.8, random_state=SEED)
Xna, Xnb, yna, ynb = train_test_split(Xn, yn, test_size=0.3, stratify=yn, random_state=SEED)
filas = []
for nombre, oof in [("fuera de pliegue (correcto)", True), ("en muestra (ingenuo)", False)]:
    p, meta = stacking_a_mano(Xna, yna, Xnb, oof)
    filas.append({"metamodelo entrenado con": nombre, "accuracy en prueba": accuracy_score(ynb, p > 0.5), **{f"coef {n}": c for (n, _), c in zip(bases(), meta.coef_[0])}})
display(pd.DataFrame(filas).set_index("metamodelo entrenado con").round(3))
coef_ing = filas[1]["coef rf"]; coef_ok = filas[0]["coef rf"]
assert coef_ing > coef_ok                                       # el ingenuo le regala mas confianza al bosque que memorizo

# (c) comparacion honesta en tres conjuntos: base vs promedio simple (soft voting) vs stacking
def comparar(nombre, X, y, repeticiones):
    cv = RepeatedStratifiedKFold(n_splits=5, n_repeats=repeticiones, random_state=SEED)
    mods = dict(bases())
    mods["promedio simple"] = VotingClassifier(bases(), voting="soft")
    mods["STACKING"] = StackingClassifier(bases(), final_estimator=LogisticRegression(max_iter=1000), cv=5)
    sc = {k: cross_val_score(m, X, y, cv=cv, n_jobs=-1) for k, m in mods.items()}
    mejor = max((k for k in bases_names), key=lambda k: sc[k].mean())
    diff = sc["STACKING"] - sc[mejor]                           # mismos pliegues: diferencia pareada
    return sc, mejor, diff


bases_names = [n for n, _ in bases()]
t0 = time.time()
conjuntos = {"Breast Cancer": (*load_breast_cancer(return_X_y=True), 2),
             "Dígitos": (*load_digits(return_X_y=True), 1),
             "Sintético ruidoso": (Xn, yn, 2)}
filas = []
for nombre, (X, y, rep) in conjuntos.items():
    sc, mejor, diff = comparar(nombre, X, y, rep)
    filas.append({"conjunto": nombre, **{k: f"{v.mean():.3f}" for k, v in sc.items()}, "mejor base": mejor,
                  "stacking − mejor base": f"{diff.mean():+.4f}", "± (desv. pliegues)": f"{diff.std():.4f}"})
tab = pd.DataFrame(filas).set_index("conjunto")
display(tab)
print(f"Tiempo de la comparación: {time.time() - t0:.0f} s")
dif_medias = [float(f["stacking − mejor base"]) for f in filas]
assert max(abs(d) for d in dif_medias) < 0.02            # el stacking queda a +-2 puntos del mejor base: ganancia pequena, sin milagros
````

---
**Lectura honesta del experimento:**

- **(a)** Nuestro *stacking* **coincide** con `StackingClassifier` hasta el error de máquina: no es magia, es *cross-validation* + un modelo logístico sobre las predicciones.
- **(b)** Con predicciones **en muestra** el metamodelo casi **duplica** el coeficiente del bosque (≈ 9.3 frente a ≈ 5.0) y **anula** al modelo lineal: el bosque memorizó el entrenamiento y «parece» perfecto, así que se lleva casi toda la confianza. Con predicciones **fuera de pliegue** el reparto refleja el desempeño **real**. En *este* conjunto la exactitud en prueba **no cambia** (≈ 0.81 en ambos casos): el daño aquí es el **peso mal asignado**, que se vuelve error cuando los modelos base difieren más. Es el mismo tipo de error que la **fuga de datos** del Módulo 00, ahora **dentro** de un ensamble.
- **(c)** La ganancia del *stacking* sobre el **mejor modelo base** es **pequeña** y de signo **variable**: ligeramente **negativa** en *Breast Cancer* y en el sintético (≈ −0.4 puntos, dentro del ruido: la desviación entre pliegues es ≈ 1.5 puntos) y de **+0.8 puntos** en los dígitos (donde los modelos fallan en ejemplos distintos y hay margen). El **promedio simple** da el mismo peso a modelos **débiles** (Naive Bayes) y queda **por debajo del mejor base** en dos de los tres conjuntos; el *stacking* **aprende** a darles poco peso. Moraleja: el *stacking* **ayuda cuando los modelos base son diversos y buenos**; **no sustituye** a mejores datos, mejores variables o un mejor modelo base, y **multiplica** el costo de entrenamiento e inferencia (y la complejidad de mantenimiento). Úsalo cuando cada décima cuente (competencias), no por defecto.

## 6. Aprendizaje Centrado en Datos: Cuando las Etiquetas Se Equivocan 🏷️

Durante años el esfuerzo se centró en **mejorar el modelo** manteniendo los datos fijos. El enfoque **centrado en datos** (*data-centric AI*) invierte la prioridad: **con un modelo razonable, mejorar la calidad de los datos suele rendir más**. Las etiquetas reales **se equivocan**: anotadores que discrepan, casos ambiguos, errores de captura o de proceso. Con ruido **simétrico** de tasa $\eta$, cada etiqueta de entrenamiento se **invierte** con probabilidad $\eta$ (en binario, el ruido simétrico hace $P(\tilde y\neq y)=\eta$); la evaluación se hace sobre etiquetas **limpias**.

Efecto esperado: los modelos que **memorizan** (1-NN, árboles profundos) **memorizan también el ruido** y degradan rápido; los modelos **de baja varianza** o **regularizados** (regresión logística) son más robustos (Frénay y Verleysen, 2014). Para **encontrar** las etiquetas dudosas implementamos una versión simplificada de **aprendizaje confiado** (*confident learning*; Northcutt et al., 2021): con un modelo que no vio el ejemplo (**predicciones fuera de pliegue**), marcamos como sospechosa toda fila cuya etiqueta tiene **probabilidad predicha < 0.5**, es decir, el modelo, **sin haber visto esa fila**, apuesta por **la otra clase**. Se eliminan las sospechosas y se **reentrena**. (La versión completa estima una *matriz de confusión* entre etiquetas observadas y verdaderas y umbrales por clase; aquí solo está la idea.)

````python
# ============================================================
# 6. Ruido de etiquetas: efecto en distintos modelos y limpieza con predicciones fuera de pliegue
# ============================================================
Xd, yd = load_breast_cancer(return_X_y=True)
fabricas = {"Regresión logística": lambda: make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)),
            "Bosque aleatorio": lambda: RandomForestClassifier(40, random_state=SEED),
            "Árbol profundo": lambda: DecisionTreeClassifier(random_state=SEED),
            "1-NN": lambda: make_pipeline(StandardScaler(), KNeighborsClassifier(1))}
tasas = [0.0, 0.1, 0.2, 0.3, 0.4]
registros = []
for semilla in range(4):
    Xa, Xb_, ya, yb_ = train_test_split(Xd, yd, test_size=0.4, stratify=yd, random_state=semilla)
    for tasa in tasas:
        rng = np.random.default_rng(semilla)
        invertida = rng.random(len(ya)) < tasa                                   # que etiquetas se corrompen (la verdad nos la sabemos)
        y_ruidosa = np.where(invertida, 1 - ya, ya)
        for nombre, fab in fabricas.items():
            registros.append((semilla, tasa, nombre, "con ruido", accuracy_score(yb_, fab().fit(Xa, y_ruidosa).predict(Xb_))))
        if tasa > 0:
            # confident learning "light": P(etiqueta dada) fuera de pliegue < 0.5  ->  sospechosa
            oof = cross_val_predict(fabricas["Regresión logística"](), Xa, y_ruidosa, cv=StratifiedKFold(5, shuffle=True, random_state=semilla), method="predict_proba")
            sospechosa = oof[np.arange(len(ya)), y_ruidosa] < 0.5
            tp = (sospechosa & invertida).sum()
            registros.append((semilla, tasa, "detector", "precisión", tp / max(sospechosa.sum(), 1)))
            registros.append((semilla, tasa, "detector", "recall", tp / max(invertida.sum(), 1)))
            for nombre, fab in fabricas.items():
                registros.append((semilla, tasa, nombre, "limpiado", accuracy_score(yb_, fab().fit(Xa[~sospechosa], y_ruidosa[~sospechosa]).predict(Xb_))))
df = pd.DataFrame(registros, columns=["semilla", "tasa", "modelo", "tipo", "valor"])
res = df.groupby(["tipo", "modelo", "tasa"]).valor.mean().unstack("tasa")
display(res.round(3))

fig, axes = plt.subplots(1, 2, figsize=(14, 4.6))
for c, nombre in zip(PALETA, fabricas):
    axes[0].plot(tasas, res.loc[("con ruido", nombre)], "o-", color=c, lw=2, label=nombre)
    axes[0].plot(tasas[1:], res.loc[("limpiado", nombre)].values[1:], "s--", color=c, lw=1.5, alpha=0.8)
axes[0].axhline(0.5, color="black", ls=":", lw=1)
axes[0].set_xlabel("Tasa de etiquetas invertidas en el entrenamiento"); axes[0].set_ylabel("Accuracy en prueba LIMPIA"); axes[0].legend(fontsize=8)
axes[0].set_title("Ruido de etiquetas (línea continua) y tras limpiar (línea discontinua)", fontweight="bold", fontsize=9)
det = res.loc["precisión"].iloc[0], res.loc["recall"].iloc[0]
axes[1].plot(tasas[1:], det[0].values[1:], "o-", color=PALETA[0], lw=2, label="precisión del detector")
axes[1].plot(tasas[1:], det[1].values[1:], "s-", color=PALETA[2], lw=2, label="recall del detector")
axes[1].set_xlabel("Tasa de ruido"); axes[1].set_ylabel("Valor"); axes[1].set_ylim(0, 1.05); axes[1].legend(fontsize=8)
axes[1].set_title("Detección de etiquetas erróneas (modelo fuera de pliegue)", fontweight="bold", fontsize=9)
plt.tight_layout(); plt.show()

assert res.loc[("con ruido", "1-NN"), 0.4] < res.loc[("con ruido", "1-NN"), 0.0] - 0.25          # el modelo que memoriza se desploma
assert res.loc[("con ruido", "Regresión logística"), 0.2] > res.loc[("con ruido", "1-NN"), 0.2] + 0.08   # el de baja varianza resiste mejor
assert res.loc[("limpiado", "1-NN"), 0.2] > res.loc[("con ruido", "1-NN"), 0.2] + 0.08                  # limpiar ayuda al que memoriza
assert res.loc[("recall", "detector"), 0.2] > 0.85
print(f"1-NN con 20 % de ruido: {res.loc[('con ruido', '1-NN'), 0.2]:.3f} → tras limpiar {res.loc[('limpiado', '1-NN'), 0.2]:.3f}. "
      f"La regresión logística pasa de {res.loc[('con ruido', 'Regresión logística'), 0.0]:.3f} (sin ruido) a {res.loc[('con ruido', 'Regresión logística'), 0.4]:.3f} (40 %).")
print(f"Detector con 20 % de ruido: recall = {res.loc[('recall', 'detector'), 0.2]:.2f}, precisión = {res.loc[('precisión', 'detector'), 0.2]:.2f}  (marca de más: algunas 'sospechosas' están bien etiquetadas).")
````

## 8. Los Límites de ML: Correlación No Es Causalidad ⚠️

Un modelo de ML **aprende asociaciones estadísticas** $P(y\mid\mathbf x)$ del conjunto de entrenamiento. No sabe **por qué** se asocian. Esto tiene consecuencias:

| Límite | Qué significa | Ejemplo |
|---|---|---|
| **Correlación ≠ causalidad** | Predecir bien **no** equivale a entender qué **causa** qué | Los helados «predicen» ahogamientos (ambos suben en verano); retirar los helados no salva a nadie |
| **Atajos espurios** (*shortcut learning*) | El modelo usa el rasgo **más fácil** que correlaciona con la etiqueta, aunque sea irrelevante | Un clasificador de radiografías que «detecta» neumonía por la **marca del hospital** (donde hay más casos) |
| **No responde a «¿qué pasaría si…?»** | Predicción ≠ intervención: cambiar una variable de entrada **no** cambia la realidad como lo hace en el modelo | Un modelo predice que quien toma cierto medicamento empeora (porque solo lo toman los más graves) |
| **Extrapolación** | Fuera del rango entrenado, la predicción **no tiene garantía** (sección 2) | Un modelo lineal usado en regiones no vistas |
| **Datos sesgados, decisiones sesgadas** | El modelo reproduce (y amplifica) los sesgos de los datos | Ver ética de la IA |

Demostramos el **atajo espurio** con un caso mínimo: una característica **causal pero ruidosa** ($x_c$, relacionada con la clase) y otra **espuria pero muy limpia** ($x_s$, que **coincide con la etiqueta el 95 %** de las veces **en el entrenamiento** porque así se recolectaron los datos). El modelo aprende a mirar la espuria. En producción esa coincidencia **se rompe**.

````python
# ============================================================
# 8. Atajo espurio: alta exactitud en el laboratorio, colapso cuando la correlacion espuria se rompe
# ============================================================
def muestra_atajo(n, acuerdo, rng):
    """x_c: causal pero ruidosa (media = y, desv. 1.5). x_s: espuria; vale 'y' con probabilidad `acuerdo`, si no vale '1-y' (ruido pequeno)."""
    y = rng.integers(0, 2, n)
    xc = y + rng.normal(0, 1.5, n)
    xs = np.where(rng.random(n) < acuerdo, y, 1 - y) + rng.normal(0, 0.1, n)
    return np.column_stack([xc, xs]), y


X_tr, y_tr = muestra_atajo(2000, 0.95, np.random.default_rng(SEED))
modelo = LogisticRegression().fit(X_tr, y_tr)
solo_causal = LogisticRegression().fit(X_tr[:, :1], y_tr)
escenarios = {"Entrenamiento (95 % coinciden)": 0.95, "Producción A (sin relación, 50 %)": 0.5, "Producción B (se invierte, 5 %)": 0.05}
filas = []
for nombre, ac in escenarios.items():
    Xt, yt = muestra_atajo(5000, ac, np.random.default_rng(1))
    filas.append({"escenario": nombre, "accuracy modelo con x_c y x_s": modelo.score(Xt, yt), "accuracy modelo solo con x_c": solo_causal.score(Xt[:, :1], yt)})
tab = pd.DataFrame(filas).set_index("escenario")
display(tab.round(3))
print(f"Coeficientes del modelo completo: x_c = {modelo.coef_[0, 0]:.2f}, x_s = {modelo.coef_[0, 1]:.2f}  (el atajo pesa ~{abs(modelo.coef_[0, 1] / modelo.coef_[0, 0]):.0f} veces más).")

fig, ax = plt.subplots(figsize=(9, 4.4))
ancho = 0.35; pos = np.arange(len(tab))
ax.bar(pos - ancho / 2, tab.iloc[:, 0], ancho, color=PALETA[1], label="Con x_c y x_s (usa el atajo)")
ax.bar(pos + ancho / 2, tab.iloc[:, 1], ancho, color=PALETA[2], label="Solo con la variable causal x_c")
ax.axhline(0.5, color="black", ls="--", lw=1, label="Azar"); ax.set_xticks(pos); ax.set_xticklabels([s.replace(" (", "\n(") for s in tab.index], fontsize=8)
ax.set_ylabel("Accuracy"); ax.set_ylim(0, 1.05); ax.legend(fontsize=8)
ax.set_title("El atajo gana en el laboratorio y pierde en el mundo real", fontweight="bold")
plt.tight_layout(); plt.show()
assert tab.iloc[0, 0] > 0.93 and tab.iloc[2, 0] < 0.1                       # 95 % en el laboratorio, ~5 % si la correlacion se invierte
assert abs(tab.iloc[0, 1] - tab.iloc[2, 1]) < 0.03 and tab.iloc[2, 1] > 0.55  # la variable causal rinde igual en los tres escenarios (aunque menos)
````

---
**Lectura.** El modelo con el atajo **gana en el laboratorio** (≈ 95 %) y **pierde de forma estrepitosa** cuando la correlación espuria se rompe o se invierte (≈ 50 % y ≈ 5 %). El modelo que usa **solo** la variable causal rinde **menos** en el laboratorio (≈ 63 %), pero **igual en todos los escenarios**: es **robusto** porque se apoya en algo que **sigue siendo cierto**. Para **elegir** entre ambos **no basta la validación** (ambos escenarios de validación «aleatoria» favorecen al atajo): se necesita **conocimiento del dominio**, **pruebas de robustez** en datos de otras fuentes y, cuando importa, **modelos causales** (fuera del alcance de este curso). Estos límites enlazan con el **pensamiento crítico sobre datos** ([Visual Analytics, Módulo 03](../../Visual%20Analytics%20and%20Critical%20Thinking/03%20-%20Pensamiento%20Critico%20en%20Datos/README.md): sesgos, errores y falacias estadísticas) y con la **ética de la IA** ([IA, Módulo 03](../../Introduccion%20a%20la%20Inteligencia%20Artificial/03%20-%20Etica%20y%20el%20Nuevo%20Paradigma%20de%20la%20IA/README.md)): un modelo que **no entiende** por qué predice puede **discriminar** o **fallar** sin que nadie lo note.

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Capítulo 3** | Rompe los supuestos «de laboratorio» (datos i.i.d., limpios, completos, equilibrados) y añade lo necesario para usar ML en la práctica: reutilizar (M08-01), explicar (M08-02), desbalance (M08-03), faltantes (M08-04) y aplicaciones (Módulo 09) |
| **Dataset shift** | $P(\mathbf x)$ (*covariate*), $P(y)$ (*label*) o $P(y\mid\mathbf x)$ (*concept*) cambian. El *label shift* **no mueve el accuracy pero hunde la precisión**; el *concept shift* puede dejar el modelo **peor que el azar** |
| **Detectar y corregir** | La **validación adversaria** detecta el *covariate shift* sin etiquetas (AUC ≈ 0.5 frente a ≈ 1.0) y señala las variables culpables. La **reponderación** $w=\hat P(d{=}1\mid\mathbf x)/\hat P(d{=}0\mid\mathbf x)\cdot n_{\text{tr}}/n_{\text{te}}$ corrige **parte** de la caída, a costa del **tamaño efectivo** de muestra |
| **Validación vs producción** | La CV que mezcla el tiempo da un número que no describe ningún momento; un monitor de **covariables** no ve el cambio de $P(y\mid\mathbf x)$: hace falta **monitorear el desempeño** con etiquetas y **reentrenar** |
| **Stacking** | Metamodelo sobre predicciones **fuera de pliegue** (verificado contra `StackingClassifier`). La ganancia sobre el mejor base es **pequeña** (décimas de punto) y el promedio simple casi iguala; usar predicciones **en muestra** es una **fuga** |
| **Centrado en datos** | El ruido en las etiquetas degrada **más** a los modelos que memorizan (1-NN, árbol profundo); las predicciones fuera de pliegue **encuentran** las etiquetas dudosas (alto *recall*, precisión moderada): **revisar, no borrar a ciegas** |
| **Reproducibilidad** | Fija semillas, **reporta la variabilidad** (mejor semilla ≠ resultado esperado), registra versiones y una huella de datos; todo el preprocesamiento en un `Pipeline`. El ciclo es datos → entrenamiento → evaluación → despliegue → **monitoreo** → datos |
| **Límites de ML** | **Correlación ≠ causalidad**: un modelo puede ganar en el laboratorio con un **atajo espurio** y colapsar en producción. La validación aleatoria no lo detecta; hace falta criterio, pruebas de robustez y ética |
