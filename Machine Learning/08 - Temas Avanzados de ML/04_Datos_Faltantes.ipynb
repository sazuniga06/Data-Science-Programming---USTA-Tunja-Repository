## 1. Mecanismos de Ausencia (MCAR, MAR, MNAR) y el Sesgo de los Casos Completos 🔬

### 1.1 Los tres mecanismos (Rubin, 1976)

Sea $\mathbf X$ la matriz de datos (completa, aunque no la veamos) y $R_{ij}=1$ si el valor $X_{ij}$ **falta**. Se separa $\mathbf X=(\mathbf X_{\text{obs}},\mathbf X_{\text{mis}})$ en lo observado y lo que falta. Los mecanismos describen **de qué depende** $R$:

| Mecanismo | Definición | Ejemplo | ¿Es «ignorable»? |
|---|---|---|---|
| **MCAR** (*missing completely at random*) | $P(R\mid\mathbf X)=P(R)$: **no depende de nada** | Un sensor falla al azar; una gota de café borra una fila | Sí: los casos completos son una **muestra aleatoria** (menos datos, **sin sesgo**) |
| **MAR** (*missing at random*) | $P(R\mid\mathbf X)=P(R\mid\mathbf X_{\text{obs}})$: depende **solo de lo observado** | Las personas **mayores** (edad observada) omiten más el ingreso | **Condicionalmente**: se puede corregir **usando las variables observadas** |
| **MNAR** (*missing not at random*) | $P(R\mid\mathbf X)$ depende **también de $\mathbf X_{\text{mis}}$** | Quienes ganan **más** omiten el ingreso **justamente por** ganar más | **No**: el sesgo **no se corrige con los datos**; exige **supuestos** o información externa |

> ⚠️ **Los datos no distinguen MAR de MNAR.** Con solo los datos observados es **imposible** saber si la ausencia depende del propio valor faltante: esa distinción se apoya en el **conocimiento del proceso** que generó los datos. Se pueden **refutar** MCAR (comparando observados y faltantes) pero nunca **demostrar** MAR.

### 1.2 Experimento: sesgo de los casos completos

Simulamos una variable de interés $X\sim\mathcal N(0,1)$ y una variable **siempre observada** $Z$ con $\text{corr}(X,Z)=0.7$. Borramos **≈ 30 %** de los valores de $X$ con cada mecanismo: **MCAR** (probabilidad fija), **MAR** (la probabilidad **crece con $Z$**, observada) y **MNAR** (la probabilidad **crece con $X$** misma). Como conocemos la verdad ($\mu=0$, $\sigma=1$, $\rho=0.7$), medimos el **sesgo** al calcular con solo los **casos completos** (se ignoran las filas con $X$ faltante), promediando 300 repeticiones. Añadimos un **estimador que aprovecha $Z$** (regresar $X$ sobre $Z$ con los casos observados y promediar sobre **todas** las filas), que **corrige el MAR pero no el MNAR**.

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas matplotlib seaborn scikit-learn scipy

import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy
import sklearn
from scipy import stats
from sklearn.datasets import load_breast_cancer, make_classification
from sklearn.model_selection import StratifiedKFold, RepeatedStratifiedKFold, cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.metrics.pairwise import nan_euclidean_distances
from sklearn.ensemble import HistGradientBoostingClassifier
try:                                                   # IterativeImputer es EXPERIMENTAL: se importa con try/except (la API puede cambiar entre versiones)
    from sklearn.experimental import enable_iterative_imputer  # noqa: F401
    from sklearn.impute import IterativeImputer
except Exception:
    IterativeImputer = None

warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (9, 5)
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"Entorno listo | numpy {np.__version__} | scipy {scipy.__version__} | scikit-learn {sklearn.__version__} | IterativeImputer: {'disponible' if IterativeImputer else 'NO disponible'}")

# ============================================================
# 1. Simulacion de los tres mecanismos y sesgo de los casos completos
# ============================================================
RHO, P_FALTA = 0.7, 0.30


def generar_xz(n, rng):
    """X ~ N(0,1) (variable de interes) y Z con corr(X, Z) = 0.7 (siempre observada)."""
    x = rng.normal(0, 1, n)
    z = RHO * x + np.sqrt(1 - RHO ** 2) * rng.normal(0, 1, n)
    return x, z


def prob_falta(score, objetivo=P_FALTA, fuerza=1.5):
    """Probabilidad logistica sigmoide(a + fuerza*score) con 'a' calibrada (biseccion) para que el promedio sea `objetivo`."""
    lo, hi = -10.0, 10.0
    for _ in range(60):
        a = (lo + hi) / 2
        if (1 / (1 + np.exp(-(a + fuerza * score)))).mean() > objetivo:
            hi = a
        else:
            lo = a
    return 1 / (1 + np.exp(-(a + fuerza * score)))


def mascara(mecanismo, x, z, rng):
    """True = X falta. MCAR: p fija | MAR: depende de Z (observada) | MNAR: depende de X (el propio valor)."""
    if mecanismo == "MCAR":
        p = np.full(len(x), P_FALTA)
    elif mecanismo == "MAR":
        p = prob_falta(z)
    else:
        p = prob_falta(x)
    return rng.random(len(x)) < p


def estimaciones(x, z, falta):
    obs = ~falta
    reg = LinearRegression().fit(z[obs, None], x[obs])                       # usa Z para predecir X con los casos observados
    return {"media": x[obs].mean(), "desv. típica": x[obs].std(ddof=1), "corr(X, Z)": np.corrcoef(x[obs], z[obs])[0, 1],
            "media usando Z": reg.predict(z[:, None]).mean(), "% casos completos": 100 * obs.mean()}


rng = np.random.default_rng(SEED)
filas = []
for mec in ["MCAR", "MAR", "MNAR"]:
    for rep in range(300):
        x, z = generar_xz(2000, rng)
        filas.append({"mecanismo": mec, **estimaciones(x, z, mascara(mec, x, z, rng))})
df = pd.DataFrame(filas)
resumen = df.groupby("mecanismo").mean().loc[["MCAR", "MAR", "MNAR"]]
verdad = {"media": 0.0, "desv. típica": 1.0, "corr(X, Z)": RHO, "media usando Z": 0.0}
sesgo = resumen[list(verdad)] - pd.Series(verdad)
display(resumen.round(3))
print("Sesgo (estimación − verdad), promedio de 300 repeticiones, n = 2000:")
display(sesgo.round(3))

fig, axes = plt.subplots(1, 3, figsize=(16, 4.4))
for ax, col, ver in zip(axes, ["media", "desv. típica", "media usando Z"], [0.0, 1.0, 0.0]):
    for mec, c in zip(["MCAR", "MAR", "MNAR"], [PALETA[2], PALETA[3], PALETA[1]]):
        ax.hist(df[df.mecanismo == mec][col], bins=30, alpha=0.65, color=c, label=mec)
    ax.axvline(ver, color="black", ls="--", label="verdad"); ax.set_xlabel(f"Estimación de la {col}"); ax.set_ylabel("Repeticiones"); ax.legend(fontsize=8)
axes[0].set_title("Casos completos: media", fontweight="bold", fontsize=10); axes[1].set_title("Casos completos: desv. típica", fontweight="bold", fontsize=10)
axes[2].set_title("Usando Z: arregla el MAR, no el MNAR", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()

assert abs(sesgo.loc["MCAR", "media"]) < 0.02 and abs(sesgo.loc["MCAR", "corr(X, Z)"]) < 0.02       # MCAR: casos completos sin sesgo
assert sesgo.loc["MAR", "media"] < -0.15 and sesgo.loc["MNAR", "media"] < sesgo.loc["MAR", "media"]   # MAR y MNAR sesgan la media; MNAR mas
assert abs(sesgo.loc["MAR", "media usando Z"]) < 0.02 and sesgo.loc["MNAR", "media usando Z"] < -0.1  # usar Z corrige el MAR pero NO el MNAR
````

---
**Lectura del experimento.** Con **MCAR** los casos completos dan estimaciones **sin sesgo** (solo con menos datos). Con **MAR** la media queda **sesgada hacia abajo** (se pierden más los $X$ altos porque se asocian con $Z$ alto), pero **usar $Z$** (que sí observamos) **corrige** el sesgo por completo: eso es lo que significa que el MAR sea «ignorable condicionalmente». Con **MNAR** el sesgo es **mayor** y **ninguna función de lo observado lo arregla**: el estimador que usa $Z$ **sigue sesgado**. Además la **desviación típica** de los casos completos queda **subestimada** en MAR y MNAR (se pierde la cola superior). **No olvides**: en datos reales **no puedes distinguir MAR de MNAR** mirando solo los datos.

## 2. Casos Completos, Media, Mediana y Moda: Costo y Distorsión 📉

### 2.1 Eliminar filas: cuánto cuesta

Con $d$ variables que fallan **independientemente** con probabilidad $p$, la fracción de filas **completas** es $(1-p)^d$. Con $p=5\,\%$ y $d=30$ queda solo $0.95^{30}\approx21\,\%$ de las filas: la **eliminación por filas** (*listwise deletion*) **desperdicia** la mayoría de los datos aun con ausencias **moderadas**.

### 2.2 Imputación simple y por qué distorsiona

La **imputación por media** (o **mediana**; **moda** para categóricas) reemplaza cada faltante por un **estadístico de la columna**:

$$\hat x_{ij}=\bar x_j=\frac1{n_j}\sum_{i:\,R_{ij}=0}x_{ij}\quad(\text{media}),\qquad \hat x_{ij}=\text{mediana}\{x_{ij}:R_{ij}=0\}\quad(\text{mediana}).$$

Es **simple y no cambia la media** (bajo MCAR), pero **distorsiona la distribución**. Si una fracción $p$ de los valores es MCAR y se imputa la media $\bar x$, los datos imputados valen **todos lo mismo**:

- **Varianza:** $\text{Var}(x_{\text{imp}})\approx(1-p)\,\text{Var}(x)$ (los $p\,n$ valores imputados aportan **cero** dispersión alrededor de la media). La desviación típica se **encoge** en un factor $\sqrt{1-p}$.
- **Covarianza con otra variable $Z$:** $\text{Cov}(x_{\text{imp}},z)\approx(1-p)\,\text{Cov}(x,z)$ (los imputados no covarían con $Z$).
- **Correlación:** $\text{corr}(x_{\text{imp}},z)=\dfrac{(1-p)\,\text{Cov}(x,z)}{\sqrt{(1-p)\text{Var}(x)}\,\sqrt{\text{Var}(z)}}=\sqrt{1-p}\;\text{corr}(x,z)$: la correlación se **atenúa** por $\sqrt{1-p}$.

Consecuencias: **errores estándar demasiado optimistas** (parece que hay menos variabilidad de la real), **relaciones debilitadas** entre variables y **picos artificiales** en el histograma. La **mediana** es más robusta a valores extremos pero produce la **misma** distorsión. Implementamos los tres imputadores **desde cero**, verificamos contra `SimpleImputer` y **comprobamos las fórmulas** con $p=30\,\%$.

````python
# ============================================================
# 2. Imputacion por media / mediana / moda desde cero (== SimpleImputer) y su efecto en varianza y correlacion
# ============================================================
def imputar_estadistico(X, estrategia="mean"):
    """Reemplaza cada NaN por la media / mediana / moda (valor MAS FRECUENTE; en empates, el MENOR, como SimpleImputer) de su columna."""
    X = np.array(X, dtype=float, copy=True)
    for j in range(X.shape[1]):
        obs = X[~np.isnan(X[:, j]), j]
        if estrategia == "mean":
            v = obs.mean()
        elif estrategia == "median":
            v = np.median(obs)
        else:
            vals, cuentas = np.unique(obs, return_counts=True)             # np.unique ordena de menor a mayor: argmax elige el menor en empates
            v = vals[np.argmax(cuentas)]
        X[np.isnan(X[:, j]), j] = v
    return X


# --- Verificacion contra SimpleImputer (media, mediana y moda con empates) ---
rng = np.random.default_rng(SEED)
Xv = rng.normal(5, 2, (200, 6))
Xv[rng.random(Xv.shape) < 0.25] = np.nan
for est, nombre_sk in [("mean", "mean"), ("median", "median")]:
    assert np.allclose(imputar_estadistico(Xv, est), SimpleImputer(strategy=nombre_sk).fit_transform(Xv))
Xc = rng.integers(0, 4, (200, 3)).astype(float)                              # categorias codificadas 0-3: hay EMPATES en la moda
Xc[rng.random(Xc.shape) < 0.25] = np.nan
assert np.array_equal(imputar_estadistico(Xc, "mode"), SimpleImputer(strategy="most_frequent").fit_transform(Xc))
print("Media, mediana y moda desde cero == SimpleImputer (la moda con empates elige el valor menor en ambos).")

# --- Costo de eliminar filas (listwise deletion): fraccion de filas completas con p = 5 % por variable ---
Xbc, ybc = load_breast_cancer(return_X_y=True)
Xbc_s = StandardScaler().fit_transform(Xbc)
M5 = np.random.default_rng(SEED).random(Xbc_s.shape) < 0.05
completas = (~M5.any(axis=1)).mean()
print(f"Breast Cancer (30 variables), 5 % MCAR por variable: filas completas = {completas:.1%}  (teoría: 0.95^30 = {0.95 ** 30:.1%}). Eliminar filas deja ~1 de cada 5.")
assert abs(completas - 0.95 ** 30) < 0.05

# --- Efecto de la imputacion por media: varianza y correlacion (MCAR 30 %) ---
rng = np.random.default_rng(SEED)
n, p = 20000, 0.30
x, z = generar_xz(n, rng)
x_falta = x.copy(); x_falta[rng.random(n) < p] = np.nan
filas = []
for nombre, X_imp in [("Datos completos (verdad)", x), ("Casos completos", x_falta[~np.isnan(x_falta)]),
                      ("Imputación por media", imputar_estadistico(x_falta[:, None], "mean")[:, 0]), ("Imputación por mediana", imputar_estadistico(x_falta[:, None], "median")[:, 0])]:
    zz = z if len(X_imp) == n else z[~np.isnan(x_falta)]
    filas.append({"método": nombre, "media": X_imp.mean(), "desv. típica": X_imp.std(ddof=1), "corr(X, Z)": np.corrcoef(X_imp, zz)[0, 1]})
tab = pd.DataFrame(filas).set_index("método")
display(tab.round(3))
teor_sd, teor_corr = np.sqrt(1 - p) * 1.0, np.sqrt(1 - p) * RHO
print(f"Teoría con p = {p}: desv. típica = sqrt(1-p) = {teor_sd:.3f} (observada {tab.loc['Imputación por media', 'desv. típica']:.3f}); correlación = sqrt(1-p)·ρ = {teor_corr:.3f} (observada {tab.loc['Imputación por media', 'corr(X, Z)']:.3f}).")
assert abs(tab.loc["Imputación por media", "desv. típica"] - teor_sd) < 0.02 and abs(tab.loc["Imputación por media", "corr(X, Z)"] - teor_corr) < 0.02
assert abs(tab.loc["Casos completos", "desv. típica"] - 1.0) < 0.02                       # bajo MCAR los casos completos conservan la varianza (con menos datos)

# --- Indicador de faltantes: cuando la AUSENCIA es informativa (y depende de si X falto) ---
rng = np.random.default_rng(SEED)
n = 4000
xi = rng.normal(0, 1, n)
falto = rng.random(n) < 0.30
yi = 2 * xi + 3 * falto + 0.5 * rng.normal(0, 1, n)                                          # el hecho de que X falte desplaza y en +3
x_obs = np.where(falto, np.nan, xi)[:, None]
sin_ind = SimpleImputer(strategy="mean").fit_transform(x_obs)
con_ind = SimpleImputer(strategy="mean", add_indicator=True).fit_transform(x_obs)
manual = np.column_stack([sin_ind[:, 0], np.isnan(x_obs[:, 0]).astype(float)])
assert np.array_equal(con_ind, manual)                                                      # nuestra columna indicadora == add_indicator
tr, te = slice(0, 3000), slice(3000, None)
r2_sin = LinearRegression().fit(sin_ind[tr], yi[tr]).score(sin_ind[te], yi[te])
r2_con = LinearRegression().fit(con_ind[tr], yi[tr]).score(con_ind[te], yi[te])
print(f"Ausencia informativa: R² con imputación por media = {r2_sin:.3f} | con media + indicador (add_indicator) = {r2_con:.3f}")
assert r2_con > r2_sin + 0.2

fig, axes = plt.subplots(1, 3, figsize=(17, 4.5))
axes[0].hist(x, bins=60, alpha=0.5, color=PALETA[9], density=True, label="verdad")
axes[0].hist(imputar_estadistico(x_falta[:, None], "mean")[:, 0], bins=60, alpha=0.6, color=PALETA[1], density=True, label="imputado por media (pico en 0)")
axes[0].set_xlabel("X"); axes[0].set_ylabel("Densidad"); axes[0].legend(fontsize=8); axes[0].set_title("La media crea un pico artificial", fontweight="bold", fontsize=10)
idx = np.random.default_rng(1).choice(20000, 1500, replace=False)
imp = imputar_estadistico(x_falta[:, None], "mean")[:, 0]
miss = np.isnan(x_falta)[idx]
axes[1].scatter(imp[idx][~miss], z[idx][~miss], s=6, color=PALETA[0], alpha=0.5, label="observados")
axes[1].scatter(imp[idx][miss], z[idx][miss], s=6, color=PALETA[1], alpha=0.7, label="imputados (línea vertical)")
axes[1].set_xlabel("X"); axes[1].set_ylabel("Z"); axes[1].legend(fontsize=8); axes[1].set_title("…y destruye la relación con Z", fontweight="bold", fontsize=10)
ps = np.linspace(0, 0.8, 9)
axes[2].plot(ps, np.sqrt(1 - ps), "o-", color=PALETA[0], lw=2, label="desv. típica relativa = $\\sqrt{1-p}$")
axes[2].plot(ps, 1 - ps, "s-", color=PALETA[3], lw=2, label="varianza relativa = $1-p$")
axes[2].set_xlabel("Fracción imputada por media $p$"); axes[2].set_ylabel("Valor relativo a la verdad"); axes[2].legend(fontsize=8); axes[2].set_title("Cuanto más se imputa, más se encoge", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

## 3. Imputación por *k* Vecinos: la Distancia Euclídea con Faltantes 👥

La **imputación por *k* vecinos** (*KNN imputation*; popularizada en bioinformática como *KNNimpute*, Troyanskaya et al., 2001) reemplaza un valor faltante por el **promedio de ese mismo atributo en las $k$ filas más parecidas**. Dos preguntas: ¿cómo medir «parecido» **si faltan valores**?, ¿y quién puede ser **donante**?

**Distancia euclídea con faltantes** (`nan_euclidean`). Para dos filas $\mathbf a,\mathbf b\in\mathbb R^d$ con coordenadas faltantes, se usan **solo las coordenadas presentes en ambas** ($P$, con $|P|=n_P$) y se **reescala** para compensar las ausentes:

$$d(\mathbf a,\mathbf b)=\sqrt{\frac{d}{n_P}\sum_{j\in P}(a_j-b_j)^2}.$$

Si $n_P=0$ (**ningún** atributo en común), la distancia **no está definida** (`NaN`).

**El algoritmo** (el de `KNNImputer`) para cada **receptor** $i$ con $x_{ij}$ faltante:
1. **Donantes:** las filas con $x_{\cdot j}$ **observado** (un donante nunca aporta un valor imputado, solo observados).
2. Calcula $d(\mathbf x_i,\mathbf x_{i'})$ con la distancia anterior (sobre las filas **originales**, con sus faltantes).
3. Toma los $k$ donantes **más cercanos** y promedia sus valores en la columna $j$: **uniforme** ($\frac1k\sum$) o **ponderado por inverso de la distancia** ($w=1/d$; si algún $d=0$, solo cuentan los donantes a distancia cero).
4. Si el receptor **no tiene ninguna distancia definida** a los donantes, se imputa la **media de la columna**.

> 🔎 **Desempates y condiciones de igualdad.** `KNNImputer` elige a los $k$ vecinos con `np.argpartition` (**no estable**). Si dos donantes empatan en la frontera del $k$-ésimo lugar, **cuál entra puede diferir** entre implementaciones. Con datos **continuos y sin valores repetidos** los empates tienen probabilidad cero y nuestra versión (que usa `argsort` estable) **debe coincidir**; con datos **discretizados** o con valores repetidos (como las mediciones redondeadas de *Breast Cancer*) **puede haber empates numéricos** y entonces **no se garantiza** la igualdad exacta en esas celdas, aunque ambas versiones son «correctas». También hay que fijar **qué se escala**: la distancia es sensible a la **escala** de las variables: **estandariza antes**.

````python
# ============================================================
# 3. nan_euclidean y KNN-imputer desde cero == sklearn (valores identicos con los mismos parametros)
# ============================================================
def nan_euclid(A, B):
    """Distancia euclidea con faltantes: sqrt( d/n_presentes * suma_{j presente en ambos} (a_j - b_j)^2 ); NaN si no hay coordenadas en comun."""
    mA, mB = ~np.isnan(A), ~np.isnan(B)
    A0, B0 = np.nan_to_num(A), np.nan_to_num(B)
    comunes = mA[:, None, :] & mB[None, :, :]                       # (n, m, d): coordenadas presentes en AMBAS filas
    ss = (((A0[:, None, :] - B0[None, :, :]) ** 2) * comunes).sum(axis=2)
    n_p = comunes.sum(axis=2)
    with np.errstate(divide="ignore", invalid="ignore"):
        D2 = ss * A.shape[1] / n_p
    D2[n_p == 0] = np.nan
    return np.sqrt(D2)


def knn_imputar(X, k=5, pesos="uniform"):
    """KNN-imputer: valor = promedio (uniforme o ponderado por 1/d) del atributo en los k donantes mas cercanos segun nan_euclid."""
    X = np.asarray(X, dtype=float)
    falta = np.isnan(X)
    D = nan_euclid(X, X)                                            # distancias sobre las filas ORIGINALES (con sus NaN)
    media_col = np.nanmean(X, axis=0)
    Xi = X.copy()
    for j in range(X.shape[1]):
        donantes = np.where(~falta[:, j])[0]
        for i in np.where(falta[:, j])[0]:
            dist = D[i, donantes]
            if np.isnan(dist).all():
                Xi[i, j] = media_col[j]; continue                    # sin ninguna distancia definida -> media de la columna
            orden = np.argsort(dist, kind="stable")[:min(k, len(donantes))]      # los NaN quedan al final; desempate estable (por indice)
            dd, v = dist[orden], X[donantes[orden], j]
            if pesos == "uniform":
                w = np.where(np.isnan(dd), 0.0, 1.0)
            else:
                w = (dd == 0).astype(float) if (dd == 0).any() else 1.0 / dd          # d = 0: solo cuentan los donantes identicos
                w = np.where(np.isnan(w), 0.0, w)
            Xi[i, j] = (w * v).sum() / w.sum()
    return Xi


rng = np.random.default_rng(SEED)
# (a) distancia: igual que sklearn (incluye filas sin coordenadas comunes -> NaN)
A = rng.normal(size=(30, 8)); A[rng.random(A.shape) < 0.35] = np.nan
A[0, :4] = np.nan; A[1, 4:] = np.nan; A[0, 4:] = 1.0; A[1, :4] = 2.0                       # filas 0 y 1: NINGUNA coordenada en comun
D_mia, D_sk = nan_euclid(A, A), nan_euclidean_distances(A, A)
assert np.allclose(D_mia, D_sk, equal_nan=True) and np.isnan(D_mia[0, 1])
print(f"(a) nan_euclid == nan_euclidean_distances de sklearn (incluida la distancia indefinida entre filas sin coordenadas comunes: {D_mia[0, 1]}).")

# (b) imputador: igual que KNNImputer en datos CONTINUOS, salvo celdas con EMPATE de distancia en la frontera del k-esimo vecino
def celdas_con_empate(X, k):
    """True en las celdas faltantes donde la distancia al k-esimo y al (k+1)-esimo donante coincide (a 1e-9): alli el desempate decide quien entra."""
    D, falta = nan_euclid(X, X), np.isnan(X)
    E = np.zeros(X.shape, dtype=bool)
    for j in range(X.shape[1]):
        donantes = np.where(~falta[:, j])[0]
        for i in np.where(falta[:, j])[0]:
            dd = np.sort(D[i, donantes][~np.isnan(D[i, donantes])])
            if len(dd) > k and abs(dd[k - 1] - dd[k]) < 1e-9:
                E[i, j] = True
    return E


Xb = StandardScaler().fit_transform(Xbc)[:, :10]
n_celdas = n_difieren = n_empates = 0
for tasa in (0.1, 0.3):
    Xm = Xb.copy(); Xm[np.random.default_rng(SEED).random(Xm.shape) < tasa] = np.nan
    for k in (1, 3, 5, 10):
        emp = celdas_con_empate(Xm, k)
        for pesos in ("uniform", "distance"):
            mio = knn_imputar(Xm, k, pesos)
            ref = KNNImputer(n_neighbors=k, weights=pesos).fit_transform(Xm)
            dif = ~np.isclose(mio, ref, atol=1e-5, rtol=0)                  # tolerancia 1e-5: sklearn calcula d con ||a||²+||b||²-2a·b (error ~1e-8) y con pesos 1/d y d≈0 lo amplifica
            assert not (dif & ~emp).any(), (tasa, k, pesos)               # fuera de los empates la igualdad se cumple a 1e-5
            n_celdas += int(np.isnan(Xm).sum()); n_difieren += int(dif.sum()); n_empates += int(emp.sum())
print(f"(b) knn_imputar == KNNImputer en {n_celdas - n_difieren:,} de {n_celdas:,} celdas imputadas (k ∈ {{1, 3, 5, 10}}, pesos uniformes y por distancia, 10 % y 30 % de faltantes). "
      f"Las {n_difieren} que difieren (tolerancia 1e-5) caen todas en celdas con empate de distancia en la frontera del k-ésimo vecino (hay {n_empates} celdas con empate en total).")

# (c) con DATOS DISCRETIZADOS hay empates: la igualdad exacta ya no esta garantizada (se documenta, no se exige)
Xd = np.round(Xb * 1.0)                                                                        # pocos valores distintos -> muchas distancias empatadas
Xd[np.random.default_rng(SEED).random(Xd.shape) < 0.3] = np.nan
mio, ref = knn_imputar(Xd, 5), KNNImputer(n_neighbors=5).fit_transform(Xd)
frac_dif = (~np.isclose(mio, ref)).mean()
print(f"(c) Con datos discretizados (empates de distancia), el {100 * frac_dif:.1f} % de las celdas difiere entre las dos implementaciones: "
      "ambas son válidas, pero el desempate (argsort estable frente a argpartition) cambia qué donantes entran.")

# (d) Efecto de k y comparacion con media: RMSE de imputacion sobre valores ocultos (MCAR 30 %)
verdad_ = Xb
M = np.random.default_rng(SEED).random(Xb.shape) < 0.30
Xm = np.where(M, np.nan, Xb)
rmse = lambda Xi: np.sqrt(((Xi - verdad_)[M] ** 2).mean())
ks = [1, 2, 3, 5, 8, 12, 20, 40]
r_knn = [rmse(knn_imputar(Xm, k)) for k in ks]
r_media = rmse(imputar_estadistico(Xm, "mean"))
fig, ax = plt.subplots(figsize=(8.5, 4.4))
ax.plot(ks, r_knn, "o-", color=PALETA[0], lw=2, label="KNN (uniforme)"); ax.axhline(r_media, color=PALETA[1], ls="--", label=f"media (RMSE = {r_media:.2f})")
ax.set_xscale("log"); ax.set_xticks(ks); ax.set_xticklabels(ks); ax.minorticks_off(); ax.set_xlabel("Número de vecinos $k$"); ax.set_ylabel("RMSE de imputación (datos estandarizados)"); ax.legend()
ax.set_title("El KNN supera a la media si las variables están correlacionadas", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
print(f"RMSE de imputación: media = {r_media:.3f}; KNN con k = 5 → {r_knn[ks.index(5)]:.3f}; mejor k en la rejilla = {ks[int(np.argmin(r_knn))]} ({min(r_knn):.3f}).")
assert min(r_knn) < r_media                                                                     # con 10 variables correlacionadas, vecinos > media
````

**Lectura del KNN.** (a) Nuestra distancia **coincide** con `nan_euclidean_distances`, incluido el caso de filas **sin coordenadas en común** (distancia `NaN`). (b) Nuestro imputador **coincide con `KNNImputer`** (diferencia ≤ $10^{-5}$; `scikit-learn` calcula las distancias con $\|\mathbf a\|^2+\|\mathbf b\|^2-2\mathbf a\cdot\mathbf b$, con error de ≈ $10^{-8}$ que los pesos $1/d$ amplifican cuando $d\approx0$, es decir, con filas casi duplicadas) en **casi todas** las celdas, para varios $k$, ambos esquemas de peso y distintas tasas de faltantes; las **pocas** que difieren están **todas** en celdas donde la distancia al $k$-ésimo y al $(k+1)$-ésimo donante **coincide numéricamente** (empate por valores repetidos en los datos): el desempate decide quién entra y `argpartition` y `argsort` no tienen por qué elegir igual. **Fuera de los empates la igualdad es exacta.** (c) Con datos **discretizados** (empates de distancia) **cambian algunos donantes** según cómo se desempate; **no es un error de ninguna de las dos**: es una ambigüedad del algoritmo que conviene conocer antes de exigir igualdad entre librerías. (d) Con variables **correlacionadas**, el KNN **reduce el error de imputación** frente a la media; **$k$ importa** (con $k=1$ hay mucha varianza, con $k$ muy grande el KNN **se parece a la media**).

## 4. Imputación Iterativa (MICE) e Idea de la Imputación Múltiple 🔁

### 4.1 Ecuaciones encadenadas

La **imputación multivariada por ecuaciones encadenadas** (*MICE*, *chained equations*; van Buuren y Groothuis-Oudshoorn, 2011), implementada en `IterativeImputer`, **modela cada variable con faltantes como función de las demás**:

1. Inicializa los faltantes (p. ej. con la media).
2. **Ciclo** sobre las columnas $j$ con faltantes: ajusta $x_j\sim f(\mathbf x_{-j})$ con las filas donde $x_j$ **está observada** y **reemplaza** sus faltantes con la predicción.
3. **Repite** hasta que las imputaciones se estabilicen (`max_iter`).

`scikit-learn` usa por defecto `BayesianRidge` como $f$. Es **más flexible** que el KNN (aprovecha relaciones lineales entre variables de forma global) y es la base de la **imputación múltiple**. Es **experimental** en `scikit-learn` (hay que habilitarlo con `sklearn.experimental`): la API puede cambiar, por eso lo importamos con `try/except`.

### 4.2 Imputación múltiple y las reglas de Rubin

Una **única** imputación (media, KNN o MICE determinista) trata los valores imputados **como si fueran reales** y **ignora la incertidumbre** de haberlos imputado: los **errores estándar** salen **demasiado pequeños**. La **imputación múltiple** (Rubin, 1987) genera $m$ conjuntos completos (con `sample_posterior=True`, **sorteando** las imputaciones de su distribución predictiva), **analiza cada uno** y **combina** con las **reglas de Rubin**:

$$\bar Q=\frac1m\sum_{l=1}^m \hat Q_l,\qquad \bar U=\frac1m\sum_{l=1}^m \hat U_l,\qquad B=\frac1{m-1}\sum_{l=1}^m(\hat Q_l-\bar Q)^2,\qquad T=\bar U+\Big(1+\frac1m\Big)B,$$

con $\hat Q_l$ la estimación del parámetro (aquí, la media) y $\hat U_l$ su varianza estimada en el conjunto $l$. El error estándar combinado es $\sqrt T$: $\bar U$ es la variabilidad **«dentro»** de cada conjunto y $B$ la **«entre» imputaciones**, es decir, **la incertidumbre por los datos faltantes**. Lo demostramos con una media bajo **MAR** (30 % faltante en $X$ según $Z$) comparando **tres errores estándar**: el **ingenuo** (imputación por media, tratada como datos reales), el de la **verdad** (datos completos, que no tendríamos) y el de **Rubin**.

````python
# ============================================================
# 4. MICE (IterativeImputer) e imputacion multiple con las reglas de Rubin (media de X bajo MAR)
# ============================================================
if IterativeImputer is None:
    print("IterativeImputer no está disponible en esta versión de scikit-learn (es experimental): se omite esta sección.")
else:
    rng = np.random.default_rng(SEED)
    n = 600
    x, z = generar_xz(n, rng)
    w2 = rng.normal(0, 1, n)                                       # una tercera variable (ruido) para que el imputador tenga contexto
    falta = mascara("MAR", x, z, rng)
    Xobs = np.column_stack([np.where(falta, np.nan, x), z, w2])

    # (a) Error estandar INGENUO: imputar la media y calcular s/sqrt(n) como si todo fuera real
    x_media = imputar_estadistico(Xobs, "mean")[:, 0]
    se_ingenuo = x_media.std(ddof=1) / np.sqrt(n)
    se_verdad = x.std(ddof=1) / np.sqrt(n)                         # con los datos completos (que en la practica NO se tienen)

    # (b) Imputacion multiple: m conjuntos con sample_posterior=True + reglas de Rubin
    m = 20
    Q, U = [], []
    for l in range(m):
        Xi = IterativeImputer(sample_posterior=True, max_iter=10, random_state=l).fit_transform(Xobs)[:, 0]
        Q.append(Xi.mean()); U.append(Xi.var(ddof=1) / n)
    Q, U = np.array(Q), np.array(U)
    Q_barra, U_barra, B = Q.mean(), U.mean(), Q.var(ddof=1)
    T = U_barra + (1 + 1 / m) * B
    se_rubin = np.sqrt(T)

    # (c) Comparacion de estimaciones de la media (verdad = media muestral completa) bajo MAR
    Xi_det = IterativeImputer(max_iter=10, random_state=0).fit_transform(Xobs)[:, 0]
    tab = pd.DataFrame({"estimación de la media": [x.mean(), x[~falta].mean(), x_media.mean(), Xi_det.mean(), Q_barra],
                        "error estándar": [se_verdad, x[~falta].std(ddof=1) / np.sqrt((~falta).sum()), se_ingenuo, Xi_det.std(ddof=1) / np.sqrt(n), se_rubin]},
                       index=["Datos completos (verdad, no disponible)", "Casos completos (sesgados bajo MAR)", "Imputación por media (SE ingenuo)", "MICE una vez (SE ingenuo)", f"Imputación múltiple (m = {m}, Rubin)"])
    display(tab.round(4))
    print(f"Descomposición de Rubin: Ū = {U_barra:.2e} (dentro) | B = {B:.2e} (entre imputaciones) | T = Ū + (1 + 1/m) B = {T:.2e}  →  SE = {se_rubin:.4f}")
    print(f"Errores estándar: ingenuo (media) = {se_ingenuo:.4f}  <  verdad (datos completos) = {se_verdad:.4f}  <  Rubin = {se_rubin:.4f}")
    assert B > 0 and T > U_barra                                    # la incertidumbre 'entre imputaciones' existe
    assert se_ingenuo < se_verdad < se_rubin                        # el SE ingenuo es demasiado optimista; Rubin es mas honesto que la verdad (se perdio informacion)
    assert abs(Q_barra - x.mean()) < abs(x[~falta].mean() - x.mean())      # la imputacion usando Z reduce el sesgo de los casos completos bajo MAR

    fig, axes = plt.subplots(1, 2, figsize=(14, 4.4))
    axes[0].hist(Q, bins=10, color=PALETA[0], alpha=0.8)
    axes[0].axvline(x.mean(), color="black", ls="--", label="media verdadera (datos completos)"); axes[0].axvline(Q_barra, color=PALETA[1], label="media combinada (Rubin)")
    axes[0].set_xlabel("Estimación de la media en cada imputación"); axes[0].set_ylabel("Imputaciones"); axes[0].legend(fontsize=8)
    axes[0].set_title("Las 20 imputaciones difieren: eso es incertidumbre", fontweight="bold", fontsize=10)
    nombres = ["ingenuo\n(media)", "verdad\n(completos)", "Rubin\n(múltiple)"]
    axes[1].bar(nombres, [se_ingenuo, se_verdad, se_rubin], color=[PALETA[1], PALETA[9], PALETA[2]]); axes[1].set_ylabel("Error estándar de la media")
    axes[1].set_title("La imputación única subestima la incertidumbre", fontweight="bold", fontsize=10)
    plt.tight_layout(); plt.show()
````

---
**Lectura de MICE e imputación múltiple.** Las 20 imputaciones dan **estimaciones distintas** de la media (la varianza «entre» $B$): eso **es** la incertidumbre por los datos faltantes. El error estándar **ingenuo** (tratar lo imputado como real) es **menor** que el de los datos completos (**falsa precisión**), mientras que el de **Rubin** es **mayor** que el de los datos completos (con razón: **se perdió información**). Además, como el imputador **usa $Z$**, la media combinada queda **mucho más cerca** de la verdad que la de los casos completos bajo MAR. En el aprendizaje automático predictivo la imputación múltiple se usa **menos** (el objetivo suele ser predecir, no inferir parámetros), pero la **idea** es importante: **una sola imputación no cuantifica su propia incertidumbre**.

## 5. Evaluar Imputadores: Ocultar Valores que Conocemos 🎭

¿Cómo saber qué imputador es mejor **en nuestros datos**? **No** se puede evaluar con los valores faltantes reales (**no los conocemos**). El protocolo estándar es:

1. Partir de un conjunto **completo**.
2. **Ocultar** valores con un mecanismo controlado (**MCAR, MAR o MNAR**) a una tasa dada.
3. **Imputar** y medir el **RMSE** (en unidades estandarizadas) **solo en las celdas ocultas**.

Así se compara en **igualdad de condiciones**. Usamos las **10 primeras variables** de *Breast Cancer* (estandarizadas: **correlacionadas** entre sí) y tres mecanismos por **columna**: **MCAR** (probabilidad fija), **MAR** (la probabilidad de que falte la columna $j$ **crece con el valor de la columna** $j+1$, que queda observada para ese caso) y **MNAR** (la probabilidad **crece con el propio valor**). Probamos a tasas de 20 % y 40 %, con 3 repeticiones por combinación, y cuatro imputadores: **media**, **mediana**, **KNN** ($k=5$, nuestra implementación) e **iterativo** (`IterativeImputer`, si está disponible).

````python
# ============================================================
# 5. RMSE de imputacion ocultando valores conocidos: mecanismo x tasa x metodo
# ============================================================
t0 = time.time()
Xtrue = StandardScaler().fit_transform(Xbc)[:, :10]


def ocultar(X, mecanismo, tasa, rng):
    """Mascara True = oculto. Por columna j: MCAR p fija; MAR: p crece con la columna (j+1) (observada); MNAR: p crece con el propio valor."""
    M = np.zeros(X.shape, dtype=bool)
    for j in range(X.shape[1]):
        if mecanismo == "MCAR":
            p = np.full(len(X), tasa)
        elif mecanismo == "MAR":
            p = prob_falta(X[:, (j + 1) % X.shape[1]], objetivo=tasa, fuerza=1.5)
        else:
            p = prob_falta(X[:, j], objetivo=tasa, fuerza=1.5)
        M[:, j] = rng.random(len(X)) < p
    return M


imputadores = {"Media": lambda Xm: imputar_estadistico(Xm, "mean"), "Mediana": lambda Xm: imputar_estadistico(Xm, "median"), "KNN (k = 5)": lambda Xm: knn_imputar(Xm, 5)}
if IterativeImputer is not None:
    imputadores["Iterativo (MICE)"] = lambda Xm: IterativeImputer(max_iter=10, random_state=0).fit_transform(Xm)
registros = []
for mec in ["MCAR", "MAR", "MNAR"]:
    for tasa in [0.2, 0.4]:
        for rep in range(3):
            M = ocultar(Xtrue, mec, tasa, np.random.default_rng(SEED + rep))
            Xm = np.where(M, np.nan, Xtrue)
            if np.isnan(Xm).all(axis=0).any():
                continue
            for nombre, f in imputadores.items():
                registros.append({"mecanismo": mec, "tasa": tasa, "método": nombre, "RMSE": np.sqrt(((f(Xm) - Xtrue)[M] ** 2).mean())})
res = pd.DataFrame(registros)
tab = res.pivot_table(index=["mecanismo", "tasa"], columns="método", values="RMSE").reindex(["MCAR", "MAR", "MNAR"], level=0)[list(imputadores)]
display(tab.round(3))
print(f"[{time.time() - t0:.0f} s]  RMSE en desviaciones típicas (1.0 = el error de predecir siempre la media si no hubiera estructura).")

fig, axes = plt.subplots(1, 2, figsize=(15, 4.6))
for ax, tasa in zip(axes, [0.2, 0.4]):
    sub = tab.xs(tasa, level="tasa")
    ancho = 0.8 / len(imputadores)
    for k, (met, c) in enumerate(zip(imputadores, [PALETA[9], PALETA[3], PALETA[0], PALETA[2]])):
        ax.bar(np.arange(3) + (k - (len(imputadores) - 1) / 2) * ancho, sub[met], ancho, color=c, label=met)
    ax.set_xticks(np.arange(3)); ax.set_xticklabels(sub.index); ax.set_ylabel("RMSE de imputación (en desv. típicas)"); ax.legend(fontsize=8)
    ax.set_title(f"{int(100 * tasa)} % de valores ocultos", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()

assert (tab["KNN (k = 5)"] < tab["Media"]).all()                                          # con variables correlacionadas, KNN < media en todos los mecanismos
if "Iterativo (MICE)" in tab:
    assert (tab["Iterativo (MICE)"] < tab["Media"]).all()
rel = (tab["Media"] - tab["KNN (k = 5)"]) / tab["Media"]
print("Reducción del RMSE del KNN frente a la media (%):", {f"{m} {int(100 * t)} %": round(100 * float(v), 0) for (m, t), v in rel.items()})
print("MNAR es el mecanismo más difícil: todos los métodos pierden precisión porque el valor ausente es justamente el extremo." if tab.loc["MNAR"].mean().mean() > tab.loc["MCAR"].mean().mean() else "")
````

---
**Lectura de la comparación.** Con variables **correlacionadas**, el **KNN** y la imputación **iterativa** reducen mucho el RMSE frente a la **media** y la **mediana** en **todos** los mecanismos, porque **aprovechan** la información de las demás columnas. La **mediana** apenas se distingue de la media en datos casi simétricos. El **MNAR** es el más difícil (el valor que falta es justo el **extremo**: las imputaciones se quedan **cortas**), y **ningún** imputador lo corrige por sí solo. **Cuidado con la conclusión:** esto vale para **estos** datos (muy correlacionados); con variables **independientes**, ningún imputador supera a la media (no hay información que aprovechar), y la tasa del 40 % degrada a todos.

## 6. Imputar Dentro del `Pipeline`, Fugas y Modelos que Aceptan `NaN` 🚰

### 6.1 La regla

Los imputadores **aprenden** (la media, la mediana, los donantes del KNN): son **transformadores con `fit`**. Deben ajustarse **solo con el entrenamiento** de cada pliegue y **aplicarse** al resto: dentro de un `Pipeline` (**[Módulo 00](../00%20-%20Introduccion%20al%20Machine%20Learning/03_Importancia_del_Preprocesamiento.ipynb)**). Si se imputa **antes** de dividir, las estadísticas de imputación usan **también** los datos de validación (fuga). La fuga de una imputación **sin la variable objetivo** suele ser **pequeña**; la de una imputación que **usa la etiqueta** es **grave** (aquí lleva la exactitud de ≈ 96.5 % a ≈ 99.3 %, y sería peor con un modelo más lejos del techo del 100 %). Medimos ambas con *Breast Cancer* y 30 % de valores **MCAR**:

- **A. Imputar fuera del `Pipeline`** (media sobre todo el conjunto, **sin** usar $y$): fuga **leve**.
- **B. Imputar dentro del `Pipeline`**: correcto.
- **C. Imputar con la media **de la clase**** (usa $y$ sobre todo el conjunto): la imputación **codifica la etiqueta** en el propio dato, un **trampa** que en producción **no se puede reproducir** (no hay $y$).

### 6.2 Modelos que aceptan `NaN` de forma nativa

Algunos modelos **no necesitan imputar**: `HistGradientBoostingClassifier` (y `HistGradientBoostingRegressor`) **aprende en cada división** a qué lado enviar los `NaN` (el valor ausente va hacia donde **más reduce la pérdida** en el entrenamiento). Ventajas: **no inventa valores** y **detecta ausencia informativa**; desventaja: **no existe** para todos los modelos. Comparamos, variando la tasa de faltantes: **(i)** media + regresión logística, **(ii)** KNN + regresión logística, **(iii)** `HistGradientBoosting` **con `NaN`** y, para la **ausencia informativa** (probabilidad de faltar distinta según la clase), **(iv)** media + **indicador** + regresión logística.

````python
# ============================================================
# 6. Fuga al imputar (A, B, C) y comparacion con HistGradientBoosting nativo; ausencia informativa e indicador
# ============================================================
t0 = time.time()
Xbc_raw, ybc = load_breast_cancer(return_X_y=True)
rng = np.random.default_rng(SEED)
Xm30 = np.where(rng.random(Xbc_raw.shape) < 0.30, np.nan, Xbc_raw)                          # 30 % MCAR
cv = RepeatedStratifiedKFold(n_splits=5, n_repeats=4, random_state=SEED)


def acc_cv(modelo, X, y):
    return cross_val_score(modelo, X, y, cv=cv, scoring="accuracy")


# A. Imputar ANTES de dividir (media de todo el conjunto, sin usar y)
Xa = SimpleImputer(strategy="mean").fit_transform(Xm30)
acc_a = acc_cv(make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000)), Xa, ybc)
# B. Imputar DENTRO del Pipeline
acc_b = acc_cv(make_pipeline(SimpleImputer(strategy="mean"), StandardScaler(), LogisticRegression(max_iter=2000)), Xm30, ybc)
# C. Imputar con la media DE LA CLASE (usa y): fuga grave
Xc = Xm30.copy()
for c in (0, 1):
    medias = np.nanmean(Xm30[ybc == c], axis=0)
    filas, cols = np.where(np.isnan(Xm30) & (ybc[:, None] == c))
    Xc[filas, cols] = medias[cols]
acc_c = acc_cv(make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000)), Xc, ybc)
tab = pd.DataFrame({"accuracy medio": [acc_a.mean(), acc_b.mean(), acc_c.mean()], "desv.": [acc_a.std(), acc_b.std(), acc_c.std()]},
                   index=["A. imputar ANTES de dividir (sin usar y)", "B. imputar DENTRO del Pipeline (correcto)", "C. imputar con la media de la CLASE (usa y)"])
display(tab.round(4))
print(f"Fuga de A frente a B: {100 * (acc_a.mean() - acc_b.mean()):+.2f} puntos (pequeña, dentro de la variabilidad {100 * acc_b.std():.1f} pts). Fuga de C: {100 * (acc_c.mean() - acc_b.mean()):+.1f} puntos (falsa).")
assert abs(acc_a.mean() - acc_b.mean()) < 0.02                                              # imputar sin y: fuga leve
assert acc_c.mean() > acc_b.mean() + 0.02                                                  # imputar con y: fuga clara

# --- Barrido de tasa de faltantes (MCAR) y escenario de AUSENCIA INFORMATIVA ---
def con_faltantes(X, y, tasa, informativa, rng):
    """MCAR con probabilidad `tasa`; si `informativa`, la probabilidad de faltar es 2.5 veces mayor en la clase 0 (maligno) que en la 1."""
    p = np.full(X.shape, tasa)
    if informativa:
        p = np.where((y == 0)[:, None], min(0.95, 1.6 * tasa), 0.4 * tasa) * np.ones(X.shape)
    return np.where(rng.random(X.shape) < p, np.nan, X)


modelos = {"Media + LR": make_pipeline(SimpleImputer(strategy="mean"), StandardScaler(), LogisticRegression(max_iter=2000)),
           "Media + indicador + LR": make_pipeline(SimpleImputer(strategy="mean", add_indicator=True), StandardScaler(), LogisticRegression(max_iter=2000)),
           "KNN + LR": make_pipeline(StandardScaler(), KNNImputer(n_neighbors=5), LogisticRegression(max_iter=2000)),
           "HistGradientBoosting (NaN nativo)": HistGradientBoostingClassifier(max_iter=60, random_state=SEED)}
cv2 = StratifiedKFold(5, shuffle=True, random_state=SEED)
tasas = [0.0, 0.1, 0.3, 0.5]
filas = []
for informativa in (False, True):
    for tasa in tasas:
        Xt = con_faltantes(Xbc_raw, ybc, tasa, informativa, np.random.default_rng(SEED))
        for nombre, m in modelos.items():
            filas.append({"ausencia": "informativa (depende de la clase)" if informativa else "MCAR", "tasa": tasa, "modelo": nombre,
                          "accuracy": cross_val_score(m, Xt, ybc, cv=cv2).mean()})
res = pd.DataFrame(filas)
for aus in res["ausencia"].unique():
    print(f"\nAusencia {aus}:")
    display(res[res.ausencia == aus].pivot(index="tasa", columns="modelo", values="accuracy")[list(modelos)].round(3))
print(f"\n[{time.time() - t0:.0f} s]")

fig, axes = plt.subplots(1, 2, figsize=(15, 4.6), sharey=True)
for ax, aus in zip(axes, res["ausencia"].unique()):
    sub = res[res.ausencia == aus]
    for (nombre, c) in zip(modelos, [PALETA[9], PALETA[4], PALETA[0], PALETA[2]]):
        s = sub[sub.modelo == nombre]
        ax.plot(100 * s["tasa"], s["accuracy"], "o-", color=c, lw=2, label=nombre)
    ax.set_xlabel("Porcentaje de valores faltantes"); ax.set_ylabel("Accuracy (validación cruzada)"); ax.set_title(f"Ausencia {aus}", fontweight="bold", fontsize=10); ax.legend(fontsize=8)
plt.tight_layout(); plt.show()

inf = res[res.ausencia.str.startswith("informativa")].pivot(index="tasa", columns="modelo", values="accuracy")
mcar = res[res.ausencia == "MCAR"].pivot(index="tasa", columns="modelo", values="accuracy")
assert inf.loc[0.3, "Media + indicador + LR"] > inf.loc[0.3, "Media + LR"] + 0.01                # con ausencia informativa, el indicador ayuda
assert mcar.loc[0.5, "KNN + LR"] > 0.8 and mcar.loc[0.5, "HistGradientBoosting (NaN nativo)"] > 0.8   # incluso con 50 % de faltantes se predice razonablemente
````

---
**Lectura de la fuga y los modelos.** (**A frente a B**) imputar **antes** de dividir, **sin** usar la etiqueta, produce una fuga **muy leve** (dentro de la variabilidad entre pliegues): conviene el `Pipeline` **por higiene y por seguridad** (la magnitud depende del imputador y del tamaño de los datos), no porque aquí cambie mucho el número. (**C**) imputar con la **media de la clase** **infla** la exactitud (≈ 99.3 % frente a ≈ 96.5 %; el techo del 100 % limita el efecto visible) porque el valor imputado **ya trae la etiqueta**; en producción no existe $y$ y el modelo **perdería** esa ventaja ficticia. **Con MCAR**, los tres enfoques (media, KNN y `HistGradientBoosting` nativo) se degradan **despacio** al crecer la tasa de faltantes gracias a la **redundancia** de las 30 variables (otra vez: resultado de **estos** datos tan correlacionados). **Con ausencia informativa**, el **indicador** (`add_indicator`) y el **modelo con `NaN` nativo** pueden **explotar** el patrón de ausencia, algo que la imputación sola **borra**; ojo: si esa ausencia **no** existirá en producción, el modelo **dependería de un artefacto**. Con **MCAR**, en cambio, el indicador **no ayuda** (añade 30 columnas de puro ruido y baja ≈ 2 puntos), y el modelo con `NaN` nativo es **competitivo pero no superior** a la media o al KNN.

## 7. Guía de Decisión: ¿Qué Hacer con los Faltantes? 🧭

### 7.1 ¿Cuánto faltante es tolerable?

No hay un umbral universal; orientaciones **heurísticas** (no reglas): **< 5 %** por variable y sin patrón: casi cualquier método razonable funciona; **5–30 %**: usa imputación que aproveche las demás variables y **evalúa con enmascaramiento** (sección 5); **> 30–50 %** en una variable: considera **descartarla** o convertirla en un **indicador** de «medido/no medido» salvo que sea **crucial**; si el faltante es **MNAR**, **ninguna** imputación basta: busca **información externa** o modela el mecanismo.

### 7.2 Árbol de decisión

| Pregunta | Si… | Entonces |
|---|---|---|
| ¿Por qué faltan los datos? | MCAR plausible | Casos completos (si sobran datos) o imputación simple; sin sesgo |
| | MAR plausible | **Imputa usando las variables relacionadas** (KNN, MICE, regresión) o modelos con `NaN` nativo |
| | MNAR plausible | Ninguna imputación corrige el sesgo: **modelo del mecanismo**, análisis de sensibilidad, **datos externos** |
| ¿Importa inferir parámetros con su incertidumbre? | Sí | **Imputación múltiple** (reglas de Rubin) |
| ¿Solo predecir? | Sí | Imputación dentro del **`Pipeline`**; evalúa por **validación cruzada** el pipeline **completo** |
| ¿La ausencia puede ser informativa? | Sí | **Indicador** (`add_indicator=True`) o modelos con **`NaN` nativo** (`HistGradientBoosting`) |
| ¿Variables **muy correlacionadas**? | Sí | **KNN** o **iterativa** superan a la media; con variables independientes, no |
| ¿Variables **categóricas**? | Sí | **Moda** o una categoría «**desconocido**» explícita; evita media sobre códigos |
| ¿El conjunto es **enorme**? | Sí | KNN cuesta $O(n^2)$ en distancias: prefiere media/mediana, MICE con pocas iteraciones o `HistGradientBoosting` |

### 7.3 Buenas prácticas

1. **Audita primero** (cuántos, dónde y con qué patrón faltan) y **formula una hipótesis** sobre el mecanismo (enlaces del inicio).
2. **Imputa siempre dentro del `Pipeline`** y **nunca uses la etiqueta**.
3. **No uses la media sin pensar**: encoge la varianza y debilita las correlaciones; **mide** el efecto.
4. **Evalúa imputadores ocultando valores conocidos** y reporta el RMSE **por mecanismo**.
5. **Guarda el indicador** cuando la ausencia pueda significar algo.
6. **Documenta** el tratamiento: es una **decisión de modelado** que afecta a las conclusiones.

Verifica a mano la distancia y el imputador de la Sección 3 con datos pequeños. Con `nan_euclid`, `knn_imputar`, `nan_euclidean_distances` y `KNNImputer`:

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Mecanismos** | **MCAR** (no depende de nada: casos completos sin sesgo), **MAR** (depende de lo observado: corregible usando las observadas), **MNAR** (depende del propio valor: **no** corregible con los datos). **No se puede distinguir MAR de MNAR** solo con los datos |
| **Casos completos** | Con $d$ variables que fallan con probabilidad $p$, quedan $(1-p)^d$ filas (0.95³⁰ ≈ 21 %). Bajo MAR y MNAR **sesgan** la media (simulado: −0.2 y peor) |
| **Media / mediana / moda** | Verificadas contra `SimpleImputer`. **Encogen** la desviación típica en $\sqrt{1-p}$ y **atenúan** la correlación en $\sqrt{1-p}$ (fórmulas verificadas con $p=30\,\%$) |
| **Indicador de faltantes** | `add_indicator=True` añade una columna «¿faltaba?»; clave cuando la **ausencia es informativa** (el $R^2$ pasa de ≈ 0.47 a ≈ 0.80 en el ejemplo) |
| **KNN** | $d=\sqrt{\frac d{n_P}\sum_{j\in P}(a_j-b_j)^2}$ con coordenadas comunes; promedio de los $k$ donantes más cercanos. **Idéntico a `KNNImputer`** salvo en las celdas con empate de distancia, donde el desempate puede diferir (verificado). Mejora a la media si las variables **se correlacionan** |
| **MICE e imputación múltiple** | Cada variable se modela con las demás (iterativo, experimental). Una sola imputación **subestima la incertidumbre**; las reglas de Rubin $T=\bar U+(1+1/m)B$ la recuperan |
| **Evaluar imputadores** | Ocultar valores conocidos y medir el **RMSE por mecanismo**: KNN e iterativa ≪ media con variables correlacionadas; el **MNAR** es el más difícil |
| **Fuga y modelos con `NaN`** | Imputar **dentro del `Pipeline`**; imputar con la **etiqueta** es una fuga grave (+2.8 pts aquí). `HistGradientBoosting` acepta `NaN` y explota la ausencia informativa |
| **Decidir** | Parte del **mecanismo** y del **objetivo** (inferir o predecir); evalúa por enmascaramiento; documenta |
