## 1. La Neurona Artificial, el Perceptrón y las Activaciones ⚡

### 1.1 La neurona artificial

Una **neurona artificial** calcula una combinación lineal de sus entradas y le aplica una **función de activación** $g$:

$$a = g\big(\mathbf{w}^\top\mathbf{x} + b\big).$$

Con $g(z) = \sigma(z)$ es **exactamente la regresión logística** del [Módulo 01](../01%20-%20Modelos%20Lineales%20Supervisados/02_Regresion_Logistica.ipynb): una red neuronal es una **red de regresiones logísticas** (o con otras activaciones) apiladas, donde la salida de unas es la entrada de otras. McCulloch y Pitts (1943) propusieron la primera neurona lógica; Rosenblatt (1958) el **perceptrón** con una regla de aprendizaje.

### 1.2 El perceptrón

El perceptrón usa la **función escalón**: $\hat{y} = \mathbf{1}[\mathbf{w}^\top\mathbf{x} + b > 0]$. Su **regla de aprendizaje** recorre los ejemplos y corrige solo cuando se equivoca ($y, \hat y \in \{0,1\}$):

$$\mathbf{w} \leftarrow \mathbf{w} + \eta\,(y - \hat{y})\,\mathbf{x}, \qquad b \leftarrow b + \eta\,(y - \hat{y}).$$

Si $y=1$ y $\hat y=0$ el vector $\mathbf{w}$ se **acerca** a $\mathbf{x}$; si $y=0$ y $\hat y = 1$ se **aleja**. **Teorema de convergencia del perceptrón** (Novikoff, 1962): si los datos son **linealmente separables** con margen $\gamma$ (para algún vector unitario $\mathbf{u}$, $y_i^\pm\,\mathbf{u}^\top\tilde{\mathbf{x}}_i \ge \gamma$ con $y^\pm=\pm1$ y $\tilde{\mathbf x} = (\mathbf x, 1)$) y $\lVert\tilde{\mathbf{x}}_i\rVert \le R$, el algoritmo (desde $\mathbf{w}=\mathbf{0}$, $\eta=1$) hace **a lo sumo $(R/\gamma)^2$ correcciones** y termina. **Pero** si los datos no son separables (el caso del **XOR**, Minsky y Papert, 1969) **nunca converge**: una sola neurona solo produce fronteras lineales.

### 1.3 Funciones de activación

Sin no linealidad, apilar capas no sirve: la composición de funciones lineales es lineal. Las activaciones más usadas:

| Activación | $g(z)$ | $g'(z)$ | Comentario |
|---|---|---|---|
| **Sigmoide** | $\dfrac{1}{1+e^{-z}}$ | $g(1-g) \le 0.25$ | Salida en $(0,1)$; **satura** y desvanece el gradiente |
| **Tanh** | $\dfrac{e^z-e^{-z}}{e^z+e^{-z}}$ | $1-g^2 \le 1$ | Centrada en 0; también satura |
| **ReLU** | $\max(0, z)$ | $\mathbf{1}[z > 0]$ | Barata, no satura para $z>0$; estándar en redes profundas; «neuronas muertas» si $z<0$ siempre |
| **Leaky ReLU** | $\max(\alpha z, z)$, $\alpha \approx 0.01$ | $1$ si $z>0$, $\alpha$ si no | Evita neuronas muertas |

**Gradiente que se desvanece.** Al retropropagar por $L$ capas se **multiplican** derivadas de activaciones; con sigmoide cada factor es $\le 0.25$, así que el gradiente puede caer como $0.25^L$. Por eso ReLU y las buenas inicializaciones permitieron entrenar redes profundas (Géron, Cap. 11).

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q torch
# (numpy, pandas, matplotlib, seaborn y scikit-learn vienen preinstalados en Google Colab)

import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.special import expit            # sigmoide numericamente estable
from sklearn.datasets import make_moons, make_classification, load_iris, load_breast_cancer, load_wine, load_digits
from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.neural_network import MLPClassifier

warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (9, 5)
print("Entorno listo.")
````

````python
# ============================================================
# 1.2 El perceptron: converge si hay separabilidad (con la cota de Novikoff) y fracasa en el XOR
# ============================================================
class Perceptron:
    """Perceptron de Rosenblatt: w <- w + eta (y - y_hat) x, con el sesgo incluido como una columna de unos."""
    def __init__(self, eta=1.0, max_epocas=100):
        self.eta, self.max_epocas = eta, max_epocas

    def fit(self, X, y):
        Xa = np.c_[X, np.ones(len(X))]                                   # x~ = (x, 1): el ultimo peso es el sesgo b
        self.w_ = np.zeros(Xa.shape[1])
        self.errores_, self.n_correcciones_ = [], 0
        for _ in range(self.max_epocas):
            errores = 0
            for xi, yi in zip(Xa, y):
                y_hat = int(xi @ self.w_ > 0)
                if y_hat != yi:                                           # corrige solo cuando se equivoca
                    self.w_ += self.eta * (yi - y_hat) * xi
                    errores += 1; self.n_correcciones_ += 1
            self.errores_.append(errores)
            if errores == 0:
                break
        return self

    def predict(self, X):
        return (np.c_[X, np.ones(len(X))] @ self.w_ > 0).astype(int)

# (a) Datos separables: Iris, setosa vs. versicolor con largo y ancho del petalo
iris = load_iris()
mascara = iris.target < 2
Xp, yp = iris.data[mascara][:, 2:4], iris.target[mascara]
per = Perceptron().fit(Xp, yp)
assert (per.predict(Xp) == yp).all()

# Cota de Novikoff: usamos como separador unitario u = w_final / ||w_final||
Xa = np.c_[Xp, np.ones(len(Xp))]
u = per.w_ / np.linalg.norm(per.w_)
gamma = np.min((2 * yp - 1) * (Xa @ u))
R = np.max(np.linalg.norm(Xa, axis=1))
print(f"Separable: convergió en {len(per.errores_)} épocas con {per.n_correcciones_} correcciones. "
      f"Cota de Novikoff (R/γ)² = ({R:.2f}/{gamma:.3f})² = {(R / gamma) ** 2:.0f}")
assert gamma > 0 and per.n_correcciones_ <= (R / gamma) ** 2

# (b) XOR: no es linealmente separable -> nunca converge
Xx = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=float); yx = np.array([0, 1, 1, 0])
per_x = Perceptron(max_epocas=100).fit(Xx, yx)
print(f"XOR: tras {len(per_x.errores_)} épocas todavía comete {per_x.errores_[-1]} errores por época; precisión final = {(per_x.predict(Xx) == yx).mean():.2f}")
assert per_x.errores_[-1] > 0 and (per_x.predict(Xx) == yx).mean() < 1

fig, axes = plt.subplots(1, 3, figsize=(18, 4.8))
g1, g2 = np.meshgrid(np.linspace(0.5, 5.5, 200), np.linspace(-0.2, 1.8, 200))
axes[0].contourf(g1, g2, per.predict(np.c_[g1.ravel(), g2.ravel()]).reshape(g1.shape), levels=[-0.5, 0.5, 1.5], colors=["#bfdbfe", "#fecaca"], alpha=0.8)
for k, nombre, color in [(0, "setosa", "#2563eb"), (1, "versicolor", "#dc2626")]:
    axes[0].scatter(Xp[yp == k, 0], Xp[yp == k, 1], color=color, edgecolor="white", label=nombre)
axes[0].set_xlabel("Largo del pétalo (cm)"); axes[0].set_ylabel("Ancho del pétalo (cm)")
axes[0].set_title("Perceptrón en datos separables", fontweight="bold"); axes[0].legend()
axes[1].plot(range(1, len(per.errores_) + 1), per.errores_, "o-", color="#16a34a", label="Iris (separable)")
axes[1].plot(range(1, 41), per_x.errores_[:40], "s-", color="#dc2626", label="XOR (no separable)")
axes[1].set_xlabel("Época"); axes[1].set_ylabel("Errores de entrenamiento en la época")
axes[1].set_title("Converge a 0 solo si hay separabilidad", fontweight="bold"); axes[1].legend()
axes[2].scatter(Xx[:, 0], Xx[:, 1], c=yx, cmap="coolwarm", s=300, edgecolor="black")
for (a, b), c in zip(Xx, yx):
    axes[2].text(a, b, str(c), ha="center", va="center", color="white", fontweight="bold", fontsize=14)
axes[2].set_xlim(-0.5, 1.5); axes[2].set_ylim(-0.5, 1.5); axes[2].set_xlabel("$x_1$"); axes[2].set_ylabel("$x_2$")
axes[2].set_title("XOR: ninguna recta separa 0s y 1s", fontweight="bold")
plt.tight_layout()
plt.show()
````

````python
# ============================================================
# 1.3 Activaciones y derivadas (con verificacion por diferencias finitas) y gradiente que se desvanece
# ============================================================
activaciones = {
    "Sigmoide":   (expit,                                  lambda z: expit(z) * (1 - expit(z))),
    "Tanh":       (np.tanh,                                lambda z: 1 - np.tanh(z) ** 2),
    "ReLU":       (lambda z: np.maximum(0, z),             lambda z: (z > 0).astype(float)),
    "Leaky ReLU": (lambda z: np.where(z > 0, z, 0.01 * z), lambda z: np.where(z > 0, 1.0, 0.01)),
}
z = np.linspace(-5, 5, 401)
colores = {"Sigmoide": "#2563eb", "Tanh": "#16a34a", "ReLU": "#dc2626", "Leaky ReLU": "#f59e0b"}
fig, axes = plt.subplots(1, 2, figsize=(15, 4.8))
for nombre, (g, dg) in activaciones.items():
    axes[0].plot(z, g(z), lw=2.5, color=colores[nombre], label=nombre)
    axes[1].plot(z, dg(z), lw=2.5, color=colores[nombre], label=nombre)
    # Verificacion: derivada analitica vs. diferencias finitas centradas (evitando el punto z = 0 de ReLU)
    zz = z[np.abs(z) > 0.05]
    numerica = (g(zz + 1e-6) - g(zz - 1e-6)) / 2e-6
    assert np.allclose(dg(zz), numerica, atol=1e-6), nombre
axes[0].set_ylim(-1.5, 3); axes[0].set_xlabel("$z$"); axes[0].set_ylabel("$g(z)$"); axes[0].set_title("Funciones de activación", fontweight="bold"); axes[0].legend()
axes[1].set_ylim(-0.1, 1.2); axes[1].set_xlabel("$z$"); axes[1].set_ylabel("$g'(z)$"); axes[1].set_title("Derivadas: la sigmoide nunca supera 0.25", fontweight="bold"); axes[1].legend()
plt.tight_layout()
plt.show()
print("Derivadas analíticas verificadas con diferencias finitas (4 activaciones).")

capas = np.array([1, 2, 5, 10, 20])
display(pd.DataFrame({"capas L": capas, "factor máximo sigmoide 0.25^L": 0.25 ** capas, "factor máximo tanh 1^L": 1.0 ** capas, "ReLU (z>0) 1^L": 1.0 ** capas}))
print("Con sigmoides, tras 10 capas el gradiente se multiplica por ≤ 1e-6: las primeras capas casi no aprenden (gradiente que se desvanece).")
````

## 2. El Perceptrón Multicapa, Forward Pass y Pérdidas 🏗️

### 2.1 Arquitectura y *forward pass*

Un **perceptrón multicapa** (MLP, *feed-forward network*) organiza neuronas en **capas**: entrada → una o más **capas ocultas** → salida. Cada capa $\ell$ aplica una transformación afín seguida de una activación; con $\mathbf{a}^{(0)} = \mathbf{x}$:

$$\mathbf{z}^{(\ell)} = W^{(\ell)\top}\mathbf{a}^{(\ell-1)} + \mathbf{b}^{(\ell)}, \qquad \mathbf{a}^{(\ell)} = g^{(\ell)}\big(\mathbf{z}^{(\ell)}\big).$$

Calcular la salida a partir de la entrada se llama ***forward pass*** (propagación hacia adelante). Con **una capa oculta de $h$ neuronas** y una salida binaria:

$$\mathbf{z}_1 = W_1^\top\mathbf{x} + \mathbf{b}_1,\quad \mathbf{a}_1 = g(\mathbf{z}_1),\quad z_2 = \mathbf{w}_2^\top\mathbf{a}_1 + b_2,\quad \hat{p} = \sigma(z_2), \qquad W_1\in\mathbb{R}^{d\times h},\; \mathbf{w}_2\in\mathbb{R}^{h}.$$

El número de parámetros es $d\,h + h + h + 1$. La capa oculta aprende una **nueva representación** $\mathbf{a}_1$ de los datos en la que el problema se vuelve (casi) linealmente separable para la capa de salida, que es una regresión logística sobre $\mathbf{a}_1$.



### 2.2 Teorema de aproximación universal

> **Teorema** (Cybenko, 1989, para sigmoides; Hornik, Stinchcombe y White, 1989, para funciones «aplastantes»; Hornik, 1991). Sea $\sigma$ una activación continua, acotada y no constante (p. ej., la sigmoide; Leshno et al., 1993, lo extienden a toda activación continua **no polinomial**, como ReLU). Para toda función continua $f$ sobre un compacto $K \subset \mathbb{R}^d$ y todo $\varepsilon > 0$, existen $h$, $\alpha_j, b_j \in \mathbb{R}$ y $\mathbf{w}_j\in\mathbb{R}^d$ tales que
> $$\sup_{\mathbf{x}\in K}\Big|f(\mathbf{x}) - \sum_{j=1}^{h}\alpha_j\,\sigma\big(\mathbf{w}_j^\top\mathbf{x} + b_j\big)\Big| < \varepsilon.$$

*(Se enuncia sin demostración.)* Cuidado con lo que **no** dice: garantiza **existencia** de una red con **una** capa oculta, pero **no** cuántas neuronas hacen falta (pueden ser exponencialmente muchas), ni que el descenso del gradiente **encuentre** esos pesos, ni que **generalice**. Las redes profundas importan porque suelen representar ciertas funciones con muchos menos parámetros que una capa oculta ancha.

### 2.3 Funciones de pérdida

La pérdida se elige según el tipo de salida (y corresponde a una **máxima verosimilitud**, como en el Módulo 01):

| Problema | Salida | Pérdida | Verosimilitud |
|---|---|---|---|
| Regresión | lineal $\hat y$ | $\frac{1}{n}\sum(y_i-\hat y_i)^2$ (MSE) | Gaussiana |
| Clasificación binaria | $\hat p=\sigma(z)$ | $-\frac1n\sum[y\ln\hat p+(1-y)\ln(1-\hat p)]$ (entropía cruzada binaria) | Bernoulli |
| Clasificación multiclase | $\operatorname{softmax}(\mathbf{z})$ | $-\frac1n\sum_i\sum_k \mathbf{1}[y_i=k]\ln \hat p_{ik}$ (entropía cruzada categórica) | Categórica |

Aquí trabajamos con la **entropía cruzada binaria**, más una penalización L2 opcional $\frac{\lambda}{2}\lVert\mathbf{W}\rVert^2$ sobre los pesos (no sobre los sesgos).

````python
# ============================================================
# 2. XOR resuelto "a mano" con una capa oculta, y aproximacion universal (anchura vs. error)
# ============================================================
# (a) XOR con 2 neuronas ocultas ReLU:  h1 = ReLU(x1 + x2),  h2 = ReLU(x1 + x2 - 1),  salida = h1 - 2*h2
W1_xor, b1_xor = np.array([[1.0, 1.0], [1.0, 1.0]]), np.array([0.0, -1.0])
w2_xor = np.array([1.0, -2.0])
oculta = np.maximum(0, Xx @ W1_xor + b1_xor)
salida = oculta @ w2_xor
print("Entradas XOR :", Xx.astype(int).tolist())
print("Capa oculta  :", oculta.tolist())
print("Salida de la red:", salida.tolist(), "-> XOR =", yx.tolist())
assert np.array_equal(salida, yx)
print("La capa oculta transforma el XOR en un problema lineal: la salida (combinación lineal de la capa oculta) lo resuelve.\n")

# (b) Aproximacion universal: neuronas ocultas tanh con pesos aleatorios fijos + salida por minimos cuadrados
rng = np.random.default_rng(SEED)
f = lambda x: np.sin(2 * np.pi * x) + 0.5 * np.cos(6 * np.pi * x)
xg = np.linspace(0, 1, 500)
anchuras = [1, 2, 3, 5, 10, 20, 50, 100, 200]
errores, ajustes = [], {}
for h in anchuras:
    W = rng.normal(0, 10, size=h); b = rng.uniform(-10, 10, size=h)            # pesos de la capa oculta, aleatorios y FIJOS
    H = np.c_[np.tanh(np.outer(xg, W) + b), np.ones_like(xg)]                   # activaciones ocultas (+ sesgo de salida)
    coef = np.linalg.lstsq(H, f(xg), rcond=None)[0]                             # solo la capa de salida se ajusta
    ajustes[h] = H @ coef
    errores.append(np.max(np.abs(ajustes[h] - f(xg))))

fig, axes = plt.subplots(1, 2, figsize=(15, 4.8))
axes[0].plot(xg, f(xg), "k", lw=2.5, label="Función objetivo $f(x)$")
for h, color in [(3, "#dc2626"), (10, "#f59e0b"), (50, "#16a34a")]:
    axes[0].plot(xg, ajustes[h], lw=1.8, color=color, label=f"{h} neuronas ocultas")
axes[0].set_xlabel("$x$"); axes[0].set_ylabel("$y$"); axes[0].set_title("Más neuronas → mejor aproximación", fontweight="bold"); axes[0].legend(fontsize=8)
axes[1].loglog(anchuras, errores, "o-", color="#2563eb")
axes[1].set_xlabel("Neuronas ocultas $h$ (escala log)"); axes[1].set_ylabel("Error máximo $\\sup_x |f - \\hat f|$ (escala log)")
axes[1].set_title("El error decrece con la anchura de la capa oculta", fontweight="bold")
plt.tight_layout()
plt.show()
print("Error máximo por anchura:", {h: round(float(e), 4) for h, e in zip(anchuras, errores)})
assert errores[-1] < errores[2] and errores[-1] < 0.05
print("(Esto ilustra la EXISTENCIA de buenas aproximaciones; entrenar todos los pesos con descenso del gradiente es otro problema.)")
````

## 3. Retropropagación: la Regla de la Cadena en Acción 🔗

Entrenar la red es minimizar la pérdida $J(\boldsymbol\theta)$ con **descenso del gradiente**, así que necesitamos $\nabla_{\boldsymbol\theta}J$ para **todos** los pesos. La **retropropagación** (*backpropagation*; Rumelhart, Hinton y Williams, 1986) no es un algoritmo de optimización distinto sino **la regla de la cadena aplicada de forma ordenada** y reutilizando cálculos (una forma de *diferenciación automática en modo inverso*).

### 3.1 Derivación para un ejemplo

Con la red de la Sección 2.1 y la pérdida de un ejemplo $L = -y\ln\hat p-(1-y)\ln(1-\hat p)$, avanzamos **de la salida hacia la entrada**. Definimos $\boldsymbol\delta_1 = \partial L/\partial\mathbf{z}_1$.

| Paso | Cantidad | Resultado | Justificación |
|---|---|---|---|
| 1 | $\dfrac{\partial L}{\partial z_2}$ | $\hat p - y$ | Entropía cruzada + sigmoide (derivación del [Módulo 01](../01%20-%20Modelos%20Lineales%20Supervisados/02_Regresion_Logistica.ipynb), Sec. 2.2) |
| 2 | $\dfrac{\partial L}{\partial \mathbf{w}_2},\ \dfrac{\partial L}{\partial b_2}$ | $(\hat p-y)\,\mathbf{a}_1,\ \ \hat p - y$ | $z_2 = \mathbf{w}_2^\top\mathbf{a}_1 + b_2$ |
| 3 | $\dfrac{\partial L}{\partial \mathbf{a}_1}$ | $(\hat p - y)\,\mathbf{w}_2$ | La salida «reparte la culpa» según los pesos |
| 4 | $\boldsymbol\delta_1 = \dfrac{\partial L}{\partial \mathbf{z}_1}$ | $(\hat p - y)\,\mathbf{w}_2\odot g'(\mathbf{z}_1)$ | $\mathbf{a}_1 = g(\mathbf{z}_1)$ actúa componente a componente |
| 5 | $\dfrac{\partial L}{\partial W_1},\ \dfrac{\partial L}{\partial \mathbf{b}_1}$ | $\mathbf{x}\,\boldsymbol\delta_1^\top,\ \ \boldsymbol\delta_1$ | $\mathbf{z}_1 = W_1^\top\mathbf{x} + \mathbf{b}_1$ |

Para una red con más capas el patrón se **repite**: $\boldsymbol\delta^{(\ell)} = \big(W^{(\ell+1)}\boldsymbol\delta^{(\ell+1)}\big)\odot g'(\mathbf{z}^{(\ell)})$ y $\partial L/\partial W^{(\ell)} = \mathbf{a}^{(\ell-1)}\boldsymbol\delta^{(\ell)\top}$.

### 3.2 Forma matricial por lotes

Para un lote de $m$ ejemplos (filas de $X \in \mathbb{R}^{m\times d}$), con $\hat{\mathbf p}, \mathbf{y}\in\mathbb{R}^m$, $A_1 = g(Z_1)\in\mathbb{R}^{m\times h}$ y la pérdida promedio más $\frac\lambda2\lVert\mathbf{W}\rVert^2$:

$$\boxed{\begin{aligned}
\mathbf{d}_2 &= \tfrac1m(\hat{\mathbf p}-\mathbf{y}), & \nabla_{\mathbf{w}_2}J &= A_1^\top\mathbf{d}_2 + \lambda\mathbf{w}_2, & \nabla_{b_2}J &= \mathbf{1}^\top\mathbf{d}_2,\\
D_1 &= (\mathbf{d}_2\mathbf{w}_2^\top)\odot g'(Z_1), & \nabla_{W_1}J &= X^\top D_1 + \lambda W_1, & \nabla_{\mathbf{b}_1}J &= D_1^\top\mathbf{1}.
\end{aligned}}$$

**Costo:** el *backward pass* cuesta del orden del *forward pass* (dos multiplicaciones matriciales por capa), a cambio de **guardar las activaciones** del forward. Si $g=\sigma$ o $\tanh$ aparecen los factores $g'\le 0.25$ o $\le 1$ que explican el desvanecimiento del gradiente.

### 3.3 Implementación desde cero y *gradient check*

Implementamos un MLP de **una capa oculta** con: activaciones `tanh`/`relu`/`sigmoide`, pérdida de entropía cruzada estable, **L2**, inicializaciones (Xavier/Glorot, He, etc.), **descenso del gradiente por mini-lotes** y *early stopping*. Primero verificamos el gradiente con **diferencias finitas centradas**:

$$\frac{\partial J}{\partial\theta_k} \approx \frac{J(\boldsymbol\theta + \epsilon\mathbf{e}_k) - J(\boldsymbol\theta - \epsilon\mathbf{e}_k)}{2\epsilon}.$$

Es **la** herramienta de depuración de la retropropagación (Raschka, Cap. 12): si el error relativo es $\lesssim 10^{-6}$ el gradiente es correcto; si es grande, hay un error en la derivación o el código. Solo se usa para verificar (es $O(\#\text{parámetros})$ evaluaciones de la pérdida: demasiado lento para entrenar).

````python
# ============================================================
# 3.3 MLP de una capa oculta desde cero (forward + backprop + mini-lotes) y gradient check
# ============================================================
ACTIVACIONES = {                                       # (g, g'(z, a)) usando la activacion ya calculada a = g(z) cuando conviene
    "tanh":     (np.tanh,                      lambda z, a: 1 - a ** 2),
    "relu":     (lambda z: np.maximum(0, z),   lambda z, a: (z > 0).astype(float)),
    "sigmoide": (expit,                        lambda z, a: a * (1 - a)),
}

class MLPDesdeCero:
    """MLP binario con UNA capa oculta, salida sigmoide, entropia cruzada + L2 y descenso del gradiente por mini-lotes."""

    def __init__(self, h=8, activacion="tanh", eta=0.5, epocas=200, batch=32, lam=0.0, init="xavier", paciencia=None, semilla=0):
        self.h, self.activacion, self.eta, self.epocas, self.batch = h, activacion, eta, epocas, batch
        self.lam, self.init, self.paciencia, self.semilla = lam, init, paciencia, semilla

    # ---------- parametros ----------
    def _inicializar(self, d):
        rng, h = np.random.default_rng(self.semilla), self.h
        if self.init == "ceros":
            self.W1, self.W2 = np.zeros((d, h)), np.zeros(h)
        elif self.init == "constante":                                  # todas las neuronas identicas -> simetria
            self.W1, self.W2 = np.full((d, h), 0.5), np.full(h, 0.5)
        elif self.init == "pequena":
            self.W1, self.W2 = rng.normal(0, 0.01, (d, h)), rng.normal(0, 0.01, h)
        elif self.init == "grande":
            self.W1, self.W2 = rng.normal(0, 5.0, (d, h)), rng.normal(0, 5.0, h)
        elif self.init == "he":                                         # N(0, 2 / fan_in)
            self.W1, self.W2 = rng.normal(0, np.sqrt(2 / d), (d, h)), rng.normal(0, np.sqrt(2 / h), h)
        else:                                                           # xavier / glorot uniforme: U(+-sqrt(6 / (fan_in + fan_out)))
            l1, l2 = np.sqrt(6 / (d + h)), np.sqrt(6 / (h + 1))
            self.W1, self.W2 = rng.uniform(-l1, l1, (d, h)), rng.uniform(-l2, l2, h)
        self.b1, self.b2 = np.zeros(h), 0.0

    def parametros(self):
        return self.W1, self.b1, self.W2, self.b2

    # ---------- forward ----------
    def _adelante(self, X, W1, b1, W2, b2):
        g, _ = ACTIVACIONES[self.activacion]
        Z1 = X @ W1 + b1                                                # (m, h)
        A1 = g(Z1)
        z2 = A1 @ W2 + b2                                               # (m,)  logits
        return Z1, A1, z2, expit(z2)

    def perdida(self, X, y, W1, b1, W2, b2, con_penalizacion=True):
        z2 = self._adelante(X, W1, b1, W2, b2)[2]
        bce = np.mean(np.maximum(z2, 0) - z2 * y + np.log1p(np.exp(-np.abs(z2))))    # = -[y ln p + (1-y) ln(1-p)], estable
        return bce + (0.5 * self.lam * (np.sum(W1 ** 2) + np.sum(W2 ** 2)) if con_penalizacion else 0.0)

    # ---------- backprop (las formulas de la Seccion 3.2) ----------
    def gradientes(self, X, y, W1, b1, W2, b2):
        _, g_prima = ACTIVACIONES[self.activacion]
        m = len(y)
        Z1, A1, z2, p = self._adelante(X, W1, b1, W2, b2)
        d2 = (p - y) / m                                                # dJ/dz2
        dW2, db2 = A1.T @ d2 + self.lam * W2, d2.sum()
        D1 = np.outer(d2, W2) * g_prima(Z1, A1)                         # dJ/dZ1 (regla de la cadena)
        dW1, db1 = X.T @ D1 + self.lam * W1, D1.sum(axis=0)
        return dW1, db1, dW2, db2

    # ---------- entrenamiento ----------
    def fit(self, X, y, X_val=None, y_val=None):
        X, y = np.asarray(X, dtype=float), np.asarray(y, dtype=float)
        n = len(y)
        self._inicializar(X.shape[1])
        rng = np.random.default_rng(self.semilla + 1)
        self.hist_ = {"train": [], "val": []}                           # perdida de DATOS (sin penalizacion) por epoca
        mejor, sin_mejora, self.mejor_epoca_ = np.inf, 0, None
        for ep in range(self.epocas):
            orden = rng.permutation(n)                                  # mini-lotes: barajar en cada epoca
            for i in range(0, n, self.batch):
                idx = orden[i:i + self.batch]
                gW1, gb1, gW2, gb2 = self.gradientes(X[idx], y[idx], *self.parametros())
                self.W1 -= self.eta * gW1; self.b1 -= self.eta * gb1
                self.W2 -= self.eta * gW2; self.b2 -= self.eta * gb2
            self.hist_["train"].append(self.perdida(X, y, *self.parametros(), con_penalizacion=False))
            if X_val is not None:
                v = self.perdida(X_val, y_val, *self.parametros(), con_penalizacion=False)
                self.hist_["val"].append(v)
                if self.paciencia is not None:                          # early stopping: guardar el mejor modelo en validacion
                    if v < mejor - 1e-9:
                        mejor, sin_mejora, self.mejor_epoca_ = v, 0, ep
                        guardado = (self.W1.copy(), self.b1.copy(), self.W2.copy(), self.b2)
                    else:
                        sin_mejora += 1
                        if sin_mejora >= self.paciencia:
                            self.W1, self.b1, self.W2, self.b2 = guardado
                            break
        return self

    def predict_proba(self, X):
        return self._adelante(np.asarray(X, dtype=float), *self.parametros())[3]

    def predict(self, X, umbral=0.5):
        return (self.predict_proba(X) >= umbral).astype(int)


# --- Datos: make_moons (no lineal), estandarizados con la media/desviacion del entrenamiento ---
X, y = make_moons(n_samples=600, noise=0.25, random_state=SEED)
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, stratify=y, random_state=SEED)
esc = StandardScaler().fit(X_tr)
Z_tr, Z_te = esc.transform(X_tr), esc.transform(X_te)

# --- Gradient check: analitico (backprop) vs. diferencias finitas centradas, para las 3 activaciones ---
filas = []
for act in ["tanh", "relu", "sigmoide"]:
    red = MLPDesdeCero(h=5, activacion=act, lam=0.1)
    rng = np.random.default_rng(1)
    theta = [rng.normal(size=(2, 5)), rng.normal(size=5), rng.normal(size=5), np.array(0.3)]   # punto arbitrario (no el optimo)
    xb, yb = Z_tr[:50], y_tr[:50].astype(float)
    analitico = red.gradientes(xb, yb, *theta)
    peor = 0.0
    for k, P in enumerate(theta):
        numerico = np.zeros(P.shape)
        for idx in np.ndindex(*P.shape):
            def J(delta):
                Q = [q.copy() for q in theta]; Q[k][idx] += delta
                return red.perdida(xb, yb, Q[0], Q[1], Q[2], float(Q[3]))
            numerico[idx] = (J(1e-5) - J(-1e-5)) / 2e-5
        a = np.asarray(analitico[k], dtype=float)
        peor = max(peor, np.max(np.abs(a - numerico) / np.maximum(np.abs(a) + np.abs(numerico), 1e-8)))
    filas.append({"activación": act, "parámetros verificados": sum(np.size(P) for P in theta), "máx. error relativo": peor})
    assert peor < 1e-6, (act, peor)
display(pd.DataFrame(filas).set_index("activación"))
print("Gradient check superado: la retropropagación (Sección 3.2) coincide con las diferencias finitas (error relativo < 1e-6).")
````

## 4. Entrenamiento por Mini-lotes y Frontera No Lineal 🌙

El **descenso del gradiente por mini-lotes** (*mini-batch SGD*) actualiza los pesos con el gradiente de un lote de $m$ ejemplos (en vez de todo el conjunto o de uno solo):

$$\boldsymbol\theta \leftarrow \boldsymbol\theta - \eta\,\nabla_{\boldsymbol\theta}J_{\text{lote}}(\boldsymbol\theta).$$

Una **época** es una pasada completa por los datos (barajados). Lotes pequeños dan gradientes ruidosos pero baratos (el ruido incluso ayuda a escapar de mesetas; los **picos** que verás en la curva de pérdida son ese ruido con una tasa $\eta$ alta); lotes grandes dan gradientes más estables y aprovechan mejor el hardware. La función de pérdida **ya no es convexa**: hay muchos mínimos locales y puntos de silla, y el resultado depende de la **inicialización** (Sección 6).

Entrenamos sobre `make_moons` (dos «lunas» entrelazadas: **no separables linealmente**) y comparamos con la **regresión logística** del Módulo 01, que solo puede trazar una recta. Después verificamos nuestras fórmulas **contra `MLPClassifier`** de `scikit-learn`:

1. Un **paso de descenso del gradiente** de `scikit-learn` desde nuestros pesos debe dar **exactamente** los mismos pesos que nuestro gradiente.
2. Cargando en nuestra red los pesos que `MLPClassifier` aprendió, el *forward pass* y la pérdida deben coincidir.

## 5. La Misma Red en PyTorch 🔥

**PyTorch** es el framework de *deep learning* del curso. Reemplaza nuestra retropropagación manual por **diferenciación automática** (`autograd`): cada operación sobre un tensor con `requires_grad=True` se registra en un **grafo computacional** y `loss.backward()` aplica la regla de la cadena en modo inverso, **acumulando** los gradientes en `.grad` de cada parámetro. Los componentes de un entrenamiento típico son:

| Pieza | En PyTorch | Rol |
|---|---|---|
| **Modelo** | subclase de `nn.Module` con capas `nn.Linear` y método `forward` | Define los parámetros y el *forward pass* |
| **Pérdida** | `nn.BCEWithLogitsLoss` | Sigmoide + entropía cruzada fusionadas (más estable numéricamente) |
| **Optimizador** | `torch.optim.SGD` / `Adam` | Actualiza los parámetros con `.grad` |
| **Bucle** | `optimizador.zero_grad()` → `loss.backward()` → `optimizador.step()` | Un paso de gradiente por lote (¡hay que poner a cero los gradientes!) |

Primero **verificamos** que `autograd` produce **exactamente** los gradientes de nuestra derivación (cargando los mismos pesos en `float64`) y luego entrenamos con un bucle por mini-lotes.

````python
# ============================================================
# 5. La misma red en PyTorch: nn.Module + autograd + bucle de entrenamiento
# ============================================================
import torch
import torch.nn as nn

torch.manual_seed(SEED)

class MLPTorch(nn.Module):
    """Misma arquitectura: d -> h (tanh) -> 1 logit."""
    def __init__(self, d=2, h=8):
        super().__init__()
        self.oculta = nn.Linear(d, h)                  # W1 (transpuesta) y b1
        self.salida = nn.Linear(h, 1)                  # w2 y b2

    def forward(self, x):
        return self.salida(torch.tanh(self.oculta(x))).squeeze(-1)        # logits (la sigmoide va dentro de la pérdida)

# (1) autograd vs. nuestra retropropagacion, con los MISMOS pesos (float64) y la misma pérdida con L2
net = MLPDesdeCero(h=8, activacion="tanh", lam=0.01)
rng = np.random.default_rng(3)
net.W1, net.b1, net.W2, net.b2 = rng.normal(0, 0.5, (2, 8)), rng.normal(0, 0.5, 8), rng.normal(0, 0.5, 8), 0.1
modelo_d = MLPTorch(2, 8).double()
with torch.no_grad():
    modelo_d.oculta.weight.copy_(torch.from_numpy(net.W1.T)); modelo_d.oculta.bias.copy_(torch.from_numpy(net.b1))
    modelo_d.salida.weight.copy_(torch.from_numpy(net.W2).reshape(1, -1)); modelo_d.salida.bias.fill_(net.b2)
xb, yb = torch.from_numpy(Z_tr).double(), torch.from_numpy(y_tr).double()
perdida = nn.BCEWithLogitsLoss()(modelo_d(xb), yb) + 0.5 * net.lam * (modelo_d.oculta.weight.pow(2).sum() + modelo_d.salida.weight.pow(2).sum())
perdida.backward()                                                         # autograd: regla de la cadena en modo inverso
gW1, gb1, gW2, gb2 = net.gradientes(Z_tr, y_tr.astype(float), *net.parametros())
dif = max(np.abs(modelo_d.oculta.weight.grad.numpy().T - gW1).max(), np.abs(modelo_d.oculta.bias.grad.numpy() - gb1).max(),
          np.abs(modelo_d.salida.weight.grad.numpy().ravel() - gW2).max(), abs(modelo_d.salida.bias.grad.item() - gb2))
print(f"Pérdida PyTorch = {perdida.item():.10f} | pérdida propia = {net.perdida(Z_tr, y_tr.astype(float), *net.parametros()):.10f}")
print(f"Diferencia máxima entre los gradientes de autograd y los de nuestra retropropagación: {dif:.2e}")
assert dif < 1e-12 and np.isclose(perdida.item(), net.perdida(Z_tr, y_tr.astype(float), *net.parametros()))

# (2) Bucle de entrenamiento por mini-lotes (float32): SGD y Adam
Xt_tr, yt_tr = torch.tensor(Z_tr, dtype=torch.float32), torch.tensor(y_tr, dtype=torch.float32)
Xt_te, yt_te = torch.tensor(Z_te, dtype=torch.float32), torch.tensor(y_te, dtype=torch.float32)

def entrenar_torch(optimizador, lr, epocas=200, batch=64, semilla=SEED):
    torch.manual_seed(semilla)
    modelo = MLPTorch(2, 8)
    opt = torch.optim.SGD(modelo.parameters(), lr=lr) if optimizador == "SGD" else torch.optim.Adam(modelo.parameters(), lr=lr)
    criterio = nn.BCEWithLogitsLoss()
    hist = []
    for _ in range(epocas):
        modelo.train()
        orden = torch.randperm(len(Xt_tr))
        for i in range(0, len(orden), batch):
            idx = orden[i:i + batch]
            opt.zero_grad()                                                # 1) limpiar gradientes acumulados
            criterio(modelo(Xt_tr[idx]), yt_tr[idx]).backward()            # 2) backward (autograd)
            opt.step()                                                     # 3) actualizar parametros
        with torch.no_grad():                                              # evaluacion sin construir el grafo
            hist.append(criterio(modelo(Xt_tr), yt_tr).item())
    modelo.eval()
    with torch.no_grad():
        acc = ((modelo(Xt_te) > 0).float() == yt_te).float().mean().item()
    return modelo, hist, acc

mod_sgd, h_sgd, acc_sgd = entrenar_torch("SGD", lr=0.5)
mod_adam, h_adam, acc_adam = entrenar_torch("Adam", lr=0.01)
print(f"PyTorch SGD  (lr=0.5)  : pérdida final = {h_sgd[-1]:.4f} | accuracy en prueba = {acc_sgd:.3f}")
print(f"PyTorch Adam (lr=0.01) : pérdida final = {h_adam[-1]:.4f} | accuracy en prueba = {acc_adam:.3f}")
red_t = MLPDesdeCero(h=8, activacion="tanh", eta=0.5, epocas=200, batch=64, semilla=SEED).fit(Z_tr, y_tr)     # misma configuración que SGD en PyTorch
acc_red_t = (red_t.predict(Z_te) == y_te).mean()
print(f"NumPy desde cero (SGD) : pérdida final = {red_t.hist_['train'][-1]:.4f} | accuracy en prueba = {acc_red_t:.3f}")
assert abs(acc_sgd - acc_red_t) <= 0.04 and acc_adam > 0.88

fig, axes = plt.subplots(1, 2, figsize=(15, 4.8))
axes[0].plot(red_t.hist_["train"], color="#2563eb", lw=2, label="NumPy desde cero (SGD)")
axes[0].plot(h_sgd, color="#16a34a", lw=2, linestyle="--", label="PyTorch (SGD)")
axes[0].plot(h_adam, color="#dc2626", lw=2, label="PyTorch (Adam)")
axes[0].set_xlabel("Época"); axes[0].set_ylabel("Entropía cruzada (entrenamiento)"); axes[0].set_title("Misma red, mismo comportamiento", fontweight="bold"); axes[0].legend()
with torch.no_grad():
    pr = torch.sigmoid(mod_adam(torch.tensor(malla, dtype=torch.float32))).numpy().reshape(g1.shape)
axes[1].contourf(g1, g2, pr, levels=np.linspace(0, 1, 11), cmap="RdBu_r", alpha=0.7)
axes[1].contour(g1, g2, pr, levels=[0.5], colors="black", linewidths=2)
axes[1].scatter(Z_tr[:, 0], Z_tr[:, 1], c=y_tr, cmap="RdBu_r", edgecolor="white", s=25)
axes[1].set_xlabel("$x_1$ (estandarizada)"); axes[1].set_ylabel("$x_2$ (estandarizada)"); axes[1].set_title(f"Frontera de la red en PyTorch (Adam) | prueba = {acc_adam:.3f}", fontweight="bold")
plt.tight_layout()
plt.show()
print(f"\nParámetros de la red: {sum(p.numel() for p in mod_adam.parameters())} (= 2·8 + 8 + 8 + 1). Nota: `torch.optim.SGD(weight_decay=λ)` aplica L2 a TODOS los parámetros (también sesgos).")
````

## 6. Inicialización, Tasa de Aprendizaje y Regularización 🎛️

### 6.1 Inicialización
- **Ceros / constantes: rompe la simetría.** Si todos los pesos de la capa oculta parten iguales, las neuronas reciben **el mismo gradiente** y se mantienen idénticas para siempre (equivalen a una sola neurona). Con todo en **cero** y $\tanh$, además las activaciones son 0 y los gradientes de los pesos se anulan: solo aprende el sesgo de salida.
- **Escala.** Pesos muy pequeños → señales y gradientes diminutos (*meseta*); pesos muy grandes → neuronas **saturadas** (en $\tanh$/sigmoide la derivada es ≈ 0). La inicialización **Glorot/Xavier** (uniforme en $\pm\sqrt{6/(n_{\text{ent}}+n_{\text{sal}})}$, Glorot y Bengio, 2010) mantiene estable la varianza de activaciones y gradientes en $\tanh$/sigmoide; **He** (varianza $2/n_{\text{ent}}$, He et al., 2015) está pensada para ReLU. (Géron, Cap. 11.)

### 6.2 Tasa de aprendizaje $\eta$
Es **el hiperparámetro más importante**: muy pequeña → convergencia lentísima; adecuada → baja rápido; demasiado grande → **oscila o diverge**. Optimizadores como **Adam** adaptan la tasa por parámetro (Kingma y Ba, 2015) y suelen ser menos sensibles.

### 6.3 Regularización
Una red con muchas neuronas puede **memorizar** el entrenamiento. Tres defensas clásicas:

| Técnica | Idea | En la práctica |
|---|---|---|
| **L2** (*weight decay*) | Penalizar $\frac\lambda2\lVert\mathbf{W}\rVert^2$: pesos pequeños → función más suave | `alpha` en `MLPClassifier`; `weight_decay` en PyTorch |
| ***Early stopping*** | Vigilar la pérdida en **validación** y detenerse (y restaurar los mejores pesos) cuando deja de mejorar | `early_stopping=True` en sklearn; implementado en nuestra clase con `paciencia` |
| ***Dropout*** | En cada paso de entrenamiento, **apagar al azar** una fracción $q$ de las neuronas (y reescalar por $1/(1-q)$); se desactiva en la evaluación. Actúa como un promedio de muchas subredes (Srivastava et al., 2014) | `nn.Dropout(q)` + `model.train()` / `model.eval()` |

A continuación: (a) el efecto de la **inicialización**, (b) el de la **tasa de aprendizaje**, y luego (c) la regularización sobre un problema pequeño y ruidoso.

````python
# ============================================================
# 6.1-6.2 Inicializacion y tasa de aprendizaje (make_moons, 100 epocas)
# ============================================================
inits = {"ceros": "ceros", "constante (0.5)": "constante", "pequeña (σ=0.01)": "pequena", "Xavier/Glorot": "xavier", "grande (σ=5)": "grande"}
fig, axes = plt.subplots(1, 2, figsize=(16, 5))
filas = []
for (nombre, init) in inits.items():
    r = MLPDesdeCero(h=8, eta=0.5, epocas=100, batch=32, init=init, semilla=SEED).fit(Z_tr, y_tr)
    axes[0].plot(r.hist_["train"], lw=2, label=nombre)
    filas.append({"inicialización": nombre, "pérdida final": r.hist_["train"][-1], "accuracy prueba": (r.predict(Z_te) == y_te).mean()})
    if init == "constante":
        assert np.allclose(r.W1, r.W1[:, [0]]), "las neuronas ocultas deberian seguir identicas"
        sim = "todas las columnas de W1 siguen IDÉNTICAS"
    if init == "ceros":
        assert np.allclose(r.W1, 0) and np.allclose(r.W2, 0)
axes[0].set_xlabel("Época"); axes[0].set_ylabel("Entropía cruzada (entrenamiento)"); axes[0].set_title("Efecto de la inicialización ($\\eta$ = 0.5)", fontweight="bold"); axes[0].legend(fontsize=8)
tabla_init = pd.DataFrame(filas).set_index("inicialización")
display(tabla_init.round(3))
print(f"Con inicialización constante: {sim} tras entrenar (la red equivale a 1 neurona). Con ceros: W1 = W2 = 0 siguen en 0 y solo cambia el sesgo.")
assert tabla_init.loc["Xavier/Glorot", "accuracy prueba"] > tabla_init.loc["constante (0.5)", "accuracy prueba"] + 0.05

etas = [0.001, 0.01, 0.1, 0.5, 2.0, 10.0, 50.0]
filas = []
for eta in etas:
    r = MLPDesdeCero(h=8, eta=eta, epocas=100, batch=32, semilla=SEED).fit(Z_tr, y_tr)
    axes[1].plot(r.hist_["train"], lw=2, label=f"$\\eta$ = {eta}")
    filas.append({"η": eta, "pérdida final": r.hist_["train"][-1], "accuracy prueba": (r.predict(Z_te) == y_te).mean()})
axes[1].set_yscale("log"); axes[1].set_xlabel("Época"); axes[1].set_ylabel("Entropía cruzada (escala log)")
axes[1].set_title("Efecto de la tasa de aprendizaje (Xavier)", fontweight="bold"); axes[1].legend(fontsize=8, ncol=2)
plt.tight_layout()
plt.show()
tabla_eta = pd.DataFrame(filas).set_index("η")
display(tabla_eta.round(3))
assert tabla_eta.loc[0.5, "pérdida final"] < tabla_eta.loc[0.001, "pérdida final"] and tabla_eta.loc[50.0, "pérdida final"] > 3 * tabla_eta.loc[0.5, "pérdida final"]
print("η muy pequeña: avanza muy despacio (100 épocas no bastan). η = 0.5: buena. η = 50: inestable, la pérdida se dispara en lugar de bajar.")
````

````python
# ============================================================
# 6.3 Regularizacion: sobreajuste, L2, early stopping y dropout
# ============================================================
# Problema pequeno y ruidoso: 60 ejemplos de entrenamiento, 45 de validacion, y un conjunto de prueba grande (2000)
Xs, ys = make_moons(n_samples=150, noise=0.35, random_state=7)
Xa, Xr, ya, yr = train_test_split(Xs, ys, test_size=0.6, stratify=ys, random_state=7)
Xv, _, yv, _ = train_test_split(Xr, yr, test_size=0.5, stratify=yr, random_state=7)
Xt_big, yt_big = make_moons(n_samples=2000, noise=0.35, random_state=99)
esc2 = StandardScaler().fit(Xa)
Xa, Xv, Xt_big = esc2.transform(Xa), esc2.transform(Xv), esc2.transform(Xt_big)
print(f"Entrenamiento: {len(ya)} | Validación: {len(yv)} | Prueba: {len(yt_big)}")

config = dict(h=100, eta=0.5, epocas=2000, batch=len(ya), init="he", semilla=SEED)     # red grande, lote completo
sin_reg = MLPDesdeCero(lam=0.0, **config).fit(Xa, ya, Xv, yv)
con_l2 = MLPDesdeCero(lam=0.003, **config).fit(Xa, ya, Xv, yv)
con_es = MLPDesdeCero(lam=0.0, paciencia=50, **config).fit(Xa, ya, Xv, yv)
mejor_ep = int(np.argmin(sin_reg.hist_["val"]))

fig, axes = plt.subplots(1, 4, figsize=(21, 4.8))
axes[0].plot(np.arange(1, len(sin_reg.hist_["train"]) + 1), sin_reg.hist_["train"], color="#2563eb", lw=2, label="Entrenamiento")
axes[0].plot(np.arange(1, len(sin_reg.hist_["val"]) + 1), sin_reg.hist_["val"], color="#dc2626", lw=2, label="Validación")
axes[0].axvline(mejor_ep + 1, color="black", linestyle=":", label=f"Mejor época = {mejor_ep + 1}")
axes[0].set_xscale("log"); axes[0].set_xlabel("Época (escala log)"); axes[0].set_ylabel("Entropía cruzada")
axes[0].set_title("Sin regularizar: la validación empeora", fontweight="bold", fontsize=10); axes[0].legend(fontsize=8)
g1b, g2b = np.meshgrid(np.linspace(-2.8, 2.8, 250), np.linspace(-2.8, 2.8, 250)); mallab = np.c_[g1b.ravel(), g2b.ravel()]
resultados = []
for ax, modelo, titulo in [(axes[1], sin_reg, "Sin regularizar (2000 épocas)"), (axes[2], con_l2, "L2 ($\\lambda$ = 0.003)"), (axes[3], con_es, f"Early stopping (época {con_es.mejor_epoca_ + 1})")]:
    acc_t, acc_p = (modelo.predict(Xa) == ya).mean(), (modelo.predict(Xt_big) == yt_big).mean()
    resultados.append({"modelo": titulo.split(" (")[0], "accuracy entrenamiento": acc_t, "accuracy prueba (2000)": acc_p})
    pr = modelo.predict_proba(mallab).reshape(g1b.shape)
    ax.contourf(g1b, g2b, pr, levels=np.linspace(0, 1, 11), cmap="RdBu_r", alpha=0.7)
    ax.contour(g1b, g2b, pr, levels=[0.5], colors="black", linewidths=2)
    ax.scatter(Xa[:, 0], Xa[:, 1], c=ya, cmap="RdBu_r", edgecolor="white", s=30)
    ax.set_xlabel("$x_1$"); ax.set_ylabel("$x_2$"); ax.set_title(f"{titulo}\nentren. = {acc_t:.2f} | prueba = {acc_p:.3f}", fontweight="bold", fontsize=10)
plt.tight_layout()
plt.show()
display(pd.DataFrame(resultados).set_index("modelo").round(3))
assert sin_reg.hist_["val"][-1] > min(sin_reg.hist_["val"]) + 0.2                      # la validacion empeoro claramente
assert resultados[1]["accuracy prueba (2000)"] > resultados[0]["accuracy prueba (2000)"] + 0.015 and resultados[2]["accuracy prueba (2000)"] > resultados[0]["accuracy prueba (2000)"] + 0.015
print("La red grande memoriza el ruido (frontera retorcida); L2 y early stopping suavizan la frontera y mejoran la generalización.")

# Dropout (solo mencion): apaga neuronas al azar en entrenamiento (reescalando por 1/(1-q)) y no hace nada en evaluacion
capa = nn.Dropout(p=0.5)
x_unos = torch.ones(1, 10)
capa.train(); salida_tr = capa(x_unos)
capa.eval(); salida_ev = capa(x_unos)
print(f"\nDropout q=0.5 en entrenamiento: {salida_tr.numpy().ravel()}  (ceros y valores reescalados a 1/(1-q) = 2)")
print(f"Dropout en evaluación         : {salida_ev.numpy().ravel()}  (identidad)")
assert torch.equal(salida_ev, x_unos) and set(salida_tr.unique().tolist()) <= {0.0, 2.0}
````

## 7. ¿Redes, Árboles o Modelos Lineales en Datos Tabulares? 🥊

No existe un ganador universal (*No Free Lunch*, [Módulo 01](../01%20-%20Modelos%20Lineales%20Supervisados/00_Fundamentos_del_Aprendizaje_Supervisado.ipynb)). Guía práctica:

| Situación | Primera opción | Por qué |
|---|---|---|
| **Pocos datos** (cientos–miles), tabulares | **Modelo lineal** regularizado o **bosque/GBM** | Las redes necesitan más datos y más ajuste fino |
| **Datos tabulares** medianos con variables heterogéneas, categorías, valores faltantes | **Bosques / *gradient boosting*** | Robustos, casi sin preprocesamiento, mejores *por defecto*; en *benchmarks* recientes (p. ej., Grinsztajn et al., 2022; Shwartz-Ziv y Armon, 2022) los modelos de árboles igualan o superan a las redes en tablas típicas *(referencias citadas de memoria: verifícalas)* |
| **Relaciones suaves/aditivas, extrapolación** | **Lineal / GAM / red** | Los árboles no extrapolan (predicen constantes) |
| **Imágenes, texto, audio, series, grafos** | **Redes** (convolucionales, recurrentes, *transformers*) | Aprenden **representaciones** a partir de datos crudos; los árboles no |
| **Muchísimos datos, salidas múltiples, componer con otros módulos diferenciables** | **Redes** | Entrenamiento por gradiente flexible, aprendizaje por transferencia |
| **Interpretabilidad / auditoría** | **Lineal o árbol pequeño** | Reglas o coeficientes legibles |

**Requisitos de las redes:** variables **estandarizadas** (los árboles no lo necesitan), codificación de categóricas, ajuste de $\eta$, arquitectura y regularización, y mayor costo computacional; a cambio, **suavidad**, composición y escalabilidad con los datos. Comparamos los tres tipos de modelo en cinco conjuntos pequeños (validación cruzada de 3 particiones; las redes y la regresión logística reciben las variables estandarizadas).

````python
# ============================================================
# 7. Comparacion lineal / bosque / red en datos tabulares (CV de 3 particiones)
# ============================================================
digits = load_digits()
Xr_, yr_ = make_classification(n_samples=800, n_features=30, n_informative=5, n_redundant=5, n_clusters_per_class=2, flip_y=0.03, random_state=SEED)
Xm_, ym_ = make_moons(n_samples=600, noise=0.3, random_state=SEED)
datasets = {"make_moons (no lineal)": (Xm_, ym_), "Breast Cancer": load_breast_cancer(return_X_y=True), "Wine": load_wine(return_X_y=True),
            "Dígitos 8x8": (digits.data, digits.target), "Sintético (30 var., 20 irrelevantes)": (Xr_, yr_)}
modelos = {"Regresión logística": make_pipeline(StandardScaler(), LogisticRegression(max_iter=3000)),
           "Random Forest": RandomForestClassifier(n_estimators=100, random_state=SEED),
           "Red neuronal (MLP 64)": make_pipeline(StandardScaler(), MLPClassifier(hidden_layer_sizes=(64,), max_iter=300, random_state=SEED))}
cv = StratifiedKFold(3, shuffle=True, random_state=SEED)
tabla = pd.DataFrame({d: {nombre: cross_val_score(m, X_, y_, cv=cv).mean() for nombre, m in modelos.items()} for d, (X_, y_) in datasets.items()})
display(tabla.round(3))

ax = tabla.T.plot(kind="bar", figsize=(14, 4.8), color=["#2563eb", "#16a34a", "#dc2626"], width=0.8)
ax.set_ylabel("Accuracy (validación cruzada)"); ax.set_ylim(0.6, 1.02); ax.set_title("Ningún modelo gana siempre", fontweight="bold")
plt.xticks(rotation=15, ha="right"); plt.legend(loc="lower right"); plt.tight_layout(); plt.show()
ganador = tabla.idxmax()
print("Mejor modelo por conjunto:", ganador.to_dict())
assert tabla.loc["Red neuronal (MLP 64)", "make_moons (no lineal)"] > tabla.loc["Regresión logística", "make_moons (no lineal)"] + 0.02
assert tabla.loc["Regresión logística", "Breast Cancer"] > 0.9
assert tabla.loc["Red neuronal (MLP 64)", "Sintético (30 var., 20 irrelevantes)"] < tabla.loc["Random Forest", "Sintético (30 var., 20 irrelevantes)"]
assert ganador.nunique() >= 2                                                       # ningun modelo gana en todos
brecha = (tabla.max() - tabla.loc["Regresión logística"]).round(3)
print("\nVentaja del mejor modelo sobre la regresión logística:", brecha.to_dict())
print("Lectura: en make_moons (frontera no lineal) la regresión logística queda claramente atrás; en las tablas casi lineales (Breast Cancer, Wine) los tres modelos\n"
      "quedan a 1-2 puntos y el modelo lineal es el más barato; con muchas variables irrelevantes (sintético) la red es la peor (sobreajusta el ruido) y gana el bosque.\n"
      "Ningún modelo es mejor en todo: los valores exactos dependen de la partición y la semilla.")
````

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Neurona** | $a = g(\mathbf{w}^\top\mathbf{x}+b)$; con sigmoide es una regresión logística. **Perceptrón**: converge solo si los datos son separables (cota de Novikoff, verificada); falla en el XOR |
| **Activaciones** | Sigmoide ($\le 0.25$), tanh ($\le 1$), ReLU: sin no linealidad las capas colapsan en una sola; las sigmoides desvanecen el gradiente |
| **MLP** | Capas de «transformación afín + activación»; la capa oculta **aprende una representación** en la que el problema es lineal |
| **Aproximación universal** | Una capa oculta suficientemente ancha aproxima cualquier función continua en un compacto (**existencia**, no aprendibilidad ni generalización) |
| **Retropropagación** | Regla de la cadena de salida a entrada: $\delta^{(\ell)} = (W^{(\ell+1)}\delta^{(\ell+1)})\odot g'(z^{(\ell)})$; costo ≈ el del *forward*. **Verificada con diferencias finitas** (error < $10^{-6}$) |
| **Entrenamiento** | Mini-lotes, no convexo; nuestro gradiente coincide **exactamente** con un paso de `MLPClassifier` y la red resuelve `make_moons` (≈ 93 %) donde el modelo lineal llega a ≈ 84 % |
| **PyTorch** | `nn.Module` + `autograd` + `optim`: los gradientes coinciden con los derivados a mano (≈ $10^{-16}$); Adam es menos sensible a $\eta$ |
| **Inicialización y $\eta$** | Romper la simetría; escala Xavier/He; $\eta$ es el hiperparámetro más crítico (muy pequeña = lenta, muy grande = diverge) |
| **Regularización** | L2, *early stopping* y *dropout* controlan la complejidad y suavizan la frontera |
| **¿Cuándo usar redes?** | Imágenes, texto, audio, secuencias y mucho dato; en **tablas** pequeñas/medianas suelen competir mejor los modelos lineales y los **árboles/GBM** |
