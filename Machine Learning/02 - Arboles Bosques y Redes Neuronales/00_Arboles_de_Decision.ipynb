## 1. Partición Recursiva e Impureza 🌳

### 1.1 La idea

Un **árbol de decisión** representa una función mediante preguntas encadenadas. Cada **nodo interno** hace una pregunta sobre **una sola variable**, $x_j \le t$ (sí → rama izquierda, no → rama derecha); cada **hoja** guarda una predicción. Equivale a **partir el espacio de entrada en $M$ regiones rectangulares** $R_1, \dots, R_M$ y predecir una constante en cada una:

$$\hat{f}(\mathbf{x}) = \sum_{m=1}^{M} c_m \,\mathbf{1}[\mathbf{x} \in R_m].$$

En **clasificación**, $c_m$ es la clase mayoritaria de la región (y las **probabilidades** son las proporciones de clase en ella); en **regresión**, es la **media** de $y$ en la región.



### 1.2 Medidas de impureza

Para decidir **qué pregunta hacer** se mide qué tan **mezclado** está un nodo. Sea $\hat{p}_k$ la proporción de la clase $k$ en el nodo ($K$ clases):

| Medida | Fórmula | Valor mínimo (nodo puro) | Máximo ($K=2$) |
|---|---|---|---|
| **Gini** | $G = 1 - \sum_{k} \hat{p}_k^{\,2}$ | $0$ | $0.5$ en $\hat p = 0.5$ |
| **Entropía** | $H = -\sum_{k} \hat{p}_k \log_2 \hat{p}_k$ (con $0\log 0 = 0$) | $0$ | $1$ bit |
| **Error de clasificación** | $E = 1 - \max_k \hat{p}_k$ | $0$ | $0.5$ |
| **MSE / varianza** (regresión) | $\dfrac{1}{n_t}\sum_{i \in t}(y_i - \bar{y}_t)^2$ | $0$ | sin tope |

El índice de **Gini** es la probabilidad de **equivocarse** al etiquetar un ejemplo del nodo con una clase sorteada según las proporciones del propio nodo. La **entropía** (Shannon) mide la incertidumbre en bits. El **error de clasificación** es poco sensible a mejoras que no cambian la clase mayoritaria, por eso casi no se usa para *crecer* árboles (sí a veces para podar).

### 1.3 Ganancia de información

Al dividir un nodo padre $S$ ($n$ ejemplos) en $S_L$ y $S_R$ ($n_L + n_R = n$), la **ganancia** es la reducción de impureza, con cada hijo ponderado por su tamaño:

$$\text{Ganancia}(S; j, t) = I(S) - \Big[\tfrac{n_L}{n}\,I(S_L) + \tfrac{n_R}{n}\,I(S_R)\Big].$$

Con $I = H$ esta cantidad es la **ganancia de información** (la **información mutua** entre la etiqueta y la respuesta a la pregunta); con $I = G$, la reducción de Gini; con $I$ = varianza, la **reducción de varianza**. Maximizar la ganancia equivale a **minimizar la impureza ponderada de los hijos**.

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas matplotlib seaborn scikit-learn

import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_moons, make_classification, load_breast_cancer, load_iris
from sklearn.model_selection import train_test_split, cross_val_score, validation_curve, StratifiedKFold
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor, export_text, plot_tree

warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (9, 5)
print("Entorno listo.")
````

````python
# ============================================================
# 1. Impureza como funcion de p y un ejemplo numerico de ganancia
# ============================================================
def gini(p):
    p = np.asarray(p, dtype=float)
    return 1.0 - np.sum(p ** 2)

def entropia(p):
    p = np.asarray(p, dtype=float)
    p = p[p > 0]                                   # convencion 0 * log(0) = 0
    return -np.sum(p * np.log2(p))

def error_clasif(p):
    return 1.0 - np.max(p)

ps = np.linspace(0.001, 0.999, 400)
fig, axes = plt.subplots(1, 2, figsize=(15, 4.8))
axes[0].plot(ps, [gini([p, 1 - p]) for p in ps], color="#2563eb", lw=2.5, label="Gini")
axes[0].plot(ps, [entropia([p, 1 - p]) for p in ps], color="#dc2626", lw=2.5, label="Entropía (bits)")
axes[0].plot(ps, [entropia([p, 1 - p]) / 2 for p in ps], color="#dc2626", lw=1.5, linestyle="--", label="Entropía / 2 (misma escala que Gini)")
axes[0].plot(ps, [error_clasif([p, 1 - p]) for p in ps], color="#16a34a", lw=2.5, label="Error de clasificación")
axes[0].set_xlabel("Proporción de la clase 1 en el nodo, $\\hat{p}$"); axes[0].set_ylabel("Impureza")
axes[0].set_title("Medidas de impureza (dos clases)", fontweight="bold"); axes[0].legend(fontsize=8)

# Ejemplo: padre con 10 positivos y 10 negativos; la pregunta produce hijos (8, 2) y (2, 8)
padre, izq, der = [10, 10], [8, 2], [2, 8]
n, nl, nr = 20, 10, 10
filas = []
for nombre, f in [("Gini", gini), ("Entropía", entropia), ("Error de clasif.", error_clasif)]:
    ip, il, ir = f(np.array(padre) / n), f(np.array(izq) / nl), f(np.array(der) / nr)
    filas.append({"medida": nombre, "I(padre)": ip, "I(izq)": il, "I(der)": ir, "ponderada hijos": nl / n * il + nr / n * ir,
                  "ganancia": ip - (nl / n * il + nr / n * ir)})
tabla = pd.DataFrame(filas).set_index("medida")
display(tabla.round(4))

# Una division 'a medias' vs. una division perfecta: la ganancia (Gini) crece al mejorar la pureza
for i, (hi, hd) in enumerate([([5, 5], [5, 5]), ([8, 2], [2, 8]), ([10, 0], [0, 10])]):
    g = gini(np.array(padre) / n) - (0.5 * gini(np.array(hi) / 10) + 0.5 * gini(np.array(hd) / 10))
    axes[1].bar(["Sin información\n(5,5) | (5,5)", "Buena\n(8,2) | (2,8)", "Perfecta\n(10,0) | (0,10)"][i], g, color=["#94a3b8", "#2563eb", "#16a34a"][i])
    axes[1].text(i, g + 0.01, f"{g:.2f}", ha="center", fontweight="bold")
axes[1].set_ylabel("Ganancia de Gini"); axes[1].set_title("Ganancia según la calidad de la pregunta", fontweight="bold")
plt.tight_layout()
plt.show()

assert np.isclose(entropia([0.5, 0.5]), 1.0) and np.isclose(gini([0.5, 0.5]), 0.5)
assert np.isclose(tabla.loc["Gini", "ganancia"], 0.5 - 0.32) and np.isclose(tabla.loc["Entropía", "ganancia"], 1 - 0.7219280948873623)
print("Comprobado: Gini(10,10)=0.5, H=1 bit; con hijos (8,2) y (2,8) la ganancia de Gini es 0.18 y la de entropía 0.278 bits.")
````

## 2. El Algoritmo CART 🔧

**CART** (*Classification And Regression Trees*, Breiman, Friedman, Olshen y Stone, 1984) construye árboles **binarios** de forma **voraz** (*greedy*) y **recursiva**. En un nodo con muestra $S$:

1. Para cada variable $j$ y cada **umbral candidato** $t$, calcular la impureza ponderada de los hijos
   $$J(j, t) = \frac{n_L}{n}\,I(S_L) + \frac{n_R}{n}\,I(S_R), \qquad S_L = \{i: x_{ij} \le t\},\; S_R = \{i: x_{ij} > t\}.$$
2. Elegir el par $(j^{\ast}, t^{\ast})$ que **minimiza** $J$ (equivale a maximizar la ganancia).
3. Repetir recursivamente en $S_L$ y $S_R$ hasta cumplir un **criterio de parada**: nodo puro, `max_depth`, `min_samples_split` (tamaño mínimo para dividir) o `min_samples_leaf` (tamaño mínimo de cada hoja).

**Umbrales candidatos.** Basta probar los **puntos medios** entre valores consecutivos *distintos* de la variable ordenada: a lo sumo $n-1$ por variable, porque cualquier umbral entre dos valores consecutivos produce la misma partición. Ordenando una vez cada variable y acumulando los conteos de clase, se evalúan todos los cortes en $O(n)$ por variable (esto hace que el costo por nodo sea $O(p\,n\log n)$).

**Es voraz.** Cada división se elige mirando solo el nodo actual, sin garantizar el árbol globalmente óptimo (hallar el árbol óptimo es **NP-completo**, Hyafil y Rivest, 1976). A cambio, el algoritmo es rápido y funciona muy bien en la práctica.

**Predicción.** Se baja por el árbol siguiendo las preguntas hasta una hoja; la clase es la mayoritaria y las **probabilidades** son las proporciones de clase de la hoja (en `scikit-learn`, `predict_proba`).

> 📝 **Familia de algoritmos:** ID3 (Quinlan, 1986) y C4.5 (Quinlan, 1993) usan divisiones **multirrama** y la *razón de ganancia*; CART siempre divide en **dos** ramas. `scikit-learn` implementa una versión optimizada de CART.

### Convenciones de nuestra implementación (y cómo se comparan con `scikit-learn`)

Para que nuestro árbol sea **comparable** con `DecisionTreeClassifier` fijamos las mismas reglas:

- Las variables se convierten a `float32` (así lo hace `scikit-learn`) y el umbral es el **punto medio** $a/2 + b/2$ de dos valores consecutivos distintos.
- Se divide **solo si** el nodo no es puro, tiene al menos `min_samples_split` ejemplos y al menos `2*min_samples_leaf`, y no se alcanzó `max_depth`; cada hijo debe tener al menos `min_samples_leaf` ejemplos.
- Predicción de hoja: clase de **mayor conteo**; si hay empate, la de **menor índice**.
- ⚠️ **Desempates entre divisiones igual de buenas:** nuestra implementación elige la **primera** encontrada (variable de menor índice, umbral menor). `scikit-learn` recorre las variables en **orden aleatorio** (controlado por `random_state`), así que en un **empate exacto** puede elegir otra. Esto solo ocurre cuando dos preguntas producen **la misma impureza** (típico en nodos diminutos o con variables redundantes) y, como veremos, **no afecta la calidad** del árbol.

### Verificación contra `DecisionTreeClassifier`

Tras la implementación entrenamos nuestro árbol y el de `scikit-learn` con **los mismos hiperparámetros** sobre `make_moons` (variables continuas: los empates exactos son improbables) y comparamos la **estructura completa** (variable y umbral de cada nodo, en preorden) y las **predicciones y probabilidades** en entrenamiento y prueba. Después estudiamos el árbol **sin límite de profundidad**, donde los nodos diminutos sí producen empates, y el caso de **Iris**, donde dos variables separan perfectamente a una clase.

````python
# ============================================================
# 2. Arbol de clasificacion CART desde cero (recursivo, NumPy)
# ============================================================
def impureza(conteos, criterio):
    """Impureza de un nodo a partir de los conteos por clase (criterio 'gini' o 'entropy')."""
    conteos = np.asarray(conteos, dtype=float)
    p = conteos / conteos.sum()
    if criterio == "gini":
        return 1.0 - np.sum(p ** 2)
    p = p[p > 0]
    return -np.sum(p * np.log2(p))


class ArbolClasificacionDesdeCero:
    """Arbol CART de clasificacion con divisiones binarias axiales (x_j <= t)."""

    def __init__(self, criterio="gini", max_depth=None, min_samples_split=2, min_samples_leaf=1):
        self.criterio, self.max_depth = criterio, max_depth
        self.min_samples_split, self.min_samples_leaf = min_samples_split, min_samples_leaf

    def fit(self, X, y):
        X = np.asarray(X, dtype=np.float32).astype(np.float64)       # igual que scikit-learn: float32
        self.clases_, yi = np.unique(y, return_inverse=True)
        self.K_ = len(self.clases_)
        self.raiz_ = self._crecer(X, yi, profundidad=0)
        return self

    def _impureza_cortes(self, conteos, n_i):
        """Impureza para TODOS los cortes a la vez (conteos: matriz [cortes, K])."""
        p = conteos / n_i[:, None]
        if self.criterio == "gini":
            return 1.0 - np.sum(p ** 2, axis=1)
        return -np.sum(np.where(p > 0, p * np.log2(np.where(p > 0, p, 1.0)), 0.0), axis=1)

    def _mejor_division(self, X, y):
        n, d = X.shape
        Y = np.eye(self.K_)[y]
        mejor = (np.inf, None, None)                                  # (impureza ponderada, variable, umbral)
        for j in range(d):
            orden = np.argsort(X[:, j], kind="stable")
            xs = X[orden, j]
            izq = np.cumsum(Y[orden], axis=0)[:-1]                    # conteos a la izquierda de cada corte
            der = Y.sum(axis=0) - izq
            n_izq = np.arange(1, n)
            valido = (xs[1:] > xs[:-1]) & (n_izq >= self.min_samples_leaf) & (n - n_izq >= self.min_samples_leaf)
            if not valido.any():
                continue
            ponderada = (n_izq * self._impureza_cortes(izq, n_izq) + (n - n_izq) * self._impureza_cortes(der, n - n_izq)) / n
            ponderada = np.where(valido, ponderada, np.inf)
            i = int(np.flatnonzero(ponderada <= ponderada.min() + 1e-12)[0])   # desempate: el primer corte
            if ponderada[i] < mejor[0] - 1e-12:                                 # desempate: la primera variable
                mejor = (ponderada[i], j, xs[i] / 2.0 + xs[i + 1] / 2.0)
        return mejor

    def _crecer(self, X, y, profundidad):
        conteos = np.bincount(y, minlength=self.K_).astype(float)
        nodo = {"conteos": conteos, "n": len(y), "impureza": impureza(conteos, self.criterio)}
        es_hoja = (nodo["impureza"] == 0 or len(y) < self.min_samples_split or len(y) < 2 * self.min_samples_leaf
                   or (self.max_depth is not None and profundidad >= self.max_depth))
        if not es_hoja:
            _, j, umbral = self._mejor_division(X, y)
            if j is not None:
                izq = X[:, j] <= umbral
                nodo.update(variable=j, umbral=umbral,
                            izq=self._crecer(X[izq], y[izq], profundidad + 1),
                            der=self._crecer(X[~izq], y[~izq], profundidad + 1))
        return nodo

    def _hoja(self, x):
        nodo = self.raiz_
        while "variable" in nodo:
            nodo = nodo["izq"] if x[nodo["variable"]] <= nodo["umbral"] else nodo["der"]
        return nodo

    def predict_proba(self, X):
        X = np.asarray(X, dtype=np.float32).astype(np.float64)
        hojas = [self._hoja(x) for x in X]
        return np.array([h["conteos"] / h["n"] for h in hojas])

    def predict(self, X):
        return self.clases_[self.predict_proba(X).argmax(axis=1)]         # empate -> menor indice (argmax)

    def estructura(self):
        """Lista (variable, umbral) en preorden; (-1, nan) para las hojas."""
        sal = []
        def rec(n):
            if "variable" in n:
                sal.append((n["variable"], n["umbral"])); rec(n["izq"]); rec(n["der"])
            else:
                sal.append((-1, np.nan))
        rec(self.raiz_)
        return sal


def estructura_sklearn(clf):
    t = clf.tree_
    return [(int(t.feature[i]), t.threshold[i]) if t.children_left[i] != -1 else (-1, np.nan) for i in range(t.node_count)]


def misma_estructura(a, b):
    return len(a) == len(b) and all(u[0] == v[0] and (np.isnan(u[1]) or np.isclose(u[1], v[1], atol=1e-6)) for u, v in zip(a, b))


# Un ejemplo pequeno a mano: ¿la primera pregunta es la que dice la teoria?
Xe = np.array([[2.0], [3.0], [10.0], [11.0], [12.0]]); ye = np.array([0, 0, 1, 1, 1])
mini = ArbolClasificacionDesdeCero().fit(Xe, ye)
print(f"Mini ejemplo: primera pregunta -> x_{mini.raiz_['variable']} <= {mini.raiz_['umbral']} (punto medio entre 3 y 10 = 6.5)")
assert mini.raiz_["umbral"] == 6.5 and np.array_equal(mini.predict(Xe), ye)
````

## 3. Fronteras de Decisión Ortogonales 🧱

Como cada pregunta es $x_j \le t$, **cada frontera es paralela a un eje** y las regiones son **hiperrectángulos**. Consecuencias:

- Con profundidad pequeña el árbol es un modelo **simple** (pocas regiones, alto sesgo); al crecer, aproxima fronteras curvas con **escalones** cada vez más finos.
- Es **invariante a transformaciones monótonas** de cada variable (estandarizar o aplicar un logaritmo no cambia el árbol), por eso **no necesita escalado**.
- Pero es **sensible a la orientación de los ejes** (Géron, Cap. 5): una frontera **diagonal**, que un modelo lineal captura con una sola recta, exige una **escalera** de muchas preguntas; rotar los datos cambia por completo el árbol.

````python
# ============================================================
# 3. Fronteras ortogonales: efecto de la profundidad y de rotar los datos
# ============================================================
def dibujar_frontera(ax, modelo, X, y, titulo, rango=None):
    x1 = np.linspace(X[:, 0].min() - 0.4, X[:, 0].max() + 0.4, 300)
    x2 = np.linspace(X[:, 1].min() - 0.4, X[:, 1].max() + 0.4, 300)
    g1, g2 = np.meshgrid(x1, x2)
    zz = modelo.predict(np.c_[g1.ravel(), g2.ravel()]).reshape(g1.shape)
    ax.contourf(g1, g2, zz, levels=[-0.5, 0.5, 1.5], colors=["#bfdbfe", "#fecaca"], alpha=0.8)
    ax.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", edgecolor="white", s=28)
    ax.set_xlabel("$x_1$"); ax.set_ylabel("$x_2$"); ax.set_title(titulo, fontweight="bold", fontsize=10)

fig, axes = plt.subplots(2, 4, figsize=(19, 8.4))
for ax, md in zip(axes[0], [1, 2, 4, None]):
    arbol = DecisionTreeClassifier(max_depth=md, random_state=SEED).fit(X_tr, y_tr)
    dibujar_frontera(ax, arbol, X_tr, y_tr, f"max_depth={md}: {arbol.get_n_leaves()} hojas | test = {arbol.score(X_te, y_te):.2f}")

# Datos con frontera alineada a los ejes (x1 > 0) y los mismos datos rotados 45 grados
rng = np.random.default_rng(SEED)
P = rng.uniform(-1, 1, size=(600, 2))
P = P[np.abs(P[:, 0]) > 0.08]                                          # un pequeno margen alrededor de la frontera
yP = (P[:, 0] > 0).astype(int)
th = np.pi / 4
R = np.array([[np.cos(th), -np.sin(th)], [np.sin(th), np.cos(th)]])
P_rot = P @ R.T
res = {}
for nombre, datos, ax_fila in [("alineados", P, axes[1][:2]), ("rotados 45°", P_rot, axes[1][2:])]:
    Ptr, Pte, ytrP, yteP = train_test_split(datos, yP, test_size=0.3, random_state=SEED, stratify=yP)
    for ax, md in zip(ax_fila, [1, 3]):
        a = DecisionTreeClassifier(max_depth=md, random_state=SEED).fit(Ptr, ytrP)
        dibujar_frontera(ax, a, Ptr, ytrP, f"Datos {nombre}, max_depth={md} | test = {a.score(Pte, yteP):.2f}")
        res[(nombre, md)] = a.score(Pte, yteP)
plt.tight_layout()
plt.show()
assert res[("alineados", 1)] > 0.99 and res[("rotados 45°", 1)] < res[("alineados", 1)] - 0.2
print(f"Datos alineados con el eje: un solo corte basta (acc. {res[('alineados', 1)]:.2f}). Rotados 45°: con 1 corte {res[('rotados 45°', 1)]:.2f}, con 3 niveles {res[('rotados 45°', 3)]:.2f} (escalera).")
````

## 4. Sobreajuste y Profundidad 📈

Un árbol **sin restricciones** sigue dividiendo hasta que cada hoja es pura: **error de entrenamiento cero** (salvo ejemplos idénticos con etiquetas distintas), es decir, **sesgo mínimo y varianza máxima**; memoriza el ruido. Es el compromiso sesgo–varianza del [Módulo 01, cuaderno 00](../01%20-%20Modelos%20Lineales%20Supervisados/00_Fundamentos_del_Aprendizaje_Supervisado.ipynb), con la **profundidad** (o el número de hojas) como perilla de complejidad.

Los hiperparámetros de **prevención** (pre-poda) más usados son:

| Hiperparámetro | Efecto |
|---|---|
| `max_depth` | Limita la profundidad del árbol |
| `min_samples_split` / `min_samples_leaf` | Exigen un tamaño mínimo para dividir / en cada hoja (suaviza las predicciones) |
| `max_leaf_nodes` | Limita el número de hojas |
| `min_impurity_decrease` | Exige una ganancia mínima por división |

Los medimos con **curvas de validación**: *accuracy* de entrenamiento y de **validación cruzada** (5 particiones) en función de la complejidad, sobre un problema con **ruido en las etiquetas** (15 %).

````python
# ============================================================
# 4. Curvas de validacion: profundidad y tamano minimo de hoja
# ============================================================
Xc, yc = make_classification(n_samples=800, n_features=10, n_informative=4, n_redundant=2, flip_y=0.15,
                             class_sep=1.0, random_state=SEED)
Xc_tr, Xc_te, yc_tr, yc_te = train_test_split(Xc, yc, test_size=0.3, stratify=yc, random_state=SEED)
cv = StratifiedKFold(5, shuffle=True, random_state=SEED)

profs = np.arange(1, 21)
tr_d, va_d = validation_curve(DecisionTreeClassifier(random_state=SEED), Xc_tr, yc_tr, param_name="max_depth", param_range=profs, cv=cv)
hojas = [1, 2, 5, 10, 20, 40, 80]
tr_h, va_h = validation_curve(DecisionTreeClassifier(random_state=SEED), Xc_tr, yc_tr, param_name="min_samples_leaf", param_range=hojas, cv=cv)

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
for ax, rango, tr, va, xlab, titulo in [(axes[0], profs, tr_d, va_d, "max_depth", "Complejidad = profundidad"),
                                        (axes[1], hojas, tr_h, va_h, "min_samples_leaf", "Complejidad inversa = tamaño mínimo de hoja")]:
    ax.plot(rango, tr.mean(axis=1), "o-", color="#2563eb", label="Entrenamiento")
    ax.plot(rango, va.mean(axis=1), "s-", color="#dc2626", label="Validación cruzada (5 particiones)")
    ax.fill_between(rango, va.mean(axis=1) - va.std(axis=1), va.mean(axis=1) + va.std(axis=1), color="#dc2626", alpha=0.15)
    ax.set_xlabel(xlab); ax.set_ylabel("Accuracy"); ax.set_title(titulo, fontweight="bold"); ax.legend(loc="center right")
axes[1].set_xscale("log")
plt.tight_layout()
plt.show()

mejor_d = profs[int(np.argmax(va_d.mean(axis=1)))]
print(f"Mejor profundidad por validación cruzada: {mejor_d} (CV = {va_d.mean(axis=1).max():.3f}). "
      f"Profundidad 20: entrenamiento = {tr_d.mean(axis=1)[-1]:.3f}, validación = {va_d.mean(axis=1)[-1]:.3f}.")
assert tr_d.mean(axis=1)[-1] > 0.99 and va_d.mean(axis=1)[-1] < va_d.mean(axis=1).max() - 0.03
print("El árbol profundo memoriza el ruido (≈100 % en entrenamiento) y pierde varios puntos en validación: sobreajuste.")
````

## 5. Poda por Complejidad de Costo ✂️

La **pre-poda** (parar antes) es miope: una división mala hoy puede habilitar una excelente mañana. Breiman et al. (1984) proponen **crecer el árbol completo y podarlo hacia atrás**, minimizando el **costo-complejidad**

$$R_\alpha(T) = R(T) + \alpha\,|\tilde{T}|, \qquad R(T) = \sum_{t \in \tilde{T}} \frac{n_t}{N}\, I(t),$$

donde $\tilde{T}$ es el conjunto de **hojas**, $|\tilde T|$ su número y $\alpha \ge 0$ el **precio por hoja**. Para un nodo interno $t$ con subárbol $T_t$, podarlo (convertirlo en hoja) aumenta el error en $R(t) - R(T_t)$ y ahorra $|\tilde{T}_t| - 1$ hojas. Su **alfa efectivo** es el precio al que le da igual podar o no:

$$\alpha_{\text{ef}}(t) = \frac{R(t) - R(T_t)}{|\tilde{T}_t| - 1}.$$

El algoritmo del **enlace más débil** (*weakest link*) poda repetidamente el nodo de menor $\alpha_{\text{ef}}$, produciendo una **sucesión anidada** de árboles $T_0 \supset T_1 \supset \dots \supset \{\text{raíz}\}$ con $\alpha$ creciente. Se elige $\alpha$ por **validación cruzada**. En `scikit-learn` es `ccp_alpha` y `cost_complexity_pruning_path` devuelve la sucesión de alfas.

A continuación (i) **verificamos a mano** el primer alfa efectivo y (ii) elegimos $\alpha$ con validación cruzada.

````python
# ============================================================
# 5. Poda por complejidad de costo: alfa efectivo a mano + eleccion de alfa por CV
# ============================================================
completo = DecisionTreeClassifier(random_state=SEED).fit(Xc_tr, yc_tr)
camino = completo.cost_complexity_pruning_path(Xc_tr, yc_tr)
alfas, imp_total = camino.ccp_alphas, camino.impurities

# (i) Alfa efectivo de cada nodo interno calculado a mano a partir del arbol completo
t = completo.tree_
N = t.weighted_n_node_samples[0]
R_nodo = t.impurity * t.weighted_n_node_samples / N                    # R(t) = I(t) * n_t / N
hoja = t.children_left == -1
def subarbol(i):                                                      # (R(T_t), numero de hojas de T_t)
    if hoja[i]:
        return R_nodo[i], 1
    rl, nl = subarbol(t.children_left[i]); rr, nr = subarbol(t.children_right[i])
    return rl + rr, nl + nr
alfa_ef = [(R_nodo[i] - subarbol(i)[0]) / (subarbol(i)[1] - 1) for i in range(t.node_count) if not hoja[i]]
print(f"Árbol completo: {t.node_count} nodos, {hoja.sum()} hojas. Alfa efectivo mínimo calculado a mano = {min(alfa_ef):.6f}")
print(f"Primeros alfas de scikit-learn (cost_complexity_pruning_path): {np.round(alfas[:4], 6)}")
assert np.isclose(min(alfa_ef), alfas[1]), "el alfa efectivo minimo debe ser el primer alfa positivo del camino"
print("Verificado: el alfa del 'enlace más débil' calculado a mano es el primer alfa positivo de scikit-learn.")

# (ii) Un arbol por alfa (sin el ultimo, que es solo la raiz), evaluados con CV y en prueba
alfas_u = np.unique(np.round(alfas[:-1], 8))
paso = max(1, len(alfas_u) // 40)
alfas_u = alfas_u[::paso]
arboles = [DecisionTreeClassifier(random_state=SEED, ccp_alpha=a).fit(Xc_tr, yc_tr) for a in alfas_u]
cv_a = np.array([cross_val_score(DecisionTreeClassifier(random_state=SEED, ccp_alpha=a), Xc_tr, yc_tr, cv=cv).mean() for a in alfas_u])
hojas_a = np.array([a.get_n_leaves() for a in arboles])
assert np.all(np.diff(hojas_a) <= 0), "mas alfa -> arbol mas pequeno (sucesion anidada)"
mejor_i = int(np.argmax(cv_a))
a_best = alfas_u[mejor_i]

fig, axes = plt.subplots(1, 3, figsize=(18, 4.8))
axes[0].step(alfas, imp_total, where="post", color="#2563eb", lw=2)
axes[0].set_xlabel("$\\alpha$"); axes[0].set_ylabel("Impureza total de las hojas $R(T)$"); axes[0].set_title("Podar más (↑$\\alpha$) aumenta el error de entrenamiento", fontweight="bold", fontsize=10)
axes[1].plot(alfas_u, hojas_a, "o-", color="#16a34a", ms=3); axes[1].set_xlabel("$\\alpha$"); axes[1].set_ylabel("Número de hojas")
axes[1].set_title("Árboles anidados: más $\\alpha$, menos hojas", fontweight="bold", fontsize=10)
axes[2].plot(alfas_u, [a.score(Xc_tr, yc_tr) for a in arboles], label="Entrenamiento", color="#2563eb")
axes[2].plot(alfas_u, cv_a, label="Validación cruzada", color="#dc2626")
axes[2].plot(alfas_u, [a.score(Xc_te, yc_te) for a in arboles], label="Prueba", color="#16a34a", linestyle="--")
axes[2].axvline(a_best, color="black", linestyle=":", label=f"Mejor $\\alpha$ por CV = {a_best:.4f}")
axes[2].set_xlabel("$\\alpha$ (`ccp_alpha`)"); axes[2].set_ylabel("Accuracy"); axes[2].set_title("Elegir $\\alpha$ por validación cruzada", fontweight="bold", fontsize=10); axes[2].legend(fontsize=8)
plt.tight_layout()
plt.show()

podado = arboles[mejor_i]
print(f"Árbol completo: {completo.get_n_leaves()} hojas | prueba = {completo.score(Xc_te, yc_te):.3f}")
print(f"Árbol podado (ccp_alpha = {a_best:.4f}): {podado.get_n_leaves()} hojas | prueba = {podado.score(Xc_te, yc_te):.3f}")
````

## 6. Árboles de Regresión 📉

Para un objetivo numérico, la hoja predice la **media** de $y$ en su región, $c_m = \bar{y}_{R_m}$, y la impureza es el **error cuadrático** (la varianza del nodo). La mejor división minimiza la **suma de cuadrados residual** ponderada:

$$\min_{j,\,t}\;\Big[\sum_{i \in S_L}(y_i - \bar{y}_L)^2 + \sum_{i \in S_R}(y_i - \bar{y}_R)^2\Big] \;\Longleftrightarrow\; \text{máxima reducción de varianza.}$$

El resultado es una **función escalonada** (constante por tramos). Dos propiedades importantes: (i) puede aproximar relaciones **no lineales** arbitrarias, y (ii) **no extrapola**: fuera del rango de entrenamiento predice el valor de la hoja extrema (plano), a diferencia de una recta o un polinomio.

Verificamos a mano la **primera división** y vemos el efecto de la profundidad sobre $y = \sin(2\pi x) + \varepsilon$.

````python
# ============================================================
# 6. Arboles de regresion: primera division a mano, profundidad y extrapolacion
# ============================================================
rng = np.random.default_rng(SEED)
xr = np.sort(rng.uniform(0, 1, 120))
yr = np.sin(2 * np.pi * xr) + rng.normal(0, 0.25, xr.size)

# Primera division a mano: probar todos los puntos medios y quedarse con el de menor SSE
xs32 = xr.astype(np.float32).astype(np.float64)
cand = (xs32[1:] + xs32[:-1]) / 2
sse = np.array([((yr[xs32 <= c] - yr[xs32 <= c].mean()) ** 2).sum() + ((yr[xs32 > c] - yr[xs32 > c].mean()) ** 2).sum() for c in cand])
c_best = cand[sse.argmin()]
stump = DecisionTreeRegressor(max_depth=1, random_state=SEED).fit(xr.reshape(-1, 1), yr)
print(f"Primera división a mano: x <= {c_best:.5f} | scikit-learn: x <= {stump.tree_.threshold[0]:.5f}")
assert np.isclose(c_best, stump.tree_.threshold[0], atol=1e-6)
sse_total = ((yr - yr.mean()) ** 2).sum()
print(f"Reducción de varianza de la primera pregunta: {1 - sse.min() / sse_total:.1%} de la variación total de y.")

xg = np.linspace(-0.3, 1.3, 400).reshape(-1, 1)
fig, axes = plt.subplots(1, 3, figsize=(18, 4.8), sharey=True)
for ax, md in zip(axes, [1, 3, None]):
    arbol_r = DecisionTreeRegressor(max_depth=md, random_state=SEED).fit(xr.reshape(-1, 1), yr)
    ax.scatter(xr, yr, color="#475569", s=22, alpha=0.7, label="Datos")
    ax.plot(xg, np.sin(2 * np.pi * xg), "k--", lw=1.2, label="Función verdadera")
    ax.plot(xg, arbol_r.predict(xg), color="#dc2626", lw=2.5, label="Árbol")
    ax.axvspan(-0.3, 0, color="gray", alpha=0.1); ax.axvspan(1, 1.3, color="gray", alpha=0.1, label="Fuera del rango de entrenamiento")
    ax.set_xlabel("$x$"); ax.set_title(f"max_depth={md}: {arbol_r.get_n_leaves()} hojas", fontweight="bold")
axes[0].set_ylabel("$y$"); axes[0].legend(fontsize=8, loc="lower left")
plt.tight_layout()
plt.show()
print("Profundidad 1: demasiado simple (sesgo). Sin límite: sigue el ruido (varianza). Fuera del rango el árbol es plano: no extrapola.")
````

## 7. Importancia de Variables (MDI) y sus Sesgos ⚖️

La **importancia por disminución media de impureza** (*Mean Decrease in Impurity*, MDI; es `feature_importances_` en `scikit-learn`) suma, sobre todos los nodos que dividen por la variable $j$, la ganancia ponderada por la fracción de ejemplos que llegan al nodo:

$$\text{Imp}_j = \sum_{t:\; v(t) = j} \frac{n_t}{N}\,\Delta I(t), \qquad \Delta I(t) = I(t) - \tfrac{n_L}{n_t}I(t_L) - \tfrac{n_R}{n_t}I(t_R),$$

y se **normaliza** para que sume 1. Es gratis (sale del propio entrenamiento) pero tiene **sesgos conocidos** (Strobl et al., 2007):

1. Se calcula **sobre los datos de entrenamiento**: un árbol profundo que memoriza ruido atribuye importancia a variables **irrelevantes**; no mide utilidad *fuera de muestra*.
2. **Favorece variables con muchos valores distintos** (continuas, identificadores): tienen más umbrales candidatos y, por azar, alguno reduce la impureza.
3. Con variables **correlacionadas**, el crédito se reparte arbitrariamente entre ellas.

La alternativa es la **importancia por permutación** (Breiman, 2001): barajar una variable en datos de *prueba* y medir cuánto **cae el desempeño**. La estudiamos a fondo en el [cuaderno 01](01_Random_Forests.ipynb); aquí la usamos solo para contrastar.

````python
# ============================================================
# 7. MDI a mano (verificada con scikit-learn) y su sesgo con variables irrelevantes
# ============================================================
def mdi_a_mano(arbol):
    t = arbol.tree_
    N = t.weighted_n_node_samples[0]
    imp = np.zeros(arbol.n_features_in_)
    for i in range(t.node_count):
        if t.children_left[i] != -1:                                   # nodo interno
            l, r = t.children_left[i], t.children_right[i]
            ganancia = (t.weighted_n_node_samples[i] * t.impurity[i]
                        - t.weighted_n_node_samples[l] * t.impurity[l] - t.weighted_n_node_samples[r] * t.impurity[r]) / N
            imp[t.feature[i]] += ganancia
    return imp / imp.sum()

# Verificacion en Breast Cancer
bc = load_breast_cancer()
a_bc = DecisionTreeClassifier(max_depth=5, random_state=SEED).fit(bc.data, bc.target)
assert np.allclose(mdi_a_mano(a_bc), a_bc.feature_importances_)
print("MDI calculada a mano == feature_importances_ de scikit-learn (Breast Cancer).")

# Sesgo: 2 variables informativas + 3 irrelevantes (continua, 'identificador' y binaria)
rng = np.random.default_rng(SEED)
n = 600
Xi2, yi2 = make_classification(n_samples=n, n_features=2, n_informative=2, n_redundant=0, n_clusters_per_class=1, class_sep=0.6, flip_y=0.10, random_state=SEED)
ruido_cont = rng.normal(size=n)
ruido_id = rng.permutation(n).astype(float)                            # 0..n-1 barajado: valores todos distintos
ruido_bin = rng.integers(0, 2, n).astype(float)
Xs = np.c_[Xi2, ruido_cont, ruido_id, ruido_bin]
nombres = ["informativa_1", "informativa_2", "ruido_continua", "ruido_id", "ruido_binaria"]
Xs_tr, Xs_te, ys_tr, ys_te = train_test_split(Xs, yi2, test_size=0.4, stratify=yi2, random_state=SEED)

arbol_s = DecisionTreeClassifier(random_state=SEED).fit(Xs_tr, ys_tr)       # sin limite: memoriza
mdi = arbol_s.feature_importances_

def permutacion(modelo, X, y, n_rep=30, seed=SEED):
    rng_p = np.random.default_rng(seed)
    base = modelo.score(X, y)
    caida = np.zeros((n_rep, X.shape[1]))
    for j in range(X.shape[1]):
        for r in range(n_rep):
            Xp = X.copy(); Xp[:, j] = rng_p.permutation(Xp[:, j])
            caida[r, j] = base - modelo.score(Xp, y)
    return caida.mean(axis=0)
perm = permutacion(arbol_s, Xs_te, ys_te)

fig, axes = plt.subplots(1, 2, figsize=(14, 4.6))
axes[0].barh(nombres[::-1], mdi[::-1], color=["#dc2626"] * 3 + ["#16a34a"] * 2)
axes[0].set_xlabel("Importancia MDI (normalizada, calculada en entrenamiento)"); axes[0].set_title("MDI: ¡las variables irrelevantes reciben crédito!", fontweight="bold", fontsize=10)
axes[1].barh(nombres[::-1], perm[::-1], color=["#dc2626"] * 3 + ["#16a34a"] * 2)
axes[1].axvline(0, color="black", lw=0.8)
axes[1].set_xlabel("Caída de accuracy al permutar (datos de prueba)"); axes[1].set_title("Permutación en prueba: el ruido no aporta nada", fontweight="bold", fontsize=10)
plt.tight_layout()
plt.show()

print(f"MDI total de las 3 variables irrelevantes: {mdi[2:].sum():.1%} | de las 2 informativas: {mdi[:2].sum():.1%}")
print(f"Permutación en prueba: informativas = {np.round(perm[:2], 3)}, irrelevantes ≈ {np.round(perm[2:], 3)}")
assert mdi[2:].sum() > 0.2 and perm[2:].max() < perm[:2].min()
````

## 8. Interpretabilidad e Inestabilidad 🔍

**Interpretabilidad.** Un árbol pequeño es un modelo **transparente**: cada predicción se explica con una cadena de reglas «si… entonces…» que un experto puede auditar, y no requiere preprocesamiento (escalado, codificación de relaciones). `export_text` imprime las reglas y `plot_tree` las dibuja; cada nodo muestra la pregunta, la impureza, el número de ejemplos y el conteo por clase. Pero la interpretabilidad **se degrada con la profundidad**: un árbol de 40 hojas ya no se lee.

**Inestabilidad (alta varianza).** Como cada división depende de un *óptimo local sobre pocos datos*, un **cambio pequeño** en la muestra (quitar el 10 % de los ejemplos) puede cambiar la **pregunta de la raíz** y con ello **todo el árbol**, aunque su *accuracy* casi no cambie. Es la misma varianza que vimos en el sobreajuste y es la **motivación** de promediar muchos árboles (**bagging y bosques aleatorios**, cuaderno siguiente).

````python
# ============================================================
# 8. Interpretabilidad (export_text / plot_tree) e inestabilidad (remuestreos del 90 %)
# ============================================================
Xb_tr, Xb_te, yb_tr, yb_te = train_test_split(bc.data, bc.target, test_size=0.3, stratify=bc.target, random_state=SEED)
base = DecisionTreeClassifier(max_depth=3, random_state=SEED).fit(Xb_tr, yb_tr)
print(export_text(base, feature_names=list(bc.feature_names), decimals=3))
print(f"Accuracy en prueba del árbol de profundidad 3: {base.score(Xb_te, yb_te):.3f}  (clases: 0 = maligno, 1 = benigno)\n")

# Inestabilidad: 60 arboles entrenados cada uno con una submuestra aleatoria del 90 % del entrenamiento
rng = np.random.default_rng(SEED)
raices, accs, firmas = [], [], []
for _ in range(60):
    idx = rng.choice(len(yb_tr), size=int(0.9 * len(yb_tr)), replace=False)
    a = DecisionTreeClassifier(max_depth=3, random_state=SEED).fit(Xb_tr[idx], yb_tr[idx])
    raices.append(bc.feature_names[a.tree_.feature[0]])
    accs.append(a.score(Xb_te, yb_te))
    firmas.append(tuple(a.tree_.feature[:3]))                               # variable de la raiz y de sus dos hijos
conteo = pd.Series(raices).value_counts()

fig = plt.figure(figsize=(19, 5.6))
ax0 = fig.add_subplot(1, 3, (1, 2))
plot_tree(base, feature_names=bc.feature_names, class_names=["maligno", "benigno"], filled=True, rounded=True, fontsize=7, ax=ax0)
ax0.set_title("Árbol de profundidad 3 (Breast Cancer)", fontweight="bold")
ax1 = fig.add_subplot(1, 3, 3)
ax1.barh(conteo.index[::-1], conteo.values[::-1], color="#2563eb")
ax1.set_xlabel("Veces que es la raíz (de 60 remuestreos)"); ax1.set_title("¿Cuál es la pregunta de la raíz?", fontweight="bold")
plt.tight_layout()
plt.show()

print(f"Raíces distintas en 60 remuestreos del 90 %: {conteo.shape[0]} | estructuras (raíz + 2 hijos) distintas: {len(set(firmas))}")
print(f"Accuracy en prueba: media = {np.mean(accs):.3f}, mínimo = {np.min(accs):.3f}, máximo = {np.max(accs):.3f}")
print("La estructura del árbol cambia con la muestra aunque el desempeño casi no cambie: un árbol es un modelo de alta varianza.")
````

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Idea** | El árbol **parte el espacio en rectángulos** con preguntas $x_j \le t$ y predice una constante (clase mayoritaria o media) por región |
| **Impureza** | Gini $1-\sum \hat p_k^2$, entropía $-\sum \hat p_k\log_2 \hat p_k$, MSE en regresión; la **ganancia** es la reducción ponderada de impureza |
| **CART** | Búsqueda **voraz** y recursiva de la mejor $(j,t)$ entre los puntos medios; no garantiza el óptimo global |
| **Implementación** | Nuestro árbol coincidió con `DecisionTreeClassifier` (estructura, predicciones y probabilidades); las únicas diferencias son **desempates exactos** entre divisiones igual de buenas |
| **Fronteras** | **Ortogonales**; invariantes a transformaciones monótonas, pero sensibles a la **rotación** de los ejes |
| **Sobreajuste** | Sin límite, el árbol memoriza: error de entrenamiento 0 y varianza alta; se controla con `max_depth`, `min_samples_leaf`… y validación cruzada |
| **Poda** | Costo-complejidad $R(T)+\alpha\lvert\tilde T\rvert$: sucesión anidada por *weakest link* y $\alpha$ elegido por CV |
| **Regresión** | Función **escalonada**; no extrapola |
| **MDI** | Útil y gratuita pero **sesgada** (entrenamiento, variables con muchos valores, correlación); contrastar con permutación |
| **Inestabilidad** | Pequeños cambios en los datos → árboles distintos. **Motivación del bagging y los bosques** |
