## 1. Bagging y Bootstrap 🎒

### 1.1 El bootstrap

Dado un conjunto de entrenamiento de $n$ ejemplos, una **muestra bootstrap** (Efron, 1979) se obtiene **sorteando $n$ ejemplos con reemplazo**. Algunos aparecen varias veces y otros ninguna. La probabilidad de que un ejemplo concreto **no** salga en ninguno de los $n$ sorteos es

$$\Big(1 - \frac{1}{n}\Big)^{n} \;\xrightarrow[n\to\infty]{}\; e^{-1} \approx 0.368,$$

de modo que una muestra bootstrap contiene en promedio una fracción $1 - (1-\tfrac1n)^n \to 1 - e^{-1} \approx \mathbf{63.2\,\%}$ de **ejemplos únicos**; el ≈ 36.8 % restante queda **fuera de la bolsa** (*out-of-bag*, OOB). El número de veces que aparece un ejemplo es aproximadamente $\text{Poisson}(1)$.

### 1.2 Bagging

**Bagging** (*Bootstrap AGGregatING*, Breiman, 1996) entrena $B$ modelos, cada uno con una muestra bootstrap distinta, y **agrega** sus predicciones: promedio en regresión y voto (o promedio de probabilidades) en clasificación,

$$\hat{f}_{\text{bag}}(\mathbf{x}) = \frac{1}{B}\sum_{b=1}^{B}\hat{f}^{*b}(\mathbf{x}).$$

Funciona especialmente bien con modelos de **alta varianza y bajo sesgo** como los árboles profundos. Si se sortea **sin reemplazo** se llama *pasting*.

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas matplotlib seaborn scikit-learn

import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_moons, make_classification, make_friedman1, load_breast_cancer, load_wine
from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.ensemble import (RandomForestClassifier, RandomForestRegressor, BaggingClassifier,
                              GradientBoostingClassifier, HistGradientBoostingClassifier)
from sklearn.inspection import permutation_importance

warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (9, 5)
print("Entorno listo.")
````

````python
# ============================================================
# 1. Verificacion numerica: ~63.2 % de ejemplos unicos en una muestra bootstrap
# ============================================================
rng = np.random.default_rng(SEED)
tamanos = [5, 10, 20, 50, 100, 500, 2000, 10000]
emp, teo = [], []
for n in tamanos:
    unicos = [len(np.unique(rng.integers(0, n, n))) / n for _ in range(300)]       # 300 muestras bootstrap por tamano
    emp.append(np.mean(unicos)); teo.append(1 - (1 - 1 / n) ** n)
tabla = pd.DataFrame({"n": tamanos, "fraccion_unicos_simulada": emp, "teoria_1-(1-1/n)^n": teo, "limite_1-1/e": 1 - np.exp(-1)})
display(tabla.round(4))
assert all(abs(e - t) < 0.015 for e, t in zip(emp, teo)) and abs(emp[-1] - (1 - np.exp(-1))) < 0.005

n = 500
unicos = np.array([len(np.unique(rng.integers(0, n, n))) / n for _ in range(2000)])
apar = np.concatenate([np.bincount(rng.integers(0, n, n), minlength=n) for _ in range(200)])
fig, axes = plt.subplots(1, 3, figsize=(18, 4.6))
axes[0].semilogx(tamanos, emp, "o-", color="#2563eb", label="Simulación (300 muestras)")
axes[0].semilogx(tamanos, teo, "k--", label="$1-(1-1/n)^n$")
axes[0].axhline(1 - np.exp(-1), color="#dc2626", linestyle=":", label="Límite $1-1/e \\approx 0.632$")
axes[0].set_xlabel("Tamaño de la muestra $n$ (escala log)"); axes[0].set_ylabel("Fracción de ejemplos únicos")
axes[0].set_title("Convergencia a 63.2 %", fontweight="bold"); axes[0].legend(fontsize=8)
axes[1].hist(unicos, bins=30, color="#16a34a", edgecolor="white")
axes[1].axvline(1 - np.exp(-1), color="#dc2626", linestyle=":", lw=2)
axes[1].set_xlabel("Fracción de únicos (n = 500)"); axes[1].set_ylabel("Frecuencia (2000 remuestreos)")
axes[1].set_title("Distribución de la fracción de únicos", fontweight="bold")
k = np.arange(0, 7)
axes[2].bar(k - 0.2, [np.mean(apar == i) for i in k], width=0.4, color="#2563eb", label="Simulado")
from math import factorial
axes[2].bar(k + 0.2, [np.exp(-1) / factorial(i) for i in k], width=0.4, color="#f59e0b", label="Poisson(1)")
axes[2].set_xlabel("Veces que un ejemplo aparece en la muestra"); axes[2].set_ylabel("Proporción")
axes[2].set_title("Apariciones ≈ Poisson(1): 36.8 % no aparece", fontweight="bold"); axes[2].legend()
plt.tight_layout()
plt.show()
print(f"Con n = 500: fracción media de únicos = {unicos.mean():.4f} (teoría 0.6322); proporción fuera de la bolsa = {np.mean(apar == 0):.4f} (teoría 0.3679).")
````

## 2. ¿Por qué Promediar Reduce la Varianza? 📉

Sean $B$ estimadores con la misma varianza $\sigma^2$ y **correlación por pares** $\rho$. La varianza de su promedio es

$$\operatorname{Var}\Big(\frac{1}{B}\sum_{b=1}^{B}\hat{f}_b\Big) = \frac{1}{B^2}\Big[B\sigma^2 + B(B-1)\rho\sigma^2\Big] = \boxed{\;\rho\,\sigma^2 + \frac{1-\rho}{B}\,\sigma^2\;}$$

Dos lecciones fundamentales:

- Si los estimadores fueran **independientes** ($\rho = 0$), la varianza caería como $\sigma^2/B$ → 0.
- Con **correlación** $\rho > 0$, al crecer $B$ el segundo término desaparece pero queda el **piso** $\rho\sigma^2$: **agregar más árboles no puede bajar de ahí**. La clave para un buen bosque es **reducir $\rho$**, es decir, que los árboles sean **lo más distintos posible** sin volverse mucho peores individualmente.

Bagging decorrela solo parcialmente: todos los árboles ven datos parecidos y, si hay una variable muy fuerte, **todos la eligen en la raíz** → árboles muy correlacionados. Los bosques aleatorios atacan justo ese problema (Sección 3). Primero verificamos la fórmula con simulación y luego con árboles reales.

````python
# ============================================================
# 2. Varianza de un promedio: formula vs. simulacion, y con arboles reales (Friedman #1)
# ============================================================
rng = np.random.default_rng(SEED)
sigma2, Bs = 1.0, np.array([1, 2, 5, 10, 25, 50, 100, 300])
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
for rho, color in zip([0.0, 0.3, 0.6, 0.9], ["#16a34a", "#2563eb", "#f59e0b", "#dc2626"]):
    cov = sigma2 * (rho * np.ones((300, 300)) + (1 - rho) * np.eye(300))              # varianza 1 y correlacion rho entre todos los pares
    muestras = rng.multivariate_normal(np.zeros(300), cov, size=4000)
    sim = [muestras[:, :b].mean(axis=1).var() for b in Bs]
    teo = rho * sigma2 + (1 - rho) * sigma2 / Bs
    assert np.allclose(sim, teo, rtol=0.12, atol=0.01), (rho, sim, teo)
    axes[0].plot(Bs, sim, "o", color=color); axes[0].plot(Bs, teo, "-", color=color, label=f"$\\rho$ = {rho}")
axes[0].set_xscale("log"); axes[0].set_xlabel("Número de estimadores $B$ (escala log)"); axes[0].set_ylabel("Varianza del promedio")
axes[0].set_title("Fórmula (línea) vs. simulación (puntos): el piso es $\\rho\\sigma^2$", fontweight="bold"); axes[0].legend()

# --- Con arboles reales: varianza de la prediccion a traves de 40 conjuntos de entrenamiento distintos ---
n_rep, n_arb, n_tr = 40, 30, 200
X_test, y_test_ver = make_friedman1(n_samples=300, noise=0.0, random_state=999)                # funcion verdadera sin ruido
pred = {"arbol": [], "bagging": [], "bosque": []}
curva = {"bagging": [], "bosque": []}
for r in range(n_rep):
    Xr, yr = make_friedman1(n_samples=n_tr, noise=1.0, random_state=1000 + r)
    pred["arbol"].append(DecisionTreeRegressor(random_state=r).fit(Xr, yr).predict(X_test))
    for nombre, mf in [("bagging", 1.0), ("bosque", 0.33)]:
        rf = RandomForestRegressor(n_estimators=n_arb, max_features=mf, random_state=r).fit(Xr, yr)
        por_arbol = np.array([a.predict(X_test) for a in rf.estimators_])                     # [arboles, puntos]
        pred[nombre].append(por_arbol.mean(axis=0))
        curva[nombre].append([por_arbol[:b].mean(axis=0) for b in (1, 2, 5, 10, 30)])
var = {k: np.mean(np.var(np.array(v), axis=0)) for k, v in pred.items()}                  # varianza a traves de conjuntos, promediada en puntos
sesgo2 = {k: np.mean((np.mean(np.array(v), axis=0) - y_test_ver) ** 2) for k, v in pred.items()}
display(pd.DataFrame({"varianza": var, "sesgo^2": sesgo2, "ECM total": {k: var[k] + sesgo2[k] for k in var}}).round(3))
bs = (1, 2, 5, 10, 30)
for nombre, color in [("bagging", "#2563eb"), ("bosque", "#16a34a")]:
    v = [np.mean(np.var(np.array(curva[nombre])[:, i, :], axis=0)) for i in range(len(bs))]
    axes[1].plot(bs, v, "o-", color=color, label="Bagging (todas las variables)" if nombre == "bagging" else "Bosque (`max_features`=1/3)")
axes[1].axhline(var["arbol"], color="#dc2626", linestyle="--", label="Un solo árbol")
axes[1].set_xlabel("Número de árboles $B$"); axes[1].set_ylabel("Varianza de la predicción")
axes[1].set_title("Árboles reales: la decorrelación baja el piso de varianza", fontweight="bold"); axes[1].legend()
plt.tight_layout()
plt.show()
assert var["bagging"] < var["arbol"] and var["bosque"] < var["bagging"]
print(f"Varianza: un árbol = {var['arbol']:.2f} → bagging = {var['bagging']:.2f} → bosque = {var['bosque']:.2f}. Promediar baja la varianza; decorrelar la baja aún más.")
print(f"Ojo: el sesgo² sube un poco ({sesgo2['bagging']:.2f} → {sesgo2['bosque']:.2f}) porque cada árbol del bosque es más débil; el ECM total depende de cuál efecto pesa más en el problema.")
````

## 3. Random Forest: Bagging + Submuestreo de Variables 🎲

Un **bosque aleatorio** (Breiman, 2001; con raíces en Ho, 1995, y Amit y Geman, 1997) es *bagging* de árboles con un ingrediente más: **en cada división** el árbol solo considera un **subconjunto aleatorio de $m$ variables** (`max_features`) en lugar de todas las $p$.

**Algoritmo.** Para $b = 1, \dots, B$:
1. Sortear una muestra bootstrap de tamaño $n$.
2. Hacer crecer un árbol **profundo** (sin poda) en esa muestra; **en cada nodo**, elegir $m$ variables al azar y buscar la mejor división **solo entre ellas**.

Predecir por **voto** o promedio de probabilidades de los $B$ árboles.

**Por qué funciona.** Si una variable domina, el *bagging* la pone en todas las raíces → $\rho$ alta. Con $m < p$, esa variable **no está disponible** en una fracción $1 - m/p$ de los nodos, forzando a los árboles a explorar otras estructuras → **$\rho$ baja**, aunque cada árbol sea algo peor (más sesgo individual). Es un **intercambio** que casi siempre conviene. Valores habituales: $m = \sqrt{p}$ en clasificación y $m = p/3$ en regresión (Breiman); `scikit-learn` usa `max_features="sqrt"` por defecto en `RandomForestClassifier` y `1.0` (todas) en `RandomForestRegressor`.

Breiman demostró que el error de generalización está acotado por una función de la **fuerza** $s$ de los árboles individuales y su **correlación media** $\bar\rho$: $\;PE^\ \le \bar\rho\,(1-s^2)/s^2$. Buenos bosques = árboles **fuertes y poco correlacionados**.

### 3.1 Error *out-of-bag* (OOB)

Como cada árbol deja fuera ≈ 36.8 % de los ejemplos, para cada ejemplo $i$ podemos **promediar solo los árboles que no lo vieron** y obtener una predicción **honesta**. El *accuracy* de esas predicciones es el **score OOB**: una estimación de generalización **sin reservar datos de prueba** (equivale en espíritu a una validación cruzada, pero gratis).

### 3.2 Implementación desde cero

Implementamos el bosque con **bootstrap + submuestreo de variables por división + OOB**. **Decisión de diseño:** como árbol base usamos `DecisionTreeClassifier(max_features="sqrt")` de `scikit-learn` (que ya aplica el submuestreo *por división*), y construimos **nosotros** el bootstrap, la agregación y el OOB; así el foco está en la lógica del *ensamble*. (El árbol propio del [cuaderno 00](00_Arboles_de_Decision.ipynb) podría usarse igual, pero sería más lento.)

````python
# ============================================================
# 3.2 Bosque aleatorio desde cero (bootstrap + max_features + OOB) vs. RandomForestClassifier
# ============================================================
class BosqueDesdeCero:
    """Random forest de clasificacion: bootstrap + arboles con submuestreo de variables por division + OOB."""
    def __init__(self, n_estimators=200, max_features="sqrt", max_depth=None, min_samples_leaf=1, random_state=0):
        self.n_estimators, self.max_features = n_estimators, max_features
        self.max_depth, self.min_samples_leaf, self.random_state = max_depth, min_samples_leaf, random_state

    def fit(self, X, y):
        X, y = np.asarray(X), np.asarray(y)
        rng = np.random.default_rng(self.random_state)
        n = len(y)
        self.clases_ = np.unique(y)
        yi = np.searchsorted(self.clases_, y)
        self.arboles_ = []
        suma_oob = np.zeros((n, len(self.clases_)))                       # suma de probabilidades OOB por ejemplo
        cuenta_oob = np.zeros(n)                                          # en cuantos arboles quedo fuera de la bolsa
        for b in range(self.n_estimators):
            idx = rng.integers(0, n, n)                                   # 1) muestra bootstrap (con reemplazo)
            fuera = np.setdiff1d(np.arange(n), idx)                       #    ejemplos fuera de la bolsa de este arbol
            arbol = DecisionTreeClassifier(max_features=self.max_features, max_depth=self.max_depth,
                                           min_samples_leaf=self.min_samples_leaf,
                                           random_state=int(rng.integers(1 << 31)))
            arbol.fit(X[idx], yi[idx])                                    # 2) arbol profundo; sorteo de variables en CADA division
            self.arboles_.append(arbol)
            if len(fuera):
                p = np.zeros((len(fuera), len(self.clases_)))
                p[:, arbol.classes_] = arbol.predict_proba(X[fuera])      # (un bootstrap podria no contener alguna clase)
                suma_oob[fuera] += p; cuenta_oob[fuera] += 1
        self.frac_oob_ = cuenta_oob.mean() / self.n_estimators             # fraccion media de arboles que NO vieron a un ejemplo
        con_oob = cuenta_oob > 0
        self.oob_proba_ = np.full_like(suma_oob, np.nan)
        self.oob_proba_[con_oob] = suma_oob[con_oob] / cuenta_oob[con_oob, None]
        self.oob_score_ = np.mean(self.clases_[self.oob_proba_[con_oob].argmax(axis=1)] == y[con_oob])
        return self

    def predict_proba(self, X):
        X = np.asarray(X)
        P = np.zeros((len(X), len(self.clases_)))
        for a in self.arboles_:
            P[:, a.classes_] += a.predict_proba(X)
        return P / len(self.arboles_)                                     # promedio de probabilidades (como scikit-learn)

    def predict(self, X):
        return self.clases_[self.predict_proba(X).argmax(axis=1)]

bc = load_breast_cancer()
Xb_tr, Xb_te, yb_tr, yb_te = train_test_split(bc.data, bc.target, test_size=0.3, stratify=bc.target, random_state=SEED)

propio = BosqueDesdeCero(n_estimators=300, random_state=SEED).fit(Xb_tr, yb_tr)
sk_rf = RandomForestClassifier(n_estimators=300, max_features="sqrt", oob_score=True, random_state=SEED).fit(Xb_tr, yb_tr)
arbol = DecisionTreeClassifier(random_state=SEED).fit(Xb_tr, yb_tr)

res = pd.DataFrame({"Un árbol": [arbol.score(Xb_tr, yb_tr), np.nan, arbol.score(Xb_te, yb_te)],
                    "Bosque desde cero": [(propio.predict(Xb_tr) == yb_tr).mean(), propio.oob_score_, (propio.predict(Xb_te) == yb_te).mean()],
                    "RandomForestClassifier": [sk_rf.score(Xb_tr, yb_tr), sk_rf.oob_score_, sk_rf.score(Xb_te, yb_te)]},
                   index=["accuracy entrenamiento", "accuracy OOB", "accuracy prueba"])
display(res.round(4))

# El bosque propio coincide con el de scikit-learn en desempeno (la aleatoriedad impide igualdad exacta)
assert abs((propio.predict(Xb_te) == yb_te).mean() - sk_rf.score(Xb_te, yb_te)) <= 0.02
assert abs(propio.oob_score_ - sk_rf.oob_score_) <= 0.02
p_prop, p_sk = propio.predict_proba(Xb_te)[:, 1], sk_rf.predict_proba(Xb_te)[:, 1]
print(f"Correlación entre las probabilidades de ambos bosques en prueba: {np.corrcoef(p_prop, p_sk)[0, 1]:.4f} | "
      f"desacuerdo en clases predichas: {np.mean(propio.predict(Xb_te) != sk_rf.predict(Xb_te)):.1%}")
assert np.corrcoef(p_prop, p_sk)[0, 1] > 0.98
# Cada ejemplo quedo fuera de la bolsa en ~36.8 % de los arboles
assert abs(propio.frac_oob_ - np.exp(-1)) < 0.01
print(f"Fracción media de árboles que dejan fuera a un ejemplo: {propio.frac_oob_:.4f} (teoría e^-1 = {np.exp(-1):.4f}).")
print(f"OOB propio = {propio.oob_score_:.4f} ≈ accuracy de prueba = {(propio.predict(Xb_te) == yb_te).mean():.4f}: el OOB estima la generalización sin usar la partición de prueba.")
````

````python
# ============================================================
# 3.3 Error OOB y de prueba en funcion del numero de arboles (warm_start)
# ============================================================
rf_w = RandomForestClassifier(max_features="sqrt", oob_score=True, warm_start=True, random_state=SEED)
Ns = list(range(10, 201, 10))
oob_err, test_err = [], []
for N in Ns:
    rf_w.set_params(n_estimators=N).fit(Xb_tr, yb_tr)                                   # agrega arboles sin reentrenar los anteriores
    oob_err.append(1 - rf_w.oob_score_); test_err.append(1 - rf_w.score(Xb_te, yb_te))

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(Ns, oob_err, color="#2563eb", lw=2, label="Error OOB")
ax.plot(Ns, test_err, color="#dc2626", lw=2, linestyle="--", label="Error en prueba")
ax.axhline(1 - arbol.score(Xb_te, yb_te), color="gray", linestyle=":", label="Un solo árbol (prueba)")
ax.set_xlabel("Número de árboles"); ax.set_ylabel("Tasa de error")
ax.set_title("El error se estabiliza: más árboles no sobreajustan", fontweight="bold"); ax.legend()
plt.tight_layout()
plt.show()
print(f"Error OOB con {Ns[0]} árboles = {oob_err[0]:.3f}; con {Ns[-1]} = {oob_err[-1]:.3f}. Error de prueba con {Ns[-1]} = {test_err[-1]:.3f}.")
print("Más árboles reducen la varianza hasta el piso ρσ² y NO causan sobreajuste (por eso 'n_estimators' se fija alto y se paga solo en tiempo).")
````

## 4. Importancia de Variables: MDI vs. Permutación 🔍

**MDI** (*Mean Decrease in Impurity*, visto en el [cuaderno 00](00_Arboles_de_Decision.ipynb)) en un bosque es el promedio de la MDI de sus árboles. Hereda sus **sesgos**: se calcula sobre el **entrenamiento**, favorece variables **continuas o de alta cardinalidad** y reparte arbitrariamente el crédito entre variables **correlacionadas**.

La **importancia por permutación** (Breiman, 2001; Kunapuli, Cap. 9) mide la **caída del desempeño** al romper la relación entre una variable $j$ y la respuesta, barajando sus valores en un conjunto de **evaluación**:

$$\text{PI}_j = s(\text{modelo};\, X, y) - \frac{1}{R}\sum_{r=1}^{R} s(\text{modelo};\, X^{(j,r)}, y),$$

con $X^{(j,r)}$ la matriz con la columna $j$ permutada en la repetición $r$. Ventajas: es **agnóstica al modelo**, se evalúa **fuera de muestra** y detecta variables irrelevantes (≈ 0, posiblemente ligeramente negativa por azar). Limitaciones: con variables **correlacionadas** la importancia se *diluye* (el modelo se apoya en la gemela), y cuesta más cómputo.

Lo comprobamos con 3 variables informativas y 3 irrelevantes (una continua, un «identificador» con valores únicos y una binaria), y luego **duplicamos** una informativa para ver el efecto de la correlación.

````python
# ============================================================
# 4. MDI (entrenamiento) vs. permutacion (prueba), y el efecto de duplicar una variable
# ============================================================
rng = np.random.default_rng(SEED)
n = 1000
Xi3, yi3 = make_classification(n_samples=n, n_features=3, n_informative=3, n_redundant=0, n_clusters_per_class=2,
                               class_sep=0.8, flip_y=0.05, random_state=SEED)
ruido = np.c_[rng.normal(size=n), rng.permutation(n).astype(float), rng.integers(0, 2, n).astype(float)]
Xd = np.c_[Xi3, ruido]
nombres = ["informativa_1", "informativa_2", "informativa_3", "ruido_continua", "ruido_id", "ruido_binaria"]
Xd_tr, Xd_te, yd_tr, yd_te = train_test_split(Xd, yi3, test_size=0.4, stratify=yi3, random_state=SEED)

rf = RandomForestClassifier(n_estimators=150, random_state=SEED).fit(Xd_tr, yd_tr)
mdi = rf.feature_importances_
pi_sk = permutation_importance(rf, Xd_te, yd_te, n_repeats=15, random_state=SEED)

# Permutacion implementada a mano (misma definicion, score = accuracy)
def permutacion_a_mano(modelo, X, y, n_rep=15, seed=SEED):
    g = np.random.default_rng(seed)
    base = modelo.score(X, y)
    caidas = np.zeros(X.shape[1])
    for j in range(X.shape[1]):
        for _ in range(n_rep):
            Xp = X.copy(); Xp[:, j] = g.permutation(Xp[:, j])
            caidas[j] += (base - modelo.score(Xp, y)) / n_rep
    return caidas
pi_mano = permutacion_a_mano(rf, Xd_te, yd_te)
assert np.allclose(pi_mano, pi_sk.importances_mean, atol=0.02)
print("Permutación a mano ≈ sklearn.inspection.permutation_importance (diferencia máxima "
      f"{np.max(np.abs(pi_mano - pi_sk.importances_mean)):.4f}).")

fig, axes = plt.subplots(1, 2, figsize=(15, 4.8))
col = ["#16a34a"] * 3 + ["#dc2626"] * 3
axes[0].barh(nombres[::-1], mdi[::-1], color=col[::-1])
axes[0].set_xlabel("MDI (normalizada, calculada con el entrenamiento)"); axes[0].set_title("MDI: el ruido continuo recibe crédito", fontweight="bold")
axes[1].barh(nombres[::-1], pi_sk.importances_mean[::-1], xerr=pi_sk.importances_std[::-1], color=col[::-1])
axes[1].axvline(0, color="black", lw=0.8)
axes[1].set_xlabel("Caída de accuracy al permutar (prueba, 15 repeticiones)"); axes[1].set_title("Permutación en prueba: el ruido ≈ 0", fontweight="bold")
plt.tight_layout()
plt.show()
print(f"MDI de las 3 variables irrelevantes: {mdi[3:].sum():.1%} del total. Permutación: informativas = {np.round(pi_sk.importances_mean[:3], 3)}, "
      f"irrelevantes = {np.round(pi_sk.importances_mean[3:], 3)}")
assert pi_sk.importances_mean[3:].max() < 0.02 < pi_sk.importances_mean[:3].min()

# Efecto de la correlacion: agregar una COPIA de informativa_1 (con una pizca de ruido)
Xdup_tr = np.c_[Xd_tr, Xd_tr[:, 0] + rng.normal(0, 0.01, len(Xd_tr))]
Xdup_te = np.c_[Xd_te, Xd_te[:, 0] + rng.normal(0, 0.01, len(Xd_te))]
rf_d = RandomForestClassifier(n_estimators=150, random_state=SEED).fit(Xdup_tr, yd_tr)
pi_d = permutation_importance(rf_d, Xdup_te, yd_te, n_repeats=15, random_state=SEED).importances_mean
print(f"\ninformativa_1 sola: MDI = {mdi[0]:.3f}, permutación = {pi_sk.importances_mean[0]:.3f}")
print(f"con una copia suya:   MDI = {rf_d.feature_importances_[0]:.3f} + {rf_d.feature_importances_[-1]:.3f} (el crédito se reparte), "
      f"permutación = {pi_d[0]:.3f} y {pi_d[-1]:.3f} (cada una se diluye: el modelo usa la gemela)")
````

## 5. Hiperparámetros Clave 🎛️

| Hiperparámetro | Efecto | Guía práctica |
|---|---|---|
| `n_estimators` ($B$) | Más árboles → menos varianza hasta el piso $\rho\sigma^2$; **no sobreajusta** | Subir hasta que el OOB se estabilice (100–500 suele bastar) |
| `max_features` ($m$) | Menos variables por división → **menos correlación** pero árboles más débiles | `sqrt` (clasificación) o $p/3$; es **el más importante** para ajustar |
| `max_depth` | Limita la complejidad de cada árbol | Árboles **profundos** (sesgo bajo) son lo habitual; limitar si hay mucho ruido |
| `min_samples_leaf` | Hojas más grandes → predicciones más suaves | Subir (5–20) en datos ruidosos o para acelerar |

Usamos el **error OOB** y el de **prueba** para leer las curvas sobre un problema de 30 variables (10 informativas, 10 redundantes —combinaciones lineales de las informativas— y 10 de ruido) con 5 % de etiquetas ruidosas.

````python
# ============================================================
# 5. Curvas de los hiperparametros clave (OOB y prueba)
# ============================================================
Xh, yh = make_classification(n_samples=1000, n_features=30, n_informative=10, n_redundant=10, n_clusters_per_class=2,
                             flip_y=0.05, class_sep=0.8, random_state=SEED)
Xh_tr, Xh_te, yh_tr, yh_te = train_test_split(Xh, yh, test_size=0.3, stratify=yh, random_state=SEED)

def barrido(param, valores, **fijos):
    oob, te = [], []
    for v in valores:
        args = dict(n_estimators=100, oob_score=True, random_state=SEED, n_jobs=1); args.update(fijos); args[param] = v
        m = RandomForestClassifier(**args).fit(Xh_tr, yh_tr)
        oob.append(1 - m.oob_score_); te.append(1 - m.score(Xh_te, yh_te))
    return np.array(oob), np.array(te)

configs = [("n_estimators", [1, 3, 10, 30, 100, 200], "Número de árboles"),
           ("max_depth", [1, 2, 3, 5, 8, 12, 20], "Profundidad máxima"),
           ("max_features", [1, 2, 3, 5, 8, 15, 30], "Variables por división"),
           ("min_samples_leaf", [1, 2, 5, 10, 20, 40], "Tamaño mínimo de hoja")]
fig, axes = plt.subplots(1, 4, figsize=(21, 4.6))
mejores, curvas = {}, {}
for ax, (param, vals, etiqueta) in zip(axes, configs):
    oob, te = barrido(param, vals)
    ax.plot(vals, oob, "o-", color="#2563eb", label="Error OOB"); ax.plot(vals, te, "s--", color="#dc2626", label="Error en prueba")
    ax.set_xlabel(etiqueta); ax.set_ylabel("Tasa de error"); ax.set_title(param, fontweight="bold")
    if param in ("n_estimators", "min_samples_leaf"): ax.set_xscale("log")
    ax.legend(fontsize=8)
    mejores[param] = (vals[int(np.argmin(oob))], oob.min()); curvas[param] = pd.Series(oob, index=vals)
plt.tight_layout()
plt.show()
for p, (v, e) in mejores.items():
    print(f"{p:18s}: mejor valor según OOB = {v} (error OOB = {e:.3f})")
print("\nError OOB según max_features (sqrt(30) ≈ 5.5):", {int(k): round(float(v), 3) for k, v in curvas["max_features"].items()})
assert curvas["n_estimators"].iloc[0] > curvas["n_estimators"].iloc[-1] + 0.05          # un solo arbol es mucho peor que 200
assert 3 <= mejores["max_features"][0] <= 8 and curvas["max_features"].loc[1] > mejores["max_features"][1] and curvas["max_features"].loc[30] > mejores["max_features"][1]
print("Lectura: con 1 árbol el error es alto; baja y se aplana con B (sin sobreajuste). 'max_features' dibuja una U: muy pequeño debilita a cada árbol y muy grande los vuelve a correlacionar; el óptimo queda cerca de √p.")
````

## 6. Árbol vs. Bosque, y un Vistazo al *Boosting* 🏁

**Comparación honesta.** Un bosque **casi siempre** supera a un árbol individual en *accuracy* y estabilidad, a cambio de perder la **interpretabilidad directa** (son cientos de árboles) y de más tiempo de cómputo y memoria; sigue siendo en gran medida «sin ajuste»: funciona bien con los hiperparámetros por defecto, no necesita escalado y maneja interacciones y no linealidades.

**Boosting como contraste.** *Bagging/bosques* entrenan árboles **en paralelo e independientes** y **reducen varianza** (árboles profundos). El ***boosting*** (AdaBoost: Freund y Schapire, 1997; *Gradient Boosting*: Friedman, 2001) entrena árboles **en secuencia**, cada uno **corrigiendo los errores del anterior** (ajustando el gradiente de la pérdida) y reduce principalmente el **sesgo** con árboles **poco profundos**:

$$F_m(\mathbf{x}) = F_{m-1}(\mathbf{x}) + \eta\, h_m(\mathbf{x}), \qquad h_m \approx -\nabla_{F} L\big(y, F_{m-1}(\mathbf{x})\big).$$

Implementaciones modernas (**XGBoost**, **LightGBM**, **CatBoost**) suelen ser el estado del arte en datos tabulares. Se estudian en el curso de [Data Science Programming — Módulo 09 (cuaderno 04: Boosting)](../../Data%20Science%20programming/09%20-%20Decision%20Trees/04_Boosting_y_Casos_Estudio_Desbalanceados.ipynb) y no se repiten aquí; solo los incluimos en la comparación con `GradientBoostingClassifier` e `HistGradientBoostingClassifier` de `scikit-learn`.

| | Árbol | *Bagging* | Random Forest | *Boosting* |
|---|---|---|---|---|
| Entrenamiento | 1 árbol | Paralelo, bootstrap | Paralelo, bootstrap + variables | **Secuencial** |
| Reduce principalmente | — | Varianza | Varianza (decorrelando) | Sesgo (y algo de varianza) |
| Árboles base | Cualquiera | Profundos | Profundos | **Poco profundos** |
| Riesgo de sobreajuste al crecer $B$ | — | No | No | **Sí** (hay que parar) |
| Interpretabilidad | Alta | Baja | Baja | Baja |

````python
# ============================================================
# 6. Comparacion arbol / bagging / bosque / boosting (CV 4 particiones) y fronteras en make_moons
# ============================================================
modelos = {"Árbol": DecisionTreeClassifier(random_state=SEED),
           "Bagging (árboles)": BaggingClassifier(DecisionTreeClassifier(), n_estimators=30, random_state=SEED),
           "Random Forest": RandomForestClassifier(n_estimators=100, random_state=SEED),
           "Gradient Boosting": GradientBoostingClassifier(random_state=SEED),
           "HistGradientBoosting": HistGradientBoostingClassifier(random_state=SEED)}
Xm, ym = make_moons(n_samples=500, noise=0.3, random_state=SEED)
Xw, yw = load_wine(return_X_y=True)
datasets = {"make_moons (ruido 0.3)": (Xm, ym), "Breast Cancer": (bc.data, bc.target), "Wine": (Xw, yw),
            "Sintético (30 var., ruido)": (Xh, yh)}
cv = StratifiedKFold(4, shuffle=True, random_state=SEED)
tabla = pd.DataFrame({d: {nombre: cross_val_score(m, X_, y_, cv=cv).mean() for nombre, m in modelos.items()} for d, (X_, y_) in datasets.items()})
display(tabla.round(3))
assert (tabla.loc["Random Forest"] >= tabla.loc["Árbol"]).all()
print(f"El bosque supera al árbol individual en los {tabla.shape[1]} conjuntos (ganancia media = {(tabla.loc['Random Forest'] - tabla.loc['Árbol']).mean():+.3f} de accuracy).")

Xm_tr, Xm_te, ym_tr, ym_te = train_test_split(Xm, ym, test_size=0.3, stratify=ym, random_state=SEED)
g1, g2 = np.meshgrid(np.linspace(Xm[:, 0].min() - 0.4, Xm[:, 0].max() + 0.4, 250), np.linspace(Xm[:, 1].min() - 0.4, Xm[:, 1].max() + 0.4, 250))
malla = np.c_[g1.ravel(), g2.ravel()]
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, (nombre, m) in zip(axes, [("Árbol", modelos["Árbol"]), ("Random Forest", modelos["Random Forest"]), ("Gradient Boosting", modelos["Gradient Boosting"])]):
    m.fit(Xm_tr, ym_tr)
    ax.contourf(g1, g2, m.predict_proba(malla)[:, 1].reshape(g1.shape), levels=np.linspace(0, 1, 11), cmap="RdBu_r", alpha=0.7)
    ax.scatter(Xm_tr[:, 0], Xm_tr[:, 1], c=ym_tr, cmap="RdBu_r", edgecolor="white", s=25)
    ax.set_xlabel("$x_1$"); ax.set_ylabel("$x_2$"); ax.set_title(f"{nombre} | prueba = {m.score(Xm_te, ym_te):.2f}", fontweight="bold")
plt.tight_layout()
plt.show()
print("El árbol traza escalones irregulares; el bosque promedia muchos escalones y obtiene una frontera mucho más suave (probabilidades graduales).")
````

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Bootstrap** | Muestreo con reemplazo: ≈ **63.2 %** de ejemplos únicos (verificado) y ≈ 36.8 % fuera de la bolsa |
| **Bagging** | Promediar $B$ modelos de alta varianza y bajo sesgo; la varianza del promedio es $\rho\sigma^2 + \frac{1-\rho}{B}\sigma^2$ |
| **Decorrelación** | El piso $\rho\sigma^2$ no baja con más árboles; el **submuestreo de variables** (`max_features`) reduce $\rho$ |
| **Random Forest** | *Bagging* + árboles profundos + $m$ variables al azar **por división**; promedio de probabilidades |
| **OOB** | Cada ejemplo se predice con los árboles que no lo vieron: generalización **sin partición de prueba** |
| **Implementación** | Nuestro bosque igualó a `RandomForestClassifier` en *accuracy*, OOB y probabilidades (hasta la aleatoriedad) |
| **Importancia** | **MDI** es sesgada (entrenamiento, variables continuas); la **permutación** en prueba detecta el ruido, pero se diluye con variables correlacionadas |
| **Hiperparámetros** | `n_estimators` alto (no sobreajusta); `max_features` el más influyente; `min_samples_leaf`/`max_depth` para suavizar |
| **Boosting** | Árboles **secuenciales** que corrigen errores (reduce sesgo); ver Data Science Programming — Módulo 09 |
