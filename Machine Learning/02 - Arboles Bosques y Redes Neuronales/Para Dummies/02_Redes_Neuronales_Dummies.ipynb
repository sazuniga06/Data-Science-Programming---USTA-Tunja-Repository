## 1. La neurona: un jurado que pondera opiniones ⚖️

Piensa en cómo decides **si sales a correr hoy**. Consideras varios factores, y **no todos pesan igual**:

| Factor | ¿Qué tan importante es para ti? |
|---|---|
| ¿Hay sol? | Importante (+3) |
| ¿Tienes ánimo? | Muy importante (+4) |
| ¿Tienes tareas pendientes? | Te frena (−5) |

Sumas **lo que pesa cada factor que sí se cumple**, y si el total **supera tu nivel de exigencia**, sales. Eso es una **neurona artificial**: recibe información, le da una **importancia (peso)** a cada dato, **suma** y **decide** si «se activa».

Los **pesos** son las **perillas** de la neurona: ¡lo que la red aprende son esos pesos!

````python
# Una neurona "a mano": importancia de cada factor (pesos) y nivel de exigencia (umbral)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

factores = ["¿hay sol?", "¿tienes ánimo?", "¿tareas pendientes?"]
pesos = np.array([3.0, 4.0, -5.0])          # las "perillas": cuanto pesa cada factor (negativo = frena)
exigencia = 2.0                              # el total debe superar este valor para salir

escenarios = {"Día perfecto (sol, ánimo, sin tareas)": [1, 1, 0],
              "Con sol pero sin ánimo y con tareas": [1, 0, 1],
              "Sin sol, con ánimo y sin tareas": [0, 1, 0],
              "Con ánimo pero con tareas pendientes": [0, 1, 1]}
filas = []
for nombre, datos in escenarios.items():
    total = np.dot(pesos, datos)                                  # suma ponderada
    filas.append({"escenario": nombre, "total": total, "¿sale a correr?": "sí" if total > exigencia else "no"})
print("Pesos:", {f: float(p) for f, p in zip(factores, pesos)}, "| exigencia:", exigencia, "\n")
print(pd.DataFrame(filas).to_string(index=False))
print("\nUna neurona = suma ponderada + decisión. Aprender es encontrar buenos pesos.")
````

## 2. Capas: un equipo de ayudantes 👥

Una sola neurona decide con **una suma y un umbral**, y eso tiene un límite: solo puede trazar **una línea recta** para separar casos. Hay problemas que **no se pueden resolver con una sola recta**. El clásico es el de la fiesta:

> *«Tu vecino va a la fiesta **solo si va exactamente uno** de sus dos amigos (ni ninguno, ni los dos)».*

(En computación se llama **XOR**, «o lo uno o lo otro, pero no ambos».) Dibuja los cuatro casos en un plano y verás que **ninguna recta separa** los que van de los que no.

**La solución: un equipo en capas.** Se ponen **ayudantes** (neuronas ocultas) entre las entradas y la decisión final. Cada ayudante aprende a detectar **algo sencillo** («¿va el amigo A?», «¿van los dos?»), y la neurona final **combina** esas pistas. La primera capa detecta cosas simples; la siguiente las **combina** en ideas más complejas. Ese es el poder de una red.

---
## 3. Aprender: ajustar las perillas según el error 🎛️

¿Cómo aprende la red sus pesos? Con un ciclo muy parecido a **afinar una guitarra**:

1. **Probar:** la red hace una predicción con las perillas que tiene (al principio, **al azar**).
2. **Medir el error:** compara su respuesta con la correcta.
3. **Repartir la culpa:** averigua **qué perillas contribuyeron más al error** (como en un trabajo en equipo mal calificado: la nota se reparte según cuánto aportó cada integrante). A este paso se le llama **retropropagación**: la culpa viaja **de atrás hacia adelante**.
4. **Ajustar:** mueve cada perilla **un poquito** en la dirección que reduce el error.

Se repite **miles de veces**. Veamos una mini-red de verdad (en menos de 30 líneas de NumPy) aprendiendo el problema de la fiesta (XOR).

````python
# Una mini-red: 2 entradas -> 4 ayudantes -> 1 decision. Aprende el problema de la fiesta (XOR).
def sigmoide(z):
    return 1 / (1 + np.exp(-z))               # convierte cualquier numero en un valor entre 0 y 1

def entrenar_minired(X, y, ayudantes=4, epocas=5000, velocidad=1.0, semilla=0):
    azar = np.random.default_rng(semilla)
    W1, b1 = azar.normal(0, 1, (X.shape[1], ayudantes)), np.zeros(ayudantes)     # perillas de los ayudantes (al azar)
    W2, b2 = azar.normal(0, 1, (ayudantes, 1)), np.zeros(1)                       # perillas de la decision final
    errores = []
    for _ in range(epocas):
        # 1) PROBAR: la informacion viaja hacia adelante
        ayudas = sigmoide(X @ W1 + b1)
        prediccion = sigmoide(ayudas @ W2 + b2)
        # 2) MEDIR EL ERROR
        error = prediccion - y
        errores.append(float(np.mean(error ** 2)))
        # 3) REPARTIR LA CULPA: de atras hacia adelante
        culpa_salida = error
        culpa_ayudantes = (culpa_salida @ W2.T) * ayudas * (1 - ayudas)
        # 4) AJUSTAR cada perilla un poquito
        W2 -= velocidad * ayudas.T @ culpa_salida / len(X);          b2 -= velocidad * culpa_salida.mean(axis=0)
        W1 -= velocidad * X.T @ culpa_ayudantes / len(X);            b1 -= velocidad * culpa_ayudantes.mean(axis=0)
    return (W1, b1, W2, b2), errores

def predecir(modelo, X):
    W1, b1, W2, b2 = modelo
    return sigmoide(sigmoide(X @ W1 + b1) @ W2 + b2)

# Los 4 casos de la fiesta: (amigo A, amigo B) -> va el vecino? (1 = si, solo si va exactamente uno)
casos = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=float)
va = np.array([[0], [1], [1], [0]], dtype=float)

modelo, errores = entrenar_minired(casos, va)
tabla = pd.DataFrame({"amigo A": casos[:, 0].astype(int), "amigo B": casos[:, 1].astype(int), "debería ir": va.ravel().astype(int),
                      "probabilidad que da la red": predecir(modelo, casos).ravel().round(3)})
print(tabla.to_string(index=False))

plt.figure(figsize=(8.5, 4.5))
plt.plot(errores, color="#f59e0b", lw=3)
plt.yscale("log")
plt.xlabel("Número de ajustes (épocas)"); plt.ylabel("Error (escala logarítmica)")
plt.title("La red va afinando sus perillas: el error baja", fontweight="bold")
plt.show()
print(f"\nError al inicio: {errores[0]:.3f}  ->  error al final: {errores[-1]:.5f}. ¡Aprendió lo que una sola recta no puede!")
````

## 4. Curvas en lugar de rectas 🌙

Los modelos que vimos antes en el curso (como la **regresión logística**) separan los casos con **una línea recta**. Pero muchos problemas reales tienen **fronteras curvas**. Probemos con dos «lunas» entrelazadas, un problema clásico: ninguna recta puede separar bien las dos lunas, pero **una red con ayudantes sí puede dibujar una frontera curva**.

Usamos **la misma mini-red** de arriba, sin cambiarle nada.

````python
# Dos lunas entrelazadas: la recta (regresion logistica) vs. la mini-red del cuaderno
from sklearn.datasets import make_moons
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

lunas, luna = make_moons(n_samples=400, noise=0.15, random_state=42)
L_ent, L_exa, l_ent, l_exa = train_test_split(lunas, luna, test_size=0.3, random_state=42, stratify=luna)
escala = StandardScaler().fit(L_ent)
L_ent, L_exa = escala.transform(L_ent), escala.transform(L_exa)

recta = LogisticRegression().fit(L_ent, l_ent)
red, _ = entrenar_minired(L_ent, l_ent.reshape(-1, 1).astype(float), ayudantes=8, epocas=4000, velocidad=1.0)
aciertos_red = np.mean((predecir(red, L_exa).ravel() > 0.5) == l_exa)

xx, yy = np.meshgrid(np.linspace(-2.8, 2.8, 250), np.linspace(-2.8, 2.8, 250))
malla = np.c_[xx.ravel(), yy.ravel()]
fig, ejes = plt.subplots(1, 2, figsize=(13, 5))
for eje, probas, titulo in [(ejes[0], recta.predict_proba(malla)[:, 1], f"Una recta (regresión logística)\naciertos en el examen: {recta.score(L_exa, l_exa):.0%}"),
                            (ejes[1], predecir(red, malla).ravel(), f"Mini-red con 8 ayudantes\naciertos en el examen: {aciertos_red:.0%}")]:
    eje.contourf(xx, yy, probas.reshape(xx.shape), levels=[0, 0.5, 1], colors=["#bfdbfe", "#fde68a"], alpha=0.8)
    eje.scatter(L_ent[:, 0], L_ent[:, 1], c=l_ent, cmap="coolwarm", edgecolor="white", s=30)
    eje.set_title(titulo, fontweight="bold"); eje.set_xlabel("medida 1"); eje.set_ylabel("medida 2")
plt.tight_layout()
plt.show()
print(f"Aciertos en el examen -> recta: {recta.score(L_exa, l_exa):.0%} | mini-red: {aciertos_red:.0%}")
print("La recta corta las lunas por la mitad y se equivoca más; la red dibuja una frontera curva que las separa mejor.")
````

## Resumen en una frase 🎯

> **Una red neuronal es un equipo de neuronas («jurados» que suman opiniones ponderadas) organizado en capas, que aprende ajustando sus perillas (pesos) un poquito cada vez para reducir el error, y por eso puede dibujar fronteras curvas que una sola recta no logra.**

### Lo que aprendiste hoy:
- ✅ Una **neurona** suma datos con **pesos** (importancia) y decide si «se activa».
- ✅ Con **ayudantes en capas** se resuelven problemas que ninguna recta puede (como el de la fiesta).
- ✅ **Aprender = probar, medir el error, repartir la culpa y ajustar las perillas**, miles de veces.
- ✅ Las redes dibujan **curvas**; pero necesitan datos, paciencia para ajustarlas y cuidado para no **memorizar**.
