## 2. Caso A: Fuga de Clientes (Clasificación Desbalanceada con Faltantes) 📉

**Problema de negocio.** Una empresa de telecomunicaciones quiere **anticipar qué clientes se irán** para ofrecerles una retención. Cada oferta **cuesta** y la fuga **también**. Planteamos:

- **Decisión:** contactar o no a cada cliente.
- **Costos (supuesto simplificado):** una fuga no detectada pierde **500**; contactar a alguien cuesta **50**, se fuga o no (suponemos que la oferta retiene al que se iba). Entonces **contactar conviene si** $p\cdot 500>50$ aun contando que si se contacta ya no se pierde: el costo esperado de contactar es $50$ y el de no contactar es $p\cdot500$, por lo que **se contacta si $p>50/500=0.10$** (el resultado del [Módulo 03, §7](../03%20-%20Metricas%20y%20Ajuste%20de%20Hiperparametros/00_Metricas_de_Clasificacion.ipynb)): **la métrica de negocio fija el umbral**, no el 0.5 por defecto.
- **Datos (sintéticos):** 8 000 clientes con variables numéricas y categóricas, **≈ 8 % de fugas** (desbalance 1:12), **valores faltantes** (ingreso: más ausente en jóvenes y por teléfono, **MAR**, con algo de **MNAR**; satisfacción: 45 % no responde) y **dos variables irrelevantes** de ruido. Como la verdad es una probabilidad conocida, podemos calcular un **techo realista** (lo mejor que se puede hacer conociendo la probabilidad verdadera) que **en la vida real no existe**.

**Partición.** Se separa una **prueba sellada estratificada** (25 %) **antes de mirar nada** ([Módulo 07, §1](../07%20-%20Seleccion%20de%20Modelos%20y%20Optimizacion/00_Evaluacion_y_Seleccion_de_Modelos.ipynb)). Todo lo demás (comparar, ajustar, calibrar, elegir el umbral) usa **solo el entrenamiento**.

````python
# ============================================================
# CASO A.1  Datos sinteticos de clientes: desbalance (~8 % de fuga) + valores faltantes (MAR/MNAR) + variables irrelevantes
# ============================================================
def generar_clientes(n=8000, semilla=7, prevalencia=0.08):
    rng = np.random.default_rng(semilla)
    plan = rng.choice(["basico", "estandar", "premium"], n, p=[0.5, 0.35, 0.15])
    canal = rng.choice(["web", "tienda", "telefono"], n, p=[0.45, 0.35, 0.2])
    region = rng.choice(["andina", "caribe", "pacifica", "orinoquia", "amazonia"], n, p=[0.4, 0.25, 0.2, 0.1, 0.05])
    edad = np.clip(rng.normal(42, 12, n), 18, 80)
    antig = np.clip(rng.gamma(2.0, 18.0, n), 1, 120)                                    # meses de antiguedad
    cargo = {"basico": 30, "estandar": 55, "premium": 90}
    cargo_m = np.array([cargo[p] for p in plan]) * rng.lognormal(0, 0.12, n)
    ingreso = np.exp(rng.normal(1.0 + 0.012 * (edad - 40) + np.array([{"basico": 0, "estandar": 0.2, "premium": 0.5}[p] for p in plan]), 0.4)) * 1.0   # millones de COP al mes
    uso = rng.lognormal(3.0 - 0.004 * (edad - 40), 0.6, n)                              # GB al mes
    reclamos = rng.poisson(np.where(plan == "basico", 1.0, 0.4) * np.where(antig < 12, 1.4, 0.8))
    atrasos = rng.poisson(0.25 + 0.5 * (ingreso < 2.0))
    satisf = np.clip(np.round(8 - 1.1 * reclamos - 0.6 * atrasos + rng.normal(0, 1.5, n)), 1, 10)
    ruido1, ruido2 = rng.normal(size=n), rng.integers(1, 29, n).astype(float)           # dos variables IRRELEVANTES (sucursal, dia de registro)
    # --- Verdad: logit con no linealidades e interacciones ---
    z = (-0.045 * antig + 0.9 * np.log1p(reclamos) * 1.0 + 0.55 * atrasos + 0.5 * (plan == "basico") - 0.3 * (plan == "premium")
         + 0.7 * np.maximum(0, cargo_m / (1 + 10 * ingreso) - 4) + 0.5 * np.log1p(reclamos) * atrasos - 0.5 * (uso > np.median(uso)) - 0.18 * (satisf - 6)
         + 0.4 * (canal == "telefono"))
    lo, hi = -15.0, 5.0
    for _ in range(60):                                                                   # biseccion: intercepto para obtener la prevalencia pedida
        b0 = (lo + hi) / 2; p = 1 / (1 + np.exp(-(b0 + z)))
        lo, hi = (b0, hi) if p.mean() < prevalencia else (lo, b0)
    y = (rng.random(n) < p).astype(int)
    df = pd.DataFrame({"edad": edad, "antiguedad_meses": antig, "plan": plan, "canal": canal, "region": region, "cargo_mensual": cargo_m, "ingreso_mm": ingreso, "uso_gb": uso,
                       "reclamos_6m": reclamos.astype(float), "cuotas_atrasadas": atrasos.astype(float), "satisfaccion": satisf, "ruido_sucursal": ruido1, "ruido_dia": ruido2})
    # --- Faltantes: ingreso (MAR: mas ausente en jovenes y por telefono; algo MNAR: ingresos altos se ocultan) y satisfaccion (encuesta: 45 % no responde, MCAR) ---
    p_falta_ing = np.clip(0.10 + 0.25 * (edad < 30) + 0.20 * (canal == "telefono") + 0.10 * (ingreso > np.quantile(ingreso, 0.8)), 0, 0.8)
    df.loc[rng.random(n) < p_falta_ing, "ingreso_mm"] = np.nan
    df.loc[rng.random(n) < 0.45, "satisfaccion"] = np.nan
    df.loc[rng.random(n) < 0.03, "uso_gb"] = np.nan
    return df, y, p


X_all, y_all, p_verdadera = generar_clientes()
print(f"Clientes: {len(X_all):,} | variables: {X_all.shape[1]} | prevalencia de fuga: {y_all.mean():.1%} (desbalance ≈ 1:{(1 - y_all.mean()) / y_all.mean():.0f})")
print("Faltantes por variable (%):", (100 * X_all.isna().mean()).round(1)[lambda s: s > 0].to_dict())
print("Datos SINTÉTICOS: la fuga sale de un modelo logístico con no linealidades e interacciones (conocemos la probabilidad verdadera de cada cliente), más 2 variables irrelevantes.")

# --- Planteamiento de negocio: costos y UMBRAL optimo teorico ---
C_FN, C_CONTACTO = 500.0, 50.0       # fuga NO detectada: se pierde el valor del cliente (500). Contactar (oferta de retencion) cuesta 50, sea fuga o no
# Politica: contactar a un cliente si el costo esperado de NO hacerlo (p * C_FN) supera el de hacerlo (C_CONTACTO)  ->  p > C_CONTACTO / C_FN  (suponiendo que la oferta retiene al que se iba)
UMBRAL_TEORICO = C_CONTACTO / C_FN
def costo_total(y, contactar):
    """Costo del negocio: fugas no contactadas x 500 + contactos x 50."""
    y = np.asarray(y); c = np.asarray(contactar).astype(bool)
    return float(((y == 1) & ~c).sum() * C_FN + c.sum() * C_CONTACTO)
print(f"Costos: fuga no detectada = {C_FN:.0f} | contacto = {C_CONTACTO:.0f} → se contacta si P(fuga) > {C_CONTACTO:.0f}/{C_FN:.0f} = {UMBRAL_TEORICO:.2f}  (umbral teórico con probabilidades calibradas; Módulo 03, §7)")

# --- Particion: prueba SELLADA estratificada (Modulo 07, §1) ---
X_tr, X_te, y_tr, y_te, p_tr, p_te = train_test_split(X_all, y_all, p_verdadera, test_size=0.25, stratify=y_all, random_state=SEED)
costo_nada, costo_todos = costo_total(y_te, np.zeros(len(y_te))), costo_total(y_te, np.ones(len(y_te)))
costo_oraculo_p = costo_total(y_te, p_te > UMBRAL_TEORICO)                                  # el mejor posible conociendo la PROBABILIDAD verdadera (techo realista)
costo_perfecto = costo_total(y_te, y_te == 1)
print(f"\nEntrenamiento {len(X_tr):,} | prueba sellada {len(X_te):,} ({y_te.sum()} fugas)")
display(pd.DataFrame({"costo en la prueba": {"No hacer nada": costo_nada, "Contactar a todos": costo_todos, "Con la probabilidad VERDADERA (techo realista; solo en sintético)": costo_oraculo_p,
                                              "Clarividente (conoce quién se fuga)": costo_perfecto}}).round(0))
print("Cualquier modelo debe superar la mejor política trivial (aquí «no hacer nada») y se mide contra el techo realista, que SOLO conocemos porque los datos son sintéticos.")
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
X_all.assign(fuga=y_all).groupby("plan").fuga.mean().plot.bar(ax=axes[0], color=PALETA[0], rot=0); axes[0].set_title("Tasa de fuga por plan", fontweight="bold", fontsize=10); axes[0].set_ylabel("proporción")
pd.concat([X_all.antiguedad_meses, pd.Series(y_all, name="fuga")], axis=1).groupby(pd.cut(X_all.antiguedad_meses, [0, 6, 12, 24, 48, 121])).fuga.mean().plot.bar(ax=axes[1], color=PALETA[1], rot=0)
axes[1].set_title("Tasa de fuga según antigüedad (meses)", fontweight="bold", fontsize=10); axes[1].set_xlabel("")
(100 * X_all.isna().mean()[lambda s: s > 0]).plot.bar(ax=axes[2], color=PALETA[3], rot=0); axes[2].set_title("% de valores faltantes", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
assert 0.06 < y_all.mean() < 0.10 and costo_oraculo_p < costo_nada and costo_nada < costo_todos
````

---
**Lectura.** La fuga es del ≈ 8 %. Las **políticas triviales** son los pisos: «no hacer nada» cuesta 78 500 en la prueba y «contactar a todos» 100 000 (peor: la campaña cuesta más que lo que se pierde); con la **probabilidad verdadera** el costo bajaría a ≈ 44 800 (**techo realista**: un ahorro máximo de ≈ 43 % respecto a no hacer nada) y un clarividente llegaría a ≈ 7 850 (imposible). Cualquier modelo real vivirá entre 78 500 y 44 800. Los gráficos muestran las señales que construimos: más fuga en el plan básico y con poca antigüedad, y los faltantes concentrados en ingreso y satisfacción.

## 3. Caso A (continuación): Línea Base, `Pipeline` y Comparación de Modelos con CV Repetida 🧪

**Preprocesamiento dentro del `Pipeline`** ([Módulo 00](../00%20-%20Introduccion%20al%20Machine%20Learning/03_Importancia_del_Preprocesamiento.ipynb)): `SimpleImputer(median, add_indicator=True)` —con **indicador de faltante** porque la ausencia puede ser informativa ([Módulo 08, cuaderno 04](../08%20-%20Temas%20Avanzados%20de%20ML/04_Datos_Faltantes.ipynb))— más estandarización para las numéricas y *one-hot* para las categóricas, todo en un `ColumnTransformer`. `HistGradientBoosting` recibe los `NaN` **tal cual** (soporte nativo).

**Comparación rigurosa** ([Módulo 07](../07%20-%20Seleccion%20de%20Modelos%20y%20Optimizacion/00_Evaluacion_y_Seleccion_de_Modelos.ipynb)): validación cruzada **estratificada y repetida** (5×$r$) con los **mismos pliegues** para todos (comparación **pareada**), métrica **PR-AUC** (con 8 % de positivos, la exactitud engaña: [Módulo 08, cuaderno 03](../08%20-%20Temas%20Avanzados%20de%20ML/03_Datasets_Desbalanceados.ipynb)) y una **prueba t pareada corregida** (Nadeau y Bengio, 2003), porque los pliegues comparten datos y la prueba t ingenua **sobrestima la significancia**.

````python
# ============================================================
# CASO A.2  Linea base, Pipeline + ColumnTransformer con imputacion, y comparacion de modelos con CV repetida + prueba estadistica
# ============================================================
num_cols = ["edad", "antiguedad_meses", "cargo_mensual", "ingreso_mm", "uso_gb", "reclamos_6m", "cuotas_atrasadas", "satisfaccion", "ruido_sucursal", "ruido_dia"]
cat_cols = ["plan", "canal", "region"]
ohe = lambda: OneHotEncoder(handle_unknown="ignore", sparse_output=False)                  # dense: lo aceptan todos los modelos de abajo


def prep_con_imputacion():
    """Mediana + INDICADOR de faltante (la ausencia puede ser informativa: Modulo 08, cuaderno 04) y estandarizacion; one-hot para categoricas."""
    return ColumnTransformer([("num", make_pipeline(SimpleImputer(strategy="median", add_indicator=True), StandardScaler()), num_cols), ("cat", ohe(), cat_cols)])


def prep_nan_nativo():
    """Para HistGradientBoosting: los NaN pasan tal cual (el arbol aprende hacia que lado enviarlos)."""
    return ColumnTransformer([("num", "passthrough", num_cols), ("cat", ohe(), cat_cols)])


modelos = {
    "Baseline: prior (clase mayoritaria)": make_pipeline(prep_con_imputacion(), DummyClassifier(strategy="prior")),
    "Regresión logística": make_pipeline(prep_con_imputacion(), LogisticRegression(max_iter=2000)),
    "Random Forest": make_pipeline(prep_con_imputacion(), RandomForestClassifier(n_estimators=150, min_samples_leaf=10, n_jobs=-1, random_state=SEED)),
    "HistGradientBoosting (NaN nativo)": make_pipeline(prep_nan_nativo(), HistGradientBoostingClassifier(max_iter=150, learning_rate=0.06, max_leaf_nodes=15, random_state=SEED)),
}
# Validacion cruzada REPETIDA y estratificada; el preprocesamiento va DENTRO del Pipeline (sin fuga: Modulos 00 y 03). Mismos pliegues para todos (comparacion PAREADA)
cv = RepeatedStratifiedKFold(n_splits=5, n_repeats=N_REP, random_state=SEED)
t0 = time.time()
puntajes = {}
for nombre, m in modelos.items():
    r = cross_validate(m, X_tr, y_tr, cv=cv, scoring={"pr": "average_precision", "roc": "roc_auc"}, n_jobs=1)
    puntajes[nombre] = r
print(f"[{time.time() - t0:.0f} s] Validación cruzada 5×{N_REP} = {5 * N_REP} ajustes por modelo")
tabla = pd.DataFrame({n: {"PR-AUC (media)": r["test_pr"].mean(), "PR-AUC (desv.)": r["test_pr"].std(), "ROC-AUC (media)": r["test_roc"].mean()} for n, r in puntajes.items()}).T
display(tabla.round(3))
print(f"Referencia: la PR-AUC de un clasificador sin información es la prevalencia = {y_tr.mean():.3f} (Módulo 03, §6); el techo con la probabilidad verdadera es:", round(average_precision_score(y_te, p_te), 3), "(en prueba; solo sintético)")


def ttest_corregido(d, n_tr, n_te):
    """Prueba t pareada CORREGIDA (Nadeau y Bengio, 2003) para diferencias de k pliegues de CV repetida: var corregida = (1/k + n_te/n_tr) * var(d). Las estimaciones de pliegues NO son independientes (comparten datos)."""
    k = len(d); t = np.mean(d) / np.sqrt((1 / k + n_te / n_tr) * np.var(d, ddof=1))
    return t, 2 * stats.t.sf(abs(t), df=k - 1)


nombres = [n for n in puntajes if not n.startswith("Baseline")]
orden = sorted(nombres, key=lambda n: -puntajes[n]["test_pr"].mean())
mejor, segundo = orden[0], orden[1]
d = puntajes[mejor]["test_pr"] - puntajes[segundo]["test_pr"]
n_tr_cv, n_te_cv = int(len(X_tr) * 0.8), int(len(X_tr) * 0.2)
t_ingenua, p_ingenua = stats.ttest_rel(puntajes[mejor]["test_pr"], puntajes[segundo]["test_pr"])
t_corr, p_corr = ttest_corregido(d, n_tr_cv, n_te_cv)
w = stats.wilcoxon(d)
print(f"\nMejor: «{mejor}» frente a «{segundo}»: diferencia media de PR-AUC = {d.mean():+.4f} (desv. {d.std(ddof=1):.4f}; ganó en {(d > 0).sum()}/{len(d)} pliegues)")
print(f"  t pareada ingenua: p = {p_ingenua:.4f} | t pareada CORREGIDA (Nadeau–Bengio): p = {p_corr:.4f} | Wilcoxon: p = {w.pvalue:.4f}")
print("  Moraleja (Módulo 07, §3): la prueba ingenua sobrestima la significancia porque los pliegues comparten datos; usa la corregida o una prueba pareada robusta.")
assert tabla.loc["Regresión logística", "PR-AUC (media)"] > 2 * y_tr.mean()
````

---
**Lectura honesta.** (1) Toda la comparación está **por encima** de la línea base (PR-AUC = 0.079 = prevalencia), pero **la regresión logística (≈ 0.43) gana a los dos modelos de árboles (≈ 0.38)** y se acerca al techo (≈ 0.44): en un mundo casi logístico con pocas fugas, el modelo **simple** aprovecha mejor los datos que uno flexible que **sobreajusta** (Módulo 02, cuaderno 02, sección 7). No hay ley que diga que el *boosting* siempre gana. (2) La ventaja es **consistente** (10/10 pliegues) y la **prueba t corregida** (p ≈ 0.03) es **mucho menos contundente** que la ingenua (p ≈ 0.001): el factor de corrección muestra cuánta confianza falsa daba la prueba ingenua. (3) Con apenas 10 ajustes y una diferencia de 4 puntos, «significativo» no es «enorme»: conviene **no** presumir de la diferencia, y el Wilcoxon, con solo 10 pares, tiene una resolución limitada.

## 4. Caso A (continuación): Random Search, Calibración y Umbral por Costo 🎯

**Ajuste de hiperparámetros** con ***Random Search*** ([Módulo 07, cuaderno 02](../07%20-%20Seleccion%20de%20Modelos%20y%20Optimizacion/02_Random_Search_y_GridSearch.ipynb); Bergstra y Bengio, 2012): distribuciones **log-uniformes** para la tasa de aprendizaje y la regularización, enteros para el tamaño de los árboles, con PR-AUC por CV. Ajustamos los **dos candidatos** más prometedores y nos quedamos con el que **mejor puntúa en CV**.

**Calibración** ([Módulo 07, §4](../07%20-%20Seleccion%20de%20Modelos%20y%20Optimizacion/00_Evaluacion_y_Seleccion_de_Modelos.ipynb)): para usar el umbral $0.10$ las probabilidades deben ser **creíbles**. `CalibratedClassifierCV(method="isotonic")` aprende una función monótona de corrección; se mide con la **puntuación de Brier** y la **curva de calibración**.

**Umbral por costo** ([Módulo 03, §7](../03%20-%20Metricas%20y%20Ajuste%20de%20Hiperparametros/00_Metricas_de_Clasificacion.ipynb)): se comparan (i) 0.5 por defecto, (ii) el **umbral teórico** costo_contacto / costo_fuga = 0.10 y (iii) el umbral que **minimiza el costo en las predicciones fuera de muestra del entrenamiento** (OOF); este último se elige **sin tocar la prueba**.

````python
# ============================================================
# CASO A.3  Ajuste con Random Search (en el entrenamiento), calibracion y umbral por COSTO
# ============================================================
from scipy.stats import loguniform, randint
t0 = time.time()
# Se ajustan los dos candidatos mas prometedores de la CV (regresion logistica y gradient boosting); el criterio es la PR-AUC por CV, SOLO en el entrenamiento (Modulo 07, cuaderno 02)
candidatos = {
    "Regresión logística": (clone(modelos["Regresión logística"]), {"logisticregression__C": loguniform(1e-3, 1e2)}, max(6, N_ITER // 2)),
    "HistGradientBoosting": (clone(modelos["HistGradientBoosting (NaN nativo)"]),
                             {"histgradientboostingclassifier__learning_rate": loguniform(0.01, 0.3), "histgradientboostingclassifier__max_leaf_nodes": randint(4, 48),
                              "histgradientboostingclassifier__min_samples_leaf": randint(10, 150), "histgradientboostingclassifier__l2_regularization": loguniform(1e-3, 10),
                              "histgradientboostingclassifier__max_iter": randint(60, 300)}, N_ITER),
}
busquedas = {}
for nombre, (est, espacio, n_it) in candidatos.items():
    bs = RandomizedSearchCV(est, espacio, n_iter=n_it, scoring="average_precision", cv=StratifiedKFold(3, shuffle=True, random_state=SEED), random_state=SEED, n_jobs=1, refit=True)
    busquedas[nombre] = bs.fit(X_tr, y_tr)
    print(f"  {nombre}: {n_it} configuraciones × 3 pliegues → mejor PR-AUC de CV = {bs.best_score_:.3f} con", {k.split('__')[1]: (round(float(v), 4) if isinstance(v, (float, np.floating)) else v) for k, v in bs.best_params_.items()})
nombre_ajustado = max(busquedas, key=lambda n: busquedas[n].best_score_)
modelo_ajustado = busquedas[nombre_ajustado].best_estimator_
print(f"[{time.time() - t0:.0f} s] Modelo elegido por CV: «{nombre_ajustado}» (PR-AUC de CV = {busquedas[nombre_ajustado].best_score_:.3f}). Aviso: el máximo de varias configuraciones es OPTIMISTA; lo honesto es la prueba sellada.")
print("Honestidad: si el modelo 'simple' gana, se queda; no se elige gradient boosting «porque es más moderno». Aquí el mundo sintético es casi logístico, y la regresión logística lo sabe aprovechar con pocas fugas.\n")

# --- Probabilidades fuera de muestra (OOF) en el entrenamiento: para calibrar y elegir el umbral SIN tocar la prueba ---
cv_oof = StratifiedKFold(5, shuffle=True, random_state=SEED)
p_oof = cross_val_predict(clone(modelo_ajustado), X_tr, y_tr, cv=cv_oof, method="predict_proba")[:, 1]
cal = CalibratedClassifierCV(clone(modelo_ajustado), method="isotonic", cv=5).fit(X_tr, y_tr)   # calibracion isotonica con CV interna (Modulo 07, §4)
p_te_raw = modelo_ajustado.predict_proba(X_te)[:, 1]; p_te_cal = cal.predict_proba(X_te)[:, 1]
print(f"\nPrueba sellada: ROC-AUC = {roc_auc_score(y_te, p_te_cal):.3f} | PR-AUC = {average_precision_score(y_te, p_te_cal):.3f} (techo con la probabilidad verdadera: {average_precision_score(y_te, p_te):.3f})")
print(f"Brier (menor es mejor): sin calibrar {brier_score_loss(y_te, p_te_raw):.4f} | calibrado (isotónica) {brier_score_loss(y_te, p_te_cal):.4f} | probabilidad verdadera {brier_score_loss(y_te, p_te):.4f} | siempre la prevalencia {brier_score_loss(y_te, np.full(len(y_te), y_tr.mean())):.4f}")

# --- Umbral por COSTO: se elige con las probabilidades OOF del entrenamiento y se APLICA una sola vez a la prueba ---
umbrales = np.linspace(0.01, 0.6, 120)
costo_oof = np.array([costo_total(y_tr, p_oof > u) for u in umbrales])
u_costo = float(umbrales[np.argmin(costo_oof)])
filas = {
    "No hacer nada": costo_nada, "Contactar a todos": costo_todos,
    "Umbral 0.5 (el valor por defecto)": costo_total(y_te, p_te_cal > 0.5),
    f"Umbral teórico {UMBRAL_TEORICO:.2f} (costo_contacto / costo_fuga)": costo_total(y_te, p_te_cal > UMBRAL_TEORICO),
    f"Umbral elegido por costo en OOF ({u_costo:.2f})": costo_total(y_te, p_te_cal > u_costo),
    "Techo: probabilidad verdadera": costo_oraculo_p,
}
tab_costo = pd.DataFrame({"costo en prueba": filas}); tab_costo["ahorro frente a no hacer nada"] = costo_nada - tab_costo["costo en prueba"]
display(tab_costo.round(0))

fig, axes = plt.subplots(1, 2, figsize=(14, 4.6))
axes[0].plot(umbrales, costo_oof, color=PALETA[0], lw=2); axes[0].axvline(u_costo, color=PALETA[1], ls="--", label=f"mínimo en OOF: {u_costo:.2f}"); axes[0].axvline(UMBRAL_TEORICO, color=PALETA[2], ls=":", lw=2, label=f"teórico: {UMBRAL_TEORICO:.2f}")
axes[0].axhline(costo_total(y_tr, np.zeros(len(y_tr))), color="gray", ls="--", label="no hacer nada"); axes[0].set_xlabel("Umbral de probabilidad"); axes[0].set_ylabel("Costo total (entrenamiento, OOF)"); axes[0].legend(fontsize=8)
axes[0].set_title("El umbral óptimo sale del costo del negocio, no de 0.5", fontweight="bold", fontsize=10)
for p_, nom, c in [(p_te_raw, "sin calibrar", PALETA[3]), (p_te_cal, "calibrada (isotónica)", PALETA[0])]:
    fr, md = calibration_curve(y_te, p_, n_bins=8, strategy="quantile"); axes[1].plot(md, fr, "o-", color=c, lw=2, label=nom)
axes[1].plot([0, 0.5], [0, 0.5], "k--", label="calibración perfecta"); axes[1].set_xlabel("Probabilidad predicha"); axes[1].set_ylabel("Frecuencia observada de fuga"); axes[1].legend(fontsize=8)
axes[1].set_title("Curva de calibración (prueba)", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
assert tab_costo.loc[f"Umbral elegido por costo en OOF ({u_costo:.2f})", "costo en prueba"] < costo_nada
assert tab_costo.loc[f"Umbral elegido por costo en OOF ({u_costo:.2f})", "costo en prueba"] < tab_costo.loc["Umbral 0.5 (el valor por defecto)", "costo en prueba"]
````

---
**Lectura honesta.** (1) El **ajuste de hiperparámetros ayuda poco**: la regresión logística pasa de 0.427 (CV 5×2) a 0.424 (CV de 3 pliegues, otras particiones: **no** es una mejora) y el *boosting* de 0.384 a 0.394: **centésimas**, y el máximo de varias configuraciones es **optimista**. (2) El modelo elegido es de nuevo **la regresión logística**: gana también tras el ajuste. (3) **La calibración no cambia nada** (Brier de ≈ 0.0587 sin calibrar frente a ≈ 0.0591 con isotónica): la regresión logística ya produce probabilidades calibradas (el [Módulo 07](../07%20-%20Seleccion%20de%20Modelos%20y%20Optimizacion/00_Evaluacion_y_Seleccion_de_Modelos.ipynb) lo anticipa); calibrar un modelo que ya lo está solo añade ruido. En la práctica **se mide antes de calibrar**. (4) **El umbral importa mucho más que el modelo**: con el umbral 0.5 por defecto el ahorro es de ≈ 14 000, mientras que con el umbral basado en costos asciende a **≈ 27 000–30 000** (casi el doble) y se acerca al techo (≈ 33 700). (5) El umbral **teórico** (0.10; ahorro ≈ 29 900) rindió **mejor** que el elegido empíricamente en OOF (0.12; ≈ 27 300): el costo empírico es una curva **ruidosa** con pocas fugas, y cuando las probabilidades están calibradas conviene **confiar en la derivación** y usar el empírico como verificación.

## 5. Caso A (continuación): Explicabilidad y Análisis de Errores 🔍

**Explicabilidad** ([Módulo 08, cuaderno 02](../08%20-%20Temas%20Avanzados%20de%20ML/02_Explicabilidad_de_Modelos.ipynb)): **importancia por permutación** sobre la prueba, con el *pipeline* completo (importa la columna **original**; cuánto cae la PR-AUC al barajarla; [Módulo 02, cuaderno 01](../02%20-%20Arboles%20Bosques%20y%20Redes%20Neuronales/01_Random_Forests.ipynb)) y **PDP** de las variables principales. Un **control negativo** natural: las dos variables de ruido deben tener importancia ≈ 0; si no, hay fuga o sobreajuste.

**Análisis de errores** ([Géron, Cap. 2: *Analyzing the Best Models and Their Errors*]): no basta una métrica global; hay que mirar **dónde** falla el modelo: por plan, por antigüedad y por si **falta el ingreso**. Si el recobrado cambia mucho entre grupos, puede haber un problema de **equidad** o de **cobertura de los datos**.

````python
# ============================================================
# CASO A.4  Explicabilidad (permutacion y PDP) y analisis de errores
# ============================================================
# --- Importancia por PERMUTACION sobre la prueba, con el pipeline completo (columnas ORIGINALES; Modulos 02 y 08) ---
t0 = time.time()
imp = permutation_importance(modelo_ajustado, X_te, y_te, scoring="average_precision", n_repeats=5, random_state=SEED, n_jobs=1)
tab_imp = pd.DataFrame({"importancia (caída de PR-AUC)": imp.importances_mean, "desv.": imp.importances_std}, index=X_te.columns).sort_values("importancia (caída de PR-AUC)", ascending=False)
print(f"[{time.time() - t0:.0f} s]"); display(tab_imp.head(8).round(4))
irrelevantes = tab_imp.loc[["ruido_sucursal", "ruido_dia"], "importancia (caída de PR-AUC)"]
print("Variables IRRELEVANTES por construcción (ruido): importancia", irrelevantes.round(4).to_dict(), "→ ≈ 0, como debe ser (si no, sospecha de fuga o de sobreajuste).")

fig, axes = plt.subplots(1, 3, figsize=(17, 4.4))
top = tab_imp.head(10)[::-1]
axes[0].barh(top.index, top["importancia (caída de PR-AUC)"], xerr=top["desv."], color=PALETA[0]); axes[0].set_xlabel("Caída de PR-AUC al permutar"); axes[0].set_title("Importancia por permutación (prueba)", fontweight="bold", fontsize=10)
PartialDependenceDisplay.from_estimator(modelo_ajustado, X_te, ["antiguedad_meses", "reclamos_6m"], ax=axes[1:], kind="average", grid_resolution=20, random_state=SEED)
axes[1].set_title("PDP: antigüedad", fontweight="bold", fontsize=10); axes[2].set_title("PDP: reclamos en 6 meses", fontweight="bold", fontsize=10)
for ax in axes[1:]:
    ax.set_ylabel("probabilidad media predicha de fuga")
plt.tight_layout(); plt.show()
print("Lectura de los PDP: la probabilidad de fuga baja con la antigüedad y sube con los reclamos, como en el generador. Recuerda: PDP/permutación describen al MODELO, no la causalidad, y fallan con variables correlacionadas (Módulo 08, cuaderno 02).")

# --- Analisis de errores: ¿donde se pierden las fugas? ---
contactar = p_te_cal > u_costo
df_err = X_te.assign(fuga=y_te, contactar=contactar, p=p_te_cal, ingreso_faltante=X_te.ingreso_mm.isna(), tramo_antig=pd.cut(X_te.antiguedad_meses, [0, 12, 36, 121], labels=["<12 m", "12-36 m", ">36 m"]))
def resumen_seg(col):
    g = df_err.groupby(col, observed=True)
    return pd.DataFrame({"clientes": g.size(), "tasa real de fuga": g.fuga.mean(), "% contactados": g.contactar.mean(), "recobrado de fugas": g.apply(lambda d: d.contactar[d.fuga == 1].mean()),
                         "fugas perdidas (FN)": g.apply(lambda d: int(((d.fuga == 1) & ~d.contactar).sum()))})
print(f"Con el umbral {u_costo:.2f}: contactados {contactar.mean():.1%} de los clientes | recobrado global {contactar[y_te == 1].mean():.1%} | precisión {y_te[contactar].mean():.1%} (prevalencia {y_te.mean():.1%})")
for col in ["plan", "tramo_antig", "ingreso_faltante"]:
    display(resumen_seg(col).round(3))
cm = confusion_matrix(y_te, contactar); print("Matriz de confusión (filas: real; columnas: contactar sí/no):", cm.tolist())
# La calibracion por segmento: ¿el modelo sobreestima o subestima en algun grupo? (revision de equidad/robustez)
print("Probabilidad media predicha frente a tasa real por plan:", df_err.groupby("plan").apply(lambda d: f"{d.p.mean():.3f} vs {d.fuga.mean():.3f}").to_dict())
assert tab_imp.index[0] in ("antiguedad_meses", "reclamos_6m", "cuotas_atrasadas", "cargo_mensual", "plan") and abs(irrelevantes).max() < 0.02
````

**Lectura honesta.** (1) Las variables más importantes son las que **construimos** como relevantes (reclamos, cuotas atrasadas, antigüedad) y las dos de **ruido** tienen importancia ≈ 0 (±0.001): el **control negativo** pasa. Los PDP tienen la forma esperada (la fuga baja con la antigüedad y sube con los reclamos), pero describen **al modelo**, no la causalidad: contactar a clientes con muchos reclamos no garantiza que dejen de irse. (2) Con el umbral elegido, se contacta a ≈ 17 % de los clientes y se recupera ≈ 56 % de las fugas con una **precisión de ≈ 26 %** (prevalencia 8 %): la mayoría de los contactados **no** se iban, y es **correcto** que así sea cuando una oferta cuesta 50 y una fuga 500. (3) Por segmentos, **el recobrado cambia mucho por plan** (≈ 0.69 en el básico y ≈ 0.26–0.27 en estándar y premium): en parte porque el plan básico tiene más fugas (la probabilidad estimada ya lo refleja) y en parte porque con el mismo umbral el modelo deja pasar más fugas donde la tasa base es baja. Es una **decisión de negocio y de equidad** que hay que **revisar** (¿se aceptan umbrales distintos por segmento?, ¿qué costo tiene?); el modelo no la toma solo. (4) La calibración por plan es buena (probabilidad media predicha ≈ tasa real).

## 6. Caso B: Precio de Vivienda (Regresión con Diagnóstico de Residuos) 🏠

**Problema.** Estimar el **precio** (millones de COP) de una vivienda a partir de área, habitaciones, antigüedad, distancia al centro, estrato y barrio. **Datos sintéticos** (3 000 viviendas): el precio es **multiplicativo** (cada característica lo cambia en un **porcentaje**), con ruido proporcional. Eso crea las dificultades típicas: **cola derecha**, **varianza que crece con el precio** y relaciones **logarítmicas**.

**Qué hacemos** ([Módulo 01, cuaderno 01](../01%20-%20Modelos%20Lineales%20Supervisados/01_Regresion_Lineal.ipynb) y [Módulo 03, cuaderno 01](../03%20-%20Metricas%20y%20Ajuste%20de%20Hiperparametros/01_Metricas_de_Regresion.ipynb)): (i) línea base (la **mediana**); (ii) regresión lineal sobre el **precio** y sobre el **logaritmo del precio** (`TransformedTargetRegressor`, que **devuelve** las predicciones a la escala original), más *gradient boosting*; (iii) **métricas en unidades reales** (RMSE, MAE, MAPE) y no solo $R^2$; (iv) **diagnóstico de residuos**: linealidad, **homocedasticidad** (¿se abre un embudo?) y **normalidad** (QQ-plot); (v) **intervalos de predicción** por regresión **cuantílica** y **verificación de su cobertura**; (vi) el **error por segmento** de precio. Con **ingeniería de variables**: añadimos $\log(\text{área})$.

````python
# ============================================================
# CASO B  Regresion: precio de vivienda (sinteticos) con diagnostico de residuos y metricas en UNIDADES REALES
# ============================================================
def generar_viviendas(n=3000, semilla=11):
    rng = np.random.default_rng(semilla)
    area = np.clip(rng.lognormal(np.log(70), 0.4, n), 25, 400)                                    # m2
    estrato = rng.choice([1, 2, 3, 4, 5, 6], n, p=[0.10, 0.25, 0.30, 0.20, 0.10, 0.05])
    barrio = rng.choice(["norte", "centro", "sur", "periferia"], n, p=[0.3, 0.25, 0.25, 0.2])
    dist = np.clip(rng.gamma(2.0, 3.0, n), 0.3, 30)                                               # km al centro
    antig = np.clip(rng.gamma(2.0, 8.0, n), 0, 60)                                                # anios
    hab = np.clip(np.round(area / 28 + rng.normal(0, 0.7, n)), 1, 7)
    ef = {"norte": 0.12, "centro": 0.0, "sur": -0.08, "periferia": -0.2}
    log_p = (3.3 + 0.85 * np.log(area) + 0.11 * estrato - 0.025 * dist - 0.006 * antig + np.array([ef[b] for b in barrio]) + 0.05 * np.log(area) * (estrato - 3) / 3)
    ruido = rng.normal(0, 0.16, n) * (1 + 0.15 * (estrato >= 5))                                  # ruido MULTIPLICATIVO (un poco mayor en estratos altos)
    precio = np.exp(log_p + ruido)                                                                # millones de COP
    return pd.DataFrame({"area_m2": area, "habitaciones": hab, "antiguedad": antig, "dist_centro_km": dist, "estrato": estrato.astype(float), "barrio": barrio}), precio


Xv, pv = generar_viviendas()
Xv["log_area"] = np.log(Xv.area_m2)                                                              # ingenieria de variables: el precio crece con el LOGARITMO del area (relacion multiplicativa)
Xv_tr, Xv_te, pv_tr, pv_te = train_test_split(Xv, pv, test_size=0.25, random_state=SEED)
print(f"Viviendas: {len(Xv):,} | precio mediano {np.median(pv):.0f} millones | percentiles 5-95: {np.percentile(pv, 5):.0f}–{np.percentile(pv, 95):.0f} | asimetría {stats.skew(pv):.2f} (cola derecha: el precio es multiplicativo)")

prep_v = lambda: ColumnTransformer([("num", StandardScaler(), ["area_m2", "log_area", "habitaciones", "antiguedad", "dist_centro_km", "estrato"]), ("cat", OneHotEncoder(drop="first"), ["barrio"])])
modelos_b = {
    "Baseline: mediana": make_pipeline(prep_v(), DummyRegressor(strategy="median")),
    "Lineal sobre el precio": make_pipeline(prep_v(), LinearRegression()),
    "Lineal sobre log(precio)": make_pipeline(prep_v(), TransformedTargetRegressor(regressor=LinearRegression(), func=np.log, inverse_func=np.exp)),
    "Gradient boosting sobre el precio": make_pipeline(prep_v(), HistGradientBoostingRegressor(max_iter=200, learning_rate=0.06, random_state=SEED)),
    "Gradient boosting sobre log(precio)": make_pipeline(prep_v(), TransformedTargetRegressor(regressor=HistGradientBoostingRegressor(max_iter=200, learning_rate=0.06, random_state=SEED), func=np.log, inverse_func=np.exp)),
}
filas, pred_b = {}, {}
for n, m in modelos_b.items():
    m.fit(Xv_tr, pv_tr); p = m.predict(Xv_te); pred_b[n] = p
    filas[n] = {"RMSE (mill. COP)": np.sqrt(mean_squared_error(pv_te, p)), "MAE (mill. COP)": mean_absolute_error(pv_te, p), "MAPE": mean_absolute_percentage_error(pv_te, p), "R²": r2_score(pv_te, p)}
tabla_b = pd.DataFrame(filas).T; display(tabla_b.round(3))
print("Métricas en UNIDADES REALES (millones de COP) y MAPE: «el error típico es de X millones» se entiende; un R² suelto no (Módulo 03, cuaderno 01). Línea base: la mediana.")

# --- Diagnostico de residuos (Modulo 01, cuaderno 01): linealidad, homocedasticidad y normalidad ---
m_lin, m_log = modelos_b["Lineal sobre el precio"], modelos_b["Lineal sobre log(precio)"]
res_lin = pv_te - pred_b["Lineal sobre el precio"]; res_log = np.log(pv_te) - np.log(pred_b["Lineal sobre log(precio)"])
rho_lin = stats.spearmanr(np.abs(res_lin), pred_b["Lineal sobre el precio"])[0]; rho_log = stats.spearmanr(np.abs(res_log), pred_b["Lineal sobre log(precio)"])[0]
fig, axes = plt.subplots(1, 3, figsize=(17, 4.5))
axes[0].scatter(pred_b["Lineal sobre el precio"], res_lin, s=7, alpha=0.5, color=PALETA[1]); axes[0].axhline(0, color="k", lw=1)
axes[0].set_xlabel("Precio predicho (mill. COP)"); axes[0].set_ylabel("Residuo (real − predicho)"); axes[0].set_title(f"Lineal sobre el precio: embudo (ρ de |res| con el ajuste = {rho_lin:.2f})", fontweight="bold", fontsize=9)
axes[1].scatter(np.log(pred_b["Lineal sobre log(precio)"]), res_log, s=7, alpha=0.5, color=PALETA[2]); axes[1].axhline(0, color="k", lw=1)
axes[1].set_xlabel("log(precio predicho)"); axes[1].set_ylabel("Residuo en escala log"); axes[1].set_title(f"Lineal sobre log(precio): varianza más estable (ρ = {rho_log:.2f})", fontweight="bold", fontsize=9)
stats.probplot(res_log, dist="norm", plot=axes[2]); axes[2].set_title("QQ-plot de los residuos (escala log)", fontweight="bold", fontsize=9)
axes[2].get_lines()[0].set_markersize(3); axes[2].get_lines()[0].set_color(PALETA[0])
plt.tight_layout(); plt.show()
print(f"Residuos del lineal sobre el precio: asimetría {stats.skew(res_lin):.2f}; sobre el log: asimetría {stats.skew(res_log):.2f}, curtosis en exceso {stats.kurtosis(res_log):.2f} (≈ 0 es normal).")

# --- Intervalos de prediccion con regresion cuantilica y su COBERTURA (honestidad: ¿el 80 % nominal se cumple?) ---
q = {a: make_pipeline(prep_v(), HistGradientBoostingRegressor(loss="quantile", quantile=a, max_iter=150, learning_rate=0.06, random_state=SEED)).fit(Xv_tr, pv_tr) for a in (0.1, 0.9)}
lo, hi = q[0.1].predict(Xv_te), q[0.9].predict(Xv_te)
cobertura = np.mean((pv_te >= lo) & (pv_te <= hi))
print(f"Intervalo [cuantil 10 %, cuantil 90 %] (cobertura nominal 80 %): cobertura real en prueba = {cobertura:.1%}; ancho mediano = {np.median(hi - lo):.0f} millones")
# --- Error por segmento de precio (el error relativo no es uniforme) ---
seg = pd.qcut(pv_te, 4, labels=["Q1 (baratas)", "Q2", "Q3", "Q4 (caras)"])
display(pd.DataFrame({"MAPE": [mean_absolute_percentage_error(pv_te[seg == s], pred_b["Lineal sobre log(precio)"][seg == s]) for s in seg.categories], "MAE (mill.)": [mean_absolute_error(pv_te[seg == s], pred_b["Lineal sobre log(precio)"][seg == s]) for s in seg.categories]}, index=seg.categories).round(3))
assert tabla_b.loc["Lineal sobre log(precio)", "RMSE (mill. COP)"] < tabla_b.loc["Baseline: mediana", "RMSE (mill. COP)"] * 0.5 and abs(rho_lin) > abs(rho_log) and 0.6 < cobertura < 0.95
````

---
**Lectura honesta.** (1) La **mediana** como línea base da un RMSE de ≈ 690 millones y $R^2<0$: **peor que nada** en términos de $R^2$, la forma de recordar que la línea base es el piso. (2) El **modelo lineal sobre el logaritmo** (con $\log(\text{área})$) **gana** (RMSE ≈ 256 millones, $R^2\approx0.85$), por delante del lineal sobre el precio (≈ 306) y del *gradient boosting* (≈ 284): en un mundo **log-lineal**, el modelo simple **bien especificado** supera al flexible, que debe **descubrir** esa estructura con 2 250 ejemplos. El *boosting* con y sin log da casi lo mismo (≈ 284 y ≈ 286): los árboles no necesitan la transformación. (3) **Diagnóstico:** el lineal sobre el precio muestra el **embudo** (correlación de Spearman entre |residuo| y el ajuste $\rho\approx0.31$, frente a $\approx0.04$ sobre el logaritmo) y residuos **muy asimétricos** (≈ 2.1); sobre el logaritmo la varianza es más estable, la asimetría desaparece (≈ 0.05) y el QQ-plot es casi una recta (curtosis en exceso ≈ 0.17). (4) **Intervalos:** el intervalo cuantílico «de 80 %» tiene una cobertura **real de ≈ 66 %**: **sub-cubre**; sin calibración de los intervalos (p. ej. **predicción conforme**), un «80 %» nominal **no** se cumple. Verificar la cobertura es obligatorio. (5) **Por segmento:** el error relativo (MAPE) es de ≈ 11–15 % y el error absoluto crece con el precio (MAE ≈ 95 millones en las baratas y ≈ 326 en las caras): reportar **un solo número** esconde esa diferencia.

## 7. Caso C: Segmentación de Clientes (PCA + K-Means, Validados) 🧩

**Problema.** Agrupar **3 000 clientes** en segmentos de comportamiento (gasto, frecuencia, recencia, canal, devoluciones) para diseñar acciones distintas. **Sin etiquetas**, validar es difícil ([Módulo 04, cuaderno 00](../04%20-%20Clustering%20No%20Supervisado/00_Introduccion_Cap2_y_No_Supervisado.ipynb)). Como los datos son **sintéticos**, conocemos los 4 segmentos ocultos y podemos **contrastar** (en la vida real no).

**Qué hacemos:** (i) **log** de las variables sesgadas y **estandarización** (la escala define la distancia); (ii) **PCA** ([Módulo 05](../05%20-%20Reduccion%20de%20Dimensionalidad/00_PCA.ipynb)) para mirar la varianza y visualizar; (iii) **K-Means** para $k=2,\dots,8$ con **silueta** ([Módulo 04, cuaderno 04](../04%20-%20Clustering%20No%20Supervisado/04_Metricas_de_Clustering.ipynb)) y **estabilidad** (ARI entre ajustes sobre remuestras *bootstrap*: ¿la segmentación se reproduce?); (iv) ¿**ayuda** PCA antes de K-Means?; (v) **perfilar** los segmentos en unidades originales para que el negocio los pueda **nombrar**.

````python
# ============================================================
# CASO C  Segmentacion de clientes (sinteticos): PCA + K-Means validados con silueta y ESTABILIDAD
# ============================================================
def generar_segmentos(n=3000, semilla=5):
    rng = np.random.default_rng(semilla)
    # 4 segmentos latentes: (gasto mensual, compras/mes, dias desde la ultima compra, % online, % devoluciones)
    centros = {0: (400, 1.5, 60, 0.15, 0.05), 1: (150, 4.0, 15, 0.80, 0.10), 2: (900, 8.0, 7, 0.55, 0.04), 3: (80, 0.7, 150, 0.30, 0.15)}
    pesos = [0.35, 0.30, 0.15, 0.20]
    z = rng.choice(4, n, p=pesos)
    g = np.array([centros[k] for k in z])
    X = pd.DataFrame({"gasto_mensual": g[:, 0] * rng.lognormal(0, 0.3, n), "compras_mes": rng.poisson(g[:, 1]) + 0.0, "dias_ultima_compra": np.maximum(1, rng.exponential(g[:, 2])),
                      "pct_online": np.clip(g[:, 3] + rng.normal(0, 0.12, n), 0, 1), "pct_devoluciones": np.clip(g[:, 4] + rng.normal(0, 0.04, n), 0, 1), "ruido": rng.normal(size=n)})
    return X, z


Xs, z_verdadero = generar_segmentos()
# preprocesamiento: log a las variables sesgadas + estandarizar (Modulo 04, cuaderno 00: la escala define la distancia)
Xs_t = Xs.copy()
for c in ["gasto_mensual", "dias_ultima_compra"]:
    Xs_t[c] = np.log(Xs_t[c])
Xs_t["compras_mes"] = np.log1p(Xs_t["compras_mes"])
Z = StandardScaler().fit_transform(Xs_t)
print(f"Clientes: {len(Xs):,} | variables: {Xs.shape[1]} (una de ellas, «ruido», es irrelevante) | segmentos reales (ocultos): 4 con tamaños {np.round(np.bincount(z_verdadero) / len(z_verdadero), 2).tolist()}")

pca = PCA().fit(Z)
print("Varianza explicada por componente:", np.round(pca.explained_variance_ratio_, 3), "| acumulada con 2:", round(pca.explained_variance_ratio_[:2].sum(), 3), "| con 3:", round(pca.explained_variance_ratio_[:3].sum(), 3))
Zp = PCA(n_components=3).fit_transform(Z)

def estabilidad(Zx, k, n_pares=8, semilla=0):
    """Estabilidad por remuestreo: se ajustan dos K-Means sobre dos submuestras bootstrap y se comparan (ARI) sus asignaciones sobre TODOS los puntos. ARI ≈ 1: segmentacion reproducible."""
    rng = np.random.default_rng(semilla); aris = []
    for _ in range(n_pares):
        a = KMeans(k, n_init=3, random_state=int(rng.integers(1e6))).fit(Zx[rng.choice(len(Zx), len(Zx), replace=True)])
        b = KMeans(k, n_init=3, random_state=int(rng.integers(1e6))).fit(Zx[rng.choice(len(Zx), len(Zx), replace=True)])
        aris.append(adjusted_rand_score(a.predict(Zx), b.predict(Zx)))
    return float(np.mean(aris)), float(np.std(aris))


filas = []
for k in range(2, 9):
    km = KMeans(k, n_init=10, random_state=SEED).fit(Z)
    est_m, est_s = estabilidad(Z, k)
    filas.append({"k": k, "inercia": km.inertia_, "silueta": silhouette_score(Z, km.labels_), "estabilidad (ARI)": est_m, "ARI vs verdad oculta": adjusted_rand_score(z_verdadero, km.labels_)})
tab_k = pd.DataFrame(filas).set_index("k"); display(tab_k.round(3))
k_sil = int(tab_k.silueta.idxmax()); k_est = int(tab_k["estabilidad (ARI)"].idxmax())
print(f"k por silueta = {k_sil}; k más estable = {k_est}; la verdad oculta (que en la vida real NO se conoce) es k = 4. Se elige k combinando silueta, estabilidad, el codo y la INTERPRETABILIDAD de los grupos para el negocio.")

K_ELEGIDO = 4 if tab_k.loc[4, "silueta"] >= 0.9 * tab_k.silueta.max() else k_sil                  # criterio explicito: k=4 si su silueta esta a menos del 10 % del maximo
km4 = KMeans(K_ELEGIDO, n_init=10, random_state=SEED).fit(Z); etq = km4.labels_
km4p = KMeans(K_ELEGIDO, n_init=10, random_state=SEED).fit(Zp)
print(f"\nk elegido = {K_ELEGIDO}. ¿PCA (3 componentes) antes de K-Means ayuda? ARI vs verdad oculta: con las {Z.shape[1]} variables = {adjusted_rand_score(z_verdadero, etq):.3f} | con 3 componentes = {adjusted_rand_score(z_verdadero, km4p.labels_):.3f} "
      f"(silueta {silhouette_score(Z, etq):.3f} vs {silhouette_score(Zp, km4p.labels_):.3f}; mediciones en espacios distintos: NO comparables entre sí).")

perfil = Xs.assign(segmento=etq).groupby("segmento").agg(clientes=("gasto_mensual", "size"), gasto_mensual=("gasto_mensual", "median"), compras_mes=("compras_mes", "mean"), dias_ultima_compra=("dias_ultima_compra", "median"),
                                                         pct_online=("pct_online", "mean"), pct_devoluciones=("pct_devoluciones", "mean"))
display(perfil.round(2))
fig, axes = plt.subplots(1, 3, figsize=(17, 4.6))
axes[0].plot(tab_k.index, tab_k.silueta, "o-", color=PALETA[0], lw=2, label="silueta"); axes[0].plot(tab_k.index, tab_k["estabilidad (ARI)"], "s-", color=PALETA[2], lw=2, label="estabilidad (ARI)")
axes[0].set_xlabel("k"); axes[0].set_ylabel("valor"); axes[0].legend(fontsize=8); axes[0].set_title("Elegir k: silueta y estabilidad", fontweight="bold", fontsize=10)
Z2 = PCA(n_components=2).fit_transform(Z)
axes[1].scatter(Z2[:, 0], Z2[:, 1], c=etq, cmap="tab10", s=6, alpha=0.6); axes[1].set_xlabel("PC1"); axes[1].set_ylabel("PC2"); axes[1].set_title(f"Segmentos de K-Means (k={K_ELEGIDO}) en el plano de PCA", fontweight="bold", fontsize=10)
tab_cm = pd.crosstab(z_verdadero, etq); sns.heatmap(tab_cm, annot=True, fmt="d", cmap="Blues", cbar=False, ax=axes[2]); axes[2].set_xlabel("Segmento encontrado"); axes[2].set_ylabel("Segmento verdadero (oculto)")
axes[2].set_title("Contraste con la verdad oculta (solo posible en sintético)", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
assert adjusted_rand_score(z_verdadero, etq) > 0.4 and tab_k.loc[2:3, "estabilidad (ARI)"].max() > 0.3
````

(2) La **silueta es máxima en $k=4$** (0.32) y la **estabilidad también** (ARI ≈ 0.99): ambos criterios apuntan al mismo $k$, **y coinciden con los 4 segmentos ocultos** (ARI frente a la verdad ≈ 0.96). **Cuidado:** esto es un **mundo sintético con grupos bien separados**; en datos reales la silueta suele ser **mucho menor**, los criterios **discrepan** y $k$ se decide con **el negocio**. Una silueta de 0.32 es «estructura razonable», no «estructura fuerte». (3) **PCA antes de K-Means no ayudó**: con 3 componentes el ARI es ≈ 0.93 frente a ≈ 0.96 con las 6 variables (descarta algo de señal con poco beneficio); sirve para **visualizar**, no siempre para **agrupar** (y las siluetas en espacios distintos **no** son comparables entre sí). (4) El perfil permite nombrar los grupos (p. ej., «alto gasto y alta frecuencia», «inactivos con muchas devoluciones»): **una segmentación solo sirve si se puede traducir en acciones**.
