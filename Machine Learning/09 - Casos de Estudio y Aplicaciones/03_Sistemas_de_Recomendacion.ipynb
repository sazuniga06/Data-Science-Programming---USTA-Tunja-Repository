## 1. El Problema: una Matriz Usuario–Ítem Casi Vacía 🎬

Sea $R\in\mathbb R^{n_u\times n_i}$ la matriz de calificaciones de $n_u$ usuarios sobre $n_i$ ítems; **solo se observa una fracción** $\Omega=\{(u,i)\text{ observados}\}$ (en la práctica, el 0.1–5 %). Hay dos tareas relacionadas pero **distintas**:

- **Predicción de calificaciones:** estimar $\hat r_{ui}$ para $(u,i)\notin\Omega$. Se evalúa con **RMSE**.
- **Recomendación *top-k*:** para cada usuario, **ordenar** los ítems que aún no ha visto y mostrar los $k$ mejores. Se evalúa con **Precision@k, Recall@k, NDCG@k**.

| Retroalimentación | Qué observamos | Ejemplo | Ojo |
|---|---|---|---|
| **Explícita** | El usuario **dice** cuánto le gusta | Estrellas, «me gusta» | Escasa; sesgo de **auto-selección** (quien califica no es representativo) |
| **Implícita** | Se **infiere** del comportamiento | Clics, compras, tiempo de reproducción | Abundante, ruidosa; **no hay negativos** (no hacer clic ≠ no gustar) y se modela como **confianza**, no como calificación (Hu et al., 2008) |

**Datos de este cuaderno (sintéticos).** Generamos calificaciones con **estructura conocida**: $r=\mu+b_u+b_i+\mathbf p_u^\top\mathbf q_i+\varepsilon$ con $k=5$ factores latentes, redondeadas a $\{1,\dots,5\}$, y solo **≈ 6 %** de la matriz observada. La **probabilidad de observar** un par crece con la popularidad del ítem, la actividad del usuario y (un poco) con lo que le gustaría al usuario: **los datos faltan *no al azar* (MNAR)**, como en la realidad. Que sean sintéticos nos permite **comparar contra la verdad** (algo imposible con datos reales), pero también implica que **los resultados no son transferibles tal cual** a un catálogo real. Hay un bloque **opcional** con MovieLens al final.

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas matplotlib seaborn scikit-learn scipy

import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy
import sklearn
from sklearn.metrics import ndcg_score

warnings.filterwarnings("ignore")
SEED = 42
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (9, 5)
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]

# ---- Banderas del cuaderno ----
RAPIDO = True              # True: 3 conjuntos de datos sinteticos en la comparacion final (~35 s en total). False: 6 conjuntos (~1 min en total)
USAR_MOVIELENS = False     # True SOLO con red (Colab): descarga MovieLens 'ml-latest-small' (~1 MB) y repite el flujo en la seccion 8 (opcional)
N_SEMILLAS = 3 if RAPIDO else 6
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | scipy {scipy.__version__} | scikit-learn {sklearn.__version__}")
print(f"RAPIDO = {RAPIDO} | conjuntos de datos en la comparación final = {N_SEMILLAS} | USAR_MOVIELENS = {USAR_MOVIELENS}")
````

````python
# ============================================================
# 1. Datos SINTETICOS de calificaciones: factores latentes + sesgos + ruido + dispersion y exposicion realistas
# ============================================================
def generar_ratings(n_usuarios=600, n_items=400, k=5, semilla=0, densidad=0.06):
    """Calificaciones 1-5 con estructura conocida:  r = mu + b_u + b_i + p_u . q_i + ruido, redondeada y recortada a [1, 5].
    Solo se OBSERVA una fraccion: la probabilidad de calificar un item crece con la popularidad del item, la actividad del usuario
    y (un poco) con lo mucho que le gustaria al usuario (los usuarios calificaban lo que consumen, y consumen lo que les gusta: MNAR)."""
    rng = np.random.default_rng(semilla)
    P, Q = rng.normal(0, 0.9, (n_usuarios, k)), rng.normal(0, 0.9, (n_items, k))
    bu, bi, mu = rng.normal(0, 0.4, n_usuarios), rng.normal(0, 0.5, n_items), 2.6
    z_pop = rng.normal(size=n_items)
    bi = bi + 0.25 * z_pop                                                        # los items populares son (algo) mejores
    gusto = mu + bu[:, None] + bi[None, :] + P @ Q.T                              # calificacion "verdadera" sin ruido
    R_cont = gusto + rng.normal(0, 0.5, gusto.shape)
    R = np.clip(np.round(R_cont), 1, 5)
    pop = np.exp(1.1 * z_pop); pop = pop / pop.mean()                             # popularidad de cola larga (log-normal)
    act = np.exp(0.7 * rng.normal(size=n_usuarios)); act = act / act.mean()       # actividad de cada usuario (log-normal)
    expo = act[:, None] * pop[None, :] * np.exp(0.3 * (gusto - mu))
    prob = np.clip(expo * densidad / expo.mean(), 0, 0.9)
    M = rng.random(gusto.shape) < prob
    for u in range(n_usuarios):                                                   # cada usuario califica al menos 12 items
        if M[u].sum() < 12:
            M[u, rng.choice(n_items, 12, replace=False)] = True
    uu, ii = np.where(M)
    df = pd.DataFrame({"usuario": uu, "item": ii, "rating": R[uu, ii], "ts": rng.random(len(uu))})
    atributos = (Q @ rng.normal(0, 1, (k, 6)) + rng.normal(0, 0.6, (n_items, 6)))   # 6 "generos/atributos" de cada item (para contenido)
    return df, dict(P=P, Q=Q, bu=bu, bi=bi, mu=mu, gusto=gusto, R=R, atributos=atributos, n_usuarios=n_usuarios, n_items=n_items, pop=pop)


df, verdad = generar_ratings(semilla=0)
NU, NI = verdad["n_usuarios"], verdad["n_items"]
dens = len(df) / (NU * NI)
print(f"Matriz usuario-ítem: {NU} usuarios × {NI} ítems = {NU * NI:,} celdas | observadas: {len(df):,} ({dens:.1%}); el {1 - dens:.1%} está VACÍO")
print("Calificaciones por usuario: mediana", int(df.groupby('usuario').size().median()), "| por ítem: mediana", int(df.groupby('item').size().reindex(range(NI), fill_value=0).median()),
      "| ítems sin ninguna calificación:", int((df.groupby('item').size().reindex(range(NI), fill_value=0) == 0).sum()))

fig, axes = plt.subplots(1, 3, figsize=(16, 4.2))
cuentas = df.groupby("item").size().reindex(range(NI), fill_value=0).sort_values(ascending=False).values
axes[0].plot(cuentas, color=PALETA[0], lw=2); axes[0].fill_between(range(NI), cuentas, color=PALETA[0], alpha=0.2)
axes[0].set_xlabel("Ítems ordenados por popularidad"); axes[0].set_ylabel("Número de calificaciones"); axes[0].set_title("Cola larga: pocos ítems concentran las calificaciones", fontweight="bold", fontsize=9)
axes[1].bar(*np.unique(df.rating, return_counts=True), color=PALETA[2]); axes[1].set_xlabel("Calificación"); axes[1].set_ylabel("Frecuencia"); axes[1].set_title(f"Distribución de las calificaciones (media {df.rating.mean():.2f})", fontweight="bold", fontsize=9)
M_vis = np.zeros((NU, NI)); M_vis[df.usuario, df.item] = 1
orden_u, orden_i = np.argsort(-M_vis.sum(1))[:80], np.argsort(-M_vis.sum(0))[:100]
axes[2].imshow(M_vis[np.ix_(orden_u, orden_i)], cmap="Blues", aspect="auto"); axes[2].set_title("Matriz (80 usuarios y 100 ítems más activos): dispersa", fontweight="bold", fontsize=9)
axes[2].set_xlabel("Ítems"); axes[2].set_ylabel("Usuarios")
plt.tight_layout(); plt.show()
assert 0.04 < dens < 0.09 and df.groupby("usuario").size().min() >= 12

# --- Particion POR USUARIO y por TIEMPO: para cada usuario, sus calificaciones mas recientes van a prueba ---
def split_usuario_tiempo(d, frac_prueba=0.2):
    d = d.sort_values(["usuario", "ts"]).copy()
    rango = d.groupby("usuario").cumcount(); n = d.groupby("usuario")["item"].transform("size")
    es_prueba = rango >= np.ceil(n * (1 - frac_prueba))
    return d[~es_prueba].reset_index(drop=True), d[es_prueba].reset_index(drop=True)


tr, te = split_usuario_tiempo(df)
tr_fit, va = split_usuario_tiempo(tr, 0.2)                                       # validacion: dentro del entrenamiento (para elegir hiperparametros)
print(f"Entrenamiento {len(tr):,} | prueba {len(te):,} ({len(te) / len(df):.0%}) | del entrenamiento: ajuste {len(tr_fit):,} + validación {len(va):,}")
assert (tr.groupby("usuario").ts.max() <= te.groupby("usuario").ts.min().reindex(tr.groupby("usuario").ts.max().index).fillna(2)).all()     # nada del futuro en el entrenamiento
print("Verificado: para cada usuario, TODA calificación de entrenamiento es anterior a las de prueba (nada del futuro en el ajuste).")
print("Nota: en estos datos sintéticos el tiempo es aleatorio, por eso este corte equivale en distribución a uno aleatorio por usuario; en datos reales el corte temporal es más duro (los gustos y el catálogo cambian).")
````

**Lectura.** La matriz tiene **94 % de celdas vacías**; la popularidad de los ítems sigue una **cola larga** (unos pocos ítems concentran la mayoría de las calificaciones) y las calificaciones se concentran en valores altos (media ≈ 3.5). La partición **por usuario y tiempo** separa, para cada usuario, sus calificaciones más antiguas (entrenamiento) de las más recientes (prueba): **nada del futuro entra al ajuste**, como en el [Módulo 07](../07%20-%20Seleccion%20de%20Modelos%20y%20Optimizacion/01_Validacion_Cruzada_Avanzada.ipynb). ## 2. Evaluación Correcta y Líneas Base 📏

**RMSE** sobre las calificaciones de prueba: $\sqrt{\tfrac1{|T|}\sum_{(u,i)\in T}(r_{ui}-\hat r_{ui})^2}$. Para el ranking, si $\text{rel}_j\in\{0,1\}$ indica si el ítem en la posición $j$ del *top-k* es **relevante** (aquí: calificación de prueba $\ge4$):

$$\text{P@}k=\frac1k\sum_{j=1}^k\text{rel}_j,\qquad \text{R@}k=\frac{\sum_{j=1}^k\text{rel}_j}{|\text{relevantes}|},\qquad \text{NDCG@}k=\frac{\text{DCG@}k}{\text{IDCG@}k},\quad \text{DCG@}k=\sum_{j=1}^k\frac{\text{rel}_j}{\log_2(j+1)} .$$

El **descuento logarítmico** premia poner lo relevante **arriba**; IDCG es el DCG del orden **perfecto**, así NDCG $\in[0,1]$. Se promedia sobre los usuarios con al menos un ítem relevante en prueba, **excluyendo** los ítems ya calificados en entrenamiento.

**Líneas base (siempre primero):** *azar*; *media global* $\mu$; *media por ítem* amortiguada; **sesgos usuario + ítem** $\hat r_{ui}=\mu+b_u+b_i$ (mínimos cuadrados regularizados), y **popularidad** (ordenar por número de calificaciones). Un modelo complicado que **no** supera a estas líneas base **no merece desplegarse**.

````python
# ============================================================
# 2. Evaluacion (RMSE y metricas de ranking desde cero) y lineas base
# ============================================================
def a_matrices(d, nu=NU, ni=NI):
    """Triplete (usuario, item, rating) -> matriz densa R (0 donde falta) y mascara booleana M de lo observado."""
    R = np.zeros((nu, ni)); M = np.zeros((nu, ni), dtype=bool)
    R[d.usuario.values, d.item.values] = d.rating.values; M[d.usuario.values, d.item.values] = True
    return R, M


def rmse(pred, d, lo=1, hi=5):
    p = np.clip(pred[d.usuario.values, d.item.values], lo, hi)                       # las calificaciones viven en [lo, hi] = [1, 5]
    return float(np.sqrt(np.mean((p - d.rating.values) ** 2)))


def metricas_ranking(S, M_tr, d_te, k=10, umbral=4, rel=None):
    """Precision@k, Recall@k y NDCG@k promedio sobre los usuarios que tienen algun item relevante en prueba.
    S (NU,NI): puntaje de cada item para cada usuario. Se EXCLUYEN los items ya calificados en entrenamiento.
    Relevante = item de prueba con calificacion >= umbral (o la matriz booleana `rel`, si se da). NDCG@k = DCG/IDCG con DCG = sum_i rel_i / log2(i + 1) (posiciones i = 1..k)."""
    if rel is None:
        rel = np.zeros_like(M_tr)
        dr = d_te[d_te.rating >= umbral]; rel[dr.usuario.values, dr.item.values] = True
    S = np.where(M_tr, -np.inf, S)
    top = np.argsort(-S, axis=1, kind="stable")[:, :k]
    aciertos = np.take_along_axis(rel, top, axis=1).astype(float)
    n_rel = rel.sum(1); ok = n_rel > 0
    desc = 1.0 / np.log2(np.arange(2, k + 2))
    dcg = (aciertos * desc).sum(1)
    idcg = np.array([desc[:min(k, n)].sum() for n in n_rel])
    return {f"P@{k}": float((aciertos.sum(1) / k)[ok].mean()), f"R@{k}": float((aciertos.sum(1) / np.maximum(n_rel, 1))[ok].mean()), f"NDCG@{k}": float((dcg[ok] / idcg[ok]).mean())}


# --- Verificacion de NDCG@k contra scikit-learn (puntajes continuos: sin empates) y de P/R a mano ---
rng = np.random.default_rng(1)
Rtr, Mtr = a_matrices(tr)
S_az = rng.normal(size=(NU, NI)); mi = metricas_ranking(S_az, Mtr, te)
rel_ = np.zeros((NU, NI)); dr_ = te[te.rating >= 4]; rel_[dr_.usuario.values, dr_.item.values] = 1
ok = np.where(rel_.sum(1) > 0)[0]
nd_sk = np.mean([ndcg_score(rel_[u:u + 1], np.where(Mtr[u], -1e9, S_az[u])[None], k=10) for u in ok])
assert np.isclose(mi["NDCG@10"], nd_sk)
S_perfecto = np.where(rel_ > 0, 10.0, 0.0) + rng.normal(0, 0.01, (NU, NI)); m_perf = metricas_ranking(S_perfecto, Mtr, te)
assert np.isclose(m_perf["NDCG@10"], 1.0)                                         # un orden perfecto da NDCG = 1
ex = metricas_ranking(np.array([[0.9, 0.8, 0.1, 0.7, 0.2]]), np.zeros((1, 5), bool), pd.DataFrame({"usuario": [0, 0], "item": [1, 4], "rating": [5, 5]}), k=3)
assert np.isclose(ex["P@3"], 1 / 3) and np.isclose(ex["R@3"], 1 / 2) and np.isclose(ex["NDCG@3"], (1 / np.log2(3)) / (1 + 1 / np.log2(3)))      # item 1 en la posicion 2; item 4 fuera del top-3
print(f"OK: NDCG@10 desde cero == sklearn.metrics.ndcg_score (azar: {mi['NDCG@10']:.4f}); orden perfecto → NDCG = 1; ejemplo a mano: P@3 = 1/3, R@3 = 1/2, NDCG@3 = {ex['NDCG@3']:.4f}.")
print(f"Un ranking AL AZAR obtiene: {mi}  (la línea base más tonta; fracción de relevantes: {len(dr_) / len(te):.0%} de las calificaciones de prueba).")

# --- Lineas base de calificacion: media global, media por item (amortiguada) y sesgos usuario + item ---
def modelo_sesgos(R, M, lam_u=10.0, lam_i=25.0, iters=10):
    """b_u y b_i por minimos cuadrados regularizados alternados:  r ≈ mu + b_u + b_i."""
    mu = R[M].mean(); bu = np.zeros(R.shape[0]); bi = np.zeros(R.shape[1])
    for _ in range(iters):
        bi = (M * (R - mu - bu[:, None])).sum(0) / (lam_i + M.sum(0))
        bu = (M * (R - mu - bi[None, :])).sum(1) / (lam_u + M.sum(1))
    return mu, bu, bi


mu, bu_, bi_ = modelo_sesgos(Rtr, Mtr)
pred_media = np.full((NU, NI), mu)
pred_item = mu + np.tile(((Mtr * (Rtr - mu)).sum(0) / (25 + Mtr.sum(0)))[None, :], (NU, 1))
pred_sesgos = mu + bu_[:, None] + bi_[None, :]
pop_cuentas = Mtr.sum(0).astype(float)

def evaluar(S_rank, pred_rating=None, k=10):
    out = metricas_ranking(S_rank, Mtr, te, k)
    out["RMSE"] = rmse(pred_rating, te) if pred_rating is not None else np.nan
    return out

filas = {
    "Al azar": evaluar(rng.normal(size=(NU, NI))),
    "Media global": evaluar(rng.normal(size=(NU, NI)), pred_media),
    "Popularidad (n.º de calificaciones)": evaluar(np.tile(pop_cuentas, (NU, 1))),
    "Media por ítem (amortiguada)": evaluar(pred_item, pred_item),
    "Sesgos usuario + ítem": evaluar(pred_sesgos, pred_sesgos),
}
tabla_base = pd.DataFrame(filas).T[["RMSE", "P@10", "R@10", "NDCG@10"]]
display(tabla_base.round(3))
print("\nLo que importa para evaluar un recomendador (honestidad):")
print("  · La RMSE mide qué tan bien se PREDICE la calificación; el ranking mide qué se RECOMIENDA. Son objetivos distintos y pueden contradecirse.")
print("  · Un recomendador real solo muestra pocos ítems: P@k, R@k y NDCG@k premian poner lo relevante ARRIBA.")
print("  · «No calificado» no significa «no le gusta»: tratamos los ítems no calificados como no relevantes, lo que SUBESTIMA todas las métricas y favorece a lo popular (se discute en la sección 7).")
assert tabla_base.loc["Sesgos usuario + ítem", "RMSE"] < tabla_base.loc["Media global", "RMSE"]
assert tabla_base.loc["Popularidad (n.º de calificaciones)", "NDCG@10"] > 3 * tabla_base.loc["Al azar", "NDCG@10"]
````

**Lectura honesta.** Nuestro **NDCG@k coincide con `sklearn.metrics.ndcg_score`**, el orden perfecto da NDCG = 1 y el ejemplo a mano se cumple. Las líneas base ya enseñan dos cosas: (1) los **sesgos usuario + ítem** bajan el RMSE de 1.41 (media global) a ≈ 1.29, pero **casi no ayudan en el ranking** (NDCG ≈ 0.13: el sesgo de un usuario no cambia el *orden* de los ítems para ese usuario); (2) la **popularidad** —sin personalizar nada— obtiene un NDCG@10 de ≈ 0.19, **más de 15 veces el del azar** (≈ 0.01). La popularidad es una línea base **muy dura** de superar, y por razones que veremos en la sección 7.

## 3. Filtrado Colaborativo por Vecinos 👥

El **filtrado colaborativo (CF)** usa **solo la matriz de interacciones** (sin atributos de usuarios ni ítems). La versión clásica es por **vecindarios** (Resnick et al., 1994; Sarwar et al., 2001):

- **Basado en usuarios:** «los usuarios *parecidos a ti* calificaron alto este ítem».
- **Basado en ítems:** «tú calificaste alto ítems *parecidos a este*».

Para medir la similitud se **centran** las calificaciones (restando una base: la media del usuario, o el modelo de sesgos) para que «califica siempre alto» no parezca gusto, y se usa el **coseno** entre los vectores centrados (equivalente a la **correlación de Pearson** cuando se centra por la media sobre lo coincidente):
$$s_{uv}=\frac{\sum_{i}\tilde r_{ui}\tilde r_{vi}}{\sqrt{\sum_i\tilde r_{ui}^2}\sqrt{\sum_i\tilde r_{vi}^2}}\cdot\frac{n_{uv}}{n_{uv}+\lambda},\qquad \tilde r_{ui}=r_{ui}-b_{ui},$$
donde el último factor es la **contracción** (*shrinkage*, «ponderación por significancia»: Herlocker et al., 2004) que **castiga similitudes calculadas con pocos ítems en común** ($n_{uv}$). La predicción es el **promedio ponderado** de las desviaciones de los $k$ vecinos más parecidos:
$$\hat r_{ui}=b_{ui}+\frac{\sum_{v\in N_k(u)}s_{uv}\,\tilde r_{vi}}{\varepsilon+\sum_{v\in N_k(u)}|s_{uv}|}\quad(\text{y análogamente por ítems con }s_{ij}).$$
Los hiperparámetros ($k$, $\lambda$) se eligen **en validación**, nunca en prueba. En ambas variantes el costo crece con la **matriz de similitudes** ($n_u^2$ o $n_i^2$): con millones de usuarios hace falta aproximación (LSH, ANN).

````python
# ============================================================
# 3. Filtrado colaborativo por vecinos (basado en usuarios y en items) desde cero
# ============================================================
def similitud_coseno_centrada(Rc, M, shrink=10.0, por_filas=True):
    """Similitud COSENO entre las FILAS (o columnas) de la matriz centrada Rc (0 donde falta), equivalente a Pearson sobre lo coincidente.
    shrink: 'contraccion' sim * n/(n + shrink), con n = n.º de coincidencias; castiga similitudes calculadas con pocos datos."""
    if not por_filas:
        Rc, M = Rc.T, M.T
    num = Rc @ Rc.T                                                                  # producto escalar de las calificaciones centradas
    nr = np.sqrt((Rc ** 2).sum(1)); den = np.outer(nr, nr)
    S = np.divide(num, den, out=np.zeros_like(num), where=den > 0)
    n_co = M.astype(float) @ M.T.astype(float)                                       # n.º de elementos calificados por AMBOS
    S = S * n_co / np.maximum(n_co + shrink, 1e-12)                                  # sin coincidencias (n=0) la similitud es 0, no 0/0 = NaN (con shrink=0)
    np.fill_diagonal(S, 0.0)
    return S


def top_k_por_fila(S, k):
    """Se queda con los k vecinos de mayor similitud POSITIVA de cada fila (el resto, 0)."""
    Sk = np.zeros_like(S)
    idx = np.argpartition(-S, k, axis=1)[:, :k]
    filas = np.arange(S.shape[0])[:, None]
    Sk[filas, idx] = np.clip(S[filas, idx], 0, None)
    return Sk


def knn_usuarios(R, M, k=40, shrink=10.0, base=None, amort=1.0):
    """Prediccion basada en USUARIOS: r_ui = base_ui + sum_v s_uv (r_vi - base_vi) / (amort + sum_v |s_uv|), v en los k vecinos mas parecidos de u que calificaron i."""
    mu_u = (R * M).sum(1) / np.maximum(M.sum(1), 1)
    B = (mu_u[:, None] + np.zeros_like(R)) if base is None else base
    Rc = (R - B) * M
    S = similitud_coseno_centrada(Rc, M, shrink)
    Sk = top_k_por_fila(S, k)
    return B + (Sk @ Rc) / (amort + np.abs(Sk) @ M.astype(float))


def knn_items(R, M, k=40, shrink=10.0, base=None, amort=1.0):
    """Prediccion basada en ITEMS: r_ui = base_ui + sum_j s_ij (r_uj - base_uj) / (amort + sum_j |s_ij|), j en los k items mas parecidos a i que u califico (coseno ajustado)."""
    mu_u = (R * M).sum(1) / np.maximum(M.sum(1), 1)
    B = (mu_u[:, None] + np.zeros_like(R)) if base is None else base
    Rc = (R - B) * M
    S = similitud_coseno_centrada(Rc, M, shrink, por_filas=False)
    Sk = top_k_por_fila(S, k)
    return B + (Rc @ Sk.T) / (amort + M.astype(float) @ np.abs(Sk).T)


# --- Verificacion de la similitud: Pearson sobre las calificaciones coincidentes a mano (un par de usuarios) ---
Rtr, Mtr = a_matrices(tr)
mu_u = (Rtr * Mtr).sum(1) / Mtr.sum(1)
Rc = (Rtr - mu_u[:, None]) * Mtr
S_u = similitud_coseno_centrada(Rc, Mtr, shrink=0.0)
u, v = np.argsort(-Mtr.sum(1))[:2]                                                    # los dos usuarios mas activos
co = Mtr[u] & Mtr[v]
a, b = Rc[u, co], Rc[v, co]
cos_a_mano = (a @ b) / np.sqrt((a @ a) * (b @ b)) if co.sum() > 1 else np.nan
# la norma del coseno de la funcion usa TODAS las calificaciones de cada usuario; la de Pearson 'a mano' solo las coincidentes: lo comprobamos por separado
nu_, nv_ = np.sqrt((Rc[u] ** 2).sum()), np.sqrt((Rc[v] ** 2).sum())
assert np.isclose(S_u[u, v], (a @ b) / (nu_ * nv_))
print(f"OK: similitud coseno centrada de los usuarios {u} y {v} (coinciden en {int(co.sum())} ítems) = {S_u[u, v]:.4f}; Pearson solo sobre coincidencias = {cos_a_mano:.4f} (el Pearson clásico normaliza solo con lo coincidente; nuestra variante usa la norma de TODAS las calificaciones de cada usuario y por eso castiga a quien coincide en poco).")

# --- Eleccion de k y del encogimiento en VALIDACION (no en prueba) ---
Rfit, Mfit = a_matrices(tr_fit)
filas = []
for k in [10, 20, 40, 80, 160]:
    for shrink in [0.0, 10.0, 30.0]:
        filas.append({"k": k, "shrink": shrink, "usuarios": rmse(knn_usuarios(Rfit, Mfit, k, shrink), va), "ítems": rmse(knn_items(Rfit, Mfit, k, shrink), va)})
val_knn = pd.DataFrame(filas)
mejor_u = val_knn.loc[val_knn["usuarios"].idxmin()]; mejor_i = val_knn.loc[val_knn["ítems"].idxmin()]
print("RMSE de validación (mejor combinación):", f"usuarios k={int(mejor_u.k)}, shrink={mejor_u.shrink:g} → {mejor_u['usuarios']:.3f}", "|", f"ítems k={int(mejor_i.k)}, shrink={mejor_i.shrink:g} → {mejor_i['ítems']:.3f}")
print(val_knn.pivot(index="k", columns="shrink").round(3).to_string())
fig, ax = plt.subplots(figsize=(8.6, 4.4))
for sh, c in zip([0.0, 10.0, 30.0], PALETA):
    g = val_knn[val_knn.shrink == sh]
    ax.plot(g.k, g["usuarios"], "o-", color=c, lw=2, label=f"usuarios, shrink={sh:g}"); ax.plot(g.k, g["ítems"], "s--", color=c, lw=2, alpha=0.6, label=f"ítems, shrink={sh:g}")
ax.set_xscale("log"); ax.set_xticks([10, 20, 40, 80, 160]); ax.set_xticklabels([10, 20, 40, 80, 160]); ax.set_xlabel("Vecinos k"); ax.set_ylabel("RMSE de validación"); ax.legend(fontsize=7, ncol=2)
ax.set_title("kNN: más vecinos reducen el error; aquí la contracción (shrink) no ayuda", fontweight="bold", fontsize=9)
plt.tight_layout(); plt.show()

# --- Evaluacion en PRUEBA (entrenando con todo el entrenamiento) ---
ku, ki = int(mejor_u.k), int(mejor_i.k)
pred_uu = knn_usuarios(Rtr, Mtr, ku, mejor_u.shrink); pred_ii = knn_items(Rtr, Mtr, ki, mejor_i.shrink)
# base mas fuerte: predecir el RESIDUO respecto al modelo de sesgos (usuario + item)
B_ses = mu + bu_[:, None] + bi_[None, :]
pred_uu_b = knn_usuarios(Rtr, Mtr, ku, mejor_u.shrink, base=B_ses); pred_ii_b = knn_items(Rtr, Mtr, ki, mejor_i.shrink, base=B_ses)
res_knn = {
    "kNN usuarios (base: media del usuario)": evaluar(pred_uu, pred_uu), "kNN ítems (base: media del usuario)": evaluar(pred_ii, pred_ii),
    "kNN usuarios (base: sesgos u+i)": evaluar(pred_uu_b, pred_uu_b), "kNN ítems (base: sesgos u+i)": evaluar(pred_ii_b, pred_ii_b),
}
display(pd.concat([tabla_base.loc[["Popularidad (n.º de calificaciones)", "Sesgos usuario + ítem"]], pd.DataFrame(res_knn).T[["RMSE", "P@10", "R@10", "NDCG@10"]]]).round(3))
````

**Lectura honesta.** La similitud coincide con el cálculo a mano (con la norma completa de cada usuario). Los vecindarios **mejoran la predicción** frente a los sesgos (RMSE de ≈ 1.29 a ≈ 1.16 con usuarios y ≈ 1.20 con ítems, centrando por el modelo de sesgos), pero **en ranking no superan a la popularidad**: el mejor kNN por ítems alcanza un NDCG@10 de ≈ 0.18 frente a ≈ 0.19 de la popularidad (y ≈ 0.13 de los sesgos). Dos lecciones: (1) en validación, **más vecinos** reducen el error (con `shrink=0`: RMSE ≈ 1.34 con k=10 frente a ≈ 1.24 con k=160 en usuarios, y de ≈ 1.35 a ≈ 1.31 en ítems), pero la **contracción NO ayudó** en este conjunto sintético: con k=160 el RMSE es ≈ 1.24 (`shrink=0`), ≈ 1.30 (`shrink=10`) y ≈ 1.34 (`shrink=30`) en usuarios, y ≈ 1.31, ≈ 1.34 y ≈ 1.36 en ítems; el mejor es `shrink=0` en ambos. Es una propiedad de *estos* datos, no una regla: aquí la norma de la similitud (que usa todas las calificaciones de cada usuario) ya castiga a quien coincide en pocos ítems, y la contracción añadida aplana también las similitudes fiables; con otros datos podría ayudar, por eso se decide en validación y no se da por supuesto. (Con `shrink=0` hay pares sin ninguna co-valoración: su factor $n/(n+0)$ sería $0/0$, y la función lo trata como similitud 0 en lugar de NaN.) (2) centrar por el **modelo de sesgos** da **mejor RMSE** que centrar por la media del usuario (usuarios: 1.16 frente a 1.18; ítems: 1.20 frente a 1.25) y, en la variante por ítems, también **mejor ranking** (NDCG@10 de ≈ 0.177 frente a ≈ 0.149); con usuarios el ranking queda igual (≈ 0.164 frente a ≈ 0.167): el sesgo del ítem es justamente lo que hace subir a los populares.

## 4. Factorización de Matrices con SGD 🧬

En lugar de comparar usuarios o ítems directamente, suponemos que **cada usuario y cada ítem viven en un espacio de $k$ factores latentes** (p. ej. «cuánto le gusta la acción», «cuánta acción tiene la película») y que la afinidad es su producto escalar. Con sesgos (Koren et al., 2009):

$$\hat r_{ui}=\mu+b_u+b_i+\mathbf p_u^\top\mathbf q_i,\qquad \min_{b,\mathbf p,\mathbf q}\ \sum_{(u,i)\in\Omega}\tfrac12\,e_{ui}^2+\tfrac{\lambda}2\big(b_u^2+b_i^2+\|\mathbf p_u\|^2+\|\mathbf q_i\|^2\big),\quad e_{ui}=r_{ui}-\hat r_{ui}.$$

Es **la misma idea de LSA y de los *embeddings*** del [cuaderno 02](02_Procesamiento_de_Lenguaje_Natural.ipynb): factorizar una matriz en dos de rango bajo. La diferencia clave es que **solo se suman las celdas observadas** y se **regulariza**. Los gradientes de la pérdida por observación son
$$\frac{\partial\ell}{\partial b_u}=-e_{ui}+\lambda b_u,\quad \frac{\partial\ell}{\partial b_i}=-e_{ui}+\lambda b_i,\quad \frac{\partial\ell}{\partial\mathbf p_u}=-e_{ui}\mathbf q_i+\lambda\mathbf p_u,\quad \frac{\partial\ell}{\partial\mathbf q_i}=-e_{ui}\mathbf p_u+\lambda\mathbf q_i,$$
y el **descenso del gradiente estocástico** recorre las observaciones (aquí en **mini-lotes** de 64, por velocidad) y actualiza $\theta\leftarrow\theta-\eta\,\nabla\ell$. Como siempre que se escribe un gradiente a mano, **se verifica con diferencias finitas**. Los hiperparámetros ($k$, $\lambda$) se eligen en **validación**.

````python
# ============================================================
# 4. Factorizacion de matrices con sesgos y regularizacion, entrenada con SGD (desde cero)
# ============================================================
# Modelo:  r_hat_ui = mu + b_u + b_i + p_u . q_i
# Perdida por observacion:  l_ui = 1/2 * e_ui^2 + lam/2 * ( b_u^2 + b_i^2 + ||p_u||^2 + ||q_i||^2 ),   e_ui = r_ui - r_hat_ui
# Gradientes:  dl/db_u = -e + lam b_u ;  dl/db_i = -e + lam b_i ;  dl/dp_u = -e q_i + lam p_u ;  dl/dq_i = -e p_u + lam q_i
def perdida_obs(params, u, i, r, mu, lam):
    bu, bi, P, Q = params
    e = r - (mu + bu[u] + bi[i] + (P[u] * Q[i]).sum(1))
    return float(np.sum(0.5 * e ** 2 + 0.5 * lam * (bu[u] ** 2 + bi[i] ** 2 + (P[u] ** 2).sum(1) + (Q[i] ** 2).sum(1))))


def grad_analitico(params, u, i, r, mu, lam):
    bu, bi, P, Q = params
    e = r - (mu + bu[u] + bi[i] + (P[u] * Q[i]).sum(1))
    gbu, gbi, gP, gQ = np.zeros_like(bu), np.zeros_like(bi), np.zeros_like(P), np.zeros_like(Q)
    np.add.at(gbu, u, -e + lam * bu[u]); np.add.at(gbi, i, -e + lam * bi[i])           # np.add.at suma las contribuciones de observaciones repetidas
    np.add.at(gP, u, -e[:, None] * Q[i] + lam * P[u]); np.add.at(gQ, i, -e[:, None] * P[u] + lam * Q[i])
    return gbu, gbi, gP, gQ


# --- Verificacion del gradiente con diferencias finitas centradas ---
rng = np.random.default_rng(0)
nu_t, ni_t, k_t = 6, 5, 3
u_t, i_t = rng.integers(0, nu_t, 20), rng.integers(0, ni_t, 20); r_t = rng.integers(1, 6, 20).astype(float)
par_t = [rng.normal(0, 0.3, nu_t), rng.normal(0, 0.3, ni_t), rng.normal(0, 0.3, (nu_t, k_t)), rng.normal(0, 0.3, (ni_t, k_t))]
g_an = grad_analitico(par_t, u_t, i_t, r_t, 3.0, 0.1)
max_err = 0.0
for arr, g in zip(par_t, g_an):
    for idx in np.ndindex(arr.shape):
        orig = arr[idx]; h = 1e-6
        arr[idx] = orig + h; fp = perdida_obs(par_t, u_t, i_t, r_t, 3.0, 0.1)
        arr[idx] = orig - h; fm = perdida_obs(par_t, u_t, i_t, r_t, 3.0, 0.1)
        arr[idx] = orig
        max_err = max(max_err, abs((fp - fm) / (2 * h) - g[idx]) / (abs(g[idx]) + 1e-8))
assert max_err < 1e-5
print(f"OK: gradiente analítico == diferencias finitas centradas en las {sum(a.size for a in par_t)} entradas de (b_u, b_i, P, Q); error relativo máximo = {max_err:.1e}")


def entrenar_mf_sgd(d, nu=NU, ni=NI, k=8, lam=0.05, lr=0.05, epocas=40, lote=64, semilla=0, d_val=None):
    """SGD por mini-lotes sobre las calificaciones OBSERVADAS. Devuelve (modelo, historial de RMSE de entrenamiento y validacion)."""
    rng = np.random.default_rng(semilla)
    u, i, r = d.usuario.values, d.item.values, d.rating.values.astype(float); mu = r.mean()
    params = [np.zeros(nu), np.zeros(ni), rng.normal(0, 0.1, (nu, k)), rng.normal(0, 0.1, (ni, k))]
    hist = []
    for ep in range(epocas):
        orden = rng.permutation(len(r))
        for s in range(0, len(r), lote):
            b = orden[s:s + lote]
            g = grad_analitico(params, u[b], i[b], r[b], mu, lam)
            for p, gp in zip(params, g):
                p -= lr * gp
        pred = mu + params[0][:, None] + params[1][None, :] + params[2] @ params[3].T
        hist.append((rmse(pred, d), rmse(pred, d_val) if d_val is not None else np.nan))
    return dict(mu=mu, bu=params[0], bi=params[1], P=params[2], Q=params[3]), np.array(hist)


def predecir_mf(m):
    return m["mu"] + m["bu"][:, None] + m["bi"][None, :] + m["P"] @ m["Q"].T


# --- Hiperparametros en VALIDACION (k y lambda) ---
t0 = time.time()
filas = []
for k in [2, 5, 10]:
    for lam in [0.02, 0.1, 0.3]:
        m, h = entrenar_mf_sgd(tr_fit, k=k, lam=lam, epocas=25)
        filas.append({"k": k, "lambda": lam, "RMSE entrenamiento": h[-1, 0], "RMSE validación": rmse(predecir_mf(m), va)})
val_mf = pd.DataFrame(filas)
mejor = val_mf.loc[val_mf["RMSE validación"].idxmin()]
display(val_mf.pivot(index="k", columns="lambda", values="RMSE validación").round(3))
print(f"Mejor en validación: k = {int(mejor.k)}, λ = {mejor['lambda']} (RMSE de validación {mejor['RMSE validación']:.3f}) [{time.time() - t0:.0f} s]")

# --- Curvas de entrenamiento y evaluacion final en PRUEBA con los hiperparametros elegidos ---
KMF, LAMMF = int(mejor.k), float(mejor["lambda"])
m_sgd, hist = entrenar_mf_sgd(tr, k=KMF, lam=LAMMF, epocas=40, d_val=te)
pred_sgd = predecir_mf(m_sgd)
res_sgd = evaluar(pred_sgd, pred_sgd)
print("MF (SGD) en prueba:", {k_: round(v, 3) for k_, v in res_sgd.items()})
fig, axes = plt.subplots(1, 2, figsize=(14, 4.4))
axes[0].plot(hist[:, 0], color=PALETA[0], lw=2, label="entrenamiento"); axes[0].plot(hist[:, 1], color=PALETA[1], lw=2, label="prueba (solo para ILUSTRAR; no se usó para elegir nada)")
axes[0].axhline(rmse(pred_sesgos, te), color="gray", ls="--", label="línea base de sesgos"); axes[0].set_xlabel("Época"); axes[0].set_ylabel("RMSE"); axes[0].legend(fontsize=8); axes[0].set_title("MF con SGD: el error de entrenamiento baja; el de prueba toca fondo", fontweight="bold", fontsize=9)
for k, c in zip([2, 5, 10], PALETA):
    g = val_mf[val_mf.k == k]; axes[1].plot(g["lambda"], g["RMSE validación"], "o-", color=c, lw=2, label=f"k = {k}")
axes[1].set_xscale("log"); axes[1].set_xlabel("Regularización λ"); axes[1].set_ylabel("RMSE de validación"); axes[1].legend(); axes[1].set_title("Más factores exigen más regularización", fontweight="bold", fontsize=9)
plt.tight_layout(); plt.show()
````

## 5. Mínimos Cuadrados Alternados (ALS) y la Conexión con la SVD 🔁

Si fijamos $Q$, el problema en $(\mathbf p_u,b_u)$ es una **regresión ridge** con **solución cerrada**; si fijamos $P$, igual para $(\mathbf q_i,b_i)$. **ALS** alterna esas dos resoluciones exactas:

$$\begin{bmatrix}\mathbf p_u\\ b_u\end{bmatrix}=\big(X_u^\top X_u+\lambda\,n_u I\big)^{-1}X_u^\top\mathbf y_u,\qquad X_u=[\,Q_{I_u}\ \ \mathbf 1\,],\quad \mathbf y_u=\mathbf r_u-\mu-\mathbf b_{I_u}.$$

Con la regularización **ponderada** por el número de observaciones ($\lambda n_u$, ALS-WR de Zhou et al., 2008), **cada paso es un mínimo exacto** del mismo objetivo, así que **el objetivo nunca aumenta** (lo verificamos). Frente a SGD: ALS no necesita tasa de aprendizaje, **paraleliza** trivialmente (cada usuario/ítem es independiente) y es el método estándar para **retroalimentación implícita** (Hu et al., 2008), donde hay que sumar sobre **toda** la matriz con pesos de confianza.

**La conexión con la SVD (Eckart–Young, 1936).** Si la matriz $R$ estuviera **completa**, la mejor aproximación de rango $k$ (en norma de Frobenius) es la **SVD truncada** $R_k=U_k\Sigma_kV_k^\top$, con error $\|R-R_k\|_F=\sqrt{\sum_{j>k}\sigma_j^2}$. ALS sin regularizar resuelve **el mismo problema**, así que debe **encontrar la misma matriz**. Pero con **datos faltantes** el problema cambia: la SVD **no sabe** qué celdas faltan y trataría los huecos como **ceros** (calificar «0 estrellas»), lo cual es un error grave; MF **ajusta solo lo observado**.

````python
# ============================================================
# 5. Minimos cuadrados alternados (ALS) y la conexion con la SVD (Eckart-Young)
# ============================================================
def als(R, M, k=5, lam=0.05, iters=15, semilla=0, mu=0.0, sesgos=True, historial=False):
    """ALS sobre las calificaciones OBSERVADAS (R, con mascara M). Con sesgos = True el modelo es r ≈ mu + b_u + b_i + p_u . q_i y minimiza
        J = sum_obs (r - mu - b_u - b_i - p_u.q_i)^2 + lam * [ sum_u n_u (||p_u||^2 + b_u^2) + sum_i n_i (||q_i||^2 + b_i^2) ]      (regularizacion ponderada, ALS-WR).
    Con Q fijo, el vector [p_u, b_u] de cada usuario es una REGRESION RIDGE con solucion cerrada: [p_u, b_u] = (X^T X + lam n_u I)^-1 X^T y, con X = [Q_Iu, 1]
    e y = r_u - mu - b_i. Lo mismo para cada item con P fijo. Cada paso es un minimo exacto, asi que J NUNCA aumenta."""
    rng = np.random.default_rng(semilla)
    nu, ni = R.shape
    P, Q = rng.normal(0, 0.1, (nu, k)), rng.normal(0, 0.1, (ni, k))
    bu, bi = np.zeros(nu), np.zeros(ni)
    nu_obs, ni_obs = M.sum(1), M.sum(0); J = []
    completa = bool(M.all()) and not sesgos                                              # sin faltantes ni sesgos: una sola resolucion para todos los usuarios (y items)
    kk = k + 1 if sesgos else k
    I_k = np.eye(kk)
    for _ in range(iters):
        if completa:
            P = np.linalg.solve(Q.T @ Q + lam * nu_obs[0] * I_k, Q.T @ R.T).T
            Q = np.linalg.solve(P.T @ P + lam * ni_obs[0] * I_k, P.T @ R).T
        else:
            for u in range(nu):
                idx = M[u]
                if idx.any():
                    X = np.hstack([Q[idx], np.ones((idx.sum(), 1))]) if sesgos else Q[idx]
                    y = R[u, idx] - mu - (bi[idx] if sesgos else 0.0)
                    th = np.linalg.solve(X.T @ X + lam * nu_obs[u] * I_k, X.T @ y); P[u] = th[:k]
                    if sesgos:
                        bu[u] = th[k]
            for i in range(ni):
                idx = M[:, i]
                if idx.any():
                    X = np.hstack([P[idx], np.ones((idx.sum(), 1))]) if sesgos else P[idx]
                    y = R[idx, i] - mu - (bu[idx] if sesgos else 0.0)
                    th = np.linalg.solve(X.T @ X + lam * ni_obs[i] * I_k, X.T @ y); Q[i] = th[:k]
                    if sesgos:
                        bi[i] = th[k]
        if historial:
            e = M * (R - mu - bu[:, None] - bi[None, :] - P @ Q.T)
            J.append(float((e ** 2).sum() + lam * ((nu_obs * ((P ** 2).sum(1) + bu ** 2)).sum() + (ni_obs * ((Q ** 2).sum(1) + bi ** 2)).sum())))
    modelo = dict(mu=mu, bu=bu, bi=bi, P=P, Q=Q)
    return (modelo, np.array(J)) if historial else modelo


# --- (a) Verificacion: cada actualizacion de ALS NO AUMENTA el objetivo (J decrece monotonamente) ---
Rtr, Mtr = a_matrices(tr)
_, J = als(Rtr, Mtr, k=5, lam=0.05, iters=15, mu=Rtr[Mtr].mean(), historial=True)
assert (np.diff(J) <= 1e-9).all()
print(f"OK: el objetivo de ALS decrece monótonamente: J[0] = {J[0]:.1f} → J[-1] = {J[-1]:.1f} (en 15 iteraciones; cada paso es un mínimo exacto).")

# --- (b) Hiperparametros en VALIDACION ---
Rf, Mf = a_matrices(tr_fit)
t0 = time.time(); filas = []
for k in [2, 5, 10]:
    for lam in [0.02, 0.05, 0.1, 0.3]:
        m = als(Rf, Mf, k=k, lam=lam, iters=12, mu=Rf[Mf].mean())
        filas.append({"k": k, "lambda": lam, "RMSE validación": rmse(predecir_mf(m), va)})
val_als = pd.DataFrame(filas); mejor_a = val_als.loc[val_als["RMSE validación"].idxmin()]
display(val_als.pivot(index="k", columns="lambda", values="RMSE validación").round(3))
KALS, LAMALS = int(mejor_a.k), float(mejor_a["lambda"])
print(f"Mejor en validación: k = {KALS}, λ = {LAMALS} (RMSE de validación {mejor_a['RMSE validación']:.3f}) [{time.time() - t0:.0f} s]")

m_als = als(Rtr, Mtr, k=KALS, lam=LAMALS, iters=20, mu=Rtr[Mtr].mean())
pred_als = predecir_mf(m_als)
res_als = evaluar(pred_als, pred_als)
print("MF (ALS) en prueba:", {k_: round(v, 3) for k_, v in res_als.items()}, "| MF (SGD):", {k_: round(v, 3) for k_, v in res_sgd.items()})

# --- (c) Eckart-Young: con la matriz COMPLETAMENTE observada, la mejor aproximacion de rango k es la SVD truncada, y ALS (sin regularizar) la reencuentra ---
Rfull = verdad["R"][:300, :200]                                                      # una submatriz COMPLETA (sin faltantes) de calificaciones
U_, S_, Vt_ = np.linalg.svd(Rfull, full_matrices=False)
for k in [3, 5]:
    A_k = (U_[:, :k] * S_[:k]) @ Vt_[:k]
    err_svd = np.linalg.norm(Rfull - A_k)
    mf_ = als(Rfull, np.ones_like(Rfull, dtype=bool), k=k, lam=1e-9, iters=400, semilla=1, sesgos=False)
    P_f, Q_f = mf_["P"], mf_["Q"]
    err_als = np.linalg.norm(Rfull - P_f @ Q_f.T)
    err_teo = np.sqrt((S_[k:] ** 2).sum())                                           # Eckart-Young: error = raiz de la suma de los valores singulares descartados al cuadrado
    P_al, Q_al = np.random.default_rng(5).normal(size=(300, k)), np.random.default_rng(6).normal(size=(200, k))
    assert np.isclose(err_svd, err_teo) and abs(err_als - err_svd) / err_svd < 1e-6 and err_als <= np.linalg.norm(Rfull - P_al @ Q_al.T)
    assert np.allclose(P_f @ Q_f.T, A_k, atol=1e-4)                                  # y la reconstruccion COINCIDE (el subespacio optimo es unico si sigma_k > sigma_{k+1})
    print(f"k = {k}: error de Frobenius SVD = {err_svd:.4f} | raíz(Σσ² descartados) = {err_teo:.4f} | ALS (λ≈0) = {err_als:.4f} → la reconstrucción de ALS == SVD truncada (σ_k/σ_(k+1) = {S_[k - 1] / S_[k]:.2f})")
print("OK (Eckart–Young): con la matriz completa, ALS sin regularizar converge a la SVD truncada; con datos faltantes ya NO son lo mismo, como muestra la siguiente tabla.")

# --- (d) Con datos faltantes, la SVD de la matriz «rellena con ceros» NO es la solucion ---
def svd_relleno(E, k):
    U, S, Vt = np.linalg.svd(E, full_matrices=False)
    return (U[:, :k] * S[:k]) @ Vt[:k]
cero_raw = svd_relleno(Rtr, KALS)                                                     # rellenar con 0 las calificaciones crudas: "0 estrellas" inexistentes
E_tr = (Rtr - (mu + bu_[:, None] + bi_[None, :])) * Mtr                              # residuo respecto a los sesgos (0 donde falta)
relleno_res = mu + bu_[:, None] + bi_[None, :] + svd_relleno(E_tr, KALS)             # rellenar con la linea base (residuo 0) = SVD de imputacion
tab = pd.DataFrame({"SVD con ceros en lugar de lo faltante": {"RMSE": rmse(cero_raw, te)}, "SVD con la línea base en lugar de lo faltante": {"RMSE": rmse(relleno_res, te)},
                    "ALS solo sobre lo observado (regularizado)": {"RMSE": rmse(pred_als, te)}, "MF con SGD (sesgos + regularización)": {"RMSE": rmse(pred_sgd, te)}}).T
display(tab.round(3))
assert tab.iloc[0, 0] > tab.iloc[2, 0] + 1.0 and tab.iloc[1, 0] > tab.iloc[2, 0]
````

---
**Lectura.** (1) El objetivo de ALS **decrece monótonamente** en cada iteración. (2) ALS y SGD, con los hiperparámetros elegidos en validación, llegan a **resultados casi idénticos** (RMSE ≈ 0.78–0.79; mismo $k=5$): son dos algoritmos para el **mismo modelo**, y la elección es de ingeniería (paralelismo, implícito, simplicidad). (3) **Eckart–Young verificado**: con la matriz completa, el error de la SVD truncada es **exactamente** $\sqrt{\sum_{j>k}\sigma_j^2}$ y ALS sin regularizar **converge a la misma reconstrucción** (aunque la brecha $\sigma_k/\sigma_{k+1}\approx1.08$ es pequeña y exige muchas iteraciones). (4) Con **datos faltantes**, la SVD «rellenando con ceros» es **desastrosa** (RMSE ≈ 2.6: peor que predecir la media), rellenar con la **línea base** mejora mucho (≈ 1.2) y solo **ajustar lo observado** con regularización llega a ≈ 0.79. *Matemáticamente*, SVD y MF coinciden **solo** sin faltantes.

## 6. La Comparación Honesta: Varios Conjuntos de Datos 🏁

## 7. Arranque en Frío, Sesgo de Popularidad y Burbuja de Filtro ❄️🫧

**Arranque en frío** (*cold start*): el CF puro **no puede** recomendar a un usuario nuevo ni un ítem nuevo, porque no tiene historial. Las salidas habituales: popularidad al inicio, **preguntar** al usuario sus gustos y **usar contenido** —atributos del ítem (género, texto, imagen) o del usuario— para estimar sus factores latentes. La **recomendación basada en contenido** compara el perfil del usuario con los atributos de los ítems; un **sistema híbrido** usa CF donde hay historia y contenido donde no. Aquí aprendemos un **ridge de atributos → factores latentes** en ítems con historia y lo aplicamos a los ítems nuevos.

**Sesgo de popularidad y burbuja de filtro.** Los datos observados **ya reflejan** lo que el sistema mostró antes (**sesgo de exposición**). Un modelo que se entrena y se evalúa sobre esos datos **refuerza lo popular** (el rico se hace más rico) y tiende a **homogeneizar** lo que ve cada usuario (**burbuja de filtro**, Pariser, 2011). Lo medimos con la **cobertura** del catálogo y la popularidad media de lo recomendado. Y, como los datos son sintéticos, podemos hacer algo que **no se puede hacer en la realidad**: evaluar **contra la verdad completa**.

````python
# ============================================================
# 7. Arranque en frio, sesgo de popularidad y recomendacion basada en contenido
# ============================================================
# Usamos el conjunto 0 (variables tr, te, m_sgd, pred_sgd... de las secciones anteriores)
Rtr, Mtr = a_matrices(tr)
n_ent_u = Mtr.sum(1)

# --- (a) ARRANQUE EN FRIO DE USUARIOS: error segun cuantas calificaciones tiene el usuario en entrenamiento ---
te2 = te.assign(n_ent=n_ent_u[te.usuario.values])
bins = pd.qcut(te2.n_ent, 4, duplicates="drop"); te2["grupo"] = bins.astype(str)
filas = []
for g, d in te2.groupby("grupo", sort=False):
    filas.append({"calificaciones del usuario en entrenamiento": g, "n. pruebas": len(d), "RMSE sesgos": rmse(pred_sesgos, d), "RMSE MF (SGD)": rmse(pred_sgd, d), "RMSE MF (ALS)": rmse(pred_als, d)})
frio_u = pd.DataFrame(filas).set_index("calificaciones del usuario en entrenamiento")
display(frio_u.round(3))

# --- (b) ARRANQUE EN FRIO DE ITEMS: items NUEVOS (sin ninguna calificacion). CF puro no puede; el contenido (atributos) si ---
rng = np.random.default_rng(3)
fríos = rng.choice(NI, size=int(0.1 * NI), replace=False); es_frio = np.zeros(NI, bool); es_frio[fríos] = True
tr_w = tr[~es_frio[tr.item.values]]; te_f = te[es_frio[te.item.values]]                  # entrenamos SIN esos items; evaluamos SOLO sobre ellos
m_w, _ = entrenar_mf_sgd(tr_w, k=KMF, lam=LAMMF, epocas=30)
pred_cf = predecir_mf(m_w)                                                              # para un item frio, p_u . q_i con q_i ~ 0 y b_i = 0: queda solo mu + b_u
from sklearn.linear_model import Ridge
X_att = verdad["atributos"]; warm = ~es_frio
# el contenido PREDICE los parametros latentes del item (sesgo y factores) a partir de sus atributos: modelo ridge aprendido SOLO con items con historia
reg = Ridge(alpha=1.0).fit(X_att[warm], np.column_stack([m_w["bi"][warm], m_w["Q"][warm]]))
est = reg.predict(X_att[es_frio])
bi_c, Q_c = m_w["bi"].copy(), m_w["Q"].copy(); bi_c[es_frio] = est[:, 0]; Q_c[es_frio] = est[:, 1:]
pred_contenido = m_w["mu"] + m_w["bu"][:, None] + bi_c[None, :] + m_w["P"] @ Q_c.T
filas = {"Media global": rmse(np.full((NU, NI), tr_w.rating.mean()), te_f), "Sesgo del usuario (sin info del ítem)": rmse(pred_cf, te_f), "Híbrido: CF + contenido para ítems nuevos": rmse(pred_contenido, te_f)}
display(pd.DataFrame({"RMSE en ítems nuevos": filas}).round(3))
r2_q = reg.score(X_att[warm], np.column_stack([m_w["bi"][warm], m_w["Q"][warm]]))
print(f"{es_frio.sum()} ítems nuevos y {len(te_f)} calificaciones de prueba sobre ellos. R² del ridge de atributos → factores (en ítems con historia): {r2_q:.2f}")
assert filas["Híbrido: CF + contenido para ítems nuevos"] < filas["Sesgo del usuario (sin info del ítem)"]

# --- (c) SESGO DE POPULARIDAD: ¿a quien recomienda cada modelo? cobertura del catalogo y popularidad media de lo recomendado ---
def lista_top(S, k=10):
    S = np.where(Mtr, -np.inf, S); return np.argsort(-S, axis=1, kind="stable")[:, :k]
pop_rank = (-Mtr.sum(0)).argsort().argsort() / (NI - 1)                                  # 0 = el mas popular, 1 = el menos popular
filas = {}
for nombre, S in [("Popularidad", np.tile(Mtr.sum(0).astype(float), (NU, 1))), ("kNN ítems", knn_items(Rtr, Mtr, ki, mejor_i.shrink, base=mu + bu_[:, None] + bi_[None, :])),
                  ("MF (SGD)", pred_sgd), ("MF (ALS)", pred_als)]:
    top = lista_top(S)
    filas[nombre] = {"cobertura del catálogo": len(np.unique(top)) / NI, "percentil de popularidad medio (0 = más popular)": pop_rank[top].mean()}
cob = pd.DataFrame(filas).T; display(cob.round(3))
print("Cobertura = fracción del catálogo que aparece en alguna lista de top-10. Una cobertura baja concentra la exposición en pocos ítems (el rico se hace más rico):")
print("  si el sistema se REENTRENA con lo que el propio sistema muestra, esa concentración se REFUERZA (bucle de retroalimentación) → burbuja de filtro y homogeneización.")
print("  Mitigaciones: reordenar por diversidad/novedad, exploración (bandidos, Módulo 06), penalizar la popularidad y MEDIR cobertura y diversidad junto con la exactitud.")

# --- (d) ¿Y si pudiéramos evaluar contra la VERDAD? Solo es posible porque los datos son sintéticos: conocemos la calificacion de TODOS los pares ---
rel_oraculo = (verdad["R"] >= 4) & ~Mtr                                                   # relevante = le gustaria (>= 4) y aun no lo ha calificado
filas = {}
for nombre, S in [("Popularidad", np.tile(Mtr.sum(0).astype(float), (NU, 1))), ("kNN ítems", knn_items(Rtr, Mtr, ki, mejor_i.shrink, base=mu + bu_[:, None] + bi_[None, :])),
                  ("MF (SGD)", pred_sgd), ("MF (ALS)", pred_als)]:
    o_prueba = metricas_ranking(S, Mtr, te, 10)
    o_verdad = metricas_ranking(S, Mtr, te, 10, rel=rel_oraculo)
    filas[nombre] = {"NDCG@10 (prueba observada)": o_prueba["NDCG@10"], "NDCG@10 (contra la verdad completa)": o_verdad["NDCG@10"], "P@10 (contra la verdad completa)": o_verdad["P@10"]}
orac = pd.DataFrame(filas).T; display(orac.round(3))
print("La evaluación offline con calificaciones OBSERVADAS premia lo popular (lo que la gente ya consume y califica); contra la verdad completa, el orden cambia.")
print("En datos reales NUNCA se tiene la verdad completa: de ahí la importancia de las pruebas A/B en línea, de medir cobertura/diversidad y de corregir el sesgo de exposición (p. ej. ponderación inversa por propensión).")
assert orac.loc["MF (SGD)", "NDCG@10 (contra la verdad completa)"] > orac.loc["Popularidad", "NDCG@10 (contra la verdad completa)"]
assert cob.loc["Popularidad", "cobertura del catálogo"] < cob.loc["MF (SGD)", "cobertura del catálogo"]
````

---
**Lectura honesta.** (a) **Usuarios fríos:** el RMSE de MF mejora con la **actividad** del usuario (más calificaciones, mejores factores), pero incluso en el grupo con pocas calificaciones **no es peor que los sesgos**: la regularización lo **encoge hacia la línea base**. (b) **Ítems nuevos:** el CF puro solo conoce el sesgo del usuario (RMSE ≈ 1.18); el **híbrido con contenido** baja a ≈ 0.83 (el ridge explica ≈ 60 % de los factores), aunque depende de que los atributos **sí** informen sobre los factores (aquí así se construyó). (c) **Cobertura:** la popularidad recomienda solo el ≈ 9 % del catálogo (y casi todo es lo más popular), el kNN por ítems ≈ 29 % y **MF ≈ 70 %**: MF **personaliza** mucho más. (d) **Contra la verdad completa**, el orden se invierte: **MF acierta ≈ 93–95 % del *top-10*** (precisión@10) frente a ≈ 42 % de la popularidad y ≈ 75 % del kNN por ítems. Entonces MF **sí** recomienda mejor, y **la evaluación offline con calificaciones observadas** lo castigó porque **solo cuenta como relevante lo que ya se había expuesto**, y eso es lo popular. En un caso real, **nunca** tendrás la verdad completa: de ahí las **pruebas A/B**, las métricas de cobertura/diversidad/novedad y las correcciones de exposición. Un resultado así **no** prueba que MF sea «mejor» en general: depende de la métrica, del protocolo y del mundo.
