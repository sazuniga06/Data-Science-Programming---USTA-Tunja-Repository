## 1. El Corpus: Reseñas Sintéticas en Español 🧪

No hay descargas en el flujo principal: **generamos un corpus de reseñas en español** a partir de **plantillas con ruido**, con **semilla fija**. Conviene decirlo sin rodeos:

> ⚠️ **El corpus es SINTÉTICO.** Los textos salen de unas pocas plantillas (simples, con negación, con doble negación, mixtas con «pero» y sarcásticas) y un vocabulario pequeño (≈ 570 palabras); por eso **las métricas serán más optimistas que en datos reales** (un clasificador puede «aprender las plantillas»). Le añadimos ruido para que **no sea trivial**: **30 % de los textos sin tildes**, **8 % de las palabras con una errata**, **5 % de etiquetas invertidas** y una frase de relleno en la mitad de los casos. Como contrapeso, en la sección 6 probamos con **frases escritas a mano** y con **otro dominio** (electrónica), y la sección 8 ofrece un bloque **opcional** con datos reales.

**Convención del generador** (documentada para poder razonar sobre los errores): en las reseñas **mixtas** (*«X es bueno pero Y es malo»*) la etiqueta la decide la cláusula **posterior al «pero»**; en el **sarcasmo** aparece una palabra positiva con sentimiento negativo.

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas matplotlib seaborn scikit-learn scipy

import re
import time
import unicodedata
import warnings
from collections import Counter
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy
import sklearn
from scipy import sparse
from scipy.special import logsumexp
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.decomposition import TruncatedSVD
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, classification_report, ConfusionMatrixDisplay

warnings.filterwarnings("ignore")
SEED = 42
rng = np.random.default_rng(SEED)
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (9, 5)
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]

# ---- Banderas del cuaderno (ambas DESACTIVADAS: las secciones opcionales necesitan red y/o bibliotecas extra) ----
USAR_20NG = False            # True SOLO con red: descarga 20 Newsgroups (~14 MB) y repite la comparacion TF-IDF/Naive Bayes/regresion logistica en la seccion 8
USAR_TRANSFORMERS = False    # True SOLO con red y `pip install transformers`: usa un modelo BERT multilingue de sentimiento en la seccion 8
print(f"Entorno listo | numpy {np.__version__} | scipy {scipy.__version__} | scikit-learn {sklearn.__version__}")
print(f"USAR_20NG = {USAR_20NG} | USAR_TRANSFORMERS = {USAR_TRANSFORMERS}")
````

````python
# ============================================================
# 1. Corpus SINTETICO de resenas en espanol (plantillas + ruido), con semilla fija
# ============================================================
ASPECTOS = {
    "restaurante": ["la comida", "el servicio", "la atención", "el lugar", "el postre", "la sopa", "el café", "los precios", "la pizza", "el mesero"],
    "electrónica": ["la batería", "la pantalla", "el celular", "el sonido", "la cámara", "el envío", "el cargador", "los audífonos", "el portátil", "la garantía"],
}
POS_GEN = ["excelente", "delicioso", "fantástico", "maravilloso", "genial", "recomendable", "impecable", "perfecto", "espectacular", "increíble"]
NEG_GEN = ["pésimo", "horrible", "malísimo", "decepcionante", "terrible", "lamentable", "deficiente", "asqueroso", "desastroso", "mediocre"]
POS_DOM = {"restaurante": ["sabroso", "fresco", "abundante", "acogedor"], "electrónica": ["rápido", "nítido", "potente", "duradero"]}
NEG_DOM = {"restaurante": ["frío", "salado", "caro", "ruidoso"], "electrónica": ["lento", "defectuoso", "frágil", "inservible"]}
RELLENO = ["fui con mi familia el sábado", "lo compré la semana pasada", "llegamos temprano", "pedí lo de siempre", "me lo recomendó un amigo",
           "estuve ahí dos veces", "es la tercera vez que voy", "lo probé por primera vez", "estaba en promoción"]
INTENS = ["", "", "muy", "realmente", "bastante"]


def ruido_texto(t, rng, p_acento=0.3, p_typo=0.08):
    """Ruido realista: a veces SIN tildes, minusculas/mayusculas al azar y algun error tipografico (letras intercambiadas)."""
    if rng.random() < p_acento:
        t = "".join(c for c in unicodedata.normalize("NFD", t) if unicodedata.category(c) != "Mn")
    palabras = t.split()
    for i, w in enumerate(palabras):
        if len(w) > 4 and rng.random() < p_typo:
            j = int(rng.integers(0, len(w) - 1)); palabras[i] = w[:j] + w[j + 1] + w[j] + w[j + 2:]
    t = " ".join(palabras)
    return t.capitalize() if rng.random() < 0.5 else t


def generar_resena(rng, dominio="restaurante", p_dom=0.35):
    """Devuelve (texto, etiqueta, tipo) con etiqueta 1 = positiva, 0 = negativa. `tipo` documenta la plantilla (util para analizar errores)."""
    pos = lambda: POS_GEN[rng.integers(len(POS_GEN))] if rng.random() > p_dom else POS_DOM[dominio][rng.integers(4)]
    neg = lambda: NEG_GEN[rng.integers(len(NEG_GEN))] if rng.random() > p_dom else NEG_DOM[dominio][rng.integers(4)]
    asp = lambda: ASPECTOS[dominio][rng.integers(10)]
    it = lambda: INTENS[rng.integers(len(INTENS))]
    y = int(rng.random() < 0.5)
    tipo = rng.choice(["simple", "simple", "simple", "negacion", "negacion_doble", "mixta", "sarcasmo"], p=[.25, .2, .15, .15, .08, .12, .05])
    if tipo == "simple":
        palabra = pos() if y else neg()
        t = rng.choice([f"{asp()} es {it()} {palabra}", f"{asp()}: {palabra}", f"me pareció {palabra} {asp()}", f"{palabra} {asp()}, {'volveré' if y else 'no vuelvo'}"])
    elif tipo == "negacion":                                                             # "no es nada bueno" -> negativa ; "no me gustó" -> negativa
        t = f"{asp()} no es nada {pos()}" if not y else f"{asp()} no es {neg()}, {pos()} en general"
        if not y and rng.random() < 0.4:
            t = f"no me gustó {asp()}, no es {pos()}"
    elif tipo == "negacion_doble":                                                       # "no está mal" -> positiva leve
        t = f"{asp()} no está nada mal, {pos()}" if y else f"no puedo decir que {asp()} sea {pos()}, más bien {neg()}"
    elif tipo == "mixta":                                                                # la clausula tras 'pero' decide la etiqueta (convencion del generador)
        a, b = (pos(), neg()) if not y else (neg(), pos())
        t = f"{asp()} es {a} pero {asp()} es {b}"
    else:                                                                                # sarcasmo: palabra POSITIVA, sentimiento NEGATIVO (solo en negativas)
        y = 0
        t = rng.choice([f"qué {pos()} que {asp()} tarde dos horas en llegar", f"claro, {asp()} es {pos()}... si te gusta esperar y que llegue frío",
                        f"{pos()} {asp()}, justo lo que quería: gastar y quedar decepcionado"])
    if rng.random() < 0.5:
        t = f"{RELLENO[rng.integers(len(RELLENO))]}, {t}"
    return ruido_texto(str(t), rng), y, str(tipo)


def generar_corpus(n, dominio, semilla, p_dom=0.35, ruido_etiqueta=0.05):
    r = np.random.default_rng(semilla); filas = [generar_resena(r, dominio, p_dom) for _ in range(n)]
    df = pd.DataFrame(filas, columns=["texto", "y", "tipo"]); df["dominio"] = dominio
    flip = r.random(n) < ruido_etiqueta                                                  # ruido de etiquetas (como en la vida real: anotadores que se equivocan)
    df.loc[flip, "y"] = 1 - df.loc[flip, "y"]
    return df


df_rest = generar_corpus(4000, "restaurante", semilla=1)
df_elec = generar_corpus(1000, "electrónica", semilla=2)                                 # OTRO dominio: sirve para medir el desplazamiento de dominio
print(f"Corpus de restaurantes: {len(df_rest)} reseñas (positivas: {df_rest.y.mean():.1%}) | de electrónica: {len(df_elec)} (positivas: {df_elec.y.mean():.1%})")
print("Longitud media (palabras):", round(df_rest.texto.str.split().str.len().mean(), 1), "| ruido de etiquetas: 5 % | 30 % de los textos SIN tildes | 8 % de palabras con una errata\n")
display(df_rest.sample(8, random_state=3)[["texto", "y", "tipo"]].reset_index(drop=True))
print("Tipos de plantilla:", df_rest.tipo.value_counts(normalize=True).round(2).to_dict())
assert 0.4 < df_rest.y.mean() < 0.6 and df_rest.texto.str.len().min() > 5
````

La reseña 1 («impecable» con etiqueta 0) es una de las **etiquetas ruidosas**: ningún modelo la acertará, y eso fija un **techo** de ≈ 95 % de exactitud.

## 2. Tokenización y Normalización 🔤

Un **token** es la unidad que el modelo cuenta (aquí, palabras). La **tokenización** parte el texto; la **normalización** reduce variantes superficiales (mayúsculas, tildes, ...). Cada decisión **cambia el vocabulario** y, con él, el modelo:

| Paso | Qué hace | Riesgo en español |
|---|---|---|
| **Minúsculas** | «EXCELENTE» = «excelente» | Casi sin riesgo (se pierde el énfasis de las mayúsculas) |
| **Quitar tildes** | «está» → «esta», «más» → «mas» | **Fusiona palabras distintas** (*él*/*el*, *sí*/*si*, *qué*/*que*); pero da robustez ante textos escritos sin tildes |
| **Quitar *stopwords*** | Elimina palabras muy frecuentes («el», «de»...) | Las listas genéricas suelen incluir **«no», «nada», «sin»**: **invierten el sentimiento** |
| ***Stemming* / lematización** | «gustaron» → «gust» | Reduce el vocabulario; puede **mezclar** significados |
| **Erratas y alargamientos** | «buenooo», «lguar» | Cada variante es un token raro; mitigables con *n*-gramas de **caracteres** o *subpalabras* (BPE) |

Los modelos de lenguaje modernos usan **subpalabras** (*byte-pair encoding*, *WordPiece*) para no depender de un vocabulario cerrado ni de la limpieza manual. Aquí usamos palabras para ver los mecanismos.

````python
# ============================================================
# 2. Tokenizacion y normalizacion (con trampas del espanol)
# ============================================================
def quitar_tildes(t):
    """'está' -> 'esta', 'canción' -> 'cancion' (descompone en letra + marca y elimina las marcas; la 'ñ' tambien pierde su virgulilla: ojo)."""
    return "".join(c for c in unicodedata.normalize("NFD", t) if unicodedata.category(c) != "Mn")


def tokenizar(texto, sin_tildes=False):
    """Minusculas -> (opcional) sin tildes -> palabras formadas por letras. Descarta puntuacion y digitos."""
    t = texto.lower()
    if sin_tildes:
        t = quitar_tildes(t)
    return re.findall(r"[a-záéíóúüñ]+", t) if not sin_tildes else re.findall(r"[a-zn]+", t)


ejemplo = "¡La comida estaba EXCELENTE! Pero el servicio… no está nada bien, y el café ¿cuánto cuesta? 3 mil pesos."
print("Texto:      ", ejemplo)
print("Tokens:     ", tokenizar(ejemplo))
print("Sin tildes: ", tokenizar(ejemplo, sin_tildes=True))

# --- Trampas del espanol: las tildes distinguen palabras; quitarlas FUSIONA significados ---
for a, b in [("está", "esta"), ("él", "el"), ("más", "mas"), ("sí", "si"), ("qué", "que"), ("sólo", "solo")]:
    assert quitar_tildes(a) == b
print("\nQuitar tildes fusiona: " + ", ".join(f"'{a}' y '{b}'" for a, b in [("está", "esta"), ("él", "el"), ("más", "mas"), ("sí", "si"), ("qué", "que")]) + " → ganamos robustez ante textos sin tildes, perdemos matices.")

# --- Vocabulario segun el preprocesamiento: la normalizacion reduce (o no) el espacio de rasgos ---
corpus = df_rest.texto.tolist()
variantes = {"solo minúsculas": lambda t: tokenizar(t), "minúsculas + sin tildes": lambda t: tokenizar(t, True)}
for nombre, f in variantes.items():
    v = Counter(w for t in corpus for w in f(t))
    print(f"Vocabulario ({nombre}): {len(v)} palabras distintas; {sum(1 for c in v.values() if c == 1)} aparecen UNA sola vez (en su mayoría erratas).")

# --- Stopwords: lista corta en espanol. OJO con 'no': quitarla invierte el sentido ---
STOP_BASICAS = {"el", "la", "los", "las", "un", "una", "de", "del", "y", "o", "en", "con", "por", "para", "es", "que", "me", "se", "lo", "al", "mi", "fue", "esta", "está", "muy"}
NEGACIONES = {"no", "nunca", "ni", "nada", "jamás", "tampoco", "sin"}
print("\nTrampa clásica: las listas genéricas de stopwords en español suelen incluir 'no', 'nada', 'sin'... que SON la señal de la negación.")
frase = "no me gustó la comida"
print(f"  '{frase}' sin stopwords (lista que incluye 'no') → {[w for w in tokenizar(frase) if w not in STOP_BASICAS | NEGACIONES]}  (¡parece positiva!)")
print(f"  conservando las negaciones                      → {[w for w in tokenizar(frase) if w not in STOP_BASICAS]}")
assert "no" not in [w for w in tokenizar(frase) if w not in STOP_BASICAS | NEGACIONES]
print("\nRegla práctica: en análisis de sentimiento, NO elimines las negaciones; mide el efecto de cada paso de limpieza con validación, no por intuición.")
````

**Lectura.** Quitar tildes **reduce el vocabulario** (573 → 519 palabras) y fusiona pares como *está*/*esta*; las palabras que aparecen **una sola vez** son sobre todo **erratas** que no aportan. Y la advertencia importante: «no me gustó la comida» **sin** la palabra «no» queda como `['gustó', 'comida']`, es decir, **parece positiva**. En la sección 6 **medimos** cuánto daña esto.

## 3. Bolsa de Palabras y TF-IDF desde Cero 🧮

**Bolsa de palabras** (*bag of words*): cada documento $d$ es un vector de **conteos** $n_{d,t}$ (cuántas veces aparece la palabra $t$), **ignorando el orden**. Resulta una matriz $X\in\mathbb N^{n\times V}$ **muy dispersa** (la mayoría de las entradas son 0).

**TF-IDF** pondera cada conteo por **qué tan específica** es la palabra: las que aparecen en casi todos los documentos («el», «la») informan poco. Con la formulación suavizada de `scikit-learn`:

$$\operatorname{idf}(t)=\ln\frac{1+n}{1+\operatorname{df}(t)}+1,\qquad \operatorname{tfidf}(d,t)=n_{d,t}\cdot\operatorname{idf}(t),\qquad \tilde{\mathbf x}_d=\frac{\operatorname{tfidf}(d,\cdot)}{\|\operatorname{tfidf}(d,\cdot)\|_2},$$

donde $n$ es el número de documentos y $\operatorname{df}(t)$ cuántos contienen $t$. El «+1» evita que una palabra presente en **todos** los documentos tenga peso cero; el «+1» dentro del logaritmo (suavizado) evita dividir por cero con palabras no vistas. La **normalización L2** pone todos los documentos a la misma norma, así el largo del texto no domina. Implementamos las tres funciones y las **verificamos con `assert`** contra `scikit-learn` (mismo tokenizador, mismo vocabulario en el mismo orden).

````python
# ============================================================
# 3. Bolsa de palabras y TF-IDF desde cero, verificados contra scikit-learn
# ============================================================
def construir_vocabulario(docs_tokens):
    """Vocabulario ordenado alfabeticamente (igual criterio que scikit-learn) -> dict palabra -> columna."""
    return {w: j for j, w in enumerate(sorted({w for d in docs_tokens for w in d}))}


def matriz_conteos(docs_tokens, vocab):
    """Bolsa de palabras: X[i, j] = numero de veces que la palabra j aparece en el documento i (matriz dispersa CSR)."""
    filas, cols, vals = [], [], []
    for i, d in enumerate(docs_tokens):
        for w, c in Counter(d).items():
            if w in vocab:
                filas.append(i); cols.append(vocab[w]); vals.append(c)
    return sparse.csr_matrix((vals, (filas, cols)), shape=(len(docs_tokens), len(vocab)), dtype=np.float64)


def tfidf_desde_cero(X_conteos, df=None):
    """TF-IDF con la misma formula (suavizada) de scikit-learn: idf_t = ln((1+n)/(1+df_t)) + 1 ; fila = tf * idf ; normalizacion L2 por documento."""
    n = X_conteos.shape[0]
    if df is None:
        df = np.asarray((X_conteos > 0).sum(axis=0)).ravel()
    idf = np.log((1 + n) / (1 + df)) + 1
    X = X_conteos.multiply(idf).tocsr()
    norma = np.sqrt(np.asarray(X.multiply(X).sum(axis=1)).ravel())
    return sparse.diags(1.0 / norma) @ X, idf


tokens_rest = [tokenizar(t) for t in df_rest.texto]
vocab = construir_vocabulario(tokens_rest)
C = matriz_conteos(tokens_rest, vocab)
T, idf = tfidf_desde_cero(C)
print(f"Matriz documento-término: {C.shape} | no ceros: {C.nnz} ({C.nnz / np.prod(C.shape):.1%}, dispersa) | vocabulario: {len(vocab)} palabras")

# --- Verificacion contra CountVectorizer y TfidfVectorizer (mismo tokenizador, sin minusculas ni patron propios) ---
cv = CountVectorizer(tokenizer=tokenizar, lowercase=False, token_pattern=None)
C_sk = cv.fit_transform(df_rest.texto)
tv = TfidfVectorizer(tokenizer=tokenizar, lowercase=False, token_pattern=None, smooth_idf=True, norm="l2")
T_sk = tv.fit_transform(df_rest.texto)
assert list(cv.get_feature_names_out()) == sorted(vocab, key=vocab.get)                  # mismo vocabulario y mismo orden
assert np.array_equal(C.toarray(), C_sk.toarray())
assert np.allclose(idf, tv.idf_)
assert np.allclose(T.toarray(), T_sk.toarray())
assert np.allclose(np.sqrt(np.asarray(T.multiply(T).sum(axis=1)).ravel()), 1.0)         # cada fila tiene norma 1
print("OK: conteos == CountVectorizer, idf == TfidfVectorizer.idf_ y TF-IDF == TfidfVectorizer (suavizado + normalización L2), sobre 4 000 documentos.")

# --- Que pesa mas: palabras raras (idf alto) o comunes (idf bajo) ---
orden = np.argsort(idf)
inv = {j: w for w, j in vocab.items()}
print("\nidf más BAJO (palabras en casi todos los documentos):", [(inv[j], round(float(idf[j]), 2)) for j in orden[:6]])
print("idf más ALTO (palabras raras, casi siempre erratas):   ", [(inv[j], round(float(idf[j]), 2)) for j in orden[-4:]])

# --- Ejemplo a mano: 3 documentos, para ver el calculo completo ---
mini = [["buen", "servicio"], ["mal", "servicio"], ["buen", "buen", "precio"]]
vm = construir_vocabulario(mini); Cm = matriz_conteos(mini, vm); Tm, idfm = tfidf_desde_cero(Cm)
print("\nMini-ejemplo   vocabulario:", list(vm), "| df =", np.asarray((Cm > 0).sum(0)).ravel(), "| idf =", np.round(idfm, 3))
print("Documento 3 ('buen buen precio'): conteos", Cm.toarray()[2], "→ TF-IDF normalizado", np.round(Tm.toarray()[2], 3))
assert np.isclose(idfm[vm["servicio"]], np.log(4 / 3) + 1) and np.isclose(idfm[vm["precio"]], np.log(4 / 2) + 1)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.2))
frec = np.asarray(C.sum(0)).ravel()
axes[0].loglog(np.arange(1, len(frec) + 1), np.sort(frec)[::-1], "o", ms=3, color=PALETA[0]); axes[0].set_xlabel("Rango de la palabra"); axes[0].set_ylabel("Frecuencia")
axes[0].set_title("Frecuencia de palabras (ley de Zipf: pocas muy comunes, muchas raras)", fontweight="bold", fontsize=9)
dfv = np.asarray((C > 0).sum(0)).ravel()
axes[1].scatter(dfv, idf, s=8, color=PALETA[1]); axes[1].set_xscale("log"); axes[1].set_xlabel("Documentos que contienen la palabra (df)"); axes[1].set_ylabel("idf suavizado")
axes[1].set_title("idf = ln((1+n)/(1+df)) + 1 : decrece con df", fontweight="bold", fontsize=9)
plt.tight_layout(); plt.show()
````

---
**Lectura.** La matriz tiene **1.4 % de entradas no nulas**: por eso se guarda en formato disperso. Nuestras tres funciones **coinciden exactamente** con `CountVectorizer` y `TfidfVectorizer` (vocabulario, conteos, idf y la matriz completa con norma L2 por fila). Las palabras con idf más bajo son las de función («el», «la», «es», «no»: aparecen en casi todos los textos) y las de idf más alto, **erratas** vistas una sola vez. Observa también la curva de Zipf: **pocas palabras muy frecuentes y una larga cola de raras**, rasgo universal del lenguaje.

## 4. Naive Bayes Multinomial desde Cero 🎲

Para clasificar un documento $d$ (conteos $n_t$) en una clase $c$, el **teorema de Bayes** da $P(c\mid d)\propto P(c)\,P(d\mid c)$. El supuesto «ingenuo» (*naive*): **las palabras son independientes dada la clase**, y siguen una distribución **multinomial** con probabilidades $\theta_{c,t}$:

$$\log P(c\mid d)=\log\pi_c+\sum_{t=1}^{V} n_t\,\log\theta_{c,t}+\text{cte},\qquad \theta_{c,t}=\frac{N_{c,t}+\alpha}{N_c+\alpha V},$$

con $\pi_c$ la proporción de documentos de la clase, $N_{c,t}$ el conteo de $t$ en la clase $c$ y $N_c=\sum_t N_{c,t}$. Dos detalles **no son opcionales**:

- **Suavizado de Laplace** ($\alpha=1$): sin él, una palabra que **nunca** apareció en la clase $c$ tendría $\theta_{c,t}=0$ y **anularía** la probabilidad de toda la clase (el $\log 0=-\infty$).
- **Log-espacio**: un producto de cientos de probabilidades minúsculas **se desborda hacia 0** (*underflow*); la suma de logaritmos, no. Para normalizar se usa **log-sum-exp**.

El supuesto de independencia es **falso** (las palabras se correlacionan), y aun así el clasificador funciona razonablemente porque solo necesita que el **orden** de las clases sea correcto, no que las probabilidades sean fiables (de hecho, sus probabilidades suelen estar **mal calibradas**: ver el [Módulo 07](../07%20-%20Seleccion%20de%20Modelos%20y%20Optimizacion/00_Evaluacion_y_Seleccion_de_Modelos.ipynb)).

````python
# ============================================================
# 4. Naive Bayes multinomial desde cero (suavizado de Laplace, en log-espacio)
# ============================================================
class NaiveBayesMultinomial:
    """P(c | d) ∝ P(c) * prod_t theta_{c,t}^{n_t(d)}  ->  log P(c|d) = log pi_c + sum_t n_t(d) * log theta_{c,t} + cte.
    theta_{c,t} = (N_{c,t} + alpha) / (N_c + alpha * V): frecuencia de la palabra t en la clase c, suavizada para que ninguna probabilidad sea 0."""
    def __init__(self, alpha=1.0):
        self.alpha = alpha

    def fit(self, X, y):
        self.clases_ = np.unique(y)
        Nct = np.vstack([np.asarray(X[y == c].sum(axis=0)).ravel() for c in self.clases_])      # conteo de cada palabra por clase
        self.log_prior_ = np.log(np.array([(y == c).mean() for c in self.clases_]))
        self.log_theta_ = np.log(Nct + self.alpha) - np.log(Nct.sum(1, keepdims=True) + self.alpha * X.shape[1])
        return self

    def log_verosimilitud_conjunta(self, X):
        return np.asarray(X @ self.log_theta_.T) + self.log_prior_                              # (n_docs, n_clases)

    def predict_log_proba(self, X):
        J = self.log_verosimilitud_conjunta(X)
        return J - logsumexp(J, axis=1, keepdims=True)                                          # normalizar EN log-espacio (log-sum-exp)

    def predict(self, X):
        return self.clases_[np.argmax(self.log_verosimilitud_conjunta(X), axis=1)]


Xc_tr, Xc_te, y_tr, y_te, txt_tr, txt_te = train_test_split(C, df_rest.y.values, df_rest.texto.values, test_size=0.25, stratify=df_rest.y.values, random_state=0)
nb_mio = NaiveBayesMultinomial(alpha=1.0).fit(Xc_tr, y_tr)
nb_sk = MultinomialNB(alpha=1.0).fit(Xc_tr, y_tr)
assert np.allclose(nb_mio.log_theta_, nb_sk.feature_log_prob_) and np.allclose(nb_mio.log_prior_, nb_sk.class_log_prior_)
assert np.allclose(nb_mio.predict_log_proba(Xc_te), nb_sk.predict_log_proba(Xc_te))
assert np.array_equal(nb_mio.predict(Xc_te), nb_sk.predict(Xc_te))
print(f"OK: Naive Bayes multinomial desde cero == MultinomialNB (log θ, log π, log-probabilidades y predicciones). Exactitud de prueba: {accuracy_score(y_te, nb_mio.predict(Xc_te)):.3f}")

# --- Por que el suavizado de Laplace: una palabra vista solo en una clase anula (o hunde) toda la evidencia del resto ---
Nc0 = np.asarray(Xc_tr[y_tr == 0].sum(axis=0)).ravel(); Nc1 = np.asarray(Xc_tr[y_tr == 1].sum(axis=0)).ravel()
inv_cv = cv.get_feature_names_out()
solo_pos = np.where((Nc0 == 0) & (Nc1 > 0))[0]                                                  # palabras vistas SOLO en reseñas positivas (p. ej. erratas)
w_raro = int(solo_pos[0])
x_neg = cv.transform(["el servicio es pésimo y la comida horrible, malísimo, lamentable"])         # una reseña clarísimamente NEGATIVA...
x_neg = x_neg + sparse.csr_matrix(([1.0], ([0], [w_raro])), shape=x_neg.shape)                   # ...con UNA palabra que en el entrenamiento solo apareció en positivas
nb_sin = NaiveBayesMultinomial(alpha=1e-12).fit(Xc_tr, y_tr)                                    # (casi) sin suavizado
print(f"Palabra «{inv_cv[w_raro]}»: aparece {int(Nc1[w_raro])} veces en positivas y 0 en negativas.")
print("Casi sin suavizado (α≈0): log-verosimilitud [neg, pos] =", np.round(nb_sin.log_verosimilitud_conjunta(x_neg)[0], 1), "→ predice", ["negativa", "positiva"][int(nb_sin.predict(x_neg)[0])], "(¡una sola palabra rara decide!)")
print("Con Laplace (α=1):        log-verosimilitud [neg, pos] =", np.round(nb_mio.log_verosimilitud_conjunta(x_neg)[0], 1), "→ predice", ["negativa", "positiva"][int(nb_mio.predict(x_neg)[0])])
assert nb_sin.predict(x_neg)[0] == 1 and nb_mio.predict(x_neg)[0] == 0

# --- Por que el log-espacio: el producto de probabilidades se underflowea en documentos largos ---
doc_largo = Xc_tr[0] * 60                                                                       # un documento "largo": mismas palabras repetidas 60 veces
p_directo = np.exp(np.asarray(doc_largo @ nb_mio.log_theta_.T)).ravel() * np.exp(nb_mio.log_prior_)
print(f"\nProducto directo de probabilidades en un documento largo: {p_directo} (¡0.0 en ambas clases! underflow)")
print("En log-espacio la suma sigue siendo perfectamente comparable:", np.round(nb_mio.log_verosimilitud_conjunta(doc_largo)[0], 1))
assert (p_directo == 0).all() and np.isfinite(nb_mio.log_verosimilitud_conjunta(doc_largo)).all()

# --- Efecto de alpha ---
alfas = [0.01, 0.1, 0.5, 1, 2, 5]
acc_alfa = [accuracy_score(y_te, NaiveBayesMultinomial(a).fit(Xc_tr, y_tr).predict(Xc_te)) for a in alfas]
print("\nExactitud de prueba según alpha:", dict(zip(alfas, np.round(acc_alfa, 3))))
````

---
**Lectura.** Nuestro Naive Bayes **coincide exactamente** con `MultinomialNB` (log-probabilidades por palabra y por clase, probabilidades posteriores y predicciones) y alcanza ≈ 0.80 de exactitud. Las dos demostraciones muestran **por qué** el suavizado y el log-espacio importan: sin suavizado, **una sola palabra rara** (vista únicamente en positivas) **invierte** la predicción de una reseña claramente negativa, mientras que con $\alpha=1$ la penalización es moderada y la predicción sigue siendo negativa; y en un documento largo el producto directo da **0.0 en ambas clases** (indistinguibles), mientras que en log-espacio la diferencia sigue siendo clara. Casi todos los valores de $\alpha$ dan lo mismo aquí (≈ 0.79–0.81): con un vocabulario pequeño y muchos datos, el suavizado importa poco para la exactitud, pero sí para la **robustez**.

## 5. Regresión Logística sobre TF-IDF y Evaluación 📊

El **modelo lineal** estándar para texto es la regresión logística ([Módulo 01](../01%20-%20Modelos%20Lineales%20Supervisados/02_Regresion_Logistica.ipynb)) sobre los vectores TF-IDF: cada palabra recibe un **coeficiente** (su peso en los *log-odds* de «positiva»), lo que la hace **interpretable**. A diferencia de Naive Bayes, **no asume independencia** y **ajusta los pesos conjuntamente**. Además, usar **1-2-gramas** (pares de palabras consecutivas) recupera un poco del **orden**: «no es», «no me gustó» pasan a ser rasgos propios, lo que ayuda con la **negación**.

**Evaluación correcta:** el vectorizador **aprende** vocabulario e idf **de los datos**; si se ajusta antes de dividir, hay **fuga** (el idf del conjunto de prueba contamina el entrenamiento). Por eso se mete dentro de un `Pipeline`, que lo **reajusta en cada pliegue** (Módulos [00](../00%20-%20Introduccion%20al%20Machine%20Learning/03_Importancia_del_Preprocesamiento.ipynb) y [03](../03%20-%20Metricas%20y%20Ajuste%20de%20Hiperparametros/02_Validacion_Cruzada_y_GridSearch.ipynb)). Las métricas son las del [Módulo 03](../03%20-%20Metricas%20y%20Ajuste%20de%20Hiperparametros/00_Metricas_de_Clasificacion.ipynb): matriz de confusión, precisión, recobrado y F1.

````python
# ============================================================
# 5. Regresion logistica sobre TF-IDF y evaluacion (el vectorizador DENTRO del Pipeline)
# ============================================================
def evaluar(modelo, Xtr, ytr, Xte, yte, nombre):
    modelo.fit(Xtr, ytr); p = modelo.predict(Xte)
    return {"modelo": nombre, "exactitud": accuracy_score(yte, p), "F1 (positiva)": f1_score(yte, p), "F1 macro": f1_score(yte, p, average="macro")}


txt_tr, txt_te = list(txt_tr), list(txt_te)
tok = lambda t: tokenizar(t)
modelos = {
    "Naive Bayes + bolsa de palabras": make_pipeline(CountVectorizer(tokenizer=tok, lowercase=False, token_pattern=None), MultinomialNB()),
    "Reg. logística + TF-IDF (1-gramas)": make_pipeline(TfidfVectorizer(tokenizer=tok, lowercase=False, token_pattern=None), LogisticRegression(max_iter=1000)),
    "Reg. logística + TF-IDF (1-2-gramas)": make_pipeline(TfidfVectorizer(tokenizer=tok, lowercase=False, token_pattern=None, ngram_range=(1, 2)), LogisticRegression(max_iter=1000)),
}
filas = [evaluar(m, txt_tr, y_tr, txt_te, y_te, n) for n, m in modelos.items()]
tabla = pd.DataFrame(filas).set_index("modelo")
display(tabla.round(3))

# Validacion cruzada CORRECTA: el vectorizador (que aprende vocabulario e idf) se reajusta en cada pliegue gracias al Pipeline (Modulos 00 y 03)
cv5 = StratifiedKFold(5, shuffle=True, random_state=0)
for n, m in modelos.items():
    s = cross_val_score(m, txt_tr, y_tr, cv=cv5, scoring="f1")
    print(f"  F1 por validación cruzada (5 pliegues) — {n}: {s.mean():.3f} ± {s.std():.3f}")

mejor = modelos["Reg. logística + TF-IDF (1-2-gramas)"]; p_te = mejor.predict(txt_te)
print("\n" + classification_report(y_te, p_te, target_names=["negativa", "positiva"], digits=3))
techo = (1 - 0.05)                                                                                  # con 5 % de etiquetas invertidas al azar, ni un oraculo supera ~95 % de exactitud
print(f"Techo teórico por el ruido de etiquetas ≈ {techo:.0%}: una exactitud cercana a él indica que el modelo aprendió casi todo lo que había (y que el corpus sintético es FÁCIL).")

fig, axes = plt.subplots(1, 2, figsize=(14, 4.8))
ConfusionMatrixDisplay(confusion_matrix(y_te, p_te), display_labels=["negativa", "positiva"]).plot(ax=axes[0], cmap="Blues", colorbar=False)
axes[0].set_title("Matriz de confusión (prueba)", fontweight="bold", fontsize=10); axes[0].grid(False)
vec = mejor.named_steps["tfidfvectorizer"]; lr = mejor.named_steps["logisticregression"]
nombres = np.array(vec.get_feature_names_out()); coef = lr.coef_[0]; idx = np.r_[np.argsort(coef)[:10], np.argsort(coef)[-10:]]
axes[1].barh(nombres[idx], coef[idx], color=[PALETA[1] if c < 0 else PALETA[2] for c in coef[idx]])
axes[1].set_xlabel("Coeficiente (log-odds de «positiva»)"); axes[1].set_title("Las 10 palabras/bigramas más negativos y más positivos", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()

# --- Analisis de errores por tipo de plantilla ---
df_te = pd.DataFrame({"texto": txt_te, "y": y_te, "pred": p_te}).merge(df_rest[["texto", "tipo"]].drop_duplicates("texto"), on="texto", how="left")
err = df_te.groupby("tipo").apply(lambda g: pd.Series({"n": len(g), "exactitud": (g.y == g.pred).mean()}))
display(err.round(3))
assert tabla["exactitud"].min() > 0.7 and tabla["exactitud"].max() > 0.9
````

---
**Lectura honesta.** (1) **Regresión logística supera con claridad a Naive Bayes** (≈ 0.89 con 1-gramas y ≈ **0.93** con 1-2-gramas, frente a ≈ 0.80): la negación y las construcciones con «pero» **necesitan pares de palabras**, y Naive Bayes, que suma evidencia palabra a palabra, no puede captarlas bien. (2) Pasar de 1-gramas a 1-2-gramas aporta **≈ 4 puntos**, el mayor salto de la tabla. (3) La validación cruzada (≈ 0.92 de F1) concuerda con la prueba (0.924): no hay señales de fuga ni de azar afortunado. (4) El error se concentra en las reseñas **mixtas** (≈ 0.86), y los coeficientes son legibles: los más negativos son palabras de crítica y «no vuelvo»; los más positivos, los elogios. (5) Ojo: aun el **sarcasmo** se acierta al ≈ 0.91, **no** porque el modelo «entienda» ironía, sino porque en nuestro generador el sarcasmo usa **frases fijas** («si te gusta esperar...») que el modelo memoriza. En datos reales **no** sería así: es la mayor limitación de un corpus sintético.

## 6. Análisis de Sentimiento: Léxico frente a Aprendizaje Supervisado, y las Trampas ⚠️

Hay dos grandes familias:

| Enfoque | Cómo funciona | Ventajas | Desventajas |
|---|---|---|---|
| **Basado en léxico** | Un diccionario de palabras con polaridad (+1/−1); se suman las del texto, con reglas (p. ej. invertir tras una negación) | **No necesita etiquetas**; transparente; estable entre dominios | Cobertura limitada; reglas frágiles; ignora ironía y contexto |
| **Aprendizaje supervisado** | Un clasificador aprende los pesos de las palabras desde ejemplos etiquetados | Se **adapta a los datos**; mucho más preciso **en su dominio** | Necesita etiquetas; **no se transfiere** a otro dominio; hereda sesgos |

Las **trampas** que medimos aquí: **negación** («no está nada mal»), **sarcasmo**, **desplazamiento de dominio** (entrenamos con restaurantes y probamos con electrónica; el vocabulario de polaridad cambia: «frío» es malo en una sopa y neutro en un portátil), **preprocesamiento destructivo** (quitar «no»), **desbalance** y **erratas/tildes**. La pregunta de fondo es la del [Módulo 08, cuaderno 00](../08%20-%20Temas%20Avanzados%20de%20ML/00_Introduccion_Cap3_y_Conceptos_Avanzados.ipynb): ¿el modelo se va a usar sobre los **mismos datos** con los que se validó?

````python
# ============================================================
# 6a. Analisis de sentimiento: lexico frente a aprendizaje supervisado, y el desplazamiento de dominio
# ============================================================
# Lexico GENERICO escrito a mano (no mira los datos): palabras de polaridad +1 / -1, con ventana de negacion de 3 palabras
LEX_POS = {"excelente", "delicioso", "fantástico", "maravilloso", "genial", "recomendable", "impecable", "perfecto", "espectacular", "increíble", "bueno", "buena", "bien", "encantó", "recomiendo", "mejor", "bonito", "amable", "rico"}
LEX_NEG = {"pésimo", "horrible", "malísimo", "decepcionante", "terrible", "lamentable", "deficiente", "asqueroso", "desastroso", "mediocre", "malo", "mala", "mal", "peor", "tarde", "grosero", "decepcionado"}
NEGADORES = {"no", "nunca", "ni", "tampoco", "sin"}


def puntaje_lexico(texto, con_negacion=True, ventana=3):
    toks = tokenizar(texto); total, cubiertas, resto_neg = 0, 0, 0
    for tok_ in toks:
        pol = (tok_ in LEX_POS) - (tok_ in LEX_NEG)
        if pol:
            cubiertas += 1
            total += -pol if (con_negacion and resto_neg > 0) else pol
        resto_neg = ventana if tok_ in NEGADORES else max(resto_neg - 1, 0)
    return total, cubiertas


def predecir_lexico(textos, con_negacion=True):
    out = [puntaje_lexico(t, con_negacion) for t in textos]
    return np.array([int(s >= 0) for s, _ in out]), np.mean([c > 0 for _, c in out])           # empates y sin cobertura -> «positiva»


# Modelo supervisado (entrenado SOLO con restaurantes)
Xt, Xv, yt, yv = train_test_split(df_rest.texto.values, df_rest.y.values, test_size=0.25, stratify=df_rest.y.values, random_state=0)
lr_pipe = make_pipeline(TfidfVectorizer(tokenizer=tok, lowercase=False, token_pattern=None, ngram_range=(1, 2)), LogisticRegression(max_iter=1000)).fit(Xt, yt)
test_rest = pd.DataFrame({"texto": Xv, "y": yv}).merge(df_rest[["texto", "tipo"]].drop_duplicates("texto"), on="texto", how="left")

filas = []
for nombre_dom, dfx in [("restaurantes (mismo dominio)", test_rest), ("electrónica (OTRO dominio)", df_elec)]:
    for nombre, pred_f in [("Léxico sin negación", lambda t: predecir_lexico(t, False)[0]), ("Léxico con negación", lambda t: predecir_lexico(t, True)[0]), ("Reg. logística TF-IDF 1-2-gramas", lambda t: lr_pipe.predict(t))]:
        p = pred_f(dfx.texto.tolist())
        filas.append({"dominio": nombre_dom, "método": nombre, "exactitud": accuracy_score(dfx.y, p), "F1 macro": f1_score(dfx.y, p, average="macro")})
res_dom = pd.DataFrame(filas).pivot(index="método", columns="dominio", values="exactitud")
display(res_dom.round(3))
cob = predecir_lexico(df_elec.texto.tolist())[1]
print(f"Cobertura del léxico (documentos con al menos una palabra del léxico) en electrónica: {cob:.0%}")

# --- Trampas: exactitud del modelo supervisado por tipo de plantilla y por dominio ---
df_elec["pred_lr"] = lr_pipe.predict(df_elec.texto.tolist()); df_elec["pred_lex"] = predecir_lexico(df_elec.texto.tolist())[0]
test_rest["pred_lr"] = lr_pipe.predict(test_rest.texto.tolist()); test_rest["pred_lex"] = predecir_lexico(test_rest.texto.tolist())[0]
por_tipo = pd.concat({"LR restaurantes": test_rest.groupby("tipo").apply(lambda g: (g.y == g.pred_lr).mean()), "LR electrónica": df_elec.groupby("tipo").apply(lambda g: (g.y == g.pred_lr).mean()),
                      "Léxico restaurantes": test_rest.groupby("tipo").apply(lambda g: (g.y == g.pred_lex).mean()), "Léxico electrónica": df_elec.groupby("tipo").apply(lambda g: (g.y == g.pred_lex).mean())}, axis=1)
print("\nExactitud por tipo de reseña (negación, mixta, sarcasmo...):")
display(por_tipo.round(3))

# --- Ablacion del preprocesamiento: que pasa al «limpiar» como en muchos tutoriales ---
variantes = {
    "Mínimo: minúsculas": lambda t: tokenizar(t),
    "Quitar stopwords (lista que incluye 'no')": lambda t: [w for w in tokenizar(t) if w not in STOP_BASICAS | NEGACIONES],
    "Quitar stopwords pero CONSERVAR negaciones": lambda t: [w for w in tokenizar(t) if w not in STOP_BASICAS],
    "Quitar tildes": lambda t: tokenizar(t, True),
}
idx_neg = test_rest.tipo.isin(["negacion", "negacion_doble"]).values
filas = []
for nombre, f in variantes.items():
    m = make_pipeline(TfidfVectorizer(tokenizer=f, lowercase=False, token_pattern=None, ngram_range=(1, 2)), LogisticRegression(max_iter=1000)).fit(Xt, yt)
    p = m.predict(Xv)
    sin_tilde = [quitar_tildes(t) for t in Xv]                                                  # robustez: la MISMA prueba pero sin tildes
    filas.append({"preprocesamiento": nombre, "exactitud (prueba)": accuracy_score(yv, p), "solo reseñas con negación": accuracy_score(yv[idx_neg], p[idx_neg]), "prueba sin tildes": accuracy_score(yv, m.predict(sin_tilde))})
abl = pd.DataFrame(filas).set_index("preprocesamiento")
display(abl.round(3))
assert abl.loc["Quitar stopwords (lista que incluye 'no')", "solo reseñas con negación"] < abl.loc["Mínimo: minúsculas", "solo reseñas con negación"] - 0.05
````

````python
# ============================================================
# 6b. Desbalance de clases y frases «escritas a mano»
# ============================================================
# --- (1) Desbalance: solo el 8 % de las reseñas son positivas (p. ej. quejas frente a elogios). La exactitud engaña (Modulos 03 y 08) ---
rng_d = np.random.default_rng(0)
idx_pos = np.where(yt == 1)[0]; idx_neg_ = np.where(yt == 0)[0]
sel = np.r_[idx_neg_, rng_d.choice(idx_pos, size=int(0.08 / 0.92 * len(idx_neg_)), replace=False)]
Xd, yd = Xt[sel], yt[sel]
pos_v = np.where(yv == 1)[0]; neg_v = np.where(yv == 0)[0]
sel_v = np.r_[neg_v, rng_d.choice(pos_v, size=int(0.08 / 0.92 * len(neg_v)), replace=False)]
Xdv, ydv = Xv[sel_v], yv[sel_v]
filas = []
for nombre, cw in [("Sin corregir", None), ("class_weight='balanced'", "balanced")]:
    m = make_pipeline(TfidfVectorizer(tokenizer=tok, lowercase=False, token_pattern=None, ngram_range=(1, 2)), LogisticRegression(max_iter=1000, class_weight=cw)).fit(Xd, yd)
    p = m.predict(Xdv)
    filas.append({"entrenamiento": nombre, "exactitud": accuracy_score(ydv, p), "F1 clase minoritaria": f1_score(ydv, p), "recobrado minoritaria": ((p == 1) & (ydv == 1)).sum() / (ydv == 1).sum()})
filas.insert(0, {"entrenamiento": "Siempre «negativa»", "exactitud": (ydv == 0).mean(), "F1 clase minoritaria": 0.0, "recobrado minoritaria": 0.0})
display(pd.DataFrame(filas).set_index("entrenamiento").round(3))
print(f"Con {ydv.mean():.0%} de positivas, «siempre negativa» ya da {(ydv == 0).mean():.0%} de exactitud: usa F1/recobrado de la clase de interés, PR-AUC y pesos o umbral (Módulo 08, cuaderno 03).\n")

# --- (2) Frases escritas a mano (NO salen de las plantillas): una prueba de realidad ---
A_MANO = [("El ceviche estaba riquísimo y el personal muy atento.", 1), ("Una experiencia para olvidar: se equivocaron con el pedido y nadie se disculpó.", 0),
          ("Superó mis expectativas, lo volvería a comprar.", 1), ("Pagué por un producto nuevo y llegó rayado.", 0), ("Súper recomendado, buen precio y entrega a tiempo.", 1),
          ("No es el peor lugar de la ciudad, pero no volvería.", 0), ("Qué bonito detalle dejarnos esperando una hora.", 0), ("Nada que envidiarle a los mejores restaurantes.", 1),
          ("El celular se calienta muchísimo y la batería dura poco.", 0), ("La pantalla es nítida y el sonido espectacular.", 1), ("Tardaron una eternidad y la comida llegó helada.", 0),
          ("Atención de lujo, repetiría sin dudarlo.", 1), ("Me cobraron de más y encima fueron groseros.", 0), ("Está bien, pero esperaba más por ese precio.", 0),
          ("No estuvo mal para el precio.", 1), ("Todo estuvo perfecto, gracias.", 1)]
txt_m, y_m = [t for t, _ in A_MANO], np.array([y for _, y in A_MANO])
nb_pipe = make_pipeline(CountVectorizer(tokenizer=tok, lowercase=False, token_pattern=None), MultinomialNB()).fit(Xt, yt)
pred = pd.DataFrame({"frase": txt_m, "real": y_m, "NB": nb_pipe.predict(txt_m), "LR": lr_pipe.predict(txt_m), "léxico": predecir_lexico(txt_m)[0]})
pred["¿LR acierta?"] = np.where(pred.LR == pred.real, "sí", "NO")
pd.set_option("display.max_colwidth", 70)
display(pred)
print(f"Frases con al menos una palabra del léxico: {np.mean([puntaje_lexico(t)[1] > 0 for t in txt_m]):.0%} | sentimiento real positivo: {y_m.mean():.0%} (el léxico predice «positiva» cuando no tiene evidencia)")
print("Aciertos sobre 16 frases (¡muestra diminuta, solo ilustrativa!):", {k: int((pred[k] == pred.real).sum()) for k in ["NB", "LR", "léxico"]})
print("Los modelos se entrenaron con plantillas: con frases reales fallan donde el vocabulario o la estructura no estaban en el corpus sintético. Por eso las métricas del corpus sintético son OPTIMISTAS.")
````

## 7. Representaciones Densas: de LSA a los *Embeddings* 🧠

La bolsa de palabras trata «excelente» y «genial» como **columnas sin relación alguna** (la distancia entre ambas es la misma que con «pizza»). Las **representaciones densas** ponen cada palabra o documento en un espacio de **pocas dimensiones** donde **la cercanía significa algo**. Todas parten de la **hipótesis distribucional**: *«una palabra se caracteriza por las compañías que mantiene»* (Firth, 1957) *(citado de memoria, verificar)*.

**LSA (análisis semántico latente)** (Deerwester et al., 1990): la **SVD truncada** de la matriz documento-término, $X\approx U_k\Sigma_k V_k^\top$ (la mejor aproximación de rango $k$, teorema de Eckart–Young, [Módulo 05](../05%20-%20Reduccion%20de%20Dimensionalidad/00_PCA.ipynb)); las filas de $U_k\Sigma_k$ son los documentos en $k$ «temas» latentes.

**Vectores de palabras por co-ocurrencia:** se cuenta cuántas veces coaparecen las palabras en una ventana y se reemplaza el conteo por la **información mutua puntual positiva**
$$\operatorname{PPMI}(w,c)=\max\Big(0,\ \ln\frac{p(w,c)}{p(w)\,p(c)}\Big),$$
cuya SVD trunca da vectores $W=U_d\Sigma_d^{1/2}$ de $d$ dimensiones. **Es el puente a *word2vec*:** Mikolov et al. (2013) aprenden vectores prediciendo contexto con una red superficial, y Levy y Goldberg (2014) mostraron que el *skip-gram* con muestreo negativo **factoriza implícitamente** una matriz PMI desplazada *(citados de memoria, verificar)*. Es decir, los *embeddings* son, en esencia, **factorización de matrices** (idea que retomaremos en el [cuaderno 03](03_Sistemas_de_Recomendacion.ipynb)).

````python
# ============================================================
# 7. Representaciones densas: LSA y vectores de palabras por co-ocurrencia + SVD (desde cero)
# ============================================================
# --- (a) LSA: SVD truncada de la matriz TF-IDF (documento-termino) ---
K = 10
U, S, Vt = np.linalg.svd(T.toarray(), full_matrices=False)                                      # SVD completa (4000 x 573: cabe sin problema)
Z_mio = U[:, :K] * S[:K]                                                                        # coordenadas de cada documento en los K «temas» latentes
svd_sk = TruncatedSVD(n_components=K, algorithm="arpack", tol=0, random_state=0).fit(T)
Z_sk = svd_sk.transform(T)
assert np.allclose(S[:K], svd_sk.singular_values_, rtol=1e-6)                                   # mismos valores singulares
assert np.allclose(np.abs(Vt[:K]), np.abs(svd_sk.components_), atol=1e-6)                       # mismas direcciones (el signo de cada vector es arbitrario)
assert np.allclose(np.abs(Z_mio), np.abs(Z_sk), atol=1e-6)
print(f"OK: LSA desde cero (SVD de la matriz TF-IDF) == TruncatedSVD(K={K}): valores singulares, componentes (salvo signo) y proyecciones coinciden.")
print("Varianza explicada por los", K, "primeros componentes:", np.round(S[:K] ** 2 / (S ** 2).sum(), 3), "→ acumulada", round(float((S[:K] ** 2).sum() / (S ** 2).sum()), 3))
print("Nota: TruncatedSVD NO centra los datos (PCA sí: es la SVD de la matriz centrada, Módulo 05); con matrices dispersas no se centra para no volverlas densas.")

inv = {j: w for w, j in vocab.items()}
for c in range(3):
    top = np.argsort(-np.abs(Vt[c]))[:7]
    print(f"Componente {c + 1}: " + ", ".join(f"{inv[j]} ({Vt[c, j]:+.2f})" for j in top))

# Clasificar con 10 rasgos densos (LSA) vs los 573 de TF-IDF: ¿cuanta informacion de sentimiento sobrevive?
idx_tr, idx_te = train_test_split(np.arange(len(df_rest)), test_size=0.25, stratify=df_rest.y.values, random_state=0)
y_all = df_rest.y.values
res = {}
for k in [2, 5, 10, 20, 50]:
    Zk = U[:, :k] * S[:k]
    res[k] = LogisticRegression(max_iter=2000).fit(Zk[idx_tr], y_all[idx_tr]).score(Zk[idx_te], y_all[idx_te])
base = LogisticRegression(max_iter=2000).fit(T[idx_tr], y_all[idx_tr]).score(T[idx_te], y_all[idx_te])
print("\nExactitud con k componentes LSA:", {k: round(v, 3) for k, v in res.items()}, f"| con los {T.shape[1]} rasgos TF-IDF (1-gramas): {base:.3f}")

# --- (b) Vectores de PALABRAS: co-ocurrencia en ventana ±2 -> PMI positivo (PPMI) -> SVD ---
V = len(vocab); ventana = 2
M = np.zeros((V, V))
for d in tokens_rest:
    ids = [vocab[w] for w in d]
    for i, wi in enumerate(ids):
        for j in range(max(0, i - ventana), min(len(ids), i + ventana + 1)):
            if j != i:
                M[wi, ids[j]] += 1
total = M.sum(); pw = M.sum(1, keepdims=True) / total; pc = M.sum(0, keepdims=True) / total
with np.errstate(divide="ignore", invalid="ignore"):
    PMI = np.log((M / total) / (pw @ pc))                                                       # log( p(w,c) / (p(w) p(c)) )
PPMI = np.where(np.isfinite(PMI) & (PMI > 0), PMI, 0.0)                                         # solo asociaciones MAS frecuentes que el azar
Uw, Sw, _ = np.linalg.svd(PPMI, full_matrices=False)
D = 20
W = Uw[:, :D] * np.sqrt(Sw[:D])                                                                 # embeddings de palabras de dimension 20
W_sk = TruncatedSVD(n_components=D, algorithm="arpack", tol=0, random_state=0).fit(PPMI)
assert np.allclose(Sw[:D], W_sk.singular_values_, rtol=1e-5)
Wn = W / (np.linalg.norm(W, axis=1, keepdims=True) + 1e-12)


def vecinos(palabra, n=6):
    s = Wn @ Wn[vocab[palabra]]; orden = np.argsort(-s)
    return [(inv[j], round(float(s[j]), 2)) for j in orden[1:n + 1]]


print(f"\nOK: vectores de palabras (SVD de PPMI, d = {D}): valores singulares == TruncatedSVD. Tamaño de la matriz de co-ocurrencia: {M.shape}")
for w in ["excelente", "horrible", "servicio", "batería"]:
    if w in vocab:
        print(f"  vecinos de «{w}»: {vecinos(w)}")
sim_mismo = np.mean([Wn[vocab[a]] @ Wn[vocab[b]] for a in ["excelente", "genial", "fantástico", "perfecto"] for b in ["excelente", "genial", "fantástico", "perfecto"] if a != b])
sim_cruz = np.mean([Wn[vocab[a]] @ Wn[vocab[b]] for a in ["excelente", "genial", "fantástico", "perfecto"] for b in ["horrible", "terrible", "pésimo", "lamentable"]])
print(f"Similitud coseno media entre elogios: {sim_mismo:.2f} | entre un elogio y una crítica: {sim_cruz:.2f}")

# --- Documento = promedio de los vectores de sus palabras (con TF-IDF como pesos) ---
doc_vec = np.asarray(T @ W)
doc_vec = doc_vec / (np.linalg.norm(doc_vec, axis=1, keepdims=True) + 1e-12)
acc_emb = LogisticRegression(max_iter=2000).fit(doc_vec[idx_tr], y_all[idx_tr]).score(doc_vec[idx_te], y_all[idx_te])
print(f"Exactitud con documentos representados por {D} dimensiones (promedio ponderado de vectores de palabras): {acc_emb:.3f}")
assert sim_mismo > sim_cruz and res[50] > res[2] + 0.2

fig, ax = plt.subplots(figsize=(8.6, 6.2))
palabras = ["excelente", "genial", "fantástico", "perfecto", "delicioso", "increíble", "horrible", "terrible", "pésimo", "lamentable", "asqueroso", "mediocre", "servicio", "comida", "atención", "lugar", "postre", "café"]
palabras = [w for w in palabras if w in vocab]
P2 = (Wn[[vocab[w] for w in palabras]] - Wn.mean(0)) @ np.linalg.svd(Wn - Wn.mean(0), full_matrices=False)[2][:2].T
col = [PALETA[2] if w in ["excelente", "genial", "fantástico", "perfecto", "delicioso", "increíble"] else PALETA[1] if w in ["horrible", "terrible", "pésimo", "lamentable", "asqueroso", "mediocre"] else PALETA[0] for w in palabras]
ax.scatter(P2[:, 0], P2[:, 1], c=col, s=60)
for (x, y), w in zip(P2, palabras):
    ax.annotate(w, (x, y), fontsize=9, xytext=(4, 4), textcoords="offset points")
ax.set_xlabel("Componente 1 de los vectores de palabras"); ax.set_ylabel("Componente 2"); ax.set_title("Palabras que ocurren en contextos parecidos quedan cerca (verde: elogios, rojo: críticas, azul: aspectos)", fontweight="bold", fontsize=9)
plt.tight_layout(); plt.show()
````

---
**Lectura honesta.** (1) Nuestro LSA **coincide con `TruncatedSVD`** (valores singulares, componentes salvo signo, proyecciones). (2) Pero **LSA no es supervisado**: los primeros componentes reflejan la **estructura de las plantillas** («es», «el», «la», «me pareció») —lo que más varía en el corpus— y **no el sentimiento**. Con 10 componentes la exactitud es solo ≈ 0.56 (casi azar) y se necesitan ≈ 50 para llegar a ≈ 0.81, todavía **por debajo** de los 573 rasgos TF-IDF (≈ 0.89). Reducir dimensionalidad **sin mirar la etiqueta** puede tirar justo la señal que importa. (3) Los **vectores de palabras** sí capturan relaciones: los vecinos de «horrible» son otras críticas («deficiente», «decepcionante», «lamentable») y los de «servicio», otros aspectos («postre», «comida»). Pero ojo: los **antónimos aparecen en contextos parecidos** («la comida es excelente / es horrible»), así que la similitud elogio–elogio (≈ 0.85) **no está tan lejos** de elogio–crítica (≈ 0.67), y «excelente» tiene entre sus vecinos a «malísimo» (por cómo está construido el corpus). Los *embeddings* por co-ocurrencia mezclan **tema** y **sentimiento**: es una limitación conocida. (4) Con un promedio de vectores de 20 dimensiones, la exactitud es ≈ 0.73: útil como compresión, no como sustituto de TF-IDF aquí.

## 8. Contexto: *Transformers* y BERT, y Dos Bloques Opcionales 🔌

Las representaciones fijas de la sección 7 asignan **un vector por palabra**, sin importar el contexto: «banco» (de sentarse) y «banco» (de dinero) comparten vector. Los ***Transformers*** (Vaswani et al., 2017) resuelven esto con **atención**: el vector de cada palabra se recalcula como una **mezcla ponderada** de las demás palabras del texto,
$$\operatorname{Atención}(Q,K,V)=\operatorname{softmax}\!\Big(\frac{QK^\top}{\sqrt{d_k}}\Big)V .$$
**BERT** (Devlin et al., 2019) se **preentrena** con enormes cantidades de texto sin etiquetar (predecir palabras enmascaradas) y se ***fine-tunea*** con una **cabeza de clasificación** y pocos ejemplos etiquetados: exactamente el esquema «*backbone* preentrenado + cabeza nueva» del [Módulo 08](../08%20-%20Temas%20Avanzados%20de%20ML/01_Transfer_Learning_y_Fine_Tuning.ipynb) (Géron, Cap. 14–15). Frente a TF-IDF + regresión logística, un modelo así suele ser **mucho mejor** con contexto, negación e ironía, a cambio de **cómputo, latencia y opacidad** (y de **heredar los sesgos** de su corpus de preentrenamiento: [Módulo 08 sobre explicabilidad](../08%20-%20Temas%20Avanzados%20de%20ML/02_Explicabilidad_de_Modelos.ipynb) y [ética de la IA](../../Introduccion%20a%20la%20Inteligencia%20Artificial/03%20-%20Etica%20y%20el%20Nuevo%20Paradigma%20de%20la%20IA/README.md)).

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Texto → vectores** | Tokenizar y normalizar (cada paso cambia el vocabulario); **bolsa de palabras** = conteos; **TF-IDF** $=n_{d,t}\cdot(\ln\frac{1+n}{1+\mathrm{df}}+1)$ con norma L2. **Verificado con `assert`** contra `CountVectorizer` y `TfidfVectorizer` |
| **Trampas del español** | Las tildes distinguen palabras (*él*/*el*); las listas de *stopwords* contienen **«no»**: quitarlas baja la exactitud en reseñas con negación de 0.955 a 0.784; quitar tildes no cuesta nada y da robustez |
| **Naive Bayes multinomial** | $\log P(c\mid d)=\log\pi_c+\sum_t n_t\log\theta_{c,t}$ con suavizado de Laplace y **log-espacio**; idéntico a `MultinomialNB`. Rápido, pero ≈ 0.80 aquí |
| **Regresión logística + TF-IDF** | ≈ 0.89 (1-gramas) y ≈ **0.93** (1-2-gramas): los pares de palabras captan la negación. Vectorizador **dentro del `Pipeline`** para no filtrar información |
| **Léxico frente a supervisado** | El léxico no necesita etiquetas y es estable entre dominios (≈ 0.66–0.68) pero poco preciso y ciego al sarcasmo; el supervisado es más preciso pero **pierde ≈ 10 puntos al cambiar de dominio** (0.93 → 0.83) |
| **Desbalance** | Con 8 % de positivas, «siempre negativa» = 92 % de exactitud; usa F1/recobrado de la clase de interés y `class_weight` o umbral |
| **LSA y *embeddings*** | SVD de TF-IDF (LSA) o de PPMI (vectores de palabras) ≡ `TruncatedSVD`; **no supervisados**: con 10 componentes LSA da ≈ 0.56 (el sentimiento no es la mayor varianza); los antónimos comparten contextos. Puente a *word2vec* (factorización implícita de PMI) |
| **Transformers/BERT** | Atención $\operatorname{softmax}(QK^\top/\sqrt{d_k})V$; preentrenar + *fine-tunear* (Módulo 08). Mucho mejores con contexto, pero más costosos y opacos |
| **Honestidad** | El corpus es **sintético**: las métricas son optimistas; con 16 frases a mano la regresión logística falla justo donde el vocabulario es nuevo |
