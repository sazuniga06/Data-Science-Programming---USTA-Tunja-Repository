## 1. Recapitulación: la Rejilla y por qué Duele 🧱

Con $d$ hiperparámetros y $m_i$ valores para el $i$-ésimo, la rejilla evalúa $\prod_i m_i$ combinaciones, cada una con $k$ pliegues. Resumen de lo visto en el [módulo 03](../03%20-%20Metricas%20y%20Ajuste%20de%20Hiperparametros/02_Validacion_Cruzada_y_GridSearch.ipynb) y de lo que se añade aquí:

| | **GridSearch** | **Random Search** |
|---|---|---|
| **Qué prueba** | Todas las combinaciones del producto cartesiano de listas | $N$ configuraciones muestreadas de **distribuciones** |
| **Costo** | $k\prod_i m_i$ ajustes: **exponencial en $d$** | $k\,N$ ajustes: **tú eliges $N$**, independiente de $d$ |
| **Valores distintos por eje** | $m_i$ (con $N=\prod m_i$: $N^{1/d}$ si todos iguales) | **$N$** en cada eje, siempre |
| **Hiperparámetros continuos** | Hay que discretizar a mano | Se muestrean con su distribución (uniforme, log-uniforme...) |
| **Hiperparámetros irrelevantes** | Multiplican el costo **sin aportar** | No cuestan nada extra |
| **Presupuesto flexible** | No: añadir un valor multiplica el costo | Sí: se puede **parar o ampliar** en cualquier momento (cada muestra es i.i.d.) |
| **Cobertura del espacio** | Regular (útil en 1–2 dimensiones finas) | Aleatoria (huecos y cúmulos, pero sin sesgos de alineación) |
| **Reproducible** | Sí, determinista | Sí, con `random_state` |

La razón de fondo es un hecho empírico sobre los problemas reales: **solo unos pocos hiperparámetros importan de verdad**, y cuáles importan **depende del conjunto de datos** (Bergstra y Bengio, 2012). Si la función de validación tiene **dimensión efectiva baja**, la rejilla desperdicia casi todo su presupuesto en combinaciones que repiten valores de los hiperparámetros relevantes.

## 2. Dimensión Efectiva y el Argumento de Bergstra y Bengio (2012) 🎯

**Formalización.** Sea $\Psi(\boldsymbol\lambda)$ la métrica de validación en función del vector de hiperparámetros $\boldsymbol\lambda\in[0,1]^d$. Se dice que $\Psi$ tiene **dimensión efectiva baja** si casi toda su variación se explica por pocos de los $d$ ejes (en el caso extremo, $\Psi(\boldsymbol\lambda)=g(\lambda_1)$). Con un presupuesto de $N$ evaluaciones:

- una **rejilla** $k\times\cdots\times k$ ($N=k^d$) prueba solo **$k=N^{1/d}$ valores distintos** de cada hiperparámetro; si solo importa $\lambda_1$, **$N-k$ evaluaciones repiten** un valor ya visto de $\lambda_1$;
- **Random Search** prueba **$N$ valores distintos** de cada hiperparámetro, incluido $\lambda_1$.

**Una cuenta útil.** Si la «región buena» ocupa una fracción $p$ del espacio, cada muestra aleatoria cae en ella con probabilidad $p$, así que con $N$ muestras

$$P(\text{al menos una en la región buena})=1-(1-p)^N,\qquad N\ge\frac{\ln(1-q)}{\ln(1-p)}\ \text{para tener confianza }q.$$

Con $p=5\,\%$ y $q=95\,\%$ hacen falta $N=59$ muestras, **sin importar $d$** (una rejilla de 64 puntos en $d=6$ prueba solo $64^{1/6}=2$ valores por eje).

**Demostración numérica propia.** Definimos una función de respuesta en $[0,1]^3$ con tres ejes de importancia muy distinta: $x_1$ con un **pico estrecho** (su ubicación es desconocida), $x_2$ con un efecto **leve y ancho** y $x_3$ **irrelevante**:

$$\Psi(\mathbf x)=0.9\,e^{-\frac{(x_1-c)^2}{2\cdot0.06^2}}+0.1\,e^{-\frac{(x_2-0.5)^2}{2\cdot0.35^2}}.$$

Repetimos 3 000 veces, cada vez con un centro $c$ aleatorio (para que la rejilla no gane ni pierda por «suerte»), y comparamos el **mejor valor encontrado** por una rejilla $k^3$ frente a $k^3$ muestras uniformes. Es un modelo de juguete: **pone la dimensión efectiva baja por construcción**; los experimentos con modelos reales vienen después.

````python
# ============================================================
# 2. Dimension efectiva: con el MISMO presupuesto, Random Search explora mas valores de lo que importa
# ============================================================
# Respuesta sintetica en [0,1]^3: x1 importa MUCHO (pico estrecho, ubicacion desconocida), x2 poco, x3 NADA.
def respuesta(Pts, centro):
    f1 = np.exp(-((Pts[..., 0] - centro) ** 2) / (2 * 0.06 ** 2))             # pico estrecho en x1 (ancho ~ 0.06)
    f2 = np.exp(-((Pts[..., 1] - 0.5) ** 2) / (2 * 0.35 ** 2))                # efecto leve y ancho de x2
    return 0.9 * f1 + 0.1 * f2                                                # x3 no aparece: dimension irrelevante

rng = np.random.default_rng(SEED)
T = 3000
centros = rng.uniform(0.15, 0.85, T)                                          # el optimo cambia en cada ensayo: se promedia la 'suerte' de la rejilla
filas = []
for k in (2, 3, 4, 5):
    N = k ** 3
    ejes = (np.arange(k) + 0.5) / k                                           # rejilla k x k x k con valores equiespaciados (centros de celda)
    rejilla = np.stack(np.meshgrid(ejes, ejes, ejes, indexing="ij"), axis=-1).reshape(-1, 3)
    mejor_rej = respuesta(rejilla[None, :, :], centros[:, None]).max(axis=1)  # (T,)
    azar = rng.uniform(size=(T, N, 3))
    mejor_az = respuesta(azar, centros[:, None]).max(axis=1)
    filas.append({"presupuesto N": N, "valores distintos de x1 (rejilla)": k, "valores distintos de x1 (azar)": N,
                  "mejor encontrado: rejilla": mejor_rej.mean(), "mejor encontrado: aleatoria": mejor_az.mean(),
                  "P(aleatoria > rejilla)": np.mean(mejor_az > mejor_rej)})
res = pd.DataFrame(filas).set_index("presupuesto N")
display(res.round(3))
assert (res["mejor encontrado: aleatoria"] > res["mejor encontrado: rejilla"]).all()          # con el mismo N, el azar gana en este problema
print("Con x1 como única dimensión importante, la rejilla gasta N evaluaciones pero solo prueba k valores de x1; Random Search prueba N valores distintos.")

# Probabilidad de que N muestras aleatorias caigan al menos una vez en el mejor p-% del espacio: 1 - (1 - p)^N
p_top = 0.05
tabla_n = []
for N in (5, 14, 29, 59, 90):
    sim = np.mean((rng.uniform(size=(20000, N)) < p_top).any(axis=1))
    tabla_n.append({"N": N, "P(al menos una en el mejor 5 %) teórica": 1 - (1 - p_top) ** N, "simulada": sim})
tabla_n = pd.DataFrame(tabla_n).set_index("N"); display(tabla_n.round(3))
assert np.allclose(tabla_n.iloc[:, 0], tabla_n.iloc[:, 1], atol=0.01)
N95 = math.ceil(math.log(0.05) / math.log(1 - p_top)); assert N95 == 59
print(f"Regla: para tener 95 % de probabilidad de caer en el mejor 5 % de la región basta N ≥ ln(0.05)/ln(0.95) = {N95} muestras, SIN importar la dimensión del espacio.")

# Figura clasica (Bergstra y Bengio, 2012): 9 evaluaciones en una rejilla 3x3 vs 9 al azar, y lo que cada una 've' del eje importante
centro0 = 0.37
g3 = (np.arange(3) + 0.5) / 3
rej9 = np.array([(a, b) for a in g3 for b in g3]); az9 = np.random.default_rng(3).uniform(size=(9, 2))
xs = np.linspace(0, 1, 400)
fig, axes = plt.subplots(1, 2, figsize=(13, 5), sharey=False)
for ax, pts, nombre, col in [(axes[0], rej9, "Rejilla 3×3", "#dc2626"), (axes[1], az9, "9 puntos aleatorios", "#16a34a")]:
    ax.scatter(pts[:, 0], pts[:, 1], s=60, color=col, zorder=3)
    ax.plot(xs, 0.12 + 0.2 * np.exp(-((xs - centro0) ** 2) / (2 * 0.06 ** 2)) , color="#2563eb", lw=2, label="Importancia de $x_1$ (pico en 0.37)")
    ax.plot(pts[:, 0], np.full(9, 0.03), "|", color=col, ms=14, mew=2, label="Valores de $x_1$ probados")
    ax.axvline(centro0, color="#2563eb", ls=":", alpha=0.6)
    ax.set_xlim(0, 1); ax.set_ylim(0, 1.2); ax.set_xlabel("$x_1$ (hiperparámetro importante)"); ax.set_ylabel("$x_2$ (poco importante)")
    ax.set_title(f"{nombre}: {len(np.unique(np.round(pts[:, 0], 6)))} valores distintos de $x_1$", fontweight="bold"); ax.legend(fontsize=8, loc="upper right")
plt.tight_layout(); plt.show()
````

## 3. Distribuciones de Muestreo: Uniforme, Log-uniforme y Enteros 📐

`RandomizedSearchCV` recibe un diccionario cuyos valores son **listas** (se elige uno al azar, uniformemente) o **distribuciones** de `scipy.stats` (cualquier objeto con método `.rvs(random_state=...)`). Las más útiles:

| Hiperparámetro | Distribución | Por qué |
|---|---|---|
| `C`, `gamma`, `alpha`, tasa de aprendizaje, $\lambda$ | `loguniform(a, b)` | Actúan **multiplicativamente**: pasar de $10^{-3}$ a $10^{-2}$ importa tanto como de $10^{2}$ a $10^{3}$ |
| `max_depth`, `n_neighbors`, `min_samples_leaf` | `randint(a, b)` (**$b$ excluido**) | Enteros |
| `max_features`, `subsample`, `l1_ratio` | `uniform(loc, scale)` (**cubre $[loc,\,loc+scale]$**) | Proporciones y mezclas |
| `kernel`, `penalty`, `class_weight` | Lista de valores | Categóricos |

**Por qué la escala logarítmica.** Una uniforme en $[10^{-3},10^{3}]$ pone el **90 %** de las muestras en la última década ($[10^2,10^3]$) y casi nada por debajo de 1: **seis órdenes de magnitud** de los que solo se exploran uno. La log-uniforme da a **cada década** la misma probabilidad ($1/6$). Su definición es $\log_{10}X\sim U(\log_{10}a,\log_{10}b)$, es decir $X=10^{U}$.

````python
# ============================================================
# 3. Distribuciones de muestreo: uniforme vs log-uniforme (scipy.stats) para C, gamma y enteros
# ============================================================
rng = np.random.RandomState(SEED)
M = 100_000
C_unif = uniform(loc=1e-3, scale=1e3 - 1e-3).rvs(M, random_state=rng)                  # U(0.001, 1000): lo que pasa si se usa 'uniform' sin pensar
C_log = loguniform(1e-3, 1e3).rvs(M, random_state=rng)                                 # log-uniforme: cada DECADA recibe la misma probabilidad
decadas = np.arange(-3, 3)
frac = lambda v: np.array([np.mean((v >= 10.0 ** d) & (v < 10.0 ** (d + 1))) for d in decadas])
tab = pd.DataFrame({"uniforme U(0.001, 1000)": frac(C_unif), "log-uniforme(0.001, 1000)": frac(C_log)},
                   index=[f"[10^{d}, 10^{d + 1})" for d in decadas])
display(tab.round(4))
assert frac(C_unif)[-1] > 0.89 and frac(C_unif)[:3].sum() < 0.003                       # uniforme: 90 % en la ultima decada y casi nada por debajo de 1
assert np.allclose(frac(C_log), 1 / 6, atol=0.01)                                       # log-uniforme: 1/6 por decada

# Equivalencia en DISTRIBUCION (no en valores) con la transformada inversa hecha a mano: C = 10^U, U ~ U(log10 a, log10 b)
manual = 10 ** np.random.RandomState(1).uniform(-3, 3, M)
ks = stats.ks_2samp(np.log10(C_log), np.log10(manual))
print(f"Prueba de Kolmogorov-Smirnov (loguniform de scipy vs 10^U a mano): estadístico = {ks.statistic:.4f}, p = {ks.pvalue:.3f} -> misma distribución.")
assert ks.pvalue > 0.01

# randint: enteros en [low, high) (el extremo superior se EXCLUYE); uniform(loc, scale): continuo en [loc, loc + scale]
d_prof = randint(2, 11).rvs(10000, random_state=rng)
assert d_prof.min() == 2 and d_prof.max() == 10                                         # randint(2, 11) -> {2, ..., 10}
print("randint(2, 11) produce enteros 2..10 (el 11 no entra); uniform(loc, scale) cubre [loc, loc + scale] (¡no [loc, scale]!).")

fig, axes = plt.subplots(1, 2, figsize=(13, 4.4))
axes[0].hist(np.log10(C_unif), bins=60, color="#dc2626", alpha=0.85, density=True); axes[0].set_title("Uniforme en $C$: casi todo cae en la última década", fontweight="bold", fontsize=10)
axes[1].hist(np.log10(C_log), bins=60, color="#16a34a", alpha=0.85, density=True); axes[1].set_title("Log-uniforme: todas las décadas igual de probables", fontweight="bold", fontsize=10)
for ax in axes: ax.set_xlabel("$\\log_{10} C$"); ax.set_ylabel("Densidad")
plt.tight_layout(); plt.show()
print("Regla: los hiperparámetros que actúan de forma MULTIPLICATIVA (C, gamma, alpha, tasa de aprendizaje, penalizaciones) se muestrean en escala logarítmica; los enteros con randint; las proporciones con uniform.")
````

## 4. Random Search desde Cero (y su Verificación) 🛠️

El algoritmo es mínimo:

1. Para $t=1,\dots,N$: muestrear $\boldsymbol\lambda_t\sim p(\boldsymbol\lambda)$ (independientes entre hiperparámetros).
2. Estimar el error de validación cruzada $\widehat{\text{Err}}_{\text{CV}}(\boldsymbol\lambda_t)$.
3. Devolver $\hat{\boldsymbol\lambda}=\arg\min_t\widehat{\text{Err}}_{\text{CV}}(\boldsymbol\lambda_t)$ y **reajustar** con todo el entrenamiento.

La reproducibilidad exacta frente a `scikit-learn` exige replicar su **protocolo de números aleatorios**: un único `RandomState(seed)`, claves **ordenadas alfabéticamente** y cada valor de `dist.rvs(random_state=rng)`. Lo implementamos y comprobamos que genera **exactamente las mismas configuraciones** y **los mismos puntajes** que `RandomizedSearchCV` al darle **las mismas particiones**. Lo que **no** puede ser exacto: cualquier otro generador (p. ej. la transformada inversa «a mano») produce valores numéricos distintos, equivalentes solo **en distribución**.

````python
# ============================================================
# 4. Random Search DESDE CERO, verificado contra RandomizedSearchCV (mismo generador, mismas particiones)
# ============================================================
Xb, yb = load_breast_cancer(return_X_y=True)
pliegues = list(StratifiedKFold(5, shuffle=True, random_state=SEED).split(Xb, yb))
pipe = make_pipeline(StandardScaler(), SVC())
distribuciones = {"svc__C": loguniform(1e-2, 1e2), "svc__gamma": loguniform(1e-6, 1e0)}
N_ITER = 20

def muestrear_desde_cero(dists, n_iter, seed):
    """Mismo protocolo que ParameterSampler: UN RandomState; claves ordenadas; cada valor sale de dist.rvs(random_state=rng)."""
    rng = np.random.RandomState(seed)
    for _ in range(n_iter):
        yield {k: dists[k].rvs(random_state=rng) for k in sorted(dists)}

def random_search_desde_cero(estimador, dists, n_iter, X, y, pliegues, seed):
    """n_iter configuraciones aleatorias x validacion cruzada; devuelve (lista de params, medias CV, mejor indice)."""
    candidatos = list(muestrear_desde_cero(dists, n_iter, seed))
    medias = np.array([np.mean([clone(estimador).set_params(**p).fit(X[a], y[a]).score(X[b], y[b]) for a, b in pliegues]) for p in candidatos])
    return candidatos, medias, int(np.argmax(medias))                              # argmax: el PRIMERO en empates (como sklearn)

t0 = time.perf_counter()
cand, medias, i_mejor = random_search_desde_cero(pipe, distribuciones, N_ITER, Xb, yb, pliegues, SEED)
t_mio = time.perf_counter() - t0
rs = RandomizedSearchCV(pipe, distribuciones, n_iter=N_ITER, cv=pliegues, random_state=SEED, scoring="accuracy").fit(Xb, yb)

assert cand == list(ParameterSampler(distribuciones, N_ITER, random_state=SEED))     # EXACTAMENTE las mismas configuraciones muestreadas
assert np.allclose(medias, rs.cv_results_["mean_test_score"])                          # exactamente las mismas puntuaciones CV
assert cand[i_mejor] == rs.best_params_ and np.isclose(medias[i_mejor], rs.best_score_)
print(f"Random Search desde cero == RandomizedSearchCV: best_score_ = {medias[i_mejor]:.4f}, C = {cand[i_mejor]['svc__C']:.3g}, gamma = {cand[i_mejor]['svc__gamma']:.3g}")
print(f"{N_ITER} configuraciones × 5 pliegues = {N_ITER * 5} ajustes (+1 de reajuste) | tiempo propio = {t_mio:.2f} s")
print("Lo que NO es exacto: la versión con transformada inversa a mano (sección 3) genera otros valores numéricos (otro flujo de números aleatorios) aunque siga la misma distribución;")
print("y con listas de valores en vez de distribuciones, sklearn muestrea la rejilla SIN reemplazo (nunca repite combinación), mientras que con distribuciones puede repetir valores discretos.")

df = pd.DataFrame(rs.cv_results_)[["param_svc__C", "param_svc__gamma", "mean_test_score", "rank_test_score"]].astype({"param_svc__C": float, "param_svc__gamma": float})
display(df.sort_values("rank_test_score").head(5).round(5))
fig, ax = plt.subplots(figsize=(8.5, 5.2))
sc_ = ax.scatter(np.log10(df["param_svc__C"]), np.log10(df["param_svc__gamma"]), c=df["mean_test_score"], cmap="viridis", s=90, edgecolor="white")
ax.scatter(np.log10(cand[i_mejor]["svc__C"]), np.log10(cand[i_mejor]["svc__gamma"]), s=260, facecolors="none", edgecolors="#dc2626", lw=2.5, label="Mejor")
plt.colorbar(sc_, label="Accuracy CV (5 pliegues)"); ax.set_xlabel("$\\log_{10} C$"); ax.set_ylabel("$\\log_{10}\\gamma$"); ax.legend()
ax.set_title("20 configuraciones log-uniformes de Random Search", fontweight="bold")
plt.tight_layout(); plt.show()
````

## 5. Mejor Puntaje vs. Presupuesto: ¿Quién Gana y Cuándo? 📈

La pregunta práctica es: **dado un presupuesto de $N$ evaluaciones, ¿qué método encuentra mejores hiperparámetros?** Como la respuesta es **aleatoria** para Random Search, se promedia sobre **muchas repeticiones**. Dos experimentos reales, con problemas de **dimensión distinta**:

- **(a) 2 hiperparámetros** (`C` y `gamma` de un SVM RBF sobre *Breast Cancer*). Se evalúa una vez una rejilla fina de 10×10 y los buscadores operan sobre esa tabla (Random Search muestrea celdas), de modo que se pueden simular **miles de búsquedas** sin reentrenar.
- **(b) 4 hiperparámetros** (profundidad, hoja mínima, `max_features` y poda `ccp_alpha` de un árbol de decisión sobre un problema sintético). Una **reserva** de 300 configuraciones i.i.d. permite simular Random Search de cualquier $N\le300$ remuestreando la reserva; la rejilla ($b^4$ puntos con $b$ cuantiles de cada distribución, para que sea **justa**) se evalúa directamente, con **3 disposiciones desplazadas** porque una rejilla es una sola tirada de suerte.

> ⚠️ **Qué se mide:** el mejor **puntaje de CV** que ve el buscador (un máximo, por tanto optimista; módulo 03 y [cuaderno 01](01_Validacion_Cruzada_Avanzada.ipynb)). Sirve para **comparar eficiencia de búsqueda**, no para reportar rendimiento.

````python
# ============================================================
# 5a. Mejor puntaje vs presupuesto en 2 dimensiones (SVM RBF, C x gamma): superficie precalculada y MUCHAS repeticiones
# ============================================================
# Se evalua UNA vez una rejilla fina de 10 x 10 (superficie de CV de 3 pliegues) y los buscadores operan sobre esa tabla:
# asi se pueden simular miles de busquedas sin reentrenar. Random Search muestrea celdas de la tabla (discretiza la log-uniforme).
logC, logG = np.linspace(-2, 2, 10), np.linspace(-6, 0, 10)
pl3 = list(StratifiedKFold(3, shuffle=True, random_state=SEED).split(Xb, yb))
def cv_svc(lc, lg):
    return np.mean([make_pipeline(StandardScaler(), SVC(C=10.0 ** lc, gamma=10.0 ** lg)).fit(Xb[a], yb[a]).score(Xb[b], yb[b]) for a, b in pl3])
t0 = time.perf_counter()
superficie = np.array([[cv_svc(lc, lg) for lg in logG] for lc in logC])
print(f"Superficie 10 × 10 evaluada en {time.perf_counter() - t0:.1f} s | máximo = {superficie.max():.4f} | mediana = {np.median(superficie):.4f} | fracción de celdas dentro de 0.5 pts del máximo = {np.mean(superficie >= superficie.max() - 0.005):.2f}")

rng = np.random.default_rng(SEED)
presupuestos = [b * b for b in range(2, 9)]
rej_mejor, az_media, az_sd, p_gana = [], [], [], []
plano = superficie.ravel()
for b in range(2, 9):
    idx = np.round(np.linspace(0, 9, b)).astype(int)                              # sub-rejilla b x b equiespaciada (incluye los bordes)
    mejor_r = superficie[np.ix_(idx, idx)].max()
    azar = plano[rng.integers(0, 100, size=(4000, b * b))].max(axis=1)           # 4000 repeticiones de Random Search con presupuesto b^2
    rej_mejor.append(mejor_r); az_media.append(azar.mean()); az_sd.append(azar.std()); p_gana.append(np.mean(azar > mejor_r))
res2d = pd.DataFrame({"rejilla b×b": rej_mejor, "aleatoria (media)": az_media, "aleatoria (desv.)": az_sd, "P(aleatoria > rejilla)": p_gana},
                     index=pd.Index(presupuestos, name="presupuesto N"))
display(res2d.round(4))

fig, axes = plt.subplots(1, 2, figsize=(14, 4.8))
im = axes[0].imshow(superficie, origin="lower", cmap="viridis", extent=[logG[0], logG[-1], logC[0], logC[-1]], aspect="auto")
plt.colorbar(im, ax=axes[0], label="Accuracy CV (3 pliegues)"); axes[0].set_xlabel("$\\log_{10}\\gamma$"); axes[0].set_ylabel("$\\log_{10} C$")
axes[0].set_title("Superficie de validación del SVM (precalculada)", fontweight="bold")
axes[1].plot(presupuestos, rej_mejor, "o-", color="#dc2626", lw=2, label="Rejilla $b\\times b$ (determinista)")
axes[1].plot(presupuestos, az_media, "s-", color="#16a34a", lw=2, label="Random Search (media de 4 000 repeticiones)")
axes[1].fill_between(presupuestos, np.array(az_media) - np.array(az_sd), np.array(az_media) + np.array(az_sd), color="#16a34a", alpha=0.15)
axes[1].axhline(superficie.max(), color="black", ls=":", label="Máximo de la tabla")
axes[1].set_xlabel("Presupuesto $N$ (evaluaciones)"); axes[1].set_ylabel("Mejor accuracy CV encontrado"); axes[1].legend(fontsize=8)
axes[1].set_ylim(0.90, 0.99); axes[1].set_title("2 hiperparámetros: ambos métodos se parecen (N = 4 queda fuera del rango)", fontweight="bold", fontsize=9)
plt.tight_layout(); plt.show()
print(f"Lectura honesta: con solo 2 hiperparámetros y una superficie suave NO hay un ganador sistemático. Con N = 4 la rejilla 2×2 solo toca esquinas (accuracy {rej_mejor[0]:.3f} frente a {az_media[0]:.3f} del azar); desde N ≈ 9 ambos están a menos de un punto del máximo de la tabla.")
print("La rejilla gana en algunos presupuestos (sus nodos caen por casualidad sobre la 'cresta') y el azar en otros (columna P(aleatoria > rejilla)). La ventaja de Random Search aparece cuando hay MÁS hiperparámetros y pocos importan (siguiente celda).")
print("Ojo: el 'mejor puntaje CV' es optimista para ambos (es un máximo): sirve para comparar buscadores, no para reportar rendimiento.")
````

````python
# ============================================================
# 5b. Mejor puntaje vs presupuesto en 4 dimensiones (arbol de decision) y el mismo presupuesto para rejilla y azar
# ============================================================
Xh, yh = make_classification(n_samples=500, n_features=20, n_informative=5, n_redundant=2, n_clusters_per_class=3, flip_y=0.05, class_sep=0.9, random_state=0)
pl_h = list(StratifiedKFold(3, shuffle=True, random_state=0).split(Xh, yh))
def cv_arbol(p):
    return np.mean([DecisionTreeClassifier(max_depth=int(p["max_depth"]), min_samples_leaf=int(p["min_samples_leaf"]), max_features=float(p["max_features"]),
                                           ccp_alpha=float(p["ccp_alpha"]), random_state=0).fit(Xh[a], yh[a]).score(Xh[b], yh[b]) for a, b in pl_h])
dist_h = {"max_depth": randint(1, 17), "min_samples_leaf": randint(1, 41), "max_features": uniform(0.2, 0.8), "ccp_alpha": loguniform(1e-5, 1e-1)}

t0 = time.perf_counter()
pool = list(ParameterSampler(dist_h, 300, random_state=0))                            # 300 configuraciones i.i.d. de la distribucion de Random Search
pool_sc = np.array([cv_arbol(p) for p in pool])
def rejilla_h(b, u):
    q = (np.arange(b) + u) / b                                                      # b cuantiles de CADA distribucion (misma cobertura); u desplaza la rejilla
    return {k: [dist_h[k].ppf(x).item() for x in q] for k in dist_h}
# Una rejilla es UNA tirada de suerte (depende de donde caen sus valores): se evaluan 3 disposiciones desplazadas (u = 0.25, 0.5, 0.75)
res_g = {b ** 4: [max(cv_arbol(p) for p in ParameterGrid(rejilla_h(b, u))) for u in (0.25, 0.5, 0.75)] for b in (2, 3)}
print(f"Evaluaciones: 300 (reserva de Random Search) + {3 * sum(b ** 4 for b in (2, 3))} (3 disposiciones de rejilla × b^4 puntos) en {time.perf_counter() - t0:.1f} s")

rng = np.random.default_rng(SEED)
Ns = [4, 8, 16, 32, 64, 81, 128]
sim = {N: pool_sc[rng.integers(0, len(pool_sc), size=(3000, N))].max(axis=1) for N in Ns}   # Random Search de presupuesto N = N muestras iid de la reserva
tabla = pd.DataFrame({"aleatoria (media)": [sim[N].mean() for N in Ns], "aleatoria (desv.)": [sim[N].std() for N in Ns],
                      "rejilla (media de 3)": [np.mean(res_g[N]) if N in res_g else np.nan for N in Ns],
                      "rejilla (mín)": [np.min(res_g[N]) if N in res_g else np.nan for N in Ns],
                      "rejilla (máx)": [np.max(res_g[N]) if N in res_g else np.nan for N in Ns],
                      "P(aleatoria > rejilla media)": [np.mean(sim[N] > np.mean(res_g[N])) if N in res_g else np.nan for N in Ns]}, index=pd.Index(Ns, name="presupuesto N"))
display(tabla.round(4))

fig, ax = plt.subplots(figsize=(10, 4.8))
ax.plot(Ns, tabla["aleatoria (media)"], "s-", color="#16a34a", lw=2, label="Random Search (media de 3 000 repeticiones)")
ax.fill_between(Ns, tabla["aleatoria (media)"] - tabla["aleatoria (desv.)"], tabla["aleatoria (media)"] + tabla["aleatoria (desv.)"], color="#16a34a", alpha=0.15)
for N, v in res_g.items():
    ax.plot([N] * 3, v, "o", color="#dc2626", ms=7, alpha=0.7, label="Rejilla $b^4$ (3 disposiciones)" if N == 16 else None)
    ax.plot([N], [np.mean(v)], "D", color="#7f1d1d", ms=9, label="Rejilla (media)" if N == 16 else None)
ax.set_xscale("log"); ax.set_xlabel("Presupuesto $N$ (evaluaciones, escala log)"); ax.set_ylabel("Mejor accuracy CV encontrado")
ax.set_title("4 hiperparámetros del árbol: mejor puntaje vs. presupuesto", fontweight="bold"); ax.legend(fontsize=8)
plt.tight_layout(); plt.show()
assert tabla.loc[16, "aleatoria (media)"] > tabla.loc[16, "rejilla (media de 3)"]                        # con presupuesto chico (N = 16) el azar supera en promedio a la rejilla 2^4
print(f"N = 16: Random Search {tabla.loc[16, 'aleatoria (media)']:.4f} frente a rejilla {tabla.loc[16, 'rejilla (media de 3)']:.4f} (media de 3 disposiciones; P(aleatoria > rejilla) = {tabla.loc[16, 'P(aleatoria > rejilla media)']:.2f}).")
print(f"N = 81: Random Search {tabla.loc[81, 'aleatoria (media)']:.4f} frente a rejilla {tabla.loc[81, 'rejilla (media de 3)']:.4f}, pero la rejilla oscila entre {tabla.loc[81, 'rejilla (mín)']:.4f} y {tabla.loc[81, 'rejilla (máx)']:.4f} según DÓNDE se coloquen sus valores: una rejilla es una tirada de suerte.")
print("Lectura: una rejilla de 3 valores por eje (81 puntos) solo prueba 3 profundidades distintas, mientras que 81 muestras aleatorias prueban 81. Con presupuestos chicos la ventaja del azar es clara; con N = 81 es un empate técnico.")
print("Honestidad: el efecto es REAL pero MODERADO aquí; los hiperparámetros del árbol no tienen una dimensión efectiva tan baja como la función de la sección 2. Matiz: la reserva de 300 muestras es finita (para N grande las repeticiones se parecen entre sí).")
````

## 6. Refinamiento «de Grueso a Fino» (*Coarse-to-Fine*) 🔍

Una estrategia práctica que combina lo mejor de ambos mundos: **(1)** Random Search **amplio** con pocas muestras para localizar la región prometedora; **(2)** construir una **caja** alrededor de los mejores puntos (con un margen) y **(3)** buscar de nuevo **solo dentro de la caja**, con la misma cantidad de puntos pero mucha más resolución. Se puede iterar. Géron (Cap. 2) ilustra la primera mitad: en su ejemplo el mejor `n_clusters` resultó ser **15, el máximo de los valores evaluados**, y advierte que *«probablemente deberías volver a buscar con valores más altos»* (traducción libre, verificada en el PDF); acotar la búsqueda alrededor de la zona prometedora es la mitad complementaria.

El coste total es el de ambas etapas; la pregunta honesta es si **gana a una sola etapa con el mismo presupuesto total** (aquí 12 + 12 frente a 24). Lo medimos con 4 semillas sobre la superficie del SVM de la sección 5a, con evaluaciones reales (no la tabla) en cada punto.

````python
# ============================================================
# 6. Refinamiento "de grueso a fino" (coarse-to-fine): busqueda amplia + busqueda en la zona prometedora
# ============================================================
def cv_log(lc, lg):                                                                # puntaje CV 3 pliegues en coordenadas (log10 C, log10 gamma)
    return cv_svc(lc, lg)

def busqueda_aleatoria(rng, n, caja):
    """n puntos uniformes en log10 dentro de caja = (lcmin, lcmax, lgmin, lgmax) -> equivale a loguniform en C y gamma."""
    P = np.c_[rng.uniform(caja[0], caja[1], n), rng.uniform(caja[2], caja[3], n)]
    return P, np.array([cv_log(*p) for p in P])

CAJA0 = (-2, 2, -6, 0)
filas, ejemplo = [], None
t0 = time.perf_counter()
for s in range(4):
    rng = np.random.default_rng(100 + s)
    P1, s1 = busqueda_aleatoria(rng, 12, CAJA0)                                    # etapa 1: amplia
    top = P1[np.argsort(-s1)[:4]]                                                  # los 4 mejores puntos de la etapa 1
    caja1 = (max(CAJA0[0], top[:, 0].min() - 0.5), min(CAJA0[1], top[:, 0].max() + 0.5), max(CAJA0[2], top[:, 1].min() - 0.5), min(CAJA0[3], top[:, 1].max() + 0.5))
    P2, s2 = busqueda_aleatoria(rng, 12, caja1)                                    # etapa 2: zoom (misma cantidad de puntos)
    P3, s3 = busqueda_aleatoria(rng, 24, CAJA0)                                    # referencia: UNA etapa con los mismos 32 puntos
    filas.append({"semilla": s, "1 etapa × 24": s3.max(), "etapa 1 (12)": s1.max(), "2 etapas (12 + 12)": max(s1.max(), s2.max())})
    if s == 0: ejemplo = (P1, s1, P2, s2, caja1)
tab = pd.DataFrame(filas).set_index("semilla"); display(tab.round(4))
print(f"Medias: 1 etapa × 24 = {tab['1 etapa × 24'].mean():.4f} | 2 etapas (12 + 12) = {tab['2 etapas (12 + 12)'].mean():.4f}  (tiempo = {time.perf_counter() - t0:.1f} s)")
assert (tab["2 etapas (12 + 12)"] >= tab["etapa 1 (12)"]).all()

P1, s1, P2, s2, caja1 = ejemplo
fig, ax = plt.subplots(figsize=(9, 5.4))
im = ax.imshow(superficie, origin="lower", cmap="viridis", extent=[logG[0], logG[-1], logC[0], logC[-1]], aspect="auto", alpha=0.85)
ax.scatter(P1[:, 1], P1[:, 0], s=60, facecolor="white", edgecolor="black", label="Etapa 1 (amplia, 12)")
ax.scatter(P2[:, 1], P2[:, 0], s=90, marker="*", color="#f59e0b", edgecolor="black", label="Etapa 2 (zoom, 12)")
ax.add_patch(plt.Rectangle((caja1[2], caja1[0]), caja1[3] - caja1[2], caja1[1] - caja1[0], fill=False, ec="#dc2626", lw=2, ls="--", label="Caja de la etapa 2"))
plt.colorbar(im, label="Accuracy CV"); ax.set_xlabel("$\\log_{10}\\gamma$"); ax.set_ylabel("$\\log_{10} C$"); ax.legend(fontsize=8, loc="lower left")
ax.set_title("Coarse-to-fine sobre la superficie del SVM", fontweight="bold")
plt.tight_layout(); plt.show()
print("Honestamente: con 4 semillas y un presupuesto tan pequeño la ganancia de la segunda etapa es modesta y puede no superar a una sola etapa; el zoom rinde cuando la región buena es MUY estrecha respecto al rango inicial.")
print("Reglas: (1) la caja de la etapa 2 se construye con los mejores puntos y un margen (aquí ± 0.5 décadas); (2) no la estrangules: el óptimo puede estar en el borde; (3) vuelve a ampliar si el mejor punto toca el borde de la caja.")
````

## 7. Reducción Sucesiva (*Successive Halving*) ✂️

**Idea** (Jamieson y Talwalkar, 2016; base de *Hyperband*, Li *et al.*, 2017): si evaluar una configuración con **pocos recursos** (pocos datos, pocas iteraciones, pocos árboles) ya permite **descartar a las malas**, no tiene sentido pagar el costo completo por todas. Con factor $\eta$ (aquí 3) y $R$ rondas:

1. Empieza con $n_0=\eta^{R-1}\cdot m$ candidatos y recurso mínimo $r_0$.
2. En la ronda $i$, evalúa los candidatos vivos con recurso $r_i=r_0\eta^i$ y conserva el **mejor $1/\eta$**.
3. La última ronda usa el recurso completo con muy pocos candidatos.

Con 27 candidatos, $\eta=3$ y tres rondas con $N/9,\,N/3,\,N$ ejemplos, el costo es $27\cdot\tfrac N9+9\cdot\tfrac N3+3N=9N$: **lo mismo que 9 configuraciones completas**, pero habiendo mirado 27. El riesgo: **el orden con pocos datos puede diferir del orden con todos** (un `gamma` óptimo depende de $n$) y entonces se descarta al ganador. La clase `HalvingRandomSearchCV` de `scikit-learn` implementa esto (con `resource='n_samples'` o cualquier parámetro del estimador, p. ej. `n_estimators`); sigue siendo **experimental** (hay que habilitarla con `enable_halving_search_cv`), por eso se importa dentro de `try/except`.

````python
# ============================================================
# 7. Reduccion sucesiva (successive halving) DESDE CERO + HalvingRandomSearchCV (experimental, opcional)
# ============================================================
Xs, ys = make_classification(n_samples=1500, n_features=20, n_informative=6, n_redundant=4, n_clusters_per_class=3, flip_y=0.05, class_sep=1.0, random_state=SEED)
X_pool, X_te, y_pool, y_te = train_test_split(Xs, ys, test_size=0.4, stratify=ys, random_state=SEED)       # prueba final: 600
X_ent, X_val, y_ent, y_val = train_test_split(X_pool, y_pool, test_size=1 / 3, stratify=y_pool, random_state=SEED)   # entrenamiento 600 / validacion 300
dist_s = {"C": loguniform(1e-2, 1e3), "gamma": loguniform(1e-5, 1e0)}
N_FULL = len(y_ent)

def puntuar_svm(p, r, perm):
    """Entrena con los primeros r ejemplos de una permutacion fija del entrenamiento y puntua en la validacion."""
    i = perm[:r]
    return make_pipeline(StandardScaler(), SVC(C=p["C"], gamma=p["gamma"])).fit(X_ent[i], y_ent[i]).score(X_val, y_val)

def successive_halving(cands, perm, eta=3, rondas=3):
    """Ronda i: evalua los candidatos vivos con r_i = N / eta^(rondas-1-i) ejemplos y conserva el mejor 1/eta. Devuelve (mejor, costo)."""
    vivos, costo = list(range(len(cands))), 0
    for i in range(rondas):
        r = N_FULL // eta ** (rondas - 1 - i)
        sc = np.array([puntuar_svm(cands[j], r, perm) for j in vivos]); costo += len(vivos) * r
        orden = np.argsort(-sc, kind="stable")
        vivos = [vivos[j] for j in orden[: max(1, len(vivos) // eta)]] if i < rondas - 1 else [vivos[orden[0]]]
    return vivos[0], costo

filas = []
t0 = time.perf_counter()
for s in range(10):
    rng = np.random.RandomState(s); perm = rng.permutation(N_FULL)
    c27 = list(ParameterSampler(dist_s, 27, random_state=s))
    j_sh, costo_sh = successive_halving(c27, perm)                                  # 27 candidatos -> 9 -> 3 -> 1 (costo ~ 9 ajustes completos)
    c9 = c27[:9]                                                                    # Random Search con el MISMO costo: 9 candidatos con todos los datos
    sc9 = [puntuar_svm(c, N_FULL, perm) for c in c9]; j_rs = int(np.argmax(sc9))
    final = lambda c: make_pipeline(StandardScaler(), SVC(C=c["C"], gamma=c["gamma"])).fit(X_ent, y_ent).score(X_te, y_te)
    filas.append({"sh_prueba": final(c27[j_sh]), "rs9_prueba": final(c9[j_rs]), "sh_costo": costo_sh, "rs9_costo": 9 * N_FULL})
res = pd.DataFrame(filas)
display(pd.DataFrame({"accuracy en prueba (media ± desv., 10 semillas)": [f"{res[c].mean():.4f} ± {res[c].std():.4f}" for c in ("sh_prueba", "rs9_prueba")],
                      "costo (ejemplos de entrenamiento procesados)": [int(res["sh_costo"].mean()), int(res["rs9_costo"].mean())]},
                     index=["Successive halving (27 → 9 → 3 → 1)", "Random Search, 9 candidatos completos"]))
print(f"(tiempo = {time.perf_counter() - t0:.1f} s)")
assert abs(res["sh_costo"].mean() - res["rs9_costo"].mean()) / res["rs9_costo"].mean() < 0.05           # mismo costo (±5 %)
print("Lectura: halving evalúa 27 candidatos pagando lo de 9. Si los puntajes con pocos datos ORDENAN igual que con todos los datos, ve más candidatos por el mismo costo;")
print("si el orden cambia con el tamaño (p. ej. el mejor gamma depende de n), puede descartar al ganador en las primeras rondas. Aquí la comparación es honesta: mira la media Y la desviación.")

# Version de scikit-learn (experimental): se importa dentro de try/except por si la API cambia entre versiones
try:
    from sklearn.experimental import enable_halving_search_cv  # noqa: F401
    from sklearn.model_selection import HalvingRandomSearchCV
    hs = HalvingRandomSearchCV(make_pipeline(StandardScaler(), SVC()), {"svc__C": loguniform(1e-2, 1e3), "svc__gamma": loguniform(1e-5, 1e0)},
                               n_candidates=27, factor=3, resource="n_samples", min_resources="exhaust", cv=3, random_state=SEED).fit(X_ent, y_ent)
    print(f"\nHalvingRandomSearchCV: candidatos por ronda = {hs.n_candidates_}, recursos (n) por ronda = {hs.n_resources_}, mejor = C {hs.best_params_['svc__C']:.3g}, gamma {hs.best_params_['svc__gamma']:.3g}")
    print(f"Accuracy en la prueba con el reajustado: {hs.score(X_te, y_te):.4f}")
except Exception as e:
    print(f"\nHalvingRandomSearchCV no está disponible en esta versión de scikit-learn ({type(e).__name__}); la implementación propia anterior cubre la idea.")
````

## 8. Paralelismo, Costos y Reproducibilidad ⚙️

- **Costo.** $\text{ajustes}=n_{\text{iter}}\times k\ (+1\ \text{reajuste})$; el tiempo es ese número por el costo de un ajuste. Con Random Search tú **decides** $n_{\text{iter}}$: es una **perilla de presupuesto** (tiempo/dinero), no una consecuencia de la rejilla.
- **Paralelismo.** Cada ajuste es independiente: `n_jobs=-1` los reparte entre procesos (`joblib`). Ayuda si cada ajuste dura **decenas de milisegundos o más**; con ajustes muy cortos el **sobrecosto** de crear procesos y copiar datos puede **empeorar** el tiempo. Cuidado con la memoria (cada proceso copia los datos) y con paralelizar dos veces (`n_jobs` del buscador y `n_jobs` del bosque).
- **Reproducibilidad.** Hay **tres** fuentes de azar: la **búsqueda** (`RandomizedSearchCV(random_state=)`: qué configuraciones), el **estimador** (`random_state` del modelo: qué modelo sale de cada una) y el **divisor** de validación cruzada. Con las tres fijas el resultado es **idéntico** aunque cambie `n_jobs`; si falta alguna, no.

````python
# ============================================================
# 8. Paralelismo (n_jobs), costos y reproducibilidad (semillas)
# ============================================================
from sklearn.datasets import load_iris
print(f"CPUs visibles: {os.cpu_count()}")
Xi, yi = load_iris(return_X_y=True)
dist_arbol = {"max_depth": randint(2, 12), "min_samples_leaf": randint(1, 10)}
dist_rf = {"max_depth": randint(2, 12), "min_samples_leaf": randint(1, 10), "max_features": uniform(0.2, 0.8)}

def buscar(estimador, dists, n_iter, n_jobs, X=Xb, y=yb, seed=SEED):
    t0 = time.perf_counter()
    r = RandomizedSearchCV(estimador, dists, n_iter=n_iter, cv=3, random_state=seed, n_jobs=n_jobs, refit=False).fit(X, y)
    return r, time.perf_counter() - t0

# (a) Costo y paralelismo: tres estimadores de costo creciente (el mismo numero de CPU, distinto costo por ajuste)
casos = [("árbol sobre Iris (150 filas, ~0.5 ms/ajuste)", DecisionTreeClassifier(random_state=SEED), dist_arbol, 40, Xi, yi),
         ("árbol sobre Breast Cancer (~2 ms/ajuste)", DecisionTreeClassifier(random_state=SEED), dist_arbol, 40, Xb, yb),
         ("bosque de 60 árboles (~100 ms/ajuste)", RandomForestClassifier(n_estimators=60, random_state=SEED), dist_rf, 8, Xb, yb)]
filas, rf_runs = [], {}
for nombre, est, dists, n_it, XX, yy in casos:
    r1, t1 = buscar(est, dists, n_it, 1, XX, yy); rN, tN = buscar(est, dists, n_it, -1, XX, yy)
    filas.append({"estimador": nombre, "ajustes": n_it * 3, "n_jobs=1 (s)": t1, "n_jobs=-1 (s)": tN, "aceleración": t1 / tN})
    rf_runs[nombre] = (r1, rN)
display(pd.DataFrame(filas).set_index("estimador").round(2))
print("Regla: el costo total es n_iter × k pliegues × (tiempo de un ajuste). n_jobs reparte los ajustes entre procesos (joblib): la aceleración depende del COSTO por ajuste y del número de núcleos;")
print("con ajustes de fracciones de milisegundo el arranque de los procesos y la copia de datos pueden anular (o invertir) la ganancia. Mide siempre con tu máquina antes de fijar n_jobs=-1.")

# (b) Reproducibilidad: random_state de la BUSQUEDA (que configuraciones), del ESTIMADOR (que modelo sale de cada una) y del divisor
r1, rN = rf_runs[casos[2][0]]
assert r1.cv_results_["params"] == rN.cv_results_["params"]
assert np.allclose(r1.cv_results_["mean_test_score"], rN.cv_results_["mean_test_score"])         # n_jobs NO cambia el resultado (con semillas fijas)
arbol_azar = lambda: DecisionTreeClassifier(max_features=0.5)                                      # SIN random_state: el arbol desempata variables al azar
otra, _ = buscar(DecisionTreeClassifier(random_state=SEED), dist_arbol, 40, 1, seed=SEED + 1)
base, _ = buscar(DecisionTreeClassifier(random_state=SEED), dist_arbol, 40, 1)
assert otra.cv_results_["params"] != base.cv_results_["params"]                                   # otra semilla de busqueda -> otras configuraciones
s1_, _ = buscar(arbol_azar(), dist_arbol, 40, 1); s2_, _ = buscar(arbol_azar(), dist_arbol, 40, 1)
assert s1_.cv_results_["params"] == s2_.cv_results_["params"]                                     # mismas configuraciones...
dif = np.abs(s1_.cv_results_["mean_test_score"] - s2_.cv_results_["mean_test_score"]).max()
assert dif > 0                                                                                    # ...pero puntajes distintos si el ESTIMADOR no tiene semilla
print("Misma semilla de búsqueda y de estimador: resultados idénticos con n_jobs=1 y n_jobs=-1 ✔ | otra semilla de búsqueda: configuraciones distintas ✔")
print(f"Estimador SIN random_state: las mismas configuraciones dan puntajes distintos entre ejecuciones (diferencia máxima = {dif:.4f}).")
print("Buenas prácticas: fija random_state en la búsqueda, en el estimador y en el divisor de validación cruzada; guarda cv_results_; y repite con varias semillas para saber cuánto de un resultado es suerte.")
````

## 9. ¿Cuándo Gana Cada Uno? Guía Práctica 🧭

| Situación | Recomendación | Por qué |
|---|---|---|
| **1–2 hiperparámetros**, rangos conocidos, ajustes **baratos** | **Rejilla** (fina, incluyendo los bordes) | Cubre todo; la sección 5a muestra que compite con el azar; produce un **mapa de calor** interpretable |
| **≥ 3–4 hiperparámetros** o no sabes cuáles importan | **Random Search** | Explora $N$ valores por eje (sección 5b: ventaja clara con presupuestos chicos); el costo lo eliges tú; ampliable sin rehacer nada |
| Hay hiperparámetros **continuos de escala amplia** (`C`, `gamma`, tasa de aprendizaje) | Random Search con **`loguniform`** | La escala logarítmica evita desperdiciar el 90 % de las muestras |
| Sabes que el óptimo está en una **zona pequeña** | **Coarse-to-fine** (aleatoria amplia → zoom) | Resolución donde importa, con el mismo presupuesto |
| Ajustar es **muy costoso** (miles de segundos) y se puede evaluar con menos datos/iteraciones | **Successive halving / Hyperband** | Descarta pronto a las malas (riesgo: el orden con poco presupuesto puede ser engañoso) |
| Ajustar es **muy costoso** y el espacio es continuo y de pocas dimensiones | **Optimización bayesiana** ([cuaderno 03](03_Optimizacion_Bayesiana.ipynb)) | Usa lo ya evaluado para decidir qué probar; sin memoria, el azar desperdicia información |
| Siempre | **Fija semillas, guarda `cv_results_`, usa validación anidada o prueba sellada** | Un número de búsqueda es optimista ([cuaderno 01](01_Validacion_Cruzada_Avanzada.ipynb)) |

> 🔑 **Idea para llevar:** Random Search no es «mejor que la rejilla» en abstracto: es **mejor cuando hay muchas dimensiones y pocas importan**, y **equivalente o peor** cuando hay pocas y el espacio es suave. Lo robusto es **medirlo** (curvas de mejor puntaje vs. presupuesto), no suponerlo.

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Rejilla** | $k\prod m_i$ ajustes (exponencial en $d$); con $N=k^d$ solo prueba $N^{1/d}$ valores distintos por eje; los hiperparámetros irrelevantes multiplican el costo |
| **Dimensión efectiva** | Si pocos hiperparámetros importan, **Random Search** prueba $N$ valores de cada uno. En la demostración propia ($d=3$, solo $x_1$ pesa) ganó a la rejilla con **todos** los presupuestos ($N=8$ a $125$) |
| **Cobertura probabilística** | $1-(1-p)^N$: **59** muestras bastan para caer en el mejor 5 % con 95 % de confianza, **sin importar $d$** |
| **Distribuciones** | `loguniform` para `C`, `gamma`, `alpha`...; `randint` (extremo superior excluido); `uniform(loc, scale)`; con listas, `scikit-learn` muestrea la rejilla **sin reemplazo** |
| **Desde cero** | Mismo `RandomState`, claves ordenadas, `dist.rvs(random_state=rng)`: **idéntico** a `RandomizedSearchCV` (configuraciones y puntajes) |
| **Resultados reales** | **2 dimensiones** (SVM): **sin ganador sistemático**; ambos a menos de un punto del máximo desde $N\approx9$. **4 dimensiones** (árbol): Random Search gana con presupuesto chico ($N=16$) y **empata** con $N=81$; la rejilla es una tirada de suerte (varía con la disposición). Efecto real pero **moderado** |
| **Coarse-to-fine** | Zoom en la región prometedora; con presupuestos pequeños la ganancia fue **modesta** |
| **Successive halving** | Evalúa 27 candidatos al costo de 9 completos; funciona si el orden con pocos recursos predice el orden completo. `HalvingRandomSearchCV` es **experimental** |
| **Operación** | `n_jobs` ayuda con ajustes lentos y **perjudica** con ajustes de ~1 ms; con `random_state` en búsqueda, estimador y divisor, el resultado es **idéntico** |
