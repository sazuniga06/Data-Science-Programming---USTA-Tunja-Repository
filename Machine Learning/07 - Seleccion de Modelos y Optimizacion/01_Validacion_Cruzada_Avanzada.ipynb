## 1. *k*-fold Repetido: la Estimación por CV También Tiene Varianza 🔁

Una validación cruzada produce **un número**, y ese número depende de **cómo se partieron los datos**: otra semilla, otro resultado. Con $n$ pequeño la diferencia entre dos particiones puede superar la diferencia entre dos modelos. El ***k*-fold repetido** (*repeated k-fold*) ejecuta $r$ veces el *k*-fold con particiones aleatorias distintas y promedia los $k\cdot r$ puntajes:

$$\widehat{\text{Err}}_{k\times r}=\frac1{r}\sum_{\rho=1}^{r}\widehat{\text{Err}}_{\text{CV}}^{(\rho)}.$$

Si las $r$ particiones fueran independientes, la **varianza por partición** bajaría como $1/r$ (la desviación como $1/\sqrt r$). Hay **dos fuentes de ruido** que no hay que confundir: (i) **la partición** (que la repetición reduce) y (ii) **el conjunto de datos finito** (que **no** se reduce repitiendo: con otro conjunto de $n$ ejemplos obtendrías otro valor). Por eso el *k*-fold repetido **estabiliza** la estimación, pero no la vuelve más «verdadera»; Bates, Hastie y Tibshirani (2023) argumentan que la CV estima mejor el error **promedio sobre conjuntos de entrenamiento** de tamaño $n(k-1)/k$ que el error del modelo concreto ajustado con *estos* datos, y que su error estándar real es **mayor** que el que sugiere el cálculo ingenuo con los pliegues *(citado de memoria, verificar)*.

Construimos `RepeatedKFold` desde cero (un `RandomState` compartido que se baraja en cada repetición, como hace `scikit-learn`) y medimos la **dispersión** de la estimación entre 20 semillas para 1, 4 y 10 repeticiones.

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas matplotlib seaborn scikit-learn scipy

import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy
import sklearn
import itertools
from sklearn.base import clone
from sklearn.datasets import load_breast_cancer, make_classification
from sklearn.model_selection import (KFold, StratifiedKFold, RepeatedKFold, RepeatedStratifiedKFold, GroupKFold,
                                     TimeSeriesSplit, ShuffleSplit, GridSearchCV, ParameterGrid, cross_val_score,
                                     train_test_split)
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import LogisticRegression, Ridge
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.metrics import average_precision_score

bc = load_breast_cancer()
X, y = bc.data, bc.target
warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (9, 5)
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"Entorno listo | numpy {np.__version__} | scipy {scipy.__version__} | scikit-learn {sklearn.__version__}")
````

````python
# ============================================================
# 1. k-fold repetido DESDE CERO (verificado) y la varianza de la estimacion por CV
# ============================================================
def repeated_kfold_desde_cero(n, k, r, seed):
    """r repeticiones de k-fold; un RandomState COMPARTIDO se baraja en cada repeticion (el mismo mecanismo de sklearn)."""
    rng = np.random.RandomState(seed)
    for _ in range(r):
        idx = np.arange(n); rng.shuffle(idx)
        tam = np.full(k, n // k); tam[: n % k] += 1
        ini = 0
        for t in tam:
            prueba = np.sort(idx[ini: ini + t]); ent = np.sort(np.concatenate([idx[:ini], idx[ini + t:]]))
            yield ent, prueba
            ini += t

for n_, k_, r_ in [(103, 5, 3), (50, 10, 2)]:
    mios = list(repeated_kfold_desde_cero(n_, k_, r_, SEED))
    sk = list(RepeatedKFold(n_splits=k_, n_repeats=r_, random_state=SEED).split(np.zeros((n_, 1))))
    assert len(mios) == len(sk) == k_ * r_ and all(np.array_equal(a1, a2) and np.array_equal(b1, b2) for (a1, b1), (a2, b2) in zip(mios, sk))
print("RepeatedKFold desde cero == sklearn.RepeatedKFold (mismos pliegues exactos, con n no divisible entre k).")

# Experimento: la estimacion CV de UN mismo modelo cambia con la particion. ¿Cuanto baja esa variabilidad al repetir?
modelo = GaussianNB()
semillas, repeticiones = range(20), (1, 4, 10)

def acc_cv(cv):
    """Accuracy medio de 'modelo' sobre los pliegues de 'cv' (bucle propio: mas rapido que cross_val_score para miles de llamadas)."""
    return np.mean([clone(modelo).fit(X[a], y[a]).score(X[b], y[b]) for a, b in cv.split(X, y)])
t0 = time.perf_counter()
est = {r: [] for r in repeticiones}; est_mc = {r: [] for r in repeticiones}
for s in semillas:
    for r in repeticiones:
        est[r].append(acc_cv(RepeatedStratifiedKFold(n_splits=5, n_repeats=r, random_state=s)))
        est_mc[r].append(acc_cv(ShuffleSplit(n_splits=5 * r, test_size=0.2, random_state=s)))   # Monte Carlo CV (seccion 5.2)
t_rep = time.perf_counter() - t0
tab = pd.DataFrame({"fits por estimación": [5 * r for r in repeticiones],
                    "media de la estimación": [np.mean(est[r]) for r in repeticiones],
                    "desv. estándar entre particiones": [np.std(est[r]) for r in repeticiones],
                    "rango (máx − mín)": [np.ptp(est[r]) for r in repeticiones]},
                   index=pd.Index([f"5-fold × {r}" for r in repeticiones], name="esquema"))
display(tab.round(4))
assert tab["desv. estándar entre particiones"].is_monotonic_decreasing                       # mas repeticiones -> menos ruido de particion

fig, ax = plt.subplots(figsize=(10, 4.6))
ax.boxplot([est[r] for r in repeticiones], widths=0.5, patch_artist=True, boxprops=dict(facecolor="#bfdbfe"), medianprops=dict(color="#dc2626", lw=2))
ax.set_xticks(range(1, len(repeticiones) + 1)); ax.set_xticklabels([f"5-fold × {r}\n({5 * r} ajustes)" for r in repeticiones])
ax.set_ylabel("Accuracy estimada por CV (Naive Bayes, Breast Cancer)")
ax.set_title("20 particiones aleatorias distintas: el número cambia con la semilla", fontweight="bold")
plt.tight_layout(); plt.show()
print(f"(tiempo = {t_rep:.1f} s) Con un solo 5-fold, dos semillas pueden diferir hasta {100 * tab.iloc[0]['rango (máx − mín)']:.1f} puntos de accuracy; repetir lo reduce (≈ 1/√r si las particiones son independientes).")
print("Pero cuidado: repetir reduce el ruido de la PARTICIÓN, no la incertidumbre por tener solo n datos. La estimación converge al promedio sobre particiones de ESTE conjunto, no al error verdadero.")
````

## 2. Validación Cruzada por Grupos: Cuando las Filas No Son Independientes 👥

El *k*-fold supone que **las filas son independientes**: cualquier fila puede ir a entrenamiento o a prueba sin que eso diga nada sobre otra. Con **medidas repetidas** (varias filas por paciente, cliente, sensor, documento o sesión) es falso: las filas de un mismo grupo comparten una **firma** común. Si algunas filas del paciente $p$ están en entrenamiento y otras en prueba, el modelo puede **reconocer al paciente** y copiar su etiqueta; la estimación mide **«¿reconozco a pacientes que ya vi?»** y no **«¿generalizo a pacientes nuevos?»**, que es lo que importa en producción.

La solución es **partir por grupos**: `GroupKFold` reparte **grupos completos**, de modo que ningún grupo aparece a la vez en entrenamiento y prueba. Algoritmo (la versión clásica, voraz): ordenar los grupos del más grande al más pequeño y asignar cada uno al pliegue con **menos filas** acumuladas. Variantes: `StratifiedGroupKFold` (además conserva la proporción de clases), `LeaveOneGroupOut` y `GroupShuffleSplit`.

Simulamos **100 pacientes con 20 mediciones** cada uno. Cada paciente tiene una firma propia grande y una **señal débil real** que depende de su estado (enfermo/sano). Comparamos el mismo bosque con `KFold` barajado (ignora grupos) y con `GroupKFold`.

````python
# ============================================================
# 2. Validacion por grupos: varias filas por paciente -> fuga si se ignoran los grupos
# ============================================================
# 100 pacientes x 20 mediciones. Cada paciente tiene una "firma" propia (offset de 10 variables) y una etiqueta a nivel de paciente
# (enfermo / sano) con una senal DEBIL real (+-0.2 en 3 variables). Un modelo que reconoce al paciente "recuerda" su etiqueta.
rng = np.random.default_rng(SEED)
P, FILAS, D = 100, 20, 10
enfermo = rng.integers(0, 2, P)
firma = rng.normal(size=(P, D))
senal = np.r_[np.ones(3), np.zeros(D - 3)]
Xg = np.vstack([firma[p] + 0.2 * (2 * enfermo[p] - 1) * senal + rng.normal(0, 0.5, (FILAS, D)) for p in range(P)])
yg, grupos = np.repeat(enfermo, FILAS), np.repeat(np.arange(P), FILAS)
print(f"{len(yg)} filas de {P} pacientes ({FILAS} por paciente); prevalencia de 'enfermo' = {yg.mean():.2f}")

def group_kfold_desde_cero(grupos, k):
    """Reparte GRUPOS completos: del mas grande al mas chico, cada uno al pliegue con menos filas (algoritmo voraz)."""
    ids, cont = np.unique(grupos, return_counts=True)
    carga, asign = np.zeros(k), {}
    for g in ids[np.argsort(-cont, kind="stable")]:
        j = int(np.argmin(carga)); asign[g] = j; carga[j] += (grupos == g).sum()
    pliegue = np.array([asign[g] for g in grupos])
    for j in range(k):
        yield np.where(pliegue != j)[0], np.where(pliegue == j)[0]

# Propiedad clave (verificada en nuestra version y en la de sklearn): NINGUN grupo aparece a la vez en entrenamiento y prueba
for generador in (group_kfold_desde_cero(grupos, 5), GroupKFold(5).split(Xg, yg, grupos)):
    tam = []
    for ent, pru in generador:
        assert len(np.intersect1d(grupos[ent], grupos[pru])) == 0
        tam.append(len(pru))
    assert max(tam) - min(tam) <= FILAS                                           # pliegues casi del mismo tamano
print("Ningún paciente queda en entrenamiento y prueba a la vez (GroupKFold propio y de scikit-learn).")

rf = RandomForestClassifier(n_estimators=50, random_state=SEED, n_jobs=1)
acc_mal = cross_val_score(rf, Xg, yg, cv=KFold(5, shuffle=True, random_state=SEED))                       # ignora los grupos
acc_bien = cross_val_score(rf, Xg, yg, cv=GroupKFold(5), groups=grupos)                                  # respeta los grupos
comparte = [len(np.intersect1d(grupos[a], grupos[b])) for a, b in KFold(5, shuffle=True, random_state=SEED).split(Xg)]
display(pd.DataFrame({"accuracy medio": [acc_mal.mean(), acc_bien.mean()], "desv. entre pliegues": [acc_mal.std(), acc_bien.std()],
                      "pacientes presentes en entrenamiento Y prueba (por pliegue)": [f"{np.mean(comparte):.0f} de {P}", "0"]},
                     index=["KFold barajado (ignora grupos)", "GroupKFold"]).round(3))
assert acc_mal.mean() > 0.75 and acc_bien.mean() < 0.65                                                   # fuga enorme frente al nivel real (azar = 0.5)

fig, ax = plt.subplots(figsize=(8, 4.4))
ax.bar(["KFold barajado\n(ignora grupos)", "GroupKFold"], [acc_mal.mean(), acc_bien.mean()], yerr=[acc_mal.std(), acc_bien.std()],
       color=["#dc2626", "#16a34a"], capsize=6)
ax.axhline(0.5, color="black", ls="--", label="Azar (prevalencia ≈ 0.5)")
ax.set_ylabel("Accuracy de validación cruzada"); ax.set_ylim(0.4, 1.0); ax.legend(fontsize=9)
ax.set_title("Mismo bosque, mismos datos: la fuga por grupos infla el accuracy", fontweight="bold")
plt.tight_layout(); plt.show()
print(f"Con KFold el bosque 'reconoce' al paciente por su firma y copia su etiqueta: {acc_mal.mean():.0%}. Para PACIENTES NUEVOS (lo que importa en producción) rinde ≈ {acc_bien.mean():.0%}.")
print("Regla: si varias filas comparten un origen (paciente, cliente, sensor, documento, sesión), la unidad de partición es el GRUPO. Variantes: StratifiedGroupKFold (conserva la proporción de clases) y LeaveOneGroupOut.")
````

## 3. Series de Tiempo: Nunca Entrenar con el Futuro ⏳

En una serie temporal las observaciones están **ordenadas y autocorrelacionadas**: el valor de hoy se parece al de ayer. Si barajas, el modelo se entrena con puntos **anteriores y posteriores** a cada punto de prueba. En producción solo tendrás el **pasado**. La validación debe **imitar el uso real**:

- ***Walk-forward* / ventana expansiva** (`TimeSeriesSplit`): en el pliegue $j$ se entrena con $[0,\,s_j)$ y se prueba en el bloque siguiente $[s_j,\,s_j+m)$; el entrenamiento **crece**. Con $n$ observaciones y $k$ pliegues, el bloque de prueba mide $m=\lfloor n/(k+1)\rfloor$.
- **Ventana deslizante** (`max_train_size`): el entrenamiento tiene tamaño fijo; útil si el proceso **cambia** con el tiempo (*concept drift*).
- ***Gap*** (`gap`): descarta periodos entre entrenamiento y prueba. Es la versión simple de la **purga** y el **embargo** de López de Prado (2018): si la etiqueta en $t$ usa información hasta $t+H$ (p. ej. el retorno a $H$ días), las observaciones de entrenamiento con etiquetas que se solapan con la prueba **filtran información** y deben eliminarse (purga); además se deja un margen tras la prueba (embargo) *(citado de memoria, verificar)*.

Lo demostramos con una serie con **tendencia, estacionalidad y ruido autocorrelacionado** y dos formas de construir las variables: **(A)** variables de calendario (el tiempo mismo) con un bosque, y **(B)** tres rezagos con una regresión Ridge. Implementamos el *walk-forward* desde cero con `gap` y ventana deslizante y lo verificamos contra `TimeSeriesSplit`.

````python
# ============================================================
# 3. Series de tiempo: walk-forward DESDE CERO (verificado) y por que barajar es trampa
# ============================================================
def walk_forward_desde_cero(n, k, gap=0, max_train=None):
    """Pliegues 'expanding window' (o ventana deslizante con max_train): entrena SIEMPRE con el pasado y prueba en un bloque futuro."""
    tam_prueba = n // (k + 1)
    for ini in range(n - k * tam_prueba, n, tam_prueba):
        fin_ent = ini - gap                                                        # 'gap' = periodos descartados entre entrenamiento y prueba
        ent = np.arange(max(0, fin_ent - max_train) if max_train else 0, fin_ent)
        yield ent, np.arange(ini, ini + tam_prueba)

for n_, k_, kw in [(103, 5, {}), (103, 5, {"gap": 3}), (103, 4, {"max_train": 40}), (60, 3, {"gap": 2, "max_train": 25})]:
    mios = list(walk_forward_desde_cero(n_, k_, kw.get("gap", 0), kw.get("max_train")))
    sk = list(TimeSeriesSplit(n_splits=k_, gap=kw.get("gap", 0), max_train_size=kw.get("max_train")).split(np.zeros((n_, 1))))
    assert len(mios) == len(sk) and all(np.array_equal(a1, a2) and np.array_equal(b1, b2) for (a1, b1), (a2, b2) in zip(mios, sk))
    assert all(a.max() < b.min() for a, b in mios)                                  # el entrenamiento SIEMPRE es anterior a la prueba
print("walk-forward desde cero == TimeSeriesSplit (expanding, con gap y con ventana deslizante); nunca se entrena con el futuro.")

# Serie sintetica: tendencia + estacionalidad (periodo 50) + ruido autocorrelacionado AR(1)
rng = np.random.default_rng(SEED)
n = 600; t = np.arange(n)
e = np.zeros(n)
for i in range(1, n): e[i] = 0.9 * e[i - 1] + rng.normal(0, 0.5)
serie = 0.01 * t + 2 * np.sin(2 * np.pi * t / 50) + e
X_cal = np.c_[t, np.sin(2 * np.pi * t / 50), np.cos(2 * np.pi * t / 50)]            # (A) variables de "calendario": el tiempo mismo
X_lag = np.c_[serie[2:-1], serie[1:-2], serie[:-3]]; y_lag = serie[3:]                # (B) tres rezagos: y_{t-1}, y_{t-2}, y_{t-3} -> y_t

def r2_por_pliegue(modelo, X, y, cv):
    out = []
    for a, b in cv.split(X):
        p = modelo.fit(X[a], y[a]).predict(X[b])
        out.append(1 - np.sum((y[b] - p) ** 2) / np.sum((y[b] - y[b].mean()) ** 2))
    return np.array(out)

filas = {}
for nombre, modelo_, Xm, ym in [("(A) calendario + bosque", RandomForestRegressor(50, random_state=SEED), X_cal, serie),
                                 ("(B) rezagos + Ridge", Ridge(1.0), X_lag, y_lag)]:
    filas[nombre] = {"KFold barajado": r2_por_pliegue(modelo_, Xm, ym, KFold(5, shuffle=True, random_state=SEED)).mean(),
                     "TimeSeriesSplit (walk-forward)": r2_por_pliegue(modelo_, Xm, ym, TimeSeriesSplit(5)).mean()}
res_ts = pd.DataFrame(filas).T
display(res_ts.round(3))
assert res_ts.loc["(A) calendario + bosque", "KFold barajado"] > 0.9 and res_ts.loc["(A) calendario + bosque", "TimeSeriesSplit (walk-forward)"] < 0.5

fig, axes = plt.subplots(1, 2, figsize=(15, 4.6), gridspec_kw={"width_ratios": [1.1, 1]})
for fila, (nombre, cv_) in enumerate([("KFold barajado", KFold(5, shuffle=True, random_state=SEED)), ("TimeSeriesSplit", TimeSeriesSplit(5))]):
    for j, (a, b) in enumerate(cv_.split(X_cal)):
        y0 = fila * 6.5 + j
        axes[0].scatter(a, np.full(len(a), y0), s=1.5, color="#93c5fd"); axes[0].scatter(b, np.full(len(b), y0), s=1.5, color="#dc2626")
axes[0].set_yticks([j for f in range(2) for j in [f * 6.5 + 2]]); axes[0].set_yticklabels(["KFold barajado", "TimeSeriesSplit"])
axes[0].set_xlabel("Tiempo $t$"); axes[0].set_title("Entrenamiento (azul) y prueba (rojo) por pliegue", fontweight="bold")
x = np.arange(len(res_ts)); w = 0.35
axes[1].bar(x - w / 2, res_ts["KFold barajado"], w, color="#dc2626", label="KFold barajado")
axes[1].bar(x + w / 2, res_ts["TimeSeriesSplit (walk-forward)"], w, color="#16a34a", label="TimeSeriesSplit")
axes[1].set_xticks(x); axes[1].set_xticklabels(res_ts.index, fontsize=8); axes[1].axhline(0, color="black", lw=0.8)
axes[1].set_ylabel("$R^2$ medio de validación"); axes[1].legend(fontsize=8); axes[1].set_title("¿Interpolar o pronosticar?", fontweight="bold")
plt.tight_layout(); plt.show()
print("Barajar deja a cada punto de prueba RODEADO de vecinos temporales en el entrenamiento: el modelo interpola. Pronosticar es EXTRAPOLAR hacia adelante; solo el walk-forward lo mide.")
print("Con variables de calendario (A) la diferencia es abismal (el bosque no extrapola la tendencia); con rezagos (B) es menor pero sigue habiendo optimismo: el resultado depende del modelo.")
print("Purga y embargo (López de Prado, 2018, citado de memoria): si la etiqueta mira H periodos hacia adelante, hay que descartar (gap ≥ H) las observaciones de entrenamiento cuya etiqueta se solapa con la prueba.")
````

## 4. Validación Cruzada Anidada: Estimar el Procedimiento, No Solo el Modelo 🪆

### 4.1 El problema

Cuando usas `GridSearchCV` y reportas `best_score_`, ese número es **el máximo** de los puntajes de CV de todas las combinaciones. Como vimos en el módulo 03, el máximo de variables aleatorias es **optimista** aunque cada una sea insesgada: los mismos pliegues se usaron para **elegir** y para **evaluar** (Varma & Simon, 2006; Cawley & Talbot, 2010).

### 4.2 La solución: dos bucles

```mermaid
flowchart TD
    D["Todos los datos"] --> OUT["BUCLE EXTERNO: k_ext pliegues"]
    OUT --> EN["Entrenamiento externo (k_ext − 1 pliegues)"]
    OUT --> VE["Pliegue externo de validación<br/>NO interviene en ninguna decisión"]
    EN --> IN["BUCLE INTERNO: CV de k_int pliegues<br/>sobre el entrenamiento externo"]
    IN --> HP["Elige hiperparámetros θ*"]
    HP --> RF["Reajusta con TODO el entrenamiento externo"]
    RF --> SC["Puntúa en el pliegue externo"]
    VE --> SC
    SC --> AV["Promedio de los k_ext puntajes =<br/>estimación del PROCEDIMIENTO completo"]
```

Cada pliegue externo repite **todo el proceso de selección** (incluida la búsqueda de hiperparámetros) con sus propios datos de entrenamiento y se evalúa en datos **que no participaron**. El promedio estima el rendimiento del **procedimiento** «elegir $\theta$ con CV interna y reajustar», que es exactamente lo que desplegarás. Cuesta $k_{\text{ext}}\times(|\Theta|\,k_{\text{int}}+1)$ ajustes. Raschka (Cap. 6) describe la variante de 5 pliegues externos × 2 internos.

> ⚠️ La CV anidada **no devuelve un modelo**: devuelve una **estimación**. Los hiperparámetros pueden **cambiar** de un pliegue externo a otro, y eso es **informativo** (si cambian mucho, la elección es inestable). El modelo final se obtiene aplicando el procedimiento a **todos** los datos.

### 4.3 Verificación y demostración cuantitativa

Primero comprobamos que nuestra versión **coincide con la composición de `scikit-learn`**, `cross_val_score(GridSearchCV(...), cv=externo)`, con las **mismas particiones**. Después **medimos el sesgo**: simulamos 30 problemas pequeños ($n=60$, 40 variables de las que solo 3 sirven) con una rejilla de 48 combinaciones, y comparamos el **mejor puntaje de CV**, el **puntaje anidado** y el **rendimiento verdadero** (medido sobre una población de 20 000 ejemplos).

````python
# ============================================================
# 4. Validacion cruzada ANIDADA desde cero == cross_val_score(GridSearchCV(...)) con las mismas particiones
# ============================================================
def busqueda_rejilla(puntuar, rejilla, X, y, pliegues):
    """Bucle INTERNO: devuelve (mejores_params, mejor_media_cv). 'puntuar(params, Xent, yent, Xval, yval)' -> accuracy."""
    mejor, mejores = -np.inf, None
    for params in ParameterGrid(rejilla):
        media = np.mean([puntuar(params, X[a], y[a], X[b], y[b]) for a, b in pliegues])
        if media > mejor: mejor, mejores = media, params                         # estricto '>' -> el primero gana los empates (como sklearn)
    return mejores, mejor

def cv_anidada_desde_cero(puntuar, rejilla, X, y, externo, interno):
    """Bucle EXTERNO: para cada pliegue externo, elige hiperparametros SOLO con su entrenamiento (validacion cruzada interna),
    reajusta con todo ese entrenamiento y puntua en el pliegue externo, que jamas intervino en la eleccion."""
    puntajes, elegidos = [], []
    for ent, val in externo.split(X, y):
        pliegues_int = list(interno.split(X[ent], y[ent]))
        mejores, _ = busqueda_rejilla(puntuar, rejilla, X[ent], y[ent], pliegues_int)
        puntajes.append(puntuar(mejores, X[ent], y[ent], X[val], y[val])); elegidos.append(mejores)
    return np.array(puntajes), elegidos

pipe = make_pipeline(StandardScaler(), SVC())
rejilla = {"svc__C": [0.1, 1, 10, 100], "svc__gamma": [1e-4, 1e-3, 1e-2, 1e-1, 1]}
puntuar_sk = lambda params, Xa, ya, Xb, yb: clone(pipe).set_params(**params).fit(Xa, ya).score(Xb, yb)
externo = StratifiedKFold(5, shuffle=True, random_state=SEED)
interno = StratifiedKFold(3, shuffle=True, random_state=SEED)

t0 = time.perf_counter()
mios, elegidos = cv_anidada_desde_cero(puntuar_sk, rejilla, X, y, externo, interno)
t_mio = time.perf_counter() - t0
t0 = time.perf_counter()
sk = cross_val_score(GridSearchCV(pipe, rejilla, cv=interno), X, y, cv=externo)         # la composicion manual de scikit-learn
t_sk = time.perf_counter() - t0
assert np.allclose(mios, sk)                                                            # MISMA estimacion, pliegue a pliegue
print(f"Anidada desde cero == cross_val_score(GridSearchCV): {np.round(mios, 4)}  (media = {mios.mean():.4f})")
print(f"Tiempo: propia = {t_mio:.1f} s | sklearn = {t_sk:.1f} s | ajustes = 5 externos × (20 combinaciones × 3 internos + 1 reajuste) = {5 * (20 * 3 + 1)}")
print("Hiperparámetros elegidos en cada pliegue externo (¡pueden cambiar de un pliegue a otro!):")
for j, p in enumerate(elegidos): print(f"  pliegue externo {j + 1}: C = {p['svc__C']:g}, gamma = {p['svc__gamma']:g}")
print("\nLa anidada NO produce un modelo final: estima el rendimiento del PROCEDIMIENTO «elegir hiperparámetros con CV y reajustar». El modelo final se obtiene aplicando ese mismo procedimiento a todos los datos.")
````

````python
# ============================================================
# 4b. Sesgo optimista: mejor puntaje de CV (no anidado) vs. anidado vs. error VERDADERO (poblacion conocida)
# ============================================================
# Problema: n = 60, 40 variables de las que solo 3 tienen senal. "Modelo": ridge sobre las m variables mas correlacionadas con y
# (la seleccion ocurre DENTRO de cada entrenamiento, sin fuga). Rejilla: m x lambda = 8 x 6 = 48 combinaciones. Todo en NumPy: muy rapido.
def generar(rng, n, p=40, delta=0.4, n_inf=3):
    yy = rng.integers(0, 2, n); XX = rng.normal(size=(n, p)); XX[:, :n_inf] += delta * (2 * yy[:, None] - 1)
    return XX, yy

def puntuar_np(params, Xa, ya, Xb, yb):
    m, lam = params["m"], params["lam"]
    yc = 2.0 * ya - 1; mu, sd = Xa.mean(0), Xa.std(0) + 1e-12; Z = (Xa - mu) / sd
    idx = np.argsort(-np.abs(Z.T @ yc))[:m]; Zs = Z[:, idx]                       # seleccion de variables con los datos de ENTRENAMIENTO
    w = np.linalg.solve(Zs.T @ Zs + lam * np.eye(m), Zs.T @ (yc - yc.mean()))
    return np.mean(((((Xb - mu) / sd)[:, idx] @ w + yc.mean()) > 0) == (yb == 1))

rejilla_np = {"m": [1, 2, 3, 5, 8, 13, 21, 40], "lam": [0.01, 0.1, 1, 10, 100, 1000]}
R_REPS, N_ENT = 30, 60
rng = np.random.default_rng(SEED)
X_pob, y_pob = generar(rng, 20000)                                                 # "poblacion" para medir el rendimiento VERDADERO
t0 = time.perf_counter()
filas = []
for r in range(R_REPS):
    Xs, ys = generar(rng, N_ENT)
    pl = list(StratifiedKFold(5, shuffle=True, random_state=r).split(Xs, ys))
    mejores, mejor_cv = busqueda_rejilla(puntuar_np, rejilla_np, Xs, ys, pl)        # (a) NO anidado: se reporta el mejor puntaje CV
    verdadero = puntuar_np(mejores, Xs, ys, X_pob, y_pob)                           # rendimiento real del modelo reajustado con todo
    anid, _ = cv_anidada_desde_cero(puntuar_np, rejilla_np, Xs, ys, StratifiedKFold(5, shuffle=True, random_state=1000 + r),
                                    StratifiedKFold(3, shuffle=True, random_state=r))
    filas.append((mejor_cv, anid.mean(), verdadero))
res = pd.DataFrame(filas, columns=["mejor CV (no anidado)", "CV anidada", "verdadero"])
sesgo = res[["mejor CV (no anidado)", "CV anidada"]].sub(res["verdadero"], axis=0)
tabla = pd.DataFrame({"media del puntaje": res.mean(), "sesgo medio (puntaje − verdadero)": [sesgo["mejor CV (no anidado)"].mean(), sesgo["CV anidada"].mean(), 0.0]})
display(tabla.round(3))
print(f"({R_REPS} conjuntos de n = {N_ENT}; {time.perf_counter() - t0:.1f} s. Para resultados más finos usa R_REPS = 200.)")
assert sesgo["mejor CV (no anidado)"].mean() > 0.04                                  # reportar el mejor CV es claramente optimista
assert abs(sesgo["CV anidada"].mean()) < 0.04 and abs(sesgo["CV anidada"].mean()) < sesgo["mejor CV (no anidado)"].mean() / 2    # la anidada esta mucho mas cerca de la verdad

fig, ax = plt.subplots(figsize=(9.5, 4.6))
ax.boxplot([sesgo["mejor CV (no anidado)"], sesgo["CV anidada"]], widths=0.5, patch_artist=True, boxprops=dict(facecolor="#bfdbfe"), medianprops=dict(color="#dc2626", lw=2))
ax.axhline(0, color="black", ls="--"); ax.set_xticks([1, 2]); ax.set_xticklabels(["Mejor puntaje CV\n(no anidado)", "CV anidada"])
ax.set_ylabel("Puntaje estimado − accuracy verdadero"); ax.set_title(f"Optimismo de la selección ({R_REPS} repeticiones, 48 combinaciones, n = {N_ENT})", fontweight="bold")
plt.tight_layout(); plt.show()
print(f"Reportar el mejor puntaje de la búsqueda sobreestima el rendimiento real en ≈ {100 * sesgo['mejor CV (no anidado)'].mean():.1f} puntos de accuracy; la anidada queda a {100 * sesgo['CV anidada'].mean():+.1f} puntos")
print("(el signo negativo es esperable: cada bucle interno entrena con menos datos que el modelo final, así que la anidada tiende a ser ligeramente PESIMISTA).")
print("Esto es el 'winner\'s curse' (módulo 03) cuantificado: cuantas más combinaciones, menos datos y más ruido, mayor el optimismo. Con n grande y pocas combinaciones, el sesgo se vuelve pequeño.")
````

## 5. Remuestreo: *Bootstrap* .632 / .632+ y CV Monte Carlo 🎒🎲

### 5.1 El estimador .632 y el .632+ (*bootstrap*)

Una **remuestra bootstrap** de tamaño $n$ (con reemplazo) contiene, en promedio, $1-(1-\tfrac1n)^n\to1-e^{-1}\approx63.2\,\%$ de los **ejemplos distintos**; el resto ($\approx36.8\,\%$) queda **fuera de la bolsa** (*out-of-bag*, OOB) y sirve para probar. Estimadores del error de clasificación a partir de $B$ remuestras:

- **Error de entrenamiento** $\overline{\text{err}}$: optimista (el modelo ya vio esos datos).
- **Error OOB** $\widehat{\text{Err}}^{(1)}$: promedio del error de cada modelo en sus ejemplos OOB; algo **pesimista**, porque cada modelo se entrenó con solo ≈ 63.2 % de los datos distintos.
- **Estimador .632** (Efron, 1983): $\ \widehat{\text{Err}}^{(.632)}=0.368\,\overline{\text{err}}+0.632\,\widehat{\text{Err}}^{(1)}$. Funciona bien con modelos poco flexibles, pero con un modelo que **memoriza** ($\overline{\text{err}}=0$) queda **demasiado optimista**.
- **Estimador .632+** (Efron y Tibshirani, 1997): corrige con la **tasa relativa de sobreajuste** $R=\dfrac{\widehat{\text{Err}}^{(1)}-\overline{\text{err}}}{\gamma-\overline{\text{err}}}$, donde $\gamma$ es la tasa de error **sin información** (predicciones independientes de $y$; con 0-1 y dos clases, $\gamma=\hat p_1(1-\hat q_1)+(1-\hat p_1)\hat q_1$), y un peso $w=\dfrac{0.632}{1-0.368\,R}$ (entre .632 y 1):
$$\widehat{\text{Err}}^{(.632+)}=(1-w)\,\overline{\text{err}}+w\,\widehat{\text{Err}}'^{(1)},\qquad \widehat{\text{Err}}'^{(1)}=\min\big(\widehat{\text{Err}}^{(1)},\gamma\big).$$

Con $R=1$ (sobreajuste total) $w=1$ y el .632+ se reduce al OOB; con $R=0$ es el .632. Lo implementamos y medimos su sesgo contra un **error verdadero** conocido. (En la práctica, el *k*-fold repetido es más sencillo y suele bastar; el .632+ es útil con muestras muy pequeñas.)

### 5.2 Validación cruzada Monte Carlo (*ShuffleSplit*)

La **CV Monte Carlo** (o *repeated random sub-sampling*, `ShuffleSplit`) repite $s$ veces: baraja los datos, separa una fracción de **prueba** (p. ej. 20 %) y entrena con el resto. A diferencia del *k*-fold, **el número de particiones $s$ y el tamaño de prueba son independientes** (útil con muchos datos o cuando se quiere una prueba grande). Su costo: **no garantiza cobertura uniforme**, pues algunos ejemplos se prueban varias veces y otros ninguna, y las pruebas **se solapan**. Es exactamente el esquema cuyo *t* ingenuo fallaba en el [cuaderno 00](00_Evaluacion_y_Seleccion_de_Modelos.ipynb) (30 remuestras 90/10).

Lo implementamos desde cero (verificado contra `ShuffleSplit`), contamos cuántas veces se prueba cada ejemplo y comparamos la **dispersión** con la del *k*-fold repetido (los resultados de la sección 1 se reutilizan, con el mismo número de evaluaciones).

````python
# ============================================================
# 5.1 Bootstrap .632 y .632+ (Efron, 1983; Efron y Tibshirani, 1997): fraccion unica, OOB y los estimadores
# ============================================================
# (a) Una remuestra bootstrap contiene en promedio 1 - (1 - 1/n)^n -> 1 - 1/e = 63.2 % de los ejemplos distintos
n_ = 500; rng = np.random.default_rng(SEED)
frac = np.mean([len(np.unique(rng.integers(0, n_, n_))) / n_ for _ in range(2000)])
assert abs(frac - (1 - (1 - 1 / n_) ** n_)) < 0.005 and abs(frac - (1 - np.exp(-1))) < 0.005
print(f"Fracción de ejemplos distintos en una remuestra bootstrap: {frac:.4f} (teoría 1 - (1-1/n)^n = {1 - (1 - 1 / n_) ** n_:.4f} → 1 - 1/e = {1 - np.exp(-1):.4f})")

def estimadores_bootstrap(fabrica, Xa, ya, B, rng):
    """Devuelve (err_entrenamiento, err_OOB, err_.632, err_.632+) para una tasa de error 0-1."""
    n = len(ya); modelo = fabrica().fit(Xa, ya); pred = modelo.predict(Xa)
    err_tr = np.mean(pred != ya)
    errs = []
    for _ in range(B):
        i = rng.integers(0, n, n); fuera = np.setdiff1d(np.arange(n), i)               # OOB: los ejemplos NO elegidos prueban el modelo
        errs.append(np.mean(fabrica().fit(Xa[i], ya[i]).predict(Xa[fuera]) != ya[fuera]))
    oob = np.mean(errs)
    e632 = 0.368 * err_tr + 0.632 * oob                                                  # promedio ponderado: el OOB es pesimista, el de entrenamiento optimista
    p1, q1 = ya.mean(), pred.mean()
    gamma = p1 * (1 - q1) + (1 - p1) * q1                                                # tasa de error "sin informacion" (predicciones independientes de y)
    oob_ = min(oob, gamma)
    R = (oob_ - err_tr) / (gamma - err_tr) if (oob_ > err_tr and gamma > err_tr) else 0.0     # tasa relativa de sobreajuste
    w = 0.632 / (1 - 0.368 * R)
    return err_tr, oob, e632, (1 - w) * err_tr + w * oob_

# (b) Poblacion conocida (error verdadero medible); n = 60 y 25 conjuntos; B = 30 remuestras
Xp, yp = make_classification(n_samples=12000, n_features=10, n_informative=4, n_redundant=0, n_clusters_per_class=2, class_sep=1.0, flip_y=0.1, random_state=SEED)
X_ev, y_ev, X_fu, y_fu = Xp[:4000], yp[:4000], Xp[4000:], yp[4000:]
rng = np.random.default_rng(SEED); filas = []
t0 = time.perf_counter()
for nombre, fab in [("1-NN (memoriza: err. entrenamiento = 0)", lambda: KNeighborsClassifier(1)), ("LDA (poco flexible)", lambda: LinearDiscriminantAnalysis())]:
    out = []
    for _ in range(25):
        i = rng.choice(len(y_fu), 60, replace=False); Xa, ya = X_fu[i], y_fu[i]
        et, oob, e632, e632p = estimadores_bootstrap(fab, Xa, ya, 30, rng)
        cv = np.mean([np.mean(fab().fit(Xa[a], ya[a]).predict(Xa[b]) != ya[b]) for a, b in KFold(10, shuffle=True, random_state=int(rng.integers(10**6))).split(Xa)])
        out.append((et, cv, oob, e632, e632p, np.mean(fab().fit(Xa, ya).predict(X_ev) != y_ev)))
    out = np.array(out)
    filas.append(dict(zip(["modelo", "entrenamiento", "10-fold CV", "OOB (bootstrap)", ".632", ".632+", "error verdadero"], [nombre, *out.mean(axis=0)])))
tab = pd.DataFrame(filas).set_index("modelo")
display(tab.round(3))
display((tab.drop(columns="error verdadero").sub(tab["error verdadero"], axis=0)).round(3).rename_axis("sesgo = estimación − error verdadero"))
print(f"(tiempo = {time.perf_counter() - t0:.1f} s; los números son MEDIAS sobre 25 conjuntos de n = 60)")
fila = tab.iloc[0]
assert fila[".632"] < fila["error verdadero"] - 0.05                                         # .632 es MUY optimista con un modelo que memoriza
assert abs(fila[".632+"] - fila["error verdadero"]) < abs(fila[".632"] - fila["error verdadero"])    # .632+ lo corrige
print("Lectura: el error de entrenamiento engaña (sesgo muy negativo); el OOB es algo pesimista (cada modelo se entrena con ≈ 63 % de los datos distintos); el .632 lo compensa mezclando ambos,")
print("pero con un modelo que MEMORIZA (1-NN, error de entrenamiento 0) queda muy por debajo del error verdadero: el .632+ corrige gran parte de ese sesgo con la 'tasa de sobreajuste' relativa a un error sin información (γ).")
print("Nota honesta: en esta simulación el 10-fold CV está igual de cerca (o más) de la verdad; el .632+ es una alternativa para muestras muy pequeñas, no un reemplazo del k-fold repetido.")
````

````python
# ============================================================
# 5.2 Monte Carlo CV (ShuffleSplit): desde cero, cobertura de ejemplos y comparacion con k-fold
# ============================================================
def shuffle_split_desde_cero(n, n_splits, test_size, seed):
    rng = np.random.RandomState(seed); n_te = int(np.ceil(test_size * n)); n_tr = n - n_te
    for _ in range(n_splits):
        perm = rng.permutation(n)
        yield perm[n_te: n_te + n_tr], perm[:n_te]                                   # (entrenamiento, prueba); mismas reglas que scikit-learn

for n_, ns_, ts_ in [(100, 7, 0.2), (103, 5, 0.25)]:
    mios = list(shuffle_split_desde_cero(n_, ns_, ts_, SEED)); sk = list(ShuffleSplit(ns_, test_size=ts_, random_state=SEED).split(np.zeros((n_, 1))))
    assert all(np.array_equal(a1, a2) and np.array_equal(b1, b2) for (a1, b1), (a2, b2) in zip(mios, sk))
print("ShuffleSplit desde cero == sklearn.ShuffleSplit (mismas particiones exactas).")

# ¿Cuantas veces entra cada ejemplo en una prueba? k-fold: EXACTAMENTE r veces. Monte Carlo: aleatorio (algunos ejemplos nunca se evaluan).
n_, r_ = len(y), 4
veces_kf = np.zeros(n_, int); veces_mc = np.zeros(n_, int)
for _, b in repeated_kfold_desde_cero(n_, 5, r_, SEED): veces_kf[b] += 1
for _, b in shuffle_split_desde_cero(n_, 5 * r_, 0.2, SEED): veces_mc[b] += 1
assert (veces_kf == r_).all() and veces_mc.sum() == 5 * r_ * int(np.ceil(0.2 * n_))             # cada evaluacion Monte Carlo prueba ceil(0.2 n) ejemplos
print(f"Con {5 * r_} evaluaciones: k-fold repetido prueba cada ejemplo exactamente {r_} veces; Monte Carlo: min = {veces_mc.min()}, máx = {veces_mc.max()}, ejemplos que NUNCA se prueban = {(veces_mc == 0).sum()}.")

sd_kf = [np.std(est[r]) for r in repeticiones]; sd_mc = [np.std(est_mc[r]) for r in repeticiones]
display(pd.DataFrame({"5-fold repetido": sd_kf, "Monte Carlo (test 20 %)": sd_mc}, index=pd.Index([f"{5 * r} evaluaciones" for r in repeticiones], name="presupuesto")).round(4))
fig, axes = plt.subplots(1, 2, figsize=(14, 4.4))
axes[0].hist(veces_mc, bins=np.arange(-0.5, veces_mc.max() + 1.5), color="#93c5fd", edgecolor="white", label="Monte Carlo (ShuffleSplit)")
axes[0].axvline(r_, color="#dc2626", lw=2, label=f"k-fold repetido: todos exactamente {r_}")
axes[0].set_xlabel("Veces que un ejemplo cae en una partición de prueba"); axes[0].set_ylabel("Número de ejemplos"); axes[0].legend(fontsize=8)
axes[0].set_title("Cobertura de la evaluación (20 evaluaciones)", fontweight="bold")
axes[1].plot([5 * r for r in repeticiones], sd_kf, "o-", color="#2563eb", lw=2, label="5-fold repetido")
axes[1].plot([5 * r for r in repeticiones], sd_mc, "s--", color="#f59e0b", lw=2, label="Monte Carlo (20 % de prueba)")
axes[1].set_xlabel("Número de evaluaciones (ajustes)"); axes[1].set_ylabel("Desv. estándar de la estimación entre semillas"); axes[1].legend(fontsize=8)
axes[1].set_title("Ruido de la partición según el presupuesto", fontweight="bold")
plt.tight_layout(); plt.show()
print("Monte Carlo CV desacopla el número de particiones del tamaño de la prueba (útil con muchos datos o pruebas grandes), pero no garantiza cobertura uniforme y, a igual presupuesto, el k-fold repetido")
print("tiende a tener menos ruido de partición. Ambos estiman el error del modelo con ≈ 80 % de los datos; ninguno elimina la incertidumbre por el tamaño de la muestra.")
````

## 6. La Regla de una Desviación Estándar 📏

Cuando varios hiperparámetros dan puntajes **estadísticamente indistinguibles**, ¿cuál elegir? La **regla de una desviación estándar** (*one-standard-error rule*; Hastie *et al.*, Cap. 7 *(citado de memoria, verificar)*) dice: **elige el modelo más simple cuyo puntaje esté a menos de un error estándar del mejor.** Con $k$ pliegues,

$$\text{EE}=\frac{s}{\sqrt k},\qquad \text{umbral}=\bar m_{\max}-\text{EE}_{\max},$$

donde $s$ es la desviación estándar de los puntajes por pliegue. (En el [módulo 03](../03%20-%20Metricas%20y%20Ajuste%20de%20Hiperparametros/02_Validacion_Cruzada_y_GridSearch.ipynb) se usó $s$ como aproximación; aquí usamos el error estándar de la **media**, $s/\sqrt k$, que es el que propone la regla.) Se necesita un **orden de complejidad**: aquí, un $C$ más pequeño significa un modelo más regularizado. Beneficio: **más estabilidad, interpretabilidad y menor sobreajuste a la selección**, a un costo de rendimiento que es indistinguible del ruido.

````python
# ============================================================
# 6. Regla de una desviacion estandar (1-SE): el modelo mas simple dentro del error estandar del mejor
# ============================================================
# Hastie et al. (Cap. 7, citado de memoria): elegir el modelo MAS SIMPLE cuyo puntaje CV este a menos de 1 error estandar del mejor.
X_dev, X_ho, y_dev, y_ho = train_test_split(X, y, test_size=0.3, stratify=y, random_state=SEED)
Cs = np.logspace(-4, 2, 13)                                                          # C pequeno = mas regularizacion = modelo mas simple
cv10 = StratifiedKFold(10, shuffle=True, random_state=SEED)
pliegues_acc = np.array([cross_val_score(make_pipeline(StandardScaler(), LogisticRegression(C=C, max_iter=2000)), X_dev, y_dev, cv=cv10) for C in Cs])
media, ee = pliegues_acc.mean(axis=1), pliegues_acc.std(axis=1, ddof=1) / np.sqrt(pliegues_acc.shape[1])      # error estandar de la MEDIA = s / sqrt(k)
i_mejor = int(np.argmax(media))
umbral = media[i_mejor] - ee[i_mejor]
i_1se = int(np.where(media >= umbral)[0].min())                                      # la C mas pequena (mas simple) que supera el umbral
assert Cs[i_1se] <= Cs[i_mejor] and media[i_1se] >= umbral

def acc_holdout(C):
    return make_pipeline(StandardScaler(), LogisticRegression(C=C, max_iter=2000)).fit(X_dev, y_dev).score(X_ho, y_ho)
display(pd.DataFrame({"C": [Cs[i_mejor], Cs[i_1se]], "CV (media)": [media[i_mejor], media[i_1se]], "error estándar": [ee[i_mejor], ee[i_1se]],
                      f"accuracy en prueba (n = {len(y_ho)})": [acc_holdout(Cs[i_mejor]), acc_holdout(Cs[i_1se])]},
                     index=["mejor puntaje CV", "regla 1-SE (más simple)"]).round(4))

fig, ax = plt.subplots(figsize=(10, 4.6))
ax.errorbar(Cs, media, yerr=ee, fmt="o-", color="#2563eb", capsize=4, label="Accuracy CV ± 1 error estándar")
ax.axhline(umbral, color="#f59e0b", ls="--", label="Mejor − 1 error estándar")
ax.scatter([Cs[i_mejor]], [media[i_mejor]], s=140, color="#dc2626", zorder=5, label=f"Mejor: C = {Cs[i_mejor]:.3g}")
ax.scatter([Cs[i_1se]], [media[i_1se]], s=140, marker="*", color="#16a34a", zorder=5, label=f"Regla 1-SE: C = {Cs[i_1se]:.3g}")
ax.set_xscale("log"); ax.set_xlabel("C de la regresión logística (escala log; menor = más regularizada)"); ax.set_ylabel("Accuracy de validación cruzada (10 pliegues)")
ax.set_title("Regla de una desviación estándar", fontweight="bold"); ax.legend(fontsize=8, loc="lower right")
plt.tight_layout(); plt.show()
err_mejor, err_1se = int(round((1 - acc_holdout(Cs[i_mejor])) * len(y_ho))), int(round((1 - acc_holdout(Cs[i_1se])) * len(y_ho)))
print(f"{int((media >= umbral).sum())} de {len(Cs)} valores de C son indistinguibles del mejor dentro de 1 error estándar: la regla elige el más simple (C = {Cs[i_1se]:.3g}).")
print(f"En la prueba, el modelo 1-SE cometió {err_1se} errores frente a {err_mejor} del 'mejor': con tan pocos errores la diferencia es ruido (no hay almuerzo gratis: la regla compra simplicidad y estabilidad, y a veces cuesta un poco de rendimiento).")
print("Matiz: el error estándar s/√k trata los pliegues como independientes (no lo son, sección 3 del cuaderno 00); es un criterio heurístico de parsimonia, no una prueba de hipótesis.")
````

## 7. Validación Cruzada con Datos Desbalanceados ⚖️

Con una clase rara surgen tres problemas específicos:

1. **Pliegues sin positivos o con muy pocos.** Con 60 positivos en 2 000 filas y 5 pliegues, un `KFold` barajado puede dejar un pliegue con 5 y otro con 20 positivos; las métricas por pliegue (recall, F1, AP) se vuelven muy ruidosas. **`StratifiedKFold`** reparte los positivos casi igual.
2. **Métrica.** El *accuracy* engaña (la línea base ya es 97 %); usa **PR-AUC/AP**, $F_\beta$ o costo (módulo 03). Con pocos positivos por pliegue, además de **promediar** la métrica de cada pliegue conviene calcularla sobre las **predicciones agrupadas** (*pooled*) de todos los pliegues.
3. **Fuga por remuestreo.** El **sobremuestreo** (duplicar o sintetizar positivos) **debe hacerse dentro de cada pliegue de entrenamiento**. Si se hace antes de partir, **copias exactas** de un mismo positivo caen en entrenamiento y validación y el modelo «reconoce» lo que ya vio: es el mismo fenómeno de la fuga por grupos de la sección 2, con las copias como «mismo paciente».

Lo medimos: pliegues de prueba en 300 particiones, dispersión de la estimación de AP y la fuga por sobremuestreo con un bosque (que memoriza).

````python
# ============================================================
# 7. CV con datos desbalanceados: estratificar, y NUNCA remuestrear antes de dividir
# ============================================================
X2, y2 = make_classification(n_samples=2000, n_features=10, n_informative=3, n_redundant=0, weights=[0.97, 0.03], class_sep=0.8, flip_y=0.01, random_state=0)
print(f"n = {len(y2)}, positivos = {y2.sum()} ({y2.mean():.1%}). Un 5-fold sin estratificar puede dar pliegues con muy pocos positivos.")

# (a) Positivos por pliegue de prueba en 300 particiones aleatorias
pos_kf, pos_sk = [], []
for s in range(300):
    pos_kf += [y2[b].sum() for _, b in KFold(5, shuffle=True, random_state=s).split(X2)]
    pos_sk += [y2[b].sum() for _, b in StratifiedKFold(5, shuffle=True, random_state=s).split(X2, y2)]
print(f"Positivos por pliegue (ideal = {y2.sum() / 5:.1f}):  KFold barajado: mín = {min(pos_kf)}, máx = {max(pos_kf)}  |  Estratificado: mín = {min(pos_sk)}, máx = {max(pos_sk)}")
assert max(pos_sk) - min(pos_sk) <= 1 and max(pos_kf) - min(pos_kf) > 4

# (b) Dispersion de la estimacion (PR-AUC) entre 20 particiones, y "media de pliegues" vs "predicciones agrupadas"
lr = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
ap_kf = [cross_val_score(lr, X2, y2, cv=KFold(5, shuffle=True, random_state=s), scoring="average_precision").mean() for s in range(20)]
ap_sk = [cross_val_score(lr, X2, y2, cv=StratifiedKFold(5, shuffle=True, random_state=s), scoring="average_precision").mean() for s in range(20)]
pooled = np.empty(len(y2)); por_pliegue = []
for a, b in StratifiedKFold(5, shuffle=True, random_state=0).split(X2, y2):
    pooled[b] = clone(lr).fit(X2[a], y2[a]).predict_proba(X2[b])[:, 1]; por_pliegue.append(average_precision_score(y2[b], pooled[b]))
print(f"PR-AUC (AP) por CV en 20 particiones:  KFold: {np.mean(ap_kf):.3f} ± {np.std(ap_kf):.3f}  |  Estratificado: {np.mean(ap_sk):.3f} ± {np.std(ap_sk):.3f}")
print("(con la media de AP de todo el CV la dispersión entre particiones es parecida; la ventaja de estratificar se ve sobre todo en el reparto de positivos por pliegue y en las métricas POR PLIEGUE.)")
print(f"Con la misma partición: media de las AP por pliegue = {np.mean(por_pliegue):.3f} (pliegues: {np.round(por_pliegue, 2)}) | AP de las predicciones agrupadas = {average_precision_score(y2, pooled):.3f}")

# (c) FUGA por remuestreo antes de dividir: sobremuestrear los positivos (duplicarlos) ANTES de la CV
def sobremuestrear(Xa, ya, rng):
    pos, neg = np.where(ya == 1)[0], np.where(ya == 0)[0]
    idx = np.r_[neg, rng.choice(pos, len(neg), replace=True)]                       # duplica positivos hasta igualar a los negativos
    return Xa[idx], ya[idx]

rf = lambda: RandomForestClassifier(n_estimators=30, random_state=SEED, n_jobs=1)
t0 = time.perf_counter(); mal, bien = [], []
for s in range(3):
    rng = np.random.default_rng(s)
    Xo, yo = sobremuestrear(X2, y2, rng)                                              # MAL: duplicados ANTES de partir
    mal.append(np.mean([average_precision_score(yo[b], rf().fit(Xo[a], yo[a]).predict_proba(Xo[b])[:, 1])
                        for a, b in StratifiedKFold(5, shuffle=True, random_state=s).split(Xo, yo)]))
    f = []                                                                            # BIEN: se parte primero y se sobremuestrea SOLO el entrenamiento
    for a, b in StratifiedKFold(5, shuffle=True, random_state=s).split(X2, y2):
        Xa, ya = sobremuestrear(X2[a], y2[a], rng)
        f.append(average_precision_score(y2[b], rf().fit(Xa, ya).predict_proba(X2[b])[:, 1]))
    bien.append(np.mean(f))
display(pd.DataFrame({"AP medio (3 semillas)": [np.mean(mal), np.mean(bien)], "línea base (prevalencia)": [y2.mean()] * 2},
                     index=["sobremuestreo ANTES de la CV (fuga)", "sobremuestreo DENTRO de cada pliegue"]).round(3))
assert np.mean(mal) > 0.95 and np.mean(bien) < 0.6
fig, ax = plt.subplots(figsize=(8, 4.2))
ax.bar(["Antes de la CV\n(fuga)", "Dentro de cada pliegue"], [np.mean(mal), np.mean(bien)], color=["#dc2626", "#16a34a"])
ax.axhline(y2.mean(), color="black", ls="--", label=f"Línea base (prevalencia = {y2.mean():.3f})"); ax.set_ylabel("PR-AUC (AP) por CV"); ax.legend(fontsize=8)
ax.set_title("Duplicar positivos antes de dividir 'memoriza' la validación", fontweight="bold")
plt.tight_layout(); plt.show()
print(f"({time.perf_counter() - t0:.1f} s) Al duplicar antes de partir, copias exactas de los positivos caen en entrenamiento Y validación: el bosque las reconoce y la AP sube a ≈ {np.mean(mal):.2f} (¡falso!).")
print("Regla: estratifica; usa métricas apropiadas (PR-AUC, F-beta); y todo remuestreo (sobremuestreo, SMOTE, submuestreo) va DENTRO del pliegue de entrenamiento (con imbalanced-learn: Pipeline de imblearn).")
````

```mermaid
    A["¿Cómo se generaron los datos?"] --> B{"¿Hay orden temporal que<br/>el modelo usará hacia adelante?"}
    B -- "Sí" --> TS["Walk-forward: TimeSeriesSplit<br/>(+ gap / purga y embargo si hay horizonte)"]
    B -- "No" --> C{"¿Varias filas comparten origen<br/>(paciente, cliente, sensor)?"}
    C -- "Sí" --> GR["GroupKFold / StratifiedGroupKFold"]
    C -- "No" --> D{"¿Clases raras<br/>o muy desbalanceadas?"}
    D -- "Sí" --> ST["StratifiedKFold repetido +<br/>remuestreo DENTRO del pliegue"]
    D -- "No" --> KF["KFold (k = 5 o 10), repetido si n es pequeño"]
    TS --> E{"¿Eliges hiperparámetros<br/>y quieres reportar rendimiento?"}
    GR --> E
    ST --> E
    KF --> E
    E -- "Sí" --> N["VALIDACIÓN ANIDADA<br/>(el mismo esquema en ambos bucles)<br/>o prueba final sellada"]
    E -- "No" --> F["CV simple + intervalo de confianza"]
```

**Reglas transversales:** (1) el esquema de partición **debe imitar el uso real** del modelo (nuevos pacientes, futuro, nuevos clientes); (2) **todo lo que aprende de los datos** va dentro del `Pipeline` (módulo 03); (3) **elegir y evaluar con los mismos pliegues** es optimista: anida o separa una prueba; (4) si dos opciones son **indistinguibles**, usa la **regla de una desviación estándar** y prefiere la más simple.

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| ***k*-fold repetido** | La estimación por CV **cambia con la partición**; repetir reduce ese ruido (≈ $1/\sqrt r$) pero **no** la incertidumbre por tener solo $n$ datos. `RepeatedKFold` desde cero coincide con `scikit-learn` |
| **Grupos** | Si varias filas comparten origen, partir por **grupo** (`GroupKFold`). Ignorarlo infló el *accuracy* de ≈ 58 % (real) a ≈ 88 % en la demostración: el modelo reconoce al paciente |
| **Series de tiempo** | *Walk-forward* (`TimeSeriesSplit`, con `gap` y ventana deslizante, verificado): **nunca entrenar con el futuro**. Barajar mide interpolación, no pronóstico; purga y embargo si la etiqueta mira hacia adelante |
| **CV anidada** | Bucle externo estima, bucle interno elige hiperparámetros. Desde cero == `cross_val_score(GridSearchCV(...))`. Estima el **procedimiento**, no produce un modelo |
| **Sesgo optimista** | Reportar el **mejor puntaje de CV** sobreestimó el rendimiento real en ≈ 7 puntos en la simulación (48 combinaciones, $n=60$); la **anidada** quedó a ≈ 2 puntos, en sentido contrario (ligeramente pesimista) |
| **.632 / .632+** | El OOB es pesimista, el de entrenamiento optimista; el .632 queda muy por debajo del error verdadero con modelos que memorizan (1-NN) y el **.632+** lo corrige con la tasa de sobreajuste; el *k*-fold repetido sigue siendo la opción simple |
| **Monte Carlo CV** | `ShuffleSplit`: desacopla el número de particiones del tamaño de prueba, pero **sin cobertura uniforme** |
| **Regla 1-SE** | Entre opciones indistinguibles (dentro de $s/\sqrt k$ del mejor), elige la **más simple**: compra estabilidad y parsimonia, sin garantía de ganar rendimiento |
| **Desbalanceo** | Estratificar; métricas PR-AUC/$F_\beta$; **remuestrear dentro del pliegue**: el sobremuestreo antes de dividir llevó la AP de ≈ 0.3 (real) a ≈ 1.0 (falso) |

> 🎓 **Siguiente paso:** con una estimación honesta del error ya podemos **buscar hiperparámetros** de forma eficiente: [Random Search y GridSearch](02_Random_Search_y_GridSearch.ipynb) y [Optimización bayesiana](03_Optimizacion_Bayesiana.ipynb).
