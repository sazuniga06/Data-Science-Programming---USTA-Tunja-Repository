## 1. De la Regresión Lineal a la Sigmoide 🔄

### 1.1 El problema de usar regresión lineal para clasificar

Si $y \in \{0, 1\}$ y ajustamos $\hat{y} = \mathbf{w}^\top\mathbf{x} + b$ por mínimos cuadrados, el modelo puede producir valores **menores que 0 o mayores que 1**, que no son probabilidades; además, la pérdida cuadrática es inadecuada para una variable de Bernoulli y los puntos extremos arrastran la recta. Queremos un modelo que siga siendo **lineal en los parámetros** pero que devuelva siempre un número en $(0, 1)$.

### 1.2 El modelo logístico

Se modela la probabilidad de la clase positiva pasando la combinación lineal $z = \mathbf{w}^\top\mathbf{x} + b$ por la **función sigmoide** (o logística):

$$P(y = 1 \mid \mathbf{x}) = \sigma(z) = \frac{1}{1 + e^{-z}}, \qquad z = \mathbf{w}^\top\mathbf{x} + b.$$

Propiedades útiles: $\sigma(z) \in (0,1)$; $\sigma(0) = 0.5$; $\sigma(-z) = 1 - \sigma(z)$; y su derivada es

$$\sigma'(z) = \sigma(z)\big(1 - \sigma(z)\big) \;\le\; \tfrac{1}{4}.$$

### 1.3 Odds, log-odds y odds ratio

Sea $p = P(y=1 \mid \mathbf{x})$. Los **odds** (momios) son $\dfrac{p}{1-p}$: cuántas veces es más probable el éxito que el fracaso. Despejando $z$ de la sigmoide:

$$\ln\!\left(\frac{p}{1-p}\right) = \operatorname{logit}(p) = \mathbf{w}^\top\mathbf{x} + b.$$

**El logit es la inversa de la sigmoide**, y el modelo afirma que los **log-odds son lineales** en las variables. Consecuencia clave para la interpretación: al aumentar $x_j$ en **una unidad** (con las demás fijas), los log-odds cambian en $w_j$ y los **odds se multiplican por** $e^{w_j}$ — el **odds ratio** (OR):

$$\text{OR}_j = e^{w_j}.$$

| Valor de $w_j$ | $\text{OR}_j = e^{w_j}$ | Efecto de aumentar $x_j$ |
|---|---|---|
| $w_j > 0$ | $> 1$ | Aumenta los odds (y la probabilidad) de $y = 1$ |
| $w_j = 0$ | $= 1$ | Sin efecto |
| $w_j < 0$ | $< 1$ | Disminuye los odds de $y = 1$ |

Ojo: el OR es **constante** (multiplica los odds siempre igual), pero el efecto sobre la **probabilidad** no lo es: depende de dónde se parta, porque la sigmoide es no lineal.

### 1.4 Regla de decisión

Se clasifica como positivo si $P(y=1 \mid \mathbf{x}) \ge \tau$ (umbral, por defecto $\tau = 0.5$). Como $\sigma$ es monótona creciente, esto equivale a $\mathbf{w}^\top\mathbf{x} + b \ge \operatorname{logit}(\tau)$; con $\tau = 0.5$, $\operatorname{logit}(\tau) = 0$ y la **frontera de decisión** es el hiperplano $\mathbf{w}^\top\mathbf{x} + b = 0$: el clasificador es **lineal**.

## 2. Estimación: Máxima Verosimilitud y Entropía Cruzada 🧮

### 2.1 Verosimilitud

Con $y_i \in \{0, 1\}$ y $p_i = \sigma(\mathbf{w}^\top\mathbf{x}_i + b)$, cada observación sigue una distribución de **Bernoulli**: $P(y_i \mid \mathbf{x}_i) = p_i^{\,y_i}(1 - p_i)^{1 - y_i}$. Con observaciones independientes, la **verosimilitud** y la **log-verosimilitud** son

$$\mathcal{L}(\mathbf{w}, b) = \prod_{i=1}^{n} p_i^{\,y_i}(1-p_i)^{1-y_i}, \qquad \ell(\mathbf{w}, b) = \sum_{i=1}^{n}\Big[y_i \ln p_i + (1 - y_i)\ln(1 - p_i)\Big].$$

Maximizar $\ell$ equivale a minimizar su negativo promediado, la **entropía cruzada binaria** (*binary cross-entropy*, *log-loss*):

$$J(\mathbf{w}, b) = -\frac{1}{n}\sum_{i=1}^{n}\Big[y_i \ln p_i + (1 - y_i)\ln(1 - p_i)\Big].$$

(Es la pérdida que se vio de forma gráfica en el [Cuaderno 00](00_Fundamentos_del_Aprendizaje_Supervisado.ipynb) como pérdida sustituta de la 0-1; y es la divergencia de Kullback–Leibler entre la distribución empírica de las etiquetas y la predicha, salvo una constante.)

### 2.2 Derivación del gradiente

Sea $z_i = \mathbf{w}^\top\mathbf{x}_i + b$ y $p_i = \sigma(z_i)$. Para un solo ejemplo, $J_i = -y\ln\sigma(z) - (1-y)\ln(1 - \sigma(z))$. Usando $\sigma'(z) = \sigma(1 - \sigma)$:

$$\frac{\partial J_i}{\partial z} = -y\,\frac{\sigma(1-\sigma)}{\sigma} + (1-y)\,\frac{\sigma(1-\sigma)}{1 - \sigma} = -y(1 - \sigma) + (1 - y)\sigma = \sigma - y = p - y.$$

Con la regla de la cadena ($\partial z_i/\partial \mathbf{w} = \mathbf{x}_i$, $\partial z_i/\partial b = 1$):

$$\boxed{\;\nabla_{\mathbf{w}} J = \frac{1}{n}X^\top(\mathbf{p} - \mathbf{y}), \qquad \frac{\partial J}{\partial b} = \frac{1}{n}\sum_i (p_i - y_i)\;}$$

Es **idéntico en forma** al gradiente de la regresión lineal, $\frac{1}{n}X^\top(\hat{\mathbf{y}} - \mathbf{y})$: solo cambia la **predicción** ($\hat{y}$ lineal vs. $p$ sigmoide). No es casualidad: ambos son *modelos lineales generalizados* con función de enlace canónica.

### 2.3 Convexidad y optimización

El Hessiano es $\nabla^2 J = \frac{1}{n}X^\top S X$ con $S = \operatorname{diag}\big(p_i(1 - p_i)\big)$, **semidefinido positivo**: $J$ es **convexa**, sin mínimos locales. Pero, a diferencia de MCO, **no existe solución cerrada**: se usan métodos iterativos (descenso del gradiente, Newton–Raphson / mínimos cuadrados reponderados iterativamente —IRLS—, L-BFGS, que es el optimizador por defecto de `scikit-learn`).

> ⚠️ **Separación perfecta:** si existe un hiperplano que separa las dos clases sin error, $J$ no tiene mínimo: se reduce indefinidamente aumentando $\lVert\mathbf{w}\rVert$ (las probabilidades tienden a 0 y 1) y el estimador de máxima verosimilitud **no existe**. La **regularización** (Sección 4) lo evita.

### 2.4 Con regularización L2

Añadir $\frac{\lambda}{2}\lVert\mathbf{w}\rVert^2$ (sin penalizar $b$) solo agrega $\lambda\mathbf{w}$ al gradiente de $\mathbf{w}$. `scikit-learn` parametriza con $C$ (inverso de la fuerza) y minimiza $\,C\sum_i \text{log-loss}_i + \tfrac12\lVert\mathbf{w}\rVert^2$, lo que equivale a la $J$ anterior con $\lambda = 1/(C\,n)$.

### 2.5 Implementación desde cero

Implementamos descenso del gradiente por lotes con **verificación numérica del gradiente** (comparación con diferencias finitas) y lo comparamos con `LogisticRegression`. Usamos un problema 2D (para poder dibujarlo), con 8 % de etiquetas ruidosas (así las clases **no** son perfectamente separables y el MLE existe), y estandarizamos las variables con la media y la desviación del conjunto de entrenamiento.

````python
# ============================================================
# 2.5 Regresion logistica desde cero (descenso del gradiente) + verificacion numerica del gradiente
# ============================================================
from sklearn.datasets import make_classification

class RegresionLogisticaDesdeCero:
    """Regresion logistica binaria con descenso del gradiente por lotes y penalizacion L2 opcional (intercepto sin penalizar)."""
    def __init__(self, eta=0.5, n_iter=20_000, lam=0.0, tol=1e-12):
        self.eta, self.n_iter, self.lam, self.tol = eta, n_iter, lam, tol

    def _costo(self, Z, y, w, b):
        p = np.clip(expit(Z @ w + b), 1e-15, 1 - 1e-15)               # clip evita log(0)
        return -np.mean(y * np.log(p) + (1 - y) * np.log(1 - p)) + 0.5 * self.lam * w @ w

    def _gradiente(self, Z, y, w, b):
        p = expit(Z @ w + b)
        return Z.T @ (p - y) / len(y) + self.lam * w, np.mean(p - y)   # (1/n) X'(p - y) + lambda w

    def fit(self, Z, y):
        Z, y = np.asarray(Z, dtype=float), np.asarray(y, dtype=float)
        self.w_, self.b_ = np.zeros(Z.shape[1]), 0.0
        self.historial_ = []
        for _ in range(self.n_iter):
            gw, gb = self._gradiente(Z, y, self.w_, self.b_)
            self.historial_.append(self._costo(Z, y, self.w_, self.b_))
            if np.sqrt(gw @ gw + gb ** 2) < self.tol:
                break
            self.w_ -= self.eta * gw
            self.b_ -= self.eta * gb
        return self

    def predict_proba(self, Z):
        return expit(np.asarray(Z, dtype=float) @ self.w_ + self.b_)

    def predict(self, Z, umbral=0.5):
        return (self.predict_proba(Z) >= umbral).astype(int)

# --- Datos 2D con ruido en las etiquetas (8 %): no son perfectamente separables ---
X2, y2 = make_classification(n_samples=300, n_features=2, n_redundant=0, n_informative=2, n_clusters_per_class=1,
                             flip_y=0.08, class_sep=1.0, random_state=SEED)
X2_tr, X2_te, y2_tr, y2_te = train_test_split(X2, y2, test_size=0.3, stratify=y2, random_state=SEED)
esc2 = StandardScaler().fit(X2_tr)                                    # media/desviacion SOLO con entrenamiento
Z_tr, Z_te = esc2.transform(X2_tr), esc2.transform(X2_te)

# --- Verificacion del gradiente analitico con diferencias finitas centradas ---
modelo_gc = RegresionLogisticaDesdeCero(lam=0.1)
w0, b0 = np.array([0.7, -0.4]), 0.2
gw_ana, gb_ana = modelo_gc._gradiente(Z_tr, y2_tr, w0, b0)
h = 1e-6
gw_num = np.array([(modelo_gc._costo(Z_tr, y2_tr, w0 + h * e, b0) - modelo_gc._costo(Z_tr, y2_tr, w0 - h * e, b0)) / (2 * h) for e in np.eye(2)])
gb_num = (modelo_gc._costo(Z_tr, y2_tr, w0, b0 + h) - modelo_gc._costo(Z_tr, y2_tr, w0, b0 - h)) / (2 * h)
print(f"Gradiente analítico  : w = {gw_ana}, b = {gb_ana:.8f}")
print(f"Gradiente numérico   : w = {gw_num}, b = {gb_num:.8f}")
assert np.allclose(gw_ana, gw_num, atol=1e-7) and np.isclose(gb_ana, gb_num, atol=1e-7)
print("Gradient check superado: la derivación (1/n) X'(p - y) es correcta.")
````

````python
# ============================================================
# 2.5 Comparacion con scikit-learn: sin regularizar (C enorme) y con L2 (C = 1)
# ============================================================
n_tr = len(y2_tr)

# (a) Sin regularizacion: el MLE existe porque las clases NO son separables
propio_mle = RegresionLogisticaDesdeCero(eta=0.5, n_iter=30_000).fit(Z_tr, y2_tr)
sk_mle = LogisticRegression(C=1e10, tol=1e-12, max_iter=100_000).fit(Z_tr, y2_tr)
assert np.allclose(propio_mle.w_, sk_mle.coef_[0], rtol=1e-4) and np.isclose(propio_mle.b_, sk_mle.intercept_[0], rtol=1e-4, atol=1e-6)
assert np.allclose(propio_mle.predict_proba(Z_te), sk_mle.predict_proba(Z_te)[:, 1], atol=1e-4)

# (b) Con L2: sklearn con C  <=>  lambda = 1 / (C * n)
C = 1.0
propio_l2 = RegresionLogisticaDesdeCero(eta=0.5, n_iter=30_000, lam=1 / (C * n_tr)).fit(Z_tr, y2_tr)
sk_l2 = LogisticRegression(C=C, tol=1e-12, max_iter=100_000).fit(Z_tr, y2_tr)
assert np.allclose(propio_l2.w_, sk_l2.coef_[0], rtol=1e-4) and np.isclose(propio_l2.b_, sk_l2.intercept_[0], rtol=1e-4, atol=1e-6)

tabla = pd.DataFrame({"desde_cero": [*propio_mle.w_, propio_mle.b_, *propio_l2.w_, propio_l2.b_],
                      "scikit_learn": [*sk_mle.coef_[0], sk_mle.intercept_[0], *sk_l2.coef_[0], sk_l2.intercept_[0]]},
                     index=pd.MultiIndex.from_product([["Sin regularizar", "L2 (C=1)"], ["w1", "w2", "b"]]))
tabla["diferencia_abs"] = (tabla["desde_cero"] - tabla["scikit_learn"]).abs()
display(tabla.round(6))
print("Descenso del gradiente desde cero == LogisticRegression (sin regularizar y con L2): verificado con assert.")

fig, ax = plt.subplots(figsize=(10, 5))
costo_opt = propio_mle.historial_[-1]
ax.plot(np.arange(1, len(propio_mle.historial_) + 1), np.maximum(np.array(propio_mle.historial_) - costo_opt, 1e-16), color="#2563eb", lw=2, label="Sin regularizar")
costo_opt2 = propio_l2.historial_[-1]
ax.plot(np.arange(1, len(propio_l2.historial_) + 1), np.maximum(np.array(propio_l2.historial_) - costo_opt2, 1e-16), color="#16a34a", lw=2, label="Con L2 (C=1)")
ax.set_xscale("log"); ax.set_yscale("log")
ax.set_xlabel("Iteración (escala log)")
ax.set_ylabel("Exceso de costo sobre el óptimo (escala log)")
ax.set_title("Convergencia del descenso del gradiente en regresión logística", fontweight="bold")
ax.legend()
plt.tight_layout()
plt.show()
````

## 3. Frontera de Decisión y Umbral 🎯

Con $\tau = 0.5$ la frontera es la recta $w_1 z_1 + w_2 z_2 + b = 0$ (en este caso 2D). Con otro umbral $\tau$, la frontera sigue siendo una **recta paralela**, desplazada hasta $\mathbf{w}^\top\mathbf{x} + b = \operatorname{logit}(\tau) = \ln\frac{\tau}{1-\tau}$:

| Umbral $\tau$ | $\operatorname{logit}(\tau)$ | Efecto |
|---|---|---|
| $0.3$ | $-0.85$ | Se predice la clase 1 con **más facilidad** (más positivos, menos falsos negativos, más falsos positivos) |
| $0.5$ | $0$ | Valor por defecto |
| $0.7$ | $+0.85$ | Se predice la clase 1 con **más exigencia** (menos falsos positivos, más falsos negativos) |

El umbral no es un parámetro del modelo sino una **decisión de negocio**: depende del costo relativo de cada tipo de error. (Cómo elegirlo con rigor —curvas ROC, precisión–exhaustividad, costos— es materia del Módulo 03.)

## 4. Regularización L1 y L2 y el Parámetro $C$ ⚖️

La regularización añade a la entropía cruzada una penalización sobre los pesos (nunca sobre el intercepto):

| Penalización | Término añadido | Efecto | Interpretación bayesiana (MAP) |
|---|---|---|---|
| **L2** (*Ridge*) | $\tfrac{1}{2C}\lVert\mathbf{w}\rVert_2^2$ | Encoge todos los pesos hacia cero, sin anularlos | *A priori* gaussiano sobre $\mathbf{w}$ |
| **L1** (*Lasso*) | $\tfrac{1}{C}\lVert\mathbf{w}\rVert_1$ | Anula pesos: **selección de variables** (modelo disperso) | *A priori* de Laplace sobre $\mathbf{w}$ |

En `scikit-learn`, $C$ es el **inverso** de la fuerza de regularización: **$C$ pequeño = regularización fuerte** (modelo simple, más sesgo), **$C$ grande = regularización débil** (modelo flexible, más varianza). Es la perilla de complejidad del [Cuaderno 00](00_Fundamentos_del_Aprendizaje_Supervisado.ipynb). Las variables **deben estar estandarizadas**; de lo contrario la penalización castiga de forma desigual a variables con distintas unidades.

Usamos el dataset **Breast Cancer Wisconsin** (569 tumores, 30 variables numéricas calculadas de imágenes de aspirados con aguja fina). Con 30 variables y solo 398 casos de entrenamiento, en esta partición las clases resultan **linealmente separables**: sin regularización el MLE no existe y $\lVert\mathbf{w}\rVert$ diverge. Comprobémoslo.

````python
# ============================================================
# 4. Caminos de regularizacion L1 / L2 en Breast Cancer (variables estandarizadas)
# (L1 con solver 'liblinear': rapido; nota: liblinear penaliza tambien el intercepto, una diferencia menor)
# ============================================================
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import cross_val_score, StratifiedKFold

bc = load_breast_cancer()
Xbc, ybc = bc.data, bc.target                                         # OJO: en sklearn 0 = maligno, 1 = benigno
Xbc_tr, Xbc_te, ybc_tr, ybc_te = train_test_split(Xbc, ybc, test_size=0.3, stratify=ybc, random_state=SEED)
esc_bc = StandardScaler().fit(Xbc_tr)
Zbc_tr, Zbc_te = esc_bc.transform(Xbc_tr), esc_bc.transform(Xbc_te)
print(f"Entrenamiento: {Zbc_tr.shape} | Prueba: {Zbc_te.shape} | 1 = benigno ({ybc_tr.mean():.1%} del entrenamiento)")

Cs = np.logspace(-1.5, 2, 22)
rutas = {"l2": [], "l1": []}
n_no_nulos, norma_l2 = [], []
for C_ in Cs:
    m2 = LogisticRegression(penalty="l2", C=C_, max_iter=10_000).fit(Zbc_tr, ybc_tr)
    m1 = LogisticRegression(penalty="l1", C=C_, solver="liblinear", max_iter=10_000).fit(Zbc_tr, ybc_tr)
    rutas["l2"].append(m2.coef_[0]); rutas["l1"].append(m1.coef_[0])
    n_no_nulos.append(int(np.sum(np.abs(m1.coef_[0]) > 1e-8))); norma_l2.append(np.linalg.norm(m2.coef_[0]))
rutas = {k: np.array(v) for k, v in rutas.items()}

# Exactitud por validacion cruzada (solo sobre entrenamiento) segun C
cv = StratifiedKFold(5, shuffle=True, random_state=SEED)
acc = {pen: [cross_val_score(LogisticRegression(penalty=pen, C=C_, solver="liblinear" if pen == "l1" else "lbfgs", max_iter=10_000),
                             Zbc_tr, ybc_tr, cv=cv).mean() for C_ in Cs] for pen in ["l2", "l1"]}

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, pen, titulo in [(axes[0], "l2", "L2: los pesos crecen suavemente con $C$"), (axes[1], "l1", "L1: los pesos se activan uno a uno")]:
    ax.plot(Cs, rutas[pen], lw=1.4)
    ax.set_xscale("log"); ax.axhline(0, color="black", lw=0.8)
    ax.set_xlabel("$C$ (escala log): ← más regularización | menos →"); ax.set_ylabel("Coeficiente (variables estandarizadas)")
    ax.set_title(titulo, fontweight="bold")
ax_dcha = axes[1].twinx()
ax_dcha.plot(Cs, n_no_nulos, "k--", lw=2, label="n.º de pesos ≠ 0 (eje derecho)")
ax_dcha.set_ylabel("Número de pesos distintos de cero"); ax_dcha.grid(False); ax_dcha.legend(loc="upper left", fontsize=8)
axes[2].plot(Cs, acc["l2"], "o-", color="#2563eb", label="L2")
axes[2].plot(Cs, acc["l1"], "s-", color="#dc2626", label="L1")
axes[2].set_xscale("log")
axes[2].set_xlabel("$C$ (escala log)"); axes[2].set_ylabel("Accuracy (validación cruzada, 5 particiones)")
axes[2].set_title("Exactitud según la fuerza de regularización", fontweight="bold"); axes[2].legend()
plt.tight_layout()
plt.show()

print(f"Pesos distintos de cero con L1:  C = {Cs[0]:.3f} -> {n_no_nulos[0]} | C = {Cs[10]:.2f} -> {n_no_nulos[10]} | C = {Cs[-1]:.0f} -> {n_no_nulos[-1]} (de 30)")
sin_reg = LogisticRegression(C=1e8, max_iter=100_000).fit(Zbc_tr, ybc_tr)
print(f"Sin regularizar (C = 1e8): accuracy en entrenamiento = {sin_reg.score(Zbc_tr, ybc_tr):.3f} y ||w|| = {np.linalg.norm(sin_reg.coef_):.0f} "
      f"(con C = 1: ||w|| = {np.linalg.norm(LogisticRegression(C=1).fit(Zbc_tr, ybc_tr).coef_):.1f})")
print("-> clases separables: el MLE no existe y la norma de w solo está acotada por la tolerancia del optimizador.")
mejor_pen = max(acc, key=lambda k: max(acc[k]))
print(f"Mejor accuracy CV: {max(max(v) for v in acc.values()):.3f} con penalización {mejor_pen.upper()}, C = {Cs[int(np.argmax(acc[mejor_pen]))]:.3f}")
````

````python
# ============================================================
# 4.1 L2 desde cero en Breast Cancer (30 variables) vs. scikit-learn, y evaluacion basica en prueba
# ============================================================
n_bc = len(ybc_tr)
propio_bc = RegresionLogisticaDesdeCero(eta=0.5, n_iter=30_000, lam=1 / (1.0 * n_bc)).fit(Zbc_tr, ybc_tr)     # C = 1
sk_bc = LogisticRegression(C=1.0, tol=1e-12, max_iter=100_000).fit(Zbc_tr, ybc_tr)

dif = np.max(np.abs(propio_bc.w_ - sk_bc.coef_[0]))
assert np.allclose(propio_bc.w_, sk_bc.coef_[0], atol=1e-4) and np.isclose(propio_bc.b_, sk_bc.intercept_[0], atol=1e-4)
assert np.array_equal(propio_bc.predict(Zbc_te), sk_bc.predict(Zbc_te))
print(f"30 coeficientes + intercepto: diferencia máxima con scikit-learn = {dif:.2e}. Predicciones idénticas en el conjunto de prueba.")

pred = propio_bc.predict(Zbc_te)
cm = confusion_matrix(ybc_te, pred)
print(f"Accuracy en prueba (modelo desde cero, C = 1): {accuracy_score(ybc_te, pred):.3f}")
print("Matriz de confusión (filas = real, columnas = predicho; 0 = maligno, 1 = benigno):")
print(pd.DataFrame(cm, index=["Real maligno", "Real benigno"], columns=["Pred. maligno", "Pred. benigno"]).to_string())
print("Nota: aquí un falso 'benigno' es mucho más grave que un falso 'maligno'; por eso el accuracy NO basta (ver Módulo 03).")
````

## 5. Más de Dos Clases: *One-vs-Rest* y *Softmax* 🌈

Para $K > 2$ clases hay dos estrategias para extender el modelo:

| Estrategia | Idea | Probabilidades | En `scikit-learn` |
|---|---|---|---|
| **One-vs-Rest** (OvR / OvA) | Entrenar $K$ clasificadores binarios independientes «clase $k$ vs. el resto»; se predice la clase con mayor probabilidad | Se **normalizan** para que sumen 1 (no son un modelo conjunto) | `OneVsRestClassifier(LogisticRegression())` |
| **Softmax** (multinomial) | Un **único modelo conjunto** con $K$ vectores de pesos $\mathbf{w}_k$ | Suman 1 por construcción | `LogisticRegression()` (por defecto con `lbfgs`) |

### Regresión softmax

$$z_k = \mathbf{w}_k^\top\mathbf{x} + b_k, \qquad P(y = k \mid \mathbf{x}) = \frac{e^{z_k}}{\sum_{j=1}^{K} e^{z_j}}, \qquad J = -\frac{1}{n}\sum_{i=1}^{n}\sum_{k=1}^{K}\mathbf{1}[y_i = k]\,\ln p_{ik}.$$

Es la **entropía cruzada categórica**. Repitiendo la derivación de la Sección 2.2 se obtiene el gradiente por clase

$$\frac{\partial J}{\partial \mathbf{w}_k} = \frac{1}{n}\sum_i (p_{ik} - \mathbf{1}[y_i = k])\,\mathbf{x}_i \;\;\Longrightarrow\;\; \nabla_W J = \frac{1}{n}X^\top(P - Y),$$

con $Y$ la matriz *one-hot* de etiquetas. Observa que es otra vez «predicción menos verdad». Dos detalles importantes:

- Con $K = 2$, la softmax se reduce a la sigmoide con $\mathbf{w} = \mathbf{w}_1 - \mathbf{w}_0$: la regresión logística binaria **es** un caso particular.
- **Estabilidad numérica:** se calcula $e^{z_k - \max_j z_j}$ (el resultado no cambia, pero evita desbordes). Además, sumar el mismo vector a todos los $\mathbf{w}_k$ deja las probabilidades intactas (**no identificabilidad**); la regularización L2 resuelve la ambigüedad.

Probamos con **Iris** usando solo largo y ancho del pétalo (para poder dibujar las regiones).

````python
# ============================================================
# 5. Softmax desde cero (L2) vs. LogisticRegression multinomial; y OvR con OneVsRestClassifier
# ============================================================
from sklearn.datasets import load_iris
from sklearn.multiclass import OneVsRestClassifier

def softmax(S):
    S = S - S.max(axis=1, keepdims=True)                              # estabilidad numerica
    E = np.exp(S)
    return E / E.sum(axis=1, keepdims=True)

class SoftmaxDesdeCero:
    """Regresion logistica multinomial con descenso del gradiente por lotes y penalizacion L2 (intercepto sin penalizar)."""
    def __init__(self, eta=0.5, n_iter=30_000, lam=0.0):
        self.eta, self.n_iter, self.lam = eta, n_iter, lam
    def fit(self, Z, y):
        n, p = Z.shape
        K = int(y.max()) + 1
        Y = np.eye(K)[y]                                              # codificacion one-hot
        self.W_, self.b_ = np.zeros((p, K)), np.zeros(K)
        self.historial_ = []
        for _ in range(self.n_iter):
            P = softmax(Z @ self.W_ + self.b_)
            self.historial_.append(-np.mean(np.sum(Y * np.log(np.clip(P, 1e-15, 1)), axis=1)) + 0.5 * self.lam * np.sum(self.W_ ** 2))
            self.W_ -= self.eta * (Z.T @ (P - Y) / n + self.lam * self.W_)      # gradiente (1/n) X'(P - Y) + lambda W
            self.b_ -= self.eta * (P - Y).mean(axis=0)
        return self
    def predict_proba(self, Z):
        return softmax(Z @ self.W_ + self.b_)
    def predict(self, Z):
        return self.predict_proba(Z).argmax(axis=1)

iris = load_iris()
Xi, yi = iris.data[:, 2:4], iris.target                               # largo y ancho del petalo
Xi_tr, Xi_te, yi_tr, yi_te = train_test_split(Xi, yi, test_size=0.3, stratify=yi, random_state=SEED)
esc_i = StandardScaler().fit(Xi_tr)
Zi_tr, Zi_te = esc_i.transform(Xi_tr), esc_i.transform(Xi_te)

soft_propio = SoftmaxDesdeCero(lam=1 / (1.0 * len(yi_tr))).fit(Zi_tr, yi_tr)                  # C = 1
soft_sk = LogisticRegression(C=1.0, tol=1e-12, max_iter=100_000).fit(Zi_tr, yi_tr)             # multinomial por defecto
ovr_sk = OneVsRestClassifier(LogisticRegression(C=1.0, max_iter=10_000)).fit(Zi_tr, yi_tr)

assert np.allclose(soft_propio.W_.T, soft_sk.coef_, atol=1e-4) and np.allclose(soft_propio.b_, soft_sk.intercept_, atol=1e-4)
assert np.allclose(soft_propio.predict_proba(Zi_te), soft_sk.predict_proba(Zi_te), atol=1e-4)
print("Softmax desde cero == LogisticRegression multinomial (pesos, intercepto y probabilidades): verificado.\n")

print(f"Accuracy en prueba -> softmax: {soft_sk.score(Zi_te, yi_te):.3f} | one-vs-rest: {ovr_sk.score(Zi_te, yi_te):.3f}")
punto = esc_i.transform([[4.8, 1.6]])                                  # un caso dificil (versicolor/virginica)
print("\nProbabilidades para un pétalo de 4.8 cm x 1.6 cm (setosa, versicolor, virginica):")
print(f"  Softmax : {np.round(soft_sk.predict_proba(punto)[0], 3)}  (suma = {soft_sk.predict_proba(punto).sum():.3f})")
print(f"  OvR     : {np.round(ovr_sk.predict_proba(punto)[0], 3)}  (OvR normaliza los K modelos binarios independientes)")
````

## 6. Interpretación de Coeficientes como Odds Ratios 🔍

Recordemos: $\text{OR}_j = e^{w_j}$ es el factor por el que se multiplican los **odds** de $y = 1$ al aumentar $x_j$ en una unidad, con las demás variables fijas. Para variables estandarizadas, la «unidad» es **una desviación estándar**, lo que permite comparar la magnitud del efecto entre variables.

Para no confundir al lector, aquí **definimos $y = 1$ como «maligno»** (invirtiendo la codificación de `scikit-learn`, donde 1 = benigno), de modo que $\text{OR} > 1$ signifique *más riesgo de malignidad*. Usamos un subconjunto de variables **poco correlacionadas** entre sí (el radio, el perímetro y el área son casi redundantes y harían inestables las interpretaciones; ver multicolinealidad en el [cuaderno anterior](01_Regresion_Lineal.ipynb)). Los **intervalos de confianza** se estiman por *bootstrap* (remuestreo con reemplazo).

````python
# ============================================================
# 6. Odds ratios con intervalos de confianza bootstrap (Breast Cancer, y = 1 = maligno)
# ============================================================
vars_ = ["mean radius", "mean texture", "mean smoothness", "mean concavity", "mean symmetry", "mean fractal dimension"]
cols = [list(bc.feature_names).index(v) for v in vars_]
X_or = Xbc[:, cols]
y_or = 1 - ybc                                                         # 1 = maligno
X_or_tr, X_or_te, y_or_tr, y_or_te = train_test_split(X_or, y_or, test_size=0.3, stratify=y_or, random_state=SEED)
esc_or = StandardScaler().fit(X_or_tr)
Z_or = esc_or.transform(X_or_tr)

modelo_or = LogisticRegression(C=10.0, max_iter=10_000).fit(Z_or, y_or_tr)    # regularizacion suave
pesos = modelo_or.coef_[0]
OR = np.exp(pesos)

# Verificacion numerica de la definicion: al subir 1 DE una variable, los odds se multiplican exactamente por e^w
z0 = Z_or[0].copy()
z1 = z0.copy(); z1[0] += 1.0
odds = lambda z: (lambda p: p / (1 - p))(modelo_or.predict_proba(z.reshape(1, -1))[0, 1])
assert np.isclose(odds(z1) / odds(z0), OR[0])
print(f"Comprobación: subir 1 DE de '{vars_[0]}' multiplica los odds por {odds(z1) / odds(z0):.3f} = e^w = {OR[0]:.3f} (para ese paciente y para cualquiera).")

# Bootstrap (300 remuestreos) de los odds ratios
rng = np.random.default_rng(SEED)
boot = []
for _ in range(300):
    idx = rng.integers(0, len(y_or_tr), len(y_or_tr))
    boot.append(np.exp(LogisticRegression(C=10.0, max_iter=10_000).fit(Z_or[idx], y_or_tr[idx]).coef_[0]))
boot = np.array(boot)
ic_inf, ic_sup = np.percentile(boot, [2.5, 97.5], axis=0)

tabla = pd.DataFrame({"peso_w": pesos, "OR_por_1DE": OR, "IC95_inf": ic_inf, "IC95_sup": ic_sup,
                      "IC_incluye_1": (ic_inf <= 1) & (ic_sup >= 1)}, index=vars_).sort_values("OR_por_1DE", ascending=False)
display(tabla.round(3))

fig, ax = plt.subplots(figsize=(10, 4.8))
orden = tabla.index[::-1]
for i, v in enumerate(orden):
    fila = tabla.loc[v]
    color = "#94a3b8" if fila["IC_incluye_1"] else ("#dc2626" if fila["OR_por_1DE"] > 1 else "#2563eb")
    ax.plot([fila["IC95_inf"], fila["IC95_sup"]], [i, i], color=color, lw=3)
    ax.plot(fila["OR_por_1DE"], i, "o", color=color, ms=9)
ax.axvline(1, color="black", linestyle="--", label="OR = 1 (sin efecto)")
ax.set_xscale("log")
ax.set_yticks(range(len(orden))); ax.set_yticklabels(orden)
ax.set_xlabel("Odds ratio por +1 desviación estándar (escala log, IC 95 % bootstrap)")
ax.set_title("Odds de malignidad: rojo = aumentan, azul = reducen, gris = no concluyente", fontweight="bold")
ax.legend()
plt.tight_layout()
plt.show()

acc_te = accuracy_score(y_or_te, modelo_or.predict(esc_or.transform(X_or_te)))
print(f"Accuracy en prueba del modelo con estas 6 variables: {acc_te:.3f}")
print("Los OR en gris tienen un IC que incluye 1: con estos datos no se puede afirmar un efecto distinto de cero.")
print("Recuerda: son asociaciones condicionadas al resto de variables del modelo, no relaciones causales.")
````

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Modelo** | $P(y=1\mid\mathbf{x}) = \sigma(\mathbf{w}^\top\mathbf{x} + b)$: un modelo **lineal en los log-odds** que siempre produce probabilidades válidas |
| **Odds y OR** | $\text{odds} = p/(1-p)$; $e^{w_j}$ es el factor multiplicativo de los odds por cada unidad de $x_j$ (el efecto sobre la probabilidad **no** es constante) |
| **Estimación** | Máxima verosimilitud de Bernoulli $\Rightarrow$ **entropía cruzada**; sin solución cerrada, función **convexa** |
| **Gradiente** | $\nabla_{\mathbf{w}}J = \frac{1}{n}X^\top(\mathbf{p}-\mathbf{y})$ (igual de simple que en regresión lineal); verificado con diferencias finitas |
| **Implementación** | El descenso del gradiente desde cero coincidió con `LogisticRegression` (binaria, con L2 y softmax) |
| **Frontera y umbral** | Frontera lineal $\mathbf{w}^\top\mathbf{x}+b=\operatorname{logit}(\tau)$; el umbral es una decisión que intercambia FP por FN |
| **Regularización** | $C$ pequeño = más regularización; L2 encoge, L1 anula; evita el MLE inexistente con clases separables |
| **Multiclase** | OvR (K modelos independientes, probabilidades normalizadas) vs. softmax (modelo conjunto, entropía cruzada categórica) |
| **Evaluación** | Aquí solo *accuracy* y matriz de confusión; las métricas de clasificación se ven a fondo en el **Módulo 03** |
