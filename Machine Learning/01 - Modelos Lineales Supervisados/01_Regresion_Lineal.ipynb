## 1. El Modelo y los Mínimos Cuadrados Ordinarios 📐

### 1.1 Modelo

La **regresión lineal múltiple** supone que la respuesta es una combinación lineal de las $p$ variables más un error:

$$y_i = \beta_0 + \beta_1 x_{i1} + \cdots + \beta_p x_{ip} + \varepsilon_i, \qquad i = 1, \ldots, n.$$

Añadiendo una columna de unos a la matriz de diseño, $X \in \mathbb{R}^{n \times (p+1)}$, el modelo completo se escribe

$$\mathbf{y} = X\boldsymbol{\beta} + \boldsymbol{\varepsilon}, \qquad \hat{\mathbf{y}} = X\boldsymbol{\beta}.$$

Es el caso de ERM (Cuaderno 00) con clase de hipótesis lineal y **pérdida cuadrática**. Los mínimos cuadrados los empleó Legendre (1805) y Gauss (1809) para ajustar órbitas astronómicas.

### 1.2 Derivación de la ecuación normal

Buscamos $\boldsymbol{\beta}$ que minimice la **suma de cuadrados de los residuos**:

$$\text{RSS}(\boldsymbol{\beta}) = \lVert \mathbf{y} - X\boldsymbol{\beta} \rVert_2^2 = (\mathbf{y} - X\boldsymbol{\beta})^\top (\mathbf{y} - X\boldsymbol{\beta}) = \mathbf{y}^\top\mathbf{y} - 2\boldsymbol{\beta}^\top X^\top \mathbf{y} + \boldsymbol{\beta}^\top X^\top X \boldsymbol{\beta}.$$

Derivando respecto de $\boldsymbol{\beta}$ e igualando a cero:

$$\nabla_{\boldsymbol{\beta}}\,\text{RSS} = -2X^\top \mathbf{y} + 2X^\top X \boldsymbol{\beta} = \mathbf{0} \;\;\Longrightarrow\;\; \boxed{\,X^\top X\,\hat{\boldsymbol{\beta}} = X^\top \mathbf{y}\,} \;\;\Longrightarrow\;\; \hat{\boldsymbol{\beta}} = (X^\top X)^{-1} X^\top \mathbf{y}.$$

**¿Es un mínimo?** El Hessiano es $2X^\top X$, y para todo vector $\mathbf{v}$: $\mathbf{v}^\top X^\top X \mathbf{v} = \lVert X\mathbf{v}\rVert^2 \ge 0$, así que RSS es **convexa**. Si las columnas de $X$ son linealmente independientes, $X^\top X$ es definida positiva, invertible, y el mínimo es **único**. Si hay colinealidad *exacta*, $X^\top X$ es singular y existen infinitas soluciones (por eso se prefiere `np.linalg.lstsq` o la pseudoinversa a invertir a ciegas).

**Interpretación geométrica.** $\hat{\mathbf{y}} = X\hat{\boldsymbol{\beta}} = H\mathbf{y}$ con $H = X(X^\top X)^{-1}X^\top$ (la **matriz sombrero**): $\hat{\mathbf{y}}$ es la **proyección ortogonal** de $\mathbf{y}$ sobre el subespacio generado por las columnas de $X$, y el vector de residuos $\mathbf{e} = \mathbf{y} - \hat{\mathbf{y}}$ es ortogonal a ese subespacio: $X^\top \mathbf{e} = \mathbf{0}$. En particular, con intercepto, **los residuos suman cero**.

### 1.3 Propiedades estadísticas

- **Máxima verosimilitud:** si $\varepsilon_i \sim \mathcal{N}(0, \sigma^2)$ independientes, la log-verosimilitud es $-\tfrac{n}{2}\ln(2\pi\sigma^2) - \tfrac{\text{RSS}}{2\sigma^2}$, de modo que **maximizarla equivale a minimizar RSS**: MCO es el estimador de máxima verosimilitud.
- **Teorema de Gauss–Markov:** si $\mathbb{E}[\boldsymbol{\varepsilon}] = \mathbf{0}$ y $\operatorname{Var}(\boldsymbol{\varepsilon}) = \sigma^2 I$ (errores no correlacionados y de varianza constante), $\hat{\boldsymbol{\beta}}$ es el **mejor estimador lineal insesgado** (BLUE). No se necesita normalidad para esto.
- **Varianza del estimador:** $\operatorname{Var}(\hat{\boldsymbol{\beta}}) = \sigma^2 (X^\top X)^{-1}$, con $\hat{\sigma}^2 = \text{RSS}/(n - p - 1)$ como estimador insesgado de $\sigma^2$. Las raíces de la diagonal son los **errores estándar** de los coeficientes.
- **Costo computacional:** formar $X^\top X$ cuesta $O(np^2)$ y resolver el sistema $O(p^3)$: excelente si $p$ es moderado, lento si $p$ es enorme (Géron, Cap. 4). Para $p$ grande o $n$ enorme se usa **descenso del gradiente** (Sección 3).

### 1.4 Caso particular: regresión simple

Con una sola variable ($p = 1$) la ecuación normal tiene solución explícita:

$$\hat{\beta}_1 = \frac{\sum_i (x_i - \bar{x})(y_i - \bar{y})}{\sum_i (x_i - \bar{x})^2} = r_{xy}\,\frac{s_y}{s_x}, \qquad \hat{\beta}_0 = \bar{y} - \hat{\beta}_1 \bar{x},$$

donde $r_{xy}$ es la correlación de Pearson y $s_x, s_y$ las desviaciones estándar. La recta siempre pasa por el punto $(\bar{x}, \bar{y})$.

````python
# ============================================================
# Configuracion inicial y carga de datos
# ============================================================
# !pip install -q numpy pandas matplotlib seaborn scikit-learn scipy

import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression

warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (9, 5)

# Dataset de diabetes (Efron et al., 2004): 442 pacientes, 10 variables basales y una medida de progresion
# de la enfermedad un año despues. scaled=False conserva las UNIDADES ORIGINALES (necesario para interpretar).
datos = load_diabetes(scaled=False, as_frame=True)
X, y = datos.data, datos.target
print(f"Dataset: {X.shape[0]} pacientes x {X.shape[1]} variables | objetivo: progresión de la enfermedad (media = {y.mean():.1f})")
display(X.describe().T[["mean", "std", "min", "max"]].round(2))
print("Variables: age=edad, sex=sexo (codificado 1/2), bmi=indice de masa corporal, bp=presion arterial media,")
print("s1..s6 = mediciones de suero sanguineo (s1 colesterol total, s2 LDL, s3 HDL, s4 colesterol/HDL, s5 log-triglicéridos, s6 glucosa)")
````

````python
# ============================================================
# 1.4 Regresion simple desde cero: bmi -> progresion de la enfermedad
# ============================================================
x = X["bmi"].to_numpy()
yv = y.to_numpy()

# Formulas cerradas de la regresion simple
beta1 = np.sum((x - x.mean()) * (yv - yv.mean())) / np.sum((x - x.mean()) ** 2)
beta0 = yv.mean() - beta1 * x.mean()

# Verificacion contra dos implementaciones independientes
lr_simple = LinearRegression().fit(x.reshape(-1, 1), yv)
b1_np, b0_np = np.polyfit(x, yv, 1)
assert np.allclose([beta0, beta1], [lr_simple.intercept_, lr_simple.coef_[0]])
assert np.allclose([beta0, beta1], [b0_np, b1_np])
# Pendiente = correlacion * (s_y / s_x)
assert np.isclose(beta1, np.corrcoef(x, yv)[0, 1] * yv.std() / x.std())

y_hat = beta0 + beta1 * x
residuos = yv - y_hat
r2 = 1 - np.sum(residuos ** 2) / np.sum((yv - yv.mean()) ** 2)
print(f"Ecuación: progresión = {beta0:.2f} + {beta1:.2f} * bmi")
print(f"Interpretación: cada punto adicional de IMC se asocia con {beta1:.2f} unidades más de progresión (en promedio).")
print(f"R² = {r2:.3f}  (= correlación² = {np.corrcoef(x, yv)[0, 1] ** 2:.3f})")
print(f"Suma de residuos = {residuos.sum():.2e}  (≈ 0, como predice la teoría)")

fig, ax = plt.subplots(figsize=(9, 5.5))
ax.scatter(x, yv, s=18, alpha=0.5, color="#475569", label="Pacientes")
idx = np.random.default_rng(SEED).choice(len(x), 35, replace=False)
ax.vlines(x[idx], y_hat[idx], yv[idx], color="#dc2626", alpha=0.5, lw=1, label="Residuos (muestra de 35)")
xs = np.linspace(x.min(), x.max(), 100)
ax.plot(xs, beta0 + beta1 * xs, color="#1e3a8a", lw=3, label="Recta de MCO")
ax.plot(x.mean(), yv.mean(), "o", color="#f59e0b", ms=11, label="Centroide $(\\bar{x}, \\bar{y})$")
ax.set_xlabel("Índice de masa corporal (bmi)")
ax.set_ylabel("Progresión de la enfermedad")
ax.set_title("Regresión simple: la recta que minimiza la suma de residuos al cuadrado", fontweight="bold")
ax.legend()
plt.tight_layout()
plt.show()
````

````python
# ============================================================
# 1.2 Regresion multiple desde cero con la ecuacion normal y comparacion con scikit-learn
# ============================================================
class MCODesdeCero:
    """Minimos cuadrados ordinarios resolviendo la ecuacion normal X'X b = X'y."""
    def fit(self, X, y):
        Xb = np.c_[np.ones(len(X)), np.asarray(X, dtype=float)]     # columna de unos = intercepto
        self.beta_ = np.linalg.solve(Xb.T @ Xb, Xb.T @ np.asarray(y, dtype=float))   # mejor que invertir X'X explicitamente
        self.intercept_, self.coef_ = self.beta_[0], self.beta_[1:]
        return self
    def predict(self, X):
        return self.intercept_ + np.asarray(X, dtype=float) @ self.coef_

mco = MCODesdeCero().fit(X, y)
sk = LinearRegression().fit(X, y)

assert np.allclose(mco.coef_, sk.coef_, rtol=1e-6, atol=1e-8)
assert np.isclose(mco.intercept_, sk.intercept_, rtol=1e-6)
assert np.allclose(mco.predict(X), sk.predict(X))
print("Ecuación normal desde cero  ==  LinearRegression de scikit-learn  (coeficientes, intercepto y predicciones)")

comparacion = pd.DataFrame({"desde_cero": np.r_[mco.intercept_, mco.coef_], "scikit_learn": np.r_[sk.intercept_, sk.coef_]},
                           index=["intercepto"] + list(X.columns))
comparacion["diferencia_abs"] = (comparacion["desde_cero"] - comparacion["scikit_learn"]).abs()
display(comparacion.round(4))

# Nota numerica: formar X'X ELEVA AL CUADRADO el numero de condicion de X
Xb = np.c_[np.ones(len(X)), X.to_numpy()]
print(f"\nNúmero de condición: cond(X) = {np.linalg.cond(Xb):.2e}  |  cond(X'X) = {np.linalg.cond(Xb.T @ Xb):.2e}  (≈ cond(X)²)")
beta_lstsq = np.linalg.lstsq(Xb, y.to_numpy(), rcond=None)[0]       # via SVD: es lo que usa scikit-learn por debajo
print(f"Diferencia máxima ecuación normal vs. lstsq (SVD): {np.max(np.abs(mco.beta_ - beta_lstsq)):.2e}")
print("Aquí no hay problema, pero con colinealidad fuerte la ecuación normal pierde precisión: de ahí que se prefiera lstsq/SVD.")
````

## 2. Descenso del Gradiente 🏔️

La ecuación normal resuelve el problema en un solo paso, pero exige $O(p^3)$ operaciones y que $X^\top X$ quepa en memoria. El **descenso del gradiente** (GD) es la alternativa iterativa que escala a millones de datos y variables, y es **la base del entrenamiento de las redes neuronales**. Para la regresión lineal usamos la función de costo

$$J(\boldsymbol{\beta}) = \frac{1}{2n}\lVert X\boldsymbol{\beta} - \mathbf{y}\rVert^2, \qquad \nabla J(\boldsymbol{\beta}) = \frac{1}{n} X^\top (X\boldsymbol{\beta} - \mathbf{y}), \qquad \boldsymbol{\beta}_{t+1} = \boldsymbol{\beta}_t - \eta\,\nabla J(\boldsymbol{\beta}_t),$$

donde $\eta > 0$ es la **tasa de aprendizaje**. (El factor $\tfrac12$ solo simplifica el gradiente; no cambia el minimizador.)

### 2.1 ¿Cuándo converge y qué tan rápido?

$J$ es una **cuadrática** con Hessiano constante $H = \tfrac{1}{n}X^\top X$. Restando el óptimo $\hat{\boldsymbol{\beta}}$ (donde $\nabla J = 0$) la recursión queda $\boldsymbol{\beta}_{t+1} - \hat{\boldsymbol{\beta}} = (I - \eta H)(\boldsymbol{\beta}_t - \hat{\boldsymbol{\beta}})$, y por tanto

$$\boldsymbol{\beta}_t - \hat{\boldsymbol{\beta}} = (I - \eta H)^t (\boldsymbol{\beta}_0 - \hat{\boldsymbol{\beta}}).$$

Si $\lambda_{\max}, \lambda_{\min}$ son los valores propios extremos de $H$:

- **Converge si y solo si** $0 < \eta < 2/\lambda_{\max}$. Por encima de ese umbral, la dirección de mayor curvatura **diverge**.
- La velocidad la fija el **número de condición** $\kappa = \lambda_{\max}/\lambda_{\min}$: la mejor tasa $\eta^* = 2/(\lambda_{\max} + \lambda_{\min})$ da un factor de contracción $\rho = (\kappa - 1)/(\kappa + 1)$ por iteración. Si $\kappa$ es enorme, $\rho \approx 1$ y la convergencia es **lentísima**.
- **El escalado de variables reduce $\kappa$**: con columnas de media 0 y desviación 1, $H$ es esencialmente la matriz de correlaciones, con valores propios de orden 1. Con variables en unidades muy distintas, $\kappa$ puede ser de millones.

### 2.2 Variantes

| Variante | Gradiente calculado con | Característica |
|---|---|---|
| **Por lotes** (*batch*) | Los $n$ datos | Trayectoria suave y estable; costosa si $n$ es enorme |
| **Estocástico** (SGD) | 1 dato por paso | Muy barata y ruidosa; puede escapar de mínimos locales (útil en problemas no convexos) |
| **Mini-batch** | Un subconjunto de $m$ datos | Compromiso estándar en *deep learning* |

Aquí implementamos el GD por lotes (Géron, Cap. 4, desarrolla las tres).

````python
# ============================================================
# 2. Descenso del gradiente por lotes desde cero, con estandarizacion de variables
# ============================================================
def descenso_gradiente(Xb, y, eta, n_iter, beta0=None):
    """GD por lotes para J = (1/2n)||Xb b - y||^2. Devuelve beta final, historial de J y de beta."""
    n = len(y)
    beta = np.zeros(Xb.shape[1]) if beta0 is None else beta0.copy()
    costo, trayectoria = [], []
    for _ in range(n_iter):
        resid = Xb @ beta - y
        costo.append(0.5 * np.mean(resid ** 2))
        trayectoria.append(beta.copy())
        beta -= eta * (Xb.T @ resid) / n                          # beta <- beta - eta * gradiente
    return beta, np.array(costo), np.array(trayectoria)

# Estandarizacion: z = (x - media) / desviacion. Se anade la columna de unos (sin escalar) para el intercepto
mu, sd = X.to_numpy().mean(axis=0), X.to_numpy().std(axis=0)
Z = (X.to_numpy() - mu) / sd
Zb = np.c_[np.ones(len(Z)), Z]
yv = y.to_numpy()

# Solucion exacta (MCO) sobre las variables estandarizadas: es el "punto de llegada" que debe alcanzar el GD
beta_ols_z = np.linalg.lstsq(Zb, yv, rcond=None)[0]
J_opt = 0.5 * np.mean((Zb @ beta_ols_z - yv) ** 2)

# Valores propios del Hessiano H = Zb'Zb / n  ->  umbral de estabilidad y mejor tasa de aprendizaje
H = Zb.T @ Zb / len(yv)
lam = np.linalg.eigvalsh(H)
lam_min, lam_max = lam[0], lam[-1]
eta_opt = 2 / (lam_max + lam_min)
print(f"Hessiano (variables estandarizadas): λ_min = {lam_min:.4f}, λ_max = {lam_max:.3f}, κ = {lam_max / lam_min:.0f}")
print(f"Umbral de estabilidad 2/λ_max = {2 / lam_max:.3f} | mejor tasa 2/(λ_max+λ_min) = {eta_opt:.3f} | factor de contracción ρ = {(lam_max / lam_min - 1) / (lam_max / lam_min + 1):.4f}")

beta_gd, costo, tray = descenso_gradiente(Zb, yv, eta=eta_opt, n_iter=20_000)

# Volver a las unidades originales: b_j = b_j^z / sd_j ; b_0 = b_0^z - sum(b_j^z * mu_j / sd_j)
coef_orig = beta_gd[1:] / sd
interc_orig = beta_gd[0] - np.sum(beta_gd[1:] * mu / sd)

assert np.allclose(coef_orig, sk.coef_, rtol=1e-4, atol=1e-6), "El GD debe converger a la solucion de MCO"
assert np.isclose(interc_orig, sk.intercept_, rtol=1e-4)
print(f"\nGD tras 20.000 iteraciones: máxima diferencia con scikit-learn (coeficientes originales) = {np.max(np.abs(coef_orig - sk.coef_)):.2e}")
print("GD desde cero == LinearRegression: verificado.")

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
axes[0].plot(np.arange(1, len(costo) + 1), np.maximum(costo - J_opt, 1e-12), color="#2563eb", lw=2, label="$J(\\beta_t) - J^*$ (descenso del gradiente)")
axes[0].set_xscale("log"); axes[0].set_yscale("log")
axes[0].set_xlabel("Iteración (escala log)")
axes[0].set_ylabel("Exceso de costo sobre el mínimo exacto (escala log)")
axes[0].set_title("Curva de convergencia del descenso del gradiente", fontweight="bold")
axes[0].legend()

dist = np.linalg.norm(tray - beta_ols_z, axis=1)
axes[1].plot(dist, color="#16a34a", lw=2)
axes[1].set_yscale("log")
axes[1].set_xlabel("Iteración")
axes[1].set_ylabel("Distancia $\\Vert \\beta_t - \\hat{\\beta}_{MCO} \\Vert$ (escala log)")
axes[1].set_title("Distancia a la solución exacta (se frena en la precisión de máquina)", fontweight="bold")
plt.tight_layout()
plt.show()
````

````python
# ============================================================
# 2.1 Efecto de la tasa de aprendizaje y del escalado de variables
# ============================================================
umbral = 2 / lam_max
factores = [0.05, 0.5, 0.95, 1.05]
colores = ["#94a3b8", "#16a34a", "#f59e0b", "#dc2626"]

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
for f, c in zip(factores, colores):
    _, cst, _ = descenso_gradiente(Zb, yv, eta=f * umbral, n_iter=300)
    etiqueta = f"η = {f:.2f} × umbral" + ("  (DIVERGE)" if f > 1 else "")
    axes[0].plot(np.maximum(cst - J_opt, 1e-6), color=c, lw=2, label=etiqueta)
axes[0].set_yscale("log")
axes[0].set_ylim(1e-1, 1e9)
axes[0].set_xlabel("Iteración")
axes[0].set_ylabel("Exceso de costo $J(\\beta_t) - J^*$ (escala log)")
axes[0].set_title("Tasa de aprendizaje: umbral de estabilidad $2/\\lambda_{max}$", fontweight="bold")
axes[0].legend(fontsize=8)

# Escalado: mismas 3000 iteraciones con variables estandarizadas vs. variables en unidades originales
Xb_raw = np.c_[np.ones(len(X)), X.to_numpy()]
beta_ols_raw = np.linalg.lstsq(Xb_raw, yv, rcond=None)[0]
J_opt_raw = 0.5 * np.mean((Xb_raw @ beta_ols_raw - yv) ** 2)
lam_raw = np.linalg.eigvalsh(Xb_raw.T @ Xb_raw / len(yv))
eta_raw = 0.95 * 2 / lam_raw[-1]                                 # lo mejor que se puede hacer sin divergir (casi en el umbral)
_, cst_raw, _ = descenso_gradiente(Xb_raw, yv, eta=eta_raw, n_iter=3000)
_, cst_z, _ = descenso_gradiente(Zb, yv, eta=eta_opt, n_iter=3000)
axes[1].plot(np.maximum(cst_z - J_opt, 1e-12), color="#2563eb", lw=2, label=f"Estandarizadas (κ ≈ {lam_max / lam_min:,.0f})")
axes[1].plot(np.maximum(cst_raw - J_opt_raw, 1e-12), color="#dc2626", lw=2, label=f"Unidades originales (κ ≈ {lam_raw[-1] / lam_raw[0]:,.0e})")
axes[1].set_yscale("log")
axes[1].set_xlabel("Iteración")
axes[1].set_ylabel("Exceso de costo $J(\\beta_t) - J^*$ (escala log)")
axes[1].set_title("Escalar las variables reduce el número de condición", fontweight="bold")
axes[1].legend()
plt.tight_layout()
plt.show()

print(f"Tras 3000 iteraciones: exceso de costo con escalado = {cst_z[-1] - J_opt:.2e} | sin escalar = {cst_raw[-1] - J_opt_raw:.2e}")
print(f"Umbral de estabilidad sin escalar: 2/λ_max = {2 / lam_raw[-1]:.2e}, {umbral / (2 / lam_raw[-1]):,.0f} veces menor que con escalado -> el GD sin escalar apenas avanza.")
````

## 3. Regresión Múltiple: Interpretación, Inferencia y Calidad del Ajuste 🔍

### 3.1 Interpretación de los coeficientes

En $\hat{y} = \hat{\beta}_0 + \sum_j \hat{\beta}_j x_j$, el coeficiente $\hat{\beta}_j$ es el cambio esperado en $y$ por **cada unidad adicional de $x_j$, manteniendo constantes las demás variables** (*ceteris paribus*). Tres advertencias:

1. Es una **asociación**, no necesariamente causalidad.
2. El valor depende de **qué otras variables hay en el modelo** (el efecto de `bmi` no es el mismo solo que acompañado de `bp` y `s5`).
3. Depende de las **unidades**: para comparar la *importancia relativa* se usan los **coeficientes estandarizados** $\hat{\beta}_j\, s_{x_j}/s_y$, que miden el cambio en desviaciones estándar de $y$ por cada desviación estándar de $x_j$.

### 3.2 Inferencia

Con errores normales, $\;\dfrac{\hat{\beta}_j - \beta_j}{\operatorname{EE}(\hat{\beta}_j)} \sim t_{\,n-p-1}$, donde $\operatorname{EE}(\hat{\beta}_j) = \hat{\sigma}\sqrt{[(X^\top X)^{-1}]_{jj}}$. De ahí salen la **prueba $t$** de $H_0\!: \beta_j = 0$ (valor $p$) y el **intervalo de confianza** $\hat{\beta}_j \pm t_{1-\alpha/2,\,n-p-1}\,\operatorname{EE}(\hat{\beta}_j)$. Todo se calcula a mano a continuación.

### 3.3 $R^2$ y $R^2$ ajustado

$$R^2 = 1 - \frac{\text{RSS}}{\text{TSS}} = 1 - \frac{\sum (y_i - \hat{y}_i)^2}{\sum (y_i - \bar{y})^2}, \qquad R^2_{\text{aj}} = 1 - (1 - R^2)\,\frac{n - 1}{n - p - 1}.$$

$R^2$ es la proporción de la varianza de $y$ explicada por el modelo. **Nunca disminuye al agregar variables** (aunque sean ruido), por lo que premia la complejidad; $R^2_{\text{aj}}$ **penaliza** los grados de libertad consumidos y sí puede bajar. En datos *no vistos* $R^2$ puede incluso ser **negativo** (un modelo peor que predecir siempre la media).

````python
# ============================================================
# 3.2 Tabla de coeficientes calculada a mano: error estandar, t, valor p, IC 95 % y coeficiente estandarizado
# ============================================================
n, p = X.shape
Xb = np.c_[np.ones(n), X.to_numpy()]
beta = np.linalg.lstsq(Xb, yv, rcond=None)[0]
resid = yv - Xb @ beta
gl = n - p - 1                                                    # grados de libertad residuales
sigma2 = resid @ resid / gl                                       # estimador insesgado de sigma^2
cov_beta = sigma2 * np.linalg.inv(Xb.T @ Xb)                      # Var(beta) = sigma^2 (X'X)^-1
ee = np.sqrt(np.diag(cov_beta))
t_val = beta / ee
p_val = 2 * stats.t.sf(np.abs(t_val), gl)
t_crit = stats.t.ppf(0.975, gl)
beta_est = np.r_[np.nan, beta[1:] * X.std().to_numpy() / yv.std(ddof=1)]    # coeficiente estandarizado

tabla = pd.DataFrame({"coef": beta, "error_est": ee, "t": t_val, "valor_p": p_val,
                      "IC95_inf": beta - t_crit * ee, "IC95_sup": beta + t_crit * ee, "coef_estandarizado": beta_est},
                     index=["intercepto"] + list(X.columns))
display(tabla.round(3))
print(f"σ̂ (error estándar residual) = {np.sqrt(sigma2):.2f} | grados de libertad = {gl}")
print("Variables con valor p < 0.05:", [c for c in X.columns if tabla.loc[c, "valor_p"] < 0.05])

# Verificacion opcional contra statsmodels (si esta instalado)
try:
    import statsmodels.api as sm
    res_sm = sm.OLS(yv, sm.add_constant(X.to_numpy())).fit()
    assert np.allclose(res_sm.params, beta) and np.allclose(res_sm.bse, ee) and np.allclose(res_sm.pvalues, p_val)
    print("Verificación con statsmodels: coeficientes, errores estándar y valores p coinciden.")
except ImportError:
    print("statsmodels no está instalado: se omite la verificación opcional (el resto del cuaderno no depende de él).")
````

````python
# ============================================================
# 3.3 R2 y R2 ajustado a mano; el R2 "premia" agregar ruido, el R2 ajustado y el R2 de prueba no
# ============================================================
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

def r2(y_real, y_pred):
    return 1 - np.sum((y_real - y_pred) ** 2) / np.sum((y_real - y_real.mean()) ** 2)

def r2_ajustado(r2_val, n_obs, n_vars):
    return 1 - (1 - r2_val) * (n_obs - 1) / (n_obs - n_vars - 1)

assert np.isclose(r2(yv, Xb @ beta), r2_score(yv, Xb @ beta))
print(f"R² (a mano) = {r2(yv, Xb @ beta):.4f} | R² ajustado = {r2_ajustado(r2(yv, Xb @ beta), n, p):.4f}  (modelo con las 10 variables, todos los datos)")

# Experimento: agregar k variables de RUIDO PURO y medir R2 de entrenamiento, R2 ajustado y R2 en prueba
X_tr, X_te, y_tr, y_te = train_test_split(X.to_numpy(), yv, test_size=0.3, random_state=SEED)
ks = [0, 5, 10, 20, 40, 80]
res = {k: [] for k in ks}
rng = np.random.default_rng(SEED)
for _ in range(30):                                               # 30 repeticiones con ruido distinto
    ruido_tr, ruido_te = rng.normal(size=(len(X_tr), max(ks))), rng.normal(size=(len(X_te), max(ks)))
    for k in ks:
        A_tr = np.c_[np.ones(len(X_tr)), X_tr, ruido_tr[:, :k]]
        A_te = np.c_[np.ones(len(X_te)), X_te, ruido_te[:, :k]]
        b = np.linalg.lstsq(A_tr, y_tr, rcond=None)[0]
        r2_tr = r2(y_tr, A_tr @ b)
        res[k].append((r2_tr, r2_ajustado(r2_tr, len(y_tr), 10 + k), r2(y_te, A_te @ b)))
tabla_r2 = pd.DataFrame({k: np.mean(v, axis=0) for k, v in res.items()}, index=["R² entrenamiento", "R² ajustado (train)", "R² prueba"]).T
tabla_r2.index.name = "variables de ruido añadidas"
display(tabla_r2.round(4))

fig, ax = plt.subplots(figsize=(9.5, 5))
for col, c, mk in zip(tabla_r2.columns, ["#2563eb", "#f59e0b", "#dc2626"], ["o", "s", "^"]):
    ax.plot(tabla_r2.index, tabla_r2[col], marker=mk, color=c, lw=2, label=col)
ax.set_xlabel("Número de variables de ruido añadidas al modelo")
ax.set_ylabel("Coeficiente de determinación")
ax.set_title("El R² siempre sube con ruido; el R² ajustado y el de prueba lo delatan", fontweight="bold")
ax.legend()
plt.tight_layout()
plt.show()
````

## 4. Supuestos del Modelo y Diagnóstico de Residuos 🩺

La **estimación** MCO siempre se puede calcular; lo que cambia es **qué tan confiables** son la inferencia (errores estándar, valores $p$, intervalos) y las predicciones. Los supuestos clásicos se recuerdan con la sigla **LINE** (más una condición sobre las variables):

| Supuesto | Qué exige | Cómo se diagnostica | Si falla |
|---|---|---|---|
| **L**inealidad | $\mathbb{E}[y \mid \mathbf{x}]$ es lineal en los parámetros | Residuos vs. ajustados: **no debe haber curvatura** | Transformar variables, añadir términos polinomiales o interacciones |
| **I**ndependencia | Errores no correlacionados entre sí | Residuos vs. orden de observación (series de tiempo); estadístico de Durbin–Watson | Modelos de series temporales, errores robustos/agrupados |
| **N**ormalidad | $\varepsilon \sim \mathcal{N}(0, \sigma^2)$ (solo necesaria para la inferencia exacta en muestras pequeñas) | Gráfico Q-Q; prueba de Shapiro–Wilk | Transformar $y$, remuestreo (*bootstrap*); con $n$ grande el TLC alivia el problema |
| **E**quivarianza (homocedasticidad) | $\operatorname{Var}(\varepsilon_i) = \sigma^2$ constante | Residuos vs. ajustados (**forma de embudo**), gráfico escala-ubicación; prueba de Breusch–Pagan | Transformar $y$ (log), mínimos cuadrados ponderados, errores estándar robustos |
| Sin multicolinealidad exacta | Columnas de $X$ linealmente independientes | VIF, número de condición (Sección 5) | Eliminar/combinar variables, regularización |

Los **residuos estandarizados (internamente estudentizados)** $r_i = e_i / \big(\hat{\sigma}\sqrt{1 - h_{ii}}\big)$ corrigen por el *apalancamiento* $h_{ii}$ (diagonal de la matriz sombrero $H$), que mide cuánta influencia tiene la observación $i$ sobre su propio ajuste. La **distancia de Cook** $D_i = r_i^2\,\dfrac{h_{ii}}{(p+1)(1 - h_{ii})}$ combina residuo y apalancamiento para señalar puntos influyentes.

Para la **prueba de Breusch–Pagan** (versión de Koenker) se regresan los residuos al cuadrado sobre las variables explicativas; el estadístico $LM = n R^2_{\text{aux}}$ sigue (bajo $H_0$: homocedasticidad) una $\chi^2_p$. La **prueba de Shapiro–Wilk** contrasta $H_0$: los datos son normales.

````python
# ============================================================
# 4. Diagnostico de residuos del modelo completo de diabetes
# ============================================================
ajustados = Xb @ beta
Q, _ = np.linalg.qr(Xb)
apalanc = (Q ** 2).sum(axis=1)                                    # h_ii = diag(H), H = X (X'X)^-1 X'
r_est = resid / np.sqrt(sigma2 * (1 - apalanc))                   # residuos estandarizados
cook = r_est ** 2 * apalanc / ((p + 1) * (1 - apalanc))           # distancia de Cook

def tendencia(x_, y_, ventana=45):
    """Media movil de y_ ordenada por x_ (suavizador simple para ver la tendencia)."""
    orden = np.argsort(x_)
    return x_[orden], pd.Series(y_[orden]).rolling(ventana, center=True, min_periods=ventana // 2).mean().to_numpy()

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
# (1) Linealidad / equivarianza
axes[0, 0].scatter(ajustados, resid, s=15, alpha=0.5, color="#475569")
axes[0, 0].plot(*tendencia(ajustados, resid), color="#dc2626", lw=2.5, label="Tendencia (media móvil)")
axes[0, 0].axhline(0, color="black", linestyle="--")
axes[0, 0].set_xlabel("Valores ajustados $\\hat{y}$"); axes[0, 0].set_ylabel("Residuos $e$")
axes[0, 0].set_title("① Residuos vs. ajustados (linealidad y varianza)", fontweight="bold"); axes[0, 0].legend()
# (2) Normalidad
(osm, osr), (pend, inter, _) = stats.probplot(r_est, dist="norm")
axes[0, 1].scatter(osm, osr, s=15, alpha=0.6, color="#2563eb")
axes[0, 1].plot(osm, pend * osm + inter, color="#dc2626", lw=2, label="Normalidad perfecta")
axes[0, 1].set_xlabel("Cuantiles teóricos $\\mathcal{N}(0,1)$"); axes[0, 1].set_ylabel("Residuos estandarizados")
axes[0, 1].set_title("② Gráfico Q-Q (normalidad)", fontweight="bold"); axes[0, 1].legend()
# (3) Escala-ubicacion
axes[1, 0].scatter(ajustados, np.sqrt(np.abs(r_est)), s=15, alpha=0.5, color="#475569")
axes[1, 0].plot(*tendencia(ajustados, np.sqrt(np.abs(r_est))), color="#dc2626", lw=2.5, label="Tendencia (media móvil)")
axes[1, 0].set_xlabel("Valores ajustados $\\hat{y}$"); axes[1, 0].set_ylabel("$\\sqrt{|r_i|}$")
axes[1, 0].set_title("③ Escala-ubicación (homocedasticidad)", fontweight="bold"); axes[1, 0].legend()
# (4) Apalancamiento e influencia
sc = axes[1, 1].scatter(apalanc, r_est, c=cook, cmap="viridis_r", s=25, alpha=0.8)
plt.colorbar(sc, ax=axes[1, 1], label="Distancia de Cook")
top = np.argsort(cook)[-3:]
for i in top:
    axes[1, 1].annotate(str(i), (apalanc[i], r_est[i]), textcoords="offset points", xytext=(5, 5), fontsize=9)
axes[1, 1].axhline(0, color="black", linestyle="--")
axes[1, 1].set_xlabel("Apalancamiento $h_{ii}$"); axes[1, 1].set_ylabel("Residuo estandarizado")
axes[1, 1].set_title("④ Apalancamiento vs. residuo (puntos influyentes)", fontweight="bold")
plt.tight_layout()
plt.show()

# Pruebas formales
W, p_sw = stats.shapiro(r_est)
aux = np.linalg.lstsq(Xb, resid ** 2, rcond=None)[0]
r2_aux = r2(resid ** 2, Xb @ aux)
LM, p_bp = n * r2_aux, stats.chi2.sf(n * r2_aux, p)
DW = np.sum(np.diff(resid) ** 2) / np.sum(resid ** 2)
veredicto = lambda pv: "se RECHAZA H0" if pv < 0.05 else "no se rechaza H0"
print(f"Shapiro-Wilk (normalidad):        W = {W:.4f}, p = {p_sw:.4f} -> {veredicto(p_sw)}")
print(f"Breusch-Pagan (homocedasticidad): LM = {LM:.2f}, p = {p_bp:.4f} -> {veredicto(p_bp)}")
print(f"Durbin-Watson (autocorrelación):  DW = {DW:.3f} (≈2 si no hay autocorrelación). Ojo: el orden de los pacientes es arbitrario,")
print("   así que aquí esta prueba no tiene sentido sustantivo; importa en datos ordenados en el tiempo.")
print(f"Observaciones más influyentes (Cook): {[int(i) for i in top[::-1]]} | D máxima = {cook.max():.3f} (regla práctica: revisar si D > 4/n = {4 / n:.3f})")
````

````python
# ============================================================
# 4.1 Como se VE un supuesto violado: bien especificado vs. mal especificado (datos sinteticos)
# ============================================================
rng = np.random.default_rng(SEED)
n_s = 300
x_s = rng.uniform(0, 10, n_s)
casos = {
    "Bien especificado:  y = 2 + 1.5x + ε,  ε ~ N(0, 2²)": 2 + 1.5 * x_s + rng.normal(0, 2, n_s),
    "Mal especificado:  y = 2 + 0.25x² + ε,  ruido que crece con x": 2 + 0.25 * x_s ** 2 + rng.normal(0, 0.25 + 0.35 * x_s, n_s),
}
fig, axes = plt.subplots(2, 2, figsize=(14, 9))
for fila, (nombre, y_s) in enumerate(casos.items()):
    A = np.c_[np.ones(n_s), x_s]
    b = np.linalg.lstsq(A, y_s, rcond=None)[0]
    e = y_s - A @ b
    aj = A @ b
    s2 = e @ e / (n_s - 2)
    h = (np.linalg.qr(A)[0] ** 2).sum(axis=1)
    r_s = e / np.sqrt(s2 * (1 - h))
    aux_s = np.linalg.lstsq(A, e ** 2, rcond=None)[0]
    p_bp_s = stats.chi2.sf(n_s * r2(e ** 2, A @ aux_s), 1)
    p_sw_s = stats.shapiro(r_s)[1]
    axes[fila, 0].scatter(aj, e, s=14, alpha=0.5, color="#475569")
    axes[fila, 0].plot(*tendencia(aj, e), color="#dc2626", lw=2.5)
    axes[fila, 0].axhline(0, color="black", linestyle="--")
    axes[fila, 0].set_title(f"{nombre}\nBreusch-Pagan p = {p_bp_s:.3g}", fontsize=10, fontweight="bold")
    axes[fila, 0].set_xlabel("Valores ajustados"); axes[fila, 0].set_ylabel("Residuos")
    (osm, osr), (pend, inter, _) = stats.probplot(r_s, dist="norm")
    axes[fila, 1].scatter(osm, osr, s=14, alpha=0.6, color="#2563eb")
    axes[fila, 1].plot(osm, pend * osm + inter, color="#dc2626", lw=2)
    axes[fila, 1].set_title(f"Gráfico Q-Q  (Shapiro-Wilk p = {p_sw_s:.3g})", fontsize=10, fontweight="bold")
    axes[fila, 1].set_xlabel("Cuantiles teóricos"); axes[fila, 1].set_ylabel("Residuos estandarizados")
plt.tight_layout()
plt.show()
print("Fila superior: nube horizontal sin estructura y Q-Q sobre la diagonal.")
print("Fila inferior: curvatura (la recta no captura la parábola) + embudo (la varianza crece) + colas pesadas en el Q-Q.")
print("Remedios: añadir x² (linealidad), transformar y con log/raíz o usar mínimos cuadrados ponderados / errores robustos (varianza).")
````

## 5. Multicolinealidad: cuando las Variables se Parecen Demasiado 👯

Hay **multicolinealidad** cuando una variable explicativa se puede predecir casi perfectamente a partir de las demás. No viola Gauss–Markov (siempre que no sea exacta), pero tiene un efecto devastador sobre la **interpretación**: $\operatorname{Var}(\hat{\beta}_j)$ se **infla**, los coeficientes se vuelven inestables (cambian de signo con pequeñas variaciones de la muestra) y los valores $p$ pierden sentido, aunque las **predicciones** del modelo conjunto sigan siendo buenas.

### Factor de inflación de la varianza (VIF)

Para cada variable $j$ se regresa $x_j$ contra todas las demás variables y se obtiene su $R_j^2$. Entonces

$$\text{VIF}_j = \frac{1}{1 - R_j^2}, \qquad \operatorname{Var}(\hat{\beta}_j) = \frac{\sigma^2}{(n-1)\,s_{x_j}^2}\cdot \text{VIF}_j.$$

Un $\text{VIF}_j = 1$ significa que $x_j$ no es explicable por las demás; $\text{VIF}_j = 10$ significa que su varianza se multiplicó por 10 respecto al caso ortogonal. **Reglas prácticas** (convenciones, no leyes): VIF > 5 merece atención y VIF > 10 indica problema serio. En el dataset de diabetes, `s1` (colesterol total) y `s2` (LDL) tienen correlación $\approx 0.9$: ¡una es casi una copia de la otra!

````python
# ============================================================
# 5. VIF calculado a mano + inestabilidad de los coeficientes (bootstrap)
# ============================================================
def calcular_vif(M):
    """VIF_j = 1 / (1 - R2_j), con R2_j de regresar la columna j contra las demas (con intercepto)."""
    M = np.asarray(M, dtype=float)
    vifs = []
    for j in range(M.shape[1]):
        otras = np.c_[np.ones(len(M)), np.delete(M, j, axis=1)]
        b = np.linalg.lstsq(otras, M[:, j], rcond=None)[0]
        vifs.append(1 / (1 - r2(M[:, j], otras @ b)))
    return np.array(vifs)

vif = pd.Series(calcular_vif(X.to_numpy()), index=X.columns)
try:                                                                # verificacion opcional con statsmodels
    from statsmodels.stats.outliers_influence import variance_inflation_factor
    vif_sm = [variance_inflation_factor(np.c_[np.ones(len(X)), X.to_numpy()], j + 1) for j in range(X.shape[1])]
    assert np.allclose(vif.to_numpy(), vif_sm)
    print("Verificación con statsmodels: VIF coincide.\n")
except ImportError:
    print("statsmodels no está instalado: se omite la verificación opcional del VIF.\n")
display(vif.round(2).to_frame("VIF").T)

# Bootstrap: 300 remuestreos -> coeficientes estandarizados de s1 y s2 (variables casi duplicadas)
rng = np.random.default_rng(SEED)
nombres = list(X.columns)
j1, j2 = nombres.index("s1"), nombres.index("s2")
boot = np.zeros((300, 2))
for b_ in range(300):
    idx = rng.integers(0, n, n)
    Zb_ = np.c_[np.ones(n), Z[idx]]
    bb = np.linalg.lstsq(Zb_, yv[idx], rcond=None)[0]
    boot[b_] = bb[1 + j1], bb[1 + j2]
print(f"Correlación entre los coeficientes bootstrap de s1 y s2: {np.corrcoef(boot.T)[0, 1]:.3f}")
print(f"Desviación estándar bootstrap -> s1: {boot[:, 0].std():.1f} | s2: {boot[:, 1].std():.1f}   (coeficientes por 1 desv. estándar de la variable)")

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
sns.heatmap(X.corr(), annot=True, fmt=".2f", cmap="RdBu_r", center=0, vmin=-1, vmax=1, ax=axes[0], annot_kws={"size": 7}, cbar=False)
axes[0].set_title("Correlaciones entre variables explicativas", fontweight="bold")
colores = ["#dc2626" if v > 10 else "#f59e0b" if v > 5 else "#2563eb" for v in vif]
axes[1].barh(vif.index, vif.values, color=colores)
axes[1].axvline(5, color="#f59e0b", linestyle="--", label="VIF = 5")
axes[1].axvline(10, color="#dc2626", linestyle="--", label="VIF = 10")
axes[1].set_xlabel("VIF"); axes[1].set_title("Factor de inflación de la varianza", fontweight="bold"); axes[1].legend()
axes[2].scatter(boot[:, 0], boot[:, 1], s=14, alpha=0.5, color="#7c3aed")
axes[2].axhline(0, color="gray", lw=0.8); axes[2].axvline(0, color="gray", lw=0.8)
axes[2].set_xlabel("Coeficiente de s1 (colesterol total)"); axes[2].set_ylabel("Coeficiente de s2 (LDL)")
axes[2].set_title("Bootstrap: s1 y s2 «se compensan»", fontweight="bold")
plt.tight_layout()
plt.show()
print("La nube alargada y de pendiente negativa indica que el modelo puede 'repartir' el efecto entre s1 y s2 de muchas maneras:")
print("la SUMA de ambos efectos está bien estimada, pero cada coeficiente por separado es inestable.")
````

## 6. Extensiones: Regresión Polinomial y Regularización (Ridge y Lasso) 🧩

### 6.1 Polinomios: "lineal" significa lineal en los **parámetros**

El modelo $y = \beta_0 + \beta_1 x + \beta_2 x^2 + \cdots + \beta_d x^d$ es **lineal en $\boldsymbol{\beta}$**: basta construir una nueva matriz de diseño con las columnas $(1, x, x^2, \ldots, x^d)$ y aplicar exactamente la misma ecuación normal. Más grado = más flexibilidad = más riesgo de **sobreajuste** (Cuaderno 00). Lo mismo ocurre con la multicolinealidad: las potencias de $x$ están fuertemente correlacionadas.

### 6.2 Regularización

Para domar la varianza se **penaliza** el tamaño de los coeficientes (sin penalizar el intercepto y con las variables **estandarizadas**):

| Método | Función objetivo | Efecto | Solución |
|---|---|---|---|
| **Ridge** (L2) | $\lVert \mathbf{y} - X\boldsymbol{\beta}\rVert^2 + \alpha\lVert\boldsymbol{\beta}\rVert_2^2$ | Encoge los coeficientes hacia cero; reparte el efecto entre variables correlacionadas | **Cerrada:** $\hat{\boldsymbol{\beta}} = (X^\top X + \alpha I)^{-1}X^\top\mathbf{y}$ |
| **Lasso** (L1) | $\tfrac{1}{2n}\lVert \mathbf{y} - X\boldsymbol{\beta}\rVert^2 + \alpha\lVert\boldsymbol{\beta}\rVert_1$ | Lleva algunos coeficientes **exactamente a cero** (selección de variables) | Sin forma cerrada: descenso por coordenadas |

Ridge tiene una ventaja didáctica: al sumar $\alpha I$, la matriz $X^\top X + \alpha I$ **siempre es invertible** (para $\alpha > 0$), lo que resuelve de raíz la singularidad por colinealidad. Es una forma de sesgo deliberado a cambio de menor varianza (Cuaderno 00, Sección 5.3). La elección de $\alpha$ se hace por validación cruzada (`RidgeCV`, `LassoCV`). El tratamiento completo está en [DSP — Regresión Polinomial y Regularización](../../Data%20Science%20programming/07%20-%20Regression/03_Regresion_Polinomial_y_Regularizacion.ipynb) y en [Data Mining — Regresión Lineal, Polinomial y Regularizada](../../Data%20Mining/02%20-%20Clasificacion%20y%20Regresion/03_Regresion_Lineal_Polinomial_y_Regularizada.ipynb).

````python
# ============================================================
# 6.2 Ridge desde cero (forma cerrada) vs. scikit-learn + trayectorias de regularizacion en diabetes
# ============================================================
from sklearn.linear_model import Ridge, Lasso
from sklearn.preprocessing import StandardScaler

def ridge_desde_cero(Z_, y_, alpha):
    """Ridge con forma cerrada sobre datos centrados: intercepto SIN penalizar."""
    Zc, yc = Z_ - Z_.mean(axis=0), y_ - y_.mean()
    b = np.linalg.solve(Zc.T @ Zc + alpha * np.eye(Z_.shape[1]), Zc.T @ yc)
    return y_.mean() - Z_.mean(axis=0) @ b, b

esc = StandardScaler().fit(X)                                    # (en un proyecto real: ajustar solo con train)
Zs = esc.transform(X)
for alpha in [0.1, 10, 1000]:
    b0, b = ridge_desde_cero(Zs, yv, alpha)
    sk_r = Ridge(alpha=alpha).fit(Zs, yv)
    assert np.allclose(b, sk_r.coef_) and np.isclose(b0, sk_r.intercept_)
print("Ridge desde cero (forma cerrada) == sklearn.linear_model.Ridge para alpha = 0.1, 10 y 1000.")

alfas_r = np.logspace(-2, 4, 40)
alfas_l = np.logspace(-2, 1.6, 40)                                # Lasso usa otra escala de alpha (divide por 2n)
camino_ridge = np.array([ridge_desde_cero(Zs, yv, a)[1] for a in alfas_r])
camino_lasso = np.array([Lasso(alpha=a, max_iter=20_000).fit(Zs, yv).coef_ for a in alfas_l])

fig, axes = plt.subplots(1, 2, figsize=(15, 5.2))
for ax, alf, camino, titulo, xlab in [(axes[0], alfas_r, camino_ridge, "Ridge: los coeficientes se encogen suavemente", "Penalización $\\alpha$ de Ridge (escala log)"),
                                       (axes[1], alfas_l, camino_lasso, "Lasso: los coeficientes se anulan uno a uno", "Penalización $\\alpha$ de Lasso (escala log)")]:
    for j, nombre in enumerate(X.columns):
        ax.plot(alf, camino[:, j], lw=2, label=nombre)
    ax.set_xscale("log"); ax.axhline(0, color="black", lw=0.8)
    ax.set_xlabel(xlab); ax.set_ylabel("Coeficiente (variables estandarizadas)")
    ax.set_title(titulo, fontweight="bold")
axes[1].legend(ncol=2, fontsize=8, loc="upper right")
plt.tight_layout()
plt.show()
print("Fíjate en s1 y s2 (colineales): Ridge los encoge juntos; Lasso tiende a conservar uno y anular el otro.")
````

````python
# ============================================================
# 6.1-6.2 Polinomio de grado alto: MCO vs. Ridge vs. Lasso (datos sinteticos, semilla fija)
# ============================================================
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import RidgeCV, LassoCV

rng = np.random.default_rng(SEED)
f_real = lambda x_: np.sin(2 * np.pi * x_)
x_ent = rng.uniform(0, 1, 20); y_ent = f_real(x_ent) + rng.normal(0, 0.3, 20)
x_pru = rng.uniform(0, 1, 2000); y_pru = f_real(x_pru) + rng.normal(0, 0.3, 2000)
GRADO = 12

modelos = {
    "MCO (sin regularizar)": make_pipeline(PolynomialFeatures(GRADO, include_bias=False), StandardScaler(), LinearRegression()),
    "Ridge (α por CV)":      make_pipeline(PolynomialFeatures(GRADO, include_bias=False), StandardScaler(), RidgeCV(alphas=np.logspace(-6, 2, 50))),
    "Lasso (α por CV)":      make_pipeline(PolynomialFeatures(GRADO, include_bias=False), StandardScaler(), LassoCV(alphas=np.logspace(-4, 0, 60), cv=5, max_iter=200_000)),
}
filas, xs = [], np.linspace(0, 1, 400)
fig, ax = plt.subplots(figsize=(11, 5.5))
ax.scatter(x_ent, y_ent, color="#475569", s=35, zorder=3, label="Datos de entrenamiento (n=20)")
ax.plot(xs, f_real(xs), "k--", lw=1.5, label="Verdad")
for (nombre, mod), c in zip(modelos.items(), ["#dc2626", "#2563eb", "#16a34a"]):
    mod.fit(x_ent.reshape(-1, 1), y_ent)
    ax.plot(xs, mod.predict(xs.reshape(-1, 1)), color=c, lw=2.3, label=nombre)
    coefs = mod[-1].coef_
    filas.append({"modelo": nombre, "MSE_entrenamiento": np.mean((y_ent - mod.predict(x_ent.reshape(-1, 1))) ** 2),
                  "MSE_prueba": np.mean((y_pru - mod.predict(x_pru.reshape(-1, 1))) ** 2),
                  "coef_no_nulos": int(np.sum(np.abs(coefs) > 1e-8)), "max|coef|": np.max(np.abs(coefs))})
ax.set_ylim(-2.5, 2.5)
ax.set_xlabel("$x$"); ax.set_ylabel("$y$")
ax.set_title(f"Polinomio de grado {GRADO} con 20 datos: la regularización doma el sobreajuste", fontweight="bold")
ax.legend(loc="lower left")
plt.tight_layout()
plt.show()
display(pd.DataFrame(filas).set_index("modelo").round(3))
print("El MSE de prueba se compara con el error irreducible sigma^2 = 0.09. Observa también la magnitud de los coeficientes.")
````

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Modelo y MCO** | $\mathbf{y} = X\boldsymbol{\beta} + \boldsymbol{\varepsilon}$; MCO minimiza RSS y su condición de primer orden es la **ecuación normal** $X^\top X\boldsymbol{\beta} = X^\top\mathbf{y}$ |
| **Geometría** | $\hat{\mathbf{y}}$ es la proyección ortogonal de $\mathbf{y}$ sobre el espacio de columnas de $X$; los residuos son ortogonales a $X$ |
| **Gauss–Markov / MLE** | MCO es BLUE con errores no correlacionados y varianza constante, y es el MLE con errores normales |
| **Descenso del gradiente** | Converge si $\eta < 2/\lambda_{\max}$; la velocidad depende de $\kappa$; **estandarizar** reduce $\kappa$ drásticamente |
| **Verificación** | Las implementaciones desde cero (ecuación normal, GD y Ridge) coincidieron con `scikit-learn` |
| **Coeficientes** | Efecto *ceteris paribus* por unidad; con error estándar, valor $p$ e IC calculados a mano; los estandarizados permiten comparar |
| **$R^2$ vs. $R^2$ ajustado** | $R^2$ solo sube con más variables (incluso ruido); $R^2_{\text{aj}}$ y el $R^2$ de prueba penalizan la complejidad |
| **Supuestos (LINE)** | Linealidad, Independencia, Normalidad, Equivarianza: se diagnostican con gráficos de residuos y pruebas |
| **Multicolinealidad** | VIF alto $\Rightarrow$ coeficientes inestables (aunque la predicción conjunta sea buena) |
| **Regularización** | Ridge (forma cerrada, encoge) y Lasso (anula): cambian un poco de sesgo por mucha menos varianza |
