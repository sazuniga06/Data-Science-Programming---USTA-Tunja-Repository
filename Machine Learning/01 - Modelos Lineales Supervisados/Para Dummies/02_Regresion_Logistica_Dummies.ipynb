## 1. Preguntas de sí o no, con probabilidad 🎲

Un profesor quiere anticipar si sus estudiantes **aprobarán o no** según las **horas** que estudian. La respuesta solo puede ser **sí (1)** o **no (0)**. Si intentáramos usar una línea recta, nos diría cosas absurdas como «tiene 130 % de probabilidad de aprobar» o «-20 %». Una probabilidad **solo puede estar entre 0 % y 100 %**.

La solución es una curva en forma de **S**:

- Con **pocas horas** de estudio, la probabilidad es **cercana a 0 %** y sube despacio.
- Hacia el **medio** sube **rápido** (cada hora extra cambia mucho las cosas).
- Con **muchas horas**, se acerca a **100 %**, pero **nunca lo pasa**.

Es como un **regulador de luz** (dimmer): va de apagado a encendido de forma suave, sin salirse de los extremos. Veámoslo con datos de 100 estudiantes.

````python
# Datos inventados: horas de estudio y si el estudiante aprobo (1) o no (0)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

rng = np.random.default_rng(42)
horas = rng.uniform(0, 10, 100)
probabilidad_real = 1 / (1 + np.exp(-(-3 + 0.7 * horas)))             # la "regla oculta": mas horas -> mas probabilidad de aprobar
aprobo = rng.binomial(1, probabilidad_real)

# Separamos un grupo para "el examen": los estudiantes con los que NO se entrena
horas_ent, horas_exa, aprobo_ent, aprobo_exa = train_test_split(horas.reshape(-1, 1), aprobo, test_size=0.3, random_state=42)

modelo = LogisticRegression()
modelo.fit(horas_ent, aprobo_ent)

x_dibujo = np.linspace(0, 10, 200).reshape(-1, 1)
plt.figure(figsize=(8.5, 5))
plt.scatter(horas_ent, aprobo_ent + rng.uniform(-0.03, 0.03, len(aprobo_ent)), color="#f59e0b", edgecolor="white", s=60, label="Estudiantes (0 = no aprobó, 1 = aprobó)")
plt.plot(x_dibujo, modelo.predict_proba(x_dibujo)[:, 1], color="#16a34a", lw=3, label="Probabilidad de aprobar (curva en S)")
plt.xlabel("Horas de estudio")
plt.ylabel("Probabilidad de aprobar")
plt.title("La curva en S nunca se sale de 0 % a 100 %")
plt.legend()
plt.show()

for h in [2, 5, 8]:
    print(f"Un estudiante que estudia {h} horas tiene {modelo.predict_proba([[h]])[0, 1]:.0%} de probabilidad de aprobar.")
````

## 2. De la probabilidad a la decisión: el punto de corte ✂️

El modelo no dice «aprobará» o «no aprobará»: dice una **probabilidad**. Para tomar una decisión hay que fijar un **punto de corte**. Es como la nota mínima para aprobar una materia:

- Con corte en **50 %**: «si la probabilidad es de la mitad o más, digo que aprobará».
- Con un corte **más bajo (30 %)**: soy **más optimista**; marco a más estudiantes como «aprobarán», pero cometo más **falsas alarmas**.
- Con un corte **más alto (70 %)**: soy **más exigente**; marco a menos, pero **se me escapan** algunos que sí aprobarían.

No hay un corte «correcto» para todo: depende de **qué error cuesta más**. Si fuera un examen médico, preferiríamos que **no se nos escape ningún enfermo**, aunque haya falsas alarmas.

> 📌 En este cuaderno solo contamos aciertos y errores de forma sencilla. Las formas serias de evaluar un modelo se estudian a fondo en el **Módulo 03** del curso.

````python
# Efecto del punto de corte, evaluado con estudiantes que el modelo NO vio (el "examen")
probabilidades = modelo.predict_proba(horas_exa)[:, 1]

filas = []
for corte in [0.3, 0.5, 0.7]:
    dice_aprueba = probabilidades >= corte
    filas.append({"punto de corte": f"{corte:.0%}",
                  "aciertos": int(np.sum(dice_aprueba == (aprobo_exa == 1))),
                  "falsas alarmas (dijo que aprobaba, pero no)": int(np.sum(dice_aprueba & (aprobo_exa == 0))),
                  "se nos escaparon (aprobó, dijo que no)": int(np.sum(~dice_aprueba & (aprobo_exa == 1)))})
print(f"Estudiantes del examen: {len(aprobo_exa)}")
print(pd.DataFrame(filas).to_string(index=False))
print("\nA medida que subes el corte, bajan las falsas alarmas y suben los que se escapan: no hay almuerzo gratis.")
````

---
## 3. Los «momios»: cuántas veces es más probable 🎰

En las rifas y las apuestas se habla de **«3 a 1»**. Eso es lo que se llama **momios** (*odds*): cuántas veces es **más probable que pase a que no pase**.

| Probabilidad de aprobar | Se dice... | Momios |
|---|---|---|
| 50 % | «mitad y mitad» | 1 a 1 |
| 75 % | «es probable» | 3 a 1 (3 veces más probable aprobar que no) |
| 90 % | «casi seguro» | 9 a 1 |
| 20 % | «poco probable» | 1 a 4 (4 veces más probable NO aprobar) |

Lo bonito de la regresión logística es esto: **cada hora extra de estudio multiplica los momios por el mismo número**, sin importar desde dónde partas. Si estudiar una hora más multiplica por 2 los momios, pasar de 1 a 2 horas o de 6 a 7 horas hace lo mismo con los **momios** (aunque en **probabilidad** el cambio no sea igual de grande). Veamos el número del modelo.

````python
# Momios = probabilidad / (1 - probabilidad). Cuanto multiplica cada hora extra a los momios?
factor_por_hora = np.exp(modelo.coef_[0, 0])
print(f"Cada hora extra de estudio MULTIPLICA los momios de aprobar por {factor_por_hora:.2f}.\n")

for h in [2, 4, 6, 8]:
    p = modelo.predict_proba([[h]])[0, 1]
    momios = p / (1 - p)
    print(f"{h} horas -> probabilidad {p:.0%} -> momios {momios:.2f} a 1")

print("\nObserva: de una fila a la siguiente (2 horas más), los momios se multiplican por el mismo factor.")
print(f"(Factor por hora = {factor_por_hora:.2f}; por dos horas = {factor_por_hora**2:.2f}.)")
````

## 4. ¿Y si hay más de dos opciones? 🌈

A veces la pregunta no es «¿sí o no?» sino «¿**cuál** de varias?»: ¿qué tipo de flor es?, ¿qué tipo de papa es? La regresión logística se extiende fácilmente: en lugar de una probabilidad, el modelo entrega **un porcentaje para cada opción, y todos suman 100 %** (como el reparto de votos en una elección). La opción con el porcentaje más alto es la que se elige.

Probemos con el famoso conjunto de datos de **flores iris** (tres especies), usando solo dos medidas del pétalo.

````python
# Mas de dos opciones: un porcentaje por cada especie (los porcentajes suman 100 %)
from sklearn.datasets import load_iris

iris = load_iris()
medidas = iris.data[:, 2:4]                                           # largo y ancho del petalo
modelo_flores = LogisticRegression(max_iter=1000).fit(medidas, iris.target)

flores_nuevas = np.array([[1.4, 0.2], [4.3, 1.3], [5.5, 2.0], [4.8, 1.7]])
porcentajes = modelo_flores.predict_proba(flores_nuevas)

tabla = pd.DataFrame(porcentajes * 100, columns=iris.target_names).round(0).astype(int).astype(str) + " %"
tabla.insert(0, "pétalo (largo x ancho, cm)", [f"{a} x {b}" for a, b in flores_nuevas])
tabla["elige"] = [iris.target_names[i] for i in porcentajes.argmax(axis=1)]
print(tabla.to_string(index=False))
print("\nLa última flor (4.8 x 1.7) es un caso difícil, entre dos especies: el modelo lo dice con sus porcentajes, no finge seguridad.")
````

## Resumen en una frase 🎯

> **La regresión logística responde preguntas de sí o no con una probabilidad (curva en S entre 0 % y 100 %); un punto de corte la convierte en decisión, los momios («3 a 1») ayudan a interpretarla y, con más de dos opciones, reparte un porcentaje a cada una.**

### Lo que aprendiste hoy:
- ✅ Para «sí o no» se usa una **curva en S**, no una recta.
- ✅ El modelo entrega **probabilidades**; el **punto de corte** las convierte en decisiones.
- ✅ Subir o bajar el corte **intercambia** falsas alarmas por casos que se escapan.
- ✅ Los **momios** dicen cuántas veces es más probable que pase a que no pase.
- ✅ Con varias opciones, el modelo da un **porcentaje por opción** (suman 100 %).
