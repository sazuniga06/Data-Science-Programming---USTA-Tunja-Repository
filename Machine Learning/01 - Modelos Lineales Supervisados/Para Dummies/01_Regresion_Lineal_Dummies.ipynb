## 1. La línea que atraviesa los puntos 🏠

Cuando alguien quiere comprar apartamento en Tunja, mira **cuánto cuestan** otros apartamentos según su **tamaño**. Si grafica los precios de apartamentos parecidos, nota algo: **a más metros cuadrados, más caro**, aunque no todos siguen la regla exactamente (la ubicación, el piso, el estado... cambian el precio).

La regresión lineal propone una **regla sencilla**:

> **precio = valor base + (lo que cuesta cada metro cuadrado) × (metros cuadrados)**

Es como la tarifa de un taxi: un **arranque fijo** más un **precio por cada kilómetro**. Para encontrar esos dos números, partimos de los apartamentos que ya conocemos.

````python
# Datos inventados: 50 apartamentos (metros cuadrados y precio en millones de pesos)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
metros = rng.uniform(30, 120, 50)
precio = 35 + 2.4 * metros + rng.normal(0, 22, 50)          # regla "real" + diferencias por ubicacion, piso, etc.

apartamentos = pd.DataFrame({"metros_cuadrados": metros.round(0), "precio_millones": precio.round(0)})
print("Primeros 5 apartamentos:")
print(apartamentos.head().to_string(index=False))

plt.figure(figsize=(8, 5))
plt.scatter(metros, precio, color="#f59e0b", edgecolor="white", s=70)
plt.xlabel("Metros cuadrados")
plt.ylabel("Precio (millones de pesos)")
plt.title("Apartamentos: a más metros, más caro (más o menos)")
plt.show()
````

---
## 2. ¿Cómo se elige «la mejor» línea? 🎯

Podríamos dibujar mil líneas distintas entre los puntos. ¿Cuál es la mejor? La que **deja los errores más pequeños**.

- El **error** de cada apartamento es la distancia vertical entre su precio real y el que predice la línea (en el dibujo, los **palitos rojos**).
- Se busca la línea que, **en total, tenga los palitos más cortos**.

La computadora la encuentra sola con una sola instrucción. Veámosla.

````python
# Encontrar la mejor linea con scikit-learn (una instruccion)
from sklearn.linear_model import LinearRegression

modelo = LinearRegression()
modelo.fit(apartamentos[["metros_cuadrados"]], apartamentos["precio_millones"])    # "aprende" la linea con los datos

valor_base = modelo.intercept_
por_metro = modelo.coef_[0]
prediccion = modelo.predict(apartamentos[["metros_cuadrados"]])

print(f"Valor base (arranque): {valor_base:.1f} millones")
print(f"Cada metro cuadrado adicional suma: {por_metro:.2f} millones")

plt.figure(figsize=(8, 5))
plt.scatter(metros, precio, color="#f59e0b", edgecolor="white", s=70, label="Apartamentos reales")
plt.vlines(apartamentos["metros_cuadrados"], prediccion, apartamentos["precio_millones"], color="#dc2626", alpha=0.5, label="Errores (palitos rojos)")
orden = np.argsort(metros)
plt.plot(metros[orden], prediccion[orden], color="#16a34a", lw=3, label="La mejor línea")
plt.xlabel("Metros cuadrados")
plt.ylabel("Precio (millones de pesos)")
plt.title("La línea que deja los palitos rojos más cortos")
plt.legend()
plt.show()
````

---
## 3. Leer los dos números y predecir 🔮

- **El aumento por cada unidad (pendiente):** «por cada metro cuadrado más, el precio sube en promedio X millones». Es lo más útil de leer.
- **El valor base (intercepto):** el precio que predice la línea cuando los metros valen cero. Ojo: un apartamento de 0 m² no existe, así que este número es solo el **punto de partida de la línea**; no hay que interpretarlo al pie de la letra.

Con la línea podemos **predecir** el precio de un apartamento que no estaba en la lista. ¡Pero con una advertencia!

> ⚠️ **No te fíes de la línea lejos de tus datos.** Solo vimos apartamentos entre 30 y 120 m². Si te preguntan por una **finca de 5.000 m²**, la línea te dará un número, pero nadie garantiza que sea razonable: la regla pudo ser distinta ahí. (Como suponer que, si un niño crece 6 cm al año, a los 40 años medirá 3 metros.)

````python
# Predecir apartamentos nuevos (dentro del rango de datos) y una "extrapolacion" peligrosa
nuevos = pd.DataFrame({"metros_cuadrados": [40, 70, 100, 400]})
nuevos["precio_predicho_millones"] = modelo.predict(nuevos[["metros_cuadrados"]]).round(0)
nuevos["comentario"] = ["dentro del rango", "dentro del rango", "dentro del rango", "¡FUERA del rango! no confiar"]
print(nuevos.to_string(index=False))
````

## 4. ¿Qué tan buena es la línea? 📝

Hay dos formas fáciles de revisarla:

1. **El error promedio:** «en promedio, nos equivocamos por X millones». Se interpreta directo, en las mismas unidades del precio.
2. **Una nota de 0 a 1** (se llama *R²*): qué parte de las diferencias de precio entre apartamentos **logra explicar** la línea. Cerca de 1 = explica casi todo; cerca de 0 = casi nada.

Y un truco de detective: **mira los errores uno por uno**. Si la línea es buena, los errores deben verse como una **nube sin forma**. Si hacen una curva o un abanico, la línea se está dejando algo importante por fuera.

````python
# Error promedio, nota de 0 a 1 y revision de los errores
from sklearn.metrics import r2_score

errores = apartamentos["precio_millones"] - prediccion
print(f"Error promedio: nos equivocamos en {np.mean(np.abs(errores)):.1f} millones por apartamento.")
print(f"Nota de 0 a 1 (R²): {r2_score(apartamentos['precio_millones'], prediccion):.2f}  -> la línea explica cerca del {r2_score(apartamentos['precio_millones'], prediccion):.0%} de las diferencias de precio.")

plt.figure(figsize=(8, 4.5))
plt.scatter(prediccion, errores, color="#6366f1", edgecolor="white", s=60)
plt.axhline(0, color="black", linestyle="--")
plt.xlabel("Precio que predice la línea (millones)")
plt.ylabel("Error (real − predicho)")
plt.title("Los errores deben verse como una nube sin forma")
plt.show()
print("Aquí los errores no hacen curvas ni abanicos: la línea es una buena descripción de estos datos.")
````

## 5. ¿Y si hay más de una variable? ➕

Un apartamento no solo tiene tamaño: también habitaciones, piso, distancia al centro... La regresión lineal **múltiple** usa la misma idea con una regla más larga: *un valor base + un aporte por cada característica*. Dos cuidados:

- **Variables repetidas:** si incluyes «metros cuadrados» y «metros cuadrados en pies» (dos formas de decir lo mismo), el modelo no sabe a cuál darle el mérito; es como **dos testigos que dicen exactamente lo mismo**. Los aportes salen inestables, aunque la predicción total siga bien.
- **Datos raros:** un apartamento de lujo con un precio absurdo puede **jalar la línea** hacia él. Conviene mirar los datos antes de modelar.

## Resumen en una frase 🎯

> **La regresión lineal dibuja la línea que deja los errores más pequeños para predecir un número: un valor base más un aumento por cada unidad; sirve dentro del rango de los datos, y se revisa mirando el error promedio y los errores uno por uno.**

### Lo que aprendiste hoy:
- ✅ Predecir un número con una **línea de tendencia**.
- ✅ La «mejor» línea es la que deja **los errores más cortos** en total.
- ✅ El aumento por cada unidad se lee como «por cada X más, el resultado sube en promedio Y».
- ✅ **No extrapolar** lejos de los datos.
- ✅ Revisar el **error promedio** y que los errores parezcan una **nube sin forma**.
