## 1. La Hipótesis de la Variedad y la Distancia Geodésica 🧶

### 1.1 Variedades

Un conjunto de datos con $d$ variables vive en $\mathbb{R}^d$ (la **dimensión ambiente**), pero a menudo sus puntos están restringidos por pocas causas independientes. La **hipótesis de la variedad** (*manifold hypothesis*) afirma que **los datos reales de alta dimensión se concentran cerca de una variedad de dimensión mucho menor** $m\ll d$ (la **dimensión intrínseca**): una superficie «doblada» que **localmente se parece a $\mathbb{R}^m$**. Pensemos en imágenes de un mismo objeto girando: miles de píxeles, pero **un solo grado de libertad** (el ángulo). El *Swiss roll* es el ejemplo de juguete: una hoja bidimensional **enrollada** en $\mathbb{R}^3$ ($d=3,\ m=2$).

### 1.2 Euclídea frente a geodésica

La **distancia geodésica** entre dos puntos de una variedad es la longitud del **camino más corto que no sale de ella**. Dos puntos en capas contiguas del rollo están **cerca en línea recta** (distancia Euclídea pequeña, «atravesando el aire») pero **lejos geodésicamente** (hay que recorrer toda la espiral). Cualquier método basado en distancias Euclídeas —PCA incluido— los confunde con vecinos.

> **Idea de Isomap:** estimar la distancia geodésica con **caminos más cortos en un grafo de vecinos** y luego hallar coordenadas de baja dimensión que **reproduzcan esas distancias**.

### 1.3 Una verdad terreno analítica

`make_swiss_roll` genera $(x,y,z)=(t\cos t,\ y,\ t\sin t)$ con $t\in[1.5\pi,\,4.5\pi]$ y $y\in[0,21]$ (más ruido gaussiano opcional). Es una **superficie desarrollable**: la curva $(t\cos t,\,t\sin t)$ es una espiral de Arquímedes cuya **longitud de arco** es

$$s(t)=\int_0^t\sqrt{1+\tau^2}\,d\tau=\tfrac12\Big(t\sqrt{1+t^2}+\operatorname{asinh}t\Big),$$

y al desenrollarla se obtiene el **rectángulo** $(s(t),\,y)$, **isométrico** al rollo: la distancia geodésica entre dos puntos es **exactamente** la distancia Euclídea entre sus coordenadas $(s,y)$ (el rectángulo es convexo, así que las geodésicas son rectas). Esto nos da una **respuesta correcta** contra la cual juzgar cualquier método. Para la curva S, $t$ ya es la longitud de arco ($\|\mathbf r'(t)\|=1$) y el desenrollado es $(t,\,y)$.

````python
# ============================================================
# Configuracion inicial y datos: Swiss roll con verdad terreno analitica
# ============================================================
# !pip install -q numpy pandas matplotlib seaborn scikit-learn scipy

import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.collections import LineCollection
from scipy.linalg import eigh
from scipy.sparse.csgraph import shortest_path, connected_components
from scipy.spatial import procrustes
from scipy.spatial.distance import pdist, squareform
from scipy.stats import spearmanr
from sklearn.datasets import make_swiss_roll, make_s_curve
from sklearn.decomposition import PCA
from sklearn.manifold import Isomap, TSNE, LocallyLinearEmbedding, trustworthiness
from sklearn.metrics import pairwise_distances
from sklearn.neighbors import kneighbors_graph

warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (9, 5)
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d"]

def longitud_arco(t):
    """Longitud de arco de la espiral (t cos t, t sin t) desde 0 hasta t: s(t) = (t sqrt(1+t^2) + asinh t) / 2."""
    return 0.5 * (t * np.sqrt(1 + t ** 2) + np.arcsinh(t))

n = 1000
X, t = make_swiss_roll(n_samples=n, noise=0.05, random_state=SEED)
U = np.c_[longitud_arco(t), X[:, 1]]                             # coordenadas DESENROLLADAS: verdad terreno (s, y)
D_verdad = squareform(pdist(U))                                  # distancia geodesica verdadera entre todos los pares
D_euclid = squareform(pdist(X))

# Comprobacion: la longitud de arco numerica coincide con la formula (integral por trapecios)
tt = np.linspace(0, 10, 200001)
assert np.isclose(np.trapezoid(np.sqrt(1 + tt ** 2), tt) if hasattr(np, "trapezoid") else np.trapz(np.sqrt(1 + tt ** 2), tt), longitud_arco(10.0), rtol=1e-8)

fig = plt.figure(figsize=(15, 5.5))
ax = fig.add_subplot(1, 2, 1, projection="3d")
ax.scatter(X[:, 0], X[:, 1], X[:, 2], c=t, cmap="Spectral", s=8)
ax.view_init(elev=7, azim=-80); ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")
ax.set_title("Swiss roll en $\\mathbb{R}^3$ (dimensión ambiente 3)", fontweight="bold", fontsize=10)
ax = fig.add_subplot(1, 2, 2)
ax.scatter(U[:, 0], U[:, 1], c=t, cmap="Spectral", s=8)
ax.set_aspect("equal"); ax.set_xlabel("longitud de arco $s(t)$"); ax.set_ylabel("$y$")
ax.set_title("Verdad terreno: el rollo desenrollado es un rectángulo (dimensión intrínseca 2)", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

````python
# ============================================================
# 1b. Euclidea vs geodesica: pares cercanos en linea recta pero lejanos sobre la variedad
# ============================================================
k = 10
G = kneighbors_graph(X, k, mode="distance"); G = G.maximum(G.T)  # grafo de k vecinos (simetrizado), pesos = longitud euclidiana de la arista
D_grafo, pred = shortest_path(G, method="D", directed=False, return_predecessors=True)    # caminos mas cortos (Dijkstra)
assert not np.isinf(D_grafo).any()                              # el grafo es conexo

fuera = ~np.eye(n, dtype=bool)
err_euclid = (np.abs(D_euclid - D_verdad)[fuera] / D_verdad[fuera])
err_grafo = (np.abs(D_grafo - D_verdad)[fuera] / D_verdad[fuera])
print(f"Error relativo medio respecto de la geodésica verdadera: distancia Euclídea = {100 * err_euclid.mean():.1f} %, "
      f"distancia por caminos en el grafo (k={k}) = {100 * err_grafo.mean():.1f} %")
print(f"Error mediano: Euclídea = {100 * np.median(err_euclid):.1f} %, grafo = {100 * np.median(err_grafo):.2f} %")
assert err_grafo.mean() < 0.25 * err_euclid.mean() and err_grafo.mean() < 0.06       # el grafo aproxima la geodesica mucho mejor que la recta

# Par "engañoso": cercano en linea recta, muy lejano sobre la variedad
cand = (D_euclid > 2.0) & (D_euclid < 8.0)
razon = np.where(cand, D_verdad / np.maximum(D_euclid, 1e-12), 0)
i0, j0 = np.unravel_index(razon.argmax(), razon.shape)
camino = [j0]
while camino[-1] != i0:
    camino.append(pred[i0, camino[-1]])
camino = np.array(camino)
print(f"Par engañoso: distancia Euclídea = {D_euclid[i0, j0]:.1f}, geodésica verdadera = {D_verdad[i0, j0]:.1f}, camino en el grafo = {D_grafo[i0, j0]:.1f}")

fig = plt.figure(figsize=(15, 5.5))
ax = fig.add_subplot(1, 2, 1)                                   # seccion transversal (x, z) del rollo: la espiral
ax.scatter(X[:, 0], X[:, 2], c="lightgray", s=8)
ax.plot(*X[[i0, j0]][:, [0, 2]].T, color=PALETA[0], lw=3, ls="--", label=f"recta (Euclídea = {D_euclid[i0, j0]:.1f})")
ax.plot(*X[camino][:, [0, 2]].T, color=PALETA[1], lw=2.5, label=f"camino geodésico en el grafo ({D_grafo[i0, j0]:.1f})")
ax.scatter(*X[[i0, j0]][:, [0, 2]].T, color="black", s=60, zorder=5)
ax.set_aspect("equal"); ax.set_xlabel("x"); ax.set_ylabel("z"); ax.legend(fontsize=8, loc="upper left")
ax.set_title("Sección (x, z) del rollo: cercanos en línea recta, lejanos sobre la variedad", fontweight="bold", fontsize=10)
ax = fig.add_subplot(1, 2, 2)
lim = D_verdad.max()
sub = np.random.RandomState(SEED).choice(len(fuera.nonzero()[0]), 6000, replace=False)
ax.scatter(D_verdad[fuera][sub], D_euclid[fuera][sub], s=3, alpha=0.3, color=PALETA[0], label="Euclídea")
ax.scatter(D_verdad[fuera][sub], D_grafo[fuera][sub], s=3, alpha=0.3, color=PALETA[1], label="camino en el grafo (k=10)")
ax.plot([0, lim], [0, lim], "k--", lw=1); ax.set_xlabel("Distancia geodésica verdadera"); ax.set_ylabel("Distancia estimada")
ax.set_title("El grafo estima la geodésica; la recta la subestima", fontweight="bold", fontsize=10)
ax.legend(markerscale=4, fontsize=8)
plt.tight_layout(); plt.show()
````

## 2. Isomap: Grafo, Caminos Más Cortos y MDS Clásico 🛠️

### 2.1 El algoritmo

> 1. **Grafo de vecinos.** Conectar $\mathbf x_i$ y $\mathbf x_j$ si uno está entre los $k$ vecinos más cercanos del otro (o a menos de $\varepsilon$), con **peso** $\|\mathbf x_i-\mathbf x_j\|$.
> 2. **Distancias geodésicas.** $D^G_{ij}$ = longitud del **camino más corto** entre $i$ y $j$ en el grafo (Dijkstra desde cada nodo, $O(n\,(E+n\log n))$, o **Floyd–Warshall**, $O(n^3)$).
> 3. **MDS clásico** sobre $D^G$: hallar $\mathbf y_i\in\mathbb{R}^m$ cuyas distancias Euclídeas reproduzcan $D^G$.

Con $n$ suficientemente grande y muestreo denso, $D^G\to$ distancia geodésica verdadera (Bernstein et al., 2000), para variedades **isométricas a un subconjunto convexo** de $\mathbb{R}^m$ (el caso del rollo).

### 2.2 MDS clásico (Torgerson–Gower)

Si $\mathbf y_i$ están **centrados** ($\sum\mathbf y_i=\mathbf 0$) y $d_{ij}^2=\|\mathbf y_i-\mathbf y_j\|^2=\|\mathbf y_i\|^2+\|\mathbf y_j\|^2-2\mathbf y_i^\top\mathbf y_j$, el **doble centrado** con $J=I-\tfrac1n\mathbf 1\mathbf 1^\top$ elimina los dos primeros términos:

$$B=-\tfrac12\,J\,D^{(2)}\,J=YY^\top\qquad(D^{(2)}_{ij}=d_{ij}^2),$$

es decir, $B$ es la **matriz de Gram** de las coordenadas buscadas. Si $B=V\Lambda V^\top$ (autodescomposición, con $\lambda_1\ge\lambda_2\ge\dots$), las coordenadas son

$$Y=V_m\,\Lambda_m^{1/2}\qquad(\text{las }m\text{ columnas de mayor autovalor}),$$

únicas **salvo rotaciones, reflexiones y traslaciones** (el signo de cada columna es arbitrario, como en PCA). Dos hechos importantes:

- **PCA es MDS sobre distancias Euclídeas.** Si $D$ son las distancias Euclídeas de $X$, entonces $B=X_cX_c^\top$ y sus autovalores no nulos son los de $X_c^\top X_c$ ($=(n-1)\lambda_j$ de PCA): las coordenadas coinciden con los **puntajes de PCA** (salvo signo). Lo verificamos abajo.
- Con $D^G$ (que **no es** Euclídea) $B$ puede tener **autovalores negativos**; MDS conserva los $m$ **mayores positivos**. `scikit-learn` implementa Isomap como un **Kernel PCA** con el núcleo precalculado $K=-\tfrac12(D^G)^{2}$, que es exactamente lo anterior.

La **dimensión intrínseca** se puede estimar con el **espectro** de $B$ (o con la *varianza residual* $1-R^2(D^G,D_Y)$ en función de $m$): hay un **codo** en $m$.

**Sobre el algoritmo de caminos más cortos.** Programamos **Floyd–Warshall** vectorizado (una línea por nodo intermedio, $O(n^3)$) y lo **verificamos contra `scipy.sparse.csgraph.shortest_path`** (Floyd–Warshall y Dijkstra); para $n=1000$ usamos **Dijkstra de SciPy**, mucho más eficiente en grafos dispersos.

````python
# ============================================================
# 2. Isomap DESDE CERO: grafo kNN + Floyd-Warshall propio (verificado vs SciPy) + MDS clasico
# ============================================================
def grafo_knn(X, k):
    """Grafo de k vecinos simetrizado (una arista si i esta en los k vecinos de j O j en los de i), peso = distancia euclidiana."""
    G = kneighbors_graph(X, k, mode="distance")
    return G.maximum(G.T)

def floyd_warshall(G):
    """Floyd-Warshall vectorizado, O(n^3): D[i,j] = min(D[i,j], D[i,m] + D[m,j]) para cada nodo intermedio m."""
    D = G.toarray().astype(float)
    D[D == 0] = np.inf                                           # sin arista = infinito (los pesos reales son > 0)
    np.fill_diagonal(D, 0.0)
    for m in range(len(D)):
        np.minimum(D, D[:, m, None] + D[None, m, :], out=D)
    return D

def mds_clasico(D, m=2):
    """MDS clasico: B = -1/2 J D^2 J (doble centrado), Y = V_m sqrt(Lambda_m). Devuelve coordenadas y los m mayores autovalores."""
    D2 = D ** 2
    B = -0.5 * (D2 - D2.mean(axis=0, keepdims=True) - D2.mean(axis=1, keepdims=True) + D2.mean())   # doble centrado sin formar J
    lam, V = eigh(B, subset_by_index=[len(B) - m, len(B) - 1])   # solo los m MAYORES autovalores (mucho mas rapido que la descomposicion completa)
    orden = np.argsort(lam)[::-1]
    lam, V = lam[orden], V[:, orden]
    return V * np.sqrt(np.clip(lam, 0, None)), lam

def isomap_desde_cero(X, k=10, m=2):
    """Isomap: grafo kNN -> caminos mas cortos (Dijkstra de SciPy) -> MDS clasico. Devuelve Y, matriz geodesica y autovalores."""
    D = shortest_path(grafo_knn(X, k), method="D", directed=False)
    D = np.minimum(D, D.T)                                       # elimina asimetrias de redondeo (1e-14) del recorrido de Dijkstra
    if np.isinf(D).any():
        raise ValueError("Grafo desconectado: aumenta k (ver seccion 3).")
    Y, lam = mds_clasico(D, m)
    return Y, D, lam

def alinear_columnas(A, B):
    """Cambia el signo de cada COLUMNA de A para que apunte como la de B (ambiguedad de signo de los autovectores)."""
    return A * np.sign((A * B).sum(axis=0))[None, :]

# (a) Floyd-Warshall propio vs SciPy (Floyd-Warshall y Dijkstra) sobre un subconjunto de 400 puntos
Gp = grafo_knn(X[:400], 8)
D_fw = floyd_warshall(Gp)
assert np.allclose(D_fw, shortest_path(Gp, method="FW", directed=False))       # igual que scipy (FW)
assert np.allclose(D_fw, shortest_path(Gp, method="D", directed=False))        # igual que scipy (Dijkstra)
print("Floyd–Warshall propio == scipy.sparse.csgraph.shortest_path (método 'FW' y método 'D') en 400 puntos.")

# (b) MDS clasico con doble centrado explicito (J) == version sin formar J
Dp = pairwise_distances(X[:300]); J = np.eye(300) - np.ones((300, 300)) / 300
assert np.allclose(-0.5 * J @ Dp ** 2 @ J, -0.5 * (Dp ** 2 - (Dp ** 2).mean(0) - (Dp ** 2).mean(1)[:, None] + (Dp ** 2).mean()))

# (c) PCA == MDS clasico sobre distancias EUCLIDEAS
Y_mds, lam_mds = mds_clasico(pairwise_distances(X), 3)
pca3 = PCA(3, svd_solver="full").fit(X)
assert np.allclose(alinear_columnas(Y_mds, pca3.transform(X)), pca3.transform(X), atol=1e-8)   # coordenadas = puntajes de PCA
assert np.allclose(lam_mds, (len(X) - 1) * pca3.explained_variance_)                           # autovalores de B = (n-1) * lambda de PCA
print("MDS clásico sobre distancias Euclídeas == puntajes de PCA (salvo signo); autovalores de B = (n-1)·λ_PCA.")
````

````python
# ============================================================
# 2b. Verificacion contra sklearn.manifold.Isomap y recuperacion del desenrollado verdadero
# ============================================================
Y_mio, D_geo, lam_iso = isomap_desde_cero(X, k=10, m=2)
iso = Isomap(n_neighbors=10, n_components=2, eigen_solver="dense", path_method="D").fit(X)    # 'dense': autodescomposicion exacta
Y_sk = iso.embedding_

assert np.allclose(alinear_columnas(Y_mio, Y_sk), Y_sk, atol=1e-6)                           # embedding igual (salvo signo de cada eje)
assert np.allclose(pdist(Y_mio), pdist(Y_sk), atol=1e-8)                                      # y sus matrices de distancias son identicas
assert np.allclose(D_geo, iso.dist_matrix_, atol=1e-8)                                        # misma matriz geodesica
print(f"Isomap propio == sklearn.manifold.Isomap: max |diferencia de coordenadas (salvo signo)| = {np.abs(alinear_columnas(Y_mio, Y_sk) - Y_sk).max():.1e}; "
      f"max |diferencia de distancias| = {np.abs(pdist(Y_mio) - pdist(Y_sk)).max():.1e}")

# Procrustes contra la verdad terreno (0 = identico salvo traslacion, escala, rotacion y reflexion; 1 = sin relacion)
Y_pca = PCA(2).fit_transform(X)
disp_iso, disp_pca = procrustes(U, Y_mio)[2], procrustes(U, Y_pca)[2]
print(f"Disparidad de Procrustes con el desenrollado verdadero: Isomap = {disp_iso:.4f} | PCA = {disp_pca:.4f}")
assert disp_iso < 0.01 and disp_pca > 0.5

# Varianza residual (Tenenbaum et al.): 1 - R^2 entre la geodesica y las distancias del embedding, segun la dimension m
Ym, lam_all = mds_clasico(D_geo, 8)
res_iso = [1 - np.corrcoef(squareform(D_geo), pdist(Ym[:, :m]))[0, 1] ** 2 for m in range(1, 9)]
res_pca = [1 - np.corrcoef(squareform(D_geo), pdist(PCA(m).fit_transform(X)))[0, 1] ** 2 for m in range(1, 4)]

fig, axes = plt.subplots(1, 3, figsize=(17, 4.8))
axes[0].scatter(*Y_pca.T, c=t, cmap="Spectral", s=8); axes[0].set_title(f"PCA 2D (Procrustes = {disp_pca:.2f}): capas mezcladas", fontweight="bold", fontsize=10)
axes[1].scatter(*Y_mio.T, c=t, cmap="Spectral", s=8); axes[1].set_aspect("equal")
axes[1].set_title(f"Isomap propio (== sklearn), Procrustes = {disp_iso:.4f}", fontweight="bold", fontsize=10)
for ax in axes[:2]: ax.set_xlabel("Componente 1"); ax.set_ylabel("Componente 2")
axes[2].plot(range(1, 9), res_iso, "o-", color=PALETA[0], lw=2, label="Isomap (dimensión del embedding)")
axes[2].plot(range(1, 4), res_pca, "s--", color=PALETA[1], lw=2, label="PCA (R² con la geodésica del grafo)")
axes[2].set_xlabel("Dimensión $m$"); axes[2].set_ylabel("Varianza residual $1-R^2$"); axes[2].legend(fontsize=8)
axes[2].set_title("Isomap: el codo en $m=2$ revela la dimensión intrínseca", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
print("Autovalores de B (mayores 5):", np.round(lam_all[:5], 1), "→ dos dominan; el resto es mucho menor.")
````

## 3. Elegir $k$: Grafos Desconectados y Cortocircuitos 🔌

$k$ es el hiperparámetro crítico de Isomap, con dos modos de fallo opuestos:

- **$k$ demasiado pequeño → grafo desconectado.** Si el grafo tiene varias componentes conexas, la distancia entre componentes es $+\infty$ y MDS no está definido. las versiones recientes de `scikit-learn` **conectan las componentes** artificialmente con un aviso (*«the number of connected components of the neighbors graph is...»*); nosotros lo **detectamos** con `connected_components` y lo tratamos como un error.
- **$k$ demasiado grande → cortocircuito (*short-circuit*).** Si algún vecino es de **otra capa** (más cerca en línea recta que en la variedad), la arista **atraviesa el vacío** y crea un **atajo** que reduce las distancias geodésicas de **muchos** pares; el mapa se «pliega» (Balasubramanian & Schwartz, 2002). **Es el fallo característico de Isomap**, y empeora con **ruido** y con **muestreo escaso**.

Gracias a la verdad terreno podemos **medir** el cortocircuito: llamamos **aristas malas** a las del grafo cuya longitud Euclídea es **menos de un tercio** de su distancia geodésica verdadera. Barremos $k$ y medimos conectividad, aristas malas, varianza residual, Procrustes y *trustworthiness*.

````python
# ============================================================
# 3. Barrido de k: desconexion (k pequeno) y cortocircuito (k grande), medidos contra la verdad terreno
# ============================================================
ks = [3, 5, 10, 15, 20, 30, 60]
filas, Ys = [], {}
for k in ks:
    Gk = grafo_knn(X, k)
    ncomp, etiq = connected_components(Gk, directed=False)
    fila = {"k": k, "componentes conexas": ncomp}
    ii, jj, largo = Gk.tocoo().row, Gk.tocoo().col, Gk.tocoo().data
    fila["aristas malas (cortocircuitos)"] = int(((D_verdad[ii, jj] > 3 * largo).sum()) // 2)
    if ncomp == 1:
        Y_k, D_k, _ = isomap_desde_cero(X, k, 2)
        Ys[k] = Y_k
        fila["varianza residual (m=2)"] = 1 - np.corrcoef(squareform(D_k), pdist(Y_k))[0, 1] ** 2
        fila["Procrustes vs verdad"] = procrustes(U, Y_k)[2]
        fila["trustworthiness (k=10)"] = trustworthiness(X, Y_k, n_neighbors=10)
    filas.append(fila)
tabla = pd.DataFrame(filas).set_index("k")
display(tabla.round(4))

# Aserciones: el comportamiento cualitativo que predice la teoria
assert tabla.loc[3, "componentes conexas"] > 1                                   # k muy pequeno: grafo desconectado
assert tabla.loc[10, "aristas malas (cortocircuitos)"] == 0 and tabla.loc[10, "Procrustes vs verdad"] < 0.01    # k razonable: sin atajos y desenrollado fiel
assert tabla.loc[30, "aristas malas (cortocircuitos)"] > 0 and tabla.loc[30, "Procrustes vs verdad"] > 0.2      # k grande: atajos y mapa plegado

# Comportamiento de sklearn con grafo desconectado (k = 3)
with warnings.catch_warnings(record=True) as avisos:
    warnings.simplefilter("always")
    try:
        emb3 = Isomap(n_neighbors=3, n_components=2).fit_transform(X)
        print(f"sklearn con k=3 NO falla: completa el grafo y devuelve un embedding {emb3.shape}; avisos emitidos: {len(avisos)}")
        if avisos: print("  Primer aviso:", str(avisos[0].message)[:160])
    except Exception as e:                                                         # versiones antiguas pueden lanzar error
        print("sklearn con k=3 lanza:", type(e).__name__, str(e)[:120])

fig, axes = plt.subplots(2, 4, figsize=(20, 9))
for ax, k in zip(axes[0], [5, 10, 20, 60]):
    ax.scatter(*Ys[k].T, c=t, cmap="Spectral", s=8); ax.set_aspect("equal")
    ax.set_title(f"k = {k}: Procrustes = {tabla.loc[k, 'Procrustes vs verdad']:.2f}, {tabla.loc[k, 'aristas malas (cortocircuitos)']} aristas malas", fontweight="bold", fontsize=10)
ax = axes[1, 0]
ax.plot(tabla.index[tabla["Procrustes vs verdad"].notna()], tabla["Procrustes vs verdad"].dropna(), "o-", color=PALETA[1], label="Procrustes vs verdad")
ax.plot(tabla.index[tabla["Procrustes vs verdad"].notna()], 1 - tabla["trustworthiness (k=10)"].dropna(), "s--", color=PALETA[0], label="1 - trustworthiness")
ax.axvspan(2.5, 4, color="gray", alpha=0.2, label="grafo desconectado")
ax.set_xlabel("k (vecinos)"); ax.set_ylabel("Error (menor es mejor)"); ax.legend(fontsize=8)
ax.set_title("Error de Isomap según k: una ventana útil", fontweight="bold", fontsize=10)
for ax, k, tit in zip(axes[1, 1:3], [20, 30], ["aristas malas, k = 20", "aristas malas, k = 30"]):
    Gk = grafo_knn(X, k).tocoo(); ii, jj, largo = Gk.row, Gk.col, Gk.data
    malas = (D_verdad[ii, jj] > 3 * largo) & (ii < jj)
    ax.scatter(U[:, 0], U[:, 1], c=t, cmap="Spectral", s=5, alpha=0.6)
    ax.add_collection(LineCollection(np.stack([U[ii[malas]], U[jj[malas]]], axis=1), colors="black", linewidths=0.4, alpha=0.6))     # aristas malas
    ax.set_aspect("equal"); ax.set_title(f"{tit}: atajos entre capas (plano desenrollado)", fontweight="bold", fontsize=9)
ncomp3, etiq3 = connected_components(grafo_knn(X, 3), directed=False)
axes[1, 3].scatter(U[:, 0], U[:, 1], c=etiq3, cmap="tab10", s=8); axes[1, 3].set_aspect("equal")
axes[1, 3].set_title(f"k = 3: {ncomp3} componentes conexas (colores)", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

## 4. Comparación: PCA frente a Isomap frente a t-SNE 🥊

### 4.1 Cómo medir una proyección

Con $k$ vecinos, denotamos $r_{ij}$ el **rango** de $j$ entre los vecinos de $i$ en el espacio original y $\hat r_{ij}$ en el mapa (1 = el más cercano). Sean $U_i^{(k)}$ los puntos que entran en la vecindad de $i$ **en el mapa pero no en el original** («intrusos») y $V_i^{(k)}$ los que están en la vecindad **original** pero no en el mapa («expulsados»). Venna & Kaski (2006):

$$\boxed{T(k)=1-\frac{2}{nk(2n-3k-1)}\sum_{i=1}^n\sum_{j\in U_i^{(k)}}\big(r_{ij}-k\big)},\qquad \boxed{C(k)=1-\frac{2}{nk(2n-3k-1)}\sum_{i=1}^n\sum_{j\in V_i^{(k)}}\big(\hat r_{ij}-k\big)}$$

- ***Trustworthiness* $T$:** ¿pueden **confiarse** las vecindades del mapa? Penaliza los **intrusos** (puntos lejanos que el mapa acerca), ponderados por **cuán lejos** estaban.
- ***Continuity* $C$:** ¿el mapa **conserva** las vecindades originales? Penaliza los **expulsados** (vecinos reales que el mapa separa). Es **$T$ con los papeles intercambiados**: $C(X,Y)=T(Y,X)$.
- Ambas son **locales** (miran $k$ vecinos). Para medir **estructura global** usamos, como aquí tenemos verdad terreno, la **correlación de rangos de Spearman** entre las distancias del mapa y las distancias geodésicas verdaderas, y la disparidad de **Procrustes** con el desenrollado.

Implementamos $T$ y $C$ desde cero y verificamos $T$ contra `sklearn.manifold.trustworthiness`.

````python
# ============================================================
# 4. Trustworthiness y continuity DESDE CERO (verificadas contra sklearn)
# ============================================================
def matriz_rangos(D):
    """R[i, j] = posicion de j en la lista de vecinos de i (1 = el mas cercano); el propio i queda en la ultima posicion (n)."""
    n = len(D)
    D = D.copy(); np.fill_diagonal(D, np.inf)
    orden = np.argsort(D, axis=1)
    R = np.empty((n, n), dtype=int)
    R[np.arange(n)[:, None], orden] = np.arange(1, n + 1)[None, :]
    return R

def trustworthiness_propia(D_alta, D_baja, k):
    """T(k): penaliza los 'intrusos' (vecinos en el mapa que NO lo eran en el espacio original), ponderados por su rango original."""
    n = len(D_alta)
    R_alta, R_baja = matriz_rangos(D_alta), matriz_rangos(D_baja)
    intrusos = (R_baja <= k) & (R_alta > k)
    return 1 - 2.0 / (n * k * (2 * n - 3 * k - 1)) * ((R_alta - k) * intrusos).sum()

def continuity_propia(D_alta, D_baja, k):
    """C(k) = T con los papeles intercambiados: penaliza los vecinos originales que el mapa separa."""
    return trustworthiness_propia(D_baja, D_alta, k)

D_a = pairwise_distances(X)
for kk in [5, 10, 30]:
    T_prop, T_sk = trustworthiness_propia(D_a, pairwise_distances(Y_pca), kk), trustworthiness(X, Y_pca, n_neighbors=kk)
    assert np.isclose(T_prop, T_sk, atol=1e-12), (kk, T_prop, T_sk)
    assert np.isclose(continuity_propia(D_a, pairwise_distances(Y_pca), kk), trustworthiness(Y_pca, X, n_neighbors=kk), atol=1e-12)    # C(X,Y) = T(Y,X) en sklearn
print("Trustworthiness propia == sklearn.manifold.trustworthiness (k = 5, 10, 30); continuity = trustworthiness con papeles intercambiados.")
````

````python
# ============================================================
# 4b. Swiss roll y curva S: PCA, Isomap, t-SNE (y LLE como referencia) con metricas locales y globales
# ============================================================
def desenrollado_s(t_, X_):
    return np.c_[t_, X_[:, 1]]                                   # curva S: t ya es la longitud de arco (|r'(t)| = 1)

Xs, ts = make_s_curve(n_samples=n, noise=0.05, random_state=SEED)
conjuntos = {"Swiss roll": (X, U, t), "Curva S": (Xs, desenrollado_s(ts, Xs), ts)}
metodos = {
    "PCA": lambda A: PCA(2).fit_transform(A),
    "Isomap (k=10)": lambda A: Isomap(n_neighbors=10, n_components=2).fit_transform(A),
    "t-SNE (perp. 30)": lambda A: (TSNE(n_components=2, perplexity=30, init="pca", learning_rate="auto", max_iter=500, random_state=SEED) if "max_iter" in TSNE().get_params()
                                   else TSNE(n_components=2, perplexity=30, init="pca", learning_rate="auto", n_iter=500, random_state=SEED)).fit_transform(A),
    "LLE (k=12)": lambda A: LocallyLinearEmbedding(n_neighbors=12, n_components=2, random_state=SEED).fit_transform(A),
}
filas, mapas = [], {}
t0 = time.time()
for nom_c, (A, U_, col) in conjuntos.items():
    D_alta, D_geo_verdad = pairwise_distances(A), pdist(U_)
    for nom_m, f in metodos.items():
        Ym = f(A); mapas[(nom_c, nom_m)] = Ym
        D_baja = pairwise_distances(Ym)
        filas.append({"conjunto": nom_c, "método": nom_m,
                      "trustworthiness (k=10)": trustworthiness_propia(D_alta, D_baja, 10),
                      "continuity (k=10)": continuity_propia(D_alta, D_baja, 10),
                      "Spearman con geodésica real (global)": spearmanr(pdist(Ym), D_geo_verdad)[0],
                      "Procrustes vs verdad (↓)": procrustes(U_, Ym)[2]})
print(f"Cálculo de las 8 proyecciones y sus métricas: {time.time() - t0:.1f} s")
res = pd.DataFrame(filas).set_index(["conjunto", "método"])
display(res.round(3))

# Aserciones: lo que establece la teoria
r1 = res.loc["Swiss roll"]
assert r1.loc["Isomap (k=10)", "Procrustes vs verdad (↓)"] < 0.01 and r1.loc["PCA", "Procrustes vs verdad (↓)"] > 0.5
assert r1.loc["Isomap (k=10)", "Spearman con geodésica real (global)"] > 0.99                                 # Isomap: estructura GLOBAL fiel
assert r1.loc["t-SNE (perp. 30)", "trustworthiness (k=10)"] > r1.loc["PCA", "trustworthiness (k=10)"] + 0.02   # t-SNE: vecindades LOCALES mucho mejores que PCA
assert r1.loc["t-SNE (perp. 30)", "Spearman con geodésica real (global)"] < r1.loc["Isomap (k=10)", "Spearman con geodésica real (global)"]   # ... pero peor global

fig, axes = plt.subplots(2, 4, figsize=(20, 9))
for fila_ax, (nom_c, (A, U_, col)) in zip(axes, conjuntos.items()):
    for ax, nom_m in zip(fila_ax, metodos):
        Ym = mapas[(nom_c, nom_m)]
        ax.scatter(*Ym.T, c=col, cmap="Spectral", s=8)
        r_ = res.loc[(nom_c, nom_m)]
        ax.set_title(f"{nom_c} · {nom_m}\nT = {r_.iloc[0]:.3f}, C = {r_.iloc[1]:.3f}, Spearman global = {r_.iloc[2]:.2f}", fontweight="bold", fontsize=9)
        ax.set_xticks([]); ax.set_yticks([])
plt.tight_layout(); plt.show()

# Datos nuevos: PCA e Isomap tienen transform; t-SNE no
tr, te = np.arange(800), np.arange(800, n)
iso_tr, pca_tr = Isomap(n_neighbors=10).fit(X[tr]), PCA(2).fit(X[tr])
print("\nDatos nuevos (200 puntos no vistos): trustworthiness de la proyección con transform():")
print(f"  PCA    = {trustworthiness(X[te], pca_tr.transform(X[te]), n_neighbors=10):.3f}")
print(f"  Isomap = {trustworthiness(X[te], iso_tr.transform(X[te]), n_neighbors=10):.3f}")
print("  t-SNE  = no tiene transform:", not hasattr(TSNE(), "transform"))
````

## 5. Tabla Comparativa Final y Guía de Elección 🧭

| Criterio | **PCA** | **Isomap** | **t-SNE** |
|---|---|---|---|
| **Tipo** | **Lineal** (proyección ortogonal) | **No lineal**, basada en variedad (geodésicas + MDS) | **No lineal**, probabilística (vecindades) |
| **Qué preserva** | Varianza global / distancias Euclídeas grandes | **Distancias geodésicas** (estructura **global**) | **Vecindades locales** (probabilidades $p_{ij}$) |
| **Estructura global** | Sí (pero solo lineal) | **Sí**, si la variedad es suave y convexa | **No fiable** (tamaños y distancias entre grupos sin significado) |
| **Estructura local** | Débil si la estructura es curva | Buena si $k$ es adecuado | **Excelente** (separa grupos) |
| **Determinista** | Sí (salvo signo) | Sí (salvo signo) | **No**: depende de la semilla, la inicialización y la perplejidad |
| **Datos nuevos (`transform`)** | **Sí** (exacto) | Sí (aproximado, vía vecinos del grafo) | **No** (`fit_transform` solamente) |
| **Costo** | $O(nd\min(n,d))$; muy rápido (hay versión aleatorizada e incremental) | $O(n^2\log n)$–$O(n^3)$ (geodésicas) + autodescomposición $n\times n$; memoria $O(n^2)$ | $O(n^2)$ exacto; $O(n\log n)$ con Barnes–Hut |
| **Hiperparámetros** | $m$ (número de componentes) | $k$ (vecinos) y $m$ | Perplejidad, iteraciones, tasa de aprendizaje, inicialización |
| **Fallos típicos** | Estructura curva; sensible a la escala; varianza $\neq$ relevancia | **Cortocircuitos** (ruido, $k$ grande), grafo desconectado, variedades no convexas o con agujeros | Estructura falsa con perplejidad baja; global engañoso; no sirve para preprocesar |
| **Cuándo usarla** | *Baseline*, compresión, decorrelación, preprocesamiento, ruido | Variedad suave y bien muestreada, cuando interesa la **geometría global** (rotación, pose, trayectoria) | **Visualizar** grupos en datos complejos; exploración |

**Otras técnicas.** ***LLE*** (*Locally Linear Embedding*; Roweis & Saul, 2000): reconstruye cada punto como combinación lineal de sus vecinos y busca coordenadas de baja dimensión que **conserven esos pesos**; es un problema de autovalores disperso (Géron, Cap. 7; Vasques, §2.5.1.4); más robusto al ruido que Isomap en algunos casos, pero sin garantía de preservar distancias (en la tabla anterior, su Spearman global es intermedio). ***UMAP*** (McInnes et al., 2018): grafo de vecinos difuso y optimización tipo t-SNE, **más rápido**, mejor estructura global y con `transform`; no está en `scikit-learn` (paquete `umap-learn`). Otras opciones: *MDS* métrico/no métrico, *Spectral Embedding* (Laplacian eigenmaps), *Kernel PCA*, **autoencoders**.

**Receta práctica:** (1) empieza con **PCA** (rápido, interpretable, base de comparación) y mira su espectro; (2) si hay estructura curva y quieres geometría global, prueba **Isomap** variando $k$ y vigilando los cortocircuitos; (3) para **visualizar** grupos usa **t-SNE o UMAP** variando perplejidad y semillas, **sin sobreinterpretar** tamaños ni distancias; (4) **valida** con *trustworthiness* y *continuity* (y con la tarea final, si la hay), no con la estética del mapa.

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Hipótesis de la variedad** | Los datos de alta dimensión se concentran cerca de una variedad de **dimensión intrínseca** baja; la distancia **geodésica** (sobre la variedad) difiere de la Euclídea |
| **Isomap** | (1) grafo de $k$ vecinos con pesos Euclídeos, (2) **caminos más cortos** como geodésicas, (3) **MDS clásico** sobre ellas |
| **MDS clásico** | $B=-\tfrac12JD^{(2)}J$ (Gram), $Y=V_m\Lambda_m^{1/2}$; **PCA = MDS con distancias Euclídeas**; Isomap = Kernel PCA con $K=-\tfrac12(D^G)^2$ |
| **Verificación** | Floyd–Warshall propio = SciPy; Isomap propio = `sklearn.manifold.Isomap` (salvo signo); desenrollado verificado con **Procrustes** contra la verdad analítica |
| **Elegir $k$** | $k$ pequeño → **grafo desconectado**; $k$ grande → **cortocircuito** (atajos entre capas, mapa plegado); hay una **ventana útil**; el ruido la estrecha |
| **Medir** | ***Trustworthiness*** (confianza en las vecindades del mapa) y ***continuity*** (conservación de las originales); implementadas desde cero = `sklearn` |
| **PCA vs Isomap vs t-SNE** | PCA: lineal, global, rápido, con `transform`. Isomap: no lineal, **global**, determinista, frágil a atajos. t-SNE: **local**, visual, no determinista, sin `transform` |
| **Más allá** | **LLE** (pesos de reconstrucción locales), **UMAP** (rápido, con `transform`), *Spectral Embedding*, autoencoders |
