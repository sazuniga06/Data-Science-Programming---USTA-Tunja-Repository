## Preparación 🧰

Ejecuta esta celda tal como está. El pasillo tiene **6 casillas** numeradas de 0 a 5. La tortuga empieza en la casilla `0` y la lechuga está en la casilla `5` (**META**). Acciones: `0` = izquierda, `1` = derecha. Cada paso cuesta **−1** (energía) y comerse la lechuga da **+10**. Chocar con la pared izquierda te deja en la misma casilla.

- `paso(s, a)`: pasillo normal (la tortuga siempre hace lo que pide) → devuelve `(s_nuevo, recompensa)`.
- `paso_resbaloso(s, a, rng)`: pasillo mojado → con probabilidad 0.8 hace lo que pide y con 0.2 se **queda quieta**.

```python
import numpy as np

META = 5
GAMMA = 0.9          # la "paciencia" de la tortuga


def listo(*valores):
    """True si ninguna variable sigue siendo None o ... (sin completar)."""
    return all(v is not None and v is not Ellipsis for v in valores)


def paso(s, a):
    s2 = min(max(s + (1 if a == 1 else -1), 0), META)
    return s2, (10 if s2 == META else -1)


def paso_resbaloso(s, a, rng):
    if rng.random() < 0.2:          # resbala: se queda quieta
        return s, -1
    return paso(s, a)


print("Pasillo de", META + 1, "casillas | primer paso a la derecha desde 0:", paso(0, 1))
```

$$
