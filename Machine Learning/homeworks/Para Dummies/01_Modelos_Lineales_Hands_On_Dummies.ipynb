## 🎈 ¿Qué vamos a hacer en este taller?

Dos problemitas cotidianos, dos modelos lineales:

```python
# ============================================================
# Preparación (no modificar): imports y datos inventados
# ============================================================
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression

# --- Puesto de jugos: 120 días ---
rng = np.random.default_rng(5)
n = 120
temperatura_c = rng.uniform(8, 26, n)
es_festivo = rng.binomial(1, 0.3, n)
vasos = 20 + 3.5 * (temperatura_c - 8) + 25 * es_festivo + rng.normal(0, 6, n)
jugos = pd.DataFrame({"temperatura_c": temperatura_c, "es_festivo": es_festivo, "vasos_vendidos": vasos})
Xj_train, Xj_test, yj_train, yj_test = train_test_split(jugos[["temperatura_c", "es_festivo"]], jugos["vasos_vendidos"],
                                                       test_size=0.25, random_state=0)

# --- Estudiantes: 240 personas ---
rng = np.random.default_rng(8)
n = 240
horas = rng.uniform(0, 10, n)
asistencia = rng.uniform(50, 100, n)
aprobo = rng.binomial(1, 1 / (1 + np.exp(-(0.7 * (horas - 5) + 0.06 * (asistencia - 75)))))
estudiantes = pd.DataFrame({"horas_estudio": horas, "asistencia_pct": asistencia, "aprobo": aprobo})
Xe_train, Xe_test, ye_train, ye_test = train_test_split(estudiantes[["horas_estudio", "asistencia_pct"]], estudiantes["aprobo"],
                                                       test_size=0.25, stratify=estudiantes["aprobo"], random_state=0)

print("Jugos:", jugos.shape, "| Estudiantes:", estudiantes.shape)
display(jugos.head(3))
display(estudiantes.head(3))
```

4. `exactitud`: el porcentaje de aciertos en el examen (`Xe_test`, `ye_test`).
5. `odds_horas`: $e^{\beta}$ del coeficiente de `horas_estudio` (los «momios»): por cada hora extra de estudio, los momios de aprobar se **multiplican** por ese número.
