## 🎈 ¿Qué vamos a hacer en este taller?

En una quesería de Boyacá quieren saber **si un queso aprobará el control de calidad**. Vas a probar tres «equipos de jurados»:

1. 🌳 **Un árbol de decisión** (el juego de las 20 preguntas): ¿qué pasa si le dejas hacer *demasiadas* preguntas? Memoriza.
2. 🌲 **Un bosque** (preguntarle a cien amigos): muchos árboles votando suelen ganarle a uno solo.
3. 🧠 **Una red neuronal** (un equipo de ayudantes que aprende): resuelve un problema donde una línea recta no sirve.

```python
# ============================================================
# Preparación (no modificar): imports y datos inventados
# ============================================================
import numpy as np
import pandas as pd
from sklearn.datasets import make_circles
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.linear_model import LogisticRegression

# --- Quesos: 600 unidades; aprueba si humedad y sal están en su rango, o si está bien madurado y con pH bajo ---
rng = np.random.default_rng(4)
n = 600
humedad = rng.uniform(40, 60, n)
sal = rng.uniform(1, 4, n)
ph = rng.normal(5.4, 0.3, n)
dias = rng.integers(5, 60, n).astype(float)
extra = rng.normal(0, 1, (n, 2))          # dos medidas sin ninguna utilidad
ok = ((np.abs(humedad - 50) < 5) & (np.abs(sal - 2.5) < 0.9)) | ((dias > 45) & (ph < 5.3))
aprobado = np.where(rng.random(n) < 0.08, 1 - ok, ok).astype(int)      # 8 % de errores humanos
quesos = pd.DataFrame({"humedad_pct": humedad, "sal_pct": sal, "ph": ph, "dias_maduracion": dias,
                       "medida_extra_1": extra[:, 0], "medida_extra_2": extra[:, 1], "aprobado": aprobado})
Xq_train, Xq_test, yq_train, yq_test = train_test_split(quesos.drop(columns="aprobado"), quesos["aprobado"],
                                                       test_size=0.25, stratify=quesos["aprobado"], random_state=0)

# --- Dos anillos concéntricos (un problema donde una línea recta no sirve) ---
Xa, ya = make_circles(n_samples=400, noise=0.12, factor=0.4, random_state=0)
Xa_train, Xa_test, ya_train, ya_test = train_test_split(Xa, ya, test_size=0.25, stratify=ya, random_state=0)

print("Quesos:", quesos.shape, "| Anillos:", Xa.shape)
display(quesos.head(3))
```

## Reto 2: Preguntarle a cien amigos 🌲

Un **bosque aleatorio** entrena muchos árboles (cada uno ve una parte distinta de los datos) y los deja **votar**. Además se autoevalúa: cada árbol puede ser examinado con los quesos que *no* vio (puntaje **OOB**, *out-of-bag*, «fuera de la bolsa»).

**Puntaje sugerido:** 35 puntos.

```python
# Casillas que debes llenar
bosque = None
acc_bosque = oob = mas_importante = None

# 💡 1) 100 árboles, con autoevaluación OOB
bosque = RandomForestClassifier(n_estimators=..., oob_score=..., random_state=0)
bosque.fit(..., ...)

# 💡 2) Examen y autoevaluación
acc_bosque = bosque.score(..., ...)
oob = bosque.oob_score_

# 💡 3) feature_importances_ da un número por columna; np.argmax te dice la posición del mayor
columnas = list(Xq_train.columns)
mas_importante = columnas[np.argmax(bosque.feature_importances_)]
```
