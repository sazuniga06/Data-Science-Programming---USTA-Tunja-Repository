```python
# ============================================================
# Preparación (no modificar): imports y los aguacates de la plaza
# ============================================================
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.dummy import DummyClassifier

rng = np.random.default_rng(3)
n = 400
firmeza = rng.normal(50, 15, n).clip(5, 95)     # 0 = blandito, 100 = durísimo
oscuridad_piel = rng.uniform(0, 10, n)          # 0 = verde claro, 10 = casi negro
peso_g = rng.normal(300, 60, n)                 # peso en gramos (¡números grandes!)
z = -0.09 * (firmeza - 50) + 0.55 * (oscuridad_piel - 5)
maduro = rng.binomial(1, 1 / (1 + np.exp(-z)))  # 1 = maduro, 0 = todavía no
df = pd.DataFrame({"firmeza": firmeza, "oscuridad_piel": oscuridad_piel, "peso_g": peso_g, "maduro": maduro})
df.loc[rng.random(n) < 0.08, "firmeza"] = np.nan   # a veces se nos olvida medir la firmeza

print("Aguacates:", df.shape, "| ¿Cuántos medidas faltan?", int(df["firmeza"].isna().sum()))
display(df.head())
```
