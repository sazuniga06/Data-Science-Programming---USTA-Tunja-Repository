## 2. Aprendizaje No Supervisado: Formalización (2.3) 🔍

### 2.1 Definición

En el aprendizaje **supervisado** disponemos de pares $(\mathbf{x}_i, y_i)$ muestreados de una distribución conjunta y queremos aproximar $P(y\mid\mathbf{x})$ (o su media). En el **no supervisado** solo tenemos

$$\mathcal{D} = \{\mathbf{x}_1, \dots, \mathbf{x}_n\}, \qquad \mathbf{x}_i \in \mathbb{R}^d, \quad \mathbf{x}_i \overset{\text{i.i.d.}}{\sim} p(\mathbf{x}),$$

y el objetivo es **aprender propiedades de la distribución $p(\mathbf{x})$** sin que nadie nos diga cuál es la «respuesta correcta». Casi siempre se postula una **estructura latente** que explica los datos: un **grupo oculto** $z_i \in \{1,\dots,K\}$ (en *clustering*), unas **pocas coordenadas ocultas** $\mathbf{h}_i \in \mathbb{R}^m$ con $m \ll d$ (en reducción de dimensionalidad), etc. Lo que observamos es $\mathbf{x}_i$; lo que queremos recuperar es esa estructura.

### 2.2 Las cuatro tareas clásicas

| Tarea | Qué se busca | Formalmente | Dónde se estudia |
|---|---|---|---|
| ***Clustering*** (agrupamiento) | Particionar los datos en grupos «parecidos» | $z:\mathbb{R}^d\to\{1,\dots,K\}$ (o $\{-1\}\cup\{1,\dots,K\}$ si hay ruido) | **Módulo 04** |
| **Reducción de dimensionalidad** | Representación compacta que conserve lo esencial | $f:\mathbb{R}^d\to\mathbb{R}^m$, $m\ll d$ | Módulo 05 |
| **Estimación de densidad** | Estimar la propia $p(\mathbf{x})$ | $\hat p(\mathbf{x})$: KDE, mezclas gaussianas | Se menciona aquí; relación con DBSCAN y *GMM* |
| **Detección de anomalías** | Encontrar puntos de $p(\mathbf{x})$ muy baja | $\text{anomalía}\iff \hat p(\mathbf{x}) < \tau$ | Se menciona aquí; DBSCAN marca ruido |

Estas tareas **no son compartimentos estancos**: DBSCAN es a la vez un método de *clustering* y de detección de anomalías, una mezcla gaussiana sirve para *clustering*, densidad y anomalías, y el *clustering* suele aplicarse **después** de reducir dimensionalidad. Veamos las cuatro sobre datos sintéticos y reales.

````python
# ============================================================
# 2. Las cuatro tareas del aprendizaje no supervisado, lado a lado
# ============================================================
rng = np.random.RandomState(SEED)
X_b, _ = make_blobs(n_samples=[150, 150, 150], centers=[[-4, -2], [4, -1], [0, 5]],
                    cluster_std=[1.0, 1.0, 0.8], random_state=SEED)
fig, axes = plt.subplots(2, 2, figsize=(12.5, 9))

# (a) Clustering: K-Means con K = 3
km = KMeans(n_clusters=3, n_init=10, random_state=SEED).fit(X_b)
for k in range(3):
    axes[0, 0].scatter(*X_b[km.labels_ == k].T, s=18, color=PALETA[k], alpha=0.8, label=f"grupo {k}")
axes[0, 0].scatter(*km.cluster_centers_.T, marker="X", s=200, color="black", label="centroides")
axes[0, 0].set_title("(a) Clustering: partir los datos en grupos", fontweight="bold")
axes[0, 0].legend(fontsize=8)

# (b) Reduccion de dimensionalidad: digitos 64D -> 2D con PCA (las etiquetas solo se usan para COLOREAR)
dig = load_digits()
Z = PCA(n_components=2, random_state=SEED).fit_transform(dig.data)
sc = axes[0, 1].scatter(Z[:, 0], Z[:, 1], c=dig.target, cmap="tab10", s=8, alpha=0.8)
axes[0, 1].set_title("(b) Reducción de dimensionalidad: 64 variables → 2 (PCA)", fontweight="bold")
axes[0, 1].set_xlabel("componente 1"); axes[0, 1].set_ylabel("componente 2")
cb = plt.colorbar(sc, ax=axes[0, 1]); cb.set_label("dígito (solo para colorear)")

# (c) Estimacion de densidad: KDE gaussiana
kde = KernelDensity(bandwidth=0.8).fit(X_b)
gx, gy = np.meshgrid(np.linspace(-9, 9, 120), np.linspace(-7, 10, 120))
dens = np.exp(kde.score_samples(np.c_[gx.ravel(), gy.ravel()])).reshape(gx.shape)
cs = axes[1, 0].contourf(gx, gy, dens, levels=12, cmap="Blues")
axes[1, 0].scatter(*X_b.T, s=6, color="black", alpha=0.5)
axes[1, 0].set_title(r"(c) Estimación de densidad: $\hat p(\mathbf{x})$ con KDE", fontweight="bold")
plt.colorbar(cs, ax=axes[1, 0]).set_label(r"$\hat p(\mathbf{x})$")

# (d) Deteccion de anomalias: Isolation Forest sobre los blobs + 12 puntos dispersos
X_o = np.vstack([X_b, rng.uniform(-9, 9, size=(12, 2))])
iso = IsolationForest(contamination=12 / len(X_o), random_state=SEED).fit(X_o)
es_anom = iso.predict(X_o) == -1
axes[1, 1].scatter(*X_o[~es_anom].T, s=14, color=PALETA[0], alpha=0.7, label="normales")
axes[1, 1].scatter(*X_o[es_anom].T, s=70, color=PALETA[1], marker="x", linewidths=2, label="anomalías")
axes[1, 1].set_title("(d) Detección de anomalías (Isolation Forest)", fontweight="bold")
axes[1, 1].legend(fontsize=8)

for ax in (axes[0, 0], axes[1, 0], axes[1, 1]):
    ax.set_xlabel("$x_1$"); ax.set_ylabel("$x_2$")
plt.tight_layout(); plt.show()
print(f"Varianza explicada por 2 componentes en los dígitos: {PCA(2).fit(dig.data).explained_variance_ratio_.sum():.1%}  |  anomalías marcadas: {es_anom.sum()}")
````

## 3. ¿Por Qué Es Tan Difícil Validar? Sin Verdad de Terreno ⚖️

En clasificación, la **etiqueta** es un árbitro externo: $\hat y\ne y$ es un error, punto. En el *clustering* **no existe ese árbitro**, y eso trae tres problemas que conviene tener presentes desde ya:

1. **No hay un único agrupamiento «correcto».** Los mismos datos pueden tener una estructura jerárquica: ¿hay 2 grandes grupos o 4 pequeños? Depende de la **escala** a la que miremos y del **objetivo** del análisis (segmentar clientes en 3 grupos accionables puede ser mejor que en 12 «verdaderos»).
2. **Un algoritmo siempre devuelve una respuesta.** *K-Means* con $K=3$ entrega tres grupos sobre **cualquier** nube de puntos, incluso sobre ruido uniforme sin estructura alguna. Que el algoritmo converja **no** prueba que los grupos existan.
3. **El criterio interno no sirve para elegir $K$.** La inercia (suma de cuadrados dentro de cada grupo) **decrece siempre** con $K$ y llega a cero cuando cada punto es su propio grupo. Hacen falta criterios que penalicen la complejidad (cuaderno 04).

Kleinberg (2002) formalizó una versión extrema: **no existe** una función de *clustering* que satisfaga a la vez *invariancia a la escala*, *riqueza* (puede producir cualquier partición) y *consistencia* (acercar puntos del mismo grupo y alejar los de grupos distintos no cambia el resultado). Es un teorema de imposibilidad *(citado de memoria, verificar)*: **toda** elección de algoritmo incorpora supuestos. Lo ilustramos.

````python
# ============================================================
# 3. (a) Dos lecturas validas de los mismos datos; (b) estructura inexistente en ruido uniforme
# ============================================================
# Cuatro blobs organizados en dos parejas: K=2 y K=4 son ambas lecturas razonables
X_j, _ = make_blobs(n_samples=400, centers=[[-5.5, -1], [-3.3, 1.2], [3.3, -1.2], [5.5, 1]], cluster_std=0.6, random_state=SEED)
X_u = np.random.RandomState(SEED).uniform(-1, 1, size=(400, 2))     # ruido uniforme: NO hay grupos

fig, axes = plt.subplots(1, 3, figsize=(16, 4.6))
for ax, (X, K, titulo) in zip(axes, [(X_j, 2, "Misma data, K = 2: dos 'continentes'"),
                                    (X_j, 4, "Misma data, K = 4: cuatro 'países'"),
                                    (X_u, 3, "Ruido uniforme con K = 3: ¡también 'encuentra' grupos!")]):
    m = KMeans(n_clusters=K, n_init=10, random_state=SEED).fit(X)
    for k in range(K):
        ax.scatter(*X[m.labels_ == k].T, s=14, color=PALETA[k], alpha=0.8)
    ax.scatter(*m.cluster_centers_.T, marker="X", s=180, color="black")
    ax.set_title(f"{titulo}\ninercia = {m.inertia_:.1f}", fontweight="bold", fontsize=10)
    ax.set_xlabel("$x_1$"); ax.set_ylabel("$x_2$")
plt.tight_layout(); plt.show()

# La inercia baja SIEMPRE con K (no sirve para elegir K): lo comprobamos
ine = [KMeans(n_clusters=k, n_init=5, random_state=SEED).fit(X_j).inertia_ for k in range(1, 11)]
assert all(a >= b - 1e-9 for a, b in zip(ine, ine[1:]))
print("Inercia para K = 1..10:", np.round(ine, 1))
print("Decrece monótonamente: minimizarla elegiría K = n. Hay que usar criterios que penalicen la complejidad (cuaderno 04).")
````

## 4. Escalado y Métrica de Distancia: Definen Qué Significa «Parecido» 📏

Todos los algoritmos de este módulo se apoyan en una noción de **disimilitud** entre observaciones. Elegirla es **ya** una decisión de modelado.

### 4.1 La familia de distancias

Para $\mathbf{x},\mathbf{y}\in\mathbb{R}^d$:

| Distancia | Fórmula | Característica |
|---|---|---|
| **Minkowski** ($p\ge 1$) | $d_p(\mathbf{x},\mathbf{y})=\Big(\sum_{j=1}^d |x_j-y_j|^p\Big)^{1/p}$ | Familia general |
| **Euclídea** ($p=2$) | $\sqrt{\sum_j (x_j-y_j)^2}$ | Sensible a la escala; la que usan *K-Means* y *Ward* |
| **Manhattan** ($p=1$) | $\sum_j |x_j-y_j|$ | Menos dominada por una sola coordenada atípica; natural con datos «en cuadrícula» |
| **Coseno** | $1-\dfrac{\mathbf{x}^\top\mathbf{y}}{\|\mathbf{x}\|\,\|\mathbf{y}\|}$ | Mide **ángulo**, ignora la **magnitud** (texto, perfiles de uso) |

Las tres primeras cumplen los axiomas de **métrica** (no negatividad, identidad, simetría, desigualdad triangular). La *distancia coseno* $1-\cos\theta$ **no** cumple la desigualdad triangular (sí la cumple la distancia angular $\arccos(\cos\theta)$), un detalle que importa para algunos algoritmos con índices espaciales.

### 4.2 Dos mecanismos que cambian el resultado

- **Escala de las variables.** Con la distancia Euclídea, la variable de mayor rango o varianza **domina** el cálculo. Medir un ingreso en pesos y una edad en años hace que la «edad» sea prácticamente invisible. Solución habitual: **estandarizar** ($z=(x-\mu)/\sigma$). No es obligatorio siempre: si las unidades son comparables y la escala *es* información, puede omitirse. (Ya vimos esto desde la óptica del preprocesamiento en el [Módulo 00, cuaderno 03](../00%20-%20Introduccion%20al%20Machine%20Learning/03_Importancia_del_Preprocesamiento.ipynb).)
- **Tipo de distancia.** Define la **geometría** de «cercano». Con coseno, dos puntos sobre la misma semirrecta que parte del origen son «idénticos» aunque estén lejos.

Probamos el primero con un caso real, el conjunto **Wine** (178 vinos, 13 variables químicas, 3 cultivares). Las variables tienen rangos muy distintos (la *prolina* ronda los cientos o miles; la mayoría de las demás, unidades). Comparamos *K-Means* con $K=3$ **sin escalar** y **escalado**, usando el *Adjusted Rand Index* (ARI; 1 = coincide con los cultivares, ≈0 = azar; lo estudiamos en el cuaderno 04) solo para **cuantificar** qué tan bien recupera los cultivares.

````python
# ============================================================
# 4a. Efecto del escalado: Wine sin escalar vs. estandarizado
# ============================================================
wine = load_wine()
Xw, yw = wine.data, wine.target
desv = pd.Series(Xw.std(axis=0), index=wine.feature_names).sort_values(ascending=False)
print("Desviación estándar por variable (las 3 mayores):")
print(desv.head(3).round(2).to_string())
print(f"Cociente entre la mayor y la menor desviación: {desv.max() / desv.min():,.0f}\n")

resultados = {}
for nombre, X in [("Sin escalar", Xw), ("Estandarizado", StandardScaler().fit_transform(Xw))]:
    m = KMeans(n_clusters=3, n_init=10, random_state=SEED).fit(X)
    resultados[nombre] = (adjusted_rand_score(yw, m.labels_), m.labels_)
    print(f"{nombre:14s}: ARI frente a los cultivares = {resultados[nombre][0]:.3f}")

# Contribucion de cada variable a la distancia cuadratica entre dos vinos al azar (sin escalar)
i, j = np.random.RandomState(SEED).choice(len(Xw), 2, replace=False)
contrib = pd.Series((Xw[i] - Xw[j]) ** 2, index=wine.feature_names)
contrib = (contrib / contrib.sum()).sort_values(ascending=False)

fig, axes = plt.subplots(1, 2, figsize=(14, 4.6))
axes[0].bar(range(len(desv)), desv.values, color=PALETA[0])
axes[0].set_yscale("log"); axes[0].set_xticks(range(len(desv)))
axes[0].set_xticklabels(desv.index, rotation=75, ha="right", fontsize=8)
axes[0].set_ylabel("Desviación estándar (escala log)"); axes[0].set_title("Wine: las escalas de las variables son muy distintas", fontweight="bold")
axes[1].bar(list(resultados.keys()), [v[0] for v in resultados.values()], color=[PALETA[1], PALETA[2]])
axes[1].set_ylim(0, 1); axes[1].set_ylabel("ARI frente a los cultivares reales")
axes[1].set_title("K-Means (K = 3) con y sin estandarizar", fontweight="bold")
for x, v in enumerate(resultados.values()):
    axes[1].text(x, v[0] + 0.02, f"{v[0]:.2f}", ha="center", fontweight="bold")
plt.tight_layout(); plt.show()
print(f"En la distancia cuadrática entre dos vinos, la variable '{contrib.index[0]}' aporta {contrib.iloc[0]:.1%} del total.")
assert resultados["Estandarizado"][0] > resultados["Sin escalar"][0] + 0.3     # escalar mejora mucho el resultado en Wine
````

### 4.3 La métrica importa: dirección contra magnitud

Imagina documentos representados por **frecuencias de palabras**: dos textos sobre el mismo tema, uno corto y otro largo, apuntan en la **misma dirección** del espacio pero con **magnitudes** distintas. La distancia Euclídea los separa por su longitud; la distancia coseno, no. Construimos dos «temas» como dos **semirrectas** que parten del origen con ángulos cercanos y magnitudes de 1 a 10, y comparamos *K-Means* sobre los puntos crudos con *K-Means* sobre los puntos **normalizados a longitud 1**.

Hay una razón matemática por la que normalizar equivale a usar coseno: si $\|\mathbf{u}\|=\|\mathbf{v}\|=1$, entonces

$$\|\mathbf{u}-\mathbf{v}\|^2 = \|\mathbf{u}\|^2+\|\mathbf{v}\|^2-2\,\mathbf{u}^\top\mathbf{v} = 2\,(1-\cos\theta),$$

es decir, **la distancia Euclídea entre vectores unitarios es una función monótona de la distancia coseno**, así que ambas inducen el mismo orden de cercanía. A *K-Means* sobre vectores normalizados se le llama a veces *spherical K-Means* (Dhillon & Modha, 2001; *citado de memoria, verificar*).

````python
# ============================================================
# 4b. Euclidea vs. coseno: dos "temas" como semirrectas desde el origen
# ============================================================
rng = np.random.RandomState(SEED)
def semirrectas(n, angulos=(15, 40), ruido_grados=2.0):
    P, y = [], []
    for g, ang in enumerate(angulos):
        r = rng.uniform(1, 10, n)                                    # magnitud (p. ej. longitud del documento)
        th = np.deg2rad(ang + rng.normal(0, ruido_grados, n))        # direccion (p. ej. el tema)
        P.append(np.c_[r * np.cos(th), r * np.sin(th)]); y += [g] * n
    return np.vstack(P), np.array(y)

Xr, yr = semirrectas(200)
Xn = Xr / np.linalg.norm(Xr, axis=1, keepdims=True)                 # normalizacion L2 por fila

# Identidad: ||u - v||^2 = 2 (1 - cos) para vectores unitarios
u, v = Xn[0], Xn[250]
assert np.isclose(np.sum((u - v) ** 2), 2 * (1 - u @ v))
assert np.isclose(cdist([Xn[0]], [Xn[250]], "euclidean")[0, 0] ** 2, 2 * cdist([Xn[0]], [Xn[250]], "cosine")[0, 0])

lab_euc = KMeans(2, n_init=10, random_state=SEED).fit_predict(Xr)
lab_cos = KMeans(2, n_init=10, random_state=SEED).fit_predict(Xn)
fig, axes = plt.subplots(1, 3, figsize=(16, 4.8))
for ax, lab, titulo in [(axes[0], yr, "Temas reales (dos direcciones)"),
                        (axes[1], lab_euc, f"K-Means Euclídea (ARI = {(lambda v: 0.0 if abs(v) < 0.005 else v)(adjusted_rand_score(yr, lab_euc)):.2f})"),
                        (axes[2], lab_cos, f"K-Means sobre vectores normalizados ≡ coseno (ARI = {adjusted_rand_score(yr, lab_cos):.2f})")]:
    for k in (0, 1):
        ax.scatter(*Xr[lab == k].T, s=14, color=PALETA[k], alpha=0.8, label=f"grupo {k}")
    ax.set_title(titulo, fontweight="bold", fontsize=10); ax.set_xlabel("$x_1$"); ax.set_ylabel("$x_2$"); ax.set_aspect("equal"); ax.legend(fontsize=8)
plt.tight_layout(); plt.show()
print("La distancia Euclídea separa por MAGNITUD (cerca/lejos del origen); el coseno separa por DIRECCIÓN (el tema).")
assert adjusted_rand_score(yr, lab_cos) > 0.9 > adjusted_rand_score(yr, lab_euc)
````

## 5. La Maldición de la Dimensionalidad 🌀

Los algoritmos basados en distancias se apoyan en que «cercano» y «lejano» se **distingan**. Cuando $d$ crece, esa distinción **se desvanece**. Dos hechos clásicos:

**(a) Concentración de distancias.** Para puntos i.i.d. con $d$ coordenadas independientes, la distancia al **vecino más lejano** y al **más cercano** tienden a ser casi iguales. Beyer et al. (1999) demuestran que, bajo condiciones generales, el **contraste relativo**

$$\text{CR}_d=\frac{D_{\max}-D_{\min}}{D_{\min}}\;\xrightarrow[d\to\infty]{\ \mathbb{P}\ }\;0,$$

donde $D_{\max}$ y $D_{\min}$ son la mayor y la menor distancia desde un punto de consulta al conjunto *(citado de memoria, verificar)*. Si todo está «igual de lejos», el vecino más cercano deja de ser informativo.

**(b) La geometría del volumen.** La fracción del hipercubo $[0,1]^d$ ocupada por la bola inscrita (radio $\tfrac12$) es

$$\frac{V_{\text{bola}}}{V_{\text{cubo}}}=\frac{\pi^{d/2}}{\Gamma\!\left(\tfrac d2+1\right)}\left(\frac12\right)^d \xrightarrow[d\to\infty]{} 0,$$

y ya en $d=10$ es menor que el 0.3 %: **casi todo el volumen está en las esquinas**, lejos de cualquier centro.

**(c) Consecuencia para el *clustering*.** Cada variable **irrelevante** añade ruido a todas las distancias y **diluye** la señal de las variables informativas. Lo medimos: dos grupos gaussianos bien separados en 2 dimensiones, a los que añadimos $d-2$ variables de ruido puro (*K-Means* estandarizado, ARI promedio sobre 5 repeticiones).

````python
# ============================================================
# 5. Concentracion de distancias, volumen de la bola inscrita y degradacion del clustering
# ============================================================
dims = [1, 2, 5, 10, 20, 50, 100, 200, 500, 1000]
rng = np.random.RandomState(SEED)
contraste = []
for d in dims:
    X = rng.rand(1000, d); Q = rng.rand(50, d)                    # 1000 puntos, 50 consultas en el hipercubo
    D = cdist(Q, X)
    contraste.append(np.mean((D.max(axis=1) - D.min(axis=1)) / D.min(axis=1)))
log10_frac = [((d / 2) * np.log(np.pi) - gammaln(d / 2 + 1) - d * np.log(2)) / np.log(10) for d in dims]   # en escala log para evitar underflow
assert np.isclose(10 ** log10_frac[1], np.pi / 4)                       # d = 2: circulo inscrito en el cuadrado = pi/4

# Clustering con variables de ruido anadidas: 2 grupos separados en 2 variables + (d - 2) variables de ruido
dims_c = [2, 5, 10, 20, 50, 100, 200, 500]
ari_d = []
for d in dims_c:
    vals = []
    for s in range(5):
        Xs, ys = make_blobs(n_samples=300, centers=[[0, 0], [4, 0]], cluster_std=1.0, random_state=s)
        Xs = np.hstack([Xs, np.random.RandomState(100 + s).normal(size=(300, d - 2))])
        lab = KMeans(2, n_init=5, random_state=SEED).fit_predict(StandardScaler().fit_transform(Xs))
        vals.append(adjusted_rand_score(ys, lab))
    ari_d.append(np.mean(vals))

fig, axes = plt.subplots(1, 3, figsize=(16, 4.4))
axes[0].semilogx(dims, contraste, "o-", color=PALETA[0], lw=2); axes[0].set_xlabel("Dimensión $d$ (escala log)")
axes[0].set_ylabel(r"Contraste relativo $(D_{\max}-D_{\min})/D_{\min}$"); axes[0].set_yscale("log")
axes[0].set_title("Las distancias se concentran al crecer $d$", fontweight="bold", fontsize=10)
axes[1].semilogx(dims, log10_frac, "s-", color=PALETA[1], lw=2); axes[1].set_xlabel("Dimensión $d$ (escala log)")
axes[1].set_ylabel(r"$\log_{10}$(volumen bola inscrita / volumen cubo)"); axes[1].set_title("Casi todo el volumen queda en las esquinas", fontweight="bold", fontsize=10)
axes[2].semilogx(dims_c, ari_d, "^-", color=PALETA[2], lw=2); axes[2].set_ylim(-0.05, 1.05); axes[2].set_xlabel("Variables totales $d$ (2 informativas + ruido; escala log)")
axes[2].set_ylabel("ARI de K-Means (K = 2)"); axes[2].set_title("El clustering se degrada con variables irrelevantes", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()

print(pd.DataFrame({"d": dims, "contraste relativo": np.round(contraste, 3), "log10(bola/cubo)": np.round(log10_frac, 1)}).to_string(index=False))
print("\nARI de K-Means según variables totales:", {d: round(float(a), 2) for d, a in zip(dims_c, ari_d)})
assert contraste[0] > contraste[3] > contraste[-1]               # el contraste decrece con d
assert ari_d[0] > 0.7 and ari_d[-1] < ari_d[0] - 0.2             # la calidad cae al anadir ruido
````

---
Las **salidas prácticas** ante la maldición de la dimensionalidad: **seleccionar** variables relevantes, **reducir** dimensionalidad antes de agrupar (Módulo 05), usar **distancias más robustas** (coseno o Manhattan suelen degradarse menos que la Euclídea en alta dimensión) o algoritmos que modelen subespacios. Y, sobre todo, **no esperar** que un algoritmo de distancia funcione bien con cientos de variables sin curar.

## 6. Taxonomía de los Algoritmos de *Clustering* 🌳

No hay un «mejor» algoritmo: cada familia asume una **idea distinta de grupo**.



### 6.1 Tabla comparativa

| Criterio | **Particional** (*K-Means*) | **Jerárquico** (aglomerativo) | **Densidad** (DBSCAN) | **Mezclas** (GMM) |
|---|---|---|---|---|
| **Idea de grupo** | Cercanía a un centroide | Árbol de fusiones sucesivas | Región conexa de alta densidad | Componente probabilístico |
| **¿Hay que fijar $K$?** | Sí | No (se **corta** el dendrograma) | No (se fijan $\varepsilon$ y `min_samples`) | Sí (o se elige por BIC/AIC) |
| **Forma de los grupos** | Convexos, esféricos, tamaño similar | Depende del enlace (*single* sigue formas largas; *Ward* ≈ esféricos) | **Arbitraria** | Elipsoides (covarianzas libres) |
| **Ruido y *outliers*** | Los absorbe y desplaza centroides | Sensible (sobre todo *complete* y *Ward*) | **Los marca** como ruido (`-1`) | Sensibles; la densidad baja los delata |
| **Asignación** | Dura | Dura (por corte) | Dura (más ruido) | **Suave** (probabilidades) |
| **Determinismo** | No (inicialización) | **Sí** | Casi (solo puntos borde ambiguos) | No (inicialización de EM) |
| **Costo típico** | $O(nKd)$ por iteración | $O(n^2)$ memoria, $O(n^2)$–$O(n^3)$ tiempo | $O(n\log n)$ con índice espacial; $O(n^2)$ peor caso | $O(nKd^2)$ por iteración (cov. completa) |
| **Escala a $n$ grande** | **Muy bien** (*Mini-Batch*) | Mal sin restricciones | Bien en baja dimensión | Aceptable |
| **Cuaderno** | [01](01_KMeans.ipynb) | [02](02_Clustering_Jerarquico.ipynb) | [03](03_DBSCAN.ipynb) | Mención en 01 |

### 6.2 Asignación dura y asignación suave

Un *clustering* **duro** asigna cada punto a exactamente un grupo ($z_i\in\{1,\dots,K\}$). Uno **suave** asigna **probabilidades** $\gamma_{ik}=P(z_i=k\mid\mathbf{x}_i)$ con $\sum_k\gamma_{ik}=1$. *K-Means* es el **caso límite** de una mezcla de gaussianas esféricas de igual varianza $\sigma^2$ cuando $\sigma^2\to0$ (las responsabilidades se vuelven 0/1): es un resultado que aprovecharemos en el cuaderno 01 para entender sus limitaciones.

### 6.3 Un flujo de trabajo razonable



La validación (cuaderno 04) **no es un sello final**, sino parte de un ciclo: si las métricas o el criterio experto no convencen, se vuelve a la distancia, a la familia o a los hiperparámetros.
