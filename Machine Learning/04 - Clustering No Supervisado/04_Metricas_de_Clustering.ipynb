## 1. ¿Cómo se Evalúa un Agrupamiento sin Etiquetas? 🧭

Los criterios de validación se agrupan en tres familias (Jain & Dubes, 1988; *citado de memoria, verificar*):

| Familia | Usa… | Responde a… | Ejemplos |
|---|---|---|---|
| **Externa** | una **partición de referencia** (etiquetas reales) | ¿Se parece mi agrupamiento a la verdad? | **ARI**, **NMI/AMI**, homogeneidad, V-measure, pureza |
| **Interna** | **solo los datos** y las etiquetas de grupo | ¿Los grupos son internamente cohesivos y están bien separados? | **Silueta**, **Calinski–Harabasz**, Davies–Bouldin, **DBCV** |
| **Relativa** | compara **varias** soluciones del mismo algoritmo | ¿Qué $k$ o qué hiperparámetros son mejores? | codo, *gap statistic*, máximos de silueta/CH |

Los criterios internos **cuantifican dos ideas** que todos compartimos intuitivamente:

- **Cohesión** (compacidad): los puntos de un grupo deben estar **cerca entre sí**.
- **Separación**: los grupos distintos deben estar **lejos unos de otros**.

La diferencia entre un índice y otro es **cómo** define «cerca» y «lejos». Y ahí está la trampa: **cada definición incorpora una idea de qué es un grupo**. Los índices basados en **centroides y distancias** (silueta, CH, Davies–Bouldin, inercia) suponen grupos **convexos**; los basados en **densidad** (DBCV) suponen grupos como **regiones densas conexas**. Ningún criterio interno es «neutral».

> ⚠️ **Dos reglas de oro.** (1) **Un índice interno no valida contra una verdad**: dice si la partición es coherente con *su* noción de grupo, y esa noción puede no ser la del problema. (2) Los valores **solo se comparan** entre soluciones sobre **los mismos datos** (y con el mismo preprocesamiento).

## 2. Coeficiente de Silueta (2.5.1) 🌅

### 2.1 Definición (Rousseeuw, 1987)

Para cada observación $i$ que pertenece al grupo $C_I$ se calculan dos cantidades con una distancia $d$ (por defecto la Euclídea):

- **Cohesión** $a(i)$: la **distancia media** de $i$ a los **demás** puntos de su propio grupo,
$$a(i)=\frac{1}{|C_I|-1}\sum_{j\in C_I,\;j\ne i}d(i,j).$$
- **Separación** $b(i)$: la distancia media de $i$ a los puntos del **grupo vecino más cercano** (el otro grupo con menor distancia media),
$$b(i)=\min_{J\ne I}\ \frac{1}{|C_J|}\sum_{j\in C_J}d(i,j).$$

La **silueta** de $i$ es

$$s(i)=\frac{b(i)-a(i)}{\max\{a(i),\,b(i)\}}\in[-1,1],\qquad s(i)=0\ \text{si }|C_I|=1 .$$

El **coeficiente de silueta** (*silhouette score*) del agrupamiento es la **media** de $s(i)$ sobre todos los puntos.

### 2.2 Interpretación

| Valor de $s(i)$ | Lectura |
|---|---|
| **cerca de 1** | $a\ll b$: el punto está **muy cerca de su grupo y lejos de los demás** |
| **cerca de 0** | $a\approx b$: el punto está en la **frontera** entre dos grupos |
| **negativo** | $a>b$: el punto está **más cerca de otro grupo** que del suyo: probablemente mal asignado |

Para la media, Kaufman & Rousseeuw (1990) proponen la lectura orientativa *(citado de memoria, verificar los umbrales)*: **0.71–1.00** estructura fuerte; **0.51–0.70** razonable; **0.26–0.50** débil o artificial; **$\le0.25$** sin estructura sustancial. Son guías, no reglas.

### 2.3 Propiedades y costos

- Requiere **al menos 2 y a lo sumo $n-1$ grupos**.
- Cuesta $O(n^2)$ (distancias entre todos los pares): para $n$ grande se **submuestrea** (`sample_size` en `scikit-learn`).
- El **gráfico de siluetas** (una barra por punto, agrupadas por grupo y ordenadas) revela cosas que la media oculta: **grupos con siluetas bajas o negativas**, grupos **de tamaños muy dispares** y si **todos** los grupos superan la media.

Lo implementamos con la matriz de distancias y lo verificamos contra `silhouette_samples` y `silhouette_score`.

````python
# ============================================================
# 2. Silueta DESDE CERO (por muestra y global), verificada contra sklearn.metrics
# ============================================================
def silueta_muestras(X, etiquetas):
    """s(i) = (b - a) / max(a, b), con a = distancia media al propio grupo (sin contarse a si mismo)
    y b = distancia media al grupo vecino mas cercano. s = 0 si el grupo tiene un solo punto."""
    etiquetas = np.asarray(etiquetas)
    D = cdist(X, X)
    clases, idx = np.unique(etiquetas, return_inverse=True)
    n, K = len(X), len(clases)
    tam = np.bincount(idx, minlength=K)
    sumas = np.column_stack([D[:, idx == k].sum(axis=1) for k in range(K)])      # suma de distancias de cada punto a cada grupo
    a = sumas[np.arange(n), idx] / np.maximum(tam[idx] - 1, 1)                   # el propio punto aporta distancia 0 a la suma
    medias = sumas / tam                                                          # distancia media a cada grupo
    medias[np.arange(n), idx] = np.inf                                            # excluir el propio grupo al buscar el vecino
    b = medias.min(axis=1)
    with np.errstate(invalid="ignore", divide="ignore"):
        s = (b - a) / np.maximum(a, b)
    s[tam[idx] == 1] = 0.0                                                        # grupos de un solo punto
    return np.nan_to_num(s)                                                       # a = b = 0 (puntos duplicados) -> 0

def silueta_score(X, etiquetas):
    return float(silueta_muestras(X, etiquetas).mean())

# --- Verificacion en varios conjuntos y particiones ---
X_a, y_a = make_blobs(n_samples=400, centers=4, cluster_std=1.0, random_state=8)
iris = load_iris(); X_i = iris.data
casos = {"blobs, K-Means K=4": (X_a, KMeans(4, n_init=10, random_state=SEED).fit_predict(X_a)),
         "blobs, K-Means K=2": (X_a, KMeans(2, n_init=10, random_state=SEED).fit_predict(X_a)),
         "blobs, partición real": (X_a, y_a),
         "iris, K-Means K=3": (X_i, KMeans(3, n_init=10, random_state=SEED).fit_predict(X_i)),
         "iris, especies": (X_i, iris.target)}
filas = []
for nombre, (X, lab) in casos.items():
    s_mio, s_sk = silueta_muestras(X, lab), skm.silhouette_samples(X, lab)
    assert np.allclose(s_mio, s_sk, atol=1e-10), nombre                           # silueta por muestra identica
    assert np.isclose(silueta_score(X, lab), skm.silhouette_score(X, lab)), nombre  # silueta global identica
    filas.append({"partición": nombre, "silueta propia": silueta_score(X, lab), "silueta sklearn": skm.silhouette_score(X, lab),
                  "min s(i)": s_mio.min(), "% con s<0": 100 * np.mean(s_mio < 0)})
print("Silueta propia == sklearn.metrics.silhouette_samples / silhouette_score (por muestra y global).")
display(pd.DataFrame(filas).set_index("partición").round(4))

# --- Grafico de siluetas por muestra para K = 2, 4 y 6 ---
def grafico_silueta(ax, X, lab, titulo):
    s = silueta_muestras(X, lab); y0 = 10
    for k, c in enumerate(np.unique(lab)):
        sc = np.sort(s[lab == c])
        ax.fill_betweenx(np.arange(y0, y0 + len(sc)), 0, sc, color=PALETA[k % len(PALETA)], alpha=0.85)
        ax.text(-0.09, y0 + len(sc) / 2, f"{c}", fontsize=9)
        y0 += len(sc) + 10
    ax.axvline(s.mean(), color="red", ls="--", label=f"media = {s.mean():.2f}")
    ax.set_xlim(-0.2, 1.0); ax.set_yticks([]); ax.set_xlabel("Silueta $s(i)$"); ax.set_title(titulo, fontweight="bold", fontsize=10); ax.legend(loc="lower right", fontsize=8)

fig, axes = plt.subplots(1, 3, figsize=(16, 4.8))
for ax, K in zip(axes, [2, 4, 6]):
    grafico_silueta(ax, X_a, KMeans(K, n_init=10, random_state=SEED).fit_predict(X_a), f"K-Means con K = {K}")
axes[0].set_ylabel("Observaciones agrupadas por grupo")
plt.tight_layout(); plt.show()
print("K = 4: los cuatro grupos tienen siluetas altas y parejas; K = 2 junta grupos distintos y K = 6 parte grupos reales (siluetas bajas o negativas en las 'fronteras' artificiales).")
````

## 3. Índice de Calinski–Harabasz (2.5.2) 📊

### 3.1 Definición (Caliński & Harabasz, 1974)

También llamado **criterio de la razón de varianzas** (*variance ratio criterion*). Con $n$ puntos, $K$ grupos, $\boldsymbol\mu$ la media global y $\boldsymbol\mu_k$ la media del grupo $C_k$, se definen la **dispersión entre grupos** y la **dispersión dentro de los grupos**:

$$B_K=\sum_{k=1}^{K}|C_k|\,\|\boldsymbol\mu_k-\boldsymbol\mu\|^2,\qquad W_K=\sum_{k=1}^{K}\sum_{i\in C_k}\|\mathbf{x}_i-\boldsymbol\mu_k\|^2\;(\text{la inercia de }K\text{-Means}).$$

Se cumple la **descomposición de la suma de cuadrados total** (como en ANOVA): $\;T=\sum_i\|\mathbf{x}_i-\boldsymbol\mu\|^2=B_K+W_K$. El índice es el cociente de las dos dispersiones **normalizadas por sus grados de libertad**:

$$\boxed{\;\text{CH}=\frac{B_K/(K-1)}{W_K/(n-K)}\;}$$

Es literalmente el **estadístico $F$ de ANOVA** aplicado a los grupos encontrados.

### 3.2 Interpretación

- **Más alto es mejor**: grupos **compactos** (W pequeño) y **bien separados** (B grande).
- **No está acotado**: va de 0 a $+\infty$, así que **su valor absoluto no significa nada**; solo sirve para **comparar** soluciones sobre los mismos datos (típicamente, el $K$ que maximiza CH).
- Es **muy rápido** ($O(nd)$), a diferencia de la silueta ($O(n^2)$).
- Como se basa en **centroides y sumas de cuadrados**, favorece **grupos convexos y esféricos**, igual que *K-Means* (de hecho $W_K$ es su función objetivo).
- Pese al nombre «razón F», **no** se interpreta con la tabla $F$: los grupos se formaron a partir de los mismos datos, así que el supuesto de ANOVA no se cumple.

````python
# ============================================================
# 3. Calinski-Harabasz DESDE CERO + descomposicion T = B + W, verificado contra sklearn
# ============================================================
def sumas_de_cuadrados(X, etiquetas):
    """Devuelve (T, B, W): total, entre grupos e intra grupos."""
    mu = X.mean(axis=0); T = float(((X - mu) ** 2).sum()); B = W = 0.0
    for c in np.unique(etiquetas):
        Xc = X[etiquetas == c]; mc = Xc.mean(axis=0)
        B += len(Xc) * float(((mc - mu) ** 2).sum())
        W += float(((Xc - mc) ** 2).sum())
    return T, B, W

def calinski_harabasz(X, etiquetas):
    n, K = len(X), len(np.unique(etiquetas))
    T, B, W = sumas_de_cuadrados(X, etiquetas)
    return (B / (K - 1)) / (W / (n - K))

filas = []
for nombre, (X, lab) in casos.items():
    T, B, W = sumas_de_cuadrados(X, lab)
    assert np.isclose(T, B + W)                                                    # descomposicion T = B + W
    assert np.isclose(calinski_harabasz(X, lab), skm.calinski_harabasz_score(X, lab)), nombre
    filas.append({"partición": nombre, "T": T, "B (entre)": B, "W (intra)": W, "CH propio": calinski_harabasz(X, lab), "CH sklearn": skm.calinski_harabasz_score(X, lab)})
print("T = B + W y Calinski-Harabasz propio == sklearn.metrics.calinski_harabasz_score en las 5 particiones.")
display(pd.DataFrame(filas).set_index("partición").round(3))
````

## 5. El Sesgo hacia Grupos Convexos: DBSCAN Penalizado Injustamente 🎭

Silueta, CH y Davies–Bouldin comparten un supuesto: **un buen grupo es compacto alrededor de un centro**. Cuando los grupos reales son **lunas** o **anillos**, un agrupamiento **perfecto** puede puntuar **peor** que uno **equivocado** que simplemente **parte el espacio en trozos compactos**. Es el caso inverso al de la sección 4: aquí el índice **contradice** la verdad.

Lo mostramos en *lunas* y *círculos concéntricos*: comparamos la partición **correcta** (DBSCAN, ARI ≈ 1) con la **incorrecta** de *K-Means* ($K=2$), midiendo silueta, CH y el ARI frente a la verdad.

### 5.1 La trampa de la etiqueta `-1`

DBSCAN etiqueta el ruido con `-1`. Las funciones de `scikit-learn` **no lo saben**: `silhouette_score` y `calinski_harabasz_score` tratan `-1` como **un grupo más** (el «grupo del ruido», disperso por todo el espacio, con siluetas negativas). Hay dos prácticas, y **hay que declararlas**: (a) **excluir** el ruido y evaluar solo los puntos asignados (a costa de no penalizar el ruido) o (b) **incluirlo** como un grupo (castiga el ruido, pero con un criterio sin sentido geométrico). Aquí calculamos ambas.

````python
# ============================================================
# 5. Silueta y CH penalizan a DBSCAN sobre formas no convexas; trampa de la etiqueta -1
# ============================================================
datos = {"Lunas": (make_moons(400, noise=0.06, random_state=SEED), 0.18, 5),
         "Círculos": (make_circles(400, factor=0.4, noise=0.05, random_state=SEED), 0.15, 5)}

def evaluar(X, lab, y):
    """Silueta (sin ruido), CH (sin ruido) y ARI/NMI frente a la verdad."""
    m = lab != -1
    return {"silueta": silueta_score(X[m], lab[m]), "CH": calinski_harabasz(X[m], lab[m]),
            "ARI": skm.adjusted_rand_score(y, lab), "NMI": skm.normalized_mutual_info_score(y, lab)}

fig, axes = plt.subplots(2, 2, figsize=(11, 9)); filas = []
for r, (nombre, ((X, y), eps, m)) in enumerate(datos.items()):
    lab_db = DBSCAN(eps=eps, min_samples=m).fit_predict(X)
    lab_km = KMeans(2, n_init=10, random_state=SEED).fit_predict(X)
    for c, (alg, lab) in enumerate([("DBSCAN", lab_db), ("K-Means", lab_km)]):
        e = evaluar(X, lab, y); filas.append({"conjunto": nombre, "algoritmo": alg, **e})
        for g in np.unique(lab):
            axes[r, c].scatter(*X[lab == g].T, s=9, color="gray" if g < 0 else PALETA[g], marker="x" if g < 0 else "o", alpha=0.8)
        axes[r, c].set_title(f"{nombre} — {alg}\nsilueta = {e['silueta']:.2f} | CH = {e['CH']:.0f} | ARI = {e['ARI']:.2f}", fontweight="bold", fontsize=10)
        axes[r, c].set_xticks([]); axes[r, c].set_yticks([])
plt.tight_layout(); plt.show()
tabla = pd.DataFrame(filas).set_index(["conjunto", "algoritmo"]); display(tabla.round(3))

for nombre in datos:
    assert tabla.loc[(nombre, "K-Means"), "silueta"] > tabla.loc[(nombre, "DBSCAN"), "silueta"]    # la silueta prefiere la particion EQUIVOCADA
    assert tabla.loc[(nombre, "K-Means"), "CH"] > tabla.loc[(nombre, "DBSCAN"), "CH"]              # CH tambien
    assert tabla.loc[(nombre, "DBSCAN"), "ARI"] > 0.95 > 0.5 > tabla.loc[(nombre, "K-Means"), "ARI"]
print("Silueta y CH puntúan MÁS ALTO a K-Means (partición equivocada) que a DBSCAN (partición correcta): miden compacidad, no 'forma real'.")

# --- La trampa de -1: lunas con ruido uniforme anadido ---
rng = np.random.RandomState(SEED)
Xr = np.vstack([make_moons(400, noise=0.06, random_state=SEED)[0], rng.uniform([-1.5, -1], [2.5, 1.5], size=(40, 2))])
lab_r = DBSCAN(eps=0.18, min_samples=6).fit_predict(Xr); m = lab_r != -1
print(f"\nLunas + 40 puntos de ruido: DBSCAN marca {int((~m).sum())} puntos como ruido (-1).")
print(f"  Silueta EXCLUYENDO el ruido        : {skm.silhouette_score(Xr[m], lab_r[m]):.3f}")
print(f"  Silueta INCLUYENDO -1 como un grupo: {skm.silhouette_score(Xr, lab_r):.3f}   (sklearn lo trata así por defecto)")
````

## 6. DBCV: Validación de *Clustering* Basada en Densidad (2.5.3) 🏞️

### 6.1 Motivación

Si queremos evaluar un agrupamiento de **formas arbitrarias y con ruido** necesitamos un índice que entienda de **densidad**, no de centroides. **DBCV** (Moulavi, Jaskowiak, Campello, Zimek & Sander, 2014) es el índice interno propuesto para ello. Su idea: **un buen grupo es una región de densidad homogénea (poco «dispersa» por dentro) y está separada de los demás por zonas de menor densidad**. Y se mide con la maquinaria de HDBSCAN: la **distancia de alcanzabilidad mutua** y el **árbol generador mínimo**.

### 6.2 Definiciones

Sea $d$ la **dimensión** de los datos y, para cada grupo $C_l$ con $n_l$ puntos:

1. **Distancia núcleo de todos los puntos** (*all-points core distance*) de $o\in C_l$:
$$\text{coredist}(o)=\left(\frac{1}{n_l-1}\sum_{o'\in C_l,\ o'\ne o}\frac{1}{d(o,o')^{\,d}}\right)^{-1/d}.$$
Es un promedio (de potencia $-d$) de las distancias inversas a **todos los demás** puntos del grupo: **pequeña** en zonas densas, **grande** para puntos aislados. Reemplaza el «$k$-ésimo vecino» de DBSCAN por una cantidad que **no depende de elegir $k$**.
2. **Distancia de alcanzabilidad mutua** (*mutual reachability*): 
$$d_{\text{mreach}}(o,o')=\max\{\text{coredist}(o),\ \text{coredist}(o'),\ d(o,o')\}.$$
En zonas ralas «infla» las distancias; en zonas densas deja la distancia original.
3. **Árbol generador mínimo** $\text{MST}_{\text{MRD}}(C_l)$ del grafo completo sobre los puntos del grupo con pesos $d_{\text{mreach}}$. Un nodo es **interno** si tiene grado $>1$ en el árbol (se descartan las hojas, que son sensibles al ruido).
4. **Dispersión de densidad** (*density sparseness*): el **peso máximo** de las aristas del MST **entre nodos internos**:
$$\text{DSPC}(C_l)=\max_{(u,v)\in\text{MST}_{\text{MRD}}(C_l),\ u,v\ \text{internos}}d_{\text{mreach}}(u,v).$$
Mide **el «hueco» más ancho dentro del grupo**: si es grande, el grupo es poco homogéneo.
5. **Separación de densidad** (*density separation*) entre dos grupos: la **mínima** $d_{\text{mreach}}$ entre nodos **internos** de uno y otro (con la distancia núcleo calculada **dentro de cada grupo**):
$$\text{DSC}(C_i,C_j)=\min_{o\in\text{int}(C_i),\ o'\in\text{int}(C_j)}d_{\text{mreach}}(o,o').$$
6. **Validez de un grupo**: compara la separación con el grupo más cercano frente a su propia dispersión,
$$V_C(C_i)=\frac{\min_{j\ne i}\text{DSC}(C_i,C_j)-\text{DSPC}(C_i)}{\max\{\min_{j\ne i}\text{DSC}(C_i,C_j),\ \text{DSPC}(C_i)\}}\in[-1,1].$$
7. **DBCV**: promedio ponderado por tamaño, **donde el denominador es el número total de puntos $|O|$ incluido el ruido**:
$$\text{DBCV}=\sum_{i=1}^{K}\frac{|C_i|}{|O|}\,V_C(C_i)\in[-1,1].$$

### 6.3 Lectura

- $V_C>0$ (y cercano a 1): el grupo es **más homogéneo por dentro que separado de su vecino**. $V_C<0$: tiene huecos internos mayores que su distancia al vecino (típico de un grupo **partido artificialmente** o de **dos grupos fusionados**).
- **El ruido no se evalúa pero sí cuenta en $|O|$**: un agrupamiento con **mucho ruido** queda penalizado (la suma ponderada se encoge), sin necesidad de tratarlo como grupo.
- Se necesita **al menos 2 grupos** (la separación no está definida con uno solo) y cada grupo debe tener **al menos 2 puntos**.
- Costo: $O(\sum_l n_l^2)$ para las distancias y los MST (Prim denso).

### 6.4 Detalles de implementación (y por qué conviene verificar)

El artículo deja abiertos detalles que cada implementación resuelve a su manera. Seguimos los de la implementación de referencia en Python (`hdbscan.validity.validity_index`): (i) las **distancias nulas** (puntos duplicados) se excluyen de la suma de inversas; (ii) **si el MST no tiene aristas entre nodos internos** (grupos muy pequeños) se usan **todas** las aristas; (iii) **los empates** son muy comunes en la distancia de alcanzabilidad mutua (muchas aristas valen exactamente la distancia núcleo de un punto) y el **MST no es único** con empates: cuáles nodos resultan internos depende de **qué árbol** se elija. Adoptamos la misma convención que `hdbscan`: se construye con Prim desde el nodo 0, y el **padre** de cada nodo nuevo es el nodo de **menor índice** ya en el árbol a esa distancia. Con esa convención esperamos coincidir con la referencia hasta redondeo; sin ella, podrían aparecer diferencias pequeñas pero reales.

### 6.5 Verificación contra la implementación de referencia

Comparamos nuestro DBCV con `hdbscan.validity.validity_index` en distintos escenarios: particiones **verdaderas** (deterministas) de lunas, círculos y grupos gaussianos en 2D y 3D, y particiones **producidas por algoritmos** (DBSCAN con y sin ruido, K-Means).

- **Si `hdbscan` está instalado**, se compara en vivo y se exige coincidencia hasta ~$10^{-9}$.
- **Si no lo está** (o no se puede compilar en tu entorno), **se dice explícitamente** y se compara contra los **valores de referencia que `hdbscan` 0.8.44 devolvió al preparar este cuaderno** (incluidos como constantes para las particiones verdaderas, que no dependen de ningún algoritmo).

````python
# ============================================================
# 6. DBCV DESDE CERO (Moulavi et al., 2014)
# ============================================================
def distancia_nucleo_todos(D, d):
    """coredist(o) = ( mean_{o' != o} d(o, o')^(-d) )^(-1/d); las distancias nulas (duplicados) no aportan, como en hdbscan."""
    n = len(D)
    inv = np.zeros_like(D); pos = D > 0
    inv[pos] = (1.0 / D[pos]) ** d
    media = inv.sum(axis=1) / (n - 1)
    if media.sum() == 0:
        return np.zeros(n)
    with np.errstate(divide="ignore"):
        return media ** (-1.0 / d)

def mst_prim(W):
    """MST de un grafo completo (matriz W). Prim desde el nodo 0. Convencion de empates: el padre es el nodo de MENOR indice
    que ya esta en el arbol a esa distancia (la misma que usa hdbscan). Devuelve aristas (padre, hijo, peso)."""
    n = len(W)
    en = np.zeros(n, dtype=bool); en[0] = True
    mejor = W[0].copy()                                         # mejor distancia de cada nodo al arbol actual
    aristas = []
    for _ in range(n - 1):
        j = int(np.argmin(np.where(en, np.inf, mejor)))         # siguiente nodo: el mas cercano al arbol (primer indice en empates)
        w = float(mejor[j])
        padres = np.flatnonzero(en & np.isclose(W[j], w))       # nodos del arbol a distancia w; se toma el de menor indice
        aristas.append((int(padres[0]), j, w))
        en[j] = True
        mejor = np.minimum(mejor, W[j])
    return aristas

def dbcv(X, etiquetas, por_grupo=False):
    """DBCV de Moulavi et al. (2014). El ruido (-1) no se evalua pero cuenta en el denominador |O|.
    Devuelve NaN si hay menos de 2 grupos; lanza ValueError si algun grupo tiene un solo punto."""
    X = np.asarray(X, dtype=float); etiquetas = np.asarray(etiquetas)
    ids = [c for c in np.unique(etiquetas) if c != -1]
    if len(ids) < 2:
        return (np.nan, {}) if por_grupo else np.nan
    d = X.shape[1]
    info = {}
    for c in ids:
        Xc = X[etiquetas == c]
        if len(Xc) < 2:
            raise ValueError("DBCV no esta definido para grupos de un solo punto")
        D = cdist(Xc, Xc)
        core = distancia_nucleo_todos(D, d)
        mr = np.maximum(D, np.maximum(core[:, None], core[None, :]))        # alcanzabilidad mutua dentro del grupo
        aristas = mst_prim(mr)
        grado = np.zeros(len(Xc), dtype=int)
        for a, b, _ in aristas:
            grado[a] += 1; grado[b] += 1
        internos = np.flatnonzero(grado > 1)
        pesos_int = [w for a, b, w in aristas if grado[a] > 1 and grado[b] > 1]
        dspc = max(pesos_int) if pesos_int else max(w for _, _, w in aristas)   # sin aristas internas: todas las aristas
        if len(internos) == 0:
            internos = np.array([0])                                         # (caso de grupos de 2 puntos, como en hdbscan)
        info[c] = {"X": Xc, "core": core, "internos": internos, "dspc": dspc}
    validez = {}
    for c in ids:
        ic = info[c]; dsc = np.inf
        for o in ids:
            if o == c:
                continue
            io = info[o]
            A, B = ic["X"][ic["internos"]], io["X"][io["internos"]]
            M = np.maximum(cdist(A, B), np.maximum(ic["core"][ic["internos"]][:, None], io["core"][io["internos"]][None, :]))
            dsc = min(dsc, M.min())                                          # separacion de densidad con el grupo mas cercano
        validez[c] = (dsc - ic["dspc"]) / max(dsc, ic["dspc"])
    total = sum(len(info[c]["X"]) / len(X) * validez[c] for c in ids)       # |O| = todos los puntos, ruido incluido
    return (total, validez) if por_grupo else total



# ============================================================
# 6b. Verificacion de DBCV contra hdbscan.validity.validity_index (o contra valores de referencia fijos)
# ============================================================
X_lu, y_lu = make_moons(400, noise=0.07, random_state=1)
X_bl, y_bl = make_blobs(500, centers=4, cluster_std=1.0, random_state=8)
X_ci, y_ci = make_circles(400, factor=0.4, noise=0.05, random_state=0)
X_3d, y_3d = make_blobs(300, n_features=3, centers=3, cluster_std=1.2, random_state=2)

# Valores que devolvio hdbscan 0.8.44 con las particiones VERDADERAS de estos conjuntos (referencia fija, sin depender del algoritmo)
REF_HDBSCAN = {"lunas (verdad)": 0.3740922449159969, "blobs (verdad)": 0.701195993153579,
               "círculos (verdad)": 0.4135859404323754, "blobs 3D (verdad)": 0.4429740794691027}
escenarios = {"lunas (verdad)": (X_lu, y_lu), "blobs (verdad)": (X_bl, y_bl), "círculos (verdad)": (X_ci, y_ci), "blobs 3D (verdad)": (X_3d, y_3d)}
algoritmicos = {"lunas, DBSCAN": (X_lu, DBSCAN(eps=0.2, min_samples=5).fit_predict(X_lu)),
                "lunas, K-Means": (X_lu, KMeans(2, n_init=10, random_state=SEED).fit_predict(X_lu)),
                "blobs, K-Means K=3": (X_bl, KMeans(3, n_init=10, random_state=SEED).fit_predict(X_bl)),
                "blobs, K-Means K=6": (X_bl, KMeans(6, n_init=10, random_state=SEED).fit_predict(X_bl)),
                "blobs, DBSCAN (con ruido)": (X_bl, DBSCAN(eps=0.6, min_samples=6).fit_predict(X_bl))}

filas = []
for nombre, (X, lab) in {**escenarios, **algoritmicos}.items():
    mio = dbcv(X, lab)
    fila = {"escenario": nombre, "grupos": len(set(lab)) - (1 if -1 in lab else 0), "ruido": int((lab == -1).sum()), "DBCV propio": mio}
    if HAY_HDBSCAN:
        ref = float(dbcv_hdbscan(X, lab)); fila["DBCV hdbscan (en vivo)"] = ref
        assert np.isclose(mio, ref, rtol=0, atol=1e-9), (nombre, mio, ref)
        fila["|diferencia|"] = abs(mio - ref)
    if nombre in REF_HDBSCAN:
        assert np.isclose(mio, REF_HDBSCAN[nombre], rtol=0, atol=1e-9), (nombre, mio, REF_HDBSCAN[nombre])
        fila["referencia fija"] = REF_HDBSCAN[nombre]
    filas.append(fila)
tabla = pd.DataFrame(filas).set_index("escenario")
with pd.option_context("display.float_format", "{:.10f}".format, "display.max_columns", None, "display.width", 220):
    display(tabla)
if HAY_HDBSCAN:
    print(f"DBCV propio == hdbscan.validity.validity_index en {len(filas)} escenarios (diferencia máxima {tabla['|diferencia|'].max():.1e}).")
else:
    print("hdbscan NO está instalado: se omitió la comparación en vivo. Se verificaron las 4 particiones verdaderas contra los valores de referencia fijos "
          "(obtenidos con hdbscan 0.8.44). Para la comparación completa: `pip install hdbscan` y reinicia el entorno.")
````

### 6.6 DBCV en acción: lo que silueta y CH no ven

Además, DBCV sirve para **ajustar hiperparámetros** de algoritmos de densidad: barremos $\varepsilon$ en DBSCAN sobre las lunas y comparamos qué valor elige cada criterio y cuál es el óptimo según la verdad (ARI). ## 7. Métricas Externas: Cuando Sí Hay Etiquetas 🏷️

Si disponemos de una **partición de referencia** $U$ (clases reales) podemos comparar el agrupamiento $V$ con ella. Como las **etiquetas de grupo son arbitrarias** (el grupo «0» de $V$ no tiene por qué ser la clase «0»), las métricas externas deben ser **invariantes a las permutaciones** de etiquetas. Ambas se calculan desde la **tabla de contingencia** $n_{ij}=|U_i\cap V_j|$, con marginales $a_i=\sum_jn_{ij}$ y $b_j=\sum_in_{ij}$.

### 7.1 *Adjusted Rand Index* (Hubert & Arabie, 1985)

Cuenta los **pares de puntos** que ambas particiones tratan igual (juntos en ambas o separados en ambas) y **corrige por azar**:

$$\text{ARI}=\frac{\sum_{ij}\binom{n_{ij}}{2}-\dfrac{\sum_i\binom{a_i}{2}\sum_j\binom{b_j}{2}}{\binom n2}}{\dfrac12\Big[\sum_i\binom{a_i}{2}+\sum_j\binom{b_j}{2}\Big]-\dfrac{\sum_i\binom{a_i}{2}\sum_j\binom{b_j}{2}}{\binom n2}}.$$

**1** = particiones idénticas; **≈0** = asignación al azar (su valor esperado es 0 para particiones aleatorias); puede ser **negativo** (peor que el azar).

### 7.2 *Normalized Mutual Information* (Strehl & Ghosh, 2002)

Mide la **información mutua** entre los dos agrupamientos y la normaliza por la media (aritmética, el valor por defecto de `scikit-learn`) de sus entropías:

$$\text{NMI}=\frac{I(U;V)}{\tfrac12\,[H(U)+H(V)]},\qquad I(U;V)=\sum_{i,j}\frac{n_{ij}}{n}\ln\frac{n\,n_{ij}}{a_i\,b_j},\qquad H(U)=-\sum_i\frac{a_i}{n}\ln\frac{a_i}{n}.$$

**Rango $[0,1]$**. Ojo: **NMI no corrige por azar** y tiende a **crecer con el número de grupos** (más grupos ⇒ más información mutua «por casualidad»). Su versión corregida es la **AMI** (*Adjusted Mutual Information*; Vinh et al., 2010), disponible en `scikit-learn`.

Implementamos ARI y NMI, las verificamos y demostramos con un experimento la **corrección por azar**: etiquetas **completamente aleatorias** con $K$ grupos; un buen índice debería dar $\approx0$ siempre.

````python
# ============================================================
# 7. ARI y NMI DESDE CERO, verificados contra sklearn; correccion por azar
# ============================================================
def contingencia(u, v):
    _, iu = np.unique(u, return_inverse=True); _, iv = np.unique(v, return_inverse=True)
    C = np.zeros((iu.max() + 1, iv.max() + 1), dtype=np.int64); np.add.at(C, (iu, iv), 1)
    return C

def comb2(x):
    return x * (x - 1) / 2.0

def ari(u, v):
    C = contingencia(u, v); n = C.sum()
    s_ij, s_a, s_b = comb2(C).sum(), comb2(C.sum(axis=1)).sum(), comb2(C.sum(axis=0)).sum()
    esperado = s_a * s_b / comb2(n)
    maximo = 0.5 * (s_a + s_b)
    if maximo == esperado:                      # ambas particiones triviales (todo junto o todo separado)
        return 1.0
    return float((s_ij - esperado) / (maximo - esperado))

def nmi(u, v):
    C = contingencia(u, v); n = C.sum()
    pij = C / n; pi, pj = pij.sum(axis=1), pij.sum(axis=0)
    nz = pij > 0
    info_mutua = float((pij[nz] * np.log(pij[nz] / np.outer(pi, pj)[nz])).sum())
    Hu = -float((pi[pi > 0] * np.log(pi[pi > 0])).sum()); Hv = -float((pj[pj > 0] * np.log(pj[pj > 0])).sum())
    if Hu == 0 and Hv == 0:
        return 1.0
    return 0.0 if info_mutua <= 1e-15 else info_mutua / (0.5 * (Hu + Hv))

rng = np.random.RandomState(SEED)
y_ver = np.repeat([0, 1, 2], 100)
pruebas = {"idénticas": y_ver.copy(), "permutadas": (y_ver + 1) % 3, "ruidosas (15 %)": np.where(rng.rand(300) < 0.15, rng.randint(0, 3, 300), y_ver),
           "aleatorias (K=3)": rng.randint(0, 3, 300), "todo en un grupo": np.zeros(300, int), "todo separado": np.arange(300)}
filas = []
for nombre, lab in pruebas.items():
    assert np.isclose(ari(y_ver, lab), skm.adjusted_rand_score(y_ver, lab)), nombre
    assert np.isclose(nmi(y_ver, lab), skm.normalized_mutual_info_score(y_ver, lab)), nombre
    filas.append({"agrupamiento": nombre, "ARI propio": ari(y_ver, lab), "ARI sklearn": skm.adjusted_rand_score(y_ver, lab),
                  "NMI propio": nmi(y_ver, lab), "NMI sklearn": skm.normalized_mutual_info_score(y_ver, lab)})
print("ARI y NMI propios == sklearn.metrics (6 agrupamientos), incluida la invariancia a permutaciones de etiquetas.")
display(pd.DataFrame(filas).set_index("agrupamiento").round(4))

# --- Correccion por azar: etiquetas aleatorias con K grupos (20 repeticiones) ---
Ks = [2, 3, 5, 10, 30, 100, 300]
res = {"ARI": [], "NMI": [], "AMI": []}
for K in Ks:
    a, nm, am = [], [], []
    for _ in range(20):
        lab = rng.randint(0, K, 300)
        a.append(ari(y_ver, lab)); nm.append(nmi(y_ver, lab)); am.append(skm.adjusted_mutual_info_score(y_ver, lab))
    res["ARI"].append(np.mean(a)); res["NMI"].append(np.mean(nm)); res["AMI"].append(np.mean(am))
fig, ax = plt.subplots(figsize=(8.5, 4.6))
for k, (nombre, v) in enumerate(res.items()):
    ax.semilogx(Ks, v, "o-", lw=2, color=PALETA[k], label=nombre)
ax.set_xlabel("Número de grupos $K$ de la asignación aleatoria (escala log)"); ax.set_ylabel("Valor medio de la métrica"); ax.set_ylim(-0.05, 0.8)
ax.set_title("Etiquetas ALEATORIAS: NMI crece con K, ARI y AMI se quedan en ≈ 0", fontweight="bold"); ax.legend()
plt.tight_layout(); plt.show()
assert max(abs(np.array(res["ARI"]))) < 0.02 and max(abs(np.array(res["AMI"]))) < 0.02 and res["NMI"][-1] > res["NMI"][0] + 0.2
print(f"Con K = {Ks[-1]} grupos aleatorios (casi uno por punto): NMI = {res['NMI'][-1]:.2f} (¡parece información!), ARI = {res['ARI'][-1]:.3f}, AMI = {res['AMI'][-1]:.3f}.")


# ============================================================
# 7b. Panorama final: metricas internas vs. externas en lunas, circulos y blobs
# ============================================================
paquetes = {"Lunas": (make_moons(400, noise=0.06, random_state=SEED), 0.18, 5, 2),
            "Círculos": (make_circles(400, factor=0.4, noise=0.05, random_state=SEED), 0.15, 5, 2),
            "Blobs": (make_blobs(400, centers=3, cluster_std=0.9, random_state=SEED), 0.7, 6, 3)}
filas = []
for nombre, ((X, y), eps, m, K) in paquetes.items():
    for alg, lab in [("K-Means", KMeans(K, n_init=10, random_state=SEED).fit_predict(X)), ("DBSCAN", DBSCAN(eps=eps, min_samples=m).fit_predict(X))]:
        msk = lab != -1
        k = len(set(lab[msk]))
        filas.append({"conjunto": nombre, "algoritmo": alg, "silueta": silueta_score(X[msk], lab[msk]) if k >= 2 else np.nan,
                      "CH": calinski_harabasz(X[msk], lab[msk]) if k >= 2 else np.nan, "DBCV": dbcv(X, lab),
                      "ARI": ari(y, lab), "NMI": nmi(y, lab)})
panorama = pd.DataFrame(filas).set_index(["conjunto", "algoritmo"]); display(panorama.round(3))
print("Lunas y círculos: silueta y CH prefieren K-Means (equivocado); DBCV, ARI y NMI prefieren DBSCAN (correcto).")
print("Blobs (grupos convexos): K-Means recupera la verdad (ARI = 1). Los índices internos de DBSCAN salen algo MÁS ALTOS porque se calculan SIN los puntos de ruido: excluir el ruido infla los índices internos, por eso hay que declararlo.")
assert panorama.loc[("Lunas", "DBSCAN"), "DBCV"] > panorama.loc[("Lunas", "K-Means"), "DBCV"]
assert panorama.loc[("Círculos", "DBSCAN"), "DBCV"] > panorama.loc[("Círculos", "K-Means"), "DBCV"]
````

## 8. Limitaciones de DBCV y Guía de Uso 🧰

### 8.1 Limitaciones de DBCV

- **Requiere al menos 2 grupos** y no admite grupos de un solo punto.
- **Costo $O(n_l^2)$** por grupo (matriz de distancias y MST): se vuelve pesado con decenas de miles de puntos.
- **Sensible a la dimensión**: la distancia núcleo usa el exponente $d$ y, en alta dimensión, sufre la concentración de distancias (cuaderno 00).
- **Detalles de implementación** (convención de empates, nodos internos): implementaciones independientes pueden diferir en el tercer decimal; por eso verificamos.
- Sigue siendo un criterio **interno**: con densidades muy desiguales entre grupos o con grupos muy elongados cercanos entre sí puede dar valores poco intuitivos; no sustituye al criterio experto.

### 8.2 Qué métrica para qué situación

| Métrica | Tipo | Rango | Mejor | Supone… | Ideal para | Cuidado con… |
|---|---|---|---|---|---|---|
| **Silueta** | Interna | $[-1,1]$ | alto | Distancia entre pares; grupos compactos y convexos | Elegir $k$ con *K-Means*/jerárquico; diagnosticar puntos mal asignados | $O(n^2)$; formas no convexas; `-1` como grupo |
| **Calinski–Harabasz** | Interna | $[0,\infty)$ | alto | Centroides; grupos esféricos de varianza parecida | Elegir $k$ rápido en datos grandes | Valor absoluto sin significado; sesgo convexo |
| **Davies–Bouldin** | Interna | $[0,\infty)$ | **bajo** | Centroides y dispersión media | Alternativa rápida | Mismo sesgo convexo |
| **DBCV** | Interna | $[-1,1]$ | alto | **Densidad** y conectividad; admite ruido | **DBSCAN/HDBSCAN**, formas arbitrarias | $O(n_l^2)$; $\ge2$ grupos; detalles de implementación |
| **ARI** | Externa | $[-1,1]$ | alto | Una partición de referencia | Comparar con clases reales o entre ejecuciones (estabilidad) | Las clases reales pueden no ser «grupos naturales» |
| **NMI / AMI** | Externa | $[0,1]$ (AMI ≈ 0 al azar) | alto | Una partición de referencia | Comparación con muchas clases | NMI sube con $K$; preferir AMI |

**Recomendaciones prácticas:**

1. **Combina** criterios y **mira los datos** (proyecciones, siluetas por muestra): ninguna cifra sustituye la inspección.
2. **Haz coincidir el criterio con la familia del algoritmo**: silueta/CH para *K-Means* y jerárquicos compactos; DBCV para densidad.
3. **Evalúa estabilidad**: repite con submuestras o semillas distintas y compara con **ARI** entre ejecuciones.
4. **Declara qué haces con el ruido** (`-1`) al calcular cualquier índice.
5. Al final, **el mejor agrupamiento es el más útil para el problema**: ¿los grupos son interpretables y accionables?

| Concepto | Mensaje Central |
|---|---|
| **Familias de criterios** | **Externos** (con verdad: ARI, NMI), **internos** (solo datos: silueta, CH, DBCV) y **relativos** (comparar soluciones); ninguno interno es neutral |
| **Silueta** | $s(i)=\frac{b-a}{\max(a,b)}\in[-1,1]$: cohesión $a$ frente a separación $b$; global = media; $O(n^2)$; el gráfico por muestra revela grupos débiles. Verificada contra `sklearn` |
