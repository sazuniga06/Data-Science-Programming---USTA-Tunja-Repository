## 1. La Idea: un Árbol de Fusiones 🌳

### 1.1 Aglomerativo y divisivo

- **Aglomerativo** (*bottom-up*, HAC): empieza con **$n$ grupos de un solo punto** y, en cada paso, **fusiona los dos grupos más cercanos**, hasta quedar uno. Es el más usado.
- **Divisivo** (*top-down*): empieza con **un solo grupo** con todos los datos y lo **divide** recursivamente (DIANA, Kaufman & Rousseeuw, 1990; en la práctica, ***Bisecting K-Means***). Hay $2^{n-1}-1$ divisiones posibles de un grupo, así que se necesitan heurísticas.

Ambos producen una **jerarquía anidada**: la partición en $K$ grupos se obtiene refinando la de $K-1$. Eso permite explorar **varias escalas** del mismo conjunto (¿2 continentes o 4 países?, cuaderno 00) con **un solo cálculo**.

### 1.2 Algoritmo aglomerativo

> **Entrada:** matriz de disimilitudes $D\in\mathbb{R}^{n\times n}$ y un criterio de **enlace** $d(A,B)$ entre grupos.
> **Inicio:** $n$ grupos unitarios $\{\mathbf{x}_1\},\dots,\{\mathbf{x}_n\}$.
> **Para** $t=1,\dots,n-1$: hallar la pareja $(A,B)$ con **menor** $d(A,B)$; **fusionarla** en $A\cup B$ a la **altura** $h_t=d(A,B)$ y **actualizar** las distancias del nuevo grupo a los demás.

El resultado es la **matriz de enlace** $Z\in\mathbb{R}^{(n-1)\times4}$ que usa `scipy`: cada fila registra los **dos grupos fusionados**, la **altura** $h_t$ y el **tamaño** del nuevo grupo. (Los grupos originales se numeran $0,\dots,n-1$ y el que nace en el paso $t$ recibe el número $n+t$.)

### 1.3 Criterios de enlace (*linkage*)

El criterio responde: **¿qué significa que dos grupos estén «cerca»?**

| Enlace | $d(A,B)$ | Idea | Tendencia |
|---|---|---|---|
| ***Single*** (vecino más cercano) | $\min_{a\in A,\,b\in B}d(a,b)$ | Basta **un par** cercano | Grupos **alargados**; sufre **encadenamiento** |
| ***Complete*** (vecino más lejano) | $\max_{a\in A,\,b\in B}d(a,b)$ | **Todos** los pares deben estar cerca | Grupos **compactos** de diámetro parecido; sensible a *outliers* |
| ***Average*** (UPGMA) | $\dfrac{1}{|A||B|}\sum_{a\in A}\sum_{b\in B}d(a,b)$ | Distancia **promedio** entre pares | Compromiso entre los dos anteriores |
| ***Ward*** | $\dfrac{|A||B|}{|A|+|B|}\|\bar{\mathbf{a}}-\bar{\mathbf{b}}\|^2$ | Fusiona el par que **menos aumenta** la inercia (suma de cuadrados intra-grupo) | Grupos **esféricos y de tamaño similar** (como *K-Means*); solo con distancia Euclídea |

### 1.4 La fórmula de Lance–Williams

No hace falta recalcular las distancias entre grupos desde cero: al fusionar $i$ y $j$, la distancia del nuevo grupo a cualquier otro $k$ se obtiene **solo de las distancias ya conocidas**:

$$d(k,\,i\cup j)=\alpha_i\,d(k,i)+\alpha_j\,d(k,j)+\beta\,d(i,j)+\gamma\,\big|d(k,i)-d(k,j)\big|.$$

| Enlace | $\alpha_i$ | $\alpha_j$ | $\beta$ | $\gamma$ |
|---|---|---|---|---|
| *Single* | $\tfrac12$ | $\tfrac12$ | $0$ | $-\tfrac12$ |
| *Complete* | $\tfrac12$ | $\tfrac12$ | $0$ | $+\tfrac12$ |
| *Average* | $\tfrac{n_i}{n_i+n_j}$ | $\tfrac{n_j}{n_i+n_j}$ | $0$ | $0$ |
| *Ward* (sobre distancias **al cuadrado**) | $\tfrac{n_i+n_k}{n_i+n_j+n_k}$ | $\tfrac{n_j+n_k}{n_i+n_j+n_k}$ | $-\tfrac{n_k}{n_i+n_j+n_k}$ | $0$ |

Para *single* y *complete* la fórmula se reduce a $\min$ y $\max$ (porque $\tfrac12(a+b)\mp\tfrac12|a-b|$ es exactamente el mínimo o el máximo). En *Ward* la cantidad que se actualiza es $2\,\Delta\text{ESS}$ (el doble del aumento de la suma de cuadrados), de modo que la **altura que reporta `scipy`** es $\sqrt{2\,\Delta\text{ESS}}=\sqrt{\tfrac{2\,n_An_B}{n_A+n_B}}\;\|\bar{\mathbf{a}}-\bar{\mathbf{b}}\|$.

### 1.5 Implementación ingenua y verificación contra SciPy

Usamos el esquema más simple: en cada paso **buscar el mínimo de toda la matriz** ($O(n^2)$) y **actualizar una fila** con Lance–Williams ($O(n)$), $n-1$ veces: **$O(n^3)$ en total** y $O(n^2)$ de memoria. Verificaremos que coincide con `scipy.cluster.hierarchy.linkage` **fila por fila** (pares fusionados, alturas y tamaños). *Desempate:* si hay varias parejas a igual distancia, `argmin` toma la primera en orden por filas; con datos continuos los empates no ocurren.

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas matplotlib seaborn scikit-learn scipy

import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.spatial.distance import cdist, pdist, squareform
from scipy.cluster.hierarchy import linkage, fcluster, dendrogram, cophenet, leaves_list
from scipy.sparse.csgraph import minimum_spanning_tree
from sklearn.cluster import AgglomerativeClustering
from sklearn.datasets import make_blobs, make_moons, make_circles
from sklearn.metrics import adjusted_rand_score

warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (9, 5)
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d"]
print("Entorno listo.")
````

````python
# ============================================================
# 1. Aglomerativo ingenuo O(n^3) con Lance-Williams, verificado contra scipy.cluster.hierarchy.linkage
# ============================================================
def aglomerativo(X, metodo="single"):
    """Clustering jerarquico aglomerativo. Devuelve la matriz de enlace Z (n-1, 4) con el formato de scipy:
    [id_grupo_a, id_grupo_b, altura, tamano]. Los grupos originales son 0..n-1 y el creado en el paso t es n+t."""
    n = len(X)
    D = cdist(X, X)
    if metodo == "ward":
        D = D ** 2                               # Lance-Williams de Ward actua sobre distancias euclidianas al cuadrado
    np.fill_diagonal(D, np.inf)                  # inf en la diagonal y en las filas/columnas de grupos ya fusionados
    tam = np.ones(n)                             # tamano de cada grupo vivo
    ident = np.arange(n)                         # numero (formato scipy) del grupo que vive en cada fila
    Z = np.zeros((n - 1, 4))
    for t in range(n - 1):
        i, j = np.unravel_index(np.argmin(D), D.shape)      # pareja mas cercana (O(n^2)); desempate: primera en orden por filas
        i, j = min(i, j), max(i, j)
        h, ni, nj = D[i, j], tam[i], tam[j]
        if metodo == "single":                    # alpha=1/2, gamma=-1/2  ->  minimo
            nuevo = np.minimum(D[i], D[j])
        elif metodo == "complete":                # alpha=1/2, gamma=+1/2  ->  maximo
            nuevo = np.maximum(D[i], D[j])
        elif metodo == "average":                 # UPGMA: media ponderada por tamanos
            nuevo = (ni * D[i] + nj * D[j]) / (ni + nj)
        elif metodo == "ward":                    # Lance-Williams de Ward
            nuevo = ((ni + tam) * D[i] + (nj + tam) * D[j] - tam * h) / (ni + nj + tam)
        else:
            raise ValueError("metodo debe ser single, complete, average o ward")
        a, b = sorted((ident[i], ident[j]))
        Z[t] = [a, b, np.sqrt(h) if metodo == "ward" else h, ni + nj]   # scipy reporta la raiz para Ward
        nuevo[i] = nuevo[j] = np.inf
        D[i, :] = nuevo; D[:, i] = nuevo          # el nuevo grupo vive en la fila i
        D[j, :] = np.inf; D[:, j] = np.inf        # la fila j queda desactivada
        tam[i], ident[i] = ni + nj, len(X) + t
    return Z

# --- Verificacion en varios conjuntos y los cuatro enlaces ---
filas = []
for semilla in range(4):
    X, _ = make_blobs(n_samples=80, centers=3, cluster_std=1.6, random_state=semilla)
    for metodo in ["single", "complete", "average", "ward"]:
        Z_mio, Z_sp = aglomerativo(X, metodo), linkage(X, method=metodo)
        assert np.allclose(Z_mio[:, 2], Z_sp[:, 2]), (semilla, metodo)         # mismas alturas de fusion
        assert np.array_equal(Z_mio[:, 3], Z_sp[:, 3]), (semilla, metodo)      # mismos tamanos
        assert np.array_equal(Z_mio[:, :2], Z_sp[:, :2]), (semilla, metodo)    # mismas parejas fusionadas, en el mismo orden
        filas.append({"semilla": semilla, "enlace": metodo, "max |dif. de alturas|": np.abs(Z_mio[:, 2] - Z_sp[:, 2]).max()})
print("Aglomerativo propio == scipy.cluster.hierarchy.linkage (pares, alturas y tamaños) en 4 conjuntos x 4 enlaces.")
print("Máxima diferencia absoluta de alturas (solo error de redondeo):")
with pd.option_context("display.float_format", "{:.1e}".format):
    display(pd.DataFrame(filas).pivot(index="semilla", columns="enlace", values="max |dif. de alturas|"))
````

## 2. El Efecto del Enlace: Misma Data, Cuatro Árboles 🔗

La elección del enlace **cambia la forma de los grupos** que el árbol favorece. Cortamos cada dendrograma en el **número correcto de grupos** (`fcluster(..., criterion="maxclust")`) y medimos el acuerdo con la verdad (ARI) en tres conjuntos: grupos **gaussianos** de tamaños distintos, **lunas** y **círculos concéntricos**.

**Qué esperar** (por la definición del enlace):

- ***Ward*** y ***complete*** minimizan «diámetro» o varianza: ganan con **grupos compactos** (blobs) y **fallan** con formas no convexas.
- ***Single*** conecta por el **par más cercano**: puede **seguir** formas arbitrarias (lunas, anillos) siempre que estén **bien separadas** de las demás... pero es muy frágil frente al ruido (sección 5).
- ***Average*** queda en medio.

````python
# ============================================================
# 2. Cuatro enlaces sobre tres conjuntos (corte en el K correcto)
# ============================================================
rng_d = np.random.RandomState(SEED)
datos = {"Blobs (tamaños distintos)": make_blobs(n_samples=[200, 100, 60], centers=[[-3.5, 0], [2.5, 2.5], [2.5, -2.5]], cluster_std=[1.3, 1.0, 0.8], random_state=SEED),
         "Lunas": make_moons(n_samples=300, noise=0.06, random_state=SEED),
         "Círculos": make_circles(n_samples=300, factor=0.45, noise=0.04, random_state=SEED)}
metodos = ["single", "complete", "average", "ward"]
fig, axes = plt.subplots(3, 4, figsize=(17, 12))
tabla = {}
for r, (nombre, (X, y)) in enumerate(datos.items()):
    K = len(np.unique(y))
    for c, metodo in enumerate(metodos):
        Z = linkage(X, method=metodo)
        lab = fcluster(Z, K, criterion="maxclust")
        a = adjusted_rand_score(y, lab); tabla.setdefault(nombre, {})[metodo] = a
        for g in np.unique(lab):
            axes[r, c].scatter(*X[lab == g].T, s=9, color=PALETA[(g - 1) % len(PALETA)], alpha=0.8)
        axes[r, c].set_title(f"{metodo}  (ARI = {0.0 if abs(a) < 0.005 else a:.2f})", fontweight="bold", fontsize=10)
        axes[r, c].set_xticks([]); axes[r, c].set_yticks([])
        if c == 0:
            axes[r, c].set_ylabel(nombre, fontweight="bold")
plt.suptitle("El enlace decide qué formas de grupo puede capturar el árbol", fontweight="bold", y=1.0)
plt.tight_layout(); plt.show()
display(pd.DataFrame(tabla).T.round(2))
print("Ward/complete/average recuperan los blobs (single falla: los puntos del borde encadenan los grupos); single sigue las lunas y los círculos, que los otros enlaces cortan mal.")
assert tabla["Lunas"]["single"] > 0.95 and tabla["Círculos"]["single"] > 0.95        # single sigue formas no convexas separadas
assert tabla["Lunas"]["ward"] < 0.7 and tabla["Blobs (tamaños distintos)"]["ward"] > 0.9 > 0.3 > tabla["Blobs (tamaños distintos)"]["single"]
````

## 3. Leer un Dendrograma: Alturas y Cortes ✂️

### 3.1 Qué se ve

En el **dendrograma** las hojas son los puntos; cada **unión** de dos ramas se dibuja a la **altura** $h_t$ a la que se fusionaron. Reglas de lectura:

- La **altura de una fusión** es la disimilitud (según el enlace) a la que los dos grupos se juntan: **fusiones bajas = grupos muy parecidos**; fusiones altas = grupos muy distintos.
- La **posición horizontal** de las hojas **no** es una distancia (se puede girar cualquier rama sin cambiar el árbol); solo importan las **alturas**.
- **Una brecha larga** entre fusiones consecutivas (una rama vertical larga) indica una **separación natural**: cortar ahí da una partición estable.

### 3.2 Cortar el árbol

Hay dos formas equivalentes de obtener una partición:

1. **Por altura** $t$: se cortan todas las ramas por encima de $t$ (`fcluster(Z, t, criterion="distance")`).
2. **Por número de grupos** $K$: se cortan las $K-1$ fusiones más altas (`fcluster(Z, K, criterion="maxclust")`).

Ambas coinciden: cortar a una altura $t$ **entre** la fusión $n-K-1$ y la $n-K$ (contando desde 0 en $Z$) deja exactamente $K$ grupos. La **heurística de la mayor brecha** elige el corte en el mayor salto entre alturas consecutivas.

### 3.3 La matriz de distancias ordenada por el árbol

Reordenar las filas y columnas de $D$ según el **orden de las hojas** del dendrograma (`leaves_list`) revela el **bloque diagonal**: los grupos aparecen como cuadrados oscuros (distancias pequeñas) sobre la diagonal. Es una verificación visual muy útil de que el agrupamiento tiene sentido.

````python
# ============================================================
# 3. Dendrograma (Ward), corte por altura y por K, heuristica de la mayor brecha, matriz D reordenada
# ============================================================
X3, y3 = make_blobs(n_samples=36, centers=[[-5, 0], [0, 4], [5, 0]], cluster_std=1.1, random_state=SEED)
n = len(X3)
Z = linkage(X3, method="ward")
h = Z[:, 2]

# Mayor brecha entre alturas consecutivas -> numero de grupos sugerido
brechas = np.diff(h)
m_star = int(np.argmax(brechas))                       # el corte va entre la fusion m_star y la m_star + 1
K_sug = n - (m_star + 1)
t_corte = (h[m_star] + h[m_star + 1]) / 2
print(f"Mayor brecha: entre h = {h[m_star]:.2f} y h = {h[m_star + 1]:.2f}  ->  K sugerido = {K_sug}; altura de corte = {t_corte:.2f}")

# Cortar por altura y por numero de grupos da la MISMA particion
lab_t = fcluster(Z, t_corte, criterion="distance")
lab_k = fcluster(Z, K_sug, criterion="maxclust")
assert len(np.unique(lab_t)) == K_sug and adjusted_rand_score(lab_t, lab_k) == 1.0
print(f"fcluster por altura == fcluster por K = {K_sug}; ARI frente a la verdad: {adjusted_rand_score(y3, lab_k):.2f}")

fig = plt.figure(figsize=(16, 5.6))
ax1 = fig.add_subplot(1, 3, (1, 2))
dendrogram(Z, ax=ax1, color_threshold=t_corte, above_threshold_color="gray", leaf_font_size=8)
ax1.axhline(t_corte, color="black", ls="--", lw=1.2, label=f"corte h = {t_corte:.1f}  ->  {K_sug} grupos")
ax1.set_xlabel("Observaciones (el orden horizontal no es una distancia)"); ax1.set_ylabel("Altura de fusión (Ward)")
ax1.set_title("Dendrograma: una brecha larga marca la separación natural", fontweight="bold"); ax1.legend()
ax2 = fig.add_subplot(1, 3, 3)
orden = leaves_list(Z)
D = squareform(pdist(X3))[np.ix_(orden, orden)]
im = ax2.imshow(D, cmap="viridis_r"); plt.colorbar(im, ax=ax2, fraction=0.046).set_label("distancia Euclídea")
ax2.set_title("Matriz de distancias reordenada\n(bloques = grupos)", fontweight="bold"); ax2.set_xticks([]); ax2.set_yticks([])
plt.tight_layout(); plt.show()
````

## 4. Correlación Cofenética: ¿Qué Tan Fiel Es el Árbol? 🌲

Un dendrograma **resume** $\binom n2$ distancias en $n-1$ alturas: inevitablemente **distorsiona**. La **distancia cofenética** entre dos puntos es **la altura de la fusión en la que quedan por primera vez en el mismo grupo**:

$$c_{ij}=h\big(\text{primer grupo que contiene a } i \text{ y a } j\big).$$

Se calcula recorriendo $Z$: en cada fusión, **todos los pares** con un punto en cada rama reciben la altura de esa fusión. La **correlación cofenética** (CPCC; Sokal & Rohlf, 1962) es la correlación de Pearson entre las distancias originales $d_{ij}$ y las cofenéticas $c_{ij}$ sobre todos los pares $i<j$:

$$\text{CPCC}=\frac{\sum_{i<j}(d_{ij}-\bar d)(c_{ij}-\bar c)}{\sqrt{\sum_{i<j}(d_{ij}-\bar d)^2\;\sum_{i<j}(c_{ij}-\bar c)^2}}\in[-1,1].$$

Un valor **cercano a 1** indica que el árbol conserva bien las distancias; sirve para **comparar enlaces** sobre los mismos datos (típicamente *average* obtiene un CPCC alto). Ojo: **no** mide si los grupos son «buenos», solo la **fidelidad del árbol**. Lo implementamos y lo verificamos contra `scipy.cluster.hierarchy.cophenet`.

````python
# ============================================================
# 4. Distancia y correlacion cofenetica DESDE CERO, verificadas contra scipy.cluster.hierarchy.cophenet
# ============================================================
def matriz_cofenetica(Z, n):
    """c_ij = altura de la fusion en la que i y j quedan por primera vez en el mismo grupo."""
    C = np.zeros((n, n))
    miembros = {i: [i] for i in range(n)}
    for t, (a, b, h, _) in enumerate(Z):
        ma, mb = miembros.pop(int(a)), miembros.pop(int(b))
        C[np.ix_(ma, mb)] = h; C[np.ix_(mb, ma)] = h        # todos los pares entre las dos ramas
        miembros[n + t] = ma + mb
    return C

def cpcc(Z, X):
    n = len(X)
    iu = np.triu_indices(n, k=1)
    d = squareform(pdist(X))[iu]; c = matriz_cofenetica(Z, n)[iu]
    return float(np.corrcoef(d, c)[0, 1])

Xc, _ = make_blobs(n_samples=120, centers=4, cluster_std=1.8, random_state=3)
filas = []
for metodo in ["single", "complete", "average", "ward"]:
    Z = linkage(Xc, method=metodo)
    c_sp, coph = cophenet(Z, pdist(Xc))
    assert np.allclose(matriz_cofenetica(Z, len(Xc))[np.triu_indices(len(Xc), 1)], coph)    # misma matriz cofenetica
    assert np.isclose(cpcc(Z, Xc), c_sp)                                                  # misma correlacion
    filas.append({"enlace": metodo, "CPCC (propio)": cpcc(Z, Xc), "CPCC (scipy)": c_sp})
print("Distancias y correlación cofenética propias == scipy.cluster.hierarchy.cophenet.")
display(pd.DataFrame(filas).set_index("enlace").round(4))
print("Interpretación: el árbol 'average' suele ser el más fiel a las distancias originales; Ward y complete deforman más porque priorizan compacidad.")
````

## 5. Encadenamiento (*Chaining*), Ruido y la Conexión con el Árbol Generador Mínimo ⛓️

### 5.1 El problema del enlace *single*

Como *single* fusiona dos grupos si **existe un solo par** de puntos cercanos, una **cadena de puntos** («puente») entre dos grupos bien separados basta para **fundirlos en uno**: es el **encadenamiento**. Unos pocos puntos de ruido pueden destruir un agrupamiento que sin ellos era perfecto. Los enlaces *complete*, *average* y *Ward*, que miran **todos** los pares o la varianza, son mucho más robustos.

Lo medimos: dos grupos gaussianos bien separados a los que añadimos $m$ puntos de **puente** equiespaciados entre ellos, y registramos el ARI (sobre los puntos originales) al cortar cada árbol en 2 grupos.

### 5.2 *Single linkage* = árbol generador mínimo

El algoritmo de **Kruskal** construye el **árbol generador mínimo** (MST) de un grafo ordenando sus aristas y uniendo componentes en orden creciente de peso. El enlace *single* hace **exactamente lo mismo** sobre el grafo completo de distancias: las **alturas** de fusión de *single linkage* son los **pesos ordenados de las aristas del MST**. Esa equivalencia permite calcularlo en $O(n^2)$ y es la base de HDBSCAN (single linkage sobre la distancia de alcanzabilidad mutua, que usaremos en el [cuaderno 04](04_Metricas_de_Clustering.ipynb) para DBCV). La verificamos con `scipy.sparse.csgraph.minimum_spanning_tree`.

````python
# ============================================================
# 5. Chaining: ARI vs. numero de puntos "puente"; single linkage == MST
# ============================================================
rng = np.random.RandomState(SEED)
A = rng.normal(loc=[-3, 0], scale=0.7, size=(100, 2))
B = rng.normal(loc=[3, 0], scale=0.7, size=(100, 2))
base = np.vstack([A, B]); y_base = np.r_[np.zeros(100), np.ones(100)]

ms = [0, 1, 2, 3, 4, 6, 8, 12]
res = {m: [] for m in ["single", "complete", "average", "ward"]}
for m_pts in ms:
    puente = np.c_[np.linspace(-2, 2, m_pts + 2)[1:-1], rng.normal(scale=0.08, size=m_pts)] if m_pts > 0 else np.empty((0, 2))
    Xp = np.vstack([base, puente])
    for metodo in res:
        lab = fcluster(linkage(Xp, method=metodo), 2, criterion="maxclust")[:200]          # ARI solo sobre los 200 puntos originales
        res[metodo].append(adjusted_rand_score(y_base, lab))

fig, axes = plt.subplots(1, 2, figsize=(14, 4.8))
for k, (metodo, v) in enumerate(res.items()):
    axes[0].plot(ms, v, "o-", lw=2, color=PALETA[k], label=metodo)
axes[0].set_xlabel("Puntos de puente entre los dos grupos"); axes[0].set_ylabel("ARI al cortar en 2 grupos"); axes[0].set_ylim(-0.05, 1.05)
axes[0].set_title("El enlace single se rompe con unos pocos puntos de ruido", fontweight="bold"); axes[0].legend()
puente = np.c_[np.linspace(-2, 2, 14)[1:-1], rng.normal(scale=0.08, size=12)]
Xp = np.vstack([base, puente])
lab_s = fcluster(linkage(Xp, method="single"), 2, criterion="maxclust")
for g in np.unique(lab_s):
    axes[1].scatter(*Xp[lab_s == g].T, s=14, color=PALETA[g - 1], alpha=0.8, label=f"grupo {g} ({np.sum(lab_s == g)} puntos)")
axes[1].scatter(*puente.T, s=40, marker="x", color="black", label="puente (12 puntos)")
axes[1].set_title("Single con 12 puntos de puente: un grupo gigante y una 'migaja'", fontweight="bold"); axes[1].legend(fontsize=8)
axes[1].set_xlabel("$x_1$"); axes[1].set_ylabel("$x_2$")
plt.tight_layout(); plt.show()
print({m: [round(float(a), 2) for a in v] for m, v in res.items()})
assert res["single"][0] > 0.99 and res["single"][-1] < 0.2 and min(res["ward"]) > 0.95        # chaining: single se derrumba, ward no

# --- single linkage == pesos ordenados del MST ---
Xm = make_blobs(n_samples=150, centers=4, cluster_std=1.5, random_state=SEED)[0]
alturas_single = linkage(Xm, method="single")[:, 2]
mst = minimum_spanning_tree(squareform(pdist(Xm)))
pesos_mst = np.sort(mst.data)
assert len(pesos_mst) == len(Xm) - 1 and np.allclose(alturas_single, pesos_mst)
print(f"\nAlturas de single linkage == pesos ordenados del MST ({len(pesos_mst)} aristas): verificado.")
````

## 6. Complejidad y Escalabilidad ⏱️

| Algoritmo | Tiempo | Memoria | Comentario |
|---|---|---|---|
| **Ingenuo** (el de este cuaderno) | $O(n^3)$ | $O(n^2)$ | Didáctico |
| **Cola de prioridad** (por grupo) | $O(n^2\log n)$ | $O(n^2)$ | Clásico para enlaces generales |
| **Cadena de vecinos más cercanos** (NN-chain, Murtagh) | $O(n^2)$ | $O(n^2)$ | Para enlaces «reducibles» (*complete*, *average*, *Ward*; también *single*, que además admite la vía del MST); es el esquema que implementan `scipy` y `fastcluster` (Müllner, 2011; *citado de memoria, verificar*) |
| ***Single*** vía MST | $O(n^2)$ (o $O(n\log n)$ en baja dimensión con índices espaciales) | $O(n)$–$O(n^2)$ | Equivalencia con el MST |

La **memoria** es el verdadero límite: la matriz de distancias ocupa $\approx 8\,\tfrac{n(n-1)}{2}$ bytes en formato condensado. Para $n=10^5$ son **unos 40 GB**. Por eso el jerárquico clásico no escala a «datos grandes» y se recurre a **conectividad** (`AgglomerativeClustering(connectivity=...)`, que restringe las fusiones a vecinos de un grafo), a **BIRCH** (resume los datos en un árbol de características) o a muestrear. Medimos el tiempo de nuestra versión ingenua frente a la de `scipy`.

> ⚠️ **Inversiones.** Los enlaces *centroid* y *median* (UPGMC/WPGMC; `method="centroid"` en `scipy`) **no son monótonos**: pueden producir dendrogramas con **inversiones** (una fusión posterior a una altura *menor*). Los cuatro enlaces de este cuaderno **sí** son monótonos: las alturas nunca decrecen.

````python
# ============================================================
# 6. Tiempo: naive O(n^3) vs scipy; memoria de la matriz de distancias
# ============================================================
tam_n = [100, 200, 400, 800]
t_mio, t_sp = [], []
for n_ in tam_n:
    Xt = np.random.RandomState(0).normal(size=(n_, 2))
    t0 = time.time(); aglomerativo(Xt, "average"); t_mio.append(time.time() - t0)
    t0 = time.time(); linkage(Xt, method="average"); t_sp.append(time.time() - t0)
pend_mio = np.polyfit(np.log(tam_n), np.log(t_mio), 1)[0]; pend_sp = np.polyfit(np.log(tam_n), np.log(t_sp), 1)[0]

fig, axes = plt.subplots(1, 2, figsize=(14, 4.6))
axes[0].loglog(tam_n, t_mio, "o-", color=PALETA[1], lw=2, label=f"propio, ingenuo (pendiente ≈ {pend_mio:.1f})")
axes[0].loglog(tam_n, t_sp, "s-", color=PALETA[0], lw=2, label=f"scipy (pendiente ≈ {pend_sp:.1f})")
axes[0].set_xlabel("Número de observaciones $n$"); axes[0].set_ylabel("Tiempo (s)"); axes[0].legend()
axes[0].set_title("Enlace average: el ingenuo crece mucho más rápido", fontweight="bold")
ns = np.array([1e3, 1e4, 5e4, 1e5, 1e6])
gb = ns * (ns - 1) / 2 * 8 / 1e9
axes[1].bar([f"{int(v):,}" for v in ns], gb, color=PALETA[2]); axes[1].set_yscale("log")
axes[1].set_xlabel("Número de observaciones $n$"); axes[1].set_ylabel("Memoria de la matriz de distancias (GB, escala log)")
axes[1].set_title("La memoria cuadrática es el límite real", fontweight="bold")
for x, v in enumerate(gb):
    axes[1].text(x, v * 1.3, f"{v:,.2g} GB", ha="center", fontsize=9)
plt.tight_layout(); plt.show()
print(f"n = 800: propio = {t_mio[-1]:.2f} s | scipy = {t_sp[-1]:.3f} s. Para n = 100 000 la matriz condensada ocupa ≈ {1e5 * (1e5 - 1) / 2 * 8 / 1e9:.0f} GB.")

# Verificacion cruzada con scikit-learn: AgglomerativeClustering(ward) corta igual que fcluster
Xv = make_blobs(n_samples=200, centers=3, cluster_std=1.5, random_state=SEED)[0]
lab_sk = AgglomerativeClustering(n_clusters=3, linkage="ward").fit_predict(Xv)
lab_sp = fcluster(linkage(Xv, method="ward"), 3, criterion="maxclust")
assert adjusted_rand_score(lab_sk, lab_sp) == 1.0
print("AgglomerativeClustering(linkage='ward') produce la misma partición que fcluster sobre linkage(method='ward').")
````

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Aglomerativo** | Parte de $n$ grupos unitarios y fusiona el par más cercano $n-1$ veces; produce la matriz de enlace $Z$ y un árbol anidado; no necesita $K$ ni inicialización aleatoria (determinista) |
| **Enlaces** | *Single* (mín.), *complete* (máx.), *average* (media), *Ward* (mínimo aumento de inercia); todos se actualizan con **Lance–Williams** |
| **Verificación** | Nuestro aglomerativo $O(n^3)$ reproduce `scipy` fila por fila (pares, alturas y tamaños) en los cuatro enlaces |
| **Forma de grupos** | *Ward*/*complete* → compactos; *single* → alargados o no convexos (si están separados); *average* → intermedio |
| **Dendrograma** | La **altura** es la disimilitud de fusión; el corte por altura o por $K$ da la misma partición; una brecha larga marca una separación natural |
| **Cofenética** | CPCC = correlación entre distancias originales y cofenéticas; mide la **fidelidad del árbol**, no la calidad de los grupos |
| **Chaining** | Unos pocos puntos de «puente» funden grupos con *single*; *single* ≡ MST (alturas = pesos ordenados de sus aristas) |
| **Escala** | $O(n^2)$ de memoria y de 2 a 3 potencias de $n$ en tiempo: no escala a $n$ grande sin conectividad, BIRCH o muestreo |
