## 1. Formulación: Inercia y Centroides 🎯

### 1.1 El problema de optimización

Dados $\mathbf{x}_1,\dots,\mathbf{x}_n\in\mathbb{R}^d$ y un número de grupos $K$, *K-Means* busca una **partición** $C_1,\dots,C_K$ y unos **centroides** $\boldsymbol\mu_1,\dots,\boldsymbol\mu_K$ que minimicen la **inercia** (también llamada **WCSS**, *within-cluster sum of squares*, o distorsión):

$$J(C,\boldsymbol\mu)=\sum_{k=1}^{K}\sum_{i\in C_k}\|\mathbf{x}_i-\boldsymbol\mu_k\|^2 .$$

Es un problema **combinatorio**: hay del orden de $K^n/K!$ particiones posibles. De hecho es **NP-difícil** en general, incluso con $K=2$ (Aloise et al., 2009) o en el plano con $K$ arbitrario (Mahajan et al., 2009) *(citado de memoria, verificar)*. Por eso se usan **heurísticas** que garantizan *buenos* mínimos locales, no el óptimo global.

### 1.2 Dos subproblemas fáciles

El problema completo es difícil, pero fijando una de las dos incógnitas el otro es **trivial**:

- **Con los centroides fijos**, la mejor partición asigna cada punto a su **centroide más cercano**: $\;z_i=\arg\min_k\|\mathbf{x}_i-\boldsymbol\mu_k\|^2$ (cada término de la suma es mínimo por separado).
- **Con la partición fija**, el mejor centroide de cada grupo es su **media**. En efecto, $J_k(\boldsymbol\mu)=\sum_{i\in C_k}\|\mathbf{x}_i-\boldsymbol\mu\|^2$ es una función cuadrática convexa de $\boldsymbol\mu$ y
$$\nabla_{\boldsymbol\mu}J_k=-2\sum_{i\in C_k}(\mathbf{x}_i-\boldsymbol\mu)=\mathbf 0\ \Longrightarrow\ \boldsymbol\mu_k=\frac{1}{|C_k|}\sum_{i\in C_k}\mathbf{x}_i=\bar{\mathbf{x}}_{C_k}.$$

Alternar estos dos pasos es el **algoritmo de Lloyd** (propuesto por Lloyd en 1957 y publicado en 1982; Forgy, 1965, dio una variante equivalente; MacQueen, 1967, acuñó el nombre *k-means*).

### 1.3 Algoritmo de Lloyd

> **Entrada:** datos $X$, número de grupos $K$, centros iniciales $\boldsymbol\mu_1^{(0)},\dots,\boldsymbol\mu_K^{(0)}$.
> **Repetir** hasta que las asignaciones no cambien:
> 1. **Asignación:** $z_i^{(t)}=\arg\min_k\|\mathbf{x}_i-\boldsymbol\mu_k^{(t-1)}\|^2$ (desempate: el índice $k$ **menor**).
> 2. **Actualización:** $\boldsymbol\mu_k^{(t)}=\operatorname{media}\{\mathbf{x}_i: z_i^{(t)}=k\}$.

Costo por iteración: $O(nKd)$. Casos delicados: un grupo puede **quedar vacío** (aquí conservamos el centroide anterior; `scikit-learn` lo reubica en el punto peor representado).

### 1.4 ¿Por qué converge? La inercia nunca sube

Sea $J^{(t)}$ la inercia tras la iteración $t$. El paso de **asignación** no puede aumentarla (para cada punto elige el mínimo). El paso de **actualización** tampoco (la media minimiza la suma de cuadrados del grupo). Por tanto

$$J(z^{(t)},\boldsymbol\mu^{(t-1)})\;\ge\;J(z^{(t)},\boldsymbol\mu^{(t)})\;\ge\;J(z^{(t+1)},\boldsymbol\mu^{(t)})\;\ge\cdots\;\ge 0.$$

Como hay un número **finito** de particiones ($K^n$) y la inercia **solo se repite si la partición se repite**, el algoritmo no puede ciclar y **termina en un número finito de pasos** en un **punto fijo**: unas asignaciones que son óptimas para sus propios centroides y viceversa. Cuidado con lo que esto **no** dice: el punto fijo es un **mínimo local** (respecto a los dos pasos), **no necesariamente el global**, y depende por completo de la inicialización. Además puede existir un punto cuyo traslado individual a otro grupo reduce $J$ aunque Lloyd ya se haya detenido (de ahí la variante de Hartigan–Wong; *citado de memoria, verificar*).

Lo implementamos y comprobamos todo esto numéricamente. **Verificación contra `scikit-learn`:** si le pasamos a `KMeans` **exactamente los mismos centros iniciales** (`init=C0`, `n_init=1`), con `algorithm="lloyd"` y `tol=0` (corre hasta que las asignaciones dejen de cambiar), debe devolver las **mismas etiquetas, los mismos centroides y la misma inercia**.

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas matplotlib seaborn scikit-learn scipy

import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs, make_moons
from sklearn.cluster import KMeans, MiniBatchKMeans
from sklearn.mixture import GaussianMixture
from sklearn.metrics import adjusted_rand_score

warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (9, 5)
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d"]
print("Entorno listo.")
````

````python
# ============================================================
# 1. Algoritmo de Lloyd DESDE CERO, verificado contra sklearn.cluster.KMeans
# ============================================================
def distancias_cuadradas(X, C):
    """||x_i - c_k||^2 para todo par (i, k): matriz (n, K)."""
    return ((X[:, None, :] - C[None, :, :]) ** 2).sum(axis=2)

def inercia(X, etiquetas, C):
    """J = suma de cuadrados de cada punto a su centroide (WCSS)."""
    return float(((X - C[etiquetas]) ** 2).sum())

def kmeans_lloyd(X, C0, max_iter=300):
    """Algoritmo de Lloyd. Devuelve etiquetas, centroides, inercia, historia de J y numero de iteraciones.
    - Desempate en la asignacion: argmin toma el indice de centroide MENOR (igual que sklearn).
    - Cluster vacio: se conserva el centroide anterior (sklearn lo reubica; no ocurre en estos datos).
    - Parada: las asignaciones no cambian entre dos iteraciones consecutivas (punto fijo)."""
    C = np.array(C0, dtype=float)
    K = len(C)
    etiquetas, historia = None, []                 # historia: J tras cada semipaso (asignar, actualizar)
    for it in range(1, max_iter + 1):
        D2 = distancias_cuadradas(X, C)
        nuevas = D2.argmin(axis=1)                                           # paso 1: asignacion
        if etiquetas is not None and np.array_equal(nuevas, etiquetas):
            break                                                            # punto fijo alcanzado
        etiquetas = nuevas
        historia.append(inercia(X, etiquetas, C))                            # J con la nueva asignacion y los centroides viejos
        for k in range(K):                                                   # paso 2: actualizacion (media)
            miembros = X[etiquetas == k]
            if len(miembros) > 0:
                C[k] = miembros.mean(axis=0)
        historia.append(inercia(X, etiquetas, C))                            # J con la misma asignacion y los centroides nuevos
    return etiquetas, C, inercia(X, etiquetas, C), np.array(historia), it

# --- Verificacion: mismos centros iniciales => mismo resultado que sklearn ---
rng = np.random.RandomState(SEED)
bancos = {"blobs (K=4)": (make_blobs(n_samples=500, centers=4, cluster_std=1.2, random_state=1)[0], 4),
          "blobs (K=6)": (make_blobs(n_samples=800, centers=6, cluster_std=1.0, random_state=2)[0], 6),
          "lunas (K=2)": (make_moons(n_samples=400, noise=0.1, random_state=3)[0], 2),
          "uniforme (K=5)": (rng.uniform(size=(600, 3)), 5)}
filas = []
for nombre, (X, K) in bancos.items():
    C0 = X[rng.choice(len(X), K, replace=False)]                              # centros iniciales: K puntos al azar
    et, C, J, hist, its = kmeans_lloyd(X, C0)
    sk = KMeans(n_clusters=K, init=C0, n_init=1, algorithm="lloyd", tol=0, max_iter=300).fit(X)
    assert np.array_equal(et, sk.labels_), nombre                             # mismas etiquetas
    assert np.allclose(C, sk.cluster_centers_, atol=1e-8), nombre             # mismos centroides
    assert np.isclose(J, sk.inertia_, rtol=1e-10), nombre                     # misma inercia
    filas.append({"conjunto": nombre, "iteraciones (propias)": its, "iteraciones (sklearn)": sk.n_iter_, "inercia propia": J, "inercia sklearn": sk.inertia_})
print("Lloyd propio == sklearn.cluster.KMeans (etiquetas, centroides e inercia) en los 4 conjuntos.\n")
display(pd.DataFrame(filas).round(4))
````

````python
# ============================================================
# 1b. La inercia decrece monotonamente en CADA semipaso (asignacion y actualizacion)
# ============================================================
X, K = make_blobs(n_samples=500, centers=5, cluster_std=1.3, random_state=7)
fig, axes = plt.subplots(1, 2, figsize=(15, 4.8))
for s in range(5):                                                            # 5 inicializaciones aleatorias distintas
    C0 = X[np.random.RandomState(s).choice(len(X), 5, replace=False)]
    et, C, J, hist, its = kmeans_lloyd(X, C0)
    assert np.all(np.diff(hist) <= 1e-9), "la inercia debe ser no creciente"     # monotonia en cada semipaso
    axes[0].step(np.arange(len(hist)), hist, where="post", color=PALETA[s], lw=2, label=f"inicio {s}: J final = {J:.0f} ({its} iter.)")
axes[0].set_xlabel("Semipaso (asignar, actualizar, asignar, ...)"); axes[0].set_ylabel("Inercia $J$")
axes[0].set_title("La inercia nunca sube; cada inicio llega a un punto fijo", fontweight="bold", fontsize=10); axes[0].legend(fontsize=8)

# Cuatro instantaneas del primer inicio: centroides y asignaciones en iteraciones 1, 2, 3 y final
C = X[np.random.RandomState(0).choice(len(X), 5, replace=False)].astype(float)
trayectoria = [C.copy()]
for _ in range(30):
    et = distancias_cuadradas(X, C).argmin(axis=1)
    Cn = np.array([X[et == k].mean(axis=0) if np.any(et == k) else C[k] for k in range(5)])
    trayectoria.append(Cn.copy())
    if np.allclose(Cn, C): break
    C = Cn
axes[1].scatter(*X.T, c=[PALETA[k] for k in et], s=12, alpha=0.5)
for k in range(5):
    tray = np.array([t[k] for t in trayectoria])
    axes[1].plot(tray[:, 0], tray[:, 1], "-o", color="black", ms=4, lw=1)
    axes[1].scatter(*tray[-1], marker="X", s=190, color=PALETA[k], edgecolor="black", zorder=5)
axes[1].set_title("Trayectoria de los centroides (puntos negros) hasta el punto fijo", fontweight="bold", fontsize=10)
axes[1].set_xlabel("$x_1$"); axes[1].set_ylabel("$x_2$")
plt.tight_layout(); plt.show()
print("Aserción verificada: J(asignación t) ≥ J(actualización t) ≥ J(asignación t+1) en todas las corridas.")
````

## 2. Sensibilidad a la Inicialización y *k-means++* 🎲

### 2.1 El problema: muchos mínimos locales

Como $J$ no es convexa, **distintos centros iniciales llevan a distintos puntos fijos**, con inercias muy diferentes. La inicialización clásica elige $K$ **puntos al azar**; si dos centros caen en el mismo grupo real y ninguno en otro, Lloyd **no sabe corregirlo**. Dos remedios estándar:

- **Reinicios múltiples** (`n_init` en `scikit-learn`): correr varias veces y **quedarse con la menor inercia**.
- **Una inicialización inteligente**: *k-means++*.

### 2.2 *k-means++* (Arthur & Vassilvitskii, 2007)

Idea: **repartir los centros iniciales** favoreciendo puntos **lejanos** a los ya elegidos, pero **sin ser determinista** (para no elegir siempre *outliers*).

> 1. Elegir el primer centro $\boldsymbol\mu_1$ **uniformemente** entre los datos.
> 2. Para cada punto, sea $D(\mathbf{x})^2$ la distancia cuadrada al **centro más cercano ya elegido**.
> 3. Elegir el siguiente centro **con probabilidad proporcional a $D(\mathbf{x})^2$**: $\;P(\mathbf{x}_i)=\dfrac{D(\mathbf{x}_i)^2}{\sum_j D(\mathbf{x}_j)^2}$.
> 4. Repetir hasta tener $K$ centros y entonces correr Lloyd.

**Garantía teórica:** en esperanza, la inercia de la inicialización cumple $\;\mathbb{E}[J]\le 8(\ln K+2)\,J_{\text{óptima}}$, es decir, es $O(\log K)$-competitiva (Arthur & Vassilvitskii, 2007; *citado de memoria, verificar la constante*), mientras que la inicialización uniforme **no tiene ninguna garantía**. (`scikit-learn` usa una variante *greedy* que prueba varios candidatos por paso; la que implementamos es la versión original.)

Verificaremos dos cosas: **(i)** que nuestra regla de muestreo reproduce las probabilidades $D^2/\sum D^2$ (comprobación estadística) y **(ii)** que mejora la distribución de inercias finales frente a la inicialización aleatoria sobre un conjunto con 8 grupos que se solapan algo.

````python
# ============================================================
# 2. k-means++ DESDE CERO + sensibilidad a la inicializacion
# ============================================================
def kmeans_pp_init(X, K, rng):
    """Inicializacion k-means++ (version original): muestreo con probabilidad proporcional a D(x)^2."""
    n = len(X)
    idx = [rng.randint(n)]                                  # 1. primer centro uniforme
    d2 = ((X - X[idx[0]]) ** 2).sum(axis=1)                 # D(x)^2 al centro mas cercano
    for _ in range(1, K):
        i = rng.choice(n, p=d2 / d2.sum())                  # 3. P(x_i) = D(x_i)^2 / sum_j D(x_j)^2
        idx.append(i)
        d2 = np.minimum(d2, ((X - X[i]) ** 2).sum(axis=1))  # actualizar D(x)^2 con el nuevo centro
    return X[idx].copy()

# --- (i) Verificacion de la regla de muestreo: frecuencia empirica del 2.o centro vs. D^2 / sum D^2 ---
Xs = make_blobs(n_samples=30, centers=3, random_state=0)[0]
rng = np.random.RandomState(SEED)
primero = 4                                                 # fijamos el primer centro para aislar el 2.o sorteo
d2 = ((Xs - Xs[primero]) ** 2).sum(axis=1); p_teorica = d2 / d2.sum()
R = 60000
sorteos = rng.choice(len(Xs), size=R, p=p_teorica)          # mismo mecanismo de muestreo que usa kmeans_pp_init
frec = np.bincount(sorteos, minlength=len(Xs)) / R
print(f"Máx |frecuencia empírica - D²/ΣD²| tras {R} sorteos: {np.abs(frec - p_teorica).max():.4f}   (el punto fijado tiene prob. {p_teorica[primero]:.1f})")
assert np.abs(frec - p_teorica).max() < 0.01 and p_teorica[primero] == 0

# --- (ii) Distribucion de inercias finales: inicializacion aleatoria vs k-means++ (100 corridas cada una) ---
X8, _ = make_blobs(n_samples=600, centers=8, cluster_std=0.9, random_state=11)
K = 8
rng = np.random.RandomState(SEED)
J_azar = [kmeans_lloyd(X8, X8[rng.choice(len(X8), K, replace=False)])[2] for _ in range(100)]
J_pp = [kmeans_lloyd(X8, kmeans_pp_init(X8, K, rng))[2] for _ in range(100)]
J_sk = [KMeans(K, init="k-means++", n_init=1, random_state=s).fit(X8).inertia_ for s in range(100)]
J_mejor_de_10 = [min(kmeans_lloyd(X8, X8[rng.choice(len(X8), K, replace=False)])[2] for _ in range(10)) for _ in range(40)]

resumen = pd.DataFrame({"mín": [min(J_azar), min(J_pp), min(J_sk), min(J_mejor_de_10)],
                        "media": [np.mean(J_azar), np.mean(J_pp), np.mean(J_sk), np.mean(J_mejor_de_10)],
                        "máx": [max(J_azar), max(J_pp), max(J_sk), max(J_mejor_de_10)]},
                       index=["Aleatoria (1 corrida)", "k-means++ propio (1 corrida)", "k-means++ sklearn (1 corrida)", "Aleatoria, mejor de 10"])
display(resumen.round(1))
assert np.mean(J_pp) < np.mean(J_azar)                      # k-means++ mejora la inercia media

fig, ax = plt.subplots(figsize=(10.5, 4.6))
bins = np.linspace(min(min(J_azar), min(J_pp)) - 20, max(max(J_azar), max(J_pp)) + 20, 40)
ax.hist(J_azar, bins=bins, alpha=0.6, color=PALETA[1], label=f"Aleatoria (media {np.mean(J_azar):.0f})")
ax.hist(J_pp, bins=bins, alpha=0.6, color=PALETA[0], label=f"k-means++ propio (media {np.mean(J_pp):.0f})")
ax.axvline(min(J_pp), color="black", ls="--", lw=1.5, label=f"mejor inercia hallada: {min(J_pp):.0f}")
ax.set_xlabel("Inercia final $J$ (una corrida de Lloyd)"); ax.set_ylabel("Número de corridas")
ax.set_title("Muchos mínimos locales: la inicialización importa (K = 8, 100 corridas)", fontweight="bold")
ax.legend(); plt.tight_layout(); plt.show()
mejor = min(min(J_azar), min(J_pp))
print(f"Corridas con inercia > 25 % por encima de la mejor hallada ({mejor:.0f}): inicio aleatorio {np.mean(np.array(J_azar) > 1.25 * mejor):.0%} | k-means++ propio {np.mean(np.array(J_pp) > 1.25 * mejor):.0%} | k-means++ sklearn {np.mean(np.array(J_sk) > 1.25 * mejor):.0%}.")
````

## 3. Elegir $k$: el Método del Codo 📉

La inercia **siempre baja** al aumentar $K$ (cuaderno 00), así que no se puede minimizar para elegir $K$. El **método del codo** busca el punto donde **añadir un grupo deja de aportar una mejora grande**: se grafica $J(K)$ y se busca un «codo». Tiene dos problemas: es **subjetivo** (¿dónde está el codo?) y **a veces no existe** (datos sin estructura clara o jerárquica). Para automatizar la lectura usamos una heurística geométrica simple: el punto de la curva **más alejado de la recta** que une sus extremos (tras normalizar ambos ejes; es la idea detrás de *Kneedle*, Satopää et al., 2011, *citado de memoria, verificar*).

Alternativas más fundamentadas, que no desarrollamos aquí: **silueta** y **Calinski–Harabasz** (cuaderno 04), el ***gap statistic*** (Tibshirani, Walther & Hastie, 2001; compara $\log J(K)$ con su valor esperado bajo una distribución de referencia sin estructura), **BIC/AIC** si se usa una mezcla gaussiana, **estabilidad** del agrupamiento entre submuestras y, sobre todo, la **utilidad para el problema**.

````python
# ============================================================
# 3. Metodo del codo en dos conjuntos: uno con estructura clara y otro ambiguo
# ============================================================
def codo_por_cuerda(ks, J):
    """K cuyo punto (K, J) esta mas lejos de la recta que une los extremos (ejes normalizados)."""
    x = (np.asarray(ks) - ks[0]) / (ks[-1] - ks[0])
    y = (np.asarray(J) - J[-1]) / (J[0] - J[-1])
    dist = np.abs(x + y - 1) / np.sqrt(2)                  # distancia a la recta que une (0,1) con (1,0)
    return ks[int(np.argmax(dist))]

ks = list(range(1, 11))
conjuntos = {"4 grupos bien separados": make_blobs(n_samples=600, centers=[[-6, -4], [6, -4], [-6, 5], [6, 5]], cluster_std=1.3, random_state=SEED)[0],
             "6 grupos que se solapan": make_blobs(n_samples=600, centers=6, cluster_std=2.2, random_state=5)[0]}
fig, axes = plt.subplots(1, 2, figsize=(14, 4.6))
for ax, (nombre, X) in zip(axes, conjuntos.items()):
    J = [KMeans(k, n_init=10, random_state=SEED).fit(X).inertia_ for k in ks]
    k_codo = codo_por_cuerda(ks, J)
    ax.plot(ks, J, "o-", color=PALETA[0], lw=2)
    ax.axvline(k_codo, color=PALETA[1], ls="--", label=f"codo sugerido: K = {k_codo}")
    ax.set_xlabel("Número de grupos $K$"); ax.set_ylabel("Inercia $J(K)$"); ax.set_xticks(ks)
    ax.set_title(f"{nombre}", fontweight="bold"); ax.legend()
    print(f"{nombre:26s} -> K sugerido por el codo = {k_codo}")
plt.tight_layout(); plt.show()
print("\nEn el conjunto nítido el codo es claro; en el ambiguo la curva baja casi suavemente y el 'codo' es discutible.")
````

## 4. Limitaciones de *K-Means* ⚠️

*K-Means* hace supuestos fuertes, aunque implícitos. Conviene verlos con claridad:

1. **Fronteras lineales (celdas de Voronoi).** Un punto va al centroide más cercano, así que la frontera entre dos grupos es el **hiperplano mediatriz** de sus centroides: cada grupo es **convexo**. No puede capturar lunas, anillos ni espirales.
2. **Grupos «esféricos» y de tamaño parecido.** Minimizar la distancia cuadrática equivale a ajustar **gaussianas esféricas de igual varianza**. Con grupos alargados (anisotrópicos), de **varianzas distintas** o de **tamaños** muy diferentes, el algoritmo **corta** el grupo grande para «equilibrar» las inercias.
3. **Sensibilidad a *outliers*.** El centroide es una **media**: un punto lejano lo arrastra (y un único *outlier* extremo puede acaparar un grupo entero).
4. **Hay que fijar $K$** y el resultado depende de la **inicialización** y del **escalado** (cuaderno 00).

### 4.1 Relación con las mezclas gaussianas (GMM)

Una mezcla gaussiana modela $p(\mathbf{x})=\sum_k\pi_k\,\mathcal{N}(\mathbf{x}\mid\boldsymbol\mu_k,\Sigma_k)$ y se ajusta con el algoritmo **EM**, que asigna **responsabilidades** suaves $\gamma_{ik}=P(z_i=k\mid\mathbf{x}_i)$. Si se fuerza $\Sigma_k=\sigma^2 I$ con el mismo $\sigma$ y $\pi_k$ iguales, entonces

$$\gamma_{ik}=\frac{\exp\!\big(-\|\mathbf{x}_i-\boldsymbol\mu_k\|^2/2\sigma^2\big)}{\sum_{j}\exp\!\big(-\|\mathbf{x}_i-\boldsymbol\mu_j\|^2/2\sigma^2\big)}\ \xrightarrow[\sigma^2\to0]{}\ \begin{cases}1&k=\arg\min_j\|\mathbf{x}_i-\boldsymbol\mu_j\|\\0&\text{en otro caso}\end{cases}$$

y EM **se convierte en el algoritmo de Lloyd**: *K-Means* es el **límite duro** de una GMM esférica. Esto explica sus supuestos, y sugiere el remedio: **liberar las covarianzas** (`GaussianMixture(covariance_type="full")`) maneja grupos elípticos y de varianzas distintas. Lo comprobamos numéricamente y sobre tres conjuntos clásicos.

````python
# ============================================================
# 4. (a) K-Means como limite duro de una GMM esferica; (b) limitaciones en tres conjuntos clasicos
# ============================================================
# (a) Responsabilidades de un punto entre dos centroides cuando sigma^2 -> 0
mu = np.array([[0.0, 0.0], [3.0, 0.0]]); x = np.array([1.2, 0.0])         # x esta mas cerca del centroide 0
d2x = ((x - mu) ** 2).sum(axis=1)
print("Responsabilidades gamma_ik del punto x = (1.2, 0) segun sigma^2:")
for s2 in [4.0, 1.0, 0.25, 0.05, 0.01]:
    w = np.exp(-(d2x - d2x.min()) / (2 * s2)); gamma = w / w.sum()          # (se resta el minimo por estabilidad numerica)
    print(f"  sigma^2 = {s2:5.2f}  ->  gamma = [{gamma[0]:.4f}, {gamma[1]:.4f}]")

# (b) Tres conjuntos: lunas, grupos alargados (anisotropicos) y varianzas distintas
Xa = make_blobs(n_samples=500, centers=3, random_state=170)[0] @ np.array([[0.6, -0.6], [-0.4, 0.8]])
datos = {"Lunas": make_moons(n_samples=500, noise=0.07, random_state=SEED),
         "Alargados (anisotrópicos)": (Xa, make_blobs(n_samples=500, centers=3, random_state=170)[1]),
         "Varianzas distintas": make_blobs(n_samples=[500, 150, 80], centers=[[-5, 0], [1, 0], [5, 3]], cluster_std=[2.4, 0.9, 0.5], random_state=SEED)}
fig, axes = plt.subplots(2, 3, figsize=(16, 9))
filas = []
for j, (nombre, (X, y)) in enumerate(datos.items()):
    K = len(np.unique(y))
    km = KMeans(K, n_init=10, random_state=SEED).fit(X)
    gm = GaussianMixture(K, covariance_type="full", n_init=5, random_state=SEED).fit(X)
    lab_gm = gm.predict(X)
    # fondo: regiones de Voronoi de K-Means (fronteras lineales)
    gx, gy = np.meshgrid(np.linspace(X[:, 0].min() - 1, X[:, 0].max() + 1, 200), np.linspace(X[:, 1].min() - 1, X[:, 1].max() + 1, 200))
    zz = km.predict(np.c_[gx.ravel(), gy.ravel()]).reshape(gx.shape)
    axes[0, j].contourf(gx, gy, zz, levels=np.arange(-0.5, K + 0.5, 1), colors=PALETA[:K], alpha=0.15)
    for k in range(K):
        axes[0, j].scatter(*X[km.labels_ == k].T, s=10, color=PALETA[k], alpha=0.8)
        axes[1, j].scatter(*X[lab_gm == k].T, s=10, color=PALETA[k], alpha=0.8)
    axes[0, j].scatter(*km.cluster_centers_.T, marker="X", s=170, color="black")
    a_km, a_gm = adjusted_rand_score(y, km.labels_), adjusted_rand_score(y, lab_gm)
    axes[0, j].set_title(f"{nombre}\nK-Means (ARI = {a_km:.2f})", fontweight="bold", fontsize=10)
    axes[1, j].set_title(f"{nombre}\nMezcla gaussiana, covarianza libre (ARI = {a_gm:.2f})", fontweight="bold", fontsize=10)
    filas.append({"conjunto": nombre, "ARI K-Means": a_km, "ARI GMM": a_gm})
for ax in axes.ravel():
    ax.set_xlabel("$x_1$"); ax.set_ylabel("$x_2$")
plt.tight_layout(); plt.show()
display(pd.DataFrame(filas).round(3))
print("K-Means corta las lunas con una línea recta, parte los grupos alargados y 'recorta' el grupo ancho. La GMM con covarianza libre arregla los dos últimos casos, pero NO las lunas (no son gaussianas): para eso necesitamos densidad (cuaderno 03).")
````

````python
# ============================================================
# 4c. Sensibilidad a outliers: unos pocos puntos extremos secuestran un grupo
# ============================================================
X3, y3 = make_blobs(n_samples=450, centers=[[-5, -3], [5, -3], [0, 5]], cluster_std=1.1, random_state=SEED)
desp = np.array([[0, 0], [1, -1], [-1, 1], [0.5, 0.5], [-0.5, -0.5]])
outliers = np.array([35, 35]) + desp                                                    # 5 puntos lejanisimos
X_con = np.vstack([X3, outliers])
km_sin = KMeans(3, n_init=10, random_state=SEED).fit(X3)
km_con = KMeans(3, n_init=10, random_state=SEED).fit(X_con)
ari_sin = adjusted_rand_score(y3, km_sin.labels_)
ari_con = adjusted_rand_score(y3, km_con.labels_[:len(X3)])                           # ARI SOLO sobre los puntos legitimos

fig, axes = plt.subplots(1, 2, figsize=(13, 4.8))
for ax, X, km, titulo in [(axes[0], X3, km_sin, f"Sin outliers (ARI = {ari_sin:.2f})"), (axes[1], X_con, km_con, f"Con 5 outliers extremos (ARI sobre los legítimos = {ari_con:.2f})")]:
    for k in range(3):
        ax.scatter(*X[km.labels_ == k].T, s=12, color=PALETA[k], alpha=0.8)
    ax.scatter(*km.cluster_centers_.T, marker="X", s=190, color="black")
    ax.set_title(titulo, fontweight="bold", fontsize=10); ax.set_xlabel("$x_1$"); ax.set_ylabel("$x_2$")
plt.tight_layout(); plt.show()
# Outliers moderados: no secuestran un grupo, pero arrastran su centroide
X_mod = np.vstack([X3, np.array([22, 22]) + desp]); km_mod = KMeans(3, n_init=10, random_state=SEED).fit(X_mod)
desplaz = max(np.min(np.linalg.norm(km_sin.cluster_centers_ - c, axis=1)) for c in km_mod.cluster_centers_)
print(f"Con K = 3, los 5 outliers extremos ocupan un grupo propio y dos grupos reales quedan fusionados: la media no es robusta.")
print(f"Con outliers moderados en (22, 22) quedan absorbidos por un grupo real y arrastran su centroide {desplaz:.2f} unidades (desviación del grupo: 1.1).")
print("Remedios: eliminar/marcar outliers antes (p. ej. con DBSCAN), usar K-Medoids (centros = puntos reales, distancias no cuadráticas) o una mezcla con una componente de ruido.")
assert ari_con < ari_sin - 0.2
````

## 5. *Mini-Batch K-Means* y Otras Variantes ⚡

### 5.1 *Mini-Batch K-Means* (Sculley, 2010)

Cuando $n$ es muy grande, recorrer **todos** los datos en cada iteración es caro. *Mini-Batch K-Means* usa en cada paso una **muestra pequeña** (un *mini-lote* $M$) y actualiza los centroides **de forma incremental** con una tasa de aprendizaje que decrece con la cantidad de puntos que ha absorbido cada centro:

> Para cada $\mathbf{x}\in M$ con centro asignado $c$: $\;v_c\leftarrow v_c+1,\quad \eta=1/v_c,\quad \boldsymbol\mu_c\leftarrow(1-\eta)\,\boldsymbol\mu_c+\eta\,\mathbf{x}.$

Con $\eta=1/v_c$, $\boldsymbol\mu_c$ es **exactamente la media corriente** de todos los puntos que se le han asignado. El resultado es **un poco peor** en inercia que Lloyd completo, pero **mucho más rápido** y con **memoria acotada**. Lo programamos y comparamos con `MiniBatchKMeans` y con `KMeans`.

### 5.2 Otras variantes (solo mención)

| Variante | Qué cambia | Cuándo |
|---|---|---|
| **K-Medoids** (PAM; Kaufman & Rousseeuw) | Los centros son **puntos reales** y se minimiza la suma de distancias (no cuadradas); admite **cualquier disimilitud** | Robustez a *outliers*, datos no vectoriales (`scikit-learn-extra`, no incluido en `scikit-learn`) |
| **K-Medians** | Centro = **mediana** por coordenada; distancia Manhattan | Robustez moderada |
| **Fuzzy C-Means** | Asignación **suave** con pesos | Cuando los grupos se solapan |
| **GMM** (EM) | Covarianzas libres y asignación suave | Grupos elípticos (sección 4.1) |
| **Bisecting K-Means** (`sklearn.cluster.BisectingKMeans`) | Divide **recursivamente** el grupo de mayor inercia (jerárquico divisivo) | Jerarquía de grupos con costo de *K-Means* |

````python
# ============================================================
# 5. Mini-Batch K-Means DESDE CERO (regla de Sculley) vs. MiniBatchKMeans de sklearn vs. KMeans completo
# ============================================================
def minibatch_kmeans(X, C0, tam_lote=256, n_pasos=150, seed=0):
    """Mini-Batch K-Means: media corriente por centro con tasa eta = 1 / (conteo acumulado del centro)."""
    rng = np.random.RandomState(seed)
    C = np.array(C0, dtype=float)
    v = np.zeros(len(C))                                               # cuantos puntos ha absorbido cada centro
    for _ in range(n_pasos):
        lote = X[rng.randint(len(X), size=tam_lote)]                 # mini-lote al azar (con reemplazo: barato aun con n enorme)
        asig = distancias_cuadradas(lote, C).argmin(axis=1)           # asignaciones con los centros al INICIO del lote
        for x, c in zip(lote, asig):
            v[c] += 1
            eta = 1.0 / v[c]
            C[c] = (1 - eta) * C[c] + eta * x                          # media corriente
    return C

# Datos grandes: 500 000 puntos en 10 dimensiones, 8 grupos
Xg, _ = make_blobs(n_samples=500_000, n_features=10, centers=8, cluster_std=2.0, random_state=SEED)
K = 8
C0 = kmeans_pp_init(Xg[:5000], K, np.random.RandomState(SEED))        # k-means++ sobre una submuestra (inicio comun)

# Verificacion de la regla: con eta = 1/v el centro es exactamente la media de lo absorbido
v_demo = 0; c_demo = np.zeros(2); datos_demo = np.random.RandomState(1).normal(size=(50, 2))
for x in datos_demo:
    v_demo += 1; c_demo = (1 - 1 / v_demo) * c_demo + (1 / v_demo) * x
assert np.allclose(c_demo, datos_demo.mean(axis=0))

def inercia_de(X, C, bloque=20000):
    """Inercia de unos centros dados, por bloques (evita crear la matriz (n, K, d) completa)."""
    return float(sum(distancias_cuadradas(X[i:i + bloque], C).min(axis=1).sum() for i in range(0, len(X), bloque)))

t0 = time.time(); C_mb = minibatch_kmeans(Xg, C0); t_mb = time.time() - t0
t0 = time.time(); sk_mb = MiniBatchKMeans(K, init=C0, n_init=1, batch_size=1024, max_iter=20, random_state=SEED).fit(Xg); t_skmb = time.time() - t0
t0 = time.time(); sk_full = KMeans(K, init=C0, n_init=1, algorithm="lloyd", random_state=SEED).fit(Xg); t_full = time.time() - t0

tabla = pd.DataFrame({"inercia": [inercia_de(Xg, C_mb), sk_mb.inertia_, sk_full.inertia_],
                      "tiempo (s)": [t_mb, t_skmb, t_full]},
                     index=["Mini-Batch propio", "MiniBatchKMeans (sklearn)", "KMeans completo (Lloyd)"])
tabla["inercia relativa"] = tabla["inercia"] / tabla.loc["KMeans completo (Lloyd)", "inercia"]
display(tabla.round(4))
assert tabla.loc["Mini-Batch propio", "inercia relativa"] < 1.10        # a lo sumo ~10 % peor que Lloyd completo
print(f"Calidad: Mini-Batch queda a {100 * (tabla.iloc[0]['inercia relativa'] - 1):.2f} % (propio) y {100 * (tabla.iloc[1]['inercia relativa'] - 1):.2f} % (sklearn) de la inercia de Lloyd completo.")
print(f"Tiempo (depende del equipo): MiniBatchKMeans de sklearn = {t_skmb:.2f} s frente a {t_full:.2f} s de KMeans completo. "
      "Nuestra versión en Python puro es más lenta que la de sklearn (bucle interpretado); la ventaja real del mini-lote aparece con n de millones o con datos que no caben en memoria (partial_fit).")
````

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Objetivo** | Minimizar la inercia $J=\sum_k\sum_{i\in C_k}\|\mathbf{x}_i-\boldsymbol\mu_k\|^2$; problema NP-difícil, se resuelve con una heurística |
| **Lloyd** | Alternar **asignación** (centroide más cercano) y **actualización** (media del grupo); coincide con `KMeans` verificado con `assert` |
| **Convergencia** | $J$ no crece en ningún semipaso y hay finitas particiones: termina en un **punto fijo**, un **mínimo local**, no el global |
| **Inicialización** | Muchos mínimos locales; ***k-means++*** (muestreo $\propto D^2$, $O(\log K)$-competitivo) y **reinicios** (`n_init`) los mitigan |
| **Elegir $k$** | $J(K)$ siempre decrece: se usa el **codo** (subjetivo), silueta/CH (cuaderno 04), *gap statistic*, BIC y criterio de negocio |
| **Limitaciones** | Fronteras lineales (Voronoi), grupos esféricos de igual varianza y tamaño, sensibilidad a *outliers*, requiere $K$ y escalado |
| **GMM** | *K-Means* es el límite duro ($\sigma^2\to0$) de una GMM esférica; liberar covarianzas arregla grupos elípticos, no formas no convexas |
| **Escala** | *Mini-Batch K-Means* usa medias corrientes con $\eta=1/v_c$: mucho más rápido con poca pérdida de inercia |
