## 1. Aprender sin profesor 🧑‍🏫🚫

Imagina que te regalan **una bolsa con 150 frutas mezcladas**, sin etiquetas, y te piden: *«ordénalas en montones»*. Nadie te dice cuántos tipos hay ni cómo se llaman. Tú mirarías **qué se parece a qué**: las pequeñas y muy dulces por un lado, las grandes y poco dulces por otro...

| | **Con profesor** (supervisado) | **Sin profesor** (no supervisado) |
|---|---|---|
| Qué recibes | Datos **y** la respuesta correcta de cada uno | **Solo los datos** |
| Qué haces | Aprender a **adivinar** la respuesta | **Descubrir** grupos, patrones o rarezas |
| Cómo sabes si acertaste | Comparas con la respuesta | **No hay con qué comparar** |

Hagámoslo con un programa: le damos las frutas (solo su **dulzor** y su **tamaño**) y le pedimos que forme **3 montones**. El programa no sabe cómo se llaman las frutas.

````python
# Una bolsa de 150 frutas SIN etiqueta: de cada una solo sabemos dulzor (0-10) y tamano (cm)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

frutas, _ = make_blobs(n_samples=150, centers=[[8, 2.5], [5, 7.5], [3, 13]], cluster_std=[0.8, 0.9, 1.0], random_state=7)
tabla = pd.DataFrame(frutas, columns=["dulzor", "tamano_cm"])

# Le pedimos al programa que forme 3 montones (el nombre del algoritmo, K-Means, lo veremos en el siguiente cuaderno)
tabla["montón"] = KMeans(n_clusters=3, n_init=10, random_state=0).fit_predict(frutas)
print(tabla.groupby("montón").agg(cantidad=("dulzor", "size"), dulzor_promedio=("dulzor", "mean"), tamano_promedio=("tamano_cm", "mean")).round(1))

plt.figure(figsize=(7, 4.5))
for m, color in zip(range(3), ["#2563eb", "#dc2626", "#16a34a"]):
    sel = tabla["montón"] == m
    plt.scatter(tabla.loc[sel, "dulzor"], tabla.loc[sel, "tamano_cm"], color=color, label=f"montón {m}", alpha=0.8)
plt.xlabel("Dulzor (0 a 10)"); plt.ylabel("Tamaño (cm)"); plt.title("El programa descubrió 3 tipos de fruta sin que nadie se lo dijera", fontweight="bold")
plt.legend(); plt.show()
print("\nCada montón tiene un 'perfil': pequeñas y dulces, medianas, grandes y poco dulces. El programa nunca supo los nombres de las frutas.")
````

## 2. Las cuatro cosas que se pueden hacer 🧰

Sin respuestas, casi todo se reduce a cuatro tareas. Piensa en una **ciudad**:

| Tarea | Pregunta | Ejemplo cotidiano |
|---|---|---|
| **Agrupar** (*clustering*) | ¿Qué cosas se parecen entre sí? | Ordenar la ropa por tipo de prenda; clientes con hábitos de compra parecidos |
| **Reducir dimensiones** | ¿Puedo **resumir** muchas columnas en pocas? | Resumir un libro de 300 páginas en un afiche; pasar de 64 columnas a 2 para dibujar un mapa |
| **Estimar densidad** | ¿**Dónde** se concentran los datos? | Un mapa de calor de dónde se junta la gente en una plaza |
| **Detectar rarezas** (anomalías) | ¿Cuál **no** encaja? | El pez extraño del acuario; una compra con tarjeta que no se parece a las demás |

En este módulo nos concentramos en la **primera**: agrupar. Las otras aparecerán más adelante (la reducción de dimensiones tiene su propio módulo).

## 3. ¿Cómo sé si el resultado es bueno? 🤔

Aquí está el gran problema: **no hay hoja de respuestas**. Y hay dos trampas:

1. **Pueden existir varios agrupamientos válidos.** Ordenar la ropa «por prenda» o «por color» son ambas formas correctas; depende de **para qué** la quieras ordenar.
2. **Un programa siempre encuentra grupos**, **incluso donde no los hay**. Mira: le damos puntos tirados completamente al azar y le pedimos 3 grupos.

````python
# Puntos al azar (sin ningun grupo real) vs. puntos con grupos reales: el programa "encuentra" 3 grupos en ambos
rng = np.random.RandomState(0)
azar = rng.uniform(0, 10, size=(300, 2))
con_grupos, _ = make_blobs(n_samples=300, centers=3, cluster_std=0.9, random_state=3)

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))
for ax, datos, titulo in [(axes[0], con_grupos, "Con grupos de verdad"), (axes[1], azar, "Puro azar: ¡no hay grupos!")]:
    etiquetas = KMeans(n_clusters=3, n_init=10, random_state=0).fit_predict(datos)
    ax.scatter(datos[:, 0], datos[:, 1], c=etiquetas, cmap="tab10", s=15)
    ax.set_title(f"{titulo}\n(el programa siempre devuelve 3 grupos)", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
print("Moraleja: que el programa 'encuentre' grupos NO prueba que los grupos existan. Siempre hay que revisar si tienen sentido.")
````

## 4. Las unidades importan: la regla y la báscula 📏

Para decidir qué se parece a qué, el programa **mide distancias**. Si una columna está en **números enormes** y otra en **números pequeños**, la enorme **manda** y la pequeña casi no cuenta. Es como comparar personas por su altura medida en **milímetros** (1 750) y su edad en **años** (30): para el programa, la diferencia de alturas pesa muchísimo más que cualquier diferencia de edad, solo por la unidad.

La solución usual es **poner todas las columnas en la misma escala** (*estandarizar*). Lo probamos con 178 vinos de tres tipos (cultivares) descritos por 13 medidas químicas. Pedimos 3 grupos y medimos el **acuerdo** con los tipos reales (1 = idéntico; 0 = como adivinar al azar).

````python
# Vinos: 13 medidas quimicas, una de ellas (prolina) con numeros mucho mas grandes que las demas
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import adjusted_rand_score

vinos = load_wine()
tipos_reales = vinos.target
sin_escalar = KMeans(n_clusters=3, n_init=10, random_state=0).fit_predict(vinos.data)
escalado = KMeans(n_clusters=3, n_init=10, random_state=0).fit_predict(StandardScaler().fit_transform(vinos.data))

print(f"Acuerdo con los tipos reales SIN poner en la misma escala: {adjusted_rand_score(tipos_reales, sin_escalar):.2f}")
print(f"Acuerdo con los tipos reales poniendo todo en la misma escala: {adjusted_rand_score(tipos_reales, escalado):.2f}")
print("\nUna sola columna con números enormes (la prolina) decidía casi todo. Al igualar las escalas, el agrupamiento mejora muchísimo.")
````

## 5. ¿Qué significa «cerca»? Tres formas de medir distancia 🚶

«Cerca» depende de **cómo midas**. Imagina que quieres ir de un punto A a un punto B en una ciudad:

- **Línea recta** (distancia **euclídea**): «como vuela el pájaro». Es la más usada.
- **Por las calles** (distancia **Manhattan**): solo puedes moverte en cuadrícula, como un taxi. Suma los pasos horizontales y verticales.
- **Por dirección** (distancia **coseno**): no importa **cuán lejos**, sino **hacia dónde** apuntas, como una brújula. Dos textos, uno corto y uno largo, que hablan del mismo tema «apuntan» en la misma dirección.

## 6. Demasiadas columnas: perder a un amigo en un edificio enorme 🏢

Buscar a un amigo en **una calle** (1 dimensión) es fácil. En **una plaza** (2 dimensiones) es más difícil. En **un edificio de 100 pisos con 100 pasillos** (muchas dimensiones), casi **todos están igual de lejos de ti**: ya no hay «los cercanos» y «los lejanos». Cuando eso pasa, agrupar por distancias deja de funcionar bien. Lo medimos al final de la celda siguiente: tomamos 1 000 puntos al azar y, desde uno cualquiera, miramos **cuántas veces más lejos está el más lejano que el más cercano**.

````python
from scipy.spatial.distance import euclidean, cityblock, cosine, cdist

# --- Parte 1: tres formas de medir distancia ---
A, B = np.array([0, 0]), np.array([3, 4])
print("De A = (0, 0) a B = (3, 4):")
print(f"  En línea recta (euclídea)      : {euclidean(A, B):.1f}   (el teorema de Pitágoras: 3-4-5)")
print(f"  Por las calles (Manhattan)     : {cityblock(A, B):.1f}   (3 pasos + 4 pasos)")

texto_corto, texto_largo = np.array([1, 2, 0]), np.array([10, 20, 0])      # mismo tema (misma proporcion), distinta longitud
print("\nUn texto corto y uno 10 veces más largo sobre el mismo tema:")
print(f"  Distancia en línea recta       : {euclidean(texto_corto, texto_largo):.1f}   (parecen muy distintos...)")
print(f"  Distancia por dirección (coseno): {cosine(texto_corto, texto_largo):.1f}   (... pero apuntan exactamente al mismo lado)")

# --- Parte 2: demasiadas columnas ---
rng = np.random.RandomState(1)
print("\nColumnas | el más lejano está ... veces más lejos que el más cercano")
for columnas in [2, 10, 100, 1000]:
    puntos = rng.rand(1000, columnas)
    desde_uno = cdist(puntos[:1], puntos[1:])[0]
    print(f"{columnas:8d} | {desde_uno.max() / desde_uno.min():8.1f} veces")
print("\nCon 2 columnas hay vecinos MUY cercanos y otros MUY lejanos. Con 1000, todos están casi a la misma distancia: 'cerca' ya no significa nada.")
print("Remedios: quedarse con las columnas útiles o resumirlas en pocas (reducción de dimensiones, módulo 05).")
````

```python
personas = pd.DataFrame({"edad":   [25, 60, 27, 58],
                         "sueldo": [2_000_000, 2_100_000, 2_900_000, 3_000_000]},
                        index=["Ana", "Beto", "Carla", "Diego"])

# 1) Distancia en linea recta desde Ana (con los numeros tal cual)
dist_cruda = np.sqrt(((personas - personas.loc["Ana"]) ** 2).sum(axis=1)).drop("Ana")
mas_parecida_cruda = dist_cruda.idxmin()
print("Distancias sin escalar:\n", dist_cruda.round(0), "\n-> la mas parecida:", mas_parecida_cruda)

# 2) Misma escala: (columna - promedio) / desviacion
escalada = (personas - personas.mean()) / personas.std(ddof=0)

# 3) Distancias de Ana con los datos en la misma escala
dist_escalada = np.sqrt(((escalada - escalada.loc["Ana"]) ** 2).sum(axis=1)).drop("Ana")
mas_parecida_escalada = dist_escalada.idxmin()
print("\nDistancias en la misma escala:\n", dist_escalada.round(2), "\n-> la mas parecida:", mas_parecida_escalada)

# 4) Comprobaciones
assert mas_parecida_cruda == "Beto" and mas_parecida_escalada == "Carla"
```
