## 1. Funciones de Valor y Ecuaciones de Bellman 📐

### 1.1 Definiciones

Dada una política $\pi$ definimos dos funciones de valor:

$$v_\pi(s)=\mathbb E_\pi\!\left[G_t\mid s_t=s\right],\qquad q_\pi(s,a)=\mathbb E_\pi\!\left[G_t\mid s_t=s,\,a_t=a\right].$$

$v_\pi$ responde «¿qué tan buena es la **situación** $s$ si sigo $\pi$?» y $q_\pi$ «¿qué tan buena es la **acción** $a$ en $s$, si después sigo $\pi$?». Se relacionan por

$$v_\pi(s)=\sum_a\pi(a\mid s)\,q_\pi(s,a),\qquad q_\pi(s,a)=r(s,a)+\gamma\sum_{s'}P(s'\mid s,a)\,v_\pi(s').$$

### 1.2 Ecuación de Bellman de expectativa (derivación)

Partimos de la recursión $G_t=r_{t+1}+\gamma G_{t+1}$ y tomamos esperanza condicionada a $s_t=s$. Por la **ley de la esperanza total** condicionando a la acción $a_t$ y al siguiente estado $s_{t+1}$:

$$v_\pi(s)=\sum_a\pi(a\mid s)\sum_{s'}P(s'\mid s,a)\Big[R(s,a,s')+\gamma\,\underbrace{\mathbb E_\pi[G_{t+1}\mid s_{t+1}=s']}_{=\,v_\pi(s')}\Big].$$

El paso clave es la **propiedad de Markov**: dado $s_{t+1}=s'$, el futuro ya no depende de $(s_t,a_t)$, por lo que la esperanza de $G_{t+1}$ es simplemente $v_\pi(s')$. Análogamente, para $q$:

$$q_\pi(s,a)=\sum_{s'}P(s'\mid s,a)\Big[R(s,a,s')+\gamma\sum_{a'}\pi(a'\mid s')\,q_\pi(s',a')\Big].$$

En forma matricial, $\mathbf v_\pi=\mathbf r_\pi+\gamma P_\pi\mathbf v_\pi$, que es el sistema lineal del cuaderno anterior. Definimos el **operador de Bellman de expectativa** $(T^\pi v)(s)=\sum_a\pi(a\mid s)\big[r(s,a)+\gamma\sum_{s'}P(s'\mid s,a)v(s')\big]$: $v_\pi$ es su **punto fijo**.

### 1.3 Optimalidad y ecuación de Bellman de optimalidad (derivación)

Una política $\pi$ es **mejor o igual** que $\pi'$ si $v_\pi(s)\ge v_{\pi'}(s)$ para todo $s$. En un MDP finito existe siempre una política **óptima** $\pi_{\ast}$ (determinista y estacionaria) que es mejor o igual que todas, y todas las óptimas comparten las mismas funciones $v_{\ast}$ y $q_{\ast}$:

$$v_{\ast}(s)=\max_\pi v_\pi(s),\qquad q_{\ast}(s,a)=\max_\pi q_\pi(s,a).$$

*Derivación.* Un agente óptimo que está en $s$ debe elegir la acción que maximice el valor **de elegir $a$ y luego actuar óptimamente**, es decir, $v_{\ast}(s)=\max_a q_{\ast}(s,a)$. Sustituyendo $q_{\ast}(s,a)=r(s,a)+\gamma\sum_{s'}P(s'\mid s,a)v_{\ast}(s')$ se obtienen las dos formas de la **ecuación de Bellman de optimalidad**:

$$\boxed{v_{\ast}(s)=\max_a\Big[r(s,a)+\gamma\sum_{s'}P(s'\mid s,a)\,v_{\ast}(s')\Big]},\qquad \boxed{q_{\ast}(s,a)=r(s,a)+\gamma\sum_{s'}P(s'\mid s,a)\,\max_{a'}q_{\ast}(s',a')}.$$

**Diferencia fundamental:** la de expectativa es **lineal** (un sistema de ecuaciones); la de optimalidad tiene un $\max$ y es **no lineal**, sin solución en forma cerrada. Una vez conocida $q_{\ast}$, la política óptima es **voraz**: $\pi_{\ast}(s)=\arg\max_a q_{\ast}(s,a)$ (¡sin necesidad del modelo!, razón por la que aprender $q_{\ast}$ es tan atractivo).

### 1.4 El operador de optimalidad es una contracción

Sea $(T^{\ast}v)(s)=\max_a\big[r(s,a)+\gamma\sum_{s'}P(s'\mid s,a)v(s')\big]$. Como $|\max_a x_a-\max_a y_a|\le\max_a|x_a-y_a|$ y $P(\cdot\mid s,a)$ es una distribución,

$$\|T^{\ast}v-T^{\ast}w\|_\infty\;\le\;\gamma\,\|v-w\|_\infty .$$

Por el **teorema del punto fijo de Banach**: $T^{\ast}$ tiene **un único punto fijo** $v_{\ast}$ y la iteración $v_{k+1}=T^{\ast}v_k$ **converge** a él desde **cualquier** $v_0$, con $\|v_k-v_{\ast}\|_\infty\le\gamma^k\|v_0-v_{\ast}\|_\infty$. (El mismo argumento sirve para $T^\pi$.) Esta es la base de la **iteración de valor** y de la convergencia de Q-learning.

````python
# ============================================================
# Configuracion inicial
# ============================================================
# (numpy, pandas, matplotlib y seaborn vienen preinstalados en Google Colab; este cuaderno solo usa NumPy: no necesita PyTorch ni Gymnasium)

import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import random as _random
from bisect import bisect_right
from IPython.display import display

warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (9, 5)
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"NumPy {np.__version__}")
````

````python
# ============================================================
# Un MDP finito DESDE CERO: tensores P[s, a, s'] y R[s, a, s']
# ============================================================
MAPAS = {
    "4x4": ["SFFF", "FHFH", "FFFH", "HFFG"],
    "8x8": ["SFFFFFFF", "FFFFFFFF", "FFFHFFFF", "FFFFFHFF", "FFFHFFFF", "FHHFFFHF", "FHFFHFHF", "FFFHFFFG"],
}
MOV = {0: (0, -1), 1: (1, 0), 2: (0, 1), 3: (-1, 0)}      # 0 = Izquierda, 1 = Abajo, 2 = Derecha, 3 = Arriba (convencion de Gymnasium)
NOMBRE_ACCION = {0: "Izquierda", 1: "Abajo", 2: "Derecha", 3: "Arriba"}


class MDP:
    """MDP finito (S, A, P, R, gamma) con estado inicial fijo y estados terminales absorbentes.

    P[s, a, s2] = probabilidad de pasar a s2 tras hacer a en s;  R[s, a, s2] = recompensa de esa transicion.
    """
    def __init__(self, P, R, gamma, s0, terminales):
        self.P, self.R, self.gamma, self.s0 = P, R, gamma, s0
        self.terminales = np.asarray(terminales, dtype=bool)
        self.nS, self.nA = P.shape[0], P.shape[1]
        self.r = (P * R).sum(axis=2)                            # r(s, a) = E[R | s, a]
        self.cdf = np.cumsum(P, axis=2); self.cdf[..., -1] = 1.0   # para muestrear s2 con una sola uniforme

    def reset(self):
        return self.s0

    def step(self, s, a, rng):
        """Muestrea (s2, recompensa, terminado) segun la dinamica del MDP."""
        s2 = min(int(np.searchsorted(self.cdf[s, a], rng.random(), side="right")), self.nS - 1)
        return s2, self.R[s, a, s2], bool(self.terminales[s2])


def frozenlake(mapa="4x4", resbaladizo=True, gamma=0.99):
    """Construye FrozenLake como MDP explicito. Resbaladizo: la accion elegida ocurre con prob. 1/3 y con 1/3 cada
    una de las dos direcciones perpendiculares. Chocar con el borde = quedarse. H y G son terminales; recompensa 1 al entrar a G."""
    filas = MAPAS[mapa]; n = len(filas); nS = n * n
    P = np.zeros((nS, 4, nS)); R = np.zeros((nS, 4, nS)); term = np.zeros(nS, dtype=bool)
    for f in range(n):
        for c in range(n):
            s = f * n + c
            if filas[f][c] in "HG":                              # terminal absorbente: se queda y no da mas recompensa
                term[s] = True; P[s, :, s] = 1.0; continue
            for a in range(4):
                direcciones = [(a - 1) % 4, a, (a + 1) % 4] if resbaladizo else [a]
                for d in direcciones:
                    f2 = min(max(f + MOV[d][0], 0), n - 1); c2 = min(max(c + MOV[d][1], 0), n - 1)
                    s2 = f2 * n + c2
                    P[s, a, s2] += 1.0 / len(direcciones)
                    R[s, a, s2] = float(filas[f2][c2] == "G")
    return MDP(P, R, gamma, 0, term)


def evaluar_exacto(mdp, pi, gamma=None):
    """V^pi exacto resolviendo (I - gamma P_pi) V = r_pi. pi: matriz (nS, nA). V(terminal) = 0."""
    g = mdp.gamma if gamma is None else gamma
    Ppi = np.einsum("sa,sat->st", pi, mdp.P); rpi = (pi * mdp.r).sum(axis=1)
    Ppi[mdp.terminales] = 0.0; rpi[mdp.terminales] = 0.0       # el valor de un estado terminal es 0
    return np.linalg.solve(np.eye(mdp.nS) - g * Ppi, rpi)

# ============================================================
# Utilidad de dibujo: mapas con valores y flechas de politica
# ============================================================
COLOR_CASILLA = {"S": "#fde68a", "F": "#e0f2fe", "H": "#1e293b", "G": "#86efac", "C": "#fecaca", "W": "#94a3b8"}
FLECHA = {0: "←", 1: "↓", 2: "→", 3: "↑"}

def pintar_mapa(ax, mapa, V=None, accion=None, titulo="", etiquetas=None, fmt="{:.2f}", vmax=None):
    """Dibuja una cuadricula. V: valores por estado (colorea y escribe); accion: accion (indice) por estado -> flecha."""
    nf, nc = len(mapa), len(mapa[0]); etiquetas = etiquetas or {}
    vmax = vmax if vmax is not None else (np.nanmax(np.abs(V)) if V is not None else 1.0)
    cmap = plt.get_cmap("Blues")
    for f in range(nf):
        for c in range(nc):
            s, ch = f * nc + c, mapa[f][c]
            color = COLOR_CASILLA[ch]
            if V is not None and ch in "SF" and vmax > 0:
                color = cmap(0.08 + 0.8 * min(max(V[s] / vmax, 0), 1))
            ax.add_patch(plt.Rectangle((c - 0.5, f - 0.5), 1, 1, facecolor=color, edgecolor="white", lw=2))
            oscuro = ch in "H" or (V is not None and ch in "SF" and V[s] / vmax > 0.6)
            if ch in "HGWC" or ch in etiquetas:
                ax.text(c, f, etiquetas.get(ch, ch), ha="center", va="center", fontsize=13, fontweight="bold", color="white" if ch == "H" else "#0f172a")
                continue
            if V is not None:
                ax.text(c, f + (0.22 if accion is not None else 0), fmt.format(V[s]), ha="center", va="center", fontsize=8.5, color="white" if oscuro else "#0f172a")
            if accion is not None:
                ax.text(c, f - (0.12 if V is not None else 0), FLECHA[int(accion[s])], ha="center", va="center", fontsize=17, color="white" if oscuro else "#b91c1c")
            if ch == "S" and V is None and accion is None:
                ax.text(c, f, "S", ha="center", va="center", fontsize=13, fontweight="bold")
    ax.set_xlim(-0.5, nc - 0.5); ax.set_ylim(nf - 0.5, -0.5); ax.set_aspect("equal")
    ax.set_xticks([]); ax.set_yticks([]); ax.grid(False); ax.set_title(titulo, fontweight="bold", fontsize=10)

# ------------------------------------------------------------
# Entorno de trabajo: FrozenLake 4x4 resbaladizo (gamma = 0.99)
# ------------------------------------------------------------
mdp = frozenlake("4x4", resbaladizo=True, gamma=0.99)
nt = ~mdp.terminales                                                   # mascara de estados no terminales
pi_aleatoria = np.full((mdp.nS, mdp.nA), 1 / mdp.nA)
print(f"MDP listo: |S| = {mdp.nS}, |A| = {mdp.nA}, gamma = {mdp.gamma}")
````

## 2. Verificar Bellman en Código ✅

Antes de construir algoritmos, comprobamos las identidades con el MDP de *FrozenLake* y la política aleatoria:

1. $v_\pi=\sum_a\pi(a\mid s)\,q_\pi(s,a)$ con $q_\pi=r+\gamma Pv_\pi$.
2. $v_\pi$ es el **punto fijo** del operador $T^\pi$ (residuo de Bellman nulo).
3. $T^{\ast}$ y $T^\pi$ son **contracciones** de módulo $\gamma$: probamos la desigualdad $\|Tv-Tw\|_\infty\le\gamma\|v-w\|_\infty$ con **miles de pares aleatorios** $(v,w)$.

````python
# ============================================================
# 2. Identidades de Bellman y contraccion del operador
# ============================================================
def q_desde_v(mdp, V):
    """q(s, a) = r(s, a) + gamma * sum_s' P(s'|s,a) V(s')   -> matriz (nS, nA)."""
    return mdp.r + mdp.gamma * mdp.P @ V

def T_pi(mdp, V, pi):
    """Operador de Bellman de expectativa: (T^pi V)(s) = sum_a pi(a|s) q(s, a)."""
    return (pi * q_desde_v(mdp, V)).sum(axis=1)

def T_opt(mdp, V):
    """Operador de Bellman de optimalidad: (T* V)(s) = max_a q(s, a)."""
    return q_desde_v(mdp, V).max(axis=1)

V_pi = evaluar_exacto(mdp, pi_aleatoria)
Q_pi = q_desde_v(mdp, V_pi)
assert np.allclose((pi_aleatoria * Q_pi).sum(axis=1), V_pi)                # v = sum_a pi q
assert np.abs(T_pi(mdp, V_pi, pi_aleatoria) - V_pi).max() < 1e-12            # v_pi es punto fijo de T^pi
print("Residuo de Bellman de expectativa  max|T^pi v_pi - v_pi| =", f"{np.abs(T_pi(mdp, V_pi, pi_aleatoria) - V_pi).max():.1e}")

# Contraccion: ||T v - T w||_inf <= gamma ||v - w||_inf para pares (v, w) arbitrarios
rng = np.random.default_rng(SEED)
razones_opt, razones_pi = [], []
for _ in range(5000):
    v, w = rng.normal(0, 3, mdp.nS), rng.normal(0, 3, mdp.nS)
    d = np.abs(v - w).max()
    razones_opt.append(np.abs(T_opt(mdp, v) - T_opt(mdp, w)).max() / d)
    razones_pi.append(np.abs(T_pi(mdp, v, pi_aleatoria) - T_pi(mdp, w, pi_aleatoria)).max() / d)
assert max(razones_opt) <= mdp.gamma + 1e-12 and max(razones_pi) <= mdp.gamma + 1e-12
print(f"Contraccion (5000 pares aleatorios): max ||T*v - T*w||/||v - w|| = {max(razones_opt):.4f} <= gamma = {mdp.gamma};  "
      f"para T^pi: {max(razones_pi):.4f} <= gamma")

# Una mirada a q_pi en algunas casillas (accion = Izquierda, Abajo, Derecha, Arriba)
tabla = pd.DataFrame(Q_pi[[0, 6, 10, 13, 14]], index=[f"s={s}" for s in [0, 6, 10, 13, 14]], columns=[NOMBRE_ACCION[a] for a in range(4)]).round(4)
display(tabla)
print("Lectura: en s=14 (junto a la meta) 'Izquierda' es la peor accion: ninguna de sus tres direcciones de deslizamiento puede llegar a G;")
print("         las otras tres dejan 1/3 de probabilidad de llegar y difieren en donde dejan al agente cuando resbala hacia otro lado.")
````

residuo_der = np.abs(T_pi(mdp, V_der, pi_der) - V_der).max()

# 3) Bellman a mano en el estado 10 (la politica solo elige la accion 2)
v10_manual = sum(mdp.P[10, 2, s2] * (mdp.R[10, 2, s2] + mdp.gamma * V_der[s2]) for s2 in range(mdp.nS))

# 4) q_pi y comprobaciones
Q_der = q_desde_v(mdp, V_der)
print(f"residuo de Bellman = {residuo_der:.1e} | v(10) exacto = {V_der[10]:.5f} | v(10) a mano = {v10_manual:.5f}")
print("q(10, .) =", {NOMBRE_ACCION[a]: round(float(Q_der[10, a]), 4) for a in range(4)})
assert residuo_der < 1e-10
assert np.isclose(v10_manual, V_der[10])
assert np.isclose(Q_der[10, 2], V_der[10])            # para la accion que elige la politica, q = v
assert Q_der[10].max() > V_der[10] + 1e-6              # otra accion es mejor: 'siempre Derecha' no es optima
print("v_pi cumple Bellman y en s=10 hay una accion mejor que seguir la politica: hay margen de mejora.")

## 3. Programación Dinámica: Resolver un MDP con el Modelo Conocido 🧮

Si conocemos $(P,R)$, resolvemos el MDP **sin interactuar**. La **programación dinámica (DP)** aplica las ecuaciones de Bellman como **reglas de actualización**:

| Algoritmo | Idea | Actualización |
|---|---|---|
| **Evaluación iterativa de política** | Aplicar $T^\pi$ hasta el punto fijo | $v_{k+1}=T^\pi v_k$ |
| **Mejora de política** | Volverse **voraz** respecto a $v_\pi$ | $\pi'(s)=\arg\max_a q_\pi(s,a)$ |
| **Iteración de política** | Alternar evaluación y mejora hasta que no cambie | $\pi_0\to v_{\pi_0}\to\pi_1\to v_{\pi_1}\to\cdots\to\pi_{\ast}$ |
| **Iteración de valor** | Aplicar $T^{\ast}$ directamente | $v_{k+1}=T^{\ast}v_k$ |

**Teorema de mejora de política.** Si $q_\pi(s,\pi'(s))\ge v_\pi(s)$ para todo $s$, entonces $v_{\pi'}\ge v_\pi$. En particular, la política voraz $\pi'$ respecto a $v_\pi$ es **al menos tan buena** que $\pi$; y si $\pi'=\pi$, entonces $v_\pi$ cumple la ecuación de optimalidad y **$\pi$ es óptima**. Como hay un número finito de políticas deterministas, la iteración de política **termina** en un número finito de iteraciones.

**Iteración de valor** equivale a una evaluación **truncada a un solo barrido** combinada con la mejora. Su error decae geométricamente: $\|v_k-v_{\ast}\|_\infty\le\gamma^k\|v_0-v_{\ast}\|_\infty$ (§1.4). Con $\gamma=0.99$ eso son **cientos** de barridos para lograr precisión de máquina; la iteración de política suele necesitar **muy pocas** iteraciones (pero cada evaluación es más cara).

Las **verificaciones** que haremos son todas **exactas**: (i) la evaluación iterativa converge a la solución del sistema lineal, (ii) iteración de política e iteración de valor dan **la misma $v_{\ast}$**, (iii) $v_{\ast}$ cumple la ecuación de optimalidad, (iv) la política voraz resultante **alcanza** $v_{\ast}$ y (v) el error de la iteración de valor respeta la **cota $\gamma^k$**.

````python
# ============================================================
# 3. Programacion dinamica DESDE CERO
# ============================================================
def evaluacion_iterativa(mdp, pi, tol=1e-12, max_it=100000):
    """Evaluacion de politica por barridos v <- T^pi v. Devuelve (V, numero de barridos)."""
    V = np.zeros(mdp.nS)
    for k in range(1, max_it + 1):
        V_nuevo = T_pi(mdp, V, pi)
        delta = np.abs(V_nuevo - V).max(); V = V_nuevo
        if delta < tol:
            return V, k
    return V, max_it

def iteracion_politica(mdp):
    """Iteracion de politica: evaluar (exacto) -> mejorar (voraz), hasta que la politica no cambie."""
    pol = np.zeros(mdp.nS, dtype=int)                                   # politica inicial: siempre 'Izquierda'
    for it in range(1, 1000):
        V = evaluar_exacto(mdp, np.eye(mdp.nA)[pol])
        Q = q_desde_v(mdp, V)
        # cambia de accion solo si la mejora es real (evita bucles por empates numericos)
        nueva = np.where(Q.max(axis=1) > Q[np.arange(mdp.nS), pol] + 1e-10, Q.argmax(axis=1), pol)
        if np.array_equal(nueva, pol):
            return pol, V, Q, it
        pol = nueva
    raise RuntimeError("no convergio")

def iteracion_valor(mdp, tol=1e-12, max_it=100000):
    """Iteracion de valor: v <- T* v. Devuelve (V*, Q*, politica voraz, historial de V)."""
    V = np.zeros(mdp.nS); hist = [V.copy()]
    for k in range(1, max_it + 1):
        V_nuevo = T_opt(mdp, V)
        delta = np.abs(V_nuevo - V).max(); V = V_nuevo; hist.append(V.copy())
        if delta < tol:
            break
    Q = q_desde_v(mdp, V)
    return V, Q, Q.argmax(axis=1), hist

# --- (i) Evaluacion iterativa == solucion exacta (politica aleatoria)
V_it, barridos = evaluacion_iterativa(mdp, pi_aleatoria)
assert np.abs(V_it - V_pi).max() < 1e-9
print(f"(i)   Evaluacion iterativa: {barridos} barridos; max|V_iter - V_exacto| = {np.abs(V_it - V_pi).max():.1e}")

# --- (ii) Iteracion de politica vs iteracion de valor
pol_ip, V_ip, Q_ip, n_ip = iteracion_politica(mdp)
V_iv, Q_iv, pol_iv, hist_iv = iteracion_valor(mdp)
assert np.abs(V_ip - V_iv).max() < 1e-9 and np.abs(Q_ip - Q_iv).max() < 1e-9
print(f"(ii)  Iteracion de politica: {n_ip} iteraciones | iteracion de valor: {len(hist_iv) - 1} barridos | max|V*_IP - V*_IV| = {np.abs(V_ip - V_iv).max():.1e}")

# --- (iii) Ecuacion de optimalidad y (iv) la politica voraz alcanza V*
assert np.abs(T_opt(mdp, V_iv) - V_iv).max() < 1e-10
V_voraz = evaluar_exacto(mdp, np.eye(mdp.nA)[pol_iv])
assert np.abs(V_voraz - V_iv).max() < 1e-9
assert (V_iv >= V_pi - 1e-12).all()                                       # V* domina a la politica aleatoria en todos los estados
print(f"(iii) Residuo de la ecuacion de optimalidad: {np.abs(T_opt(mdp, V_iv) - V_iv).max():.1e};  (iv) V de la politica voraz == V*: {np.abs(V_voraz - V_iv).max():.1e}")

# --- (v) cota geometrica ||V_k - V*|| <= gamma^k ||V_0 - V*||
err_k = np.array([np.abs(h - V_iv).max() for h in hist_iv])
assert (err_k[1:] <= mdp.gamma ** np.arange(1, len(err_k)) * err_k[0] + 1e-12).all()
print(f"(v)   Cota de contraccion respetada en los {len(err_k) - 1} barridos. V*(inicio) = {V_iv[mdp.s0]:.4f} frente a {V_pi[mdp.s0]:.4f} de la politica aleatoria.")

# Probabilidad de llegar a la meta con la politica optima (sin descuento: gamma = 1)
p_opt = evaluar_exacto(mdp, np.eye(mdp.nA)[pol_iv], gamma=1.0)[0]
p_rand = evaluar_exacto(mdp, pi_aleatoria, gamma=1.0)[0]
print(f"Probabilidad de alcanzar la meta: politica optima = {p_opt:.3f} | aleatoria = {p_rand:.3f}")
assert p_opt > 0.7 and p_rand < 0.02
````

## 4. Sin Modelo: Predicción por Monte Carlo y por Diferencia Temporal TD(0) 🎲➡️🧩

Si **no** conocemos $(P,R)$ solo podemos **muestrear** experiencia. Primero el problema de **predicción**: estimar $v_\pi$ de una política **fija** (la aleatoria) con dos filosofías.

**Monte Carlo (MC).** Esperar al **final del episodio** y promediar el retorno observado: $v_\pi(s)\approx$ media de los $G_t$ de cada visita (primera visita, aquí). No tiene sesgo, pero tiene **alta varianza** (un retorno depende de **toda** la trayectoria) y solo aprende al terminar un episodio.

**Diferencia temporal, TD(0).** Aprender **en cada paso** *bootstrappeando*: se usa la **estimación actual** del siguiente estado en lugar del retorno completo,

$$v(s_t)\leftarrow v(s_t)+\alpha\,\underbrace{\big[r_{t+1}+\gamma\,v(s_{t+1})-v(s_t)\big]}_{\text{error TD }\delta_t}.$$

El **objetivo TD** $r_{t+1}+\gamma v(s_{t+1})$ es una muestra de $T^\pi v$, así que TD es un **método de punto fijo estocástico** (la versión muestral de la evaluación iterativa). Tiene **sesgo** (el objetivo usa una estimación aún imperfecta) pero **baja varianza** y funciona en tareas continuas.

| | **Monte Carlo** | **TD(0)** |
|---|---|---|
| Objetivo | $G_t$ (retorno real completo) | $r_{t+1}+\gamma v(s_{t+1})$ (*bootstrapping*) |
| Sesgo | **Ninguno** | Sí (se desvanece al converger) |
| Varianza | **Alta** | Baja |
| Actualiza | Al terminar el episodio | **En cada paso** |
| Tareas continuas | No directamente | **Sí** |
| Hiperparámetro | — (promedio muestral $1/N$) | **$\alpha$** (con $\alpha$ constante nunca converge del todo: oscila alrededor del valor) |

Comparamos el error RMS frente a $v_\pi$ **exacta** (§2) en función del número de episodios, con **10 repeticiones** independientes por método (15 000 episodios). Cada episodio empieza en un estado no terminal elegido al azar para cubrir todos los estados.

````python
# ============================================================
# 4. Prediccion: Monte Carlo (primera visita) vs TD(0), contra el V_pi exacto
# ============================================================
def tablas_rapidas(mdp):
    """Listas de Python (mas rapidas que NumPy para pasos sueltos): CDF de transicion, recompensas y terminales."""
    cdf = [[mdp.cdf[s, a].tolist() for a in range(mdp.nA)] for s in range(mdp.nS)]
    return cdf, mdp.R.tolist(), mdp.terminales.tolist()

def prediccion(mdp, V_ref, metodo, episodios, alpha=0.05, semilla=0, cada=100, max_pasos=100):
    """Evalua la politica aleatoria uniforme con 'mc' o 'td'. Devuelve el error RMS (vs V_ref, estados no terminales) cada 'cada' episodios."""
    rnd = _random.Random(semilla); rand = rnd.random
    cdf, R, term = tablas_rapidas(mdp)
    nS, nA, gamma = mdp.nS, mdp.nA, mdp.gamma
    iniciales = [s for s in range(nS) if not term[s]]
    V = [0.0] * nS; N = [0] * nS; errores = []
    for ep in range(1, episodios + 1):
        s = iniciales[int(rand() * len(iniciales))]
        estados, recs = [], []
        for _ in range(max_pasos):
            a = int(rand() * nA)                                           # politica aleatoria uniforme
            s2 = min(bisect_right(cdf[s][a], rand()), nS - 1)
            r = R[s][a][s2]
            if metodo == "td":
                V[s] += alpha * (r + gamma * V[s2] - V[s])                 # error TD (V de terminales = 0)
            estados.append(s); recs.append(r); s = s2
            if term[s2]:
                break
        if metodo == "mc":
            G, primera = 0.0, {}
            for t in range(len(estados) - 1, -1, -1):                      # retornos hacia atras; la ultima asignacion es la PRIMERA visita
                G = recs[t] + gamma * G
                primera[estados[t]] = G
            for st, g in primera.items():
                N[st] += 1; V[st] += (g - V[st]) / N[st]                   # media incremental
        if ep % cada == 0:
            errores.append(float(np.sqrt(np.mean((np.array(V)[nt] - V_ref[nt]) ** 2))))
    return np.array(errores)

EPIS_PRED, REPS = 15000, 10
t0 = time.time()
curvas = {}
for nombre, metodo, alpha in [("Monte Carlo (primera visita)", "mc", None), ("TD(0), alpha = 0.01", "td", 0.01), ("TD(0), alpha = 0.05", "td", 0.05), ("TD(0), alpha = 0.2", "td", 0.2)]:
    curvas[nombre] = np.array([prediccion(mdp, V_pi, metodo, EPIS_PRED, alpha=alpha, semilla=k) for k in range(REPS)])
print(f"Prediccion ({REPS} repeticiones x 4 metodos x {EPIS_PRED} episodios): {time.time() - t0:.1f} s")

eje = np.arange(1, EPIS_PRED // 100 + 1) * 100
fig, ax = plt.subplots(figsize=(10, 5))
for (nombre, c), color in zip(curvas.items(), [PALETA[0], PALETA[2], PALETA[3], PALETA[1]]):
    m, sd = c.mean(axis=0), c.std(axis=0)
    ax.plot(eje, m, color=color, lw=2.3, label=nombre); ax.fill_between(eje, m - sd, m + sd, color=color, alpha=0.15)
ax.set_xlabel("Episodios"); ax.set_ylabel(r"Error RMS de $\hat v$ frente a $v_\pi$ exacta (media ± desv. entre 10 repeticiones)")
ax.set_yscale("log"); ax.set_title("Predicción sin modelo: MC y TD(0) convergen al valor exacto", fontweight="bold"); ax.legend(fontsize=9)
plt.tight_layout(); plt.show()

fin = {n: c[:, -1].mean() for n, c in curvas.items()}
inicio = {n: c[:, 0].mean() for n, c in curvas.items()}
for n in curvas:
    print(f"  {n:<30} error RMS: tras 100 episodios = {inicio[n]:.4f} | tras {EPIS_PRED} episodios = {fin[n]:.4f}")
assert fin["Monte Carlo (primera visita)"] < 0.01 and fin["TD(0), alpha = 0.01"] < 0.02            # ambos convergen cerca del valor exacto
assert fin["TD(0), alpha = 0.2"] > fin["TD(0), alpha = 0.05"] > fin["TD(0), alpha = 0.01"]         # el 'piso de ruido' de TD crece con alpha constante
print("Lectura: con alpha constante TD(0) se estanca en un 'piso' de ruido que CRECE con alpha (con 0.2 ni siquiera mejora: oscila); MC promedia con 1/N y sigue bajando.")
````

## 5. Control sin Modelo: SARSA y Q-learning 🎯

Para **controlar** (hallar la política óptima) aprendemos directamente $q(s,a)$, pues la política voraz respecto a $q$ **no necesita el modelo**. Dada una transición $(s,a,r,s')$ y la acción siguiente $a'$:

$$\textbf{SARSA:}\quad q(s,a)\leftarrow q(s,a)+\alpha\Big[r+\gamma\,q(s',a')-q(s,a)\Big]\qquad(a'\text{ la elige la política que se está ejecutando})$$

$$\textbf{Q-learning:}\quad q(s,a)\leftarrow q(s,a)+\alpha\Big[r+\gamma\,\max_{a'}q(s',a')-q(s,a)\Big]$$

SARSA debe su nombre a la quíntupla $(s,a,r,s',a')$. Son idénticos salvo el **objetivo**:

- **SARSA es *on-policy***: evalúa y mejora la política que **realmente ejecuta** (la $\varepsilon$-*greedy*, con sus tropiezos exploratorios incluidos). Aprende el valor de ser *«óptimo pero explorador»*.
- **Q-learning es *off-policy***: **aprende sobre la política voraz** (el $\max$) mientras **se comporta** con otra (la $\varepsilon$-*greedy*). Su objetivo es una muestra del operador $T^{\ast}$ y por eso apunta directamente a $q_{\ast}$, **sin importar cómo explore** (con tal de que visite todo par $(s,a)$ infinitas veces).

**Convergencia.** Watkins & Dayan (1992) demostraron que Q-learning tabular converge a $q_{\ast}$ con probabilidad 1 si todo par $(s,a)$ se visita infinitas veces y las tasas $\alpha_t$ cumplen las **condiciones de Robbins–Monro**: $\sum_t\alpha_t=\infty$ y $\sum_t\alpha_t^2<\infty$ (p. ej. $\alpha=1/N(s,a)^{p}$ con $p\in(0.5,1]$). SARSA converge a $q_{\ast}$ si además la política se vuelve voraz en el límite (**GLIE**: *greedy in the limit with infinite exploration*), por ejemplo con $\varepsilon\to0$.

**Exploración $\varepsilon$-*greedy* con decaimiento.** Con probabilidad $\varepsilon_k$ elegimos una acción al azar y con $1-\varepsilon_k$ la voraz. Empezamos con $\varepsilon_0=1$ (exploración pura) y lo reducimos geométricamente, $\varepsilon_{k+1}=\max(\varepsilon_{\min},\,\lambda\,\varepsilon_k)$. **Dejamos un piso $\varepsilon_{\min}>0$** (aquí 0.1) para que todos los pares sigan visitándose; es el compromiso práctico habitual, y tiene consecuencias que veremos (SARSA aprenderá el valor de la política **con** ese $\varepsilon$).

**Tasa de aprendizaje.** Usamos $\alpha(s,a)=\alpha_0/N(s,a)^{0.7}$ (cumple Robbins–Monro) con $\alpha_0=1$: los primeros updates sobrescriben casi todo y luego se estabiliza. La alternativa, $\alpha$ constante, se ensaya en el §7.

````python
# ============================================================
# 5. SARSA y Q-learning tabulares DESDE CERO (listas de Python: ~1.6 microsegundos por paso)
# ============================================================
def control_td(mdp, episodios, algoritmo="q", alpha0=1.0, expo=0.7, alpha_fijo=None,
               eps0=1.0, eps_min=0.1, eps_decay=0.99995, max_pasos=100, semilla=0, cada=None):
    """SARSA ('sarsa') o Q-learning ('q') tabulares con epsilon-greedy decreciente.
    alpha = alpha0 / N(s,a)^expo, o 'alpha_fijo' si se da. Devuelve (Q, retornos por episodio, copias de Q cada 'cada' episodios)."""
    rnd = _random.Random(semilla); rand = rnd.random
    cdf, R, term = tablas_rapidas(mdp)
    nS, nA, gamma = mdp.nS, mdp.nA, mdp.gamma
    Q = [[0.0] * nA for _ in range(nS)]; N = [[0] * nA for _ in range(nS)]
    eps, retornos, copias = eps0, [], []

    def elegir(s):                                                         # epsilon-greedy con desempate aleatorio
        if rand() < eps:
            return int(rand() * nA)
        q = Q[s]; m = max(q)
        if q.count(m) == 1:
            return q.index(m)
        return rnd.choice([i for i in range(nA) if q[i] == m])

    for ep in range(1, episodios + 1):
        s = mdp.s0; a = elegir(s); G, desc = 0.0, 1.0
        for _ in range(max_pasos):
            s2 = min(bisect_right(cdf[s][a], rand()), nS - 1)
            r, fin = R[s][a][s2], term[s2]
            a2 = elegir(s2)                                                # SARSA la usa en el objetivo; Q-learning solo la usa para actuar
            N[s][a] += 1
            alpha = alpha_fijo if alpha_fijo is not None else alpha0 / N[s][a] ** expo
            if fin:
                objetivo = r                                               # terminal: no hay futuro (V = 0)
            elif algoritmo == "q":
                objetivo = r + gamma * max(Q[s2])                          # off-policy: max sobre las acciones del siguiente estado
            else:
                objetivo = r + gamma * Q[s2][a2]                           # on-policy: la accion que realmente se tomara
            Q[s][a] += alpha * (objetivo - Q[s][a])
            G += desc * r; desc *= gamma
            s, a = s2, a2
            if fin:
                break
        retornos.append(G); eps = max(eps_min, eps * eps_decay)
        if cada and ep % cada == 0:
            copias.append(np.array(Q))
    return np.array(Q), np.array(retornos), copias

print("control_td definido. Ejemplo de uso: Q, retornos, copias = control_td(mdp, 1000, 'q', semilla=0)")
````

````python
# ============================================================
# 5. Q-learning y SARSA en FrozenLake resbaladizo: convergencia hacia Q*
# ============================================================
V_opt, Q_opt, pol_opt, _ = iteracion_valor(mdp)                            # verdad terreno (programacion dinamica)
EPIS, CADA, SEMILLAS = 100_000, 5000, [0, 1]
t0 = time.time()
resultados = {alg: [control_td(mdp, EPIS, alg, semilla=k, cada=CADA) for k in SEMILLAS] for alg in ("q", "sarsa")}
print(f"Entrenamiento ({len(SEMILLAS)} semillas x 2 algoritmos x {EPIS:,} episodios): {time.time() - t0:.1f} s")

def error_q(Q): return np.abs(Q - Q_opt)[nt].max()                           # error maximo vs Q* en estados no terminales
def valor_politica_voraz(Q):                                                # valor EXACTO en el estado inicial de la politica voraz de Q
    return evaluar_exacto(mdp, np.eye(mdp.nA)[Q.argmax(axis=1)])[mdp.s0]

filas = []
for alg, nombre in [("q", "Q-learning"), ("sarsa", "SARSA")]:
    for k, (Q, ret, copias) in zip(SEMILLAS, resultados[alg]):
        filas.append({"algoritmo": nombre, "semilla": k, "max|Q - Q*|": error_q(Q), "V(inicio) de la politica voraz": valor_politica_voraz(Q),
                      "V*(inicio)": V_opt[0], "retorno medio (ultimos 5000 ep., con exploracion)": ret[-5000:].mean()})
tabla = pd.DataFrame(filas)
display(tabla.round(4))

for k, (Q, ret, copias) in zip(SEMILLAS, resultados["q"]):
    assert error_q(Q) < 0.08                                                 # Q-learning converge a Q* (margen: observado ~0.03)
    assert valor_politica_voraz(Q) > V_opt[0] - 0.01                         # y su politica voraz es optima (en valor)
for k, (Q, ret, copias) in zip(SEMILLAS, resultados["sarsa"]):
    assert valor_politica_voraz(Q) > V_opt[0] - 0.05                         # SARSA tambien produce una politica voraz (casi) optima
print(f"\nQ-learning: max|Q-Q*| = {[round(float(error_q(Q)), 3) for Q, _, _ in resultados['q']]} (tolerancia 0.08) | politica voraz optima en valor: OK")

fig, axes = plt.subplots(1, 3, figsize=(19, 4.9))
puntos = np.arange(1, EPIS // CADA + 1) * CADA
for alg, nombre, color in [("q", "Q-learning", PALETA[0]), ("sarsa", "SARSA", PALETA[1])]:
    errs = np.array([[error_q(Qc) for Qc in copias] for _, _, copias in resultados[alg]])
    axes[0].plot(puntos, errs.mean(axis=0), color=color, lw=2.4, label=nombre); axes[0].fill_between(puntos, errs.min(axis=0), errs.max(axis=0), color=color, alpha=0.15)
    ret_suave = np.array([np.convolve(ret, np.ones(2000) / 2000, mode="valid") for _, ret, _ in resultados[alg]])
    axes[1].plot(np.arange(len(ret_suave[0])) + 2000, ret_suave.mean(axis=0), color=color, lw=2, label=nombre)
axes[0].set_yscale("log"); axes[0].set_xlabel("Episodios"); axes[0].set_ylabel(r"$\max_{s,a}|Q-Q_*|$ (log; banda = min-máx)")
axes[0].set_title(r"Q-learning converge a $Q_*$; SARSA converge a otra cosa", fontweight="bold", fontsize=10); axes[0].legend(fontsize=9)
axes[1].axhline(V_opt[0], color="black", ls=":", label=r"$v_*(s_0)$ (sin exploración)"); axes[1].legend(fontsize=9)
axes[1].set_xlabel("Episodios"); axes[1].set_ylabel("Retorno descontado (media móvil de 2000)")
axes[1].set_title(r"Retorno durante el entrenamiento (con $\varepsilon\geq0.1$)", fontweight="bold", fontsize=10)
pintar_mapa(axes[2], MAPAS["4x4"], V=resultados["q"][0][0].max(axis=1), accion=resultados["q"][0][0].argmax(axis=1), titulo=r"Política y $\max_a Q(s,a)$ aprendidos por Q-learning (semilla 0)", fmt="{:.3f}")
plt.tight_layout(); plt.show()
mismas = (resultados["q"][0][0].argmax(axis=1) == pol_opt)[nt].mean()
print(f"La politica voraz de Q-learning coincide con pi* en {100 * mismas:.0f}% de los estados no terminales (puede diferir donde dos acciones empatan o son irrelevantes).")
print("Nota: SARSA aprende el valor de la politica epsilon-greedy con epsilon >= 0.1, por eso su Q no tiende a Q* (aunque su politica voraz sea buena).")
````

## 6. *On-policy* frente a *Off-policy*: el Acantilado (*Cliff Walking*) 🧗

El ejemplo clásico (Sutton & Barto, Ejemplo 6.6) hace visible la diferencia entre SARSA y Q-learning. Una cuadrícula de $4\times12$: el agente empieza en la esquina inferior izquierda (`S`) y debe llegar a la inferior derecha (`G`). La fila inferior entre ambas es un **acantilado** (`C`): pisarlo da **−100** y devuelve al inicio. Cada paso cuesta **−1**. El entorno es determinista y **no descontado** ($\gamma=1$). Es una versión propia, equivalente a `CliffWalking` de Gymnasium.

- El **camino óptimo** pega al acantilado: **13 pasos** (retorno −13).
- Pero con $\varepsilon=0.1$ el agente a veces **se tropieza** y cae: cuanto más cerca del borde, más caro sale explorar.

**SARSA** *sabe* que explora (su objetivo incluye la acción $a'$ exploratoria) y aprende a **alejarse del borde** (camino **seguro**, más largo). **Q-learning** aprende el valor de la política **voraz** (sin tropiezos) y se pega al acantilado, de modo que **durante el entrenamiento** cae más veces y obtiene **peor recompensa en línea**, aunque su política voraz sea la **óptima**. Mantenemos $\varepsilon=0.1$ **constante** y $\alpha=0.1$ constante, como en el libro, con 8 semillas.

````python
# ============================================================
# 6. Cliff Walking propio: SARSA (camino seguro) vs Q-learning (camino optimo)
# ============================================================
def acantilado(filas=4, cols=12, gamma=1.0):
    """Cuadricula determinista; fila inferior entre S (inicio) y G (meta) = acantilado: -100 y vuelta al inicio. Paso = -1."""
    nS = filas * cols
    P = np.zeros((nS, 4, nS)); R = np.zeros((nS, 4, nS)); term = np.zeros(nS, dtype=bool)
    inicio, meta = (filas - 1) * cols, filas * cols - 1
    term[meta] = True
    for s in range(nS):
        if term[s]:
            P[s, :, s] = 1.0; continue
        f, c = divmod(s, cols)
        for a in range(4):
            f2 = min(max(f + MOV[a][0], 0), filas - 1); c2 = min(max(c + MOV[a][1], 0), cols - 1)
            s2, r = f2 * cols + c2, -1.0
            if f2 == filas - 1 and 1 <= c2 <= cols - 2:                    # cae por el acantilado
                s2, r = inicio, -100.0
            P[s, a, s2] = 1.0; R[s, a, s2] = r
    return MDP(P, R, gamma, inicio, term)

MAPA_CLIFF = ["FFFFFFFFFFFF", "FFFFFFFFFFFF", "FFFFFFFFFFFF", "SCCCCCCCCCCG"]
cliff = acantilado()
V_c, Q_c, pol_c, _ = iteracion_valor(cliff)
assert np.isclose(V_c[cliff.s0], -13.0)                                    # el camino optimo mide 13 pasos

def ruta_voraz(mdp_det, Q, max_pasos=100):
    """Ruta que sigue la politica voraz de Q en un entorno determinista (se corta si entra en un ciclo)."""
    s, ruta = mdp_det.s0, [mdp_det.s0]
    for _ in range(max_pasos):
        s = int(np.argmax(mdp_det.P[s, int(Q[s].argmax())])); ruta.append(s)
        if mdp_det.terminales[s] or ruta.count(s) > 1:
            break
    return ruta, bool(mdp_det.terminales[ruta[-1]])

def pi_eps_greedy(Q, eps):
    """Matriz de la politica epsilon-greedy asociada a Q (empates repartidos)."""
    pi = np.full(Q.shape, eps / Q.shape[1])
    for s in range(Q.shape[0]):
        mejores = np.flatnonzero(Q[s] == Q[s].max()); pi[s, mejores] += (1 - eps) / len(mejores)
    return pi

N_SEM, EPIS_C, EPS = 8, 2000, 0.1
t0 = time.time()
corridas = {alg: [control_td(cliff, EPIS_C, alg, alpha_fijo=0.1, eps0=EPS, eps_min=EPS, max_pasos=500, semilla=k) for k in range(N_SEM)] for alg in ("sarsa", "q")}
print(f"Cliff Walking ({N_SEM} semillas x 2 algoritmos x {EPIS_C} episodios): {time.time() - t0:.1f} s")

resumen = {}
for alg, nombre in [("sarsa", "SARSA"), ("q", "Q-learning")]:
    longitudes, valores, llegan = [], [], []
    for Q, ret, _ in corridas[alg]:
        ruta, ok = ruta_voraz(cliff, Q); longitudes.append(len(ruta) - 1); llegan.append(ok)
        valores.append(evaluar_exacto(cliff, pi_eps_greedy(Q, EPS))[cliff.s0])        # retorno ESPERADO al ejecutar la politica epsilon-greedy (exacto)
    resumen[alg] = (np.array(longitudes), np.array(valores), np.array(llegan))
    print(f"{nombre:<11} | camino voraz: longitudes {sorted(set(longitudes))} (optimo = 13) | retorno esperado con epsilon={EPS}: media {np.mean(valores):.1f} (min {np.min(valores):.1f}, max {np.max(valores):.1f})")

L_s, V_s, ok_s = resumen["sarsa"]; L_q, V_q, ok_q = resumen["q"]
assert ok_s.all() and ok_q.all()
assert (L_q == 13).all()                                                    # Q-learning encuentra SIEMPRE el camino optimo (13 pasos), pegado al acantilado
assert (L_s > 13).all()                                                     # SARSA siempre elige un camino mas largo (mas seguro)
assert V_s.mean() > V_q.mean() + 15                                         # y su retorno esperado CON exploracion es claramente mayor (observado: ~ -22 vs ~ -51)

fig, axes = plt.subplots(1, 3, figsize=(20, 4.4), gridspec_kw={"width_ratios": [1.4, 1.2, 1.2]})
for alg, nombre, color in [("sarsa", "SARSA (on-policy)", PALETA[1]), ("q", "Q-learning (off-policy)", PALETA[0])]:
    suave = np.array([np.convolve(ret, np.ones(20) / 20, mode="valid") for _, ret, _ in corridas[alg]])
    axes[0].plot(np.arange(suave.shape[1]) + 20, suave.mean(axis=0), color=color, lw=2.2, label=nombre)
    axes[0].fill_between(np.arange(suave.shape[1]) + 20, suave.mean(axis=0) - suave.std(axis=0), suave.mean(axis=0) + suave.std(axis=0), color=color, alpha=0.12)
axes[0].set_ylim(-150, 0); axes[0].axhline(-13, color="black", ls=":", label="Óptimo sin tropiezos (−13)")
axes[0].set_xlabel("Episodio"); axes[0].set_ylabel("Suma de recompensas (media móvil de 20 ep.)")
axes[0].set_title("Rendimiento en línea (media de 8 semillas ± desv., $\\varepsilon$=0.1): SARSA cae menos", fontweight="bold", fontsize=10); axes[0].legend(fontsize=8, loc="lower right")
for ax, alg, nombre in [(axes[1], "sarsa", "SARSA: camino seguro"), (axes[2], "q", "Q-learning: camino óptimo (borde del acantilado)")]:
    Q = corridas[alg][0][0]; ruta, _ = ruta_voraz(cliff, Q)
    pintar_mapa(ax, MAPA_CLIFF, titulo=f"{nombre} ({len(ruta) - 1} pasos, semilla 0)")
    ax.plot([s % 12 for s in ruta], [s // 12 for s in ruta], color=PALETA[1] if alg == "sarsa" else PALETA[0], lw=3.5, marker="o", ms=4)
plt.tight_layout(); plt.show()
print("Lectura: Q-learning APRENDE el camino optimo (13 pasos) pero, como explora con epsilon, se cae mas durante el entrenamiento;")
print("         SARSA incorpora sus propios tropiezos en el valor y por eso elige el rodeo. Si epsilon -> 0, ambos tienden al mismo camino.")
````

## 7. Hiperparámetros: $\alpha$ y el Decaimiento de $\varepsilon$ (con Honestidad) 🎛️

Dos decisiones dominan el comportamiento de Q-learning tabular. Las medimos en *FrozenLake* (20 000 episodios, 2 semillas por configuración; el error es $\max|Q-Q_{\ast}|$ sobre estados no terminales):

1. **Tasa de aprendizaje.** $\alpha$ **constante** es simple y adaptable (útil si el entorno cambia), pero **no converge del todo**: oscila alrededor de $Q_{\ast}$ con una amplitud que crece con $\alpha$ (como el SGD con paso fijo). $\alpha$ **decreciente** (Robbins–Monro) sí converge, aunque **más lento** al inicio.
2. **Decaimiento de $\varepsilon$.** Si baja **demasiado rápido**, el agente se vuelve voraz antes de haber visto la meta; en un entorno de **recompensa escasa** esto puede dejarlo con una política pobre (en *FrozenLake* 4×4, que es pequeño, el efecto resulta moderado). Si baja **demasiado lento**, se desperdician episodios explorando.

> ⚠️ **Honestidad sobre el experimento:** son pocas semillas y un solo entorno; las **tendencias** son informativas, los números exactos no. En la práctica estos hiperparámetros se ajustan **por problema**.

````python
# ============================================================
# 7. Barrido de hiperparametros en FrozenLake (20 000 episodios, 2 semillas)
# ============================================================
EP_H, SEM_H = 20_000, [0, 1]
t0 = time.time()
configs = [
    ("alpha = 1/N^0.7 (decreciente)", dict(alpha0=1.0, expo=0.7), 0.99985),
    ("alpha = 0.5 (constante)",       dict(alpha_fijo=0.5), 0.99985),
    ("alpha = 0.1 (constante)",       dict(alpha_fijo=0.1), 0.99985),
    ("alpha = 0.01 (constante)",      dict(alpha_fijo=0.01), 0.99985),
    ("eps decae RAPIDO (0.99)",       dict(alpha0=1.0, expo=0.7), 0.99),
    ("eps decae LENTO (0.99995)",     dict(alpha0=1.0, expo=0.7), 0.99995),
]
filas = []
for nombre, kw, decay in configs:
    errs, vals = [], []
    for k in SEM_H:
        Q, _, _ = control_td(mdp, EP_H, "q", eps_decay=decay, semilla=k, **kw)
        errs.append(error_q(Q)); vals.append(valor_politica_voraz(Q))
    filas.append({"configuracion": nombre, "decaimiento eps": decay, "max|Q-Q*| (media)": np.mean(errs), "V(inicio) politica voraz (media)": np.mean(vals), "V(inicio) por semilla": [round(v, 3) for v in vals]})
display(pd.DataFrame(filas).set_index("configuracion").round(4))
print(f"({time.time() - t0:.1f} s)   V*(inicio) = {V_opt[0]:.4f}")

e = {f["configuracion"]: f["max|Q-Q*| (media)"] for f in filas}
assert e["alpha = 0.5 (constante)"] > e["alpha = 1/N^0.7 (decreciente)"]               # alpha constante grande deja un piso de ruido mayor
print("\nLectura honesta: (1) alpha constante grande (0.5) deja un error de Q mucho mayor que alpha decreciente (piso de ruido) y en una semilla la politica voraz quedo mala (V = 0.165);")
print("alpha muy pequeno (0.01) casi no propaga el valor en 20 000 episodios (error alto). (2) Con epsilon que decae rapido el error de Q es mayor (visita menos los pares raros),")
print("aunque aqui la politica voraz salio casi optima: en este entorno pequeno el efecto es moderado; en entornos grandes o de recompensa escasa suele ser decisivo.")
````

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Funciones de valor** | $v_\pi(s)=\mathbb E_\pi[G_t\mid s]$ y $q_\pi(s,a)$; $v_\pi=\sum_a\pi\,q_\pi$ y $q_\pi=r+\gamma Pv_\pi$ |
| **Bellman de expectativa** | $v_\pi=T^\pi v_\pi$ (lineal): sale de $G_t=r_{t+1}+\gamma G_{t+1}$ y la propiedad de Markov |
| **Bellman de optimalidad** | $v_{\ast}(s)=\max_a[r+\gamma\sum P\,v_{\ast}]$ (no lineal); $\pi_{\ast}$ es **voraz** respecto a $q_{\ast}$ |
| **Contracción** | $\|T^{\ast}v-T^{\ast}w\|_\infty\le\gamma\|v-w\|_\infty$ → punto fijo único y convergencia geométrica $\gamma^k$ (verificado) |
| **Programación dinámica** | Evaluación iterativa, **iteración de política** (pocas iteraciones) e **iteración de valor** (muchos barridos baratos): **coinciden** con la solución exacta |
| **MC vs TD(0)** | MC: sin sesgo, alta varianza, espera al final. TD(0): *bootstrapping*, baja varianza, paso a paso; $\alpha$ constante deja un piso de ruido |
| **SARSA vs Q-learning** | *On-policy* (objetivo con $a'$ real) vs *off-policy* (objetivo con $\max$); solo el segundo apunta a $q_{\ast}$ **sin importar la exploración** |
| **Convergencia de Q-learning** | Visitas infinitas + Robbins–Monro: aquí, $\max\lvert Q-Q_{\ast}\rvert<0.08$ tras 100 000 episodios y política voraz óptima |
| **Cliff Walking** | Q-learning → camino óptimo (13 pasos) pero **más caídas** al entrenar; SARSA → camino **seguro** y mejor rendimiento en línea |
| **Límites** | La tabla $Q$ crece con $\lvert\mathcal S\rvert\cdot\lvert\mathcal A\rvert$: no escala (cuaderno 03); la recompensa escasa exige muchos episodios; $\alpha$ y $\varepsilon$ se ajustan por problema |
