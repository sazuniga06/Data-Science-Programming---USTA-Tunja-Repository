## 1. ¿Por Qué Optimizar la Política Directamente? 🎛️

Los métodos *value-based* (Q-learning, DQN) aprenden $q(s,a)$ y actúan con $\arg\max_a q(s,a)$. Funcionan muy bien, pero tienen tres limitaciones que los **gradientes de política** resuelven:

1. **Acciones continuas.** Si $a\in\mathbb R^d$ (el torque de un brazo robótico), el $\arg\max_a q(s,a)$ es en sí un problema de optimización **en cada paso**. Con una política paramétrica, p. ej. gaussiana $a\sim\mathcal N(\mu_\theta(s),\sigma^2)$, se **muestrea** directamente.
2. **Políticas estocásticas óptimas.** Cuando hay **observabilidad parcial** o **adversarios** (piedra-papel-tijera, póker), la mejor política puede ser **aleatoria**; un $\arg\max$ solo produce políticas deterministas. El ejemplo del **pasillo corto** (abajo) lo demuestra con números.
3. **Suavidad.** En un método *value-based* un cambio minúsculo de $q$ puede **cambiar bruscamente** la acción elegida; en uno *policy-based* las probabilidades cambian **de forma continua** con $\theta$, lo que da mejores garantías de convergencia (a un óptimo **local**).

### Parametrización *softmax*

Para acciones **discretas** usamos preferencias $\theta_{s,a}$ (o la salida de una red) y una **softmax**:

$$\pi_\theta(a\mid s)=\frac{e^{\theta_{s,a}}}{\sum_b e^{\theta_{s,b}}},\qquad J(\theta)=\mathbb E_{\pi_\theta}\!\Big[\sum_{t\ge0}\gamma^t r_{t+1}\Big]=\sum_s\rho_0(s)\,v_{\pi_\theta}(s).$$

El **objetivo** es $J(\theta)$ y la **regla** es el ascenso del gradiente $\theta\leftarrow\theta+\eta\,\nabla_\theta J(\theta)$. La dificultad: $J$ depende de $\theta$ **a través de la distribución de estados que visita el agente**, que no podemos derivar directamente porque depende de la dinámica $P$ (desconocida). El teorema del §2 resuelve exactamente eso.

````python
# ============================================================
# Configuracion inicial
# ============================================================
# (numpy, pandas, matplotlib y seaborn vienen preinstalados en Google Colab; este cuaderno solo usa NumPy: no necesita PyTorch ni Gymnasium)

import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

warnings.filterwarnings("ignore")

SEED = 42
np.random.seed(SEED)
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (9, 5)
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"NumPy {np.__version__}")
````

````python
# ============================================================
# 1. El pasillo corto (Sutton & Barto, Ej. 13.1): la politica optima es ESTOCASTICA
# ============================================================
# Tres casillas con el MISMO aspecto para el agente (aliasing): no puede distinguirlas, asi que usa UNA sola probabilidad p = P(derecha).
#   s0: derecha -> s1, izquierda -> se queda | s1: acciones INVERTIDAS (derecha -> s0, izquierda -> s2) | s2: derecha -> META, izquierda -> s1
# Recompensa -1 por paso, sin descuento: J(p) = v(s0) = -(pasos esperados).
def J_pasillo(p):
    A = np.array([[p, -p, 0.0], [-p, 1.0, -(1 - p)], [0.0, -(1 - p), 1.0]])          # sistema de Bellman de expectativa: A v = -1
    return np.linalg.solve(A, -np.ones(3))[0]

ps = np.linspace(0.01, 0.99, 99); Js = np.array([J_pasillo(p) for p in ps])
p_opt, J_opt = ps[Js.argmax()], Js.max()
assert 0.55 < p_opt < 0.63 and -11.8 < J_opt < -11.5                                  # optimo estocastico: p ~ 0.59, J ~ -11.6
assert J_pasillo(0.99) < -100 and J_pasillo(0.01) < -100                              # las politicas casi deterministas son pesimas
print(f"Mejor politica: P(derecha) = {p_opt:.2f}, J = {J_opt:.2f} pasos esperados.")
print(f"Casi deterministas: p=0.01 -> J = {J_pasillo(0.01):.0f};  p=0.99 -> J = {J_pasillo(0.99):.0f}  (ambas SIEMPRE se atascan o dan vueltas)")

fig, ax = plt.subplots(figsize=(9, 4.4))
ax.plot(ps, Js, color=PALETA[0], lw=2.8)
ax.scatter([p_opt], [J_opt], color=PALETA[1], s=90, zorder=5, label=f"Óptimo estocástico: p = {p_opt:.2f}, J = {J_opt:.1f}")
ax.set_ylim(-60, 0); ax.set_xlabel("p = P(acción derecha) en cualquier casilla"); ax.set_ylabel("J(p) = −(pasos esperados hasta la meta)")
ax.set_title("Pasillo corto: ninguna política determinista es buena; la óptima es aleatoria", fontweight="bold"); ax.legend(fontsize=9)
plt.tight_layout(); plt.show()
print("Un metodo basado en q con argmax produce politicas deterministas (o epsilon-greedy casi deterministas): no puede llegar a p = 0.59. Un gradiente de politica si.")
````

## 2. El Teorema del Gradiente de la Política y el Truco de la Log-Derivada 📐

### 2.1 Derivación (REINFORCE)

Sea $\tau=(s_0,a_0,r_1,s_1,a_1,\dots)$ una trayectoria y $G(\tau)=\sum_t\gamma^t r_{t+1}$ su retorno descontado. Su probabilidad es

$$p_\theta(\tau)=\rho_0(s_0)\prod_{t}\pi_\theta(a_t\mid s_t)\,P(s_{t+1}\mid s_t,a_t),\qquad J(\theta)=\sum_\tau p_\theta(\tau)\,G(\tau).$$

**Truco de la log-derivada** ($\nabla p=p\,\nabla\log p$):

$$\nabla_\theta J=\sum_\tau\nabla_\theta p_\theta(\tau)\,G(\tau)=\sum_\tau p_\theta(\tau)\,\nabla_\theta\log p_\theta(\tau)\,G(\tau)=\mathbb E_{\tau\sim p_\theta}\!\big[\nabla_\theta\log p_\theta(\tau)\,G(\tau)\big].$$

Como $\log p_\theta(\tau)=\log\rho_0+\sum_t\log\pi_\theta(a_t\mid s_t)+\sum_t\log P(s_{t+1}\mid s_t,a_t)$ y **solo el término de la política depende de $\theta$**, la dinámica $P$ **desaparece** del gradiente:

$$\nabla_\theta\log p_\theta(\tau)=\sum_t\nabla_\theta\log\pi_\theta(a_t\mid s_t).$$

**Causalidad.** Una acción $a_t$ no puede influir en recompensas **anteriores**, y $\mathbb E[\nabla_\theta\log\pi_\theta(a\mid s)]=\sum_a\nabla_\theta\pi_\theta(a\mid s)=\nabla_\theta 1=0$, así que los términos «pasado × acción presente» tienen esperanza nula y se pueden **eliminar**, quedando solo las recompensas **futuras**:

$$\boxed{\;\nabla_\theta J(\theta)=\mathbb E_{\pi_\theta}\!\Big[\sum_{t\ge0}\gamma^t\,G_t\,\nabla_\theta\log\pi_\theta(a_t\mid s_t)\Big]\;},\qquad G_t=\sum_{k\ge0}\gamma^k r_{t+k+1}.$$

Es el estimador **REINFORCE** (Williams, 1992): **sube la probabilidad** de las acciones cuyo retorno posterior fue alto y la **baja** si fue bajo. Se estima por Monte Carlo con episodios simulados, **sin conocer $P$**. (Muchas implementaciones omiten el factor $\gamma^t$; es una práctica común que cambia ligeramente el objetivo. Aquí lo mantenemos para que el estimador sea **exactamente** el gradiente de $J$.)

### 2.2 Forma del teorema y gradiente exacto en un MDP finito

Agrupando por estados (Sutton et al., 2000):

$$\nabla_\theta J=\sum_s d_\pi(s)\sum_a q_\pi(s,a)\,\nabla_\theta\pi_\theta(a\mid s),\qquad d_\pi(s)=\sum_{t\ge0}\gamma^t\Pr(s_t=s).$$

$d_\pi$ es la **ocupación descontada** (no está normalizada: en una tarea continua sumaría $1/(1-\gamma)$). En forma matricial: $d_\pi^\top=\rho_0^\top(I-\gamma P_\pi)^{-1}$. Con la **softmax**, $\partial\pi(a\mid s)/\partial\theta_{s,b}=\pi(a\mid s)(\delta_{ab}-\pi(b\mid s))$ y se obtiene la **fórmula cerrada**

$$\frac{\partial J}{\partial\theta_{s,b}}=d_\pi(s)\,\pi(b\mid s)\,\big(q_\pi(s,b)-v_\pi(s)\big)=d_\pi(s)\,\pi(b\mid s)\,A_\pi(s,b),$$

con la **ventaja** $A_\pi=q_\pi-v_\pi$: se refuerzan las acciones con ventaja **positiva**, ponderadas por **cuánto visita el agente ese estado** y por su probabilidad actual. Y para $\nabla_\theta\log\pi(a\mid s)$ con softmax: $\nabla_{\theta_{s,\cdot}}\log\pi(a\mid s)=\mathbf e_a-\pi(\cdot\mid s)$ (vector *one-hot* menos las probabilidades).

### 2.3 Plan de verificación

Como en un MDP finito conocemos $P$, podemos calcular el gradiente **tres** maneras que **deben coincidir**: (1) la **fórmula cerrada** (arriba), (2) **diferencias finitas** sobre $J(\theta)$ exacta, y (3) el **estimador REINFORCE** promediado sobre muchos episodios (con error $\propto1/\sqrt N$).

````python
# ============================================================
# Un MDP finito DESDE CERO: tensores P[s, a, s'] y R[s, a, s']
# ============================================================
MAPAS = {
    "4x4": ["SFFF", "FHFH", "FFFH", "HFFG"],
    "8x8": ["SFFFFFFF", "FFFFFFFF", "FFFHFFFF", "FFFFFHFF", "FFFHFFFF", "FHHFFFHF", "FHFFHFHF", "FFFHFFFG"],
}
MOV = {0: (0, -1), 1: (1, 0), 2: (0, 1), 3: (-1, 0)}      # 0 = Izquierda, 1 = Abajo, 2 = Derecha, 3 = Arriba (convencion de Gymnasium)
NOMBRE_ACCION = {0: "Izquierda", 1: "Abajo", 2: "Derecha", 3: "Arriba"}


class MDP:
    """MDP finito (S, A, P, R, gamma) con estado inicial fijo y estados terminales absorbentes.

    P[s, a, s2] = probabilidad de pasar a s2 tras hacer a en s;  R[s, a, s2] = recompensa de esa transicion.
    """
    def __init__(self, P, R, gamma, s0, terminales):
        self.P, self.R, self.gamma, self.s0 = P, R, gamma, s0
        self.terminales = np.asarray(terminales, dtype=bool)
        self.nS, self.nA = P.shape[0], P.shape[1]
        self.r = (P * R).sum(axis=2)                            # r(s, a) = E[R | s, a]
        self.cdf = np.cumsum(P, axis=2); self.cdf[..., -1] = 1.0   # para muestrear s2 con una sola uniforme

    def reset(self):
        return self.s0

    def step(self, s, a, rng):
        """Muestrea (s2, recompensa, terminado) segun la dinamica del MDP."""
        s2 = min(int(np.searchsorted(self.cdf[s, a], rng.random(), side="right")), self.nS - 1)
        return s2, self.R[s, a, s2], bool(self.terminales[s2])


def frozenlake(mapa="4x4", resbaladizo=True, gamma=0.99):
    """Construye FrozenLake como MDP explicito. Resbaladizo: la accion elegida ocurre con prob. 1/3 y con 1/3 cada
    una de las dos direcciones perpendiculares. Chocar con el borde = quedarse. H y G son terminales; recompensa 1 al entrar a G."""
    filas = MAPAS[mapa]; n = len(filas); nS = n * n
    P = np.zeros((nS, 4, nS)); R = np.zeros((nS, 4, nS)); term = np.zeros(nS, dtype=bool)
    for f in range(n):
        for c in range(n):
            s = f * n + c
            if filas[f][c] in "HG":                              # terminal absorbente: se queda y no da mas recompensa
                term[s] = True; P[s, :, s] = 1.0; continue
            for a in range(4):
                direcciones = [(a - 1) % 4, a, (a + 1) % 4] if resbaladizo else [a]
                for d in direcciones:
                    f2 = min(max(f + MOV[d][0], 0), n - 1); c2 = min(max(c + MOV[d][1], 0), n - 1)
                    s2 = f2 * n + c2
                    P[s, a, s2] += 1.0 / len(direcciones)
                    R[s, a, s2] = float(filas[f2][c2] == "G")
    return MDP(P, R, gamma, 0, term)


def evaluar_exacto(mdp, pi, gamma=None):
    """V^pi exacto resolviendo (I - gamma P_pi) V = r_pi. pi: matriz (nS, nA). V(terminal) = 0."""
    g = mdp.gamma if gamma is None else gamma
    Ppi = np.einsum("sa,sat->st", pi, mdp.P); rpi = (pi * mdp.r).sum(axis=1)
    Ppi[mdp.terminales] = 0.0; rpi[mdp.terminales] = 0.0       # el valor de un estado terminal es 0
    return np.linalg.solve(np.eye(mdp.nS) - g * Ppi, rpi)

# ============================================================
# Utilidad de dibujo: mapas con valores y flechas de politica
# ============================================================
COLOR_CASILLA = {"S": "#fde68a", "F": "#e0f2fe", "H": "#1e293b", "G": "#86efac", "C": "#fecaca", "W": "#94a3b8"}
FLECHA = {0: "←", 1: "↓", 2: "→", 3: "↑"}

def pintar_mapa(ax, mapa, V=None, accion=None, titulo="", etiquetas=None, fmt="{:.2f}", vmax=None):
    """Dibuja una cuadricula. V: valores por estado (colorea y escribe); accion: accion (indice) por estado -> flecha."""
    nf, nc = len(mapa), len(mapa[0]); etiquetas = etiquetas or {}
    vmax = vmax if vmax is not None else (np.nanmax(np.abs(V)) if V is not None else 1.0)
    cmap = plt.get_cmap("Blues")
    for f in range(nf):
        for c in range(nc):
            s, ch = f * nc + c, mapa[f][c]
            color = COLOR_CASILLA[ch]
            if V is not None and ch in "SF" and vmax > 0:
                color = cmap(0.08 + 0.8 * min(max(V[s] / vmax, 0), 1))
            ax.add_patch(plt.Rectangle((c - 0.5, f - 0.5), 1, 1, facecolor=color, edgecolor="white", lw=2))
            oscuro = ch in "H" or (V is not None and ch in "SF" and V[s] / vmax > 0.6)
            if ch in "HGWC" or ch in etiquetas:
                ax.text(c, f, etiquetas.get(ch, ch), ha="center", va="center", fontsize=13, fontweight="bold", color="white" if ch == "H" else "#0f172a")
                continue
            if V is not None:
                ax.text(c, f + (0.22 if accion is not None else 0), fmt.format(V[s]), ha="center", va="center", fontsize=8.5, color="white" if oscuro else "#0f172a")
            if accion is not None:
                ax.text(c, f - (0.12 if V is not None else 0), FLECHA[int(accion[s])], ha="center", va="center", fontsize=17, color="white" if oscuro else "#b91c1c")
            if ch == "S" and V is None and accion is None:
                ax.text(c, f, "S", ha="center", va="center", fontsize=13, fontweight="bold")
    ax.set_xlim(-0.5, nc - 0.5); ax.set_ylim(nf - 0.5, -0.5); ax.set_aspect("equal")
    ax.set_xticks([]); ax.set_yticks([]); ax.grid(False); ax.set_title(titulo, fontweight="bold", fontsize=10)

# ============================================================
# Entorno: el mundo 4x3 de Russell & Norvig (+ utilidades de politica softmax y gradiente EXACTO)
# ============================================================
def mundo_4x3(gamma=0.99, costo=-0.04, p_ok=0.8, bono_terminal=0.0):
    """Cuadricula 3x4. (0,3) = meta +1 (G), (1,3) = trampa -1 (H), (1,1) = pared (W, inalcanzable). Cada paso cuesta 'costo'.
    La accion elegida ocurre con prob. p_ok y con (1-p_ok)/2 cada una de las dos direcciones perpendiculares; chocar = quedarse.
    bono_terminal: constante que se SUMA a la recompensa de entrar a la meta o a la trampa (solo para el experimento de varianza)."""
    filas, cols = 3, 4; nS = filas * cols
    P = np.zeros((nS, 4, nS)); R = np.zeros((nS, 4, nS)); term = np.zeros(nS, dtype=bool)
    pared, meta, trampa = 1 * cols + 1, 0 * cols + 3, 1 * cols + 3
    term[[pared, meta, trampa]] = True
    for s in range(nS):
        if term[s]:
            P[s, :, s] = 1.0; continue
        f, c = divmod(s, cols)
        for a in range(4):
            for d, p in [(a, p_ok), ((a - 1) % 4, (1 - p_ok) / 2), ((a + 1) % 4, (1 - p_ok) / 2)]:
                f2, c2 = f + MOV[d][0], c + MOV[d][1]
                if not (0 <= f2 < filas and 0 <= c2 < cols) or f2 * cols + c2 == pared:
                    f2, c2 = f, c                                          # choque con borde o pared: se queda
                s2 = f2 * cols + c2
                P[s, a, s2] += p
                R[s, a, s2] = (1.0 + bono_terminal) if s2 == meta else ((-1.0 + bono_terminal) if s2 == trampa else costo)
    return MDP(P, R, gamma, 2 * cols + 0, term)                           # inicio: esquina inferior izquierda

MAPA_4x3 = ["FFFG", "FWFH", "SFFF"]
ETIQ_4x3 = {"G": "+1", "H": "-1"}

def iteracion_valor(mdp, tol=1e-12):
    """V*, Q* por iteracion de valor (version compacta del cuaderno 01)."""
    V = np.zeros(mdp.nS)
    while True:
        V2 = (mdp.r + mdp.gamma * mdp.P @ V).max(axis=1)
        if np.abs(V2 - V).max() < tol:
            V = V2; break
        V = V2
    return V, mdp.r + mdp.gamma * mdp.P @ V

def softmax_filas(theta):
    z = np.exp(theta - theta.max(axis=1, keepdims=True))
    return z / z.sum(axis=1, keepdims=True)

def J_exacto(mdp, theta):
    """J(theta) = v_{pi_theta}(s0), exacto (sistema lineal)."""
    return evaluar_exacto(mdp, softmax_filas(theta))[mdp.s0]

def ocupacion_descontada(mdp, pi):
    """d_pi(s) = sum_t gamma^t Pr(s_t = s) = [rho0^T (I - gamma P_pi)^-1]_s (los terminales no propagan)."""
    Ppi = np.einsum("sa,sat->st", pi, mdp.P); Ppi[mdp.terminales] = 0.0
    rho0 = np.zeros(mdp.nS); rho0[mdp.s0] = 1.0
    return np.linalg.solve((np.eye(mdp.nS) - mdp.gamma * Ppi).T, rho0)

def grad_analitico(mdp, theta):
    """Teorema del gradiente con softmax: dJ/dtheta[s,b] = d(s) * pi(b|s) * (q(s,b) - v(s))."""
    pi = softmax_filas(theta)
    V = evaluar_exacto(mdp, pi); Q = mdp.r + mdp.gamma * mdp.P @ V
    return ocupacion_descontada(mdp, pi)[:, None] * pi * (Q - V[:, None])

def grad_diferencias(mdp, theta, h=1e-5):
    """Gradiente de J por diferencias finitas centrales (derivando J_exacto)."""
    g = np.zeros_like(theta)
    for idx in np.ndindex(*theta.shape):
        tp, tm = theta.copy(), theta.copy(); tp[idx] += h; tm[idx] -= h
        g[idx] = (J_exacto(mdp, tp) - J_exacto(mdp, tm)) / (2 * h)
    return g

mdp = mundo_4x3(gamma=0.99)
V_opt, Q_opt = iteracion_valor(mdp); J_opt = V_opt[mdp.s0]; nt = ~mdp.terminales
print(f"Mundo 4x3: |S| = {mdp.nS} (la pared es un estado inalcanzable), gamma = {mdp.gamma}; J* = v*(inicio) = {J_opt:.4f}")

# --- Verificacion (1) vs (2): gradiente analitico == diferencias finitas, en un theta aleatorio
rng = np.random.default_rng(SEED)
theta0 = rng.normal(0, 0.5, (mdp.nS, mdp.nA))
g_an, g_fd = grad_analitico(mdp, theta0), grad_diferencias(mdp, theta0)
dif = np.abs(g_an - g_fd).max()
assert dif < 1e-8
print(f"Gradiente analitico vs diferencias finitas ({theta0.size} componentes): max|diferencia| = {dif:.1e} (escala del gradiente: {np.abs(g_an).max():.3f})")
assert np.allclose(g_an[mdp.terminales], 0.0)                              # los terminales no influyen en J

fig, axes = plt.subplots(1, 3, figsize=(18, 4.6))
pintar_mapa(axes[0], MAPA_4x3, V=V_opt, titulo=r"$v_*$ (iteración de valor)", etiquetas=ETIQ_4x3, fmt="{:.2f}")
d0 = ocupacion_descontada(mdp, softmax_filas(theta0))
pintar_mapa(axes[1], MAPA_4x3, V=d0, titulo=r"Ocupación descontada $d_\pi(s)$ de la política inicial", etiquetas=ETIQ_4x3, fmt="{:.2f}", vmax=d0.max())
lim = np.abs(g_an).max() * 1.1
axes[2].plot([-lim, lim], [-lim, lim], color="black", ls=":", label="y = x")
axes[2].scatter(g_fd.ravel(), g_an.ravel(), color=PALETA[0], s=35, alpha=0.8, label=f"{theta0.size} componentes")
axes[2].set_xlabel(r"$\partial J/\partial\theta$ por diferencias finitas"); axes[2].set_ylabel(r"$\partial J/\partial\theta$ por el teorema (fórmula cerrada)")
axes[2].set_title("El teorema del gradiente coincide con las diferencias finitas", fontweight="bold", fontsize=10); axes[2].legend(fontsize=9)
plt.tight_layout(); plt.show()
````

````python
# ============================================================
# 2. Estimador REINFORCE (Monte Carlo) vs gradiente exacto
# ============================================================
def retornos_hacia_atras(recompensas, gamma):
    G = np.zeros(len(recompensas)); acumulado = 0.0
    for t in range(len(recompensas) - 1, -1, -1):
        acumulado = recompensas[t] + gamma * acumulado; G[t] = acumulado
    return G

def muestrear_episodio(mdp, cdf_pi, rng, max_pasos=500):
    """Un episodio con la politica dada por la CDF de sus acciones en cada estado. Devuelve (estados, acciones, recompensas)."""
    s, S, A, Rw = mdp.s0, [], [], []
    for _ in range(max_pasos):
        a = min(int(np.searchsorted(cdf_pi[s], rng.random(), side="right")), mdp.nA - 1)
        s2, r, fin = mdp.step(s, a, rng)
        S.append(s); A.append(a); Rw.append(r); s = s2
        if fin:
            break
    return np.array(S), np.array(A), np.array(Rw)

def gradiente_episodio(mdp, pi, S, A, Rw, b=0.0):
    """Estimador REINFORCE de UN episodio: sum_t gamma^t (G_t - b(s_t)) grad log pi(a_t|s_t).  b: escalar o vector por estado."""
    G = retornos_hacia_atras(Rw, mdp.gamma)
    w = mdp.gamma ** np.arange(len(Rw)) * (G - (b[S] if np.ndim(b) else b))
    g = np.zeros_like(pi)
    np.add.at(g, (S, A), w)                                                # + w * e_a
    np.add.at(g, S, -w[:, None] * pi[S])                                   # - w * pi(.|s)
    return g

def muestras_gradiente(mdp, theta, N, base, rng):
    """N estimaciones de gradiente (una por episodio), matriz (N, nS*nA). base: 'ninguna' | 'constante' (b = J(theta)) | 'valor' (b(s) = v_pi(s))."""
    pi = softmax_filas(theta); cdf = np.cumsum(pi, axis=1)
    V = evaluar_exacto(mdp, pi)
    b = {"ninguna": 0.0, "constante": V[mdp.s0], "valor": V}[base]         # las baselines usan valores EXACTOS (ideales) para aislar el efecto
    G = np.zeros((N, theta.size))
    for i in range(N):
        S, A, Rw = muestrear_episodio(mdp, cdf, rng)
        G[i] = gradiente_episodio(mdp, pi, S, A, Rw, b).ravel()
    return G

t0 = time.time()
rng = np.random.default_rng(SEED + 1)
N_MC = 20000
G_mc = muestras_gradiente(mdp, theta0, N_MC, "ninguna", rng)
mu, ee = G_mc.mean(axis=0), G_mc.std(axis=0, ddof=1) / np.sqrt(N_MC)
z = (mu - g_an.ravel()) / np.maximum(ee, 1e-12)
cos = mu @ g_an.ravel() / (np.linalg.norm(mu) * np.linalg.norm(g_an))
print(f"REINFORCE con {N_MC:,} episodios ({time.time() - t0:.1f} s): similitud coseno con el gradiente exacto = {cos:.4f}; max |z| = {np.abs(z[ee > 0]).max():.2f}")
assert cos > 0.99                                                          # misma direccion
assert np.abs(z[ee > 0]).max() < 4.5                                       # cada componente dentro de ~4.5 errores estandar (insesgado)

# Error vs N: debe decaer como 1/sqrt(N)
Ns = np.unique(np.logspace(2, np.log10(N_MC), 25).astype(int))
err_N = np.array([np.linalg.norm(G_mc[:n].mean(axis=0) - g_an.ravel()) for n in Ns])
pendiente = np.polyfit(np.log(Ns), np.log(err_N), 1)[0]
assert -0.75 < pendiente < -0.3
print(f"Pendiente log-log del error frente a N: {pendiente:.2f} (teoria: -0.5)")

fig, axes = plt.subplots(1, 2, figsize=(14, 4.8))
lim = np.abs(g_an).max() * 1.15
axes[0].plot([-lim, lim], [-lim, lim], color="black", ls=":", label="y = x")
axes[0].errorbar(g_an.ravel(), mu, yerr=2 * ee, fmt="o", color=PALETA[0], ms=4, alpha=0.8, label=f"REINFORCE ({N_MC:,} ep.) ± 2 EE")
axes[0].set_xlabel("Gradiente exacto (teorema)"); axes[0].set_ylabel("Estimación Monte Carlo")
axes[0].set_title("El estimador REINFORCE es insesgado: coincide con el exacto", fontweight="bold", fontsize=10); axes[0].legend(fontsize=9)
axes[1].loglog(Ns, err_N, "o-", color=PALETA[1], lw=2, label=r"$\|\hat g_N-g\|_2$")
axes[1].loglog(Ns, err_N[0] * np.sqrt(Ns[0] / Ns), color="black", ls="--", label=r"$\propto 1/\sqrt{N}$")
axes[1].set_xlabel("Episodios $N$"); axes[1].set_ylabel("Error del estimador"); axes[1].legend(fontsize=9)
axes[1].set_title(f"El error se encoge como 1/√N (pendiente {pendiente:.2f})", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

## 3. Varianza del Estimador y *Baselines* 🎚️

REINFORCE es **insesgado** pero **ruidoso**: el retorno $G_t$ de una sola trayectoria suma el efecto de **todas** las decisiones y del azar del entorno, y **multiplica** un vector $\nabla\log\pi$ que apunta en direcciones arbitrarias. Con $N$ episodios la varianza decae como $1/N$ (lo vimos arriba), pero cada episodio cuesta una simulación.

### La *baseline*

Para cualquier función $b(s)$ que **no dependa de la acción**,

$$\mathbb E_{a\sim\pi}\big[b(s)\,\nabla_\theta\log\pi_\theta(a\mid s)\big]=b(s)\sum_a\nabla_\theta\pi_\theta(a\mid s)=b(s)\,\nabla_\theta\!\underbrace{\sum_a\pi_\theta(a\mid s)}_{=1}=0,$$

por lo que restar $b(s_t)$ a $G_t$ **no cambia la esperanza** del estimador (sigue **insesgado**):

$$\nabla_\theta J=\mathbb E_{\pi_\theta}\!\Big[\sum_t\gamma^t\,\big(G_t-b(s_t)\big)\,\nabla_\theta\log\pi_\theta(a_t\mid s_t)\Big].$$

*¿Por qué reduce la varianza?* Si $G_t$ es grande y positivo **siempre** (p. ej. recompensas todas positivas), el estimador **empuja hacia arriba todas las acciones** y la señal útil (cuál es *relativamente* mejor) queda sepultada en ruido. Restar un valor típico **centra** $G_t$. La elección natural es $b(s)=v_\pi(s)$ (la que casi minimiza la varianza): entonces $G_t-v_\pi(s_t)$ es una muestra insesgada de la **ventaja** $A_\pi(s_t,a_t)$, y el algoritmo refuerza una acción solo si fue **mejor que lo esperado en ese estado**.

### Experimento

Medimos la **varianza total** del estimador, $\operatorname{tr}\operatorname{Cov}(\hat g)=\sum_i\operatorname{Var}(\hat g_i)$, para tres versiones —sin *baseline*, *baseline* constante $b=J(\theta)$ y *baseline* de valor $b(s)=v_\pi(s)$ (exactas, para aislar el efecto)— en **tres políticas** (inicial, intermedia y casi óptima, alcanzadas por ascenso del gradiente **exacto**) y en una variante del entorno con un **bono de $+20$ al terminar el episodio** (los retornos pasan a ser **grandes y de un solo signo**), donde el efecto debería ser mucho mayor. Además comprobamos que **ninguna *baseline* introduce sesgo**.

````python
# ============================================================
# 3. Varianza del estimador con y sin baseline (y comprobacion de que no hay sesgo)
# ============================================================
def ascenso_exacto(mdp, theta, pasos, lr=1.0):
    """Ascenso del gradiente con el gradiente EXACTO (solo para fabricar politicas de referencia)."""
    th = theta.copy()
    for _ in range(pasos):
        th += lr * grad_analitico(mdp, th)
    return th

politicas = {"inicial (uniforme)": np.zeros((mdp.nS, mdp.nA))}
politicas["intermedia"] = ascenso_exacto(mdp, politicas["inicial (uniforme)"], 40)
politicas["casi optima"] = ascenso_exacto(mdp, politicas["intermedia"], 400)
for nombre, th in politicas.items():
    print(f"  politica {nombre:<20} J(theta) = {J_exacto(mdp, th):.4f}  (J* = {J_opt:.4f})")

N_VAR = 3000
t0 = time.time()
filas, z_ini = [], {}
for nombre, th in politicas.items():
    g_ex = grad_analitico(mdp, th).ravel()
    for base in ("ninguna", "constante", "valor"):
        G = muestras_gradiente(mdp, th, N_VAR, base, np.random.default_rng(SEED + 7))
        ee = G.std(axis=0, ddof=1) / np.sqrt(N_VAR)
        if nombre.startswith("inicial"):                                   # insesgadez: en la politica inicial todos los estados se visitan MUCHO y el z es fiable
            z_max = np.abs((G.mean(axis=0) - g_ex)[ee > 0] / ee[ee > 0]).max()
            assert z_max < 4.5, (base, z_max)                              # NINGUNA baseline introduce sesgo detectable
            z_ini[base] = z_max
        filas.append({"politica": nombre, "baseline": base, "varianza total": G.var(axis=0, ddof=1).sum()})
tab = pd.DataFrame(filas)
display(tab.pivot(index="politica", columns="baseline", values="varianza total").loc[list(politicas)][["ninguna", "constante", "valor"]].round(3))
print(f"({time.time() - t0:.1f} s)  Insesgadez en la politica inicial (max |z| de las 36 componentes, 3000 episodios): { {b: round(float(v), 2) for b, v in z_ini.items()} } (< 4.5)")
print("(En politicas casi optimas el gradiente es diminuto y algunos estados se visitan muy rara vez: el z-score de esas componentes no es fiable, por eso la prueba de sesgo se hace donde hay senal.)")
# Identidad exacta detras de la insesgadez: E_{a~pi}[grad log pi(a|s)] = sum_a pi(a|s) (e_a - pi(.|s)) = 0 en todo estado, para toda politica
for th in politicas.values():
    pi_ = softmax_filas(th)
    assert np.allclose((pi_[:, :, None] * (np.eye(mdp.nA)[None, :, :] - pi_[:, None, :])).sum(axis=1), 0.0)
print("Identidad E[grad log pi] = 0 verificada exactamente en las tres politicas.")

pv = tab.pivot(index="politica", columns="baseline", values="varianza total")
assert (pv["valor"] < pv["ninguna"]).all()                                 # la baseline de valor reduce la varianza en las tres politicas

# Variante con retornos de escala grande y de un solo signo: +20 al terminar el episodio (sea por la meta o por la trampa)
mdp_b = mundo_4x3(gamma=0.99, bono_terminal=20.0)
fb = []
for base in ("ninguna", "constante", "valor"):
    G = muestras_gradiente(mdp_b, np.zeros((mdp_b.nS, mdp_b.nA)), N_VAR, base, np.random.default_rng(SEED + 9))
    fb.append({"baseline": base, "varianza total": G.var(axis=0, ddof=1).sum()})
fb = pd.DataFrame(fb).set_index("baseline")
display(fb.round(2).T)
red_b = fb.loc["ninguna", "varianza total"] / fb.loc["valor", "varianza total"]
red_a = pv.loc["inicial (uniforme)", "ninguna"] / pv.loc["inicial (uniforme)", "valor"]
print(f"Reduccion de varianza con la baseline de valor: entorno original (politica inicial) = {red_a:.1f}x | con bono terminal de +20 = {red_b:.1f}x")
assert red_b > 3 * red_a and red_b > 8                                     # el efecto es MUCHO mayor cuando los retornos son grandes y de un solo signo

fig, axes = plt.subplots(1, 2, figsize=(14, 4.6))
ancho, x = 0.26, np.arange(3)
for k, (base, color) in enumerate(zip(("ninguna", "constante", "valor"), (PALETA[1], PALETA[3], PALETA[2]))):
    axes[0].bar(x + (k - 1) * ancho, [pv.loc[n, base] for n in politicas], ancho, color=color, label=f"baseline: {base}")
axes[0].set_xticks(x); axes[0].set_xticklabels(list(politicas)); axes[0].set_ylabel("Varianza total del estimador (3000 episodios)")
axes[0].set_title("Varianza de REINFORCE según la baseline y la política", fontweight="bold", fontsize=10); axes[0].legend(fontsize=9)
axes[1].bar(np.arange(3), fb["varianza total"].values, color=(PALETA[1], PALETA[3], PALETA[2]))
axes[1].set_xticks(np.arange(3)); axes[1].set_xticklabels(["ninguna", "constante", "valor"]); axes[1].set_yscale("log")
axes[1].set_ylabel("Varianza total (escala log)"); axes[1].set_title(f"Retornos grandes y positivos (bono +20): la baseline reduce la varianza {red_b:.0f} veces", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

## 4. REINFORCE Completo desde Cero 🔁

Con el estimador y la *baseline* ya verificados, el algoritmo es un **bucle de tres pasos**:

1. **Muestrear** un lote de $B$ episodios con la política actual $\pi_\theta$.
2. **Calcular** el gradiente estimado: promedio de $\sum_t\gamma^t(G_t-b(s_t))\nabla\log\pi_\theta(a_t\mid s_t)$.
3. **Actualizar** $\theta\leftarrow\theta+\eta\,\hat g$ (ascenso).

La *baseline* de valor ahora se **aprende** a la vez: $\hat v(s)$ es un promedio incremental de los retornos observados desde $s$ (es una *baseline*, **no** un crítico: se estima con retornos completos de Monte Carlo, sin *bootstrapping*). Como conocemos $P$, **seguimos el objetivo exacto** $J(\theta_k)$ en cada iteración (en la práctica solo se ve el retorno ruidoso de los episodios). Comparamos las tres variantes con **8 semillas** y un **lote pequeño** ($B=2$ episodios por actualización), donde el ruido del estimador sí importa; los resultados se informan **tal cual**, también la variabilidad entre semillas.

````python
# ============================================================
# 4. REINFORCE tabular (softmax) DESDE CERO, con tres baselines, 8 semillas
# ============================================================
def reinforce(mdp, iteraciones, lote, lr, base, semilla):
    """Devuelve (theta final, J exacto en cada iteracion). base: 'ninguna' | 'constante' (media movil de retornos) | 'valor' (V(s) aprendido por Monte Carlo)."""
    rng = np.random.default_rng(semilla)
    theta = np.zeros((mdp.nS, mdp.nA))
    V_hat, visitas, b_const = np.zeros(mdp.nS), np.zeros(mdp.nS), 0.0
    historia = []
    for it in range(iteraciones):
        historia.append(J_exacto(mdp, theta))
        pi = softmax_filas(theta); cdf = np.cumsum(pi, axis=1)
        g = np.zeros_like(theta); retornos_0 = []
        for _ in range(lote):
            S, A, Rw = muestrear_episodio(mdp, cdf, rng)
            b = {"ninguna": 0.0, "constante": b_const, "valor": V_hat}[base]            # la baseline usa SOLO informacion previa a este episodio
            g += gradiente_episodio(mdp, pi, S, A, Rw, b)
            G = retornos_hacia_atras(Rw, mdp.gamma); retornos_0.append(G[0])
            if base == "valor":                                                         # actualizar V(s) con los retornos de este episodio
                for t, s in enumerate(S):
                    visitas[s] += 1; V_hat[s] += (G[t] - V_hat[s]) / visitas[s] ** 0.6
        b_const = np.mean(retornos_0) if it == 0 else 0.9 * b_const + 0.1 * np.mean(retornos_0)
        theta += lr * g / lote
    historia.append(J_exacto(mdp, theta))
    return theta, np.array(historia)

ITER, LOTE, LR, SEMS = 400, 2, 0.5, range(8)
t0 = time.time()
corridas = {b: [reinforce(mdp, ITER, LOTE, LR, b, k) for k in SEMS] for b in ("ninguna", "constante", "valor")}
print(f"REINFORCE ({len(SEMS)} semillas x 3 baselines x {ITER} iteraciones, lote = {LOTE}): {time.time() - t0:.1f} s")

filas = []
for b, lista in corridas.items():
    J = np.array([h for _, h in lista])
    filas.append({"baseline": b, "J a las 100 it.": J[:, 100].mean(), "J a las 200 it.": J[:, 200].mean(), "J final (media)": J[:, -1].mean(),
                  "J final (min)": J[:, -1].min(), "J final (desv.)": J[:, -1].std(), "% de J* (media final)": 100 * J[:, -1].mean() / J_opt})
display(pd.DataFrame(filas).set_index("baseline").round(3))

Jn, Jv = np.array([h for _, h in corridas["ninguna"]]), np.array([h for _, h in corridas["valor"]])
assert Jv[:, -1].mean() > 0.93 * J_opt and Jn[:, -1].mean() > 0.85 * J_opt        # ambas aprenden una politica cercana a la optima
assert Jv[:, 100].mean() > Jn[:, 100].mean() + 0.03                               # la baseline de valor acelera el aprendizaje (observado ~ +0.08)
assert Jv[:, -1].std() < Jn[:, -1].std()                                          # y reduce la dispersion entre semillas

fig, axes = plt.subplots(1, 2, figsize=(15, 4.9), gridspec_kw={"width_ratios": [1.5, 1]})
for (b, lista), color in zip(corridas.items(), (PALETA[1], PALETA[3], PALETA[2])):
    J = np.array([h for _, h in lista]); m, sd = J.mean(axis=0), J.std(axis=0)
    axes[0].plot(m, color=color, lw=2.3, label=f"baseline: {b}"); axes[0].fill_between(np.arange(len(m)), m - sd, m + sd, color=color, alpha=0.15)
axes[0].axhline(J_opt, color="black", ls=":", label=r"$J_*$ (óptimo, iteración de valor)")
axes[0].set_xlabel("Iteración (cada una = 2 episodios)"); axes[0].set_ylabel(r"$J(\theta)$ exacto de la política actual (media ± desv., 8 semillas)")
axes[0].set_title("REINFORCE: la baseline de valor acelera y estabiliza", fontweight="bold", fontsize=10); axes[0].legend(fontsize=9, loc="lower right"); axes[0].set_ylim(0.2, 0.75)
theta_fin = corridas["valor"][0][0]; pi_fin = softmax_filas(theta_fin)
pintar_mapa(axes[1], MAPA_4x3, V=evaluar_exacto(mdp, pi_fin), accion=pi_fin.argmax(axis=1), etiquetas=ETIQ_4x3, fmt="{:.2f}",
            titulo=r"$v_{\pi_\theta}$ y acción más probable (REINFORCE + baseline, semilla 0)")
plt.tight_layout(); plt.show()
voraz = pi_fin.argmax(axis=1); optima = Q_opt.argmax(axis=1)
print(f"La accion mas probable de REINFORCE (semilla 0) coincide con la optima en {int((voraz == optima)[nt & (np.arange(mdp.nS) != 5)].sum())} de {int((nt & (np.arange(mdp.nS) != 5)).sum())} estados alcanzables.")
print("Observacion honesta: la baseline CONSTANTE (aqui una media movil de retornos, estimada y con retraso) NO mejoro el aprendizaje medio aunque, con valores exactos,")
print("sí reduce la varianza (tabla del §3); la baseline que depende del ESTADO fue la unica que mejoro de forma clara la velocidad y la dispersion entre semillas.")
````

## 5. Actor-Crítico y la Ventaja ($A=q-v$) 🎭

REINFORCE espera al **final del episodio** para conocer $G_t$ y tiene **alta varianza**. El **actor-crítico** combina los dos mundos:

- **Actor**: la política $\pi_\theta$ (se actualiza con el gradiente de política).
- **Crítico**: una función de valor $V_w$ (se aprende por **diferencia temporal**, como en el cuaderno 01).

En lugar del retorno completo, el actor usa el **error TD** del crítico como estimación de la ventaja:

$$\delta_t=r_{t+1}+\gamma\,V_w(s_{t+1})-V_w(s_t),\qquad \theta\leftarrow\theta+\eta\,\gamma^t\,\delta_t\,\nabla_\theta\log\pi_\theta(a_t\mid s_t).$$

*¿Por qué funciona?* Si el crítico fuera **exacto** ($V_w=v_\pi$), entonces
$\mathbb E[\delta_t\mid s_t,a_t]=r(s,a)+\gamma\sum_{s'}P\,v_\pi(s')-v_\pi(s)=q_\pi(s,a)-v_\pi(s)=A_\pi(s,a)$: el error TD es una **muestra insesgada de la ventaja**. Y como el teorema del gradiente se escribe con $A_\pi$ en lugar de $q_\pi$ (la *baseline* no lo cambia), el actor sigue un gradiente correcto. El precio: un crítico **aproximado** introduce **sesgo** (a cambio de mucha menos varianza), y hay que entrenar **dos** objetos.

Verificamos los dos hechos con el MDP exacto: (i) $\mathbb E_{a\sim\pi}[A_\pi(s,a)]=0$ en todo estado y (ii) $\mathbb E[\delta\mid s,a]=A_\pi(s,a)$ cuando el crítico es exacto, y (iii) que el teorema escrito con $A$ da **el mismo gradiente** que con $q$. **No implementamos el algoritmo completo** (A2C, PPO y compañía se mencionan en el cuaderno 03).

````python
# ============================================================
# 5. La ventaja A = q - v: propiedades exactas y relacion con el error TD
# ============================================================
th = politicas["intermedia"]
pi = softmax_filas(th); V = evaluar_exacto(mdp, pi); Q = mdp.r + mdp.gamma * mdp.P @ V
A = Q - V[:, None]

assert np.allclose((pi * A).sum(axis=1), 0.0)                              # (i) E_pi[A(s, .)] = 0 en todo estado
print("(i)   E_pi[A(s,.)] = 0 en todos los estados: max |.| =", f"{np.abs((pi * A).sum(axis=1)).max():.1e}")

# (ii) E[delta | s, a] = sum_s' P(s'|s,a) [R(s,a,s') + gamma V(s') - V(s)] = A(s, a)  (con el critico exacto)
delta_esperado = np.einsum("sat,sat->sa", mdp.P, mdp.R + mdp.gamma * V[None, None, :] - V[:, None, None])
assert np.allclose(delta_esperado, A)
print("(ii)  E[error TD | s, a] == A(s, a) con el critico exacto: max |diferencia| =", f"{np.abs(delta_esperado - A).max():.1e}")

# (iii) el teorema escrito con A (o con q) da el mismo gradiente
d = ocupacion_descontada(mdp, pi)
g_con_A = d[:, None] * pi * A
g_con_Q = d[:, None] * pi * (Q - (pi * Q).sum(axis=1, keepdims=True))      # restar la baseline v(s) = sum_a pi q  (identica a A)
g_sin_baseline = np.zeros_like(th)                                          # gradiente con q SIN restar nada: d(s) sum_a q(s,a) dpi(a|s)/dtheta_b
for b in range(mdp.nA):
    g_sin_baseline[:, b] = d * ((Q * (-pi[:, [b]] * pi)).sum(axis=1) + Q[:, b] * pi[:, b])
assert np.allclose(g_con_A, grad_analitico(mdp, th)) and np.allclose(g_con_A, g_con_Q)
assert np.allclose(g_sin_baseline, g_con_A)                                 # restar v(s) no cambia el gradiente exacto (baseline)
print("(iii) grad con A == grad con q (sin y con baseline) == grad analitico: OK")

fig, axes = plt.subplots(1, 3, figsize=(18, 4.6))
pintar_mapa(axes[0], MAPA_4x3, V=V, titulo=r"$v_\pi(s)$ (política intermedia)", etiquetas=ETIQ_4x3, fmt="{:.2f}")
mejor_a = A.argmax(axis=1)
pintar_mapa(axes[1], MAPA_4x3, V=A.max(axis=1), accion=mejor_a, titulo=r"$\max_a A_\pi(s,a)$ y la acción con mayor ventaja", etiquetas=ETIQ_4x3, fmt="{:.3f}", vmax=max(A.max(), 1e-9))
est = 8                                                                      # estado inicial
axes[2].bar(range(4), A[est], color=[PALETA[2] if x > 0 else PALETA[1] for x in A[est]])
axes[2].axhline(0, color="black", lw=1); axes[2].set_xticks(range(4)); axes[2].set_xticklabels([NOMBRE_ACCION[a] for a in range(4)])
axes[2].set_ylabel(r"$A_\pi(s_0,a)=q_\pi-v_\pi$"); axes[2].set_title(f"Ventaja de cada acción en el estado inicial (π(a|s₀) = {np.round(pi[est], 2)})", fontweight="bold", fontsize=9)
plt.tight_layout(); plt.show()
print("La ventaja es positiva para las acciones MEJORES que el promedio de la politica y negativa para las peores; el gradiente refuerza las primeras.")
````

## 6. Comparación: *Value-Based*, *Policy-Based* y Actor-Crítico 📊

| Criterio | **Value-based** (Q-learning, DQN) | **Policy-based** (REINFORCE) | **Actor-crítico** (A2C, PPO, SAC…) |
|---|---|---|---|
| **Qué se aprende** | $q(s,a)$; la política es $\arg\max$ (implícita) | La política $\pi_\theta$ (explícita) | Política **y** función de valor |
| **Acciones continuas** | Difícil: $\arg\max_a q$ en cada paso | **Natural**: política gaussiana, etc. | **Natural** |
| **Políticas estocásticas** | Solo vía $\varepsilon$-*greedy* (artificial) | **Sí** (puede ser la óptima, ej. pasillo corto) | **Sí** |
| **Eficiencia de muestras** | **Alta** (*off-policy*, memoria de repetición) | Baja (*on-policy*: cada dato se usa una vez) | Media (PPO: varias pasadas por lote; SAC/TD3: *off-policy*) |
| **Varianza del gradiente** | Baja (objetivo TD) | **Alta** (retornos completos) | Media (crítico reduce la varianza) |
| **Sesgo** | Sesgo del *bootstrapping*; riesgo de divergencia con aproximación (*tríada letal*) | Insesgado | Sesgo del crítico aproximado |
| **Convergencia** | $Q\to q_{\ast}$ (tabular); con redes, **sin garantías** | Óptimo **local**, cambios suaves | Óptimo local; más estable con *clipping*/regiones de confianza |
| **Sensibilidad a hiperparámetros** | Alta (tasa, $\varepsilon$, red objetivo…) | Alta (tamaño del lote, tasa, *baseline*) | Alta, pero hay recetas robustas (PPO) |
| **Ejemplos** | Atari con DQN | CartPole con REINFORCE | Control continuo (SAC), ajuste de LLM con RLHF (PPO) |

**Regla práctica.** Acciones **discretas** y pocos datos disponibles por interacción costosa → *value-based* con repetición de experiencia. Acciones **continuas** o política **estocástica** necesaria → *policy-based* / actor-crítico. En la práctica moderna domina el **actor-crítico** (en especial **PPO**, por su simplicidad y robustez) y, en control continuo, variantes *off-policy* como **SAC** o **TD3**. Estos algoritmos (A2C/A3C: Mnih et al., 2016; TRPO: Schulman et al., 2015; PPO: Schulman et al., 2017; SAC: Haarnoja et al., 2018 *—citados de memoria, verificar—*) se mencionan pero **no** se implementan aquí.

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Por qué política directa** | Acciones continuas, **políticas estocásticas óptimas** (pasillo corto: $p^{\ast}\approx0.59$ vs. deterministas pésimas) y cambios **suaves** |
| **Parametrización** | Softmax $\pi_\theta(a\mid s)\propto e^{\theta_{s,a}}$; objetivo $J(\theta)=\mathbb E_{\pi_\theta}[\sum_t\gamma^t r_{t+1}]$; ascenso del gradiente |
| **Truco log-derivada** | $\nabla p=p\,\nabla\log p$ → $\nabla J=\mathbb E[\sum_t\gamma^tG_t\nabla\log\pi(a_t\mid s_t)]$; la **dinámica $P$ desaparece** del gradiente |
| **Teorema (forma cerrada)** | $\partial J/\partial\theta_{s,b}=d_\pi(s)\,\pi(b\mid s)\,A_\pi(s,b)$ con $d_\pi$ la ocupación descontada |
| **Verificación triple** | Fórmula analítica = diferencias finitas ($<10^{-8}$) = Monte Carlo (error $\propto1/\sqrt N$, sin sesgo detectable) |
| **Varianza y *baseline*** | Restar $b(s)$ **no sesga** ($\mathbb E[\nabla\log\pi]=0$) y reduce la varianza; con retornos grandes y de un solo signo la reducción es **muy grande**; $b=v_\pi$ ~ ventaja |
| **REINFORCE** | Muestrear → estimar gradiente → ascender; **con baseline de valor** aprende antes y con menos dispersión entre semillas; sin ella es **inestable** con lotes pequeños |
| **Actor-crítico** | El error TD del crítico estima la ventaja ($\mathbb E[\delta\mid s,a]=A_\pi$ con crítico exacto): menos varianza, **algo de sesgo** |
| **Elegir enfoque** | *Value-based*: discreto y eficiente en datos; *policy-based*: continuo/estocástico; **actor-crítico** (PPO, SAC): lo habitual hoy |
