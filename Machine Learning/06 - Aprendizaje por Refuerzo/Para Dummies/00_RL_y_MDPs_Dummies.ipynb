## 1. El perro, el entrenador y los premios 🐶

Piensa en cómo entrenas a un perro a darte la pata. No puedes **explicarle** la tarea: él prueba cosas, y cuando hace algo bueno recibe una galleta. Con el tiempo, **repite lo que le dio galletas**. En el lenguaje del aprendizaje por refuerzo:

| En la historia del perro | En el aprendizaje por refuerzo | ¿Qué es? |
|---|---|---|
| El **perro** | **Agente** | Quien decide qué hacer |
| La **casa y el entrenador** | **Entorno** | El mundo que reacciona a lo que haces |
| Sentarse, dar la pata, ladrar | **Acciones** | Lo que el agente puede hacer |
| La **galleta** (o el regaño) | **Recompensa** | Un puntaje que solo dice «bien» o «mal», **no** dice qué hacer |
| Cómo se porta el perro en cada situación | **Política** | La «regla» con la que el agente decide |

**Lo especial:** nadie le dice al perro *«en esta situación, haz esto»*. Solo recibe premios, y **muchas veces tarde** (la galleta llega *después* de la pata, no *durante* los 10 intentos fallidos). Entender **cuál de las cosas que hizo** causó el premio es lo difícil. Además, **lo que el perro hace cambia lo que ve después**: él mismo genera su «material de estudio».

> 🧠 Un *sistema de aprendizaje* normal (como el de los cuadernos anteriores) recibe una tabla con las respuestas correctas. Aquí **no hay tabla**: solo hay **experiencia y puntos**.

## 2. Un premio que llega tarde vale menos: la *paciencia* ⏳

Te ofrecen **20 pesos hoy** o **50 pesos dentro de 10 días**. ¿Qué eliges? Depende de qué tan **paciente** seas. Un agente de refuerzo tiene una perilla de paciencia entre 0 y 1:

- **Paciencia baja (cerca de 0):** casi solo le importa lo de **ahora**; «pan para hoy».
- **Paciencia alta (cerca de 1):** valora casi igual lo que llega **más adelante**.

Cada día de espera «descuenta» el premio: con paciencia 0.9, un premio que llega mañana vale el 90% de hoy, el de pasado mañana el 90% de eso, y así. Veamos cuánto valen hoy 100 pesos que llegan dentro de varios días, y qué elegiría cada agente en tu oferta.

````python
import numpy as np
import matplotlib.pyplot as plt

# Cuanto vale HOY un premio de 100 pesos que llega dentro de k dias, segun la paciencia
dias = np.arange(0, 31)
fig, axes = plt.subplots(1, 2, figsize=(13, 4.4))
for paciencia, color in [(0.5, "#dc2626"), (0.9, "#2563eb"), (0.99, "#16a34a")]:
    axes[0].plot(dias, 100 * paciencia ** dias, color=color, lw=2.5, label=f"paciencia {paciencia}")
axes[0].set_xlabel("Días que hay que esperar"); axes[0].set_ylabel("Cuánto vale HOY un premio de 100 pesos")
axes[0].set_title("Cuanto más impaciente, más rápido se «derrite» el premio", fontweight="bold", fontsize=10); axes[0].legend()

# Tu oferta: 20 pesos hoy o 50 pesos dentro de 10 dias
for paciencia in [0.5, 0.9, 0.99]:
    valor_esperar = 50 * paciencia ** 10
    decision = "ESPERA los 50" if valor_esperar > 20 else "toma los 20 hoy"
    print(f"Paciencia {paciencia:<4}: los 50 pesos de dentro de 10 días valen hoy {valor_esperar:5.1f} -> {decision}")
colores = ["#dc2626" if 50 * p ** 10 <= 20 else "#16a34a" for p in [0.5, 0.9, 0.99]]
axes[1].bar(["0.5", "0.9", "0.99"], [50 * p ** 10 for p in [0.5, 0.9, 0.99]], color=colores)
axes[1].axhline(20, color="black", ls="--", label="Lo que te dan hoy: 20 pesos"); axes[1].legend()
axes[1].set_xlabel("Paciencia del agente"); axes[1].set_ylabel("Lo que valen hoy los 50 pesos de dentro de 10 días")
axes[1].set_title("Verde: espera. Rojo: se conforma con los 20 de hoy", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
assert 50 * 0.5 ** 10 < 20 < 50 * 0.99 ** 10                            # el impaciente toma 20; el paciente espera
````

## 3. El lago congelado resbaloso 🧊

Ahora el escenario: un lago congelado de **4×4 casillas**. Empiezas arriba a la izquierda (`S`) y quieres llegar al tesoro abajo a la derecha (`G`). Algunas casillas son **huecos** (`H`): si caes, se acabó el juego.

```
S F F F        S = salida      F = hielo firme
F H F H        H = hueco       G = tesoro (+1 punto)
F F F H
H F F G
```

Ganas **1 punto** si llegas al tesoro y 0 en cualquier otro caso. Pero el hielo está **mojado**: cuando decides ir en una dirección, **solo 1 de cada 3 veces** vas ahí; las otras 2 veces **resbalas hacia uno de los lados**. Es como caminar sobre un piso recién trapeado: **quieres ir recto, pero tu pie se va a un lado**.

Probemos tres «jugadores»: uno que **camina al azar**, y otro que sigue un **camino perfecto dibujado en el mapa** (bajar 2, ir a la derecha 2, bajar 1 y a la derecha 1). Los dejamos jugar 5 000 partidas, primero en **piso seco** y luego en **piso resbaloso**.

````python
rng = np.random.default_rng(42)
MAPA = ["SFFF", "FHFH", "FFFH", "HFFG"]
MOV = [(0, -1), (1, 0), (0, 1), (-1, 0)]                   # 0 = izquierda, 1 = abajo, 2 = derecha, 3 = arriba

def paso(f, c, accion, resbaloso):
    """Un movimiento. Piso resbaloso: 1/3 haces lo que querias, 1/3 resbalas a cada lado."""
    direcciones = [(accion - 1) % 4, accion, (accion + 1) % 4] if resbaloso else [accion]
    d = direcciones[rng.integers(len(direcciones))]
    return min(max(f + MOV[d][0], 0), 3), min(max(c + MOV[d][1], 0), 3)

def jugar(regla, resbaloso, max_pasos=100):
    """Devuelve 1 si llega al tesoro, 0 si cae en un hueco o se queda sin tiempo."""
    f = c = 0
    for _ in range(max_pasos):
        f, c = paso(f, c, regla(f, c), resbaloso)
        if MAPA[f][c] == "G": return 1
        if MAPA[f][c] == "H": return 0
    return 0

# El "camino perfecto" del mapa: en cada casilla de la ruta, que direccion tomar (en las demas, derecha)
RUTA = {(0, 0): 1, (1, 0): 1, (2, 0): 2, (2, 1): 2, (2, 2): 1, (3, 2): 2}
al_azar = lambda f, c: int(rng.integers(4))
camino_perfecto = lambda f, c: RUTA.get((f, c), 2)

N = 5000
resultados = {}
for nombre, regla in [("Al azar", al_azar), ("Camino perfecto", camino_perfecto)]:
    for suelo, resb in [("piso seco", False), ("piso resbaloso", True)]:
        resultados[(nombre, suelo)] = 100 * np.mean([jugar(regla, resb) for _ in range(N)])
        print(f"{nombre:<16} en {suelo:<14}: llega al tesoro en {resultados[(nombre, suelo)]:5.1f}% de las {N} partidas")

fig, ax = plt.subplots(figsize=(9, 4.4))
x = np.arange(2); ancho = 0.35
ax.bar(x - ancho / 2, [resultados[("Al azar", "piso seco")], resultados[("Al azar", "piso resbaloso")]], ancho, color="#f59e0b", label="Al azar")
ax.bar(x + ancho / 2, [resultados[("Camino perfecto", "piso seco")], resultados[("Camino perfecto", "piso resbaloso")]], ancho, color="#2563eb", label="Camino perfecto del mapa")
ax.set_xticks(x); ax.set_xticklabels(["Piso seco", "Piso resbaloso"]); ax.set_ylabel("% de partidas que llegan al tesoro"); ax.legend()
ax.set_title("El mismo plan perfecto: gana siempre en seco y casi nunca en resbaloso", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
assert resultados[("Camino perfecto", "piso seco")] == 100.0                      # sin resbalar, el plan perfecto siempre funciona
assert resultados[("Camino perfecto", "piso resbaloso")] < 20 and resultados[("Camino perfecto", "piso resbaloso")] > resultados[("Al azar", "piso resbaloso")]
print("\nLección: en un mundo con azar, un plan fijo no basta: hace falta saber, en CADA casilla, cuál es la mejor acción teniendo en cuenta que puedes resbalar.")
````

## 4. ¿Probar algo nuevo o repetir lo que funciona? 🍽️

Llegas a una ciudad nueva y vas a comer 30 días. El primer restaurante que pruebas es **bueno**. ¿Qué haces?

- **Repetir (explotar):** vas siempre al mismo; aseguras una comida buena, pero quizá **te pierdas uno excelente**.
- **Probar (explorar):** cada tanto pruebas otro; a veces comes mal, pero **descubres** mejores opciones.

Un agente de refuerzo vive **este dilema en cada decisión**. Si explora de más, desperdicia puntos; si no explora nada, puede quedarse con una **estrategia mediocre** para siempre. La solución típica es parecida a lo que haría una persona sensata: **explorar mucho al principio** (cuando no sabe nada) y **cada vez menos** a medida que aprende. Este dilema ya lo viste en pequeño con las «máquinas tragamonedas» del [cuaderno de paradigmas](../../00%20-%20Introduccion%20al%20Machine%20Learning/Para%20Dummies/01_Paradigmas_Dummies.ipynb); en el lago es más difícil porque **una decisión de hoy afecta qué casillas puedes pisar mañana**.

En el siguiente cuaderno verás cómo un agente **apunta notas** en una libreta para aprender, casilla por casilla, qué hacer, y descubre una estrategia que sí llega al tesoro **más de 8 de cada 10 veces**, aun con el piso resbaloso.

## Resumen en una frase 🎯

> **En el aprendizaje por refuerzo un *agente* prueba *acciones* en un *entorno*, recibe *premios* (a menudo tarde) y aprende una *regla de decisión* que le dé más premios; vale más lo pronto que lo tardío, el azar del mundo arruina los planes fijos, y siempre hay que repartirse entre probar cosas nuevas y repetir lo que funciona.**

### Lo que aprendiste hoy:

- ✅ Nadie da la respuesta correcta: solo hay **puntos**, y llegan **tarde**.
- ✅ La **paciencia** decide si prefieres 20 hoy o 50 después.
- ✅ Con **azar** en el mundo (piso resbaloso), un plan perfecto en papel puede fallar casi siempre.
- ✅ **Explorar** (probar) y **explotar** (repetir) deben equilibrarse.
