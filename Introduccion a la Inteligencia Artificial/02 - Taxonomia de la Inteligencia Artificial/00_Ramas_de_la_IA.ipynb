## 2. Aprendizaje Automático (*Machine Learning*) 📈

Es la rama dedicada a construir sistemas que **mejoran su desempeño a partir de datos y experiencia**, en lugar de seguir reglas escritas explícitamente a mano.

| Paradigma | ¿Cómo aprende? | Ejemplo típico |
|---|---|---|
| **Aprendizaje Supervisado** | A partir de ejemplos etiquetados (entrada → salida conocida) | Predecir si un correo es spam a partir de correos ya clasificados |
| **Aprendizaje No Supervisado** | Encontrando estructura en datos sin etiquetas | Agrupar clientes con hábitos de compra similares |
| **Aprendizaje por Refuerzo** | Mediante prueba, error y recompensas | Un agente que aprende a jugar ajedrez jugando contra sí mismo |

A continuación implementamos, desde cero y sin librerías externas, un clasificador supervisado sencillo del tipo **k-vecinos más cercanos (k-NN)**.

````python
# Conjunto de entrenamiento: fruta -> (peso_gramos, textura_rugosidad_0a10)
frutas_entrenamiento = [
    ("manzana", 150, 2), ("manzana", 130, 3), ("manzana", 160, 2),
    ("naranja", 170, 6), ("naranja", 180, 7), ("naranja", 165, 6),
    ("uva", 5, 1), ("uva", 6, 1), ("uva", 4, 2),
]

def distancia_euclidiana(p1, p2):
    return ((p1[0] - p2[0]) ** 2 + (p1[1] - p2[1]) ** 2) ** 0.5

def clasificar_knn(nueva_fruta, k=3):
    vecinos = sorted(
        frutas_entrenamiento,
        key=lambda ejemplo: distancia_euclidiana((ejemplo[1], ejemplo[2]), nueva_fruta)
    )[:k]
    etiquetas = [v[0] for v in vecinos]
    return max(set(etiquetas), key=etiquetas.count), vecinos

fruta_misteriosa = (155, 2.5)  # peso, textura
prediccion, vecinos_usados = clasificar_knn(fruta_misteriosa, k=3)

print(f"🍏 Fruta desconocida: peso={fruta_misteriosa[0]}g, rugosidad={fruta_misteriosa[1]}")
print(f"🔍 Los {len(vecinos_usados)} vecinos más cercanos son: {[v[0] for v in vecinos_usados]}")
print(f"✅ Predicción del modelo de Aprendizaje Automático (k-NN): {prediccion}")
````

## 3. Procesamiento de Lenguaje Natural (NLP) 🗣️

Estudia cómo lograr que un computador **comprenda, interprete y genere lenguaje humano** (texto y voz).

| Tarea de NLP | Pregunta que resuelve | Ejemplo de uso |
|---|---|---|
| Análisis de sentimiento | ¿El texto expresa una opinión positiva o negativa? | Monitoreo de reseñas de producto |
| Traducción automática | ¿Cómo se dice este texto en otro idioma? | Google Translate |
| Reconocimiento de entidades | ¿Qué nombres propios, lugares o fechas aparecen? | Extracción de datos de contratos |
| Generación de texto | ¿Cómo redactar una respuesta coherente? | Chatbots y asistentes virtuales |

Construyamos un analizador de sentimiento simple basado en un **léxico de palabras clave**, una de las técnicas más antiguas de NLP.

````python
import re

lexico_sentimiento = {
    "excelente": 2, "increíble": 2, "bueno": 1, "útil": 1,
    "malo": -1, "terrible": -2, "lento": -1, "confuso": -1,
}

def tokenizar(texto):
    return re.findall(r"[a-zñáéíóúü]+", texto.lower())

def analizar_sentimiento(texto):
    tokens = tokenizar(texto)
    puntaje = sum(lexico_sentimiento.get(t, 0) for t in tokens)
    if puntaje > 0:
        etiqueta = "positivo 😀"
    elif puntaje < 0:
        etiqueta = "negativo 😞"
    else:
        etiqueta = "neutral 😐"
    return tokens, puntaje, etiqueta

resena = "El curso es excelente y muy útil, aunque el material es un poco confuso al inicio."
tokens, puntaje, etiqueta = analizar_sentimiento(resena)

print("Reseña original:", resena)
print("Tokens detectados:", tokens)
print(f"Puntaje de sentimiento: {puntaje} -> Clasificación: {etiqueta}")
````

## 4. Visión por Computador 👁️

Busca que los sistemas de IA **interpreten y extraigan información de imágenes y video**, tal como lo hace el sistema visual humano.

| Tarea de Visión | Pregunta que resuelve | Ejemplo de uso |
|---|---|---|
| Clasificación de imágenes | ¿Qué objeto contiene esta imagen? | Identificar razas de perro en una foto |
| Detección de objetos | ¿Dónde están los objetos y cuántos hay? | Conteo de vehículos en una vía |
| Segmentación | ¿Qué píxeles pertenecen a cada objeto? | Delimitar tumores en imágenes médicas |
| Detección de bordes | ¿Dónde cambia bruscamente la intensidad de la imagen? | Paso previo en visión industrial |

Antes de que existieran las redes neuronales convolucionales, la detección de bordes se hacía con **filtros de convolución manuales**, como el operador de Sobel. Implementémoslo sobre una imagen pequeña representada como una matriz de intensidades.

````python
imagen = [
    [10, 10, 10, 200, 200],
    [10, 10, 10, 200, 200],
    [10, 10, 10, 200, 200],
    [10, 10, 10, 200, 200],
    [10, 10, 10, 200, 200],
]

nucleo_sobel_horizontal = [
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0, 1],
]

def aplicar_convolucion(imagen, nucleo):
    alto, ancho = len(imagen), len(imagen[0])
    salida = [[0] * ancho for _ in range(alto)]
    for i in range(1, alto - 1):
        for j in range(1, ancho - 1):
            suma = 0
            for ki in range(-1, 2):
                for kj in range(-1, 2):
                    suma += imagen[i + ki][j + kj] * nucleo[ki + 1][kj + 1]
            salida[i][j] = suma
    return salida

bordes = aplicar_convolucion(imagen, nucleo_sobel_horizontal)

print("Imagen original (matriz de intensidades de gris 0-255):")
for fila in imagen:
    print(fila)

print("\nMapa de bordes detectado (filtro Sobel horizontal aplicado manualmente):")
for fila in bordes:
    print(fila)
````

## 6. Sistemas Expertos 🧑‍⚕️📜

Son programas que **codifican el conocimiento de un experto humano como un conjunto de reglas** ("si-entonces") y usan un motor de inferencia para llegar a conclusiones. Fueron una de las primeras aplicaciones comerciales exitosas de la IA (décadas de 1970-1980).

| Componente | Función |
|---|---|
| **Base de conocimiento** | Conjunto de hechos y reglas del dominio, capturadas de expertos humanos |
| **Motor de inferencia** | Evalúa qué reglas se activan dado un conjunto de hechos observados |
| **Base de hechos** | Los datos concretos del caso actual que se están evaluando |

Implementemos un mini sistema experto de **triage médico básico** con un puñado de reglas.

````python
hechos_paciente = {
    "fiebre": True,
    "tos": True,
    "dificultad_respirar": False,
    "dolor_pecho": False,
    "dias_sintomas": 2,
}

reglas = [
    {
        "nombre": "Regla de Emergencia Respiratoria",
        "condicion": lambda h: h.get("dificultad_respirar") or h.get("dolor_pecho"),
        "conclusion": "🚨 Remitir a urgencias de inmediato.",
    },
    {
        "nombre": "Regla de Infección Respiratoria Común",
        "condicion": lambda h: h.get("fiebre") and h.get("tos") and h.get("dias_sintomas", 0) <= 5,
        "conclusion": "🩺 Probable infección respiratoria leve: reposo, líquidos y control en 48h.",
    },
    {
        "nombre": "Regla de Seguimiento Prolongado",
        "condicion": lambda h: h.get("tos") and h.get("dias_sintomas", 0) > 5,
        "conclusion": "📋 Tos persistente: remitir a valoración médica presencial.",
    },
]

def motor_inferencia(hechos, base_reglas):
    conclusiones_activadas = []
    for regla in base_reglas:
        if regla["condicion"](hechos):
            conclusiones_activadas.append((regla["nombre"], regla["conclusion"]))
    return conclusiones_activadas

resultados = motor_inferencia(hechos_paciente, reglas)

print("Hechos del paciente:", hechos_paciente)
print("\nReglas activadas por el motor de inferencia:")
for nombre_regla, conclusion in resultados:
    print(f"• {nombre_regla} -> {conclusion}")
````

## 7. Planificación (*Planning*) 🗺️✅

| Elemento | Descripción | Ejemplo (logística de entregas) |
|---|---|---|
| Estado inicial | Situación de partida | Todos los paquetes están en la bodega |
| Estado meta | Situación deseada | Todos los paquetes fueron entregados |
| Operadores | Acciones disponibles con precondiciones y efectos | "Cargar paquete", "Conducir a destino", "Descargar paquete" |
| Plan | Secuencia de operadores que conecta inicio y meta | Cargar → Conducir → Descargar (repetido por cada paquete) |

## 📌 Resumen / Puntos Clave

* La IA se organiza en **ramas o subcampos** con objetivos y técnicas propias: Aprendizaje Automático, NLP, Visión por Computador, Robótica, Sistemas Expertos, Planificación e IA Generativa.
* El **Aprendizaje Automático** aprende patrones a partir de datos y experiencia (supervisado, no supervisado, por refuerzo).
* El **NLP** y la **Visión por Computador** permiten a los sistemas comprender lenguaje e imágenes, respectivamente.
* Los **Sistemas Expertos** codifican conocimiento humano como reglas explícitas, mientras que la **Planificación** busca secuencias de acciones hacia una meta.
* La **IA Generativa** crea contenido nuevo y suele combinarse con otras ramas (especialmente NLP) para funcionar.
* En sistemas reales, estas ramas casi nunca operan en aislamiento: se **integran** para resolver problemas complejos.
