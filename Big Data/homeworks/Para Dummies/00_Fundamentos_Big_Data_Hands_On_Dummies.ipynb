Además, en la hora pico **llegan 3 clientes por minuto**: una caja atiende 2 por minuto y dos cajas atienden 4. La fila de cada minuto es: *fila nueva = lo que había + los que llegan - los que atiendo* (nunca menor que 0).

1. `mes_mil`: el primer mes en que los pedidos **superan 1 000** (el mes 1 tiene 40).
2. `fila_una`: la fila tras **10 minutos** con una caja (atiende 2 por minuto).
3. `fila_dos`: la fila tras 10 minutos con dos cajas (atienden 4 por minuto).
