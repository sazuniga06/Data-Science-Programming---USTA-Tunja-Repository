```python
# ============================================================
# Preparación (no modificar): imports y datos inventados
# ============================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.text import Annotation

SEMILLA = 21
rng_rifa = np.random.default_rng(SEMILLA)


def fila_del_cine(n):
    """La fila llega persona por persona (un generador): no se puede contar de antemano ni volver atrás."""
    for i in range(1, n + 1):
        yield f"Persona {i}"


# 20 000 ventas de «La Esquina», con dos cajas: la Caja 1 vende cosas baratas y la Caja 2 cosas caras
rng = np.random.default_rng(SEMILLA + 1)
n1, n2 = 15_000, 5_000
peso_hora = np.array([0.2, 0.1, 0.1, 0.1, 0.1, 0.3, 1, 2, 3, 3, 4, 6, 7, 10, 6, 4, 4, 5, 6, 4, 3, 2, 1, 0.5])
horas = rng.choice(24, size=n1 + n2, p=peso_hora / peso_hora.sum())
ventas = pd.DataFrame({
    "caja": ["Caja 1"] * n1 + ["Caja 2"] * n2,
    "minuto": horas * 60 + rng.uniform(0, 60, n1 + n2),                                   # minuto del día
    "monto": np.concatenate([rng.gamma(4, 2000, n1), rng.gamma(4, 5000, n2)]).round(),    # pesos
}).sample(frac=1, random_state=SEMILLA).reset_index(drop=True)

print("Ventas:", ventas.shape)
display(ventas.head(3))
```

## Reto 1: La rifa justa en la fila del cine 🎟️

La taquilla quiere sortear **3 bonos** entre quienes pasen hoy por la fila, pero nadie sabe cuántas personas llegarán y no se puede anotar a todas. El truco del módulo (*muestreo de reservorio*) es como una **bolsa de $k$ cupos**:

- Las primeras $k$ personas **entran** a la bolsa.
- La persona número $n$ (contando desde 1) **entra con probabilidad $k/n$**: sacas un número al azar entre $0$ y $n-1$; si cae en uno de los $k$ cupos (es menor que $k$), esa persona **reemplaza** a quien ocupaba ese cupo.

**Pasos**
1. Completa la función `rifa(fila, k, rng)`: recorre `fila` **una sola vez** (¡sin `len` ni listas con todos!) y devuelve la lista de $k$ premiados.
2. `ganadores`: el resultado de rifar 3 bonos en una fila de 500 personas con `rng_rifa`.
3. `prob_teorica`: la probabilidad de que una persona cualquiera de las 500 resulte premiada ($k/n$).

**Puntaje sugerido:** 30 puntos.

```python
# Casillas que debes llenar (empiezan vacías; no las borres)
ganadores = prob_teorica = None


# 💡 1) La bolsa de k cupos
def rifa(fila, k, rng):
    premios = []
    vistos = 0
    for persona in fila:
        vistos += 1
        if len(premios) < ...:            # 💡 mientras haya cupos libres, todos entran
            premios.append(persona)
        else:
            j = rng.integers(0, ...)      # 💡 un número al azar entre 0 y vistos - 1 (el segundo límite no se incluye)
            if j < ...:                   # 💡 solo si cae en uno de los k cupos
                premios[j] = persona
    return premios


# 💡 2) Rifar 3 bonos entre 500 personas
ganadores = rifa(fila_del_cine(500), ..., rng_rifa)

# 💡 3) Cada persona tiene la misma oportunidad: k / n
prob_teorica = ... / 500
```

```python
# Autoverificación del Reto 1 (no modificar)
def _rifar(n, k, rng_):
    try:
        return list(rifa(fila_del_cine(n), k, rng_))
    except Exception as e:
        raise AssertionError(f"Pista 🧸: tu función falló ({type(e).__name__}). Revisa que no queden «...» y que recorras la fila con un for, sin len().")

assert ganadores is not None and len(ganadores) == 3 and len(set(ganadores)) == 3, "Pista 🧸: ganadores debe tener 3 personas distintas: rifa(fila_del_cine(500), 3, rng_rifa)."
assert all(str(g).startswith("Persona ") for g in ganadores), "Los ganadores deben ser personas de la fila."
assert prob_teorica is not None and abs(prob_teorica - 3 / 500) < 1e-12, "prob_teorica = k / n = 3 / 500."
assert sorted(_rifar(2, 3, np.random.default_rng(0))) == ["Persona 1", "Persona 2"], "Si llegan menos personas que bonos, todas ganan: la bolsa no se llena."
_rng = np.random.default_rng(99)
_veces = {f"Persona {i}": 0 for i in range(1, 13)}
for _ in range(6000):
    _g = _rifar(12, 3, _rng)
    assert len(_g) == 3 and len(set(_g)) == 3, "Cada rifa debe devolver 3 personas distintas."
    for _p in _g:
        _veces[_p] += 1
_frec = np.array(list(_veces.values())) / 6000
assert np.all(np.abs(_frec - 3 / 12) < 0.04), f"En una rifa justa cada una de las 12 personas gana ≈ 25 % de las veces; la tuya: de {_frec.min():.0%} a {_frec.max():.0%}. Pista 🧸: ¿usas la probabilidad k / n de la persona número n (y no siempre la misma)?"
print(f"✅ ¡Reto 1 superado! Cada persona de una fila de 12 ganó entre {_frec.min():.0%} y {_frec.max():.0%} de las veces (lo justo: 25 %). Tus ganadores de hoy: {ganadores}")
```

## Reto 2: Contar por hora, pesar los promedios y contar el cuento 🗂️📖

La tienda de la esquina tiene **20 000 ventas**. Dibujar un alfiler por venta sería una mancha ilegible: mejor **contar por hora** (como contar casas por manzana) y dibujar **24 barras**. Además hay dos cajas («ayudantes») que te entregan **su promedio**, y hay que combinarlos con justicia.

**Pasos**
1. `por_caja`: para cada caja, su promedio (`mean`) y cuántas ventas contó (`size`). `promedio_ingenuo` es el promedio de los dos promedios (¡sin pesar!) y `promedio_justo` pesa cada promedio por su número de ventas.
2. Agrupa por hora: `ventas["hora"]` (los minutos entre 60 dan la hora), `ventas_por_hora` (cuántas ventas hay en cada hora) y `hora_pico` (la hora con más ventas).
3. Dibuja `fig, ax`: 24 barras donde **solo la hora pico** tenga otro color; un **título que diga la conclusión** (nombrando la hora pico, no «Ventas por hora»); los **ejes con nombre**; y **una anotación con flecha** sobre la barra más alta.

**Puntaje sugerido:** 45 puntos.

```python
# Casillas que debes llenar
por_caja = promedio_ingenuo = promedio_justo = None
ventas_por_hora = hora_pico = fig = ax = None

# 💡 1) Cada caja es un «ayudante»: su promedio y cuántas ventas contó
por_caja = ventas.groupby("caja")["monto"].agg(["mean", "size"])
promedio_ingenuo = por_caja["mean"].mean()                                        # promedio de promedios, sin pesar
promedio_justo = (por_caja["mean"] * por_caja[...]).sum() / por_caja[...].sum()   # 💡 cada promedio pesa según su columna "size"

# 💡 2) Contar por hora
ventas["hora"] = (ventas["minuto"] // ...).astype(int)     # 💡 una hora tiene 60 minutos
ventas_por_hora = ventas.groupby(...).size()               # 💡 agrupa por la columna "hora"
hora_pico = ventas_por_hora.idxmax()                       # idxmax() da la ETIQUETA (la hora) del valor más alto

# 💡 3) El cuento: barras grises y la hora pico en naranja
fig, ax = plt.subplots(figsize=(8, 4))
colores = ["#f59e0b" if h == hora_pico else "#cbd5e1" for h in ventas_por_hora.index]
ax.bar(ventas_por_hora.index, ventas_por_hora.values, color=colores)
ax.set_title("...")        # 💡 una frase con la conclusión, p. ej. «La tienda vende más a las <hora> h: ...»
ax.set_xlabel("...")       # 💡 ¿qué hay en el eje horizontal? (con unidad)
ax.set_ylabel("...")       # 💡 ¿qué se cuenta en el eje vertical?
ax.annotate("...", xy=(hora_pico, ventas_por_hora.max()), xytext=(hora_pico + 3, ventas_por_hora.max()),
            arrowprops=dict(arrowstyle="->"))      # 💡 el texto de la anotación debe decir cuántas ventas hay en la hora pico
plt.show()
```

```python
# Autoverificación del Reto 2 (no modificar)
_pc = ventas.groupby("caja")["monto"].agg(["mean", "size"])
assert isinstance(por_caja, pd.DataFrame) and np.allclose(por_caja.sort_index()["mean"], _pc["mean"]) and (por_caja.sort_index()["size"] == _pc["size"]).all(), "Pista 🧸: por_caja = ventas.groupby('caja')['monto'].agg(['mean', 'size'])."
assert promedio_justo is not None and abs(promedio_justo - ventas["monto"].mean()) < 1e-6, "Pista 🧸: promedio_justo = (promedios × número de ventas).sum() / total de ventas; debe dar el promedio de las 20 000 ventas."
assert abs(promedio_ingenuo - _pc["mean"].mean()) < 1e-6 and abs(promedio_ingenuo - promedio_justo) > 500, "promedio_ingenuo es por_caja['mean'].mean(); como la Caja 1 tiene el triple de ventas, no puede coincidir con el justo."
assert ventas["hora"].notna().all() and np.array_equal(ventas["hora"].to_numpy(), (ventas["minuto"] // 60).astype(int).to_numpy()), "Pista 🧸: ventas['hora'] = (ventas['minuto'] // 60).astype(int)."
assert ventas_por_hora is not None and len(ventas_por_hora) == 24 and int(ventas_por_hora.sum()) == len(ventas), "Pista 🧸: ventas_por_hora debe tener 24 horas y sumar las 20 000 ventas: ventas.groupby('hora').size()."
assert hora_pico == ventas_por_hora.idxmax(), "hora_pico es la ETIQUETA de la hora con más ventas: ventas_por_hora.idxmax()."
# ---- El gráfico (se revisan sus propiedades, no los píxeles) ----
assert isinstance(fig, plt.Figure) and ax in fig.axes, "Pista 🧸: fig, ax = plt.subplots(...) y dibuja las barras en ax."
assert len(ax.patches) == 24, "El gráfico debe tener 24 barras (una por hora), no 20 000 puntos."
_col = [p.get_facecolor() for p in ax.patches]
assert _col.count(_col[hora_pico]) == 1 and len(set(_col)) == 2, "Solo la barra de la hora pico debe tener otro color (resaltar lo importante)."
_t = ax.get_title()
assert _t.strip() not in ("", "...") and len(_t.split()) >= 4 and str(hora_pico) in _t, "Pista 🧸: el título debe decir la conclusión (al menos 4 palabras) y nombrar la hora pico."
assert ax.get_xlabel().strip() not in ("", "...") and ax.get_ylabel().strip() not in ("", "..."), "Pista 🧸: ponle nombre a los dos ejes."
_an = [t for t in ax.texts if isinstance(t, Annotation)]
assert len(_an) == 1 and _an[0].get_text().strip() not in ("", "..."), "Pon exactamente UNA anotación con texto (ax.annotate)."
assert abs(_an[0].xy[0] - hora_pico) <= 0.5 and abs(_an[0].xy[1] - ventas_por_hora.max()) <= 1, "La flecha debe apuntar a la barra más alta: xy=(hora_pico, ventas_por_hora.max())."
assert ax.get_ylim()[0] <= 0, "En barras el eje vertical empieza en 0 (si no, la regla torcida exagera las diferencias)."
print(f"✅ ¡Reto 2 superado! Promedio ingenuo ${promedio_ingenuo:,.0f} frente al justo ${promedio_justo:,.0f}; la hora pico es las {hora_pico} h con {ventas_por_hora.max():,} ventas.")
```
