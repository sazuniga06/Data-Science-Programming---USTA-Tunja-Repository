## Reto 2: El libro de actas a prueba de borrones 📖

Cada hoja del libro tiene tres cosas: el **texto**, la **huella de la hoja anterior** y **su propia huella**, que se calcula con el texto *y* la huella anterior (así todas las hojas quedan «amarradas»). La huella se obtiene con `hashlib.sha256` sobre el texto `anterior + "|" + texto` y se escribe en hexadecimal.

**Pasos**
1. Completa `huella(anterior, texto)` y `armar_libro(actas)`: una lista de hojas `{"texto", "anterior", "huella"}`; la primera hoja usa `INICIO` como «anterior».
2. Completa `libro_ok(libro, ancla=None)`: revisa hoja por hoja que su «anterior» sea la huella de la hoja de antes (o `INICIO`) y que su huella coincida con la que recalculas. Si te dan un `ancla` (la última huella, guardada **afuera** del libro), también debe ser igual a la última huella del libro.
3. `libro = armar_libro(ACTAS)` y `ancla` = la huella de la última hoja.

**Puntaje sugerido:** 30 puntos.

```python
# Casillas que debes llenar
libro = ancla = None


# 💡 1) La huella digital y el libro encadenado
def huella(anterior, texto):
    return hashlib.sha256((... + "|" + ...).encode("utf-8")).hexdigest()      # 💡 anterior, una barra «|» y el texto


def armar_libro(actas):
    libro = []
    anterior = INICIO
    for texto in actas:
        h = huella(anterior, ...)                                              # 💡 la huella de ESTA hoja
        libro.append({"texto": texto, "anterior": anterior, "huella": h})
        anterior = ...                                                         # 💡 la huella de esta hoja es el «anterior» de la siguiente
    return libro


# 💡 2) Revisar el libro
def libro_ok(libro, ancla=None):
    anterior = INICIO
    for hoja in libro:
        if hoja["anterior"] != anterior or hoja["huella"] != huella(anterior, hoja[...]):    # 💡 ¿coincide lo que dice la hoja con lo que calculas?
            return False
        anterior = hoja["huella"]
    if ancla is not None and anterior != ...:                                  # 💡 la última huella debe ser igual al ancla guardada afuera
        return False
    return True


libro = armar_libro(ACTAS)
ancla = libro[-1][...]                                                         # 💡 la huella de la última hoja
```

```python
# Autoverificación del Reto 2 (no modificar)
import copy
_h = lambda a, t: hashlib.sha256((a + "|" + t).encode("utf-8")).hexdigest()
_ref, _a = [], INICIO
for _t in ACTAS:
    _ref.append({"texto": _t, "anterior": _a, "huella": _h(_a, _t)})
    _a = _ref[-1]["huella"]
def _try(f, *a):
    try:
        return f(*a)
    except Exception as e:
        raise AssertionError(f"Pista 🧸: tu función falló ({type(e).__name__}). Revisa que no queden «...».")


assert _try(huella, "a", "b") == hashlib.sha256(b"a|b").hexdigest(), "Pista 🧸: huella(anterior, texto) = sha256(anterior + '|' + texto) en hexadecimal (.hexdigest())."
assert libro is not None and libro == _ref, "Pista 🧸: cada hoja es {'texto', 'anterior', 'huella'}; la primera usa INICIO y cada 'anterior' es la huella de la hoja de antes."
assert ancla == _ref[-1]["huella"], "Pista 🧸: ancla = la huella de la última hoja (libro[-1]['huella'])."
assert libro_ok(libro) is True and libro_ok(libro, ancla) is True, "Pista 🧸: un libro sin tocar debe salir bien."


def _rehacer(lib, desde):                                       # un tramposo que recalcula TODAS las huellas
    for i in range(desde, len(lib)):
        lib[i]["anterior"] = INICIO if i == 0 else lib[i - 1]["huella"]
        lib[i]["huella"] = _h(lib[i]["anterior"], lib[i]["texto"])
    return lib


_l = copy.deepcopy(libro); _l[1]["texto"] = "Se pagó la bomba usada"
assert libro_ok(_l, ancla) is False, "Pista 🧸: cambiar el texto de una hoja debe notarse (la huella ya no coincide)."
_l = copy.deepcopy(libro); del _l[2]
assert libro_ok(_l, ancla) is False and libro_ok(_l) is False, "Pista 🧸: arrancar una hoja del medio debe notarse (la siguiente apunta a una huella que ya no está)."
_l = copy.deepcopy(libro); _l[4]["huella"] = "f" * 64
assert libro_ok(_l, ancla) is False, "Pista 🧸: una huella falsificada debe notarse."
_l = copy.deepcopy(libro); _l[3]["anterior"] = "x" * 64
assert libro_ok(_l, ancla) is False, "Pista 🧸: el «anterior» escrito en cada hoja también debe coincidir con la huella de la hoja de antes."
_l = _rehacer(copy.deepcopy(libro), 0); _l[1]["texto"] = "Se pagó la bomba usada"; _l = _rehacer(_l, 1)
assert libro_ok(_l) is True, "Pista 🧸: si el tramposo recalcula TODAS las huellas, el libro por sí solo se ve «bien» (¡esa es su debilidad!)."
assert libro_ok(_l, ancla) is False, "Pista 🧸: ...pero la huella guardada afuera (el ancla) delata la trampa: la última huella ya no coincide."
_l = copy.deepcopy(libro)[:-2]
assert libro_ok(_l) is True and libro_ok(_l, ancla) is False, "Pista 🧸: arrancar las últimas hojas solo lo delata el ancla."
print(f"✅ ¡Reto 2 superado! El libro de {len(libro)} hojas detecta que cambien, arranquen o falsifiquen hojas; si alguien recalcula todo, solo la huella guardada afuera ({ancla[:10]}…) lo delata.")
```

## Reto 3: El camino de los datos — ¿qué se afecta y de dónde viene? 🧭

En la Junta, unos datos sirven para armar otros: de `lecturas_medidor` sale la `grafica_consumo`, y `SIGUIENTES` dice, para cada dato, qué otras cosas arma. Hay que contestar dos preguntas de gobernanza: **¿qué se daña si falla un dato?** (hacia adelante) y **¿de dónde viene lo que se publica?** (hacia atrás).

**Pasos**
1. Completa `recorrer(mapa, inicio)`: todos los datos a los que se llega siguiendo las flechas desde `inicio` (sin incluir `inicio`), sin repetir ninguno. Usa una lista de «pendientes» como una pila.
2. `si_falla`: lo que se afecta si falla `lecturas_medidor` (`recorrer(SIGUIENTES, ...)`).
3. `ANTERIORES`: el mapa inverso (para cada dato, de cuáles sale directamente). `viene_de`: de dónde viene `lista_deudores`.
4. `registro_usuarios` es **sensible** (tiene nombres). `hereda_sensible`: todo lo que se arma a partir de él. `riesgo`: ¿la `cartelera_publica` quedó entre ellos?

**Puntaje sugerido:** 25 puntos.

```python
# Casillas que debes llenar
si_falla = ANTERIORES = viene_de = hereda_sensible = riesgo = None


# 💡 1) Seguir las flechas
def recorrer(mapa, inicio):
    vistos = set()
    pendientes = [inicio]
    while pendientes:
        actual = pendientes.pop()
        for sig in mapa.get(actual, []):
            if sig not in ...:                  # 💡 solo si no lo habías visto
                vistos.add(...)
                pendientes.append(...)          # 💡 hay que seguir las flechas que salen de ese dato
    return vistos


# 💡 2) Hacia adelante
si_falla = recorrer(SIGUIENTES, "...")          # 💡 el dato que falla

# 💡 3) Hacia atrás: invertimos cada flecha (si a -> b, en ANTERIORES b -> a)
ANTERIORES = defaultdict(list)
for origen, destinos in SIGUIENTES.items():
    for destino in destinos:
        ANTERIORES[...].append(...)             # 💡 el destino «viene de» el origen
viene_de = recorrer(ANTERIORES, "lista_deudores")

# 💡 4) Lo sensible se hereda hacia adelante
hereda_sensible = recorrer(SIGUIENTES, "registro_usuarios")
riesgo = "cartelera_publica" in ...             # 💡 ¿está la cartelera entre lo que hereda lo sensible?
```
