## 1. La Comparación, Criterio por Criterio ⚖️

| Criterio | **Data Warehouse** (almacén) | **Data Lake** (lago) |
|---|---|---|
| **Esquema** | **En escritura** (*schema-on-write*): se define antes de cargar; lo que no encaja **no entra** | **En lectura** (*schema-on-read*): se guarda tal cual y el esquema se aplica **al consultar** |
| **Tipos de datos** | Principalmente **estructurados** (tablas); semiestructurados con límites | **Todos**: estructurados, semiestructurados (JSON, XML) y **no estructurados** (texto, imágenes, audio, *logs*) |
| **Procesamiento** | **ETL**: se transforma **antes** de cargar | **ELT**: se carga primero y se transforma **después**, según la necesidad |
| **Calidad del dato** | **Alta por construcción** (restricciones, tipos, dimensiones conformadas) | **Variable**: depende de las capas, las reglas y la disciplina; la capa cruda **no** se garantiza |
| **Usuarios típicos** | Analistas de negocio, directivos, herramientas de BI (SQL, tableros) | Ingenieros de datos, científicos de datos, también analistas sobre las capas curadas |
| **Costo** | Almacenamiento y cómputo **más caros por unidad**, y trabajo de **modelado** previo; suele guardarse **solo lo modelado** | Almacenamiento **más barato por unidad** (objetos, formatos abiertos), pero se guarda **más** y hay costo de **gobierno** y de cómputo por consulta |
| **Gobernanza y seguridad** | **Madura**: permisos por tabla/columna, auditoría, linaje del modelo | **Más difícil**: permisos por archivo o carpeta, catálogo a cargo del equipo; sin disciplina se vuelve un **pantano** |
| **Rendimiento** | **Alto y predecible** para consultas **modeladas** (columnar, índices, estadísticas) | **Variable**: depende del **formato** (JSON crudo frente a Parquet particionado) y del motor |
| **Agilidad ante el cambio** | **Baja/media**: un campo nuevo exige **cambiar el modelo**, el ETL y recargar | **Alta al escribir** (se guarda lo que llegue); el costo se **traslada al lector** |
| **Casos típicos** | Informes financieros y regulatorios, tableros de KPI, historia auditada | *Machine learning*, analítica exploratoria, *logs*, texto e imágenes, datos aún sin modelar |

**Cuatro advertencias de rigor.** (1) **No son excluyentes**: muchas organizaciones tienen **ambos** (el lago alimenta al almacén) o una arquitectura **lakehouse** (sección 4). (2) Las **diferencias de costo y de rendimiento dependen del proveedor, del motor, del formato y de la carga**: la tabla describe tendencias **cualitativas**, no cifras (no damos precios ni porcentajes porque cambian y **no se pueden generalizar**). (3) «Almacén» y «lago» describen **estilos de almacenamiento**, no productos: un motor columnar leyendo Parquet puede comportarse **como un almacén** (lo veremos en el experimento). (4) Los criterios **se acoplan**: la flexibilidad del esquema al leer **es** la causa de que la calidad sea variable.

**El experimento.** Servimos **el mismo conjunto de ventas** (sintético) así:

| Servicio | Dónde | Esquema | Formato |
|---|---|---|---|
| **Almacén (DW)** | Tabla `ventas` en DuckDB + dimensión `dim_producto` | **Estricto al escribir**: tipos, `NOT NULL`, `CHECK (cantidad > 0)`, llave foránea | Columnar interno |
| **Lago crudo** | `crudo/ventas.jsonl` | **Ninguno**: texto JSON tal como lo emitió el origen (con un campo extra, `canal`, que el almacén **no modeló**) | JSON por líneas |
| **Lago curado** | `ventas/periodo=AAAA-MM/*.parquet` | **En lectura**, con tipos del propio archivo | Parquet particionado |

````python
# !pip install -q duckdb        # (en Google Colab suele venir preinstalado; descomenta si falta)
import json, tempfile, time, sqlite3, hashlib, os, shutil
from pathlib import Path
import numpy as np
import pandas as pd
import duckdb
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

pd.set_option("display.width", 140); pd.set_option("display.max_columns", 20)
RAPIDO = True                                                    # True: 300 000 ventas (rapido). False: 1 500 000
N = 300_000 if RAPIDO else 1_500_000


def generar_ventas(n, semilla=7):
    """Ventas sinteticas. 'canal' es un campo que el origen emite pero que el almacen NO modelo."""
    rng = np.random.default_rng(semilla)
    precio = rng.integers(15, 250, 60) * 100
    prod = rng.integers(1, 61, n)
    df = pd.DataFrame({"id_venta": np.arange(1, n + 1), "fecha": np.datetime64("2025-01-01") + rng.integers(0, 365, n).astype("timedelta64[D]"),
                       "tienda": [f"T{i:02d}" for i in rng.integers(1, 9, n)], "producto_id": prod, "cliente_id": rng.integers(1, 5001, n),
                       "cantidad": rng.integers(1, 7, n)})
    df["monto"] = df.cantidad * precio[prod - 1]
    df["canal"] = rng.choice(["web", "app", "tienda"], n)
    dim = pd.DataFrame({"producto_id": np.arange(1, 61), "categoria": np.repeat(["Lácteos", "Despensa", "Bebidas", "Aseo"], 15), "precio": precio})
    return df, dim


ventas, dim_producto = generar_ventas(N)
ventas["periodo"] = ventas.fecha.dt.strftime("%Y-%m")
directorio = tempfile.TemporaryDirectory(prefix="dw_vs_lago_")
RAIZ = Path(directorio.name); LAGO = RAIZ / "lago"

# ---------------- LAGO: crudo (JSON por lineas) y curado (Parquet particionado por periodo) ----------------
t0 = time.perf_counter()
(LAGO / "crudo").mkdir(parents=True)
ventas.drop(columns="periodo").assign(fecha=ventas.fecha.dt.strftime("%Y-%m-%d")).to_json(LAGO / "crudo" / "ventas.jsonl", orient="records", lines=True)
t_json = time.perf_counter() - t0
t0 = time.perf_counter()
for periodo, g in ventas.groupby("periodo"):
    d = LAGO / "ventas" / f"periodo={periodo}"; d.mkdir(parents=True)
    g.drop(columns="periodo").to_parquet(d / "part-0.parquet", index=False)
dim_producto.to_parquet(LAGO / "dim_producto.parquet", index=False)
t_parquet = time.perf_counter() - t0

# ---------------- ALMACEN: esquema estricto al escribir ----------------
dw = duckdb.connect(":memory:")
dw.execute("""
CREATE TABLE dim_producto (producto_id INTEGER PRIMARY KEY, categoria VARCHAR NOT NULL, precio INTEGER NOT NULL);
CREATE TABLE ventas (id_venta BIGINT PRIMARY KEY, fecha DATE NOT NULL, tienda VARCHAR NOT NULL,
                     producto_id INTEGER NOT NULL REFERENCES dim_producto(producto_id), cliente_id INTEGER NOT NULL,
                     cantidad INTEGER NOT NULL CHECK (cantidad > 0), monto BIGINT NOT NULL);""")
dw.register("t_dim", dim_producto); dw.register("t_ventas", ventas)
t0 = time.perf_counter()
dw.execute("INSERT INTO dim_producto SELECT producto_id, categoria, precio FROM t_dim")
dw.execute("INSERT INTO ventas SELECT id_venta, fecha, tienda, producto_id, cliente_id, cantidad, monto FROM t_ventas")      # 'canal' NO se modelo: no entra
t_dw = time.perf_counter() - t0

mb = lambda ps: round(sum(p.stat().st_size for p in ps) / 1e6, 1)
print(f"{N:,} ventas | JSON crudo {mb([LAGO / 'crudo' / 'ventas.jsonl'])} MB ({t_json:.1f} s de escritura) | "
      f"Parquet {mb(list((LAGO / 'ventas').rglob('*.parquet')))} MB en {len(list((LAGO / 'ventas').rglob('*.parquet')))} archivos ({t_parquet:.1f} s) | carga en el almacen {t_dw:.1f} s")
assert dw.sql("SELECT COUNT(*) FROM ventas").fetchone()[0] == N == len(ventas)
assert "canal" not in dw.sql("SELECT * FROM ventas LIMIT 1").df().columns          # el almacen solo conoce lo que se modelo
assert (LAGO / "crudo" / "ventas.jsonl").stat().st_size > 5 * sum(p.stat().st_size for p in (LAGO / "ventas").rglob("*.parquet"))      # el JSON crudo pesa varias veces mas que Parquet
````

## 2. Experimento 1: la Misma Consulta Analítica, Tres Formas de Servir el Dato ⏱️

La pregunta de negocio es **fija**: *ventas totales por categoría y por mes* (una unión con la dimensión de producto y una agregación). La ejecutamos sobre (a) la **tabla del almacén**, (b) el **lago curado en Parquet** y (c) el **lago crudo en JSON**, con el **mismo motor (DuckDB)** para que la comparación sea justa en lo único que cambia: **cómo está guardado el dato**. Verificamos que **las tres respuestas son idénticas** y medimos la **mediana de 7 ejecuciones** (3 en el JSON, que es lento). Después probamos la **rigidez**: una pregunta sobre un campo que **el almacén no modeló** (`canal`) y un dato **inválido** que llega a cada sistema.

**Lo que NO se puede concluir de este experimento:** que «el lago es más rápido» o «el almacén es más rápido». Medimos **un motor, una máquina y un conjunto de tamaño modesto**; el resultado cambia con el motor, el tamaño, el formato y las estadísticas. Lo que sí muestra es **qué factor pesa**.

````python
GLOB = f"{LAGO}/ventas/*/*.parquet"
Q_DW = """SELECT p.categoria, strftime(v.fecha, '%Y-%m') AS mes, CAST(SUM(v.monto) AS BIGINT) AS monto
          FROM ventas v JOIN dim_producto p USING (producto_id) GROUP BY 1, 2 ORDER BY 1, 2"""
Q_LAGO = f"""SELECT p.categoria, strftime(v.fecha, '%Y-%m') AS mes, CAST(SUM(v.monto) AS BIGINT) AS monto
             FROM read_parquet('{GLOB}') v JOIN read_parquet('{LAGO}/dim_producto.parquet') p USING (producto_id) GROUP BY 1, 2 ORDER BY 1, 2"""
Q_JSON = f"""SELECT p.categoria, strftime(CAST(v.fecha AS DATE), '%Y-%m') AS mes, CAST(SUM(v.monto) AS BIGINT) AS monto
             FROM read_json_auto('{LAGO}/crudo/ventas.jsonl', format = 'newline_delimited') v
             JOIN read_parquet('{LAGO}/dim_producto.parquet') p USING (producto_id) GROUP BY 1, 2 ORDER BY 1, 2"""


def mediana_ms(con, q, veces):
    con.sql(q).fetchall()                                              # calentamiento
    t = []
    for _ in range(veces):
        t0 = time.perf_counter(); con.sql(q).fetchall(); t.append((time.perf_counter() - t0) * 1000)
    return float(np.median(t))


r_dw, r_lago, r_json = dw.sql(Q_DW).df(), dw.sql(Q_LAGO).df(), dw.sql(Q_JSON).df()
pd.testing.assert_frame_equal(r_dw, r_lago); pd.testing.assert_frame_equal(r_dw, r_json)          # MISMA respuesta
assert len(r_dw) == 4 * 12 and int(r_dw.monto.sum()) == int(ventas.monto.sum())
tiempos = pd.Series({"Almacén (tabla)": mediana_ms(dw, Q_DW, 7), "Lago curado (Parquet)": mediana_ms(dw, Q_LAGO, 7), "Lago crudo (JSON)": mediana_ms(dw, Q_JSON, 3)})
print("Mediana (ms):"); print(tiempos.round(1).to_string()); print("Razón frente al almacén:", (tiempos / tiempos.iloc[0]).round(1).to_dict())
assert tiempos["Lago crudo (JSON)"] > tiempos["Lago curado (Parquet)"]                           # el factor decisivo es el FORMATO, no la etiqueta

# ---------------- Rigidez 1: una pregunta sobre un campo que el almacen NO modelo ----------------
try:
    dw.sql("SELECT canal, SUM(monto) FROM ventas GROUP BY 1").fetchall(); almacen_responde = True
except duckdb.BinderException as e:
    almacen_responde = False; msg = str(e).splitlines()[0]
por_canal = dw.sql(f"SELECT canal, CAST(SUM(monto) AS BIGINT) AS monto FROM read_parquet('{GLOB}') GROUP BY 1 ORDER BY 1").df()
print(f"\nVentas por canal en el ALMACEN: {'responde' if almacen_responde else 'ERROR -> ' + msg}")
print("Ventas por canal en el LAGO (mismo dato, esquema al leer):", por_canal.set_index("canal").monto.to_dict())
assert not almacen_responde and list(por_canal.canal) == ["app", "tienda", "web"] and int(por_canal.monto.sum()) == int(ventas.monto.sum())

# ---------------- Rigidez 2: llega un dato INVALIDO (cantidad = 0) ----------------
malo = ventas.head(1).assign(id_venta=99_000_001, cantidad=0)
dw.register("t_malo", malo.drop(columns=["canal", "periodo"]))
try:
    dw.execute("INSERT INTO ventas SELECT * FROM t_malo"); rechazo_dw = None
except duckdb.ConstraintException as e:
    rechazo_dw = str(e).splitlines()[0]
(LAGO / "crudo" / "llegada_tardia.jsonl").write_text(malo.drop(columns="periodo").assign(fecha="2025-03-01").to_json(orient="records", lines=True), encoding="utf-8")   # el lago lo acepta
en_lago = dw.sql(f"SELECT COUNT(*) FROM read_json_auto('{LAGO}/crudo/llegada_tardia.jsonl', format='newline_delimited') WHERE cantidad <= 0").fetchone()[0]
print(f"\nDato invalido -> ALMACEN lo rechaza al escribir: {rechazo_dw[:80] if rechazo_dw else 'no'}...")
print(f"                 -> LAGO lo guardo sin chistar; solo una consulta de validacion lo detecta ({en_lago} fila con cantidad <= 0)")
assert rechazo_dw is not None and en_lago == 1 and dw.sql("SELECT COUNT(*) FROM ventas").fetchone()[0] == N
````

## 3. Experimento 2: Llegan Datos Nuevos con Columnas Inesperadas 🧩

Un mes después, la tienda en línea empieza a emitir **dos campos nuevos**: `cupon` (código de descuento) y `region_envio`. Nadie avisó al equipo de datos. Veamos el **costo de cambio** en cada enfoque, **paso a paso y con tiempos**:

- **Almacén:** el lote **no cabe** en la tabla. El «arreglo» obvio (insertar solo las columnas conocidas) **funciona sin error pero pierde información en silencio**. Conservar los campos exige **cambiar el modelo** (`ALTER TABLE`), **cambiar el ETL** y **recargar**.
- **Lago:** el lote se **escribe como un archivo nuevo** en una partición nueva, sin tocar nada. Pero el costo **se traslada al lector**: por defecto, DuckDB lee el esquema del **primer** archivo y **ignora** columnas nuevas; hay que pedir explícitamente `union_by_name = true`. Y si un lote cambia un **tipo** (`monto` llega como texto con comas), **la consulta falla al leer**.

Registramos cada paso en una **bitácora** (qué se hizo, si funcionó, cuánto tardó). Los milisegundos aquí son pequeños porque es **una sola tabla en una base embebida**; en una organización real el costo está en **coordinar** a quienes dependen del modelo (informes, tableros, otros equipos), que **no se puede medir** con este código.

````python
def lote_nuevo(n, desde_id, fecha, periodo, semilla=21):
    rng = np.random.default_rng(semilla)
    b = ventas.sample(n, random_state=semilla).reset_index(drop=True).drop(columns="periodo")
    b["id_venta"] = np.arange(desde_id, desde_id + n); b["fecha"] = np.datetime64(fecha)
    b["cupon"] = rng.choice(["DESC10", "DESC20", None], n); b["region_envio"] = rng.choice(["Andina", "Caribe", "Orinoquía"], n)
    return b


bitacora = []
def paso(enfoque, descripcion, fn):
    t0 = time.perf_counter()
    try:
        fn(); ok, detalle = True, ""
    except Exception as e:
        ok, detalle = False, str(e).splitlines()[0][:70]
    bitacora.append({"enfoque": enfoque, "paso": descripcion, "funciono": ok, "ms": round((time.perf_counter() - t0) * 1000, 1), "detalle": detalle})
    return ok


lote = lote_nuevo(5_000, 10_000_001, "2026-01-15", "2026-01")
dw.register("t_lote", lote)

# ---- ALMACEN ----
paso("almacén", "INSERT directo del lote (SELECT *)", lambda: dw.execute("INSERT INTO ventas SELECT * FROM t_lote"))
paso("almacén", "INSERT solo de las columnas conocidas", lambda: dw.execute(
    "INSERT INTO ventas SELECT id_venta, fecha, tienda, producto_id, cliente_id, cantidad, monto FROM t_lote"))
perdidas = [c for c in lote.columns if c not in dw.sql("SELECT * FROM ventas LIMIT 1").df().columns]
dw.execute("DELETE FROM ventas WHERE id_venta >= 10000001")                                   # deshacer el insert que perdia columnas
paso("almacén", "ALTER TABLE: agregar canal, cupon, region_envio", lambda: [dw.execute(f"ALTER TABLE ventas ADD COLUMN {c} VARCHAR") for c in ["canal", "cupon", "region_envio"]])
paso("almacén", "INSERT completo tras ajustar el modelo", lambda: dw.execute(
    "INSERT INTO ventas (id_venta, fecha, tienda, producto_id, cliente_id, cantidad, monto, canal, cupon, region_envio) "
    "SELECT id_venta, fecha, tienda, producto_id, cliente_id, cantidad, monto, canal, cupon, region_envio FROM t_lote"))
# ---- LAGO ----
def escribir_lote():
    d = LAGO / "ventas" / "periodo=2026-01"; d.mkdir(parents=True, exist_ok=True)
    lote.to_parquet(d / "part-0.parquet", index=False)
paso("lago", "escribir el lote como archivo nuevo", escribir_lote)
cuenta = lambda opcion: dw.sql(f"SELECT COUNT(*) AS filas, COUNT(cupon) AS con_cupon FROM read_parquet('{GLOB}' {opcion})").fetchone()
paso("lago", "leer cupon con la lectura por defecto", lambda: cuenta(""))
paso("lago", "leer cupon con union_by_name = true", lambda: cuenta(", union_by_name = true"))
filas_tot, con_cupon = cuenta(", union_by_name = true")

# ---- Conflicto de TIPO: 'monto' llega como texto con comas ----
conflicto = lote_nuevo(500, 20_000_001, "2026-02-10", "2026-02", semilla=22).assign(monto=lambda d: d.monto.map(lambda x: f"{x:,}"))
d = LAGO / "ventas" / "periodo=2026-02"; d.mkdir(parents=True, exist_ok=True); conflicto.to_parquet(d / "part-0.parquet", index=False)
paso("lago", "SUM(monto) sobre todos los archivos (hay un lote con monto como texto)", lambda: dw.sql(f"SELECT SUM(monto) FROM read_parquet('{GLOB}', union_by_name = true)").fetchall())
paso("lago", "reparar en el LECTOR: TRY_CAST(REPLACE(monto, ',', ''))", lambda: dw.sql(
    f"SELECT SUM(TRY_CAST(REPLACE(CAST(monto AS VARCHAR), ',', '') AS BIGINT)) FROM read_parquet('{GLOB}', union_by_name = true)").fetchall())
bit = pd.DataFrame(bitacora); print(bit.to_string(index=False))

# ---- Verificaciones ----
b = bit.set_index("paso").funciono
assert not b["INSERT directo del lote (SELECT *)"] and b["INSERT solo de las columnas conocidas"] and perdidas == ["canal", "cupon", "region_envio"]     # 'funciona' pero pierde 3 columnas
assert b["ALTER TABLE: agregar canal, cupon, region_envio"] and b["INSERT completo tras ajustar el modelo"]
assert b["escribir el lote como archivo nuevo"] and b["leer cupon con union_by_name = true"]            # (la lectura por defecto fallo en esta version de DuckDB; ver la bitacora)
assert filas_tot == N + 5_000 and con_cupon == int(lote.cupon.notna().sum())
assert not b["SUM(monto) sobre todos los archivos (hay un lote con monto como texto)"] and b["reparar en el LECTOR: TRY_CAST(REPLACE(monto, ',', ''))"]
print(f"\nColumnas que el INSERT 'silencioso' habria perdido: {perdidas}. En el lago, tras union_by_name: {filas_tot:,} filas, {con_cupon:,} con cupon (las demas, NULL).")
````

## 4. El *Lakehouse*: Intentar Quedarse con lo Mejor de los Dos 🏠🌊

Los experimentos muestran un patrón: el **almacén** da **garantías** (tipos, restricciones, transacciones) a cambio de **rigidez**; el **lago** da **flexibilidad y bajo costo de almacenamiento** a cambio de **garantías débiles** (un archivo mal escrito, un tipo cambiado, dos escritores a la vez). El ***lakehouse*** (Armbrust et al., 2021; *citado de memoria, verificar*) propone **una capa de tabla** sobre los archivos del lago (Parquet en almacenamiento de objetos) que añade lo que le faltaba:

| Capacidad | Qué aporta | Problema del lago que resuelve |
|---|---|---|
| **Transacciones ACID** | Una escritura **ocurre completa o no ocurre**; lecturas consistentes mientras alguien escribe | Lectores que ven **archivos a medias**, escrituras concurrentes que se pisan |
| **Imposición y evolución de esquema** | El esquema vive en la **metadata de la tabla**; se rechazan escrituras incompatibles y se permite **agregar columnas** de forma controlada | La «columna sorpresa» y el conflicto de tipos que sufrimos en el experimento |
| **Viaje en el tiempo** (*time travel*) | Consultar la tabla **como estaba** en una versión anterior | Auditoría y reproducibilidad |
| **Actualizaciones y borrados (*upserts*)** | `MERGE`/`UPDATE`/`DELETE` sobre archivos que antes eran **inmutables** | Cumplir **supresión de datos** y corregir errores sin reescribir todo a mano |
| **Metadata eficiente** | Estadísticas y listas de archivos en un **registro**, no listando directorios | La **lentitud** de listar miles de archivos (el problema de archivos pequeños) |

Los tres formatos de tabla abiertos más conocidos son **Delta Lake**, **Apache Iceberg** y **Apache Hudi** (a nivel **conceptual** aquí; **no se instalan** ni se ejecutan en este curso). Comparten la idea: **archivos de datos inmutables + un registro de metadatos** (en Delta, un **registro de transacciones** de *commits*; en Iceberg, **instantáneas** y **manifiestos**; los detalles internos de cada uno *verificar* en su documentación). Cada **commit** crea una **versión**: leer «la versión 12» es leer **la lista de archivos que esa versión declara**. Matices de honestidad: (1) **no es magia**: sigue habiendo costo de mantenimiento (compactar, limpiar versiones viejas) y de gobierno; (2) **las garantías dependen del formato y del motor** que escriba (no todos los motores soportan todas las funciones); (3) es una **tendencia reciente** y las capacidades **cambian rápido**: verifica la documentación vigente antes de decidir.

Dibujamos las **tres arquitecturas** lado a lado.

## 5. Una Guía de Decisión y un Modelo de Costo (con Supuestos Explícitos) 🧭

**No existe una respuesta única**; existen **criterios**. Los convertimos en dos herramientas pequeñas, **ambas con supuestos que tú puedes cambiar**:

1. **Una regla heurística** (`recomendar`): puntúa **almacén**, **lago** y ***lakehouse*** según el perfil de la organización (qué fracción de sus datos **no es estructurada**, quiénes consultan, cuántos **cambios de esquema** esperan al mes, si exigen **auditoría y transacciones**, cuánto pueden invertir en **gobernanza**). Es una **guía de conversación**, no un oráculo.
2. **Un modelo de costo mensual** (`costo_mensual`) con tres partes: **almacenamiento**, **cómputo por consulta** e **ingeniería** (horas por cambio de esquema y de gobierno). **Todos los parámetros son FICTICIOS** (unidades monetarias «u.m.», **no precios reales**: los precios reales dependen del proveedor y cambian). Lo único que tiene valor es la **estructura**: qué costos **crecen con el volumen**, cuáles **con las consultas** y cuáles son **fijos**. Con ella calculamos el **punto de equilibrio**: el número de consultas al mes a partir del cual la arquitectura de **costo fijo alto y costo por consulta bajo** (almacén de capacidad reservada) **supera en conveniencia** a la de costo fijo bajo y costo por consulta mayor (lago). Como el costo es lineal en las consultas $q$, con $C_i = F_i + v_i\,q$ el equilibrio es $q^{\ast} = (F_{\mathrm{DW}} - F_{\mathrm{lago}})/(v_{\mathrm{lago}} - v_{\mathrm{DW}})$, y se **verifica numéricamente**.

````python
SUPUESTOS = dict(                                   # TODO FICTICIO (unidades monetarias por mes): cambia los numeros y mira la estructura
    alm_dw_por_tb=60.0, alm_lago_por_tb=20.0,        # almacenamiento por TB-mes
    comp_dw_por_tb=0.5, comp_lago_por_tb=0.6,        # computo por TB leido
    frac_modelado=0.30,                              # fraccion del dato crudo que se modela y guarda en el almacen
    esc_dw=0.20, esc_lago=0.25, esc_lh=0.15,         # fraccion de lo almacenado que lee una consulta tipica (poda, estadisticas, formato)
    fijo_dw=15000.0, fijo_lago=1000.0, fijo_lh=3000.0,   # capacidad reservada / plataforma / catalogo
    h_cambio_dw=16, h_cambio_lago=4, h_cambio_lh=6,  # horas de ingenieria por cambio de esquema
    h_gob_lago=20, h_gob_lh=12, costo_hora=50.0)     # horas/mes de gobierno del lago y valor de la hora


def costo_mensual(arq, tb_crudo, consultas, cambios, s=SUPUESTOS):
    """Costo mensual (u.m.) = fijo + almacenamiento + computo por consulta + ingenieria. Devuelve (total, desglose)."""
    if arq == "almacén":
        tb = tb_crudo * s["frac_modelado"]; alm = tb * s["alm_dw_por_tb"]; comp = consultas * tb * s["esc_dw"] * s["comp_dw_por_tb"]
        ing = cambios * s["h_cambio_dw"] * s["costo_hora"]; fijo = s["fijo_dw"]
    elif arq == "lago":
        tb = tb_crudo; alm = tb * s["alm_lago_por_tb"]; comp = consultas * tb * s["esc_lago"] * s["comp_lago_por_tb"]
        ing = (cambios * s["h_cambio_lago"] + s["h_gob_lago"]) * s["costo_hora"]; fijo = s["fijo_lago"]
    else:                                                                                            # lakehouse
        tb = tb_crudo; alm = tb * s["alm_lago_por_tb"]; comp = consultas * tb * s["esc_lh"] * s["comp_lago_por_tb"]
        ing = (cambios * s["h_cambio_lh"] + s["h_gob_lh"]) * s["costo_hora"]; fijo = s["fijo_lh"]
    d = {"fijo": fijo, "almacenamiento": alm, "computo": comp, "ingenieria": ing}
    return sum(d.values()), d


def punto_equilibrio(tb_crudo, cambios, s=SUPUESTOS):
    """Consultas/mes a partir de las cuales el almacen cuesta menos que el lago (costo lineal en las consultas)."""
    f_dw, d_dw = costo_mensual("almacén", tb_crudo, 0, cambios, s)[0], costo_mensual("almacén", tb_crudo, 1, cambios, s)[0] - costo_mensual("almacén", tb_crudo, 0, cambios, s)[0]
    f_lg, d_lg = costo_mensual("lago", tb_crudo, 0, cambios, s)[0], costo_mensual("lago", tb_crudo, 1, cambios, s)[0] - costo_mensual("lago", tb_crudo, 0, cambios, s)[0]
    return (f_dw - f_lg) / (d_lg - d_dw)


TB, CAMBIOS = 100, 2
q_star = punto_equilibrio(TB, CAMBIOS)
c_dw, c_lg = costo_mensual("almacén", TB, q_star, CAMBIOS)[0], costo_mensual("lago", TB, q_star, CAMBIOS)[0]
print(f"Con {TB} TB de dato crudo y {CAMBIOS} cambios de esquema al mes, el punto de equilibrio es q* = {q_star:,.0f} consultas/mes (costo en q*: almacén {c_dw:,.0f} u.m. = lago {c_lg:,.0f} u.m.)")
assert abs(c_dw - c_lg) < 1e-6                                              # verificacion numerica del equilibrio analitico
assert costo_mensual("almacén", TB, 0.5 * q_star, CAMBIOS)[0] > costo_mensual("lago", TB, 0.5 * q_star, CAMBIOS)[0]       # pocas consultas: gana el lago
assert costo_mensual("almacén", TB, 2 * q_star, CAMBIOS)[0] < costo_mensual("lago", TB, 2 * q_star, CAMBIOS)[0]           # muchas consultas: gana el almacen
qs = np.logspace(0, 4, 200); fig, (a1, a2) = plt.subplots(1, 2, figsize=(13, 4.3), gridspec_kw={"width_ratios": [1.3, 1]})
for arq, col in [("almacén", "#1e3a8a"), ("lago", "#a16207"), ("lakehouse", "#0f766e")]:
    a1.plot(qs, [costo_mensual(arq, TB, q, CAMBIOS)[0] / 1000 for q in qs], label=arq, color=col, lw=2)
a1.axvline(q_star, color="#64748b", ls="--"); a1.text(q_star * 0.92, a1.get_ylim()[1] * 0.9, f"q* ≈ {q_star:,.0f}", color="#475569", ha="right")
a1.set_xscale("log"); a1.set_xlabel("consultas al mes (escala log)"); a1.set_ylabel("costo mensual (miles de u.m., FICTICIAS)"); a1.legend(); a1.set_title("Costo mensual frente a consultas", fontweight="bold")
desg = pd.DataFrame({arq: costo_mensual(arq, TB, 3000, CAMBIOS)[1] for arq in ["almacén", "lago", "lakehouse"]}).T / 1000
desg.plot(kind="bar", stacked=True, ax=a2, color=["#94a3b8", "#0ea5e9", "#f59e0b", "#16a34a"]); a2.set_ylabel("miles de u.m."); a2.set_xlabel("")
a2.set_title("Desglose con 3 000 consultas/mes", fontweight="bold"); a2.tick_params(axis="x", rotation=0)
plt.tight_layout(); plt.show()


# ---------------- Guia heuristica de decision ----------------
def recomendar(no_estructurados, usuarios_bi, cambios_mes, exige_auditoria, gobernanza):
    """Puntua almacen / lago / lakehouse. no_estructurados y usuarios_bi en [0, 1]; gobernanza 'bajo' o 'alto'. Devuelve (ranking, razones)."""
    p, razones = {"almacén": 0, "lago": 0, "lakehouse": 0}, []
    def suma(arq, pts, texto):
        p[arq] += pts; razones.append(f"{arq} {pts:+d}: {texto}")
    if usuarios_bi > 0.6: suma("almacén", 2, "la mayoría son usuarios de BI/SQL")
    if exige_auditoria: suma("almacén", 2, "exige auditoría y garantías transaccionales")
    if no_estructurados > 0.5: suma("almacén", -2, "más de la mitad de los datos no es tabular"); suma("lago", 2, "muchos datos no estructurados")
    if cambios_mes >= 2: suma("lago", 1, "el esquema cambia seguido"); suma("lakehouse", 1, "el esquema cambia seguido y se quiere control")
    if gobernanza == "bajo": suma("lago", -2, "poca capacidad de gobierno: riesgo de pantano"); suma("almacén", 1, "el modelo impone orden"); suma("lakehouse", 1, "la capa de tabla impone esquema y versiones donde falta disciplina")
    if 0.2 <= no_estructurados <= 0.8 and 0.3 <= usuarios_bi <= 0.7: suma("lakehouse", 2, "perfil mixto: BI y ciencia de datos sobre los mismos datos")
    if exige_auditoria and no_estructurados > 0.3: suma("lakehouse", 1, "auditoría sobre datos que no son solo tablas")
    return sorted(p.items(), key=lambda kv: -kv[1]), razones


perfiles = {"Banco: informes regulatorios": (0.05, 0.8, 0, True, "alto"), "Startup de visión por computador": (0.9, 0.1, 4, False, "alto"),
            "Empresa mixta con BI y ML": (0.5, 0.5, 3, True, "alto"), "Equipo pequeño sin gobierno de datos": (0.8, 0.2, 3, False, "bajo")}
decision = {}
for nombre, args in perfiles.items():
    ranking, razones = recomendar(*args); decision[nombre] = ranking[0][0]
    print(f"{nombre:42s} -> {ranking[0][0]:10s} {dict(ranking)}")
assert decision == {"Banco: informes regulatorios": "almacén", "Startup de visión por computador": "lago", "Empresa mixta con BI y ML": "lakehouse",
                    "Equipo pequeño sin gobierno de datos": "lakehouse"}
print("\nNota: el último perfil muestra la advertencia clave: un lago SIN gobierno no es la mejor opción aunque los datos sean no estructurados (una capa de tabla impone parte del orden que falta).")
````

```python
assert q_200 < q_star and q_cambios > q_star
print(f"q* = {q_star:,.0f} | con 200 TB: {q_200:,.0f} | con 8 cambios/mes: {q_cambios:,.0f}")
print("Con estos supuestos: con mas datos cada consulta del lago lee mas y cuesta mas, asi que el almacen (que guarda y lee solo lo modelado) compensa ANTES (q baja); y los cambios de esquema cuestan 16 h en el almacen frente a 4 h en el lago, asi que con mas cambios el almacen necesita MAS consultas para compensar (q sube).")
```

## 6. El «Pantano de Datos» (*Data Swamp*) y Cómo Evitarlo 🐊

Un lago **sin gobierno** se degrada hasta volverse un **pantano**: hay datos, pero **nadie sabe qué son, de quién son, si son confiables ni si ya existen en otro lado**. Los síntomas típicos y su causa:

| Síntoma | Causa | Control que lo evita |
|---|---|---|
| **No se encuentra nada** (nombres como `export_final_v2.csv`) | No hay **catálogo** ni convenciones de nombres | Catálogo obligatorio con **descripción**; nombres y rutas estandarizados |
| **Nadie sabe de quién es** | Sin **dueño** | Campo **dueño** obligatorio (en código: `NOT NULL`) |
| **Mismos datos, varias copias** | Copias manuales para «no dañar el original» | Detección de **duplicados** (huella *hash* del contenido) y una sola fuente de verdad |
| **Datos abandonados** que nadie usa pero que cuestan y exponen riesgos | Sin política de **retención** | Revisión periódica; **archivar** lo que no se usa (ver [ciclo de vida, Módulo 00](../00%20-%20Fundamentos%20de%20Big%20Data/03_Ciclo_de_Vida_de_los_Datos.ipynb)) |
| **Datos sensibles sin protección** | Sin **clasificación** | Campo de **sensibilidad** obligatorio y permisos según ella |
| **Calidad desconocida** | Sin capas ni validaciones | Capas **cruda → curada → de negocio** y pruebas de calidad (cuaderno 01) |

Lo **medimos**: generamos un «lago desordenado» **sintético** de 36 archivos con los defectos típicos y calculamos cuatro **métricas de salud**: porcentaje **en el catálogo**, porcentaje con **metadatos completos** (dueño, descripción y sensibilidad), **duplicados exactos** (mismo contenido, otro nombre) y porcentaje de **archivos viejos** (más de un año sin tocar). Luego aplicamos los controles —**catálogo con restricciones**, **eliminación de duplicados** y **archivado** de lo viejo— y **volvemos a medir**. En la vida real, asignar dueños y descripciones **requiere a las personas**: el código solo puede **obligar** a que existan y **medir** si existen.

````python
def generar_pantano(raiz, semilla=11):
    """36 archivos con nombres criptico: 30 distintos + 6 copias exactas; 14 con fecha de modificacion de hace mas de un anio."""
    rng = np.random.default_rng(semilla); raiz = Path(raiz); raiz.mkdir(parents=True, exist_ok=True)
    prefijos = ["export_final", "tmp", "datos", "backup", "ventas_nuevo", "test"]
    nombres = [f"{prefijos[i % 6]}_{'v2' if i % 5 == 0 else i}.csv" for i in range(30)]
    for i, n in enumerate(nombres):
        pd.DataFrame({"id": np.arange(40), "x": rng.integers(0, 1000, 40), "origen": i}).to_csv(raiz / n, index=False)
    for j, i in enumerate([2, 7, 11, 16, 22, 27]):                              # copias exactas con otro nombre
        shutil.copy(raiz / nombres[i], raiz / f"copia_de_{nombres[i][:-4]}_{j}.csv")
    archivos = sorted(raiz.glob("*.csv")); ahora = time.time()
    for k, p in enumerate(archivos):
        edad_dias = 400 + 30 * k if k % 5 in (1, 3) else 5 + k                   # 14 archivos viejos (mas de un anio), el resto recientes
        os.utime(p, (ahora - edad_dias * 86400, ahora - edad_dias * 86400))
    db = sqlite3.connect(":memory:")                                            # catalogo INICIAL: laxo, sin restricciones
    db.execute("CREATE TABLE datasets (ruta TEXT PRIMARY KEY, dueno TEXT, descripcion TEXT, sensibilidad TEXT)")
    for k, p in enumerate(archivos[:9]):
        db.execute("INSERT INTO datasets VALUES (?, ?, ?, ?)", (p.name, "ana" if k % 2 == 0 else None, "ventas" if k % 3 else None, "interna" if k % 4 else None))
    db.commit()
    return archivos, db


def auditar(raiz, db):
    """Metricas de salud del lago (solo archivos activos, fuera de la carpeta 'archivo')."""
    activos = sorted(p for p in Path(raiz).glob("*.csv"))
    reg = {r[0]: r for r in db.execute("SELECT ruta, dueno, descripcion, sensibilidad FROM datasets")}
    en_cat = [p for p in activos if p.name in reg]
    completos = [p for p in en_cat if all(reg[p.name][1:])]
    hashes = [hashlib.sha256(p.read_bytes()).hexdigest() for p in activos]
    viejos = [p for p in activos if (time.time() - p.stat().st_mtime) / 86400 > 365]
    n = len(activos)
    return {"archivos": n, "% en el catálogo": round(100 * len(en_cat) / n, 1), "% con metadatos completos": round(100 * len(completos) / n, 1),
            "duplicados exactos": n - len(set(hashes)), "% viejos (> 1 año)": round(100 * len(viejos) / n, 1)}


PANTANO = RAIZ / "pantano"
archivos_p, db_p = generar_pantano(PANTANO)
antes = auditar(PANTANO, db_p)
print("ANTES :", antes)
assert antes["archivos"] == 36 and antes["duplicados exactos"] == 6 and antes["% en el catálogo"] == 25.0 and antes["% con metadatos completos"] < antes["% en el catálogo"]

# ---------------- Controles: catalogo CON restricciones, deduplicar, archivar lo viejo ----------------
db_p.execute("""CREATE TABLE datasets_gobernado (ruta TEXT PRIMARY KEY, dueno TEXT NOT NULL CHECK (length(dueno) > 0),
                descripcion TEXT NOT NULL CHECK (length(descripcion) > 0), sensibilidad TEXT NOT NULL CHECK (sensibilidad IN ('publica', 'interna', 'confidencial')))""")
try:
    db_p.execute("INSERT INTO datasets_gobernado VALUES ('x.csv', NULL, 'sin dueno', 'interna')"); rechazo_sin_dueno = False
except sqlite3.IntegrityError:
    rechazo_sin_dueno = True                                                    # el catalogo NO deja registrar un dato sin dueño
assert rechazo_sin_dueno

vistos, duplicados = {}, []
for p in sorted(PANTANO.glob("*.csv")):                                         # 1) eliminar duplicados exactos (se conserva el primero por nombre)
    h = hashlib.sha256(p.read_bytes()).hexdigest()
    if h in vistos: duplicados.append(p)
    else: vistos[h] = p
for p in duplicados: p.unlink()
(PANTANO / "archivo").mkdir(exist_ok=True)
for p in sorted(PANTANO.glob("*.csv")):                                         # 2) archivar (NO borrar) lo que lleva mas de un anio sin tocarse
    if (time.time() - p.stat().st_mtime) / 86400 > 365:
        shutil.move(str(p), PANTANO / "archivo" / p.name)
EQUIPOS = ["ventas", "logística", "finanzas"]                                    # 3) registrar lo que queda, con dueño, descripcion y sensibilidad OBLIGATORIOS
for k, p in enumerate(sorted(PANTANO.glob("*.csv"))):
    db_p.execute("DELETE FROM datasets WHERE ruta = ?", (p.name,))
    db_p.execute("INSERT INTO datasets_gobernado VALUES (?, ?, ?, ?)", (p.name, EQUIPOS[k % 3], f"Conjunto sintético {p.stem}", "interna"))
db_p.execute("DROP TABLE datasets"); db_p.execute("ALTER TABLE datasets_gobernado RENAME TO datasets")
despues = auditar(PANTANO, db_p)
print("DESPUÉS:", despues)
print(f"Se eliminaron {len(duplicados)} duplicados y se archivaron {len(list((PANTANO / 'archivo').glob('*.csv')))} archivos viejos (siguen disponibles en 'archivo/').")
assert despues["duplicados exactos"] == 0 and despues["% en el catálogo"] == 100.0 and despues["% con metadatos completos"] == 100.0 and despues["% viejos (> 1 año)"] == 0.0
assert despues["archivos"] == antes["archivos"] - len(duplicados) - len(list((PANTANO / "archivo").glob("*.csv")))
````

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Comparación** | Almacén: esquema **al escribir**, datos estructurados, ETL, calidad alta, usuarios de BI, gobierno maduro, rendimiento predecible **en lo modelado**, cambio **costoso**. Lago: esquema **al leer**, todo tipo de datos, ELT, calidad variable, usuarios técnicos, gobierno **difícil**, rendimiento **según el formato**, cambio **barato al escribir**. Las diferencias de costo y rendimiento **dependen** del proveedor, el motor y la carga |
| **Experimento 1 (rapidez)** | Mismo motor, misma respuesta en los tres servicios; el **JSON crudo** fue **varias veces más lento** que el almacén y que **Parquet** (en nuestra ejecución ≈ 6 y ≈ 13 veces; varía con el equipo), y el **Parquet curado** fue **tan rápido o más** que la tabla del almacén (en esta máquina y a este tamaño): el factor decisivo es el **formato y el motor**, no la etiqueta |
| **Experimento 1 (rigidez)** | El almacén **no puede** responder por `canal` (no se modeló) y **rechaza** un dato inválido al escribir; el lago responde por `canal` y **acepta** el dato inválido (solo una validación posterior lo detecta): la flexibilidad **traslada el error al lector** |
| **Experimento 2 (cambio)** | Columnas nuevas: el almacén **no admite** el lote; el «arreglo rápido» **pierde 3 columnas en silencio**; conservarlas exige `ALTER TABLE`, ETL y recarga. El lago escribe sin cambios pero **el lector** debe pedir `union_by_name`, y un **conflicto de tipos** rompe la consulta hasta repararla en el lector |
| ***Lakehouse*** | **Archivos abiertos + capa de tabla** (Delta Lake, Iceberg, Hudi): ACID, esquema controlado, versiones, *upserts*, metadata eficiente. **No es magia**: tiene costos de mantenimiento y las capacidades dependen del formato y del motor |
| **Decisión y costo** | La decisión depende de **datos no estructurados, usuarios, cambios de esquema, auditoría y capacidad de gobierno**. Un modelo de costo (con supuestos **ficticios**) muestra **estructura**: fijo, almacenamiento, cómputo por consulta e ingeniería; el **punto de equilibrio** (q* ≈ 1 167 consultas/mes con esos supuestos) cambia con el volumen y con los cambios de esquema |
| **Pantano de datos** | Sin catálogo, dueño, clasificación, deduplicación y retención, un lago se degrada (en el ejemplo: 25 % en el catálogo, 6 duplicados, 39 % de archivos viejos). Los controles **en código** (`NOT NULL`, `CHECK`, *hash*, archivado) lo mejoran, pero **asignar dueños y describir datos requiere personas** |
