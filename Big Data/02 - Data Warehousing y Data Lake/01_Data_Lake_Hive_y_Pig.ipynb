## 1. ¿Qué es un Lago de Datos? Las Tres Capas 🏞️

Un **lago de datos** es un repositorio que almacena datos en su **formato original o casi original** (estructurados, semiestructurados y no estructurados), normalmente sobre **almacenamiento de objetos o un sistema de archivos distribuido** y con **formatos abiertos** (CSV, JSON, Parquet, ORC). El término se atribuye a James Dixon (*citado de memoria, verificar*). Sus tres ideas:

1. **Guardar primero, decidir después:** no hace falta modelar antes de ingerir; el dato crudo se conserva **intacto** (se puede **reprocesar** con mejores reglas, como en el [ciclo de vida del Módulo 00](../00%20-%20Fundamentos%20de%20Big%20Data/03_Ciclo_de_Vida_de_los_Datos.ipynb)).
2. **Esquema en lectura** (*schema-on-read*): el esquema se **aplica al leer**, y distintos consumidores pueden leer **el mismo dato con esquemas distintos**.
3. **Desacople de almacenamiento y cómputo:** los archivos viven aparte y **varios motores** (Spark, Hive, DuckDB, Presto/Trino...) los consultan.

**Capas.** Para que el lago **no se vuelva un pantano**, el dato se organiza en zonas de calidad creciente. La nomenclatura «bronce, plata, oro» (**arquitectura medallón**, popularizada por Databricks; *citado de memoria, verificar*) equivale a **cruda, curada y de negocio**:

| Capa | Contenido | Formato típico | Esquema | Quién la usa |
|---|---|---|---|---|
| **Cruda / bronce** | El dato **tal como llegó**, **inmutable**, con lote y fecha de ingesta (incluso lo malformado) | JSON, CSV, *logs*, binarios | **En lectura** (ninguno impuesto) | Ingeniería de datos (para **reprocesar** y auditar) |
| **Curada / plata** | **Limpia, deduplicada, con tipos**; lo inválido aparte, en **cuarentena** | **Parquet particionado** | Esquema **estable y registrado** en el catálogo | Analistas de datos, científicos de datos |
| **De negocio / oro** | **Agregados y tablas listas** para un uso (tablero, modelo, informe) | Parquet / tablas pequeñas | Esquema **de negocio**, documentado | Directivos, tableros, aplicaciones |

Dos piezas hacen que esto funcione a escala: las **particiones** (el dato se reparte en directorios por columnas como `anio` y `mes`, para **leer solo lo necesario**) y el **catálogo o *metastore*** (el «índice» que dice **qué tablas y particiones existen y dónde**). Dibujamos el flujo que vamos a construir.

## 2. Construir el Lago: Cruda → Curada → de Negocio 🏗️

Simulamos los **pedidos de una plataforma ficticia** de venta de productos del campo («Mercado Boyacá»: **todo inventado**, productos, municipios y precios incluidos) durante 2025: unos 36 000 pedidos, llegados en **lotes semanales** de líneas JSON. Como en la realidad, el dato crudo trae problemas que **sembramos a propósito** para poder comprobar que el pipeline los encuentra:

| Defecto sembrado | Cantidad | Qué hacemos en la capa plata |
|---|---|---|
| Línea **JSON malformada** (cortada) | 40 | **Cuarentena** (`json_malformado`): no hay forma segura de arreglarla |
| Pedido **duplicado** (misma `id_pedido`) | 30 | **Cuarentena** (`duplicado`), nos quedamos con la primera |
| `valor_unit` **cero o negativo** | 25 | **Cuarentena** (`valor_no_positivo`) |
| `cantidad` **como texto** (`"3"` en lugar de `3`) | 200 | **Corregir** (convertir a entero): el significado es inequívoco |
| Campo **`canal`** que **solo aparece desde abril** | desde 2025-04-01 | Esquema que **evoluciona**: la plata lo declara **nulo** en los meses anteriores |

La capa **plata** se escribe como **Parquet particionado estilo Hive**: la ruta `plata/pedidos/anio=2025/mes=03/part-....parquet` **contiene** el valor de `anio` y `mes`, por lo que **no se guardan dentro del archivo** (se recuperan del nombre del directorio). Cada lote semanal produce **un archivo por mes que toca** (una semana que cruza dos meses se parte en dos); eso dará decenas de archivos pequeños, un detalle que **importa** (el «problema de los archivos pequeños», más adelante). La capa **oro** son dos tablas pequeñas de agregados. Al final verificamos la **conservación de filas** (`líneas crudas = plata + cuarentena`) y que **oro cuadra con plata**.

````python
SEMILLA = 2025
MUNICIPIOS = ["Tunja", "Duitama", "Sogamoso", "Chiquinquirá", "Paipa", "Villa de Leyva", "Samacá", "Ráquira"]
PRECIOS = {"papa pastusa": 2200, "cebolla larga": 3000, "zanahoria": 1800, "arveja": 4500, "maíz": 2500, "fresa": 6000,
           "tomate de árbol": 3500, "queso campesino": 18000, "leche": 2800, "miel": 22000}          # pesos por unidad (inventados)
SCHEMA_PLATA = pa.schema([("id_pedido", pa.string()), ("ts", pa.timestamp("ms")), ("municipio", pa.string()), ("producto", pa.string()),
                          ("cantidad", pa.int32()), ("valor_unit", pa.int32()), ("total", pa.int64()), ("canal", pa.string())])
SIEMBRA = {"json_malformado": 40, "duplicado": 30, "valor_no_positivo": 25, "cantidad_como_texto": 200}


def generar_bronce(raiz, semilla=SEMILLA, por_dia=100):
    """CAPA CRUDA: escribe lotes semanales de lineas JSON (con defectos sembrados). Devuelve el numero de lineas escritas."""
    rng = np.random.default_rng(semilla)
    dias = pd.date_range("2025-01-01", "2025-12-31")
    n_dia = rng.poisson(por_dia * (1 + 0.3 * np.sin(dias.dayofyear.to_numpy() / 365 * 2 * np.pi)))
    idx = np.repeat(np.arange(len(dias)), n_dia); n = len(idx)
    ts = (dias[idx] + pd.to_timedelta(rng.integers(0, 86400, n), unit="s")).sort_values()
    prods = list(PRECIOS)
    pop = rng.choice(len(prods), n, p=np.array([0.22, 0.12, 0.1, 0.07, 0.1, 0.08, 0.08, 0.08, 0.1, 0.05]))
    df = pd.DataFrame({"id_pedido": [f"PED-{i:06d}" for i in range(1, n + 1)], "ts": ts.strftime("%Y-%m-%dT%H:%M:%S"),
                       "municipio": rng.choice(MUNICIPIOS, n), "producto": [prods[i] for i in pop],
                       "cantidad": rng.integers(1, 9, n), "valor_unit": [PRECIOS[prods[i]] for i in pop]})
    registros = df.to_dict("records")
    canal = rng.choice(["web", "app", "whatsapp"], n, p=[0.5, 0.3, 0.2])
    for i, r in enumerate(registros):
        if ts[i] >= pd.Timestamp("2025-04-01"):
            r["canal"] = str(canal[i])                                            # el campo NUEVO aparece solo desde abril
    perm = [int(i) for i in rng.permutation(n)]; pos = 0
    def tomar(k):
        nonlocal pos
        sel = perm[pos:pos + k]; pos += k
        return sel
    for i in tomar(SIEMBRA["cantidad_como_texto"]): registros[i]["cantidad"] = str(registros[i]["cantidad"])
    for j, i in enumerate(tomar(SIEMBRA["valor_no_positivo"])): registros[i]["valor_unit"] = [0, -500][j % 2]
    lineas = [json.dumps(r, ensure_ascii=False) for r in registros]
    for i in tomar(SIEMBRA["json_malformado"]): lineas[i] = lineas[i][: len(lineas[i]) // 2]          # linea cortada
    dup_de = tomar(SIEMBRA["duplicado"])                                          # copias de filas sin otros defectos
    semana = pd.Series(ts).dt.to_period("W-SUN").dt.start_time.dt.strftime("%Y-%m-%d").to_numpy()
    lotes = {}
    for i, l in enumerate(lineas):
        lotes.setdefault(semana[i], []).append(l)
    for i in dup_de:
        lotes[semana[i]].append(lineas[i])
    carpeta = Path(raiz) / "bronce" / "pedidos"; carpeta.mkdir(parents=True, exist_ok=True)
    for lunes, ls in lotes.items():
        (carpeta / f"lote_{lunes}.jsonl").write_text("\n".join(ls) + "\n", encoding="utf-8")
    return sum(len(ls) for ls in lotes.values())


def bronce_a_plata(raiz):
    """CAPA CURADA: valida, corrige, deduplica y escribe Parquet particionado estilo Hive. Devuelve (filas_plata, cuarentena)."""
    raiz = Path(raiz); cuarentena, vistos, filas = [], set(), []
    for f in sorted((raiz / "bronce" / "pedidos").glob("lote_*.jsonl")):
        for nlinea, linea in enumerate(f.read_text(encoding="utf-8").splitlines(), start=1):
            try:
                r = json.loads(linea)
            except json.JSONDecodeError:
                cuarentena.append({"lote": f.stem, "linea": nlinea, "motivo": "json_malformado", "contenido": linea[:60]}); continue
            if r["id_pedido"] in vistos:
                cuarentena.append({"lote": f.stem, "linea": nlinea, "motivo": "duplicado", "contenido": r["id_pedido"]}); continue
            if int(r["valor_unit"]) <= 0:
                cuarentena.append({"lote": f.stem, "linea": nlinea, "motivo": "valor_no_positivo", "contenido": r["id_pedido"]}); continue
            vistos.add(r["id_pedido"])
            filas.append({**r, "cantidad": int(r["cantidad"]), "valor_unit": int(r["valor_unit"]), "canal": r.get("canal"), "_lote": f.stem})
    df = pd.DataFrame(filas)
    df["ts"] = pd.to_datetime(df.ts); df["total"] = df.cantidad * df.valor_unit
    df["anio"], df["mes"] = df.ts.dt.year, df.ts.dt.month
    for (lote, anio, mes), g in df.groupby(["_lote", "anio", "mes"]):
        d = raiz / "plata" / "pedidos" / f"anio={anio}" / f"mes={mes:02d}"; d.mkdir(parents=True, exist_ok=True)
        tabla = pa.Table.from_pandas(g[[c.name for c in SCHEMA_PLATA]], schema=SCHEMA_PLATA, preserve_index=False)   # anio/mes NO van dentro del archivo
        pq.write_table(tabla, d / f"part-{lote}.parquet")
    pd.DataFrame(cuarentena).to_json(raiz / "plata" / "_cuarentena_pedidos.jsonl", orient="records", lines=True, force_ascii=False)
    return df, pd.DataFrame(cuarentena)


def plata_a_oro(raiz, plata):
    """CAPA DE NEGOCIO: agregados listos para consumir."""
    oro = Path(raiz) / "oro"; oro.mkdir(parents=True, exist_ok=True)
    mensual = plata.groupby(["anio", "mes", "producto"], as_index=False).agg(pedidos=("id_pedido", "count"), unidades=("cantidad", "sum"), ingresos=("total", "sum"))
    por_muni = plata.groupby(["municipio"], as_index=False).agg(pedidos=("id_pedido", "count"), ingresos=("total", "sum"))
    mensual.to_parquet(oro / "ventas_mensuales_por_producto.parquet", index=False); por_muni.to_parquet(oro / "ingresos_por_municipio.parquet", index=False)
    return mensual, por_muni


directorio = tempfile.TemporaryDirectory(prefix="lago_")            # se borra al cerrar el nucleo
RAIZ = Path(directorio.name)
t0 = time.time(); n_lineas = generar_bronce(RAIZ); t_b = time.time() - t0
t0 = time.time(); plata, cuarentena = bronce_a_plata(RAIZ); t_p = time.time() - t0
oro_mensual, oro_muni = plata_a_oro(RAIZ, plata)

# --- Verificaciones: conservacion de filas y defectos encontrados EXACTAMENTE ---
motivos = cuarentena.motivo.value_counts().to_dict()
assert n_lineas == len(plata) + len(cuarentena)                                                         # nada desaparece sin explicacion
assert motivos == {k: v for k, v in SIEMBRA.items() if k != "cantidad_como_texto"}, motivos
assert plata.id_pedido.is_unique and (plata.valor_unit > 0).all() and (plata.cantidad > 0).all()
assert oro_mensual.ingresos.sum() == plata.total.sum() == oro_muni.ingresos.sum()                       # oro cuadra con plata
assert plata.canal.notna().sum() == (plata.ts >= "2025-04-01").sum() and plata[plata.ts < "2025-04-01"].canal.isna().all()   # canal solo desde abril

archivos = sorted((RAIZ / "plata" / "pedidos").rglob("*.parquet"))
resumen = pd.DataFrame({"capa": ["bronce (JSON)", "plata (Parquet)", "oro (Parquet)"],
                        "archivos": [len(list((RAIZ / "bronce" / "pedidos").glob("*.jsonl"))), len(archivos), 2],
                        "filas/lineas": [n_lineas, len(plata), len(oro_mensual) + len(oro_muni)],
                        "MB": [round(sum(p.stat().st_size for p in (RAIZ / "bronce").rglob("*") if p.is_file()) / 1e6, 2),
                               round(sum(p.stat().st_size for p in archivos) / 1e6, 2),
                               round(sum(p.stat().st_size for p in (RAIZ / "oro").glob("*.parquet")) / 1e6, 3)]})
print(resumen.to_string(index=False))
print(f"\nBronce {t_b:.2f} s | bronce->plata {t_p:.2f} s | cuarentena: {motivos} | cantidades corregidas (texto -> entero): {SIEMBRA['cantidad_como_texto']}")
print("Particiones en disco:", sorted({str(p.parent.relative_to(RAIZ / 'plata' / 'pedidos')) for p in archivos})[:3], "...")
print("Un archivo de ejemplo:", archivos[0].relative_to(RAIZ), "| columnas dentro del archivo:", pq.read_schema(archivos[0]).names)
assert "anio" not in pq.read_schema(archivos[0]).names and len(archivos) > 50
````

## 3. Esquema en Lectura: el Mismo Dato Crudo, Varios Esquemas 📖

En el almacén (cuaderno anterior) el esquema se impone **al escribir**: lo que no encaja, **no entra**. En el lago el esquema se aplica **al leer**: los bytes de la capa cruda **no cambian**, y cada consumidor decide **qué campos le interesan y con qué tipos**. Dos consecuencias que vamos a comprobar:

- **Dos lectores, dos esquemas, un solo dato:** un esquema «ventas» (producto, cantidad, valor) y un esquema «geografía» (municipio, hora) leen **los mismos archivos**, sin que nadie transforme el origen.
- **Evolución del esquema:** desde abril los lotes traen el campo `canal`. Un lector **antiguo** (que no lo conoce) **sigue funcionando** sin cambios; un lector **nuevo** (que lo declara) obtiene **nulo** para los meses anteriores. Además, el lector es **tolerante**: lo que no se puede interpretar **se cuenta** y se reporta, no tumba la lectura. En una base con esquema al escribir esto habría requerido un `ALTER TABLE` y decidir qué hacer con el histórico.

El costo del esquema en lectura es que **los errores aparecen al leer**, y cada consumidor puede interpretar **distinto** (lo veremos en el cuaderno de comparación). Implementamos un lector genérico y lo contrastamos con el lector de **DuckDB** (`read_json_auto`) sobre los mismos archivos.

````python
def leer_bronce(raiz, esquema):
    """Lector con ESQUEMA EN LECTURA: esquema = {campo: conversor}. Campo ausente -> None; linea ilegible o valor no convertible -> se cuenta."""
    filas, ilegibles, no_convertibles = [], 0, 0
    for f in sorted((Path(raiz) / "bronce" / "pedidos").glob("lote_*.jsonl")):
        for linea in f.read_text(encoding="utf-8").splitlines():
            try:
                r = json.loads(linea)
            except json.JSONDecodeError:
                ilegibles += 1; continue
            fila = {}
            for campo, conv in esquema.items():
                v = r.get(campo)
                try:
                    fila[campo] = None if v is None else conv(v)
                except (TypeError, ValueError):
                    fila[campo] = None; no_convertibles += 1
            filas.append(fila)
    return pd.DataFrame(filas), ilegibles, no_convertibles


ESQUEMA_VENTAS = {"id_pedido": str, "producto": str, "cantidad": int, "valor_unit": int}
ESQUEMA_GEO = {"municipio": str, "ts": pd.Timestamp}
ESQUEMA_VENTAS_V2 = {**ESQUEMA_VENTAS, "canal": str}                 # el esquema NUEVO declara el campo que aparece desde abril

ventas_l, mal1, nc1 = leer_bronce(RAIZ, ESQUEMA_VENTAS)
geo_l, mal2, nc2 = leer_bronce(RAIZ, ESQUEMA_GEO)
nuevo_l, mal3, _ = leer_bronce(RAIZ, ESQUEMA_VENTAS_V2)
print(f"Lector 'ventas'   : {len(ventas_l)} filas, columnas {list(ventas_l.columns)}, lineas ilegibles {mal1}, valores no convertibles {nc1}")
print(f"Lector 'geografia': {len(geo_l)} filas, columnas {list(geo_l.columns)}  (MISMOS archivos, otro esquema)")
print(f"Lector 'ventas v2': canal no nulo en {int(nuevo_l.canal.notna().sum())} de {len(nuevo_l)} filas; nulo en las 5 000 primeras (enero-febrero): {bool(nuevo_l.canal.iloc[:5000].isna().all())}")
assert mal1 == mal2 == mal3 == SIEMBRA["json_malformado"] and nc1 == 0 and len(ventas_l) == n_lineas - SIEMBRA["json_malformado"]
assert list(ventas_l.columns) == list(ESQUEMA_VENTAS) and "canal" not in ventas_l.columns            # el lector viejo sigue funcionando
assert nuevo_l.canal.iloc[:5000].isna().all() and nuevo_l.canal.notna().sum() >= plata.canal.notna().sum()

# --- el mismo dato crudo, leido por un MOTOR con inferencia de esquema (DuckDB) ---
patron = str(RAIZ / "bronce" / "pedidos" / "*.jsonl")
duck = duckdb.sql(f"SELECT * FROM read_json_auto('{patron}', format='newline_delimited', union_by_name=true, ignore_errors=true, sample_size=-1)").df()
print("\nDuckDB infirio las columnas:", dict(zip(duck.columns, [str(t) for t in duck.dtypes])))
vacias = int(duck.id_pedido.isna().sum()); validas_duck = len(duck) - vacias
print(f"filas leidas por DuckDB: {len(duck)} | filas vacias: {vacias} (nuestro lector conto {mal1} lineas ilegibles) | filas con datos: {validas_duck} frente a {len(ventas_l)} de nuestro lector")
assert "canal" in duck.columns and vacias >= SIEMBRA["json_malformado"] and validas_duck <= len(ventas_l)            # cada motor trata el error a su manera
assert abs(int(duck.canal.notna().sum()) - int(nuevo_l.canal.notna().sum())) <= 10                                  # los dos lectores coinciden (casi) en el campo nuevo
assert str(duck.cantidad.dtype) == "object"                                                  # ¡inferencia por mayoria de tipos: 'cantidad' llega como TEXTO por las 200 filas con "3"!
````

## 4. Un Catálogo (*Metastore*) Mínimo, desde Cero, y la Poda de Particiones 🗃️

Los archivos Parquet **no saben** a qué tabla pertenecen ni cuántas particiones hay. Por eso los lagos tienen un **catálogo** (en Hive, el ***Hive Metastore***, una base relacional —MySQL, PostgreSQL o Derby— que guarda tablas, columnas, formato, ubicación y particiones; el mismo catálogo lo reutilizan Spark y otros motores; *citado de memoria, verificar*). Construimos uno **mínimo con SQLite**:

| Tabla del catálogo | Qué guarda |
|---|---|
| `tablas` | nombre, ubicación en disco, formato, columnas de partición y **esquema** (JSON) |
| `particiones` | por cada directorio `anio=2025/mes=03`: valores, ubicación, nº de archivos, **nº de filas** (leído del pie del Parquet, sin abrir los datos) y bytes |

Su método `reparar` hace lo que `MSCK REPAIR TABLE` en Hive: **recorre el directorio, descubre las carpetas `clave=valor` y registra las particiones que faltan** (es idempotente: ejecutarlo dos veces no duplica). Con él, un lector **planea** qué archivos abrir **antes** de leer ningún dato.

**Poda de particiones** (*partition pruning*): si la consulta filtra por una **columna de partición** (`mes = 3`), el motor **solo abre los directorios que cumplen** el filtro: se leen menos archivos y menos bytes. Si filtra por una columna **que no es de partición** (por ejemplo `month(ts) = 3`), el motor **no sabe** en qué directorios mirar y debe abrir **todos** (aunque las estadísticas internas del Parquet pueden evitar leer parte del contenido). Además medimos el **problema de los archivos pequeños**: muchos archivos diminutos cuestan por **abrirlos**, no por su contenido; la **compactación** (un archivo por partición) lo mitiga.

````python
class Metastore:
    """Catalogo minimo estilo Hive Metastore, desde cero con SQLite: tablas y particiones."""
    DDL = """
    CREATE TABLE IF NOT EXISTS tablas (nombre TEXT PRIMARY KEY, ubicacion TEXT NOT NULL, formato TEXT NOT NULL,
                                       columnas_particion TEXT NOT NULL, esquema TEXT NOT NULL);
    CREATE TABLE IF NOT EXISTS particiones (tabla TEXT NOT NULL REFERENCES tablas(nombre), valores TEXT NOT NULL, ubicacion TEXT NOT NULL,
                                            n_archivos INTEGER, n_filas INTEGER, bytes INTEGER, PRIMARY KEY (tabla, ubicacion));"""

    def __init__(self, ruta=":memory:"):
        self.db = sqlite3.connect(ruta)
        self.db.executescript(self.DDL)

    def crear_tabla(self, nombre, ubicacion, formato, columnas_particion, esquema):
        self.db.execute("INSERT INTO tablas VALUES (?, ?, ?, ?, ?)", (nombre, str(ubicacion), formato, json.dumps(columnas_particion), json.dumps(esquema)))
        self.db.commit()

    def reparar(self, nombre):
        """Como MSCK REPAIR TABLE: descubre carpetas clave=valor con archivos y registra las particiones que faltan. Devuelve cuantas agrego."""
        ubicacion, cols = self.db.execute("SELECT ubicacion, columnas_particion FROM tablas WHERE nombre = ?", (nombre,)).fetchone()
        cols, nuevas = json.loads(cols), 0
        for d in sorted(Path(ubicacion).rglob("*")):
            archivos = list(d.glob("*.parquet")) if d.is_dir() else []
            if archivos:
                valores = dict(p.split("=", 1) for p in d.relative_to(ubicacion).parts)
                assert list(valores) == cols, f"las carpetas {d.relative_to(ubicacion)} no siguen las columnas de particion {cols}"
                n_filas = sum(pq.ParquetFile(a).metadata.num_rows for a in archivos)         # solo el pie del archivo, sin leer datos
                cur = self.db.execute("INSERT OR IGNORE INTO particiones VALUES (?, ?, ?, ?, ?, ?)",
                                      (nombre, json.dumps(valores), str(d), len(archivos), n_filas, sum(a.stat().st_size for a in archivos)))
                nuevas += cur.rowcount
        self.db.commit()
        return nuevas

    def particiones(self, nombre, **filtro):
        """Particiones que cumplen el filtro (valor o lista de valores por columna de particion)."""
        def cumple(v):
            for col, ok in filtro.items():
                ok = {int(x) for x in (ok if isinstance(ok, (list, tuple, set, range)) else [ok])}
                if int(v[col]) not in ok:
                    return False
            return True
        filas = self.db.execute("SELECT valores, ubicacion, n_archivos, n_filas, bytes FROM particiones WHERE tabla = ? ORDER BY ubicacion", (nombre,)).fetchall()
        return [{"valores": json.loads(v), "ubicacion": u, "n_archivos": a, "n_filas": f, "bytes": b} for v, u, a, f, b in filas if cumple(json.loads(v))]

    def archivos(self, nombre, **filtro):
        return sorted(str(a) for p in self.particiones(nombre, **filtro) for a in Path(p["ubicacion"]).glob("*.parquet"))

    def plan(self, nombre, **filtro):
        ps = self.particiones(nombre, **filtro)
        return {"particiones": len(ps), "archivos": sum(p["n_archivos"] for p in ps), "filas": sum(p["n_filas"] for p in ps), "MB": round(sum(p["bytes"] for p in ps) / 1e6, 3)}


meta = Metastore()
esquema_tabla = {**{c.name: str(c.type) for c in SCHEMA_PLATA}, "anio": "int", "mes": "int"}
meta.crear_tabla("pedidos", RAIZ / "plata" / "pedidos", "parquet", ["anio", "mes"], esquema_tabla)
n1 = meta.reparar("pedidos"); n2 = meta.reparar("pedidos")                       # la segunda vez no agrega nada (idempotente)
print(f"MSCK REPAIR: {n1} particiones registradas; segunda ejecucion: {n2}")
assert n1 == 12 and n2 == 0 and meta.plan("pedidos")["filas"] == len(plata) and meta.plan("pedidos")["archivos"] == len(archivos)

casos = {"sin filtro": {}, "mes = 3": {"mes": 3}, "trimestre 1 (mes 1-3)": {"mes": [1, 2, 3]}, "anio = 2025 y mes = 12": {"anio": 2025, "mes": 12}, "anio = 2024": {"anio": 2024}}
tabla_plan = pd.DataFrame({k: meta.plan("pedidos", **f) for k, f in casos.items()}).T
print("\nPlan de lectura segun el filtro (decidido SOLO con el catalogo, sin abrir datos):"); tabla_plan = tabla_plan.astype({"particiones": int, "archivos": int, "filas": int})
print(tabla_plan.to_string())
total_archivos = len(archivos)
assert tabla_plan.loc["sin filtro", "archivos"] == total_archivos and tabla_plan.loc["anio = 2024", "archivos"] == 0
assert tabla_plan.loc["mes = 3", "archivos"] < total_archivos / 6 and tabla_plan.loc["mes = 3", "filas"] == int((plata.mes == 3).sum())

# --- Poda medida: los mismos resultados, muchos menos archivos ---
con_lago = duckdb.connect()
def lista_sql(fs):
    return "[" + ", ".join(f"'{a}'" for a in fs) + "]"


def mediana_ms(fn, veces=7):
    t = []
    for _ in range(veces):
        t0 = time.perf_counter(); fn(); t.append((time.perf_counter() - t0) * 1000)
    return round(float(np.median(t)), 1)


todos, podados = meta.archivos("pedidos"), meta.archivos("pedidos", mes=3)
q_sin = lambda: con_lago.sql(f"SELECT COUNT(*), SUM(total) FROM read_parquet({lista_sql(todos)}) WHERE month(ts) = 3").fetchone()       # filtro NO de particion
q_con = lambda: con_lago.sql(f"SELECT COUNT(*), SUM(total) FROM read_parquet({lista_sql(podados)})").fetchone()                       # solo las particiones de mes=3
assert q_sin() == q_con() == (int((plata.mes == 3).sum()), int(plata[plata.mes == 3].total.sum()))                                   # MISMO resultado
print(f"\nSin poda: abre {len(todos)} archivos -> {mediana_ms(q_sin)} ms | Con poda: abre {len(podados)} archivos -> {mediana_ms(q_con)} ms (mediana de 7)")

# --- Problema de los archivos pequenos: compactar a un archivo por particion ---
compacta = RAIZ / "plata_compacta"
for p in meta.particiones("pedidos"):
    destino = compacta / "/".join(f"{k}={v}" for k, v in p["valores"].items()); destino.mkdir(parents=True, exist_ok=True)
    pq.write_table(pa.concat_tables([pq.read_table(a, schema=SCHEMA_PLATA) for a in sorted(Path(p["ubicacion"]).glob("*.parquet"))]), destino / "part-0.parquet")
comp_archivos = sorted(str(a) for a in compacta.rglob("*.parquet"))
q_pequenos = lambda: con_lago.sql(f"SELECT SUM(total) FROM read_parquet({lista_sql(todos)})").fetchone()
q_compacta = lambda: con_lago.sql(f"SELECT SUM(total) FROM read_parquet({lista_sql(comp_archivos)})").fetchone()
assert q_pequenos() == q_compacta() == (int(plata.total.sum()),) and len(comp_archivos) == 12
print(f"Leer todo: {len(todos)} archivos pequenos -> {mediana_ms(q_pequenos)} ms | {len(comp_archivos)} archivos compactados -> {mediana_ms(q_compacta)} ms (las diferencias a este tamano son pequenas y dependen del equipo: el efecto crece con miles de archivos y con almacenamiento remoto)")
````

## 5. Consultas Estilo HiveQL: Spark SQL, DuckDB y pandas 🐝

**Hive** puso **SQL sobre archivos** en Hadoop: se declara una **tabla externa** (los datos ya existen en un directorio) y se consulta con **HiveQL**, un SQL casi estándar. El DDL de nuestra tabla en HiveQL sería (**no se ejecuta aquí**: requiere un *metastore* y un entorno Hive):

```sql
CREATE EXTERNAL TABLE pedidos (
  id_pedido STRING, ts TIMESTAMP, municipio STRING, producto STRING,
  cantidad INT, valor_unit INT, total BIGINT, canal STRING)
PARTITIONED BY (anio INT, mes INT)
STORED AS PARQUET
LOCATION '/lago/plata/pedidos';

MSCK REPAIR TABLE pedidos;      -- registra las particiones que ya estan en disco
SHOW PARTITIONS pedidos;
```

Hoy la forma habitual de ejecutar ese SQL es **Spark SQL** (y motores como DuckDB, Trino o Presto, sobre los mismos archivos). Spark SQL entiende gran parte del dialecto HiveQL, incluidos `MSCK REPAIR TABLE` y `SHOW PARTITIONS` (no **todo** HiveQL; *verificar* la lista actual en la documentación). Hacemos **tres cosas** con *Spark* en modo local (una sesión que **se detiene al final**):

1. **Descubrir las particiones** de nuestra tabla con `CREATE TABLE ... LOCATION` + `MSCK REPAIR TABLE` y comprobar que **coinciden con las de nuestro catálogo**.
2. Ejecutar **la misma consulta** (los 3 productos con más ingresos de cada mes del primer trimestre, con una función de ventana) en **Spark SQL**, **DuckDB** y **pandas**, y verificar que dan **el mismo resultado**.
3. Leer en el **plan de ejecución** de Spark si hubo **poda de particiones** (`PartitionFilters`) o no (filtro sobre una columna que no es de partición).

````python
from pyspark.sql import SparkSession, functions as F

spark = (SparkSession.builder.master("local[2]").appName("lago").config("spark.ui.enabled", "false").config("spark.ui.showConsoleProgress", "false")
         .config("spark.sql.shuffle.partitions", "4").config("spark.sql.warehouse.dir", str(RAIZ / "spark_warehouse"))
         .config("spark.local.dir", str(RAIZ / "spark_tmp")).getOrCreate())
spark.sparkContext.setLogLevel("ERROR")
ruta_plata = str(RAIZ / "plata" / "pedidos")

# 1) Descubrimiento de particiones: CREATE TABLE ... LOCATION + MSCK REPAIR TABLE (como en Hive)
spark.sql(f"CREATE TABLE pedidos_lago USING PARQUET LOCATION '{ruta_plata}'")
spark.sql("MSCK REPAIR TABLE pedidos_lago")
particiones_spark = sorted(r.partition for r in spark.sql("SHOW PARTITIONS pedidos_lago").collect())
particiones_mias = sorted("/".join(f"{k}={v}" for k, v in p["valores"].items()) for p in meta.particiones("pedidos"))
print("Particiones en Spark:", particiones_spark[:3], "...", len(particiones_spark), "| en nuestro catalogo:", len(particiones_mias))
assert particiones_spark == particiones_mias                                      # el mismo mapa de particiones, descubierto por dos catalogos

# 2) La MISMA consulta en tres motores
SQL_TOP3 = """
WITH m AS (SELECT mes, producto, SUM(total) AS ingresos FROM pedidos WHERE anio = 2025 AND mes <= 3 GROUP BY mes, producto)
SELECT mes, producto, ingresos, rk FROM (SELECT *, ROW_NUMBER() OVER (PARTITION BY mes ORDER BY ingresos DESC, producto) AS rk FROM m) WHERE rk <= 3 ORDER BY mes, rk"""
spark.read.parquet(ruta_plata).createOrReplaceTempView("pedidos")
r_spark = [(r.mes, r.producto, int(r.ingresos), r.rk) for r in spark.sql(SQL_TOP3).collect()]

# DuckDB conserva el nombre de carpeta 'mes=03' como TEXTO ('03'); Spark lo infiere como entero: aqui lo convertimos explicitamente
con_lago.execute(f"""CREATE OR REPLACE VIEW pedidos AS SELECT * REPLACE (CAST(anio AS INTEGER) AS anio, CAST(mes AS INTEGER) AS mes)
                     FROM read_parquet('{ruta_plata}/*/*/*.parquet', hive_partitioning = true)""")
r_duck = [(int(m), p, int(i), int(k)) for m, p, i, k in con_lago.sql(SQL_TOP3).fetchall()]

g = plata[(plata.anio == 2025) & (plata.mes <= 3)].groupby(["mes", "producto"], as_index=False).agg(ingresos=("total", "sum"))
g = g.sort_values(["mes", "ingresos", "producto"], ascending=[True, False, True]); g["rk"] = g.groupby("mes").cumcount() + 1
r_pandas = [(int(m), p, int(i), int(k)) for m, p, i, k in g[g.rk <= 3][["mes", "producto", "ingresos", "rk"]].itertuples(index=False)]
print("Top 3 por mes (Spark SQL):", r_spark[:3], "...")
assert r_spark == r_duck == r_pandas and len(r_spark) == 9                          # tres motores, el mismo resultado

# 3) ¿Hubo poda de particiones? Se lee en el plan fisico
def plan(df):
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        df.explain()
    return buf.getvalue()


plan_poda = plan(spark.sql("SELECT SUM(total) FROM pedidos WHERE mes = 3"))                  # filtro sobre columna de PARTICION
plan_sin = plan(spark.sql("SELECT SUM(total) FROM pedidos WHERE month(ts) = 3"))            # filtro sobre columna que NO es de particion
linea = lambda t: [l.strip() for l in t.splitlines() if "PartitionFilters" in l][0]
pf = lambda t: linea(t).split("PartitionFilters:")[1].split("PushedFilters")[0].strip().rstrip(",")
print("\nCon filtro mes = 3        -> PartitionFilters:", pf(plan_poda))
print("Con filtro month(ts) = 3  -> PartitionFilters:", pf(plan_sin))
assert "mes" in pf(plan_poda) and pf(plan_sin) == "[]"
spark.stop()
assert SparkSession.getActiveSession() is None
print("\nSesion de Spark detenida.")
````

## 6. Un Flujo Estilo Pig Latin: de `LOAD` a `STORE` 🐷

**Apache Pig** introdujo **Pig Latin**, un lenguaje **de flujo de datos** (procedural, paso a paso) para analizar datos grandes en Hadoop **sin escribir MapReduce a mano** (Olston et al., 2008; *citado de memoria, verificar*). Cada línea define una **relación** (una colección de tuplas) a partir de la anterior, y Pig **compila** el script a una secuencia de trabajos (originalmente **MapReduce**; versiones posteriores añadieron otros motores como Tez, *verificar*). El script que implementaremos (**no se ejecuta aquí**: requiere instalar Pig y Hadoop) responde: *¿qué cinco municipios generaron más ingresos en el primer semestre?*

```pig
pedidos   = LOAD 'pedidos.csv' USING PigStorage(',')
            AS (id_pedido:chararray, mes:int, municipio:chararray, producto:chararray, cantidad:int, valor_unit:int);
validos   = FILTER pedidos BY mes <= 6 AND valor_unit > 0;
con_total = FOREACH validos GENERATE municipio, cantidad * valor_unit AS total;
por_muni  = GROUP con_total BY municipio;
resumen   = FOREACH por_muni GENERATE group AS municipio, COUNT(con_total) AS pedidos, SUM(con_total.total) AS ingresos;
ordenado  = ORDER resumen BY ingresos DESC;
top5      = LIMIT ordenado 5;
STORE top5 INTO 'salida_top5' USING PigStorage(',');
```

Lo implementamos **de dos maneras**: (a) como un **mini-Pig en Python** (una función por operador: `LOAD`, `FILTER`, `FOREACH_GENERATE`, `GROUP`, `ORDER`, `LIMIT`, `STORE`, con relaciones como listas de diccionarios), y (b) con la **API de DataFrames de Spark**, y verificamos que **dan lo mismo**.

**Honestidad: ¿qué es hoy «heredado»?** Pig y Hive sobre MapReduce marcaron una época porque permitieron analizar datos masivos **sin programar en Java**. Hoy, para un proyecto nuevo, lo habitual es **Spark SQL** o los `DataFrame` de Spark (más rápidos en general porque evitan escribir a disco entre etapas, como vimos en el [Módulo 01](../01%20-%20Procesamiento%20y%20Almacenamiento%20Distribuido/README.md), y con una sola API para SQL, programación y *streaming*). Lo que sí **perdura** de Hive es su **catálogo (*Hive Metastore*)**, que otros motores siguen usando, y sus **ideas** (tablas externas, particiones, SQL sobre archivos). Del estado de los proyectos (versiones y actividad actuales de Apache Pig y de Hive en MapReduce) **no afirmamos cifras ni fechas: *verificar* en los sitios oficiales**.

````python
# ---------------- Mini-Pig en Python: una funcion por operador ----------------
def LOAD(ruta, campos):
    with open(ruta, newline="", encoding="utf-8") as f:
        return [{n: c(v) for (n, c), v in zip(campos, fila)} for fila in csv.reader(f)]

def FILTER(rel, pred):
    return [t for t in rel if pred(t)]

def FOREACH_GENERATE(rel, fn):
    return [fn(t) for t in rel]

def GROUP(rel, campo):
    grupos = {}
    for t in rel:
        grupos.setdefault(t[campo], []).append(t)
    return [{"group": k, "bag": bolsa} for k, bolsa in grupos.items()]

def ORDER(rel, clave):
    return sorted(rel, key=clave)

def LIMIT(rel, n):
    return rel[:n]

def STORE(rel, ruta):
    Path(ruta).parent.mkdir(parents=True, exist_ok=True)
    with open(ruta, "w", newline="", encoding="utf-8") as f:
        w = csv.writer(f)
        for t in rel:
            w.writerow(list(t.values()))


entrada = RAIZ / "pig" / "pedidos.csv"; entrada.parent.mkdir(exist_ok=True)
plata[["id_pedido", "mes", "municipio", "producto", "cantidad", "valor_unit"]].to_csv(entrada, index=False, header=False)    # PigStorage(',') no usa encabezado
CAMPOS = [("id_pedido", str), ("mes", int), ("municipio", str), ("producto", str), ("cantidad", int), ("valor_unit", int)]

t0 = time.time()
pedidos = LOAD(entrada, CAMPOS)
validos = FILTER(pedidos, lambda t: t["mes"] <= 6 and t["valor_unit"] > 0)
con_total = FOREACH_GENERATE(validos, lambda t: {"municipio": t["municipio"], "total": t["cantidad"] * t["valor_unit"]})
por_muni = GROUP(con_total, "municipio")
resumen_pig = FOREACH_GENERATE(por_muni, lambda g: {"municipio": g["group"], "pedidos": len(g["bag"]), "ingresos": sum(t["total"] for t in g["bag"])})
ordenado = ORDER(resumen_pig, lambda t: (-t["ingresos"], t["municipio"]))
top5 = LIMIT(ordenado, 5)
STORE(top5, RAIZ / "pig" / "salida_top5" / "part-0.csv")
t_py = time.time() - t0
print(f"Pipeline Pig en Python ({len(pedidos)} tuplas, {t_py:.2f} s). Top 5 municipios (1.er semestre):")
for t in top5:
    print(f"  {t['municipio']:15s} pedidos={t['pedidos']:5d} ingresos=${t['ingresos']:,}")

# Verificacion independiente con pandas sobre la capa plata
ref = plata[plata.mes <= 6].groupby("municipio").agg(pedidos=("id_pedido", "count"), ingresos=("total", "sum")).reset_index()
ref = ref.sort_values(["ingresos", "municipio"], ascending=[False, True]).head(5)
esperado = [(r.municipio, int(r.pedidos), int(r.ingresos)) for r in ref.itertuples()]
resultado_py = [(t["municipio"], t["pedidos"], t["ingresos"]) for t in top5]
assert resultado_py == esperado
with open(RAIZ / "pig" / "salida_top5" / "part-0.csv", newline="", encoding="utf-8") as f:
    guardado = [tuple(fila) for fila in csv.reader(f)]
assert guardado == [(m, str(p), str(i)) for m, p, i in esperado]                                    # lo que STORE escribio en disco

# ---------------- El equivalente en Spark (API de DataFrames) ----------------
spark = (SparkSession.builder.master("local[2]").appName("pig_en_spark").config("spark.ui.enabled", "false").config("spark.ui.showConsoleProgress", "false")
         .config("spark.sql.shuffle.partitions", "4").config("spark.sql.warehouse.dir", str(RAIZ / "spark_warehouse"))
         .config("spark.local.dir", str(RAIZ / "spark_tmp")).getOrCreate())
spark.sparkContext.setLogLevel("ERROR")
t0 = time.time()
ped = spark.read.csv(str(entrada), schema="id_pedido STRING, mes INT, municipio STRING, producto STRING, cantidad INT, valor_unit INT")        # LOAD
top5_spark = (ped.filter((F.col("mes") <= 6) & (F.col("valor_unit") > 0))                                                                      # FILTER
              .withColumn("total", F.col("cantidad") * F.col("valor_unit"))                                                                    # FOREACH ... GENERATE
              .groupBy("municipio").agg(F.count("*").alias("pedidos"), F.sum("total").alias("ingresos"))                                       # GROUP + FOREACH
              .orderBy(F.desc("ingresos"), "municipio").limit(5))                                                                              # ORDER + LIMIT
resultado_spark = [(r.municipio, int(r.pedidos), int(r.ingresos)) for r in top5_spark.collect()]
top5_spark.write.mode("overwrite").csv(str(RAIZ / "pig" / "salida_spark"))                                                                    # STORE
t_sp = time.time() - t0
print(f"\nEquivalente en Spark ({t_sp:.1f} s, incluye arrancar tareas): resultado identico = {resultado_spark == resultado_py}")
assert resultado_spark == resultado_py and any((RAIZ / "pig" / "salida_spark").glob("part-*.csv"))
spark.stop()
assert SparkSession.getActiveSession() is None
print("Sesion de Spark detenida. Con 36 mil filas el Python puro gana en tiempo (Spark paga el arranque); la ventaja de Spark aparece con volumenes que no caben en una maquina.")
````

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Lago de datos** | Guarda el dato **crudo e inmutable** en **formatos abiertos** sobre almacenamiento barato; el **esquema se aplica al leer**; varios motores consultan los mismos archivos. Sin organización se vuelve un **pantano** (cuaderno siguiente) |
| **Capas** | **Cruda/bronce** (tal cual llegó, reprocesable), **curada/plata** (limpia, con tipos, particionada, con cuarentena) y **de negocio/oro** (agregados listos para consumir). Verificamos **conservación de filas** (36 509 = 36 414 + 95) y que **oro cuadra con plata** |
| **Parquet particionado estilo Hive** | `anio=2025/mes=03/`: el **nombre del directorio es el dato** de la partición (no se guarda dentro del archivo); permite **leer solo lo necesario** |
| **Esquema en lectura** | Los mismos bytes se leen con **esquemas distintos**; un campo nuevo (`canal`) no rompe al lector antiguo; los errores **aparecen al leer** (DuckDB infirió `cantidad` como texto por 200 filas con `"3"`) |
| **Catálogo / *metastore*** | Tablas y particiones registradas (aquí, **SQLite desde cero**); `reparar` ≈ `MSCK REPAIR TABLE`; el **plan de lectura se decide con el catálogo**, sin abrir datos; Spark descubrió **las mismas 12 particiones** |
| **Poda de particiones** | Filtrar por la **columna de partición** abre solo los directorios necesarios (aquí **6 de 62** archivos para `mes = 3`, mismo resultado); filtrar por una columna **no particionada** no poda (el plan de Spark lo muestra). Muchos **archivos pequeños** cuestan: se **compactan** |
| **HiveQL / Spark SQL** | El mismo SQL (con función de ventana) da **el mismo resultado** en Spark SQL, DuckDB y pandas; hoy Spark SQL es la forma habitual de ejecutarlo |
| **Pig Latin** | Lenguaje de **flujo de datos** (`LOAD`, `FILTER`, `GROUP`, `FOREACH ... GENERATE`, `ORDER`, `STORE`); lo implementamos en Python y en Spark con el mismo resultado. Pig y Hive sobre MapReduce son **herramientas heredadas** frente a Spark; perduran el **metastore** y las **ideas** |
| **Honestidad** | Datos **sintéticos**; con 36 mil filas el Python puro **le gana en tiempo** a Spark (arranque de la JVM): Spark vale la pena cuando el dato **no cabe** en una máquina; las diferencias de milisegundos en la poda dependen del equipo |
