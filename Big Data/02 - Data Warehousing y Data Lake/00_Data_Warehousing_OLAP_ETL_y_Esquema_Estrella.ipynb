## 1. De las Bases Transaccionales al Almacén de Datos 🏬

Una base **OLTP** (*Online Transaction Processing*) atiende la operación diaria: registrar una venta, actualizar un saldo. Una base **OLAP** (*Online Analytical Processing*) atiende preguntas como «¿cuánto vendimos por categoría y trimestre en cada ciudad?». Son cargas de trabajo **opuestas**, y por eso se diseñan distinto:

| Aspecto | **OLTP** (sistema operacional) | **OLAP / DW** (sistema analítico) |
|---|---|---|
| **Propósito** | Ejecutar el negocio (registrar, modificar) | Entender el negocio (consultar, decidir) |
| **Operación típica** | Muchas escrituras pequeñas: `INSERT`/`UPDATE` de 1 fila | Pocas lecturas enormes: agregar millones de filas |
| **Consulta típica** | «Datos del pedido 123» | «Ventas por categoría, trimestre y ciudad» |
| **Modelo** | Normalizado (3FN), muchas tablas pequeñas | **Dimensional** (estrella), pocas tablas anchas |
| **Historia** | Estado **actual** (se sobrescribe) | **Histórico** (se conservan versiones) |
| **Almacenamiento** | Por **filas** (leer un registro completo es rápido) | Por **columnas** (leer pocas columnas de muchas filas es rápido) |
| **Usuarios** | Operadores, aplicaciones | Analistas, directivos, tableros |

**¿Qué es un DW?** La definición clásica de **Inmon** (*citado de memoria, verificar*) lo describe como una colección de datos **orientada a temas** (ventas, clientes: no a aplicaciones), **integrada** (varias fuentes unificadas con nombres, unidades y códigos consistentes), **variante en el tiempo** (conserva el histórico) y **no volátil** (lo cargado no se modifica ni se borra en la operación normal). Existen dos grandes escuelas de arquitectura: la de **Inmon** (un DW corporativo normalizado y, de él, *data marts*) y la de **Kimball** (el DW es la unión de *data marts* dimensionales con dimensiones conformadas); aquí seguimos el **modelado dimensional de Kimball**, el más usado para consulta.

**Modelado dimensional en cuatro decisiones (Kimball):** (1) elegir el **proceso de negocio** (aquí, las *ventas*); (2) declarar la **granularidad** («una fila = una línea de venta»); (3) identificar las **dimensiones** (los «por»: por fecha, producto, cliente, tienda); (4) identificar los **hechos** (las medidas numéricas: cantidad, monto, costo). La **tabla de hechos** guarda medidas y claves foráneas; las **dimensiones** guardan atributos descriptivos. Detalles que usaremos:

- **Clave sustituta** (*surrogate key*): un entero sin significado de negocio que identifica cada fila de la dimensión; **desacopla** el DW de las claves de los sistemas fuente y **permite guardar varias versiones** de una misma entidad.
- **Dimensión degenerada:** un identificador sin atributos propios que vive en la tabla de hechos (aquí, `id_venta`).
- **Aditividad:** `monto` y `cantidad` son **aditivos** (se pueden sumar por cualquier dimensión); un precio unitario o un saldo **no** lo son.
- **Estrella frente a copo de nieve:** en la **estrella** cada dimensión es **una sola tabla desnormalizada**; en el **copo de nieve** las dimensiones se **normalizan** en varias tablas (producto → subcategoría → categoría). Lo veremos medido en la sección 5.
- **ETL frente a ELT:** *Extract-Transform-Load* transforma **antes** de cargar; *ELT* carga primero el dato crudo y transforma **dentro** del almacén (típico de almacenes en la nube y de lagos de datos, cuaderno siguiente).

**Cubo y OLAP.** Con $n$ dimensiones sin jerarquías, un cubo tiene $2^{n}$ posibles agrupaciones («cuboides»), desde el detalle (todas las dimensiones) hasta el total (ninguna); `CUBE` de SQL las genera **todas** y `ROLLUP` genera solo las $n+1$ **jerárquicas** (año → trimestre → mes → total). Las operaciones OLAP son: ***roll-up*** (consolidar: subir un nivel de la jerarquía), ***drill-down*** (detallar: bajar un nivel), ***slice*** (rebanada: fijar **una** dimensión), ***dice*** (cubo menor: restringir **varias** dimensiones) y ***pivot*** (rotar ejes). Primero **dibujamos** el modelo que vamos a construir.

````python
# !pip install -q duckdb        # (en Google Colab suele venir preinstalado; descomenta si falta)
import json, tempfile, time
from datetime import date, timedelta
from pathlib import Path
import numpy as np
import pandas as pd
import duckdb
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

pd.set_option("display.width", 140); pd.set_option("display.max_columns", 20)


def caja(ax, x, y, w, titulo, campos, color):
    """Dibuja una tabla como caja con titulo y lista de campos (alto segun el numero de campos)."""
    h = 0.45 + 0.27 * len(campos)
    ax.add_patch(FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.02,rounding_size=0.06", fc="white", ec=color, lw=2))
    ax.add_patch(FancyBboxPatch((x, y + h - 0.38), w, 0.38, boxstyle="round,pad=0.02,rounding_size=0.06", fc=color, ec=color, lw=2))
    ax.text(x + w / 2, y + h - 0.19, titulo, ha="center", va="center", color="white", fontweight="bold", fontsize=9)
    for i, c in enumerate(campos):
        ax.text(x + 0.1, y + h - 0.6 - 0.27 * i, c, ha="left", va="center", fontsize=7.3, family="monospace")
    return (x + w / 2, y + h / 2, h)


def une(ax, p, q):
    ax.annotate("", xy=q, xytext=p, arrowprops=dict(arrowstyle="-|>", lw=1.4, color="#475569", shrinkA=2, shrinkB=2))


fig, (a1, a2) = plt.subplots(1, 2, figsize=(15.5, 6.2), gridspec_kw={"width_ratios": [1, 1.12]})
for a in (a1, a2):
    a.set_aspect("equal"); a.axis("off"); a.grid(False)

# --- Panel 1: ESTRELLA ---
a1.set_xlim(0, 11); a1.set_ylim(0, 8.2)
f = caja(a1, 3.55, 2.1, 3.9, "fact_ventas  (HECHOS)", ["id_venta (degenerada)", "sk_fecha  → dim_fecha", "sk_producto → dim_producto", "sk_cliente → dim_cliente", "sk_tienda → dim_tienda", "cantidad  (medida)", "monto     (medida)", "costo_total (medida)"], "#1e3a8a")
d1 = caja(a1, 0.2, 5.5, 3.1, "dim_fecha", ["sk_fecha", "fecha, anio, trimestre", "mes, nombre_mes", "dia_semana", "es_fin_de_semana"], "#0ea5e9")
d2 = caja(a1, 7.7, 5.5, 3.1, "dim_producto", ["sk_producto", "cod_producto, nombre", "categoria", "subcategoria", "precio_lista, costo"], "#16a34a")
d3 = caja(a1, 0.2, 0.15, 3.1, "dim_cliente (SCD2)", ["sk_cliente", "documento, nombre", "ciudad, segmento", "vigente_desde/hasta", "es_actual"], "#f59e0b")
d4 = caja(a1, 7.7, 0.15, 3.1, "dim_tienda", ["sk_tienda", "cod_tienda", "nombre_tienda", "ciudad, m2"], "#dc2626")
for p, q in [((1.8, 5.5), (4.7, 4.74)), ((9.2, 5.5), (6.3, 4.74)), ((1.8, 1.97), (3.55, 3.0)), ((9.2, 1.70), (7.45, 3.0))]:
    une(a1, p, q)
a1.set_title("ESTRELLA: dimensiones desnormalizadas (1 union por dimension)", fontweight="bold", fontsize=10.5)

# --- Panel 2: COPO DE NIEVE (la dimension producto se normaliza) ---
a2.set_xlim(0, 12.2); a2.set_ylim(0, 8.2)
f2 = caja(a2, 0.2, 2.2, 3.4, "fact_ventas", ["sk_fecha", "sk_producto", "sk_cliente", "sk_tienda", "cantidad, monto", "costo_total"], "#1e3a8a")
p1 = caja(a2, 4.5, 4.6, 3.0, "sf_producto", ["sk_producto", "nombre, precio, costo", "sk_subcategoria"], "#16a34a")
p2 = caja(a2, 8.3, 4.6, 3.6, "sf_subcategoria", ["sk_subcategoria", "nombre", "sk_categoria"], "#65a30d")
p3 = caja(a2, 8.3, 1.2, 3.6, "sf_categoria", ["sk_categoria", "nombre"], "#a16207")
for p, q in [((3.6, 4.0), (4.5, 5.3)), ((7.5, 5.4), (8.3, 5.4)), ((10.1, 4.6), (10.1, 2.55))]:
    une(a2, p, q)
a2.text(2.0, 1.1, "(dim_fecha, dim_cliente y dim_tienda\nse mantienen como en la estrella)", ha="center", fontsize=8, color="#475569", style="italic")
a2.set_title("COPO DE NIEVE: dimension normalizada (3 uniones para llegar a la categoria)", fontweight="bold", fontsize=10.5)
plt.tight_layout(); plt.show()
````

## 2. Los «Sistemas Fuente» (con Inconsistencias a Propósito) 🗂️

Un DW **integra** datos de sistemas que **no se pusieron de acuerdo**. Simulamos cuatro, cada uno con su formato y sus defectos (todo **inventado, con semilla fija**):

| Sistema | Archivo | Formato | Contenido | Defectos que sembramos |
|---|---|---|---|---|
| **POS** (caja registradora) | `pos_ventas.csv` | CSV | 4 000 líneas de venta de 2025 | fechas en dos formatos (`2025-03-07` y `07/03/2025`), fechas imposibles (`2025-02-30`), cantidades `0` o negativas, precio vacío, cliente vacío o con espacios, productos/tiendas/clientes inexistentes y **60 líneas duplicadas** |
| **ERP** (catálogo) | `erp_catalogo.csv`, `erp_tiendas.csv` | CSV | 60 productos (4 categorías × 3 subcategorías × 5) y 8 tiendas | (referencia confiable) |
| **CRM** (clientes) | `crm_clientes.json` | JSON | 150 clientes (+ 8 repetidos) | ciudad en minúsculas, mayúsculas o con espacios; algunos con el campo anidado `contacto` (semiestructurado) |
| **CRM** (novedades) | `crm_novedades.json` | JSON | cambios de ciudad o segmento de clientes durante el año | 2 «novedades» que **no cambian nada** (trae el mismo valor) |

Como conocemos exactamente **cuántos defectos de cada tipo sembramos** (`inyectado`), después podremos **comprobar que el ETL los encuentra todos** y ni uno más.

````python
SEMILLA = 2025
CIUDADES = ["Tunja", "Duitama", "Sogamoso", "Chiquinquirá", "Paipa", "Villa de Leyva", "Bogotá", "Yopal"]
CATEGORIAS = {"Lácteos": ["Leche", "Queso", "Yogur"], "Despensa": ["Arroz", "Lentejas", "Aceite"],
              "Bebidas": ["Jugo", "Gaseosa", "Agua"], "Aseo": ["Jabón", "Detergente", "Papel"]}
MARCAS = ["Andina", "Boyacá", "Sol", "Campo", "Real"]
INICIO_DW, FIN_DW = "2025-01-01", "2025-12-31"


def generar_fuentes(carpeta, n_ventas=4000, semilla=SEMILLA):
    """Escribe los 5 archivos de los sistemas fuente y devuelve cuantos defectos de cada tipo se sembraron."""
    rng = np.random.default_rng(semilla)
    carpeta = Path(carpeta); carpeta.mkdir(parents=True, exist_ok=True)
    # --- ERP: tiendas y catalogo (precios en pesos, inventados) ---
    tiendas = pd.DataFrame({"cod_tienda": [f"T{i:02d}" for i in range(1, 9)], "nombre_tienda": [f"Mercado {c}" for c in CIUDADES],
                            "ciudad": CIUDADES, "m2": rng.integers(80, 400, 8)})
    tiendas.to_csv(carpeta / "erp_tiendas.csv", index=False)
    filas, k = [], 0
    for cat, subs in CATEGORIAS.items():
        for sub in subs:
            for marca in MARCAS:
                k += 1
                precio = int(rng.integers(15, 250)) * 100
                filas.append((f"P{k:03d}", f"{sub} {marca}", cat, sub, precio, int(precio * 0.7)))
    catalogo = pd.DataFrame(filas, columns=["cod_producto", "nombre_producto", "categoria", "subcategoria", "precio_lista", "costo"])
    catalogo.to_csv(carpeta / "erp_catalogo.csv", index=False)
    # --- CRM: clientes (ficticios) con ciudad sucia y duplicados ---
    n_cli = 150
    docs = [str(d) for d in rng.choice(np.arange(1_000_000_000, 1_100_000_000), n_cli, replace=False)]
    ciudad = [str(c) for c in rng.choice(CIUDADES, n_cli)]
    segmento = [str(s) for s in rng.choice(["Hogar", "Negocio", "Premium"], n_cli, p=[0.6, 0.3, 0.1])]
    crm = []
    for i in range(n_cli):
        reg = {"documento": docs[i], "nombre": f"Cliente {i + 1:03d}", "ciudad": ciudad[i], "segmento": segmento[i]}
        if rng.random() < 0.5:
            reg["contacto"] = {"email": f"cliente{i + 1:03d}@example.com"}                  # campo anidado: semiestructurado
        crm.append(reg)
    for i in rng.choice(n_cli, 25, replace=False):                                           # ciudad sucia (minusculas, mayusculas, espacios)
        c = crm[i]["ciudad"]; crm[i]["ciudad"] = [c.lower(), c.upper(), f"  {c} "][int(rng.integers(0, 3))]
    for i in rng.choice(n_cli, 8, replace=False):                                            # clientes repetidos
        crm.append({**crm[i], "ciudad": crm[i]["ciudad"].upper()})
    (carpeta / "crm_clientes.json").write_text(json.dumps(crm, ensure_ascii=False, indent=1), encoding="utf-8")
    # --- CRM: novedades (cambios de atributos con fecha efectiva) ---
    elegidos = [int(i) for i in rng.choice(n_cli, 20, replace=False)]
    nov = []
    def fecha_ef(base="2025-03-01", span=150):
        return (date.fromisoformat(base) + timedelta(days=int(rng.integers(0, span)))).isoformat()
    for i in elegidos[:12]:                                                                  # 12 mudanzas de ciudad
        nueva = str(rng.choice([c for c in CIUDADES if c != ciudad[i]]))
        nov.append({"documento": docs[i], "fecha_efectiva": fecha_ef(), "ciudad": nueva, "segmento": segmento[i]})
    otros = {"Hogar": "Negocio", "Negocio": "Premium", "Premium": "Negocio"}
    for i in elegidos[12:18]:                                                                # 6 cambios de segmento
        nov.append({"documento": docs[i], "fecha_efectiva": fecha_ef(), "ciudad": ciudad[i], "segmento": otros[segmento[i]]})
    for i in elegidos[18:]:                                                                  # 2 novedades que no cambian nada
        nov.append({"documento": docs[i], "fecha_efectiva": fecha_ef(), "ciudad": ciudad[i], "segmento": segmento[i]})
    for n_ in list(nov[:2]):                                                                 # 2 clientes cambian DOS veces (cadena de versiones)
        nov.append({"documento": n_["documento"], "fecha_efectiva": (date.fromisoformat(n_["fecha_efectiva"]) + timedelta(days=60)).isoformat(),
                    "ciudad": n_["ciudad"], "segmento": "Premium"})
    (carpeta / "crm_novedades.json").write_text(json.dumps(nov, ensure_ascii=False, indent=1), encoding="utf-8")
    # --- POS: ventas ---
    dias = pd.date_range(INICIO_DW, FIN_DW)
    peso = np.where(dias.dayofweek >= 5, 1.5, 1.0) * np.where(dias.month == 12, 1.6, 1.0)
    fechas = dias[rng.choice(len(dias), n_ventas, p=peso / peso.sum())]
    pop = 1 / (np.arange(60) + 4.0)
    prod = rng.choice(60, n_ventas, p=pop / pop.sum())
    precio_lista = catalogo.precio_lista.to_numpy()[prod]
    precio = np.where(rng.random(n_ventas) < 0.1, precio_lista * 9 // 10 // 100 * 100, precio_lista)   # 10 % con descuento
    v = pd.DataFrame({"id_venta": [f"V{i:06d}" for i in range(1, n_ventas + 1)], "fecha": fechas.strftime("%Y-%m-%d"),
                      "cod_tienda": [f"T{int(t):02d}" for t in rng.choice(np.arange(1, 9), n_ventas, p=np.array([2, 2, 1.5, 1, 1, 1, 3, 1]) / 12.5)],
                      "cod_producto": catalogo.cod_producto.to_numpy()[prod], "documento_cliente": [docs[int(j)] for j in rng.integers(0, n_cli, n_ventas)],
                      "cantidad": rng.integers(1, 7, n_ventas).astype(str), "precio_unit": precio.astype(str)})
    perm = rng.permutation(n_ventas); pos = 0
    def tomar(k):
        nonlocal pos
        idx = perm[pos:pos + k]; pos += k
        return idx
    inyectado = {}
    ix = tomar(40); v.loc[ix, "fecha"] = pd.to_datetime(v.loc[ix, "fecha"]).dt.strftime("%d/%m/%Y"); inyectado["fecha_dd/mm/aaaa"] = 40
    ix = tomar(20); v.loc[ix, "precio_unit"] = ""; inyectado["precio_vacio"] = 20
    ix = tomar(30); v.loc[ix, "documento_cliente"] = ""; inyectado["cliente_vacio"] = 30
    ix = tomar(20); v.loc[ix, "documento_cliente"] = " " + v.loc[ix, "documento_cliente"] + " "; inyectado["cliente_con_espacios"] = 20
    ix = tomar(12); v.loc[ix, "fecha"] = [["2025-02-30", "2025-04-31", "2025-13-01", "2025-06-31"][j % 4] for j in range(12)]; inyectado["fecha_invalida"] = 12
    ix = tomar(15); v.loc[ix, "cod_producto"] = "P999"; inyectado["producto_inexistente"] = 15
    ix = tomar(10); v.loc[ix, "cod_tienda"] = "T99"; inyectado["tienda_inexistente"] = 10
    ix = tomar(25); v.loc[ix, "cantidad"] = [["0", "-1", "-2"][j % 3] for j in range(25)]; inyectado["cantidad_no_positiva"] = 25
    ix = tomar(10); v.loc[ix, "documento_cliente"] = [str(2_000_000_000 + j) for j in range(10)]; inyectado["cliente_no_registrado"] = 10
    dup = v.iloc[perm[pos:pos + 60]]; inyectado["duplicado"] = 60                                # copias de filas SIN otros defectos
    v = pd.concat([v, dup]).sample(frac=1, random_state=semilla).reset_index(drop=True)
    v.to_csv(carpeta / "pos_ventas.csv", index=False)
    return inyectado


directorio = tempfile.TemporaryDirectory(prefix="dw_")           # se borra al cerrar el nucleo
FUENTES = Path(directorio.name) / "fuentes"
inyectado = generar_fuentes(FUENTES)
print("Archivos fuente:", sorted(p.name for p in FUENTES.iterdir()))
print("Defectos sembrados:", inyectado)
print(pd.read_csv(FUENTES / "pos_ventas.csv", dtype=str, keep_default_na=False).head(4).to_string(index=False))
assert (FUENTES / "pos_ventas.csv").exists() and len(pd.read_csv(FUENTES / "pos_ventas.csv")) == 4000 + inyectado["duplicado"]
````

## 3. ETL: Extraer, Transformar, Cargar (con Calidad y Cuarentena) 🧹

El ETL tiene tres fases y en cada una hay **decisiones de calidad**:

1. **Extract:** leer cada fuente **sin interpretar de más** (todo como texto) para no perder información en la lectura.
2. **Transform:** aplicar **correcciones** (arreglar lo que tiene arreglo **inequívoco**: normalizar un formato, quitar espacios, completar un precio con el de lista) y **reglas de rechazo** (lo que **no se puede** arreglar con seguridad: fecha imposible, producto que no existe, cantidad negativa). Lo rechazado **no se descarta en silencio**: va a **cuarentena con su motivo** para que alguien lo revise y lo corrija en el origen. Si una fila viola varias reglas, cuenta el **primer** motivo que falla (el orden está documentado).
3. **Load:** insertar en el almacén (siguiente sección).

Una invariante que siempre debe cumplirse: **filas de entrada = filas cargadas + filas en cuarentena**. Y como sembramos los defectos, podemos comprobar que cada regla detecta **exactamente** lo sembrado. (Corregir **no** es lo mismo que acertar: completar un precio vacío con el precio de lista es una **regla de negocio** que alguien debe aprobar; aquí queda **contada y trazada**.)

````python
def extraer(carpeta):
    """EXTRACT: lee todo como texto para no perder informacion en la lectura."""
    ventas = pd.read_csv(carpeta / "pos_ventas.csv", dtype=str, keep_default_na=False)
    catalogo = pd.read_csv(carpeta / "erp_catalogo.csv")
    tiendas = pd.read_csv(carpeta / "erp_tiendas.csv")
    crm = pd.DataFrame(json.loads((carpeta / "crm_clientes.json").read_text(encoding="utf-8")))
    novedades = pd.DataFrame(json.loads((carpeta / "crm_novedades.json").read_text(encoding="utf-8")))
    return ventas, catalogo, tiendas, crm, novedades


def limpiar_clientes(crm):
    """TRANSFORM (clientes): normaliza la ciudad contra la lista canonica y elimina repetidos por documento."""
    canon = {c.casefold(): c for c in CIUDADES}
    c = crm.copy()
    c["ciudad"] = c.ciudad.str.strip().str.casefold().map(canon)
    assert c.ciudad.notna().all(), "ciudad que no esta en la lista canonica"
    n_antes = len(c)
    c = c.drop_duplicates("documento", keep="first").reset_index(drop=True)
    return c[["documento", "nombre", "ciudad", "segmento"]], n_antes - len(c)


def transformar_ventas(ventas, catalogo, tiendas, clientes):
    """TRANSFORM (ventas): correcciones inequivocas + reglas de rechazo en orden. Devuelve (limpias, cuarentena, corregidas)."""
    v = ventas.copy()
    corregidas = {}
    v["documento_cliente"] = v.documento_cliente.str.strip()                                   # CORRECCION 1: espacios
    corregidas["cliente_con_espacios"] = int((ventas.documento_cliente != v.documento_cliente).sum())
    iso = pd.to_datetime(v.fecha, format="%Y-%m-%d", errors="coerce")                         # CORRECCION 2: dos formatos de fecha
    latino = pd.to_datetime(v.fecha, format="%d/%m/%Y", errors="coerce")
    corregidas["fecha_dd/mm/aaaa"] = int((iso.isna() & latino.notna()).sum())
    v["fecha_dt"] = iso.fillna(latino)
    lista = v.cod_producto.map(catalogo.set_index("cod_producto").precio_lista)
    precio = pd.to_numeric(v.precio_unit.replace("", np.nan), errors="coerce")
    corregidas["precio_vacio"] = int((precio.isna() & lista.notna()).sum())                     # CORRECCION 3: precio vacio -> precio de lista
    v["precio"] = precio.fillna(lista)
    v["cant"] = pd.to_numeric(v.cantidad, errors="coerce")
    sin_cliente = v.documento_cliente == ""                                                     # CORRECCION 4: sin cliente -> miembro 'No identificado'
    corregidas["cliente_vacio"] = int(sin_cliente.sum())
    # REGLAS DE RECHAZO, en este orden (el primer motivo que falla es el que se registra)
    reglas = [
        ("duplicado", v.duplicated("id_venta", keep="first")),
        ("fecha_invalida", v.fecha_dt.isna()),
        ("producto_inexistente", ~v.cod_producto.isin(catalogo.cod_producto)),
        ("tienda_inexistente", ~v.cod_tienda.isin(tiendas.cod_tienda)),
        ("cantidad_no_positiva", ~(v.cant > 0)),
        ("cliente_no_registrado", (~sin_cliente) & (~v.documento_cliente.isin(clientes.documento))),
    ]
    v["motivo"] = ""
    for nombre, viola in reglas:
        v.loc[(v.motivo == "") & viola, "motivo"] = nombre
    cuarentena = ventas.loc[v.motivo != ""].assign(motivo=v.motivo[v.motivo != ""])
    ok = v[v.motivo == ""]
    limpias = pd.DataFrame({"id_venta": ok.id_venta, "fecha": ok.fecha_dt, "cod_tienda": ok.cod_tienda, "cod_producto": ok.cod_producto,
                            "documento": ok.documento_cliente.where(ok.documento_cliente != "", "SIN-DOC"),
                            "cantidad": ok.cant.astype(int), "precio_unit": ok.precio.astype(int)}).reset_index(drop=True)
    return limpias, cuarentena.reset_index(drop=True), corregidas


ventas_raw, catalogo, tiendas, crm, novedades = extraer(FUENTES)
clientes, n_clientes_repetidos = limpiar_clientes(crm)
t0 = time.time()
ventas_limpias, cuarentena, corregidas = transformar_ventas(ventas_raw, catalogo, tiendas, clientes)
t_etl = time.time() - t0

motivos = cuarentena.motivo.value_counts()
print(f"Entrada: {len(ventas_raw)} | cargables: {len(ventas_limpias)} | cuarentena: {len(cuarentena)}  ({t_etl:.2f} s)")
print("Cuarentena por motivo:", motivos.to_dict())
print("Correcciones aplicadas (no son rechazos):", corregidas)
print("Clientes: 158 registros en el CRM ->", len(clientes), "unicos (", n_clientes_repetidos, "repetidos eliminados )")

# --- Verificaciones: conservacion de filas y cada regla encuentra EXACTAMENTE lo sembrado ---
assert len(ventas_raw) == len(ventas_limpias) + len(cuarentena)                                 # nada desaparece sin explicacion
for motivo in ["duplicado", "fecha_invalida", "producto_inexistente", "tienda_inexistente", "cantidad_no_positiva", "cliente_no_registrado"]:
    assert motivos[motivo] == inyectado[motivo], (motivo, motivos[motivo], inyectado[motivo])
for corr in ["fecha_dd/mm/aaaa", "precio_vacio", "cliente_vacio", "cliente_con_espacios"]:
    assert corregidas[corr] == inyectado[corr], corr                                              # las correcciones tambien coinciden con lo sembrado
assert n_clientes_repetidos == 8 and len(clientes) == 150 and clientes.documento.is_unique
assert ventas_limpias.id_venta.is_unique and (ventas_limpias.cantidad > 0).all() and (ventas_limpias.precio_unit > 0).all()

fig, ax = plt.subplots(figsize=(8.5, 3.2))
ax.barh(motivos.index[::-1], motivos.values[::-1], color="#dc2626")
for i, val in enumerate(motivos.values[::-1]):
    ax.text(val + 0.5, i, str(val), va="center", fontsize=9)
ax.set_xlabel("filas enviadas a cuarentena"); ax.set_title(f"Cuarentena por motivo ({len(cuarentena)} de {len(ventas_raw)} filas)", fontweight="bold")
plt.tight_layout(); plt.show()
````

## 4. Carga: el Esquema Estrella en DuckDB y la Dimensión de Variación Lenta Tipo 2 ⭐

**Carga.** Creamos las tablas con su **DDL** (llaves primarias y foráneas, `CHECK`) en **DuckDB**, una base de datos analítica **embebida y columnar** (comparte la idea de fondo con los motores de los almacenes en la nube, pero corre en un solo nodo y dentro de tu proceso). Construimos primero las **dimensiones** y después la **tabla de hechos**, buscando para cada venta las **claves sustitutas**.

**El problema de la historia.** Un cliente se **muda de Tunja a Duitama en julio**. Si el DW simplemente **sobrescribe** la ciudad (**SCD tipo 1**), las ventas de enero a junio —hechas cuando vivía en Tunja— pasarían a figurar como ventas de Duitama: **se reescribe el pasado**. La dimensión de **variación lenta tipo 2 (SCD2)** conserva **una fila por versión** con un intervalo de vigencia (`vigente_desde`, `vigente_hasta`) y una marca `es_actual`. La tabla de hechos apunta a la **versión vigente en la fecha de la venta** (*point-in-time*). Reglas que verificaremos:

1. Cada cliente tiene **exactamente una** versión actual.
2. Las vigencias de un cliente son **contiguas y no se traslapan** (la siguiente empieza el día después de que termina la anterior).
3. Una novedad que **no cambia ningún atributo vigilado** (`ciudad`, `segmento`) **no crea versión**.
4. Cada venta se une a **una sola versión** (la carga no multiplica ni pierde filas).

Además existe el miembro `sk_cliente = 0` («**No identificado**») para las ventas sin cliente: es una práctica estándar para no perder hechos cuando falta una dimensión. Las fechas de vigencia se guardan como texto ISO `AAAA-MM-DD` (que **se ordena como fecha**) para poder usar el año `9999` como «sin fin» sin los límites de `datetime64[ns]`.

**Qué veremos al final de la celda:** el **total** de ventas **no cambia** entre SCD2 y tipo 1 (la suma de las diferencias por ciudad es 0), pero el **reparto por ciudad sí**: con el tipo 1 la ciudad actual del cliente «se adueña» de ventas hechas cuando vivía en otra. El costo de SCD2 es real: la dimensión **crece** (150 → 170 versiones), las uniones necesitan la condición de **fecha** y el ETL es más complejo. Hay otros tipos (el tipo 3 guarda solo el «valor anterior» en otra columna; el tipo 0 nunca cambia; el tipo 6 combina 1, 2 y 3; *citado de memoria, verificar*) y **no existe uno «mejor»**: se elige **atributo por atributo** según si la historia importa para el análisis.

````python
FIN = "9999-12-31"
ATRIBUTOS = ["ciudad", "segmento"]                                   # atributos vigilados por el SCD2


def dia_antes(f):
    return (date.fromisoformat(f) - timedelta(days=1)).isoformat()


def dim_cliente_inicial(clientes):
    d = clientes.copy()
    d["vigente_desde"], d["vigente_hasta"], d["es_actual"] = INICIO_DW, FIN, True
    return d


def aplicar_scd2(dim, novedades):
    """SCD2: por cada novedad (en orden cronologico) cierra la version actual y abre otra. Devuelve (dim, versiones_nuevas, sin_cambio)."""
    dim, filas_nuevas, sin_cambio = dim.copy(), [], 0
    for nov in novedades.sort_values(["fecha_efectiva", "documento"]).itertuples():
        m = (dim.documento == nov.documento) & dim.es_actual
        assert m.sum() == 1, "cada cliente debe tener exactamente una version actual"
        actual = dim[m].iloc[0]
        if all(actual[a] == getattr(nov, a) for a in ATRIBUTOS):
            sin_cambio += 1                                           # regla 3: sin cambio real no se crea version
            continue
        assert nov.fecha_efectiva > actual.vigente_desde
        dim.loc[m, "vigente_hasta"] = dia_antes(nov.fecha_efectiva)    # cierra la version anterior el dia anterior
        dim.loc[m, "es_actual"] = False
        nueva = actual.to_dict(); nueva.update({a: getattr(nov, a) for a in ATRIBUTOS})
        nueva.update(vigente_desde=nov.fecha_efectiva, vigente_hasta=FIN, es_actual=True)
        dim = pd.concat([dim, pd.DataFrame([nueva])], ignore_index=True)
        filas_nuevas.append(nov.documento)
    return dim.sort_values(["documento", "vigente_desde"]).reset_index(drop=True), len(filas_nuevas), sin_cambio


def verificar_scd2(dim):
    """Las reglas 1 y 2 como funcion reutilizable (se usa tambien en la Practica 1)."""
    assert (dim.groupby("documento").es_actual.sum() == 1).all(), "regla 1"
    for _, g in dim.groupby("documento"):
        g = g.sort_values("vigente_desde")
        assert g.vigente_desde.iloc[0] == INICIO_DW and g.vigente_hasta.iloc[-1] == FIN
        for a, b in zip(g.vigente_hasta.iloc[:-1], g.vigente_desde.iloc[1:]):
            assert dia_antes(b) == a, "regla 2: vigencias contiguas y sin traslape"


dim_cli_df, n_nuevas, n_sin_cambio = aplicar_scd2(dim_cliente_inicial(clientes), novedades)
verificar_scd2(dim_cli_df)
n_reales = len(novedades) - n_sin_cambio
print(f"Novedades: {len(novedades)} -> versiones nuevas: {n_nuevas}, sin cambio real: {n_sin_cambio}. Filas de dim_cliente: {len(clientes)} -> {len(dim_cli_df)}")
assert n_sin_cambio == 2 and n_nuevas == n_reales == 20 and len(dim_cli_df) == len(clientes) + n_nuevas
ejemplo = dim_cli_df[dim_cli_df.documento == novedades.documento.iloc[0]]
print(ejemplo[["documento", "ciudad", "segmento", "vigente_desde", "vigente_hasta", "es_actual"]].to_string(index=False))

# ---------------- Dimensiones restantes (con claves sustitutas) ----------------
MESES = ["enero", "febrero", "marzo", "abril", "mayo", "junio", "julio", "agosto", "septiembre", "octubre", "noviembre", "diciembre"]
DIAS = ["lunes", "martes", "miércoles", "jueves", "viernes", "sábado", "domingo"]
fechas = pd.date_range(INICIO_DW, FIN_DW)
dim_fecha_df = pd.DataFrame({"sk_fecha": fechas.strftime("%Y%m%d").astype(int), "fecha": fechas.strftime("%Y-%m-%d"), "anio": fechas.year,
                             "trimestre": fechas.quarter, "mes": fechas.month, "nombre_mes": [MESES[m - 1] for m in fechas.month],
                             "dia": fechas.day, "dia_semana": [DIAS[d] for d in fechas.dayofweek], "es_fin_de_semana": fechas.dayofweek >= 5})
dim_producto_df = catalogo.rename(columns={"cod_producto": "cod_producto"}).assign(sk_producto=lambda d: np.arange(1, len(d) + 1))
dim_tienda_df = tiendas.assign(sk_tienda=lambda d: np.arange(1, len(d) + 1))
dim_cliente_sql = pd.concat([pd.DataFrame([{"documento": "SIN-DOC", "nombre": "No identificado", "ciudad": "Desconocida", "segmento": "Desconocido",
                                            "vigente_desde": "1900-01-01", "vigente_hasta": FIN, "es_actual": True}]), dim_cli_df], ignore_index=True)
dim_cliente_sql.insert(0, "sk_cliente", np.arange(0, len(dim_cliente_sql)))

# ---------------- DDL y carga en DuckDB ----------------
con = duckdb.connect(":memory:")
con.execute("""
CREATE TABLE dim_fecha (sk_fecha INTEGER PRIMARY KEY, fecha DATE NOT NULL, anio SMALLINT, trimestre SMALLINT, mes SMALLINT, nombre_mes VARCHAR,
                        dia SMALLINT, dia_semana VARCHAR, es_fin_de_semana BOOLEAN);
CREATE TABLE dim_producto (sk_producto INTEGER PRIMARY KEY, cod_producto VARCHAR UNIQUE NOT NULL, nombre_producto VARCHAR, categoria VARCHAR,
                           subcategoria VARCHAR, precio_lista INTEGER, costo INTEGER);
CREATE TABLE dim_tienda (sk_tienda INTEGER PRIMARY KEY, cod_tienda VARCHAR UNIQUE NOT NULL, nombre_tienda VARCHAR, ciudad VARCHAR, m2 INTEGER);
CREATE TABLE dim_cliente (sk_cliente INTEGER PRIMARY KEY, documento VARCHAR NOT NULL, nombre VARCHAR, ciudad VARCHAR, segmento VARCHAR,
                          vigente_desde DATE NOT NULL, vigente_hasta DATE NOT NULL, es_actual BOOLEAN NOT NULL);
CREATE TABLE fact_ventas (id_venta VARCHAR PRIMARY KEY,
                          sk_fecha INTEGER NOT NULL REFERENCES dim_fecha(sk_fecha), sk_producto INTEGER NOT NULL REFERENCES dim_producto(sk_producto),
                          sk_cliente INTEGER NOT NULL REFERENCES dim_cliente(sk_cliente), sk_tienda INTEGER NOT NULL REFERENCES dim_tienda(sk_tienda),
                          cantidad INTEGER NOT NULL CHECK (cantidad > 0), monto BIGINT NOT NULL CHECK (monto > 0), costo_total BIGINT NOT NULL);
""")
con.register("t_fecha", dim_fecha_df); con.register("t_prod", dim_producto_df); con.register("t_tienda", dim_tienda_df); con.register("t_cli", dim_cliente_sql)
con.execute("INSERT INTO dim_fecha SELECT sk_fecha, CAST(fecha AS DATE), anio, trimestre, mes, nombre_mes, dia, dia_semana, es_fin_de_semana FROM t_fecha")
con.execute("INSERT INTO dim_producto SELECT sk_producto, cod_producto, nombre_producto, categoria, subcategoria, precio_lista, costo FROM t_prod")
con.execute("INSERT INTO dim_tienda SELECT sk_tienda, cod_tienda, nombre_tienda, ciudad, m2 FROM t_tienda")
con.execute("INSERT INTO dim_cliente SELECT sk_cliente, documento, nombre, ciudad, segmento, CAST(vigente_desde AS DATE), CAST(vigente_hasta AS DATE), es_actual FROM t_cli")

# Hechos: para cada venta se busca la clave sustituta; el cliente se une por la VERSION VIGENTE EN LA FECHA DE LA VENTA
con.register("t_ventas", ventas_limpias)
con.execute("""
INSERT INTO fact_ventas
SELECT v.id_venta, CAST(strftime(v.fecha, '%Y%m%d') AS INTEGER), p.sk_producto, COALESCE(c.sk_cliente, 0), t.sk_tienda,
       v.cantidad, v.cantidad * v.precio_unit, v.cantidad * p.costo
FROM t_ventas v
JOIN dim_producto p ON p.cod_producto = v.cod_producto
JOIN dim_tienda t ON t.cod_tienda = v.cod_tienda
LEFT JOIN dim_cliente c ON c.documento = v.documento AND CAST(v.fecha AS DATE) BETWEEN c.vigente_desde AND c.vigente_hasta
""")

# ---------------- Verificaciones de la carga ----------------
n_hechos = con.sql("SELECT COUNT(*) FROM fact_ventas").fetchone()[0]
assert n_hechos == len(ventas_limpias), "la union con SCD2 no debe multiplicar ni perder ventas"
huerfanos = con.sql("""SELECT COUNT(*) FROM fact_ventas f LEFT JOIN dim_cliente c USING (sk_cliente) WHERE c.sk_cliente IS NULL""").fetchone()[0]
assert huerfanos == 0
monto_sql = con.sql("SELECT SUM(monto) FROM fact_ventas").fetchone()[0]
monto_pandas = int((ventas_limpias.cantidad * ventas_limpias.precio_unit).sum())
assert monto_sql == monto_pandas                                      # conciliacion: DW = datos limpios
sin_doc = int((ventas_limpias.documento == "SIN-DOC").sum())
assert con.sql("SELECT COUNT(*) FROM fact_ventas WHERE sk_cliente = 0").fetchone()[0] == sin_doc
print(f"Hechos cargados: {n_hechos} | monto total: ${monto_sql:,} (conciliado con pandas) | ventas 'No identificado': {sin_doc}")
print(con.sql("SELECT 'dim_fecha' t, COUNT(*) n FROM dim_fecha UNION ALL SELECT 'dim_producto', COUNT(*) FROM dim_producto UNION ALL SELECT 'dim_cliente', COUNT(*) FROM dim_cliente UNION ALL SELECT 'dim_tienda', COUNT(*) FROM dim_tienda UNION ALL SELECT 'fact_ventas', COUNT(*) FROM fact_ventas").df().to_string(index=False))

# ---------------- SCD2 frente a tipo 1: ¿quien reescribe el pasado? ----------------
q_hist = """SELECT c.ciudad, CAST(SUM(f.monto) AS BIGINT) AS monto FROM fact_ventas f JOIN dim_cliente c USING (sk_cliente) WHERE c.sk_cliente > 0 GROUP BY 1 ORDER BY 1"""
q_t1 = """SELECT a.ciudad, CAST(SUM(f.monto) AS BIGINT) AS monto FROM fact_ventas f JOIN dim_cliente c USING (sk_cliente)
          JOIN dim_cliente a ON a.documento = c.documento AND a.es_actual WHERE c.sk_cliente > 0 GROUP BY 1 ORDER BY 1"""
hist, t1 = con.sql(q_hist).df(), con.sql(q_t1).df()
comp = hist.merge(t1, on="ciudad", suffixes=("_scd2", "_tipo1"))
comp["diferencia"] = comp.monto_tipo1 - comp.monto_scd2
print("\nVentas por ciudad del cliente: historia (SCD2) frente a sobrescribir (tipo 1)"); print(comp.to_string(index=False))
assert comp.monto_scd2.sum() == comp.monto_tipo1.sum()               # el total no cambia...
assert comp.diferencia.abs().sum() > 0 and comp.diferencia.sum() == 0   # ...pero el reparto por ciudad SI (el tipo 1 reescribe el pasado)
````

## 5. Estrella frente a Copo de Nieve ❄️

El **copo de nieve** normaliza las dimensiones: en lugar de una tabla `dim_producto` con `categoria` y `subcategoria` repetidas en cada una de las 60 filas, hay tres tablas encadenadas (`sf_producto` → `sf_subcategoria` → `sf_categoria`). **Ventaja:** menos texto repetido y una sola «fuente» para cada nombre de categoría. **Costo:** **más uniones** en cada consulta y un modelo más difícil de leer para el analista. La recomendación habitual del modelado dimensional (Kimball, *citado de memoria, verificar*) es **preferir la estrella** salvo razones concretas, porque las dimensiones son **pequeñas frente a la tabla de hechos**: el ahorro de espacio al normalizarlas casi nunca compensa la complejidad.

Lo comprobamos con el **mismo dato**: construimos el copo de nieve en el mismo DuckDB, ejecutamos la **misma pregunta** (ventas por categoría y subcategoría) en las dos versiones, verificamos que dan **exactamente el mismo resultado** y medimos uniones, texto almacenado y tiempo (a este tamaño las diferencias de tiempo son **ruido**: lo honesto es decirlo).

````python
# --- Copo de nieve: normalizar dim_producto en tres tablas ---
sf_categoria = pd.DataFrame({"sk_categoria": np.arange(1, len(CATEGORIAS) + 1), "nombre": list(CATEGORIAS)})
subs = dim_producto_df[["categoria", "subcategoria"]].drop_duplicates().reset_index(drop=True)
subs = subs.merge(sf_categoria.rename(columns={"nombre": "categoria"}), on="categoria").assign(sk_subcategoria=lambda d: np.arange(1, len(d) + 1))
sf_subcategoria = subs[["sk_subcategoria", "subcategoria", "sk_categoria"]].rename(columns={"subcategoria": "nombre"})
sf_producto = dim_producto_df.merge(subs[["categoria", "subcategoria", "sk_subcategoria"]], on=["categoria", "subcategoria"])[
    ["sk_producto", "nombre_producto", "precio_lista", "costo", "sk_subcategoria"]]
for nombre, df in [("sf_categoria", sf_categoria), ("sf_subcategoria", sf_subcategoria), ("sf_producto", sf_producto)]:
    con.register(nombre + "_df", df)
    con.execute(f"CREATE TABLE {nombre} AS SELECT * FROM {nombre}_df")

Q_ESTRELLA = """SELECT p.categoria, p.subcategoria, CAST(SUM(f.monto) AS BIGINT) AS monto
                FROM fact_ventas f JOIN dim_producto p USING (sk_producto) GROUP BY 1, 2 ORDER BY 1, 2"""
Q_COPO = """SELECT c.nombre AS categoria, s.nombre AS subcategoria, CAST(SUM(f.monto) AS BIGINT) AS monto
            FROM fact_ventas f JOIN sf_producto p USING (sk_producto) JOIN sf_subcategoria s USING (sk_subcategoria)
            JOIN sf_categoria c USING (sk_categoria) GROUP BY 1, 2 ORDER BY 1, 2"""


def mediana_ms(q, veces=30):
    t = []
    for _ in range(veces):
        t0 = time.perf_counter(); con.sql(q).fetchall(); t.append((time.perf_counter() - t0) * 1000)
    return float(np.median(t))


r_est, r_copo = con.sql(Q_ESTRELLA).df(), con.sql(Q_COPO).df()
pd.testing.assert_frame_equal(r_est, r_copo)                                    # MISMO resultado
assert len(r_est) == 12 and r_est.monto.sum() == monto_sql

texto_estrella = 2 * len(dim_producto_df)                                       # categoria y subcategoria repetidas en cada producto
texto_copo = len(sf_categoria) + len(sf_subcategoria)                           # cada nombre una sola vez
resumen = pd.DataFrame({"uniones para llegar a la categoria": [1, 3], "tablas del modelo de producto": [1, 3],
                        "nombres de categoria/subcategoria almacenados": [texto_estrella, texto_copo],
                        "mediana de la consulta (ms)": [round(mediana_ms(Q_ESTRELLA), 2), round(mediana_ms(Q_COPO), 2)]}, index=["estrella", "copo de nieve"])
print(resumen.T.astype(object).to_string())
assert texto_estrella == 120 and texto_copo == 16
print("\nMismo resultado, 3 uniones en vez de 1; el ahorro de texto (120 -> 16 valores) es irrelevante frente a 3 928 hechos. Las diferencias de milisegundos son ruido a este tamaño.")
````

## 6. Operaciones OLAP: *Roll-up*, *Drill-down*, *Slice*, *Dice*, *Pivot* 🧊

Sobre la estrella creamos una **vista desnormalizada** `v_ventas` (hechos + dimensiones). Cada operación se calcula **dos veces** con herramientas distintas y se **verifica que coinciden**: en **SQL (DuckDB)** y en **pandas** sobre una copia en memoria del cubo.

| Operación | Idea | En SQL | En pandas |
|---|---|---|---|
| ***Roll-up*** | **Consolidar**: subir un nivel de la jerarquía (mes → trimestre → año → total) | `GROUP BY ROLLUP(anio, trimestre, mes)` | varios `groupby` con menos claves |
| ***Drill-down*** | **Detallar**: bajar un nivel (categoría → subcategoría) | agregar una columna al `GROUP BY` | `groupby` con una clave más |
| ***Slice*** | **Rebanada**: fijar **una** dimensión (solo Tunja) | `WHERE ciudad_tienda = 'Tunja'` | filtro booleano |
| ***Dice*** | **Dados**: sub-cubo restringiendo **varias** dimensiones | `WHERE a IN (...) AND b IN (...)` | filtro con `isin` |
| ***Pivot*** | **Rotar** ejes: filas ↔ columnas | `SUM(CASE WHEN ...)` o `PIVOT` | `pivot_table` |
| ***Cube*** | **Todas** las combinaciones de agrupación ($2^{n}$) | `GROUP BY CUBE(a, b)` | (se compone con `groupby`) |

En los resultados de `ROLLUP`/`CUBE`, las filas de **subtotal** muestran `NULL` en la dimensión consolidada; como un `NULL` también podría ser un dato real, se usa la función `GROUPING(...)`, que devuelve un número que indica **qué columnas se consolidaron** (0 = detalle).

````python
con.execute("""
CREATE VIEW v_ventas AS
SELECT f.id_venta, d.fecha, d.anio, d.trimestre, d.mes, d.nombre_mes, d.dia_semana, d.es_fin_de_semana,
       p.categoria, p.subcategoria, p.nombre_producto, t.ciudad AS ciudad_tienda, c.ciudad AS ciudad_cliente, c.segmento,
       f.cantidad, f.monto, f.costo_total
FROM fact_ventas f JOIN dim_fecha d USING (sk_fecha) JOIN dim_producto p USING (sk_producto)
JOIN dim_tienda t USING (sk_tienda) JOIN dim_cliente c USING (sk_cliente)""")


def sql(q):
    return con.sql(q).df()


cubo = sql("SELECT * FROM v_ventas")                                  # el mismo cubo, en memoria, para verificar con pandas
assert len(cubo) == n_hechos and int(cubo.monto.sum()) == monto_sql
# Conciliacion INDEPENDIENTE del DW: ventas limpias + catalogo (sin pasar por la estrella)
indep = ventas_limpias.merge(catalogo, on="cod_producto").assign(m=lambda d: d.cantidad * d.precio_unit).groupby("categoria").m.sum()
assert indep.to_dict() == cubo.groupby("categoria").monto.sum().to_dict()

# ---------- ROLL-UP: mes -> trimestre -> anio -> total ----------
ru = sql("""SELECT anio, trimestre, mes, GROUPING(anio, trimestre, mes) AS nivel, CAST(SUM(monto) AS BIGINT) AS monto
            FROM v_ventas GROUP BY ROLLUP (anio, trimestre, mes) ORDER BY anio, trimestre, mes""")
esperado = {0: cubo.groupby(["anio", "trimestre", "mes"]).monto.sum(), 1: cubo.groupby(["anio", "trimestre"]).monto.sum(), 3: cubo.groupby("anio").monto.sum()}
for nivel, serie in esperado.items():
    filas = ru[ru.nivel == nivel]
    claves = {0: ["anio", "trimestre", "mes"], 1: ["anio", "trimestre"], 3: ["anio"]}[nivel]
    assert filas.set_index(claves).monto.to_dict() == serie.to_dict(), nivel
assert ru[ru.nivel == 7].monto.iloc[0] == cubo.monto.sum() and len(ru) == 12 + 4 + 1 + 1       # 18 filas: n+1 = 4 niveles de consolidacion
print("ROLL-UP (primeras filas; nivel 0 = mes, 1 = subtotal de trimestre, 3 = anio, 7 = total):")
print(ru.head(5).to_string(index=False)); print(ru.tail(2).to_string(index=False))

# ---------- DRILL-DOWN: de categoria a subcategoria (cuarto trimestre) ----------
dd0 = sql("SELECT categoria, CAST(SUM(monto) AS BIGINT) AS monto FROM v_ventas WHERE trimestre = 4 GROUP BY 1 ORDER BY 1")
dd1 = sql("SELECT categoria, subcategoria, CAST(SUM(monto) AS BIGINT) AS monto FROM v_ventas WHERE trimestre = 4 GROUP BY 1, 2 ORDER BY 1, 2")
assert dd1.groupby("categoria").monto.sum().to_dict() == dd0.set_index("categoria").monto.to_dict()      # detallar CONSERVA el total
assert dd1.set_index(["categoria", "subcategoria"]).monto.to_dict() == cubo[cubo.trimestre == 4].groupby(["categoria", "subcategoria"]).monto.sum().to_dict()
print("\nDRILL-DOWN T4:", dd0.set_index("categoria").monto.to_dict(), "->", len(dd1), "subcategorias; suma conservada")


# ---------- SLICE, DICE y PIVOT (categoria x trimestre) ----------
def pivot_sql(filtro="TRUE"):
    cols = ", ".join(f"CAST(COALESCE(SUM(CASE WHEN trimestre = {t} THEN monto END), 0) AS BIGINT) AS T{t}" for t in (1, 2, 3, 4))
    return sql(f"SELECT categoria, {cols} FROM v_ventas WHERE {filtro} GROUP BY categoria ORDER BY categoria").set_index("categoria")


def pivot_pd(df):
    p = df.pivot_table(index="categoria", columns="trimestre", values="monto", aggfunc="sum", fill_value=0).reindex(columns=[1, 2, 3, 4], fill_value=0)
    p.columns = [f"T{t}" for t in p.columns]
    return p.astype("int64")


def igual(a, b):
    pd.testing.assert_frame_equal(a, b, check_dtype=False, check_names=False)


slice_sql = pivot_sql("ciudad_tienda = 'Tunja'")                                                     # SLICE: una dimension fija
igual(slice_sql, pivot_pd(cubo[cubo.ciudad_tienda == "Tunja"]))
dice_filtro = "categoria IN ('Lácteos', 'Bebidas') AND ciudad_tienda IN ('Tunja', 'Duitama', 'Sogamoso') AND trimestre IN (3, 4)"
dice_sql = pivot_sql(dice_filtro)[["T3", "T4"]]                                                      # DICE: varias dimensiones
mask = cubo.categoria.isin(["Lácteos", "Bebidas"]) & cubo.ciudad_tienda.isin(["Tunja", "Duitama", "Sogamoso"]) & cubo.trimestre.isin([3, 4])
igual(dice_sql, pivot_pd(cubo[mask])[["T3", "T4"]])
pivot_full = pivot_sql(); igual(pivot_full, pivot_pd(cubo))                                         # PIVOT: categoria x trimestre
nativo = sql("PIVOT v_ventas ON trimestre USING CAST(SUM(monto) AS BIGINT) GROUP BY categoria ORDER BY categoria").set_index("categoria")
igual(nativo.rename(columns=lambda c: f"T{c}")[["T1", "T2", "T3", "T4"]], pivot_full)                # la sintaxis PIVOT de DuckDB da lo mismo
assert pivot_full.to_numpy().sum() == cubo.monto.sum()
print("\nSLICE (solo Tunja), categoria x trimestre:"); print(slice_sql.to_string())
print("\nDICE (Lácteos+Bebidas, 3 ciudades, T3-T4):"); print(dice_sql.to_string())
print("\nPIVOT categoria x trimestre (pesos):"); print(pivot_full.to_string())

# ---------- CUBE: las 2^n combinaciones de agrupacion ----------
cu = sql("""SELECT categoria, trimestre, GROUPING(categoria, trimestre) AS nivel, CAST(SUM(monto) AS BIGINT) AS monto
            FROM v_ventas GROUP BY CUBE (categoria, trimestre)""")
assert cu.nivel.nunique() == 2 ** 2 and len(cu) == (4 + 1) * (4 + 1)                                # 4 niveles de agrupacion, 25 filas
assert (cu.groupby("nivel").monto.sum() == cubo.monto.sum()).all()                                  # cada nivel reparte el MISMO total
assert cu[cu.nivel == 0].set_index(["categoria", "trimestre"]).monto.to_dict() == cubo.groupby(["categoria", "trimestre"]).monto.sum().to_dict()
assert cu[cu.nivel == 1].set_index("categoria").monto.to_dict() == cubo.groupby("categoria").monto.sum().to_dict()
assert cu[cu.nivel == 2].set_index("trimestre").monto.to_dict() == cubo.groupby("trimestre").monto.sum().to_dict()
print(f"\nCUBE(categoria, trimestre): {cu.nivel.nunique()} niveles (2^2) y {len(cu)} filas = (4+1) x (4+1); cada nivel suma ${int(cubo.monto.sum()):,}")

# ---------- Grafico: roll-up mensual y pivot ----------
fig, (a1, a2) = plt.subplots(1, 2, figsize=(13, 4.2), gridspec_kw={"width_ratios": [1.25, 1]})
m = ru[ru.nivel == 0]
a1.bar(m.mes, m.monto / 1e6, color=[["#0ea5e9", "#16a34a", "#f59e0b", "#dc2626"][t - 1] for t in m.trimestre])
for t, (x0, x1) in enumerate([(1, 3), (4, 6), (7, 9), (10, 12)], start=1):
    a1.text((x0 + x1) / 2, m.monto.max() / 1e6 * 1.08, f"T{t}: ${ru[(ru.nivel == 1) & (ru.trimestre == t)].monto.iloc[0] / 1e6:.1f} M", ha="center", fontsize=9, fontweight="bold")
a1.set_xticks(range(1, 13)); a1.set_xlabel("mes (2025)"); a1.set_ylabel("ventas (millones de pesos)"); a1.set_ylim(0, m.monto.max() / 1e6 * 1.2)
a1.set_title("Roll-up: de mes a trimestre", fontweight="bold")
im = a2.imshow(pivot_full.to_numpy() / 1e6, cmap="Blues", aspect="auto")
a2.set_xticks(range(4)); a2.set_xticklabels(["T1", "T2", "T3", "T4"]); a2.set_yticks(range(4)); a2.set_yticklabels(pivot_full.index)
for i in range(4):
    for j in range(4):
        a2.text(j, i, f"{pivot_full.iloc[i, j] / 1e6:.1f}", ha="center", va="center", fontsize=9, color="white" if pivot_full.iloc[i, j] > 0.6 * pivot_full.to_numpy().max() else "black")
a2.set_title("Pivot: categoria x trimestre (millones)", fontweight="bold"); a2.grid(False)
plt.tight_layout(); plt.show()
````

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **OLTP frente a OLAP** | OLTP registra (muchas escrituras pequeñas, estado actual, modelo normalizado); un **DW** responde preguntas analíticas sobre el **histórico** (pocas lecturas enormes, modelo dimensional, almacenamiento columnar). Inmon: **orientado a temas, integrado, variante en el tiempo, no volátil** |
| **Esquema estrella** | Una **tabla de hechos** (medidas aditivas + claves foráneas a dimensiones + dimensión degenerada) rodeada de **dimensiones desnormalizadas**; se decide por **proceso, granularidad, dimensiones y hechos**; las **claves sustitutas** desacoplan el DW de las fuentes |
| **ETL con calidad** | **Corregir** lo inequívoco, **rechazar a cuarentena con motivo** lo demás, y verificar la invariante **entrada = cargadas + cuarentena** (4 060 = 3 928 + 132); cada regla encontró **exactamente** lo sembrado |
| **SCD tipo 2** | Una fila por **versión** con vigencia (`vigente_desde/hasta`, `es_actual`); los hechos apuntan a la versión **vigente en la fecha**; el tipo 1 **reescribe el pasado** (mismo total, distinto reparto por ciudad). Cuesta espacio y complejidad: se elige **por atributo** |
| **Estrella frente a copo de nieve** | **Mismo resultado**; el copo usa **3 uniones** en lugar de 1 y ahorra texto repetido (120 → 16 valores), irrelevante frente a la tabla de hechos: se prefiere la **estrella** por simplicidad |
| **OLAP** | *Roll-up* y *drill-down* suben y bajan la jerarquía (**conservan el total**), *slice* fija una dimensión, *dice* varias, *pivot* rota ejes; `ROLLUP` da $n+1$ niveles y `CUBE` $2^{n}$; `GROUPING()` distingue subtotales de `NULL` reales. **SQL y pandas coinciden** |
| **Honestidad** | Datos **sintéticos**; DuckDB es un motor **embebido de un solo nodo** (los DW en la nube —*Redshift*, *BigQuery*, *Snowflake*— usan ideas similares a gran escala, *citado de memoria, verificar*); las diferencias de milisegundos a este tamaño son ruido; **corregir no es acertar** |
