## 1. Del Dato al Valor: la Pirámide DIKW 🔺

Un dato aislado **no sirve para decidir**. El valor aparece cuando se **transforma** a lo largo de una jerarquía que la literatura llama **DIKW** (*Data → Information → Knowledge → Wisdom*; a veces se cierra con la **decisión** o la **acción**; la atribución y los límites exactos de cada nivel varían según el autor):

| Nivel | Qué es | Ejemplo (estación meteorológica de un cultivo de papa) | Quién lo produce |
|---|---|---|---|
| **Dato** | Símbolos o medidas **sin contexto** | `63.5`, `11.2`, `63.5`… (cientos de números) | Sensores, formularios, sistemas |
| **Información** | Datos con **contexto, unidades y resumen** | «Humedad relativa media del día: 78 %; máxima: 94 %; temperatura media: 12 °C» | Limpieza, agregación, consultas |
| **Conocimiento** | **Patrones y reglas** que relacionan información con consecuencias | «Muchas horas seguidas de humedad muy alta con frío moderado favorecen enfermedades fúngicas de la papa» | Análisis, modelos, expertos del dominio |
| **Decisión / sabiduría** | **Qué hacer** con ese conocimiento, ponderando contexto y costos | «Hoy conviene revisar el lote y, si procede, aplicar un tratamiento preventivo» | Personas (agrónomo, agricultor) apoyadas por el sistema |

> ⚠️ **Idea clave:** el Big Data aporta sobre todo en los **dos primeros saltos** (convertir un mar de datos en información y patrones); el último salto, **decidir**, sigue siendo humano, y **es donde se crea (o se pierde) el valor**. Un modelo muy preciso que nadie usa en una decisión **vale cero**.

La celda siguiente **ejecuta** la pirámide con datos **sintéticos** de dos semanas de una estación (horaria). La regla de «hora de riesgo» (humedad ≥ 90 % con temperatura entre 10 y 20 °C, y alerta si hay ≥ 6 horas de riesgo en el día) es **solo ilustrativa: no es una recomendación agronómica**; para un cultivo real debe definirla un especialista y validarse con datos locales.

````python
# ============================================================
# Configuracion inicial
# ============================================================
# !pip install -q numpy pandas matplotlib scikit-learn

import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.patches import Polygon
from sklearn.cluster import KMeans
from sklearn.ensemble import IsolationForest
from sklearn.metrics import adjusted_rand_score, average_precision_score, silhouette_score
from sklearn.preprocessing import StandardScaler

warnings.filterwarnings("ignore")
SEED = 42
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__}")
````

````python
# ============================================================
# 1. La piramide DIKW ejecutada: de lecturas horarias (dato) a una alerta (decision), con datos SINTETICOS
# ============================================================
def generar_estacion(dias=14, semilla=SEED):
    """Lecturas horarias sinteticas de humedad relativa (%) y temperatura (C); los dias 8, 9 y 10 son lluviosos (mas humedad, mas frio)."""
    rng = np.random.default_rng(semilla)
    h = np.arange(dias * 24)
    dia, hora = h // 24, h % 24
    lluvia = ((dia >= 8) & (dia <= 10)).astype(float)
    ciclo = np.sin(2 * np.pi * (hora - 9) / 24)                                # maximo de calor hacia las 15 h
    temp = 13 + 4 * ciclo - 1.5 * lluvia + rng.normal(0, 0.7, h.size)
    hum = np.clip(72 - 12 * ciclo + 26 * lluvia + rng.normal(0, 3, h.size), 0, 100)
    return pd.DataFrame({"dia": dia, "hora": hora, "humedad": hum.round(1), "temperatura": temp.round(1)})


def a_informacion(d):
    """DATO -> INFORMACION: resumen diario con unidades (media y maxima)."""
    return d.groupby("dia").agg(hum_media=("humedad", "mean"), hum_max=("humedad", "max"), temp_media=("temperatura", "mean")).round(1)


def a_conocimiento(d, hum_min=90.0, t_lo=10.0, t_hi=20.0):
    """INFORMACION -> CONOCIMIENTO: regla (ilustrativa) 'hora de riesgo' = humedad alta con frio moderado; se cuentan por dia."""
    riesgo = (d.humedad >= hum_min) & d.temperatura.between(t_lo, t_hi)
    return riesgo.groupby(d.dia).sum().rename("horas_de_riesgo")


def a_decision(horas_riesgo, umbral=6):
    """CONOCIMIENTO -> DECISION: dias en que se recomienda revisar el lote."""
    return list(horas_riesgo.index[horas_riesgo >= umbral])


crudo = generar_estacion()
info = a_informacion(crudo)
conocimiento = a_conocimiento(crudo)
alertas = a_decision(conocimiento)
print(f"DATO: {len(crudo)} lecturas horarias (ej.: {crudo.humedad.iloc[0]}, {crudo.humedad.iloc[1]}, {crudo.humedad.iloc[2]} ...)")
print("\nINFORMACION (primeros 3 dias y dias lluviosos):"); print(info.loc[[0, 1, 2, 8, 9, 10]].to_string())
print("\nCONOCIMIENTO (horas de riesgo por dia):", conocimiento.to_dict())
print("DECISION: revisar el lote los dias", alertas)
assert len(crudo) == 14 * 24 and len(info) == 14
assert alertas and set(alertas) <= {8, 9, 10}                                   # solo los dias lluviosos disparan alerta
assert conocimiento[[0, 1, 2, 3]].sum() == 0                                     # los dias secos no tienen horas de riesgo

# --- Dibujo: la piramide (izquierda) y las horas de riesgo por dia (derecha) ---
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(13, 5), gridspec_kw={"width_ratios": [1.15, 1]})
capas = [("Dato", "336 lecturas horarias", PALETA[3]), ("Información", "resumen diario con unidades", PALETA[2]),
         ("Conocimiento", "regla: horas de riesgo por día", PALETA[0]), ("Decisión", "¿revisar el lote hoy?", PALETA[1])]
for i, (nombre, ejemplo, col) in enumerate(capas):
    y0, y1 = i, i + 0.92
    w0, w1 = 4.2 - 0.9 * i, 4.2 - 0.9 * (i + 1) + 0.0
    ax.add_patch(Polygon([(-w0, y0), (w0, y0), (w1 + 0.08, y1), (-w1 - 0.08, y1)], closed=True, color=col, alpha=0.9))
    ax.text(0, y0 + 0.46, nombre, ha="center", va="center", color="white", fontweight="bold", fontsize=11)
    ax.text(4.6, y0 + 0.46, ejemplo, ha="left", va="center", fontsize=9)
ax.set_xlim(-4.5, 9.5); ax.set_ylim(-0.1, 4.1); ax.axis("off")
ax.set_title("Pirámide DIKW: cada capa filtra y da sentido a la anterior", fontweight="bold", fontsize=10)
colores = [PALETA[1] if d in alertas else PALETA[0] for d in conocimiento.index]
ax2.bar(conocimiento.index, conocimiento.values, color=colores)
ax2.axhline(6, color="black", ls="--", lw=1); ax2.text(0, 6.3, "umbral de alerta (ilustrativo)", fontsize=8)
ax2.set_xlabel("día"); ax2.set_ylabel("horas de riesgo"); ax2.set_title("Del conocimiento a la decisión: días con alerta en rojo", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lectura.** Las 336 lecturas se vuelven **14 filas de información** (con unidades y contexto), luego **14 números de conocimiento** (horas de riesgo por día) y por último **una lista corta de decisiones** (días 8, 9 y 10, los lluviosos por construcción). En cada salto **se descarta casi todo el volumen y se gana significado**. Si el umbral o la regla están mal elegidos, el sistema **alerta de más (fatiga, desconfianza) o de menos (cosecha perdida)**: por eso la regla debe nacer del **experto** y medirse con datos reales.

### 💰 La economía del dato

El dato **no se comporta como un bien físico**, y eso explica por qué unas organizaciones capturan tanto valor:

- **No rival:** usarlo no lo agota; el mismo dato sirve a muchos análisis a la vez.
- **Costo de copia casi nulo**, pero **costo de producirlo y curarlo alto** (sensores, limpieza, personas); lo escaso no es el dato en bruto sino el **dato de calidad con contexto**.
- **Complementariedad:** dos conjuntos de datos **combinados** valen más que por separado (cruzar clima con ventas).
- **Efectos de red y de escala:** más usuarios generan más datos, que mejoran el producto, que atrae más usuarios.
- **Perecedero:** muchos datos **pierden valor con el tiempo** (la ubicación de hace un mes sirve poco para tráfico) y **generan costos** mientras se guardan (almacenamiento, seguridad, cumplimiento).
- **Externalidades:** el costo de una filtración recae también en las **personas**, no solo en la organización.

### ⛓️ La cadena de valor del dato

| Eslabón | Qué se añade | Dónde se pierde valor |
|---|---|---|
| **1. Generación** | Se produce el dato (sensor, transacción, formulario) | Datos que **nunca se capturan** o se capturan mal |
| **2. Captura e ingesta** | Se mueve a donde puede usarse | Silos incompatibles, pérdidas en la transmisión |
| **3. Almacenamiento** | Se conserva con seguridad y estructura | Sin catálogo ni metadatos nadie encuentra nada |
| **4. Procesamiento** | Se limpia, integra y transforma | Calidad pobre: «basura entra, basura sale» |
| **5. Análisis y modelado** | Se extraen patrones y predicciones | Preguntas mal planteadas, sobreajuste |
| **6. Decisión y producto** | Se actúa: una política, una alerta, una oferta | **El análisis no llega a quien decide** |
| **7. Retroalimentación** | Se mide el efecto y se corrige | No se mide el resultado, no se aprende |

Las tres demostraciones siguientes recorren los eslabones **5 y 6** con datos **sintéticos**: lo que ves es el **mecanismo**, no el rendimiento esperable en un caso real.

## 3. Demostración A: Segmentación de Clientes (Comercio) 🛍️

**Pregunta:** ¿hay grupos de clientes con comportamientos distintos que merezcan tratos distintos? Generamos 3 000 clientes con tres comportamientos ocultos —**ocasionales**, **regulares** y **fieles**— descritos por **recencia** (días desde la última compra), **frecuencia** (compras al año) y **ticket medio**, y aplicamos **K-Means** sobre las variables transformadas (logaritmo y estandarización). Como los datos son sintéticos, conocemos la **verdad** y podemos medir con el **índice de Rand ajustado** (ARI) cuánto se parece la segmentación a ella; en la realidad **no tendrías esa verdad**.

````python
# ============================================================
# 3. Segmentacion de clientes: K-Means sobre recencia, frecuencia y ticket (datos SINTETICOS con 3 segmentos ocultos)
# ============================================================
def generar_clientes(n=3000, semilla=3):
    """3 segmentos ocultos (60 % ocasionales, 30 % regulares, 10 % fieles); la 'verdad' queda en segmento_real."""
    rng = np.random.default_rng(semilla)
    seg = rng.choice(3, n, p=[0.60, 0.30, 0.10])
    rec = np.array([rng.lognormal(np.log(120), 0.5, n), rng.lognormal(np.log(45), 0.5, n), rng.lognormal(np.log(12), 0.5, n)])
    fre = np.array([rng.poisson(2, n) + 1, rng.poisson(8, n) + 1, rng.poisson(24, n) + 1])
    tic = np.array([rng.lognormal(np.log(30), 0.4, n), rng.lognormal(np.log(45), 0.4, n), rng.lognormal(np.log(80), 0.4, n)])
    i = np.arange(n)
    d = pd.DataFrame({"recencia_dias": rec[seg, i].round(0), "compras_anio": fre[seg, i], "ticket_medio": tic[seg, i].round(2), "segmento_real": seg})
    d["ingreso_anual"] = d.compras_anio * d.ticket_medio
    return d


clientes = generar_clientes()
X = StandardScaler().fit_transform(np.column_stack([np.log1p(clientes.recencia_dias), np.log(clientes.compras_anio), np.log(clientes.ticket_medio)]))

filas = []
for k in range(2, 7):
    etiquetas = KMeans(k, n_init=10, random_state=SEED).fit_predict(X)
    filas.append({"k": k, "silueta": silhouette_score(X, etiquetas), "ARI vs verdad (solo sintetico)": adjusted_rand_score(clientes.segmento_real, etiquetas)})
barrido = pd.DataFrame(filas).set_index("k")
print(barrido.round(3).to_string())
k_sil, k_ari = int(barrido.silueta.idxmax()), int(barrido.iloc[:, 1].idxmax())
print(f"\nLa silueta elige k = {k_sil}; la verdad conocida (ARI) se recupera mejor con k = {k_ari}.")

km = KMeans(3, n_init=10, random_state=SEED).fit(X)
clientes["cluster"] = km.labels_
perfil = clientes.groupby("cluster").agg(clientes=("ingreso_anual", "size"), recencia_mediana=("recencia_dias", "median"), compras_mediana=("compras_anio", "median"),
                                          ticket_mediano=("ticket_medio", "median"), ingreso_total=("ingreso_anual", "sum"))
perfil["% clientes"] = (100 * perfil.clientes / perfil.clientes.sum()).round(1)
perfil["% ingreso"] = (100 * perfil.ingreso_total / perfil.ingreso_total.sum()).round(1)
perfil = perfil.sort_values("ingreso_total", ascending=False)
print("\n" + perfil.drop(columns="ingreso_total").round(1).to_string())

mejor = perfil.iloc[0]
assert barrido.silueta.idxmax() == 2 and barrido.iloc[:, 1].idxmax() == 3        # honesto: la silueta prefiere 2, la estructura real es de 3
assert adjusted_rand_score(clientes.segmento_real, clientes.cluster) > 0.7
assert mejor["% clientes"] < 15 and mejor["% ingreso"] > 40                       # concentracion: pocos clientes, mucho ingreso
assert abs(perfil["% ingreso"].sum() - 100) < 0.5

fig, ax = plt.subplots(1, 2, figsize=(12, 4.5))
for c in perfil.index:
    m = clientes.cluster == c
    ax[0].scatter(np.log1p(clientes.recencia_dias[m]), np.log(clientes.compras_anio[m]), s=8, alpha=0.5, label=f"grupo {c}")
ax[0].set_xlabel("log(1 + recencia en días)"); ax[0].set_ylabel("log(compras por año)"); ax[0].legend(); ax[0].set_title("Los tres grupos en recencia y frecuencia", fontweight="bold", fontsize=10)
ancho = 0.38; xs = np.arange(len(perfil))
ax[1].bar(xs - ancho / 2, perfil["% clientes"], ancho, color=PALETA[0], label="% de clientes"); ax[1].bar(xs + ancho / 2, perfil["% ingreso"], ancho, color=PALETA[1], label="% del ingreso")
ax[1].set_xticks(xs); ax[1].set_xticklabels([f"grupo {c}" for c in perfil.index]); ax[1].legend(); ax[1].set_title("Concentración del ingreso por segmento", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lectura honesta.** (1) La **silueta** más alta es la de $k=2$ (≈ 0.44), pero la estructura real tiene **tres** grupos: con $k=3$ la segmentación coincide mucho más con la verdad (ARI ≈ 0.77, frente a ≈ 0.70 con $k=2$) aunque su silueta sea menor (≈ 0.36). La silueta premia grupos **compactos y separados** y aquí los dos segmentos «de menor valor» se solapan: **una métrica interna no sustituye el criterio de negocio** (¿necesito tres tratos distintos?). (2) El grupo de mayor ingreso reúne **≈ 11 % de los clientes pero ≈ 56 % del ingreso**: la concentración es lo que **justifica tratos diferenciados** (programa de fidelización para pocos, reactivación para los inactivos). Es el valor: **dónde poner la atención escasa**. (3) Los grupos de K-Means **no son «verdad»**: otra semilla o variables distintas darían otros límites; antes de actuar se **validan con el negocio** y con una prueba (por ejemplo, un piloto controlado).

## 4. Demostración B: Detección de Anomalías en Transacciones (Finanzas) 🚨

**Pregunta:** de 40 000 transacciones de 500 clientes, ≈ 1 % son **fraudulentas** (**sintéticas**): la mitad con un **monto inusualmente alto para ese cliente** (4–10 veces su valor típico) y la otra mitad **de madrugada** (1–4:30 h). Un equipo de revisión solo puede mirar **el 1 % de las transacciones** (400 casos), así que lo que importa es **qué tan puros** son esos 400 alertas. Comparamos cuatro puntajes:

1. **Monto global:** alertar las transacciones de mayor monto (el «umbral» ingenuo).
2. **$z$ robusto por cliente:** qué tan lejos está el $\log$ del monto de la **mediana de ese cliente**, en unidades de la **MAD** (desviación absoluta mediana, escalada por $1.4826$), $z=\dfrac{\log m-\operatorname{mediana}_c}{1.4826\,\operatorname{MAD}_c}$.
3. ***Isolation Forest*** sobre $(\log m,\ \text{hora})$ (sin contexto del cliente).
4. ***Isolation Forest*** sobre $(z,\ \text{hora})$: con **contexto del cliente** y de la **hora**.

Se evalúa con **precisión en el 1 % superior** y con la **precisión promedio** (área bajo la curva precisión–exhaustividad, AP). La **línea base** del azar es la tasa de fraude (≈ 1 %).

````python
# ============================================================
# 4. Deteccion de anomalias: umbral global vs z robusto por cliente vs Isolation Forest (datos SINTETICOS, ~1 % de fraude)
# ============================================================
def generar_transacciones_fraude(n_clientes=500, n=40_000, tasa=0.01, semilla=11):
    rng = np.random.default_rng(semilla)
    cli = rng.integers(0, n_clientes, n)
    tipico = rng.lognormal(np.log(60), 0.6, n_clientes)                        # monto tipico de cada cliente
    monto = rng.lognormal(np.log(tipico[cli]), 0.35)
    hora = np.clip(rng.normal(14, 3.5, n), 6, 23.5)
    fraude = rng.random(n) < tasa
    grande = fraude & (rng.random(n) < 0.5)                                    # mitad: monto muy alto para ESE cliente
    noche = fraude & ~grande                                                   # mitad: madrugada (y monto algo mayor)
    monto[grande] *= rng.uniform(4, 10, grande.sum())
    monto[noche] *= rng.uniform(1, 2, noche.sum()); hora[noche] = rng.uniform(1, 4.5, noche.sum())
    return pd.DataFrame({"cliente": cli, "monto": monto.round(2), "hora": hora, "fraude": fraude.astype(int)})


tx = generar_transacciones_fraude()
lm = np.log(tx.monto)
mediana_c = lm.groupby(tx.cliente).transform("median")
mad_c = (lm - mediana_c).abs().groupby(tx.cliente).transform("median") * 1.4826
z = (lm - mediana_c) / mad_c                                                   # z robusto: resiste a los propios fraudes (mediana y MAD)
hora_xy = np.column_stack([np.sin(2 * np.pi * tx.hora / 24), np.cos(2 * np.pi * tx.hora / 24)])   # la hora es circular: 23 h esta cerca de 0 h
X_ctx = np.column_stack([z, hora_xy])
X_sin = np.column_stack([lm, tx.hora])

puntajes = {
    "monto global": lm.values,
    "z robusto por cliente": z.abs().values,
    "Isolation Forest (monto, hora)": -IsolationForest(n_estimators=200, random_state=SEED).fit(X_sin).score_samples(X_sin),
    "Isolation Forest (z, hora)": -IsolationForest(n_estimators=200, random_state=SEED).fit(X_ctx).score_samples(X_ctx),
}
K = int(0.01 * len(tx))                                                        # presupuesto de revision: 1 % de las transacciones
base = tx.fraude.mean()
filas = []
for nombre, s in puntajes.items():
    top = np.argsort(-s)[:K]
    filas.append({"puntaje": nombre, "precisión en el 1 % superior": tx.fraude.values[top].mean(), "AP": average_precision_score(tx.fraude, s)})
res = pd.DataFrame(filas).set_index("puntaje")
res["veces mejor que el azar"] = res["precisión en el 1 % superior"] / base
print(f"Transacciones: {len(tx):,} | fraude: {tx.fraude.sum()} ({base:.2%}) | presupuesto de revisión: {K} casos")
print(res.round(3).to_string())

assert abs(base - 0.01) < 0.003 and K == 400
assert res.loc["z robusto por cliente", "AP"] > res.loc["monto global", "AP"]                     # el contexto del cliente ayuda
assert res.loc["Isolation Forest (z, hora)", "AP"] > res.loc["z robusto por cliente", "AP"]       # y la hora capta el fraude 'de madrugada'
assert res["veces mejor que el azar"].min() > 20                                                    # todos superan con claridad al azar

fig, ax = plt.subplots(figsize=(8.5, 4.5))
ax.barh(res.index[::-1], res["AP"].values[::-1], color=PALETA[0]); ax.axvline(base, color="black", ls="--", lw=1)
ax.text(base + 0.01, 0.1, "azar (tasa de fraude)", fontsize=8)
ax.set_xlabel("precisión promedio (AP)"); ax.set_title("El contexto (cliente y hora) mejora la detección", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lectura honesta.** El umbral global de monto —lo más ingenuo— ya supera por mucho al azar (≈ 34 % de precisión en el 1 % superior frente a ≈ 1 % de la tasa base), porque **un fraude grande es grande para casi todos**. Pero **olvida el contexto**: un cliente que gasta poco y de pronto gasta 8 veces su costumbre **no** es llamativo en términos absolutos, y a un cliente de montos altos se le alerta de más. El **$z$ robusto por cliente** sube la precisión a ≈ 50 % y el *Isolation Forest* con **$z$ y hora** a ≈ 52 % (AP ≈ 0.58, frente a ≈ 0.32 del monto global); el mismo *Isolation Forest* **sin** el contexto del cliente se queda en ≈ 41 % (AP ≈ 0.36): el modelo más sofisticado **no** gana por ser sofisticado, sino por recibir **mejores variables**. Tres advertencias: (1) **cada alerta es una decisión humana**, no una sentencia; con 1 % de presupuesto aún **casi la mitad de los alertas son falsas**; (2) el *Isolation Forest* no «entiende» fraude, solo **aísla lo raro**: funciona porque lo raro **coincide** con el fraude en este mundo sintético; (3) las estadísticas por cliente se calcularon **con todos los datos** (incluidos los fraudes): en producción se calculan **solo con el pasado** de cada cliente, y los clientes **nuevos** no tienen historia (*arranque en frío*). El valor para el negocio no es «detectar todo», sino **concentrar la atención escasa de los revisores** donde es más probable el fraude.

## 5. Demostración C: Pronóstico de Demanda (Comercio / Agro) 📈

**Pregunta:** ¿cuántas unidades de un producto se pedirán en las **próximas 8 semanas**? Generamos **≈ 3 años de demanda diaria sintética** con **tendencia creciente**, **patrón semanal** (fines de semana más altos) y **estacionalidad anual**, más ruido. Comparamos, sobre los últimos **56 días** (que el modelo **no ve**), cuatro pronósticos, de menos a más elaborado:

| Pronóstico | Regla |
|---|---|
| **Ingenuo** | «Mañana igual que el último día observado» |
| **Promedio móvil de 7 días** | La media de la última semana, constante |
| **Estacional ingenuo** | «Cada día igual que el mismo día de la última semana observada» |
| **Regresión** | Mínimos cuadrados con **tendencia + día de la semana + ciclo anual** ($\sin$ y $\cos$ de $2\pi t/365.25$) |

Además del error absoluto medio (**MAE**) y el **sMAPE**, traducimos el error a **costo** con una regla de **inventario** (parámetros **ilustrativos**): si pides de más pagas **1** por cada unidad sobrante; si pides de menos pierdes **3** por cada unidad que no pudiste vender. Es la lectura de negocio: **un error por defecto cuesta más que uno por exceso**.

````python
# ============================================================
# 5. Pronostico de demanda: lineas base vs regresion con tendencia y estacionalidad (datos SINTETICOS) y su costo de inventario
# ============================================================
def generar_demanda(dias=1099, semilla=5):
    rng = np.random.default_rng(semilla)
    t = np.arange(dias)
    semana = np.array([0.90, 0.88, 0.93, 1.00, 1.15, 1.35, 1.10])              # lun..dom: el fin de semana vende mas
    base = (300 + 0.12 * t) * semana[t % 7] * (1 + 0.10 * np.sin(2 * np.pi * t / 365.25))
    return pd.Series(np.maximum(0, base + rng.normal(0, 18, dias)), index=pd.date_range("2022-01-03", periods=dias, freq="D"))   # el 3-ene-2022 es lunes


def disenio(t, dia_semana):
    """Columnas: constante, tendencia, 6 indicadoras de dia (lunes = base) y ciclo anual."""
    D = np.eye(7)[dia_semana][:, 1:]
    return np.column_stack([np.ones(len(t)), t, D, np.sin(2 * np.pi * t / 365.25), np.cos(2 * np.pi * t / 365.25)])


demanda = generar_demanda()
H = 56
y_train, y_test = demanda.iloc[:-H], demanda.iloc[-H:]
t_all, dow = np.arange(len(demanda)), demanda.index.dayofweek.values
beta = np.linalg.lstsq(disenio(t_all[:-H], dow[:-H]), y_train.values, rcond=None)[0]

pronosticos = {
    "ingenuo": np.full(H, y_train.iloc[-1]),
    "promedio móvil 7 d": np.full(H, y_train.iloc[-7:].mean()),
    "estacional ingenuo": np.tile(y_train.iloc[-7:].values, H // 7),
    "regresión": disenio(t_all[-H:], dow[-H:]) @ beta,
}
C_SOBRA, C_FALTA = 1.0, 3.0                                                    # costo por unidad sobrante / faltante (ILUSTRATIVOS)
filas = []
for nombre, p in pronosticos.items():
    e = y_test.values - p
    costo = np.where(p > y_test.values, C_SOBRA * (p - y_test.values), C_FALTA * (y_test.values - p)).sum()
    filas.append({"pronóstico": nombre, "MAE": np.abs(e).mean(), "sMAPE": np.mean(2 * np.abs(e) / (np.abs(y_test.values) + np.abs(p))), "costo de inventario": costo})
tabla = pd.DataFrame(filas).set_index("pronóstico")
tabla["costo relativo al ingenuo"] = tabla["costo de inventario"] / tabla.loc["ingenuo", "costo de inventario"]
print(tabla.round(3).to_string())

assert len(y_train) == 1043 and len(y_test) == 56 and y_test.index[0].dayofweek == 0     # el horizonte empieza en lunes: 8 semanas completas
assert tabla.MAE["regresión"] < tabla.MAE["estacional ingenuo"] < min(tabla.MAE["promedio móvil 7 d"], tabla.MAE["ingenuo"])   # lo que ignora la semana pierde con lo que la usa
assert tabla.loc["regresión", "costo relativo al ingenuo"] < 0.25

fig, ax = plt.subplots(figsize=(10, 4.5))
ax.plot(demanda.index[-120:], demanda.values[-120:], color="#9ca3af", lw=1, label="demanda real")
ax.plot(y_test.index, y_test.values, color="black", lw=1.5, label="prueba (no vista)")
for (nombre, p), c in zip(pronosticos.items(), [PALETA[1], PALETA[3], PALETA[2], PALETA[0]]):
    ax.plot(y_test.index, p, color=c, lw=2 if nombre == "regresión" else 1.2, label=nombre)
ax.axvline(y_test.index[0], color="black", ls=":"); ax.legend(ncol=3, fontsize=8); ax.set_ylabel("unidades por día")
ax.set_title("Pronosticar las últimas 8 semanas: la estructura (semana, tendencia) es la que aporta", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lectura honesta.** El pronóstico ingenuo tiene un MAE ≈ 53 unidades y el promedio móvil de 7 días, ≈ 54: **el promedio móvil no mejora al ingenuo** (ambos son planos e **ignoran la semana**; que el ingenuo no sea peor depende de que el entrenamiento terminó en domingo, un día alto). El estacional ingenuo (que **aprovecha el patrón semanal**) baja a ≈ 30 y la regresión a ≈ 13.5, **casi en el límite** de este mundo (con ruido de desviación 18, el MAE mínimo alcanzable es ≈ 14). En **costo de inventario**, la regresión gasta ≈ **21 %** de lo que gasta el ingenuo y el estacional ingenuo ≈ 71 %. Pero cuidado con leerlo como «el aprendizaje automático gana»: lo que más aportó fue **explotar la estructura** (pasar de ingenuo a estacional ingenuo cuesta **una línea de código** y recorta casi la mitad del error) y la regresión la gana porque **conoce la forma exacta con que se generaron los datos** (tendencia lineal, un ciclo anual). Con datos reales hay **promociones, festivos, clima y cambios de régimen** que ninguna de estas reglas ve. Y los parámetros de costo (1 y 3) son **ilustrativos**: con otros, la decisión óptima cambia (véase la Autoevaluación).

## 6. Riesgos y Ética: el Otro Lado del Valor ⚖️

Cada demostración anterior tiene su cara oscura: la segmentación puede **discriminar** (¿quién queda en el grupo «de poco valor» y qué trato recibe?), la detección de anomalías puede **bloquear injustamente** a quien se sale del patrón, y el pronóstico puede **reforzar** desigualdades históricas. Un mapa de riesgos del Big Data:

| Riesgo | Qué ocurre | Mitigación inicial |
|---|---|---|
| **Re-identificación** | Cruzar datos «anónimos» con otras fuentes revela personas (ubicación, fecha de nacimiento, ocupación…) | **Minimizar** datos, generalizar, medir **k-anonimato**, controles de acceso |
| **Sesgo y discriminación** | Los datos históricos reflejan desigualdades; un modelo las **automatiza** a gran escala | Auditar por subgrupos, revisión humana de decisiones con impacto |
| **Seguridad** | Más datos, más superficie de ataque; una filtración daña a las personas | Cifrado, control de acceso, auditoría (**Módulo 04**) |
| **Opacidad y rendición de cuentas** | «Lo decidió el algoritmo» sin explicación ni apelación | Explicabilidad, trazabilidad, responsable humano identificado |
| **Calidad y uso fuera de contexto** | Decisiones sobre datos erróneos, desactualizados o usados para otro fin | Gobernanza, catálogo de datos, principio de **finalidad** |
| **Asimetría de poder** | Quien concentra los datos concentra el poder de decidir | Datos abiertos, derechos de acceso y portabilidad, regulación |

> 🇨🇴 **Marco colombiano (conceptual):** la protección de datos personales en Colombia se rige por la **Ley 1581 de 2012** (*Habeas Data*) y sus normas reglamentarias; la autoridad de control es la **Superintendencia de Industria y Comercio** *(citado de memoria; verifica en el texto oficial vigente antes de aplicarlo)*. Los **datos sensibles** (por ejemplo, de salud) reciben un tratamiento más restrictivo. El detalle de los principios y de la retención está en el [cuaderno 03](03_Ciclo_de_Vida_de_los_Datos.ipynb), y la discusión de fondo, en [Consideraciones éticas de la IA](../../Introduccion%20a%20la%20Inteligencia%20Artificial/03%20-%20Etica%20y%20el%20Nuevo%20Paradigma%20de%20la%20IA/00_Consideraciones_Eticas_de_la_IA.ipynb) y en [Ética, privacidad y gobernanza en minería de datos](../../Data%20Mining/00%20-%20Introduccion%20al%20Data%20Mining/04_Etica_Privacidad_y_Gobernanza_en_Mineria_de_Datos.ipynb).

**Una demostración concreta: «sin nombre» no es «anónimo».** Un registro es **único** si nadie más comparte sus **cuasi-identificadores** (atributos que por separado parecen inocuos, como municipio, edad y sexo). El **$k$-anonimato** exige que **cada combinación aparezca al menos $k$ veces**. Medimos cuántos de 5 000 beneficiarios **sintéticos** de un programa son únicos y cómo baja esa cifra **generalizando** (edad en bandas de 10 años y municipios en 3 «zonas» **arbitrarias** del ejemplo).

````python
# ============================================================
# 6. k-anonimato: fraccion de registros unicos segun los cuasi-identificadores y la generalizacion (5 000 personas SINTETICAS)
# ============================================================
MUNICIPIOS = ["Tunja", "Duitama", "Sogamoso", "Chiquinquirá", "Paipa", "Villa de Leyva", "Moniquirá", "Samacá", "Ramiriquí", "Garagoa"]
ZONA = dict(zip(MUNICIPIOS, [1, 2, 2, 3, 2, 3, 3, 1, 1, 1]))                    # agrupacion ARBITRARIA, solo para el ejemplo


def generar_beneficiarios(n=5000, semilla=7):
    rng = np.random.default_rng(semilla)
    p = np.array([.22, .16, .15, .10, .08, .06, .07, .06, .05, .05])
    d = pd.DataFrame({"municipio": rng.choice(MUNICIPIOS, n, p=p / p.sum()), "edad": rng.integers(18, 91, n), "sexo": rng.choice(["F", "M"], n),
                      "ocupacion": rng.choice(["agricultor", "comerciante", "docente", "conductor", "estudiante", "salud", "otro"], n)})
    d["banda_edad"] = (d.edad // 10) * 10
    d["zona"] = d.municipio.map(ZONA)
    return d


def tamanos_de_grupo(d, cols):
    """Para cada registro, cuantos registros comparten SUS cuasi-identificadores (el k de ese registro)."""
    return d.groupby(cols)[cols[0]].transform("size")


benef = generar_beneficiarios()
esquemas = {
    "municipio + edad + sexo": ["municipio", "edad", "sexo"],
    "municipio + edad + sexo + ocupación": ["municipio", "edad", "sexo", "ocupacion"],
    "zona + banda de edad + sexo (generalizado)": ["zona", "banda_edad", "sexo"],
    "zona + banda de edad + sexo + ocupación (generalizado)": ["zona", "banda_edad", "sexo", "ocupacion"],
}
filas = []
for nombre, cols in esquemas.items():
    k = tamanos_de_grupo(benef, cols)
    filas.append({"cuasi-identificadores": nombre, "k mínimo": int(k.min()), "% únicos (k=1)": 100 * (k == 1).mean(), "% en grupos con k<5": 100 * (k < 5).mean()})
tabla_k = pd.DataFrame(filas).set_index("cuasi-identificadores")
print(tabla_k.round(1).to_string())

u = tabla_k["% únicos (k=1)"]
assert u.iloc[1] > u.iloc[0] > u.iloc[2]                                             # mas atributos -> mas unicos; generalizar -> menos
assert tabla_k["k mínimo"].iloc[2] >= 5                                              # con zona + banda + sexo se cumple 5-anonimato
assert u.iloc[1] > 40 and u.iloc[3] < u.iloc[1]
````

---
**Lectura.** Solo con **municipio, edad y sexo**, ≈ **5 %** de las personas ya son **únicas** (y ≈ 44 % están en grupos de menos de 5); al añadir la **ocupación**, **más de la mitad (≈ 56 %) son únicas**: bastaría saber esos cuatro datos de un vecino para ubicar su registro, **aunque no haya nombre ni cédula**. Al **generalizar** (zona y banda de edad) el grupo más pequeño tiene **6 personas** (5-anonimato) y nadie es único; con la ocupación vuelve a haber ≈ 0.4 % de únicos. El **costo** es perder **resolución analítica**: no puedes preguntar por la edad exacta ni por el municipio. Generalizar **reduce el riesgo, no lo elimina** (el *k*-anonimato no protege frente a quien conoce valores sensibles homogéneos dentro de un grupo, ni frente a cruces con otras fuentes). En un proyecto real se aplican **varias capas**: minimización, acceso restringido, evaluación de impacto y gobernanza.

|---|---|
| **Del dato al valor** | **DIKW:** dato → información → conocimiento → decisión; el Big Data pesa sobre todo en los primeros saltos, pero **el valor se crea (o se pierde) al decidir**. En el ejemplo agrícola, 336 lecturas terminaron en una lista de 3 días de alerta |

| **Economía del dato** | No rival, copia casi gratis pero **curarlo cuesta**, complementario, perecedero y con externalidades; lo escaso es el **dato de calidad con contexto**. La cadena de valor se rompe donde el análisis **no llega a quien decide** |
| **Segmentación** | K-Means encontró 3 grupos; **≈ 11 % de los clientes aportan ≈ 56 % del ingreso** (sintético). La **silueta prefirió $k=2$** y la estructura real era 3: una métrica interna no reemplaza al criterio de negocio |
| **Anomalías** | El **contexto del cliente y la hora** sube la precisión del umbral global de ≈ 34 % a ≈ 52 % en el 1 % superior (AP 0.58 frente a 0.32); aun así ≈ 5 de cada 10 alertas son falsas: **decisión humana** |
| **Pronóstico** | Las **líneas base** hacen casi todo: ingenuo → estacional ingenuo reduce el MAE de ≈ 53 a ≈ 30 con una línea de código (el promedio móvil **no** mejoró al ingenuo); la regresión (≈ 13.5) gana porque **conoce la estructura** del mundo sintético. El error se traduce a **costo** |
