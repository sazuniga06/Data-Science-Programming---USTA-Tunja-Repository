## ¿Qué vamos a aprender aquí? 🎈

Este cuaderno es la versión **«para no ingenieros»** del [cuaderno 03 del Módulo 00 (Ciclo de vida de los datos)](../03_Ciclo_de_Vida_de_los_Datos.ipynb). Un dato **hace un viaje**, igual que la **leche de una finca boyacense** hasta tu mesa:

| El viaje de la leche | El viaje del dato |
|---|---|
| **Ordeño** | **Generación:** el dato nace (un sensor, una venta, un formulario) |
| **Recolección en cantinas** | **Ingesta:** llega adonde se va a usar; se revisa que venga bien |
| **Tanque de enfriamiento** | **Almacenamiento:** se guarda de forma segura |
| **Pasteurizar y empacar** | **Procesamiento:** se limpia y se ordena |
| **Venta y consumo** | **Análisis:** se usa para decidir |
| **Fecha de vencimiento** | **Archivado y eliminación:** se guarda lo necesario y se **bota** lo que ya no se debe tener |

Al terminar podrás explicar:
1. Las **seis etapas** del viaje de un dato y el **riesgo** de cada una.
2. Cómo un **recibo** (cuántos datos entran y salen en cada paso) ayuda a **no perder nada sin saber**.
3. Por qué **botar datos a tiempo** y **borrar los de una persona que lo pide** también es parte del trabajo.

## 1. El Mapa del Viaje 🗺️

Primero, un dibujo sencillo de las seis estaciones. Fíjate en que la **fecha de vencimiento** (el final) no es un detalle: es una parte **igual de importante** que el ordeño.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

estaciones = [("Ordeño", "generación", "#f59e0b"), ("Cantinas", "ingesta", "#dc2626"), ("Tanque", "almacenamiento", "#7c3aed"),
              ("Pasteurizar", "procesamiento", "#2563eb"), ("Venta", "análisis", "#16a34a"), ("Vencimiento", "archivo / borrado", "#78350f")]
assert len(estaciones) == 6

fig, ax = plt.subplots(figsize=(11, 2.6))
ax.set_xlim(0, 6); ax.set_ylim(0, 1); ax.axis("off"); ax.grid(False)
for i, (leche, dato, color) in enumerate(estaciones):
    ax.add_patch(FancyBboxPatch((i + 0.08, 0.2), 0.84, 0.6, boxstyle="round,pad=0.02,rounding_size=0.06", fc=color, ec="white", lw=2))
    ax.text(i + 0.5, 0.58, leche, ha="center", va="center", color="white", fontweight="bold", fontsize=9.5)
    ax.text(i + 0.5, 0.36, dato, ha="center", va="center", color="white", fontsize=8)
    if i < 5:
        ax.annotate("", xy=(i + 1.07, 0.5), xytext=(i + 0.93, 0.5), arrowprops=dict(arrowstyle="-|>", lw=2, color="#334155"))
ax.set_title("El viaje de un dato (como el de la leche): seis estaciones", fontweight="bold", fontsize=11)
plt.tight_layout(); plt.show()
````

---
**Lección:** cada estación **puede dañar** lo que viene después: si el **ordeño** se hace sucio, ninguna pasteurización lo arregla del todo; si **se pierde** leche en las cantinas, nadie lo nota hasta que falta; y si **nadie mira la fecha de vencimiento**, se vende leche mala. Con los datos es igual. Por eso en cada estación hay que **revisar y anotar** lo que entra y lo que sale.

## 2. El Recibo de Cada Estación 🧾

Una **finca** ordeña a sus vacas varios días y anota los litros. Los datos de abajo son **inventados** y traen los problemas de la vida real: un renglón **vacío**, un número **imposible** (−3 litros y 250 litros: ninguna vaca da tanto) y un renglón **repetido** por error.

Vamos a pasar los datos por **seis funciones**, una por estación, y a imprimir un **recibo** en cada una: *¿cuántos renglones entraron y cuántos salieron?* Lo que se **bota** no desaparece en silencio: queda **anotado con su motivo**.

````python
def ordenar():
    """1. Generacion: la finca anota los litros de cada ordeño (con los problemas de la vida real)."""
    return [
        {"dia": 1, "vaca": "Margarita", "litros": 18.5}, {"dia": 1, "vaca": "Estrella", "litros": 21.0}, {"dia": 1, "vaca": "Luna", "litros": 19.2},
        {"dia": 1, "vaca": "Margarita", "litros": 18.5},                                    # repetido por error
        {"dia": 2, "vaca": "Margarita", "litros": 17.9}, {"dia": 2, "vaca": "Estrella", "litros": 250.0},      # 250 litros: imposible
        {"dia": 2, "vaca": "Luna", "litros": None},                                           # no se anoto
        {"dia": 2, "vaca": "Pinta", "litros": -3.0},                                          # negativo: imposible
        {"dia": 3, "vaca": "Margarita", "litros": 18.8}, {"dia": 3, "vaca": "Estrella", "litros": 20.4}, {"dia": 3, "vaca": "Luna", "litros": 19.9}, {"dia": 3, "vaca": "Pinta", "litros": 16.7},
    ]


def recoger(registros):
    """2. Ingesta: llegan a la oficina; revisamos que cada renglon tenga lo basico (dia, vaca, litros)."""
    buenos = [r for r in registros if {"dia", "vaca", "litros"} <= set(r)]
    return buenos


def guardar(registros):
    """3. Almacenamiento: guardamos una COPIA sin tocar (el original), por si hay que revisar."""
    return [dict(r) for r in registros]


def limpiar(registros):
    """4. Procesamiento: botamos repetidos, vacios e imposibles, pero ANOTANDO el motivo."""
    vistos, limpios, botados = set(), [], []
    for r in registros:
        clave = (r["dia"], r["vaca"])
        if clave in vistos:
            botados.append((r, "repetido"))
        elif r["litros"] is None:
            botados.append((r, "vacio"))
        elif not (0 < r["litros"] <= 60):                                                   # sentido comun: una vaca no da mas de 60 litros al dia
            botados.append((r, "imposible"))
        else:
            vistos.add(clave); limpios.append(r)
    return limpios, botados


def analizar(limpios):
    """5. Analisis: promedio de litros por vaca."""
    return pd.DataFrame(limpios).groupby("vaca").litros.mean().round(1)


recibo = []
crudos = ordenar();            recibo.append(("1. Ordeño (generación)", 0, len(crudos)))
recogidos = recoger(crudos);   recibo.append(("2. Cantinas (ingesta)", len(crudos), len(recogidos)))
guardados = guardar(recogidos); recibo.append(("3. Tanque (almacenamiento)", len(recogidos), len(guardados)))
limpios, botados = limpiar(guardados); recibo.append(("4. Pasteurizar (procesamiento)", len(guardados), len(limpios)))
promedios = analizar(limpios); recibo.append(("5. Venta (análisis)", len(limpios), len(promedios)))

print(f"{'Estación':32s}{'entran':>8s}{'salen':>8s}")
for nombre, e, s in recibo:
    print(f"{nombre:32s}{e:>8d}{s:>8d}")
print("\nLo que se botó (con motivo):", [(b["vaca"], b["dia"], motivo) for b, motivo in botados])
print("Promedio de litros por vaca:", promedios.to_dict())

assert len(guardados) == len(limpios) + len(botados)                                          # NADA desaparece sin explicacion: 12 = 8 + 4
assert sorted(m for _, m in botados) == ["imposible", "imposible", "repetido", "vacio"]
assert promedios["Estrella"] == 20.7 and len(promedios) == 4
````

---
**Lección:** entraron **12 renglones** y quedaron **8 limpios**; los otros **4 se botaron con su motivo** (1 repetido, 1 vacío y 2 imposibles). Lo importante es la **cuenta que cierra**: *12 = 8 + 4*; si no cerrara, alguien se estaría **comiendo datos sin avisar**. Dos detalles de honestidad: (1) la regla «una vaca no da más de 60 litros» es de **sentido común**, no una verdad absoluta: la debe definir **quien conoce las vacas**; (2) **limpiar no es lo mismo que acertar**: un litraje de 17.0 que en realidad fue 27.0 pasa todas las revisiones. Por eso se guarda también la **copia original** (el tanque): para poder **revisar y volver a limpiar** con mejores reglas.

## 3. La Fecha de Vencimiento y el «Borren mis Datos» 🗑️

La última estación es la que **más se olvida**. Guardar datos «por si acaso» **no es gratis**: ocupan espacio, hay que protegerlos y, si se filtran, **se hace daño a personas**. Por eso se define una **política**: *¿cuánto tiempo guardo cada dato y qué hago cuando vence?*

Además, en Colombia la **Ley 1581 de 2012** (protección de datos personales, o *Habeas Data*) reconoce que las personas pueden **pedir que se corrijan o se supriman sus datos** en ciertos casos (aquí solo lo vemos como idea general; para aplicarlo en serio se consulta el texto oficial y a un especialista). Simulamos las dos cosas con una lista de **clientes ficticios** de la finca: (1) **botar lo que lleva más de 30 días** guardado y (2) **borrar todo lo de una persona** que lo pide.

````python
HOY = 100                                                      # 'hoy' es el dia 100 (numeros inventados, para no usar fechas)
PLAZO = 30                                                     # guardamos los datos por 30 dias (plazo ILUSTRATIVO, lo define la organizacion)
registros = pd.DataFrame({
    "cliente": ["Ana", "Beto", "Ana", "Carla", "Beto", "Ana", "Diego", "Carla"],
    "compra": ["leche", "queso", "queso", "leche", "leche", "arequipe", "queso", "leche"],
    "dia_registro": [40, 55, 68, 80, 90, 95, 98, 99],          # cuando se anoto cada compra
})

# (1) Vencimiento: lo que lleva mas de PLAZO dias guardado sale de la tabla 'activa'
registros["antiguedad"] = HOY - registros.dia_registro
vencidos = registros[registros.antiguedad > PLAZO]
activos = registros[registros.antiguedad <= PLAZO]
print(f"Vencidos ({len(vencidos)}): dias {vencidos.dia_registro.tolist()} | Activos ({len(activos)}): dias {activos.dia_registro.tolist()}")
assert len(vencidos) == 3 and len(activos) == 5 and len(vencidos) + len(activos) == len(registros)

# (2) Derecho a que borren mis datos: Ana lo pide
def borrar_a(persona, tabla):
    return tabla[tabla.cliente != persona]


activos_2 = borrar_a("Ana", activos)
print(f"Despues de borrar a Ana: quedan {len(activos_2)} registros activos, y clientes = {sorted(activos_2.cliente.unique())}")
assert "Ana" not in activos_2.cliente.values and len(activos_2) == len(activos) - (activos.cliente == "Ana").sum()
assert set(activos_2.cliente) == {"Beto", "Carla", "Diego"}

# Pero Ana tambien estaba en los VENCIDOS (y en copias de seguridad...): borrar de UN solo lugar no basta
lugares = {"tabla activa (ya borrada)": activos_2, "tabla de vencidos (archivo)": vencidos}
donde_esta_ana = {nombre: int((t.cliente == "Ana").sum()) for nombre, t in lugares.items()}
print("¿Dónde más está Ana?", donde_esta_ana)
assert donde_esta_ana == {"tabla activa (ya borrada)": 0, "tabla de vencidos (archivo)": 2}
````

---
**Lección:** la **fecha de vencimiento** sacó **3 de las 8 compras** de la tabla activa (las que llevaban más de 30 días), y el **«borren mis datos»** de Ana quitó su compra de la tabla activa. Pero fíjate en lo difícil: **Ana también estaba en el archivo de vencidos** (2 compras más) y, en la vida real, estaría en **copias de seguridad, hojas de cálculo que alguien exportó y reportes viejos**. Por eso, para **borrar de verdad**, hay que saber **dónde está cada dato** (un inventario o «catálogo»). Y un recordatorio sobre qué significa «anonimizar»: **tachar el nombre no basta** si quedan datos que permiten reconocer a la persona (barrio, edad, profesión...).

## Resumen en una frase 🎯

> **Un dato nace, viaja, se guarda, se limpia, se usa y debe morir a tiempo: en cada estación se anota lo que entra y lo que sale para no perder nada sin saber, y al final se botan o se anonimizan los datos que ya no se necesitan o que alguien pide borrar.**

### Lo que aprendiste hoy:
- ✅ **Seis estaciones:** generación, ingesta, almacenamiento, procesamiento, análisis y archivado/eliminación (como ordeño → cantinas → tanque → pasteurizar → venta → vencimiento).
- ✅ **El recibo:** entraron 12 renglones, quedaron 8 y se botaron 4 **con su motivo** (12 = 8 + 4); los datos **no desaparecen en silencio**.
- ✅ **Guardar el original:** la copia sin tocar permite **volver a limpiar** con mejores reglas.
- ✅ **Vencimiento y borrado:** botar lo viejo reduce riesgos; **borrar a una persona** exige encontrarla **en todos los lugares** (tabla activa, archivo, copias).
