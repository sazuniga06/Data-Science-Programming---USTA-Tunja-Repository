## 1. La Tendera que Conoce a sus Clientes 🛍️

Doña Marta tiene una tienda con **20 clientes** (inventados). Anota cuántas veces compra cada uno **al mes** y cuánto gasta **en promedio** (en miles de pesos). Sin ningún programa raro, hace tres grupos con una regla simple:

- **Fieles:** compran **8 veces o más** al mes.
- **Regulares:** entre **3 y 7** veces.
- **Ocasionales:** **1 o 2** veces.

Y se pregunta: ¿qué **parte de los clientes** es cada grupo, y qué **parte del dinero** deja?

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

clientes = pd.DataFrame({
    "cliente": [f"C{i:02d}" for i in range(1, 21)],
    "compras_al_mes": [1, 2, 1, 3, 1, 2, 9, 1, 4, 2, 10, 1, 3, 2, 1, 12, 2, 1, 5, 1],
    "gasto_medio": [20, 35, 15, 40, 25, 30, 60, 18, 45, 28, 70, 22, 38, 26, 19, 80, 31, 17, 50, 21],      # miles de pesos por compra
})
clientes["dinero_al_mes"] = clientes.compras_al_mes * clientes.gasto_medio


def grupo(c):
    return "fiel" if c >= 8 else "regular" if c >= 3 else "ocasional"


clientes["grupo"] = clientes.compras_al_mes.map(grupo)
res = clientes.groupby("grupo").agg(clientes=("cliente", "size"), dinero=("dinero_al_mes", "sum")).loc[["fiel", "regular", "ocasional"]]
res["% de clientes"] = (100 * res.clientes / res.clientes.sum()).round(0)
res["% del dinero"] = (100 * res.dinero / res.dinero.sum()).round(0)
print(res.to_string())
assert res.loc["fiel", "% de clientes"] == 15
assert res.loc["fiel", "% del dinero"] > 2 * res.loc["fiel", "% de clientes"]                 # pocos clientes, mucho dinero

fig, ax = plt.subplots(figsize=(7, 3.8))
x = np.arange(3)
ax.bar(x - 0.2, res["% de clientes"], 0.4, color="#2563eb", label="% de clientes"); ax.bar(x + 0.2, res["% del dinero"], 0.4, color="#dc2626", label="% del dinero")
ax.set_xticks(x); ax.set_xticklabels(res.index); ax.legend(); ax.set_title("Los clientes fieles son pocos, pero dejan mucho", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lección:** los **3 clientes fieles** (el 15 %) dejan **≈ 66 % del dinero** (dos de cada tres pesos), mientras que los 13 ocasionales (el 65 % de la lista) dejan solo **≈ 14 %**. Eso **cambia lo que Marta hace**: cuida a los fieles (un detalle de agradecimiento) y piensa cómo **animar a los ocasionales** a volver. Esto es **segmentar clientes**; con miles de clientes y muchas variables, una computadora encuentra los grupos **sola** (el cuaderno estándar usa *K-Means*). Una advertencia: los grupos **no son «verdad»**: son una **forma útil de mirar**; con otra regla saldrían otros.

## 2. La Alarma de la Tarjeta 🚨

Cuando pagas con tarjeta, el banco mira si esa compra **se parece a lo que tú sueles hacer**. Una idea muy simple: **¿esta compra es mucho más grande que lo normal para esta persona?** Usamos «lo normal» = **la mediana** (el valor del medio de la lista, que no se deja engañar por una compra gigante) y alertamos si una compra es **más de 3 veces** lo normal.

Las compras de Ana (en miles de pesos) incluyen dos que **se salen**: una de 900 que **no fue de ella** (fraude) y una de 120 en diciembre (**regalos de Navidad**: legítima).

````python
compras_ana = pd.Series([32, 45, 28, 51, 36, 22, 40, 47, 30, 38, 55, 26, 42, 35, 29, 44, 33, 41, 37, 900, 27, 48, 31, 120], name="valor_miles")
normal = compras_ana.median()
umbral = 3 * normal
alertas = compras_ana[compras_ana > umbral]
print(f"Lo normal para Ana (mediana): {normal:.1f} mil pesos -> alerta si una compra pasa de {umbral:.1f} mil")
print("Compras que disparan la alarma:", alertas.to_dict())
assert sorted(alertas.values) == [120, 900]

verdadero_fraude = {19}                                                           # la posicion 19 (el 900) es el fraude real; la de 120 es legitima
aciertos = len(set(alertas.index) & verdadero_fraude)
print(f"De {len(alertas)} alertas, {aciertos} fue un fraude de verdad y {len(alertas) - aciertos} fue una falsa alarma (la compra de Navidad).")
assert aciertos == 1 and len(alertas) - aciertos == 1

fig, ax = plt.subplots(figsize=(8, 3.6))
colores = ["#dc2626" if v > umbral else "#2563eb" for v in compras_ana]
ax.bar(range(len(compras_ana)), compras_ana.values, color=colores); ax.axhline(umbral, color="black", ls="--", lw=1)
ax.text(0, umbral + 25, "umbral: 3 veces lo normal", fontsize=9); ax.set_xlabel("compra n.º"); ax.set_ylabel("miles de pesos")
ax.set_title("La alarma detecta lo raro, no lo malo: a veces se equivoca", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lección:** la regla **atrapó el fraude de 900**, pero también **hizo sonar la alarma** con la compra legítima de 120 (los regalos de diciembre): una **falsa alarma**. Ninguna alarma es perfecta: si la haces **más sensible**, atrapas más fraudes **y** molestas a más clientes honrados; si la haces **menos sensible**, dejas pasar fraudes. Por eso las alertas **no son sentencias**: alguien (una persona o una llamada al cliente) **revisa** antes de bloquear nada. Los bancos reales usan más información (la hora, el lugar, el tipo de comercio), pero la **idea** es la misma: *comparar con lo que es normal para esa persona*.

## 3. ¿Cuánta Papa Pido? 🥔

Don Álvaro vende papa y debe decidir **cuántos kilos pedir** para la semana. Tiene lo que vendió en **tres semanas** (datos inventados). Quiere adivinar la **semana 3** usando **solo** las semanas 1 y 2, y probar dos formas:

1. **«Todos los días igual»:** el promedio de todo lo que vendió en las semanas 1 y 2.
2. **«Como el mismo día de la semana pasada»:** para el sábado, lo que vendió el sábado de la semana 2; para el lunes, lo del lunes...

Mide **cuánto se equivoca en promedio** (en kilos) cada forma.

````python
dias = ["lun", "mar", "mié", "jue", "vie", "sáb", "dom"]
ventas = pd.DataFrame({"semana 1": [120, 110, 125, 130, 180, 240, 160],
                       "semana 2": [125, 105, 130, 128, 190, 250, 150],
                       "semana 3": [118, 112, 128, 135, 185, 245, 158]}, index=dias)       # kilos vendidos (inventados)

todos_igual = pd.Series(ventas[["semana 1", "semana 2"]].values.mean(), index=dias)         # el mismo numero cada dia
mismo_dia = ventas["semana 2"]                                                              # lo que paso ese mismo dia la semana pasada
error = lambda pred: (pred - ventas["semana 3"]).abs().mean()
mae_todos, mae_mismo = error(todos_igual), error(mismo_dia)
print(f"Error promedio: 'todos los dias igual' = {mae_todos:.1f} kg | 'mismo dia de la semana pasada' = {mae_mismo:.1f} kg")
assert mae_mismo < mae_todos / 4

fig, ax = plt.subplots(figsize=(8, 3.8))
ax.plot(dias, ventas["semana 3"], "o-", color="black", lw=2, label="lo que se vendió de verdad")
ax.plot(dias, mismo_dia, "s--", color="#16a34a", label="como el mismo día de la semana pasada"); ax.plot(dias, todos_igual, "^--", color="#dc2626", label="todos los días igual")
ax.legend(fontsize=8); ax.set_ylabel("kilos"); ax.set_title("El patrón de la semana (el sábado se vende más) vale oro", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lección:** «todos los días igual» se equivoca **≈ 35 kilos al día en promedio** porque ignora que **el sábado se vende el doble que el martes**; «como el mismo día de la semana pasada» se equivoca **≈ 6 kilos**. Es decir, **lo que más ayudó fue aprovechar un patrón** (la semana tiene ritmo), no un programa sofisticado: por eso en el cuaderno estándar el pronóstico «ingenuo por día de la semana» es la **línea base** que cualquier modelo debe superar.

## Resumen en una frase 🎯

> **Los datos valen cuando se convierten en decisiones: agrupar clientes, detectar lo raro y anticipar la demanda son formas de pasar de «tomates» a «salsa»... siempre cuidando a las personas detrás de los datos.**

### Lo que aprendiste hoy:
- ✅ **De los datos al valor:** datos → información → conocimiento → **decisión**; lo importante es la decisión.
- ✅ **Segmentar:** 3 clientes fieles (15 %) dejaron **≈ 66 %** del dinero (dos de cada tres pesos); eso cambia cómo atiende la tendera.
- ✅ **Detectar lo raro:** la regla «más de 3 veces lo normal» atrapó el fraude, pero **también una falsa alarma**; se necesita una persona que revise.
- ✅ **Pronosticar:** «como el mismo día de la semana pasada» redujo el error de **≈ 35 kg a ≈ 6 kg**.

### ⚠️ Un cuidado:
Los mismos datos que ayudan pueden **hacer daño**. Una lista de clientes con **nombre no** es lo único que identifica a una persona: **tu municipio, tu edad y tu sexo** juntos pueden bastar para encontrarte en una tabla pequeña. Por eso se guarda **solo lo necesario**, se pide permiso y se protege (en Colombia, la **Ley 1581 de 2012** de protección de datos personales; en el cuaderno estándar y en el 03 de este módulo lo vemos con más cuidado).
