## 1. Volumen: ¿Cuántas Fotos Caben en el Disco? 📸

Un disco de computador se mide en **GB** (gigabytes) y **TB** (terabytes; 1 TB son 1 000 GB). Cada cosa que guardas ocupa un espacio: una foto del celular, una canción, una película. Los tamaños de abajo son **aproximados** (dependen de la calidad), pero sirven para tener **una idea de orden de magnitud**.

Hacemos dos cuentas: (1) **cuántas cosas caben** en un disco y (2) qué pasa si **cada mes guardas el doble** que el mes anterior.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Tamanos APROXIMADOS (en megabytes; 1 GB = 1000 MB para simplificar)
FOTO_MB = 4            # una foto de celular, ~4 MB
CANCION_MB = 5         # una cancion, ~5 MB
PELICULA_MB = 2000     # una pelicula en buena calidad, ~2 GB
discos = {"disco de 500 GB": 500_000, "disco de 1 TB": 1_000_000}      # capacidad en MB

filas = [{"disco": nombre, "fotos": mb // FOTO_MB, "canciones": mb // CANCION_MB, "películas": mb // PELICULA_MB} for nombre, mb in discos.items()]
tabla = pd.DataFrame(filas).set_index("disco")
print(tabla.to_string())
assert tabla.loc["disco de 1 TB", "fotos"] == 250_000

# Si cada mes guardas el DOBLE que el mes anterior (empezando con 1 GB), ¿en que mes se llena un disco de 1 TB?
gb, mes, historial = 1, 0, [1]
while gb < 1000:
    gb, mes = gb * 2, mes + 1
    historial.append(gb)
print(f"\nDuplicando cada mes, un disco de 1 TB se llena en el mes {mes} (pasaste de 1 GB a {gb} GB).")
assert mes == 10 and historial[5] == 32                                        # a los 5 meses solo tienes 32 GB: se ve inofensivo...

fig, ax = plt.subplots(figsize=(7.5, 4))
ax.bar(range(len(historial)), historial, color=["#16a34a" if h < 1000 else "#dc2626" for h in historial])
ax.axhline(1000, color="black", ls="--", lw=1); ax.set_ylim(0, 1200); ax.text(0.2, 1040, "disco de 1 TB (1 000 GB)", fontsize=9)
ax.set_xlabel("mes"); ax.set_ylabel("GB guardados"); ax.set_title("Duplicar cada mes parece tranquilo... hasta que de pronto no cabe", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lección:** en un disco de **1 TB caben unas 250 000 fotos** (con 4 MB cada una) o **500 películas** (con 2 GB): suena a mucho. Pero **duplicar cada mes** lleva de **1 GB a un disco lleno en solo 10 meses**; observa que a los 5 meses solo tienes 32 GB y *parece* inofensivo. Esa es la trampa del **crecimiento exponencial**: la curva es tranquila... hasta que deja de serlo. Con **muchos** discos y **muchas** empresas, hablamos de volúmenes que ya no caben en un solo computador: eso es el **volumen** del Big Data.

## 2. Velocidad: la Fila de la Caja 🧾

No solo importa **cuánto** hay, sino **qué tan rápido llega**. Piensa en una caja de supermercado que atiende a **1 cliente por minuto**:

- Normalmente llegan **menos** de uno por minuto (la fila no existe).
- A la hora del almuerzo llegan **más** de uno por minuto durante media hora (**la fila crece**).
- Luego baja otra vez, pero la fila **tarda en vaciarse**.

Simulamos 90 minutos con **una caja** y con **dos cajas**.

````python
# Clientes que llegan en cada minuto: 20 min tranquilos (1 cada 2 min), 30 min de almuerzo (3 cada 2 min) y 40 min tranquilos otra vez
llegan = [1, 0] * 10 + [2, 1] * 15 + [0, 1] * 20
assert len(llegan) == 90


def fila_en_cada_minuto(llegadas, atiende_por_minuto):
    """Cuantas personas esperan al final de cada minuto (nadie hace fila si hay caja libre)."""
    espera, historia = 0, []
    for nuevos in llegadas:
        espera = max(0, espera + nuevos - atiende_por_minuto)
        historia.append(espera)
    return historia


una_caja = fila_en_cada_minuto(llegan, 1)
dos_cajas = fila_en_cada_minuto(llegan, 2)
minuto_vacia = next(m for m in range(50, 90) if una_caja[m] == 0) + 1                        # primer minuto, despues del almuerzo, sin fila
print(f"Con UNA caja, la fila llega a {max(una_caja)} personas y recien se vacia en el minuto {minuto_vacia} (el almuerzo termino en el 50).")
print(f"Con DOS cajas, la fila maxima es {max(dos_cajas)} personas.")
assert max(una_caja) == 15 and max(dos_cajas) == 0 and minuto_vacia == 79

fig, ax = plt.subplots(figsize=(8, 3.8))
ax.plot(una_caja, color="#dc2626", lw=2, label="una caja"); ax.plot(dos_cajas, color="#16a34a", lw=2, label="dos cajas")
ax.axvspan(20, 50, color="#f59e0b", alpha=0.2); ax.text(21, 13, "hora del almuerzo", fontsize=9)
ax.set_xlabel("minuto"); ax.set_ylabel("personas esperando"); ax.legend(); ax.set_title("La fila se acumula cuando llegan más de los que puedes atender", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lección:** con **una caja**, durante el almuerzo la fila sube hasta **15 personas** y **tarda ≈ 30 minutos en vaciarse** aun cuando el almuerzo ya pasó; con **dos cajas** nunca hay fila. Con los datos es igual: si llegan **más rápido de lo que los procesas**, la cola de datos **crece**, y la información llega **tarde** (cuando ya no sirve para decidir). Un sistema que debe responder en el acto (por ejemplo, avisar que una compra con tarjeta es sospechosa) necesita **capacidad de sobra** para los picos.

## 3. Muchas Manos: las Cocineras y el Único Horno 👩‍🍳

Hay que preparar **100 empanadas**. Amasar y armar cada una toma **1 minuto** (y eso **sí se puede repartir** entre cocineras). Pero **hornear** toma **20 minutos** y la casa tiene **un solo horno**: eso **no se puede repartir**. Si contratas más cocineras, ¿se demora cada vez menos? Calculamos el tiempo total con 1, 2, 4, 8... cocineras.

````python
EMPANADAS = 100
MIN_ARMAR = 1          # por empanada: se puede repartir entre cocineras
MIN_HORNO = 20         # el horno es UNO: esta parte no se reparte


def minutos_totales(cocineras):
    return EMPANADAS * MIN_ARMAR / cocineras + MIN_HORNO


cocineras = [1, 2, 4, 8, 16, 32, 100]
tiempos = [minutos_totales(c) for c in cocineras]
tabla = pd.DataFrame({"cocineras": cocineras, "minutos": tiempos}).set_index("cocineras")
tabla["veces más rápido"] = tiempos[0] / tabla["minutos"]
print(tabla.round(1).to_string())
assert minutos_totales(1) == 120 and minutos_totales(100) >= MIN_HORNO                  # nunca baja de lo que tarda el horno
assert minutos_totales(1) - minutos_totales(2) > 10 * (minutos_totales(16) - minutos_totales(32))   # la 2.a cocinera ayuda MUCHO mas que la 32.a

fig, ax = plt.subplots(figsize=(7.5, 4))
ax.plot(range(len(cocineras)), tiempos, "o-", color="#2563eb", lw=2)
ax.axhline(MIN_HORNO, color="#dc2626", ls="--"); ax.text(0.1, MIN_HORNO + 3, "piso: el horno único (20 min)", fontsize=9, color="#dc2626")
ax.set_xticks(range(len(cocineras))); ax.set_xticklabels(cocineras); ax.set_xlabel("cocineras"); ax.set_ylabel("minutos en total")
ax.set_title("Más cocineras ayudan, pero con rendimientos cada vez menores", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

---
**Lección:** pasar de **1 a 2 cocineras** ahorra **50 minutos**; pasar de **16 a 32** apenas ahorra **≈ 3 minutos**; y con **100 cocineras** el total sigue siendo **más de 20 minutos**: **el horno manda**. Las computadoras en grupo (un *clúster*) funcionan igual: repartir el trabajo entre muchas máquinas **acelera**, pero **la parte que no se puede repartir** pone un **techo**, y cada máquina nueva **ayuda menos**. Por eso un buen ingeniero **no compra máquinas sin medir**: primero busca **cuál es «el horno»** de su problema. *(Los ingenieros llaman a esto «ley de Amdahl»; el cuaderno estándar la mide con datos reales.)*

## Resumen en una frase 🎯

> **El Big Data aparece cuando los datos son tantos, llegan tan rápido o son tan variados que ya no caben ni se atienden a tiempo en un solo computador; repartir el trabajo ayuda, pero con límites, y lo que de verdad importa es el valor que se saca de ellos.**

### Lo que aprendiste hoy:
- ✅ **Volumen:** 1 TB guarda unas 250 000 fotos; **duplicar cada mes** llena un disco de 1 TB en **10 meses**, aunque a mitad de camino parezca poco.
- ✅ **Velocidad:** si llegan **más datos de los que atiendes**, **la fila crece** (hasta 15 personas con una caja) y la respuesta llega tarde.
- ✅ **Muchas manos:** 1 → 2 cocineras ahorra 50 minutos, 16 → 32 apenas 3; **el horno único** pone el techo.
- ✅ **Variedad, veracidad y valor:** distintos formatos, datos en los que hay que poder confiar y, sobre todo, **una decisión que mejora**.
