## 1. Tres Familias de Datos: un Espectro 🌈

La diferencia entre las familias es **cuánta estructura se conoce y se impone de antemano**. Por **estructura** entendemos el **esquema**: los nombres de los campos, sus **tipos**, sus **relaciones** y las **restricciones** (qué valores son válidos).

| | **Estructurados** | **Semiestructurados** | **No estructurados** |
|---|---|---|---|
| **Esquema** | **Fijo y declarado** antes de guardar | **Autodescriptivo** (cada registro trae sus nombres de campo) y **flexible** | Ninguno explícito: el significado **hay que inferirlo** |
| **Ejemplos** | Tablas de una base de datos relacional, hojas de cálculo bien formadas, CSV regulares | **JSON**, **XML**, YAML, eventos de una aplicación, *logs* con patrón | Texto libre, imágenes, audio, video, documentos escaneados |
| **Almacenamiento típico** | Bases de datos **relacionales** (SQL), *data warehouse* | Bases **documentales** y de clave-valor (NoSQL), archivos en un *data lake* | Almacenamiento de **objetos** y *data lake*; índices de búsqueda |
| **Consulta** | **SQL**, declarativa y optimizable | Rutas (`a.b.c`), lenguajes de documentos; a menudo se **aplana** a tabla | Requiere **extraer características** (regex, NLP, visión) antes de analizar |
| **Fortaleza** | Integridad, rendimiento, herramientas maduras | Flexibilidad: los campos pueden **cambiar** sin migrar | Es **lo más rico**: contiene lo que las personas dicen y ven |
| **Costo** | Rígido: cambiar el esquema cuesta | Esquema variable: más trabajo al leer, nulos por todas partes | Procesamiento caro; el valor **no está en el archivo, sino en la inferencia** |

> 🧠 **Es un espectro, no tres cajas.** Un *log* es texto libre y a la vez tiene un patrón; un CSV puede estar tan sucio que no es estructurado en la práctica; un PDF «con tablas» es no estructurado aunque contenga una. La estructura es una propiedad de **cómo vas a usar el dato**. Suele afirmarse que la **mayor parte** de los datos generados hoy es no estructurada; las cifras citadas varían según la fuente y no existe una medición única, así que **no usaremos ninguna**.

**¿Cuándo se impone la estructura?** Dos estrategias opuestas, que veremos en la sección 5:

- ***Schema-on-write*** (esquema al escribir): el dato **se valida y se transforma al guardarlo**; lo que no cumple se rechaza. Es el modelo de las bases relacionales y de un *data warehouse*.
- ***Schema-on-read*** (esquema al leer): el dato **se guarda tal cual** y **cada lectura aplica su propio esquema**. Es el modelo típico de un *data lake* y de Hadoop/Spark sobre archivos.

## 2. Datos Estructurados: el Mundo Tabular y SQL 🧮

En el modelo **relacional** los datos viven en **tablas** (filas = registros, columnas = atributos con **tipo**), se identifican con **claves primarias** y se relacionan con **claves foráneas**. El esquema permite que la base de datos **garantice** reglas (*restricciones*: `NOT NULL`, `CHECK`, `FOREIGN KEY`) y **optimice** consultas (**índices**, planes de ejecución). Además, las transacciones ofrecen garantías **ACID** (atomicidad, consistencia, aislamiento, durabilidad): o se aplican **todos** los cambios de una operación o ninguno.

Construimos con `sqlite3` (incluida en Python, sin servidor) una mini base de **clientes y pedidos sintéticos**, y verificamos cuatro cosas: (1) un `JOIN` + `GROUP BY` equivale al `merge` + `groupby` de *pandas*; (2) el esquema **rechaza datos inválidos al escribir**; (3) un **índice** cambia el plan de ejecución; (4) una **transacción** fallida **no deja rastro**.

````python
# ============================================================
# 2. Datos estructurados: sqlite3 (esquema, restricciones, JOIN, indice y transaccion) y su equivalente en pandas
# ============================================================
MUNICIPIOS = ["Tunja", "Duitama", "Sogamoso", "Paipa", "Villa de Leyva", "Chiquinquirá"]
PRODUCTOS = {"papa criolla": 2.8, "cebolla larga": 3.1, "arveja": 5.2, "maiz": 2.2, "zanahoria": 1.9}      # precio base por kg (ficticio)

rng = np.random.default_rng(SEED)
clientes_df = pd.DataFrame({"id": np.arange(1, 201), "nombre": [f"Cliente {i:03d}" for i in range(1, 201)], "municipio": rng.choice(MUNICIPIOS, 200)})
n_ped = 2000
pedidos_df = pd.DataFrame({
    "id": np.arange(1, n_ped + 1),
    "cliente_id": rng.integers(1, 201, n_ped),
    "fecha": pd.date_range("2025-01-01", periods=n_ped, freq="3h").strftime("%Y-%m-%d"),
    "producto": rng.choice(list(PRODUCTOS), n_ped),
    "cantidad": rng.integers(1, 50, n_ped),
})
pedidos_df["valor"] = (pedidos_df.cantidad * pedidos_df["producto"].map(PRODUCTOS) * 1000).round(0)          # en pesos (ficticio)

con = sqlite3.connect(":memory:")
con.execute("PRAGMA foreign_keys = ON")                                       # SQLite exige activar las claves foraneas
con.executescript("""
CREATE TABLE clientes (id INTEGER PRIMARY KEY, nombre TEXT NOT NULL, municipio TEXT NOT NULL);
CREATE TABLE pedidos (
    id         INTEGER PRIMARY KEY,
    cliente_id INTEGER NOT NULL REFERENCES clientes(id),
    fecha      TEXT    NOT NULL,
    producto   TEXT    NOT NULL,
    cantidad   INTEGER NOT NULL CHECK (cantidad > 0),
    valor      REAL    NOT NULL CHECK (valor >= 0)
);
""")
con.executemany("INSERT INTO clientes VALUES (?, ?, ?)", clientes_df.itertuples(index=False, name=None))
con.executemany("INSERT INTO pedidos VALUES (?, ?, ?, ?, ?, ?)", pedidos_df.itertuples(index=False, name=None))
con.commit()

# (1) JOIN + GROUP BY en SQL == merge + groupby en pandas
consulta = """SELECT c.municipio, COUNT(*) AS pedidos, SUM(p.valor) AS ventas
              FROM pedidos p JOIN clientes c ON c.id = p.cliente_id
              GROUP BY c.municipio ORDER BY ventas DESC"""
sql_df = pd.read_sql_query(consulta, con)
pd_df = (pedidos_df.merge(clientes_df, left_on="cliente_id", right_on="id").groupby("municipio")
         .agg(pedidos=("valor", "size"), ventas=("valor", "sum")).reset_index().sort_values("ventas", ascending=False).reset_index(drop=True))
print(sql_df.to_string(index=False))
assert (sql_df.municipio.values == pd_df.municipio.values).all() and np.allclose(sql_df.ventas, pd_df.ventas) and (sql_df.pedidos.values == pd_df.pedidos.values).all()

# (2) schema-on-write: el esquema RECHAZA lo invalido al escribir
rechazos = {}
for nombre, fila in {"cantidad negativa": (9001, 1, "2025-06-01", "arveja", -5, 100.0),
                     "cliente inexistente": (9002, 9999, "2025-06-01", "arveja", 5, 100.0),
                     "campo obligatorio vacio": (9003, 1, None, "arveja", 5, 100.0)}.items():
    try:
        con.execute("INSERT INTO pedidos VALUES (?, ?, ?, ?, ?, ?)", fila)
    except sqlite3.IntegrityError as e:
        rechazos[nombre] = str(e)
print("\nRechazos del esquema:", rechazos)
assert len(rechazos) == 3

# (3) un indice cambia el plan de ejecucion de una consulta selectiva
def plan(q):
    return " | ".join(r[3] for r in con.execute("EXPLAIN QUERY PLAN " + q))
q_sel = "SELECT SUM(valor) FROM pedidos WHERE cliente_id = 17"
plan_antes = plan(q_sel)
con.execute("CREATE INDEX idx_pedidos_cliente ON pedidos(cliente_id)")
plan_despues = plan(q_sel)
print("\nPlan sin indice:", plan_antes, "\nPlan con indice:", plan_despues)
assert "SCAN" in plan_antes and "idx_pedidos_cliente" in plan_despues

# (4) transaccion atomica: si falla UNA operacion, no se aplica NINGUNA
antes = con.execute("SELECT COUNT(*) FROM pedidos").fetchone()[0]
try:
    with con:                                                                 # 'with con' = commit si todo sale bien, rollback si hay una excepcion
        con.execute("INSERT INTO pedidos VALUES (9101, 1, '2025-06-02', 'maiz', 10, 22000)")      # valida
        con.execute("INSERT INTO pedidos VALUES (9102, 1, '2025-06-02', 'maiz', -1, 22000)")     # invalida -> IntegrityError
except sqlite3.IntegrityError:
    pass
despues = con.execute("SELECT COUNT(*) FROM pedidos").fetchone()[0]
print(f"\nPedidos antes: {antes} | despues de la transaccion fallida: {despues}")
assert antes == despues == n_ped                                              # el insert valido tambien se revirtio
con.close()
````

---
**Lectura.** La misma pregunta (ventas por municipio) da **el mismo resultado** en SQL y en *pandas*: son dos interfaces sobre el mismo modelo tabular. Lo que aporta una **base de datos** y no un `DataFrame` es la **garantía al escribir**: el esquema rechazó un valor negativo, un cliente inexistente y un campo obligatorio vacío; un **índice** convirtió un barrido completo (`SCAN`) en una búsqueda; y la **transacción** fallida revirtió también la fila válida (atomicidad). El precio es la **rigidez**: para añadir un campo nuevo hay que **migrar el esquema**, y los datos que no encajan en la tabla **no entran**. Esa tensión es la que explica la siguiente familia.

## 3. Datos Semiestructurados: JSON y XML 🧩

Los formatos **semiestructurados** son **autodescriptivos**: cada registro **lleva los nombres de sus campos**, que pueden **anidarse** (objetos dentro de objetos, listas) y **cambiar de un registro a otro**. **JSON** (el de las API web y las aplicaciones) y **XML** (documentos, intercambio entre sistemas) son los más comunes. Ventaja: añadir un campo **no rompe** a los lectores antiguos. Costo: para analizar hay que **parsear**, **aplanar** y decidir qué hacer con los campos que **solo algunos registros tienen**.

La celda genera **300 eventos JSON sintéticos** de tres tipos de fuente (sensor de temperatura, GPS y compra), con campos **distintos** entre tipos y **opcionales** dentro de cada tipo:

1. `json.dumps` / `json.loads`: la **ida y vuelta** (serialización) conserva los datos.
2. `pd.json_normalize`: **aplana** el anidamiento (`pos.lat`) y revela el **esquema variable** (porcentaje de nulos por columna).
3. `record_path`: **explota las listas** (un renglón por ítem de compra).
4. **XML** con `ElementTree`: los mismos datos del sensor, ahora con **atributos y etiquetas**, convergen al mismo `DataFrame`.

````python
# ============================================================
# 3. Semiestructurados: eventos JSON de esquema variable, aplanado con json_normalize, y el mismo sensor en XML
# ============================================================
def generar_eventos(n=300, semilla=SEED):
    rng = np.random.default_rng(semilla)
    ts = pd.date_range("2025-03-01", periods=n, freq="min").strftime("%Y-%m-%dT%H:%M:%S")
    eventos = []
    for i in range(n):
        tipo = rng.choice(["temperatura", "gps", "compra"], p=[0.45, 0.30, 0.25])
        e = {"id": int(i), "tipo": str(tipo), "ts": ts[i]}
        if tipo == "temperatura":
            e.update({"valor": round(float(rng.normal(13, 3)), 2), "unidad": "C"})
            if rng.random() < 0.2:
                e["bateria"] = round(float(rng.uniform(5, 100)), 0)                     # campo opcional: solo algunos sensores lo envian
        elif tipo == "gps":
            e.update({"pos": {"lat": round(float(5.53 + rng.normal(0, 0.01)), 5), "lon": round(float(-73.36 + rng.normal(0, 0.01)), 5)}, "velocidad_kmh": round(float(rng.gamma(2, 12)), 1)})
        else:
            items = [{"sku": f"P{rng.integers(1, 6)}", "precio": int(rng.integers(2, 90)) * 100, "cantidad": int(rng.integers(1, 5))} for _ in range(int(rng.integers(1, 5)))]
            e.update({"usuario": {"id": int(rng.integers(1, 40)), "ciudad": str(rng.choice(["Tunja", "Duitama", "Sogamoso"]))}, "items": items})
            if rng.random() < 0.3:
                e["cupon"] = "PROMO10"
        eventos.append(e)
    return eventos


eventos = generar_eventos()
texto_json = json.dumps(eventos, ensure_ascii=False)
assert json.loads(texto_json) == eventos                                          # (1) ida y vuelta sin perdida
print(f"JSON: {len(eventos)} eventos, {len(texto_json) / 1024:.0f} KiB. Un evento de compra:\n{json.dumps(next(e for e in eventos if e['tipo'] == 'compra'), ensure_ascii=False)}")

# (2) aplanado: cada ruta anidada se vuelve una columna; lo que un tipo no trae queda nulo
plano = pd.json_normalize(eventos)
nulos = (100 * plano.isna().mean()).round(1).sort_values()
print(f"\nColumnas tras aplanar: {plano.shape[1]} (filas: {plano.shape[0]})")
print("% de nulos por columna:\n" + nulos.to_string())
claves_por_tipo = {t: sorted({k for e in eventos if e["tipo"] == t for k in e}) for t in ["temperatura", "gps", "compra"]}
print("\nClaves de primer nivel por tipo:", claves_por_tipo)
assert plano.shape[0] == len(eventos) and {"pos.lat", "pos.lon", "usuario.ciudad"} <= set(plano.columns)
assert nulos["id"] == 0 and nulos["tipo"] == 0 and nulos["pos.lat"] > 50              # el esquema VARIABLE produce muchos nulos

# (3) listas: un renglon por item de compra (record_path) conservando metadatos del evento
compras = [e for e in eventos if e["tipo"] == "compra"]
items = pd.json_normalize(compras, record_path="items", meta=["id", ["usuario", "ciudad"]])
assert len(items) == sum(len(e["items"]) for e in compras)
total_items = (items.precio * items.cantidad).sum()
print(f"\nCompras: {len(compras)} -> {len(items)} lineas de items; valor total = {total_items:,.0f} pesos (ficticios)")

# (4) el MISMO sensor de temperatura en XML: etiquetas y atributos en lugar de claves
temp = [e for e in eventos if e["tipo"] == "temperatura"]
raiz = ET.Element("lecturas")
for e in temp:
    l = ET.SubElement(raiz, "lectura", id=str(e["id"]), ts=e["ts"])
    ET.SubElement(l, "valor", unidad=e["unidad"]).text = str(e["valor"])
    if "bateria" in e:
        ET.SubElement(l, "bateria").text = str(e["bateria"])
texto_xml = ET.tostring(raiz, encoding="unicode")
leido = ET.fromstring(texto_xml)
xml_df = pd.DataFrame([{"id": int(l.get("id")), "ts": l.get("ts"), "valor": float(l.findtext("valor")), "unidad": l.find("valor").get("unidad"),
                        "bateria": float(l.findtext("bateria")) if l.find("bateria") is not None else np.nan} for l in leido])
json_df = pd.json_normalize(temp)[["id", "ts", "valor", "unidad", "bateria"]]
assert xml_df.equals(json_df.astype(xml_df.dtypes.to_dict())) or (np.allclose(xml_df.valor, json_df.valor) and (xml_df.id.values == json_df.id.values).all())
print(f"XML: {len(texto_xml) / 1024:.0f} KiB para los mismos {len(temp)} eventos de temperatura que en JSON ocupan {len(json.dumps(temp)) / 1024:.0f} KiB (tamaños comparables: ninguno gana siempre).")
print(xml_df.head(3).to_string(index=False))
````

**Lectura.** (1) La serialización JSON es **sin pérdida** y legible por personas y máquinas. ## 4. Datos No Estructurados: Texto, Imágenes y *Logs* 🖼️

Aquí **no hay esquema** y el valor aparece **al extraer características**. Tres ejemplos con datos **sintéticos**:

- **Texto libre** (reseñas de clientes): se **limpia y tokeniza**, se extraen **entidades con patrón** (correos, fechas) con **expresiones regulares** y se **cuenta** (frecuencia de palabras, tamaño del vocabulario). El lenguaje natural más rico (sentimiento, temas) es tarea de NLP.
- **Imagen:** para el computador es un **arreglo de números** (alto × ancho × canales). Un píxel **no significa nada por sí solo**: el significado está en **las relaciones entre píxeles**. Pesa mucho más que una fila de tabla.
- ***Log*** de un servidor: texto con un **patrón** (IP, instante, petición, estado, bytes). Se parsea con una expresión regular **con grupos con nombre**; las líneas que **no encajan** hay que **separarlas, no ignorarlas**.

````python
# ============================================================
# 4. No estructurados: texto (regex + conteos), imagen (arreglo de numeros) y logs (patron de linea)
# ============================================================
# ---------- (a) Texto libre ----------
resenas = [
    "Excelente la papa criolla, llegó fresca el 12/03/2025. Escribanme a juan.perez@ejemplo.com si hay más.",
    "El empaque venía roto y la arveja estaba húmeda. Pedí reembolso de $ 25.000 el 14/03/2025.",
    "Buen precio, buena calidad, volveré a comprar. Contacto: maria.lopez@ejemplo.com",
    "Pésimo servicio: la entrega llegó tarde, el 15/03/2025, y nadie respondió. Quiero mi dinero de vuelta.",
    "La cebolla larga estaba fresca y el precio fue justo. Excelente atención del vendedor.",
    "No volveré a comprar: la papa estaba verde. Precio alto y mala calidad.",
]
PARADAS = {"el", "la", "los", "las", "de", "del", "y", "a", "en", "que", "un", "una", "si", "mi", "me", "al", "fue", "con", "por", "hay", "más", "mas", "o", "estaba", "era", "venía", "llegó"}
RE_CORREO = re.compile(r"[\w.+-]+@[\w-]+\.[\w.]+")
RE_FECHA = re.compile(r"\b(\d{1,2})/(\d{1,2})/(\d{4})\b")
RE_MONTO = re.compile(r"\$\s?([\d.]+)")
correos = [m for r in resenas for m in RE_CORREO.findall(r)]
fechas = [m.group(0) for r in resenas for m in RE_FECHA.finditer(r)]
montos = [int(m.replace(".", "")) for r in resenas for m in RE_MONTO.findall(r)]
tokens = [t for r in resenas for t in re.findall(r"[a-záéíóúñü]+", RE_CORREO.sub(" ", r).lower()) if t not in PARADAS]       # sin correos ni palabras vacias
conteo = Counter(tokens)
print("Correos:", correos, "| Fechas:", fechas, "| Montos:", montos)
print(f"Tokens: {len(tokens)} | vocabulario: {len(conteo)} | mas frecuentes: {conteo.most_common(6)}")
assert len(correos) == 2 and len(fechas) == 3 and montos == [25000]
assert conteo["precio"] == 3 and len(conteo) < len(tokens)                       # el vocabulario es menor que el numero de tokens (se repiten palabras)

# ---------- (b) Imagen como arreglo ----------
rng = np.random.default_rng(SEED)
yy, xx = np.indices((32, 32))
circulo = ((yy - 16) ** 2 + (xx - 16) ** 2 <= 8 ** 2).astype(float)
gris = np.clip(0.15 + 0.7 * circulo + rng.normal(0, 0.05, (32, 32)), 0, 1)                 # imagen en escala de grises (0 = negro, 1 = blanco)
rgb = np.stack([gris, gris * 0.6, gris * 0.3], axis=-1)                                      # version a color: 3 canales
rgb8 = (rgb * 255).astype(np.uint8)
reducida = gris.reshape(16, 2, 16, 2).mean(axis=(1, 3))                                      # reduccion 2x2 por promedio de bloques
print(f"\nImagen gris: forma {gris.shape}, dtype {gris.dtype}, {gris.nbytes} bytes | a color (uint8): forma {rgb8.shape}, {rgb8.nbytes} bytes | vector de rasgos: {gris.ravel().shape}")
fila_tabla = np.zeros(8, dtype=np.float64)
print(f"Una fila de tabla de 8 numeros pesa {fila_tabla.nbytes} bytes; esta miniatura a color, {rgb8.nbytes / fila_tabla.nbytes:.0f} veces mas.")
assert gris.shape == (32, 32) and rgb8.shape == (32, 32, 3) and reducida.shape == (16, 16) and rgb8.nbytes == 3072
assert gris[16, 16] > 0.6 and gris[0, 0] < 0.4                                                # el centro es claro y la esquina oscura: la informacion esta en la RELACION entre pixeles

# ---------- (c) Logs ----------
def generar_log(n=600, semilla=SEED):
    rng = np.random.default_rng(semilla)
    rutas = ["/tienda/papa", "/tienda/arveja", "/carrito", "/pago", "/login", "/api/precios"]
    estados = rng.choice([200, 301, 404, 500], n, p=[0.86, 0.04, 0.07, 0.03])
    lineas = []
    for i in range(n):
        seg = i % 60; minuto = (i // 60) % 60
        lineas.append(f'10.0.{rng.integers(0, 4)}.{rng.integers(1, 250)} - - [01/Mar/2025:10:{minuto:02d}:{seg:02d} -0500] "GET {rng.choice(rutas)} HTTP/1.1" {estados[i]} {rng.integers(200, 5000)}')
    return lineas


lineas = generar_log() + ["linea corrupta sin patron", '10.0.1.7 - - [01/Mar/2025:10:59:59 -0500] "GET /pago HTTP/1.1" 200']          # 2 lineas mal formadas
RE_LOG = re.compile(r'(?P<ip>\S+) \S+ \S+ \[(?P<ts>[^\]]+)\] "(?P<metodo>\w+) (?P<ruta>\S+) [^"]+" (?P<estado>\d{3}) (?P<bytes>\d+)')
ok, cuarentena = [], []
for l in lineas:
    m = RE_LOG.fullmatch(l)
    (ok.append(m.groupdict()) if m else cuarentena.append(l))                                  # lo que no encaja NO se descarta: va a cuarentena para revisarlo
log_df = pd.DataFrame(ok).astype({"estado": int, "bytes": int})
print(f"\nLog: {len(lineas)} lineas -> {len(log_df)} parseadas, {len(cuarentena)} en cuarentena: {cuarentena}")
por_estado = log_df.estado.value_counts().sort_index()
print(por_estado.to_dict(), f"| proporcion de errores 5xx: {(log_df.estado >= 500).mean():.1%}")
assert len(log_df) == 600 and len(cuarentena) == 2 and (log_df.estado >= 500).sum() > 0

fig, ax = plt.subplots(1, 3, figsize=(14, 3.8))
mc = conteo.most_common(8)
ax[0].barh([p for p, _ in mc][::-1], [c for _, c in mc][::-1], color=PALETA[0]); ax[0].set_title("Texto: palabras más frecuentes", fontsize=10, fontweight="bold")
ax[1].imshow(gris, cmap="gray", vmin=0, vmax=1); ax[1].set_title("Imagen: 32×32 = 1 024 números", fontsize=10, fontweight="bold"); ax[1].axis("off")
ax[2].bar(por_estado.index.astype(str), por_estado.values, color=[PALETA[2] if e < 400 else PALETA[3] if e < 500 else PALETA[1] for e in por_estado.index])
ax[2].set_title("Log: peticiones por código de estado", fontsize=10, fontweight="bold")
plt.tight_layout(); plt.show()
````

## 5. *Schema-on-write* frente a *Schema-on-read* ⚖️

Con los **mismos datos «sucios»**, comparamos las dos estrategias. Generamos **2 000 pagos** en JSON por líneas (**sintéticos**) con los defectos típicos de la realidad: el `valor` a veces llega como **texto con coma decimal** (`"12,5"`), como `"N/A"` o como `null`; la `fecha` viene en dos formatos (ISO `2025-03-05` y `05/03/2025`); y algunos registros traen un **campo nuevo** (`canal`).

- **Schema-on-write** (cargar a una tabla estricta): una función **valida y convierte antes de guardar**; lo que no cumple va a una **cola de rechazados** (*dead-letter*). Rápido de consultar después, pero **lo rechazado se pierde** si nadie lo recupera, y si el esquema cambia hay que **reingestar**.
- **Schema-on-read** (guardar el crudo): se almacenan las líneas **sin tocar**; cada lectura aplica **su** esquema. Dos lectores con reglas distintas obtienen **resultados distintos de los mismos bytes**, y se puede **volver al crudo** cuando mejora la regla.

````python
# ============================================================
# 5. schema-on-write vs schema-on-read con los MISMOS datos sucios (2 000 pagos sinteticos en JSON por lineas)
# ============================================================
def generar_pagos(n=2000, semilla=SEED):
    rng = np.random.default_rng(semilla)
    pagos = []
    for i in range(n):
        v = round(float(rng.lognormal(10, 0.6)), 1)                                        # valor en pesos (ficticio)
        u = rng.random()
        valor = v if u < 0.86 else str(v).replace(".", ",") if u < 0.94 else "N/A" if u < 0.98 else None
        d = date(2025, 3, 1 + int(rng.integers(0, 28)))
        fecha = d.isoformat() if rng.random() < 0.85 else d.strftime("%d/%m/%Y")
        p = {"id": i, "valor": valor, "fecha": fecha}
        if rng.random() < 0.3:
            p["canal"] = str(rng.choice(["web", "app", "tienda"]))
        pagos.append(p)
    return pagos


crudo = [json.dumps(p) for p in generar_pagos()]                                           # lo que 'llega': lineas de texto JSON


# ---------- schema-on-write: validar y convertir ANTES de guardar ----------
def validar(r):
    if not isinstance(r["valor"], (int, float)) or r["valor"] < 0:
        raise ValueError("valor no numerico")
    try:
        date.fromisoformat(r["fecha"])                                                     # solo acepta ISO
    except ValueError:
        raise ValueError("fecha no ISO")
    return (r["id"], float(r["valor"]), r["fecha"])


aceptados, rechazados, motivos = [], [], Counter()
for linea in crudo:
    r = json.loads(linea)
    try:
        aceptados.append(validar(r))
    except ValueError as e:
        rechazados.append(linea); motivos[str(e)] += 1
con = sqlite3.connect(":memory:")
con.execute("CREATE TABLE pagos (id INTEGER PRIMARY KEY, valor REAL NOT NULL, fecha TEXT NOT NULL)")
con.executemany("INSERT INTO pagos VALUES (?, ?, ?)", aceptados)
suma_write = con.execute("SELECT SUM(valor) FROM pagos").fetchone()[0]
con.close()
print(f"SCHEMA-ON-WRITE : {len(aceptados)} aceptados, {len(rechazados)} rechazados ({len(rechazados) / len(crudo):.1%}) | motivos: {dict(motivos)}")
print(f"                  suma de valores = {suma_write:,.0f}")

# ---------- schema-on-read: el crudo queda intacto; cada lector aplica SU esquema ----------
df_crudo = pd.DataFrame([json.loads(l) for l in crudo])                                    # el 'esquema' se descubre al leer (canal solo existe en algunos)


def leer_v1(d):
    """Lector ESTRICTO: numero tal cual y fecha ISO; lo demas es nulo."""
    return pd.DataFrame({"valor": pd.to_numeric(d.valor, errors="coerce"), "fecha": pd.to_datetime(d.fecha, format="%Y-%m-%d", errors="coerce")})


def leer_v2(d):
    """Lector TOLERANTE: acepta coma decimal y dos formatos de fecha; ignora 'N/A'."""
    valor = pd.to_numeric(d.valor.astype("string").str.replace(",", ".", regex=False), errors="coerce")
    fecha = pd.to_datetime(d.fecha, format="%Y-%m-%d", errors="coerce").fillna(pd.to_datetime(d.fecha, format="%d/%m/%Y", errors="coerce"))
    return pd.DataFrame({"valor": valor, "fecha": fecha})


v1, v2 = leer_v1(df_crudo).dropna(), leer_v2(df_crudo).dropna()                          # 'util' = fila con valor Y fecha interpretables
util1, util2 = len(v1), len(v2)
print(f"SCHEMA-ON-READ  : lector estricto -> {util1} filas utiles, suma = {v1.valor.sum():,.0f} | lector tolerante -> {util2} filas utiles, suma = {v2.valor.sum():,.0f}")
print(f"                  el crudo conserva las {len(crudo)} lineas; 'canal' aparece en {df_crudo.canal.notna().mean():.0%} de ellas")

tabla = pd.DataFrame({"filas utiles": [len(aceptados), util1, util2], "% del total": [100 * len(aceptados) / len(crudo), 100 * util1 / len(crudo), 100 * util2 / len(crudo)]},
                     index=["on-write (estricto)", "on-read v1 (estricto)", "on-read v2 (tolerante)"]).round(1)
print("\n" + tabla.to_string())
assert len(aceptados) + len(rechazados) == len(crudo) == 2000
assert len(aceptados) == util1                                                              # un lector estricto equivale al esquema al escribir...
assert util2 > util1 and v2.valor.sum() > v1.valor.sum()                                   # ...pero un lector mejor RECUPERA datos del mismo crudo
assert abs(suma_write - v1.valor.sum()) < 1e-6 and "canal" in df_crudo.columns
````

---
**Lectura.** El **esquema al escribir** aceptó **1 453 de 2 000 pagos (72.6 %)** y **rechazó 547 (27.4 %)**: 287 por valor no numérico y 260 por fecha que no era ISO. Su suma de valores (≈ 38.5 millones) **subestima** la realidad, y quien consulte esa tabla **no sabe** cuánto se quedó fuera. Un lector **estricto** sobre el crudo da **exactamente lo mismo** (mismas filas, misma suma), pero el crudo **sigue ahí**: el lector **tolerante** (que entiende comas decimales y dos formatos de fecha) **recupera** hasta **1 889 filas (94.4 %)** y una suma de ≈ 49.9 millones (**≈ 30 % más**) **sin reingestar nada**. Lo que ni siquiera el lector tolerante recupera (los «N/A» y los nulos) es **información que nunca existió**. El compromiso es evidente: *schema-on-write* **garantiza calidad al consultar**, pero **pierde o retrasa** lo que no encaja; *schema-on-read* **conserva todo** y da flexibilidad, pero **traslada el trabajo y el riesgo a cada lector**: dos analistas pueden obtener cifras distintas **de los mismos bytes** si no comparten el esquema. En la práctica se **combinan** (zonas «crudo» y «curado»; ver el [cuaderno 03](03_Ciclo_de_Vida_de_los_Datos.ipynb) y el Módulo 02).

## 6. Formatos de Archivo: CSV, JSON por Líneas y Parquet 📦

El **formato** decide el **tamaño en disco**, la **velocidad de lectura** y si se **conservan los tipos**. Dos familias:

| Familia | Formatos | Idea | Fortaleza | Debilidad |
|---|---|---|---|---|
| **Por filas** (*row-oriented*) | CSV, JSON por líneas (*JSON Lines*), Avro | Un registro seguido de otro | Fácil de escribir y de **añadir** registros; bueno si se lee **cada registro completo** | Para leer **una columna** hay que recorrer **todo** |
| **Columnares** | Parquet, ORC | Los valores de **una columna** se guardan juntos, con **tipos**, **compresión** y **estadísticas** por bloque | Lee **solo las columnas** que necesitas; **comprime** mucho (valores parecidos juntos); conserva **tipos** | Más costoso de escribir y de modificar un registro |

CSV y JSON son **texto**: legibles por personas, pero **sin tipos** (todo es texto hasta que se interpreta) y con **más bytes por valor**. Parquet es **binario**, **tipado** y con compresión. Medimos con una tabla **sintética** de **200 000 lecturas** de sensores (identificador, instante, sensor, valor, estado y lote): tamaño en disco, tiempo de escritura, tiempo de leer **todo** y tiempo de leer **una sola columna** (`valor`), tomando el **mejor de 3 intentos** (`RAPIDO = False` sube a 1 000 000 de filas). *(Los tiempos dependen de tu máquina, de su carga y del disco: lo robusto son las **proporciones**.)*

````python
# ============================================================
# 6. CSV vs JSON por lineas vs Parquet: tamano, escritura, lectura total y lectura de UNA columna (medido)
# ============================================================
rng = np.random.default_rng(SEED)
N = N_FORMATOS
datos = pd.DataFrame({
    "id": np.arange(N),
    "ts": pd.date_range("2025-01-01", periods=N, freq="s"),
    "sensor": rng.choice([f"sensor_{i:02d}" for i in range(50)], N),
    "valor": rng.normal(20, 5, N).round(3),
    "estado": rng.choice(["ok", "alerta", "falla", "mantenimiento", "desconocido"], N, p=[0.8, 0.1, 0.04, 0.04, 0.02]),
    "lote": rng.integers(0, 1000, N).astype("int32"),
})


def mejor_de(f, n=3):
    mejor, r = np.inf, None
    for _ in range(n):
        t0 = time.perf_counter(); r = f(); mejor = min(mejor, time.perf_counter() - t0)
    return mejor, r


def mb(p):
    return os.path.getsize(p) / 2 ** 20


rutas = {"csv": os.path.join(TMP, "datos.csv"), "jsonl": os.path.join(TMP, "datos.jsonl"), "parquet": os.path.join(TMP, "datos.parquet")}
res = {}
t, _ = mejor_de(lambda: datos.to_csv(rutas["csv"], index=False), 1)
res["CSV"] = {"MB": mb(rutas["csv"]), "escribir (s)": t,
              "leer todo (s)": mejor_de(lambda: pd.read_csv(rutas["csv"]))[0], "leer 1 columna (s)": mejor_de(lambda: pd.read_csv(rutas["csv"], usecols=["valor"]))[0]}
t, _ = mejor_de(lambda: datos.to_json(rutas["jsonl"], orient="records", lines=True, date_format="iso"), 1)
t_json = mejor_de(lambda: pd.read_json(rutas["jsonl"], lines=True), 1)[0]
res["JSON por líneas"] = {"MB": mb(rutas["jsonl"]), "escribir (s)": t, "leer todo (s)": t_json, "leer 1 columna (s)": t_json}          # no hay atajo: se parsea todo
if HAY_ARROW:
    t, _ = mejor_de(lambda: datos.to_parquet(rutas["parquet"]), 1)                       # compresion por defecto (snappy)
    res["Parquet"] = {"MB": mb(rutas["parquet"]), "escribir (s)": t,
                      "leer todo (s)": mejor_de(lambda: pd.read_parquet(rutas["parquet"]))[0], "leer 1 columna (s)": mejor_de(lambda: pd.read_parquet(rutas["parquet"], columns=["valor"]))[0]}
tabla_fmt = pd.DataFrame(res).T
print(f"Filas: {N:,} | columnas: {datos.shape[1]}")
print(tabla_fmt.round(3).to_string())

if HAY_ARROW:
    csv_ida = pd.read_csv(rutas["csv"]); pq_ida = pd.read_parquet(rutas["parquet"])
    print("\nTipos tras la ida y vuelta -> CSV:", {c: str(t) for c, t in csv_ida.dtypes.items() if c in ("ts", "lote")}, "| Parquet:", {c: str(t) for c, t in pq_ida.dtypes.items() if c in ("ts", "lote")})
    meta = pq.ParquetFile(rutas["parquet"]).metadata
    print(f"Parquet: {meta.num_row_groups} grupo(s) de filas, {meta.num_columns} columnas; estadistica por bloque de 'valor' (min, max): {meta.row_group(0).column(3).statistics.min:.2f}, {meta.row_group(0).column(3).statistics.max:.2f}")
    assert res["Parquet"]["MB"] < res["CSV"]["MB"] < res["JSON por líneas"]["MB"]                    # tamano: Parquet < CSV < JSON por lineas
    assert res["Parquet"]["leer 1 columna (s)"] < res["CSV"]["leer 1 columna (s)"] < res["JSON por líneas"]["leer 1 columna (s)"]
    assert not str(csv_ida["ts"].dtype).startswith("datetime64") and str(pq_ida["ts"].dtype).startswith("datetime64")     # el CSV pierde el tipo fecha
    assert str(pq_ida["lote"].dtype) == "int32" and pd.testing.assert_frame_equal(pq_ida, datos) is None                   # Parquet conserva tipos y datos
    ratio_tam = res["CSV"]["MB"] / res["Parquet"]["MB"]; ratio_col = res["CSV"]["leer 1 columna (s)"] / res["Parquet"]["leer 1 columna (s)"]
    print(f"\nParquet ocupa {ratio_tam:.1f} veces menos que el CSV y lee una columna {ratio_col:.0f} veces mas rapido que `usecols` de CSV.")
    fig, ax = plt.subplots(1, 2, figsize=(11, 3.8))
    ax[0].bar(tabla_fmt.index, tabla_fmt["MB"], color=[PALETA[3], PALETA[1], PALETA[2]]); ax[0].set_ylabel("MB en disco"); ax[0].set_title("Tamaño en disco", fontsize=10, fontweight="bold")
    ax[1].bar(tabla_fmt.index, tabla_fmt["leer 1 columna (s)"], color=[PALETA[3], PALETA[1], PALETA[2]]); ax[1].set_yscale("log"); ax[1].set_ylabel("segundos (escala log)")
    ax[1].set_title("Leer UNA columna", fontsize=10, fontweight="bold"); plt.tight_layout(); plt.show()
else:
    print("pyarrow no esta instalado: se omite Parquet (en Colab: `!pip install -q pyarrow`).")
````

**Lectura.** Con 200 000 filas, **Parquet ocupa ≈ 2.9 veces menos que el CSV** (≈ 3.4 MB frente a ≈ 9.8 MB) y **≈ 6 veces menos que el JSON por líneas** (≈ 20.3 MB), que repite el nombre de cada campo en **cada** registro. Al **leer una sola columna**, Parquet es **entre 10 y 16 veces más rápido** (según la ejecución) que `usecols` de CSV (que igual tiene que recorrer y *tokenizar* todas las líneas) y el JSON por líneas **no tiene atajo**. Además, el CSV **pierde los tipos** (el instante vuelve como texto y el entero de 32 bits como de 64) y Parquet los **conserva** (la ida y vuelta es idéntica). Los límites: Parquet **no es legible por personas**, es más caro de **modificar** (se reescribe un bloque) y para conjuntos pequeños que **se leen completos** y se editan a mano, el CSV sigue siendo razonable. *Tus tiempos serán otros; las proporciones, similares.*

## 7. Guía Rápida: ¿Qué Almacenar Dónde? 🧭

| Necesidad | Tipo de dato | Dónde y en qué formato | Por qué |
|---|---|---|---|
| Transacciones con **integridad** (pedidos, pagos, inventario) | Estructurado | Base de datos **relacional** (SQL) | ACID, restricciones, *joins*, índices |
| **Análisis histórico** de grandes tablas (reportes, BI) | Estructurado | ***Data warehouse*** o archivos **Parquet/ORC** en un *data lake* | Lectura de pocas columnas, compresión, escaneos masivos |
| Eventos de aplicaciones, catálogos con campos que **cambian** | Semiestructurado | Base **documental** (NoSQL) o **JSON por líneas** en el *lake*; al analizar, **aplanar** a Parquet | Esquema flexible; el crudo se conserva |
| **Intercambio** entre sistemas, documentos con estructura | Semiestructurado | **XML** o **JSON** (API) | Autodescriptivo, estándar |
| Sensores y métricas (**series de tiempo**) | Estructurado / semi | Base de series de tiempo o Parquet **particionado por fecha** | Escritura continua, consultas por ventana |
| Imágenes, audio, video, PDF | No estructurado | **Almacenamiento de objetos** + **metadatos** en una tabla | Binarios grandes; se busca por metadatos o por características extraídas |
| Texto (reseñas, correos) | No estructurado | Objetos o columnas de texto + **índice de búsqueda**; *embeddings* para similitud | Búsqueda y NLP |
| ***Logs*** | Semiestructurado (patrón) | Archivos comprimidos en el *lake* + parseo a tabla | Volumen alto; se parsea **al leer** |

> 🎯 **Regla de oro:** **guarda el crudo, publica lo curado, y documenta el esquema**. Lo crudo permite rehacer; lo curado permite confiar; el esquema documentado permite que otros entiendan.

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Tres familias** | **Estructurados** (esquema fijo, SQL), **semiestructurados** (autodescriptivos: JSON, XML, *logs*) y **no estructurados** (texto, imágenes, audio: el significado se **infiere**). Es un **espectro**, no tres cajas |
| **Estructurados** | SQL y *pandas* dan **el mismo resultado**; la base de datos añade **garantías al escribir** (restricciones, claves foráneas), **índices** y **transacciones atómicas** a cambio de **rigidez** |
| **Semiestructurados** | Esquema variable ⇒ al **aplanar** aparecen **muchos nulos que no son faltantes** (campos que no aplican); las listas exigen decidir la **granularidad** (`record_path`) |
| **No estructurados** | Se necesita **extraer características**: *regex* y conteos para texto, **arreglos** para imágenes (una miniatura a color pesa cientos de veces una fila), **patrones** para *logs* con **cuarentena** de líneas mal formadas |
| ***Schema-on-write* frente a *on-read*** | Al escribir: calidad garantizada al consultar, pero se **rechaza** (y se pierde) lo que no encaja. Al leer: se **conserva todo** y cada lector aplica su esquema (dos lectores, dos cifras). En la práctica se **combinan** (crudo + curado) |
| **Formatos de archivo** | **Parquet** (columnar, binario, tipado, comprimido) ocupó **≈ 2.9 veces menos** que el CSV y leyó **una columna entre 10 y 16 veces más rápido** (200 000 filas, esta máquina); el CSV **pierde tipos**; el JSON por líneas es el más grande |
| **Guía** | Transacciones → SQL; histórico analítico → *warehouse*/Parquet; eventos cambiantes → documentos/JSON; binarios → objetos + metadatos; **guarda el crudo, publica lo curado, documenta el esquema** |
