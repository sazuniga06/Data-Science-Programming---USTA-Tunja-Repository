## 1. Las Seis Etapas del Ciclo de Vida 🔄

| # | Etapa | Pregunta que responde | Qué ocurre | Riesgo típico | Métrica / control |
|---|---|---|---|---|---|
| 1 | **Generación** | ¿De dónde viene y con qué finalidad? | Sensores, transacciones, formularios, personas, otros sistemas producen el dato | Capturar de más (**sin finalidad**), capturar mal, no tener consentimiento | Cobertura de fuentes; consentimiento registrado |
| 2 | **Ingesta** | ¿Cómo llega a donde se usa? | Lotes o flujo; se valida el formato; lo malformado se aparta | Pérdida, **duplicados**, orden alterado, datos corruptos | Filas recibidas frente a aceptadas; **cuarentena** |
| 3 | **Almacenamiento** | ¿Dónde y con qué estructura se guarda? | Zonas **cruda** (intacta) y **curada**; formato y particiones; cifrado | Silos, sin metadatos, accesos excesivos, sin respaldo | Tamaño, costo, tiempo de acceso, accesos auditados |
| 4 | **Procesamiento** | ¿Cómo se limpia y transforma? | Limpieza, deduplicación, tipos, integración, derivación de variables | **Errores silenciosos**, transformaciones no reproducibles | **Calidad** (completitud, unicidad, validez); pruebas |
| 5 | **Análisis y visualización** | ¿Qué decisión apoya? | Agregaciones, modelos, tableros, informes | Conclusiones sin linaje; **sesgo**; cifras que nadie puede reproducir | Reproducibilidad; **linaje** hasta la cifra mostrada |
| 6 | **Archivado y eliminación** | ¿Cuánto tiempo se conserva y cómo se destruye? | Se mueve a almacenamiento frío o se **borra/anonimiza** según una política | **Conservar de más** (riesgo legal y de seguridad), borrar de menos (copias olvidadas) | Registros vencidos; **borrado verificable** |

**No es una línea, es un ciclo:** las decisiones del análisis **generan nuevos datos** (una campaña produce respuestas, una alerta produce acciones), y el **archivado** de unos datos puede ser la **fuente** de otros (históricos para entrenar). Atraviesan todas las etapas cuatro hilos: **calidad**, **seguridad**, **linaje** (trazabilidad: ¿de dónde sale esta cifra?) y **gobernanza** (quién es responsable del dato).

La celda siguiente **dibuja el ciclo** con `matplotlib`, sin herramientas externas: cajas (`FancyBboxPatch`) y flechas (`annotate`).

## 2. Un *Pipeline* Mínimo: una Función por Etapa 🛠️

Implementamos el ciclo completo con **datos sintéticos** de **6 estaciones de monitoreo de cultivos** (humedad relativa y temperatura horarias durante 7 días). Los datos **traen los defectos de la realidad**, para que el *pipeline* tenga algo que hacer:

- **~4 %** de lecturas de humedad **vacías** y **~2 %** **fuera de rango** (humedad de 140 % o negativa: falla del sensor).
- **~3 %** de lecturas **duplicadas** (el sensor reintenta el envío).
- Una estación («E03») con el sensor **«pegado»**: repite **63.0 %** durante 12 horas.
- **5 líneas corruptas** (JSON truncado).
- Llegada **desordenada**.

**Diseño (una función por etapa):**

| Función | Etapa | Entrada → Salida | Qué hace |
|---|---|---|---|
| `generar()` | 1 | — → `origen/lecturas.jsonl` | Simula la fuente: escribe líneas JSON |
| `ingerir()` | 2 | archivo → `DataFrame` + líneas rechazadas | Lee línea a línea; **aparta lo no parseable** |
| `almacenar()` | 3 | `DataFrame` → zona **cruda** | Guarda **sin modificar** y comprueba la ida y vuelta |
| `procesar()` | 4 | cruda → zona **curada** + **cuarentena** | Tipos, **deduplicación**, reglas de validez; lo que no cumple va a cuarentena **con motivo** |
| `analizar()` | 5 | curada → tabla de resultados | Resumen diario por estación |
| `archivar()` | 6 | curada → **archivo** frío | Mueve lo **más antiguo que el plazo de retención** y lo comprime |

Cada función **registra en la bitácora de linaje** cuántas filas recibió, cuántas entregó, y una **huella** (*hash*) del resultado, de modo que se pueda **reconstruir qué pasó** y comprobar que **ninguna fila desaparece sin explicación** (conservación de filas).

````python
# ============================================================
# 2. Pipeline minimo: generar -> ingerir -> almacenar -> procesar -> analizar -> archivar, con bitacora de linaje
# ============================================================
ZONAS = {z: os.path.join(TMP, z) for z in ["origen", "cruda", "curada", "archivo"]}
for ruta in ZONAS.values():
    os.makedirs(ruta, exist_ok=True)
BITACORA = []                                                                  # linaje: una entrada por paso
OBLIGATORIAS = ["estacion", "ts", "humedad", "temperatura"]
CLAVE = ["estacion", "ts"]


def huella(df):
    """Huella de 8 hex del CONTENIDO (no del orden de las filas): sirve para detectar que algo cambio entre dos pasos."""
    h = pd.util.hash_pandas_object(df.reset_index(drop=True), index=False).sort_values().values
    return hashlib.sha256(h.tobytes()).hexdigest()[:8]


def registrar(etapa, funcion, n_in, n_out, df=None, nota=""):
    BITACORA.append({"paso": len(BITACORA) + 1, "etapa": etapa, "funcion": funcion, "filas_entrada": n_in, "filas_salida": n_out,
                     "huella": huella(df) if df is not None else "-", "nota": nota})


def escribir(df, nombre, zona):
    ruta = os.path.join(ZONAS[zona], f"{nombre}.{EXT}")
    df.to_parquet(ruta, index=False) if EXT == "parquet" else df.to_csv(ruta, index=False)
    return ruta


def leer(ruta):
    return pd.read_parquet(ruta) if ruta.endswith(".parquet") else pd.read_csv(ruta)


# ---------- 1. Generacion: la fuente (sensores) escribe lineas JSON con los defectos de la realidad ----------
def generar(n_estaciones=6, dias=7, semilla=SEED):
    rng = np.random.default_rng(semilla)
    horas = pd.date_range("2025-03-01", periods=dias * 24, freq="h")
    ciclo = np.sin(2 * np.pi * (horas.hour.values - 9) / 24)
    filas = []
    for e in range(1, n_estaciones + 1):
        hum = 72 - 12 * ciclo + rng.normal(0, 3, len(horas)) + rng.normal(0, 4)
        temp = 13 + 4 * ciclo + rng.normal(0, 0.7, len(horas))
        if e == 3:
            hum[60:72] = 63.0                                                  # sensor 'pegado': 12 horas con el MISMO valor
        for t, h, c in zip(horas, hum, temp):
            filas.append({"estacion": f"E{e:02d}", "ts": t.strftime("%Y-%m-%dT%H:%M:%S"), "humedad": round(float(np.clip(h, 5, 99)), 1), "temperatura": round(float(c), 1)})
    n = len(filas)
    for i in rng.choice(n, int(0.04 * n), replace=False):
        filas[i]["humedad"] = None                                             # vacias
    for i in rng.choice(n, int(0.02 * n), replace=False):
        filas[i]["humedad"] = float(rng.choice([140.0, -3.0]))                 # fuera de rango fisico
    filas += [dict(filas[i]) for i in rng.choice(n, int(0.03 * n), replace=False)]    # duplicados por reintento
    lineas = [json.dumps(f) for f in filas] + ['{"estacion": "E01", "ts": "2025-03-0', '{estacion: E02}', 'basura', '{"estacion": ', 'null null']
    lineas = [lineas[i] for i in rng.permutation(len(lineas))]                 # llegada desordenada
    ruta = os.path.join(ZONAS["origen"], "lecturas.jsonl")
    with open(ruta, "w", encoding="utf-8") as f:
        f.write("\n".join(lineas) + "\n")
    registrar("1. Generación", "generar", 0, len(lineas), nota="lineas escritas en la fuente (incluye 5 corruptas y duplicados)")
    return ruta


# ---------- 2. Ingesta: leer linea a linea y APARTAR lo no parseable ----------
def ingerir(ruta):
    buenas, rechazadas = [], []
    with open(ruta, encoding="utf-8") as f:
        for linea in f:
            linea = linea.strip()
            try:
                r = json.loads(linea)
                if not isinstance(r, dict) or not set(CLAVE) <= set(r):
                    raise ValueError("no es un registro de lectura")
                buenas.append(r)
            except ValueError:
                rechazadas.append(linea)
    df = pd.DataFrame(buenas)
    registrar("2. Ingesta", "ingerir", len(buenas) + len(rechazadas), len(df), df, f"{len(rechazadas)} lineas rechazadas -> cuarentena de ingesta")
    return df, rechazadas


# ---------- 3. Almacenamiento: zona CRUDA, sin modificar, con verificacion de ida y vuelta ----------
def almacenar(df):
    ruta = escribir(df, "lecturas_crudas", "cruda")
    de_vuelta = leer(ruta)
    assert huella(de_vuelta[df.columns]) == huella(df), "la zona cruda no coincide con lo ingerido"
    registrar("3. Almacenamiento", "almacenar", len(df), len(de_vuelta), de_vuelta, f"zona cruda: {os.path.basename(ruta)}")
    return ruta


# ---------- 4. Procesamiento: tipos, deduplicacion y reglas de validez; lo invalido va a CUARENTENA con motivo ----------
HUM_MIN, HUM_MAX, TEMP_MIN, TEMP_MAX = 0.0, 100.0, -5.0, 45.0                  # rango fisico plausible (ilustrativo)


def procesar(df):
    n0 = len(df)
    d = df.assign(ts=pd.to_datetime(df.ts), humedad=pd.to_numeric(df.humedad, errors="coerce"), temperatura=pd.to_numeric(df.temperatura, errors="coerce"))
    sin_dup = d.drop_duplicates(CLAVE, keep="first")
    n_dup = n0 - len(sin_dup)
    nulo = sin_dup[["humedad", "temperatura"]].isna().any(axis=1)
    fuera = ~nulo & ~(sin_dup.humedad.between(HUM_MIN, HUM_MAX) & sin_dup.temperatura.between(TEMP_MIN, TEMP_MAX))
    cuarentena = pd.concat([sin_dup[nulo].assign(motivo="valor_nulo"), sin_dup[fuera].assign(motivo="fuera_de_rango")])
    curado = sin_dup[~nulo & ~fuera].sort_values(CLAVE).reset_index(drop=True)
    assert n0 == len(curado) + len(cuarentena) + n_dup                         # CONSERVACION DE FILAS: nada desaparece sin explicacion
    escribir(curado, "lecturas_curadas", "curada")
    registrar("4. Procesamiento", "procesar", n0, len(curado), curado, f"{n_dup} duplicados eliminados, {len(cuarentena)} a cuarentena ({ {k: int(v) for k, v in cuarentena.motivo.value_counts().items()} })")
    return curado, cuarentena, n_dup


# ---------- 5. Analisis: resumen diario por estacion ----------
def analizar(curado):
    res = (curado.assign(dia=curado.ts.dt.date).groupby(["estacion", "dia"]).agg(hum_media=("humedad", "mean"), hum_max=("humedad", "max"), temp_media=("temperatura", "mean"), n=("humedad", "size")).round(2).reset_index())
    registrar("5. Análisis", "analizar", len(curado), len(res), res, "resumen diario por estacion (no contiene datos personales)")
    return res


# ---------- 6. Archivado: lo mas antiguo que la retencion se mueve a la zona de archivo (comprimido) y sale de la zona curada ----------
def archivar(curado, hoy, retencion_dias):
    corte = hoy - pd.Timedelta(days=retencion_dias)
    viejo = curado[curado.ts < corte]
    activo = curado[curado.ts >= corte].reset_index(drop=True)
    ruta = os.path.join(ZONAS["archivo"], "lecturas_archivo.csv.gz")
    with gzip.open(ruta, "wt", encoding="utf-8") as f:
        viejo.to_csv(f, index=False)
    registrar("6. Archivado", "archivar", len(curado), len(activo), activo, f"{len(viejo)} filas anteriores a {corte.date()} -> archivo comprimido ({os.path.getsize(ruta) / 1024:.0f} KiB)")
    return activo, viejo


# ---------- Ejecucion de punta a punta ----------
HOY, RETENCION_DIAS = pd.Timestamp("2025-03-10"), 5                              # 'hoy' FIJO (reproducible) y plazo ILUSTRATIVO
ruta_origen = generar()
crudo_df, rechazadas = ingerir(ruta_origen)
almacenar(crudo_df)
curado_df, cuarentena_df, n_dup = procesar(crudo_df)
resumen_df = analizar(curado_df)
activo_df, archivo_df = archivar(curado_df, HOY, RETENCION_DIAS)

bitacora = pd.DataFrame(BITACORA)
print(bitacora[["paso", "etapa", "funcion", "filas_entrada", "filas_salida", "huella"]].to_string(index=False))
print("\nNotas:"); [print(f"  {r.paso}. {r.nota}") for r in bitacora.itertuples()]
assert len(bitacora) == 6 and bitacora.funcion.tolist() == ["generar", "ingerir", "almacenar", "procesar", "analizar", "archivar"]
assert len(rechazadas) == 5 and n_dup > 0 and len(cuarentena_df) > 0
assert len(activo_df) + len(archivo_df) == len(curado_df)                          # lo archivado + lo activo = lo curado
assert activo_df.ts.min() >= HOY - pd.Timedelta(days=RETENCION_DIAS) and archivo_df.ts.max() < HOY - pd.Timedelta(days=RETENCION_DIAS)
````

## 3. Calidad y Linaje: Medir lo que se Mueve 📏

Tres métricas simples y **verificables** (se calculan sobre un `DataFrame` y se pueden **automatizar como pruebas** en cada ejecución). Con $n$ filas y $m$ columnas obligatorias:

$$\text{Completitud}=1-\frac{\#\text{celdas obligatorias vacías}}{n\cdot m},\qquad \text{Unicidad}=\frac{\#\text{claves distintas}}{n},\qquad \text{Validez}=\frac{\#\text{filas que cumplen todas las reglas}}{n}.$$

La **clave** de una lectura es (`estacion`, `ts`); las **reglas** son el rango físico plausible de humedad (0–100 %) y de temperatura (−5 a 45 °C), **ilustrativo**: debe definirlo quien conoce el sensor. Medimos la calidad **antes** (zona cruda) y **después** (zona curada) del procesamiento, y miramos también la **huella de linaje** de cada paso.

> ⚠️ **Las métricas no lo dicen todo.** Las tres miden lo que **sabes** que puede fallar. Un **sensor «pegado»** (repite 63.0 durante 12 horas) **cumple todas las reglas de rango** y no es duplicado ni vacío: seguirá en la zona curada. Eso es un problema de **veracidad**, que requiere **otras reglas** (por ejemplo, detectar rachas de valores idénticos) o conocimiento del dominio. Lo comprobamos en la práctica.

````python
# ============================================================
# 3. Metricas de calidad antes/despues, y una mirada al linaje y al sensor 'pegado'
# ============================================================
def calidad(df):
    """Completitud, unicidad y validez de un DataFrame de lecturas."""
    n = len(df)
    vacias = df[OBLIGATORIAS].isna().sum().sum()
    completitud = 1 - vacias / (n * len(OBLIGATORIAS))
    unicidad = len(df.drop_duplicates(CLAVE)) / n
    h, t = pd.to_numeric(df.humedad, errors="coerce"), pd.to_numeric(df.temperatura, errors="coerce")
    validez = (h.between(HUM_MIN, HUM_MAX) & t.between(TEMP_MIN, TEMP_MAX)).mean()
    return {"completitud": completitud, "unicidad": unicidad, "validez": validez}


q_crudo, q_curado = calidad(crudo_df), calidad(curado_df)
tabla_q = pd.DataFrame({"zona cruda (tras ingesta)": q_crudo, "zona curada (tras procesar)": q_curado}).T
print((100 * tabla_q).round(1).to_string())
print(f"\nFilas: ingeridas {len(crudo_df):,} -> curadas {len(curado_df):,} | duplicados eliminados {n_dup} | cuarentena {len(cuarentena_df)} | lineas ilegibles {len(rechazadas)}")
assert all(q_curado[k] == 1.0 for k in q_curado)                                   # tras procesar, las tres metricas valen 1 POR CONSTRUCCION
assert all(q_crudo[k] < q_curado[k] for k in q_crudo)                              # y en la zona cruda las tres estan por debajo de 1

# --- el sensor 'pegado' pasa TODAS las reglas: rachas de valores identicos por estacion ---
def racha_maxima(serie):
    cambio = serie.ne(serie.shift()).cumsum()
    return serie.groupby(cambio).transform("size")


curado_df["racha"] = curado_df.groupby("estacion", group_keys=False).humedad.apply(racha_maxima)
pegado = curado_df[curado_df.racha >= 6]
print(f"Filas en rachas de >= 6 valores identicos que SUPERARON la validacion de rango: {len(pegado)} (estaciones: {sorted(pegado.estacion.unique())})")
assert set(pegado.estacion) == {"E03"} and len(pegado) >= 8

fig, ax = plt.subplots(1, 3, figsize=(15, 4))
tabla_q.T.plot.bar(ax=ax[0], color=[PALETA[1], PALETA[2]], legend=True); ax[0].set_ylim(0.9, 1.04); ax[0].set_title("Calidad antes y después de procesar (el eje y empieza en 0.90)", fontsize=9, fontweight="bold"); ax[0].tick_params(axis="x", labelrotation=0)
ax[0].legend(fontsize=7, loc="upper center", ncol=2)
e03 = curado_df[curado_df.estacion == "E03"].set_index("ts").humedad
ax[1].plot(e03.index, e03.values, color=PALETA[0], lw=1); ax[1].plot(pegado.ts, pegado.humedad, "o", color=PALETA[1], ms=4, label="racha de valores idénticos")
ax[1].set_title("Estación E03: pasa la validación y está 'pegada'", fontsize=10, fontweight="bold"); ax[1].legend(fontsize=8); ax[1].tick_params(axis="x", labelrotation=30, labelsize=7)
pasos = bitacora.set_index("etapa").filas_salida
ax[2].barh(pasos.index[::-1], pasos.values[::-1], color=PALETA[0]); ax[2].set_title("Filas que salen de cada etapa (bitácora)", fontsize=10, fontweight="bold")
plt.tight_layout(); plt.show()
````

**Lectura honesta.** (1) En la zona cruda, la **completitud** baja a ≈ 99.1 % (humedades vacías), la **unicidad** a ≈ 97.1 % (duplicados de reintento) y la **validez** a ≈ 94.3 % (vacíos y valores fuera de rango); tras procesar, las tres valen **100 % por construcción**: **las métricas miden lo que las reglas atrapan, no la verdad**. (2) Por eso mismo el sensor **«pegado» de E03** (12 horas en 63.0; 9 de esas filas llegaron a la zona curada, las demás cayeron en duplicados o cuarentena por otros defectos) **sobrevivió** a todo: es un valor «válido», no vacío y no duplicado, y solo una regla **de dominio** distinta (rachas de valores idénticos) lo atrapa. Si ese sensor alimenta una alerta de plagas, **la alerta saldrá mal con datos «de calidad 100 %»**. (3) La **bitácora** permite responder, para cualquier cifra del análisis, *¿de cuántas filas viene, qué se descartó y por qué?*, y la **huella** permite detectar que **dos ejecuciones produjeron el mismo resultado** (reproducibilidad) o **no**.

## 4. Retención, Borrado y Protección de Datos Personales 🗑️🇨🇴

La **última etapa** es la más descuidada: conservar datos **«por si acaso»** parece gratis, pero **crece el riesgo** (una filtración expone más personas) y el costo (almacenamiento, respaldos, cumplimiento). Una **política de retención** responde: *¿cuánto tiempo se necesita cada conjunto, para qué finalidad, y qué se hace al vencer?*

**Marco colombiano (a nivel conceptual).** La **Ley 1581 de 2012** (régimen general de protección de datos personales, desarrollo del derecho de ***Habeas Data***) y sus decretos reglamentarios establecen, entre otros, principios como **finalidad** (los datos se usan para el fin informado), **libertad** (tratamiento con **autorización previa e informada** del titular, salvo excepciones), **veracidad o calidad**, **transparencia**, **acceso y circulación restringida**, **seguridad** y **confidencialidad**; y derechos del titular como **conocer, actualizar y rectificar** sus datos, **solicitar prueba de la autorización**, **revocarla** y **solicitar la supresión** cuando no se respeten los principios o las normas. La autoridad de control es la **Superintendencia de Industria y Comercio** *(todo lo anterior citado de memoria y de forma resumida: este cuaderno **no es asesoría jurídica**; consulta el texto oficial vigente y a un especialista antes de aplicarlo, y ten presente que otras normas pueden **obligar a conservar** ciertos datos por plazos determinados)*.

| Acción | Qué hace | ¿El dato sigue siendo personal? | Reversible |
|---|---|---|---|
| **Eliminar** (suprimir) | Borra el registro **de todas las zonas** (y de copias y respaldos) | No existe | No |
| **Anonimizar** | Quita o generaliza los identificadores de forma que **no se pueda re-identificar** a la persona | **No** (si está bien hecho; véase el [cuaderno 01](01_Importancia_del_Big_Data_en_el_Mundo_Actual.ipynb): «sin nombre» no basta) | No |
| **Seudonimizar** | Reemplaza el identificador por un **código** (alias) y guarda la tabla de equivalencias **aparte** y protegida | **Sí** (se puede re-vincular con la tabla) | Sí, con la tabla |
| **Archivar con acceso restringido** | Mueve a almacenamiento frío y limita quién puede leerlo | Sí | Sí |

La celda siguiente implementa un **motor de retención** sobre 400 titulares **ficticios** con tres reglas **ilustrativas** (los plazos los define la organización según su finalidad y la normativa; **no** son valores legales): (R1) **sin consentimiento registrado → eliminar**; (R2) con consentimiento y **más de 730 días desde el alta** → **anonimizar** (se quitan nombre y documento y se generaliza la fecha al año); (R3) con **obligación legal de conservación** → **conservar restringido** (no se toca, pero se marca). Verifica que **todo registro cae en exactamente una categoría**, que lo anonimizado **no conserva identificadores** y que **un identificador numérico corto «protegido» con un *hash* sin sal se revierte por fuerza bruta** (por qué seudonimizar ≠ anonimizar).

````python
# ============================================================
# 4. Politica de retencion: eliminar / anonimizar / conservar restringido, y por que un hash sin sal NO anonimiza
# ============================================================
def generar_titulares(n=400, semilla=SEED):
    """Titulares FICTICIOS (nombres y documentos inventados) con fecha de alta, consentimiento y marca de obligacion legal."""
    rng = np.random.default_rng(semilla)
    return pd.DataFrame({
        "id": np.arange(1, n + 1),
        "nombre": [f"Titular {i:03d}" for i in range(1, n + 1)],
        "documento": [f"DOC-{rng.integers(1_000_000, 9_999_999)}" for _ in range(n)],
        "municipio": rng.choice(["Tunja", "Duitama", "Sogamoso", "Paipa"], n),
        "fecha_alta": pd.Timestamp("2022-01-01") + pd.to_timedelta(rng.integers(0, 1150, n), unit="D"),
        "consentimiento": rng.random(n) < 0.95,
        "obligacion_legal": rng.random(n) < 0.05,
    })


RET_PERSONAL_DIAS = 730                                                           # ILUSTRATIVO: define la organizacion, no la ley
HOY_RET = pd.Timestamp("2025-06-30")


def aplicar_politica(df, hoy=HOY_RET, ret=RET_PERSONAL_DIAS):
    """Cada registro cae en EXACTAMENTE una categoria (R3 prevalece sobre R1 y R2; luego R1; luego R2)."""
    edad = (hoy - df.fecha_alta).dt.days
    restringido = df.obligacion_legal
    eliminar = ~restringido & ~df.consentimiento
    anonimizar = ~restringido & df.consentimiento & (edad > ret)
    activo = ~restringido & df.consentimiento & (edad <= ret)
    anon = df[anonimizar].drop(columns=["nombre", "documento", "id"]).assign(anio_alta=lambda d: d.fecha_alta.dt.year).drop(columns="fecha_alta")
    return {"activos": df[activo], "anonimizados": anon, "restringidos": df[restringido], "eliminados_ids": df.loc[eliminar, "id"].tolist()}


titulares = generar_titulares()
zonas_ret = aplicar_politica(titulares)
conteo = {k: (len(v)) for k, v in zonas_ret.items()}
print("Resultado de la politica:", conteo)
assert sum(conteo.values()) == len(titulares)                                      # cada registro en una sola categoria: nada se pierde ni se duplica
assert not ({"nombre", "documento", "id", "fecha_alta"} & set(zonas_ret["anonimizados"].columns))     # lo anonimizado no conserva identificadores directos
assert zonas_ret["restringidos"].obligacion_legal.all() and conteo["eliminados_ids"] > 0

# --- por que un hash sin sal de un identificador corto NO protege: se invierte por fuerza bruta ---
def seudonimo(i, sal=""):
    return hashlib.sha256((sal + str(i)).encode()).hexdigest()[:12]


ids_chicos = np.arange(100_000)                                                   # identificadores de 5 digitos (espacio muy pequeno)
tabla_inversa = {seudonimo(i): i for i in ids_chicos}                              # el atacante prueba TODOS los candidatos
alias = seudonimo(73_452)
print(f"\nAlias sin sal: {alias} -> recuperado por fuerza bruta: {tabla_inversa[alias]} (probo {len(ids_chicos):,} candidatos)")
assert tabla_inversa[alias] == 73_452
alias_sal = seudonimo(73_452, sal="secreto-largo-y-aleatorio-guardado-aparte")      # con una sal SECRETA el atacante ya no puede precalcular la tabla
assert alias_sal != alias and alias_sal not in tabla_inversa
print("Con una sal secreta (guardada aparte) la tabla de busqueda del atacante deja de servir; aun asi el alias sigue siendo un dato PERSONAL (seudonimizado, no anonimo).")
````

---
**Lectura.** El motor reparte los 400 titulares en categorías **disjuntas** (la suma coincide, verificada con `assert`): unos **activos**, otros **anonimizados** (sin nombre, documento ni identificador, y con la fecha generalizada al año), otros **eliminados** (sin consentimiento) y unos pocos **restringidos** por obligación legal. Dos advertencias: (1) **anonimizar bien es difícil**: quitar el nombre **no basta** (con municipio y año de alta y pocos registros, alguien puede ser único; véase el [k-anonimato](01_Importancia_del_Big_Data_en_el_Mundo_Actual.ipynb) del cuaderno 01), así que antes de publicar un conjunto «anonimizado» **se mide el riesgo**; (2) el *hash* sin sal de un identificador corto **se revierte en segundos** probando todos los candidatos: **seudonimizar** reduce la exposición, pero **el alias sigue siendo dato personal** y su tabla de equivalencias debe protegerse como tal. Y lo más difícil de **eliminar** no es la tabla principal, sino las **copias olvidadas**: respaldos, cachés, exportaciones a hojas de cálculo, entornos de pruebas. Un borrado **verificable** exige **inventariar dónde está cada dato** (catálogo y linaje).

## 5. Buenas Prácticas por Etapa ✅

| Etapa | Buenas prácticas |
|---|---|
| **Generación** | Define la **finalidad** *antes* de recoger; **minimiza** (solo lo necesario); registra **consentimiento** y origen; documenta las unidades y el significado de cada campo |
| **Ingesta** | Valida el formato al entrar; **aparta lo malformado a cuarentena** (nunca lo descartes en silencio); haz la ingesta **idempotente** (reintentar no duplica); registra cuántas filas entraron |
| **Almacenamiento** | Conserva la **zona cruda intacta** y publica una **zona curada**; usa formatos tipados y comprimidos (Parquet); **cifra** en reposo y en tránsito; separa **identificadores** de **mediciones**; mantén un **catálogo** con propietario y esquema |
| **Procesamiento** | **Reproducible** (mismo código + mismos datos = mismo resultado, verificable con una huella); **pruebas de calidad automáticas**; versiona el código y las reglas; **conserva las filas** (conservación verificable) |
| **Análisis** | Guarda el **linaje** de cada cifra; documenta supuestos y **sesgos**; revisa **veracidad** (reglas de dominio) además de las métricas genéricas; limita el acceso a lo necesario (**mínimo privilegio**) |
| **Archivado y eliminación** | Política de retención **escrita** con plazos y acciones; **automatiza** el vencimiento; **verifica el borrado** en todas las zonas y copias; deja **evidencia** (qué se borró y cuándo, sin conservar el dato) |
| **Transversales** | **Gobernanza**: un responsable (propietario y custodio) por conjunto de datos; **auditoría** de accesos; **respuesta a incidentes**; capacitación (Módulo 04) |

##### 🛠️ Práctica 1: El sensor «pegado»: una regla de veracidad

La validación de rango no detecta un sensor que repite el mismo valor. Con `curado_df` (lecturas curadas con columnas `estacion`, `ts`, `humedad`, `temperatura`) y la idea de **racha** (valores consecutivos idénticos):

1. Ordena por `estacion` y `ts` y, **por cada estación**, calcula la longitud de la racha de humedades idénticas a la que pertenece cada fila (puedes reutilizar `racha_maxima` con `groupby("estacion")`). Guarda esa columna como `racha_h`.

def suprimir_titular(id_titular, titulares, zonas):
    informe = {"bloqueado_por_obligacion_legal": False}
    en_restr = zonas["restringidos"].id == id_titular
    if en_restr.any():                                                                    # una obligacion legal de conservacion prevalece: se BLOQUEA el borrado
        informe["bloqueado_por_obligacion_legal"] = True
    titulares_sin = titulares[titulares.id != id_titular]
    activos_sin = zonas["activos"][zonas["activos"].id != id_titular]
    restr_sin = zonas["restringidos"] if informe["bloqueado_por_obligacion_legal"] else zonas["restringidos"][~en_restr]
    informe.update({"titulares": len(titulares) - len(titulares_sin), "activos": len(zonas["activos"]) - len(activos_sin),
                    "restringidos": len(zonas["restringidos"]) - len(restr_sin)})
    return titulares_sin, activos_sin, restr_sin, informe
