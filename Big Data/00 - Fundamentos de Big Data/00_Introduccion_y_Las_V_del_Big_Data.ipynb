## 1. ¿Por Qué Existe el Big Data? 🧱

Una computadora, por buena que sea, tiene **cuatro recursos finitos**. El Big Data aparece cuando alguno se convierte en el **cuello de botella** de un problema:

| Recurso | Límite típico | Síntoma cuando se agota |
|---|---|---|
| **Memoria (RAM)** | Los datos no caben | `MemoryError`, el sistema empieza a *paginar* a disco y todo se vuelve lentísimo |
| **Disco** | No cabe el archivo, o leerlo toma demasiado (ancho de banda de lectura) | Cargar el archivo tarda horas; un solo disco es un **punto único de falla** |
| **CPU** | Un núcleo (o unos pocos) no alcanzan para el cálculo | El proceso tarda días; el resultado llega cuando ya no sirve |
| **Red** | Mover los datos hacia el cómputo es más caro que calcular | Los datos «pesan» más que el programa: conviene **llevar el programa a los datos** |

**Definición operativa (la que usaremos):** *Big Data* son datos cuyo **volumen, velocidad de llegada o variedad** exceden lo que una **sola máquina con herramientas tradicionales** puede almacenar o procesar **en un tiempo útil**. Fíjate en lo que **no** dice: no fija un número (p. ej. «más de X terabytes»). Lo que era «masivo» para un servidor hace unos años cabe hoy en un portátil; **la frontera se mueve con el hardware**, y por eso lo correcto es razonar con **órdenes de magnitud y con el cuello de botella**, no con cifras mágicas.

**La idea central de la solución.** En vez de comprar una máquina cada vez más grande y cara (***scale-up***, escalar *hacia arriba*), se **reparte** el almacenamiento y el trabajo entre **muchas máquinas corrientes** (***scale-out***, escalar *hacia afuera*) y se **mueve el cómputo hacia donde están los datos**. Esa idea, popularizada por los artículos de Google sobre su sistema de archivos distribuido (GFS, 2003) y sobre MapReduce (2004), y por su implementación de código abierto (Hadoop) *(citados de memoria, verificar)*, es la que recorre todo el curso.

## 2. Volumen: ¿Cuánto es «mucho»? 📦

El **volumen** es la V más intuitiva: la cantidad de datos. Dos precisiones antes de calcular:

- **Unidades SI (base 1000) frente a binarias (base 1024).** Un fabricante vende «1 TB» pensando en $10^{12}$ bytes; el sistema operativo suele mostrar $1024^3$ por GB y por eso tu disco «pierde» espacio. Las unidades **binarias** tienen nombre propio: KiB, MiB, GiB, TiB (*kibi*, *mebi*, *gibi*, *tebi*).
- **Siempre hay que hablar de órdenes de magnitud**: cada escalón (KB → MB → GB → TB → PB) es **×1000**. Un correo breve es del orden de KB, una fotografía del orden de MB, una película del orden de GB, el disco de un computador de escritorio del orden de TB, y el almacenamiento de un gran centro de datos del orden de PB o más *(órdenes de magnitud aproximados; verificar para un caso concreto)*.

**Crecimiento.** Si un conjunto de datos crece una fracción constante $r$ por periodo, su tamaño sigue $V(t)=V_0\,(1+r)^t$, un **crecimiento exponencial**: se **duplica** cada $t_2=\dfrac{\ln 2}{\ln(1+r)}$ periodos. Y en un sistema distribuido hay que sumar la **replicación**: si cada bloque se guarda $k$ veces (en HDFS el valor habitual por defecto es $k=3$ *(citado de memoria, verificar)*), el almacenamiento **bruto** necesario es $k\,V$.

La celda siguiente construye la calculadora y la usa con una **empresa ficticia** (los parámetros son **supuestos ilustrativos, no datos reales**).

````python
# ============================================================
# 2. Calculadora de almacenamiento: unidades, crecimiento compuesto y replicacion
# ============================================================
SI = ["B", "KB", "MB", "GB", "TB", "PB", "EB", "ZB"]          # base 1000
BIN = ["B", "KiB", "MiB", "GiB", "TiB", "PiB", "EiB", "ZiB"]  # base 1024


def a_bytes(valor, unidad):
    """Convierte (valor, unidad) a bytes; acepta unidades SI (KB, MB...) o binarias (KiB, MiB...)."""
    if unidad in SI:
        return valor * 1000 ** SI.index(unidad)
    return valor * 1024 ** BIN.index(unidad)


def formatear_bytes(n, binario=False):
    """Bytes -> texto legible con la unidad adecuada (por defecto SI)."""
    base, nombres = (1024, BIN) if binario else (1000, SI)
    i = 0
    while n >= base and i < len(nombres) - 1:
        n /= base
        i += 1
    return f"{n:.2f} {nombres[i]}" if i else f"{int(n)} B"


def crecimiento_compuesto(v0, tasa, periodos):
    """V(t) = V0 (1 + r)^t, para t = 0..periodos."""
    return v0 * (1 + tasa) ** np.arange(periodos + 1)


def tiempo_de_duplicacion(tasa):
    return np.log(2) / np.log1p(tasa)


# --- Comprobaciones de la calculadora ---
assert a_bytes(1, "TB") == 10 ** 12 and a_bytes(1, "TiB") == 2 ** 40
assert formatear_bytes(1_500_000) == "1.50 MB" and formatear_bytes(2 ** 20, binario=True) == "1.00 MiB"
print(f"Un disco de '1 TB' (SI) son {formatear_bytes(a_bytes(1, 'TB'), binario=True)}: por eso el sistema operativo muestra menos de lo que prometia la caja.")
assert abs(a_bytes(1, "TB") / 2 ** 30 - 931.32) < 0.01

# Escala: cuantos registros de 1 KB caben en un disco de 1 TB, y cuanto pesaria guardarlos 'por duplicado'
registros = a_bytes(1, "TB") // a_bytes(1, "KB")
print(f"Registros de 1 KB en 1 TB: {registros:,} (mil millones = 10^9)")
assert registros == 10 ** 9

# --- Empresa ficticia (SUPUESTOS ilustrativos): hoy 20 TB, crece 3 %, 5 % u 8 % mensual; un servidor tiene 50 TB utiles; replicacion 3 ---
V0_TB, CAP_SERVIDOR_TB, K_REPLICAS, MESES = 20.0, 50.0, 3, 36
tasas = [0.03, 0.05, 0.08]
filas = []
fig, ax = plt.subplots(figsize=(9, 5))
for r, color in zip(tasas, PALETA):
    v = crecimiento_compuesto(V0_TB, r, MESES)
    t_cruce = np.log(CAP_SERVIDOR_TB / V0_TB) / np.log1p(r)                       # mes en que se supera la capacidad de UN servidor
    assert np.argmax(v >= CAP_SERVIDOR_TB) == int(np.ceil(t_cruce))             # la formula coincide con la simulacion mes a mes
    filas.append({"crecimiento mensual": f"{r:.0%}", "se duplica cada (meses)": round(tiempo_de_duplicacion(r), 1),
                  "supera 1 servidor en el mes": int(np.ceil(t_cruce)), "tamaño a 24 meses (TB)": round(v[24], 1),
                  "almacenamiento bruto a 24 m con 3 réplicas (TB)": round(K_REPLICAS * v[24], 1)})
    ax.plot(v, color=color, lw=2, label=f"{r:.0%} mensual")
ax.axhline(CAP_SERVIDOR_TB, color="black", ls="--", lw=1)
ax.text(0.5, CAP_SERVIDOR_TB * 1.05, "capacidad de UN servidor (supuesto: 50 TB)", fontsize=9)
ax.set_yscale("log"); ax.set_xlabel("meses"); ax.set_ylabel("TB (escala logarítmica)")
ax.set_title("Crecimiento exponencial: tarde o temprano un solo servidor no alcanza (datos ficticios)", fontweight="bold", fontsize=10)
ax.legend(); plt.tight_layout(); plt.show()
tabla_crec = pd.DataFrame(filas)
print(tabla_crec.to_string(index=False))
assert tabla_crec["se duplica cada (meses)"].is_monotonic_decreasing                # a mayor tasa, menos tiempo para duplicarse
assert abs(tiempo_de_duplicacion(0.05) - 14.2067) < 1e-3
````

**Lectura.** La calculadora reproduce el hecho cotidiano del disco «de 1 TB» (≈ 931 GiB) y confirma que 1 TB contiene $10^9$ registros de 1 KB. En la empresa ficticia, con un **crecimiento mensual del 5 %** el tamaño **se duplica cada ≈ 14 meses** y **supera la capacidad de un servidor** (50 TB) en el **mes 19**; con 8 % ocurre antes (mes 12) y con 3 % más tarde (mes 31). Con **replicación 3**, el almacenamiento bruto a 24 meses es **tres veces** el lógico. Dos lecciones: (1) el crecimiento exponencial **sorprende** (la curva parece tranquila al principio y de pronto no cabe); (2) la **replicación** que da tolerancia a fallos **multiplica el costo de almacenamiento**, una de las razones por las que se estudian códigos de borrado y compresión.

### 2.1. ¿Por qué *pandas* deja de alcanzar? 🐼

*pandas* guarda **todo el `DataFrame` en memoria**, y lo hace con un costo por celda mayor que el del archivo: las columnas de **texto** (`object`) almacenan un objeto de Python por celda. Medimos dos cosas con un archivo CSV **sintético** de transacciones:

1. **El perfil de memoria** del `DataFrame` completo frente al tamaño del archivo, y cuánto se ahorra con tipos más compactos (`category`, `float32`, `int32`).
2. **El pico de memoria** al calcular un agregado (suma y conteo de `monto` por `categoria`) **cargando todo** frente a **procesar por trozos** (`chunksize`): se lee un trozo, se agrega, se descarta y se **combinan los agregados parciales**. Esta es la **semilla de MapReduce**: *procesa cada parte por separado y combina los resultados parciales*.

> ⚠️ **Honestidad sobre la medición:** `tracemalloc` mide las asignaciones de Python y de NumPy, pero **no** la memoria interna del analizador de C de *pandas*. El pico real del proceso es mayor; la **tendencia** (lineal frente a plana) es lo que importa.

````python
# ============================================================
# 2.1 Datos sinteticos en CSV, perfil de memoria de pandas y procesamiento por trozos (chunksize)
# ============================================================
CIUDADES = ["Tunja", "Duitama", "Sogamoso", "Chiquinquirá", "Paipa", "Villa de Leyva", "Bogotá", "Medellín", "Cali", "Barranquilla", "Bucaramanga", "Cúcuta"]
CATEGORIAS = ["alimentos", "hogar", "ropa", "tecnologia", "salud", "transporte", "ocio", "educacion"]
CANALES = ["tienda", "web", "app"]


def generar_transacciones(n, semilla=SEED):
    """Transacciones SINTETICAS: monto lognormal con una tendencia suave en el tiempo (el monto medio sube ~50 % a lo largo del archivo)."""
    rng = np.random.default_rng(semilla)
    tendencia = 1 + 0.5 * np.arange(n) / n
    return pd.DataFrame({
        "id": np.arange(n),
        "cliente_id": rng.integers(0, 20_000, n),
        "ciudad": rng.choice(CIUDADES, n),
        "categoria": rng.choice(CATEGORIAS, n, p=[.30, .15, .12, .10, .10, .09, .08, .06]),
        "canal": rng.choice(CANALES, n, p=[.55, .30, .15]),
        "monto": np.round(rng.lognormal(3.5, 0.9, n) * tendencia, 2),
    })


TMP = tempfile.mkdtemp(prefix="bd00_")
atexit.register(shutil.rmtree, TMP, ignore_errors=True)                        # el CSV temporal se borra al cerrar el kernel
ruta_csv = os.path.join(TMP, "transacciones.csv")
generar_transacciones(N_FILAS).to_csv(ruta_csv, index=False)
tam_archivo = os.path.getsize(ruta_csv)
print(f"Archivo CSV sintetico: {N_FILAS:,} filas, {formatear_bytes(tam_archivo)} en disco")

# --- (a) Perfil de memoria del DataFrame completo ---
df = pd.read_csv(ruta_csv)
mem = df.memory_usage(deep=True)
tam_ram = int(mem.sum())
df_opt = df.assign(**{c: df[c].astype("category") for c in ["ciudad", "categoria", "canal"]}).astype({"id": "int32", "cliente_id": "int32", "monto": "float32"})
tam_opt = int(df_opt.memory_usage(deep=True).sum())
perfil = pd.DataFrame({"dtype": df.dtypes.astype(str), "MB en RAM": (mem.drop("Index") / 2 ** 20).round(1)})
print(perfil.to_string())
print(f"\nRAM del DataFrame: {formatear_bytes(tam_ram)}  ({tam_ram / tam_archivo:.1f} veces el archivo) | con tipos compactos: {formatear_bytes(tam_opt)}  (-{1 - tam_opt / tam_ram:.0%})")
assert mem.drop("Index")[["ciudad", "categoria", "canal"]].sum() > mem.drop("Index")[["id", "cliente_id", "monto"]].sum()   # el texto domina
assert tam_opt < 0.5 * tam_ram
del df, df_opt


# --- (b) Pico de memoria: cargar todo frente a procesar por trozos ---
def agregado_completo(ruta, nrows):
    d = pd.read_csv(ruta, nrows=nrows)
    return d.groupby("categoria")["monto"].agg(["sum", "count"])


def agregado_por_trozos(ruta, nrows, tam_trozo=25_000):
    """MAP: agrega cada trozo; REDUCE: suma los parciales (suma y conteo, nunca promedios de promedios)."""
    parcial = None
    for trozo in pd.read_csv(ruta, nrows=nrows, chunksize=tam_trozo):
        g = trozo.groupby("categoria")["monto"].agg(["sum", "count"])
        parcial = g if parcial is None else parcial.add(g, fill_value=0)
    return parcial


def pico_mb(fn, *args):
    tracemalloc.start()
    t0 = time.perf_counter()
    resultado = fn(*args)
    seg = time.perf_counter() - t0
    _, pico = tracemalloc.get_traced_memory()
    tracemalloc.stop()
    return resultado, pico / 2 ** 20, seg


filas = []
for n in [N_FILAS // 6, N_FILAS // 3, N_FILAS // 2, N_FILAS]:
    r_todo, p_todo, t_todo = pico_mb(agregado_completo, ruta_csv, n)
    r_trozos, p_trozos, t_trozos = pico_mb(agregado_por_trozos, ruta_csv, n)
    assert np.allclose(r_todo["sum"], r_trozos["sum"]) and (r_todo["count"].values == r_trozos["count"].values).all()   # mismo resultado
    filas.append({"filas": n, "pico cargando todo (MB)": p_todo, "pico por trozos (MB)": p_trozos, "segundos todo": t_todo, "segundos trozos": t_trozos})
medidas = pd.DataFrame(filas)
print(medidas.round(2).to_string(index=False))

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(medidas["filas"], medidas["pico cargando todo (MB)"], "o-", color=PALETA[1], lw=2, label="cargar todo con read_csv")
ax.plot(medidas["filas"], medidas["pico por trozos (MB)"], "s-", color=PALETA[2], lw=2, label="procesar por trozos de 25 000 filas")
ax.set_xlabel("filas leidas"); ax.set_ylabel("pico de memoria (MB, tracemalloc)")
ax.set_title("Cargar todo crece con los datos; por trozos, el pico queda plano", fontweight="bold", fontsize=10)
ax.legend(); plt.tight_layout(); plt.show()

pendiente = np.polyfit(medidas["filas"], medidas["pico cargando todo (MB)"], 1)[0]            # MB por fila
filas_1e9 = 1_000_000_000
print(f"Extrapolacion LINEAL (supuesto): ~{formatear_bytes(pendiente * filas_1e9 * 2 ** 20)} de RAM solo para cargar {filas_1e9:,} filas de este tipo; por trozos el pico seria el de un trozo.")
assert medidas["pico por trozos (MB)"].iloc[-1] < medidas["pico cargando todo (MB)"].iloc[-1]
assert medidas["pico por trozos (MB)"].max() < 2.0 * medidas["pico por trozos (MB)"].min()   # plano: no crece con las filas
````

**Lectura honesta.** (a) El archivo de 300 000 filas pesa ≈ 12 MB, pero el `DataFrame` ocupa **≈ 59 MB de RAM (≈ 4.7 veces el archivo)**: las tres columnas de texto aportan casi todo (un objeto de Python por celda). Con `category` y tipos de 32 bits baja a **≈ 4.5 MB (−92 %)**: *antes de escalar a un clúster, escala tu código*. (b) **Cargar todo** hace que el pico de memoria **crezca linealmente con las filas** (≈ 5 MB con 50 000 filas y ≈ 28 MB con 300 000), mientras que el procesamiento **por trozos** mantiene un pico **plano de ≈ 4 MB**: el resultado es **idéntico** (verificado con `assert`) pero la memoria deja de depender del tamaño. Extrapolar la pendiente a mil millones de filas da ≈ 93 GB solo para cargar (**supuesto lineal**, orden de magnitud). Y el costo, dicho sin adornos: por trozos fue **más lento** (≈ 0.7 s frente a ≈ 0.2 s con 300 000 filas, medido bajo `tracemalloc`): se paga tiempo por memoria. Otros límites: el procesamiento por trozos **sigue siendo una sola máquina** y un solo hilo, solo sirve para operaciones que se pueden **combinar** (sumas, conteos, máximos…; no una mediana exacta), y la lectura del disco sigue siendo secuencial. Cuando el disco o la CPU son el cuello de botella, ya hay que **repartir** entre máquinas (Módulo 01).

## 3. Velocidad: Datos que No Esperan ⚡

La **velocidad** tiene dos caras: **qué tan rápido llegan** los datos (eventos por segundo) y **qué tan rápido hay que responder** (latencia). Un *dashboard* de ventas que se actualiza cada noche y un sistema antifraude que debe decidir **mientras** el cliente paga tienen requisitos muy distintos. Vocabulario clave:

- **Ventana (*window*).** Los flujos no tienen fin: para agregar se cortan en ventanas de tiempo. **Tumbling** (sin solapamiento, cada evento en una sola ventana) o **sliding** (solapadas, p. ej. «últimos 30 s, recalculado cada 5 s»).
- **Tiempo de evento frente a tiempo de procesamiento.** Un evento ocurre en $t_e$ pero **llega** en $t_p=t_e+\text{retraso}$. Si agrupas por $t_p$, un pago hecho a las 23:59:59 que llega a las 00:00:03 cae en el día equivocado. Agrupar por $t_e$ exige esperar a los **eventos tardíos**: la **marca de agua** (*watermark*) dice «espero hasta $w$ segundos de retraso y luego cierro la ventana».
- **Contrapresión (*backlog*).** Si la tasa de llegada $\lambda$ supera la tasa de servicio $\mu$, **la cola crece sin límite**; con $\lambda<\mu$ promedio pero ráfagas, la cola se acumula durante la ráfaga y se drena después.

El experimento simula **10 minutos** de un flujo **sintético** de eventos (proceso de Poisson no homogéneo: tasa con una oscilación suave y una **ráfaga** de 30 s) y mide cada concepto.

````python
# ============================================================
# 3. Simulacion de un flujo de eventos: tasa, ventanas, latencia, eventos tardios y cola (backlog)
# ============================================================
rng = np.random.default_rng(SEED)
DUR = 600.0                                                                       # 10 minutos


def tasa(t, base=200.0):
    """Eventos por segundo: base * (oscilacion suave) * (rafaga x3 entre los segundos 300 y 330)."""
    return base * (1 + 0.5 * np.sin(2 * np.pi * t / DUR)) * np.where((t >= 300) & (t < 330), 3.0, 1.0)


# Proceso de Poisson no homogeneo por 'adelgazamiento' (thinning): se generan candidatos con la tasa maxima y se aceptan con prob. tasa(t)/tasa_max
TASA_MAX = 200.0 * 1.5 * 3.0
cand = np.sort(rng.uniform(0, DUR, rng.poisson(TASA_MAX * DUR)))
t_evento = cand[rng.uniform(0, 1, cand.size) < tasa(cand) / TASA_MAX]
n_ev = t_evento.size
# Retraso de red: casi todos < 1 s, pero el 5 % llega muy tarde (entre 5 y 20 s) => eventos desordenados
retraso = rng.lognormal(np.log(0.15), 0.6, n_ev)
tarde = rng.random(n_ev) < 0.05
retraso[tarde] = rng.uniform(5, 20, tarde.sum())
t_llegada = t_evento + retraso
print(f"Eventos generados: {n_ev:,} en {DUR:.0f} s ({n_ev / DUR:.0f} por segundo en promedio) | tardios (5-20 s de retraso): {tarde.mean():.1%}")

# --- (a) eventos por segundo y ventanas ---
por_seg = np.bincount(t_evento.astype(int), minlength=int(DUR))
W = 10
tumbling = np.bincount((t_evento // W).astype(int), minlength=int(DUR // W))      # ventanas fijas de 10 s
assert tumbling.sum() == n_ev and por_seg.sum() == n_ev                            # cada evento cae en UNA sola ventana
rejilla = np.arange(30, DUR + 1, 5.0)                                              # ventana deslizante: ultimos 30 s, cada 5 s
sliding = np.searchsorted(t_evento, rejilla, "right") - np.searchsorted(t_evento, rejilla - 30, "left")
assert (sliding == [por_seg[int(t - 30):int(t)].sum() for t in rejilla]).all()    # la ventana deslizante equivale a sumar 30 conteos por segundo

# --- (b) latencia segun la estrategia de procesamiento (la computacion tarda 0.5 s: SUPUESTO) ---
PROC = 0.5
filas = []
for nombre, B in [("lote cada 60 s", 60.0), ("lote cada 10 s", 10.0), ("micro-lote cada 1 s", 1.0)]:
    lat = np.ceil(t_llegada / B) * B + PROC - t_evento                           # el evento espera al cierre del lote y luego a la computacion
    filas.append({"estrategia": nombre, "latencia media (s)": lat.mean(), "p95 (s)": np.percentile(lat, 95), "maxima (s)": lat.max()})
    assert B / 2 * 0.8 < lat.mean() < B / 2 + PROC + 8                             # aprox. B/2 + retraso medio + computo (los tardios inflan la media)
lat_stream = retraso + 0.05                                                       # evento a evento: retraso de red + 50 ms de computo (supuesto)
filas.append({"estrategia": "evento a evento (streaming)", "latencia media (s)": lat_stream.mean(), "p95 (s)": np.percentile(lat_stream, 95), "maxima (s)": lat_stream.max()})
tabla_lat = pd.DataFrame(filas)
print(tabla_lat.round(2).to_string(index=False))
assert tabla_lat["latencia media (s)"].is_monotonic_decreasing

# --- (c) tiempo de evento vs tiempo de llegada, y marca de agua ---
v_evento = (t_evento // W).astype(int)
v_llegada = (t_llegada // W).astype(int)
desacuerdo = (v_evento != v_llegada).mean()
print(f"\nEventos que caen en una ventana de {W} s DISTINTA si se agrupa por llegada y no por tiempo de evento: {desacuerdo:.1%}")
perdida = {w: float((retraso > w).mean()) for w in [0, 1, 5, 20]}
print("Fraccion de eventos descartados segun la marca de agua w:", {w: f"{p:.1%}" for w, p in perdida.items()})
assert desacuerdo > 0 and perdida[0] > perdida[1] >= perdida[5] > perdida[20] == 0.0 and retraso.max() <= 20.0   # esperar mas retiene mas eventos, a costa de latencia

# --- (d) cola (backlog): llegan 'por_seg' eventos cada segundo y el sistema procesa mu por segundo ---
def cola(llegadas, mu):
    b, out = 0.0, []
    for a in llegadas:
        b = max(0.0, b + a - mu)                                                  # recursion de Lindley
        out.append(b)
    return np.array(out)


mu_pico, mu_medio, mu_bajo = por_seg.max() * 1.05, 0.8 * por_seg.max(), por_seg.mean() * 0.9
colas = {f"mu = {m:.0f}/s": cola(por_seg, m) for m in [mu_pico, mu_medio, mu_bajo]}
etiquetas = list(colas)
assert colas[etiquetas[0]].max() == 0                                              # capacidad >= pico: nunca hay cola
assert colas[etiquetas[2]][-1] > colas[etiquetas[1]][-1] and colas[etiquetas[2]].max() > 10_000   # por debajo del promedio la cola se acumula y no se drena a tiempo

fig, ax = plt.subplots(2, 2, figsize=(12, 7.5))
ax[0, 0].plot(por_seg, color=PALETA[0], lw=0.8); ax[0, 0].set_title("Eventos por segundo (con ráfaga)", fontsize=10, fontweight="bold"); ax[0, 0].set_xlabel("segundo")
ax[0, 1].bar(np.arange(len(tumbling)) * W, tumbling, width=W * 0.9, align="edge", color=PALETA[0], alpha=0.5, label=f"tumbling {W} s")
ax[0, 1].plot(rejilla - 15, sliding / 3, color=PALETA[1], lw=2, label="sliding 30 s / 3 (centrada)")
ax[0, 1].set_title("Ventanas: fijas y deslizantes", fontsize=10, fontweight="bold"); ax[0, 1].legend(fontsize=8); ax[0, 1].set_xlabel("segundo")
ax[1, 0].bar(tabla_lat["estrategia"].str.replace(" (streaming)", "", regex=False), tabla_lat["latencia media (s)"], color=[PALETA[1], PALETA[3], PALETA[2], PALETA[0]])
ax[1, 0].set_yscale("log"); ax[1, 0].set_title("Latencia media por estrategia (s, escala log)", fontsize=10, fontweight="bold"); ax[1, 0].tick_params(axis="x", labelrotation=15, labelsize=8)
for (k, v), c in zip(colas.items(), [PALETA[2], PALETA[3], PALETA[1]]):
    ax[1, 1].plot(v, color=c, lw=2, label=k)
ax[1, 1].set_title("Cola acumulada según la capacidad de procesamiento", fontsize=10, fontweight="bold"); ax[1, 1].set_xlabel("segundo"); ax[1, 1].set_ylabel("eventos en espera"); ax[1, 1].legend(fontsize=8)
plt.tight_layout(); plt.show()
print({k: int(v.max()) for k, v in colas.items()}, "<- cola maxima (eventos)")
````

---
**Lectura.** (a) Las ventanas *tumbling* conservan el total (cada evento en una sola ventana) y las *sliding* suavizan la serie; la ráfaga de 30 s es nítida en ambas. (b) La **latencia media** de un procesamiento por lotes es ≈ **mitad del lote** más el cómputo y el retraso: un lote de 60 s entrega información con **≈ 33 s** de edad en promedio, uno de 10 s con **≈ 6 s**, un micro-lote de 1 s con **≈ 1.8 s** y el *streaming* evento a evento con **≈ 0.9 s** (la media la inflan los eventos tardíos: su máximo sigue siendo ≈ 20 s). **Menor latencia no es gratis:** exige infraestructura que corra siempre, manejo de estado y de eventos tardíos. (c) Agrupar por **llegada** en vez de por **tiempo de evento** manda a otra ventana de 10 s a **≈ 6 %** de los eventos (los tardíos y los cercanos al borde); con una **marca de agua** de 1 s se siguen descartando ≈ 5 % (los tardíos de 5-20 s) y solo con **20 s** se capturan todos, **a cambio de esperar 20 s más**. (d) Si la capacidad cubre el pico (≈ 655 eventos/s) no hay cola; con una capacidad de ≈ 499/s la cola **aparece en la ráfaga (máximo ≈ 1 760 eventos) y se drena después**; con ≈ 197/s, por debajo del promedio (≈ 219/s), la cola **crece durante casi toda la prueba** (máximo ≈ 31 000) y al terminar los 10 minutos **aún quedan ≈ 13 000 eventos sin procesar**: esa cola es retraso acumulado. Es la cara operativa de la «V» de velocidad. *(Esto es el aperitivo: el Módulo 05 retoma el procesamiento de flujos y en Data Mining ya existe un cuaderno de [Spark Structured Streaming](../../Data%20Mining/07%20-%20Mineria%20de%20Datos%20con%20Big%20Data/04_Streaming_y_Mineria_en_Tiempo_Real.ipynb).)*

## 4. Variedad y las Demás «V» 🧩

La **variedad** es la heterogeneidad de las fuentes y los formatos. Un negocio típico recibe a la vez: tablas de su base de datos transaccional, eventos JSON de una aplicación móvil, lecturas XML o CSV de sensores, registros (*logs*) de servidores, texto libre de opiniones de clientes, imágenes y audios. Cada fuente trae **su propio esquema, su propia calidad y su propio ritmo**. El trabajo del ingeniero de datos es **armonizarlas** en una representación común (el detalle de cada tipo está en el [cuaderno 02](02_Tipos_de_Datos_Estructurados_No_Estructurados_Semiestructurados.ipynb)):

| Fuente | Quién la produce | Formato típico | Estructura |
|---|---|---|---|
| Ventas del punto de venta | Sistema transaccional | Tablas / CSV | **Estructurada** |
| Eventos de una aplicación | Aplicación móvil o web | JSON | **Semiestructurada** |
| Sensores de una estación | Dispositivos IoT | XML, CSV, JSON | Semiestructurada |
| Registros del servidor | Servidores web | Líneas de texto | Semiestructurada (patrón de línea) |
| Reseñas y redes sociales | Personas | Texto libre | **No estructurada** |
| Fotos, audio, video | Cámaras, micrófonos | Binarios | No estructurada |

**Las demás «V».** Laney (2001) propuso tres (volumen, velocidad, variedad); luego se añadieron otras. **No hay una lista canónica** (según el autor son 4, 5, 7 o más), así que conviene entender las más citadas como **preguntas de calidad y de negocio**, no como dogma:

| «V» | Pregunta que plantea | Ejemplo | Dónde la veremos |
|---|---|---|---|
| **Veracidad** (*veracity*) | ¿Puedo **confiar** en el dato? (ruido, duplicados, sesgos, datos faltantes) | Un sensor descalibrado reporta 140 % de humedad | Calidad en el [cuaderno 03](03_Ciclo_de_Vida_de_los_Datos.ipynb); gobernanza en el Módulo 04 |
| **Valor** (*value*) | ¿Qué **decisión** mejora? ¿Cuesta menos que lo que aporta? | Un modelo de demanda que reduce el desperdicio de producto fresco | [Cuaderno 01](01_Importancia_del_Big_Data_en_el_Mundo_Actual.ipynb) |
| **Variabilidad** | ¿Cambia el **significado** o el ritmo del dato en el tiempo? | Un campo cuyo código cambia de sentido tras una migración | Esquema en el [cuaderno 02](02_Tipos_de_Datos_Estructurados_No_Estructurados_Semiestructurados.ipynb) |
| **Visualización** | ¿Se puede **comunicar** sin engañar? | Un gráfico con ejes truncados | Módulo 03 |

> 🎯 **La V más importante es el valor:** sin una pregunta de negocio, el volumen es un **costo** (almacenamiento, seguridad, cumplimiento), no un activo.

La celda siguiente **armoniza cinco fuentes heterogéneas** (inventadas) en un único `DataFrame` con un esquema común.

````python
# ============================================================
# 4. Variedad: cinco fuentes heterogeneas (CSV, JSON, XML, log y texto) unificadas en un esquema comun
# ============================================================
import csv
import io
import json
import re
import xml.etree.ElementTree as ET

fuente_csv = "id_venta,fecha,producto,valor\n1,2025-03-01,papa criolla,12500\n2,2025-03-01,cebolla larga,4300\n"
fuente_json = '{"evento": "compra", "usuario": {"id": 7, "ciudad": "Tunja"}, "items": [{"sku": "P1", "precio": 4300}, {"sku": "P2", "precio": 8100}], "ts": "2025-03-01T10:15:00"}'
fuente_xml = "<lectura estacion='E3'><hora>2025-03-01T10:20:00</hora><humedad>63.5</humedad><temperatura>11.2</temperatura></lectura>"
fuente_log = '192.168.0.7 - - [01/Mar/2025:10:16:02 -0500] "GET /tienda/papa HTTP/1.1" 200 512'
fuente_txt = "Excelente la papa, llegó fresca pero el empaque venía roto"

registros = []                                                                     # esquema comun: fuente, instante, entidad, medida, valor
for fila in csv.DictReader(io.StringIO(fuente_csv)):                              # (1) estructurada: cada fila ya trae columnas
    registros.append({"fuente": "csv_ventas", "instante": fila["fecha"], "entidad": fila["producto"], "medida": "valor_venta", "valor": float(fila["valor"])})
ev = json.loads(fuente_json)                                                       # (2) semiestructurada: anidada, con listas
registros.append({"fuente": "json_app", "instante": ev["ts"], "entidad": f"usuario {ev['usuario']['id']} ({ev['usuario']['ciudad']})", "medida": "total_compra", "valor": float(sum(i["precio"] for i in ev["items"]))})
raiz = ET.fromstring(fuente_xml)                                                   # (3) semiestructurada: etiquetas y atributos
registros.append({"fuente": "xml_sensor", "instante": raiz.findtext("hora"), "entidad": raiz.get("estacion"), "medida": "humedad_%", "valor": float(raiz.findtext("humedad"))})
m = re.match(r'(\S+) \S+ \S+ \[([^\]]+)\] "(\w+) (\S+) [^"]+" (\d{3}) (\d+)', fuente_log)  # (4) log: hay que 'descubrir' el patron con una expresion regular
ip, ts_log, metodo, ruta, estado, nbytes = m.groups()
registros.append({"fuente": "log_servidor", "instante": pd.to_datetime(ts_log, format="%d/%b/%Y:%H:%M:%S %z").tz_localize(None).isoformat(), "entidad": f"{metodo} {ruta}", "medida": "bytes", "valor": float(nbytes)})
palabras = re.findall(r"\w+", fuente_txt.lower())                                   # (5) no estructurada: hay que INFERIR algo (aqui, solo contar palabras)
registros.append({"fuente": "texto_resena", "instante": None, "entidad": "resena", "medida": "n_palabras", "valor": float(len(palabras))})

unificado = pd.DataFrame(registros)
unificado["instante"] = pd.to_datetime(unificado["instante"], format="ISO8601")
print(unificado.to_string(index=False))
assert unificado["fuente"].nunique() == 5 and unificado["valor"].notna().all()
assert unificado["instante"].notna().sum() == 5 and unificado["instante"].isna().sum() == 1   # el texto libre NO trae marca de tiempo: la variedad tambien es calidad desigual
assert unificado.loc[unificado.fuente == "json_app", "valor"].iloc[0] == 12400.0   # 4300 + 8100 (la lista anidada se agrego)
````

## 5. Escalar Hacia Arriba o Hacia Afuera: *Scale-up* frente a *Scale-out* 🏗️

| Criterio | ***Scale-up*** (vertical) | ***Scale-out*** (horizontal) |
|---|---|---|
| Idea | Una máquina **más grande** (más RAM, más núcleos, discos más rápidos) | **Más máquinas** corrientes trabajando juntas |
| Ventaja | Simple: **no cambia el programa**; sin red entre nodos | Crece casi sin tope; usa hardware de bajo costo; **tolera fallos** replicando |
| Desventaja | Tope físico y de precio (el costo sube más rápido que la capacidad); **punto único de falla** | Programación **más difícil** (datos repartidos, comunicación por red, fallos parciales) |
| Cuándo | Cuando el problema **cabe** en una máquina grande | Cuando el volumen, la disponibilidad o el costo lo exigen |

> 🧠 **Regla práctica (y honesta):** escala hacia arriba **mientras puedas**. Un clúster es costoso de operar; muchos problemas que parecen «Big Data» caben en una máquina bien usada (por ejemplo, con el procesamiento por trozos de arriba).

**¿Cuánto se gana al repartir? La ley de Amdahl.** Si una fracción $f$ del trabajo **se puede paralelizar** y la fracción $(1-f)$ **es secuencial** (leer la entrada, repartir, combinar), con $p$ procesadores la aceleración máxima es

$$S(p)=\frac{1}{(1-f)+\dfrac{f}{p}},\qquad S(\infty)=\frac{1}{1-f}.$$

La parte secuencial **fija un techo** sin importar cuántas máquinas añadas: con $f=0.95$ el techo es $20\times$. (La visión complementaria de Gustafson —cuando el problema *crece* con los procesadores— se cita en los recursos.) En la práctica hay **más costos** que la ley no modela: arrancar procesos, **serializar y copiar** datos entre ellos, esperar al más lento.

**Más máquinas, más fallos.** Si cada máquina falla con probabilidad $q$ en un periodo, la probabilidad de que **al menos una** de $N$ falle es $1-(1-q)^N$: con $q$ pequeña, en un clúster grande los fallos **son la norma, no la excepción**. Por eso los sistemas distribuidos **replican** los datos y **reintentan** tareas.

La celda siguiente (a) calcula la ley de Amdahl y la verifica, (b) verifica la probabilidad de fallo con una simulación y (c) hace un **experimento real**: repartir un cálculo CPU-bound (pasos de la **conjetura de Collatz** sobre números sintéticos) entre $p$ procesos con `multiprocessing`, comprobar que el resultado es el mismo y **ajustar la fracción paralelizable efectiva**.

````python
# ============================================================
# 5. Ley de Amdahl, probabilidad de fallos y un reparto REAL de trabajo con multiprocessing
# ============================================================
def amdahl(f, p):
    """Aceleracion maxima teorica con fraccion paralelizable f y p procesadores."""
    return 1.0 / ((1.0 - f) + f / np.asarray(p, dtype=float))


# --- (a) Amdahl: tabla y grafico ---
ps = np.array([1, 2, 4, 8, 16, 64, 1024])
tabla_amdahl = pd.DataFrame({f"f = {f}": np.round(amdahl(f, ps), 2) for f in [0.5, 0.9, 0.95, 0.99]}, index=ps)
tabla_amdahl.index.name = "p"
print(tabla_amdahl.to_string())
for f in [0.5, 0.9, 0.95, 0.99]:
    s = amdahl(f, ps)
    assert abs(s[0] - 1.0) < 1e-12 and (np.diff(s) > 0).all() and s[-1] < 1 / (1 - f)          # S(1) = 1, creciente y siempre BAJO el techo
    assert abs(amdahl(f, 10 ** 9) - 1 / (1 - f)) < 1e-3                                          # S(p -> infinito) = 1/(1-f)

# --- (b) Fallos: P(al menos una falla) = 1 - (1 - q)^N, verificada por simulacion (q = 0.1 % por dia: SUPUESTO ilustrativo) ---
q = 0.001
rng_f = np.random.default_rng(SEED)
filas = []
for N in [1, 10, 100, 1000]:
    teorica = 1 - (1 - q) ** N
    simulada = (rng_f.random((20_000, N)) < q).any(axis=1).mean()
    assert abs(teorica - simulada) < 0.015
    filas.append({"maquinas N": N, "P(al menos una falla) teorica": round(teorica, 3), "simulada": round(simulada, 3)})
print("\n" + pd.DataFrame(filas).to_string(index=False))


# --- (c) Experimento real: repartir pasos de Collatz entre p procesos ---
def pasos_collatz(n):
    s = 0
    while n != 1:
        n = n // 2 if n % 2 == 0 else 3 * n + 1
        s += 1
    return s


def trabajar(bloque):
    return [pasos_collatz(int(x)) for x in bloque]


numeros = np.random.default_rng(SEED).integers(1, 3_000_000, N_NUMEROS)
t0 = time.perf_counter()
ref = np.array(trabajar(numeros))                                                 # version secuencial (p = 1, sin proceso hijo)
t_seq = time.perf_counter() - t0
HAY_FORK = "fork" in mp.get_all_start_methods()                                    # en Windows/macOS se usa 'spawn' y las funciones de un cuaderno no se pueden enviar

medido = {}
if HAY_FORK:
    ctx = mp.get_context("fork")
    for p in [2, 4, 8]:
        if p > 2 * (os.cpu_count() or 1):
            continue
        t0 = time.perf_counter()
        with ctx.Pool(p) as pool:                                                  # incluye arrancar los procesos (parte SECUENCIAL)
            partes = pool.map(trabajar, np.array_split(numeros, 4 * p))          # MAP: 4p bloques repartidos entre p procesos (equilibra la carga)
        res = np.concatenate([np.array(x) for x in partes])                       # REDUCE: se concatenan los resultados
        medido[p] = time.perf_counter() - t0
        assert (res == ref).all()                                                  # el resultado paralelo es IDENTICO al secuencial
    aceler = {p: t_seq / t for p, t in medido.items()}
    # fraccion paralelizable efectiva: la f que mejor reproduce las aceleraciones medidas (minimos cuadrados sobre una rejilla)
    rej = np.linspace(0, 1, 1001)
    f_ef = rej[np.argmin([sum((amdahl(f, p) - s) ** 2 for p, s in aceler.items()) for f in rej])]
    print(f"\nSecuencial: {t_seq:.2f} s | " + " | ".join(f"p={p}: {t:.2f} s (x{aceler[p]:.2f})" for p, t in medido.items()))
    print(f"Fraccion paralelizable efectiva (ajuste de Amdahl): f = {f_ef:.3f}  ->  techo teorico {1 / (1 - f_ef) if f_ef < 1 else float('inf'):.1f}x")
    assert 0.0 < f_ef <= 1.0

    ps_m = np.array([1] + list(medido))
    fig, ax = plt.subplots(figsize=(7.5, 4.5))
    ax.plot(ps_m, ps_m, "k--", lw=1, label="ideal (lineal)")
    ax.plot(ps_m, amdahl(f_ef, ps_m), color=PALETA[1], lw=2, label=f"Amdahl con f = {f_ef:.2f}")
    ax.plot(ps_m, [1.0] + [aceler[p] for p in medido], "o-", color=PALETA[0], lw=2, label="medido")
    ax.set_xlabel("procesos p"); ax.set_ylabel("aceleración respecto a 1 proceso")
    ax.set_title("Repartir el trabajo acelera, pero por debajo del ideal", fontweight="bold", fontsize=10)
    ax.legend(); plt.tight_layout(); plt.show()
else:
    print("Este sistema no soporta 'fork' (p. ej. Windows): se omite el experimento de multiprocessing; la ley de Amdahl de arriba si aplica.")
````

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Por qué existe** | Una máquina tiene **RAM, disco, CPU y red finitos**; Big Data = datos cuyo volumen, velocidad o variedad exceden lo que **una máquina con herramientas tradicionales** procesa en **tiempo útil**. La frontera **se mueve** con el hardware |
| **Solución de fondo** | ***Scale-out***: repartir datos y trabajo entre muchas máquinas corrientes y **llevar el cómputo a los datos** (GFS, MapReduce, Hadoop) |
| **Volumen** | Unidades **SI (×1000) frente a binarias (×1024)**; crecimiento **exponencial** $V_0(1+r)^t$; la **replicación** (×3) multiplica el almacenamiento bruto. *pandas* ocupa **más RAM que el archivo**; por **trozos** el pico queda **plano** con resultado idéntico |
| **Velocidad** | Eventos/s, **ventanas** (*tumbling*/*sliding*), **latencia** ≈ mitad del lote + cómputo, **tiempo de evento frente a llegada** y **marca de agua**; si $\lambda>\mu$ la **cola crece** |
| **Variedad** | CSV, JSON, XML, *logs*, texto, imágenes: cada fuente exige **su propia lectura** y trae calidad desigual; se armoniza en un **esquema común** |
| **Otras «V»** | **Veracidad** (¿confío?) y **valor** (¿mejora una decisión?); no hay lista canónica de «V»: el valor es lo que justifica el costo |
| ***Scale-up* frente a *scale-out*** | Arriba mientras quepa; afuera cuando el volumen, la disponibilidad o el costo lo exijan. **Amdahl:** $S(p)=1/((1-f)+f/p)$, con techo $1/(1-f)$; la aceleración medida queda **bajo el ideal**. Más máquinas ⇒ **más fallos** ($1-(1-q)^N$) |
