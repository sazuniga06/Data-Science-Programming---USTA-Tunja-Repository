## 2. Los Cuatro Niveles de Analítica 🔍

Los mismos datos pueden contestar **cuatro preguntas distintas**, en orden creciente de valor y de dificultad:

| Nivel | Pregunta | Técnicas típicas | Ejemplo (tienda de víveres) | Salida |
|---|---|---|---|---|
| **Descriptiva** | ¿Qué pasó? | Agregaciones, tablas, gráficos, indicadores | Ventas por mes y por sede | Un resumen |
| **Diagnóstica** | ¿Por qué pasó? | Desagregar (*drill-down*), contribuciones, comparación entre grupos, correlación | La caída del último trimestre viene de **una sede** y de los días **sin** promoción | Una explicación (con cuidado: **correlación no es causa**) |
| **Predictiva** | ¿Qué pasará? | Regresión, series de tiempo, clasificación (aprendizaje automático) | Ventas de **mañana** por sede | Una cifra con su incertidumbre |
| **Prescriptiva** | ¿Qué debo hacer? | Optimización, simulación, decisión bajo incertidumbre | **Cuánto pedir** para minimizar el costo esperado | Una decisión |

Cada nivel **se apoya en el anterior**: no se predice bien lo que no se describió, ni se prescribe sin un pronóstico. Y aunque a mayor nivel mayor valor potencial, **no siempre** el nivel más alto es el que se necesita: muchas organizaciones ganan lo más con un buen tablero descriptivo.

**Un mismo conjunto de datos.** Una tienda de víveres **ficticia** (*«La Esquina»*) con tres sedes (Tunja, Duitama y Sogamoso) vende **tomate** durante 730 días. Los datos son **sintéticos**: hay efecto del día de la semana (más ventas el sábado), promociones y una **tendencia suave**; y en **Sogamoso**, desde el día 640, abre un competidor que **recorta las ventas de los días sin promoción** (el «secreto» que la diagnóstica debe encontrar).

````python
# ============================================================
# 2.a  Datos sinteticos + DESCRIPTIVA + DIAGNOSTICA
# ============================================================
SEDES = {"Tunja": 260, "Duitama": 190, "Sogamoso": 170}              # kg/dia base
EFECTO_DIA = np.array([0.00, -0.02, 0.00, 0.03, 0.12, 0.30, 0.18])   # lun..dom (el sabado hay plaza)
DIA_COMPETIDOR = 640


def generar_ventas(semilla=SEED, dias=730):
    rng = np.random.default_rng(semilla)
    fechas = pd.date_range("2024-01-01", periods=dias, freq="D")
    filas = []
    for sede, base in SEDES.items():
        promo = (rng.random(dias) < 0.18).astype(int)
        t = np.arange(dias)
        mu = base * (1 + 0.0003 * t) * (1 + EFECTO_DIA[fechas.dayofweek.values]) * (1 + 0.20 * promo)
        if sede == "Sogamoso":
            mu = np.where(t >= DIA_COMPETIDOR, mu * np.where(promo == 1, 0.95, 0.70), mu)    # competidor: golpea los dias SIN promo
        filas.append(pd.DataFrame({"fecha": fechas, "t": t, "sede": sede, "promo": promo, "kg": rng.poisson(mu)}))
    return pd.concat(filas, ignore_index=True)


ventas = generar_ventas()
assert ventas.shape == (3 * 730, 5) and ventas.kg.min() >= 0

# ---- DESCRIPTIVA: que paso --------------------------------------------------------------
ventas["mes"] = ventas.fecha.dt.to_period("M")
mensual = ventas.pivot_table(index="mes", columns="sede", values="kg", aggfunc="sum")
print("DESCRIPTIVA - kg vendidos por mes (ultimos 6 meses):")
print(mensual.tail(6).to_string())
ult, prev = ventas[ventas.t >= 730 - 90], ventas[(ventas.t >= 730 - 180) & (ventas.t < 730 - 90)]
cambio_total = ult.kg.sum() / prev.kg.sum() - 1
print(f"\nUltimos 90 dias vs 90 anteriores: {ult.kg.sum():,} vs {prev.kg.sum():,} kg  ({cambio_total:+.1%})")

# ---- DIAGNOSTICA: por que paso ------------------------------------------------------------
delta = (ult.groupby("sede").kg.sum() - prev.groupby("sede").kg.sum()).sort_values()
print("\nDIAGNOSTICA - variacion en kg por sede:\n" + delta.to_string())
sog = ventas[ventas.sede == "Sogamoso"]
antes_s, desp_s = sog[sog.t < DIA_COMPETIDOR], sog[sog.t >= DIA_COMPETIDOR]
por_promo = pd.DataFrame({"antes (kg/dia)": antes_s.groupby("promo").kg.mean(), "despues (kg/dia)": desp_s.groupby("promo").kg.mean()})
por_promo["cambio"] = por_promo["despues (kg/dia)"] / por_promo["antes (kg/dia)"] - 1
print("\nSogamoso, antes y despues del dia 640, segun haya promocion (0 = sin, 1 = con):\n" + por_promo.round(3).to_string())

assert delta.index[0] == "Sogamoso" and (delta.iloc[1:] > 0).all()                                    # solo Sogamoso cae; las otras crecen
assert por_promo.loc[0, "cambio"] < -0.20 and por_promo.loc[1, "cambio"] > -0.15                      # la caida esta en los dias SIN promocion
````

````python
# ============================================================
# 2.b  PREDICTIVA (pronostico a un dia) + PRESCRIPTIVA (cuanto pedir)
# ============================================================
from sklearn.ensemble import HistGradientBoostingRegressor

d = ventas.sort_values(["sede", "t"]).copy()
d["dow"] = d.fecha.dt.dayofweek
d["media7"] = d.groupby("sede").kg.transform(lambda s: s.shift(1).rolling(7).mean())               # SOLO el pasado (sin fuga)
d["lag7"] = d.groupby("sede").kg.shift(7)                                                          # igual dia de la semana pasada = linea base ingenua
d["sede_id"] = d.sede.map({s: i for i, s in enumerate(SEDES)})
d = d.dropna().reset_index(drop=True)
FEATS = ["sede_id", "dow", "promo"]                                  # el modelo predice el COCIENTE kg / media7 (se adapta a cambios de nivel)

# division CRONOLOGICA en tres: entrenar / calibrar la incertidumbre / probar
entr, cal, prueba = d[d.t < 610], d[(d.t >= 610) & (d.t < 702)], d[d.t >= 702]
modelo = HistGradientBoostingRegressor(max_iter=150, learning_rate=0.08, categorical_features=[0, 1], random_state=SEED).fit(entr[FEATS], entr.kg / entr.media7)
pred_cal, pred_test = cal.media7.values * modelo.predict(cal[FEATS]), prueba.media7.values * modelo.predict(prueba[FEATS])
mae = lambda y, p: float(np.mean(np.abs(y - p)))
res_pred, res_ing = mae(prueba.kg, pred_test), mae(prueba.kg, prueba.lag7)
print(f"PREDICTIVA - MAE en los ultimos 28 dias (kg/dia/sede): modelo = {res_pred:.1f} | ingenuo (mismo dia, semana pasada) = {res_ing:.1f}")
print("  MAE por sede:", {s: round(mae(g.kg, pred_test[prueba.sede.values == s]), 1) for s, g in prueba.groupby("sede")})
assert np.isfinite(res_pred) and np.isfinite(res_ing)

# ---- PRESCRIPTIVA: cuanto pedir (problema del 'vendedor de periodicos') ---------------------
PRECIO_VENTA, COSTO_COMPRA = 2800, 1900                     # COP/kg, ilustrativos (inventados)
res_cal = cal.kg.values - pred_cal                           # incertidumbre medida en el tramo de calibracion


def costo_total(q, y, cu, co):
    """Costo de pedir q cuando la demanda fue y: cu por kg de venta perdida (margen), co por kg que sobra (se pierde la compra)."""
    return float(np.sum(cu * np.maximum(y - q, 0) + co * np.maximum(q - y, 0)))


cu, co = PRECIO_VENTA - COSTO_COMPRA, COSTO_COMPRA
razon_critica = cu / (cu + co)                               # fraccion de la demanda que conviene cubrir
q_media = pred_test                                           # pedir lo que se espera vender
q_opt = pred_test + np.quantile(res_cal, razon_critica)       # pedir el cuantil critico de la demanda
c_media, c_opt = costo_total(q_media, prueba.kg.values, cu, co), costo_total(q_opt, prueba.kg.values, cu, co)
print(f"\nPRESCRIPTIVA - razon critica cu/(cu+co) = {razon_critica:.3f}  (pedir un poco MENOS que el pronostico)")
print(f"  costo en 28 dias x 3 sedes: pedir el pronostico = {c_media:,.0f} COP | pedir el cuantil optimo = {c_opt:,.0f} COP ({c_opt / c_media - 1:+.1%})")
assert c_opt <= c_media                                       # con costos asimetricos, el pronostico NO es la mejor decision
````

**Lectura.** (a) **Descriptiva y diagnóstica.** En los últimos 90 días se vendieron **70 603 kg frente a 73 696** de los 90 anteriores (**−4.2 %**). La descriptiva dice *que* bajó; la diagnóstica dice *dónde*: **solo Sogamoso cae** (−4 443 kg) mientras Duitama (+513) y Tunja (+837) crecen, y dentro de Sogamoso la caída está en los **días sin promoción** (de ≈ 201 a ≈ 157 kg/día, **−21.8 %**), mientras que con promoción las ventas **no caen** (≈ 247 → ≈ 255 kg/día). Con esta pista el negocio puede **probar una hipótesis** (¿llegó un competidor?), pero la diagnóstica **sugiere, no demuestra causas**. (b) **Predictiva.** Un modelo de *gradient boosting* que predice el **cociente** entre la venta y la media de los 7 días previos tiene un error absoluto medio de **≈ 16.9 kg/día por sede** en los últimos 28 días, frente a **≈ 31.2** de la línea base ingenua (repetir el mismo día de la semana pasada, que arrastra el ruido de esa semana y su estado de promoción). Por sede el error es **mayor en Sogamoso (≈ 21.3)** que en Duitama (≈ 13.3) o Tunja (≈ 16.1): el modelo se entrenó **antes** del cambio de régimen y se adapta solo a través de la media reciente. Por eso se predice un **cociente sobre el pasado reciente** y no el nivel absoluto: es una forma barata de ser menos frágil ante cambios de nivel, aunque no los elimina. (c) **Prescriptiva:** el pronóstico **no es la decisión**. el pronóstico **no es la decisión**. ## 3. Escalar la Analítica: Muestrear, Resumir, Repartir 📏

Cuando los datos **no caben** o **no alcanza el tiempo**, hay tres salidas (y se combinan):

| Estrategia | Idea | Cuándo sirve | Costo |
|---|---|---|---|
| **Muestrear** | Analizar una muestra aleatoria | Estimar medias, proporciones, tendencias | Error $\propto 1/\sqrt{n}$; **eventos raros** se pierden |
| **Resumir con estructuras aproximadas** (*sketches*) | Un resumen **de tamaño fijo** del flujo que responde una pregunta concreta | Conteo de frecuencias, de elementos distintos, pertenencia | Error **acotado** pero distinto de cero; solo responde **esa** pregunta |
| **Repartir** | Dividir los datos y combinar resultados parciales (*MapReduce*, *Spark*) | Cuando se necesita el resultado **exacto** | Infraestructura, comunicación, fallos (Módulo 00 y 01) |

**Muestrear.** Si la población tiene media $\mu$ y desviación $\sigma$, la media de una muestra de $n$ elementos tiene error típico
$$\mathrm{SE}(\bar{x}) = \frac{\sigma}{\sqrt{n}},$$
que **no depende del tamaño de la población** $N$ (si $n \ll N$). Para estimar con un error relativo $e$ y confianza del 95 % hacen falta $n \approx (1.96\,\sigma/(e\,\mu))^2$ elementos: **la misma muestra** sirve para 10 000 o para 10 000 millones de filas. Lo que **no** se logra con una muestra uniforme es ver los **eventos raros** (se verá en el cuaderno de visualización).

**Muestreo de reservorio (*Algorithm R*).** ¿Cómo tomar una muestra uniforme de $k$ elementos de un flujo **de longitud desconocida** que no cabe en memoria? Se guardan los primeros $k$; al llegar el elemento número $i > k$ (contando desde 1), se **sustituye** uno al azar del reservorio con probabilidad $k/i$. Por inducción, **cada elemento** vistos hasta ahora queda en el reservorio con probabilidad $k/N$ al terminar, usando **memoria $O(k)$** y **una sola pasada**.

````python
# ============================================================
# 3.a  Error de muestreo ~ 1/sqrt(n)   3.b  Muestreo de reservorio (desde cero)
# ============================================================
rng = np.random.default_rng(SEED)
N_POB = 1_000_000 if RAPIDO else 4_000_000
poblacion = rng.lognormal(mean=8.0, sigma=1.0, size=N_POB)            # 'montos' sinteticos, muy asimetricos
mu, sigma = poblacion.mean(), poblacion.std()

tamanos, REPS = [100, 300, 1_000, 3_000, 10_000, 30_000], 300
obs = []
for n in tamanos:
    medias = np.array([poblacion[rng.integers(0, N_POB, n)].mean() for _ in range(REPS)])
    obs.append(np.sqrt(np.mean((medias - mu) ** 2)))
teo = sigma / np.sqrt(np.array(tamanos))
tabla = pd.DataFrame({"n": tamanos, "error observado (RMSE)": obs, "teoria sigma/sqrt(n)": teo, "razon": obs / teo}).set_index("n")
print(tabla.round(2).to_string())
pendiente = np.polyfit(np.log(tamanos), np.log(obs), 1)[0]
print(f"\nPendiente log-log del error frente a n: {pendiente:.3f}  (teoria: -0.5)")
n_95_1pct = int(np.ceil((1.96 * sigma / (0.01 * mu)) ** 2))
print(f"Muestra para error relativo <= 1% al 95%: n = {n_95_1pct:,} (sobre una poblacion de {N_POB:,}; con N de 10 000 millones seria el mismo n)")
assert ((tabla["razon"] > 0.8) & (tabla["razon"] < 1.2)).all()         # el error observado sigue a la teoria
assert -0.57 < pendiente < -0.43


def muestreo_reservorio(flujo, k, rng):
    """Muestra uniforme de k elementos de un iterable de longitud desconocida: una pasada, memoria O(k)."""
    reservorio = []
    for i, x in enumerate(flujo, start=1):                             # i = numero de elementos vistos (desde 1)
        if i <= k:
            reservorio.append(x)
        elif rng.random() < k / i:                                      # el elemento i entra con probabilidad k/i...
            reservorio[rng.integers(0, k)] = x                          # ...y desplaza a uno al azar
    return reservorio


# Verificacion de uniformidad: N = 50 elementos, k = 10 -> cada uno debe quedar con probabilidad k/N = 0.2
N_, K_, ENSAYOS = 50, 10, 20_000
cuenta = np.zeros(N_)
for _ in range(ENSAYOS):
    cuenta[muestreo_reservorio(range(N_), K_, rng)] += 1
esperado, sd = ENSAYOS * K_ / N_, np.sqrt(ENSAYOS * (K_ / N_) * (1 - K_ / N_))
print(f"\nReservorio N={N_}, k={K_}: cada elemento deberia salir {esperado:.0f} +/- {sd:.0f} veces; observado min = {cuenta.min():.0f}, max = {cuenta.max():.0f}")
assert np.all(np.abs(cuenta - esperado) < 5 * sd)                       # ningun elemento esta favorecido

# Flujo grande generado 'al vuelo': nunca existe completo en memoria
def flujo_montos(n, semilla):
    g = np.random.default_rng(semilla)
    for _ in range(n // 10_000):
        yield from g.lognormal(8.0, 1.0, 10_000)

N_FLUJO = 500_000 if RAPIDO else 2_000_000
t0 = time.perf_counter()
res = np.array(muestreo_reservorio(flujo_montos(N_FLUJO, 7), 2_000, rng))
mu_flujo = np.mean(list(flujo_montos(N_FLUJO, 7)))                      # solo para comparar (un reservorio real no hace esta segunda pasada)
print(f"Flujo de {N_FLUJO:,} montos con reservorio de 2 000 en {time.perf_counter() - t0:.1f} s: media de la muestra = {res.mean():,.0f} | media real = {mu_flujo:,.0f} ({res.mean() / mu_flujo - 1:+.1%})")
assert len(res) == 2_000 and abs(res.mean() / mu_flujo - 1) < 0.10
````

---
**Lectura.** (1) **Muestreo.** El error observado sigue a la teoría $\sigma/\sqrt{n}$ con cocientes de **0.94 a 1.01** y una pendiente log-log de **−0.502** (teoría: −0.5): para **dividir el error a la mitad hay que muestrear cuatro veces más**. Para un error relativo ≤ 1 % al 95 % bastan **≈ 67 000** elementos (los montos son muy asimétricos, con coeficiente de variación ≈ 1.3), y serían los mismos con una población 10 000 veces mayor. (2) **Reservorio.** Con $N=50$ y $k=10$, cada elemento salió entre **3 803 y 4 142 veces** en 20 000 ensayos (esperado: $4\,000 \pm 57$): ninguno está favorecido. Con un flujo de 500 000 montos y un reservorio de 2 000, la media de la muestra (≈ 4 811) quedó a **−2.1 %** de la real (≈ 4 915), en ≈ 3 s: es un bucle de Python puro; en producción esto lo hace código compilado. *(Los tiempos dependen de tu máquina.)*

**Más allá de una muestra: *sketches*.** Un ***sketch*** es una estructura de **tamaño fijo** que se actualiza en una pasada y responde **una pregunta** con **error acotado**. Dos clásicos:

**1. *Count-Min Sketch* (CMS): ¿cuántas veces apareció cada elemento?** Una tabla de $d$ filas y $w$ columnas, con $d$ funciones *hash* distintas $h_1,\dots,h_d$. Al llegar el elemento $x$ se suma 1 en la celda $(j, h_j(x))$ de **cada** fila; para consultar se toma el **mínimo** de las $d$ celdas. Como las colisiones solo **suman**, la estimación **nunca subestima**: $\hat{f}(x) \ge f(x)$. Eligiendo $w = \lceil e/\varepsilon \rceil$ y $d = \lceil \ln(1/\delta) \rceil$, con $N$ el número total de apariciones:
$$\Pr\big[\hat{f}(x) \le f(x) + \varepsilon N\big] \ge 1-\delta.$$
Es excelente para los **elementos frecuentes** (*heavy hitters*); para los raros el error absoluto $\varepsilon N$ puede ser **mayor que su conteo real** (hay que decirlo con honestidad).

**2. *HyperLogLog* (HLL): ¿cuántos elementos distintos hay?** Se aplica un *hash* de 64 bits a cada elemento; los primeros $p$ bits eligen uno de $m = 2^p$ **registros** y en cada registro se guarda el **máximo** $\rho$ (posición del primer 1) visto en los bits restantes. Un registro con $\rho$ grande indica que se vio algo «raro» (muchos ceros iniciales), lo que ocurre con más probabilidad cuantos más elementos distintos haya. La **media armónica** de los registros da la estimación
$$\hat{n} = \alpha_m\, m^2 \Big/ \sum_{j=1}^{m} 2^{-M_j}, \qquad \alpha_m \approx \frac{0.7213}{1 + 1.079/m},$$
con **error estándar relativo** $\approx 1.04/\sqrt{m}$. Con $p=12$ ($m=4096$ registros de 1 byte = **4 KB**) el error típico es ≈ 1.6 %, **sin importar** si hay mil o mil millones de distintos. Nuestra versión es **simplificada** (con la corrección de rango pequeño por conteo lineal, pero sin las mejoras de *HLL++*). *Spark* ofrece `approx_count_distinct`, que usa HLL++ *(según la documentación de Spark; verificar detalles)*.

````python
# ============================================================
# 3.c  Count-Min Sketch (desde cero)
# ============================================================
P_PRIMO = (1 << 31) - 1                                            # primo de Mersenne 2^31 - 1: a*x + b cabe en int64


class CountMinSketch:
    """Tabla d x w de contadores; hash universal ((a*x + b) mod p) mod w. Claves = enteros < 2^31."""

    def __init__(self, eps, delta, semilla=0):
        self.eps, self.delta = eps, delta
        self.w, self.d = int(np.ceil(np.e / eps)), int(np.ceil(np.log(1 / delta)))
        g = np.random.default_rng(semilla)
        self.a = g.integers(1, P_PRIMO, self.d, dtype=np.int64)
        self.b = g.integers(0, P_PRIMO, self.d, dtype=np.int64)
        self.tabla = np.zeros((self.d, self.w), dtype=np.int64)
        self.n = 0

    def _idx(self, xs):
        xs = np.asarray(xs, dtype=np.int64)
        return ((self.a[:, None] * xs[None, :] + self.b[:, None]) % P_PRIMO) % self.w        # forma (d, len(xs))

    def agregar(self, xs):
        idx = self._idx(xs)
        for j in range(self.d):
            self.tabla[j] += np.bincount(idx[j], minlength=self.w)
        self.n += len(idx[0])

    def consultar(self, xs):
        idx = self._idx(xs)
        return self.tabla[np.arange(self.d)[:, None], idx].min(axis=0)


# Flujo sintetico: terminos de busqueda con frecuencias tipo Zipf (pocos muy frecuentes, cola larga)
rng = np.random.default_rng(SEED)
N_CLAVES, N_FLUJO_CMS = 50_000, (300_000 if RAPIDO else 1_500_000)
llaves = np.unique(rng.integers(1, P_PRIMO, N_CLAVES + 500))[:N_CLAVES]             # identificadores enteros distintos
prob = 1.0 / np.arange(1, len(llaves) + 1) ** 1.1
prob /= prob.sum()
flujo = llaves[rng.choice(len(llaves), size=N_FLUJO_CMS, p=prob)]                   # el flujo (en la practica, llega en trozos)
claves, real = np.unique(flujo, return_counts=True)
conteo_real = dict(zip(claves.tolist(), real.tolist()))

EPS, DELTA = 0.001, 0.01
cms = CountMinSketch(EPS, DELTA, semilla=1)
for trozo in np.array_split(flujo, 10):                                             # se alimenta por trozos, como un flujo
    cms.agregar(trozo)
est = cms.consultar(claves)
error = est - real
exceden = np.mean(error > EPS * cms.n)
mem_cms, mem_dict = cms.tabla.nbytes, len(claves) * 16                              # 16 B por par clave-conteo (muy optimista para un dict)
print(f"CMS: w = {cms.w}, d = {cms.d}, memoria = {mem_cms / 1024:.0f} KiB | claves distintas = {len(claves):,}, N = {cms.n:,}")
print(f"  cota eps*N = {EPS * cms.n:.0f} | error maximo observado = {error.max()} | claves que exceden la cota = {exceden:.4%} (delta = {DELTA:.0%})")
top = np.argsort(-real)[:5]
print("  5 mas frecuentes (real -> estimado):", [(int(real[i]), int(est[i])) for i in top])
cola = np.argsort(real)[:5]
print("  5 mas raros      (real -> estimado):", [(int(real[i]), int(est[i])) for i in cola], "<- el error absoluto puede ser mayor que el conteo")
assert np.all(est >= real)                                                           # NUNCA subestima
assert exceden <= DELTA                                                              # la cota eps*N se cumple para >= 1-delta de las claves
rel_top = np.abs(est[top] - real[top]) / real[top]
assert rel_top.max() < 0.05                                                          # frecuentes: error relativo pequeno

# ============================================================
# 3.d  HyperLogLog simplificado (desde cero)
# ============================================================
def splitmix64(x):
    """Mezcla de 64 bits (SplitMix64): convierte enteros en 'hashes' bien repartidos."""
    z = x.astype(np.uint64) + np.uint64(0x9E3779B97F4A7C15)
    z = (z ^ (z >> np.uint64(30))) * np.uint64(0xBF58476D1CE4E5B9)
    z = (z ^ (z >> np.uint64(27))) * np.uint64(0x94D049BB133111EB)
    return z ^ (z >> np.uint64(31))


class HyperLogLog:
    def __init__(self, p=12):
        self.p, self.m = p, 1 << p
        self.M = np.zeros(self.m, dtype=np.uint8)
        self.alfa = 0.7213 / (1 + 1.079 / self.m)

    def agregar(self, xs):
        h = splitmix64(np.asarray(xs))
        idx = (h >> np.uint64(64 - self.p)).astype(np.int64)                        # primeros p bits: registro
        resto = h & np.uint64((1 << (64 - self.p)) - 1)                              # bits restantes
        _, e = np.frexp(resto.astype(np.float64))                                    # e = longitud en bits (exacto: resto < 2^53)
        rho = (64 - self.p) - e + 1                                                  # posicion del primer 1 (resto == 0 -> 64-p+1)
        np.maximum.at(self.M, idx, rho.astype(np.uint8))

    def estimar(self):
        E = self.alfa * self.m ** 2 / np.sum(2.0 ** (-self.M.astype(np.float64)))
        ceros = int(np.sum(self.M == 0))
        if E <= 2.5 * self.m and ceros > 0:                                          # rango pequeno: conteo lineal
            E = self.m * np.log(self.m / ceros)
        return E


filas = []
for n in [100, 1_000, 10_000, 100_000, 1_000_000]:
    hll = HyperLogLog(p=12)
    elementos = np.arange(n) * 7919 + 13
    hll.agregar(np.concatenate([elementos, elementos, elementos[: n // 2]]))        # con REPETIDOS: no deben cambiar el conteo
    filas.append((n, hll.estimar(), hll.estimar() / n - 1))
tab_hll = pd.DataFrame(filas, columns=["distintos reales", "estimado", "error relativo"]).set_index("distintos reales")
print("\nHyperLogLog (p = 12, m = 4096, memoria = 4 KiB):\n" + tab_hll.round(4).to_string())
teoria_hll = 1.04 / np.sqrt(4096)
assert (tab_hll["error relativo"].abs() < 4 * teoria_hll).all()                      # dentro de ~4 desviaciones tipicas

errores = []
for r in range(40):                                                                   # repeticiones con conjuntos distintos
    h = HyperLogLog(p=12); h.agregar(np.arange(50_000) + r * 1_000_003); errores.append(h.estimar() / 50_000 - 1)
print(f"Desviacion tipica del error (40 repeticiones, n = 50 000): {np.std(errores):.4f} | teoria 1.04/sqrt(m) = {teoria_hll:.4f}")
assert 0.5 * teoria_hll < np.std(errores) < 1.6 * teoria_hll
print(f"Memoria exacta (set de Python con 1 000 000 de enteros: solo el contenedor) = {sys.getsizeof(set(range(1_000_000))) / 2**20:.0f} MiB frente a {hll.M.nbytes / 1024:.0f} KiB del HLL")
````

```python
print(f"memoria: {mem2 / 1024:.0f} KiB (antes {cms.tabla.nbytes / 1024:.0f} KiB) | claves fuera de cota: {frac:.4%} | error maximo: {int((est2 - real).max())}")

assert np.all(est2 >= real)
assert frac <= 0.001
print("Bajar eps (w = e/eps) cuesta memoria LINEAL en 1/eps; bajar delta (d = ln(1/delta)) solo LOGARITMICA: la confianza sale barata, la precision no.")
```

| ***Count-Min Sketch*** | Memoria $O((1/\varepsilon)\ln(1/\delta))$; **nunca subestima**; error $\le \varepsilon N$ con prob. $\ge 1-\delta$; bueno para frecuentes, **malo para raros** |
