## 1. La Estructura de una Historia con Datos 🧱

Una **historia** no es una lista de hallazgos: tiene **un mensaje**, **un orden** y **una invitación a actuar**. Dos esquemas útiles (y compatibles):

| Esquema | Pasos | Pregunta en cada paso |
|---|---|---|
| **Contexto-conflicto-resolución** (el arco clásico) | **1. Contexto:** dónde estamos y qué es «normal». **2. Conflicto:** lo que **cambió o no encaja** (el hallazgo). **3. Resolución:** qué hacer y qué se espera | ¿Qué debe saber mi audiencia para entender? · ¿Por qué debería importarle? · ¿Qué le pido? |
| **Contexto-foco-acción** | **Contexto:** la situación y el objetivo. **Foco:** lo **único** que merece atención (se resalta; lo demás se atenúa). **Acción:** la decisión o el siguiente paso | ¿A quién le hablo y qué sabe ya? · ¿Qué dato lo demuestra? · ¿Qué hacemos el lunes? |

**La «gran idea».** Antes de dibujar nada, escribe en **una frase** lo que la audiencia debe llevarse (*«Estamos creciendo, pero perdemos dinero en cada pedido de Sogamoso; con un pedido mínimo recuperamos el margen»*). Si no cabe en una frase, la historia tiene **demasiados mensajes**. **Una figura, un mensaje.** Todo lo que no sirve a esa frase es ruido.

**Audiencia y propósito.** Un gráfico **exploratorio** es para **ti** (muchas variables, para descubrir); uno **explicativo** es para **otro** (una conclusión, para convencer o informar). El mismo dato produce ambos, pero se **diseñan** distinto. En Big Data hay una trampa adicional: los resultados salen de **agregaciones** muy grandes, y la audiencia **no ve el trabajo**: hay que **declarar** qué se agregó, qué se descartó y **qué supuestos** tiene cualquier simulación.

**Nuestro caso (sintético).** *AgroCanasta* es una empresa **ficticia** que vende canastas de productos agrícolas a domicilio en **Tunja, Duitama y Sogamoso**. Durante 24 meses (2024-2025) los pedidos **crecen mucho**; la dirección celebra, pero el **margen por pedido** (ingreso menos costo del producto y del domicilio) se comporta distinto según la ciudad. **Todos los datos y cifras son inventados** y generados con semilla fija: la historia ilustra el **método**, no una empresa real.

````python
# ============================================================
# Configuracion inicial + datos sinteticos de AgroCanasta (nivel de pedido, ~36 000 filas)
# ============================================================
# !pip install -q numpy pandas matplotlib

import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")
SEED = 42
plt.rcParams.update({"figure.figsize": (9, 5), "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
GRIS, ROJO, AZUL, OSCURO = "#9ca3af", "#dc2626", "#2563eb", "#0f172a"

CIUDADES = ["Tunja", "Duitama", "Sogamoso"]
MESES = pd.period_range("2024-01", periods=24, freq="M")
PEDIDOS_0 = {"Tunja": 500, "Duitama": 300, "Sogamoso": 180}                       # pedidos en el mes 0
CRECIMIENTO = {"Tunja": 0.030, "Duitama": 0.035, "Sogamoso": 0.055}               # crecimiento mensual compuesto
TICKET = {"Tunja": (68_000, 70_000), "Duitama": (60_000, 58_000), "Sogamoso": (52_000, 34_000)}   # valor MEDIANO del pedido: mes 0 -> mes 23
DOMICILIO = {"Tunja": (5_500, 6_400), "Duitama": (6_500, 8_000), "Sogamoso": (8_000, 17_000)}     # costo medio del domicilio por pedido (COP)
COSTO_PRODUCTO = 0.62                                                              # fraccion del valor del pedido


def generar_pedidos(semilla=SEED):
    rng = np.random.default_rng(semilla)
    filas = []
    for c in CIUDADES:
        for m in range(24):
            n = rng.poisson(PEDIDOS_0[c] * (1 + CRECIMIENTO[c]) ** m)
            f = m / 23
            mediana = TICKET[c][0] + f * (TICKET[c][1] - TICKET[c][0])
            valor = np.round(rng.lognormal(np.log(mediana), 0.45, n), -2)
            dom = np.clip(DOMICILIO[c][0] + f * (DOMICILIO[c][1] - DOMICILIO[c][0]) + rng.normal(0, 600, n), 2_000, None).round(-2)
            filas.append(pd.DataFrame({"ciudad": c, "mes": m, "valor": valor, "costo_dom": dom}))
    d = pd.concat(filas, ignore_index=True)
    d["margen"] = d.valor * (1 - COSTO_PRODUCTO) - d.costo_dom                     # margen por pedido (COP)
    return d


pedidos = generar_pedidos()
pedidos_mes = pedidos.pivot_table(index="mes", columns="ciudad", values="valor", aggfunc="size")[CIUDADES]
margen_mes = pedidos.pivot_table(index="mes", columns="ciudad", values="margen", aggfunc="mean")[CIUDADES]
pedidos_mes.index = margen_mes.index = MESES.astype(str)                          # etiquetas 'AAAA-MM'
total_mes = pedidos_mes.sum(axis=1)
print(f"{len(pedidos):,} pedidos | pedidos del mes 1: {total_mes.iloc[0]:,} -> mes 24: {total_mes.iloc[-1]:,} ({total_mes.iloc[-1] / total_mes.iloc[0]:.2f} veces)")
print("Margen medio por pedido (COP), mes 1 y mes 24:\n" + margen_mes.iloc[[0, -1]].round(0).astype(int).to_string())
assert total_mes.iloc[-1] > 2 * total_mes.iloc[0]
assert (margen_mes.iloc[0] > 0).all() and margen_mes.iloc[-1]["Sogamoso"] < 0 and (margen_mes.iloc[-1][["Tunja", "Duitama"]] > 0).all()
````

## 2. De un Gráfico Exploratorio a uno Explicativo 🎯

Dibujamos primero el gráfico que sale **por defecto** al explorar (el «antes») y luego el mismo dato **rediseñado para explicar** (el «después»). Transformaciones aplicadas:

| Transformación | Antes (exploratorio) | Después (explicativo) | Por qué |
|---|---|---|---|
| **Título** | «Margen por ciudad» (un **tema**) | «El margen por pedido de Sogamoso se volvió negativo en 2025-09» (una **conclusión**) | Quien lee solo el título debe llevarse el mensaje |
| **Color con propósito** | 3 colores de la paleta por defecto | **Gris** para el contexto y **un color de alerta** para el foco | El color dirige la atención; si todo es de color, nada destaca |
| **Etiquetas** | Leyenda aparte (el ojo viaja de la línea a la leyenda y vuelta) | **Etiqueta directa** al final de cada línea | Menos esfuerzo de lectura |
| **Anotaciones** | Ninguna | Una **flecha y una frase** en el punto clave (cuándo cruza el cero) | Se dice **qué mirar** |
| **Ruido** | Rejilla, marco, ejes sin unidades | Sin rejilla ni marco; **línea del cero** resaltada; **unidades** en el eje | Menos tinta que no informa |
| **Procedencia** | Nada | **Fuente y nota** («datos sintéticos») | Credibilidad y reproducibilidad |

````python
# ============================================================
# 2. Antes / despues: el MISMO dato, dos disenos (y una auditoria automatica de cada figura, mas abajo)
# ============================================================
def mes_negativo(serie):
    """Primer mes desde el cual la serie queda negativa de ahi en adelante (None si no ocurre)."""
    neg = serie < 0
    for m in serie.index:
        if neg.loc[m:].all():
            return m
    return None


MES_NEG = mes_negativo(margen_mes["Sogamoso"])
pos_neg = list(margen_mes.index).index(MES_NEG)
print(f"El margen por pedido de Sogamoso queda negativo desde {MES_NEG} (mes {pos_neg + 1} de 24)")


def grafico_antes(ax=None):
    fig, ax = plt.subplots(figsize=(8, 4.2))
    for c in CIUDADES:
        ax.plot(margen_mes.index, margen_mes[c], label=c)                           # colores por defecto de matplotlib
    ax.grid(True); ax.legend(); ax.set_title("Margen por ciudad")
    ax.set_xticks(range(0, 24, 2)); ax.set_xticklabels(margen_mes.index[::2], rotation=45, fontsize=7)
    return fig


def grafico_despues():
    fig, ax = plt.subplots(figsize=(8, 4.2))
    x = np.arange(24)
    for c in ("Tunja", "Duitama"):
        ax.plot(x, margen_mes[c] / 1000, color=GRIS, lw=2)
        ax.text(23.3, margen_mes[c].iloc[-1] / 1000, c, color="#6b7280", va="center", fontsize=9)
    ax.plot(x, margen_mes["Sogamoso"] / 1000, color=ROJO, lw=3)
    ax.text(23.3, margen_mes["Sogamoso"].iloc[-1] / 1000, "Sogamoso", color=ROJO, va="center", fontsize=10, fontweight="bold")
    ax.axhline(0, color=OSCURO, lw=1.2)
    ax.annotate(f"Desde {MES_NEG} cada pedido\nde Sogamoso pierde plata", xy=(pos_neg, margen_mes["Sogamoso"].iloc[pos_neg] / 1000), xytext=(pos_neg - 12, -4.5), va="center",
                arrowprops=dict(arrowstyle="->", color=ROJO), color=ROJO, fontsize=9)
    ax.text(0, 27, "Al inicio, las 3 ciudades tenían margen positivo", color="#374151", fontsize=8.5)
    ax.set_title(f"El margen por pedido de Sogamoso se volvió negativo en {MES_NEG}", loc="left", fontweight="bold", fontsize=12)
    ax.set_ylabel("margen por pedido (miles de COP)")
    ax.set_xticks([0, 6, 12, 18, 23]); ax.set_xticklabels([margen_mes.index[i] for i in (0, 6, 12, 18, 23)], fontsize=9)
    ax.set_xlim(-0.5, 26); ax.set_ylim(-7, 29)
    for lado in ("top", "right"): ax.spines[lado].set_visible(False)
    ax.grid(False)
    fig.text(0.01, 0.01, "Fuente: AgroCanasta (empresa ficticia), datos sintéticos. Margen = valor del pedido - costo del producto (62 %) - domicilio.", fontsize=7, color="#6b7280")
    fig.tight_layout(rect=(0, 0.04, 1, 1))
    return fig


fig_a, fig_d = grafico_antes(), grafico_despues()
plt.show()
````

---
**Lectura.** El «antes» cuenta lo mismo pero obliga al lector a **trabajar**: tres colores con el mismo peso, una leyenda que hay que cruzar con las líneas, un título que solo nombra el tema y una rejilla que compite con los datos. El «después» deja **un solo color vivo** (Sogamoso) y atenúa el contexto en gris, rotula las líneas **al final**, anota **cuándo** el margen cruza el cero (**2025-09**, mes 21 de 24) y declara la **fuente**. *No se cambió un solo dato:* el margen medio por pedido del mes 24 es de **22 981** COP en Tunja, **15 511** en Duitama y **−2 932** en Sogamoso (que partió de **14 593** en el mes 1). La auditoría de la sección 5 mide estas diferencias.

## 3. Una Historia de Tres Figuras 🎬

Con la «gran idea» fijada —*«Crecemos rápido, pero en Sogamoso cada pedido pierde dinero; un pedido mínimo lo corrige»*—, la historia usa **una figura por paso del arco**:

1. **Contexto** — *«AgroCanasta ya procesa más del doble de los pedidos de hace dos años»*: pedidos mensuales totales (una línea, un mensaje).
2. **Conflicto** — *«…pero Sogamoso perdió su margen»*: la figura de la sección anterior, con el foco en una ciudad.
3. **Resolución** — *«Un pedido mínimo recupera margen»*: el efecto esperado de **distintos pedidos mínimos** sobre el margen de Sogamoso, con el **supuesto explícito** de qué hacen los clientes cuyo pedido queda por debajo del mínimo.

**La simulación de la resolución** (con **supuestos ilustrativos, no medidos**): para un pedido mínimo $M$ y los pedidos de Sogamoso de los **últimos 6 meses**, un cliente cuyo pedido vale $v < M$ **completa hasta $M$** con probabilidad $p$ **solo si le falta a lo sumo el 50 %** de su pedido ($v \ge M/1.5$); si le falta más, o no completa, **se pierde** (margen 0). El margen esperado $G(M)$ es
$$G(M) = \sum_{v_i \ge M} \big(0.38\,v_i - d_i\big) \;+\; \sum_{M/1.5 \,\le\, v_i < M} p\,\big(0.38\,M - d_i\big),$$
con $d_i$ el costo del domicilio de ese pedido y $0.38 = 1 - 0.62$. Se toma $p = 0.5$ como caso base y luego se hace un **análisis de sensibilidad** sobre $p$. *Lo que este modelo **no** incluye:* el valor futuro de los clientes que se pierden, la reacción de la competencia ni otras palancas (cobrar el domicilio, consolidar rutas).

````python
# ============================================================
# 3. Simulacion de la resolucion (pedido minimo) + las tres figuras de la historia
# ============================================================
sog = pedidos[(pedidos.ciudad == "Sogamoso") & (pedidos.mes >= 18)]                   # ultimos 6 meses
MINIMOS = np.arange(0, 60_001, 2_500)                                                 # minimos plausibles (hasta ~2 veces el pedido mediano)


FLEX = 1.5                                                                           # un cliente solo completa si le falta a lo sumo el 50 % de su pedido (v >= M / 1.5)


def margen_esperado(M, p, d=sog):
    ya = d[d.valor >= M]                                                               # ya cumplen el minimo: siguen igual
    ventana = d[(d.valor < M) & (d.valor * FLEX >= M)]                                 # completan con probabilidad p (y pagan SU domicilio)
    return float(ya.margen.sum() + (p * ((1 - COSTO_PRODUCTO) * M - ventana.costo_dom)).sum())    # los demas (v < M / 1.5) se pierden: 0 de margen


base = margen_esperado(0, 0.5)
curva = np.array([margen_esperado(M, 0.5) for M in MINIMOS])
M_opt = int(MINIMOS[curva.argmax()])
gan = curva.max() - base
print(f"Sogamoso, ultimos 6 meses: margen sin minimo = {base:,.0f} COP | mejor minimo (p = 0.5) = {M_opt:,} COP -> margen = {curva.max():,.0f} COP ({gan:,.0f} COP mas)")
perdidos = int((sog.valor < M_opt).sum())
completan = int(((sog.valor < M_opt) & (sog.valor * FLEX >= M_opt)).sum())
print(f"Pedidos por debajo del minimo: {perdidos:,} de {len(sog):,} ({perdidos / len(sog):.0%}); de ellos {completan:,} estan en la ventana de 'completar' (con p = 0.5, ~{completan // 2:,} lo harian) y {perdidos - completan:,} se perderian")
sens = pd.DataFrame([(p, int(MINIMOS[np.argmax([margen_esperado(M, p) for M in MINIMOS])]), max(margen_esperado(M, p) for M in MINIMOS) - margen_esperado(0, p)) for p in (0.0, 0.2, 0.5, 0.8)],
                    columns=["p (clientes que completan)", "mejor minimo (COP)", "ganancia frente a sin minimo (COP)"]).set_index("p (clientes que completan)")
print("\nSensibilidad al supuesto p:\n" + sens.round(0).astype(int).to_string())
assert M_opt > 0 and gan > 0 and base < 0                                           # hoy Sogamoso pierde dinero en conjunto y el minimo ayuda
assert (sens["ganancia frente a sin minimo (COP)"] >= 0).all()                      # nunca empeora frente a no hacer nada (en este modelo)
assert sens["mejor minimo (COP)"].is_monotonic_increasing                           # cuanto mas completan, mas alto conviene el minimo

# ---------------- FIGURA 1: contexto ----------------
f1, ax = plt.subplots(figsize=(8, 3.8))
x = np.arange(24)
ax.plot(x, total_mes.values, color=AZUL, lw=3)
ax.fill_between(x, total_mes.values, color=AZUL, alpha=0.08)
ax.annotate(f"{total_mes.iloc[0]:,} pedidos", xy=(0, total_mes.iloc[0]), xytext=(1.5, total_mes.iloc[0] + 250), arrowprops=dict(arrowstyle="->", color="#374151"), fontsize=9)
ax.annotate(f"{total_mes.iloc[-1]:,} pedidos", xy=(23, total_mes.iloc[-1]), xytext=(15.5, total_mes.iloc[-1] + 80), arrowprops=dict(arrowstyle="->", color="#374151"), fontsize=9)
ax.set_title(f"AgroCanasta ya procesa {total_mes.iloc[-1] / total_mes.iloc[0]:.1f} veces los pedidos de hace dos años", loc="left", fontweight="bold", fontsize=12)
ax.set_ylabel("pedidos por mes"); ax.set_xticks([0, 6, 12, 18, 23]); ax.set_xticklabels([margen_mes.index[i] for i in (0, 6, 12, 18, 23)])
ax.set_ylim(0, total_mes.max() * 1.25); ax.spines[["top", "right"]].set_visible(False)
f1.text(0.01, 0.01, "Fuente: AgroCanasta (ficticia), datos sintéticos.", fontsize=7, color="#6b7280"); f1.tight_layout(rect=(0, 0.04, 1, 1))

# ---------------- FIGURA 2: conflicto (la figura 'despues') ----------------
f2 = fig_d

# ---------------- FIGURA 3: resolucion ----------------
f3, ax = plt.subplots(figsize=(8, 4))
colores = [ROJO if M == M_opt else GRIS for M in MINIMOS]
ax.bar(MINIMOS / 1000, curva / 1e6, width=2.1, color=colores)
ax.axhline(0, color=OSCURO, lw=1.2)
M_txt = f"{M_opt / 1000:g}"
ax.annotate(f"Mínimo de ${M_txt} mil: margen de {curva.max() / 1e6:,.1f} millones\n(sin mínimo: {base / 1e6:,.1f} millones)", xy=(M_opt / 1000, curva.max() / 1e6), xytext=(1, curva.max() / 1e6 * 0.78),
            arrowprops=dict(arrowstyle="->", color=ROJO), color=ROJO, fontsize=9)
ax.set_title(f"Un pedido mínimo de ${M_txt} mil devolvería el margen positivo a Sogamoso", loc="left", fontweight="bold", fontsize=12)
ax.set_xlabel("pedido mínimo (miles de COP)"); ax.set_ylabel("margen esperado, últimos 6 meses\n(millones de COP)")
ax.spines[["top", "right"]].set_visible(False)
f3.text(0.01, 0.01, "Supuesto (ilustrativo, no medido): con pedido >= mínimo/1.5, la mitad completa el pedido; el resto se pierde. Datos sintéticos.", fontsize=7, color="#6b7280")
f3.tight_layout(rect=(0, 0.04, 1, 1))
plt.show()
````

## 4. Cuando una Historia Engaña: Dos Errores Éticos Frecuentes ⚖️

Contar una historia es **elegir** qué mostrar; ese poder exige responsabilidad. Dos errores clásicos, que **no** requieren mala fe y que **sí** cambian las decisiones:

**(a) El eje truncado.** En un gráfico de **barras**, la **altura** representa la magnitud; si el eje no empieza en cero, las alturas ya **no son proporcionales** a los valores y una diferencia pequeña parece enorme. Tufte propuso el **factor de mentira** (*lie factor*): el cociente entre el **efecto mostrado** y el **efecto en los datos**,
$$\text{LF} = \frac{(h_2 - h_1)/h_1}{(v_2 - v_1)/v_1},$$
con $h$ la altura en el gráfico y $v$ el valor. Con eje desde cero, $h \propto v$ y $\text{LF} = 1$. Si el eje empieza en $b$, $h \propto v - b$ y $\text{LF} = v_1/(v_1 - b)$ (para ejes desde cero no hay engaño). *(En gráficos de **líneas** un eje que no empieza en cero es aceptable, siempre que se vea.)*

**(b) El sesgo de confirmación y la ventana a conveniencia.** Si ya «sabemos» lo que queremos decir, es fácil **elegir el intervalo** que lo confirma. En una serie ruidosa **sin tendencia**, entre todas las ventanas de 6 meses **habrá** unas que suben y otras que bajan: quien muestra solo una cuenta **la historia que prefiera**. La defensa: **decidir la ventana antes de mirar**, mostrar la **serie completa** y reportar **cuántas** ventanas apoyan cada conclusión.

La celda siguiente **cuantifica ambos**: el factor de mentira de un eje truncado (con el valor medio de los pedidos de 2024 frente a 2025) y la ventana a conveniencia sobre una serie **sin tendencia** (el porcentaje de pedidos entregados a tiempo, **sintético**).

````python
# ============================================================
# 4. (a) Eje truncado y factor de mentira   (b) Ventana elegida a conveniencia
# ============================================================
ticket_2024 = float(pedidos[pedidos.mes < 12].valor.mean())
ticket_2025 = float(pedidos[pedidos.mes >= 12].valor.mean())
print(f"(a) Valor medio del pedido: 2024 = {ticket_2024:,.0f} COP | 2025 = {ticket_2025:,.0f} COP ({ticket_2025 / ticket_2024 - 1:+.1%})")


def altura_barras(ylim_inf):
    fig, ax = plt.subplots(figsize=(3.2, 3.6))
    b = ax.bar(["2024", "2025"], [ticket_2024, ticket_2025], color=[GRIS, AZUL])
    ax.set_ylim(ylim_inf, max(ticket_2024, ticket_2025) * 1.03)
    fig.canvas.draw()
    h = [ax.transData.transform((0, r.get_height()))[1] - ax.transData.transform((0, ylim_inf))[1] for r in b]      # altura VISIBLE (en pixeles) de cada barra
    plt.close(fig)
    return h


def factor_mentira(h, v1, v2):
    return ((h[1] - h[0]) / h[0]) / ((v2 - v1) / v1)


lf_cero = factor_mentira(altura_barras(0), ticket_2024, ticket_2025)
base_trunc = 0.95 * min(ticket_2024, ticket_2025)
lf_trunc = factor_mentira(altura_barras(base_trunc), ticket_2024, ticket_2025)
teo = ticket_2024 / (ticket_2024 - base_trunc)
print(f"    eje desde 0: LF = {lf_cero:.2f} | eje desde {base_trunc:,.0f}: LF = {lf_trunc:.1f} (teoria v1/(v1-b) = {teo:.1f}): el grafico exagera la diferencia unas {lf_trunc:.0f} veces")
assert abs(lf_cero - 1) < 0.02 and abs(lf_trunc / teo - 1) < 0.02 and lf_trunc > 5

fig, ax = plt.subplots(1, 2, figsize=(7.5, 3.4))
for a, inf, t in ((ax[0], base_trunc, "Eje truncado (engaña)"), (ax[1], 0, "Eje desde cero (honesto)")):
    a.bar(["2024", "2025"], [ticket_2024, ticket_2025], color=[GRIS, AZUL]); a.set_ylim(inf, max(ticket_2024, ticket_2025) * 1.03); a.set_title(t, fontsize=10)
    a.spines[["top", "right"]].set_visible(False); a.set_ylabel("valor medio del pedido (COP)", fontsize=8)
plt.tight_layout(); plt.show()

# (b) Serie SIN tendencia: % de pedidos entregados a tiempo (media 90 %, ruido de 2 puntos)
rng = np.random.default_rng(7)
a_tiempo = 90 + rng.normal(0, 2.0, 24)
pend_total = np.polyfit(np.arange(24), a_tiempo, 1)[0]
VENT = 6
pend = np.array([np.polyfit(np.arange(VENT), a_tiempo[i:i + VENT], 1)[0] for i in range(24 - VENT + 1)])
i_mejor, i_peor = int(pend.argmax()), int(pend.argmin())
print(f"\n(b) Pendiente de la serie COMPLETA: {pend_total:+.3f} puntos/mes (sin tendencia real)")
print(f"    {len(pend)} ventanas de {VENT} meses: {int((pend > 0).sum())} suben y {int((pend < 0).sum())} bajan | la mejor sube {pend.max():+.2f} pts/mes (meses {i_mejor + 1}-{i_mejor + VENT}) y la peor baja {pend.min():+.2f} pts/mes (meses {i_peor + 1}-{i_peor + VENT})")
assert abs(pend_total) < 0.25 and pend.max() > 4 * abs(pend_total) and pend.min() < -4 * abs(pend_total)     # con la misma serie se cuenta 'mejora' o 'deterioro'
assert (pend > 0).any() and (pend < 0).any()

fig, ax = plt.subplots(figsize=(8, 3.4))
ax.plot(range(1, 25), a_tiempo, color=GRIS, lw=1.5)
ax.plot(range(i_mejor + 1, i_mejor + VENT + 1), a_tiempo[i_mejor:i_mejor + VENT], color=AZUL, lw=3, label=f"«mejora»: {pend.max():+.2f} pts/mes")
ax.plot(range(i_peor + 1, i_peor + VENT + 1), a_tiempo[i_peor:i_peor + VENT], color=ROJO, lw=3, label=f"«deterioro»: {pend.min():+.2f} pts/mes")
ax.axhline(90, color=OSCURO, lw=0.8, ls=":"); ax.legend(frameon=False, fontsize=9); ax.set_xlabel("mes"); ax.set_ylabel("% entregas a tiempo")
ax.set_title("Misma serie sin tendencia: según la ventana se cuenta «mejora» o «deterioro»", loc="left", fontweight="bold", fontsize=11)
ax.spines[["top", "right"]].set_visible(False); plt.tight_layout(); plt.show()
````

**Lectura.** (a) **Eje truncado:** el valor medio del pedido pasó de **67 972** COP (2024) a **64 457** (2025), una caída de **5.2 %**. Con el eje desde cero, el factor de mentira es **1.00**; con el eje empezando en **61 234** (el 95 % del menor valor) es **10.1**, igual que la fórmula $v_1/(v_1-b)$: la barra de 2025 se ve **≈ 52 % más corta** que la de 2024 cuando la diferencia real es de 5.2 %. (b) **Ventana a conveniencia:** la serie de entregas a tiempo **no tiene tendencia** (pendiente de la serie completa: **−0.078** puntos/mes). De las **19 ventanas** de 6 meses, **7 suben y 12 bajan**; la «mejor» (**+0.72** pts/mes, meses 19-24) y la «peor» (**−0.80**, meses 16-21) cuentan **historias opuestas** con los mismos datos. Quien elige la ventana **después** de ver la serie puede sostener cualquiera de las dos.

## 5. Lista de Verificación para una Figura que Explica ✅

| # | Pregunta | Señal de alerta |
|---|---|---|
| 1 | ¿El **título** dice la **conclusión** (no solo el tema)? | «Ventas por mes», «Margen por ciudad» |
| 2 | ¿Hay **un mensaje por figura** y la audiencia sabe **qué mirar**? | Seis series del mismo peso visual |
| 3 | ¿El **color** tiene un propósito (foco frente a contexto)? ¿Se distingue en escala de grises y con daltonismo? | Paleta arcoíris sin criterio |
| 4 | ¿Hay **anotaciones** en el punto clave y **etiquetas directas** en lugar de leyendas lejanas? | Leyenda al otro lado del gráfico |
| 5 | ¿Se eliminó el **ruido** (rejilla densa, marco, 3D, decoración)? | Efectos que no informan |
| 6 | ¿Los **ejes** tienen unidades y los gráficos de **barras** empiezan en **cero**? | Barras con eje truncado |
| 7 | ¿Se muestra la **serie completa** y la ventana se decidió **antes** de mirar? | Ventana que justo confirma lo que queríamos |
| 8 | ¿Se declaran la **fuente**, la **agregación**, los **supuestos** y la **incertidumbre**? | Una cifra sola, sin su procedencia ni su rango |
| 9 | ¿Se muestra lo que **contradice** o **limita** el mensaje? | Solo evidencia a favor |
| 10 | ¿La historia termina en una **acción** concreta y con responsable? | «Es interesante» como conclusión |

La celda siguiente **automatiza una parte** de la lista (los puntos que se pueden **leer del objeto** `Figure` de *matplotlib*: título, colores, leyenda, anotaciones, ruido, unidades, fuente y base cero de las barras) y la aplica a las figuras de este cuaderno. **Es una ayuda, no un juez**: no sabe si el título es *verdadero*, ni si el mensaje es *el correcto*.

````python
# ============================================================
# 5. Auditoria automatica de una figura (lo que SE PUEDE leer del objeto Figure)
# ============================================================
def auditar_figura(fig):
    ax = fig.axes[0]
    colores = set()
    for l in ax.get_lines():
        if len(l.get_xdata()) >= 3:                                              # solo series (no lineas de referencia de 2 puntos)
            colores.add(l.get_color())
    barras = [p for p in ax.patches if p.get_width() > 0 and p.get_height() != 0]
    colores |= {tuple(np.round(p.get_facecolor(), 2)) for p in barras}
    titulo = ax.get_title(loc="left") or ax.get_title()
    texto_fig = " ".join(t.get_text() for t in fig.texts)
    return {
        "titulo_dice_conclusion": len(titulo.split()) >= 6,
        "pocos_colores (<=3)": len(colores) <= 3,
        "sin_leyenda_y_con_etiquetas": ax.get_legend() is None and len(ax.texts) >= 1,
        "con_anotaciones (>=2)": len(ax.texts) >= 2,
        "sin_marco_ni_rejilla": not ax.spines["top"].get_visible() and not ax.spines["right"].get_visible() and not any(g.get_visible() for g in ax.get_ygridlines()),
        "eje_con_unidades": len(ax.get_ylabel()) > 0,
        "con_fuente": "Fuente" in texto_fig or "fuente" in texto_fig or "Supuesto" in texto_fig,
        "barras_desde_cero": (not barras) or ax.get_ylim()[0] <= 0,
    }


def figura_barras_truncada():
    fig, ax = plt.subplots(figsize=(3.5, 3.4))
    ax.bar(["2024", "2025"], [ticket_2024, ticket_2025], color=[GRIS, AZUL]); ax.set_ylim(base_trunc, max(ticket_2024, ticket_2025) * 1.03)
    ax.set_title("Valor del pedido"); return fig


resultados = {"antes (exploratorio)": auditar_figura(fig_a), "despues (explicativo)": auditar_figura(fig_d),
              "figura 1 (contexto)": auditar_figura(f1), "figura 3 (resolucion)": auditar_figura(f3), "barras con eje truncado": auditar_figura(figura_barras_truncada())}
tabla = pd.DataFrame(resultados).T
tabla["puntaje (de 8)"] = tabla.sum(axis=1)
print(tabla.astype(int).to_string())
assert tabla.loc["despues (explicativo)", "puntaje (de 8)"] >= tabla.loc["antes (exploratorio)", "puntaje (de 8)"] + 4
assert not tabla.loc["barras con eje truncado", "barras_desde_cero"]                        # la auditoria SI detecta el eje truncado
assert tabla.loc["despues (explicativo)", "puntaje (de 8)"] >= 7
plt.close("all")
````

**Lectura de la auditoría.** El rediseño sube de **2/8** a **8/8**; la figura 1 (contexto) cumple **8/8**; la figura 3 (resolución), **7/8** (tiene una sola anotación y el criterio pide al menos dos); y las barras con el eje truncado obtienen **2/8** y fallan «barras desde cero»: la auditoría **sí detecta** el eje truncado. Pero las 8 señales son **mecánicas**: un gráfico con un título **falso** también obtendría 8/8. La lista automatiza lo que se puede **leer del objeto** `Figure`, no lo que **significa**.

| **Exploratorio frente a explicativo** | El primero es para ti (descubrir); el segundo, para otro (convencer o informar). Se **diseñan** distinto aunque el dato sea el mismo |
| **Rediseño** | Título que **dice la conclusión**, color con propósito (gris para el contexto, un color para el foco), **etiquetas directas**, **anotaciones**, sin rejilla ni marco, unidades y **fuente** |
| **Historia en tres figuras** | Contexto (crecimiento), conflicto (Sogamoso pierde margen), resolución (un pedido mínimo); la **simulación** declara sus **supuestos** y se somete a **sensibilidad** |
| **Eje truncado** | En barras, $\text{LF} = v_1/(v_1-b)$: aun un recorte pequeño **multiplica** la diferencia mostrada; las barras empiezan en cero |
| **Sesgo de confirmación** | Con ruido, hay ventanas que suben y que bajan; **decide la ventana antes**, muestra la serie completa y cuenta cuántas apoyan cada mensaje |
| **Lista de verificación** | Partes **automatizables** (título, colores, leyenda, anotaciones, ruido, unidades, fuente, base cero) y partes **humanas** (si el mensaje es verdadero, completo y útil) |
