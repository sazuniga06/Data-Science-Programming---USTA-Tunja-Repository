## 1. Qué Cambia al Escalar el Aprendizaje Automático 📐

Un algoritmo de ML clásico asume que **todo cabe en memoria** y que se puede **recorrer varias veces** y **en cualquier orden**. A escala se rompen esas tres cosas, y cada solución sacrifica algo:

| Problema | Solución | Qué se sacrifica |
|---|---|---|
| Los datos **no caben en RAM** | **Aprendizaje por tandas** (*out-of-core*): el modelo ve **un trozo cada vez** (`partial_fit`) | Exactitud ligeramente menor, hiperparámetros extra (tasa, épocas), **sensibilidad al orden** |
| Los datos **sí caben en el clúster** | **Entrenamiento distribuido**: cada nodo trabaja una partición y se combinan **estadísticos o gradientes** (*Spark MLlib*) | Comunicación en cada iteración; en **una sola máquina es más lento** que *scikit-learn* |
| El **modelo** es grande (aprendizaje profundo) | **Paralelismo de datos** (réplicas del modelo, fragmentos de datos, gradientes promediados), GPU, a veces paralelismo de modelo | Sincronización, memoria por dispositivo, ajustes de la tasa de aprendizaje |
| El **vocabulario** es enorme (texto) | **Truco del hashing**: índice = `hash(palabra) mod m` | **Colisiones** y se pierde la interpretabilidad |

**Aprendizaje por tandas.** La **regresión logística** con descenso de gradiente estocástico (SGD) actualiza los pesos con cada ejemplo o minilote: $\mathbf{w} \leftarrow \mathbf{w} - \eta\,\nabla \ell(\mathbf{w};\, \mathbf{x}_i, y_i)$. Como solo mira **un trozo a la vez**, la memoria es $O(\text{trozo})$ y no $O(N)$, a cambio de recorrer los datos en varias **épocas** y de que el resultado dependa del **orden**. Un modelo **lineal** con SGD es el caso más sencillo; los **árboles** o el *k*-medias requieren variantes específicas.

La celda siguiente prepara el entorno y **genera un archivo CSV por trozos**: nunca existe completo en memoria durante la generación.

````python
# ============================================================
# Configuracion inicial + datos sinteticos generados POR TROZOS (CSV en un directorio temporal)
# ============================================================
# !pip install -q numpy pandas matplotlib scikit-learn pyspark

import atexit
import os
import shutil
import sys
import tempfile
import time
import warnings
import zlib

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, HashingVectorizer, TfidfTransformer, TfidfVectorizer
from sklearn.linear_model import LogisticRegression, SGDClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.utils import murmurhash3_32

warnings.filterwarnings("ignore")
SEED = 42
RAPIDO = True                                   # True: 120 000 filas y 6 000 documentos (segundos); False: 5 veces mas
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
TMP = tempfile.mkdtemp(prefix="bd03_nb01_")
atexit.register(shutil.rmtree, TMP, ignore_errors=True)          # los archivos temporales se borran al cerrar el kernel

HAY_SPARK = False
try:
    import pyspark  # noqa: F401
    HAY_SPARK = shutil.which("java") is not None
except ImportError:
    pass


def iniciar_spark(nombre):
    """Sesion local de Spark con 2 hilos, sin interfaz web y con todo en el directorio temporal."""
    os.environ.setdefault("PYSPARK_PYTHON", sys.executable)
    from pyspark.sql import SparkSession
    s = (SparkSession.builder.master("local[2]").appName(nombre).config("spark.ui.enabled", "false")
         .config("spark.sql.shuffle.partitions", "4").config("spark.sql.warehouse.dir", TMP + "/warehouse").getOrCreate())
    s.sparkContext.setLogLevel("ERROR")
    return s


# --- Datos de clasificacion binaria: 20 variables, las etiquetas siguen un modelo logistico con pesos conocidos ---
D = 20
N_FILAS = 120_000 if RAPIDO else 600_000
CHUNK = 20_000
columnas = [f"x{i}" for i in range(D)]
g0 = np.random.default_rng(123)
W_REAL = g0.normal(0, 1, D) * (g0.random(D) < 0.7) * 0.6        # ~30 % de variables sin efecto
ESCALAS, CENTROS = np.exp(g0.uniform(np.log(0.5), np.log(8), D)), g0.normal(0, 3, D)    # escalas muy distintas: obliga a estandarizar


def generar_trozo(i, n):
    g = np.random.default_rng([SEED, i])
    X = g.normal(size=(n, D)) * ESCALAS + CENTROS
    z = ((X - CENTROS) / ESCALAS) @ W_REAL
    y = (g.random(n) < 1 / (1 + np.exp(-z))).astype(int)
    d = pd.DataFrame(X, columns=columnas).round(4)
    d.insert(0, "id", np.arange(i * CHUNK, i * CHUNK + n))
    d["y"] = y
    return d


ruta_csv = f"{TMP}/datos.csv"
t0 = time.perf_counter()
for i in range(N_FILAS // CHUNK):                                  # se escribe trozo a trozo: la RAM nunca ve el archivo completo
    generar_trozo(i, CHUNK).to_csv(ruta_csv, mode="a", header=(i == 0), index=False)
N_TRAIN = int(0.8 * N_FILAS)                                       # entrenamiento: id < N_TRAIN ; prueba: el resto
print(f"CSV: {N_FILAS:,} filas x {D} variables, {os.path.getsize(ruta_csv) / 2**20:.1f} MiB, generado en {time.perf_counter() - t0:.1f} s | entrenamiento = {N_TRAIN:,}")
assert sum(1 for _ in open(ruta_csv)) == N_FILAS + 1
````

````python
# ============================================================
# 2. Aprendizaje fuera de memoria: SGDClassifier.partial_fit por trozos frente a ajuste completo
# ============================================================
def trozos_train():
    """Iterador de trozos de ENTRENAMIENTO (nunca se carga el archivo completo)."""
    for t in pd.read_csv(ruta_csv, chunksize=CHUNK):
        t = t[t.id < N_TRAIN]
        if len(t):
            yield t


prueba = pd.read_csv(ruta_csv, skiprows=range(1, N_TRAIN + 1))          # solo las filas de prueba (20 %)
X_te, y_te = prueba[columnas].to_numpy(), prueba.y.to_numpy()
assert len(prueba) == N_FILAS - N_TRAIN

# techo (Bayes): con los pesos reales, ningun modelo puede acertar mas que esto (en promedio)
acc_techo = np.mean(((((X_te - CENTROS) / ESCALAS) @ W_REAL) > 0).astype(int) == y_te)

# (a) Ajuste COMPLETO: todo el entrenamiento en memoria
t0 = time.perf_counter()
tr = pd.concat(trozos_train())
esc_full = StandardScaler().fit(tr[columnas])
lr_full = LogisticRegression(max_iter=500).fit(esc_full.transform(tr[columnas]), tr.y)
t_full = time.perf_counter() - t0
acc_full = lr_full.score(esc_full.transform(X_te), y_te)
mem_full = tr.memory_usage(deep=True).sum() / 2**20
del tr

# (b) Por TANDAS: 1.a pasada: media y varianza con partial_fit del escalador; 2.a: SGD con partial_fit (varias epocas)
CFG_OK = dict(learning_rate="constant", eta0=0.001)                  # tasa de aprendizaje constante y pequena (para datos estandarizados)


def sgd_por_tandas(epocas, **cfg):
    t0 = time.perf_counter()
    esc = StandardScaler()
    for t in trozos_train():
        esc.partial_fit(t[columnas])
    clf = SGDClassifier(loss="log_loss", alpha=1e-6, random_state=SEED, **(cfg or CFG_OK))
    mem_max = 0.0
    for _ in range(epocas):
        for t in trozos_train():
            mem_max = max(mem_max, t.memory_usage(deep=True).sum() / 2**20)
            clf.partial_fit(esc.transform(t[columnas]), t.y, classes=[0, 1])
    return clf, esc, time.perf_counter() - t0, mem_max


res = {"ajuste completo (LogisticRegression)": (lr_full.score(esc_full.transform(X_te), y_te), t_full, mem_full)}
for ep in (1, 3):
    clf, esc, t_sgd, mem_sgd = sgd_por_tandas(ep)
    res[f"SGD por tandas, {ep} epoca(s)"] = (clf.score(esc.transform(X_te), y_te), t_sgd, mem_sgd)
clf_d, esc_d, t_d, mem_d = sgd_por_tandas(1, learning_rate="optimal")                          # la tasa POR DEFECTO de scikit-learn
res["SGD por tandas, 1 epoca, tasa 'optimal' (defecto)"] = (clf_d.score(esc_d.transform(X_te), y_te), t_d, mem_d)
tabla = pd.DataFrame(res, index=["exactitud (prueba)", "segundos", "MiB de datos a la vez"]).T
print(f"Techo teorico (pesos reales): {acc_techo:.4f}\n")
print(tabla.round(4).to_string())

acc_s1, acc_s3 = res["SGD por tandas, 1 epoca(s)"][0], res["SGD por tandas, 3 epoca(s)"][0]
assert abs(acc_s3 - acc_full) < 0.01                               # por tandas ~ ajuste completo
assert res["SGD por tandas, 3 epoca(s)"][2] < res["ajuste completo (LogisticRegression)"][2] / 3        # y con una fraccion de la memoria
acc_def = res["SGD por tandas, 1 epoca, tasa 'optimal' (defecto)"][0]
assert acc_def < acc_s1 - 0.02                                      # una tasa mal elegida cuesta exactitud: el SGD NO es 'plug and play'

# La trampa del ORDEN: si el archivo llega ORDENADO por clase (p. ej. por una columna de estado), el modelo 'olvida' lo anterior
tr_o = pd.concat(trozos_train()).sort_values("y")                   # (solo para la demostracion: aqui SI se carga todo y se ordena)
esc_o = StandardScaler().fit(tr_o[columnas])
clf_o = SGDClassifier(loss="log_loss", alpha=1e-6, random_state=SEED, **CFG_OK)
for ix in np.array_split(np.arange(len(tr_o)), 6):                  # 6 trozos: los 3 primeros solo con y = 0 y los 3 ultimos con y = 1
    clf_o.partial_fit(esc_o.transform(tr_o.iloc[ix][columnas]), tr_o.y.iloc[ix], classes=[0, 1])
acc_orden = clf_o.score(esc_o.transform(X_te), y_te)
del tr_o
print(f"\nMismo SGD de 1 epoca con el archivo ORDENADO por clase: exactitud = {acc_orden:.4f} (con orden aleatorio: {acc_s1:.4f})")
assert acc_orden < acc_s1 - 0.02
````

**Lectura.** (1) El **techo** teórico (con los pesos reales) es **0.7298**; la regresión logística de ajuste completo llega a **0.7288** y el SGD por tandas a **0.7285**: la diferencia es de **0.03 puntos** de exactitud, pero el SGD mantiene en memoria **≈ 3.4 MiB** (un trozo de 20 000 filas) frente a **≈ 16.8 MiB** del ajuste completo: **la memoria queda acotada por el tamaño del trozo**, no por el del archivo. Con este tamaño **todo cabe en RAM** y el ajuste completo es igual de rápido: el beneficio aparece cuando el archivo **no cabe**. Tres épocas dan lo mismo que una (**0.7285**): el problema es sencillo y está bien escalado. (2) **Los hiperparámetros importan:** con la tasa de aprendizaje por defecto de *scikit-learn* (`'optimal'`) la exactitud cae a **0.6652**; el SGD **no** es *plug and play*. (3) **El orden importa:** con el archivo **ordenado por clase** el mismo SGD queda en **0.5034** (azar), frente a **0.7285** con orden aleatorio: el modelo «olvida» lo que vio al principio. *(Los tiempos de esta tabla son del orden de un segundo y dependen de la máquina.)*

## 3. *Spark MLlib*: Regresión Logística sobre un *DataFrame* Distribuido ⚡

En *Spark*, los datos son un ***DataFrame*** dividido en **particiones**; cada tarea trabaja una partición y el controlador combina los resultados. Un **`Pipeline`** encadena **transformadores** (ensamblar columnas en un vector, estandarizar) y un **estimador** (regresión logística): se ajusta **una vez** y se aplica igual a datos nuevos.

**Cómo entrena MLlib la regresión logística** (resumen conceptual): cada iteración del optimizador (L-BFGS en la configuración por defecto) calcula el **gradiente** de la pérdida sumando **contribuciones de cada partición** (un `treeAggregate`) y el controlador actualiza los pesos. Es exactamente el **paralelismo de datos** de la sección 4, con **datos fijos en los trabajadores** y **pesos que viajan** en cada iteración. Para comparar de forma justa, usamos las **mismas filas** de entrenamiento y prueba que *scikit-learn* y **la misma regresión sin regularización** (en ambos).

````python
# ============================================================
# 3. Pipeline de Spark MLlib (VectorAssembler -> StandardScaler -> LogisticRegression) frente a scikit-learn
# ============================================================
if HAY_SPARK:
    from pyspark.ml import Pipeline
    from pyspark.ml.classification import LogisticRegression as SparkLR
    from pyspark.ml.evaluation import BinaryClassificationEvaluator, MulticlassClassificationEvaluator
    from pyspark.ml.feature import StandardScaler as SparkScaler, VectorAssembler
    from pyspark.sql import functions as F

    spark = iniciar_spark("bd03_nb01")
    df = spark.read.csv(ruta_csv, header=True, inferSchema=True)
    train_sp = df.filter(F.col("id") < N_TRAIN).repartition(2).cache()       # 2 particiones (= 2 hilos); cache(): el optimizador recorre los datos MUCHAS veces
    train_sp.count()                                                          # materializa el cache (sin esto, cada iteracion volveria a leer y a interpretar el CSV)
    test_sp = df.filter(F.col("id") >= N_TRAIN)
    pipe = Pipeline(stages=[
        VectorAssembler(inputCols=columnas, outputCol="crudo"),
        SparkScaler(inputCol="crudo", outputCol="x", withMean=True, withStd=True),
        SparkLR(featuresCol="x", labelCol="y", maxIter=40, tol=1e-5, regParam=0.0, standardization=False),
    ])
    t0 = time.perf_counter()
    modelo_sp = pipe.fit(train_sp)
    t_spark = time.perf_counter() - t0
    pred_sp = modelo_sp.transform(test_sp)
    acc_spark = MulticlassClassificationEvaluator(labelCol="y", predictionCol="prediction", metricName="accuracy").evaluate(pred_sp)
    auc_spark = BinaryClassificationEvaluator(labelCol="y", metricName="areaUnderROC").evaluate(pred_sp)

    # scikit-learn con la MISMA regresion (sin penalizacion) y los mismos datos
    tr = pd.concat(trozos_train())
    esc_s = StandardScaler().fit(tr[columnas])
    t0 = time.perf_counter()
    lr_np = LogisticRegression(penalty=None, max_iter=1000).fit(esc_s.transform(tr[columnas]), tr.y)
    t_sk = time.perf_counter() - t0
    acc_sk = lr_np.score(esc_s.transform(X_te), y_te)
    coef_sp = np.array(modelo_sp.stages[-1].coefficients)
    coef_sk = lr_np.coef_.ravel()
    coseno = float(coef_sp @ coef_sk / (np.linalg.norm(coef_sp) * np.linalg.norm(coef_sk)))
    del tr

    print(pd.DataFrame({"exactitud (prueba)": [acc_spark, acc_sk], "segundos de ajuste": [t_spark, t_sk]}, index=["Spark MLlib (local[2], 2 particiones)", "scikit-learn (1 proceso)"]).round(4).to_string())
    print(f"AUC de Spark = {auc_spark:.4f} | iteraciones = {modelo_sp.stages[-1].summary.totalIterations} | particiones del entrenamiento = {train_sp.rdd.getNumPartitions()} | coseno entre vectores de coeficientes = {coseno:.6f}")
    assert abs(acc_spark - acc_sk) < 0.005                                  # mismo modelo, misma exactitud
    assert coseno > 0.999                                                    # y practicamente los mismos coeficientes
    spark.stop()                                                             # se detiene la sesion al terminar la seccion
    print("Sesion de Spark detenida.")
else:
    print("Spark/Java no disponibles: se omite esta seccion (el resto del cuaderno no depende de ella).")
````

##### 🛠️ Práctica 1: Media y varianza exactas por trozos

## 4. Aprendizaje Profundo a Escala: Paralelismo de Datos desde Cero 🧠

Entrenar una red neuronal es minimizar una pérdida promedio $L(\theta) = \frac{1}{N}\sum_{i=1}^{N} \ell(\theta;\, \mathbf{x}_i, y_i)$ por descenso del gradiente. El **paralelismo de datos** (el esquema dominante) funciona así: se **replica** el modelo en $K$ trabajadores (GPU o máquinas), se **reparte** el lote en $K$ fragmentos, cada trabajador calcula el gradiente de su fragmento y se **promedian** los gradientes (*all-reduce*) antes de actualizar $\theta$ **igual** en todos. Si el fragmento $k$ tiene $n_k$ ejemplos y su gradiente (de la pérdida **promedio** del fragmento) es $\mathbf{g}_k$, el gradiente del lote completo de $N=\sum_k n_k$ ejemplos es **exactamente**
$$\nabla L(\theta) \;=\; \sum_{k=1}^{K} \frac{n_k}{N}\, \mathbf{g}_k .$$
Con fragmentos de **igual tamaño** el peso es $1/K$ (promedio simple); con fragmentos **desiguales** el promedio simple **ya no es** el gradiente del lote (hay que ponderar). Por eso **un paso de SGD síncrono con $K$ trabajadores equivale a un paso con un lote $K$ veces mayor**: se paraleliza el **cálculo**, no se cambia el algoritmo.

Implementamos una red de **una capa oculta** ($\tanh$) para clasificación binaria con **retropropagación escrita a mano**, **verificamos el gradiente con diferencias finitas**, comprobamos la equivalencia $K=1$ frente a $K=4$ y entrenamos. Los «trabajadores» aquí son **simulados** en un solo proceso (calcular secuencialmente el gradiente de cada fragmento): lo que se verifica es la **matemática** del *all-reduce*, no una aceleración real.

````python
# ============================================================
# 4. SGD sincrono con paralelismo de datos (K trabajadores simulados), retropropagacion a mano
# ============================================================
# Datos NO lineales: la clase depende de la distancia al origen en las 2 primeras variables (4 variables mas son ruido)
g = np.random.default_rng(SEED)
N_DL = 20_000
X_dl = g.normal(size=(N_DL, 6))
y_dl = (X_dl[:, 0] ** 2 + X_dl[:, 1] ** 2 + 0.3 * g.normal(size=N_DL) > 1.4).astype(float)
Xtr_dl, ytr_dl, Xte_dl, yte_dl = X_dl[:16_000], y_dl[:16_000], X_dl[16_000:], y_dl[16_000:]
H = 16                                                                       # neuronas ocultas


def init_params(d, h, semilla):
    r = np.random.default_rng(semilla)
    return np.concatenate([(r.normal(0, 1 / np.sqrt(d), (d, h))).ravel(), np.zeros(h), r.normal(0, 1 / np.sqrt(h), h), [0.0]])


def desempaquetar(p, d, h):
    W1 = p[: d * h].reshape(d, h); b1 = p[d * h: d * h + h]; w2 = p[d * h + h: d * h + 2 * h]; b2 = p[-1]
    return W1, b1, w2, b2


def perdida_y_grad(p, X, y):
    """Perdida logistica PROMEDIO del fragmento y su gradiente (retropropagacion manual)."""
    d = X.shape[1]
    W1, b1, w2, b2 = desempaquetar(p, d, H)
    A = np.tanh(X @ W1 + b1)
    z = A @ w2 + b2
    prob = 1 / (1 + np.exp(-z))
    perdida = float(np.mean(np.logaddexp(0, z) - y * z))
    dz = (prob - y) / len(y)                                                 # derivada de la perdida promedio respecto a z
    dA = np.outer(dz, w2)
    dZ1 = dA * (1 - A ** 2)
    grad = np.concatenate([(X.T @ dZ1).ravel(), dZ1.sum(axis=0), A.T @ dz, [dz.sum()]])
    return perdida, grad


def grad_paralelo(p, X, y, K, ponderar=True, tamanos=None):
    """Cada 'trabajador' calcula el gradiente de su fragmento; luego se combinan (all-reduce)."""
    trozos = np.array_split(np.arange(len(y)), K) if tamanos is None else np.split(np.arange(len(y)), np.cumsum(tamanos)[:-1])
    grads = [perdida_y_grad(p, X[ix], y[ix])[1] for ix in trozos]
    pesos = np.array([len(ix) for ix in trozos], dtype=float)
    pesos = pesos / pesos.sum() if ponderar else np.full(K, 1 / K)
    return sum(w * gk for w, gk in zip(pesos, grads))


p0 = init_params(6, H, 0)
xb, yb = Xtr_dl[:1_000], ytr_dl[:1_000]
_, g_completo = perdida_y_grad(p0, xb, yb)

# (1) Verificacion por diferencias finitas (comprueba que la retropropagacion esta bien escrita)
rng = np.random.default_rng(1)
for j in rng.choice(len(p0), 6, replace=False):
    e = np.zeros_like(p0); e[j] = 1e-6
    num = (perdida_y_grad(p0 + e, xb, yb)[0] - perdida_y_grad(p0 - e, xb, yb)[0]) / 2e-6
    assert abs(num - g_completo[j]) < 1e-6 * max(1.0, abs(num)), (j, num, g_completo[j])

# (2) El gradiente combinado de K trabajadores = gradiente del lote completo (fragmentos iguales o NO)
for K in (1, 2, 4, 7):                                                        # 1000 no es divisible por 7: fragmentos casi iguales
    assert np.allclose(grad_paralelo(p0, xb, yb, K), g_completo, atol=1e-12), K
DESIGUAL = [100, 900]                                                          # un trabajador con 100 ejemplos y otro con 900
assert np.allclose(grad_paralelo(p0, xb, yb, 2, tamanos=DESIGUAL), g_completo, atol=1e-12)               # ponderado: igual al lote completo
dif_simple = np.abs(grad_paralelo(p0, xb, yb, 2, ponderar=False, tamanos=DESIGUAL) - g_completo).max()   # promedio simple: NO
print(f"K = 1, 2, 4, 7 y fragmentos [100, 900] (ponderado por n_k/N): igual al gradiente del lote completo (tol. 1e-12) | promedio SIMPLE con [100, 900]: dif. maxima = {dif_simple:.2e}")
assert dif_simple > 1e-3

# (3) Entrenamiento: K = 1 (lote 256) frente a K = 4 trabajadores (4 x 64 = 256): trayectorias IDENTICAS
def entrenar(K, pasos=400, lote=256, eta=0.5, semilla=0):
    p, r, curva = init_params(6, H, 0), np.random.default_rng(semilla), []
    for _ in range(pasos):
        ix = r.integers(0, len(ytr_dl), lote)                                    # el MISMO lote global, repartido en K fragmentos
        p = p - eta * grad_paralelo(p, Xtr_dl[ix], ytr_dl[ix], K)
        curva.append(perdida_y_grad(p, Xte_dl, yte_dl)[0])
    return p, np.array(curva)


p1, curva1 = entrenar(1)
p4, curva4 = entrenar(4)
assert np.allclose(p1, p4, atol=1e-10)                                           # misma trayectoria: K no cambia el algoritmo
def exactitud(p, X, y):
    W1, b1, w2, b2 = desempaquetar(p, X.shape[1], H)
    return float(np.mean(((np.tanh(X @ W1 + b1) @ w2 + b2) > 0) == (y == 1)))
acc_dl = exactitud(p4, Xte_dl, yte_dl)
print(f"Tras 400 pasos: perdida de prueba = {curva4[-1]:.3f} (K=1: {curva1[-1]:.3f}) | exactitud de prueba (K=4) = {acc_dl:.4f}")
assert acc_dl > 0.85

# Modelo de costo ILUSTRATIVO (parametros SUPUESTOS, no medidos): tiempo por paso = computo/K + comunicacion
K_s = np.array([1, 2, 4, 8, 16, 32, 64])
T_COMPUTO, T_COM = 1.0, 0.04                                                      # unidades arbitrarias; la comunicacion crece con K
paso = T_COMPUTO / K_s + T_COM * (K_s > 1) * (1 + np.log2(K_s))
aceleracion = paso[0] / paso

fig, ax = plt.subplots(1, 2, figsize=(12, 4.3))
ax[0].plot(curva1, color=PALETA[0], lw=4, alpha=0.5, label="K = 1 (lote 256)")
ax[0].plot(curva4, color=PALETA[1], lw=1.5, ls="--", label="K = 4 (4 x 64)")
ax[0].set(xlabel="paso de SGD", ylabel="perdida logistica de prueba", title="Mismo algoritmo, mismas curvas"); ax[0].legend()
ax[1].plot(K_s, aceleracion, "o-", color=PALETA[2], label="modelo (supuestos)"); ax[1].plot(K_s, K_s, ":", color="#64748b", label="ideal lineal")
ax[1].set(xscale="log", yscale="log", xlabel="trabajadores K", ylabel="aceleracion por paso", title="Modelo ilustrativo: la comunicacion limita")
ax[1].set_xticks(K_s); ax[1].set_xticklabels(K_s); ax[1].legend()
plt.tight_layout(); plt.show()
print("Aceleracion del modelo ilustrativo:", dict(zip(K_s.tolist(), aceleracion.round(1).tolist())))
````

**Lectura.** (1) **La retropropagación escrita a mano es correcta:** la comprobación con diferencias finitas sobre 6 coordenadas al azar pasó. (2) **El *all-reduce* equivale al lote completo:** con $K = 1, 2, 4, 7$ y con fragmentos desiguales $[100, 900]$ el gradiente ponderado por $n_k/N$ coincide con el del lote completo (tolerancia $10^{-12}$), mientras que el **promedio simple** con $[100, 900]$ se desvía hasta **≈ 0.03** en alguna coordenada del gradiente (dependiendo de los datos): no es un error de redondeo. (3) **$K$ no cambia la trayectoria:** el entrenamiento con $K=1$ y con $K=4$ trabajadores (mismo lote global de 256) llega a los **mismos parámetros** (tolerancia $10^{-10}$) y a una pérdida de prueba de **0.257**, con una exactitud de **0.916**. (4) El **modelo de costo** (con parámetros **supuestos**, no medidos) muestra lo esencial: la aceleración por paso **sube de 1 a ≈ 3.8** con $K=16$ y **luego baja** por el costo de comunicar los gradientes: más trabajadores **no** son siempre mejores.

### GPU y entrenamiento distribuido: panorama

| Aspecto | Idea | Costo / límite |
|---|---|---|
| **GPU** | Miles de núcleos simples y gran ancho de banda de memoria: muy buenos para **multiplicaciones de matrices** (el corazón de una red neuronal) | La **memoria de la GPU** limita el tamaño del modelo y del lote; entrenar necesita **más memoria que inferir** (activaciones y gradientes; Géron, Cap. 12) |
| **Paralelismo de datos** | Réplicas del modelo, fragmentos de datos, **gradientes promediados** (lo que implementamos) | Cada paso exige **comunicar** un vector del tamaño del modelo; con lotes globales enormes hay que **reajustar la tasa de aprendizaje** *(Goyal et al., 2017; verificar)* |
| **Paralelismo de modelo / de tuberías** | El modelo **no cabe** en un dispositivo: se parte por capas o por tensores | Dependencias entre dispositivos, «burbujas» de inactividad |
| **Servidor de parámetros** (asíncrono) frente a ***all-reduce*** (síncrono) | Asíncrono: cada trabajador actualiza sin esperar (gradientes **desactualizados**); síncrono: todos esperan al más lento | El asíncrono converge de forma menos predecible; el síncrono sufre a los **rezagados** |
| **Dónde se hace** | Bibliotecas de aprendizaje profundo con módulos de entrenamiento distribuido y su integración con *Spark* *(p. ej. PyTorch DDP, Horovod, TorchDistributor; citado de memoria, verificar disponibilidad y versiones)* | Infraestructura, costos de GPU, tolerancia a fallos |

*Spark MLlib* cubre muy bien **modelos clásicos** (lineales, árboles, *k*-medias, ALS) sobre datos masivos; el aprendizaje profundo suele hacerse en **marcos propios** que leen los datos preparados por *Spark* (el patrón habitual *(verificar según tu arquitectura)*). En este curso **no** usamos *PyTorch* ni *TensorFlow* para no depender de instalaciones pesadas: la `MLPClassifier` de *scikit-learn* y el código de arriba ilustran la **misma matemática**.

````python
# ============================================================
# 4.b  MLPClassifier de scikit-learn: donde un modelo LINEAL no sirve (y partial_fit tambien existe para redes)
# ============================================================
lin = LogisticRegression(max_iter=500).fit(Xtr_dl, ytr_dl)
t0 = time.perf_counter()
mlp = MLPClassifier(hidden_layer_sizes=(H,), activation="tanh", solver="adam", max_iter=150, random_state=SEED).fit(Xtr_dl, ytr_dl)
t_mlp = time.perf_counter() - t0

mlp_tandas = MLPClassifier(hidden_layer_sizes=(H,), activation="tanh", solver="adam", random_state=SEED)
for ep in range(15):                                                                # 15 epocas, trozos de 2 000 ejemplos
    for ix in np.array_split(np.random.default_rng(ep).permutation(len(ytr_dl)), 8):
        mlp_tandas.partial_fit(Xtr_dl[ix], ytr_dl[ix], classes=[0.0, 1.0])
tabla = pd.DataFrame({"exactitud (prueba)": [lin.score(Xte_dl, yte_dl), mlp.score(Xte_dl, yte_dl), mlp_tandas.score(Xte_dl, yte_dl), acc_dl]},
                     index=["regresion logistica (lineal)", "MLPClassifier (ajuste completo)", "MLPClassifier (partial_fit por trozos)", "red en NumPy (K=4, desde cero)"])
print(tabla.round(4).to_string(), f"\nTiempo del MLPClassifier completo: {t_mlp:.1f} s")
assert tabla.iloc[0, 0] < 0.60                                                       # un modelo lineal no puede con una frontera circular
assert tabla.iloc[1, 0] > 0.90 and tabla.iloc[3, 0] > 0.85
````

## 5. NLP a Escala: TF-IDF y el Truco del *Hashing* 📝

Para clasificar texto hay que convertirlo en números. La **bolsa de palabras** cuenta cuántas veces aparece cada palabra del **vocabulario** en cada documento; **TF-IDF** pondera: una palabra pesa más si es **frecuente en el documento** (TF, *term frequency*) y **rara en el corpus** (IDF, *inverse document frequency*):
$$\mathrm{tfidf}(t, d) = \mathrm{tf}(t, d) \cdot \mathrm{idf}(t), \qquad \mathrm{idf}(t) = \ln\frac{N+1}{\mathrm{df}(t)+1} \quad \text{(Spark ML)},$$
con $N$ documentos y $\mathrm{df}(t)$ los que contienen $t$ *(scikit-learn usa una variante suavizada, $\ln\frac{1+N}{1+\mathrm{df}(t)}+1$; las puntuaciones difieren por esa constante)*. Una palabra que aparece en **todos** los documentos aporta casi nada.

**El problema a escala.** Un vocabulario explícito requiere **una pasada previa** para construirlo, una **tabla palabra → columna** que viaja a todos los trabajadores y crece con la variedad del texto (errores de digitación, nombres propios, códigos…), y **descarta lo que no vio** al entrenar. El **truco del *hashing*** lo evita: la columna es $\mathrm{hash}(t) \bmod m$ con $m$ fijo. **Sin pasada previa, sin tabla, memoria constante y 100 % paralelizable**; el costo son las **colisiones** (dos palabras comparten columna) y **no poder volver de la columna a la palabra**. Con $V$ palabras distintas y $m$ columnas, el número esperado de columnas ocupadas es $m\,(1-e^{-V/m})$. *Detalle práctico:* usa una función *hash* **estable** (*MurmurHash*; `zlib.crc32` también es estable pero, como se mide abajo, reparte peor claves parecidas), **no** la función `hash()` de Python, que **cambia entre ejecuciones** para las cadenas (aleatorización por proceso).

**Corpus sintético.** Generamos **6 000 «documentos» en español** de cuatro temas (*salud, campo, transporte, finanzas*) con vocabularios temáticos, palabras comunes y una **cola larga de términos raros** (códigos y errores de digitación) que **infla el vocabulario**. **Es sintético**: no son textos reales y la exactitud describe el mecanismo, no el desempeño sobre lenguaje natural.

````python
# ============================================================
# 5.a  Corpus sintetico en espanol + vocabulario explicito frente a hashing (colisiones medidas)
# ============================================================
TEMAS = {
    "salud": "paciente consulta medico hospital sintomas tratamiento vacuna fiebre diagnostico enfermera urgencias receta examen cirugia clinica dolor terapia medicina contagio cita".split(),
    "campo": "cosecha cultivo papa siembra riego abono lluvia finca tractor fruta ganado semilla vereda campesino plaga suelo invernadero mercado ordeño parcela".split(),
    "transporte": "bus ruta pasajero carretera terminal conductor peaje vehiculo trafico viaje horario via camion semaforo estacion tiquete equipaje moto carga transito".split(),
    "finanzas": "credito banco ahorro interes cuenta prestamo cuota inversion deuda tarjeta saldo impuesto factura pago cliente ganancia presupuesto seguro bolsa cartera".split(),
}
COMUNES = "el la de en y que un una los las por con para su al se del es muy mas hoy esta fue ayer nuevo bien tambien sobre entre desde".split()
NOMBRES_TEMAS = list(TEMAS)


def generar_corpus(n, semilla=SEED, cola=20_000):
    g = np.random.default_rng(semilla)
    docs, etiquetas = [], []
    for _ in range(n):
        t = g.integers(0, 4)
        largo = g.integers(12, 31)
        toks = []
        for _ in range(largo):
            u = g.random()
            if u < 0.30:
                toks.append(g.choice(TEMAS[NOMBRES_TEMAS[t]]))                          # palabra del tema
            elif u < 0.38:
                toks.append(g.choice(TEMAS[NOMBRES_TEMAS[g.integers(0, 4)]]))             # palabra de OTRO tema (confusion)
            elif u < 0.80:
                toks.append(g.choice(COMUNES))
            else:
                toks.append(f"ref{g.integers(0, cola)}")                                  # cola larga: codigos / errores de digitacion
        docs.append(toks); etiquetas.append(t)
    return docs, np.array(etiquetas)


N_DOCS = 6_000 if RAPIDO else 30_000
docs, y_doc = generar_corpus(N_DOCS)
textos = [" ".join(d) for d in docs]
n_ent = int(0.8 * N_DOCS)
print("Ejemplo:", textos[0][:110], "...", f"| tema = {NOMBRES_TEMAS[y_doc[0]]}")

# --- vocabulario EXPLICITO ---
cv = CountVectorizer().fit(textos[:n_ent])
V = len(cv.vocabulary_)
tabla_vocab = sys.getsizeof(cv.vocabulary_) / 2**10
print(f"\nVocabulario explicito: {V:,} palabras distintas; solo el contenedor del diccionario ocupa ~{tabla_vocab:.0f} KiB y debe viajar a cada trabajador")
vistas_en_prueba = set(w for d in docs[n_ent:] for w in d)
desconocidas = len(vistas_en_prueba - set(cv.vocabulary_)) / len(vistas_en_prueba)
print(f"Palabras de prueba que el vocabulario NO conoce (se descartan en silencio): {desconocidas:.1%}")

# --- HASHING con distintos m: colisiones y exactitud ---
def bucket(palabra, m):
    return murmurhash3_32(palabra, positive=True) % m                                  # la misma familia de hash que usa HashingVectorizer


def bucket_crc(palabra, m):
    return zlib.crc32(palabra.encode("utf-8")) % m                                     # estable pero de menor calidad para claves parecidas

filas = []
for k in (6, 8, 10, 12, 14, 16):
    m = 2 ** k
    ocupadas = len({bucket(w, m) for w in cv.vocabulary_})
    ocupadas_crc = len({bucket_crc(w, m) for w in cv.vocabulary_})
    esperadas = m * (1 - np.exp(-V / m))
    hv = HashingVectorizer(n_features=m, alternate_sign=False, norm=None)
    tf = TfidfTransformer().fit(hv.transform(textos[:n_ent]))
    clf = LogisticRegression(solver="newton-cg", max_iter=100).fit(tf.transform(hv.transform(textos[:n_ent])), y_doc[:n_ent])
    acc = clf.score(tf.transform(hv.transform(textos[n_ent:])), y_doc[n_ent:])
    filas.append((m, ocupadas, ocupadas_crc, esperadas, 1 - ocupadas / V, acc))
tab_h = pd.DataFrame(filas, columns=["m (columnas)", "columnas ocupadas", "ocupadas con crc32", "esperadas m(1-e^(-V/m))", "palabras que pierden columna propia", "exactitud (prueba)"]).set_index("m (columnas)")
print("\nHashing (MurmurHash3, el de HashingVectorizer):\n" + tab_h.round(4).to_string())
assert (np.abs(tab_h["columnas ocupadas"] / tab_h["esperadas m(1-e^(-V/m))"] - 1) < 0.05).all()          # la teoria de ocupacion se cumple
assert tab_h["palabras que pierden columna propia"].is_monotonic_decreasing                              # mas columnas, menos colisiones

# Explicito (referencia): mismo clasificador con vocabulario
tv = TfidfVectorizer().fit(textos[:n_ent])
clf_exp = LogisticRegression(solver="newton-cg", max_iter=100).fit(tv.transform(textos[:n_ent]), y_doc[:n_ent])
acc_exp = clf_exp.score(tv.transform(textos[n_ent:]), y_doc[n_ent:])
print(f"\nTF-IDF con vocabulario explicito ({V:,} columnas): exactitud = {acc_exp:.4f}")
assert acc_exp > 0.85 and tab_h["exactitud (prueba)"].iloc[-1] > 0.85


# ============================================================
# 5.b  Spark ML: Tokenizer -> HashingTF -> IDF -> LogisticRegression (y variante con vocabulario: CountVectorizer)
# ============================================================
if HAY_SPARK:
    from pyspark.ml import Pipeline
    from pyspark.ml.classification import LogisticRegression as SparkLR
    from pyspark.ml.evaluation import MulticlassClassificationEvaluator
    from pyspark.ml.feature import IDF, CountVectorizer as SparkCV, HashingTF, Tokenizer

    spark = iniciar_spark("bd03_nb01_nlp")
    pdf = pd.DataFrame({"texto": textos, "label": y_doc.astype(float)})
    tr_sp, te_sp = spark.createDataFrame(pdf.iloc[:n_ent]).repartition(2), spark.createDataFrame(pdf.iloc[n_ent:])
    lr = SparkLR(featuresCol="tfidf", labelCol="label", maxIter=15, regParam=1e-3, standardization=False)      # regularizacion mas fuerte: converge en pocas iteraciones
    pipe_hash = Pipeline(stages=[Tokenizer(inputCol="texto", outputCol="palabras"), HashingTF(inputCol="palabras", outputCol="tf", numFeatures=2 ** 12),
                                 IDF(inputCol="tf", outputCol="tfidf"), lr])
    pipe_vocab = Pipeline(stages=[Tokenizer(inputCol="texto", outputCol="palabras"), SparkCV(inputCol="palabras", outputCol="tf", vocabSize=2 ** 12),
                                  IDF(inputCol="tf", outputCol="tfidf"), lr])
    ev = MulticlassClassificationEvaluator(labelCol="label", predictionCol="prediction", metricName="accuracy")
    filas = []
    for nombre, pipe in (("HashingTF (2^12 = 4 096 columnas)", pipe_hash), ("CountVectorizer (vocabSize = 4 096, las mas frecuentes)", pipe_vocab)):
        t0 = time.perf_counter()
        mod = pipe.fit(tr_sp)
        t_fit = time.perf_counter() - t0
        filas.append((nombre, ev.evaluate(mod.transform(te_sp)), t_fit, mod.stages[-1].numFeatures))
    tab_sp = pd.DataFrame(filas, columns=["pipeline de Spark", "exactitud (prueba)", "segundos de ajuste", "columnas"]).set_index("pipeline de Spark")
    print(tab_sp.round(4).to_string())
    print(f"scikit-learn con vocabulario explicito: exactitud = {acc_exp:.4f} | con hashing 2^12: {tab_h.loc[2 ** 12, 'exactitud (prueba)']:.4f}")
    assert (tab_sp["exactitud (prueba)"] > 0.85).all()
    assert abs(tab_sp["exactitud (prueba)"].iloc[0] - acc_exp) < 0.03                       # mismo problema, misma exactitud (+/- el ruido)
    spark.stop()
    print("Sesion de Spark detenida.")
else:
    print("Spark/Java no disponibles: se omite esta seccion.")
````

**Lectura.** (1) **El vocabulario explícito** tiene **12 822** palabras distintas y su diccionario ya pesa **≈ 405 KiB** solo como contenedor; además, el **35.6 %** de las palabras distintas que aparecen en los documentos de prueba **no existen** en él y se descartan en silencio (con texto real, el porcentaje depende del corpus). (2) **Ocupación y colisiones:** con MurmurHash3 las columnas ocupadas **coinciden con la teoría** $m(1-e^{-V/m})$ (p. ej. **3 908** frente a **3 917** con $m=4\,096$); con `crc32` **no** (4 041, 9 337 y 10 852 en $m = 4\,096$, $16\,384$ y $65\,536$ frente a ≈ 3 917, 8 893 y 11 646 esperadas): una función *hash* estable **no** es necesariamente buena para claves parecidas (`ref123`, `ref124`…). (3) **Exactitud:** **0.848** con $m=64$ (casi todas las palabras colisionan), **0.981** con 256 y **≈ 0.99** desde 4 096, igual que el vocabulario explícito (**0.988**): aun con **69.5 %** de las palabras compartiendo columna con otra en $m=4\,096$, la exactitud casi no cambia, porque en este corpus **sintético** casi todas las colisiones ocurren entre palabras «raras» (códigos `ref…`) que **no** informan sobre el tema; con texto real el efecto puede ser distinto. (4) ***Spark*:** `HashingTF` ($2^{12}$ columnas) y `CountVectorizer` (limitado a las 4 096 palabras más frecuentes) logran **0.9867** de exactitud en la prueba, cercana a la de *scikit-learn* (**0.988-0.990**; la regularización y el optimizador no son idénticos: la comparación es de **orden de magnitud**). *(Los tiempos de ajuste de Spark, de unos 10 s por *pipeline* en una máquina compartida, dependen del arranque en frío de la JVM y de un trabajo por iteración.)*

def hashing_tf(tokens, m):
    v = np.zeros(m)
    for w in tokens:
        v[murmurhash3_32(w, positive=True) % m] += 1
    return v


d0 = docs[0]
v = hashing_tf(d0, 1024)
assert v.sum() == len(d0)
assert np.array_equal(v, hashing_tf(d0, 1024))
assert np.count_nonzero(v) <= len(set(d0))


def hay_colision(tokens, m):
    cols = [murmurhash3_32(w, positive=True) % m for w in set(tokens)]
    return len(cols) != len(set(cols))


frac_64 = np.mean([hay_colision(d, 64) for d in docs[:1000]])
frac_4096 = np.mean([hay_colision(d, 4096) for d in docs[:1000]])
print(f"Documentos con al menos una colision: m = 64 -> {frac_64:.1%} | m = 4096 -> {frac_4096:.1%}")
assert frac_64 > frac_4096

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Qué cambia al escalar** | Los datos no caben, el modelo no cabe o el vocabulario no cabe; cada solución **sacrifica** algo (exactitud, comunicación, interpretabilidad) |
| **Aprendizaje por tandas** | `partial_fit` entrena **un trozo a la vez** con memoria $O(\text{trozo})$; cuesta épocas e hiperparámetros y es **sensible al orden** (datos ordenados por clase degradan el modelo) |
| **Estadísticos combinables** | Media y varianza **sí** se combinan exactamente por trozos (n, media, M2); la mediana exacta **no** |
| ***Spark MLlib*** | *Pipeline* = transformadores + estimador; entrena sumando gradientes por partición. En **una máquina** *scikit-learn* es **más rápido**: *Spark* se justifica cuando los datos **no caben** |
| **Paralelismo de datos** | $\nabla L = \sum_k (n_k/N)\,\mathbf{g}_k$: el *all-reduce* de gradientes **equivale** al lote completo; con fragmentos desiguales el promedio simple **no**. Se paraleliza el **cálculo**, no el algoritmo |
| **Aprendizaje profundo y GPU** | GPU = multiplicación de matrices masiva con memoria limitada; distribuir exige **comunicar** gradientes y reajustar tasas con lotes grandes |
| **TF-IDF y *hashing*** | IDF baja el peso de lo común; el *hashing* elimina vocabulario y pasada previa a cambio de **colisiones** (columnas ocupadas $\approx m(1-e^{-V/m})$) y de la **interpretabilidad**; usa *hash* **estable** |
