## 1. Por Qué Visualizar a Escala es Distinto 🔭

Un gráfico es una **función de los datos a píxeles**. Una pantalla tiene del orden de **un millón de píxeles** (≈ 2 megapíxeles en una pantalla Full HD) y un ojo humano distingue **pocas decenas de formas y colores**. Cuando los datos pasan de miles a millones de filas, se rompen tres supuestos:

| Problema | Qué pasa | Cómo se nota | Remedio |
|---|---|---|---|
| **Sobredibujo** (*overplotting*) | Muchos puntos caen en el **mismo píxel**; el último dibujado tapa a los demás | Una mancha uniforme que **oculta la densidad** (y el orden de dibujo cambia lo que se ve) | **Agregar**: *binning*, histograma 2D, *hexbin*; transparencia como paliativo |
| **Latencia** | Dibujar $N$ marcas cuesta $O(N)$ en el navegador o en el motor de la herramienta | El tablero tarda segundos en cada filtro | **Agregar en el servidor** y enviar solo el resumen; precalcular |
| **Memoria y red** | Mover millones de filas del almacén a la herramienta | Fallas por falta de memoria, descargas enormes | **Empujar el cómputo a los datos** (agregar donde viven); extractos filtrados |
| **Pérdida de lo raro** | El muestreo uniforme **borra lo poco frecuente** (un pico de un solo punto, una categoría del 0.5 %) | Un fraude, una falla de sensor o un grupo minoritario **desaparece** del gráfico | Muestreo **estratificado**, decimación que **conserva extremos** (min-máx) |

**Regla de oro:** *agrega antes de graficar* **y** *haz explícito qué agregación hiciste*. Un histograma 2D con 2 500 celdas representa los mismos 500 000 puntos con **fidelidad** y se dibuja en milisegundos; un resumen **sin** documentar es otra forma de engañar (se retoma en el cuaderno de *storytelling*).

La celda siguiente genera un conjunto sintético de **viajes** (coordenadas con varias zonas densas y una categoría rara) y una **serie de sensor** con un único pico.

````python
# ============================================================
# Configuracion inicial + datos sinteticos
# ============================================================
# !pip install -q numpy pandas matplotlib

import atexit
import io
import json
import os
import shutil
import tempfile
import time
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.colors import LogNorm

warnings.filterwarnings("ignore")
SEED = 42
RAPIDO = True                                  # True: 400 000 puntos (segundos); False: 2 000 000
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
TMP = tempfile.mkdtemp(prefix="bd03_nb02_")
atexit.register(shutil.rmtree, TMP, ignore_errors=True)          # los archivos exportados se borran al cerrar el kernel

N = 400_000 if RAPIDO else 2_000_000
rng = np.random.default_rng(SEED)
# 'Viajes' (coordenadas en km): 4 zonas densas de tamanos muy distintos + fondo disperso
centros = np.array([[0.0, 0.0], [3.0, 2.0], [-2.5, 3.5], [4.5, -3.0]])
desv = np.array([0.35, 0.8, 0.5, 1.2])
peso = np.array([0.40, 0.28, 0.12, 0.10])
z = rng.choice(5, size=N, p=np.append(peso, 1 - peso.sum()))
x = np.where(z < 4, centros[np.minimum(z, 3), 0] + desv[np.minimum(z, 3)] * rng.normal(size=N), rng.uniform(-6, 7, N))
y = np.where(z < 4, centros[np.minimum(z, 3), 1] + desv[np.minimum(z, 3)] * rng.normal(size=N), rng.uniform(-6, 6, N))
valor = np.round(3000 + 1800 * np.hypot(x, y) + rng.gamma(2.0, 900, N), -1)                 # tarifa (COP, ilustrativa)
cat = rng.choice(["urbano", "periferia", "intermunicipal", "aeropuerto", "alerta"], size=N, p=[0.60, 0.25, 0.12, 0.025, 0.005])
viajes = pd.DataFrame({"x": x, "y": y, "valor": valor, "categoria": cat})

# Serie de sensor: senal suave + ruido + UN solo pico de 1 muestra
N_T = 1_000_000 if RAPIDO else 4_000_000
t = np.arange(N_T)
serie = np.sin(2 * np.pi * t / 120_000) + 0.25 * np.sin(2 * np.pi * t / 7_000) + 0.08 * rng.normal(size=N_T)
POS_PICO = int(0.6137 * N_T)
serie[POS_PICO] += 5.0
print(f"viajes: {len(viajes):,} filas, {viajes.memory_usage(deep=True).sum() / 2**20:.0f} MiB en pandas | serie: {N_T:,} muestras con un pico de 1 punto en t = {POS_PICO:,}")
assert abs(peso.sum() + 0.0 - 0.90) < 1e-12 and viajes.categoria.eq("alerta").mean() < 0.01
````

````python
# ============================================================
# 2.a  Sobredibujo: dispersion con TODOS los puntos frente a histograma 2D, hexbin y muestra
# ============================================================
RANGO = [[-6, 7], [-6, 6]]
NB = 120                                                              # 120 x 120 = 14 400 celdas


def medir(dibujar):
    """Dibuja en una figura nueva, fuerza el renderizado y devuelve (figura, segundos)."""
    fig, ax = plt.subplots(figsize=(4.6, 4.2))
    t0 = time.perf_counter()
    dibujar(ax)
    fig.canvas.draw()
    return fig, ax, time.perf_counter() - t0


tiempos, figs = {}, {}
def d_scatter(ax):
    ax.scatter(viajes.x, viajes.y, s=1, color=PALETA[0], rasterized=True); ax.set_title(f"A. Dispersion, {N:,} puntos", fontsize=9)
def d_muestra(ax):
    m = viajes.sample(2_000, random_state=SEED); ax.scatter(m.x, m.y, s=4, color=PALETA[0], alpha=0.6); ax.set_title("B. Muestra de 2 000", fontsize=9)
def d_hist2d(ax):
    H, xe, ye = np.histogram2d(viajes.x, viajes.y, bins=NB, range=RANGO)                     # el trabajo pesado: UNA pasada con NumPy
    ax.pcolormesh(xe, ye, np.where(H > 0, H, np.nan).T, norm=LogNorm(), cmap="viridis"); ax.set_title(f"C. Histograma 2D ({NB}x{NB}, color log)", fontsize=9)
def d_hexbin(ax):
    ax.hexbin(viajes.x, viajes.y, gridsize=60, bins="log", cmap="viridis", extent=(-6, 7, -6, 6)); ax.set_title("D. Hexbin (log)", fontsize=9)

for nombre, f in (("A", d_scatter), ("B", d_muestra), ("C", d_hist2d), ("D", d_hexbin)):
    figs[nombre], ax, tiempos[nombre] = medir(f)
    ax.set_xlim(-6, 7); ax.set_ylim(-6, 6); ax.grid(False)

fig, axs = plt.subplots(1, 4, figsize=(16, 4))
for a, nombre in zip(axs, "ABCD"):
    figs[nombre].canvas.draw()
    a.imshow(np.asarray(figs[nombre].canvas.buffer_rgba())); a.axis("off"); plt.close(figs[nombre])
plt.tight_layout(); plt.show()

H, _, _ = np.histogram2d(viajes.x, viajes.y, bins=NB, range=RANGO)
ocupadas = H[H > 0]
print(f"Tiempos de dibujo (s): " + " | ".join(f"{k} = {v:.2f}" for k, v in tiempos.items()))
en_densas = ocupadas[ocupadas > 50].sum() / H.sum()
print(f"Celdas ocupadas del histograma 2D: {len(ocupadas):,} de {NB * NB:,} | puntos por celda ocupada: mediana = {np.median(ocupadas):.0f}, maximo = {ocupadas.max():.0f} | puntos en celdas con mas de 50 puntos: {en_densas:.1%}")
assert H.sum() <= N and H.sum() > 0.97 * N                           # el rango deja fuera muy pocos puntos
assert ocupadas.max() > 100 * np.median(ocupadas) and en_densas > 0.5   # hay sobredibujo: la mayoria de los puntos cae en celdas muy cargadas
assert tiempos["C"] < tiempos["A"]                                    # el agregado se dibuja mas rapido que la dispersion completa
````

---
**Lectura.** (1) **La dispersión con los 400 000 puntos (A)** es un bloque azul casi uniforme: se distinguen los contornos de las zonas pero **no cuál es más densa**, y lo que se dibuja **al final tapa lo dibujado antes**. Los números lo confirman: de las 14 400 celdas del histograma, **13 970** están ocupadas, la **mediana** es de **4** puntos por celda ocupada y el **máximo** de **2 162**, y el **74.6 %** de los puntos cae en celdas con **más de 50** puntos: eso es sobredibujo. (2) **La muestra de 2 000 (B)** sí deja ver las zonas y se dibuja en ≈ 0.3 s, pero **pierde el detalle** (¿hay una zona más densa dentro de la grande?) y el azar la hace **distinta cada vez**. (3) **El histograma 2D (C) y el *hexbin* (D)** muestran la **jerarquía de densidades** con el color en escala logarítmica (la zona pequeña y compacta del centro es la **más densa**, aunque no la más grande) y los conteos suman **exactamente** los puntos dibujados; el fondo disperso aparece como celdas oscuras y unas **430 celdas vacías** (en blanco). Los tiempos de dibujo (≈ 1 s para la dispersión completa frente a ≈ 0.3-0.5 s para el resumen, medidos en una máquina compartida) **crecen con $N$** en la dispersión y **no** en el histograma, cuyo dibujo depende solo del número de celdas (aunque contar sí cuesta $O(N)$, una sola vez).

## 2. Agregar Antes de Graficar: *Binning*, Decimación y Muestreo Estratificado 🧮

**Histograma 2D (*binning*).** Se divide el plano en una rejilla y se **cuenta** cuántos puntos caen en cada celda: un cálculo $O(N)$ que se hace **una vez** (en NumPy, en SQL con `GROUP BY floor(x/ancho)`, en *Spark*) y deja **$B^2$ números** que se dibujan al instante. Lo que se dibuja ya **no son los datos** sino un **resumen**: el ancho de celda es una decisión de diseño (celdas muy anchas ocultan estructura; muy finas devuelven el ruido). *Hexbin* es lo mismo con celdas hexagonales (más isotrópicas). Como la densidad varía en **órdenes de magnitud**, el color suele ir en **escala logarítmica**, y hay que **decirlo**.

**Series de tiempo: decimación min-máx.** Una línea de 1 000 000 de puntos en una pantalla de ~1 000 píxeles de ancho: ¿qué se dibuja? Un **muestreo uniforme** (cada $k$-ésimo punto) **pierde los picos**. La **decimación min-máx** divide la serie en $W$ cubos (uno por columna de píxeles) y conserva, **por cubo, el mínimo y el máximo**: $2W$ puntos que reproducen la **envolvente** y **nunca borran un extremo** *(la familia de técnicas tipo M4, Jugel et al. 2014; verificar)*.

**Categorías raras: muestreo estratificado.** Con una muestra uniforme de tamaño $n$, una categoría con proporción $p$ aporta $\approx np$ puntos (con $p = 0.5\,\%$ y $n = 2\,000$: unos 10). El **muestreo estratificado** toma hasta $n_c$ puntos **de cada categoría** y asigna a cada punto el **peso** $w_c = N_c/n_c$ para que los **totales** sigan siendo correctos (estimador de Horvitz–Thompson para un diseño estratificado). Sin pesos, el gráfico **sobrerrepresenta** lo raro: es una decisión legítima si se **declara**.

````python
# ============================================================
# 2.b  Decimacion min-max (series) y muestreo estratificado con pesos (categorias raras)
# ============================================================
def minmax_decimar(serie, W):
    """Conserva el minimo y el maximo de cada uno de W cubos (en orden temporal): 2W puntos, nunca pierde un extremo."""
    n = (len(serie) // W) * W
    cubos = serie[:n].reshape(W, -1)
    i_min, i_max = cubos.argmin(axis=1), cubos.argmax(axis=1)
    base = np.arange(W) * cubos.shape[1]
    idx = np.sort(np.concatenate([base + i_min, base + i_max]))
    return idx, serie[idx]


W_PIX = 1_000                                                          # ~ columnas de pixeles de un grafico
idx_mm, y_mm = minmax_decimar(serie, W_PIX)
idx_us = np.sort(np.random.default_rng(1).choice(N_T, 2 * W_PIX, replace=False))             # muestreo uniforme con el MISMO numero de puntos
y_us = serie[idx_us]
pico_mm, pico_us = y_mm.max(), y_us.max()
print(f"Pico real = {serie[POS_PICO]:.2f} | maximo en min-max ({len(idx_mm):,} puntos) = {pico_mm:.2f} | maximo en muestra uniforme ({len(idx_us):,} puntos) = {pico_us:.2f}")
assert pico_mm == serie.max() == serie[POS_PICO]                       # min-max conserva el pico
assert pico_us < serie[POS_PICO] - 1                                   # la muestra uniforme lo pierde (con esta semilla)
assert len(idx_mm) == 2 * W_PIX

t0 = time.perf_counter(); fig_a, ax_a = plt.subplots(figsize=(5, 3)); ax_a.plot(serie, lw=0.4); fig_a.canvas.draw(); t_total = time.perf_counter() - t0; plt.close(fig_a)
t0 = time.perf_counter(); fig_b, ax_b = plt.subplots(figsize=(5, 3)); ax_b.plot(idx_mm, y_mm, lw=0.6); fig_b.canvas.draw(); t_mm = time.perf_counter() - t0; plt.close(fig_b)
print(f"Dibujar la linea completa ({N_T:,} puntos): {t_total:.2f} s | dibujar la decimada ({len(idx_mm):,} puntos): {t_mm:.2f} s")

fig, ax = plt.subplots(1, 2, figsize=(12, 3.6), sharey=True)
ax[0].plot(idx_us, y_us, lw=0.7, color=PALETA[1]); ax[0].set_title(f"Muestra uniforme de {len(idx_us):,}: el pico desaparece", fontsize=10)
ax[1].plot(idx_mm, y_mm, lw=0.7, color=PALETA[2]); ax[1].set_title(f"Decimacion min-max ({len(idx_mm):,} puntos): el pico se conserva", fontsize=10)
ax[1].annotate("pico de 1 muestra", xy=(POS_PICO, serie[POS_PICO]), xytext=(POS_PICO * 0.55, serie[POS_PICO] - 0.2), arrowprops=dict(arrowstyle="->", color="#334155"), fontsize=9)
for a in ax: a.set_xlabel("muestra")
plt.tight_layout(); plt.show()

# --- Categorias raras ---
real = viajes.categoria.value_counts()
n_total = 2_000
unif = viajes.sample(n_total, random_state=SEED).categoria.value_counts().reindex(real.index, fill_value=0)
N_POR_CAT = 400
estr = viajes.groupby("categoria", group_keys=False).apply(lambda g: g.sample(min(N_POR_CAT, len(g)), random_state=SEED))
estr["peso"] = estr.categoria.map(real) / estr.categoria.map(estr.categoria.value_counts())          # w_c = N_c / n_c
tabla = pd.DataFrame({"filas reales": real, "muestra uniforme (2 000)": unif, "muestra estratificada": estr.categoria.value_counts().reindex(real.index)})
print("\n" + tabla.to_string())
tot_ponderado = estr.groupby("categoria").peso.sum().reindex(real.index)
media_real = viajes.valor.mean()
media_ponderada = np.average(estr.valor, weights=estr.peso)
media_sin_peso = estr.valor.mean()
print(f"\nTotales ponderados = reales: {np.allclose(tot_ponderado, real)} | media de 'valor': real = {media_real:,.0f}, estratificada con pesos = {media_ponderada:,.0f} ({media_ponderada / media_real - 1:+.2%}), estratificada SIN pesos = {media_sin_peso:,.0f} ({media_sin_peso / media_real - 1:+.2%})")
assert unif["alerta"] < estr.categoria.value_counts()["alerta"] / 4                                  # la uniforme casi no ve 'alerta'
assert np.allclose(tot_ponderado, real)                                                               # con pesos, los totales se recuperan
assert abs(media_ponderada / media_real - 1) < 0.03 and abs(media_sin_peso - media_real) > abs(media_ponderada - media_real)
````

**Lectura.** (1) **Series:** el pico de **1 muestra** (valor real **5.31**) **se conserva** con la decimación min-máx y **desaparece** en una muestra uniforme con el mismo número de puntos (su máximo es **1.37**): con 2 000 puntos de 1 000 000, la probabilidad de que una muestra aleatoria acierte justo ese punto es del orden de 0.2 %. Dibujar la línea completa tardó **≈ 0.7 s** y la decimada **≈ 0.3 s** (*matplotlib* ya simplifica los trazos largos, por lo que la diferencia aquí es **modesta**; en un navegador con elementos SVG sería mucho mayor, pero **no lo medimos**). (2) **Categorías raras:** la muestra uniforme de 2 000 deja **11** filas de la categoría `alerta` (0.5 % de los datos) y **47** de `aeropuerto`; la estratificada, **400** de cada una. Con los **pesos** $w_c = N_c/n_c$ los totales se recuperan **exactamente** y la media del valor es **9 779** (igual a la real), mientras que **sin pesos** sale **9 825 (+0.47 %)**: el sesgo es **pequeño** aquí porque `valor` casi no depende de la categoría; sería mayor si las categorías raras fueran muy distintas.

## 3. Tableau y Power BI frente a Datos Masivos (Conceptual) 🔌

Esta sección es **conceptual**: describe **cómo suelen** conectarse estas herramientas a datos grandes. No ejecutamos Tableau ni Power BI; los detalles de licencias, límites y versiones **cambian**, y lo que no es seguro está marcado *(verificar)*. Para el panorama de las herramientas, véase [Visual Analytics, Módulo 02](../../Visual%20Analytics%20and%20Critical%20Thinking/02%20-%20Fundamentos%20de%20Visualizacion/02_Herramientas_Tableau_PowerBI_D3.ipynb).

| Idea | Tableau | Power BI | Compromiso |
|---|---|---|---|
| **Copiar los datos a la herramienta** | **Extracto** (`.hyper`): copia columnar comprimida, **rápida** y con filtros/agregación al crearla *(Mastering Tableau, Cap. 12)* | **Modo de importación**: los datos se cargan en un modelo en memoria y se **refrescan** de forma programada *(Architecting Power BI, Cap. 5)* | **Rápido**, funciona sin conexión; pero los datos pueden estar **desactualizados** y deben **caber** en el modelo |
| **Preguntar al origen en cada interacción** | **Conexión en vivo** a una base o almacén (p. ej. BigQuery; Mastering Tableau, Cap. 6) | **`DirectQuery`**: cada visual lanza consultas al origen *(Architecting Power BI, Cap. 5)* | **Siempre al día** y sin duplicar, pero el **rendimiento depende del origen** y del número de visuales; hay **límites y funciones restringidas** *(verificar según versión)* |
| **Mezclar** | Extractos parciales y fuentes combinadas *(verificar cómo se combinan)* | **Modelo compuesto**: tablas en importación y otras en `DirectQuery` | Flexible, pero **más complejo** de razonar y de mantener |
| **Precalcular resúmenes** | Extractos **agregados** a un nivel de detalle y filtros de extracto | **Tablas de agregaciones** (manuales o automáticas): las consultas se responden desde un resumen pequeño y solo se baja al detalle cuando hace falta *(Cap. 5; la disponibilidad de las automáticas depende de la licencia: verificar)* | **Mucho más rápido**, pero solo contesta **lo que se precalculó** |

**El principio común** es el de esta unidad: *llevar la agregación adonde viven los datos y dibujar un resumen*. Los límites concretos (filas máximas por consulta de `DirectQuery`, tamaño máximo de un modelo, cuántas visuales por página) **dependen de la licencia y de la versión**: **verifícalos en la documentación vigente** antes de diseñar un tablero.

**Lo que una agregación no puede contestar.** Una tabla de **sumas y conteos** responde sumas, conteos y **promedios** (= suma / conteo), mínimos y máximos. **No** responde, de forma exacta, **conteos de distintos** (un cliente que compró en dos regiones se cuenta dos veces si sumas los distintos por región), **medianas** ni **percentiles**. Lo simulamos con 500 000 transacciones y comparamos «tabla de detalle» frente a «tabla agregada» en tiempos y resultados, y mostramos el efecto de un **extracto desactualizado**.

````python
# ============================================================
# 3. Simulacion con pandas: detalle frente a agregacion precalculada, y extracto frente a 'en vivo'
# ============================================================
rng = np.random.default_rng(SEED)
M = 500_000 if RAPIDO else 3_000_000
REGIONES = ["Tunja", "Duitama", "Sogamoso", "Chiquinquira", "Paipa", "Villa de Leyva"]
detalle = pd.DataFrame({
    "fecha": pd.Timestamp("2025-01-01") + pd.to_timedelta(rng.integers(0, 365, M), unit="D"),
    "region": rng.choice(REGIONES, M, p=[0.30, 0.22, 0.18, 0.12, 0.10, 0.08]),
    "producto": rng.integers(0, 20, M),
    "cliente": rng.integers(0, 120_000, M),
    "monto": np.round(rng.lognormal(8.0, 0.9, M), 0),
})
detalle["mes"] = detalle.fecha.dt.month

# Tabla de agregaciones: UN nivel de detalle (fecha x region x producto) con medidas ADITIVAS (suma y conteo)
t0 = time.perf_counter()
agg = detalle.groupby(["fecha", "region", "producto"], as_index=False).agg(suma=("monto", "sum"), n=("monto", "size"))
agg["mes"] = agg.fecha.dt.month
t_precalculo = time.perf_counter() - t0
print(f"Detalle: {len(detalle):,} filas ({detalle.memory_usage(deep=True).sum() / 2**20:.0f} MiB) | agregada: {len(agg):,} filas ({agg.memory_usage(deep=True).sum() / 2**20:.1f} MiB), precalculo = {t_precalculo:.2f} s (una vez)")


def consultas_detalle(d):                                      # 3 'interacciones' del tablero contra el DETALLE
    return (d.groupby("region").monto.sum(), d.groupby("mes").monto.mean(), d.groupby(["region", "mes"]).monto.size())


def consultas_agg(a):                                          # las mismas, contra la AGREGADA (promedio = suma / conteo)
    por_mes = a.groupby("mes")[["suma", "n"]].sum()
    return (a.groupby("region").suma.sum(), por_mes.suma / por_mes.n, a.groupby(["region", "mes"]).n.sum())


REP = 5
t0 = time.perf_counter(); [consultas_detalle(detalle) for _ in range(REP)]; t_det = (time.perf_counter() - t0) / REP
t0 = time.perf_counter(); [consultas_agg(agg) for _ in range(REP)]; t_agg = (time.perf_counter() - t0) / REP
rd, ra = consultas_detalle(detalle), consultas_agg(agg)
iguales = [np.allclose(a.sort_index().to_numpy(), b.sort_index().to_numpy()) for a, b in zip(rd, ra)]
print(f"3 consultas: detalle = {t_det * 1000:.0f} ms | agregada = {t_agg * 1000:.0f} ms ({t_det / t_agg:.0f}x mas rapido) | resultados iguales: {iguales}")
assert all(iguales) and t_agg < t_det

# Lo que la agregacion NO puede responder: clientes DISTINTOS por region
correcto = detalle.groupby("region").cliente.nunique().sum()                                   # distintos por region, sumados
total_distintos = detalle.cliente.nunique()                                                     # distintos en toda la empresa
suma_de_distintos = detalle.groupby(["region", "mes"]).cliente.nunique().groupby("region").sum().sum()     # lo que dariamos si 'sumamos' desde una agregada por mes
print(f"Clientes distintos en toda la empresa = {total_distintos:,} | suma de distintos por region = {correcto:,} | suma de distintos por region y mes = {suma_de_distintos:,}  <- NO es aditivo")
assert suma_de_distintos > correcto > total_distintos
mediana_real = detalle.monto.median()
mediana_mal = agg.assign(prom=agg.suma / agg.n).prom.median()
print(f"Mediana del monto: real = {mediana_real:,.0f} | 'mediana de promedios' desde la agregada = {mediana_mal:,.0f} ({mediana_mal / mediana_real - 1:+.1%})")
assert abs(mediana_mal - mediana_real) > 1

# Extracto frente a 'en vivo': llegan datos nuevos despues de crear el extracto
extracto_total = detalle.monto.sum()                                                            # el extracto es una FOTO
nuevos = pd.DataFrame({"monto": rng.lognormal(8.0, 0.9, 20_000).round(0)})
en_vivo_total = extracto_total + nuevos.monto.sum()                                             # 'en vivo' consulta la base actual
print(f"Extracto: {extracto_total:,.0f} | en vivo: {en_vivo_total:,.0f} | el extracto esta desactualizado por {en_vivo_total / extracto_total - 1:.2%} hasta el proximo refresco")
assert en_vivo_total > extracto_total
````

---
**Lectura.** (1) **La agregación precalculada** reduce **500 000** filas (**44 MiB**) a **43 760** (**3.9 MiB**) y responde las tres consultas **idénticas** (comparadas con `np.allclose`) en **≈ 25 ms frente a ≈ 250 ms** del detalle (**≈ 10 veces** más rápido; el precálculo, **0.23 s**, se paga **una vez**). A escala real, la brecha entre «detalle» y «resumen» es mucho mayor. (2) **Lo que no se puede responder:** hay **118 141** clientes distintos en la empresa; sumar los distintos **por región** da **343 021** y por **región y mes**, **483 168**: **no es aditivo** (un cliente aparece en varias regiones y meses). La «mediana de promedios» desde la agregada da **4 199** frente a la mediana real de **2 985** (**+40.7 %**): una agregación de sumas y conteos **no** sirve para medianas. (3) **Extracto frente a conexión en vivo:** con 20 000 transacciones nuevas el extracto queda **4.00 % desactualizado** hasta el siguiente refresco; es el precio de la velocidad. *(Todo esto es una **simulación en pandas**: no sustituye las pruebas en la herramienta real ni la lectura de su documentación.)*

## 4. D3.js: el Modelo de Enlace de Datos y las Escalas 🧩

**D3.js** (*Data-Driven Documents*) no es una biblioteca de gráficos prefabricados: es un conjunto de piezas para **enlazar datos con elementos del documento** (normalmente SVG) y **transformar datos en atributos visuales**. Dos ideas lo definen:

1. **Enlace de datos (*data join*).** Se **enlaza** un arreglo de datos con una selección de elementos mediante una **clave**. El resultado se parte en tres: **`enter`** (datos **sin** elemento: hay que crear uno), **`update`** (datos con elemento: se actualiza) y **`exit`** (elementos **sin** dato: se eliminan). Es la base de las transiciones y de la interacción: cambian los datos y el gráfico **se reconcilia** con ellos.
2. **Escalas (*scales*).** Funciones que **mapean un dominio** (valores del dato) **a un rango** (píxeles, colores, radios): `scaleLinear`, `scaleBand` (categorías → posiciones), `scaleSqrt` (para que el **área** sea proporcional al valor), `scaleSequential` (colores)... y los **ejes** (`axisBottom`, `axisLeft`) que dibujan marcas generadas por la escala.

En JavaScript, un gráfico de barras completo se escribe así *(este bloque **no** se ejecuta en el cuaderno: necesita un navegador y la biblioteca D3 v7; verifica la API en la documentación)*:

```javascript
const x = d3.scaleBand().domain(datos.map(d => d.zona)).range([0, ancho]).padding(0.1);
const y = d3.scaleLinear().domain([0, d3.max(datos, d => d.n)]).nice().range([alto, 0]);   // y invertido: el SVG crece hacia abajo

svg.selectAll("rect")
  .data(datos, d => d.zona)                      // enlace con CLAVE
  .join(
    enter  => enter.append("rect"),              // datos nuevos -> crear
    update => update,                            // datos que siguen -> actualizar
    exit   => exit.remove()                      // elementos sin dato -> eliminar
  )
  .attr("x", d => x(d.zona)).attr("y", d => y(d.n))
  .attr("width", x.bandwidth()).attr("height", d => alto - y(d.n));

svg.append("g").attr("transform", `translate(0,${alto})`).call(d3.axisBottom(x));
svg.append("g").call(d3.axisLeft(y));
```

**Para entender qué hace D3 sin salir de Python**, implementamos las escalas lineal, de bandas y raíz cuadrada, la generación de *ticks* «bonitos» (pasos 1, 2, 5 × $10^k$) y el enlace *enter/update/exit*; y dibujamos un gráfico de barras **colocando cada rectángulo con nuestras escalas en coordenadas de píxel** (eje $y$ invertido, como en SVG).

````python
# ============================================================
# 4. Las piezas de D3 implementadas en Python: escalas, ticks, enlace de datos y un grafico en 'pixeles'
# ============================================================
def paso_bonito(a, b, cuenta):
    """Paso 'bonito' entre ticks: 1, 2, 5 x 10^k (la idea de d3.ticks)."""
    paso = (b - a) / max(cuenta, 1)
    p = np.floor(np.log10(paso))
    err = paso / 10 ** p
    return (10 if err >= np.sqrt(50) else 5 if err >= np.sqrt(10) else 2 if err >= np.sqrt(2) else 1) * 10 ** p


def ticks(a, b, cuenta=5):
    s = paso_bonito(a, b, cuenta)
    return np.round(np.arange(np.ceil(a / s - 1e-9), np.floor(b / s + 1e-9) + 1) * s, 12) + 0.0          # +0.0 evita el '-0'


class ScaleLinear:
    def __init__(self, dominio, rango, expo=1.0):
        self.d, self.r, self.e = list(dominio), list(rango), expo           # expo = 0.5 -> scaleSqrt
    def __call__(self, v):
        t = (np.asarray(v, float) ** self.e - self.d[0] ** self.e) / (self.d[1] ** self.e - self.d[0] ** self.e)
        return self.r[0] + t * (self.r[1] - self.r[0])
    def invert(self, px):
        t = (np.asarray(px, float) - self.r[0]) / (self.r[1] - self.r[0])
        return (self.d[0] ** self.e + t * (self.d[1] ** self.e - self.d[0] ** self.e)) ** (1 / self.e)
    def nice(self, cuenta=5):
        s = paso_bonito(self.d[0], self.d[1], cuenta)
        self.d = [np.floor(self.d[0] / s) * s, np.ceil(self.d[1] / s) * s]
        return self
    def ticks(self, cuenta=5):
        return ticks(self.d[0], self.d[1], cuenta)


class ScaleBand:
    def __init__(self, dominio, rango, padding=0.1):
        self.d, self.r, n = list(dominio), rango, len(dominio)
        self.paso = (rango[1] - rango[0]) / (n - padding)                          # n bandas y n-1 huecos (relleno interno, sin externo)
        self.bandwidth = self.paso * (1 - padding)
    def __call__(self, categoria):
        return self.r[0] + self.d.index(categoria) * self.paso


def enlazar(elementos, datos, clave="zona"):
    """data join: devuelve (enter, update, exit) por CLAVE."""
    claves_datos = {d[clave] for d in datos}
    return sorted(claves_datos - set(elementos)), sorted(claves_datos & set(elementos)), sorted(set(elementos) - claves_datos)


# --- Pruebas de las escalas ---
s = ScaleLinear([0, 97], [0, 400])
assert s(0) == 0 and s(97) == 400 and np.isclose(s.invert(s(31.4)), 31.4)
assert list(ticks(0, 97, 5)) == [0, 20, 40, 60, 80] and list(ScaleLinear([0.13, 97.2], [0, 1]).nice(5).d) == [0, 100]
area = ScaleLinear([0, 100], [0, 30], expo=0.5)                                  # scaleSqrt: el RADIO crece como sqrt -> el AREA es proporcional al valor
assert np.isclose((area(64) / area(16)) ** 2, 4.0)                               # un valor 4 veces mayor -> area 4 veces mayor (radio 2 veces mayor)
banda = ScaleBand(["a", "b", "c", "d"], [0, 400], padding=0.1)
assert np.isclose(banda("d") + banda.bandwidth, 400)                          # la ultima banda termina justo en el borde del rango
assert np.isclose(banda("b") - banda("a"), banda.paso)

# --- Enlace de datos ---
antes = ["Tunja", "Duitama", "Sogamoso"]
despues = [{"zona": "Duitama", "n": 5}, {"zona": "Sogamoso", "n": 3}, {"zona": "Paipa", "n": 2}]
enter, update, exit_ = enlazar(antes, despues)
print(f"enter (crear) = {enter} | update (actualizar) = {update} | exit (eliminar) = {exit_}")
assert enter == ["Paipa"] and update == ["Duitama", "Sogamoso"] and exit_ == ["Tunja"]
assert set(enter) | set(update) == {d["zona"] for d in despues} and not (set(enter) & set(update))

# --- Un grafico de barras 'en pixeles' (como lo haria D3 con SVG): conteos por celdas de la dispersion ---
conteo = viajes.assign(zona=pd.cut(viajes.x, [-6, -2, 0, 2, 4, 7], labels=["A", "B", "C", "D", "E"])).groupby("zona", observed=True).size()
datos = [{"zona": k, "n": int(v)} for k, v in conteo.items()]
ANCHO, ALTO = 480, 260
sx = ScaleBand([d["zona"] for d in datos], [0, ANCHO], padding=0.15)
sy = ScaleLinear([0, max(d["n"] for d in datos)], [ALTO, 0]).nice(5)                        # rango invertido: 0 abajo (y crece hacia abajo en SVG)
fig, ax = plt.subplots(figsize=(7, 4.2))
ax.set_xlim(-50, ANCHO + 20); ax.set_ylim(ALTO + 40, -20); ax.axis("off")                    # lienzo en PIXELES, con y invertido
for d in datos:
    ax.add_patch(plt.Rectangle((sx(d["zona"]), sy(d["n"])), sx.bandwidth, ALTO - sy(d["n"]), color=PALETA[0]))
    ax.text(sx(d["zona"]) + sx.bandwidth / 2, ALTO + 14, d["zona"], ha="center", fontsize=9)
for tk in sy.ticks(5):
    ax.plot([-6, 0], [sy(tk), sy(tk)], color="#334155", lw=1); ax.text(-10, sy(tk), f"{tk:,.0f}", ha="right", va="center", fontsize=8)
ax.plot([0, 0], [sy(sy.d[0]), sy(sy.d[1])], color="#334155", lw=1); ax.plot([0, ANCHO], [ALTO, ALTO], color="#334155", lw=1)
ax.set_title("Barras dibujadas con NUESTRAS escalas (coordenadas de pixel)", fontsize=10)
plt.tight_layout(); plt.show()
print("Ticks del eje y:", sy.ticks(5).tolist(), "| dominio tras nice():", [float(v) for v in sy.d])
````

````python
# ============================================================
# 5. Exportar AGREGADOS (no el detalle) a JSON y CSV, listos para D3 / Power BI / Tableau (directorio temporal)
# ============================================================
# (1) Mapa de calor: rejilla 60 x 60 con las celdas NO vacias, mas metadatos
NX = NY = 60
Hm, xe, ye = np.histogram2d(viajes.x, viajes.y, bins=[NX, NY], range=RANGO)
ix, iy = np.nonzero(Hm)
mapa = {"meta": {"fuente": "datos SINTETICOS (semilla 42)", "n_total": int(N), "nx": NX, "ny": NY, "x_min": RANGO[0][0], "x_max": RANGO[0][1],
                 "y_min": RANGO[1][0], "y_max": RANGO[1][1], "agregacion": "conteo de puntos por celda", "celdas_vacias_omitidas": True},
        "celdas": [{"x0": round(float(xe[i]), 4), "y0": round(float(ye[j]), 4), "n": int(Hm[i, j])} for i, j in zip(ix, iy)]}
ruta_json = f"{TMP}/mapa_calor.json"
with open(ruta_json, "w", encoding="utf-8") as f:
    json.dump(mapa, f, ensure_ascii=False)

# (2) Serie por hora (CSV): promedio, minimo y maximo -> lo que pide una herramienta de BI
horas = (viajes.assign(hora=np.arange(N) % 24).groupby("hora").valor.agg(promedio="mean", minimo="min", maximo="max", n="size").round(1).reset_index())
ruta_csv = f"{TMP}/valor_por_hora.csv"
horas.to_csv(ruta_csv, index=False)

# (3) El detalle crudo, SOLO para comparar tamanos
ruta_crudo = f"{TMP}/viajes_crudo.csv"
viajes.to_csv(ruta_crudo, index=False, float_format="%.4f")
tam = {k: os.path.getsize(p) for k, p in (("crudo.csv", ruta_crudo), ("mapa_calor.json", ruta_json), ("valor_por_hora.csv", ruta_csv))}
print({k: f"{v / 1024:,.1f} KiB" for k, v in tam.items()}, f"| celdas no vacias: {len(mapa['celdas']):,} de {NX * NY:,}")

# Verificacion: se relee el JSON y se comprueba que no se perdio nada (y que se documento que el rango deja fuera algunos puntos)
with open(ruta_json, encoding="utf-8") as f:
    relei = json.load(f)
n_en_celdas = sum(c["n"] for c in relei["celdas"])
dentro = int(((viajes.x >= RANGO[0][0]) & (viajes.x <= RANGO[0][1]) & (viajes.y >= RANGO[1][0]) & (viajes.y <= RANGO[1][1])).sum())
print(f"Puntos en celdas = {n_en_celdas:,} | puntos dentro del rango = {dentro:,} | total = {N:,}  (los {N - dentro:,} restantes quedan fuera del rango y hay que declararlo)")
assert n_en_celdas == dentro and relei["meta"]["n_total"] == N
assert tam["mapa_calor.json"] < tam["crudo.csv"] / 50 and tam["valor_por_hora.csv"] < 2_000                   # el agregado pesa una fraccion minima
assert list(pd.read_csv(ruta_csv).columns) == ["hora", "promedio", "minimo", "maximo", "n"] and pd.read_csv(ruta_csv).n.sum() == N

# (4) Una pagina minima de D3 (se escribe en el directorio temporal; NO se ejecuta aqui: necesita navegador y red para el CDN)
html = """<!doctype html><meta charset="utf-8"><title>Mapa de calor</title>
<script src="https://cdn.jsdelivr.net/npm/d3@7"></script>
<script>
d3.json("mapa_calor.json").then(d => {
  const W = 480, H = 480, m = d.meta;
  const svg = d3.select("body").append("svg").attr("width", W).attr("height", H);
  const x = d3.scaleLinear().domain([m.x_min, m.x_max]).range([0, W]);
  const y = d3.scaleLinear().domain([m.y_min, m.y_max]).range([H, 0]);
  const c = d3.scaleSequentialLog(d3.interpolateYlGnBu).domain([1, d3.max(d.celdas, e => e.n)]);
  const w = W / m.nx, h = H / m.ny;
  svg.selectAll("rect").data(d.celdas).join("rect")
     .attr("x", e => x(e.x0)).attr("y", e => y(e.y0) - h).attr("width", w).attr("height", h).attr("fill", e => c(e.n));
});
</script>"""
with open(f"{TMP}/index.html", "w", encoding="utf-8") as f:
    f.write(html)
print("Archivos listos en el directorio temporal:", sorted(os.listdir(TMP)), "\n(para verlo: `python -m http.server` en esa carpeta y abrir index.html; no probado aqui)")
````

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Visualizar a escala** | Una pantalla tiene ~$10^6$ píxeles: dibujar $N \gg 10^6$ marcas **sobredibuja, tarda y no cabe**. **Agrega antes de graficar** y **declara** la agregación |
| **Histograma 2D / *hexbin*** | Contar por celda es $O(N)$ una vez; luego se dibujan $B^2$ números. Color en **escala logarítmica** (y decirlo). El ancho de celda es una decisión de diseño |
| **Series: min-máx** | Conserva el mínimo y el máximo por cubo de píxeles: **no pierde picos**; el muestreo uniforme sí |
| **Categorías raras** | Muestreo **estratificado con pesos** $w_c = N_c/n_c$: ve lo raro y recupera los totales; sin pesos sobrerrepresenta |
| **Tableau y Power BI** | Extracto/importación (rápido, desactualizable, debe caber) frente a conexión en vivo/`DirectQuery` (fresco, depende del origen) y **modelos compuestos** y **agregaciones**. Límites y licencias: **verificar** |
| **Agregaciones precalculadas** | Muy rápidas para medidas **aditivas** (suma, conteo, promedio = suma/conteo); **no** responden distintos, medianas ni percentiles |
| **D3.js** | **Enlace de datos** (`enter`/`update`/`exit` por clave) + **escalas** (dominio → rango) + ejes; control total a costa de código. Lo importante: **enviar agregados**, no el detalle |
| **Exportar** | JSON/CSV **pequeños con metadatos** (fuente, agregación, rango, puntos fuera del rango) para D3, Power BI o Tableau |
