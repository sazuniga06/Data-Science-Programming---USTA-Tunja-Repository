## 1. Un Alfiler por Casa: la Mancha 📍

Un pueblo (inventado) tiene **20 000 casas**: un **centro** muy denso, un **barrio** más abierto y casas **dispersas** por el campo. Dibujamos el plano de dos maneras: **un punto por casa** (izquierda) y **contando casas por manzana** en una cuadrícula de 10 × 10 (derecha).

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
x = np.concatenate([rng.normal(0, 0.6, 9000), rng.normal(4, 1.0, 7000), rng.uniform(-5, 8, 4000)])
y = np.concatenate([rng.normal(0, 0.6, 9000), rng.normal(3, 0.8, 7000), rng.uniform(-4, 6, 4000)])
assert len(x) == 20_000

H, xe, ye = np.histogram2d(x, y, bins=10)                          # cuantas casas caen en cada manzana
fig, ax = plt.subplots(1, 2, figsize=(11, 4.2))
ax[0].scatter(x, y, s=18, color="#2563eb"); ax[0].set_title("Un punto por casa: ¡una mancha!", fontsize=10, fontweight="bold")
im = ax[1].imshow(H.T, origin="lower", extent=[xe[0], xe[-1], ye[0], ye[-1]], cmap="YlOrRd", aspect="auto")
for i in range(10):
    for j in range(10):
        if H[i, j] > 800:
            ax[1].text((xe[i] + xe[i + 1]) / 2, (ye[j] + ye[j + 1]) / 2, f"{int(H[i, j]):,}", ha="center", va="center", fontsize=7)
ax[1].set_title("Casas por manzana: se ve dónde vive la gente", fontsize=10, fontweight="bold")
fig.colorbar(im, ax=ax[1], label="casas por manzana")
plt.tight_layout(); plt.show()

top = pd.Series(H.ravel()).sort_values(ascending=False)
print("Las 3 manzanas más pobladas tienen:", [int(v) for v in top.head(3)], "casas | la manzana típica (mediana de las ocupadas):", int(np.median(top[top > 0])))
print(f"El 10 % de las manzanas concentra el {top.head(10).sum() / top.sum():.0%} de las casas.")
assert H.sum() <= 20_000 and H.sum() > 19_000
assert top.iloc[0] > 10 * np.median(top[top > 0])
````

---
**Lección:** con **un punto por casa**, el centro es una mancha azul donde **no se sabe si hay 100 casas o 9 000**; con **conteos por manzana**, se ve de inmediato **dónde está la gente** (unas pocas manzanas concentran la mayoría) y la figura **pesa lo mismo** si el pueblo tiene 20 000 casas o 20 millones, porque solo dibuja **100 cuadritos**. El precio: ya **no ves cada casa**; ves un **resumen**. Por eso hay que **decirlo** («cada cuadro cuenta las casas de la manzana») y que **el tamaño de la manzana** es una decisión de quien dibuja.

## 2. El Libro de Caja del Mes 📒

Un almacén de Duitama hizo **30 000 ventas** en el mes (inventadas). Para que el tablero **no se demore**, alguien prepara un **libro de caja**: **un renglón por día y por producto**, con el **total vendido** y **cuántas ventas fueron**. ¿Se pierde algo? Depende de **qué pregunta** hagas.

````python
rng = np.random.default_rng(7)
ventas = pd.DataFrame({
    "dia": rng.integers(1, 31, 30_000),
    "producto": rng.choice(["papa", "cebolla", "arveja", "zanahoria", "mora"], 30_000, p=[0.35, 0.25, 0.15, 0.15, 0.10]),
    "cliente": rng.integers(1, 3001, 30_000),
    "valor": rng.integers(2, 60, 30_000) * 100,
})
libro = ventas.groupby(["dia", "producto"], as_index=False).agg(total=("valor", "sum"), ventas=("valor", "size"))
print(f"{len(ventas):,} ventas se resumen en {len(libro)} renglones del libro de caja.")

# Pregunta 1: ¿cuanto se vendio de cada producto? -> el libro responde EXACTO
del_detalle = ventas.groupby("producto").valor.sum().sort_index()
del_libro = libro.groupby("producto").total.sum().sort_index()
print("\n¿Cuánto se vendió de cada producto?\n", pd.DataFrame({"del detalle": del_detalle, "del libro": del_libro}).to_string())
assert (del_detalle == del_libro).all()

# Pregunta 2: ¿cuantos clientes DISTINTOS compraron cada dia? Un ejemplo chiquito primero
mini = pd.DataFrame({"dia": ["lunes", "lunes", "martes", "martes"], "cliente": ["Ana", "Beto", "Ana", "Carla"]})
por_dia = mini.groupby("dia").cliente.nunique()
print("\nEjemplo chiquito: clientes distintos por día:", por_dia.to_dict(), "| suma de los días =", int(por_dia.sum()), "| clientes distintos de verdad =", mini.cliente.nunique())
assert por_dia.sum() == 4 and mini.cliente.nunique() == 3

dia_sum = ventas.groupby("dia").cliente.nunique().sum()
verdad = ventas.cliente.nunique()
print(f"En el almacén: suma de clientes distintos por día = {dia_sum:,} | clientes distintos del mes = {verdad:,}")
assert dia_sum > verdad
````

---
**Lección:** el libro de caja **responde exacto** preguntas que **se suman** (cuánto se vendió, cuántas ventas, el promedio = total ÷ ventas) y lo hace **al instante** (150 renglones en vez de 30 000). Pero hay preguntas que **no se pueden sumar**: *«¿cuántos clientes distintos compraron?»*: Ana compró lunes **y** martes, así que sumar los clientes de cada día la cuenta **dos veces** (4 en lugar de 3). Y lo mismo pasa con la **mediana**. Por eso, al preparar un resumen, hay que **pensar de antemano qué preguntas** se van a hacer. *(Los ingenieros usan tablas resumen en Power BI o extractos en Tableau por esta razón: son rápidas, pero solo contestan lo que se preparó.)*

## 3. La Regla que Convierte Litros en Centímetros 📏

Dibujar una barra es **convertir una medida**: si **500 litros** deben ocupar **10 cm** de papel, entonces **250 litros** deben ocupar **5 cm**. A esa regla la llaman **escala**. **Tableau y Power BI** traen las escalas **ya puestas** (como una plantilla de bordado: rápido y bonito, pero con el diseño que trae); **D3.js** te da las **piezas sueltas** (como Lego: puedes armar lo que quieras, pero hay que armarlo). Veamos la regla en acción con la leche recogida en cinco veredas (datos inventados).

````python
def escala(litros, dominio=(0, 500), rango=(0, 10)):
    """Convierte litros (dominio) en centimetros de barra (rango): una regla de tres."""
    (d0, d1), (r0, r1) = dominio, rango
    return r0 + (litros - d0) / (d1 - d0) * (r1 - r0)


assert escala(0) == 0 and escala(500) == 10 and escala(250) == 5

veredas = pd.Series({"El Salitre": 420, "La Concepción": 275, "Pantano de Vargas": 130, "San Isidro": 360, "Monquirá": 85})
cm = veredas.map(escala)                                            # largo de cada barra, en centimetros
print(pd.DataFrame({"litros": veredas, "centímetros de barra": cm.round(1)}).to_string())

fig, ax = plt.subplots(figsize=(7, 3.4))
ax.barh(list(veredas.index), cm, color="#2563eb")                    # el largo sale SOLO de la regla
for nombre, c, l in zip(veredas.index, cm, veredas):
    ax.text(c + 0.1, nombre, f"{l} L", va="center", fontsize=9)
ax.set_xlim(0, 11); ax.set_xlabel("centímetros de papel"); ax.set_title("Cada barra mide lo que dice la regla", fontsize=10, fontweight="bold")
ax.spines[["top", "right"]].set_visible(False); plt.tight_layout(); plt.show()

assert abs(cm["El Salitre"] - 8.4) < 1e-9 and cm.max() <= 10
````

---
**Lección:** cada barra mide **exactamente lo que dice la regla**: 420 litros → 8.4 cm; 85 litros → 1.7 cm. Cuando cambian los datos, **solo se vuelve a aplicar la regla**. D3.js hace esto con **escalas** para posiciones, colores y tamaños; Tableau y Power BI lo hacen por ti. Si el dibujo tiene **millones de datos**, primero se **resumen** (por cuadrícula, por día…) y **luego** se aplica la regla a los pocos números que quedan.

## Resumen en una frase 🎯

> **Cuando hay demasiados datos no se dibuja uno por uno: se cuenta por cuadrícula o se prepara un libro de caja, y luego una regla (escala) convierte cada número en un tamaño, color o posición; las herramientas listas ahorran trabajo y D3 da libertad total.**

### Lo que aprendiste hoy:
- ✅ **Sobredibujo:** un alfiler por casa tapa el centro; **contar por manzana** muestra dónde está la gente.
- ✅ **Libro de caja:** responde **exacto y rápido** lo que se suma (totales, promedios) pero **no** los clientes distintos (4 ≠ 3) ni la mediana.
- ✅ **Escala:** una regla de tres convierte litros en centímetros; es la idea central de D3, Tableau y Power BI.
