## 1. Las Cuatro Preguntas de Don Álvaro 🍅

Don Álvaro anotó los **kilos de tomate** que vendió durante **4 semanas** (datos **inventados**). En la semana 3, el sábado **llovió** y casi nadie salió al mercado. Vamos a contestar sus cuatro preguntas **con la misma tabla**.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

dias = ["lun", "mar", "mié", "jue", "vie", "sáb", "dom"]
kilos = [40, 42, 41, 45, 60, 95, 70,        # semana 1
         41, 43, 40, 46, 62, 97, 72,        # semana 2
         39, 41, 42, 44, 58, 66, 69,        # semana 3  (el sabado llovio)
         42, 44, 43, 47, 63, 99, 74]        # semana 4
ventas = pd.DataFrame({"semana": np.repeat([1, 2, 3, 4], 7), "dia": dias * 4, "kilos": kilos})
ventas["llovio"] = (ventas.semana == 3) & (ventas.dia == "sáb")
assert len(ventas) == 28

# 1) DESCRIPTIVA: que paso
por_semana = ventas.groupby("semana").kilos.sum()
print("1) ¿Cuántos kilos vendí por semana?\n", por_semana.to_string())

# 2) DIAGNOSTICA: por que la semana 3 fue la mas baja
sabados = ventas[ventas.dia == "sáb"].set_index("semana")
print("\n2) ¿Por qué bajó la semana 3? Miremos los sábados (kilos):\n", sabados.kilos.to_string())
print("   El sábado de la semana 3 vendí", int(sabados.loc[3, "kilos"]), "kg: ese día llovió.")

# 3) PREDICTIVA: el proximo sabado = promedio de los sabados SIN lluvia
sabados_secos = sabados[~sabados.llovio].kilos
prediccion = sabados_secos.mean()
print(f"\n3) Próximo sábado: lo normal es vender unos {prediccion:.0f} kg (promedio de los sábados sin lluvia: {sabados_secos.tolist()})")

# 4) PRESCRIPTIVA: cuanto pedir? Si sobra tomate se pierde 1.900 pesos por kilo; si falta, se pierden 900 pesos de ganancia por kilo
def costo(pedido, vendido):
    return sum(900 * max(v - pedido, 0) + 1900 * max(pedido - v, 0) for v in vendido)

pedidos = list(range(70, 111, 5))
costos = {p: costo(p, sabados.kilos.tolist()) for p in pedidos}                 # probamos cada pedido con los 4 sabados que ya vimos
mejor = min(costos, key=costos.get)
print(f"\n4) Costo de equivocarse en los 4 sábados, según cuánto pido (pesos): {costos}")
print(f"   El pedido con menor costo es {mejor} kg, y no el promedio ({prediccion:.0f} kg): sobrar sale más caro que quedarse corto.")

assert por_semana.idxmin() == 3
assert prediccion > 96 and mejor <= prediccion
````

---
**Lección:** las cuatro preguntas usan **la misma tabla**, pero van **subiendo**: contar (**¿qué pasó?**), buscar la causa (**¿por qué?**: la lluvia del sábado 3), calcular lo normal (**¿qué pasará?**: unos 97 kg) y **decidir** (**¿qué hago?**: pedir **un poco menos** que ese promedio (95 kg en lugar de 97), porque lo que sobra se daña y sale más caro que lo que se deja de vender). Fíjate en que **predecir no es decidir**: la mejor decisión depende de **cuánto cuesta equivocarse** hacia cada lado.

## 2. Preguntar a Pocos para Conocer a Muchos 🗳️

En la plaza de mercado hay **10 000 personas** (inventadas) y se quiere saber **cuántas prefieren pagar en efectivo** (en realidad, **alrededor del 62 %**; eso lo sabe la computadora, no nosotros). Preguntar a todos es carísimo. Entonces **elegimos al azar** a 10, 100 o 1 000 personas. Repetimos cada encuesta **200 veces** para ver **cuánto se equivoca**.

````python
rng = np.random.default_rng(42)
poblacion = (rng.random(10_000) < 0.62).astype(int)            # 1 = prefiere efectivo, 0 = otro medio
real = poblacion.mean()
print(f"Porcentaje REAL de la plaza (lo sabe la computadora): {real:.1%}")

filas = []
for n in (10, 100, 1000):
    estimados = np.array([rng.choice(poblacion, n, replace=False).mean() for _ in range(200)])
    filas.append((n, estimados.mean(), np.sqrt(np.mean((estimados - real) ** 2))))
tabla = pd.DataFrame(filas, columns=["personas preguntadas", "promedio de las encuestas", "error típico"]).set_index("personas preguntadas")
print(tabla.round(3).to_string())
print(f"\nPreguntar a 100 en vez de a 10 divide el error por ≈ {tabla.iloc[0, 1] / tabla.iloc[1, 1]:.1f}; a 1 000, por ≈ {tabla.iloc[0, 1] / tabla.iloc[2, 1]:.1f}.")

fig, ax = plt.subplots(figsize=(6, 3.4))
ax.bar([str(n) for n in tabla.index], tabla["error típico"] * 100, color=["#dc2626", "#f59e0b", "#16a34a"])
ax.set_xlabel("personas preguntadas"); ax.set_ylabel("error típico (puntos porcentuales)")
ax.set_title("Más encuestados, menos error (pero cada vez menos)", fontsize=10, fontweight="bold")
plt.tight_layout(); plt.show()

assert tabla.iloc[0, 1] > tabla.iloc[1, 1] > tabla.iloc[2, 1]
assert tabla.iloc[0, 1] / tabla.iloc[2, 1] > 5                  # 100 veces mas gente -> el error baja unas 10 veces (no 100)
````

---
**Lección:** con **10 personas** la encuesta se equivoca mucho (**≈ 15 puntos**); con **100**, **≈ 5**; con **1 000**, **≈ 1.4**. Fíjate en lo importante: **100 veces más gente** reduce el error **unas 10 veces**, no 100: para ser **el doble de preciso** hay que preguntar a **cuatro veces más**. Y otra cosa notable: da casi igual si la plaza tiene 10 000 personas o 10 millones: lo que manda es **cuántos preguntas**, no **cuántos hay**. Por eso en Big Data **muchas veces basta una muestra bien tomada**.

## 3. La Rifa Justa cuando No Sabes Cuántos Van a Llegar 🎟️

En la feria llegan personas **una por una** y nadie sabe **cuántas** serán. Hay que tener **siempre 3 posibles ganadoras** (en una bolsa de 3 puestos) y, al final, **cada persona debe haber tenido la misma oportunidad**. El truco (se llama *muestreo de reservorio*): las 3 primeras entran de una; a la persona que llega en el lugar **n**, se le da **3 de n** posibilidades de entrar y, si entra, **saca a una al azar**. Probemos con 10 personas, repitiendo la rifa **6 000 veces**.

````python
def rifa(personas, puestos, rng):
    bolsa = []
    for n, p in enumerate(personas, start=1):               # n = cuantas personas han llegado
        if n <= puestos:
            bolsa.append(p)
        elif rng.random() < puestos / n:                    # entra con probabilidad puestos / n
            bolsa[rng.integers(0, puestos)] = p             # y saca a una al azar
    return bolsa


rng = np.random.default_rng(7)
personas = ["Ana", "Beto", "Carla", "Diego", "Elena", "Fabio", "Gloria", "Hugo", "Inés", "Jorge"]
veces = pd.Series(0, index=personas)
for _ in range(6000):
    for ganadora in rifa(personas, 3, rng):
        veces[ganadora] += 1
frecuencia = veces / 6000
print((frecuencia.round(3)).to_string())
print(f"\nLo justo sería 3 de 10 = 0.300 para cada una | mínimo = {frecuencia.min():.3f}, máximo = {frecuencia.max():.3f}")
assert frecuencia.between(0.27, 0.33).all()                  # nadie esta favorecido (con 6 000 repeticiones)
````

---
**Lección:** aunque Ana llegó **primera** y Jorge **última**, ambos ganan **cerca del 30 %** de las veces: la rifa es **justa** sin saber de antemano cuántos llegarán y **sin guardar a todos** (solo 3 puestos de memoria). Esa es la idea de las **estructuras aproximadas** de Big Data: **resumir un flujo enorme** con una **memoria pequeña y fija**, aceptando un **error pequeño y controlado**.

## Resumen en una frase 🎯

> **Con unas pocas notas se contestan cuatro preguntas (qué pasó, por qué, qué pasará y qué hacer), y cuando son demasiadas, basta con preguntar a una muestra bien tomada o con resumir el flujo con una memoria pequeña, aceptando un error que se puede medir.**

### Lo que aprendiste hoy:
- ✅ **Las cuatro preguntas:** contar, buscar la causa (la lluvia), predecir (≈ 97 kg el próximo sábado) y **decidir** (pedir un poco menos, porque sobrar cuesta más que faltar).
- ✅ **Preguntar a pocos:** 100 veces más gente baja el error **unas 10 veces**; el tamaño de la plaza **casi no importa**.
- ✅ **La rifa justa:** con 3 puestos y una regla sencilla, **todos** tienen la misma oportunidad.
