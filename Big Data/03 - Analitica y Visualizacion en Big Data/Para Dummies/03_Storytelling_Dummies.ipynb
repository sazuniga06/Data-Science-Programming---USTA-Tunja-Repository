Es como contar un **cuento a los niños**: tiene un **inicio**, un **nudo** y un **desenlace**, y quien lo cuenta **no inventa** ni **esconde** cosas.

| En un cuento | En una historia con datos |
|---|---|
| **Inicio:** quién y dónde | **Contexto:** cómo estábamos |
| **Nudo:** lo que pasó y cambia todo | **Conflicto:** el hallazgo (lo que no cuadra) |
| **Desenlace:** cómo se arregla | **Resolución:** qué hacemos |
| **El título del cuento** | **El título del gráfico:** debe decir la **conclusión** |
| **No hacer trampa al contar** | **No usar reglas torcidas** (eje truncado) ni **escoger solo los meses que convienen** |

## 1. La Panadería de Doña Rosa 🥐

Doña Rosa tiene una panadería en **Villa de Leyva** y anotó las **unidades vendidas por mes** de tres productos durante un año (datos **inventados**). Quiere contarle a su hijo **qué está pasando**. Primero el gráfico «de explorar» (tres colores y una leyenda) y luego el «de explicar» (un solo color vivo, el título dice la conclusión y la etiqueta va junto a la línea).

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

meses = ["ene", "feb", "mar", "abr", "may", "jun", "jul", "ago", "sep", "oct", "nov", "dic"]
panaderia = pd.DataFrame({
    "pan":        [820, 805, 812, 798, 830, 815, 808, 822, 810, 800, 818, 826],
    "pandebono":  [410, 395, 405, 420, 400, 415, 408, 402, 412, 398, 406, 411],
    "almojábana": [300, 318, 335, 352, 380, 401, 425, 450, 478, 503, 531, 560],
}, index=meses)
crecimiento = (panaderia.iloc[-3:].mean() / panaderia.iloc[:3].mean() - 1) * 100       # ultimos 3 meses frente a los primeros 3
print("Crecimiento (últimos 3 meses frente a los primeros 3), en %:\n", crecimiento.round(1).to_string())
assert crecimiento.idxmax() == "almojábana" and crecimiento["almojábana"] > 50 and abs(crecimiento["pan"]) < 3

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
# --- ANTES: para explorar ---
for col in panaderia:
    ax[0].plot(panaderia.index, panaderia[col], label=col)
ax[0].legend(); ax[0].set_title("Ventas"); ax[0].grid(True)
# --- DESPUES: para explicar ---
for col in ("pan", "pandebono"):
    ax[1].plot(panaderia.index, panaderia[col], color="#9ca3af", lw=2)
    ax[1].text(11.2, panaderia[col].iloc[-1], col, color="#6b7280", va="center", fontsize=9)
ax[1].plot(panaderia.index, panaderia["almojábana"], color="#dc2626", lw=3.5)
ax[1].text(11.2, panaderia["almojábana"].iloc[-1], "almojábana", color="#dc2626", va="center", fontsize=10, fontweight="bold")
ax[1].annotate("de 300 a 560 en un año", xy=(8, panaderia["almojábana"].iloc[8]), xytext=(1.2, 600), arrowprops=dict(arrowstyle="->", color="#dc2626"), color="#dc2626", fontsize=9)
ax[1].set_title(f"La almojábana creció {crecimiento['almojábana']:.0f} % y lo demás no se movió", loc="left", fontsize=9.5, fontweight="bold")
ax[1].set_ylim(0, 950); ax[1].set_xlim(-0.3, 13); ax[1].spines[["top", "right"]].set_visible(False)
plt.tight_layout(); plt.show()
````

---
**Lección:** los dos dibujos tienen **exactamente los mismos números**, pero el de la derecha **cuenta algo**: el **título** ya dice la conclusión (*«la almojábana creció 67 % y lo demás no se movió»*), **un solo color** muestra a la protagonista y el resto va en **gris**, los **nombres van junto a las líneas** (sin leyenda para buscar) y una **flecha** dice qué mirar. Recuerda la regla: **un gráfico, un mensaje**. En el cuento de Doña Rosa: *inicio* (el pan y el pandebono se venden parejo), *nudo* (la almojábana se dispara), *desenlace* (¿preparar más masa y más horno?).

## 2. La Regla Torcida 📏

En abril la panadería vendió **98** pasteles y en mayo **100**: **2 más**, una diferencia **chiquita**. Dibujamos dos gráficos de barras con los mismos dos números: uno con la regla **empezando en 0** y otro con la regla **cortada en 96** (el dibujo solo muestra desde 96 para arriba). ¿Cuántas veces **más alta** parece la barra de mayo en cada uno?

````python
abril, mayo = 98, 100

def parece_veces(base):
    """Cuantas veces mas alta SE VE la barra de mayo, si la regla empieza en 'base'."""
    return (mayo - base) / (abril - base)

print(f"Regla desde 0:  la barra de mayo se ve {parece_veces(0):.2f} veces la de abril (en realidad: {mayo / abril:.2f})")
print(f"Regla desde 96: la barra de mayo se ve {parece_veces(96):.2f} veces la de abril (¡el doble por solo 2 pasteles!)")
assert abs(parece_veces(0) - mayo / abril) < 1e-12 and parece_veces(96) == 2.0

fig, ax = plt.subplots(1, 2, figsize=(8, 3.4))
for a, base, titulo in ((ax[0], 96, "Regla cortada en 96 (engaña)"), (ax[1], 0, "Regla desde 0 (honesta)")):
    a.bar(["abril", "mayo"], [abril, mayo], color=["#9ca3af", "#2563eb"]); a.set_ylim(base, 102); a.set_title(titulo, fontsize=10)
    a.spines[["top", "right"]].set_visible(False); a.set_ylabel("pasteles vendidos")
plt.tight_layout(); plt.show()
````

---
**Lección:** con la regla **desde cero**, la barra de mayo es apenas **2 % más alta** (como en la realidad); con la regla **cortada en 96**, **parece el doble**. Nadie dijo una mentira con palabras, pero **el dibujo miente**. Por eso, en **gráficos de barras**, la regla **siempre empieza en cero**. *(En un gráfico de **líneas** se puede empezar más arriba, siempre que se vea claro.)*

## 3. Escoger los Meses que Convienen 🗓️

Una tienda mide cada mes **cuántas entregas llegaron a tiempo** (en %). La verdad es que **no hay tendencia**: unos meses salen un poco mejor y otros un poco peor **por puro azar** (datos inventados). Veamos qué pasa si alguien **escoge cuatro meses seguidos**: ¿se puede contar «estamos mejorando» y también «estamos empeorando» con **los mismos datos**?

````python
rng = np.random.default_rng(7)
entregas = 90 + rng.normal(0, 2.0, 12)                               # 12 meses, promedio 90 %, sin tendencia
ventanas = {f"meses {i + 1}-{i + 4}": entregas[i + 3] - entregas[i] for i in range(9)}     # cambio entre el primer y el ultimo mes de cada ventana
sube = [k for k, v in ventanas.items() if v > 0]
baja = [k for k, v in ventanas.items() if v < 0]
mejor, peor = max(ventanas, key=ventanas.get), min(ventanas, key=ventanas.get)
print(f"De 9 ventanas de 4 meses, {len(sube)} 'suben' y {len(baja)} 'bajan'.")
print(f"Podría decir «mejoramos {ventanas[mejor]:+.1f} puntos» (mirando {mejor}) o «empeoramos {ventanas[peor]:+.1f} puntos» (mirando {peor}).")
print(f"Pero en todo el año: de {entregas[0]:.1f} % a {entregas[-1]:.1f} % ({entregas[-1] - entregas[0]:+.1f} puntos), con un promedio de {entregas.mean():.1f} %.")
assert len(sube) > 0 and len(baja) > 0
assert ventanas[mejor] > 2 and ventanas[peor] < -1

fig, ax = plt.subplots(figsize=(7.5, 3.2))
ax.plot(range(1, 13), entregas, color="#9ca3af", lw=1.8)
i, j = int(mejor.split()[1].split("-")[0]) - 1, int(peor.split()[1].split("-")[0]) - 1
ax.plot(range(i + 1, i + 5), entregas[i:i + 4], color="#2563eb", lw=3.5, label="«mejoramos»")
ax.plot(range(j + 1, j + 5), entregas[j:j + 4], color="#dc2626", lw=3.5, label="«empeoramos»")
ax.axhline(90, color="#374151", ls=":", lw=1); ax.legend(frameon=False); ax.set_xlabel("mes"); ax.set_ylabel("% de entregas a tiempo")
ax.set_title("Los mismos datos cuentan dos historias opuestas", loc="left", fontsize=10, fontweight="bold"); ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout(); plt.show()
````

---
**Lección:** con los **mismos 12 datos** se puede sostener que **«mejoramos»** o que **«empeoramos»**, **según los 4 meses que se escojan**. Es muy fácil caer en esto **sin mala intención**: ya sabemos lo que queremos decir y buscamos los meses que lo confirman. La defensa es sencilla: **decidir antes** qué meses se van a mirar, **mostrar el año completo** y **decir cuántas ventanas** apoyan cada conclusión.
