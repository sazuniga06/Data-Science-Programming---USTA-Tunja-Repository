## 1. El Modelo MapReduce 🗂️

Un trabajo MapReduce transforma **pares clave-valor**. El programador escribe:

$$\text{map}:\ (k_1, v_1) \;\to\; \text{lista}\,(k_2, v_2) \qquad\qquad \text{reduce}:\ \big(k_2,\ \text{lista}(v_2)\big) \;\to\; \text{lista}\,(k_3, v_3)$$

y el *framework* ejecuta estas **fases**:

| # | Fase | Qué hace | Dónde vive el resultado |
|---|---|---|---|
| 1 | **Partición de la entrada** (*splits*) | Parte la entrada en trozos que se procesan en paralelo (en Hadoop, ≈ un bloque de HDFS) | Archivos en HDFS |
| 2 | **`map`** | Cada tarea lee un trozo y emite pares $(k_2, v_2)$ | Disco **local** del nodo |
| 3 | ***Combiner*** (opcional) | **Reduce localmente** la salida del `map` antes de enviarla (un «mini-`reduce`» por tarea) | Disco local |
| 4 | **Partición por clave** | Decide a qué *reducer* va cada clave: $\text{particion}(k) = h(k) \bmod R$, con $R$ reductores | — |
| 5 | ***Shuffle* y ordenamiento** | Cada *reducer* **trae por la red** sus pares de **todos** los `map` y los agrupa por clave | Red + disco del *reducer* |
| 6 | **`reduce`** | Recibe $(k_2, [v_2, \dots])$ y emite la salida final | Archivos en HDFS (**replicados**) |

**Tolerancia a fallos:** como `map` y `reduce` son **funciones deterministas sin efectos laterales**, si una tarea falla se **vuelve a ejecutar** en otro nodo; si una es muy lenta (un *rezagado*, *straggler*), se lanza una **copia especulativa** y se usa la que termine primero. Esa simplicidad es la gracia del modelo.

**Condición del *combiner*:** como el *framework* **puede aplicarlo cero, una o varias veces**, la operación debe ser **asociativa y conmutativa** y su salida debe tener el **mismo tipo** que la entrada del `reduce`. La **suma** y el **máximo** sirven; el **promedio** directo **no** (lo veremos). Un buen *combiner* reduce el volumen que viaja por la red, que suele ser el cuello de botella.

La celda siguiente **implementa un motor MapReduce** (`mapreduce`) con `multiprocessing` (una tarea por trozo, `Pool` de procesos) y lo **ejecuta sobre tres líneas de texto** para **dibujar** la ejecución real: lo que sale de cada `map` (con *combiner*), cómo se **reparten las claves** entre dos reductores y qué produce cada uno. Para que los reductores sean reproducibles se usa un *hash* **estable** (`hashlib.blake2b`) como función de partición, **no** `hash()`, que en Python cambia entre ejecuciones.

````python
# ============================================================
# Configuracion inicial + motor MapReduce desde cero + diagrama de una ejecucion real
# ============================================================
# !pip install -q numpy pandas matplotlib networkx pyspark

import contextlib
import multiprocessing as mp
import os
import pickle
import shutil
import sys
import tempfile
import time
import warnings
import hashlib
from collections import Counter, defaultdict

import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
import pandas as pd
from matplotlib.patches import FancyBboxPatch

warnings.filterwarnings("ignore")
SEED = 42
RAPIDO = True                                                                  # True: tamanos reducidos (el cuaderno completo corre en ~1 min)
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
TMP = tempfile.mkdtemp(prefix="bd11_")
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | RAPIDO={RAPIDO} | metodos de arranque de procesos: {mp.get_all_start_methods()}")

# ----------------------------------------------------------------------------
# Motor MapReduce: 'mapper' y 'reducer' son funciones del usuario; el motor hace lo demas
# ----------------------------------------------------------------------------
def h64(x):
    """Hash estable de 64 bits (blake2b): igual en cualquier proceso y ejecucion, a diferencia de hash() de Python."""
    return int.from_bytes(hashlib.blake2b(repr(x).encode(), digest_size=8).digest(), "big")

def particion(clave, R):
    """Particionador por hash: la misma clave va siempre al mismo reductor."""
    return h64(clave) % R

def _tarea_map(args):
    trozo, mapper, combiner, R, part = args
    emitidos = [kv for registro in trozo for kv in mapper(registro)]            # 2. map
    n_emitidos = len(emitidos)
    if combiner is not None:                                                    # 3. combiner: reduce LOCAL por clave
        g = defaultdict(list)
        for k, v in emitidos: g[k].append(v)
        emitidos = [kv2 for k, vs in g.items() for kv2 in combiner(k, vs)]
    cubetas = [[] for _ in range(R)]
    for k, v in emitidos: cubetas[part(k, R)].append((k, v))                    # 4. particion por clave
    return n_emitidos, cubetas

def _tarea_reduce(args):
    pares, reducer = args
    g = defaultdict(list)
    for k, v in pares: g[k].append(v)                                           # agrupar por clave (en Hadoop: ordenar y mezclar)
    salida = [kv for k in sorted(g, key=repr) for kv in reducer(k, g[k])]       # 6. reduce, clave por clave
    return len(pares), salida

_CTX = mp.get_context("fork") if "fork" in mp.get_all_start_methods() else None

def mapreduce(entradas, mapper, reducer, combiner=None, n_map=4, n_reduce=4, procesos=2, part=particion):
    """Ejecuta un trabajo MapReduce. Devuelve (lista de (clave, valor) de salida, estadisticas).
    'procesos > 1' usa un Pool con 'fork' (Linux/Colab); si no existe 'fork' (Windows/macOS) corre secuencialmente con el mismo resultado."""
    trozos = [entradas[i::n_map] for i in range(n_map)]                         # 1. particion de la entrada
    pool = _CTX.Pool(procesos) if (_CTX and procesos > 1) else contextlib.nullcontext()
    with pool as p:
        aplicar = p.map if p else map
        res_map = list(aplicar(_tarea_map, [(t, mapper, combiner, n_reduce, part) for t in trozos]))
        por_reductor = [[kv for _, cubetas in res_map for kv in cubetas[r]] for r in range(n_reduce)]   # 5. shuffle: cada reductor junta su cubeta de TODOS los map
        res_red = list(aplicar(_tarea_reduce, [(pares, reducer) for pares in por_reductor]))
    salida = [kv for _, s in res_red for kv in s]
    stats = {"emitidos": sum(r[0] for r in res_map), "barajados": sum(len(p_) for p_ in por_reductor),
             "carga_reductores": [r[0] for r in res_red], "cubetas": [cub for _, cub in res_map]}
    return salida, stats

# Conteo de palabras: las dos funciones del usuario
def m_palabras(linea):
    return [(palabra, 1) for palabra in linea.split()]

def r_suma(clave, valores):
    return [(clave, sum(valores))]

# ----------------------------------------------------------------------------
# Ejecucion minima con 3 lineas, 3 tareas map y 2 reductores (se ve TODO)
# ----------------------------------------------------------------------------
lineas_demo = ["papa sal papa", "sal papa", "maíz sal"]
salida_demo, st_demo = mapreduce(lineas_demo, m_palabras, r_suma, combiner=r_suma, n_map=3, n_reduce=2, procesos=1)
assert dict(salida_demo) == Counter(" ".join(lineas_demo).split()) == {"papa": 3, "sal": 3, "maíz": 1}
assert st_demo["emitidos"] == 7 and st_demo["barajados"] == 6                   # el combiner pasa de 7 pares a 6: (papa,1)+(papa,1) -> (papa,2)
print("Salida:", dict(salida_demo), "| pares emitidos por los map:", st_demo["emitidos"], "-> barajados (tras el combiner):", st_demo["barajados"])

def caja(ax, x, y, w, h, texto, color, fs=8.5, tc="white"):
    ax.add_patch(FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.03,rounding_size=0.12", fc=color, ec="white", lw=2))
    ax.text(x + w / 2, y + h / 2, texto, ha="center", va="center", color=tc, fontsize=fs, fontweight="bold")

fig, ax = plt.subplots(figsize=(12.5, 5.6))
ax.set_xlim(0, 13); ax.set_ylim(0, 8); ax.axis("off"); ax.grid(False)
ys = [5.7, 3.4, 1.1]
for i, linea in enumerate(lineas_demo):
    caja(ax, 0.1, ys[i], 2.3, 1.4, f'split {i}\n"{linea}"', PALETA[9])
    emit = [(k, v) for k, v in [(w, 1) for w in linea.split()]]
    comb = [kv for cub in st_demo["cubetas"][i] for kv in cub]
    txt = "map: " + " ".join(f"({k},{v})" for k, v in emit) + "\ncombiner: " + " ".join(f"({k},{v})" for k, v in sorted(comb))
    caja(ax, 3.1, ys[i], 3.5, 1.4, txt, PALETA[0], fs=7.8)
    ax.annotate("", xy=(3.05, ys[i] + 0.7), xytext=(2.45, ys[i] + 0.7), arrowprops=dict(arrowstyle="-|>", lw=2, color="#334155"))
for r, y in enumerate([4.5, 1.9]):
    contenido = sorted(kv for cub in st_demo["cubetas"] for kv in cub[r])
    resultado = sorted(kv for kv in salida_demo if particion(kv[0], 2) == r)
    caja(ax, 9.2, y, 3.6, 1.9, f"reductor {r}\nrecibe: " + " ".join(f"({k},{v})" for k, v in contenido) + "\nsalida: " + ", ".join(f"{k}={v}" for k, v in resultado), PALETA[2], fs=7.8)
    for i in range(3):
        if st_demo["cubetas"][i][r]:
            ax.annotate("", xy=(9.15, y + 0.95), xytext=(6.65, ys[i] + 0.7), arrowprops=dict(arrowstyle="-|>", lw=1.3, color=PALETA[3], alpha=0.9))
ax.text(7.8, 7.5, "shuffle: partición por hash(clave) mod 2", ha="center", fontsize=9.5, fontweight="bold", color=PALETA[3])
ax.set_title("MapReduce sobre 3 líneas: cada flecha naranja es una cubeta enviada a un reductor", fontsize=11, fontweight="bold")
plt.tight_layout(); plt.show()
````

## 2. Cuatro Trabajos Clásicos y su Verificación ✅

El motor es **genérico**: cada problema se reduce a elegir **qué clave** agrupa los datos y **qué se hace con cada grupo**. Escribimos cuatro:

1. **Conteo de palabras** (el «Hola mundo»): `map` emite $(\text{palabra}, 1)$; `reduce` suma. Con *combiner* el volumen barajado se desploma.
2. **Promedio por clave:** el *combiner* **no puede** ser «el promedio» (el **promedio de promedios no es el promedio**: pesa igual a trozos de tamaños distintos); se emite el par **$(\text{suma}, \text{cuenta})$**, que **sí** es asociativo, y solo al final se divide.
3. ***Reduce-side join*:** para unir dos tablas por una clave, el `map` **etiqueta** cada registro con su tabla de origen y lo emite con la **clave de unión**; el `reduce` recibe **todos** los registros de ambas tablas con esa clave y forma el producto cruzado.
4. **Índice invertido:** `map` emite $(\text{palabra}, \text{id del documento})$; `reduce` junta la lista (sin repetidos) de documentos de cada palabra; es la base de un buscador.

Cada resultado se **verifica con `assert`** contra una implementación independiente (`Counter`, `groupby` y `merge` de *pandas*, o un recorrido por fuerza bruta).

````python
# ============================================================
# 2a. Conteo de palabras (con y sin combiner) y promedio por clave
# ============================================================
rng = np.random.default_rng(SEED)
N_LIN = 12_000 if RAPIDO else 60_000
vocab = np.array([f"p{i:04d}" for i in range(2000)])
pz = 1 / np.arange(1, 2001) ** 1.1; pz /= pz.sum()                              # frecuencias tipo Zipf: pocas palabras muy comunes
lineas = [" ".join(rng.choice(vocab, 12, p=pz)) for _ in range(N_LIN)]
referencia = Counter(w for l in lineas for w in l.split())

t0 = time.perf_counter(); out_sin, st_sin = mapreduce(lineas, m_palabras, r_suma, combiner=None); t_sin = time.perf_counter() - t0
t0 = time.perf_counter(); out_con, st_con = mapreduce(lineas, m_palabras, r_suma, combiner=r_suma); t_con = time.perf_counter() - t0
t0 = time.perf_counter(); Counter(w for l in lineas for w in l.split()); t_counter = time.perf_counter() - t0
assert dict(out_sin) == referencia and dict(out_con) == referencia              # el resultado es identico a Counter
assert st_con["emitidos"] == st_sin["emitidos"] == 12 * N_LIN
assert st_con["barajados"] < st_sin["barajados"] / 10                            # el combiner reduce > 10 veces lo que viaja por la red
print(f"Conteo de palabras ({12 * N_LIN:,} palabras, {len(referencia):,} distintas): pares barajados SIN combiner = {st_sin['barajados']:,} | CON combiner = {st_con['barajados']:,} ({st_sin['barajados'] / st_con['barajados']:.0f} veces menos)")
print(f"Tiempos: motor sin combiner {t_sin:.2f} s | con combiner {t_con:.2f} s | Counter en un solo proceso {t_counter:.3f} s -> en una laptop el motor NO gana: su valor es la semantica y el escalado a muchos nodos.")

# --- Promedio por clave: la trampa del promedio de promedios ---
N_L = 20_000
est = rng.choice([f"E{i:02d}" for i in range(12)], N_L, p=np.r_[np.full(6, 0.02), np.full(6, 0.88 / 6)])      # estaciones con muy distinto numero de lecturas
temp = rng.normal(15, 5, N_L).round(2)
lecturas_mr = list(zip(est.tolist(), temp.tolist()))                           # (estacion, temperatura): se reutiliza en la practica
ref_prom = pd.DataFrame(lecturas_mr, columns=["e", "t"]).groupby("e").t.mean()

def m_temp(rec): return [(rec[0], rec[1])]
def r_prom_directo(k, vs): return [(k, sum(vs) / len(vs))]                      # promedio directo (sirve como reducer, NO como combiner)
out_mal, _ = mapreduce(lecturas_mr, m_temp, r_prom_directo, combiner=r_prom_directo)
mal = pd.Series(dict(out_mal)).sort_index()
assert not np.allclose(mal.values, ref_prom.sort_index().values, rtol=0, atol=1e-6)    # promedio de promedios: RESULTADO INCORRECTO
error_max = float((mal - ref_prom.sort_index()).abs().max())

def m_par(rec): return [(rec[0], (rec[1], 1))]                                   # (suma, cuenta)
def c_par(k, vs): return [(k, (sum(a for a, _ in vs), sum(b for _, b in vs)))]   # asociativo y conmutativo
def r_par(k, vs): s, n = c_par(k, vs)[0][1]; return [(k, s / n)]
out_ok, st_ok = mapreduce(lecturas_mr, m_par, r_par, combiner=c_par)
bien = pd.Series(dict(out_ok)).sort_index()
assert np.allclose(bien.values, ref_prom.sort_index().values, atol=1e-9)
print(f"\nPromedio por estacion: con el combiner 'promedio' el error maximo es {error_max:.3f} grados (INCORRECTO); con pares (suma, cuenta) coincide con pandas (error < 1e-9). Pares barajados: {st_ok['barajados']} de {N_L:,}.")
````

## 3. Sesgo de Claves (*Skew*): el Reductor que Nunca Termina 🐌

El modelo asume que el trabajo se **reparte parejo** entre reductores. Pero **las claves de los datos reales no están equilibradas** (casi siempre siguen una ley de potencias: unas pocas claves concentran mucha de la actividad: páginas muy visitadas, un cliente corporativo, la palabra «de»). Como **todos los pares de una misma clave van al mismo reductor**, ese reductor recibe muchísimo más trabajo que los demás: **el trabajo entero tarda lo que tarda el reductor más lento** (el *rezagado*).

Medimos el desbalance con el cociente **$\text{desbalance} = \max_r(\text{carga}_r) \,/\, \overline{\text{carga}}$** (1 = perfecto) y la **eficiencia** $= 1/\text{desbalance}$: la fracción del tiempo ideal que se aprovecha. Un trabajo con $R$ reductores y desbalance $d$ acelera solo $R/d$ veces, no $R$.

**Qué hacer:**

- **Combiner:** si la operación es combinable (suma, máximo), cada `map` envía **un solo par** por clave y el sesgo casi desaparece (lo vimos en el conteo de palabras).
- **Sal de claves (*salting*):** si **no** hay *combiner* posible (p. ej. **visitantes distintos** por página con listas de usuarios), se **parte la clave caliente** en $S$ subclaves $(k, s)$ con $s = h(\text{valor}) \bmod S$, de modo que sus pares caigan en **varios reductores**; una **segunda ronda** barata junta los resultados parciales. Cuesta **una ronda extra** (con su escritura a disco...) pero elimina el rezagado. Para saber **qué claves son calientes** se **muestrea** (aquí el 5 %) en lugar de contar todo.

La celda siguiente mide el sesgo con **200 000 visitas** (página, usuario) con claves tipo Zipf y 8 reductores, aplica la sal **solo a las claves calientes** y **verifica** el resultado contra *pandas* (`nunique`).

````python
# ============================================================
# 3. Sesgo de claves: desbalance entre reductores y mitigacion con sal (salting)
# ============================================================
rng = np.random.default_rng(SEED)
N_VIS, N_PAG, R = (80_000 if RAPIDO else 400_000), 400, 8
pz = 1 / np.arange(1, N_PAG + 1) ** 1.2; pz /= pz.sum()
paginas = rng.choice(N_PAG, N_VIS, p=pz)
usuarios = rng.integers(0, 30_000, N_VIS)
visitas = list(zip(paginas.tolist(), usuarios.tolist()))                       # (pagina, usuario)
share_top = Counter(paginas.tolist()).most_common(1)[0][1] / N_VIS
print(f"{N_VIS:,} visitas a {N_PAG} paginas | la pagina mas visitada concentra el {share_top:.1%} de todas las visitas")

def m_vis(rec): return [(rec[0], rec[1])]                                         # (pagina, usuario)
def r_distintos(k, usuarios_): return [(k, len(set(usuarios_)))]                  # visitantes DISTINTOS (sin combiner de suma posible)

def desbalance(cargas): return max(cargas) / (sum(cargas) / len(cargas))

out_a, st_a = mapreduce(visitas, m_vis, r_distintos, n_reduce=R)
ref_nun = pd.DataFrame(visitas, columns=["p", "u"]).groupby("p").u.nunique().to_dict()
assert dict(out_a) == ref_nun
d_a = desbalance(st_a["carga_reductores"])
assert max(st_a["carga_reductores"]) >= Counter(paginas.tolist()).most_common(1)[0][1]    # ningun reductor puede cargar menos que la clave mas caliente
print(f"Sin mitigacion: carga por reductor = {st_a['carga_reductores']} | desbalance = {d_a:.2f} | eficiencia = {1 / d_a:.0%} (aceleracion {R / d_a:.1f}x de {R}x ideal)")

# --- sal: se muestrea el 5 % para hallar las claves calientes ---
muestra = Counter(p for p, _ in visitas[::20])
calientes = {p for p, c in muestra.items() if c / (N_VIS / 20) > 1 / R / 2}      # claves con > media reductor de carga ideal (1/(2R) del total)
S = R
def m_sal(rec):
    p, u = rec
    return [((p, h64(u) % S if p in calientes else 0), u)]   # subclave (pagina, sal): usuarios repetidos caen en la MISMA subclave
def r_parcial(k, us): return [(k[0], len(set(us)))]                              # ronda 1: distintos por subclave
def m_id(rec): return [rec]
def r_total(k, vs): return [(k, sum(vs))]                                        # ronda 2: sumar parciales (los conjuntos de usuarios son disjuntos)

parcial, st_b1 = mapreduce(visitas, m_sal, r_parcial, n_reduce=R)
out_b, st_b2 = mapreduce(parcial, m_id, r_total, combiner=r_total, n_reduce=R)
assert dict(out_b) == ref_nun                                                    # exactitud: la sal no cambia la respuesta
d_b = desbalance(st_b1["carga_reductores"])
assert d_b < d_a
print(f"Con sal en {len(calientes)} claves calientes: carga = {st_b1['carga_reductores']} | desbalance = {d_b:.2f} | aceleracion {R / d_b:.1f}x | pares de la ronda 2: {st_b2['barajados']} (casi gratis)")

# --- el desbalance crece con el numero de reductores si hay una clave dominante ---
Rs = [2, 4, 8, 16, 32]
cuenta_pag = Counter(paginas.tolist())
def cargas_hash(r):
    c = [0] * r
    for p, n_p in cuenta_pag.items(): c[particion(p, r)] += n_p
    return c
d_por_R = {r: desbalance(cargas_hash(r)) for r in Rs}
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
x = np.arange(R); w = 0.4
ax[0].bar(x - w / 2, st_a["carga_reductores"], w, color=PALETA[1], label=f"sin mitigación (desbalance {d_a:.1f})")
ax[0].bar(x + w / 2, st_b1["carga_reductores"], w, color=PALETA[2], label=f"con sal (desbalance {d_b:.1f})")
ax[0].axhline(N_VIS / R, color="#334155", ls="--", lw=1); ax[0].set_xlabel("reductor"); ax[0].set_ylabel("pares recibidos"); ax[0].set_title("Carga por reductor (línea: reparto perfecto)", fontsize=10, fontweight="bold"); ax[0].legend(fontsize=8)
ax[1].plot(Rs, [d_por_R[r] for r in Rs], "o-", color=PALETA[0]); ax[1].plot(Rs, [1] * len(Rs), "--", color="#334155")
ax[1].set_xscale("log", base=2); ax[1].set_xlabel("número de reductores R"); ax[1].set_ylabel("desbalance (máx / media)")
ax[1].set_title("Más reductores empeoran el desbalance con una clave dominante", fontsize=10, fontweight="bold")
plt.tight_layout(); plt.show()
print("Desbalance por numero de reductores:", {r: round(v, 2) for r, v in d_por_R.items()})
````

## 4. Las Limitaciones de MapReduce 🧱

El modelo es elegante, pero sus decisiones de diseño (pensadas para **fiabilidad en miles de máquinas baratas**) tienen un **costo**. Medimos cuatro, en orden de importancia práctica:

| Limitación | Causa | Quién sufre | Cómo se mide aquí |
|---|---|---|---|
| **E/S a disco entre etapas** | La salida de cada trabajo se **escribe a HDFS (replicada ×3)** y el siguiente la **lee de nuevo** | Trabajos **encadenados** | Guardamos el resultado de cada ronda a disco (con `fsync`) frente a mantenerlo en memoria |
| **Algoritmos iterativos** | Cada iteración es **un trabajo completo** (arranque + E/S + *shuffle*) | PageRank, K-Means, regresión por gradiente | PageRank como 10 rondas MapReduce frente a NumPy y Spark con y sin `cache()` |
| **Archivos pequeños** | Cada archivo es **una tarea `map`** y un objeto en la memoria del NameNode | Datos que llegan en millones de archivos diminutos | 2 000 archivos pequeños frente a 1 archivo con el mismo contenido |
| **Latencia** | Arranque de tareas, escritura de intermedios y *shuffle*: **segundos o minutos incluso con datos mínimos** | Consultas interactivas, tiempo real | El costo de lanzar un trabajo con 100 palabras |

**PageRank** (Brin y Page) asigna a cada nodo de un grafo una importancia proporcional a la de quienes lo apuntan. Con $d = 0.85$ (factor de amortiguación) y $N$ nodos, se itera hasta converger:

$$ r^{(t+1)}(v) \;=\; \frac{1-d}{N} \;+\; d \sum_{u \to v} \frac{r^{(t)}(u)}{\text{grado\_salida}(u)} $$

En MapReduce cada iteración es **un trabajo**: el `map` envía a cada vecino $v$ la **contribución** $r(u)/\text{grado}(u)$ (y reenvía la lista de vecinos para no perder el grafo), y el `reduce` suma las contribuciones recibidas. Hay que **hacerlo 10, 20 o más veces**, y entre rondas Hadoop **escribe todo a HDFS**. Spark, en cambio, mantiene el grafo **en memoria** entre iteraciones (`cache()`). En la primera celda medimos la **E/S entre rondas**; en la segunda, la comparación completa.

````python
# ============================================================
# 4a. Costo de escribir a disco entre rondas: PageRank encadenado en MapReduce
# ============================================================
rng = np.random.default_rng(7)
N_NOD, N_ARI = (1_500, 9_000) if RAPIDO else (6_000, 40_000)
peso = 1 / np.arange(1, N_NOD + 1) ** 0.8; peso /= peso.sum()                    # algunos nodos reciben muchos enlaces
cand = {(int(a), int(b)) for a, b in zip(rng.integers(0, N_NOD, N_ARI), rng.choice(N_NOD, N_ARI, p=peso)) if a != b}
cand |= {(n, (n + 1) % N_NOD) for n in range(N_NOD)}                             # todo nodo tiene al menos una salida (sin nodos colgantes)
aristas = sorted(cand)
salidas = defaultdict(list)
for a, b in aristas: salidas[a].append(b)
D, ITER = 0.85, 10
print(f"Grafo sintetico: {N_NOD:,} nodos, {len(aristas):,} aristas | PageRank con d={D}, {ITER} iteraciones")

def m_pr(rec):
    nodo, (rank, vecinos) = rec
    yield nodo, ("L", vecinos)                                                   # reenvia el grafo para no perderlo
    for v in vecinos: yield v, ("C", rank / len(vecinos))                        # contribucion a cada vecino

def r_pr(nodo, valores):
    vecinos, suma = [], 0.0
    for tag, x in valores:
        if tag == "L": vecinos = x
        else: suma += x
    yield nodo, ((1 - D) / N_NOD + D * suma, vecinos)

def pagerank_numpy(it):
    src = np.array([a for a, _ in aristas]); dst = np.array([b for _, b in aristas]); grado = np.bincount(src, minlength=N_NOD)
    r = np.full(N_NOD, 1 / N_NOD)
    for _ in range(it):
        c = np.zeros(N_NOD); np.add.at(c, dst, r[src] / grado[src]); r = (1 - D) / N_NOD + D * c
    return r

def pagerank_mapreduce(it, disco=False):
    """Cada iteracion = un trabajo MapReduce. Con disco=True el resultado de cada ronda se escribe (fsync) y se vuelve a leer, como HDFS."""
    datos = [(n, (1 / N_NOD, salidas[n])) for n in range(N_NOD)]
    t_es, bytes_ronda = 0.0, []
    for i in range(it):
        datos, _ = mapreduce(datos, m_pr, r_pr, n_map=4, n_reduce=4)
        if disco:
            t0 = time.perf_counter()
            ruta = f"{TMP}/ronda_{i}.pkl"
            with open(ruta, "wb") as f:
                pickle.dump(datos, f); f.flush(); os.fsync(f.fileno())           # durabilidad: no basta con la cache del sistema
            bytes_ronda.append(os.path.getsize(ruta))
            with open(ruta, "rb") as f: datos = pickle.load(f)                   # la ronda siguiente lee de nuevo
            t_es += time.perf_counter() - t0
    return np.array([r for _, (r, _) in sorted(datos)]), t_es, bytes_ronda

r_np = pagerank_numpy(ITER)
def mejor_de(fn, reps=3):
    mejor = None
    for _ in range(reps):
        t0 = time.perf_counter(); salida = fn(); t = time.perf_counter() - t0
        if mejor is None or t < mejor[0]: mejor = (t, salida)
    return mejor
t_mem, (r_mem, _, _) = mejor_de(lambda: pagerank_mapreduce(ITER), reps=2)
t_dis, (r_dis, t_es, bytes_ronda) = mejor_de(lambda: pagerank_mapreduce(ITER, disco=True), reps=2)
assert np.allclose(r_mem, r_np, atol=1e-12) and np.allclose(r_dis, r_np, atol=1e-12)         # las tres versiones calculan LO MISMO
assert abs(r_np.sum() - 1) < 1e-9 and min(bytes_ronda) > 0
# contra una referencia independiente: networkx converge al mismo PageRank (con muchas iteraciones de NumPy)
ref_nx = nx.pagerank(nx.DiGraph(aristas), alpha=D, tol=1e-12)
assert np.allclose(pagerank_numpy(200), [ref_nx[i] for i in range(N_NOD)], atol=1e-9)
mb = np.mean(bytes_ronda) / 1e6
print(f"Resultados identicos (NumPy = MapReduce en memoria = MapReduce con disco = networkx tras converger).")
print(f"Cada ronda escribe ~{mb:.2f} MB; en HDFS seria x3 replicas por la red = ~{3 * mb:.1f} MB por ronda ({ITER} rondas: ~{3 * mb * ITER:.0f} MB solo en intermedios).")
print(f"Tiempo de las {ITER} rondas: en memoria {t_mem:.2f} s | con escritura+lectura a disco {t_dis:.2f} s (E/S pura: {t_es:.2f} s = {t_es / t_dis:.0%} del total).")
# Con datos tan pequenos la E/S casi no se nota: lo que importa es su PROPORCION con el volumen. Medimos el rendimiento bruto de este disco:
bloque = np.random.default_rng(0).bytes((64 if RAPIDO else 256) * 1024 * 1024)
t0 = time.perf_counter()
with open(f"{TMP}/bloque.bin", "wb") as f:
    f.write(bloque); f.flush(); os.fsync(f.fileno())
t_w = time.perf_counter() - t0
t0 = time.perf_counter()
with open(f"{TMP}/bloque.bin", "rb") as f: leido = f.read()
t_r = time.perf_counter() - t0
assert leido == bloque
mbs = len(bloque) / 1e6 / (t_w + t_r)                                            # MB/s de 'escribir y volver a leer'
print(f"Rendimiento bruto de este disco (escribir con fsync + leer): {mbs:,.0f} MB/s. Regla de tres: una ronda con 100 GB de intermedios tardaria ~{100_000 / mbs:,.0f} s SOLO en esta E/S, y eso por cada ronda.")
print("Aviso: si tu /tmp es memoria (tmpfs) o un SSD rapido esto SUBESTIMA el costo; un disco mecanico, la red de un cluster y la replicacion x3 lo multiplican (extrapolacion lineal, solo orientativa).")
````

````python
# ============================================================
# 4b. PageRank iterativo: NumPy frente a MapReduce frente a Spark (sin y con cache)
# ============================================================
os.environ["PYSPARK_PYTHON"] = sys.executable
from pyspark.sql import SparkSession

t0 = time.perf_counter()
spark = (SparkSession.builder.master("local[2]").appName("bd11").config("spark.ui.enabled", "false")
         .config("spark.sql.shuffle.partitions", "4").config("spark.default.parallelism", "4")
         .config("spark.sql.warehouse.dir", f"{TMP}/warehouse").getOrCreate())
sc = spark.sparkContext; sc.setLogLevel("ERROR")
print(f"Sesion Spark local[2] lista en {time.perf_counter() - t0:.1f} s")

# (1) El MISMO modelo en otro motor: el conteo de palabras en RDDs (flatMap + map + reduceByKey) da el resultado de nuestro motor
sub = lineas[:3_000]
wc = dict(sc.parallelize(sub, 4).flatMap(lambda l: l.split()).map(lambda w: (w, 1)).reduceByKey(lambda a, b: a + b).collect())
assert wc == dict(Counter(w for l in sub for w in l.split())) == dict(mapreduce(sub, m_palabras, r_suma, combiner=r_suma)[0])
print("Conteo de palabras en Spark == motor MapReduce == Counter (reduceByKey ya aplica un 'combiner' dentro de cada particion).")

# (2) PageRank en RDDs: 'enlaces' se reutiliza en CADA iteracion -> candidato a cache()
lineas_ari = sc.parallelize([f"{a} {b}" for a, b in aristas], 4)
def pagerank_spark(it, cache):
    enlaces = lineas_ari.map(lambda l: tuple(map(int, l.split()))).groupByKey().mapValues(list)
    if cache: enlaces = enlaces.cache()
    ranks = enlaces.mapValues(lambda _: 1.0 / N_NOD)
    for _ in range(it):
        contrib = enlaces.join(ranks).flatMap(lambda kv: [(d, kv[1][1] / len(kv[1][0])) for d in kv[1][0]])
        ranks = contrib.reduceByKey(lambda x, y: x + y).mapValues(lambda s: (1 - D) / N_NOD + D * s)
    res = dict(ranks.collect())
    if cache: enlaces.unpersist()
    return np.array([res.get(i, (1 - D) / N_NOD) for i in range(N_NOD)])

pagerank_spark(1, False)                                                         # calentamiento (JIT, trabajadores de Python): no se cuenta
t_sp_sin, r_sp_sin = mejor_de(lambda: pagerank_spark(ITER, False), reps=1)
t_sp_con, r_sp_con = mejor_de(lambda: pagerank_spark(ITER, True), reps=1)
assert np.allclose(r_sp_sin, r_np, atol=1e-9) and np.allclose(r_sp_con, r_np, atol=1e-9)
spark.stop()                                                                     # se detiene Spark al final de la seccion

t_np, _ = mejor_de(lambda: pagerank_numpy(ITER), reps=5)
tab = pd.Series({"NumPy (1 proceso, en memoria)": t_np, "MapReduce propio (en memoria)": t_mem, "MapReduce propio (con disco)": t_dis,
                 "Spark local, sin cache()": t_sp_sin, "Spark local, con cache()": t_sp_con}).sort_values()
print(f"\nPageRank, {ITER} iteraciones ({N_NOD:,} nodos, {len(aristas):,} aristas): todos dan el MISMO vector.")
print(tab.round(3).to_string())
fig, ax = plt.subplots(figsize=(9, 3.2))
ax.barh(tab.index, tab.values, color=[PALETA[i] for i in range(len(tab))]); ax.set_xscale("log"); ax.invert_yaxis(); ax.set_xlabel("segundos (escala logarítmica)")
ax.set_title("PageRank iterativo: cuando el grafo cabe en memoria, un solo proceso gana por mucho", fontsize=10, fontweight="bold")
plt.tight_layout(); plt.show()
print("Lecturas honestas: (1) a este tamano NumPy en un solo proceso gana por ordenes de magnitud (la idea del articulo 'COST' de McSherry et al.: un sistema distribuido debe superar a UN hilo);")
print("(2) cache() ayuda poco aqui: el grafo es pequeno (y, posiblemente, Spark ya reutiliza los archivos del shuffle); su ventaja crece con linajes caros de recomputar (cuaderno 02).")
print("(3) la ventaja de Spark sobre Hadoop MapReduce no es 'ser mas rapido en una laptop' sino NO escribir cada ronda a HDFS y reutilizar datos en memoria en un cluster.")
````

````python
# ============================================================
# 4c. Archivos pequenos y latencia minima de un trabajo
# ============================================================
rng = np.random.default_rng(SEED)
N_REG = 100_000 if RAPIDO else 400_000
N_ARCH = 2_000
registros = [f"{rng.integers(0, 50):02d},{rng.integers(0, 1000)}" for _ in range(N_REG)]

dir_uno, dir_muchos = f"{TMP}/uno", f"{TMP}/muchos"
os.makedirs(dir_uno); os.makedirs(dir_muchos)
t0 = time.perf_counter()
with open(f"{dir_uno}/todo.csv", "w") as f: f.write("\n".join(registros))
t_esc_uno = time.perf_counter() - t0
t0 = time.perf_counter()
por_arch = N_REG // N_ARCH
for i in range(N_ARCH):
    with open(f"{dir_muchos}/p{i:05d}.csv", "w") as f: f.write("\n".join(registros[i * por_arch:(i + 1) * por_arch]))
t_esc_muchos = time.perf_counter() - t0

def contar(directorio):
    c = Counter()
    for nombre in sorted(os.listdir(directorio)):
        with open(f"{directorio}/{nombre}") as f:
            for linea in f: c[linea.split(",")[0]] += 1
    return c
t_uno, c1 = mejor_de(lambda: contar(dir_uno)); t_muchos, c2 = mejor_de(lambda: contar(dir_muchos))
assert c1 == c2 and sum(c1.values()) == N_REG                                    # mismo contenido, mismo resultado
print(f"Mismos {N_REG:,} registros: 1 archivo vs {N_ARCH:,} archivos de {por_arch} lineas")
print(f"  escritura: {t_esc_uno:.3f} s vs {t_esc_muchos:.3f} s | lectura y conteo: {t_uno:.3f} s vs {t_muchos:.3f} s ({t_muchos / t_uno:.1f}x)")
print(f"  En Hadoop cada archivo pequeno ademas = 1 tarea map (arranque de tarea) y objetos en la memoria del NameNode: con {N_ARCH:,} archivos -> {N_ARCH:,} tareas en vez de ~1.")
print("  (Orden de magnitud citado de memoria, verificar: ~150 bytes de memoria del NameNode por archivo/bloque -> 10 millones de archivos pequenos ~ 3 GB solo de metadatos.)")

# --- latencia minima: lanzar un trabajo MapReduce con 100 palabras ---
pocas = ["papa sal papa maíz sal"] * 20
t_motor, _ = mejor_de(lambda: mapreduce(pocas, m_palabras, r_suma, combiner=r_suma, n_map=4, n_reduce=4), reps=5)
t_counter_min, _ = mejor_de(lambda: Counter(w for l in pocas for w in l.split()), reps=5)
assert dict(mapreduce(pocas, m_palabras, r_suma)[0]) == {"papa": 40, "sal": 40, "maíz": 20}
print(f"\nLatencia minima con 100 palabras: motor MapReduce (Pool + fases) {t_motor * 1000:.1f} ms vs Counter {t_counter_min * 1e6:.0f} microsegundos ({t_motor / t_counter_min:,.0f} veces).")
print("En Hadoop real el arranque de un trabajo se mide en segundos (JVM, planificacion, escritura de intermedios): por eso MapReduce no sirve para consultas interactivas.")
shutil.rmtree(TMP, ignore_errors=True)
````

## 5. ¿Cuándo Sigue Siendo Razonable MapReduce? ⚖️

Después de tanta crítica, una conclusión **equilibrada**:

| Situación | ¿MapReduce (Hadoop)? | Mejor alternativa si no |
|---|---|---|
| Un **solo paso** sobre **terabytes** (conteos, filtros, agregaciones, ETL por lotes) | **Sí, es razonable**: lectura secuencial, paralelismo masivo, tolerancia a fallos probada | Spark SQL (mismo plan, menos E/S) |
| **Heredado:** ya hay miles de trabajos y datos en HDFS | **Sí**: migrar cuesta | Migración gradual a Spark |
| Memoria **escasa** frente al dato (el *spill* a disco es inevitable) | **Sí**: el modelo está pensado para ello | Spark también *derrama* a disco, con más memoria disponible |
| **Algoritmos iterativos** (PageRank, K-Means, regresión) | **No**: cada ronda paga E/S y arranque | **Spark** (`cache()`), motores de grafos (módulo 05) |
| **Consultas interactivas** o de baja latencia | **No** | Spark SQL, motores columnares (DuckDB, Trino) |
| **Tiempo real** / flujos continuos | **No** | Spark Structured Streaming, Flink, Kafka Streams |
| Datos que **caben en una máquina** | **No** | *pandas*, DuckDB, Polars |

> 🧬 **Lo que sobrevive del modelo:** las ideas de `map` (transformar en paralelo), *shuffle* (agrupar por clave: **el paso caro**), `reduce` y *combiner* (reducir antes de mover datos) están **dentro** de Spark (`map`, `reduceByKey`, `groupBy`), de SQL distribuido (`GROUP BY` = *shuffle* + agregación) y de motores como Beam o Flink. Aprender MapReduce es aprender **qué hace un `groupBy` en un clúster** y **por qué mover datos es lo caro**.

---
## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **El modelo** | Dos funciones: `map` $(k_1, v_1) \to \text{lista}(k_2, v_2)$ y `reduce` $(k_2, \text{lista}(v_2)) \to \text{lista}(k_3, v_3)$; el *framework* parte la entrada, **baraja** por clave ($h(k) \bmod R$) y **reintenta** fallos porque las funciones son deterministas |
| **Combiner** | Reduce **localmente** antes de mover datos; exige operación **asociativa y conmutativa**. En el conteo bajó los pares barajados **más de 10 veces**; el **promedio de promedios** dio un resultado **incorrecto** y el par (suma, cuenta) lo corrigió |
| **Cuatro trabajos** | Conteo de palabras, promedio por clave, ***reduce-side join*** (todo viaja por la red; igual a `pandas.merge`) e **índice invertido**: todos **verificados** contra *Counter* / *pandas* / fuerza bruta |
| **Sesgo** | Una clave dominante fija el **piso** de la carga de un reductor: **más reductores no aceleran**. Mitigación: **combiner**, o **sal** de claves calientes (detectadas por **muestreo**) y una ronda extra; exacto y con desbalance mucho menor |
| **E/S entre rondas** | Cada ronda en Hadoop **escribe a HDFS ×3** y la siguiente la **lee**; aquí se midió el costo (subestimado si tu disco es memoria o SSD) con **resultados idénticos** |
| **Iterativos** | PageRank como 10 rondas: a este tamaño **NumPy en un proceso ganó por órdenes de magnitud**; `cache()` ayudó poco con datos pequeños. La ventaja de Spark es **no escribir cada ronda a disco** y escalar cuando el dato **no cabe** |
| **Archivos pequeños y latencia** | Miles de archivos pequeños = miles de tareas y metadatos; lanzar un trabajo con 100 palabras cuesta **órdenes de magnitud más** que `Counter` |
| **Cuándo usarlo** | **Un paso** sobre **mucho** dato, o cuando ya está desplegado; **no** para iterativos, consultas interactivas, tiempo real ni datos que caben en una máquina |
