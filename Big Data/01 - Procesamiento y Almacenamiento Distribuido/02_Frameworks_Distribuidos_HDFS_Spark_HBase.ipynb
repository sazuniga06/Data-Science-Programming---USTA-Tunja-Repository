## 1. HDFS: Archivos Gigantes en Bloques Replicados 🗄️

**HDFS** (*Hadoop Distributed File System*, inspirado en el GFS de Google) guarda archivos **muy grandes** (de GB a TB), que se escriben **una vez** y se leen **muchas veces** en secuencia, sobre **máquinas corrientes que fallan con frecuencia**. Sus ideas:

| Idea | Cómo funciona |
|---|---|
| **Bloques** | Cada archivo se parte en **bloques grandes** (por defecto **128 MB** en Hadoop 2 y 3 *(citado de memoria, verificar)*) que se reparten por el clúster. Bloques grandes = pocos metadatos y lecturas secuenciales eficientes; el último bloque ocupa solo lo que necesita |
| **NameNode** (maestro) | Guarda **solo metadatos**, en memoria: el **espacio de nombres** (`ruta → [bloques]`) y **dónde está cada réplica** (`bloque → DataNodes`). **No** almacena datos |
| **DataNodes** (trabajadores) | Guardan los bloques en sus discos, envían un ***heartbeat*** periódico y un **informe de bloques** al NameNode |
| **Replicación** | Cada bloque se copia en **`factor` DataNodes** (por defecto **3**). Con ***rack awareness***: 1.ª réplica en un nodo, 2.ª en **otro *rack***, 3.ª en **otro nodo del mismo *rack* que la 2.ª** (sobrevive a la caída de un *rack* completo sin gastar tanto ancho de banda entre *racks*) |
| **Falla de un DataNode** | Si dejan de llegar sus *heartbeats*, el NameNode lo da por muerto, detecta los bloques **sub-replicados** y ordena **copiarlos** desde réplicas vivas hasta recuperar el factor |
| **Lectura** | El cliente pregunta al NameNode **dónde** están los bloques y los **lee directamente de los DataNodes** (preferiblemente de uno cercano) |

**Costo y beneficio:** con factor 3, un archivo de 1 TB ocupa **3 TB**, a cambio de sobrevivir a **2 fallas simultáneas** de nodos y de ofrecer **varias fuentes de lectura** (útil para la localidad de datos de Spark). El **NameNode** es el punto delicado (su memoria limita cuántos archivos y bloques caben: por eso los **archivos pequeños** son un problema, ver el cuaderno 01; en producción se usa NameNode con alta disponibilidad).

La celda siguiente implementa `MiniHDFS` (en memoria; bloques de **64 KB** para que el ejemplo sea pequeño) con seis DataNodes en **dos *racks***, **carga dos archivos** y **verifica con `assert`** todas las propiedades: réplicas en nodos distintos y en al menos dos *racks*, almacenamiento = factor × datos, el último bloque más corto y lectura idéntica al original.

````python
# ============================================================
# Configuracion + MiniHDFS desde cero (NameNode + DataNodes + replicacion con racks)
# ============================================================
# !pip install -q numpy pandas matplotlib pyspark

import bisect
import contextlib
import hashlib
import io
import itertools
import math
import os
import shutil
import sys
import tempfile
import time
import warnings
from collections import Counter

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.patches import FancyBboxPatch

warnings.filterwarnings("ignore")
SEED = 42
RAPIDO = True                                                                  # True: tamanos reducidos (el cuaderno completo corre en ~1 min)
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
TMP = tempfile.mkdtemp(prefix="bd12_")
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | RAPIDO={RAPIDO}")

def h64(x):
    """Hash estable de 64 bits (igual en todos los procesos y ejecuciones)."""
    return int.from_bytes(hashlib.blake2b(repr(x).encode(), digest_size=8).digest(), "big")

class DataNode:
    def __init__(self, nombre, rack):
        self.nombre, self.rack, self.vivo, self.bloques = nombre, rack, True, {}     # bloques: id -> bytes (los datos viven AQUI)

class MiniHDFS:
    """NameNode (solo metadatos: 'nombres' y 'ubic') + DataNodes (los bytes). Colocacion con conciencia de racks."""
    def __init__(self, topologia, tam_bloque=64 * 1024, factor=3):
        self.dn = {n: DataNode(n, r) for n, r in topologia.items()}
        self.tam_bloque, self.factor = tam_bloque, factor
        self.nombres, self.ubic, self.tam, self._sig = {}, {}, {}, 0                 # NameNode: ruta->[bloques], bloque->{DataNodes}, bloque->bytes

    def _vivos(self): return [d for d in self.dn.values() if d.vivo]

    def _elegir(self, k, existentes=(), rack_inicial=None):
        """Elige k DataNodes nuevos: 1.o el menos cargado (del rack del escritor); 2.o en OTRO rack; 3.o en el rack del 2.o; despues el menos cargado."""
        sel = list(existentes)
        for _ in range(k):
            cand = [d for d in self._vivos() if d.nombre not in sel]
            if not cand: break
            racks = {self.dn[n].rack for n in sel}
            if not sel and rack_inicial is not None: pref = [d for d in cand if d.rack == rack_inicial] or cand      # el "escritor" esta en este rack
            elif len(racks) == 1: pref = [d for d in cand if d.rack not in racks] or cand           # todas en un solo rack: forzar otro
            elif len(sel) == 2: pref = [d for d in cand if d.rack == self.dn[sel[-1]].rack] or cand   # tercera: mismo rack que la segunda
            else: pref = cand
            sel.append(min(pref, key=lambda d: (len(d.bloques), d.nombre)).nombre)
        return sel[len(existentes):]

    def crear(self, ruta, datos):
        ids = []
        for i in range(0, max(len(datos), 1), self.tam_bloque):
            trozo = datos[i:i + self.tam_bloque]
            bid, self._sig = self._sig, self._sig + 1
            racks = sorted({d.rack for d in self.dn.values()})
            nodos = self._elegir(self.factor, rack_inicial=racks[bid % len(racks)])     # los escritores se reparten entre los racks
            for n in nodos: self.dn[n].bloques[bid] = trozo
            self.ubic[bid], self.tam[bid] = set(nodos), len(trozo); ids.append(bid)
        self.nombres[ruta] = ids

    def leer(self, ruta):
        partes = []
        for bid in self.nombres[ruta]:
            vivas = sorted(n for n in self.ubic[bid] if self.dn[n].vivo)
            if not vivas: raise IOError(f"bloque {bid} perdido (sin replicas vivas)")
            partes.append(self.dn[vivas[0]].bloques[bid])                           # el cliente lee DIRECTO del DataNode
        return b"".join(partes)

    def fallar(self, nombre):
        d = self.dn[nombre]; d.vivo, d.bloques = False, {}                           # el nodo muere (y su disco con el)
        for bid in self.ubic:
            self.ubic[bid].discard(nombre)                                           # el NameNode lo detecta por heartbeats perdidos

    def fallar_rack(self, rack):
        for n, d in list(self.dn.items()):
            if d.rack == rack: self.fallar(n)

    def subreplicados(self):
        return {b: len(u) for b, u in self.ubic.items() if len(u) < self.factor}
    def perdidos(self): return [b for b, u in self.ubic.items() if not u]
    def replicas(self): return {b: len(u) for b, u in self.ubic.items()}

    def re_replicar(self):
        copias = 0
        for bid in sorted(self.subreplicados()):
            if not self.ubic[bid]: continue                                          # sin ninguna replica viva: irrecuperable
            origen = sorted(self.ubic[bid])[0]
            for n in self._elegir(self.factor - len(self.ubic[bid]), existentes=sorted(self.ubic[bid])):
                self.dn[n].bloques[bid] = self.dn[origen].bloques[bid]; self.ubic[bid].add(n); copias += 1
        return copias

    def bytes_almacenados(self): return sum(len(b) for d in self.dn.values() for b in d.bloques.values())

TOPOLOGIA = {"dn1": "R1", "dn2": "R1", "dn3": "R1", "dn4": "R2", "dn5": "R2", "dn6": "R2"}
rng = np.random.default_rng(SEED)
ventas = rng.bytes(1_000_000)                                                    # "archivo" de ~1 MB
pagos = rng.bytes(200_000)                                                       # "archivo" de ~200 KB (el ultimo bloque es parcial)

def nuevo_cluster(factor=3, tam_bloque=64 * 1024):
    fs = MiniHDFS(TOPOLOGIA, tam_bloque=tam_bloque, factor=factor)
    fs.crear("/datos/ventas.bin", ventas); fs.crear("/datos/pagos.bin", pagos)
    return fs

fs = nuevo_cluster()
n_bloques = {r: len(b) for r, b in fs.nombres.items()}
assert n_bloques == {"/datos/ventas.bin": math.ceil(1_000_000 / 65536), "/datos/pagos.bin": math.ceil(200_000 / 65536)}   # ceil(tamano / bloque)
assert fs.leer("/datos/ventas.bin") == ventas and fs.leer("/datos/pagos.bin") == pagos                                   # lectura identica
for bid, nodos in fs.ubic.items():
    racks = {fs.dn[n].rack for n in nodos}
    assert len(nodos) == 3 and len(racks) == 2                                    # 3 replicas en nodos distintos y en 2 racks
assert fs.bytes_almacenados() == 3 * (len(ventas) + len(pagos))                   # almacenamiento = factor x datos
assert fs.tam[fs.nombres["/datos/pagos.bin"][-1]] == 200_000 - 3 * 65536 < 65536   # el ultimo bloque es parcial: no desperdicia espacio
carga = [len(d.bloques) for d in fs.dn.values()]
print(f"NameNode: {len(fs.nombres)} archivos, {len(fs.ubic)} bloques ({n_bloques}) | replicas por bloque = 3 en 2 racks | datos {len(ventas) + len(pagos):,} B -> almacenados {fs.bytes_almacenados():,} B (x3)")
print("Bloques por DataNode:", dict(zip(fs.dn, carga)), "| diferencia maxima:", max(carga) - min(carga))
assert max(carga) - min(carga) <= 2                                               # el reparto es casi uniforme
````

````python
# ============================================================
# 1b. Falla de un DataNode, re-replicacion, perdida de datos segun el factor y tamano de bloque
# ============================================================
def matriz(fs, ruta):
    nodos = sorted(fs.dn)
    return np.array([[1 if n in fs.ubic[b] else 0 for n in nodos] for b in fs.nombres[ruta]]), nodos

M0, nodos = matriz(fs, "/datos/ventas.bin")
fs.fallar("dn2")                                                                 # cae un DataNode
bajos = fs.subreplicados(); M1, _ = matriz(fs, "/datos/ventas.bin")
assert len(bajos) > 0 and all(v == 2 for v in bajos.values()) and fs.perdidos() == []        # bloques con 2 replicas, ninguno perdido
assert fs.leer("/datos/ventas.bin") == ventas and fs.leer("/datos/pagos.bin") == pagos       # los datos SIGUEN legibles (replicas vivas)
copias = fs.re_replicar(); M2, _ = matriz(fs, "/datos/ventas.bin")
assert copias == len(bajos) and fs.subreplicados() == {}                          # el NameNode ordeno copiar cada bloque sub-replicado
assert all(v == fs.factor for v in fs.replicas().values())                       # NINGUN bloque con menos replicas que el factor
assert all(fs.dn[n].vivo for u in fs.ubic.values() for n in u) and not fs.dn["dn2"].bloques
assert fs.leer("/datos/ventas.bin") == ventas
print(f"Cayo dn2: {len(bajos)} bloques quedaron con 2 replicas (ninguno perdido; la lectura sigue funcionando) -> re-replicacion hizo {copias} copias -> todos los bloques vuelven a tener 3 replicas")

fs2 = nuevo_cluster(); bloques_r1 = sum(1 for b in fs2.ubic if all(fs2.dn[n].rack == "R1" for n in fs2.ubic[b]))
fs2.fallar_rack("R1")                                                            # se pierde un RACK completo (3 nodos)
assert fs2.perdidos() == [] and fs2.leer("/datos/ventas.bin") == ventas and bloques_r1 == 0
print(f"Cae el rack R1 completo (dn1, dn2, dn3): perdidos = {len(fs2.perdidos())} bloques; sobrevive porque cada bloque tiene replicas en los 2 racks. Re-replicar a 3 copias con solo 3 nodos: {fs2.re_replicar()} copias, replicas minimas = {min(fs2.replicas().values())}")
assert min(fs2.replicas().values()) == 3

# --- Perdida de datos segun el factor de replicacion (todas las combinaciones de fallas simultaneas) ---
filas = []
for factor in (1, 2, 3):
    f = nuevo_cluster(factor=factor)
    for k in (1, 2, 3):
        combos = list(itertools.combinations(sorted(f.dn), k))
        perdidas = sum(1 for c in combos if any(u <= set(c) for u in f.ubic.values()))      # algun bloque con TODAS sus replicas en los nodos caidos
        filas.append({"factor": factor, "fallas_simultaneas": k, "combinaciones": len(combos), "pierden_datos": perdidas, "fraccion": perdidas / len(combos), "almacenamiento_x": factor})
tabla = pd.DataFrame(filas)
print("\nFraccion de combinaciones de fallas simultaneas (de 6 nodos) que PIERDEN algun bloque:")
print(tabla.pivot(index="factor", columns="fallas_simultaneas", values="fraccion").round(2).to_string())
piv = tabla.set_index(["factor", "fallas_simultaneas"]).fraccion
assert piv[(1, 1)] > 0 and piv[(2, 1)] == 0 and piv[(3, 1)] == piv[(3, 2)] == 0   # factor f tolera f-1 fallas; el 3 tolera 2
assert piv[(3, 3)] > 0                                                           # pero 3 fallas SIMULTANEAS ya pueden perder datos

# --- Tamano de bloque y metadatos del NameNode ---
def objetos_namenode(fs): return len(fs.nombres) + len(fs.ubic)                   # cada archivo y cada bloque = un objeto en memoria del NameNode
datos_1mb = ventas[:1_048_576 - 48_576]
a = MiniHDFS(TOPOLOGIA, tam_bloque=64 * 1024); a.crear("/x", datos_1mb)
b = MiniHDFS(TOPOLOGIA, tam_bloque=8 * 1024); b.crear("/x", datos_1mb)
c = MiniHDFS(TOPOLOGIA, tam_bloque=64 * 1024)
for i in range(125): c.crear(f"/pequenos/{i}", datos_1mb[i * 8000:(i + 1) * 8000])          # los MISMOS bytes en 125 archivos pequenos
assert sum(len(c.leer(f"/pequenos/{i}")) for i in range(125)) == 1_000_000
print(f"\nMismo 1 MB: 1 archivo con bloques de 64 KB -> {objetos_namenode(a)} objetos en el NameNode | bloques de 8 KB -> {objetos_namenode(b)} | en 125 archivos pequenos -> {objetos_namenode(c)}")
assert objetos_namenode(a) < objetos_namenode(b) and objetos_namenode(a) < objetos_namenode(c)
print("Bloques grandes y pocos archivos = pocos metadatos; por eso HDFS prefiere archivos gigantes y los 'archivos pequenos' son un problema.")

fig, ax = plt.subplots(1, 3, figsize=(13, 4.4), sharey=True)
for a_, M, tit in zip(ax, (M0, M1, M2), ("Estado normal", "Cae dn2 (columna roja)", "Tras re-replicar")):
    a_.imshow(M, cmap="Blues", vmin=0, vmax=1.4, aspect="auto"); a_.set_xticks(range(len(nodos))); a_.set_xticklabels(nodos, fontsize=8)
    a_.set_title(f"{tit}\nréplicas por bloque: min {M.sum(axis=1).min()}, máx {M.sum(axis=1).max()}", fontsize=9.5, fontweight="bold"); a_.grid(False)
    a_.set_xlabel("DataNode (dn1-dn3: rack R1, dn4-dn6: rack R2)", fontsize=8)
ax[0].set_ylabel("bloque de ventas.bin"); ax[1].get_xticklabels()[1].set_color("#dc2626")
plt.tight_layout(); plt.show()
````

## 2. Spark: Arquitectura, DAG, Evaluación Perezosa y Caché ⚡

**Apache Spark** es un motor de **cómputo distribuido en memoria**. Un programa Spark tiene tres actores:

| Actor | Papel |
|---|---|
| **Driver** | Ejecuta tu programa (`SparkSession`/`SparkContext`); **construye el DAG** de transformaciones, lo parte en **etapas** y **tareas** y las programa |
| **Cluster manager** | Reparte recursos (YARN, Kubernetes, *standalone*; en nuestro modo local, el propio proceso) |
| **Ejecutores** (*executors*) | Procesos en los nodos que **ejecutan tareas** (una por **partición**) y **guardan en memoria** los datos en caché |

**Conceptos clave:**

- **Partición:** trozo de un conjunto de datos; **una tarea procesa una partición**. Más particiones = más paralelismo (hasta el número de núcleos) pero más sobrecarga.
- **Transformaciones** (`map`, `filter`, `groupBy`, `join`...) son **perezosas**: solo **registran** el linaje; **acciones** (`count`, `collect`, `write`) **disparan** la ejecución. Gracias a ello Spark ve **todo el plan** y puede optimizarlo.
- **Dependencias estrechas** (*narrow*: `map`, `filter`; cada partición de salida depende de **una** de entrada) se **encadenan en una misma etapa** sin mover datos. **Dependencias anchas** (*wide*: `reduceByKey`, `groupBy`, `join`, `repartition`) exigen un ***shuffle*** (mover datos por la red/disco) y **cortan** la etapa. **El número de *shuffles* determina el costo.**
- **Tolerancia a fallos por linaje:** si se pierde una partición, Spark **la recalcula** repitiendo las transformaciones que la produjeron (por eso no replica los datos como HDFS).
- ***Cache***/`persist`: guarda el resultado en memoria para **no recalcular** el linaje en cada acción; es **perezoso** (se materializa en la primera acción).
- **Catalyst:** el optimizador de DataFrames/SQL: reordena operaciones, **empuja filtros** hacia la lectura (*pushdown*) y elige el tipo de *join*; `explain()` muestra el plan físico.

Primero, **un mini-RDD desde cero** (`MiniRDD`: particiones, transformaciones perezosas con linaje, *shuffle* en `reduceByKey` y `cache()`), que **cuenta cuántas veces se ejecuta cada función de usuario**. Así se **ve** que (1) nada corre hasta la acción, (2) sin `cache()` cada acción **recalcula** el linaje, (3) un *shuffle* se **reutiliza** entre acciones (como los archivos de *shuffle* de Spark) y (4) las **etapas** se cortan en las dependencias anchas.

````python
# ============================================================
# 2a. MiniRDD desde cero: perezoso, linaje, etapas, shuffle y cache (con contador de llamadas)
# ============================================================
class MiniRDD:
    llamadas = Counter()                                                         # cuantas veces se ejecuto cada funcion de usuario

    def __init__(self, n, calcular, padre=None, nombre="parallelize", ancha=False):
        self.n, self._calcular, self.padre, self.nombre, self.ancha = n, calcular, padre, nombre, ancha
        self._usar_cache, self._cache, self._baraje = False, {}, None

    @classmethod
    def paralelizar(cls, datos, n=4):
        trozos = [list(datos[i::n]) for i in range(n)]
        return cls(n, lambda i: list(trozos[i]))

    @staticmethod
    def _llamar(f, *args):
        MiniRDD.llamadas[f.__name__] += 1; return f(*args)

    def _particion(self, i):
        if self._usar_cache and i in self._cache: return self._cache[i]
        res = self._calcular(i)
        if self._usar_cache: self._cache[i] = res
        return res

    # --- transformaciones (perezosas: solo construyen un nuevo MiniRDD con su linaje) ---
    def map(self, f): return MiniRDD(self.n, lambda i: [self._llamar(f, x) for x in self._particion(i)], self, "map")
    def filter(self, f): return MiniRDD(self.n, lambda i: [x for x in self._particion(i) if self._llamar(f, x)], self, "filter")
    def reduceByKey(self, f, n=None):
        n = n or self.n
        nuevo = MiniRDD(n, None, self, "reduceByKey", ancha=True)
        def calc(j):
            if nuevo._baraje is None:                                            # el shuffle se calcula UNA vez y se reutiliza (como los archivos de shuffle)
                cubetas = [{} for _ in range(n)]
                for i in range(self.n):
                    for k, v in self._particion(i):
                        c = cubetas[h64(k) % n]; c[k] = self._llamar(f, c[k], v) if k in c else v
                nuevo._baraje = cubetas; MiniRDD.llamadas["<shuffle>"] += 1
            return sorted(nuevo._baraje[j].items())
        nuevo._calcular = calc
        return nuevo
    def cache(self): self._usar_cache = True; return self

    # --- acciones (disparan la ejecucion) ---
    def collect(self): return [x for i in range(self.n) for x in self._particion(i)]
    def count(self): return sum(len(self._particion(i)) for i in range(self.n))

    # --- inspeccion ---
    def cadena(self):
        r, x = [], self
        while x is not None: r.append(x); x = x.padre
        return r[::-1]
    def etapas(self): return 1 + sum(1 for x in self.cadena() if x.ancha)
    def linaje(self): return " -> ".join(("[SHUFFLE] " if x.ancha else "") + x.nombre for x in self.cadena())

def costosa(x):                                                                  # la "funcion cara" de usuario
    return x * x % 97

def es_par(x):
    return x % 2 == 0

def a_par(x):
    return (x % 10, 1)

def sumar(a, b):
    return a + b

MiniRDD.llamadas.clear()
base = MiniRDD.paralelizar(list(range(1000)), 4)
canal = base.map(costosa).filter(es_par)                                         # solo se REGISTRA el linaje
assert sum(MiniRDD.llamadas.values()) == 0                                       # nada se ejecuto todavia: evaluacion perezosa
n1 = canal.count(); llamadas1 = MiniRDD.llamadas["costosa"]
canal.count(); llamadas2 = MiniRDD.llamadas["costosa"]
assert llamadas1 == 1000 and llamadas2 == 2000                                   # sin cache: cada accion RECALCULA el linaje
MiniRDD.llamadas.clear()
canal_c = MiniRDD.paralelizar(list(range(1000)), 4).map(costosa).filter(es_par).cache()
canal_c.count(); canal_c.count(); canal_c.collect()
assert MiniRDD.llamadas["costosa"] == 1000                                       # con cache: 1000 llamadas aunque haya 3 acciones
print(f"Perezoso: 0 llamadas al construir | sin cache: {llamadas1} -> {llamadas2} llamadas tras 2 acciones | con cache: {MiniRDD.llamadas['costosa']} tras 3 acciones")

MiniRDD.llamadas.clear()
conteo = MiniRDD.paralelizar(list(range(1000)), 4).map(costosa).map(a_par).reduceByKey(sumar).map(lambda kv: kv)
assert conteo.etapas() == 2 and conteo.n == 4
print("Linaje:", conteo.linaje(), "| etapas:", conteo.etapas())
r1 = conteo.collect(); costosa_1 = MiniRDD.llamadas["costosa"]
r2 = conteo.collect(); costosa_2 = MiniRDD.llamadas["costosa"]
assert r1 == r2 and costosa_1 == costosa_2 == 1000 and MiniRDD.llamadas["<shuffle>"] == 1      # el shuffle se reutiliza: la 2.a accion NO recalcula 'costosa'
ref = Counter()
for x in range(1000): ref[costosa(x) % 10] += 1
assert dict(r1) == dict(ref)                                                     # resultado correcto vs Counter
print(f"Con un shuffle en el linaje: 2 acciones y 'costosa' se ejecuto {costosa_2} veces (el shuffle se reutilizo: etapa omitida).")

# --- Dibujo: arquitectura (izquierda) y DAG de etapas del mini-pipeline (derecha) ---
def caja(ax, x, y, w, h, texto, color, fs=8.5):
    ax.add_patch(FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.03,rounding_size=0.12", fc=color, ec="white", lw=2))
    ax.text(x + w / 2, y + h / 2, texto, ha="center", va="center", color="white", fontsize=fs, fontweight="bold")
fig, ax = plt.subplots(1, 2, figsize=(13, 4.8))
a = ax[0]; a.set_xlim(0, 10); a.set_ylim(0, 8); a.axis("off"); a.grid(False)
caja(a, 1.8, 6.2, 6.4, 1.4, "DRIVER\nSparkSession · DAG · planificador", PALETA[0])
caja(a, 1.8, 3.6, 6.4, 1.1, "CLUSTER MANAGER\n(YARN / K8s / standalone / local)", PALETA[9])
for k, x in enumerate((0.2, 5.3)):
    caja(a, x, 0.4, 4.4, 2.2, f"WORKER {k + 1}\nEjecutor: tareas + caché\nparticiones " + ("0 y 1" if k == 0 else "2 y 3"), PALETA[2])
    a.annotate("", xy=(x + 2.2, 2.65), xytext=(5.0, 3.55), arrowprops=dict(arrowstyle="-|>", lw=1.6, color="#334155"))
a.annotate("", xy=(5.0, 4.75), xytext=(5.0, 6.15), arrowprops=dict(arrowstyle="<|-|>", lw=1.6, color="#334155"))
a.text(5.1, 5.45, "pide recursos", fontsize=8, color="#334155")
a.set_title("Arquitectura de Spark", fontsize=10.5, fontweight="bold")
b = ax[1]; b.set_xlim(0, 12); b.set_ylim(0, 6); b.axis("off"); b.grid(False)
posiciones = [(0.2, "parallelize"), (2.5, "map"), (4.8, "map")]
for x, t in posiciones:
    caja(b, x, 3.4, 2.0, 1.1, t, PALETA[0])
b.add_patch(plt.Rectangle((0.0, 3.1), 7.0, 1.7, fc="none", ec=PALETA[0], lw=2, ls="--")); b.text(3.5, 4.95, "ETAPA 0 (dependencias estrechas: se encadenan, sin mover datos)", ha="center", fontsize=8, color=PALETA[0], fontweight="bold")
caja(b, 7.7, 3.4, 2.0, 1.1, "reduceByKey", PALETA[1]); caja(b, 10.0, 3.4, 1.8, 1.1, "map", PALETA[1])
b.add_patch(plt.Rectangle((7.5, 3.1), 4.5, 1.7, fc="none", ec=PALETA[1], lw=2, ls="--")); b.text(9.75, 4.95, "ETAPA 1", ha="center", fontsize=8, color=PALETA[1], fontweight="bold")
for x0, x1 in [(2.2, 2.5), (4.5, 4.8), (6.8, 7.7), (9.7, 10.0)]:
    b.annotate("", xy=(x1, 3.95), xytext=(x0, 3.95), arrowprops=dict(arrowstyle="-|>", lw=1.8, color="#334155"))
b.text(7.25, 2.6, "SHUFFLE\n(corta la etapa)", ha="center", fontsize=8.5, color=PALETA[3], fontweight="bold")
b.text(6.0, 1.2, f"{conteo.n} particiones -> {conteo.n} tareas por etapa;\nlas acciones disparan el DAG: el linaje es el plan", ha="center", fontsize=8.5)
b.set_title("DAG del mini-pipeline: 2 etapas", fontsize=10.5, fontweight="bold")
plt.tight_layout(); plt.show()
````

````python
# ============================================================
# 2b. PySpark local: perezoso, particiones, shuffle, explain() y cache() (medido)
# ============================================================
os.environ["PYSPARK_PYTHON"] = sys.executable
from pyspark.sql import SparkSession, functions as F

t0 = time.perf_counter()
spark = (SparkSession.builder.master("local[2]").appName("bd12").config("spark.ui.enabled", "false")
         .config("spark.sql.shuffle.partitions", "4").config("spark.default.parallelism", "4")
         .config("spark.sql.warehouse.dir", f"{TMP}/warehouse").getOrCreate())
sc = spark.sparkContext; sc.setLogLevel("ERROR")
print(f"Sesion Spark local[2] lista en {time.perf_counter() - t0:.1f} s | version {spark.version}")

def plan(df):
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf): df.explain()
    return buf.getvalue()

rng = np.random.default_rng(SEED)
n = 300_000 if RAPIDO else 2_000_000
pdf = pd.DataFrame({"cliente": rng.integers(0, 2_000, n), "region": rng.choice(list("ABCDEFGH"), n), "monto": rng.gamma(2.0, 50.0, n).round(2)})
pdf.to_parquet(f"{TMP}/ventas.parquet")
sdf = spark.read.parquet(f"{TMP}/ventas.parquet"); sdf.count()                    # la lectura de metadatos tambien cuesta: se hace antes de medir

# (1) Evaluacion perezosa: construir el plan es instantaneo; ejecutarlo no
def armar():
    return sdf.filter(F.col("monto") > 50).groupBy("region").agg(F.count("*").alias("n"), F.avg("monto").alias("promedio"), F.sum("monto").alias("total"))
t0 = time.perf_counter()
consulta = armar()
t_plan = time.perf_counter() - t0
t0 = time.perf_counter(); res = consulta.toPandas().sort_values("region").reset_index(drop=True); t_accion = time.perf_counter() - t0
ref = pdf[pdf.monto > 50].groupby("region").monto.agg(n="count", promedio="mean", total="sum").reset_index()
assert list(res.region) == list(ref.region) and np.array_equal(res.n, ref.n) and np.allclose(res.promedio, ref.promedio) and np.allclose(res.total, ref.total)
assert t_plan < t_accion
print(f"Perezoso: construir la consulta {t_plan * 1000:.1f} ms | ejecutarla (accion) {t_accion * 1000:.0f} ms | resultado igual a pandas")

# (2) explain(): ¿cuantos shuffles (Exchange) tiene cada plan?
p_filtro = plan(sdf.filter(F.col("monto") > 50).select("cliente", "monto"))
p_grupo = plan(armar())                                                          # un DataFrame NUEVO: aun no ejecutado (si no, AQE muestra tambien el plan final)
print("\n--- Plan de filter + select (sin Exchange; fijate en PushedFilters) ---"); print(p_filtro.strip())
print("\n--- Plan del groupBy (un Exchange = un shuffle, entre la agregacion parcial y la final) ---"); print(p_grupo.strip())
assert p_filtro.count("Exchange") == 0 and "PushedFilters" in p_filtro and "GreaterThan" in p_filtro      # el filtro se EMPUJA a la lectura del Parquet
assert p_grupo.count("Exchange") == 1 and p_grupo.count("HashAggregate") == 2                               # agregacion parcial -> shuffle -> agregacion final

# (3) Particiones, shuffle y linaje con RDDs
rdd = sc.parallelize(range(100_000), 4)
assert rdd.getNumPartitions() == 4 and rdd.glom().map(len).collect() == [25_000] * 4
rep = rdd.repartition(8); coal = rdd.coalesce(2)
assert rep.getNumPartitions() == 8 and coal.getNumPartitions() == 2
assert "ShuffledRDD" in rep.toDebugString().decode() and "ShuffledRDD" not in coal.toDebugString().decode()      # repartition hace shuffle; coalesce no
pares = rdd.map(lambda x: (x % 5, 1)).reduceByKey(lambda a, b: a + b)
assert dict(pares.collect()) == {k: 20_000 for k in range(5)}
print("\nLinaje de un reduceByKey (la sangria marca la frontera de etapa; ShuffledRDD = el shuffle):")
for linea in pares.toDebugString().decode().splitlines()[:5]: print("   ", linea[:105])
print(f"Particiones: parallelize=4 (25 000 elementos cada una) | repartition(8) hace shuffle | coalesce(2) NO hace shuffle")

# (4) cache(): recalcular un linaje caro en cada accion frente a materializarlo una vez
import math
def cara(x):
    s = 0.0
    for i in range(1, 160): s += math.sqrt(x + i)
    return s
N_C = 30_000 if RAPIDO else 150_000
datos_c = sc.parallelize(range(N_C), 4).map(cara)
def acciones(rdd_):
    return [rdd_.count(), round(rdd_.sum(), 3), round(rdd_.max(), 3), round(rdd_.min(), 3), round(rdd_.mean(), 3)]
t0 = time.perf_counter(); r_sin = acciones(datos_c); t_sin = time.perf_counter() - t0
datos_cc = sc.parallelize(range(N_C), 4).map(cara).cache()
t0 = time.perf_counter(); r_con = acciones(datos_cc); t_con = time.perf_counter() - t0
assert np.allclose(r_sin, r_con) and datos_cc.is_cached
datos_cc.unpersist()
print(f"\ncache(): 5 acciones sobre un linaje caro -> sin cache {t_sin:.2f} s | con cache {t_con:.2f} s ({t_sin / t_con:.1f} veces mas rapido; la primera accion paga el calculo y las otras 4 leen de memoria)")
assert t_con < t_sin
spark.stop()                                                                      # detiene los procesos de Spark al final de la seccion
shutil.rmtree(TMP, ignore_errors=True)                                           # borra el Parquet temporal
print("Sesion Spark detenida y archivos temporales borrados.")
````

## 3. HBase: el Modelo de Datos de una Tabla Ordenada por Clave 🧮

**HBase** (inspirada en **Bigtable** de Google) es una base de datos de **familias de columnas** que corre sobre HDFS y está pensada para **acceso aleatorio de baja latencia** a **miles de millones de filas**. Su modelo de datos es un **mapa ordenado, disperso y multidimensional**:

$$ (\text{clave de fila},\ \text{familia}{:}\text{calificador},\ \text{timestamp}) \;\longrightarrow\; \text{valor} $$

| Concepto | Qué es | Consecuencia práctica |
|---|---|---|
| **Clave de fila** | Cadena de bytes; las filas se guardan **ordenadas lexicográficamente** por ella | El **escaneo por rango** es barato; **diseñar la clave es diseñar las consultas** |
| **Familia de columnas** | Grupo de columnas que se **guardan juntas** y se declaran al crear la tabla | Se leen juntas (solo lees la familia que necesitas); tienen su propia política de versiones |
| **Calificador** | Nombre de la columna **dentro** de la familia; se crea al escribir | Filas **dispersas**: no se guardan columnas vacías |
| **Versiones** | Cada celda guarda **varios valores con *timestamp***, hasta un máximo por familia | *Historial* y lecturas «en el tiempo»; el borrado usa **marcas** (*tombstones*) |
| **Región** | **Rango contiguo de claves** `[inicio, fin)` de una tabla. Cuando una región **crece demasiado se divide** en dos | La tabla se **reparte** entre *RegionServers*; el cliente localiza la región por la clave |
| **RegionServer / HMaster** | Los servidores atienden las regiones; el maestro las asigna y balancea | Un *RegionServer* con una región «caliente» se convierte en cuello de botella |

**La trampa clásica, los puntos calientes (*hotspotting*):** si las claves **crecen monótonamente** (marcas de tiempo, contadores), **todas las escrituras nuevas caen en la última región**, es decir, en **un solo servidor**, mientras los demás descansan. Soluciones: **prefijar la clave con una «sal»** (un cubo de hash), **invertir** la clave o **pre-dividir** la tabla; **el costo** es que un rango de tiempo ya no es contiguo y hay que **consultar cada cubo**.

La celda siguiente implementa `MiniHBase` (mapa ordenado con `bisect`, versiones con *tombstones*, `scan` por rango o prefijo, **regiones que se dividen** al superar un umbral —las hijas nacen en el mismo servidor y un **balanceador** las reparte— y *RegionServers*) y **verifica con `assert`**: orden, rangos contra fuerza bruta, límite de versiones, lecturas históricas, borrado, familias inválidas y que las **regiones cubren todo el espacio de claves sin huecos ni solapes**.

````python
# ============================================================
# 3a. MiniHBase desde cero: filas ordenadas, familias, versiones, escaneo por rango y regiones
# ============================================================
_BORRADO = object()                                                              # marca de borrado (tombstone)

class MiniHBase:
    def __init__(self, familias, max_filas_region=200, servidores=("rs1", "rs2", "rs3", "rs4")):
        self.familias, self.umbral, self.servidores = dict(familias), max_filas_region, list(servidores)   # familias: nombre -> max. versiones
        self.reloj, self.divisiones = 0, 0
        self.regiones = [{"inicio": "", "fin": None, "claves": [], "filas": {}, "servidor": self.servidores[0]}]
        self.regiones_visitadas = 0                                              # cuantas regiones toco la ultima consulta

    def _region(self, fila):                                                     # equivale a consultar la tabla META: ¿que region tiene esta clave?
        inicios = [r["inicio"] for r in self.regiones]
        return self.regiones[bisect.bisect_right(inicios, fila) - 1]

    def servidor_de(self, fila): return self._region(fila)["servidor"]

    def put(self, fila, familia, calificador, valor, ts=None):
        if familia not in self.familias: raise KeyError(f"familia desconocida: {familia}")
        self.reloj += 1; ts = self.reloj if ts is None else ts
        reg = self._region(fila)
        if fila not in reg["filas"]:
            reg["filas"][fila] = {}; bisect.insort(reg["claves"], fila)           # la clave entra EN ORDEN
        celda = reg["filas"][fila].setdefault((familia, calificador), [])
        celda.append((ts, valor)); celda.sort(key=lambda t: -t[0]); del celda[self.familias[familia]:]   # mas nuevo primero; solo N versiones
        if len(reg["claves"]) > self.umbral: self._dividir(reg)
        return ts

    def delete(self, fila, familia, calificador):
        self.reloj += 1; reg = self._region(fila)
        if fila in reg["filas"] and (familia, calificador) in reg["filas"][fila]:
            reg["filas"][fila][(familia, calificador)].insert(0, (self.reloj, _BORRADO))   # marca: oculta TODO lo anterior

    def _dividir(self, reg):
        m = len(reg["claves"]) // 2; mitad = reg["claves"][m]                    # se divide por la clave del medio
        nueva = {"inicio": mitad, "fin": reg["fin"], "claves": reg["claves"][m:], "filas": {k: reg["filas"].pop(k) for k in reg["claves"][m:]}, "servidor": reg["servidor"]}
        reg["claves"], reg["fin"] = reg["claves"][:m], mitad                      # la region hija nace en el MISMO servidor (en HBase real, igual)
        pos = next(i for i, r in enumerate(self.regiones) if r is reg)
        self.regiones.insert(pos + 1, nueva); self.divisiones += 1

    def equilibrar(self):
        """Balanceador: mueve regiones del servidor con mas regiones al que menos tiene, hasta que difieran en 1 (en HBase real corre cada cierto tiempo)."""
        while True:
            carga = {s: [r for r in self.regiones if r["servidor"] == s] for s in self.servidores}
            mayor, menor = max(self.servidores, key=lambda s: (len(carga[s]), s)), min(self.servidores, key=lambda s: (len(carga[s]), s))
            if len(carga[mayor]) - len(carga[menor]) <= 1: return
            carga[mayor][0]["servidor"] = menor

    def _celdas(self, fila, familias, calificador, versiones, hasta_ts):
        out = {}
        for (fam, cal), vers in self._region(fila)["filas"].get(fila, {}).items():
            if (familias and fam not in familias) or (calificador and cal != calificador): continue
            visibles = []
            for ts, val in vers:                                                 # vers esta ordenado de mas nuevo a mas viejo
                if hasta_ts is not None and ts > hasta_ts: continue              # lectura "en el tiempo"
                if val is _BORRADO: break                                        # una marca de borrado oculta todo lo mas viejo
                visibles.append((ts, val))
            if visibles: out[(fam, cal)] = visibles[:versiones]
        return out

    def get(self, fila, familia=None, calificador=None, versiones=1, hasta_ts=None):
        return self._celdas(fila, [familia] if familia else None, calificador, versiones, hasta_ts)

    def scan(self, inicio="", fin=None, prefijo=None, familias=None, versiones=1, limite=None):
        if prefijo is not None:
            inicio = max(inicio, prefijo); fin = prefijo + "￿" if fin is None else min(fin, prefijo + "￿")
        self.regiones_visitadas, emitidas = 0, 0
        for reg in self.regiones:                                                # solo se visitan las regiones que se solapan con el rango
            if (reg["fin"] is not None and inicio >= reg["fin"]) or (fin is not None and reg["inicio"] >= fin): continue
            self.regiones_visitadas += 1
            i = bisect.bisect_left(reg["claves"], inicio)
            while i < len(reg["claves"]) and (fin is None or reg["claves"][i] < fin):
                fila = reg["claves"][i]; i += 1
                celdas = self._celdas(fila, familias, None, versiones, None)
                if celdas:
                    yield fila, celdas; emitidas += 1
                    if limite is not None and emitidas >= limite: return

# ---- Tabla de perfiles de usuario: dos familias con politicas de versiones distintas ----
rng = np.random.default_rng(SEED)
tabla = MiniHBase({"datos": 3, "actividad": 1}, max_filas_region=100)
N_USU = 450
for u in rng.permutation(N_USU):                                                 # se insertan en DESORDEN: HBase los guarda ordenados
    fila = f"usuario#{u:04d}"
    tabla.put(fila, "datos", "nombre", f"Usuario {u}")
    tabla.put(fila, "datos", "correo", f"u{u}@ejemplo.co")
    tabla.put(fila, "actividad", "ultimo_login", int(rng.integers(1, 1000)))
    if u % 3 == 0: tabla.put(fila, "datos", "telefono", f"300{u:07d}")           # columnas dispersas: solo 1 de cada 3 usuarios
todas = [f for f, _ in tabla.scan()]
assert todas == sorted(todas) == [f"usuario#{u:04d}" for u in range(N_USU)]      # el escaneo completo sale ORDENADO
assert [f for f, _ in tabla.scan("usuario#0100", "usuario#0110")] == [f"usuario#{u:04d}" for u in range(100, 110)]     # rango [inicio, fin) contra fuerza bruta
assert [f for f, _ in tabla.scan(prefijo="usuario#03")] == [f"usuario#{u:04d}" for u in range(300, 400)]
assert tabla.regiones_visitadas < len(tabla.regiones) and len(tabla.regiones) > 4   # el escaneo por prefijo NO toca todas las regiones

# versiones: el correo de un usuario cambia 4 veces y solo se guardan las 3 versiones mas nuevas (max_versiones de 'datos' = 3)
for i in range(4): tabla.put("usuario#0007", "datos", "correo", f"nuevo{i}@ejemplo.co", ts=1000 + i)
vers = tabla.get("usuario#0007", "datos", "correo", versiones=10)[("datos", "correo")]
assert [v for _, v in vers] == ["nuevo3@ejemplo.co", "nuevo2@ejemplo.co", "nuevo1@ejemplo.co"]                   # solo 3, de la mas nueva a la mas vieja
assert tabla.get("usuario#0007", "datos", "correo", hasta_ts=1001)[("datos", "correo")][0][1] == "nuevo1@ejemplo.co"   # lectura "en el tiempo"
assert tabla.get("usuario#0007", "datos", "correo")[("datos", "correo")][0][1] == "nuevo3@ejemplo.co"                # por defecto: la version mas nueva
tabla.delete("usuario#0007", "datos", "correo")
assert ("datos", "correo") not in tabla.get("usuario#0007", versiones=10)                                           # la marca oculta todo
tabla.put("usuario#0007", "datos", "correo", "otra@ejemplo.co")
assert tabla.get("usuario#0007", "datos", "correo")[("datos", "correo")][0][1] == "otra@ejemplo.co"                 # y una escritura posterior reaparece
try:
    tabla.put("usuario#0007", "perfil", "foto", "x.png"); raise AssertionError("debio rechazar la familia")
except KeyError:
    pass
fila_sin = tabla.get("usuario#0001")                                           # usuario 1 no es multiplo de 3: sin 'telefono' (columna dispersa)
assert ("datos", "telefono") not in fila_sin and ("datos", "telefono") in tabla.get("usuario#0003")
# regiones: cubren el espacio de claves SIN huecos ni solapes y ninguna supera el umbral
tabla.equilibrar()                                                               # las hijas nacen en el mismo servidor; el balanceador las reparte
regs = tabla.regiones
assert max(Counter(r["servidor"] for r in regs).values()) - min(Counter(r["servidor"] for r in regs).get(s, 0) for s in tabla.servidores) <= 1
assert regs[0]["inicio"] == "" and regs[-1]["fin"] is None and all(regs[i]["fin"] == regs[i + 1]["inicio"] for i in range(len(regs) - 1))
assert all(len(r["claves"]) <= tabla.umbral for r in regs) and sum(len(r["claves"]) for r in regs) == N_USU
assert all(r["inicio"] <= k and (r["fin"] is None or k < r["fin"]) for r in regs for k in r["claves"])
print(f"MiniHBase OK: {N_USU} filas ordenadas | {len(regs)} regiones tras {tabla.divisiones} divisiones | servidores: {dict(Counter(r['servidor'] for r in regs))}")
print("Escaneo por prefijo 'usuario#03':", len(list(tabla.scan(prefijo="usuario#03"))), "filas, regiones visitadas:", tabla.regiones_visitadas, "de", len(regs))
````

````python
# ============================================================
# 3b. Puntos calientes: claves monotonas frente a claves con sal (y el costo de salar)
# ============================================================
def cargar(clave, n_inicial=4_000, n_nuevas=2_000):
    t = MiniHBase({"m": 1}, max_filas_region=500)
    for i in range(n_inicial): t.put(clave(i), "m", "v", i)                       # tabla ya poblada: se dividio en varias regiones
    t.equilibrar()                                                                # el balanceador ya repartio esas regiones entre los servidores
    escrituras = Counter()
    for i in range(n_inicial, n_inicial + n_nuevas):                              # las NUEVAS escrituras: ¿a que servidor van?
        escrituras[t.servidor_de(clave(i))] += 1; t.put(clave(i), "m", "v", i)
    return t, escrituras

monotona = lambda i: f"{i:08d}"                                                   # clave = marca de tiempo creciente
con_sal = lambda i: f"{i % 8}-{i:08d}"                                            # prefijo 0..7 (cubo) + marca de tiempo
tA, wA = cargar(monotona); tB, wB = cargar(con_sal)
parte_max_A, parte_max_B = max(wA.values()) / 2_000, max(wB.values()) / 2_000
assert parte_max_A == 1.0 and len(wA) == 1                                        # TODAS las escrituras nuevas van a UN solo servidor
assert parte_max_B < 0.5 and len(wB) >= 3                                         # con sal se reparten entre varios servidores
print(f"Escrituras nuevas por servidor (2 000): clave monotona = {dict(wA)} (max {parte_max_A:.0%}) | con sal = {dict(sorted(wB.items()))} (max {parte_max_B:.0%})")

# El costo de la sal: consultar un rango de tiempo [3000, 3100) toca 1 region sin sal y varias con sal (hay que buscar en cada cubo)
filasA = [f for f, _ in tA.scan(monotona(3000), monotona(3100))]; reg_A = tA.regiones_visitadas
filasB, reg_B = [], 0
for cubo in range(8):
    filasB += [f for f, _ in tB.scan(f"{cubo}-{3000:08d}", f"{cubo}-{3100:08d}")]; reg_B += tB.regiones_visitadas
assert len(filasA) == 100 and sorted(int(f.split("-")[-1]) for f in filasB) == list(range(3000, 3100))      # mismas 100 filas
assert reg_A == 1 and reg_B > reg_A
print(f"Rango de 100 filas consecutivas: sin sal visita {reg_A} region(es) | con sal hay que lanzar 8 escaneos (uno por cubo) que visitan {reg_B} regiones en total.")

fig, ax = plt.subplots(1, 2, figsize=(12, 3.9))
srv = sorted(set(tA.servidores))
x = np.arange(len(srv)); w = 0.4
ax[0].bar(x - w / 2, [wA.get(s, 0) for s in srv], w, color=PALETA[1], label="clave monótona (punto caliente)")
ax[0].bar(x + w / 2, [wB.get(s, 0) for s in srv], w, color=PALETA[2], label="clave con sal (8 cubos)")
ax[0].set_xticks(x); ax[0].set_xticklabels(srv); ax[0].set_ylabel("escrituras nuevas recibidas"); ax[0].legend(fontsize=8)
ax[0].set_title("Quién recibe las 2 000 escrituras nuevas", fontsize=10, fontweight="bold")
for k, (t, titulo) in enumerate(((tA, "monótona"), (tB, "con sal"))):
    for j, r in enumerate(t.regiones):
        ax[1].barh(k, len(r["claves"]), left=sum(len(q["claves"]) for q in t.regiones[:j]), color=PALETA[srv.index(r["servidor"]) % len(PALETA)], edgecolor="white")
ax[1].set_yticks([0, 1]); ax[1].set_yticklabels(["monótona", "con sal"]); ax[1].set_xlabel("filas (cada bloque = una región; el color indica el RegionServer)")
from matplotlib.patches import Patch
ax[1].legend(handles=[Patch(color=PALETA[i], label=s) for i, s in enumerate(srv)], fontsize=7, ncol=4, loc="upper center", bbox_to_anchor=(0.5, -0.2))
ax[1].set_title("Regiones de cada tabla (6 000 filas)", fontsize=10, fontweight="bold")
plt.tight_layout(); plt.show()
````

## 4. HDFS, Spark y HBase: Comparación y Cuándo Usar Cada Uno 🧭

Las tres piezas **no compiten**: resuelven problemas distintos y suelen usarse **juntas** (HBase guarda sus archivos en HDFS; Spark lee y escribe HDFS y puede leer HBase).

| | **HDFS** | **Spark** | **HBase** |
|---|---|---|---|
| **¿Qué es?** | Sistema de **archivos** distribuido | Motor de **cómputo** distribuido en memoria | Base de datos de **familias de columnas** (tabla ordenada por clave) |
| **Unidad básica** | **Bloque** de un archivo (128 MB por defecto *(verificar)*) | **Partición** de un RDD/DataFrame | **Celda** `(fila, familia:columna, timestamp)`; **región** = rango de claves |
| **Patrón de acceso** | **Lectura secuencial** de archivos enormes; escribir una vez | **Análisis** de grandes volúmenes: SQL, ETL, ML, *streaming* | **Acceso aleatorio** por clave y **escaneo por rango**, con baja latencia |
| **Latencia** | Alta para archivos pequeños; pensado para *throughput* | Segundos a minutos por trabajo (milisegundos en *streaming* continuo) | **Milisegundos** por lectura o escritura |
| **Tolerancia a fallos** | **Replicación** de bloques (×3) y re-replicación | **Linaje**: recalcula particiones perdidas | Regiones se **reasignan**; el registro de escritura y los archivos viven en HDFS |
| **Escalado** | Añadir DataNodes | Añadir ejecutores | Añadir RegionServers; las regiones se dividen y se reparten |
| **Modelo de consistencia** | Escritura única, lectura múltiple | Sin estado persistente propio | Fuerte **por fila** (sin transacciones multi-fila) |
| **Punto débil** | Archivos pequeños; NameNode | Memoria; *shuffles* caros; costo de arranque | Diseño de la clave de fila (puntos calientes); sin `JOIN`s ni consultas por columna |
| **Úsalo cuando...** | Necesitas **almacenar** datos crudos y masivos, baratos y duraderos (lago de datos) | Necesitas **transformar o analizar** esos datos en paralelo | Necesitas **servir** lecturas y escrituras puntuales de miles de millones de filas (perfiles, mensajes, series de tiempo) |
| **En este cuaderno** | `MiniHDFS` | `MiniRDD` + PySpark | `MiniHBase` |

> 🧩 **Una arquitectura típica:** los datos crudos llegan a **HDFS** (o a almacenamiento de objetos en la nube); **Spark** los limpia y agrega; los resultados que deben **consultarse por clave en milisegundos** se escriben en **HBase** (o en otro almacén NoSQL del [cuaderno 00](00_Opciones_de_Procesamiento_Hadoop_Spark_NoSQL_SQL.ipynb)); los informes de gestión se sirven desde un almacén analítico (Módulo 02). Para profundizar en Spark: [Data Mining, Módulo 07](../../Data%20Mining/07%20-%20Mineria%20de%20Datos%20con%20Big%20Data/README.md).

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **HDFS** | **NameNode** (metadatos: `archivo → bloques → DataNodes`) + **DataNodes** (bytes). Factor **3** con *racks*: 3 réplicas en nodos distintos y 2 *racks*. Al caer un DataNode, los datos siguen legibles y el NameNode **re-replica**; verificado que **ningún bloque queda con menos réplicas** que el factor. Factor $f$ tolera $f-1$ fallas simultáneas; **costo: $f\times$ el espacio** |
| **Bloques y metadatos** | Bloques grandes y pocos archivos = pocos objetos en el NameNode; **muchos archivos pequeños** lo saturan |
| **Spark** | **Driver** + *cluster manager* + **ejecutores**. Transformaciones **perezosas** (linaje/DAG); las **dependencias anchas** (*shuffle*) cortan **etapas**. `cache()` evita recalcular (medido: varias veces más rápido con un linaje caro); `explain()` muestra `Exchange` (un *shuffle*) y el empuje de filtros (*pushdown*) |
| **HBase** | Mapa **ordenado y disperso**: `(fila, familia:calificador, timestamp) → valor`; **versiones** por familia, escaneo por **rango** y **regiones** que se dividen. La **clave de fila** es el diseño: con claves monótonas **todas** las escrituras caen en un servidor; la **sal** lo reparte pero **encarece** los rangos |
| **Cuándo usar cada uno** | **HDFS** almacena, **Spark** procesa, **HBase** sirve consultas por clave; se **combinan** |
