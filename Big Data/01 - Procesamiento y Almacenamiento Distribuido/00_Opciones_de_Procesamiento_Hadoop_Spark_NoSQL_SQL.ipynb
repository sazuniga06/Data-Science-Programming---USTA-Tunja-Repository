## 1. Procesamiento por Lotes y por Flujo ⏱️

La primera decisión no es de producto sino de **cuándo** se procesa el dato:

| | **Por lotes** (*batch*) | **Por flujo** (*stream*) |
|---|---|---|
| **Entrada** | Conjunto **acotado** y ya guardado (los pedidos de ayer) | Secuencia **no acotada** que llega sin parar (cada pedido, al ocurrir) |
| **Cuándo se calcula** | Periódicamente, sobre **todo** el lote | De forma **continua**, evento a evento o por micro-lotes |
| **Latencia de la respuesta** | De minutos a horas (la del intervalo del lote) | De milisegundos a segundos |
| **Rendimiento (*throughput*)** | Muy alto: planifica y recorre los datos en bloque | Más limitado por evento; exige estado y tolerancia a fallos |
| **Reintentos** | Sencillos: se **repite** el lote completo | Difíciles: duplicados, orden y **tiempo de evento** |
| **Ejemplos** | Facturación mensual, reportes, entrenamiento de modelos | Alertas de fraude, tableros en vivo, telemetría |

**No son rivales:** muchos sistemas combinan ambos (una capa de lotes con el histórico exacto y una capa de flujo con lo reciente). Una idea útil: **un lote es un flujo con una ventana enorme**; si el lote cierra cada $W$ segundos, un evento espera en promedio $W/2$ antes de entrar en una respuesta, y la latencia media tiende a **cero solo cuando $W \to 0$** (flujo). La celda siguiente lo **mide** con 12 000 eventos sintéticos: el **total final es idéntico** en ambos modos; lo que cambia es **cuánto tarda en estar disponible** la respuesta.

````python
# ============================================================
# Configuracion inicial y demostracion lote vs flujo
# ============================================================
# !pip install -q numpy pandas matplotlib duckdb pyarrow pyspark

import itertools
import json
import os
import shutil
import sqlite3
import sys
import tempfile
import time
import warnings
from collections import Counter, defaultdict

import duckdb
import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
import pandas as pd
from matplotlib.patches import FancyBboxPatch

warnings.filterwarnings("ignore")
SEED = 42
RAPIDO = True                                                                  # True: tamanos reducidos (el cuaderno completo corre en ~1 min)
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
TMP = tempfile.mkdtemp(prefix="bd10_")                                         # todo lo que se escribe va aqui y se borra al final
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | duckdb {duckdb.__version__} | RAPIDO={RAPIDO}")

# --- 12 000 eventos sinteticos en 1 hora: (instante en segundos, monto) ---
rng = np.random.default_rng(SEED)
N_EV, HORIZONTE = 12_000, 3600
t_ev = np.sort(rng.uniform(0, HORIZONTE, N_EV))
monto_ev = rng.gamma(2.0, 50.0, N_EV).round(2)

def total_por_lotes(t, m, W):
    """Lote que cierra cada W segundos: devuelve el total y el retraso (cierre de ventana - instante del evento) de cada evento."""
    cierre = (np.floor(t / W) + 1) * W
    return m.sum(), cierre - t

def total_por_flujo(t, m):
    """Flujo: el acumulado se actualiza al llegar cada evento (retraso ~ 0 en el modelo)."""
    acumulado = 0.0
    for x in m:
        acumulado += x
    return acumulado, np.zeros_like(t)

tot_flujo, lag_flujo = total_por_flujo(t_ev, monto_ev)
ventanas = [10, 60, 300, 900, 3600]
retrasos = {}
for W in ventanas:
    tot_lote, lag = total_por_lotes(t_ev, monto_ev, W)
    assert np.isclose(tot_lote, tot_flujo)                                      # mismo resultado final
    assert abs(lag.mean() - W / 2) / (W / 2) < 0.06                              # el retraso medio de un lote es ~ W/2
    retrasos[W] = float(lag.mean())
print("Total de montos (identico en ambos modos):", round(tot_flujo, 2))
print("Retraso medio hasta que el evento aparece en una respuesta (s):", {W: round(v, 1) for W, v in retrasos.items()}, "| flujo ~ 0")

fig, ax = plt.subplots(1, 2, figsize=(12, 4.2))
ax[0].eventplot(t_ev[t_ev < 300][::6], colors="#475569", lineoffsets=0.5, linelengths=0.6)
for k in range(0, 5):
    ax[0].axvline(k * 60, color=PALETA[1], lw=1.4, ls="--")
ax[0].axvspan(0, 60, color=PALETA[0], alpha=0.10)
ax[0].set_xlim(0, 300); ax[0].set_ylim(0, 1.2); ax[0].set_yticks([])
ax[0].set_xlabel("tiempo (s)"); ax[0].set_title("Eventos (trazos) y cierres de un lote de W = 60 s (líneas rojas)\nflujo: se actualiza en cada trazo", fontsize=10, fontweight="bold")
ax[1].plot(ventanas, [retrasos[W] for W in ventanas], "o-", color=PALETA[0], label="medido")
ax[1].plot(ventanas, [W / 2 for W in ventanas], "--", color=PALETA[1], label="teoría: W/2")
ax[1].set_xscale("log"); ax[1].set_yscale("log"); ax[1].set_xlabel("ventana del lote W (s)"); ax[1].set_ylabel("retraso medio (s)")
ax[1].set_title("Latencia = precio de procesar en bloque", fontsize=10, fontweight="bold"); ax[1].legend()
plt.tight_layout(); plt.show()
print("Un lote de 60 s ya implica ~30 s de espera media; si la decision es una alerta de fraude, 30 s pueden ser demasiado; si es un informe mensual, son irrelevantes.")
````

## 2. El Modelo Relacional: SQL, ACID y Dos Motores 🗃️

El **modelo relacional** guarda los datos en **tablas** con esquema fijo, relaciona tablas con **claves** y se consulta con **SQL** (declarativo: dices *qué* quieres y el motor decide *cómo*). Su gran aporte son las transacciones **ACID**:

| Letra | Garantía | Significado práctico |
|---|---|---|
| **A** — Atomicidad | Todo o nada | Una transferencia que falla a medias **se revierte** completa |
| **C** — Consistencia | Las reglas siempre se cumplen | Un saldo nunca queda negativo si hay un `CHECK`; las claves foráneas apuntan a filas que existen |
| **I** — Aislamiento | Transacciones concurrentes no se «ven» a medias | Dos operaciones simultáneas se comportan como si fueran una tras otra (según el nivel) |
| **D** — Durabilidad | Lo confirmado sobrevive a una caída | Tras el `COMMIT`, el dato está en disco (registro de escritura anticipada) |

Dos motores de la biblioteca estándar y de uso común en ciencia de datos, con **diseños distintos**: **SQLite** guarda **filas** (pensado para muchas operaciones pequeñas de lectura y escritura: **OLTP**) y **DuckDB** guarda **columnas** y ejecuta de forma **vectorizada** (pensado para escanear y agregar muchas filas: **OLAP**). Ambos entienden SQL; la celda siguiente (1) demuestra **atomicidad y consistencia** con una transferencia que falla y (2) comprueba que **la misma consulta** da el mismo resultado en SQLite, DuckDB y *pandas*.

````python
# ============================================================
# 2. SQLite: transaccion ACID (atomicidad + consistencia) y la misma consulta en 3 motores
# ============================================================
con = sqlite3.connect(":memory:")
con.executescript("""
CREATE TABLE cuentas (id INTEGER PRIMARY KEY, titular TEXT NOT NULL, saldo INTEGER NOT NULL CHECK (saldo >= 0));
INSERT INTO cuentas VALUES (1, 'Ana', 1000), (2, 'Beto', 500), (3, 'Carla', 300);
""")

def total_saldos():
    return con.execute("SELECT SUM(saldo) FROM cuentas").fetchone()[0]

def transferir(origen, destino, monto):
    """Dos UPDATE dentro de UNA transaccion: o se aplican ambos o ninguno."""
    try:
        with con:                                                               # BEGIN ... COMMIT (o ROLLBACK si hay excepcion)
            con.execute("UPDATE cuentas SET saldo = saldo - ? WHERE id = ?", (monto, origen))
            con.execute("UPDATE cuentas SET saldo = saldo + ? WHERE id = ?", (monto, destino))
        return True
    except sqlite3.IntegrityError:                                              # el CHECK (saldo >= 0) fallo
        return False

antes = total_saldos()
assert transferir(1, 2, 400) is True                                            # transferencia valida
assert dict(con.execute("SELECT titular, saldo FROM cuentas")) == {"Ana": 600, "Beto": 900, "Carla": 300}
assert transferir(3, 1, 999) is False                                           # Carla solo tiene 300: el primer UPDATE viola el CHECK
assert dict(con.execute("SELECT titular, saldo FROM cuentas")) == {"Ana": 600, "Beto": 900, "Carla": 300}   # nada cambio: atomicidad
assert total_saldos() == antes == 1800                                          # invariante de consistencia: el dinero total se conserva
print("Saldos tras una transferencia valida y una invalida:", dict(con.execute("SELECT titular, saldo FROM cuentas")), "| total =", total_saldos())

# --- La misma consulta analitica en SQLite, DuckDB y pandas (datos sinteticos) ---
n_ped = 60_000 if RAPIDO else 400_000
rng = np.random.default_rng(SEED)
municipios = ["Tunja", "Duitama", "Sogamoso", "Chiquinquirá", "Paipa", "Villa de Leyva", "Samacá", "Ventaquemada"]
pedidos = pd.DataFrame({
    "pedido": np.arange(n_ped),
    "municipio": rng.choice(municipios, n_ped),
    "monto": rng.lognormal(mean=10.0, sigma=0.6, size=n_ped).round(0),
})
pedidos.to_sql("pedidos", con, index=False)
SQL = "SELECT municipio, COUNT(*) AS n, ROUND(AVG(monto), 2) AS promedio, SUM(monto) AS total FROM pedidos WHERE monto > 20000 GROUP BY municipio ORDER BY municipio"

res_sqlite = pd.read_sql_query(SQL, con)
con_duck = duckdb.connect()
con_duck.register("pedidos", pedidos)                                           # DuckDB lee el DataFrame sin copiarlo
res_duck = con_duck.execute(SQL).df()
g = pedidos[pedidos.monto > 20000].groupby("municipio").monto.agg(n="count", promedio="mean", total="sum").reset_index().sort_values("municipio")
res_pandas = g.assign(promedio=g.promedio.round(2)).reset_index(drop=True)

for otro in (res_duck, res_pandas):
    assert list(otro.municipio) == list(res_sqlite.municipio)
    assert np.array_equal(otro.n.values, res_sqlite.n.values)
    assert np.allclose(otro.promedio.values, res_sqlite.promedio.values) and np.allclose(otro.total.values, res_sqlite.total.values)
print("\nSQLite, DuckDB y pandas devuelven EXACTAMENTE el mismo resultado:")
print(res_sqlite.to_string(index=False))
print(f"\n{res_sqlite.n.sum():,} de {n_ped:,} pedidos superan los 20 000 (filtro WHERE) | SQL declarativo: el motor decide el plan, no tu.")
````

## 3. NoSQL: Cuatro Familias Construidas desde Cero 🧱

«NoSQL» (*not only SQL*) agrupa almacenes que **renuncian a parte del modelo relacional** (esquema rígido, `JOIN`s, a veces transacciones multi-fila) para ganar **escalabilidad horizontal**, **flexibilidad** o **rendimiento en un patrón de acceso concreto**. El libro (Cao, Cap. 4) los organiza en tres modelos principales y los complementa con el de documentos; aquí vemos las **cuatro familias**:

| Familia | Modelo | Operación estrella | Ejemplos de producto | Cuándo sí | Cuándo no |
|---|---|---|---|---|---|
| **Clave-valor** | `clave → valor` opaco | `get`/`set` por clave en $O(1)$, con expiración | Redis, DynamoDB, Memcached | Cachés, sesiones, contadores, carritos | Consultar por el **contenido** del valor |
| **Documentos** | Documentos JSON anidados, de esquema variable | Consultar por **campos**, con índices secundarios | MongoDB, CouchDB | Catálogos con atributos distintos por ítem, eventos | Muchas relaciones y transacciones entre documentos |
| **Familia de columnas** | Filas con **familias** de columnas dispersas, ordenadas por clave | Escaneo por **rango de clave**, escrituras masivas | HBase, Cassandra, Bigtable | Series de tiempo, mensajería, histórico a gran escala | Consultas *ad hoc* sobre cualquier columna |
| **Grafos** | Nodos y aristas (listas de adyacencia) | **Recorrer** vecindades y caminos | Neo4j, JanusGraph | Redes sociales, recomendaciones, rutas, fraude | Agregaciones masivas sobre tablas planas |

Las cuatro implementaciones siguientes son **pedagógicas** (en memoria, sin concurrencia ni persistencia), pero reproducen la **interfaz y la lógica** esenciales de cada familia, y cada una se **verifica con `assert`** contra una referencia independiente (un recorrido a fuerza bruta, *pandas* o `networkx`). Para el modelo de **familia de columnas** con versiones y regiones (HBase), el cuaderno 02 construye una versión más completa.

````python
# ============================================================
# 3a. Clave-valor con TTL y almacen de documentos con indice secundario
# ============================================================
class RelojManual:
    """Reloj controlable para probar expiraciones sin esperar (en produccion seria time.monotonic)."""
    def __init__(self): self.t = 0.0
    def __call__(self): return self.t
    def avanzar(self, s): self.t += s

class AlmacenClaveValor:
    """Clave -> valor opaco, con expiracion (TTL) 'perezosa': una clave vencida se elimina cuando alguien la consulta."""
    def __init__(self, reloj=time.monotonic):
        self._d, self._reloj = {}, reloj
    def set(self, clave, valor, ttl=None):
        self._d[clave] = (valor, None if ttl is None else self._reloj() + ttl)
    def _vigente(self, clave):
        if clave not in self._d: return False
        exp = self._d[clave][1]
        if exp is not None and self._reloj() >= exp:
            del self._d[clave]; return False                                   # expiracion perezosa
        return True
    def get(self, clave, defecto=None):
        return self._d[clave][0] if self._vigente(clave) else defecto
    def incr(self, clave, n=1):
        valor = self.get(clave, 0) + n
        self._d[clave] = (valor, self._d[clave][1] if clave in self._d else None)   # conserva la expiracion previa
        return valor
    def ttl(self, clave):
        return None if not self._vigente(clave) or self._d[clave][1] is None else self._d[clave][1] - self._reloj()
    def purgar(self):
        muertas = [k for k in list(self._d) if not self._vigente(k)]; return len(muertas)   # barrido activo
    def __len__(self): return len(self._d)

reloj = RelojManual()
kv = AlmacenClaveValor(reloj)
kv.set("sesion:ana", {"usuario": "ana"}, ttl=30)
kv.set("config:iva", 0.19)                                                      # sin expiracion
kv.set("visitas:home", 0)
for _ in range(5): kv.incr("visitas:home")
assert kv.get("sesion:ana") == {"usuario": "ana"} and kv.get("visitas:home") == 5
reloj.avanzar(20); assert kv.ttl("sesion:ana") == 10 and kv.get("sesion:ana") is not None
reloj.avanzar(11)
assert kv.get("sesion:ana") is None and kv.get("config:iva") == 0.19            # la sesion vencio; la configuracion no
kv.set("tmp:a", 1, ttl=1); kv.set("tmp:b", 2, ttl=1); reloj.avanzar(2)
assert len(kv) == 4 and kv.purgar() == 2 and len(kv) == 2                      # 'tmp:*' ocupaban sitio hasta que alguien barrio
print("Clave-valor OK: TTL, incr y purga | claves vivas:", len(kv))

class AlmacenDocumentos:
    """Documentos JSON de esquema libre. Filtros: igualdad, {'$gt','$gte','$lt','$lte','$in'}, campos anidados 'a.b' y listas."""
    def __init__(self):
        self._docs, self._sig, self._indices = {}, 0, {}
        self.examinados = 0                                                      # cuantos documentos reviso la ultima consulta
    @staticmethod
    def _campo(doc, ruta):
        for p in ruta.split("."):
            if not isinstance(doc, dict) or p not in doc: return None
            doc = doc[p]
        return doc
    def insertar(self, doc):
        self._sig += 1; doc = json.loads(json.dumps(doc)); doc["_id"] = self._sig; self._docs[self._sig] = doc
        for campo, ind in self._indices.items():
            v = self._campo(doc, campo)
            for x in (v if isinstance(v, list) else [v]):
                if x is not None: ind[x].add(self._sig)
        return self._sig
    def crear_indice(self, campo):
        self._indices[campo] = defaultdict(set)
        for d in self._docs.values():
            v = self._campo(d, campo)
            for x in (v if isinstance(v, list) else [v]):
                if x is not None: self._indices[campo][x].add(d["_id"])
    @staticmethod
    def _cumple(valor, cond):
        if isinstance(cond, dict):
            ops = {"$gt": lambda a, b: a > b, "$gte": lambda a, b: a >= b, "$lt": lambda a, b: a < b, "$lte": lambda a, b: a <= b, "$in": lambda a, b: a in b}
            return valor is not None and all(ops[o](valor, x) for o, x in cond.items())
        return valor == cond or (isinstance(valor, list) and cond in valor)
    def buscar(self, filtro):
        candidatos = None                                                        # si hay indice por igualdad, parte de ahi
        for campo, cond in filtro.items():
            if campo in self._indices and not isinstance(cond, dict):
                candidatos = set(self._indices[campo].get(cond, ())); break
        pool = self._docs.values() if candidatos is None else (self._docs[i] for i in sorted(candidatos))
        salida = []; self.examinados = 0
        for d in pool:
            self.examinados += 1
            if all(self._cumple(self._campo(d, c), v) for c, v in filtro.items()): salida.append(d)
        return salida

rng = np.random.default_rng(SEED)
N_DOC = 5_000
PRODUCTOS = ["papa", "cebolla", "arveja", "maíz", "trigo", "fresa"]
docs = []
for i in range(N_DOC):                                                           # catalogo con atributos DISTINTOS por producto (esquema variable)
    p = str(rng.choice(PRODUCTOS))
    d = {"producto": p, "finca": {"municipio": str(rng.choice(municipios)), "altitud_m": int(rng.integers(2400, 3300))},
         "precio_kg": int(rng.integers(800, 4000)), "etiquetas": [str(x) for x in rng.choice(["orgánico", "fresco", "empacado", "mayorista"], int(rng.integers(0, 3)), replace=False)]}
    if p == "papa": d["variedad"] = str(rng.choice(["pastusa", "criolla", "capiro"]))
    if p == "fresa": d["en_invernadero"] = bool(rng.integers(0, 2))
    docs.append(d)

dbdoc = AlmacenDocumentos()
for d in docs: dbdoc.insertar(d)
filtro = {"producto": "papa", "finca.municipio": "Tunja", "precio_kg": {"$gte": 1500, "$lt": 2500}}
sin_indice = dbdoc.buscar(filtro); n_sin = dbdoc.examinados
dbdoc.crear_indice("producto"); dbdoc.crear_indice("etiquetas")
con_indice = dbdoc.buscar(filtro); n_con = dbdoc.examinados
referencia = [i + 1 for i, d in enumerate(docs) if d["producto"] == "papa" and d["finca"]["municipio"] == "Tunja" and 1500 <= d["precio_kg"] < 2500]
assert [d["_id"] for d in sin_indice] == [d["_id"] for d in con_indice] == referencia and len(referencia) > 0
assert n_sin == N_DOC and n_con < n_sin / 4                                      # el indice evita revisar todo el almacen
org = dbdoc.buscar({"etiquetas": "orgánico", "producto": "fresa", "en_invernadero": True})
assert all("orgánico" in d["etiquetas"] and d["en_invernadero"] for d in org) and len(org) > 0
con_variedad = sum(1 for d in docs if "variedad" in d)                           # esquema variable: solo la papa tiene 'variedad'
print(f"Documentos OK: {len(con_indice)} resultados | revisados sin indice = {n_sin:,}, con indice = {n_con:,} | campos opcionales: {con_variedad:,} de {N_DOC:,} docs tienen 'variedad'")
````

````python
# ============================================================
# 3b. Familia de columnas (tabla ancha y dispersa) y grafo con lista de adyacencia
# ============================================================
class TablaAncha:
    """Estilo Bigtable/Cassandra simplificado: fila -> {familia: {calificador: valor}}. Las FAMILIAS se declaran al crear
    la tabla; los CALIFICADORES (columnas) se crean sobre la marcha, y las filas son DISPERSAS (no guardan 'NULL')."""
    def __init__(self, familias):
        self.familias, self._filas = set(familias), {}
    def poner(self, fila, familia, calificador, valor):
        if familia not in self.familias: raise KeyError(f"familia desconocida: {familia}")
        self._filas.setdefault(fila, {}).setdefault(familia, {})[calificador] = valor
    def obtener(self, fila, familia=None):
        f = self._filas.get(fila, {})
        return {k: dict(v) for k, v in f.items()} if familia is None else dict(f.get(familia, {}))
    def escanear(self, prefijo=""):
        for fila in sorted(k for k in self._filas if k.startswith(prefijo)):    # las filas se leen ORDENADAS por clave
            yield fila, self.obtener(fila)
    def celdas(self):
        return sum(len(q) for f in self._filas.values() for q in f.values())

lecturas = TablaAncha(familias=["meta", "medidas"])
rng = np.random.default_rng(SEED)
estaciones = ["E01", "E02", "E03"]
for est in estaciones:                                                           # clave de fila = estacion#dia
    for dia in range(1, 8):
        fila = f"{est}#2026-03-{dia:02d}"
        lecturas.poner(fila, "meta", "municipio", {"E01": "Tunja", "E02": "Duitama", "E03": "Paipa"}[est])
        lecturas.poner(fila, "medidas", "temp_max", round(float(rng.normal(19, 2)), 1))
        if est == "E02": lecturas.poner(fila, "medidas", "humedad", int(rng.integers(55, 90)))    # solo E02 mide humedad
        if est == "E03" and dia % 2 == 0: lecturas.poner(fila, "medidas", "lluvia_mm", round(float(rng.gamma(2, 3)), 1))   # E03 reporta lluvia en dias pares
filas = [f for f, _ in lecturas.escanear("E02#")]
assert len(filas) == 7 and filas == sorted(filas) and all(f.startswith("E02#") for f in filas)
assert lecturas.obtener("E01#2026-03-01", "medidas").keys() == {"temp_max"} and "humedad" in lecturas.obtener("E02#2026-03-03", "medidas")
try:
    lecturas.poner("E01#2026-03-09", "extra", "x", 1); raise AssertionError("debio rechazar la familia")
except KeyError:
    pass
celdas_reales = lecturas.celdas(); columnas_distintas = {q for _, r in lecturas.escanear() for f in r.values() for q in f}
celdas_relacional = 21 * len(columnas_distintas)                                 # una tabla plana tendria TODAS las columnas en TODAS las filas
assert celdas_reales < celdas_relacional
print(f"Tabla ancha OK: {celdas_reales} celdas guardadas frente a {celdas_relacional} en una tabla plana con NULL ({len(columnas_distintas)} columnas distintas, 21 filas) | escaneo por prefijo 'E02#': {len(filas)} filas ordenadas")

class Grafo:
    """Grafo no dirigido con lista de adyacencia: nodo -> conjunto de vecinos. Recorrer vecindades es barato (no hay JOIN)."""
    def __init__(self): self.ady = defaultdict(set)
    def arista(self, u, v): self.ady[u].add(v); self.ady[v].add(u)
    def vecinos(self, u): return self.ady[u]
    def camino_corto(self, origen, destino):
        previo, cola = {origen: None}, [origen]                                  # BFS por niveles
        for u in cola:
            if u == destino: break
            for v in sorted(self.ady[u]):
                if v not in previo: previo[v] = u; cola.append(v)
        if destino not in previo: return None
        ruta = [destino]
        while previo[ruta[-1]] is not None: ruta.append(previo[ruta[-1]])
        return ruta[::-1]
    def amigos_de_amigos(self, u):
        cuenta = Counter(w for v in self.ady[u] for w in self.ady[v] if w != u and w not in self.ady[u])
        return cuenta                                                            # candidato -> amigos en comun

N_PERSONAS = 60
rng = np.random.default_rng(SEED)
grafo = Grafo()
pares = {tuple(sorted(map(int, rng.choice(N_PERSONAS, 2, replace=False)))) for _ in range(150)}
for a, b in pares: grafo.arista(a, b)
G_nx = nx.Graph(list(pares))
for u, v in [(0, 1), (3, 40), (7, 55)]:                                          # camino corto: la misma longitud que networkx
    if u in G_nx and v in G_nx and nx.has_path(G_nx, u, v):
        assert len(grafo.camino_corto(u, v)) - 1 == nx.shortest_path_length(G_nx, u, v)
u_prueba = max(grafo.ady, key=lambda x: len(grafo.ady[x]))                       # el mas conectado
fof = grafo.amigos_de_amigos(u_prueba)
con_sql = sqlite3.connect(":memory:")                                            # lo mismo en SQL: un autoJOIN de la tabla de aristas
con_sql.execute("CREATE TABLE amistad (a INT, b INT)")
con_sql.executemany("INSERT INTO amistad VALUES (?, ?)", [(a, b) for a, b in pares] + [(b, a) for a, b in pares])
filas_sql = con_sql.execute("""SELECT f2.b AS candidato, COUNT(*) AS en_comun FROM amistad f1 JOIN amistad f2 ON f1.b = f2.a
                               WHERE f1.a = ? AND f2.b <> ? AND f2.b NOT IN (SELECT b FROM amistad WHERE a = ?) GROUP BY f2.b""", (u_prueba, u_prueba, u_prueba)).fetchall()
assert dict(filas_sql) == dict(fof) and len(fof) > 0
print(f"Grafo OK: {len(pares)} aristas, {N_PERSONAS} personas | persona {u_prueba} tiene {len(grafo.vecinos(u_prueba))} amigos y {len(fof)} candidatos de 'amigo de amigo' (igual que el autoJOIN en SQL)")
print("A dos saltos el SQL ya necesita un autoJOIN; a cinco saltos, cinco: en un grafo el recorrido es la operacion natural.")
````

## 4. ACID frente a BASE y el Teorema CAP 🌐✂️

Al repartir los datos en **varias máquinas** (réplicas) aparece una pregunta incómoda: **¿qué pasa cuando la red entre ellas se parte?** El **teorema CAP** (conjetura de Brewer, demostrada por Gilbert y Lynch) dice que un almacén distribuido **no puede garantizar a la vez** tres propiedades cuando hay una **partición** de red:

- **C — Consistencia (linealizable):** toda lectura ve la **última** escritura confirmada.
- **A — Disponibilidad:** toda petición a un nodo que funciona recibe **respuesta** (no un error).
- **P — Tolerancia a particiones:** el sistema sigue funcionando aunque la red pierda mensajes entre nodos.

**Las particiones ocurren** (cables, conmutadores, saturación), así que en la práctica **la elección es entre C y A *durante la partición***: un sistema **CP** prefiere **rechazar** peticiones antes que dar un dato desactualizado; uno **AP** prefiere **responder siempre** y reconciliar después. (Sin partición se pueden tener ambas; el trade-off cotidiano pasa a ser **latencia frente a consistencia**, lo que formaliza PACELC.)

| | **ACID** (relacional clásico) | **BASE** (muchos NoSQL) |
|---|---|---|
| Lema | Atomicidad, Consistencia, Aislamiento, Durabilidad | ***B**asically **A**vailable, **S**oft state, **E**ventual consistency* |
| Prioridad | **Corrección** de cada transacción | **Disponibilidad** y escala horizontal |
| Tras una escritura | Todos ven el nuevo valor | Las réplicas **convergen con el tiempo** |
| Costo | Coordinación; menos escalable horizontalmente | Lecturas **obsoletas** y **conflictos** que la aplicación debe tolerar |

La celda siguiente **simula** tres réplicas de un mismo dato (`stock`, unidades de un producto) y una **partición** que aísla a **una** de ellas durante 100 de 300 operaciones. Dos políticas: **CP** (lecturas y escrituras exigen **quórum**: mayoría de réplicas alcanzables) y **AP** (cada réplica atiende con lo que tiene y se reconcilia al sanar, con *última escritura gana*). Medimos tres cosas: peticiones **rechazadas**, lecturas **obsoletas** y escrituras **confirmadas pero perdidas**. (Los números dependen de la semilla; lo que se verifica son las **garantías** de cada política.)

````python
# ============================================================
# 4. Simulacion CAP: 3 replicas, una se aisla; politicas CP (quorum) y AP (siempre responde)
# ============================================================
class NoDisponible(Exception):
    pass

class ClusterReplicado:
    """n replicas de UNA clave: cada una guarda (valor, version). 'grupos' describe quien se ve con quien durante la particion."""
    def __init__(self, n=3, modo="CP"):
        self.n, self.modo, self.reloj_global = n, modo, 0
        self.rep = [(0, 0) for _ in range(n)]                                    # (valor, version) de cada replica
        self.grupos = [set(range(n))]                                            # sin particion: todas se ven
        self.ultimo_confirmado = (0, 0)                                          # (valor, version) de la ultima escritura confirmada a un cliente
    def lado(self, r):
        return next(g for g in self.grupos if r in g)
    def particionar(self, grupos): self.grupos = [set(g) for g in grupos]
    def quorum(self): return self.n // 2 + 1
    def escribir(self, coord, valor):
        lado = self.lado(coord)
        if self.modo == "CP" and len(lado) < self.quorum(): raise NoDisponible("sin quorum")
        self.reloj_global += 1
        for r in lado: self.rep[r] = (valor, self.reloj_global)                  # solo las replicas alcanzables se enteran
        self.ultimo_confirmado = (valor, self.reloj_global)
        return self.reloj_global
    def leer(self, coord):
        lado = self.lado(coord)
        if self.modo == "CP":
            if len(lado) < self.quorum(): raise NoDisponible("sin quorum")
            return max(self.rep[r] for r in lado)                                # lectura de quorum: la version mas nueva
        return self.rep[coord]                                                   # AP: lo que tenga la replica local
    def sanar(self):
        self.grupos = [set(range(self.n))]
        mejor = max(self.rep, key=lambda t: t[1])                                # reconciliacion: ultima escritura gana (version mayor)
        self.rep = [mejor] * self.n
        return mejor

def simular(modo, grupos_particion, ventana=(100, 200), n_ops=300, p_escritura=0.3, semilla=1):
    rng = np.random.default_rng(semilla)
    cl = ClusterReplicado(3, modo)
    rechazadas = obsoletas = servidas = perdidas = 0
    en_ventana = []                                                              # (version, lado) de las escrituras confirmadas durante la particion
    for i in range(n_ops):
        if i == ventana[0]:
            cl.particionar(grupos_particion)
        if i == ventana[1]:
            ganadora = cl.sanar()                                                # reconciliacion: version mayor gana
            lado_ganador = next((lado for v, lado in en_ventana if v == ganadora[1]), None)
            perdidas = sum(1 for v, lado in en_ventana if lado != lado_ganador)  # confirmadas en OTRA rama: se descartan en silencio
        coord = int(rng.integers(0, 3))
        try:
            if rng.random() < p_escritura:
                v = cl.escribir(coord, int(rng.integers(1, 10_000)))
                if ventana[0] <= i < ventana[1]: en_ventana.append((v, frozenset(cl.lado(coord))))
            else:
                _, version = cl.leer(coord)
                if version < cl.ultimo_confirmado[1]: obsoletas += 1             # leyo algo anterior a la ultima escritura confirmada
            servidas += 1
        except NoDisponible:
            rechazadas += 1
    cl.sanar()
    assert len({r for r in cl.rep}) == 1                                         # tras sanar, todas las replicas son identicas
    return {"modo": modo, "servidas": servidas, "rechazadas": rechazadas, "lecturas_obsoletas": obsoletas, "escrituras_perdidas": perdidas}

# Quorum: con n=3, R=W=2 todo quorum de lectura corta a todo quorum de escritura (R + W > N): se verifica por fuerza bruta
n_rep = 3
assert all(set(q_l) & set(q_e) for q_l in itertools.combinations(range(n_rep), 2) for q_e in itertools.combinations(range(n_rep), 2))
assert any(not (set(q_l) & set(q_e)) for q_l in itertools.combinations(range(n_rep), 1) for q_e in itertools.combinations(range(n_rep), 2))   # con R=1, W=2 NO siempre se cortan

particion = [{0, 1}, {2}]                                                       # la replica 2 queda aislada; {0, 1} es la mayoria
res = [simular(m, particion) for m in ("CP", "AP")]
tabla = pd.DataFrame(res).set_index("modo")
print(tabla.to_string())

cp, ap = tabla.loc["CP"], tabla.loc["AP"]
assert cp.lecturas_obsoletas == 0 and cp.escrituras_perdidas == 0 and cp.rechazadas > 0      # CP: nunca obsoleto, pero rechaza
assert ap.rechazadas == 0 and ap.lecturas_obsoletas > 0                                      # AP: nunca rechaza, pero lee dato viejo
assert (cp.servidas + cp.rechazadas) == (ap.servidas + ap.rechazadas) == 300
print(f"\nCP rechazó {cp.rechazadas} de 300 peticiones (las que cayeron en la réplica aislada) y no sirvió ningún dato viejo.")
print(f"AP respondió las 300, pero {ap.lecturas_obsoletas} lecturas fueron obsoletas y {ap.escrituras_perdidas} escrituras confirmadas se perdieron al reconciliar (última escritura gana).")

fig, ax = plt.subplots(1, 2, figsize=(11, 3.8))
x = np.arange(3); w = 0.38
for k, (m, color) in enumerate((("CP", PALETA[0]), ("AP", PALETA[3]))):
    ax[0].bar(x + (k - 0.5) * w, [tabla.loc[m, c] for c in ("rechazadas", "lecturas_obsoletas", "escrituras_perdidas")], w, label=m, color=color)
ax[0].set_xticks(x); ax[0].set_xticklabels(["peticiones\nrechazadas", "lecturas\nobsoletas", "escrituras\nperdidas"])
ax[0].set_title("Qué se sacrifica durante la partición", fontsize=10, fontweight="bold"); ax[0].legend()
ax[1].axis("off"); ax[1].grid(False); ax[1].set_xlim(0, 10); ax[1].set_ylim(0, 6)
for (cx, cy, txt, col) in [(1.4, 4.4, "réplica 0", PALETA[0]), (4.6, 4.4, "réplica 1", PALETA[0]), (8.2, 4.4, "réplica 2\n(aislada)", PALETA[1])]:
    ax[1].add_patch(FancyBboxPatch((cx - 1.0, cy - 0.8), 2.0, 1.6, boxstyle="round,pad=0.05", fc=col, ec="white", lw=2))
    ax[1].text(cx, cy, txt, ha="center", va="center", color="white", fontweight="bold", fontsize=9)
ax[1].annotate("", xy=(3.55, 4.4), xytext=(2.45, 4.4), arrowprops=dict(arrowstyle="<->", lw=2, color="#334155"))
ax[1].plot([6.0, 6.0], [2.9, 5.8], color=PALETA[1], lw=3, ls="--"); ax[1].text(6.0, 2.5, "partición de red", ha="center", color=PALETA[1], fontsize=9, fontweight="bold")
ax[1].text(3.0, 1.5, "mayoría {0,1}: tiene quórum", ha="center", fontsize=9); ax[1].text(8.2, 1.5, "minoría {2}: sin quórum", ha="center", fontsize=9)
ax[1].set_title("Escenario simulado (operaciones 100 a 200)", fontsize=10, fontweight="bold")
plt.tight_layout(); plt.show()
````

## 5. Panorama de Hadoop y Spark 🐘⚡

**Hadoop** nació para guardar y procesar archivos enormes en **clústeres de máquinas baratas**. Es un **ecosistema** de capas, no un solo producto; **Spark** reemplaza principalmente la capa de **cómputo** (y se apoya en el resto):

```
        ┌────────────────────────────────────────────────────────────────────────────┐
 Acceso │  Hive (SQL)   Pig (scripts)   Spark SQL / DataFrames   MLlib   Streaming   │
        ├────────────────────────────────────────────────────────────────────────────┤
 Cómputo│  MapReduce (lotes, escribe a disco)   |   Spark (en memoria, DAG de tareas) │
        ├────────────────────────────────────────────────────────────────────────────┤
 Recursos│                 YARN (asigna CPU y memoria a las aplicaciones)             │
        ├────────────────────────────────────────────────────────────────────────────┤
 Almacén│   HDFS (archivos en bloques replicados)        HBase (tablas por clave)     │
        └────────────────────────────────────────────────────────────────────────────┘
```

| Pieza | Papel | Idea clave | Se desarrolla en |
|---|---|---|---|
| **HDFS** | Sistema de archivos distribuido | Archivos en **bloques** replicados en varios nodos | [Cuaderno 02](02_Frameworks_Distribuidos_HDFS_Spark_HBase.ipynb) |
| **MapReduce** | Modelo de cómputo por lotes | `map` → barajado (*shuffle*) → `reduce`, con E/S a disco entre fases | [Cuaderno 01](01_MapReduce_y_sus_Limitaciones.ipynb) |
| **YARN** | Gestor de recursos | Reparte CPU y memoria del clúster entre aplicaciones | Conceptual |
| **Hive / Pig** | SQL y *scripts* sobre datos en HDFS | Traducen consultas a trabajos del motor de cómputo | Módulo 02 |
| **Spark** | Motor de cómputo general | **DAG** perezoso, datos en memoria, API de DataFrames, *streaming* y ML | [Cuaderno 02](02_Frameworks_Distribuidos_HDFS_Spark_HBase.ipynb) y [Data Mining 07](../../Data%20Mining/07%20-%20Mineria%20de%20Datos%20con%20Big%20Data/01_Arquitectura_Apache_Spark_y_PySpark.ipynb) |
| **HBase** | Almacén por familias de columnas sobre HDFS | Filas ordenadas por clave, **versiones** y **regiones** | [Cuaderno 02](02_Frameworks_Distribuidos_HDFS_Spark_HBase.ipynb) |

## 6. Un Mismo Cálculo en Cuatro Motores 📏

> 🧭 **Una advertencia honesta antes de medir:** Hadoop y Spark están hechos para datos que **no caben en una máquina** o para trabajos que exigen **tolerancia a fallos en un clúster**. Con datos que **sí caben** en la memoria de un portátil, un motor de una sola máquina suele ganar, porque no paga la **coordinación** (arranque de la JVM, serialización entre Python y Java, planificación de tareas). La celda siguiente mide un mismo cálculo —**conteo, promedio y suma por municipio de los pedidos con monto mayor a 50 000**— en cuatro motores y **verifica que todos dan el mismo resultado**. Los tiempos dependen de la máquina y de la carga: **solo se discuten las conclusiones cualitativas**. Spark corre en modo local con 2 hilos (`local[2]`), sin interfaz web, y se **detiene** al terminar.

````python
# ============================================================
# 6. Mismo calculo en pandas, DuckDB, SQLite y Spark local (con tiempos medidos)
# ============================================================
os.makedirs(TMP, exist_ok=True)
os.environ["PYSPARK_PYTHON"] = sys.executable                                    # los trabajadores de Spark usan este mismo Python
from pyspark.sql import SparkSession, functions as F

n = 400_000 if RAPIDO else 3_000_000
rng = np.random.default_rng(SEED)
dfb = pd.DataFrame({
    "pedido": np.arange(n),
    "municipio": rng.choice([f"M{i:02d}" for i in range(30)], n),
    "categoria": rng.choice(list("ABCDEFGH"), n),
    "monto": rng.lognormal(mean=10.5, sigma=0.7, size=n).round(0),
})
ruta_pq = f"{TMP}/pedidos.parquet"
dfb.to_parquet(ruta_pq)
UMBRAL = 50_000
print(f"Datos: {n:,} filas | parquet = {os.path.getsize(ruta_pq) / 1e6:.1f} MB | filas con monto > {UMBRAL:,}: {(dfb.monto > UMBRAL).sum():,}")

def mediana_s(fn, reps=3):
    ts = []
    for _ in range(reps):
        t0 = time.perf_counter(); salida = fn(); ts.append(time.perf_counter() - t0)
    return float(np.median(ts)), salida

def norm(df):
    df = df.rename(columns=str.lower).sort_values("municipio").reset_index(drop=True)
    return df[["municipio", "n", "promedio", "total"]].astype({"n": "int64"})

resultados, tiempos = {}, {}

# pandas
tiempos["pandas"], r = mediana_s(lambda: dfb[dfb.monto > UMBRAL].groupby("municipio").monto.agg(n="count", promedio="mean", total="sum").reset_index())
resultados["pandas"] = norm(r)

# DuckDB (columnar, vectorizado) sobre el mismo DataFrame y sobre el archivo Parquet
cd = duckdb.connect(); cd.register("pedidos", dfb)
SQL_D = f"SELECT municipio, COUNT(*) AS n, AVG(monto) AS promedio, SUM(monto) AS total FROM pedidos WHERE monto > {UMBRAL} GROUP BY municipio"
tiempos["DuckDB"], r = mediana_s(lambda: cd.execute(SQL_D).df())
resultados["DuckDB"] = norm(r)

# SQLite (filas): hay que CARGAR primero; se reporta aparte
cs = sqlite3.connect(":memory:")
t0 = time.perf_counter(); dfb.to_sql("pedidos", cs, index=False); carga_sqlite = time.perf_counter() - t0
SQL_S = f"SELECT municipio, COUNT(*) AS n, AVG(monto) AS promedio, SUM(monto) AS total FROM pedidos WHERE monto > {UMBRAL} GROUP BY municipio"
tiempos["SQLite"], r = mediana_s(lambda: pd.read_sql_query(SQL_S, cs))
resultados["SQLite"] = norm(r)

# Spark local: sesion, lectura del Parquet, consulta y recoleccion al driver
t0 = time.perf_counter()
spark = (SparkSession.builder.master("local[2]").appName("bd10").config("spark.ui.enabled", "false")
         .config("spark.sql.shuffle.partitions", "4").config("spark.default.parallelism", "4")
         .config("spark.sql.warehouse.dir", f"{TMP}/warehouse").getOrCreate())
spark.sparkContext.setLogLevel("ERROR")
arranque_spark = time.perf_counter() - t0
sdf = spark.read.parquet(ruta_pq)
def consulta_spark():
    return (sdf.filter(F.col("monto") > UMBRAL).groupBy("municipio")
               .agg(F.count("*").alias("n"), F.avg("monto").alias("promedio"), F.sum("monto").alias("total")).toPandas())
t0 = time.perf_counter(); r = consulta_spark(); primera_spark = time.perf_counter() - t0
tiempos["Spark local"], r = mediana_s(consulta_spark)
resultados["Spark local"] = norm(r)
spark.stop()                                                                     # se detiene la sesion y se liberan los procesos

ref = resultados["pandas"]
for motor, res in resultados.items():
    assert list(res.municipio) == list(ref.municipio) and np.array_equal(res.n.values, ref.n.values), motor
    assert np.allclose(res.promedio.values, ref.promedio.values) and np.allclose(res.total.values, ref.total.values), motor
print("\nLos cuatro motores devuelven el MISMO resultado (verificado con assert).")
print(f"Arranque de la sesion Spark: {arranque_spark:.1f} s | primera consulta Spark: {primera_spark:.2f} s | carga a SQLite: {carga_sqlite:.2f} s")

res_t = pd.Series(tiempos).sort_values()
print("\nTiempo mediano de la consulta (s, ya con los datos cargados):")
print(res_t.round(4).to_string())
fig, ax = plt.subplots(figsize=(8.5, 3.4))
orden = res_t.index.tolist()
ax.barh(orden, res_t.values, color=[PALETA[i] for i in range(len(orden))])
ax.set_xscale("log"); ax.invert_yaxis(); ax.set_xlabel("segundos (escala logarítmica)")
ax.set_title(f"Mismo conteo/promedio/suma por municipio, {n:,} filas", fontsize=10, fontweight="bold")
plt.tight_layout(); plt.show()
print("\nLectura cualitativa: con datos que caben en memoria, los motores columnares de una sola máquina (DuckDB, pandas) no pagan arranque ni coordinación;")
print("SQLite recorre filas y además exige cargar los datos; Spark local paga la JVM y la serialización y no gana aquí. Su ventaja aparece con datos que NO caben en una máquina.")
shutil.rmtree(TMP, ignore_errors=True)
````

## 7. Matriz de Decisión: ¿Qué Opción Uso según el Caso? 🧭

No existe «la mejor» opción; existe la que **mejor cumple los requisitos que importan en tu caso**. Antes de elegir producto, responde cinco preguntas: **(1)** ¿cuánto dato y cuánto crece?, **(2)** ¿qué **patrón de acceso** domina (por clave, por rango, por contenido, recorridos, agregaciones)?, **(3)** ¿necesito **transacciones** y consistencia fuerte o tolero consistencia eventual?, **(4)** ¿qué **latencia** se espera (milisegundos o minutos)?, **(5)** ¿el **esquema** es estable o variable?

| Caso de uso | Opción recomendada | Por qué | Cuidado con |
|---|---|---|---|
| Pagos, inventario, reservas (dinero que **no** puede perderse) | **Relacional** (SQLite/PostgreSQL) | Transacciones **ACID**, restricciones, `JOIN`s | Escala vertical; réplicas con consistencia eventual rompen el invariante |
| Sesiones de usuario, cachés, contadores en vivo | **Clave-valor** (Redis) | Acceso por clave en $O(1)$, **TTL** nativo | No se consulta por contenido; datos volátiles |
| Catálogo de productos con atributos distintos por ítem, perfiles | **Documentos** (MongoDB) | Esquema flexible, consultas por campo, un documento = una unidad de lectura | `JOIN`s y transacciones entre documentos son limitados |
| Series de tiempo, mensajería, histórico de miles de millones de filas por clave | **Familia de columnas** (HBase/Cassandra) | Escrituras masivas y escaneo por **rango de clave** | Solo consultas diseñadas de antemano; mal diseño de clave = punto caliente |
| Redes sociales, recomendación, detección de fraude por relaciones | **Grafos** (Neo4j) | Recorridos de varios saltos sin `JOIN`s en cascada | Agregaciones masivas sobre tablas planas |
| Análisis exploratorio de unos cuantos GB en un portátil | ***pandas*** o **DuckDB** (SQL columnar) | Sin coordinación; rapidísimo si cabe en memoria | Memoria de una sola máquina |
| Informe nocturno sobre cientos de GB o TB en un clúster | **Spark** (sobre HDFS u objeto en la nube) | Paralelismo, tolerancia a fallos, SQL/DataFrames | Costo operativo y de arranque; mala idea para volúmenes pequeños |
| Trabajo por lotes heredado ya escrito en Hadoop | **MapReduce** sobre HDFS | Estable y probado | Escribe a disco entre fases: lento en algoritmos iterativos (cuaderno 01) |
| Alertas en segundos sobre un flujo de eventos | **Procesamiento de flujo** (Spark Structured Streaming, Flink, Kafka Streams) | Estado y ventanas por tiempo de evento | Duplicados, orden y complejidad operativa (Módulo 05) |

> ⚖️ **Mezclar es normal (persistencia políglota):** una tienda en línea puede guardar pedidos en una base **relacional**, sesiones en **clave-valor**, el catálogo en **documentos**, las recomendaciones en un **grafo** y enviar todo a un **lago de datos** donde Spark prepara los informes (Módulo 02). Cada herramienta hace lo que mejor sabe; el costo es **operar varias** y mantener sus datos **coherentes**.

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Lotes frente a flujo** | Un lote cerrado cada $W$ s hace esperar a cada evento **$W/2$ s en promedio** (medido); el flujo lo lleva a ≈ 0. Ambos dan el **mismo total final**: cambia **cuándo** está disponible la respuesta |
| **Relacional** | Esquema, `JOIN`s y **ACID**: una transferencia que viola un `CHECK` se **revierte completa** y el total se conserva. **SQLite** (filas, OLTP) y **DuckDB** (columnas, OLAP) dieron **el mismo resultado** que *pandas* |
| **Cuatro familias NoSQL** | **Clave-valor** (get/set, TTL), **documentos** (consultas por campo; un índice redujo los documentos revisados de 5 000 a una fracción), **familia de columnas** (tabla dispersa ordenada por clave: menos celdas que una tabla plana con `NULL`) y **grafo** (el «amigo de amigo» es un recorrido natural; en SQL es un autoJOIN) |
| **ACID frente a BASE** | ACID prioriza la **corrección**; BASE, la **disponibilidad** y la escala, aceptando **consistencia eventual** |
| **CAP** | Con una partición hay que elegir: **CP** rechaza peticiones (cero lecturas obsoletas); **AP** responde siempre pero lee dato viejo y **pierde escrituras** al reconciliar con *última escritura gana*. Con R + W > N los quórums siempre se cortan |
| **Hadoop y Spark** | Hadoop = **HDFS + YARN + MapReduce** (y Hive, HBase, ...); Spark reemplaza el cómputo con un DAG en memoria. Con datos que **caben en memoria**, DuckDB y *pandas* **ganaron** a Spark local; Spark paga arranque y serialización |
| **Cómo elegir** | Por **patrón de acceso**, **consistencia**, **latencia**, **volumen** y **esquema**; en la práctica se combinan varios (**persistencia políglota**) |
