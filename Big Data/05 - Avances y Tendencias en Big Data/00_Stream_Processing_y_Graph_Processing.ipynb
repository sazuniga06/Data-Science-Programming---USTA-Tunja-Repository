## 1. Tiempo de Evento, Tiempo de Procesamiento y Ventanas ⏱️

En un flujo cada dato lleva **dos relojes**:

| Reloj | Qué es | Quién lo fija |
|---|---|---|
| **Tiempo de evento** $t_e$ | Cuándo **ocurrió** la cosa (la lectura del sensor, el clic, la venta) | El **productor** del dato |
| **Tiempo de procesamiento** $t_p$ | Cuándo el sistema **lo recibe y lo procesa** | El **reloj del servidor** |

Siempre $t_p \ge t_e$ y la diferencia $t_p - t_e$ (el **retraso**) **no es constante**: redes lentas, teléfonos sin señal, reintentos. Por eso los datos llegan **desordenados** respecto a $t_e$. Si agregas por $t_p$, el resultado cambia según el día de la red; si agregas por $t_e$, el resultado es el **correcto**, pero ahora debes decidir **cuánto esperar** a los rezagados (sección 2).

Para agregar un flujo infinito se agrupa por **ventanas** de tiempo (en tiempo de evento):

| Tipo | Definición | Un evento pertenece a... |
|---|---|---|
| ***Tumbling*** (saltos, sin solape) | Intervalos $[k\,T,\ (k+1)\,T)$ | **una** ventana |
| ***Sliding*** (deslizante, con solape) | Ventanas de tamaño $T$ que avanzan cada $S<T$: $[k\,S,\ k\,S + T)$ | $T/S$ ventanas |
| **De sesión** | Se abre con un evento y se **extiende** mientras el hueco entre eventos sea menor que una **brecha** $g$; un silencio $\ge g$ la cierra | **una** sesión (por clave) |

La celda siguiente **genera un flujo sintético** (10 minutos de lecturas de 6 estaciones; el 6 % de los eventos llega con un retraso grande, como un teléfono que recupera la señal), **dibuja** la relación $t_e$ frente a $t_p$, **implementa las tres ventanas** desde cero, las **verifica** con ejemplos hechos a mano y contra `pandas`, y las dibuja. Observa que **el orden de llegada no es el orden de los eventos**.

````python
# ============================================================
# Configuracion + flujo sintetico + tiempo de evento vs. de procesamiento + ventanas desde cero
# ============================================================
# !pip install -q numpy pandas matplotlib networkx pyspark

import hashlib
import math
import os
import shutil
import tempfile
import time
import warnings
from collections import Counter, defaultdict

import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
import pandas as pd
from matplotlib.patches import FancyBboxPatch

warnings.filterwarnings("ignore")
SEED = 42
RAPIDO = True                                                                  # True: tamanos reducidos (el cuaderno completo corre en ~1 min)
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
TMP = tempfile.mkdtemp(prefix="bd50_")
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | networkx {nx.__version__} | RAPIDO={RAPIDO}")

# ----------------------------------------------------------------------------
# Flujo sintetico: lecturas de 6 estaciones durante 10 minutos (600 s)
# ----------------------------------------------------------------------------
rng = np.random.default_rng(SEED)
N = 3_000 if RAPIDO else 12_000
DURACION = 600.0
t_evento = np.sort(rng.uniform(0, DURACION, N)).round(3)                        # cuando OCURRIO (s desde el inicio)
estacion = rng.choice([f"E{i}" for i in range(1, 7)], N)
valor = rng.gamma(2.0, 5.0, N).round(2)
tardio = rng.random(N) < 0.06                                                   # 6 %: llega con mucho retraso (sin senal)
retraso = np.where(tardio, rng.uniform(20, 90, N), rng.exponential(2.0, N))    # s de retraso hasta que el sistema lo ve
t_proc = t_evento + retraso                                                     # cuando LLEGO al sistema
eventos = (pd.DataFrame({"t_evento": t_evento, "t_proc": t_proc, "estacion": estacion, "valor": valor, "tardio": tardio})
           .sort_values("t_proc", kind="stable").reset_index(drop=True))        # el sistema los ve en orden de LLEGADA

fuera_de_orden = int((eventos.t_evento.diff() < 0).sum())
print(f"{N:,} eventos | retraso medio {retraso.mean():.1f} s | retraso maximo {retraso.max():.1f} s | "
      f"{fuera_de_orden:,} eventos llegan antes que otro que ocurrio ANTES ({fuera_de_orden / N:.0%} del flujo desordenado)")
assert (eventos.t_proc >= eventos.t_evento).all() and fuera_de_orden > 0
assert eventos.t_proc.is_monotonic_increasing and not eventos.t_evento.is_monotonic_increasing

# ----------------------------------------------------------------------------
# Ventanas desde cero (tiempo de evento)
# ----------------------------------------------------------------------------
def ventana_tumbling(x, tam):
    """Inicio de la unica ventana [k*tam, (k+1)*tam) que contiene a x."""
    return math.floor(x / tam) * tam

def ventanas_sliding(x, tam, paso):
    """Inicios de TODAS las ventanas [k*paso, k*paso+tam) que contienen a x (son tam/paso)."""
    k_max = math.floor(x / paso)
    k_min = math.floor((x - tam) / paso) + 1                                    # k*paso > x - tam  <=>  x < k*paso + tam
    return [k * paso for k in range(k_min, k_max + 1)]

def sesiones(tiempos, brecha):
    """Sesiones de UNA clave: un hueco >= brecha cierra la sesion. Devuelve [(inicio, ultimo_evento, n_eventos)]."""
    sal = []
    for x in sorted(tiempos):
        if sal and x - sal[-1][1] < brecha:
            sal[-1][1] = x; sal[-1][2] += 1
        else:
            sal.append([x, x, 1])
    return [tuple(s) for s in sal]

# --- Verificacion con un ejemplo hecho a mano ---
ej = [1, 2, 7, 8, 9, 15]
assert Counter(ventana_tumbling(x, 5) for x in ej) == Counter({0: 2, 5: 3, 15: 1})
memb = [ventanas_sliding(x, 10, 5) for x in ej]
assert all(len(m) == 2 for m in memb)                                           # cada evento esta en tam/paso = 2 ventanas
cs = Counter(s for m in memb for s in m)
assert cs[0] == 5 and cs[5] == 3 and cs[-5] == 2 and cs[10] == 1 and cs[15] == 1 and sum(cs.values()) == 2 * len(ej)
assert sesiones(ej, 3) == [(1, 2, 2), (7, 9, 3), (15, 15, 1)]
assert sesiones(ej, 6) == [(1, 9, 5), (15, 15, 1)] and sesiones(ej, 1) == [(i, i, 1) for i in ej]

# --- Verificacion a gran escala contra pandas ---
tum = Counter(ventana_tumbling(x, 60) for x in eventos.t_evento)
ref = (eventos.t_evento // 60 * 60).value_counts().to_dict()
assert dict(tum) == ref and sum(tum.values()) == N
sli = Counter(s for x in eventos.t_evento for s in ventanas_sliding(x, 60, 20))
assert sum(sli.values()) == 3 * N                                               # 60 / 20 = 3 ventanas por evento
ses = sesiones(eventos.loc[eventos.estacion == "E1", "t_evento"], 3.0)
assert sum(s[2] for s in ses) == (eventos.estacion == "E1").sum()
print(f"Tumbling 60 s: {len(tum)} ventanas (coincide con pandas) | Sliding 60 s cada 20 s: {len(sli)} ventanas, 3 por evento | "
      f"Sesiones de E1 con brecha 3 s: {len(ses)} (la mayor tiene {max(s[2] for s in ses)} eventos)")

# --- Graficos: relojes y tipos de ventana ---
fig, axs = plt.subplots(1, 2, figsize=(12, 4.6))
ax = axs[0]
n_dib = 600
sub = eventos.iloc[:n_dib]
ax.scatter(sub.t_proc, sub.t_evento, s=7, c=np.where(sub.tardio, PALETA[1], PALETA[0]))
ax.plot([0, sub.t_proc.max()], [0, sub.t_proc.max()], "k--", lw=1, label="$t_e = t_p$ (sin retraso)")
ax.set_xlabel("tiempo de procesamiento $t_p$ (s)"); ax.set_ylabel("tiempo de evento $t_e$ (s)")
ax.set_title("Cada punto cae por debajo de la diagonal: llegan tarde\n(rojo = rezagados con retraso de 20 a 90 s)", fontsize=10, fontweight="bold"); ax.legend(loc="upper left")
ax = axs[1]
ax.hist(retraso, bins=60, color=PALETA[0], alpha=0.85)
ax.set_yscale("log"); ax.set_xlabel("retraso $t_p - t_e$ (s)"); ax.set_ylabel("eventos (escala log)")
ax.set_title("Dos poblaciones: red normal (cola exponencial)\ny rezagados (meseta de 20 a 90 s)", fontsize=10, fontweight="bold")
plt.tight_layout(); plt.show()

fig, axs = plt.subplots(3, 1, figsize=(10, 5.4), sharex=True)
etiquetas = ["Tumbling (T = 5)", "Sliding (T = 10, S = 5)", "Sesion (brecha = 3)"]
for ax, et in zip(axs, etiquetas):
    ax.scatter(ej, [0.5] * len(ej), c="k", zorder=3, s=30); ax.set_ylim(0, 3.2); ax.set_yticks([]); ax.set_ylabel(et, rotation=0, ha="right", va="center", fontsize=9); ax.grid(False)
for i, ini in enumerate(sorted(set(ventana_tumbling(x, 5) for x in ej))):
    axs[0].add_patch(FancyBboxPatch((ini + 0.1, 1.0), 4.8, 0.8, boxstyle="round,pad=0.02", fc=PALETA[i % 3], alpha=0.35, ec=PALETA[i % 3]))
for i, ini in enumerate(sorted(cs)):
    axs[1].add_patch(FancyBboxPatch((ini + 0.1, 1.0 + 0.9 * (i % 2)), 9.8, 0.7, boxstyle="round,pad=0.02", fc=PALETA[i % 5], alpha=0.30, ec=PALETA[i % 5]))
for i, (a, b, _) in enumerate(sesiones(ej, 3)):
    axs[2].add_patch(FancyBboxPatch((a - 0.1, 1.0), (b - a) + 3.2, 0.8, boxstyle="round,pad=0.02", fc=PALETA[i + 2], alpha=0.35, ec=PALETA[i + 2]))
axs[2].set_xlabel("tiempo de evento (s); puntos negros = eventos en 1, 2, 7, 8, 9 y 15 s"); axs[0].set_xlim(-6, 26)
axs[0].set_title("Una misma serie de eventos, tres formas de agruparla", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

## 2. Marcas de Agua, Datos Tardíos y Semánticas de Entrega 💧

**El problema.** Una ventana de $[0, 60)$ s ¿cuándo está «completa»? Si esperas poco, **sueltas el resultado rápido pero te pierdes a los rezagados**; si esperas mucho, **el resultado es exacto pero llega tarde** y guardas más estado en memoria. La **marca de agua** (*watermark*) formaliza el compromiso con una regla simple:

$$\text{marca}(t) = \max_{\text{eventos vistos hasta } t} t_e \;-\; L$$

donde $L$ es el **retraso permitido** (*allowed lateness*). Una ventana $[a, b)$ se **cierra y emite** cuando $\text{marca} \ge b$; un evento cuya ventana **ya se cerró** es **tardío**: en el modelo más simple (el de Spark Structured Streaming) **se descarta**. Propiedades que verificaremos:

- La marca **nunca retrocede** (el máximo de $t_e$ visto solo crece).
- Subir $L$ **nunca** descarta más eventos y **nunca** adelanta la emisión: **más completitud, más latencia**.
- Si $L \ge$ (el mayor retraso del flujo), **no se descarta nada**: el resultado es exacto.

**Semánticas de entrega.** Cuando un consumidor puede caerse, hay tres garantías (conceptuales, ordenadas de menos a más cara):

| Garantía | Cómo se logra | Qué pasa si hay caída |
|---|---|---|
| **A lo sumo una vez** (*at-most-once*) | **Confirmar** el *offset* **antes** de procesar | El mensaje en curso **se pierde** |
| **Al menos una vez** (*at-least-once*) | **Procesar y luego confirmar** | Al reiniciar se **relee**: hay **duplicados** |
| **Exactamente una vez** (*exactly-once*) | *Al menos una vez* **más** un sumidero **idempotente** o una transacción que escribe el resultado **y** el *offset* a la vez (Kafka, Flink y Spark ofrecen variantes con condiciones; *verificar cada caso*) | El **efecto** se aplica una sola vez, aunque el mensaje se **procese** más de una |

> ⚠️ **«Exactamente una vez» es sobre el *efecto*, no sobre el *procesamiento***: en un sistema distribuido que puede fallar, un mensaje puede **procesarse** varias veces; lo que se garantiza es que **el resultado visible** (la fila escrita, el saldo) cambie **una sola vez**.

La celda siguiente implementa (1) un **motor de ventanas con marca de agua** que procesa el flujo **en orden de llegada** y barre $L$, y (2) una **simulación de caídas** con las tres garantías sobre un log de 1 000 mensajes.

````python
# ============================================================
# Motor de ventanas tumbling con marca de agua + simulacion de semanticas de entrega
# ============================================================
def motor_marca_agua(ev, tam, L):
    """Procesa 'ev' en orden de LLEGADA. Devuelve (emitidas {inicio: n}, descartados, latencia_media_s).
    Latencia = (tiempo de procesamiento en que se EMITE la ventana) - (fin de la ventana)."""
    te, tp = ev.t_evento.to_numpy(), ev.t_proc.to_numpy()
    abiertas, emitidas, lat = defaultdict(int), {}, []
    max_te, descartados = -np.inf, 0
    for x, ahora in zip(te, tp):
        inicio = ventana_tumbling(x, tam)
        if inicio + tam <= max_te - L:                                          # su ventana ya se cerro con la marca previa
            descartados += 1
        else:
            abiertas[inicio] += 1
        max_te = max(max_te, x)
        marca = max_te - L                                                      # la marca solo crece
        for ini in [i for i in abiertas if i + tam <= marca]:                   # cerrar y emitir
            emitidas[ini] = abiertas.pop(ini); lat.append(ahora - (ini + tam))
    for ini in sorted(abiertas):                                                # fin del flujo: vaciar lo que queda
        emitidas[ini] = abiertas[ini]; lat.append(tp[-1] - (ini + tam))
    return emitidas, descartados, float(np.mean(lat))

exacto = Counter(ventana_tumbling(x, 60) for x in eventos.t_evento)             # lo que daria esperar para siempre
maximo_retraso = float(np.ceil(retraso.max()))
barrido = sorted({0, 2, 5, 10, 30, 60, 90, int(maximo_retraso)})
filas = []
for L in barrido:
    em, desc, lat = motor_marca_agua(eventos, 60, L)
    assert sum(em.values()) + desc == N                                         # cada evento se cuenta o se descarta, jamas se pierde en silencio
    filas.append({"L (s)": L, "emitidos": sum(em.values()), "descartados": desc, "completitud": sum(em.values()) / N, "latencia media (s)": lat})
tabla = pd.DataFrame(filas)
print(tabla.to_string(index=False, formatters={"completitud": "{:.1%}".format, "latencia media (s)": "{:.1f}".format}))

comp, lats = tabla["completitud"].to_numpy(), tabla["latencia media (s)"].to_numpy()
assert np.all(np.diff(comp) >= 0) and np.all(np.diff(lats) >= -1e-9)           # mas L: mas completitud y mas latencia
assert comp[-1] == 1.0 and dict(motor_marca_agua(eventos, 60, maximo_retraso)[0]) == dict(exacto)   # L >= retraso maximo: exacto
assert tabla.loc[0, "descartados"] > 0                                          # con L = 0 se pierden rezagados
print(f"\nCon L = 0 se pierde el {1 - comp[0]:.1%} del flujo; con L = {int(maximo_retraso)} s (el retraso maximo) el resultado es EXACTO, "
      f"pero cada ventana tarda ~{lats[-1]:.0f} s mas en salir (frente a ~{lats[0]:.0f} s).")

# ----------------------------------------------------------------------------
# Semanticas de entrega: un consumidor que se cae con probabilidad p entre dos pasos
# ----------------------------------------------------------------------------
def consumir(n_msgs, estrategia, p_falla=0.02, seed=5, idempotente=False):
    """Lee un log de n_msgs mensajes (el mensaje i vale i+1). 'estrategia': orden de 'procesar' y 'confirmar offset'."""
    gen = np.random.default_rng(seed)
    confirmado, lista, dic, caidas = 0, [], {}, 0
    while confirmado < n_msgs:
        off = confirmado                                                        # tras reiniciar, se lee desde el offset CONFIRMADO
        cae = gen.random() < p_falla                                            # la caida ocurre entre los dos pasos
        if estrategia == "a lo sumo una vez":
            confirmado = off + 1                                                # 1) confirmar
            if cae: caidas += 1; continue                                       # 2) se cae antes de procesar: el mensaje se pierde
            lista.append(off + 1); dic[off] = off + 1
        else:                                                                   # "al menos una vez"
            lista.append(off + 1); dic[off] = off + 1                           # 1) procesar (escribe en el sumidero)
            if cae: caidas += 1; continue                                       # 2) se cae antes de confirmar: se releera
            confirmado = off + 1
    efectos = list(dic.values()) if idempotente else lista                      # sumidero idempotente (por clave) o de solo anexar
    return caidas, efectos

NMSG = 1_000
esperado = NMSG * (NMSG + 1) // 2
res = []
for nombre, estr, idem in [("a lo sumo una vez", "a lo sumo una vez", False), ("al menos una vez", "al menos una vez", False),
                           ("al menos una vez + sumidero idempotente", "al menos una vez", True)]:
    caidas, efectos = consumir(NMSG, estr, idempotente=idem)
    res.append({"garantia": nombre, "caidas": caidas, "perdidos": NMSG - len(set(efectos)), "duplicados": len(efectos) - len(set(efectos)), "suma (esperada 500500)": sum(efectos)})
tab2 = pd.DataFrame(res); print("\n" + tab2.to_string(index=False))
a_sumo, al_menos, exactamente = res
assert a_sumo["perdidos"] > 0 and a_sumo["duplicados"] == 0 and a_sumo["suma (esperada 500500)"] < esperado
assert al_menos["perdidos"] == 0 and al_menos["duplicados"] > 0 and al_menos["suma (esperada 500500)"] > esperado
assert exactamente["perdidos"] == 0 and exactamente["duplicados"] == 0 and exactamente["suma (esperada 500500)"] == esperado
assert a_sumo["caidas"] == a_sumo["perdidos"] and al_menos["caidas"] == al_menos["duplicados"]   # cada caida pierde (o duplica) exactamente un mensaje

fig, axs = plt.subplots(1, 2, figsize=(12, 4.3))
ax = axs[0]
ax.plot(tabla["L (s)"], tabla["completitud"] * 100, "o-", color=PALETA[2], label="completitud (%)")
ax.set_xlabel("retraso permitido $L$ (s)"); ax.set_ylabel("% de eventos contados", color=PALETA[2]); ax.set_ylim(80, 101)
ax2 = ax.twinx(); ax2.plot(tabla["L (s)"], tabla["latencia media (s)"], "s--", color=PALETA[1], label="latencia (s)"); ax2.set_ylabel("latencia media de emision (s)", color=PALETA[1]); ax2.grid(False)
ax.set_title("Marca de agua: mas espera = mas exactitud y mas latencia", fontsize=10, fontweight="bold")
ax = axs[1]
x = np.arange(3); w = 0.38
ax.bar(x - w / 2, [r["perdidos"] for r in res], w, color=PALETA[1], label="perdidos"); ax.bar(x + w / 2, [r["duplicados"] for r in res], w, color=PALETA[3], label="duplicados")
ax.set_xticks(x); ax.set_xticklabels(["a lo sumo\nuna vez", "al menos\nuna vez", "al menos una\n+ idempotente"], fontsize=9); ax.set_ylabel("mensajes (de 1 000)"); ax.legend()
ax.set_title("Cada caida cuesta un mensaje... salvo que el efecto sea idempotente", fontsize=10, fontweight="bold")
plt.tight_layout(); plt.show()
````

## 3. Un Log Particionado tipo Kafka 📜

**Apache Kafka** (Kreps et al., 2011 *(citado de memoria, verificar)*) es el «sistema circulatorio» de muchas arquitecturas de flujo. Su idea central es tan simple que cabe en 40 líneas: un **log de solo anexar** (*append-only*), dividido en **particiones**:

| Concepto | Qué es | Garantía |
|---|---|---|
| **Tema** (*topic*) | Un flujo con nombre, dividido en $P$ **particiones** | — |
| **Partición** | Un log **ordenado** e inmutable; cada mensaje recibe un ***offset*** consecutivo (0, 1, 2...) | **Orden total dentro de la partición**; **ninguno entre particiones** |
| **Clave** | El productor elige la partición con $h(\text{clave}) \bmod P$ | Todos los mensajes de **una clave** van a **la misma partición** → su orden se conserva |
| **Grupo de consumidores** | Varios consumidores que **se reparten las particiones** (cada partición, a **un solo** miembro del grupo) | Escalado horizontal; **máximo $P$** miembros útiles |
| ***Offset* confirmado** (*commit*) | Hasta dónde ha leído **el grupo** | Tras una caída, otro miembro **retoma desde el último confirmado** |
| **Varios grupos** | Cada grupo tiene sus **propios** *offsets* | Cada grupo recibe **todos** los mensajes (difusión) |

Dos consecuencias importantes: (1) **leer no borra**: se puede **volver a leer** (repetición, *replay*) moviendo el *offset*; (2) el **retraso del consumidor** (*lag*) es simplemente `fin del log − offset confirmado`. La celda construye `Tema` y `Grupo` **desde cero** y **comprueba cada garantía** con `assert`: orden por clave, reparto sin duplicados, difusión entre grupos, **reequilibrio** (*rebalance*) cuando un miembro se cae, repetición y *lag*. Usamos un *hash* **estable** (`blake2b`), no `hash()` de Python, que cambia entre ejecuciones.

````python
# ============================================================
# Mini-Kafka desde cero: Tema (particiones + offsets) y Grupo (consumidores, commit, reequilibrio)
# ============================================================
def h64(x):
    """Hash estable de 64 bits: igual en cualquier proceso y ejecucion."""
    return int.from_bytes(hashlib.blake2b(repr(x).encode(), digest_size=8).digest(), "big")

class Tema:
    """Un 'topic': P logs de solo anexar. El offset es la posicion dentro de la particion."""
    def __init__(self, nombre, n_part):
        self.nombre, self.part = nombre, [[] for _ in range(n_part)]
    def produce(self, clave, valor):
        p = h64(clave) % len(self.part)                                         # misma clave -> misma particion
        self.part[p].append((clave, valor))
        return p, len(self.part[p]) - 1                                         # (particion, offset asignado)
    def fin(self, p):
        return len(self.part[p])                                                # 'log end offset'

class Grupo:
    """Grupo de consumidores. El reparto de particiones y los offsets confirmados viven en el grupo."""
    def __init__(self, tema, nombre):
        self.tema, self.nombre = tema, nombre
        self.confirmado = {p: 0 for p in range(len(tema.part))}
        self.miembros, self.asignacion, self.pos = [], {}, {}
    def _reequilibrar(self):
        self.asignacion = {m: [] for m in self.miembros}
        for p in range(len(self.tema.part)):                                    # reparto por turnos (round-robin)
            if self.miembros: self.asignacion[self.miembros[p % len(self.miembros)]].append(p)
        self.pos = {(m, p): self.confirmado[p] for m in self.miembros for p in self.asignacion[m]}   # todos arrancan en lo CONFIRMADO
    def unirse(self, m): self.miembros.append(m); self._reequilibrar()
    def salir(self, m): self.miembros.remove(m); self._reequilibrar()           # lo leido y no confirmado se relee
    def poll(self, m, max_n=50):
        sal = []
        for p in self.asignacion[m]:
            ini = self.pos[(m, p)]
            lote = self.tema.part[p][ini:ini + max_n]
            sal += [(p, ini + i, c, v) for i, (c, v) in enumerate(lote)]
            self.pos[(m, p)] = ini + len(lote)
        return sal
    def commit(self, m):
        for p in self.asignacion[m]: self.confirmado[p] = self.pos[(m, p)]
    def retroceder(self, p, offset):                                            # 'seek': repeticion desde un offset
        self.confirmado[p] = offset; self._reequilibrar()
    def lag(self):
        return {p: self.tema.fin(p) - self.confirmado[p] for p in self.confirmado}

def drenar(g, m):
    """Lee y confirma hasta agotar las particiones del miembro m."""
    todo = []
    while True:
        lote = g.poll(m);
        if not lote: break
        todo += lote; g.commit(m)
    return todo

# --- Productor: los eventos de las estaciones, en orden de llegada, con la estacion como clave ---
NPART = 3
tema = Tema("lecturas", NPART)
asignados = [tema.produce(e, i) for i, e in enumerate(eventos.estacion)]       # el valor es el numero de orden de produccion
total = len(asignados)
for p in range(NPART):
    offs = [o for (pp, o) in asignados if pp == p]
    assert offs == list(range(len(offs))) and len(offs) == tema.fin(p)          # offsets consecutivos 0, 1, 2... por particion
assert sum(tema.fin(p) for p in range(NPART)) == total
for e in eventos.estacion.unique():                                             # cada clave en UNA sola particion y en su orden
    ps = {p for p in range(NPART) for (c, _) in tema.part[p] if c == e}
    vals = [v for (c, v) in tema.part[ps.copy().pop()] if c == e]
    assert len(ps) == 1 and vals == sorted(vals)
print("Mensajes por particion:", {p: tema.fin(p) for p in range(NPART)}, "| claves por particion:", {p: sorted({c for c, _ in tema.part[p]}) for p in range(NPART)})

# --- Grupo A con 2 miembros: reparto sin duplicados ---
A = Grupo(tema, "A"); A.unirse("a1"); A.unirse("a2")
assert {m: len(ps) for m, ps in A.asignacion.items()} == {"a1": 2, "a2": 1}                    # 3 particiones entre 2 miembros: 2 + 1
leidos = drenar(A, "a1") + drenar(A, "a2")
claves_a = [(p, o) for (p, o, _, _) in leidos]
assert len(claves_a) == len(set(claves_a)) == total                             # cada mensaje, UNA vez en todo el grupo
assert all(v == 0 for v in A.lag().values())

# --- Grupo B independiente: recibe TODO de nuevo (difusion entre grupos) ---
B = Grupo(tema, "B"); B.unirse("b1")
assert len(drenar(B, "b1")) == total

# --- Mas miembros que particiones: sobra uno ---
G5 = Grupo(tema, "G5")
for k in range(5): G5.unirse(f"m{k}")
assert sorted(len(ps) for ps in G5.asignacion.values()) == [0, 0, 1, 1, 1]      # maximo 3 miembros utiles

# --- Caida de un miembro: lo no confirmado se re-entrega (al menos una vez) ---
C_ = Grupo(tema, "C"); C_.unirse("c1"); C_.unirse("c2")
l1 = C_.poll("c1", 30); C_.commit("c1")                                         # c1 lee y confirma
l2 = C_.poll("c2", 30)                                                          # c2 lee... y se cae SIN confirmar
C_.salir("c2")                                                                  # reequilibrio: c1 hereda sus particiones
assert sorted(C_.asignacion["c1"]) == list(range(NPART))
rel = []
while True:
    lote = C_.poll("c1", 30)
    if not lote: break
    rel += lote; C_.commit("c1")
vistos1 = {(p, o) for (p, o, _, _) in l1}
vistos2 = {(p, o) for (p, o, _, _) in l2}
vistos_rel = [(p, o) for (p, o, _, _) in rel]
assert vistos2 <= set(vistos_rel)                                               # lo que c2 no confirmo se volvio a entregar
assert not (vistos1 & set(vistos_rel)) and len(vistos_rel) == len(set(vistos_rel))   # y lo confirmado por c1 NO
assert vistos1 | set(vistos_rel) == {(p, o) for p in range(NPART) for o in range(tema.fin(p))}
print(f"c2 leyo {len(l2)} mensajes sin confirmar y se cayo -> el grupo los re-entrego todos ({len(vistos2)} de {len(vistos_rel)} re-lecturas); nada se perdio.")

# --- Repeticion (replay): volver al inicio de una particion y volver a leer ---
A.retroceder(0, 0)
nuevo = drenar(A, A.miembros[0]) + drenar(A, A.miembros[1])
assert sorted(p for (p, _, _, _) in nuevo) == [0] * tema.fin(0)
assert [o for (p, o, _, _) in nuevo if p == 0] == list(range(tema.fin(0)))      # relee la particion 0 completa y en orden

# --- Lag: el productor sigue escribiendo y el grupo A queda atras ---
for i in range(40): tema.produce(f"E{1 + i % 6}", total + i)
assert sum(A.lag().values()) == 40
print("Lag del grupo A tras 40 mensajes nuevos:", A.lag())

# --- Diagrama: las particiones, 28 primeros mensajes de cada una, y los offsets confirmados por un grupo a medio camino ---
D = Grupo(tema, "D"); D.unirse("d1"); D.unirse("d2"); D.poll("d1", 12); D.poll("d2", 20); D.commit("d1"); D.commit("d2")
col = {e: PALETA[i] for i, e in enumerate(sorted(eventos.estacion.unique()))}
fig, ax = plt.subplots(figsize=(11, 3.4))
for p in range(NPART):
    for o, (c, _) in enumerate(tema.part[p][:28]):
        ax.add_patch(FancyBboxPatch((o * 0.9 + 0.05, NPART - 1 - p + 0.1), 0.8, 0.7, boxstyle="round,pad=0.02", fc=col[c], ec="k", lw=0.4))
    ax.text(-0.4, NPART - 1 - p + 0.45, f"Partición {p}", ha="right", va="center", fontsize=9, fontweight="bold")
    ax.annotate("", xy=(D.confirmado[p] * 0.9, NPART - 1 - p - 0.05), xytext=(D.confirmado[p] * 0.9, NPART - 1 - p - 0.45), arrowprops=dict(arrowstyle="->", color="k", lw=1.5))
    ax.text(D.confirmado[p] * 0.9 + 0.1, NPART - 1 - p - 0.45, f"offset confirmado = {D.confirmado[p]}", fontsize=7, va="top")
ax.text(14, NPART + 0.25, "offsets  0 → 27 (el color indica la estación que es la clave)", ha="center", fontsize=9)
ax.set_xlim(-4.2, 26); ax.set_ylim(-0.7, NPART + 0.6); ax.axis("off")
ax.set_title(f"Un tema Kafka: {NPART} logs de solo anexar; un grupo de 2 consumidores a medio camino", fontweight="bold", fontsize=10)
plt.tight_layout(); plt.show()
````

## 4. Spark Structured Streaming Real ⚡

**Structured Streaming** (Armbrust et al., 2018 *(citado de memoria, verificar)*) trata un flujo como una **tabla sin límites a la que se le agregan filas**: escribes la **misma consulta** que harías sobre un `DataFrame` estático (aquí `groupBy(window(...))`) y Spark la ejecuta **incrementalmente**, en micro-lotes. Tres decisiones definen una consulta:

| Decisión | Opciones | Ejemplo en esta sección |
|---|---|---|
| **Fuente** | `kafka`, archivos (`json`, `parquet`...), `rate` (sintética, para pruebas) | `rate` y archivos JSON |
| **Modo de salida** | `append` (solo filas **finales**), `update` (filas cambiadas), `complete` (toda la tabla) | `complete` (sin marca) y `append` (con marca) |
| **Sumidero** y *trigger* | `memory` (pruebas), archivos, `kafka`, `foreachBatch`; `processingTime`, `availableNow` | `memory`; micro-lote cada 1 s y `availableNow` |

Dos cosas que verificamos con **PySpark local real**:

1. **Fuente `rate` + sumidero en memoria:** el flujo corre unos **7 segundos** (hasta ver 4 ventanas) y se detiene con `query.stop()`. Una ventana *tumbling* de 2 s con 50 filas/s debe contar **exactamente 100** filas en cada ventana completa (la primera y la última son parciales).
2. **Fuente de archivos + `availableNow`:** escribimos **nuestros 3 000 eventos** en 6 archivos JSON y Spark los procesa **como flujo** (`availableNow` = procesa lo disponible en micro-lotes y termina). El resultado por ventana de 60 s debe ser **idéntico** al de nuestra implementación desde cero. Luego añadimos una **marca de agua** (`withWatermark`) en modo `append` con un archivo por micro-lote y vemos que **descarta rezagados** y que su resultado **no coincide fila a fila** con el de nuestro motor (el reloj de la marca avanza **por lote**, no por evento).

> ⏱️ Esta celda arranca Spark (≈ 6–15 s) y corre tres consultas de flujo (≈ 25–40 s en total). **Necesita Java 17+** (*verificar*). Al detener la primera consulta con `query.stop()` verás en la salida mensajes de error de Spark («*data source write ... is aborting*»): es el micro-lote que estaba en curso y se **cancela a propósito**, no una falla. Sin Spark, el resto del cuaderno funciona igual; basta con saltar esta celda.

````python
# ============================================================
# Spark Structured Streaming local: rate + memoria, y archivos + availableNow (+ marca de agua)
# ============================================================
from pyspark.sql import SparkSession, functions as F, types as T

t0 = time.perf_counter()
spark = (SparkSession.builder.master("local[2]").appName("bd50").config("spark.ui.enabled", "false")
         .config("spark.sql.shuffle.partitions", "4").config("spark.default.parallelism", "4").config("spark.ui.showConsoleProgress", "false")
         .config("spark.sql.warehouse.dir", f"{TMP}/warehouse").getOrCreate())
spark.sparkContext.setLogLevel("ERROR")
print(f"Sesion Spark local[2] lista en {time.perf_counter() - t0:.1f} s | version {spark.version}")

# --- (1) Fuente 'rate' (50 filas/s) -> ventana tumbling de 2 s -> sumidero en memoria, hasta 4 ventanas, query.stop() ---
TASA, VENT = 50, 2
flujo = spark.readStream.format("rate").option("rowsPerSecond", TASA).load()                 # columnas: timestamp, value
conteo = flujo.groupBy(F.window("timestamp", f"{VENT} seconds")).count()
q = (conteo.writeStream.format("memory").queryName("conteo_rate").outputMode("complete")
     .trigger(processingTime="1 second").option("checkpointLocation", f"{TMP}/ck_rate").start())
t_ini = time.time()
while time.time() - t_ini < 90:                                                              # espera a ver al menos 4 ventanas (maximo 90 s, por si el equipo esta cargado)
    time.sleep(0.5)
    try: n_ventanas = spark.sql("SELECT COUNT(*) FROM conteo_rate").first()[0]
    except Exception: n_ventanas = 0
    if n_ventanas >= 4: break
assert q.isActive
q.stop()                                                                                     # detener la consulta limpiamente
assert not q.isActive and len(spark.streams.active) == 0
rate_df = spark.sql("SELECT window.start AS inicio, count FROM conteo_rate ORDER BY inicio").toPandas()
print("\nFuente rate (ventanas de 2 s):"); print(rate_df.to_string(index=False))
interior = rate_df["count"].to_numpy()[1:-1]                                                 # la primera y la ultima ventana son parciales
assert len(rate_df) >= 3 and len(interior) >= 1 and (interior == TASA * VENT).all()
print(f"-> {len(interior)} ventana(s) completa(s) con exactamente {TASA * VENT} filas; primera y ultima parciales: {rate_df['count'].iloc[0]} y {rate_df['count'].iloc[-1]}.")

# --- (2) Fuente de archivos JSON + availableNow: comparacion exacta con nuestro motor ---
BASE = 1_735_689_600                                                                         # 2025-01-01 00:00:00 UTC (multiplo de 60 s)
dir_ev = f"{TMP}/eventos_json"; os.makedirs(dir_ev)
partes = np.array_split(np.arange(len(eventos)), 6)                                          # 6 archivos, en orden de llegada
for i, idx in enumerate(partes):
    d = eventos.iloc[idx][["t_evento", "estacion", "valor"]].copy()
    d["ts"] = BASE + d.pop("t_evento")
    d[["ts", "estacion", "valor"]].to_json(f"{dir_ev}/parte_{i}.json", orient="records", lines=True)
esquema = T.StructType([T.StructField("ts", T.DoubleType()), T.StructField("estacion", T.StringType()), T.StructField("valor", T.DoubleType())])

ev_s = spark.readStream.schema(esquema).json(dir_ev).withColumn("t", F.col("ts").cast("timestamp"))
agg = ev_s.groupBy(F.window("t", "60 seconds")).agg(F.count("*").alias("n"), F.round(F.sum("valor"), 2).alias("suma"))
q2 = (agg.writeStream.format("memory").queryName("conteo_json").outputMode("complete")
      .trigger(availableNow=True).option("checkpointLocation", f"{TMP}/ck_json").start())
q2.awaitTermination(); q2.stop()
sp = spark.sql(f"SELECT CAST(window.start AS DOUBLE) - {BASE} AS inicio, n, suma FROM conteo_json ORDER BY inicio").toPandas()
mio = (eventos.assign(inicio=[ventana_tumbling(x, 60) for x in eventos.t_evento]).groupby("inicio")
       .agg(n=("valor", "size"), suma=("valor", "sum")).reset_index())
assert len(sp) == len(mio) and (sp.inicio.to_numpy() == mio.inicio.to_numpy()).all()
assert (sp.n.to_numpy() == mio.n.to_numpy()).all() and np.allclose(sp.suma, mio.suma.round(2), atol=0.011)
print(f"\nSpark (complete, sin marca): {len(sp)} ventanas de 60 s, {int(sp.n.sum())} eventos -> IDENTICO a la implementacion desde cero (conteos y sumas).")

# --- (3) Con marca de agua de 10 s, modo append, un archivo por micro-lote ---
ev_w = spark.readStream.schema(esquema).option("maxFilesPerTrigger", 1).json(dir_ev).withColumn("t", F.col("ts").cast("timestamp"))
agg_w = ev_w.withWatermark("t", "10 seconds").groupBy(F.window("t", "60 seconds")).count()
q3 = (agg_w.writeStream.format("memory").queryName("conteo_wm").outputMode("append")
      .trigger(availableNow=True).option("checkpointLocation", f"{TMP}/ck_wm").start())
q3.awaitTermination(); q3.stop()
wm = spark.sql(f"SELECT CAST(window.start AS DOUBLE) - {BASE} AS inicio, count FROM conteo_wm ORDER BY inicio").toPandas()
em_mio, desc_mio, _ = motor_marca_agua(eventos, 60, 10)
cont_spark = int(wm["count"].sum())
print(f"Spark con marca de 10 s (append, 1 archivo por lote): emitio {len(wm)} ventanas con {cont_spark:,} de {N:,} eventos ({cont_spark / N:.1%}).")
print(f"Nuestro motor con L = 10 s (marca por evento):        emitio {len(em_mio)} ventanas con {sum(em_mio.values()):,} eventos ({sum(em_mio.values()) / N:.1%}).")
assert 0 < cont_spark <= N and (wm["count"].to_numpy() <= sp.set_index("inicio").n.reindex(wm.inicio).to_numpy()).all()   # nunca cuenta de mas
print("-> Ambos descartan rezagados, pero NO coinciden fila a fila: Spark avanza la marca por lote y el motor propio por evento (y en 'append' las ultimas ventanas pueden no emitirse).")

fig, axs = plt.subplots(1, 2, figsize=(12, 3.9))
axs[0].bar(range(len(rate_df)), rate_df["count"], color=[PALETA[1]] + [PALETA[0]] * (len(rate_df) - 2) + [PALETA[1]])
axs[0].set_xlabel("ventana de 2 s (n.º de orden)"); axs[0].set_ylabel("filas"); axs[0].set_title("Fuente rate: ventanas completas = 100 filas (rojo = parciales)", fontsize=10, fontweight="bold")
w = 0.4; xs = np.arange(len(sp))
axs[1].bar(xs - w / 2, sp.n, w, color=PALETA[0], label="Spark, sin marca (= motor propio)")
axs[1].bar(xs[: len(wm)] + w / 2, wm.set_index("inicio")["count"].reindex(sp.inicio[: len(wm)]).fillna(0).to_numpy(), w, color=PALETA[3], label="Spark, con marca de 10 s (append)")
axs[1].set_xlabel("ventana de 60 s (n.º de orden)"); axs[1].set_ylabel("eventos"); axs[1].legend(fontsize=8); axs[1].set_title("La marca de agua cuesta rezagados", fontsize=10, fontweight="bold")
plt.tight_layout(); plt.show()

spark.stop()                                                                                  # libera los procesos de Spark
shutil.rmtree(TMP, ignore_errors=True)                                                        # borra archivos y checkpoints temporales
print("Spark detenido y directorio temporal borrado.")
````

## 5. PageRank por Iteración de Potencias 🔗

**PageRank** (Page et al., 1999 *(citado de memoria, verificar)*) mide la **importancia** de cada nodo de un grafo dirigido con una idea recursiva: *un nodo es importante si lo apuntan nodos importantes*. Imagina un **navegante aleatorio** que, en cada paso, con probabilidad $\alpha$ (el **factor de amortiguación**, típicamente $0{,}85$) sigue un enlace saliente al azar y con probabilidad $1-\alpha$ **se teletransporta** a un nodo cualquiera. El PageRank es la **fracción de tiempo** que pasa en cada nodo a largo plazo:

$$PR(v) \;=\; \frac{1-\alpha}{n} \;+\; \alpha \sum_{u \to v} \frac{PR(u)}{d^{+}(u)}$$

con $n$ nodos y $d^{+}(u)$ el grado de salida de $u$. Los nodos **colgantes** (*dangling*, sin enlaces de salida) reparten su masa **uniformemente** entre todos. En forma matricial es el **vector propio dominante** de la «matriz de Google»; se calcula por **iteración de potencias**: se parte de $PR_0 = 1/n$ y se repite $PR_{k+1} = G\,PR_k$ hasta que $\lVert PR_{k+1} - PR_k \rVert_1 < \varepsilon$. Como $G$ es una **contracción** de factor $\alpha$ en la norma $L_1$:

$$\lVert PR_{k+1} - PR_k \rVert_1 \;\le\; \alpha\,\lVert PR_k - PR_{k-1} \rVert_1 \qquad\Rightarrow\qquad \text{error tras } k \text{ pasos} = O(\alpha^k)$$

La celda genera un **grafo tipo web** (enlace preferencial: los nodos nuevos apuntan más a menudo a los ya populares; 300 nodos y algunos colgantes), calcula PageRank **desde cero**, lo **verifica contra `networkx.pagerank`**, comprueba la contracción $\alpha$ y mide cómo cambia el número de iteraciones al subir $\alpha$.

> 🔗 En el [Módulo 01](../01%20-%20Procesamiento%20y%20Almacenamiento%20Distribuido/01_MapReduce_y_sus_Limitaciones.ipynb) calculamos PageRank como **rondas encadenadas de MapReduce** y vimos que escribir a disco entre rondas es lo caro. **Pregel** (sección 6) nace para resolver justo eso.

````python
# ============================================================
# PageRank por iteracion de potencias desde cero + verificacion contra networkx
# ============================================================
def grafo_web(n, m, seed):
    """Enlace preferencial dirigido: cada nodo nuevo apunta a m anteriores, elegidos con prob. proporcional a (entradas + 1)."""
    g = np.random.default_rng(seed)
    entradas = np.ones(n); aristas = set()
    for v in range(1, n):
        p = entradas[:v] / entradas[:v].sum()
        for d in g.choice(v, size=min(m, v), replace=False, p=p):
            aristas.add((v, int(d))); entradas[d] += 1
    colgantes = g.choice(np.arange(1, n), size=8, replace=False)                # 8 nodos mas pierden sus enlaces de salida
    return sorted((u, v) for (u, v) in aristas if u not in set(colgantes.tolist()))

def pagerank(aristas, n, alpha=0.85, tol=1e-12, max_iter=500):
    """Iteracion de potencias. Devuelve (vector PageRank, historial del error L1 entre iteraciones)."""
    src = np.array([u for u, _ in aristas]); dst = np.array([v for _, v in aristas])
    sal = np.bincount(src, minlength=n).astype(float)
    r = np.full(n, 1.0 / n); hist = []
    for _ in range(max_iter):
        contrib = np.zeros(n)
        np.add.at(contrib, dst, r[src] / sal[src])                              # cada nodo reparte su PR entre sus vecinos de salida
        colgante = r[sal == 0].sum()                                            # masa de los nodos sin salida: se reparte uniforme
        nuevo = (1 - alpha) / n + alpha * (contrib + colgante / n)
        err = np.abs(nuevo - r).sum(); r = nuevo; hist.append(err)
        if err < tol: break
    return r, hist

NP = 300
aristas = grafo_web(NP, 3, seed=7)
G = nx.DiGraph(); G.add_nodes_from(range(NP)); G.add_edges_from(aristas)
n_colg = sum(1 for v in G if G.out_degree(v) == 0)
pr, hist = pagerank(aristas, NP)
pr_nx = nx.pagerank(G, alpha=0.85, tol=1e-14, max_iter=1000)
v_nx = np.array([pr_nx[i] for i in range(NP)])
print(f"Grafo: {NP} nodos, {len(aristas)} aristas, {n_colg} nodos colgantes | {len(hist)} iteraciones hasta error < 1e-12")
print(f"Diferencia maxima con networkx.pagerank: {np.abs(pr - v_nx).max():.2e}")
assert abs(pr.sum() - 1) < 1e-9 and (pr > 0).all() and n_colg >= 8
assert np.abs(pr - v_nx).max() < 1e-9                                           # coincide con la referencia
assert all(hist[i + 1] <= 0.85 * hist[i] + 1e-15 for i in range(len(hist) - 1))  # contraccion de factor alfa en L1

# --- La importancia NO es lo mismo que el numero de enlaces ---
ent = np.array([G.in_degree(v) for v in range(NP)])
orden = np.argsort(-pr)[:8]
print("\nTop 8 por PageRank:"); print(pd.DataFrame({"nodo": orden, "PageRank": pr[orden].round(5), "enlaces de entrada": ent[orden]}).to_string(index=False))
rho = pd.Series(pr).corr(pd.Series(ent), method="spearman")
print(f"Correlacion de Spearman entre PageRank y grado de entrada: {rho:.4f} (casi 1, pero el top NO es el mismo orden)")
assert 0.5 < rho < 1.0

# --- Mas alfa, mas iteraciones ---
its = {a: len(pagerank(aristas, NP, alpha=a, tol=1e-10)[1]) for a in [0.5, 0.7, 0.85, 0.95, 0.99]}
print("\nIteraciones hasta error < 1e-10 segun alfa:", its, "(la cota alfa^k es de peor caso: en un grafo bien conectado el aumento es moderado)")
vals = list(its.values()); assert vals == sorted(vals) and vals[-1] > vals[0]               # sube con alfa (la cota alfa^k es de PEOR caso)

fig, axs = plt.subplots(1, 2, figsize=(12, 4.3))
ax = axs[0]
ax.semilogy(range(1, len(hist) + 1), hist, "o-", ms=3, color=PALETA[0], label="error $L_1$ medido")
ax.semilogy(range(1, len(hist) + 1), hist[0] * 0.85 ** np.arange(len(hist)), "--", color=PALETA[1], label=r"cota $\alpha^k$ con $\alpha=0{,}85$")
ax.set_xlabel("iteración"); ax.set_ylabel("error $L_1$ entre iteraciones consecutivas"); ax.legend(); ax.set_title("Convergencia geométrica de la iteración de potencias", fontsize=10, fontweight="bold")
ax = axs[1]
pos = nx.spring_layout(G, seed=3, k=0.35)
tam = 60 + 9000 * pr
nx.draw_networkx_edges(G, pos, ax=ax, alpha=0.12, arrows=False, width=0.4)
nodos = nx.draw_networkx_nodes(G, pos, ax=ax, node_size=tam, node_color=pr, cmap="viridis")
nx.draw_networkx_labels(G, pos, labels={int(v): str(int(v)) for v in orden[:5]}, font_size=8, font_color="white", ax=ax)
ax.set_title("Grafo tipo web: el tamaño es el PageRank (etiquetados los 5 primeros)", fontsize=10, fontweight="bold"); ax.axis("off")
plt.tight_layout(); plt.show()
````

## 6. Pregel: «Piensa como un Vértice» y el Modelo BSP 🕸️

**Pregel** (Malewicz et al., 2010 *(citado de memoria, verificar)*) es el modelo de Google para grafos masivos, basado en el modelo **BSP** (*Bulk Synchronous Parallel*, Valiant, 1990). En vez de programar el grafo completo, escribes **una función que corre en cada vértice**:

1. El cómputo avanza en **superpasos** (*supersteps*) separados por una **barrera** global.
2. En cada superpaso, cada vértice **activo** lee los **mensajes** que le llegaron en el superpaso anterior, **actualiza su valor** y **envía mensajes** a sus vecinos (que se entregan en el siguiente).
3. Un vértice puede **votar por detenerse** (*vote to halt*); un mensaje lo **reactiva**.
4. El cómputo **termina** cuando **todos están detenidos y no hay mensajes en tránsito**.

Un **combinador** (como el de MapReduce) reduce mensajes al mismo destino (`min`, `sum`...) **antes** de enviarlos. Frente a MapReduce encadenado, el **estado de los vértices se queda en memoria** entre superpasos y **solo se mueven mensajes**, justo lo que hace caro a MapReduce en los algoritmos iterativos.

| Algoritmo | Valor del vértice | Qué envía | Cuándo se detiene |
|---|---|---|---|
| **Componentes conexas** | la **menor etiqueta** vista | su etiqueta si **bajó** | nadie mejora |
| **Caminos más cortos** desde una fuente | la **mejor distancia** conocida | `distancia + peso` si **mejoró** | nadie mejora (como Bellman-Ford distribuido) |
| **PageRank** | el PageRank actual | `PR / grado de salida`, durante $K$ superpasos fijos | tras $K$ superpasos |

La celda implementa el **motor** `pregel(...)` (≈ 25 líneas) y los **tres programas de vértice**; los **verifica** contra `networkx` (componentes, Dijkstra, PageRank) y contra nuestro PageRank de la sección 5; mide **mensajes y vértices activos por superpaso** (con y sin combinador); y dibuja el modelo BSP.

#### GraphX y GraphFrames (solo conceptual, no se ejecuta aquí)

| | **GraphX** | **GraphFrames** |
|---|---|---|
| Qué es | Biblioteca de grafos de Spark sobre RDD | Paquete de grafos sobre **DataFrames** |
| Lenguajes | **Scala/Java** (no tiene API de Python) | **Scala, Java y Python** (`pip`/`--packages`; *verificar versión compatible con tu Spark*) |
| Operadores | `pageRank`, `connectedComponents`, `triangleCount` y el operador **`pregel`** (`vprog`, `sendMsg`, `mergeMsg` = nuestro programa, mensajes y combinador) | `pageRank`, `connectedComponents`, `shortestPaths`, `bfs`, `aggregateMessages` y búsqueda de **patrones** («motifs») |
| Cuándo | Grafos sobre RDD ya existentes | Cuando quieres mezclar **consultas relacionales** y de grafo en el mismo `DataFrame` |

Una consulta en GraphFrames se vería así (**ilustrativo, no se ejecuta**; la API exacta, el parámetro de punto de control y la versión del paquete deben **verificarse** en la documentación):

> ⚠️ **Honestidad sobre escala:** en un solo proceso, NumPy o `networkx` le ganan a cualquiera de estos motores en grafos que caben en memoria (lo vimos con PageRank en el Módulo 01). Pregel, GraphX y GraphFrames valen cuando **el grafo no cabe en una máquina**.

````python
# ============================================================
# Motor Pregel/BSP desde cero + componentes conexas + caminos mas cortos + PageRank
# ============================================================
def pregel(adj, valor_inicial, programa, combinador=None, max_pasos=500):
    """adj: {vertice: [(destino, peso), ...]}. programa(v, valor, mensajes, superpaso, aristas) -> (nuevo_valor, [(destino, msg)], detenerse)."""
    valores = dict(valor_inicial)
    activos, bandeja = set(adj), {}                                             # superpaso 0: todos activos, sin mensajes
    info = {"pasos": 0, "activos": [], "enviados": [], "entregados": []}
    for s in range(max_pasos):
        ejecutar = activos | set(bandeja)                                       # activos + los reactivados por un mensaje
        if not ejecutar: break
        info["activos"].append(len(ejecutar))
        nuevas, enviados, activos = defaultdict(list), 0, set()
        for v in ejecutar:
            valores[v], salientes, parar = programa(v, valores[v], bandeja.get(v, []), s, adj[v])
            for d, m in salientes: nuevas[d].append(m); enviados += 1
            if not parar: activos.add(v)
        # ---- BARRERA: los mensajes se entregan en el superpaso siguiente ----
        bandeja = {d: [combinador(ms)] for d, ms in nuevas.items()} if combinador else dict(nuevas)
        info["enviados"].append(enviados); info["entregados"].append(sum(len(ms) for ms in bandeja.values())); info["pasos"] = s + 1
    return valores, info

# ---- Un grafo con varias componentes: aleatorio + camino de 15 + clique de 5 + 3 aislados; pesos enteros 1..9 ----
g0 = np.random.default_rng(3)
G_cc = nx.disjoint_union_all([nx.gnm_random_graph(120, 200, seed=3), nx.path_graph(15), nx.complete_graph(5), nx.empty_graph(3)])
for u, v in G_cc.edges: G_cc[u][v]["peso"] = int(g0.integers(1, 10))
adj_cc = {u: [(v, G_cc[u][v]["peso"]) for v in G_cc[u]] for u in G_cc}          # no dirigido: cada arista en ambos sentidos
INF = float("inf")

# ---- Programa 1: componentes conexas (propagar la menor etiqueta) ----
def prog_cc(v, val, msgs, s, out):
    nuevo = min([val] + list(msgs))
    if s == 0 or nuevo < val:
        return nuevo, [(d, nuevo) for d, _ in out], True
    return val, [], True

etq, info_cc = pregel(adj_cc, {v: v for v in adj_cc}, prog_cc, combinador=min)
comps = list(nx.connected_components(G_cc))
assert len({etq[v] for v in G_cc}) == len(comps)
for c in comps: assert len({etq[v] for v in c}) == 1 and etq[next(iter(c))] == min(c)
_, info_cc_sin = pregel(adj_cc, {v: v for v in adj_cc}, prog_cc)
print(f"Componentes conexas: {len(comps)} (coincide con networkx) en {info_cc['pasos']} superpasos | mensajes enviados {sum(info_cc['enviados']):,} "
      f"| entregados con combinador min: {sum(info_cc['entregados']):,} (sin combinador: {sum(info_cc_sin['entregados']):,})")
assert sum(info_cc["entregados"]) < sum(info_cc_sin["entregados"]) and sum(info_cc["enviados"]) == sum(info_cc_sin["enviados"])

# ---- Programa 2: caminos mas cortos desde una fuente (Bellman-Ford distribuido) ----
fuente = min(max(comps, key=len))
def prog_sssp(v, val, msgs, s, out):
    nuevo = min(msgs) if msgs else INF
    if s == 0 and v == fuente: nuevo = 0
    if nuevo < val:
        return nuevo, [(d, nuevo + w) for d, w in out], True
    return val, [], True

dist, info_sp = pregel(adj_cc, {v: INF for v in adj_cc}, prog_sssp, combinador=min)
ref = nx.single_source_dijkstra_path_length(G_cc, fuente, weight="peso")
assert {v for v in dist if dist[v] < INF} == set(ref) and all(dist[v] == ref[v] for v in ref)
excentricidad_saltos = max(nx.single_source_shortest_path_length(G_cc, fuente).values())
print(f"Caminos mas cortos desde {fuente}: {len(ref)} alcanzables de {G_cc.number_of_nodes()} (coincide con Dijkstra) | {info_sp['pasos']} superpasos "
      f"(radio en saltos desde la fuente: {excentricidad_saltos}) | distancia maxima {max(ref.values())}")
assert info_sp["pasos"] >= excentricidad_saltos                                  # no puede ir mas rapido que el diametro del grafo
assert min(info_sp["activos"][1:]) < info_sp["activos"][0]                       # el frente de ondas se encoge

# ---- Programa 3: PageRank en Pregel (K superpasos fijos) sobre un grafo SIN nodos colgantes ----
n_pr, K, alfa = 150, 80, 0.85
g1 = np.random.default_rng(11)
ar_pr = sorted({(u, int(v)) for u in range(n_pr) for v in g1.choice([x for x in range(n_pr) if x != u], size=3, replace=False)})
adj_pr = {u: [] for u in range(n_pr)}
for u, v in ar_pr: adj_pr[u].append((v, 1))
def prog_pr(v, val, msgs, s, out):
    if s > 0: val = (1 - alfa) / n_pr + alfa * sum(msgs)
    if s < K: return val, [(d, val / len(out)) for d, _ in out], False
    return val, [], True
pr_pg, info_pr = pregel(adj_pr, {u: 1.0 / n_pr for u in adj_pr}, prog_pr, combinador=sum)
vec_pg = np.array([pr_pg[u] for u in range(n_pr)])
vec_pot, _ = pagerank(ar_pr, n_pr, alpha=alfa, tol=0, max_iter=K)                # K iteraciones de potencias, sin parar antes
Gd = nx.DiGraph(); Gd.add_edges_from(ar_pr)
vec_nx = np.array([nx.pagerank(Gd, alpha=alfa, tol=1e-14, max_iter=1000)[u] for u in range(n_pr)])
print(f"PageRank en Pregel ({info_pr['pasos']} superpasos): |Pregel - potencias| = {np.abs(vec_pg - vec_pot).max():.1e} | |Pregel - networkx| = {np.abs(vec_pg - vec_nx).max():.1e}")
assert np.abs(vec_pg - vec_pot).max() < 1e-12 and np.abs(vec_pg - vec_nx).max() < 1e-6 and abs(vec_pg.sum() - 1) < 1e-9
assert info_pr["pasos"] == K + 1 and all(e == len(ar_pr) for e in info_pr["enviados"][:K])   # cada superpaso envia un mensaje por arista

# ---- Graficos: el modelo BSP y la actividad por superpaso ----
fig, axs = plt.subplots(1, 2, figsize=(13, 4.6), gridspec_kw={"width_ratios": [1.15, 1]})
ax = axs[0]; ax.axis("off"); ax.set_xlim(0, 12); ax.set_ylim(0, 6)
for s in range(3):
    x0 = 0.3 + s * 4
    ax.add_patch(FancyBboxPatch((x0, 3.3), 3.2, 2.0, boxstyle="round,pad=0.05", fc="#dbeafe", ec=PALETA[0]))
    ax.text(x0 + 1.6, 5.0, f"Superpaso {s}", ha="center", fontweight="bold", fontsize=9)
    for j in range(3):
        ax.add_patch(plt.Circle((x0 + 0.6 + j * 1.0, 4.0), 0.3, fc="white", ec=PALETA[0])); ax.text(x0 + 0.6 + j * 1.0, 4.0, f"v{j + 1}", ha="center", va="center", fontsize=8)
    ax.text(x0 + 1.6, 3.55, "cada vértice: lee, calcula, envía", ha="center", fontsize=7)
    ax.add_patch(FancyBboxPatch((x0, 1.5), 3.2, 0.8, boxstyle="round,pad=0.05", fc="#fef3c7", ec=PALETA[3]))
    ax.text(x0 + 1.6, 1.9, "mensajes en tránsito", ha="center", va="center", fontsize=8)
    ax.annotate("", xy=(x0 + 1.6, 2.35), xytext=(x0 + 1.6, 3.25), arrowprops=dict(arrowstyle="->", color="k"))
    if s < 2:
        ax.plot([x0 + 3.6, x0 + 3.6], [0.7, 5.6], color=PALETA[1], lw=3); ax.text(x0 + 3.6, 0.35, "barrera", color=PALETA[1], ha="center", fontsize=8, fontweight="bold")
        ax.annotate("", xy=(x0 + 4.2, 1.9), xytext=(x0 + 3.2, 1.9), arrowprops=dict(arrowstyle="->", color=PALETA[3], lw=1.5))
ax.set_title("BSP: superpasos separados por barreras; los mensajes se entregan en el superpaso siguiente", fontsize=9, fontweight="bold")
ax = axs[1]
xs = np.arange(len(info_sp["activos"]))
ax.bar(xs - 0.2, info_sp["activos"], 0.4, color=PALETA[0], label="vértices activos")
ax.bar(xs + 0.2, info_sp["enviados"], 0.4, color=PALETA[3], label="mensajes enviados")
ax.set_xlabel("superpaso"); ax.set_ylabel("cantidad"); ax.legend()
ax.set_title(f"Caminos más cortos desde {fuente}: una onda que avanza y se apaga", fontsize=10, fontweight="bold")
plt.tight_layout(); plt.show()

# puntaje distinto por nodo (lo usa la Practica 2)
puntaje = {v: int(x) for v, x in zip(G_cc.nodes, np.random.default_rng(1).permutation(G_cc.number_of_nodes()))}
````

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Dos relojes** | El **tiempo de evento** (cuándo ocurrió) no es el de **procesamiento** (cuándo llegó): aquí el 6 % de los eventos llegó con 20 a 90 s de retraso y el flujo llegó **desordenado**. Agregar por $t_e$ da el resultado correcto, pero obliga a decidir cuánto esperar |
| **Ventanas** | *Tumbling* (una ventana por evento), *sliding* ($T/S$ ventanas por evento) y de **sesión** (se extienden mientras el hueco sea menor que la brecha); implementadas desde cero y verificadas contra `pandas` |
| **Marca de agua** | $\max t_e - L$: **más $L$ = más completitud y más latencia**, nunca al revés; con $L \ge$ el retraso máximo el resultado es **exacto** (verificado) y con $L=0$ se **pierden** rezagados |
| **Semánticas de entrega** | *A lo sumo una vez* **pierde**, *al menos una vez* **duplica**, y *exactamente una vez* es **al menos una vez + efecto idempotente**: cada caída costó exactamente un mensaje perdido o duplicado |
| **Log tipo Kafka** | Particiones con orden total **solo dentro de la partición**, clave → partición, grupos que se reparten las particiones (máximo $P$ útiles), *offsets* confirmados, reequilibrio al caer un miembro (se **re-entrega** lo no confirmado), repetición y *lag* |
| **Structured Streaming** | La consulta es la misma que la estática; con archivos y `availableNow` el resultado fue **idéntico** a nuestro motor; con marca de agua y `append` Spark descarta rezagados, pero **no coincide fila a fila** con la marca por evento (la marca avanza por lote). Se detuvo con `query.stop()` |
| **PageRank** | Iteración de potencias: contracción de factor $\alpha$ (error $O(\alpha^k)$); coincide con `networkx.pagerank` a $10^{-9}$; **correlaciona fuertemente** con el grado de entrada pero **no es igual** (un nodo con más enlaces de entrada puede tener menor PageRank si los enlaces vienen de nodos poco importantes); más $\alpha$, más iteraciones (moderadamente en este grafo) |
| **Pregel/BSP** | «Piensa como un vértice»: superpasos, mensajes, votar por detenerse y combinadores. Componentes conexas, caminos más cortos y PageRank **verificados** contra `networkx`; el estado queda en memoria y solo viajan mensajes |
| **GraphX/GraphFrames** | GraphX (Scala/Java, operador `pregel`) y GraphFrames (DataFrames, Python): **solo conceptual**. Valen cuando el grafo **no cabe** en una máquina; si cabe, `networkx` o NumPy ganan |
