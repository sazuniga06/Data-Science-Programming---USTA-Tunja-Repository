## 1. Analíticas Empoderadas con IA (I): Detección de Anomalías 🚨

La «analítica aumentada» o «empoderada con IA» no es magia: es **automatizar con modelos** lo que antes hacía una regla o un analista mirando un tablero. El caso de uso más común en Big Data es **detectar lo raro**: pagos fraudulentos, sensores averiados, accesos sospechosos. Como casi nunca hay etiquetas, se usan métodos **no supervisados**.

| Método | Idea | Fortaleza | Debilidad |
|---|---|---|---|
| **Puntuación $z$ robusta** (*modified z-score*) | $z = 0{,}6745\,(x - \text{mediana}) / \text{MAD}$; alerta si $\lvert z \rvert > 3{,}5$ (regla de Iglewicz y Hoaglin *(verificar)*) | Simple, explicable, barata | **Una variable a la vez**: no ve combinaciones raras |
| ***Isolation Forest*** (Liu et al., 2008 *(citado de memoria, verificar)*) | Árboles que **aíslan** puntos con cortes al azar sobre variables elegidas al azar: lo raro se aísla en **pocos cortes** (rama corta) | Multivariado, escala bien, sin supuestos de distribución | Menos explicable; exige elegir cuánta contaminación esperar; una anomalía que lo es en **una sola** variable entre muchas **se diluye** |

El **pipeline** correcto encadena **escalado + modelo** en un solo objeto (así, al reentrenar o validar, el escalado se ajusta **solo** con datos de entrenamiento: sin fuga de datos). Generamos **5 000 pagos sintéticos** con cuatro variables (monto, hora —representada con **seno y coseno**, porque 23:59 y 00:01 están pegadas—, distancia a la ubicación habitual y número de ítems) e inyectamos **75 anomalías (1,5 %)** de **dos tipos**: **A)** montos altísimos y **B)** pagos de **madrugada y muy lejos**, con **monto normal**. Esperamos (y comprobaremos) que **cada método vea un tipo distinto**. Evaluamos con **precisión promedio** (área bajo la curva precisión-exhaustividad, apropiada con clases muy desbalanceadas) y con la **exhaustividad por tipo** si el equipo solo puede revisar 75 alertas.

````python
# ============================================================
# Configuracion + deteccion de anomalias: z robusta desde cero vs. Pipeline(StandardScaler + IsolationForest)
# ============================================================
# !pip install -q numpy pandas matplotlib scikit-learn

import hashlib
import json
import math
import random
import re
import time
import unicodedata
import warnings
from collections import Counter

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.patches import FancyBboxPatch
from sklearn.ensemble import IsolationForest
from sklearn.metrics import average_precision_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

warnings.filterwarnings("ignore")
SEED = 42
RAPIDO = True                                                                  # True: presupuestos reducidos (el cuaderno completo corre en ~1 min)
plt.rcParams.update({"figure.figsize": (9, 5), "axes.grid": True, "grid.alpha": 0.3, "axes.spines.top": False, "axes.spines.right": False})
PALETA = ["#2563eb", "#dc2626", "#16a34a", "#f59e0b", "#7c3aed", "#0891b2", "#db2777", "#65a30d", "#78350f", "#475569"]
print(f"Entorno listo | numpy {np.__version__} | pandas {pd.__version__} | RAPIDO={RAPIDO}")

# ----------------------------------------------------------------------------
# Datos sinteticos: 5 000 pagos, 75 anomalias de dos tipos (A: monto enorme; B: madrugada + lejos, monto normal)
# ----------------------------------------------------------------------------
rng = np.random.default_rng(SEED)
n, n_a, n_b = 5_000, 45, 30
n_norm = n - n_a - n_b
hora = np.where(rng.random(n_norm) < 0.8, rng.normal(14, 4, n_norm), rng.normal(20, 1.5, n_norm)) % 24      # actividad diurna y de la tarde-noche
monto = rng.lognormal(10.5, 0.6, n_norm)                                                                      # en pesos (inventados)
dist = rng.exponential(5, n_norm)                                                                             # km a la ubicacion habitual
items = rng.poisson(3, n_norm) + 1
hora = np.r_[hora, rng.normal(14, 4, n_a) % 24, rng.uniform(2, 4.5, n_b)]
monto = np.r_[monto, rng.lognormal(10.5 + 2.8, 0.3, n_a), rng.lognormal(10.5, 0.6, n_b)]
dist = np.r_[dist, rng.exponential(5, n_a), rng.uniform(150, 400, n_b)]
items = np.r_[items, rng.poisson(3, n_a) + 1, rng.poisson(3, n_b) + 1]
tipo = np.array(["normal"] * n_norm + ["A"] * n_a + ["B"] * n_b)
df = pd.DataFrame({"hora": hora, "monto": monto, "dist_km": dist, "items": items, "tipo": tipo}).sample(frac=1, random_state=SEED).reset_index(drop=True)
y = (df.tipo != "normal").to_numpy()
assert df.shape[0] == n and y.sum() == n_a + n_b and abs(y.mean() - 0.015) < 1e-9

def matriz(d):
    """Variables del modelo: log del monto, hora como (seno, coseno), log de la distancia e items."""
    ang = 2 * np.pi * d.hora.to_numpy() / 24
    return np.c_[np.log(d.monto), np.sin(ang), np.cos(ang), np.log1p(d.dist_km), d["items"]]

# --- Metodo 1 (desde cero): puntuacion z robusta sobre UNA variable (log del monto) ---
def z_robusta(x):
    med = np.median(x); mad = np.median(np.abs(x - med))
    return 0.6745 * (x - med) / mad
z = np.abs(z_robusta(np.log(df.monto.to_numpy())))
alertas_z = z > 3.5
# --- Metodo 2: Pipeline(StandardScaler + IsolationForest) ---
pipe = Pipeline([("escalado", StandardScaler()), ("iforest", IsolationForest(n_estimators=200, contamination=0.015, random_state=SEED))])
pipe.fit(matriz(df))
puntaje_if = -pipe.score_samples(matriz(df))                                  # mayor = mas anomalo
K = int(y.sum())                                                              # presupuesto: revisar K alertas
alertas_if = np.zeros(n, bool); alertas_if[np.argsort(-puntaje_if)[:K]] = True
alertas_z_top = np.zeros(n, bool); alertas_z_top[np.argsort(-z)[:K]] = True   # mismo presupuesto para comparar con justicia

def reporte(nombre, alertas, puntaje):
    return {"metodo": nombre, "precision promedio": average_precision_score(y, puntaje), f"precision en top-{K}": (alertas & y).sum() / K,
            "exhaustividad tipo A": (alertas & (df.tipo == "A")).sum() / n_a, "exhaustividad tipo B": (alertas & (df.tipo == "B")).sum() / n_b}
tabla = pd.DataFrame([reporte("z robusta (solo el monto)", alertas_z_top, z), reporte("Pipeline(StandardScaler + IsolationForest)", alertas_if, puntaje_if)])
print(tabla.to_string(index=False, float_format=lambda v: f"{v:.3f}"))
print(f"\nLa regla |z| > 3,5 sobre el monto lanza {alertas_z.sum()} alertas por si sola (umbral fijo, sin presupuesto).")

# --- Cada metodo ve lo que su diseno le permite; combinarlos cubre ambos tipos ---
assert tabla.iloc[0, 3] >= 0.9 and tabla.iloc[0, 4] <= 0.1                    # la z robusta ve TODO el tipo A (monto) y nada del B
assert tabla.iloc[1, 4] >= 0.9 and tabla.iloc[1, 3] < 0.5                     # IsolationForest ve el tipo B (3 variables raras) pero poco del A (1 de 5 variables)
alertas_union = alertas_z | (pipe.predict(matriz(df)) == -1)                  # regla |z| > 3,5 O alerta de IsolationForest (contaminacion 1,5 %)
rec_a_u, rec_b_u = (alertas_union & (df.tipo == "A")).sum() / n_a, (alertas_union & (df.tipo == "B")).sum() / n_b
prec_u = (alertas_union & y).sum() / alertas_union.sum()
assert rec_a_u >= 0.9 and rec_b_u >= 0.9 and alertas_union.sum() < 2 * K
print(f"Union de ambos detectores: {alertas_union.sum()} alertas | precision {prec_u:.2f} | exhaustividad A {rec_a_u:.2f}, B {rec_b_u:.2f}")
print("-> Ninguno gana en todo: la regla univariada ve lo que es raro en UNA variable; Isolation Forest ve lo raro en MUCHAS a la vez y se diluye cuando solo una de cinco variables lo es. Combinarlos cubre ambos tipos.")

fig, axs = plt.subplots(1, 2, figsize=(12, 4.4))
ax = axs[0]
colores = {"normal": "#cbd5e1", "A": PALETA[1], "B": PALETA[3]}
for t in ["normal", "A", "B"]:
    m = (df.tipo == t).to_numpy()
    ax.scatter(df.hora[m], np.log10(df.dist_km[m] + 1), s=8 if t == "normal" else 28, c=colores[t], label={"normal": "normal", "A": "A: monto enorme", "B": "B: madrugada y lejos"}[t], zorder=3 if t != "normal" else 1)
ax.scatter(df.hora[alertas_if], np.log10(df.dist_km[alertas_if] + 1), s=90, facecolors="none", edgecolors="k", lw=0.8, label=f"alerta de IsolationForest ({K})")
ax.set_xlabel("hora del día"); ax.set_ylabel("log10(distancia + 1)"); ax.set_ylim(-0.1, 3.5); ax.legend(fontsize=7, loc="upper left", ncol=2); ax.set_title("El tipo B (naranja) solo se ve combinando hora y distancia", fontsize=10, fontweight="bold")
ax = axs[1]
x = np.arange(2); w = 0.38
ax.bar(x - w, [tabla.iloc[0, 3], tabla.iloc[0, 4]], w, color=PALETA[4], label="z robusta (monto)")
ax.bar(x, [tabla.iloc[1, 3], tabla.iloc[1, 4]], w, color=PALETA[0], label="IsolationForest")
ax.bar(x + w, [rec_a_u, rec_b_u], w, color=PALETA[2], label="unión de ambos")
ax.set_xticks(x); ax.set_xticklabels(["tipo A (monto enorme)", "tipo B (madrugada y lejos)"]); ax.set_ylabel(f"exhaustividad (top-{K}; la unión usa {alertas_union.sum()} alertas)"); ax.set_ylim(0, 1.1); ax.legend()
ax.set_title("¿Cuántas anomalías de cada tipo se atrapan?", fontsize=10, fontweight="bold")
plt.tight_layout(); plt.show()
````

## 2. Analíticas Empoderadas con IA (II): AutoML como Búsqueda Aleatoria 🎲

**AutoML** automatiza la elección del **modelo y de sus ajustes**. En su versión más honesta y sencilla es una **búsqueda** sobre un **espacio de *pipelines***: cada configuración es una tupla (*escalador*, *selección de variables*, *modelo*, *hiperparámetros*); se **evalúa por validación cruzada** y se queda con la mejor. Bergstra y Bengio (2012) *(citado de memoria, verificar)* mostraron que la **búsqueda aleatoria** suele ser **tan buena como la de rejilla con mucho menos costo**, porque casi siempre **pocos hiperparámetros importan** y la rejilla desperdicia pruebas repitiendo valores de los que no importan.

Tres principios que aplicaremos:

1. **Todo dentro del *pipeline*:** el escalador y la selección de variables se **ajustan dentro de cada pliegue**; si se ajustaran antes de partir los datos habría **fuga de datos** (*data leakage*).
2. **Un conjunto de prueba que la búsqueda nunca toca:** la puntuación de validación cruzada del **mejor** de $T$ intentos es **optimista** (*maldición del ganador*: elegiste el máximo de $T$ ruidos); solo la **prueba** da una estimación honesta.
3. **Curva «mejor hasta ahora»:** mide cuánto **rinde** cada intento adicional. Para suavizar el azar del orden la **promediamos sobre 300 reordenamientos** de los mismos intentos (sin reentrenar nada).

La celda genera un problema de clasificación sintético (1 500 filas, 20 variables, 5 informativas), define el espacio, **corre la búsqueda desde cero** (24 intentos con `RAPIDO = True`, validación cruzada de 3 pliegues, AUC) y la compara con una línea base (`StandardScaler` + `LogisticRegression`).

````python
# ============================================================
# AutoML = busqueda aleatoria sobre pipelines de scikit-learn (desde cero) + linea base + conjunto de prueba intocado
# ============================================================
from sklearn.datasets import make_classification
from sklearn.ensemble import HistGradientBoostingClassifier, RandomForestClassifier
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import cross_val_score, train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import MinMaxScaler, RobustScaler

Xa, ya = make_classification(n_samples=1_500, n_features=20, n_informative=5, n_redundant=3, n_clusters_per_class=2, flip_y=0.05, class_sep=0.8, random_state=SEED)
Xa_ent, Xa_pru, ya_ent, ya_pru = train_test_split(Xa, ya, test_size=0.3, stratify=ya, random_state=SEED)

def muestrear(r):
    """Una configuracion al azar del espacio de pipelines."""
    cfg = {"escalador": r.choice(["ninguno", "estandar", "minmax", "robusto"]), "k": r.choice([0, 6, 10, 15]), "modelo": r.choice(["logreg", "rf", "hgb", "knn"])}
    if cfg["modelo"] == "logreg": cfg["hp"] = {"C": round(10 ** r.uniform(-3, 2), 5)}
    elif cfg["modelo"] == "rf": cfg["hp"] = {"n_estimators": r.choice([50, 100]), "max_depth": r.choice([3, 6, 12, None])}
    elif cfg["modelo"] == "hgb": cfg["hp"] = {"max_depth": r.choice([2, 3, 5]), "learning_rate": round(10 ** r.uniform(-2, -0.5), 4), "max_iter": 60}
    else: cfg["hp"] = {"n_neighbors": r.choice([3, 7, 15, 31])}
    return cfg

def construir(cfg):
    pasos = []
    esc = {"estandar": StandardScaler, "minmax": MinMaxScaler, "robusto": RobustScaler}.get(cfg["escalador"])
    if esc: pasos.append(("escalador", esc()))
    if cfg["k"]: pasos.append(("seleccion", SelectKBest(f_classif, k=cfg["k"])))                # DENTRO del pipeline: se ajusta en cada pliegue
    m = cfg["modelo"]
    modelo = (LogisticRegression(max_iter=500, **cfg["hp"]) if m == "logreg" else RandomForestClassifier(random_state=SEED, n_jobs=1, **cfg["hp"]) if m == "rf"
              else HistGradientBoostingClassifier(random_state=SEED, **cfg["hp"]) if m == "hgb" else KNeighborsClassifier(**cfg["hp"]))
    return Pipeline(pasos + [("modelo", modelo)])

def evaluar(cfg):
    return float(cross_val_score(construir(cfg), Xa_ent, ya_ent, cv=3, scoring="roc_auc").mean())

PRESUPUESTO = 24 if RAPIDO else 60
r = random.Random(SEED)
t0 = time.perf_counter()
intentos = []
for i in range(PRESUPUESTO):
    cfg = muestrear(r)
    intentos.append({**cfg, "cv_auc": evaluar(cfg)})
t_busq = time.perf_counter() - t0
tab = pd.DataFrame(intentos)
mejor = tab.loc[tab.cv_auc.idxmax()]
print(f"{PRESUPUESTO} pipelines evaluados en {t_busq:.1f} s | AUC en validacion cruzada: min {tab.cv_auc.min():.3f}, mediana {tab.cv_auc.median():.3f}, max {tab.cv_auc.max():.3f}")
print(f"Mejor configuracion: escalador={mejor.escalador}, k={mejor.k}, modelo={mejor.modelo}, hp={mejor.hp}")
print("AUC medio de CV por familia de modelo:", tab.groupby("modelo").cv_auc.mean().round(3).to_dict())

# --- Linea base y evaluacion honesta en el conjunto de PRUEBA (que la busqueda jamas vio) ---
base = Pipeline([("escalador", StandardScaler()), ("modelo", LogisticRegression(max_iter=500))])
cv_base = float(cross_val_score(base, Xa_ent, ya_ent, cv=3, scoring="roc_auc").mean())
auc_base = roc_auc_score(ya_pru, base.fit(Xa_ent, ya_ent).predict_proba(Xa_pru)[:, 1])
final = construir(mejor.to_dict()).fit(Xa_ent, ya_ent)
auc_mejor = roc_auc_score(ya_pru, final.predict_proba(Xa_pru)[:, 1])
print(f"\nLinea base (estandar + regresion logistica): CV {cv_base:.3f} | PRUEBA {auc_base:.3f}")
print(f"Mejor de la busqueda:                        CV {mejor.cv_auc:.3f} | PRUEBA {auc_mejor:.3f}  (brecha CV - prueba = {mejor.cv_auc - auc_mejor:+.3f})")
print("-> " + ("La busqueda mejoro la linea base en prueba. " if auc_mejor > auc_base else "La busqueda NO mejoro la linea base en prueba: ocurre, y hay que decirlo. ")
      + ("La CV del ganador sobreestimo (maldicion del ganador). " if mejor.cv_auc > auc_mejor else "Esta vez la CV del ganador NO sobreestimo: la prueba (450 filas) tambien tiene ruido. ")
      + "Solo la prueba intocada es la cifra honesta; la maldicion del ganador es un RIESGO, no una constante.")

# --- Curva 'mejor hasta ahora' promediada sobre 300 reordenamientos de los MISMOS intentos ---
g = np.random.default_rng(SEED)
curvas = np.array([np.maximum.accumulate(g.permutation(tab.cv_auc.to_numpy())) for _ in range(300)])
media = curvas.mean(axis=0)
assert np.all(np.diff(media) >= -1e-12) and abs(media[-1] - tab.cv_auc.max()) < 1e-12        # monotona y termina en el maximo global
assert tab.cv_auc.max() >= tab.cv_auc.median() and PRESUPUESTO == len(tab)
print(f"Con 1 intento se espera AUC {media[0]:.3f}; con 5, {media[4]:.3f}; con {PRESUPUESTO}, {media[-1]:.3f}: los rendimientos decrecen.")

fig, axs = plt.subplots(1, 2, figsize=(12, 4.2))
axs[0].plot(range(1, PRESUPUESTO + 1), media, "o-", color=PALETA[0], ms=3, label="mejor hasta ahora (promedio de 300 órdenes)")
axs[0].axhline(cv_base, color=PALETA[1], ls="--", label="línea base (CV)")
axs[0].set_xlabel("intentos (pipelines evaluados)"); axs[0].set_ylabel("AUC en validación cruzada"); axs[0].legend(fontsize=8); axs[0].set_title("Rendimientos decrecientes de la búsqueda aleatoria", fontsize=10, fontweight="bold")
for i, (m, d) in enumerate(tab.groupby("modelo").cv_auc):
    axs[1].scatter([i] * len(d) + np.random.default_rng(i).uniform(-0.12, 0.12, len(d)), d, s=22, color=PALETA[i], alpha=0.8)
axs[1].set_xticks(range(tab.modelo.nunique())); axs[1].set_xticklabels(sorted(tab.modelo.unique())); axs[1].set_ylabel("AUC en CV"); axs[1].set_title("Cada punto = un pipeline: la familia importa, y el ajuste fino también", fontsize=10, fontweight="bold")
plt.tight_layout(); plt.show()
````

## 3. Monitoreo de Deriva de Datos: el Índice de Estabilidad Poblacional (PSI) 📉

Un modelo en producción **se degrada** porque el mundo cambia. Hay dos tipos de cambio que conviene separar:

| Tipo | Qué cambia | ¿Se ve **sin etiquetas**? |
|---|---|---|
| **Deriva de covariables** (*data drift* / *covariate shift*) | La distribución de las **entradas** $P(X)$ | **Sí**: comparas las entradas de hoy con las de entrenamiento |
| **Deriva de concepto** (*concept drift*) | La **relación** $P(Y \mid X)$ | **No**: las entradas pueden verse idénticas; solo lo delatan las **etiquetas** (cuando llegan) |

El **PSI** compara la distribución de una variable en una **base** (entrenamiento) y en una ventana **actual**. Se dividen los valores en $B$ **cajones** (aquí, los **deciles de la base**) y, con $p_i$ y $q_i$ las proporciones de base y actual en el cajón $i$:

$$\text{PSI} \;=\; \sum_{i=1}^{B} (q_i - p_i)\,\ln\frac{q_i}{p_i} \;=\; D_{KL}(p \,\Vert\, q) + D_{KL}(q \,\Vert\, p)$$

es decir, la divergencia de Kullback-Leibler **simetrizada** (propiedad que verificaremos). Siempre es $\ge 0$ y vale $0$ si no hay cambio. Se acota $p_i, q_i \ge \varepsilon$ para evitar $\ln 0$. **Reglas de dedo** habituales (nacieron en el *scoring* de crédito y son **convenciones**, no leyes; *verificar*): **< 0,1** estable; **0,1 a 0,25** cambio moderado; **> 0,25** cambio **significativo**.

La celda: (1) implementa `psi`, la **verifica** (cero contra sí misma, simetría con KL de `scipy`, sensibilidad al tamaño del desplazamiento); (2) simula un **modelo en producción** con **8 semanas** en las que la variable $x_1$ se desplaza cada vez más (deriva de covariables) y un modelo **lineal** se va degradando porque la verdad es no lineal ($\sin x_1$); y (3) muestra el **punto ciego**: una semana **sin cambio en las entradas pero con otra regla** (deriva de concepto), donde el PSI dice «todo bien» y la exactitud se derrumba.

````python
# ============================================================
# PSI desde cero + modelo en produccion con deriva de covariables + deriva de concepto (el punto ciego del PSI)
# ============================================================
from scipy.stats import entropy
from sklearn.linear_model import LogisticRegression as LR

def psi(base, actual, cajones=10, eps=1e-4):
    """Indice de estabilidad poblacional con cajones por cuantiles de la BASE."""
    cortes = np.quantile(base, np.linspace(0, 1, cajones + 1)); cortes[0], cortes[-1] = -np.inf, np.inf
    p = np.histogram(base, cortes)[0] / len(base)
    q = np.histogram(actual, cortes)[0] / len(actual)
    p, q = np.clip(p, eps, None), np.clip(q, eps, None)
    return float(np.sum((q - p) * np.log(q / p)))

g = np.random.default_rng(SEED)
base_x = g.normal(0, 1, 5_000)
# --- Verificaciones ---
assert psi(base_x, base_x) == 0.0                                              # sin cambio, PSI = 0
mismo = psi(base_x, g.normal(0, 1, 5_000))
assert 0 <= mismo < 0.1                                                        # otra muestra de la MISMA distribucion: estable
desp = [psi(base_x, g.normal(d, 1, 5_000)) for d in [0.1, 0.25, 0.5, 1.0, 2.0]]
assert desp == sorted(desp) and desp[0] < 0.1 and desp[-1] > 0.25              # crece con el desplazamiento
# simetria: con los MISMOS cajones, PSI = KL(p||q) + KL(q||p)
cortes = np.quantile(base_x, np.linspace(0, 1, 11)); cortes[0], cortes[-1] = -np.inf, np.inf
act = g.normal(0.5, 1, 5_000)
p = np.clip(np.histogram(base_x, cortes)[0] / len(base_x), 1e-4, None); q = np.clip(np.histogram(act, cortes)[0] / len(act), 1e-4, None)
assert abs(psi(base_x, act) - (entropy(p, q) + entropy(q, p))) < 1e-12
print(f"PSI base vs. misma distribucion: {mismo:.4f} | con desplazamientos de 0.1, 0.25, 0.5, 1 y 2 desviaciones: {[round(d, 3) for d in desp]}")

# ----------------------------------------------------------------------------
# Modelo en produccion: y = 1 si sin(x1) + 0.5*x2 + ruido > 0; el modelo es LINEAL (aprende 'x1 grande -> 1'), la verdad no lo es
# ----------------------------------------------------------------------------
def datos(mu1, n, semilla, signo_x2=1.0):
    gg = np.random.default_rng(semilla)
    x1, x2 = gg.normal(mu1, 1, n), gg.normal(0, 1, n)
    yy = (np.sin(x1) + signo_x2 * 0.5 * x2 + 0.3 * gg.normal(size=n) > 0).astype(int)
    return np.c_[x1, x2], yy

X0, y0 = datos(0.0, 4_000, 100)
modelo = LR().fit(X0, y0)
punt0 = modelo.predict_proba(X0)[:, 1]
filas = []
for w in range(0, 9):
    Xw, yw = datos(0.4 * w, 2_000, 200 + w)                                    # semana w: x1 se desplaza 0.4 desv. por semana
    filas.append({"semana": w, "psi_x1": psi(X0[:, 0], Xw[:, 0]), "psi_x2": psi(X0[:, 1], Xw[:, 1]),
                  "psi_puntaje": psi(punt0, modelo.predict_proba(Xw)[:, 1]), "exactitud": float((modelo.predict(Xw) == yw).mean())})
tabla_semanal = pd.DataFrame(filas)
print("\n" + tabla_semanal.to_string(index=False, float_format=lambda v: f"{v:.3f}"))
assert tabla_semanal.psi_x1.is_monotonic_increasing and tabla_semanal.psi_x1.iloc[0] < 0.1
assert tabla_semanal.psi_x2.max() < 0.1                                        # x2 NO cambia: su PSI se mantiene bajo
assert tabla_semanal.exactitud.iloc[-1] < tabla_semanal.exactitud.iloc[0] - 0.15   # el modelo se degrada de verdad

# --- Punto ciego: deriva de concepto (mismas entradas, otra regla: x2 cambia de signo) ---
Xc, yc = datos(0.0, 2_000, 999, signo_x2=-1.0)
psi_c = {"x1": psi(X0[:, 0], Xc[:, 0]), "x2": psi(X0[:, 1], Xc[:, 1]), "puntaje": psi(punt0, modelo.predict_proba(Xc)[:, 1])}
acc_c = float((modelo.predict(Xc) == yc).mean())
print(f"\nDeriva de CONCEPTO: PSI x1={psi_c['x1']:.3f}, x2={psi_c['x2']:.3f}, puntaje={psi_c['puntaje']:.3f} (todo 'estable') | exactitud: {acc_c:.3f} frente a {tabla_semanal.exactitud.iloc[0]:.3f} en la semana 0")
assert max(psi_c.values()) < 0.1 and acc_c < tabla_semanal.exactitud.iloc[0] - 0.15
print("-> El PSI vigila las ENTRADAS; no ve un cambio de regla. Para eso hacen falta etiquetas (o proxies de ellas) y monitoreo del desempeno.")

fig, axs = plt.subplots(1, 2, figsize=(12, 4.2))
ax = axs[0]
ax.plot(tabla_semanal.semana, tabla_semanal.psi_x1, "o-", color=PALETA[1], label="PSI de $x_1$ (se desplaza)")
ax.plot(tabla_semanal.semana, tabla_semanal.psi_x2, "s-", color=PALETA[2], label="PSI de $x_2$ (no cambia)")
ax.axhline(0.1, color="#94a3b8", ls=":"); ax.axhline(0.25, color="#94a3b8", ls="--"); ax.text(0.1, 0.26, "0,25: cambio significativo", fontsize=8, color="#475569")
ax.set_xlabel("semana en producción"); ax.set_ylabel("PSI"); ax.legend(fontsize=8)
ax2 = ax.twinx(); ax2.plot(tabla_semanal.semana, tabla_semanal.exactitud, "^--", color=PALETA[0], label="exactitud del modelo"); ax2.set_ylabel("exactitud", color=PALETA[0]); ax2.set_ylim(0.4, 1.0); ax2.grid(False)
ax.set_title("El PSI sube antes (o a la vez) que la exactitud cae", fontsize=10, fontweight="bold")
ax = axs[1]
etq = ["PSI x1", "PSI x2", "PSI puntaje"]; vals = [psi_c["x1"], psi_c["x2"], psi_c["puntaje"]]
ax.bar(etq, vals, color=PALETA[4]); ax.axhline(0.1, color="#94a3b8", ls=":"); ax.set_ylim(0, 0.3); ax.set_ylabel("PSI")
ax.set_title(f"Deriva de concepto: PSI ≈ 0 pero la exactitud cae a {acc_c:.2f}", fontsize=10, fontweight="bold")
plt.tight_layout(); plt.show()
````

## 4. Recuperación Aumentada por Generación (RAG) a Nivel Conceptual 📚🔎

Un **modelo de lenguaje grande (LLM)** sabe lo que vio al entrenarse, **no conoce tus documentos** y puede **inventar** respuestas con total seguridad. **RAG** (Lewis et al., 2020 *(citado de memoria, verificar)*) separa el problema en dos pasos:

1. **Recuperar:** dada la pregunta, **buscar** los $k$ fragmentos más relevantes de **tu** base de conocimiento.
2. **Generar:** pasar al modelo la pregunta **más esos fragmentos** («responde **solo** con este contexto y cita la fuente»).

```text
pregunta ──► [recuperador] ──► k fragmentos ──► [prompt = contexto + pregunta] ──► [LLM] ──► respuesta con fuentes
              (índice sobre tus documentos)                                        (aquí NO se llama a ninguno)
```

La calidad de un sistema RAG depende **más de la recuperación que del modelo**: si el fragmento correcto no llega al *prompt*, ningún LLM lo arregla. Aquí construimos **solo el recuperador**, **desde cero** y con la técnica clásica **TF-IDF**: cada documento es un vector con un peso por palabra

$$\text{tfidf}(t, d) = \text{tf}(t, d)\cdot\text{idf}(t), \qquad \text{idf}(t) = \ln\frac{1 + n}{1 + \text{df}(t)} + 1,$$

con $\text{tf}$ el conteo del término en el documento, $n$ el número de documentos y $\text{df}(t)$ cuántos documentos contienen $t$ (la versión «suavizada» de `scikit-learn`); los vectores se normalizan en $L_2$, de modo que el **producto punto** es la **similitud coseno**. Palabras raras pesan más que las comunes. Verificamos nuestra implementación **contra `TfidfVectorizer`** y medimos la recuperación con *hit@1*, *hit@3* y el **rango recíproco medio** (MRR) sobre **14 documentos y 10 preguntas sintéticos** (frases cortas escritas para este cuaderno sobre temas del curso). **Ojo:** las preguntas se escribieron **reutilizando palabras de los documentos**, así que un acierto casi perfecto es un **techo optimista**, no lo que verías con usuarios reales.

> ⚠️ **TF-IDF es léxico:** solo encuentra **palabras que se repiten**. Si la pregunta usa **sinónimos** («caídas de discos» frente a «fallas»), no encuentra nada. Los **vectores semánticos** (*embeddings*) resuelven parte de eso, a cambio de un modelo y de una infraestructura de índices vectoriales (conceptual aquí).

````python
# ============================================================
# Recuperador TF-IDF desde cero (verificado contra scikit-learn) + 'prompt' RAG + el limite lexical. Sin LLM, sin red.
# ============================================================
from sklearn.feature_extraction.text import TfidfVectorizer

DOCS = [
    "HDFS divide los archivos en bloques grandes y guarda varias réplicas de cada bloque en distintos nodos para tolerar fallas.",
    "MapReduce procesa datos en dos fases: map transforma los registros en pares clave-valor y reduce agrega los valores de cada clave.",
    "Un data warehouse almacena datos estructurados con un esquema definido de antemano y se consulta con SQL para generar reportes.",
    "Un data lake guarda datos crudos en su formato original y aplica el esquema al leerlos, lo que se conoce como schema-on-read.",
    "El teorema CAP dice que ante una partición de red un sistema distribuido debe elegir entre consistencia y disponibilidad.",
    "PageRank mide la importancia de un nodo según los enlaces que recibe de otros nodos importantes.",
    "Una ventana tumbling agrupa los eventos de un flujo en intervalos fijos de tiempo que no se solapan.",
    "La marca de agua decide cuánto tiempo esperar a los datos tardíos antes de cerrar una ventana.",
    "El k-anonimato exige que cada registro sea indistinguible de al menos k-1 registros respecto a sus cuasi-identificadores.",
    "El control de acceso basado en roles asigna permisos a los roles y los roles a los usuarios.",
    "El aprendizaje federado entrena un modelo en varios clientes y solo comparte los parámetros, no los datos crudos.",
    "Una cadena de bloques enlaza cada bloque con el hash del anterior, de modo que modificar un bloque invalida los siguientes.",
    "El formato Parquet almacena los datos por columnas, lo que acelera las consultas analíticas que leen pocas columnas.",
    "El data leakage ocurre cuando información del futuro o del objetivo se filtra al entrenamiento y produce métricas optimistas.",
]
PREGUNTAS = [   # (pregunta, indice del documento correcto)
    ("¿Cómo se toleran las fallas guardando réplicas de los bloques?", 0),
    ("¿Qué hace la fase reduce con los valores de cada clave?", 1),
    ("¿Qué es guardar datos crudos y aplicar el esquema al leerlos?", 3),
    ("¿Qué debe elegir un sistema ante una partición de red?", 4),
    ("¿Cómo se mide la importancia de un nodo por los enlaces que recibe?", 5),
    ("¿Cuánto esperar a los datos tardíos antes de cerrar la ventana?", 7),
    ("¿Cómo entrenar un modelo sin compartir los datos crudos?", 10),
    ("¿Por qué modificar un bloque invalida los siguientes?", 11),
    ("¿Qué formato guarda por columnas para consultas analíticas?", 12),
    ("¿Qué asigna permisos a los roles y los roles a los usuarios?", 9),
]
PARADAS = set("de la el los las un una unos unas y o en que se su sus por para con del al lo como cada es son no a ante entre hace".split())

def tokenizar(texto):
    t = unicodedata.normalize("NFD", texto.lower())
    t = "".join(c for c in t if unicodedata.category(c) != "Mn")                # sin tildes
    return [w for w in re.findall(r"[a-z0-9]+", t) if w not in PARADAS and len(w) > 1]

class TfIdf:
    """TF-IDF suavizado y normalizado L2 (igual que scikit-learn con sus valores por defecto)."""
    def ajustar(self, docs):
        toks = [tokenizar(d) for d in docs]
        self.vocab = sorted({w for t in toks for w in t}); self.ind = {w: i for i, w in enumerate(self.vocab)}
        df = np.zeros(len(self.vocab))
        for t in toks:
            for w in set(t): df[self.ind[w]] += 1
        self.idf = np.log((1 + len(docs)) / (1 + df)) + 1
        return self
    def transformar(self, docs):
        M = np.zeros((len(docs), len(self.vocab)))
        for i, d in enumerate(docs):
            for w, c in Counter(tokenizar(d)).items():
                if w in self.ind: M[i, self.ind[w]] = c
        M *= self.idf
        nrm = np.linalg.norm(M, axis=1, keepdims=True); nrm[nrm == 0] = 1
        return M / nrm

tf = TfIdf().ajustar(DOCS); M_docs = tf.transformar(DOCS)
sk = TfidfVectorizer(tokenizer=tokenizar, lowercase=False, token_pattern=None); M_sk = sk.fit_transform(DOCS).toarray()
assert list(sk.get_feature_names_out()) == tf.vocab and np.allclose(M_docs, M_sk)       # IDENTICO a scikit-learn
assert np.allclose(np.linalg.norm(M_docs, axis=1), 1)
print(f"Vocabulario: {len(tf.vocab)} terminos | matriz {M_docs.shape} | coincide con TfidfVectorizer (allclose)")

def recuperar(pregunta, k=3):
    q = tf.transformar([pregunta])[0]
    sims = M_docs @ q                                                          # similitud coseno (vectores normalizados)
    orden = np.argsort(-sims, kind="stable")[:k]
    return [(int(i), float(sims[i])) for i in orden]

rangos = []
for preg, gold in PREGUNTAS:
    orden = [i for i, _ in recuperar(preg, k=len(DOCS))]
    rangos.append(orden.index(gold) + 1)
rangos = np.array(rangos)
hit1, hit3, mrr = (rangos == 1).mean(), (rangos <= 3).mean(), (1 / rangos).mean()
print(f"hit@1 = {hit1:.0%} | hit@3 = {hit3:.0%} | MRR = {mrr:.3f} | rangos del documento correcto: {rangos.tolist()}")
assert hit3 >= 0.9 and hit1 >= 0.7 and 0.8 <= mrr <= 1.0
print("(Techo optimista: las preguntas reutilizan palabras de los documentos. Observa abajo que el segundo fragmento del prompt, [doc 11], solo comparte la palabra 'bloques'.)")

# --- El 'prompt' de RAG (se muestra; NO se envia a ningun modelo) y una 'generacion' extractiva simulada ---
def construir_prompt(pregunta, k=2):
    ctx = "\n".join(f"[doc {i}] {DOCS[i]}" for i, _ in recuperar(pregunta, k))
    return f"Responde SOLO con el contexto y cita el documento.\n\nContexto:\n{ctx}\n\nPregunta: {pregunta}\nRespuesta:"
print("\n--- Prompt armado para la primera pregunta (no se envia a ningun LLM) ---\n" + construir_prompt(PREGUNTAS[0][0]))
i_top = recuperar(PREGUNTAS[6][0], 1)[0][0]
print(f"\n--- 'Respuesta' extractiva SIMULADA (sin LLM) a '{PREGUNTAS[6][0]}' ---\nSegun [doc {i_top}]: {DOCS[i_top]}")

# --- El limite lexical: una pregunta con sinonimos no comparte NINGUNA palabra util con los documentos ---
sin = "¿Qué pasa con la tolerancia a caídas de discos?"
sims_sin = M_docs @ tf.transformar([sin])[0]
print(f"\nPregunta con sinonimos: '{sin}' -> similitud maxima {sims_sin.max():.3f}: el recuperador lexical devuelve basura (todos empatan en 0).")
assert sims_sin.max() == 0.0

fig, ax = plt.subplots(figsize=(10, 3.5))
S = np.array([M_docs @ tf.transformar([p])[0] for p, _ in PREGUNTAS])
im = ax.imshow(S, cmap="Blues", aspect="auto")
for i, (_, gold) in enumerate(PREGUNTAS): ax.scatter(gold, i, marker="o", s=60, facecolors="none", edgecolors=PALETA[1], lw=1.6)
ax.set_xlabel("documento (círculo rojo = el correcto)"); ax.set_ylabel("pregunta"); ax.set_xticks(range(len(DOCS))); ax.set_yticks(range(len(PREGUNTAS)))
ax.set_title("Similitud coseno TF-IDF entre cada pregunta y cada documento", fontsize=10, fontweight="bold"); ax.grid(False)
plt.colorbar(im, ax=ax, label="similitud"); plt.tight_layout(); plt.show()
````

## 5. Aprendizaje Federado: Promedio Federado (FedAvg) 🤝

A veces los datos **no pueden juntarse** (hospitales, bancos, teléfonos, o simplemente ley y confianza). El **aprendizaje federado** (McMahan et al., 2017 *(citado de memoria, verificar)*) invierte el flujo: **el modelo viaja a los datos, no los datos al modelo**. En cada **ronda**:

1. El servidor envía los parámetros globales $w$ a los clientes.
2. Cada cliente $k$ entrena **localmente** $E$ épocas con **sus** datos y obtiene $w_k$.
3. El servidor **promedia ponderando por tamaño**: $\;w \leftarrow \sum_k \frac{n_k}{n}\, w_k\;$ (**FedAvg**).

Solo viajan **parámetros**; los datos crudos **no salen** del cliente. Tres hechos que comprobamos con regresión logística escrita **desde cero** en NumPy:

- **Equivalencia exacta:** con **una sola época** de gradiente completo por ronda, FedAvg ponderado **es idéntico** al descenso de gradiente **centralizado** (el promedio ponderado de gradientes locales es el gradiente global).
- **Caso IID:** con varias épocas locales y datos repartidos **al azar**, FedAvg queda **cerca** del centralizado.
- **Caso no IID** (cada cliente ve **sobre todo una clase**): con **pocas** épocas locales casi no se nota (la regresión logística es convexa), pero con **muchas** cada cliente se acerca a **su** óptimo sesgado (*deriva de clientes*) y el promedio **empeora**; y un cliente aislado queda **muy lejos** del modelo global.

> ⚠️ **Federado no es privado por sí solo:** los parámetros y gradientes pueden **filtrar información** de los datos locales; las garantías formales exigen técnicas adicionales como **agregación segura** y **privacidad diferencial** (conceptual aquí; *verificar* el alcance de cada una). Además cada ronda cuesta **comunicación**: $K \times d$ números por ronda.

````python
# ============================================================
# FedAvg desde cero (regresion logistica) vs. centralizado: IID, no IID y clientes aislados
# ============================================================
def sigmoide(z): return 1 / (1 + np.exp(-z))

def grad(w, X, yy, l2): return X.T @ (sigmoide(X @ w) - yy) / len(yy) + l2 * w
def perdida(w, X, yy):
    p = np.clip(sigmoide(X @ w), 1e-12, 1 - 1e-12)
    return float(-np.mean(yy * np.log(p) + (1 - yy) * np.log(1 - p)))
def exactitud(w, X, yy): return float(((sigmoide(X @ w) > 0.5) == yy).mean())

def entrenar_central(X, yy, pasos, lr, l2, w0):
    w = w0.copy()
    for _ in range(pasos): w -= lr * grad(w, X, yy, l2)
    return w

def fedavg(clientes, rondas, epocas, lr, l2, w0, eval_fn=None):
    """Cada ronda: cada cliente da 'epocas' pasos de gradiente completo desde w; el servidor promedia ponderando por n_k."""
    w, n_tot, hist = w0.copy(), sum(len(yy) for _, yy in clientes), []
    for _ in range(rondas):
        w = sum(len(yy) * entrenar_central(X, yy, epocas, lr, l2, w) for X, yy in clientes) / n_tot
        if eval_fn: hist.append(eval_fn(w))
    return w, hist

# --- Datos sinteticos: 10 variables + sesgo; 4 000 de entrenamiento repartidos en 5 clientes y 2 000 de prueba ---
g = np.random.default_rng(SEED)
D, K_CLI = 10, 5
w_verd = g.normal(0, 1, D + 1)
def generar(n):
    X = np.c_[g.normal(0, 1, (n, D)), np.ones(n)]
    return X, (g.random(n) < sigmoide(X @ w_verd)).astype(float)
X_tr, y_tr = generar(4_000); X_te, y_te = generar(2_000)
L2, LR_, RONDAS, EPOCAS = 1e-3, 0.5, 40, 5
w0 = np.zeros(D + 1)

def partir_iid(X, yy):
    idx = np.random.default_rng(1).permutation(len(yy)); return [(X[i], yy[i]) for i in np.array_split(idx, K_CLI)]
def partir_no_iid(X, yy, alfa=0.3, semilla=2):
    """Reparto por Dirichlet(alfa) POR CLASE: cuanto menor alfa, mas sesgado hacia una clase tiene cada cliente."""
    gg = np.random.default_rng(semilla)
    while True:
        parte = [[] for _ in range(K_CLI)]
        for c in (0.0, 1.0):
            ix = gg.permutation(np.where(yy == c)[0]); prop = gg.dirichlet([alfa] * K_CLI)
            for k, trozo in enumerate(np.split(ix, (np.cumsum(prop)[:-1] * len(ix)).astype(int))): parte[k] += trozo.tolist()
        if min(len(p) for p in parte) >= 50: return [(X[p], yy[p]) for p in parte]

cli_iid, cli_no = partir_iid(X_tr, y_tr), partir_no_iid(X_tr, y_tr)
assert sum(len(yy) for _, yy in cli_iid) == sum(len(yy) for _, yy in cli_no) == len(y_tr)
print("Clientes IID    (n, % de clase 1):", [(len(yy), f"{yy.mean():.0%}") for _, yy in cli_iid])
print("Clientes no IID (n, % de clase 1):", [(len(yy), f"{yy.mean():.0%}") for _, yy in cli_no])

# --- (1) Equivalencia EXACTA: 1 epoca de gradiente completo por ronda == descenso centralizado ---
pasos = 30
w_fed1, _ = fedavg(cli_iid, pasos, 1, LR_, L2, w0)
w_cen1 = entrenar_central(X_tr, y_tr, pasos, LR_, L2, w0)
w_fed1n, _ = fedavg(cli_no, pasos, 1, LR_, L2, w0)                              # tambien con datos NO IID, porque el gradiente global se descompone igual
assert np.allclose(w_fed1, w_cen1, atol=1e-10) and np.allclose(w_fed1n, w_cen1, atol=1e-10)
print(f"\nEquivalencia exacta (1 epoca, {pasos} rondas): max |w_FedAvg - w_central| = {np.abs(w_fed1 - w_cen1).max():.1e} (IID) y {np.abs(w_fed1n - w_cen1).max():.1e} (no IID)")

# --- (2) Varias epocas locales: IID frente a no IID, contra centralizado con los MISMOS pasos de gradiente ---
ev = lambda w: (perdida(w, X_te, y_te), exactitud(w, X_te, y_te))
w_cen = entrenar_central(X_tr, y_tr, RONDAS * EPOCAS, LR_, L2, w0)
w_iid, h_iid = fedavg(cli_iid, RONDAS, EPOCAS, LR_, L2, w0, ev)
w_no, h_no = fedavg(cli_no, RONDAS, EPOCAS, LR_, L2, w0, ev)
locales_iid = [ev(entrenar_central(X, yy, RONDAS * EPOCAS, LR_, L2, w0)) for X, yy in cli_iid]
locales_no = [ev(entrenar_central(X, yy, RONDAS * EPOCAS, LR_, L2, w0)) for X, yy in cli_no]
fila = lambda nom, p, a: {"modelo": nom, "perdida en prueba": p, "exactitud en prueba": a}
tab = pd.DataFrame([fila("centralizado (todos los datos juntos)", *ev(w_cen)), fila("FedAvg, datos IID", *ev(w_iid)), fila("FedAvg, datos no IID", *ev(w_no)),
                    fila("cliente aislado (IID, promedio de 5)", *np.mean(locales_iid, axis=0)), fila("cliente aislado (no IID, promedio de 5)", *np.mean(locales_no, axis=0))])
print("\n" + tab.to_string(index=False, float_format=lambda v: f"{v:.4f}"))
acc = tab["exactitud en prueba"].to_numpy(); perd = tab["perdida en prueba"].to_numpy()
assert abs(acc[1] - acc[0]) <= 0.03                                            # IID: FedAvg ~ centralizado
assert acc[2] > acc[4] and perd[2] < perd[4]                                   # el promedio federado le gana a un cliente sesgado y solo
print(f"-> Con E = {EPOCAS} epocas locales, FedAvg IID queda a {abs(acc[1] - acc[0]):.3f} de exactitud del centralizado; con datos no IID la perdida casi no cambia ({perd[2]:.4f} frente a {perd[1]:.4f}): regresion logistica es convexa y 5 epocas no bastan para que los clientes se alejen. Un cliente solo (sesgado) llega a {perd[4]:.4f}.")

# --- (3) Cuantas epocas locales? Mismos 200 pasos de gradiente por cliente, repartidos en menos rondas (menos comunicacion) ---
filas = []
for E in [1, 5, 20, 100]:
    R = 200 // E
    li, ln = ev(fedavg(cli_iid, R, E, LR_, L2, w0)[0]), ev(fedavg(cli_no, R, E, LR_, L2, w0)[0])
    filas.append({"epocas locales E": E, "rondas": R, "perdida IID": li[0], "perdida no IID": ln[0], "exactitud IID": li[1], "exactitud no IID": ln[1]})
barrido = pd.DataFrame(filas)
print("\n" + barrido.to_string(index=False, float_format=lambda v: f"{v:.4f}"))
assert abs(barrido["perdida IID"].iloc[0] - perd[0]) < 1e-9 and abs(barrido["perdida no IID"].iloc[0] - perd[0]) < 1e-9     # con E = 1 ambos son el centralizado
assert barrido["perdida no IID"].iloc[-1] > barrido["perdida no IID"].iloc[0] + 0.005                                        # E grande: la deriva de clientes cuesta
assert barrido["perdida no IID"].iloc[-1] > barrido["perdida IID"].iloc[-1] + 0.005                                          # y cuesta mas con datos no IID
print("-> El costo del no IID aparece cuando los clientes entrenan MUCHO por su cuenta: cada uno se acerca a SU optimo sesgado (deriva de clientes) y el promedio empeora. Mas epocas = menos comunicacion, pero peor modelo.")
print(f"   Comunicacion: {RONDAS} rondas x {K_CLI} clientes x {D + 1} parametros x 2 sentidos = {RONDAS * K_CLI * (D + 1) * 2:,} numeros; el centralizado movio {X_tr.size:,} (los datos).")

fig, axs = plt.subplots(1, 2, figsize=(12, 4.1))
for h, nom, c in [(h_iid, "FedAvg IID (E = 5)", PALETA[0]), (h_no, "FedAvg no IID (E = 5)", PALETA[1])]:
    axs[0].plot(range(1, RONDAS + 1), [p for p, _ in h], color=c, label=nom)
axs[0].axhline(perd[0], color="k", ls="--", label="centralizado")
axs[0].set_xlabel("ronda"); axs[0].set_ylabel("pérdida (log-loss) en prueba"); axs[0].legend(); axs[0].set_title("Con E = 5, IID y no IID convergen casi igual", fontsize=10, fontweight="bold")
axs[1].plot(barrido["epocas locales E"], barrido["perdida IID"], "o-", color=PALETA[0], label="IID")
axs[1].plot(barrido["epocas locales E"], barrido["perdida no IID"], "s-", color=PALETA[1], label="no IID")
axs[1].set_xscale("log"); axs[1].set_xlabel("épocas locales E (mismos 200 pasos por cliente)"); axs[1].set_ylabel("pérdida en prueba"); axs[1].legend()
axs[1].set_title("Más épocas locales = menos comunicación, pero el no IID se resiente", fontsize=10, fontweight="bold")
plt.tight_layout(); plt.show()
````

## 6. Blockchain: Cadena de Hashes, Prueba de Trabajo y Árbol de Merkle ⛓️

Una **cadena de bloques** es, en esencia, un **registro de solo anexar y a prueba de manipulación** construido con tres ideas de criptografía elemental:

1. **Función *hash* criptográfica** (aquí SHA-256 de `hashlib`): convierte cualquier dato en una huella de 256 bits; **cambiar un solo bit** del dato cambia la huella por completo, y es inviable fabricar un dato con una huella dada.
2. **Cadena:** cada bloque guarda el *hash* del anterior. Modificar un bloque cambia su *hash*, que ya no coincide con el que guarda el siguiente: **la manipulación se detecta** en el eslabón roto. Para «reparar» hay que **rehacer todos los bloques posteriores**.
3. **Prueba de trabajo** (*proof of work*): para que rehacer sea **caro**, un bloque solo es válido si su *hash* empieza con $d$ ceros (en hexadecimal). Como el *hash* es impredecible, hay que probar valores del contador *nonce* hasta acertar: en promedio $16^{d}$ intentos. Verificar cuesta **un solo** *hash*.

Y un **árbol de Merkle**: los *hashes* de las transacciones se **emparejan y vuelven a hacerse hash** nivel a nivel hasta una única **raíz** que se guarda en el bloque. Con la raíz se puede **probar que una transacción está incluida** entregando solo $\lceil \log_2 n \rceil$ *hashes* (la «prueba de inclusión»), **sin revelar las demás**.

#### Su uso en auditoría y gobernanza (y sus límites)

| Aspecto | Lo que sí aporta | Lo que **no** resuelve (límites reales) |
|---|---|---|
| **Auditoría** | Un **registro de accesos o cambios** que **nadie puede alterar en silencio**: la manipulación **se nota**. Con la raíz publicada, un tercero puede verificar una entrada con una prueba de Merkle | **Detecta** la manipulación; **no la impide** ni dice si el dato era **verdadero** al entrar («basura entra, basura queda inmutable»). Un registro con firma y *hash* encadenado **sin** consenso distribuido ya da buena parte del beneficio en una sola organización |
| **Gobernanza / linaje** | Fijar *hashes* de conjuntos de datos y de versiones de modelos para probar **qué se usó y cuándo** | Es **un ingrediente** del linaje, no el linaje: faltan metadatos, responsables y políticas (Módulo 04) |
| **Rendimiento** | — | Un consenso distribuido exige que **muchos nodos** acuerden: el **rendimiento es mucho menor** que el de una base de datos centralizada (en cadenas públicas como Bitcoin, unas pocas transacciones por segundo, frente a miles en sistemas de pago centralizados; *aproximado, verificar*) |
| **Energía** | — | La prueba de trabajo **gasta computación a propósito**: el costo crece con la dificultad (lo medimos abajo en intentos). Algunas redes migraron a otros mecanismos de consenso (p. ej. Ethereum a prueba de participación, 2022; *verificar*) |
| **Derecho al olvido** | — | Lo **inmutable choca** con el derecho a **suprimir** datos personales (el RGPD europeo, art. 17, y la Ley 1581 de 2012 en Colombia, que reconoce la supresión; *verificar el texto vigente y consultar asesoría jurídica*). **Práctica común:** guardar en la cadena solo un ***hash*** y los datos personales **fuera**; pero un *hash* de un dato **fácil de adivinar** (un código de pocos dígitos) se **revierte por fuerza bruta** (lo demostramos), y si el *hash* sigue enlazado a una persona, es **debatible** que cuente como supresión |

La celda implementa **bloques con raíz de Merkle y prueba de trabajo**, el **verificador de cadenas**, la **prueba de inclusión**, la **detección de manipulación** (y cuánto cuesta rehacer la cadena), el **costo de la prueba de trabajo medido en intentos** y la **demostración de fuerza bruta sobre un *hash* sin sal**. Todo con *asserts*.

````python
# ============================================================
# Blockchain desde cero: hash, Merkle (prueba de inclusion), bloques con prueba de trabajo, deteccion de manipulacion
# ============================================================
import copy
import os

def sha(x):
    return hashlib.sha256(x.encode() if isinstance(x, str) else x).hexdigest()

# ------------------------- Arbol de Merkle -------------------------
def merkle_niveles(hojas):
    nivel = [sha(h) for h in hojas]; niveles = [nivel]
    while len(nivel) > 1:
        if len(nivel) % 2: nivel = nivel + [nivel[-1]]; niveles[-1] = nivel         # si es impar, se duplica el ultimo (convencion de Bitcoin)
        nivel = [sha(nivel[i] + nivel[i + 1]) for i in range(0, len(nivel), 2)]; niveles.append(nivel)
    return niveles

def raiz_merkle(hojas): return merkle_niveles(hojas)[-1][0]

def prueba_inclusion(hojas, i):
    """Lista de (hash_hermano, lado) desde la hoja i hasta la raiz."""
    niveles, idx, prueba = merkle_niveles(hojas), i, []
    for nivel in niveles[:-1]:
        prueba.append((nivel[idx ^ 1], "der" if idx % 2 == 0 else "izq")); idx //= 2
    return prueba

def verificar_inclusion(hoja, prueba, raiz):
    h = sha(hoja)
    for hermano, lado in prueba: h = sha(h + hermano) if lado == "der" else sha(hermano + h)
    return h == raiz

tx = [f"tx{i}: usuario=u{i % 3};accion={['lectura', 'escritura', 'borrado'][i % 3]};tabla=ventas" for i in range(8)]
raiz = raiz_merkle(tx)
assert raiz_merkle(tx[:1]) == sha(tx[0]) and len(raiz) == 64
for i in range(len(tx)):
    pr = prueba_inclusion(tx, i)
    assert len(pr) == 3 and verificar_inclusion(tx[i], pr, raiz)                  # log2(8) = 3 hashes bastan
    assert not verificar_inclusion(tx[i] + " ", pr, raiz)                          # cambiar un caracter rompe la prueba
tx_mod = tx.copy(); tx_mod[5] = tx_mod[5].replace("borrado", "lectura")
assert raiz_merkle(tx_mod) != raiz                                                 # cualquier cambio altera la raiz
for n_tx in [1, 2, 3, 5, 7, 8, 9, 100]:                                            # incluye cantidades impares
    hs = [f"t{i}" for i in range(n_tx)]; rz = raiz_merkle(hs)
    assert all(verificar_inclusion(hs[i], prueba_inclusion(hs, i), rz) for i in range(n_tx))
    assert len(prueba_inclusion(hs, 0)) == math.ceil(math.log2(n_tx)) if n_tx > 1 else True
print(f"Merkle: raiz de 8 transacciones = {raiz[:16]}... | prueba de inclusion = 3 hashes (ceil(log2 8)); 100 transacciones -> {len(prueba_inclusion([f't{i}' for i in range(100)], 0))} hashes")

# ------------------------- Bloques y prueba de trabajo -------------------------
def hash_bloque(indice, prev, raiz, nonce): return sha(f"{indice}|{prev}|{raiz}|{nonce}")

def crear_bloque(indice, prev, datos, dif):
    """Mina un bloque: busca el nonce con hash que empiece con 'dif' ceros hexadecimales. Devuelve el bloque (con n.º de intentos)."""
    rz, nonce, objetivo = raiz_merkle(datos), 0, "0" * dif
    while not hash_bloque(indice, prev, rz, nonce).startswith(objetivo): nonce += 1
    return {"indice": indice, "prev": prev, "datos": list(datos), "raiz": rz, "nonce": nonce, "hash": hash_bloque(indice, prev, rz, nonce), "intentos": nonce + 1}

def verificar_cadena(cadena, dif):
    """Indice del PRIMER bloque invalido, o None si la cadena es valida (un solo hash por bloque: verificar es barato)."""
    for i, b in enumerate(cadena):
        prev_esperado = "0" * 64 if i == 0 else cadena[i - 1]["hash"]
        if (b["prev"] != prev_esperado or b["raiz"] != raiz_merkle(b["datos"])
                or b["hash"] != hash_bloque(b["indice"], b["prev"], b["raiz"], b["nonce"]) or not b["hash"].startswith("0" * dif)):
            return i
    return None

DIF = 3
cadena_demo, prev = [], "0" * 64
for i in range(6):
    datos_i = [f"bloque{i}-tx{j}: usuario=u{(i + j) % 4};accion={['lectura', 'escritura'][j % 2]};tabla=ventas" for j in range(4)]
    cadena_demo.append(crear_bloque(i, prev, datos_i, DIF)); prev = cadena_demo[-1]["hash"]
assert verificar_cadena(cadena_demo, DIF) is None
assert all(b["hash"].startswith("000") for b in cadena_demo)
print(f"Cadena de {len(cadena_demo)} bloques (dificultad {DIF}): intentos por bloque = {[b['intentos'] for b in cadena_demo]} | valida: {verificar_cadena(cadena_demo, DIF) is None}")

# ------------------------- Deteccion de manipulacion -------------------------
manip = copy.deepcopy(cadena_demo)
manip[2]["datos"][1] = manip[2]["datos"][1].replace("escritura", "borrado")        # alguien reescribe una entrada de auditoria del bloque 2
assert verificar_cadena(manip, DIF) == 2                                           # se detecta en el bloque manipulado
# ...y si ademas recalcula el hash de ese bloque SIN minar de nuevo, falla la prueba de trabajo o el enlace del siguiente
manip[2]["raiz"] = raiz_merkle(manip[2]["datos"]); manip[2]["hash"] = hash_bloque(2, manip[2]["prev"], manip[2]["raiz"], manip[2]["nonce"])
assert verificar_cadena(manip, DIF) in (2, 3)
# Para ocultar el cambio hay que REHACER (minar de nuevo) el bloque 2 y TODOS los posteriores
rehecha, prev, costo_ataque = manip[:2], manip[1]["hash"], 0
for b in manip[2:]:
    nuevo = crear_bloque(b["indice"], prev, b["datos"], DIF); rehecha.append(nuevo); prev = nuevo["hash"]; costo_ataque += nuevo["intentos"]
assert verificar_cadena(rehecha, DIF) is None                                      # el atacante lo logra... pagando el trabajo
costo_honesto = sum(b["intentos"] for b in cadena_demo)
print(f"Manipular el bloque 2 se detecta (primer bloque invalido: 2). Ocultarlo exige re-minar {len(manip) - 2} bloques: {costo_ataque:,} intentos "
      f"(construir toda la cadena costo {costo_honesto:,}); verificar cuesta {len(cadena_demo)} hashes.")
print("-> En una red real el atacante, ademas, tendria que ganarle al resto de la red (idea del ataque del 51 %), no solo a si mismo.")

# ------------------------- Costo de la prueba de trabajo: intentos por dificultad -------------------------
dificultades = [1, 2, 3, 4]
medias = []
for d in dificultades:
    n_b = 80 if d <= 2 else (40 if d == 3 else 16)
    medias.append(np.mean([crear_bloque(i, sha(f"semilla{d}-{i}"), [f"x{i}"], d)["intentos"] for i in range(n_b)]))
for d, m in zip(dificultades, medias):
    assert 0.35 * 16 ** d < m < 2.8 * 16 ** d                                      # ~16^d intentos, con la variabilidad de una muestra pequena
assert medias == sorted(medias)
print("Intentos medios por bloque segun la dificultad:", {d: round(m) for d, m in zip(dificultades, medias)}, "| esperado 16^d:", {d: 16 ** d for d in dificultades})
print("Cada cero hexadecimal mas multiplica el trabajo x16 (4 bits); verificar sigue costando UN hash.")

# ------------------------- 'Guardar solo el hash': por que un hash sin sal de un dato facil de adivinar NO es anonimo -------------------------
codigo = "48213"                                                                   # codigo de estudiante inventado, de 5 digitos
en_cadena = sha(codigo)                                                            # lo que quedaria 'inmutable' en la cadena
t0 = time.perf_counter()
recuperado = next(f"{c:05d}" for c in range(100_000) if sha(f"{c:05d}") == en_cadena)
t_fb = time.perf_counter() - t0
sal = os.urandom(16).hex()                                                         # sal aleatoria guardada FUERA de la cadena
en_cadena_sal = sha(sal + codigo)
sin_sal = next((f"{c:05d}" for c in range(100_000) if sha(f"{c:05d}") == en_cadena_sal), None)       # el atacante no conoce la sal
con_sal = next((f"{c:05d}" for c in range(100_000) if sha(sal + f"{c:05d}") == en_cadena_sal), None)  # si la sal se filtra, vuelve a ser trivial
assert recuperado == codigo and sin_sal is None and con_sal == codigo
print(f"\nFuerza bruta sobre sha256 SIN sal: codigo recuperado en {t_fb * 1000:.0f} ms ({recuperado}). Con sal desconocida: {sin_sal}. Si la sal se filtra: {con_sal}.")
print("-> Un hash de un dato de poco espacio se revierte; con sal guardada fuera y BORRABLE se puede 'olvidar'... pero que eso cumpla la ley es una pregunta juridica abierta (verificar).")

# ------------------------- Graficos -------------------------
fig, axs = plt.subplots(2, 1, figsize=(12, 4.6))
def dibujar_cadena(ax, cad, malos, titulo):
    ax.axis("off"); ax.set_xlim(-0.3, 6 * 2.05); ax.set_ylim(0, 2)
    for i, b in enumerate(cad):
        x0 = i * 2.05; c = PALETA[1] if i in malos else PALETA[0]
        ax.add_patch(FancyBboxPatch((x0, 0.25), 1.75, 1.35, boxstyle="round,pad=0.03", fc="#fee2e2" if i in malos else "#dbeafe", ec=c, lw=1.6))
        ax.text(x0 + 0.875, 1.4, f"Bloque {i}", ha="center", fontweight="bold", fontsize=9)
        ax.text(x0 + 0.875, 1.05, f"prev {b['prev'][:6]}…", ha="center", fontsize=7.5, family="monospace")
        ax.text(x0 + 0.875, 0.78, f"raíz {b['raiz'][:6]}…", ha="center", fontsize=7.5, family="monospace")
        ax.text(x0 + 0.875, 0.5, f"hash {b['hash'][:6]}…", ha="center", fontsize=7.5, family="monospace", color=c)
        if i < len(cad) - 1: ax.annotate("", xy=(x0 + 2.0, 0.9), xytext=(x0 + 1.78, 0.9), arrowprops=dict(arrowstyle="->", lw=1.5))
    ax.set_title(titulo, fontsize=10, fontweight="bold")
dibujar_cadena(axs[0], cadena_demo, set(), "Cadena válida: cada bloque guarda el hash del anterior (todos empiezan con 000)")
dibujar_cadena(axs[1], manip, {verificar_cadena(manip, DIF)}, "Tras manipular el bloque 2: el verificador señala el primer bloque inválido (en rojo)")
plt.tight_layout(); plt.show()

fig, axs = plt.subplots(1, 2, figsize=(12, 4.3))
ax = axs[0]; ax.axis("off"); ax.set_xlim(0, 8); ax.set_ylim(-0.6, 4.4)
niveles = merkle_niveles(tx); i_hoja = 5; camino = [(0, i_hoja)]
for L in range(1, len(niveles)): camino.append((L, camino[-1][1] // 2))
hermanos = [(L, idx ^ 1) for L, idx in camino[:-1]]
for L, nivel in enumerate(niveles):
    paso = 8 / len(nivel)
    for j, h in enumerate(nivel):
        x = paso * (j + 0.5); y = L * 1.1
        en_camino, es_hermano = (L, j) in camino, (L, j) in hermanos
        fc = "#fecaca" if en_camino else ("#fde68a" if es_hermano else "#e2e8f0")
        ax.add_patch(FancyBboxPatch((x - 0.38, y - 0.2), 0.76, 0.4, boxstyle="round,pad=0.02", fc=fc, ec="#475569"))
        ax.text(x, y, h[:4], ha="center", va="center", fontsize=7, family="monospace")
        if L > 0:
            for hijo in (2 * j, 2 * j + 1):
                px = (8 / len(niveles[L - 1])) * (hijo + 0.5); ax.plot([x, px], [y - 0.2, (L - 1) * 1.1 + 0.2], color="#94a3b8", lw=0.7, zorder=0)
ax.text(4, 4.2, "raíz de Merkle", ha="center", fontsize=9, fontweight="bold")
ax.set_title("Prueba de inclusión de la tx5: rojo = camino; amarillo = los 3 hashes que se entregan", fontsize=9, fontweight="bold")
ax = axs[1]
ax.semilogy(dificultades, medias, "o-", color=PALETA[1], label="intentos medios medidos")
ax.semilogy(dificultades, [16 ** d for d in dificultades], "k--", label="$16^d$ (esperado)")
ax.set_xticks(dificultades); ax.set_xlabel("dificultad $d$ (ceros hexadecimales)"); ax.set_ylabel("intentos por bloque (escala log)"); ax.legend()
ax.set_title("Prueba de trabajo: cada cero multiplica el trabajo por 16", fontsize=10, fontweight="bold")
plt.tight_layout(); plt.show()
````

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Detección de anomalías** | **Ninguno gana en todo:** la **z robusta** (desde cero) vio el **tipo A** (monto enorme) y fue ciega al B (madrugada **y** lejos); `StandardScaler` + `IsolationForest` vio el **tipo B** pero apenas el A (una variable rara entre cinco se diluye). La **unión** cubrió ambos. Con pocas etiquetas, mide con **precisión promedio** y **exhaustividad por tipo** |
| **AutoML** | Es **búsqueda aleatoria** sobre *pipelines* evaluados con validación cruzada **dentro del *pipeline*** (sin fuga). Rinde **cada vez menos** por intento y el **ganador está sobreestimado** en CV: solo el conjunto de **prueba** intocado es la cifra honesta (y a veces **no supera** la línea base) |
| **Deriva y PSI** | $\text{PSI} = \sum (q_i - p_i)\ln(q_i/p_i)$ = KL simetrizada (verificado); **0,1** y **0,25** son convenciones. Vio la **deriva de covariables** antes del daño, pero fue **ciego a la deriva de concepto** (exactitud cayó con PSI $\approx 0$) |
| **RAG** | Recuperar + generar. TF-IDF desde cero **idéntico a `scikit-learn`**; sirve con **vocabulario compartido** y falla con **sinónimos** (similitud 0). Aquí no se llamó a ningún LLM: el *prompt* solo se muestra |
| **Aprendizaje federado** | FedAvg ponderado con 1 época **es** el gradiente centralizado (error $<10^{-10}$); con varias épocas y datos IID queda **cerca** del centralizado; con datos **no IID** el costo **solo aparece con muchas épocas locales** (deriva de clientes); un cliente aislado sesgado, mucho peor. **No es privado por sí solo** |
| **Blockchain** | Cadena de *hashes* + prueba de trabajo ($\approx 16^d$ intentos medidos) + Merkle (prueba de $\lceil\log_2 n\rceil$ *hashes*): la manipulación **se detecta** y ocultarla exige **rehacer todos los bloques posteriores** |
| **Límites** | **Detecta, no impide** ni garantiza la verdad del dato; **menor rendimiento** y **costo de energía** por consenso/prueba de trabajo; **tensión con el derecho al olvido**: un *hash* sin sal de un dato adivinable se **revierte en milisegundos** |
