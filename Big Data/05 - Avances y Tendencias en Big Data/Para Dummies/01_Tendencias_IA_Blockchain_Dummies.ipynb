## 1. La Báscula que se Descalibra ⚖️

En la bodega, los **bultos de papa** deben pesar **unos 10 kg** (con una variación normal de unos 0,3 kg hacia arriba o hacia abajo). Los datos son **inventados**. Hay **dos alarmas**:

1. **Alarma de bulto raro:** suena si un bulto se aleja del peso normal **más de 4 veces la variación habitual**.
2. **Alarma de corrimiento:** suena si el **promedio de hoy** se aleja del promedio habitual **más de lo que el azar suele mover un promedio**.

Probamos tres días: un **día normal**, un **día con un bulto mal pesado** (uno de 14 kg) y un **mes después con la báscula descalibrada** (todo pesa 0,4 kg de más).

````python
import random
import statistics as st

import matplotlib.pyplot as plt

random.seed(42)
habitual = [random.gauss(10.0, 0.3) for _ in range(200)]                          # pesos "de siempre" (inventados)
media_h, desv_h = st.mean(habitual), st.stdev(habitual)


def alarma_bulto_raro(pesos):
    """Bultos que se alejan del peso habitual mas de 4 veces la variacion normal."""
    return [round(p, 2) for p in pesos if abs(p - media_h) > 4 * desv_h]


def alarma_corrimiento(pesos):
    """True si el promedio de hoy se aleja mas de lo que el azar mueve un promedio (3 'errores estandar')."""
    return abs(st.mean(pesos) - media_h) > 3 * desv_h / len(pesos) ** 0.5


dia_normal = [random.gauss(10.0, 0.3) for _ in range(40)]
dia_bulto = [random.gauss(10.0, 0.3) for _ in range(39)] + [14.0]
dia_descalibrado = [random.gauss(10.4, 0.3) for _ in range(40)]                     # todo pesa 0,4 kg de mas

resultados = {}
for nombre, pesos in [("día normal", dia_normal), ("un bulto de 14 kg", dia_bulto), ("báscula descalibrada", dia_descalibrado)]:
    resultados[nombre] = (alarma_bulto_raro(pesos), alarma_corrimiento(pesos))
    print(f"{nombre:<22} | bultos raros: {resultados[nombre][0]} | alarma de corrimiento: {resultados[nombre][1]}")

assert resultados["día normal"] == ([], False)
assert resultados["un bulto de 14 kg"][0] == [14.0] and resultados["un bulto de 14 kg"][1] is False   # un solo bulto casi no mueve el promedio
assert resultados["báscula descalibrada"][1] is True and resultados["báscula descalibrada"][0] == []   # nadie "se ve raro", pero el promedio se corrio

fig, axs = plt.subplots(1, 2, figsize=(11, 3.6), sharey=True)
for ax, (nombre, pesos) in zip(axs, [("un bulto de 14 kg", dia_bulto), ("báscula descalibrada", dia_descalibrado)]):
    ax.hist(habitual, bins=20, alpha=0.5, color="#2563eb", density=True, label="habitual")
    ax.hist(pesos, bins=20, alpha=0.5, color="#dc2626", density=True, label="hoy")
    ax.set_xlabel("peso del bulto (kg)"); ax.set_title(nombre, fontsize=10, fontweight="bold"); ax.legend(fontsize=8)
plt.tight_layout(); plt.show()
````

---
**Lección:** son **dos problemas distintos**. El **bulto de 14 kg** dispara la alarma de **bulto raro** pero **no** mueve el promedio. La **báscula descalibrada** hace lo contrario: **ningún bulto se ve raro**, pero **todo el conjunto se corrió** y solo la alarma de **corrimiento** lo nota. En el mundo real, la primera es la **detección de anomalías** y la segunda el **monitoreo de deriva**; un modelo en producción necesita **las dos**. (Y ninguna de las dos te dice si la **regla del negocio** cambió: para eso hay que mirar si el modelo **sigue acertando**.)

## 2. Tres Colegios, un Solo Promedio 🏫

Tres colegios de Boyacá quieren el **promedio general** de una prueba, pero **no pueden compartir las notas** de sus estudiantes (son datos personales). Entonces **cada colegio calcula solo dos cosas**: **su promedio** y **cuántos estudiantes** tiene, y **solo esos dos números** viajan al centro. El centro combina los promedios **dándole más peso al colegio más grande**. (Es la idea del **aprendizaje federado**: viaja el resultado, no los datos. Aquí el «modelo» es lo más simple que hay: un promedio.)

````python
random.seed(7)
colegios = {"Colegio A": [random.gauss(3.8, 0.5) for _ in range(30)],
            "Colegio B": [random.gauss(3.2, 0.5) for _ in range(120)],
            "Colegio C": [random.gauss(4.1, 0.5) for _ in range(50)]}

# Lo que viaja al centro: (promedio, numero de estudiantes). Las notas se quedan en cada colegio.
resumen = {c: (st.mean(notas), len(notas)) for c, notas in colegios.items()}
for c, (m, n) in resumen.items():
    print(f"{c}: promedio {m:.3f} con {n} estudiantes")

total = sum(n for _, n in resumen.values())
promedio_pesado = sum(m * n for m, n in resumen.values()) / total                  # cada colegio pesa segun su tamano
promedio_simple = sum(m for m, _ in resumen.values()) / len(resumen)               # el error: tratar igual a un colegio de 30 y a uno de 120
promedio_real = st.mean([x for notas in colegios.values() for x in notas])          # lo que daria juntar TODAS las notas (que no se puede)

print(f"\nPromedio real (juntando todas las notas) : {promedio_real:.4f}")
print(f"Promedio pesado por tamaño (federado)    : {promedio_pesado:.4f}")
print(f"Promedio simple de los 3 promedios       : {promedio_simple:.4f}   <- se equivoca")
assert abs(promedio_pesado - promedio_real) < 1e-12                                 # igual al real, sin mover ninguna nota
assert abs(promedio_simple - promedio_real) > 0.1                                   # el promedio de promedios engana

fig, ax = plt.subplots(figsize=(7, 3.6))
ax.bar(["real\n(juntando)", "pesado\n(federado)", "simple\n(de promedios)"], [promedio_real, promedio_pesado, promedio_simple], color=["#475569", "#16a34a", "#dc2626"])
ax.set_ylim(3.0, 4.0); ax.set_ylabel("promedio general")
ax.set_title("Pesar por tamaño da el resultado real sin mover las notas", fontsize=10, fontweight="bold")
plt.tight_layout(); plt.show()
````

---
**Lección:** el promedio **pesado** da **exactamente** el mismo resultado que juntar todas las notas, **sin que ninguna nota salga de su colegio**. El promedio **simple** de los tres promedios **se equivoca** porque trata igual al colegio de 30 estudiantes y al de 120. Es la misma trampa del «promedio de promedios» del Módulo 01. Un detalle importante: viajar solo **resultados** ayuda, pero **no garantiza por sí solo la privacidad** (con pocos estudiantes, un promedio puede delatar a alguien); para eso hay técnicas adicionales que se ven en el cuaderno estándar.

## 3. El Cuaderno de Actas con Sellos 📒

Una junta de acción comunal lleva un **cuaderno de actas**. Cada hoja trae al final un **sello** calculado a partir de **lo escrito en la hoja** y **del sello de la hoja anterior**. Si alguien **cambia una palabra** de una hoja vieja, **su sello ya no coincide** y, además, **todos los sellos siguientes dejan de cuadrar**. Esto es una **cadena de bloques** en miniatura. (El «sello» es una **huella digital** del texto, calculada con una función llamada *hash*: cualquier cambio, por pequeño que sea, produce una huella totalmente distinta.)

````python
import hashlib


def sello(texto, sello_anterior):
    """Huella digital (los primeros 8 caracteres) del texto de la hoja + el sello de la hoja anterior."""
    return hashlib.sha256((sello_anterior + texto).encode()).hexdigest()[:8]


actas = ["Se aprueba arreglar el puente de la vereda", "Cuota de $20.000 por familia", "Se elige a Carla como tesorera",
         "Se compra cemento para el puente", "Se entrega el informe de gastos"]


def armar_cuaderno(textos):
    cuaderno, anterior = [], "00000000"
    for t in textos:
        anterior = sello(t, anterior); cuaderno.append({"texto": t, "sello": anterior})
    return cuaderno


def primera_hoja_rota(cuaderno):
    """Indice de la primera hoja cuyo sello no cuadra con su texto y la hoja anterior (None si todo cuadra)."""
    anterior = "00000000"
    for i, hoja in enumerate(cuaderno):
        if hoja["sello"] != sello(hoja["texto"], anterior):
            return i
        anterior = hoja["sello"]
    return None


cuaderno = armar_cuaderno(actas)
for i, h in enumerate(cuaderno):
    print(f"Hoja {i}: sello {h['sello']} | {h['texto']}")
assert primera_hoja_rota(cuaderno) is None

# Alguien cambia la cuota de la hoja 1 SIN tocar los sellos
manipulado = [dict(h) for h in cuaderno]
manipulado[1]["texto"] = "Cuota de $2.000 por familia"
print("\nTras alterar la hoja 1, la primera hoja rota es:", primera_hoja_rota(manipulado))
assert primera_hoja_rota(manipulado) == 1

# Si ademas recalcula el sello de la hoja 1, la hoja 2 deja de cuadrar: hay que rehacer TODAS las siguientes
manipulado[1]["sello"] = sello(manipulado[1]["texto"], manipulado[0]["sello"])
print("Si recalcula el sello de la hoja 1, ahora la primera hoja rota es:", primera_hoja_rota(manipulado))
assert primera_hoja_rota(manipulado) == 2

# Un cambio minimo cambia todo el sello
a, b = sello("Cuota de $20.000 por familia", "00000000"), sello("Cuota de $20.001 por familia", "00000000")
print(f"Un solo digito distinto: {a} frente a {b}")
assert a != b
````
