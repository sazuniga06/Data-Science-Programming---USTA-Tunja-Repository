## 1. Propiedad, Responsabilidad y Transparencia: los Tres Pilares 🧱

**Gobernanza de datos** (*data governance*) es el sistema de **decisiones, roles, reglas y procesos** que define **quién puede hacer qué, con qué datos, bajo qué reglas y cómo se rinde cuentas**. No es una herramienta ni un proyecto de TI: es una **función de la organización**. Se distingue de la **gestión** de datos (*data management*): la gobernanza **decide y vigila**; la gestión **ejecuta** (ingesta, almacenamiento, calidad, seguridad). El libro de la asignatura la organiza en tres pilares:

| Pilar | Pregunta | Se concreta en… | En este cuaderno |
|---|---|---|---|
| **Propiedad** (*ownership*) | ¿De quién es este dato y quién decide sobre él? | Un **propietario** (*owner*) por conjunto de datos que responde por su definición, su uso y **aprueba accesos**; un **custodio** (*steward*) que cuida su calidad y sus definiciones día a día | §1 (RACI), §2 (catálogo) |
| **Responsabilidad** (*accountability*) | ¿Quién responde si algo sale mal y cómo se prueba? | Matriz **RACI**, políticas, **auditoría**, medición de calidad y de cumplimiento | §1, §4, §5 |
| **Transparencia** (*transparency*) | ¿Se sabe qué datos hay, de dónde vienen y para qué se usan? | **Catálogo**, **linaje**, contratos, información a los titulares | §2, §3, §5 |

**Roles** (los nombres varían entre autores y organizaciones: *verificar* el glosario de tu marco de referencia, p. ej. DAMA-DMBOK):

| Rol | Responsabilidad típica |
|---|---|
| **Propietario / dueño** (*data owner*) | **Rinde cuentas** por un dominio de datos: lo clasifica, define sus reglas de uso y **aprueba o niega accesos**. Suele ser un cargo de **negocio**, no de TI |
| **Custodio** (*data steward*) | **Cuida** el dato día a día: definiciones, **calidad**, metadatos, resolución de dudas. Conoce el negocio y el dato |
| **Custodio técnico** (TI / plataforma) | **Opera** el almacenamiento, respaldos y controles técnicos (cifrado, permisos). Ejecuta, **no decide** quién accede |
| **Oficial de protección de datos** | Vigila el **cumplimiento** normativo y atiende a los titulares |
| **Comité de gobernanza** | Aprueba políticas y resuelve conflictos entre dominios |

La **matriz RACI** asigna, para cada actividad, quién es **R**esponsable (hace), **A**utoridad / quien **rinde cuentas** (**exactamente una** persona o rol), **C**onsultado e **I**nformado. Dos reglas verificables: **una y solo una A** por actividad y **al menos una R**. Si hay dos "A", nadie responde; si no hay "R", nadie lo hace.

````python
# ============================================================
# 1. Matriz RACI como tabla y validador (una sola A, al menos una R) con asserts
# ============================================================
# !pip install -q numpy pandas matplotlib networkx

import warnings
from datetime import datetime, timedelta

import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")
SEED = 42
plt.rcParams.update({"figure.figsize": (9, 5), "axes.spines.top": False, "axes.spines.right": False})

ROLES = ["Comité de gobernanza", "Propietario", "Custodio (steward)", "Custodio técnico (TI)", "Oficial de protección", "Analista (usuario)"]
ACTIVIDADES = ["Definir la política de clasificación", "Aprobar el acceso a un conjunto de datos", "Corregir errores de calidad",
               "Operar almacenamiento y respaldos", "Responder a un incidente de fuga", "Publicar catálogo y linaje"]
RACI = pd.DataFrame([["A", "C", "R", "I", "C", "I"],
                     ["I", "A", "C", "R", "C", "I"],
                     ["I", "A", "R", "C", "-", "I"],
                     ["I", "C", "I", "A/R", "-", "-"],
                     ["I", "C", "C", "R", "A", "I"],
                     ["I", "C", "A", "R", "I", "I"]], index=ACTIVIDADES, columns=ROLES)


def validar_raci(raci):
    """Devuelve la lista de problemas: actividades sin exactamente una A o sin ninguna R, y roles sin ninguna asignacion."""
    problemas = []
    for act, fila in raci.iterrows():
        nA = fila.str.startswith("A").sum()                                                        # "A/R" cuenta como A y como R
        if nA != 1:
            problemas.append(f"'{act}': {nA} autoridades (A) en lugar de 1")
        if not fila.str.contains("R").any():
            problemas.append(f"'{act}': nadie es responsable (R)")
    for rol in raci.columns:
        if (raci[rol] == "-").all():
            problemas.append(f"El rol '{rol}' no participa en ninguna actividad")
    return problemas


print(RACI.to_string())
assert validar_raci(RACI) == []
roto = RACI.copy(); roto.iloc[1, 2] = "A"; roto.iloc[2, [1, 2]] = ["C", "I"]                   # dos A en la actividad 2 y ninguna R/A valida en la 3
print("\nMatriz con defectos ->", validar_raci(roto))
assert len(validar_raci(roto)) == 3                                                           # 2 A en 'Aprobar acceso' + (0 A y 0 R) en 'Corregir errores'

fig, ax = plt.subplots(figsize=(11, 3.6)); ax.axis("off")
col = {"A": "#fca5a5", "R": "#fde68a", "C": "#bfdbfe", "I": "#e2e8f0", "-": "white"}   # "A/R" se pinta como A
tb = ax.table(cellText=RACI.values, rowLabels=ACTIVIDADES, colLabels=ROLES, cellColours=[[col[v[0]] for v in fila] for fila in RACI.values], loc="center", cellLoc="center")
tb.auto_set_font_size(False); tb.set_fontsize(8); tb.scale(1, 1.5)
ax.set_title("Matriz RACI del caso (A = rinde cuentas, R = ejecuta, C = consultado, I = informado)", fontsize=10, fontweight="bold"); plt.tight_layout(); plt.show()
````

---
**Lectura.** La matriz de ejemplo cumple las dos reglas: cada actividad tiene **una sola A** y al menos una R (la notación `A/R` indica que el mismo rol rinde cuentas y ejecuta). Observa una decisión de gobernanza: el **custodio técnico** (TI) **opera** los respaldos y es quien **ejecuta** la aprobación de un acceso (R), pero **no** es quien lo aprueba (la A es del **propietario**): TI aplica, el negocio decide. La matriz con defectos muestra tres problemas típicos detectados automáticamente (dos "A" en una actividad; ninguna "A" y ninguna "R" en otra). Una matriz RACI que **nadie valida** se pudre; por eso conviene tratarla como **datos con pruebas**, no como un PDF.

## 2. Catálogo de Datos y Metadatos 📚

Un **catálogo de datos** es el "inventario con ficha" de lo que existe: sin él no hay transparencia ni propietarios, y la seguridad no sabe qué proteger. Cada ficha son **metadatos** (datos sobre los datos) de cuatro clases:

| Clase de metadato | Ejemplos | Quién lo mantiene |
|---|---|---|
| **Técnico** | Esquema (columnas y tipos), formato, ubicación, tamaño | Custodio técnico / automático |
| **De negocio** | Descripción, glosario, definición de cada columna, etiquetas | Custodio (*steward*) |
| **Operativo** | Frecuencia de carga, calidad medida, última actualización, linaje | Automático (pipelines) |
| **De gobernanza** | **Propietario**, **clasificación** de sensibilidad, **retención**, columnas con **PII**, finalidad | Propietario / oficial de protección |

Abajo implementamos un **catálogo mínimo desde cero** (didáctico; en producción: DataHub, Amundsen, Unity Catalog, OpenMetadata, Glue Data Catalog *(verificar)*). Reglas de **validación al registrar**: campos de gobernanza obligatorios, clasificación válida, **propietario y custodio distintos** y retención definida. Incluye `perfilar(df)`, que **infiere** los metadatos técnicos de un `DataFrame`. Se registran 18 conjuntos de datos de una organización **ficticia** (nombres de áreas inventados): fuentes, zonas intermedias, productos analíticos, un modelo y tableros.

````python
# ============================================================
# 2. Catalogo de datos minimo, desde cero (DIDACTICO), con validacion y busqueda
# ============================================================
CLASES = ["publico", "interno", "confidencial", "restringido"]        # de menor a mayor sensibilidad
NIVEL = {c: i for i, c in enumerate(CLASES)}


def perfilar(df):
    """Metadatos tecnicos inferidos de un DataFrame: tipo, % de nulos y valores distintos por columna."""
    return {c: {"tipo": str(df[c].dtype), "nulos_pct": round(float(100 * df[c].isna().mean()), 1), "distintos": int(df[c].nunique())} for c in df.columns}


class Catalogo:
    OBLIGATORIOS = ("descripcion", "propietario", "custodio", "clasificacion", "retencion_dias")

    def __init__(self):
        self.fichas = {}

    def registrar(self, nombre, **meta):
        faltan = [k for k in self.OBLIGATORIOS if meta.get(k) in (None, "")]
        assert not faltan, f"{nombre}: faltan campos obligatorios {faltan}"
        assert meta["clasificacion"] in CLASES, f"{nombre}: clasificacion invalida"
        assert meta["propietario"] != meta["custodio"], f"{nombre}: propietario y custodio deben ser distintos"
        meta.setdefault("columnas", {}); meta.setdefault("etiquetas", ()); meta.setdefault("tipo", "dato")
        self.fichas[nombre] = meta

    def buscar(self, texto=None, etiqueta=None, clasificacion_min=None):
        out = []
        for n, m in self.fichas.items():
            if texto and texto.lower() not in (n + " " + m["descripcion"]).lower(): continue
            if etiqueta and etiqueta not in m["etiquetas"]: continue
            if clasificacion_min and NIVEL[m["clasificacion"]] < NIVEL[clasificacion_min]: continue
            out.append(n)
        return out

    def tiene_pii(self, nombre):
        return any(c.get("pii") for c in self.fichas[nombre]["columnas"].values())


cat = Catalogo()
COM, FIN, RRH, OPS = "Dirección Comercial", "Dirección Financiera", "Gestión Humana", "Observatorio Ambiental"
S1, S2, S3, S4 = "Equipo de datos comercial", "Equipo de datos financiero", "Equipo de datos de RR. HH.", "Equipo de datos ambientales"
cols_cli = {"id_cliente": {"tipo": "int64", "pii": False}, "nombre": {"tipo": "object", "pii": True}, "email": {"tipo": "object", "pii": True}, "ciudad": {"tipo": "object", "pii": False}}
FICHAS = [  # (nombre, tipo, descripcion, propietario, custodio, clasificacion, retencion_dias, columnas, etiquetas)
    ("crm_clientes", "fuente", "Clientes tal como salen del CRM", COM, S1, "restringido", 1095, cols_cli, ("clientes",)),
    ("pos_ventas", "fuente", "Ventas del punto de venta", COM, S1, "interno", 1825, {"id_venta": {"tipo": "int64", "pii": False}}, ("ventas",)),
    ("sensores_estaciones", "fuente", "Lecturas de estaciones meteorologicas", OPS, S4, "publico", 3650, {}, ("clima",)),
    ("nomina_rrhh", "fuente", "Nomina de empleados", RRH, S3, "restringido", 3650, {"empleado": {"tipo": "object", "pii": True}, "salario": {"tipo": "float64", "pii": True}}, ("personas",)),
    ("stg_clientes", "staging", "Clientes limpios y tipados", COM, S1, "restringido", 365, cols_cli, ("clientes",)),
    ("stg_ventas", "staging", "Ventas deduplicadas", COM, S1, "interno", 365, {}, ("ventas",)),
    ("stg_sensores", "staging", "Lecturas validadas", OPS, S4, "publico", 365, {}, ("clima",)),
    ("stg_nomina", "staging", "Nomina validada", RRH, S3, "restringido", 365, {"empleado": {"tipo": "object", "pii": True}}, ("personas",)),
    ("dim_cliente", "curado", "Dimension de clientes", COM, S1, "restringido", 1095, cols_cli, ("clientes",)),
    ("fact_ventas", "curado", "Hechos de ventas (con el cliente)", COM, S1, "restringido", 1825, {}, ("ventas",)),
    ("mart_ventas_ciudad", "mart", "Ventas agregadas por ciudad y mes", COM, S1, "interno", 1825, {}, ("ventas",)),
    ("mart_segmentos", "mart", "Segmentos de clientes con su nombre", COM, S1, "interno", 730, {"nombre_cliente": {"tipo": "object", "pii": True}, "segmento": {"tipo": "object", "pii": False}}, ("clientes",)),
    ("mart_clima", "mart", "Indicadores climaticos por estacion", OPS, S4, "publico", 3650, {}, ("clima",)),
    ("mart_costos_personal", "mart", "Costo de personal por area", RRH, S3, "interno", 1825, {}, ("personas",)),
    ("modelo_churn", "modelo", "Modelo de abandono de clientes", COM, S1, "interno", 1095, {}, ("clientes", "ml")),
    ("tablero_gerencia", "tablero", "Tablero de la gerencia", FIN, S2, "interno", 365, {}, ("ventas", "personas")),
    ("tablero_marketing", "tablero", "Tablero de marketing", COM, S1, "interno", 365, {}, ("clientes",)),
    ("tablero_clima", "tablero", "Tablero publico de clima", OPS, S4, "publico", 365, {}, ("clima",)),
]
for n, tipo, d, p, c, k, r, cols, tags in FICHAS:
    cat.registrar(n, tipo=tipo, descripcion=d, propietario=p, custodio=c, clasificacion=k, retencion_dias=r, columnas=cols, etiquetas=tags)

print(f"{len(cat.fichas)} fichas registradas | por clasificación:", pd.Series([m['clasificacion'] for m in cat.fichas.values()]).value_counts().reindex(CLASES, fill_value=0).to_dict())
print("buscar(etiqueta='clientes'):", cat.buscar(etiqueta="clientes"))
print("buscar(clasificacion_min='confidencial'):", cat.buscar(clasificacion_min="confidencial"))
print("con PII declarada:", [n for n in cat.fichas if cat.tiene_pii(n)])
ej = pd.DataFrame({"id_cliente": [1, 2, 3, 4], "email": ["a@x.co", None, "c@x.co", "c@x.co"], "ciudad": ["Tunja", "Duitama", "Tunja", None]})
print("perfilar(ejemplo):", perfilar(ej))

# validaciones al registrar
for malo in (dict(descripcion="x", propietario="", custodio="a", clasificacion="interno", retencion_dias=10), dict(descripcion="x", propietario="a", custodio="a", clasificacion="interno", retencion_dias=10),
             dict(descripcion="x", propietario="a", custodio="b", clasificacion="secreto", retencion_dias=10)):
    try:
        cat.registrar("prueba", **malo); raise RuntimeError("debio fallar")
    except AssertionError:
        pass
assert "prueba" not in cat.fichas and len(cat.fichas) == 18
assert cat.buscar(texto="NOMINA") == ["nomina_rrhh", "stg_nomina"] and cat.buscar(etiqueta="ml") == ["modelo_churn"]
assert perfilar(ej)["email"] == {"tipo": "object", "nulos_pct": 25.0, "distintos": 2}
````

---
**Lectura.** El catálogo **rechaza** fichas sin propietario, con propietario igual al custodio o con una clasificación inventada: la gobernanza empieza **negándose a registrar** lo que no tiene responsable. Con 18 fichas ya se puede **buscar** por etiqueta o por sensibilidad y saber **dónde hay PII**. Pero el catálogo es una **fotografía**: dice **qué** hay, no **de dónde viene** ni **qué se rompe si cambia**. Para eso está el linaje.

## 3. Linaje de Datos como Grafo Dirigido 🕸️

El **linaje** (*lineage*) registra **de dónde viene cada conjunto de datos y qué transformación lo produjo**: es un **grafo dirigido acíclico** (DAG: nodos = conjuntos de datos, aristas = "se derivó de"; el modelo **PROV** del W3C lo formaliza con *entidades*, *actividades* y *agentes*; ver el cap. 5 del libro de Cao). Tres usos:

| Pregunta | Operación en el grafo | Ejemplo |
|---|---|---|
| **Análisis de impacto** (aguas abajo): *¿qué se rompe si cambio o corrijo este dato?* | **Descendientes** del nodo | Si la fuente `pos_ventas` falla, ¿qué tableros se afectan? |
| **Causa raíz** (aguas arriba): *¿de dónde sale este número?* | **Ancestros** del nodo | Un tablero muestra un valor raro: ¿qué fuentes lo alimentan? |
| **Propagación de políticas** | Recorrer el grafo **heredando** atributos | Un derivado de datos **restringidos** no puede ser **público** a menos que se **anonimice** |

Se construye con `networkx.DiGraph`; `nx.is_directed_acyclic_graph` verifica que **no haya ciclos** (un dato no puede derivarse de sí mismo), y `nx.topological_generations` da las **capas** para dibujarlo (de izquierda a derecha, de la fuente al consumo). Las aristas que **anonimizan** (una agregación con grupos grandes) se marcan, porque **cortan** la herencia de sensibilidad.

````python
# ============================================================
# 3. Linaje como DiGraph: construccion, dibujo por capas, analisis de impacto y de origen
# ============================================================
ARISTAS = [  # (origen, destino, transformacion, anonimiza)
    ("crm_clientes", "stg_clientes", "limpieza y tipado", False), ("pos_ventas", "stg_ventas", "deduplicacion", False),
    ("sensores_estaciones", "stg_sensores", "validacion de rangos", False), ("nomina_rrhh", "stg_nomina", "validacion", False),
    ("stg_clientes", "dim_cliente", "modelo dimensional", False), ("stg_ventas", "fact_ventas", "claves sustitutas", False), ("dim_cliente", "fact_ventas", "union por cliente", False),
    ("fact_ventas", "mart_ventas_ciudad", "agregacion por ciudad y mes", True), ("dim_cliente", "mart_segmentos", "segmentacion por cliente", False), ("fact_ventas", "mart_segmentos", "ventas por cliente", False),
    ("stg_sensores", "mart_clima", "indicadores por estacion", False), ("stg_nomina", "mart_costos_personal", "agregacion por area", True),
    ("mart_segmentos", "modelo_churn", "entrenamiento", False), ("mart_ventas_ciudad", "tablero_gerencia", "visualizacion", False), ("mart_costos_personal", "tablero_gerencia", "visualizacion", False),
    ("mart_segmentos", "tablero_marketing", "visualizacion", False), ("mart_clima", "tablero_clima", "visualizacion", False),
]
G = nx.DiGraph()
for n, m in cat.fichas.items():
    G.add_node(n, tipo=m["tipo"], clasificacion=m["clasificacion"])
for a, b, t, anon in ARISTAS:
    G.add_edge(a, b, transformacion=t, anonimiza=anon)
assert nx.is_directed_acyclic_graph(G) and G.number_of_nodes() == len(cat.fichas) and set(G.nodes) == set(cat.fichas)
H = G.copy(); H.add_edge("tablero_gerencia", "crm_clientes")                                     # un ciclo imposible: un tablero alimentando a la fuente
assert not nx.is_directed_acyclic_graph(H)
print(f"Linaje: {G.number_of_nodes()} nodos, {G.number_of_edges()} aristas | DAG: {nx.is_directed_acyclic_graph(G)} | añadir tablero -> fuente crea un ciclo: {not nx.is_directed_acyclic_graph(H)}")


def impacto(g, nodo, tipo=None):
    """Conjuntos que dependen (directa o indirectamente) de `nodo`; opcionalmente solo de un tipo (p. ej. 'tablero')."""
    d = nx.descendants(g, nodo)
    return sorted(n for n in d if tipo is None or g.nodes[n]["tipo"] == tipo)


def origen(g, nodo, tipo="fuente"):
    return sorted(n for n in nx.ancestors(g, nodo) if g.nodes[n]["tipo"] == tipo)


print("Impacto de un fallo en pos_ventas (tableros):", impacto(G, "pos_ventas", "tablero"))
print("Impacto de corregir crm_clientes (todo):", impacto(G, "crm_clientes"))
print("Fuentes de tablero_gerencia:", origen(G, "tablero_gerencia"))
n_desc = pd.Series({n: len(nx.descendants(G, n)) for n in G}).sort_values(ascending=False)
print("Conjuntos más críticos (más dependientes aguas abajo):", n_desc.head(3).to_dict())
assert impacto(G, "pos_ventas", "tablero") == ["tablero_gerencia", "tablero_marketing"] and origen(G, "tablero_gerencia") == ["crm_clientes", "nomina_rrhh", "pos_ventas"]
assert "modelo_churn" in impacto(G, "crm_clientes") and impacto(G, "sensores_estaciones", "tablero") == ["tablero_clima"] and n_desc.index[0] in ("crm_clientes", "dim_cliente")

# ---- dibujo por capas (generaciones topologicas), nodos coloreados por clasificacion; panel 2: el impacto de un fallo en crm_clientes
capas = list(nx.topological_generations(G))
pos = {}
for i, capa in enumerate(capas):
    for j, n in enumerate(sorted(capa)):
        pos[n] = (i, -(j - (len(capa) - 1) / 2))
COLOR = {"publico": "#bbf7d0", "interno": "#bfdbfe", "confidencial": "#fde68a", "restringido": "#fecaca"}
fig, axs = plt.subplots(1, 2, figsize=(14, 5.6))
for ax, resaltado, titulo in ((axs[0], set(), "Linaje (color = clasificación declarada)"), (axs[1], set(impacto(G, "crm_clientes")) | {"crm_clientes"}, "Impacto aguas abajo de un cambio en crm_clientes")):
    cols = ["#ef4444" if n in resaltado else COLOR[G.nodes[n]["clasificacion"]] for n in G]
    nx.draw_networkx_edges(G, pos, ax=ax, arrows=True, arrowsize=10, edge_color="#64748b", node_size=900, width=1.1)
    nx.draw_networkx_nodes(G, pos, ax=ax, node_color=cols, node_size=900, edgecolors="#334155")
    nx.draw_networkx_labels(G, pos, ax=ax, font_size=6.5)
    ax.set_title(titulo, fontsize=10, fontweight="bold"); ax.axis("off")
axs[0].text(0, -2.9, "verde público · azul interno · amarillo confidencial · rojo restringido", fontsize=8, color="#334155")
plt.tight_layout(); plt.show()
````

## 4. Calidad de Datos: un Motor de Reglas ✅

La **calidad de datos** se mide con **reglas** agrupadas en **dimensiones**; gobernar es **acordar las reglas, medirlas y actuar**. Dimensiones usuales (las listas varían entre marcos, *verificar*):

| Dimensión | Pregunta | Ejemplo de regla |
|---|---|---|
| **Completitud** | ¿Están los valores obligatorios? | `email` no nulo |
| **Unicidad** | ¿Hay duplicados donde no debe haberlos? | `id_cliente` único |
| **Validez** | ¿Respeta formato, rango o dominio? | `edad` entre 0 y 110; `email` con formato válido; `ciudad` en la lista oficial |
| **Consistencia** | ¿Coincide entre tablas? | Todo `id_municipio` existe en el catálogo de municipios |
| **Oportunidad** (*timeliness*) | ¿Está actualizado y es coherente en el tiempo? | `fecha_registro` no está en el futuro |

Un **motor de reglas** separa **qué** se exige (reglas declarativas, de dominio) de **cómo** se ejecuta (el motor): cada regla devuelve una máscara de **violaciones**; el motor calcula el porcentaje, compara con un **umbral** y envía a **cuarentena** las filas que violan reglas **bloqueantes**. Probamos con una tabla de 5 000 clientes **sintética** con defectos **sembrados** (para poder verificar que el motor los encuentra **todos**).

````python
# ============================================================
# 4. Motor de reglas de calidad: reglas declarativas, informe por dimension, cuarentena y conservacion de filas
# ============================================================
import re

MUNICIPIOS_OK = ["Tunja", "Duitama", "Sogamoso", "Paipa", "Chiquinquirá", "Villa de Leyva", "Samacá", "Garagoa", "Soatá", "Aquitania", "Cómbita", "Motavita"]


def generar_clientes(n=5000, semilla=SEED):
    rng = np.random.default_rng(semilla)
    df = pd.DataFrame({"id_cliente": np.arange(1, n + 1), "nombre": [f"Cliente {i:05d}" for i in range(1, n + 1)],
                       "email": [f"cliente{i}@example.com" for i in range(1, n + 1)], "edad": rng.integers(18, 90, n),
                       "ciudad": rng.choice(MUNICIPIOS_OK, n), "id_municipio": 0,
                       "fecha_registro": pd.Timestamp("2025-01-01") + pd.to_timedelta(rng.integers(0, 600, n), unit="D")})
    df["id_municipio"] = df.ciudad.map({m: i + 1 for i, m in enumerate(MUNICIPIOS_OK)})
    idx = rng.permutation(n)                                                                      # conjuntos de filas DISJUNTOS para cada defecto sembrado
    d = {"email_nulo": idx[:150], "email_malo": idx[150:230], "edad_fuera": idx[230:270], "fecha_futura": idx[270:295], "ciudad_mal": idx[295:325], "fk_huerfana": idx[325:345]}
    df.loc[d["email_nulo"], "email"] = None
    df.loc[d["email_malo"], "email"] = "correo-sin-arroba.com"
    df.loc[d["edad_fuera"], "edad"] = rng.choice([-5, 130, 200], len(d["edad_fuera"]))
    df.loc[d["fecha_futura"], "fecha_registro"] = pd.Timestamp("2027-06-01")
    df.loc[d["ciudad_mal"], "ciudad"] = "Tunjaa"
    df.loc[d["fk_huerfana"], "id_municipio"] = 99
    limpias = np.setdiff1d(np.arange(n), np.concatenate(list(d.values())))
    dup = df.iloc[rng.choice(limpias, 60, replace=False)].copy()                                    # 60 filas duplicadas (mismo id_cliente) tomadas de filas SIN otros defectos
    return pd.concat([df, dup], ignore_index=True), {k: len(v) for k, v in d.items()} | {"duplicados": 60}


clientes, sembrados = generar_clientes()
HOY = pd.Timestamp("2026-10-08")
REGEX_EMAIL = re.compile(r"^[\w.+-]+@[\w-]+(\.[\w-]+)+$")
REGLAS = [  # (nombre, dimension, funcion -> mascara de VIOLACIONES, umbral_max_pct, bloqueante)
    ("email_no_nulo", "completitud", lambda d: d.email.isna(), 1.0, False),
    ("id_cliente_unico", "unicidad", lambda d: d.duplicated("id_cliente", keep="first"), 0.0, True),
    ("email_formato", "validez", lambda d: d.email.notna() & ~d.email.fillna("").str.match(REGEX_EMAIL), 1.0, False),
    ("edad_rango_0_110", "validez", lambda d: ~d.edad.between(0, 110), 0.0, True),
    ("ciudad_en_lista_oficial", "validez", lambda d: ~d.ciudad.isin(MUNICIPIOS_OK), 0.5, False),
    ("municipio_existe", "consistencia", lambda d: ~d.id_municipio.between(1, len(MUNICIPIOS_OK)), 0.0, True),
    ("fecha_no_futura", "oportunidad", lambda d: d.fecha_registro > HOY, 0.0, True),
]


def ejecutar_reglas(df, reglas):
    filas, masks = [], {}
    for nombre, dim, f, umbral, bloq in reglas:
        m = f(df).fillna(False).astype(bool)
        masks[nombre] = m
        pct = 100 * m.mean()
        filas.append({"regla": nombre, "dimensión": dim, "violaciones": int(m.sum()), "% filas": round(pct, 2), "umbral %": umbral, "bloqueante": bloq, "estado": "OK" if pct <= umbral else "FALLA"})
    return pd.DataFrame(filas), masks


informe, masks = ejecutar_reglas(clientes, REGLAS)
print(f"{len(clientes):,} filas (5 000 + 60 duplicadas) | defectos sembrados: {sembrados}\n")
print(informe.to_string(index=False))

bloq = pd.concat([masks[n] for n, _, _, _, b in REGLAS if b], axis=1).any(axis=1)
curado, cuarentena = clientes[~bloq], clientes[bloq]
print(f"\nCuarentena (reglas bloqueantes): {len(cuarentena)} filas | curado: {len(curado):,} | conservación: {len(curado)} + {len(cuarentena)} = {len(clientes):,}")
print("Violaciones por dimensión:", informe.groupby("dimensión")["violaciones"].sum().to_dict())

viol = informe.set_index("regla")["violaciones"]
assert viol["email_no_nulo"] == sembrados["email_nulo"] and viol["email_formato"] == sembrados["email_malo"] and viol["edad_rango_0_110"] == sembrados["edad_fuera"]
assert viol["fecha_no_futura"] == sembrados["fecha_futura"] and viol["ciudad_en_lista_oficial"] == sembrados["ciudad_mal"] and viol["municipio_existe"] == sembrados["fk_huerfana"] and viol["id_cliente_unico"] == 60
assert len(curado) + len(cuarentena) == len(clientes) and informe.set_index("regla").loc["email_no_nulo", "estado"] == "FALLA"
````

**Lectura.** El motor encuentra **exactamente** todos los defectos sembrados (150 correos nulos, 80 mal formados, 40 edades fuera de rango, 25 fechas futuras, 30 ciudades con error de digitación, 20 municipios inexistentes y 60 duplicados), cada uno por la regla de **su dimensión**. Las reglas con **umbral** (`email_no_nulo` ≤ 1 %) distinguen **tolerar** de **bloquear**: casi el 3 % de correos nulos incumple su umbral aunque no impida usar el registro. Las **bloqueantes** envían la fila a **cuarentena** (nunca se descarta en silencio: se conserva con su motivo, como en el [ciclo de vida del Módulo 00](../00%20-%20Fundamentos%20de%20Big%20Data/03_Ciclo_de_Vida_de_los_Datos.ipynb)) y la **conservación de filas** (curado + cuarentena = total) se verifica. Un límite honesto: **las reglas solo cazan lo que alguien pensó**; un dato plausible pero falso (una edad de 37 donde era 73) las pasa todas.

## 5. Políticas como Código y Contratos de Datos 📜

Una **política** escrita en un PDF **no se cumple sola**. **Política como código** (*policy as code*) expresa la regla como **una función que se evalúa automáticamente** sobre el catálogo, las bitácoras o los accesos, y entrega una lista de **incumplimientos**: se ejecuta en cada cambio, queda **versionada** y **auditable**. Cuatro políticas del caso (los valores son **ilustrativos**: cada organización los define; **no son** requisitos legales):

| Política | Regla | Controla |
|---|---|---|
| **POL-01 Propiedad** | Todo conjunto tiene propietario y custodio **distintos** | Responsabilidad |
| **POL-02 Retención** | La retención no supera el máximo de su clasificación (restringido ≤ 730 días, confidencial ≤ 1 095, interno ≤ 1 825; público sin tope) | Limitación de conservación |
| **POL-03 Clasificación** | Si una columna tiene **PII**, la clasificación es **al menos confidencial** | Confidencialidad |
| **POL-04 Accesos** | Un rol solo recibe conjuntos con clasificación **no mayor** a su habilitación | Mínimo privilegio |

Un **contrato de datos** (*data contract*) es el acuerdo **explícito** entre quien **produce** un conjunto y quienes lo **consumen**: esquema (columnas, tipos, nulos), reglas de calidad, frescura y propietario, **versionado** (semántico). Verificarlo en cada carga evita "se cambió una columna y se rompieron cinco tableros"; y comparar versiones detecta los **cambios incompatibles** (*breaking changes*): quitar una columna, cambiar un tipo, permitir nulos donde antes no.

````python
# ============================================================
# 5a. Politicas como codigo sobre el catalogo + accesos; deteccion y correccion
# ============================================================
MAX_RETENCION = {"publico": None, "interno": 1825, "confidencial": 1095, "restringido": 730}           # ILUSTRATIVOS (cada organizacion los define)
HABILITACION = {"analista": "interno", "analista_clientes": "confidencial", "oficial_proteccion": "restringido", "visitante": "publico"}
GRANTS = pd.DataFrame([("analista", "mart_ventas_ciudad", "read"), ("analista", "tablero_clima", "read"), ("oficial_proteccion", "fact_ventas", "read"),
                       ("analista", "mart_segmentos", "read"), ("analista", "dim_cliente", "read"), ("oficial_proteccion", "nomina_rrhh", "read"), ("visitante", "tablero_clima", "read")],
                      columns=["rol", "conjunto", "accion"])


def pol_propiedad(cat, grants):
    return [n for n, m in cat.fichas.items() if not m["propietario"] or not m["custodio"] or m["propietario"] == m["custodio"]]


def pol_retencion(cat, grants):
    return [n for n, m in cat.fichas.items() if MAX_RETENCION[m["clasificacion"]] is not None and m["retencion_dias"] > MAX_RETENCION[m["clasificacion"]]]


def pol_clasificacion(cat, grants):
    return [n for n, m in cat.fichas.items() if cat.tiene_pii(n) and NIVEL[m["clasificacion"]] < NIVEL["confidencial"]]


def pol_accesos(cat, grants):
    return sorted({f"{r.rol} -> {r.conjunto}" for r in grants.itertuples() if NIVEL[cat.fichas[r.conjunto]["clasificacion"]] > NIVEL[HABILITACION[r.rol]]})


POLITICAS_COD = {"POL-01 propiedad": pol_propiedad, "POL-02 retención": pol_retencion, "POL-03 clasificación": pol_clasificacion, "POL-04 accesos": pol_accesos}


def evaluar_politicas(cat, grants):
    return {nombre: f(cat, grants) for nombre, f in POLITICAS_COD.items()}


def indicadores(cat, grants, G, informe):
    """Indicadores de gobernanza (0-100) calculados sobre los propios artefactos."""
    n = len(cat.fichas); viol = evaluar_politicas(cat, grants)
    con_reglas = informe.shape[0]
    return {"responsables definidos": 100 * sum(1 for m in cat.fichas.values() if m["propietario"] and m["custodio"] and m["propietario"] != m["custodio"]) / n,
            "clasificación declarada": 100 * sum(1 for m in cat.fichas.values() if m["clasificacion"] in CLASES) / n,
            "linaje documentado": 100 * sum(1 for x in cat.fichas if G.degree(x) > 0) / n,
            "calidad: reglas que cumplen": 100 * (informe["estado"] == "OK").mean(),
            "políticas sin incumplimientos": 100 * sum(1 for v in viol.values() if not v) / len(viol),
            "retención dentro del máximo": 100 * (1 - len(viol["POL-02 retención"]) / n)}


antes = evaluar_politicas(cat, GRANTS)
ind_antes = indicadores(cat, GRANTS, G, informe)
print("Incumplimientos ANTES:")
for k, v in antes.items():
    print(f"  {k}: {v}")
assert antes["POL-01 propiedad"] == [] and antes["POL-02 retención"] == ["crm_clientes", "nomina_rrhh", "dim_cliente", "fact_ventas"]
assert antes["POL-03 clasificación"] == ["mart_segmentos"] and antes["POL-04 accesos"] == ["analista -> dim_cliente"]

# ---- correcciones decididas por los propietarios (se aplican al catalogo y se vuelve a evaluar)
for n in antes["POL-02 retención"]:                                                    # bajar la retencion al maximo de su clase
    cat.fichas[n]["retencion_dias"] = MAX_RETENCION[cat.fichas[n]["clasificacion"]]
cat.fichas["mart_segmentos"]["clasificacion"] = "confidencial"                         # tiene PII: sube de clasificacion...
intermedio = evaluar_politicas(cat, GRANTS)
print("\nTras subir mart_segmentos a 'confidencial' APARECE un nuevo incumplimiento de accesos:", intermedio["POL-04 accesos"])
assert intermedio["POL-03 clasificación"] == [] and intermedio["POL-04 accesos"] == ["analista -> dim_cliente", "analista -> mart_segmentos"]
GRANTS = GRANTS[~((GRANTS.rol == "analista") & (GRANTS.conjunto == "dim_cliente"))].copy()                                   # se revoca el acceso a dim_cliente
GRANTS.loc[(GRANTS.rol == "analista") & (GRANTS.conjunto == "mart_segmentos"), "rol"] = "analista_clientes"                  # y mart_segmentos pasa a un rol habilitado
despues = evaluar_politicas(cat, GRANTS)
print("Incumplimientos DESPUÉS:", despues)
assert all(v == [] for v in despues.values())
````

````python
# ============================================================
# 5b. Contratos de datos: verificacion de un lote y deteccion de cambios incompatibles entre versiones
# ============================================================
CONTRATO_V1 = {"conjunto": "fact_ventas", "version": "1.0.0", "propietario": COM, "frescura_max_horas": 24, "filas_min": 100,
               "columnas": {"id_venta": {"tipo": "int64", "nulos": False, "unico": True}, "id_cliente": {"tipo": "int64", "nulos": False},
                            "valor": {"tipo": "float64", "nulos": False, "min": 0}, "ciudad": {"tipo": "object", "nulos": False, "permitidos": MUNICIPIOS_OK}}}


def verificar_contrato(df, contrato, ultima_carga, ahora):
    """Lista de incumplimientos de un lote frente a su contrato (esquema, nulos, unicidad, rango, dominio, volumen y frescura)."""
    inc = []
    for col, r in contrato["columnas"].items():
        if col not in df.columns:
            inc.append(f"falta la columna '{col}'"); continue
        if str(df[col].dtype) != r["tipo"]: inc.append(f"'{col}': tipo {df[col].dtype} en lugar de {r['tipo']}")
        if not r.get("nulos", True) and df[col].isna().any(): inc.append(f"'{col}': {int(df[col].isna().sum())} nulos")
        if r.get("unico") and df[col].duplicated().any(): inc.append(f"'{col}': valores repetidos")
        if "min" in r and (df[col].dropna() < r["min"]).any(): inc.append(f"'{col}': valores por debajo de {r['min']}")
        if "permitidos" in r and (~df[col].dropna().isin(r["permitidos"])).any(): inc.append(f"'{col}': valores fuera del dominio")
    if len(df) < contrato["filas_min"]: inc.append(f"volumen: {len(df)} filas < {contrato['filas_min']}")
    if ahora - ultima_carga > timedelta(hours=contrato["frescura_max_horas"]): inc.append("frescura: el lote esta desactualizado")
    return inc


def comparar_contratos(viejo, nuevo):
    """Cambios INCOMPATIBLES para los consumidores: quitar columna, cambiar tipo, permitir nulos donde antes no."""
    rompe = []
    for col, r in viejo["columnas"].items():
        if col not in nuevo["columnas"]: rompe.append(f"se quita la columna '{col}'"); continue
        n = nuevo["columnas"][col]
        if n["tipo"] != r["tipo"]: rompe.append(f"'{col}': el tipo pasa de {r['tipo']} a {n['tipo']}")
        if not r.get("nulos", True) and n.get("nulos", True): rompe.append(f"'{col}': ahora admite nulos")
    return rompe


rng = np.random.default_rng(SEED)
lote = pd.DataFrame({"id_venta": np.arange(1000), "id_cliente": rng.integers(1, 500, 1000), "valor": rng.uniform(5_000, 300_000, 1000).round(0), "ciudad": rng.choice(MUNICIPIOS_OK, 1000)})
ahora = datetime(2026, 10, 8, 12, 0)
assert verificar_contrato(lote, CONTRATO_V1, ahora - timedelta(hours=3), ahora) == []
malo = lote.copy(); malo.loc[3, "id_cliente"] = None; malo.loc[5, "valor"] = -10; malo.loc[7, "ciudad"] = "Atlantis"; malo["id_venta"] = malo["id_venta"].astype(str); malo = malo.head(80)
inc = verificar_contrato(malo, CONTRATO_V1, ahora - timedelta(hours=40), ahora)
print("Lote malo ->", *inc, sep="\n  - ")
assert len(inc) == 7

V2 = {**CONTRATO_V1, "version": "2.0.0", "columnas": {k: dict(v) for k, v in CONTRATO_V1["columnas"].items() if k != "ciudad"}}
V2["columnas"]["valor"]["tipo"] = "object"; V2["columnas"]["id_cliente"]["nulos"] = True; V2["columnas"]["canal"] = {"tipo": "object", "nulos": True}   # agrega 'canal' (compatible)
print("\nCambios incompatibles v1 -> v2:", *comparar_contratos(CONTRATO_V1, V2), sep="\n  - ")
assert len(comparar_contratos(CONTRATO_V1, V2)) == 3 and comparar_contratos(CONTRATO_V1, {**CONTRATO_V1, "columnas": {**CONTRATO_V1["columnas"], "canal": {"tipo": "object", "nulos": True}}}) == []
````

## 6. Cumplimiento, Madurez y Cómo se Mide 🧭

**Cumplimiento a nivel conceptual (no es asesoría jurídica).** En Colombia la protección de datos personales se rige por el **Habeas Data** (derecho constitucional) y la **Ley 1581 de 2012** y su reglamentación (la autoridad es la Superintendencia de Industria y Comercio); en la Unión Europea, por el **GDPR**. Sin citar artículos de memoria (**verificar** el texto vigente en la fuente oficial), ambos comparten ideas que se traducen en **controles técnicos** que ya construimos:

| Idea (formulación general, *verificar*) | Control técnico en este módulo |
|---|---|
| **Finalidad** y **minimización**: usar solo lo necesario para un propósito declarado | Catálogo con **descripción y finalidad**; enmascarar/seudonimizar ([cuaderno 01](01_Tecnicas_de_Seguridad_Encriptacion_Acceso_Auditoria.ipynb)) |
| **Calidad / veracidad / exactitud** de los datos | **Motor de reglas** de calidad (§4) y cuarentena |
| **Transparencia** hacia el titular y la organización | **Catálogo**, **linaje** (§2 y §3), contratos |
| **Limitación de conservación** (no guardar indefinidamente) | **POL-02 retención** y ***crypto-shredding*** |
| **Seguridad y confidencialidad** | Cifrado, **RBAC/ABAC**, **POL-03 y POL-04** |
| **Responsabilidad demostrable** (*accountability*) | **RACI**, **bitácora inviolable**, informes de políticas |
| **Derechos del titular** (acceso, rectificación, supresión, oposición o revocatoria; *verificar* la lista) | El **linaje** localiza **dónde** está el dato para ejercerlos; *crypto-shredding* para suprimir |

**Modelo de madurez.** Los marcos suelen describir niveles de **madurez** de la gobernanza (p. ej. de **1 · inicial** —ad hoc, sin responsables— a **5 · optimizado** —medido y mejorado de forma continua—; los nombres y criterios varían: *verificar* el marco que se adopte). Aquí no inventamos un estándar: **calculamos indicadores** sobre los artefactos del caso (qué porcentaje de conjuntos tiene responsables, clasificación, linaje; qué reglas de calidad cumplen; qué políticas están limpias) y los **comparamos antes y después** de aplicar las correcciones. El mapeo a "nivel" usa umbrales **ilustrativos** que definimos para el ejercicio.

````python
# ============================================================
# 6. Madurez: indicadores calculados sobre los artefactos del caso, antes y despues de corregir
# ============================================================
ind_despues = indicadores(cat, GRANTS, G, informe)
tabla = pd.DataFrame({"antes": ind_antes, "después": ind_despues}).round(1)
print(tabla.to_string())


def nivel_madurez(valores):
    """Mapeo ILUSTRATIVO (no estandar): promedio de indicadores -> nivel 1 (inicial) a 5 (optimizado)."""
    prom = float(np.mean(list(valores.values())))
    for corte, nivel in ((95, 5), (80, 4), (60, 3), (40, 2)):
        if prom >= corte:
            return nivel, prom
    return 1, prom


n_ant, p_ant = nivel_madurez(ind_antes); n_des, p_des = nivel_madurez(ind_despues)
print(f"\nPromedio de indicadores: antes {p_ant:.1f} (nivel {n_ant}) -> después {p_des:.1f} (nivel {n_des})")
assert tabla.loc["políticas sin incumplimientos", "antes"] < tabla.loc["políticas sin incumplimientos", "después"] == 100.0
assert tabla.loc["calidad: reglas que cumplen", "antes"] == tabla.loc["calidad: reglas que cumplen", "después"] < 100      # calidad no se arregla con politicas: las reglas fallan igual
assert p_des > p_ant and n_des >= n_ant

fig, ax = plt.subplots(figsize=(9, 4.2))
y = np.arange(len(tabla)); ax.barh(y + 0.2, tabla["antes"], 0.38, color="#9ca3af", label="antes"); ax.barh(y - 0.2, tabla["después"], 0.38, color="#2563eb", label="después")
ax.set_yticks(y); ax.set_yticklabels(tabla.index, fontsize=8.5); ax.invert_yaxis(); ax.set_xlim(0, 105); ax.set_xlabel("% (0-100)")
ax.set_title("Indicadores de gobernanza del caso, antes y después de aplicar las políticas", fontsize=10, fontweight="bold"); ax.legend(loc="lower right"); plt.tight_layout(); plt.show()
````

---
**Lectura.** Las correcciones **suben** los indicadores que dependen de **decisiones de gobernanza** (políticas sin incumplimientos, retención) pero **no** el de **calidad**: las reglas de calidad siguen fallando porque los **datos** siguen sucios; arreglar eso exige trabajo del **custodio** sobre las fuentes (§4), no una política. Eso ilustra por qué la gobernanza es **sistema** y no un documento: cada indicador apunta a **un rol** que debe actuar. Y una advertencia sobre los indicadores: **miden lo que alguien pensó medir**; un 100 % en "políticas limpias" dice que se cumplen **las políticas que escribimos**, no que estén completas o sean las correctas.

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Gobernanza frente a gestión** | La gobernanza **decide y vigila** (roles, reglas, cuentas); la gestión **ejecuta**. Se apoya en **propiedad, responsabilidad y transparencia** |
| **Roles y RACI** | Propietario (decide, aprueba accesos), custodio/*steward* (cuida calidad y definiciones), custodio técnico (opera). Una y solo una **A** por actividad; al menos una **R**; la matriz se **valida** como datos |
| **Catálogo de datos** | Metadatos técnicos, de negocio, operativos y de **gobernanza**; el catálogo **rechaza** fichas sin responsable o con clasificación inválida; es la base de la transparencia |
| **Linaje** | Grafo dirigido acíclico (`networkx`): **impacto** aguas abajo (descendientes), **origen** aguas arriba (ancestros), **herencia** de sensibilidad (salvo aristas que anonimizan) |
| **Calidad de datos** | Reglas por dimensión (completitud, unicidad, validez, consistencia, oportunidad) con umbrales; **cuarentena** conserva las filas; las reglas solo cazan lo que alguien pensó |
| **Políticas y contratos como código** | La política se evalúa **automáticamente** y es auditable; el contrato fija esquema, calidad y frescura, y su comparación detecta **cambios incompatibles** (versionado semántico) |
| **Cumplimiento y madurez** | Habeas Data/Ley 1581 y GDPR comparten ideas que se traducen en controles técnicos (**verificar** el texto oficial; no es asesoría jurídica); la madurez se **mide con indicadores**, no se declara |
