## 1. La Tríada CIA y el Modelo de Amenazas (STRIDE) 🔺

**Seguridad de la información** = proteger tres propiedades de los datos:

| Propiedad | Pregunta | Se rompe cuando… | Ejemplo en Big Data |
|---|---|---|---|
| **Confidencialidad** (*confidentiality*) | ¿Solo quien debe puede **verlo**? | Alguien no autorizado lee los datos (**brecha**, fuga) | Un *bucket* del lago de datos queda abierto; un analista copia la tabla de clientes |
| **Integridad** (*integrity*) | ¿Los datos son **correctos y no fueron alterados**? | Se modifican o corrompen (por error o por ataque) | Alguien edita la tabla de precios; un *pipeline* sobrescribe una partición |
| **Disponibilidad** (*availability*) | ¿Están **cuando se necesitan**? | El servicio cae o los datos se bloquean | Un ataque de denegación de servicio; un secuestro con cifrado (*ransomware*) |

Un **modelo de amenazas** responde: *¿qué podría salir mal, dónde y qué hacemos?* **STRIDE** (de Microsoft) clasifica las amenazas en seis familias, cada una **viola una propiedad** de seguridad:

| Letra | Amenaza | Propiedad violada | Mitigación típica |
|---|---|---|---|
| **S** | ***Spoofing*** (suplantación de identidad) | Autenticidad | Autenticación fuerte, MFA, firmas |
| **T** | ***Tampering*** (alteración) | Integridad | Controles de escritura, *hashes*, cifrado autenticado |
| **R** | ***Repudiation*** (repudio: "yo no fui") | No repudio | **Bitácoras de auditoría** a prueba de alteración |
| **I** | ***Information disclosure*** (divulgación) | Confidencialidad | Cifrado, control de acceso, enmascaramiento |
| **D** | ***Denial of service*** (denegación de servicio) | Disponibilidad | Límites de tasa, redundancia, copias de seguridad |
| **E** | ***Elevation of privilege*** (escalada de privilegios) | Autorización | **Privilegio mínimo**, separación de funciones |

El método es **dibujar el flujo de los datos** (quién los pide, por dónde pasan, dónde descansan), marcar las **fronteras de confianza** (donde cambia quién controla qué) y preguntar por cada flecha y cada caja qué letras de STRIDE aplican. Abajo, el diagrama de una arquitectura típica de analítica (dibujado con `matplotlib`; **no es** una arquitectura de ninguna empresa real).

````python
# ============================================================
# Configuracion inicial + diagrama de flujo de datos con fronteras de confianza y letras STRIDE
# ============================================================
# !pip install -q numpy pandas matplotlib scikit-learn

import re
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.patches import FancyBboxPatch, Rectangle
from sklearn.ensemble import IsolationForest
from sklearn.metrics import roc_auc_score

warnings.filterwarnings("ignore")
SEED = 42
RAPIDO = True                    # True: 300 usuarios y 20 000 personas (segundos). False: el doble de usuarios y 5 veces mas personas
plt.rcParams.update({"figure.figsize": (9, 5), "axes.spines.top": False, "axes.spines.right": False})


def caja(ax, x, y, w, h, texto, fc="#eff6ff", ec="#1e3a8a", fs=9):
    ax.add_patch(FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.03,rounding_size=0.12", fc=fc, ec=ec, lw=1.6))
    ax.text(x + w / 2, y + h / 2, texto, ha="center", va="center", fontsize=fs, color="#0f172a")


def flecha(ax, p, q, letras, dy=0.22):
    ax.annotate("", xy=q, xytext=p, arrowprops=dict(arrowstyle="-|>", lw=1.6, color="#334155"))
    ax.text((p[0] + q[0]) / 2, (p[1] + q[1]) / 2 + dy, letras, ha="center", fontsize=10, fontweight="bold", color="#b91c1c")


fig, ax = plt.subplots(figsize=(11, 4.8))
ax.set_xlim(0, 11); ax.set_ylim(0, 5); ax.axis("off")
ax.add_patch(Rectangle((0.05, 0.35), 2.3, 4.2, fc="#fef2f2", ec="#b91c1c", ls="--", lw=1.4)); ax.text(1.2, 4.05, "Internet\n(no confiable)", ha="center", fontsize=9, color="#b91c1c")
ax.add_patch(Rectangle((2.65, 0.35), 8.2, 4.2, fc="#f0fdf4", ec="#15803d", ls="--", lw=1.4)); ax.text(6.7, 4.05, "Red interna de la organización (frontera de confianza)", ha="center", fontsize=9, color="#15803d")
caja(ax, 0.4, 1.9, 1.6, 1.0, "Usuario /\naplicación\ncliente", fc="#fff7ed", ec="#c2410c")
caja(ax, 3.0, 1.9, 1.7, 1.0, "API / portal\nde consulta")
caja(ax, 5.5, 1.9, 1.9, 1.0, "Servicio de\nprocesamiento")
caja(ax, 8.3, 1.9, 2.0, 1.0, "Lago de datos /\nalmacén (en reposo)", fc="#fef9c3", ec="#a16207")
caja(ax, 5.5, 0.5, 1.9, 0.8, "Bitácora de\nauditoría", fc="#f1f5f9", ec="#475569")
flecha(ax, (2.0, 2.4), (3.0, 2.4), "S  D")
flecha(ax, (4.7, 2.4), (5.5, 2.4), "T  E")
flecha(ax, (7.4, 2.4), (8.3, 2.4), "I  T")
flecha(ax, (6.45, 1.9), (6.45, 1.3), "R", dy=0.0)
ax.text(5.5, 0.05, "S suplantación · T alteración · R repudio · I divulgación · D denegación de servicio · E escalada de privilegios", ha="center", fontsize=8.5, color="#334155")
ax.set_title("Modelo de amenazas: flujo de datos, fronteras de confianza y letras de STRIDE (ilustrativo)", fontsize=11, fontweight="bold")
plt.tight_layout(); plt.show()
````

## 2. Brechas de Datos: Detección en una Bitácora Grande 🔍

Una **brecha de datos** (*data breach*) es el acceso o la salida **no autorizada** de datos protegidos. Rara vez se ve como una película: se ve como **filas en una bitácora** (muchos intentos de ingreso fallidos, un usuario que descarga cien veces más de lo usual a las 3 de la madrugada). Con millones de eventos por día, nadie lo lee: se **automatiza** la detección con dos tipos de herramienta: **reglas con umbral** (explícitas, explicables) y **modelos de anomalía** (aprenden lo "normal").

**Datos (sintéticos, semilla fija).** `generar_bitacora` crea la bitácora de acceso de 14 días de una organización ficticia, con comportamiento **normal** (ingresos en horario laboral, lecturas moderadas, uno que otro error de digitación) y **amenazas sembradas** con etiqueta, para poder **medir** la detección:

| Sembrado | Qué es | Etiqueta |
|---|---|---|
| **Fuerza bruta** | 4 IP externas que prueban claves contra **una** cuenta (35, 60, 90 y 120 intentos en 3 minutos); dos terminan con un ingreso exitoso | `ataque_fb` |
| **Ráfaga legítima** | 30 usuarios de la **misma IP de oficina** fallan una vez a las 8:00 (cambio de clave): parece ataque, **no lo es** | (normal) |
| **Día anómalo** | 12 usuario-día de exfiltración por alguien **con permiso**, de cuatro tipos (3 de cada uno): **A** madrugada con volumen enorme, **B** madrugada con volumen **normal** (solo el horario delata), **C** ráfaga de exportaciones, **D** ráfaga diurna de lecturas moderadas | `anomalo` |
| **Violación de privilegio** | Un rol accede a un recurso que su rol **no** debería tocar | `violacion` |

### 2.1 Fuerza bruta: umbrales por ventana de tiempo

Un ataque de **fuerza bruta** (*brute force*) prueba muchas claves contra una cuenta. La regla clásica: **si una misma clave de agrupación (IP, o IP + usuario) acumula $\geq u$ ingresos fallidos dentro de una ventana de $w$ minutos, se genera una alerta**. Los dos parámetros **trasladan costos**: un $u$ bajo atrapa más ataques (más **recobrado**) pero dispara **falsas alarmas** (menos **precisión**); un $u$ alto hace lo contrario. Con `rolling('5min')` de `pandas` el conteo por ventana es una línea. Medimos contra las 4 IP atacantes sembradas (**nivel IP**: ¿se alertó a la IP correcta?).

### 2.2 Accesos anómalos: `IsolationForest` sobre usuario-día

Las reglas detectan **lo que ya conocemos**. Para lo demás (un usuario con permiso que **descarga de más**) se usa un modelo de **anomalías**: `IsolationForest` aísla los puntos que quedan **solos** con pocas particiones aleatorias (puntaje más bajo = más anómalo). Resumimos cada **usuario-día** con cinco variables: número de eventos, filas leídas (en escala log), fracción **fuera de horario** (antes de las 7 y después de las 20 horas), número de exportaciones y recursos distintos. Evaluamos con las **12 combinaciones usuario-día sembradas**: **precisión@$k$** (¿cuántas de las $k=12$ más sospechosas lo son?), **recobrado@$2k$** y el **AUC** del puntaje; y lo comparamos con una **línea base simple** (el $z$-score de las filas leídas).

Primero armamos la bitácora y la **matriz de control de acceso** (qué puede hacer cada rol); después corren las dos detecciones.

````python
# ============================================================
# Bitacora sintetica de acceso (14 dias) + matriz de control de acceso por rol
# ============================================================
ROLES = ["analista", "ingeniero", "finanzas", "soporte"]
RECURSOS = ["ventas", "clientes", "nomina", "modelos", "bitacora_seg"]
# Matriz de control de acceso: (rol, recurso) -> permisos concedidos (privilegio minimo)
MATRIZ = {("analista", "ventas"): {"read", "export"}, ("analista", "clientes"): {"read"}, ("analista", "modelos"): {"read"},
          ("ingeniero", "ventas"): {"read"}, ("ingeniero", "modelos"): {"read", "export"}, ("ingeniero", "bitacora_seg"): {"read"},
          ("finanzas", "ventas"): {"read", "export"}, ("finanzas", "nomina"): {"read", "export"}, ("finanzas", "clientes"): {"read"},
          ("soporte", "clientes"): {"read"}}
PARES_OK = {r: [(rec, acc) for (rr, rec), perm in MATRIZ.items() if rr == r for acc in sorted(perm)] for r in ROLES}


def generar_bitacora(semilla=SEED, n_usuarios=300, dias=14):
    rng = np.random.default_rng(semilla)
    usuarios = np.array([f"u{i:04d}" for i in range(n_usuarios)])
    rol_de = dict(zip(usuarios, rng.choice(ROLES, n_usuarios, p=[0.4, 0.25, 0.15, 0.2])))
    ip_de = {u: f"10.1.{i // 200}.{i % 200 + 10}" for i, u in enumerate(usuarios)}
    n_ev = rng.poisson(12, (n_usuarios, dias)).ravel()
    ud = np.repeat(np.arange(n_usuarios * dias), n_ev)
    u, d = usuarios[ud // dias], ud % dias
    hora = np.where(rng.random(len(ud)) < 0.03, rng.uniform(0, 24, len(ud)), rng.uniform(8, 19, len(ud)))
    ts = pd.Timestamp("2025-03-03") + pd.to_timedelta(d, unit="D") + pd.to_timedelta(hora * 3600, unit="s")
    acc = rng.choice(["login_ok", "login_fail", "read", "export"], len(ud), p=[0.07, 0.03, 0.88, 0.02])
    df = pd.DataFrame({"ts": ts, "usuario": u, "rol": [rol_de[x] for x in u], "ip": [ip_de[x] for x in u], "accion": acc, "recurso": "auth", "filas": 0})
    for r in ROLES:                                                    # lecturas/exportaciones SOLO sobre pares permitidos por la matriz
        for a in ("read", "export"):
            m = (df.rol == r) & (df.accion == a)
            recs = [rec for rec, ac in PARES_OK[r] if ac == a]
            if not recs:                                               # el rol no tiene exportacion permitida: sus "export" pasan a "read"
                df.loc[m, "accion"] = "read"; m = (df.rol == r) & (df.accion == "read") & (df.recurso == "auth")
                recs = [rec for rec, ac in PARES_OK[r] if ac == "read"]
            df.loc[m, "recurso"] = rng.choice(recs, m.sum())
    lee, exp = df.accion == "read", df.accion == "export"
    df.loc[lee, "filas"] = rng.lognormal(np.log(200), 0.8, lee.sum()).astype(int)
    df.loc[exp, "filas"] = rng.lognormal(np.log(1500), 0.6, exp.sum()).astype(int)
    for c in ("ataque_fb", "anomalo", "violacion"):
        df[c] = 0
    df["tipo"] = ""
    ext = []
    # (1) fuerza bruta: 4 IP externas contra una cuenta
    for k, (n, ok) in enumerate([(35, 0), (60, 1), (90, 0), (120, 1)]):
        victima = usuarios[10 + 7 * k]; ip = f"203.0.113.{20 + k}"; t0 = pd.Timestamp("2025-03-03") + pd.Timedelta(days=2 + 3 * k, hours=2, minutes=15)
        t = t0 + pd.to_timedelta(np.sort(rng.uniform(0, 180, n)), unit="s")
        ev = pd.DataFrame({"ts": t, "usuario": victima, "rol": rol_de[victima], "ip": ip, "accion": "login_fail", "recurso": "auth", "filas": 0, "ataque_fb": 1, "anomalo": 0, "violacion": 0})
        if ok:
            ev.loc[len(ev) - 1, "accion"] = "login_ok"               # el ultimo intento acierta: cuenta comprometida
        ext.append(ev)
    # (2) rafaga legitima: 30 usuarios de la oficina fallan una vez a las 8:00 del dia 3 (misma IP)
    of = usuarios[100:130]
    ext.append(pd.DataFrame({"ts": pd.Timestamp("2025-03-06 08:00") + pd.to_timedelta(rng.uniform(0, 240, 30), unit="s"), "usuario": of,
                             "rol": [rol_de[x] for x in of], "ip": "172.16.0.9", "accion": "login_fail", "recurso": "auth", "filas": 0, "ataque_fb": 0, "anomalo": 0, "violacion": 0}))
    # (3) dias anomalos: 12 usuarios CON permiso, 4 tipos x 3: A madrugada+volumen, B madrugada+volumen normal, C rafaga de eventos, D rafaga diurna moderada
    for k, ui in enumerate(rng.choice(np.arange(150, n_usuarios), 12, replace=False)):
        x = usuarios[ui]; dia = pd.Timestamp("2025-03-03") + pd.Timedelta(days=int(rng.integers(0, dias)))
        rec, _ = PARES_OK[rol_de[x]][0]
        tipo = "ABCD"[k // 3]
        n = {"A": int(rng.integers(8, 20)), "B": int(rng.integers(10, 16)), "C": int(rng.integers(30, 41)), "D": int(rng.integers(10, 15))}[tipo]
        horas = rng.uniform(1.5, 4, n) if tipo in "AB" else rng.uniform(10, 11, n)
        mediana = {"A": 8000, "B": 200, "C": 200, "D": 1500}[tipo]
        ext.append(pd.DataFrame({"ts": dia + pd.to_timedelta(horas * 3600, unit="s"), "usuario": x, "rol": rol_de[x], "ip": ip_de[x], "accion": "read", "recurso": rec,
                                 "filas": rng.lognormal(np.log(mediana), 0.5, n).astype(int), "ataque_fb": 0, "anomalo": 1, "violacion": 0, "tipo": tipo}))
    # (4) violaciones de privilegio: roles que tocan recursos/acciones fuera de la matriz
    prohib = [(r, rec, acc) for r in ROLES for rec in RECURSOS for acc in ("read", "export") if acc not in MATRIZ.get((r, rec), set())]
    cand = rng.choice(np.arange(n_usuarios), 8, replace=False)
    for ui in cand:
        x = usuarios[ui]; opciones = [p for p in prohib if p[0] == rol_de[x]]
        for j in rng.integers(0, len(opciones), 4):
            r, rec, acc = opciones[j]
            ext.append(pd.DataFrame({"ts": [pd.Timestamp("2025-03-03") + pd.Timedelta(days=int(rng.integers(0, dias)), hours=float(rng.uniform(9, 17)))], "usuario": x, "rol": r,
                                     "ip": ip_de[x], "accion": acc, "recurso": rec, "filas": int(rng.integers(50, 500)), "ataque_fb": 0, "anomalo": 0, "violacion": 1}))
    out = pd.concat([df] + ext, ignore_index=True).sort_values("ts").reset_index(drop=True)
    out["tipo"] = out["tipo"].fillna("")
    return out


N_USU = 300 if RAPIDO else 600
bitacora = generar_bitacora(n_usuarios=N_USU)
print(f"{len(bitacora):,} eventos de {bitacora.usuario.nunique()} usuarios entre {bitacora.ts.min():%Y-%m-%d} y {bitacora.ts.max():%Y-%m-%d}")
print("Acciones:", bitacora.accion.value_counts().to_dict())
print("Sembrado -> eventos de fuerza bruta:", int(bitacora.ataque_fb.sum()), "| eventos de dias anomalos:", int(bitacora.anomalo.sum()), "| violaciones de privilegio:", int(bitacora.violacion.sum()))
assert bitacora.ts.is_monotonic_increasing and bitacora.ataque_fb.sum() == 35 + 60 + 90 + 120 and bitacora.violacion.sum() == 32
assert bitacora[bitacora.violacion == 0].apply(lambda r: r.accion in MATRIZ.get((r.rol, r.recurso), set()) or r.recurso == "auth", axis=1).all()   # lo normal respeta la matriz
bitacora.head(3)
````

````python
# ============================================================
# 2.1 Fuerza bruta: conteo de fallos por ventana deslizante y barrido del umbral
# ============================================================
ATACANTES = set(bitacora.loc[bitacora.ataque_fb == 1, "ip"])


def alertas_fuerza_bruta(df, umbral, ventana="5min", por=("ip",)):
    """Claves (IP o IP+usuario) con >= umbral fallos dentro de una ventana deslizante de tiempo."""
    f = df[df.accion == "login_fail"].assign(uno=1).set_index("ts").sort_index()
    mx = f.groupby(list(por))["uno"].rolling(ventana).sum().groupby(level=list(range(len(por)))).max()
    return set(mx[mx >= umbral].index)


def precision_recobrado(alertadas, verdaderas):
    tp = len(alertadas & verdaderas)
    return (tp / len(alertadas) if alertadas else 1.0), tp / len(verdaderas)


filas = []
for u in (5, 10, 20, 30, 40, 70, 100):
    p, r = precision_recobrado(alertas_fuerza_bruta(bitacora, u), ATACANTES)
    filas.append({"umbral (fallos en 5 min)": u, "alertas (IP)": len(alertas_fuerza_bruta(bitacora, u)), "precisión": round(p, 2), "recobrado": round(r, 2)})
barrido = pd.DataFrame(filas)
print(barrido.to_string(index=False))

print("\nIP alertadas con umbral 20 (clave = IP):", sorted(alertas_fuerza_bruta(bitacora, 20)))
# la clave de agrupacion (IP, usuario) separa "30 cuentas fallando una vez" de "35+ intentos contra UNA cuenta"
filas2 = []
for u in (5, 10, 20, 30, 40):
    ips = {ip for ip, usuario in alertas_fuerza_bruta(bitacora, u, por=("ip", "usuario"))}
    p, r = precision_recobrado(ips, ATACANTES); filas2.append({"umbral": u, "precisión (IP+usuario)": round(p, 2), "recobrado (IP+usuario)": round(r, 2)})
print("\nClave = (IP, usuario):\n" + pd.DataFrame(filas2).to_string(index=False))
cuentas_comprometidas = bitacora[(bitacora.ataque_fb == 1) & (bitacora.accion == "login_ok")].usuario.unique()
print("Cuentas con ingreso exitoso tras la rafaga de fallos (comprometidas, a rotar de inmediato):", [str(c) for c in cuentas_comprometidas])
assert "172.16.0.9" in alertas_fuerza_bruta(bitacora, 20) and "172.16.0.9" not in alertas_fuerza_bruta(bitacora, 40)   # la oficina es falsa alarma si el umbral es bajo
assert precision_recobrado(alertas_fuerza_bruta(bitacora, 40), ATACANTES)[1] < 1                                      # y un umbral alto deja pasar al ataque de 35 intentos
assert len(cuentas_comprometidas) == 2
assert all(f["precisión (IP+usuario)"] == 1.0 for f in filas2[:4]) and all(f["recobrado (IP+usuario)"] == 1.0 for f in filas2[:4])   # umbrales 5-30: sin falsas alarmas y sin perder ataques
````

````python
# ============================================================
# 2.2 IsolationForest sobre usuario-dia, evaluado contra las 12 anomalias sembradas
# ============================================================
acc = bitacora[bitacora.accion.isin(["read", "export"])].copy()
acc["dia"] = acc.ts.dt.floor("D")
acc["fuera_horario"] = ((acc.ts.dt.hour < 7) | (acc.ts.dt.hour >= 20)).astype(int)
ud = acc.groupby(["usuario", "dia"]).agg(n_eventos=("filas", "size"), filas=("filas", "sum"), fuera=("fuera_horario", "mean"),
                                         n_export=("accion", lambda s: (s == "export").sum()), n_recursos=("recurso", "nunique"),
                                         anomalo=("anomalo", "max"), tipo=("tipo", "max")).reset_index()
X = np.column_stack([ud.n_eventos, np.log1p(ud.filas), ud.fuera, ud.n_export, ud.n_recursos])
iforest = IsolationForest(n_estimators=200, random_state=SEED).fit(X)
ud["puntaje"] = -iforest.score_samples(X)                         # mayor = mas anomalo
z = (np.log1p(ud.filas) - np.log1p(ud.filas).mean()) / np.log1p(ud.filas).std()
ud["z_filas"] = z

k = int(ud.anomalo.sum())
def prec_en_k(col, kk):
    top = ud.nlargest(kk, col)
    return top.anomalo.sum() / kk
def recobrado_en_k(col, kk):
    return ud.nlargest(kk, col).anomalo.sum() / ud.anomalo.sum()

res = pd.DataFrame({"IsolationForest": [prec_en_k("puntaje", k), recobrado_en_k("puntaje", 2 * k), roc_auc_score(ud.anomalo, ud.puntaje)],
                    "z-score de filas": [prec_en_k("z_filas", k), recobrado_en_k("z_filas", 2 * k), roc_auc_score(ud.anomalo, ud.z_filas)]},
                   index=[f"precisión@{k}", f"recobrado@{2 * k}", "AUC"]).round(3)
print(f"{len(ud):,} usuario-días | anomalías sembradas: {k}")
print(res.to_string())
# recobrado@2k por TIPO de anomalia (A madrugada+volumen, B madrugada+volumen normal, C rafaga de eventos, D rafaga diurna moderada)
dentro = {col: set(ud.nlargest(2 * k, col).index) for col in ("puntaje", "z_filas")}
por_tipo = pd.DataFrame({col: {t: round(len(set(ud[ud.tipo == t].index) & dentro[col]) / 3, 2) for t in "ABCD"} for col in dentro}).rename(columns={"puntaje": "IsolationForest", "z_filas": "z-score de filas"})
print("\nRecobrado@%d por tipo de anomalía:\n" % (2 * k), por_tipo.to_string())
union = dentro["puntaje"] | dentro["z_filas"]
rec_union = len(union & set(ud[ud.anomalo == 1].index)) / k
print(f"Unión de ambas listas ({len(union)} usuario-días a revisar): recobrado {rec_union:.2f}")
assert res.loc["AUC", "IsolationForest"] > 0.9 and por_tipo.loc["B", "IsolationForest"] > por_tipo.loc["B", "z-score de filas"]
assert rec_union > max(res.loc[f"recobrado@{2 * k}"])
````

**Lectura del barrido.** El barrido muestra el **compromiso**: con umbrales de 5 a 30 se atrapan los cuatro ataques, pero la **ráfaga legítima de la oficina** (30 fallos desde una misma IP en 4 minutos) dispara una **falsa alarma** (precisión 0.80); con umbral 40 la oficina deja de molestar pero **se escapa el ataque más lento** (35 intentos; recobrado 0.75). **No existe un umbral perfecto con la clave "IP"**: el problema no es el número sino la **clave de agrupación**. Una cuenta que recibe 35 fallos de **una** IP es sospechosa; 30 fallos de **30 cuentas distintas** es otra cosa. Con la clave `(IP, usuario)` los umbrales de 5 a 30 dan **precisión y recobrado de 1.0** (los 30 fallos de la oficina son 30 claves con un solo fallo cada una). Costo: esa regla **no vería** un *rociado de contraseñas* (muchas cuentas, uno o dos intentos cada una, desde una misma IP); por eso se **combinan** ambas reglas. Y los dos ingresos exitosos tras la ráfaga marcan **cuentas comprometidas** que hay que rotar de inmediato.

**Lectura del modelo.** Con 12 anomalías sembradas entre 4 200 usuario-días, ambos detectores **ordenan bien** (AUC 0.995 para `IsolationForest` y 0.972 para el $z$-score), pero **ninguno es perfecto** y **se equivocan en cosas distintas**: `IsolationForest` recupera todos los casos de madrugada (tipos A y B, incluido el de volumen **normal** que el $z$-score no ve porque solo mira filas), pero pierde las ráfagas **diurnas** (C y D), que se parecen a un día muy activo; el $z$-score hace lo contrario. Su precisión@12 es 0.50 frente a 0.67 del $z$-score: **el modelo "sofisticado" no gana** en esta métrica. Lo que sí funciona es **combinar**: la unión de las dos listas (40 usuario-días para revisar de 4 200) recupera el 92 % de las anomalías sembradas. Dos lecciones operativas: (1) la salida de un detector es una **lista priorizada para que un analista revise**, no un veredicto; (2) las señales **se complementan**: reglas baratas y explicables más un modelo que mira varias variables a la vez. Y una anomalía **no es** una violación de acceso: estos usuarios **tenían permiso**, por eso el control de acceso, que vemos ahora, no basta por sí solo.

## 3. Accesos No Autorizados: Matriz de Control de Acceso 🔑

Un **acceso no autorizado** es una acción sobre un recurso que la política **no concede** al sujeto. El principio guía es el de **privilegio mínimo** (*least privilege*): cada rol recibe **solo** los permisos que necesita. La herramienta básica es la **matriz de control de acceso**: filas = roles (sujetos), columnas = recursos (objetos), celdas = acciones permitidas. Ya la definimos arriba (`MATRIZ`); abajo la mostramos como tabla y **auditamos la bitácora**: todo evento cuya acción no esté permitida para (rol, recurso) es una **violación**. Comparamos con las 32 violaciones sembradas.

````python
# ============================================================
# 3. Matriz de control de acceso (tabla) y deteccion de violaciones en la bitacora
# ============================================================
tabla = pd.DataFrame({rec: {r: "+".join(sorted(MATRIZ.get((r, rec), set()))) or "-" for r in ROLES} for rec in RECURSOS})
print("Matriz de control de acceso (- = sin acceso):\n", tabla.to_string())


def es_violacion(rol, recurso, accion):
    """Un evento de lectura/exportacion es una violacion si la matriz no concede esa accion a ese rol sobre ese recurso."""
    return accion in ("read", "export") and accion not in MATRIZ.get((rol, recurso), set())


obj = bitacora[bitacora.accion.isin(["read", "export"])].copy()
obj["detectada"] = [es_violacion(r, rec, a) for r, rec, a in zip(obj.rol, obj.recurso, obj.accion)]
tp = int(((obj.detectada) & (obj.violacion == 1)).sum()); fp = int(((obj.detectada) & (obj.violacion == 0)).sum()); fn = int(((~obj.detectada) & (obj.violacion == 1)).sum())
print(f"\nViolaciones detectadas: {obj.detectada.sum()} | verdaderas {tp} | falsas {fp} | no vistas {fn}")
por_usuario = obj[obj.detectada].groupby(["usuario", "rol"]).size().sort_values(ascending=False)
print("\nUsuarios con violaciones (a investigar):\n", por_usuario.head(5).to_string())
print("\nRecursos más tocados sin permiso:", obj[obj.detectada].recurso.value_counts().to_dict())
assert (tp, fp, fn) == (32, 0, 0) and por_usuario.sum() == 32
````

## 4. *Data Leakage*: Fuga de Información Personal 💧

La **fuga de datos** (*data leakage* en seguridad) es la salida de información sensible **fuera de su ámbito**: un archivo compartido por error, un correo con adjuntos, un *log* con datos personales, una tabla "anonimizada" que se puede **reidentificar**. Tres frentes, todos con datos ficticios:

1. **Buscar PII en texto** (cédulas, correos, teléfonos) con expresiones regulares, y medir qué tan bien (¡y qué falsos positivos!).
2. **Reidentificar** a una persona **enlazando** una tabla "anónima" con una pública.
3. **Medir** el riesgo con el **$k$-anonimato** y reducirlo generalizando.

### 4.1 PII en texto con expresiones regulares

Se generan 600 **mensajes de soporte** ficticios, parte con PII *fabricada* (cédulas, correos `@example.com`/`@example.org` —dominios reservados para ejemplos— y teléfonos) y todos con **distractores numéricos** (números de pedido, facturas y montos) que *parecen* cédulas. Probamos un patrón **ingenuo** (cualquier número largo) y uno **contextual** (con palabras clave y formato), y medimos **precisión** y **recobrado** a nivel de valores (normalizados a solo dígitos).

````python
# ============================================================
# 4.1 Deteccion de PII (cedulas, correos, telefonos) con regex: ingenuo vs contextual + redaccion
# ============================================================
def generar_mensajes(n=600, semilla=SEED):
    rng = np.random.default_rng(semilla)
    nombres = ["Camila", "Andrés", "Luz Marina", "Jhon", "Paola", "Edwin", "Sandra", "Fabián"]
    textos, verdad = [], []
    for i in range(n):
        v = set()
        partes = [f"Ticket {i}: {rng.choice(nombres)} reporta un problema con el pedido {int(rng.integers(10_000_000, 99_999_999))}."]
        if rng.random() < 0.9:
            partes.append(f"La factura {int(rng.integers(1_000_000, 9_999_999))} por un valor de ${int(rng.integers(10, 999))}.{int(rng.integers(100, 999))}.{int(rng.integers(100, 999))} no llega.")
        if rng.random() < 0.35:
            cc = int(rng.integers(1_000_000, 1_099_999_999))
            fmt = rng.choice(["C.C. {c}", "cédula {c}", "documento {p}"])
            partes.append("Titular: " + fmt.format(c=cc, p=f"{cc:,}".replace(",", ".")) + "."); v.add(("cc", str(cc)))
        if rng.random() < 0.3:
            c = f"{rng.choice(['ana', 'jorge', 'lucia', 'pedro'])}{int(rng.integers(1, 999))}@{rng.choice(['example.com', 'example.org'])}"
            partes.append(f"Escribir a {c} por favor."); v.add(("mail", c))
        if rng.random() < 0.3:
            t = f"3{int(rng.integers(0, 3))}{int(rng.integers(0, 10))}{int(rng.integers(1_000_000, 9_999_999))}"
            t = t[:10]; txt = rng.choice([t, f"+57 {t[:3]} {t[3:6]} {t[6:]}", f"{t[:3]}-{t[3:6]}-{t[6:]}"])
            partes.append(f"Llamar al {txt}."); v.add(("tel", t))
        textos.append(" ".join(partes)); verdad.append(v)
    return textos, verdad


PAT_MAIL = re.compile(r"[\w.+-]+@[\w-]+(?:\.[\w-]+)+")
PAT_TEL = re.compile(r"(?<!\d)(?:\+57[\s-]?)?(3\d{2})[\s-]?(\d{3})[\s-]?(\d{4})(?!\d)")
PAT_CC = re.compile(r"(?:C\.?\s?C\.?|c[eé]dula|documento)\s*:?\s*(\d{1,3}(?:\.\d{3}){1,3}|\d{6,10})", re.I)
PAT_NAIVE = re.compile(r"(?<![\d.$])\d{1,3}(?:\.\d{3}){1,3}(?!\d)|(?<![\d.$])\d{6,10}(?!\d)")


def detectar_contextual(txt):
    out = {("mail", m.group()) for m in PAT_MAIL.finditer(txt)}
    out |= {("tel", "".join(m.groups())) for m in PAT_TEL.finditer(txt)}
    out |= {("cc", re.sub(r"\D", "", m.group(1))) for m in PAT_CC.finditer(txt)}
    return out


def detectar_ingenuo(txt):
    out = {("mail", m.group()) for m in PAT_MAIL.finditer(txt)}
    out |= {("tel", "".join(m.groups())) for m in PAT_TEL.finditer(txt)}
    out |= {("cc", re.sub(r"\D", "", m.group())) for m in PAT_NAIVE.finditer(txt)}         # cualquier numero largo
    return out


def evaluar(detector, textos, verdad):
    tp = fp = fn = 0
    for t, v in zip(textos, verdad):
        d = detector(t); tp += len(d & v); fp += len(d - v); fn += len(v - d)
    return tp / (tp + fp), tp / (tp + fn), tp, fp, fn


def redactar(txt):
    txt = PAT_MAIL.sub("[EMAIL]", txt)
    txt = PAT_TEL.sub("[TEL]", txt)
    return PAT_CC.sub(lambda m: m.group(0).replace(m.group(1), "[CC]"), txt)


textos, verdad = generar_mensajes()
tot = sum(len(v) for v in verdad)
print(f"{len(textos)} mensajes | {tot} valores de PII sembrados (cédulas, correos, teléfonos) | ejemplo:\n  «{textos[0][:150]}»")
for nombre, det in (("ingenuo (cualquier número largo)", detectar_ingenuo), ("contextual (palabra clave + formato)", detectar_contextual)):
    p, r, tp, fp, fn = evaluar(det, textos, verdad)
    print(f"{nombre:38s} precisión {p:.3f} | recobrado {r:.3f} | TP {tp} FP {fp} FN {fn}")

ejemplo = next(t for t, v in zip(textos, verdad) if len(v) >= 3)
print("\nOriginal :", ejemplo[:200]); print("Redactado:", redactar(ejemplo)[:200])
residual = sum(len(detectar_contextual(redactar(t))) for t in textos)
assert residual == 0 and all(redactar(redactar(t)) == redactar(t) for t in textos)         # redactar elimina lo detectable y es idempotente
assert evaluar(detectar_contextual, textos, verdad)[1] > 0.99 and evaluar(detectar_ingenuo, textos, verdad)[0] < evaluar(detectar_contextual, textos, verdad)[0]
````

**Lectura.** El patrón **ingenuo** tiene recobrado de 1.0 pero **precisión de 0.325** (1 220 falsos positivos frente a 588 aciertos: números de pedido, facturas y montos largos se confunden con cédulas); el **contextual** exige palabra clave y formato y en estos datos no se equivoca (precisión y recobrado de 1.0). La contraparte honesta: **los datos los escribimos nosotros** con esas mismas palabras clave, así que el 1.0 es optimista; y este último **solo ve** cédulas con la palabra clave al lado; una cédula suelta ("el 1052345678 no pasa") se le escaparía. En la práctica se combinan reglas con **validaciones adicionales** (longitud, prefijos, dígitos verificadores cuando existen) y modelos de **reconocimiento de entidades**; y se **redacta** antes de que el texto llegue a registros, tableros o conjuntos de entrenamiento. Detectar PII **no sustituye** controlar de dónde sale: es la **red de seguridad**.

### 4.2 Reidentificación por enlace y $k$-anonimato desde cero

"Le quité el nombre, es anónimo" es una **falacia**. Los **cuasi-identificadores** (*quasi-identifiers*: fecha de nacimiento, sexo, municipio…) no identifican por sí solos, pero **combinados** son casi únicos, y existen **tablas públicas** (padrones, directorios, redes sociales) que los asocian con nombres. El **ataque de enlace** (*linkage attack*, el que popularizó el trabajo de Latanya Sweeney; ver [Data Mining, Módulo 00](../../Data%20Mining/00%20-%20Introduccion%20al%20Data%20Mining/04_Etica_Privacidad_y_Gobernanza_en_Mineria_de_Datos.ipynb)) une ambas por esos atributos. Lo hacemos **a escala de DataFrame** (vectorizado, miles de registros):

- **Publicación "anónima"** (salud): fecha de nacimiento, sexo, municipio y **diagnóstico** (el dato sensible), sin nombre.
- **Padrón público** (ficticio): nombre + los mismos tres atributos.

Un registro de salud queda **reidentificado** si su combinación de atributos coincide con **una sola** persona del padrón. El **$k$-anonimato** exige que **cada combinación** de cuasi-identificadores aparezca en **al menos $k$ registros**; se logra **generalizando** (año en vez de fecha, quinquenio, provincia en vez de municipio) o **suprimiendo** registros raros. Medimos $k$ (el tamaño de la clase más pequeña), el porcentaje de registros **únicos**, y la **tasa de reidentificación** del enlace, para una **escalera de generalización**.

````python
# ============================================================
# 4.2 Ataque de enlace + k-anonimato desde cero, con escalera de generalizacion
# ============================================================
PROVINCIA = {"Tunja": "Centro", "Motavita": "Centro", "Cómbita": "Centro", "Paipa": "Tundama", "Duitama": "Tundama", "Sogamoso": "Sugamuxi",
             "Aquitania": "Sugamuxi", "Chiquinquirá": "Occidente", "Villa de Leyva": "Ricaurte", "Samacá": "Centro", "Garagoa": "Neira", "Soatá": "Norte"}
MUNICIPIOS = list(PROVINCIA)                                    # provincias: agrupacion ILUSTRATIVA, no la division oficial
DIAGNOSTICOS = ["diabetes", "hipertensión", "asma", "depresión", "VIH", "ninguno"]


def generar_poblacion(n, semilla=SEED):
    rng = np.random.default_rng(semilla)
    nac = pd.Timestamp("1955-01-01") + pd.to_timedelta(rng.integers(0, 18_250, n), unit="D")
    pob = pd.DataFrame({"persona": [f"Persona {i:06d}" for i in range(n)], "nacimiento": nac, "sexo": rng.choice(["F", "M"], n),
                         "municipio": rng.choice(MUNICIPIOS, n, p=np.r_[[0.22, 0.04, 0.03, 0.08, 0.17, 0.2, 0.02, 0.07, 0.04, 0.03, 0.05, 0.05]]),
                         "diagnostico": rng.choice(DIAGNOSTICOS, n, p=[0.12, 0.2, 0.1, 0.08, 0.02, 0.48])})
    # correlacion sembrada (ficticia): los hombres de Soata nacidos antes de 1965 comparten diagnostico -> clases homogeneas
    pob.loc[(pob.municipio == "Soatá") & (pob.sexo == "M") & (pob.nacimiento.dt.year < 1965), "diagnostico"] = "diabetes"
    return pob


def generalizar(df, nivel):
    """Escalera: 0 exacto | 1 anio + municipio | 2 quinquenio + provincia | 3 decada + provincia."""
    g = pd.DataFrame({"sexo": df.sexo.values}, index=df.index)
    anio = df.nacimiento.dt.year
    g["nacimiento"] = {0: df.nacimiento.dt.strftime("%Y-%m-%d"), 1: anio.astype(str), 2: (anio // 5 * 5).astype(str), 3: (anio // 10 * 10).astype(str)}[nivel]
    g["lugar"] = df.municipio if nivel <= 1 else df.municipio.map(PROVINCIA)
    return g


QI = ["nacimiento", "sexo", "lugar"]
def tamanos_de_clase(g, qis=QI):
    return g.groupby(qis).size()
def k_anonimato(g, qis=QI):
    """k = tamano de la clase de equivalencia mas pequena."""
    return int(tamanos_de_clase(g, qis).min())


N_POB = 20_000 if RAPIDO else 100_000
poblacion = generar_poblacion(N_POB)
salud = poblacion.sample(N_POB // 2, random_state=SEED)                                   # la publicacion "anonima": la mitad de la poblacion, con diagnostico
filas = []
for nivel, etiqueta in enumerate(["0 exacto", "1 año + municipio", "2 quinquenio + provincia", "3 década + provincia"]):
    gs, gp = generalizar(salud, nivel), generalizar(poblacion, nivel)
    cand = gs.merge(tamanos_de_clase(gp).rename("candidatos").reset_index(), on=QI, how="left")["candidatos"]   # personas del padron que encajan con cada registro
    filas.append({"nivel": etiqueta, "k (publicación)": k_anonimato(gs), "% registros únicos": round(100 * (gs.groupby(QI).transform("size") == 1).mean(), 1),
                  "% reidentificados (1 candidato)": round(100 * (cand == 1).mean(), 1), "prob. media de acierto": round((1 / cand).mean(), 3), "clases": gs.groupby(QI).ngroups})
escalera = pd.DataFrame(filas)
print(f"Publicación de salud: {len(salud):,} registros | padrón público: {len(poblacion):,} personas\n")
print(escalera.to_string(index=False))

# supresion: quitar los registros de clases con menos de 5 para lograr k >= 5 SIN generalizar mas (nivel 1)
g1 = generalizar(salud, 1); tam = g1.groupby(QI)["sexo"].transform("size")
print(f"\nPara lograr k>=5 en el nivel 1 hay que suprimir el {100 * (tam < 5).mean():.1f} % de los registros ({int((tam < 5).sum())} de {len(g1):,}); el resto queda con k = {k_anonimato(g1[tam >= 5])}.")
assert k_anonimato(g1[tam >= 5]) >= 5
assert escalera["% reidentificados (1 candidato)"].is_monotonic_decreasing and escalera.loc[0, "% reidentificados (1 candidato)"] > 50
assert escalera.loc[3, "% reidentificados (1 candidato)"] == 0
````

Es el **compromiso privacidad-utilidad**. Y el $k$-anonimato tiene un **límite conocido**: protege contra *"¿cuál fila eres tú?"*, **no** contra *"qué tienes"*. Si todas las personas de una clase comparten el mismo diagnóstico, el atacante lo sabe aunque no sepa cuál fila eres (**ataque de homogeneidad**); en nuestros datos sembramos una correlación ficticia (los hombres de Soatá nacidos antes de 1965 comparten diagnóstico) y ocurre aun con $k \geq 11$. La corrección es la **$l$-diversidad** (cada clase con al menos $l$ valores distintos del atributo sensible). Lo medimos en la práctica siguiente.

## 5. Dos "Fugas" que No Hay que Confundir ⚠️

El mismo nombre en inglés, **dos problemas distintos**:

| | *Data leakage* de **seguridad** (este módulo) | *Data leakage* en **Machine Learning** |
|---|---|---|
| **Qué es** | Información sensible **sale** de donde debía estar | Información del **futuro o del conjunto de prueba** **entra** al entrenamiento |
| **Quién sufre** | Los **titulares** de los datos (privacidad) y la organización | La **validez del modelo**: métricas infladas que no se sostienen en producción |
| **Ejemplo** | Un *log* con cédulas; una tabla "anónima" reidentificable | Normalizar con **todos** los datos antes de dividir; usar una variable calculada **después** del evento a predecir |
| **Defensa** | Control de acceso, cifrado, minimización, redacción, $k$-anonimato | Dividir primero y ajustar el preprocesamiento **solo** con el entrenamiento (`Pipeline`) |

> 📘 La fuga de ML (contaminación entre entrenamiento y prueba, *target leakage*) se desarrolla con código en [Machine Learning, Módulo 00 (importancia del preprocesamiento)](../../Machine%20Learning/00%20-%20Introduccion%20al%20Machine%20Learning/03_Importancia_del_Preprocesamiento.ipynb); aquí solo la distinguimos. Hay un punto de **contacto**: un modelo entrenado con datos personales puede **memorizar** y filtrar ejemplos (ataques de inferencia de pertenencia), de modo que lo que se **publica** (también modelos) debe pasar por el mismo análisis de riesgo.

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Tríada CIA** | Confidencialidad, integridad y disponibilidad: lo que hay que **proteger** en los datos |
| **STRIDE** | Seis familias de amenazas (suplantación, alteración, repudio, divulgación, denegación de servicio, escalada) para aplicar **sobre cada flecha y caja** del flujo de datos |
| **Brecha de datos** | Se ve como **filas en una bitácora**: la detección combina **reglas con umbral por ventana** (explicables) y **modelos de anomalía** (lista priorizada para revisar) |
| **Fuerza bruta** | El umbral solo no basta: la **clave de agrupación** (IP + usuario) elimina la falsa alarma de la oficina sin perder ataques |
| **Acceso no autorizado** | Matriz de control de acceso + **privilegio mínimo**; con la matriz, la detección es exacta; lo difícil es **mantenerla** y **registrar** |
| **PII en texto** | Regex **contextuales** superan a las ingenuas; ninguna es perfecta; se **redacta** antes de que el texto circule |
| **Reidentificación** | "Sin nombre" **no es** anónimo: el enlace por cuasi-identificadores reidentifica a casi todos; **$k$-anonimato** y generalización reducen el riesgo a cambio de utilidad; la **homogeneidad** exige $l$-diversidad |
| **Dos fugas** | *Data leakage* de **seguridad** (sale información) $\neq$ de **ML** (entra información del futuro) |
