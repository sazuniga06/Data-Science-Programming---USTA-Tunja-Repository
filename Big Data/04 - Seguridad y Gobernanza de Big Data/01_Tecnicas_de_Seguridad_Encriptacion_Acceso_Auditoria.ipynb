## 1. Cifrado: Simétrico, Asimétrico y por Sobre 🔑

**Cifrar** transforma datos legibles (*texto plano*) en ilegibles (*texto cifrado*) con una **clave**; sin la clave no se recupera. Dos familias:

| | **Simétrico** (misma clave) | **Asimétrico** (par de claves) |
|---|---|---|
| **Idea** | Una sola clave cifra y descifra | Clave **pública** (cifra o verifica) y **privada** (descifra o firma) |
| **Ejemplos** | AES (con Fernet en `cryptography`) | RSA con OAEP (cifrar); RSA con PSS (firmar) |
| **Velocidad** | **Muy rápido** (sirve para gigabytes) | **Lento** y limitado a mensajes **cortos** |
| **Problema** | ¿Cómo compartir la clave sin que la vean? | Resuelve el reparto de claves y permite **firmas** |
| **Uso típico** | Cifrar los datos en el lago o el disco | Intercambiar claves, firmar, identificar servidores |

**Fernet** (de `cryptography`) es una receta **autenticada**: cifra con AES y añade una etiqueta HMAC, de modo que **alterar** el texto cifrado se **detecta** al descifrar (`InvalidToken`). Cifrar **sin autenticar** permite alterar datos sin que se note; por eso se usan siempre modos autenticados. **RSA-OAEP** cifra mensajes cortos con la clave pública; **RSA-PSS** firma: quien tiene la clave privada firma, cualquiera con la pública verifica **integridad y origen**.

**Cifrado por sobre** (*envelope encryption*): lo estándar en Big Data. Se combinan ambas ideas para manejar claves a escala:
1. Cada dato (o archivo, o partición) se cifra con su **clave de datos** (**DEK**, *data encryption key*), simétrica y rápida.
2. Cada DEK se **cifra ("envuelve")** con una **clave maestra** (**KEK**, *key encryption key*) que vive en un **KMS/HSM** y **nunca sale** de él.
3. Se guarda el dato cifrado junto con la DEK **envuelta**. Para leer: el KMS desenvuelve la DEK y con ella se descifra el dato.

Ventajas: **rotar** la KEK solo exige **re-envolver las DEK** (bytes diminutos) en lugar de re-cifrar petabytes; y se puede **borrar un dato destruyendo su DEK** (***crypto-shredding***), útil para el derecho de supresión cuando hay copias en respaldos.

**En reposo y en tránsito (conceptual).**

| Dónde están los datos | Amenaza principal | Defensa típica | Qué **no** cubre |
|---|---|---|---|
| **En reposo** (discos, lago, respaldos) | Robo de discos o respaldos; acceso al almacenamiento sin pasar por la aplicación | Cifrado de volumen/almacenamiento; cifrado por columna o por archivo; cifrado por sobre | Un usuario **autenticado** con permiso de lectura **ve** los datos descifrados |
| **En tránsito** (red) | Interceptar o alterar mensajes entre servicios | TLS (con verificación de certificados), VPN, *mutual TLS* entre servicios | Los datos **ya descifrados** en el origen o el destino |
| **En uso** (memoria/CPU) | Volcados de memoria, administradores del servidor | Mínimo privilegio, enclaves de hardware, cifrado homomórfico *(avanzado)* | Casi nada es práctico a escala todavía |

> ⚠️ Cifrar **no reemplaza** el control de acceso ni la auditoría: protege contra **un tipo** de ataque (acceso a medios), no contra el **abuso de un permiso legítimo** (que vimos en el cuaderno 00).

````python
# ============================================================
# 1. Fernet (simetrico), RSA-OAEP (cifrado asimetrico) y RSA-PSS (firma) con la biblioteca `cryptography`
# ============================================================
# !pip install -q cryptography numpy pandas matplotlib

import hashlib
import hmac
import json
import secrets
import time
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from cryptography.exceptions import InvalidSignature
from cryptography.fernet import Fernet, InvalidToken
from cryptography.hazmat.primitives import hashes
from cryptography.hazmat.primitives.asymmetric import padding, rsa

warnings.filterwarnings("ignore")
SEED = 42
plt.rcParams.update({"figure.figsize": (9, 5), "axes.spines.top": False, "axes.spines.right": False})

# ---- simetrico: Fernet (AES + HMAC)
clave = Fernet.generate_key()
f = Fernet(clave)
mensaje = "Reporte confidencial: ventas de la sucursal de Tunja".encode()
token = f.encrypt(mensaje)
assert f.decrypt(token) == mensaje and token != mensaje
alterado = bytearray(token); alterado[40] ^= 1                                # un solo bit cambiado
for etiqueta, t, ff in (("token alterado", bytes(alterado), f), ("clave equivocada", token, Fernet(Fernet.generate_key()))):
    try:
        ff.decrypt(t); print(etiqueta, "-> descifró (¡mal!)"); raise AssertionError
    except InvalidToken:
        print(f"{etiqueta:18s} -> InvalidToken (se detecta)")

# ---- asimetrico: RSA-2048 con OAEP; firma con PSS
privada = rsa.generate_private_key(public_exponent=65537, key_size=2048)
publica = privada.public_key()
oaep = padding.OAEP(mgf=padding.MGF1(hashes.SHA256()), algorithm=hashes.SHA256(), label=None)
secreto = b"clave-de-sesion-0123456789"
assert privada.decrypt(publica.encrypt(secreto, oaep), oaep) == secreto
maximo = 2048 // 8 - 2 * hashes.SHA256.digest_size - 2                       # tamano maximo del mensaje con OAEP-SHA256
try:
    publica.encrypt(b"x" * (maximo + 1), oaep); raise AssertionError
except ValueError:
    print(f"RSA-2048/OAEP solo cifra mensajes de hasta {maximo} bytes: no sirve para cifrar datos grandes")
pss = padding.PSS(mgf=padding.MGF1(hashes.SHA256()), salt_length=padding.PSS.MAX_LENGTH)
firma = privada.sign(mensaje, pss, hashes.SHA256())
publica.verify(firma, mensaje, pss, hashes.SHA256())                           # no lanza excepcion: firma valida
try:
    publica.verify(firma, mensaje + b"!", pss, hashes.SHA256()); raise AssertionError
except InvalidSignature:
    print("Firma: un mensaje alterado NO verifica (integridad + origen)")

# ---- velocidad: por que no se cifran datos grandes con RSA
datos = secrets.token_bytes(1_000_000)
t0 = time.perf_counter(); f.decrypt(f.encrypt(datos)); mb_s_fernet = 2 / (time.perf_counter() - t0)       # cifrar + descifrar 1 MB = 2 MB procesados
bloques = [publica.encrypt(secrets.token_bytes(maximo), oaep) for _ in range(20)]
t0 = time.perf_counter(); [privada.decrypt(b, oaep) for b in bloques]; mb_s_rsa = 20 * maximo / 1e6 / (time.perf_counter() - t0)
print(f"\nRendimiento medido (esta máquina): Fernet ≈ {mb_s_fernet:,.0f} MB/s | RSA-2048 (descifrar) ≈ {mb_s_rsa:.2f} MB/s -> Fernet es ≈ {mb_s_fernet / mb_s_rsa:,.0f} veces más rápido")
assert mb_s_fernet > 50 * mb_s_rsa
````

````python
# ============================================================
# 1b. DEMO DIDACTICA de cifrado por sobre (envelope encryption): KMS de juguete + DEK por registro
#     (en produccion: KMS/HSM de la nube o Vault; la KEK NUNCA sale de ese servicio)
# ============================================================
class KMSDidactico:
    """Servicio de claves de juguete en memoria. Guarda las KEK (por version) y solo expone envolver/desenvolver."""
    def __init__(self):
        self._keks, self.version = {}, 0
        self.rotar()

    def rotar(self):
        self.version += 1
        self._keks[self.version] = Fernet.generate_key()

    def envolver(self, dek):
        return self.version, Fernet(self._keks[self.version]).encrypt(dek)

    def desenvolver(self, version, dek_envuelta):
        return Fernet(self._keks[version]).decrypt(dek_envuelta)

    def reenvolver(self, version, dek_envuelta):
        return self.envolver(self.desenvolver(version, dek_envuelta))

    def destruir(self, version):
        del self._keks[version]


kms = KMSDidactico()
rng = np.random.default_rng(SEED)
registros = {f"cliente-{i:03d}": json.dumps({"nombre": f"Cliente {i}", "saldo": int(rng.integers(0, 5_000_000)), "nota": "x" * 2000}).encode() for i in range(200)}

lago, boveda = {}, {}                                          # el lago guarda el dato CIFRADO; la tabla de claves, las DEK ENVUELTAS (otro sistema)
for k, dato in registros.items():
    dek = Fernet.generate_key()                                # una DEK por registro
    lago[k] = Fernet(dek).encrypt(dato)
    boveda[k] = kms.envolver(dek)


def leer(k):
    version, dek_env = boveda[k]
    return Fernet(kms.desenvolver(version, dek_env)).decrypt(lago[k])


assert all(leer(k) == registros[k] for k in registros)
print(f"{len(lago)} registros cifrados, cada uno con su propia DEK | tamaño del lago: {sum(map(len, lago.values())) / 1e3:,.0f} kB | tamaño de las DEK envueltas: {sum(len(v[1]) for v in boveda.values()) / 1e3:,.1f} kB")

# --- rotacion de la KEK: solo se re-envuelven las DEK; los datos cifrados NO se tocan
lago_antes = dict(lago)
kms.rotar()
boveda = {k: kms.reenvolver(*v) for k, v in boveda.items()}
assert lago == lago_antes and all(v[0] == 2 for v in boveda.values()) and all(leer(k) == registros[k] for k in registros)
print(f"Rotación de KEK (v1 -> v2): se reescribieron {sum(len(v[1]) for v in boveda.values()) / 1e3:,.1f} kB de DEK; los {sum(map(len, lago.values())) / 1e3:,.0f} kB de datos quedaron idénticos")

# --- crypto-shredding: borrar la DEK envuelta de UN cliente deja su dato ilegible aunque el cifrado siga en el lago y en respaldos
respaldo_lago = dict(lago)                                     # un respaldo del lago que nadie puede "borrar"
del boveda["cliente-007"]
try:
    leer("cliente-007"); raise AssertionError
except KeyError:
    print("cliente-007: sin su DEK el dato cifrado (también el del respaldo) es irrecuperable -> supresión efectiva")
assert leer("cliente-008") == registros["cliente-008"] and respaldo_lago["cliente-007"] == lago["cliente-007"]

# --- destruir la KEK vieja: cualquier copia de una DEK envuelta con la v1 ya no se puede abrir
kms.destruir(1)
try:
    kms.desenvolver(1, Fernet(Fernet.generate_key()).encrypt(b"x")); raise AssertionError
except KeyError:
    print("KEK v1 destruida: las copias antiguas de DEK envueltas con ella dejan de ser útiles")
assert leer("cliente-001") == registros["cliente-001"]        # lo vigente (envuelto con v2) sigue funcionando
````

---
**Lectura.** `Fernet` detecta la alteración de un solo bit y la clave equivocada (ambos lanzan `InvalidToken`), RSA-OAEP solo cifra mensajes cortos (190 bytes con RSA-2048 y SHA-256) y es **cientos de veces más lento** que Fernet; por eso lo que se hace en la práctica es **cifrar los datos con una clave simétrica y proteger esa clave** (cifrado por sobre). En la demostración, rotar la KEK reescribe solo las DEK envueltas (kilobytes) y los datos cifrados quedan **idénticos**; y destruir la DEK de un cliente vuelve su dato **irrecuperable** aun en el respaldo (*crypto-shredding*). El KMS es de **juguete**: lo que se aprende es la **lógica**, no el servicio.

## 2. Contraseñas, *Hashing* con Sal y Derivación de Claves 🧂

Una función ***hash*** convierte cualquier entrada en una huella de tamaño fijo, **determinista** y (idealmente) **irreversible**. Para **contraseñas** hay una trampa: las funciones como **MD5**, **SHA-1** o **SHA-256** están diseñadas para ser **rápidas**, y la rapidez es lo **peor** que puede ofrecerle uno al atacante que robó la tabla de claves (cada intento cuesta casi nada). Tres defensas, acumulativas:

| Defensa | Qué evita | Qué **no** evita |
|---|---|---|
| **Sal** (*salt*) aleatoria y única por usuario | Tablas precalculadas (*rainbow tables*) y que dos usuarios con la misma clave tengan el mismo *hash* | El **ataque de diccionario** por usuario con una función rápida |
| **Función lenta** con muchas iteraciones (`pbkdf2_hmac`; también existen bcrypt, scrypt y Argon2, **verificar** cuál ofrece tu plataforma) | Hace cada intento **costoso** (miles de veces más lento) | Claves tan débiles que están en el diccionario |
| **Política de claves** + MFA | Claves triviales; el robo de una sola clave | — |

La derivación de claves es la misma idea en otro uso: convertir una contraseña en una **clave criptográfica** (`pbkdf2_hmac` con sal e iteraciones). Reglas: sal de `secrets.token_bytes` (nunca de `random`), comparación en **tiempo constante** (`hmac.compare_digest`) y **nunca** implementar el esquema a mano para uso real. Para el experimento fabricamos una **tabla de contraseñas filtrada** de 300 usuarios **ficticios** (85 % con claves débiles de un diccionario propio, 15 % aleatorias) y medimos cuánto le cuesta a un atacante **hipotético** (nuestro propio script) recuperarlas con cada esquema.

````python
# ============================================================
# 2. Tabla de contrasenas filtrada (SINTETICA): SHA-1 sin sal vs SHA-256 con sal vs PBKDF2 con sal
# ============================================================
def tasa(fn, n):
    t0 = time.perf_counter()
    for _ in range(n):
        fn()
    return n / (time.perf_counter() - t0)


pw, sal = b"contrasena-de-ejemplo", secrets.token_bytes(16)
ITER_REAL = 100_000                                               # orden de magnitud tipico; verificar la recomendacion vigente (OWASP/NIST)
tasas = pd.Series({"MD5": tasa(lambda: hashlib.md5(pw).digest(), 100_000), "SHA-1": tasa(lambda: hashlib.sha1(pw).digest(), 100_000),
                   "SHA-256": tasa(lambda: hashlib.sha256(pw).digest(), 100_000),
                   f"PBKDF2-HMAC-SHA256 ({ITER_REAL:,} iter.)": tasa(lambda: hashlib.pbkdf2_hmac("sha256", pw, sal, ITER_REAL), 5)})
costo = pd.DataFrame({"intentos por segundo (1 núcleo)": tasas.round(0).astype(int), "horas para 10^9 intentos": (1e9 / tasas / 3600).round(2)})
print(costo.to_string())

# --- funciones de produccion (con biblioteca estandar): sal aleatoria, iteraciones, comparacion en tiempo constante
def hash_clave(clave: str, iteraciones=ITER_REAL):
    s = secrets.token_bytes(16)
    h = hashlib.pbkdf2_hmac("sha256", clave.encode(), s, iteraciones)
    return f"pbkdf2_sha256${iteraciones}${s.hex()}${h.hex()}"


def verificar_clave(clave: str, guardado: str):
    _, it, s, h = guardado.split("$")
    nuevo = hashlib.pbkdf2_hmac("sha256", clave.encode(), bytes.fromhex(s), int(it))
    return hmac.compare_digest(nuevo.hex(), h)


g = hash_clave("tunja2025")
assert verificar_clave("tunja2025", g) and not verificar_clave("tunja2026", g) and hash_clave("tunja2025") != g     # la sal hace distintos dos hashes de la misma clave

# --- la "tabla filtrada": 300 usuarios, 85 % con claves debiles de un diccionario
DICC = ["123456", "password", "colombia", "tunja2024", "boyaca123", "qwerty", "admin123", "futbol", "santafe", "nacional", "123456789", "abc123",
        "contrasena", "bienvenido", "tunja", "boyaca", "duitama", "sogamoso", "paipa", "miamor", "familia", "dios123", "12345678", "iloveyou", "pass1234"]
CANDIDATAS = DICC + [w + s for w in DICC[:12] for s in ("1", "123")]   # diccionario del "atacante" (nuestro propio script): 49 candidatas
rng = np.random.default_rng(SEED)
usuarios = [f"u{i:03d}" for i in range(300)]
claves = [str(rng.choice(CANDIDATAS)) if rng.random() < 0.85 else secrets.token_urlsafe(9) for _ in usuarios]
debiles = [c in set(CANDIDATAS) for c in claves]

# Esquema A: SHA-1 SIN sal -> el atacante precalcula UNA tabla de candidatas y busca
tabla_a = {u: hashlib.sha1(c.encode()).hexdigest() for u, c in zip(usuarios, claves)}
duplicados_a = len(tabla_a) - len(set(tabla_a.values()))
pre = {hashlib.sha1(c.encode()).hexdigest(): c for c in CANDIDATAS}
rotas_a = sum(h in pre for h in tabla_a.values())

# Esquema B: SHA-256 CON sal por usuario -> hay que probar cada candidata CONTRA cada usuario (rapido igual)
sales = {u: secrets.token_bytes(16) for u in usuarios}
tabla_b = {u: hashlib.sha256(sales[u] + c.encode()).hexdigest() for u, c in zip(usuarios, claves)}
duplicados_b = len(tabla_b) - len(set(tabla_b.values()))
t0 = time.perf_counter()
rotas_b = sum(any(hashlib.sha256(sales[u] + c.encode()).hexdigest() == h for c in CANDIDATAS) for u, h in tabla_b.items())
t_b = time.perf_counter() - t0

# Esquema C: PBKDF2 con sal -> mismo ataque, pero cada intento es mucho mas caro. Se ejecuta de verdad con pocas iteraciones y se extrapola
ITER_DEMO = 5_000
t0 = time.perf_counter()
muestra = usuarios[:6]
tabla_c = {u: hashlib.pbkdf2_hmac("sha256", claves[usuarios.index(u)].encode(), sales[u], ITER_DEMO) for u in muestra}
rotas_c = sum(any(hashlib.pbkdf2_hmac("sha256", c.encode(), sales[u], ITER_DEMO) == tabla_c[u] for c in CANDIDATAS) for u in muestra)
t_c = time.perf_counter() - t0
intentos_total = len(usuarios) * len(CANDIDATAS)
est_real = intentos_total / tasas.iloc[-1] / 60

res = pd.DataFrame({"A: SHA-1 sin sal": [duplicados_a, len(CANDIDATAS), rotas_a], "B: SHA-256 + sal": [duplicados_b, intentos_total, rotas_b]},
                   index=["hashes repetidos en la tabla", "cálculos de hash del atacante", "claves recuperadas (de 300)"])
print(f"\nClaves débiles en la tabla: {sum(debiles)} de {len(usuarios)}\n" + res.to_string())
print(f"B tomó {t_b:.2f} s para {intentos_total:,} intentos: la sal NO frenó el ataque de diccionario con un hash rápido.")
print(f"C: con {ITER_DEMO:,} iteraciones se rompieron {rotas_c} de {len(muestra)} usuarios de muestra (los demás tienen claves aleatorias) en {t_c:.1f} s; con {ITER_REAL:,} iteraciones el mismo ataque a los 300 usuarios costaría ≈ {est_real:.0f} min (estimado con la tasa medida).")
assert rotas_a == sum(debiles) and rotas_b == sum(debiles) and duplicados_a > 50 and duplicados_b == 0 and rotas_c == sum(debiles[:6]) and tasas.iloc[0] > 100 * tasas.iloc[-1]
````

---
**Lectura.** (1) Con **SHA-1 sin sal**, el atacante hace **49 cálculos** (una tabla de candidatas) y recupera **todas las claves débiles**, y además la tabla delata quién comparte clave (hashes repetidos). (2) Con **sal**, no hay repetidos y se acaban las tablas precalculadas, pero con un *hash* **rápido** el atacante igual recupera **las mismas** claves (probar 14 700 combinaciones tarda una fracción de segundo): **la sal no basta**. (3) Con una **función lenta** cada intento cuesta miles de veces más; con 100 000 iteraciones el mismo ataque pasa de una fracción de segundo a **del orden de diez minutos** (estimado con la tasa medida; varía con la máquina) y **crece** con la cantidad de candidatas y de usuarios. Pero **ninguna** función frena la clave que **está en el diccionario**: la última defensa es la **política de claves** y el MFA. Eso explica por qué MD5/SHA-1 **sin sal** no sirven para contraseñas: son rápidos y, sin sal, hasta se **buscan** en una tabla.

## 3. Enmascaramiento, Seudonimización y Tokenización 🎭

Hay que **usar** datos sensibles sin **exponerlos**. Tres técnicas con propiedades distintas:

| Técnica | ¿Reversible? | ¿Enlazable entre tablas? | Cómo | Uso típico |
|---|---|---|---|---|
| **Enmascarar** (*masking*) | No | No | Ocultar parte del valor (`c***@dominio.com`, `*****678`) | **Mostrar** datos a personas (pantallas, reportes) |
| **Seudonimizar** con **HMAC** (clave secreta) | No (sin la clave no se recupera ni se adivina) | **Sí**: el mismo valor da el mismo seudónimo | `hmac(clave, valor)` | **Análisis** que necesita unir tablas sin ver la identidad |
| **Tokenizar** con bóveda | **Sí**, con acceso a la bóveda | Sí (si el mismo valor reutiliza el token) | Token aleatorio + tabla `token → valor` **en otro sistema** | Procesar pagos o documentos y **recuperar** el original solo cuando se debe |

**Por qué `hash(valor)` a secas no seudonimiza.** Un identificador de **pocos valores posibles** (una cédula, un teléfono) se puede **adivinar probándolos todos**: se calcula el *hash* de cada valor posible y se compara. Un **HMAC con clave secreta** no lo permite porque el atacante **no puede** calcular los candidatos sin la clave. (El [módulo 00 de Data Mining](../../Data%20Mining/00%20-%20Introduccion%20al%20Data%20Mining/04_Etica_Privacidad_y_Gobernanza_en_Mineria_de_Datos.ipynb) usa sal; con identificadores de pocos valores, una **clave secreta compartida** (no una sal por registro, que impediría el enlace) es lo que mantiene el seudónimo estable **y** seguro, mientras la clave esté protegida.) La seudonimización **sigue siendo dato personal** bajo la normativa (la clave permite, en principio, reconstruir el vínculo): se trata en el [cuaderno 02](02_Gobernanza_de_Datos.ipynb).

````python
# ============================================================
# 3. Enmascarar, seudonimizar con HMAC (clave secreta) y tokenizar con boveda; ataque de enumeracion sobre un hash simple
# ============================================================
def enmascarar_email(e):
    usuario, dominio = e.split("@")
    return usuario[0] + "*" * (len(usuario) - 1) + "@" + dominio


def enmascarar_cc(c):
    return "*" * (len(c) - 3) + c[-3:]


assert enmascarar_email("camila.rojas@example.com") == "c" + "*" * 11 + "@example.com" and enmascarar_cc("1052345678") == "*******678"

CLAVE_SEUDONIMO = secrets.token_bytes(32)                           # en produccion vive en un KMS, NO junto a los datos


def seudonimo(valor, clave=CLAVE_SEUDONIMO):
    return hmac.new(clave, str(valor).encode(), hashlib.sha256).hexdigest()[:20]


class BovedaTokens:
    """Tokenizacion con boveda (DIDACTICA): token aleatorio sin relacion matematica con el valor; la tabla vive en OTRO sistema."""
    def __init__(self):
        self._v2t, self._t2v = {}, {}

    def tokenizar(self, valor):
        if valor not in self._v2t:
            t = "tok_" + secrets.token_hex(8)
            self._v2t[valor], self._t2v[t] = t, valor
        return self._v2t[valor]

    def destokenizar(self, token):
        return self._t2v[token]


rng = np.random.default_rng(SEED)
cedulas = rng.choice(np.arange(1_000_000, 1_200_000), 3000, replace=False)                 # espacio PEQUENO de cedulas (200 000 posibles): como un barrio
clientes = pd.DataFrame({"cc": cedulas, "ciudad": rng.choice(["Tunja", "Duitama", "Sogamoso"], 3000)})
compras = pd.DataFrame({"cc": rng.choice(cedulas, 9000), "monto": rng.integers(5_000, 400_000, 9000)})
union_original = len(clientes.merge(compras, on="cc"))

# Seudonimizacion con HMAC: ambas tablas con la misma clave -> el JOIN sigue funcionando
c_ps, p_ps = clientes.assign(cc=clientes.cc.map(seudonimo)), compras.assign(cc=compras.cc.map(seudonimo))
assert len(c_ps.merge(p_ps, on="cc")) == union_original
print(f"JOIN original: {union_original:,} filas | JOIN con seudónimos HMAC: {len(c_ps.merge(p_ps, on='cc')):,} filas (enlazable, sin mostrar la cédula)")

# Ataque de enumeracion (defensivo, sobre nuestros propios datos): probar las 200 000 cedulas posibles
todas = np.arange(1_000_000, 1_200_000)
hash_simple = clientes.cc.map(lambda c: hashlib.sha256(str(c).encode()).hexdigest())           # "seudonimo" ingenuo: SHA-256 a secas
t0 = time.perf_counter(); tabla_inv = {hashlib.sha256(str(c).encode()).hexdigest(): c for c in todas}; t_enum = time.perf_counter() - t0
recuperadas_simple = int((hash_simple.map(tabla_inv) == clientes.cc).sum())
tabla_inv_hmac = {hashlib.sha256(str(c).encode()).hexdigest()[:20]: c for c in todas}          # el atacante NO tiene la clave: usa SHA-256 a secas
recuperadas_hmac = int(sum(s in tabla_inv_hmac for s in c_ps.cc))
print(f"SHA-256 a secas: el atacante recupera {recuperadas_simple:,} de {len(clientes):,} cédulas en {t_enum:.2f} s | HMAC con clave: recupera {recuperadas_hmac}")
assert recuperadas_simple == len(clientes) and recuperadas_hmac == 0
assert seudonimo(1000001, secrets.token_bytes(32)) != seudonimo(1000001)                       # con otra clave el seudonimo cambia

# Tokenizacion con boveda
b = BovedaTokens()
t1, t2 = b.tokenizar("1052345678"), b.tokenizar("1052345678")
assert t1 == t2 and b.destokenizar(t1) == "1052345678" and t1 != b.tokenizar("1052345679")
print("\nToken:", t1, "-> original (solo con acceso a la bóveda):", b.destokenizar(t1))
print("Enmascarado para pantalla:", enmascarar_email("camila.rojas@example.com"), "|", enmascarar_cc("1052345678"))
````

---
**Lectura.** Con 200 000 cédulas posibles, **un SHA-256 a secas se invierte por completo en una fracción de segundo** (el atacante calcula los 200 000 *hashes* y busca): no seudonimiza. El **HMAC con clave** mantiene el enlace entre tablas (mismo `JOIN`) y el atacante, sin la clave, no recupera **ninguna** cédula. La **tokenización** da reversibilidad controlada (solo quien accede a la bóveda) y el **enmascaramiento** es solo para **mostrar**. Ninguna es magia: **la clave y la bóveda se convierten en el activo a proteger**, y la tabla seudonimizada **sigue** siendo dato personal.

## 4. Control de Acceso: RBAC y ABAC 🧩

Decidir **quién puede hacer qué** con qué datos. Dos modelos que se implementan a continuación **desde cero** (didáctico; en producción se delegan a un proveedor de identidad y a un motor de políticas):

| | **RBAC** (basado en roles) | **ABAC** (basado en atributos) |
|---|---|---|
| **Regla** | El usuario tiene **roles**; los roles tienen **permisos** | Se evalúan **atributos** del sujeto, del recurso, de la acción y del entorno (hora, red) |
| **Ejemplo** | "Los *analistas* pueden leer `ventas`" | "Puede leer un recurso si es de **su** departamento, no es confidencial y está en la red interna" |
| **Ventaja** | Simple, auditable, fácil de explicar | **Fino y contextual**; evita la explosión de roles |
| **Riesgo** | **Explosión de roles** (un rol por cada combinación) y permisos acumulados | Reglas más difíciles de razonar y probar; hay que **probarlas** |

Principios comunes: **privilegio mínimo**, **denegar por defecto** (*default deny*) y que **una denegación explícita gane** a cualquier permiso (*deny overrides*). En arquitecturas de **cero confianza** (*Zero Trust*; ver el cap. 5 del libro de Cao) cada solicitud se evalúa con esos atributos y **no se confía** en la ubicación de red por sí sola.

````python
# ============================================================
# 4. RBAC (con herencia de roles) y ABAC (default deny, deny overrides), desde cero y con asserts. DIDACTICO.
# ============================================================
# ---------- RBAC
ROLES_PERMISOS = {"lector": {("ventas", "read")}, "analista": {("clientes", "read"), ("ventas", "export")}, "admin_datos": {("clientes", "export"), ("nomina", "read")}}
HEREDA = {"analista": ["lector"], "admin_datos": ["analista"]}              # analista incluye lector; admin_datos incluye analista
USUARIO_ROLES = {"ana": ["lector"], "beto": ["analista"], "carla": ["admin_datos"], "dario": []}


def permisos_de_rol(rol):
    p = set(ROLES_PERMISOS.get(rol, set()))
    for padre in HEREDA.get(rol, []):
        p |= permisos_de_rol(padre)
    return p


def rbac_permitido(usuario, recurso, accion):
    return any((recurso, accion) in permisos_de_rol(r) for r in USUARIO_ROLES.get(usuario, []))


casos_rbac = [("ana", "ventas", "read", True), ("ana", "clientes", "read", False), ("beto", "ventas", "read", True), ("beto", "ventas", "export", True),
              ("beto", "nomina", "read", False), ("carla", "ventas", "read", True), ("carla", "nomina", "read", True), ("dario", "ventas", "read", False), ("nadie", "ventas", "read", False)]
for u, r, a, esperado in casos_rbac:
    assert rbac_permitido(u, r, a) == esperado, (u, r, a)
print(f"RBAC: {len(casos_rbac)} casos verificados (incluye herencia y usuario sin roles / inexistente -> denegado)")

# ---------- ABAC: una politica es (nombre, efecto, condicion(sujeto, recurso, accion, entorno))
POLITICAS = [
    ("P1 propio departamento, no confidencial", "permitir",
     lambda s, r, a, e: a == "read" and s["depto"] == r["depto"] and r["clasificacion"] in ("publico", "interno")),
    ("P2 confidencial: habilitacion >= 2 y red interna", "permitir",
     lambda s, r, a, e: r["clasificacion"] == "confidencial" and s["habilitacion"] >= 2 and e["red"] == "interna"),
    ("P3 contratistas: solo lo publico", "denegar",
     lambda s, r, a, e: s.get("contratista", False) and r["clasificacion"] != "publico"),
]


def abac_decidir(politicas, sujeto, recurso, accion, entorno):
    """Denegar por defecto; si alguna regla 'denegar' aplica, gana; si no, basta una 'permitir'. Devuelve (decision, regla)."""
    permitidas = []
    for nombre, efecto, cond in politicas:
        if cond(sujeto, recurso, accion, entorno):
            if efecto == "denegar":
                return "denegar", nombre
            permitidas.append(nombre)
    return ("permitir", permitidas[0]) if permitidas else ("denegar", "default deny")


ana = {"depto": "ventas", "habilitacion": 1}; gerente = {"depto": "ventas", "habilitacion": 3}; externo = {"depto": "ventas", "habilitacion": 3, "contratista": True}
rv_interno = {"depto": "ventas", "clasificacion": "interno"}; rv_conf = {"depto": "ventas", "clasificacion": "confidencial"}; rr_publico = {"depto": "rrhh", "clasificacion": "publico"}
rr_interno = {"depto": "rrhh", "clasificacion": "interno"}; oficina, casa = {"red": "interna", "hora": 10}, {"red": "externa", "hora": 22}
casos_abac = [(ana, rv_interno, "read", oficina, "permitir"), (ana, rr_interno, "read", oficina, "denegar"),       # otro departamento
              (ana, rv_conf, "read", oficina, "denegar"), (gerente, rv_conf, "read", oficina, "permitir"),         # habilitacion
              (gerente, rv_conf, "read", casa, "denegar"),                                                         # red externa
              (externo, rv_interno, "read", oficina, "denegar"), (externo, rr_publico, "read", oficina, "denegar"),   # P3 gana... o no hay permiso
              (ana, rv_interno, "export", oficina, "denegar")]                                                     # ninguna regla permite exportar
for s, r, a, e, esperado in casos_abac:
    assert abac_decidir(POLITICAS, s, r, a, e)[0] == esperado
assert abac_decidir(POLITICAS, externo, rv_interno, "read", oficina) == ("denegar", "P3 contratistas: solo lo publico")      # deny overrides
print(f"ABAC: {len(casos_abac)} casos verificados | ejemplo -> {abac_decidir(POLITICAS, gerente, rv_conf, 'read', oficina)}")

# ---------- La "explosion de roles": cuantos roles RBAC hacen falta para expresar P1 (departamento x clasificacion) que ABAC expresa con UNA regla
DEPTOS, CLASES = ["ventas", "rrhh", "finanzas", "logistica", "ti", "legal"], ["publico", "interno"]
roles_rbac = [f"lector_{d}_{c}" for d in DEPTOS for c in CLASES]
print(f"Para expresar P1 con RBAC: {len(roles_rbac)} roles (1 por departamento x clasificación) y crece con cada atributo nuevo | con ABAC: 1 regla")
assert len(roles_rbac) == 12
````

## 5. Auditoría y Monitoreo: una Bitácora Inviolable 🧾

La auditoría responde *quién hizo qué, cuándo y con qué resultado*, y sirve para **detectar** abusos y **probarlos** después (la **R** de STRIDE: no repudio). Una bitácora que el administrador puede **editar sin dejar rastro** vale poco. Una **bitácora inviolable** (*tamper-evident*: "a prueba de manipulación **evidente**", que **no** significa que no se pueda tocar sino que **se nota**) se logra encadenando *hashes*: cada entrada guarda el *hash* de la anterior, de modo que **alterar, borrar o insertar** una entrada **rompe la cadena en ese punto**: la entrada alterada ya no cuadra con su *hash* (o la siguiente deja de apuntar a la anterior), y para **ocultarlo** el atacante tendría que **recalcular todos los *hashes* posteriores** (es la misma idea de las cadenas de bloques que veremos en el Módulo 05).

$$h_i = \text{SHA-256}\left(h_{i-1} \,\Vert\, \text{entrada}_i\right), \qquad h_0 = \text{génesis}$$

**Límites honestos** (los medimos abajo): (1) un atacante que **recalcula todos los *hashes* posteriores** produce una cadena **internamente válida**; la defensa es **anclar** el último *hash* fuera del alcance del atacante (otro sistema, almacenamiento de solo-escritura, un tercero) o **firmarlo**; (2) **truncar** el final de la bitácora tampoco se detecta sin ese ancla. Además se **monitorea** la bitácora con reglas de alerta (por ejemplo, muchas denegaciones seguidas del mismo usuario).

````python
# ============================================================
# 5. Bitacora tamper-evident con cadena de hashes (SHA-256): construir, verificar, alterar, ancla y alertas. DIDACTICO.
# ============================================================
from matplotlib.patches import FancyBboxPatch

GENESIS = "0" * 64


def hash_entrada(prev, entrada):
    return hashlib.sha256((prev + json.dumps(entrada, sort_keys=True, ensure_ascii=False)).encode()).hexdigest()


def construir_log(entradas):
    log, prev = [], GENESIS
    for e in entradas:
        h = hash_entrada(prev, e)
        log.append({"entrada": e, "prev": prev, "hash": h}); prev = h
    return log


def verificar(log, ancla=None):
    """Devuelve None si la cadena es valida; si no, el indice de la primera entrada inconsistente ('cola' si falla el ancla)."""
    prev = GENESIS
    for i, x in enumerate(log):
        if x["prev"] != prev or x["hash"] != hash_entrada(prev, x["entrada"]):
            return i
        prev = x["hash"]
    if ancla is not None and prev != ancla:
        return "cola"
    return None


def mi_log(n=20):
    r = np.random.default_rng(SEED)
    usuarios = ["ana", "beto", "carla", "dario"]
    return [{"seq": i, "usuario": str(r.choice(usuarios)), "accion": str(r.choice(["read", "export", "login"])), "recurso": str(r.choice(["ventas", "clientes", "nomina"])),
             "resultado": str(r.choice(["ok", "ok", "ok", "denegado"]))} for i in range(n)]


log = construir_log(mi_log())
ancla = log[-1]["hash"]                                       # el "ancla": se publica/guarda FUERA del alcance de quien administra la bitacora
assert verificar(log, ancla) is None

import copy
# (1) alterar el contenido de la entrada 7
l1 = copy.deepcopy(log); l1[7]["entrada"]["resultado"] = "ok" if l1[7]["entrada"]["resultado"] == "denegado" else "denegado"
# (2) borrar la entrada 5
l2 = copy.deepcopy(log); del l2[5]
# (3) el atacante ALTERA la entrada 7 y RECALCULA todos los hashes siguientes (cadena internamente valida)
l3 = copy.deepcopy(log); l3[7]["entrada"]["accion"] = "borrado_de_evidencia"
prev = l3[7]["prev"]
for x in l3[7:]:
    x["prev"] = prev; x["hash"] = hash_entrada(prev, x["entrada"]); prev = x["hash"]
# (4) truncar las ultimas 3 entradas
l4 = copy.deepcopy(log)[:-3]
res = {"alterar entrada 7": (verificar(l1), verificar(l1, ancla)), "borrar entrada 5": (verificar(l2), verificar(l2, ancla)),
       "alterar y recalcular (sin ancla / con ancla)": (verificar(l3), verificar(l3, ancla)), "truncar el final (sin ancla / con ancla)": (verificar(l4), verificar(l4, ancla))}
print(pd.DataFrame(res, index=["verificación sin ancla", "verificación con ancla"]).T.to_string())
assert verificar(l1) == 7 and verificar(l2) == 5                    # alterar y borrar se detectan solo con la cadena
assert verificar(l3) is None and verificar(l3, ancla) == "cola"      # recalcular engana a la cadena, no al ancla
assert verificar(l4) is None and verificar(l4, ancla) == "cola"      # truncar: solo el ancla lo detecta


# ---- alertas de monitoreo sobre la bitacora: >= 3 denegaciones del mismo usuario en una ventana de 8 entradas
def alertas(log, ventana=8, umbral=3):
    out = set()
    ent = [x["entrada"] for x in log]
    for i in range(len(ent)):
        w = ent[max(0, i - ventana + 1): i + 1]
        for u in {e["usuario"] for e in w}:
            if sum(e["usuario"] == u and e["resultado"] == "denegado" for e in w) >= umbral:
                out.add(u)
    return sorted(out)


sospechoso = construir_log(mi_log() + [{"seq": 20 + i, "usuario": "dario", "accion": "read", "recurso": "nomina", "resultado": "denegado"} for i in range(4)])
print("Alertas en la bitácora normal:", alertas(log), "| tras 4 denegaciones seguidas de dario:", alertas(sospechoso))
assert "dario" in alertas(sospechoso) and verificar(sospechoso) is None

# ---- diagrama de la cadena: la entrada 7 alterada deja de cuadrar
fig, ax = plt.subplots(figsize=(11, 2.8)); ax.set_xlim(0, 11); ax.set_ylim(0, 3); ax.axis("off")
for k in range(5):
    x0 = 0.2 + k * 2.15; fc, ec = ("#fee2e2", "#b91c1c") if k == 2 else (("#fef9c3", "#a16207") if k > 2 else ("#dcfce7", "#15803d"))
    ax.add_patch(FancyBboxPatch((x0, 0.9), 1.7, 1.2, boxstyle="round,pad=0.03,rounding_size=0.1", fc=fc, ec=ec, lw=1.6))
    ax.text(x0 + 0.85, 1.75, f"entrada {5 + k}", ha="center", fontsize=9, fontweight="bold")
    ax.text(x0 + 0.85, 1.3, "hash = H(prev ‖ dato)", ha="center", fontsize=7.5)
    if k < 4:
        ax.annotate("", xy=(x0 + 2.15, 1.5), xytext=(x0 + 1.7, 1.5), arrowprops=dict(arrowstyle="-|>", lw=1.4, color="#334155"))
ax.text(5.5, 0.35, "Se altera la entrada 7 (rojo): ya no cuadra con su hash; para ocultarlo habría que recalcular también las siguientes (amarillo)", ha="center", fontsize=9, color="#b91c1c")
ax.set_title("Bitácora encadenada: alterar una entrada se nota", fontsize=11, fontweight="bold"); plt.tight_layout(); plt.show()
````

## 6. Privacidad Diferencial (Introducción) 🎲

Enmascarar y seudonimizar protegen **filas**; pero las **estadísticas** también filtran: si publicas "hay 41 personas con diabetes en Soatá" y después "hay 40 si se excluye a Juan", sabes algo de Juan. La **privacidad diferencial** (*differential privacy*, Dwork *et al.*) da una **garantía matemática**: el resultado publicado es **casi igual** haya o no una persona cualquiera en los datos. Un mecanismo $\mathcal{M}$ es **$\varepsilon$-diferencialmente privado** si, para cualesquiera dos bases de datos $D$ y $D'$ que difieren en **una** persona y cualquier conjunto de salidas $S$:

$$\Pr[\mathcal{M}(D) \in S] \;\leq\; e^{\varepsilon}\,\Pr[\mathcal{M}(D') \in S]$$

El parámetro $\varepsilon$ (**presupuesto de privacidad**) mide **cuánta información puede filtrarse**: **menor $\varepsilon$ = más privacidad** (más ruido). El **mecanismo de Laplace** para un **conteo** (cambia en a lo más 1 al agregar o quitar una persona: **sensibilidad** $\Delta = 1$) publica $\tilde{c} = c + \text{Laplace}(0, \Delta/\varepsilon)$. El error absoluto esperado del ruido de Laplace de escala $b$ es exactamente $b$, así que

$$\mathbb{E}\,|\tilde{c} - c| \;=\; \frac{\Delta}{\varepsilon} \;=\; \frac{1}{\varepsilon} \quad \text{(para un conteo)}.$$

Dos hechos prácticos: (1) cada consulta **gasta** presupuesto y por **composición** secuencial los $\varepsilon$ **se suman**, de modo que preguntar muchas veces lo mismo permite **promediar el ruido** y recuperar la verdad (hay que **llevar la cuenta**); (2) el **posprocesamiento** (redondear, recortar a $\geq 0$) **no** gasta presupuesto. A diferencia del $k$-anonimato, la garantía **no depende** de qué más sepa el atacante. *(Aviso: este es un modelo introductorio con `numpy`; los generadores de ruido con números reales tienen sutilezas numéricas y en producción se usan bibliotecas especializadas como OpenDP o `diffprivlib`, **verificar**.)*

````python
# ============================================================
# 6. Mecanismo de Laplace sobre un conteo: error vs epsilon, la garantia e^eps y el costo de repetir la pregunta
# ============================================================
rng = np.random.default_rng(SEED)
N = 5000
persona_diabetes = rng.random(N) < 0.12                          # base sintetica: 12 % con "diabetes" (dato ficticio)
c_real = int(persona_diabetes.sum())
print(f"Conteo real (sintético): {c_real} de {N:,}")


def conteo_dp(c, eps, tam=None, rng=rng):
    """Mecanismo de Laplace para un conteo (sensibilidad 1)."""
    return c + rng.laplace(0, 1 / eps, tam)


# (1) el error medio observado es 1/eps
EPS = np.array([0.01, 0.05, 0.1, 0.5, 1.0, 2.0, 5.0])
error_emp = np.array([np.abs(conteo_dp(c_real, e, 20_000) - c_real).mean() for e in EPS])
tabla = pd.DataFrame({"epsilon": EPS, "error absoluto medio (simulado)": error_emp.round(2), "1/epsilon (teoría)": (1 / EPS).round(2)})
print(tabla.to_string(index=False))
assert np.allclose(error_emp, 1 / EPS, rtol=0.05)

# (2) la garantia: para vecinos con conteos c y c+1, la razon de densidades nunca supera e^eps
def pdf_laplace(x, mu, b):
    return np.exp(-np.abs(x - mu) / b) / (2 * b)
x = np.linspace(c_real - 200, c_real + 200, 4001)
for e in (0.1, 1.0):
    razon = pdf_laplace(x, c_real, 1 / e) / pdf_laplace(x, c_real + 1, 1 / e)
    assert razon.max() <= np.exp(e) + 1e-9 and razon.min() >= np.exp(-e) - 1e-9
print("Razón de densidades entre vecinos acotada por e^ε (verificada en una malla para ε = 0.1 y 1)")

# (3) el costo de repetir: promediar muchas respuestas con ruido independiente recupera la verdad
eps_consulta = 0.1
una_vez = np.abs(conteo_dp(c_real, eps_consulta, 5000) - c_real).mean()
promedio_200 = np.abs(np.array([conteo_dp(c_real, eps_consulta, 200).mean() for _ in range(2000)]) - c_real).mean()
print(f"ε = {eps_consulta} por consulta: error medio con 1 consulta ≈ {una_vez:.1f} | promediando 200 consultas ≈ {promedio_200:.2f} (gasto total ε = {200 * eps_consulta:.0f}, la privacidad prácticamente se pierde)")
assert promedio_200 < una_vez / 5

# (4) un conteo publicado con posprocesamiento (no negativo, entero)
pub = max(0, round(float(conteo_dp(c_real, 1.0))))
print("Conteo publicado con ε = 1 (redondeado y no negativo):", pub, "| real:", c_real)

fig, ax = plt.subplots(figsize=(7, 4))
ax.loglog(EPS, 1 / EPS, "-", color="#2563eb", label="teoría: 1/ε"); ax.loglog(EPS, error_emp, "o", color="#dc2626", label="simulación (20 000 repeticiones)")
ax.set_xlabel("ε (menor = más privacidad)"); ax.set_ylabel("error absoluto medio del conteo"); ax.set_title("Más privacidad cuesta más error", fontweight="bold"); ax.legend(); plt.tight_layout(); plt.show()
````

---
**Lectura.** El error del conteo sigue la curva $1/\varepsilon$: con $\varepsilon = 0.1$ el conteo se desvía unas **10 personas** en promedio; con $\varepsilon = 1$, **una**; con $\varepsilon = 0.01$, unas **100** (en una base de 5 000, ya casi inútil). Para conteos **grandes** (miles) el ruido es pequeño en proporción, por eso la privacidad diferencial funciona bien con **grandes volúmenes** y mal con grupos pequeños. Y repetir la pregunta **gasta** presupuesto: promediando 200 respuestas con $\varepsilon = 0.1$, el error baja de unas 10 personas a menos de 1, pero el gasto total ($\varepsilon = 20$) significa que casi no hay privacidad. Elegir $\varepsilon$ es una **decisión de gobernanza** (compromiso privacidad-utilidad), no solo técnica; no hay un valor "correcto" universal.

## Resumen y Puntos Clave 🎯

| Concepto | Mensaje Central |
|---|---|
| **Cifrado simétrico y asimétrico** | Fernet (AES + HMAC) es **rápido y autenticado**; RSA-OAEP solo cifra mensajes cortos y es mucho más lento; RSA-PSS **firma**. Siempre con bibliotecas auditadas |
| **Cifrado por sobre** | DEK por dato, KEK en un KMS: rotar la KEK reescribe solo las DEK; destruir una DEK da *crypto-shredding* |
| **Contraseñas** | MD5/SHA-1 sin sal son rápidos y **buscables**; la sal evita tablas precalculadas pero no el diccionario con *hash* rápido; hace falta una **función lenta** (`pbkdf2_hmac`, `secrets`, `compare_digest`) y política de claves |
| **Enmascarar, seudonimizar, tokenizar** | Enmascarar = mostrar; **HMAC con clave** = analizar enlazando; **bóveda** = poder revertir. Un *hash* a secas de un identificador pequeño se invierte por enumeración |
| **RBAC y ABAC** | RBAC es simple pero sufre explosión de roles; ABAC es contextual. **Denegar por defecto**, que **denegar gane** y **probar** las políticas |
| **Bitácora inviolable** | Cadena de *hashes*: alterar o borrar se **nota**; recalcular o truncar **solo** lo detecta un **ancla** externa |
| **Privacidad diferencial** | $\tilde{c} = c + \text{Laplace}(1/\varepsilon)$; error esperado $1/\varepsilon$; los $\varepsilon$ **se suman** al repetir consultas |
